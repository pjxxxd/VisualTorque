```python
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell', dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True)

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 14 tracked cells.')
```

```python
%%cell_progress 6/14 Load the frozen discovery examples
cell_stage('Loading activation discovery manifest')
import hashlib
import json
from collections import Counter

import pandas as pd
from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

activation_manifest = read_jsonl(ACTIVATION_MANIFEST)
manifest_scene_ids = {str(record['scene_id']) for record in activation_manifest}
assert len(manifest_scene_ids) == 254, len(manifest_scene_ids)
assert {str(record['split']) for record in activation_manifest} <= {'probe_train', 'probe_dev'}
assert all(str(record['split']) != 'probe_test' for record in activation_manifest)
manifest_by_scene = {str(record['scene_id']): record for record in activation_manifest}

cell_stage('Reconstructing the exact image-rule outcome prompts')
example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=(SCRATCH_ROOT / 'unused_activation_example_output'),
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev'),
    batch_size=BATCH_SIZE,
    generation_every=0,
    seed=SEED,
)
discovery_examples = tuple(
    example for example in load_behavior_examples(example_config)
    if example.scene_id in manifest_scene_ids
)
assert len(discovery_examples) == 254 * 6, len(discovery_examples)
counts = Counter(example.scene_id for example in discovery_examples)
assert set(counts.values()) == {6}, Counter(counts.values())
assert {example.split for example in discovery_examples} <= {'probe_train', 'probe_dev'}
print('Discovery scenes:', len(counts))
print('Discovery examples:', len(discovery_examples))
print('Scene splits:', Counter(record['split'] for record in activation_manifest))
```

```python
%%cell_progress 7/14 Load model and locate decoder layers
cell_stage('Loading processor and model weights')
adapter = HuggingFaceMultimodalAdapter(
    MODEL_ID,
    revision=MODEL_REVISION,
    dtype=MODEL_DTYPE,
    device_map=DEVICE_MAP,
)
display(adapter.metadata)

cell_stage('Resolving language-decoder block list')
def resolve_module_path(root, path):
    value = root
    for component in path.split('.'):
        value = getattr(value, component)
    return value

def discover_decoder_layers(model, override=None):
    candidates = [override] if override else [
        'model.language_model.layers',
        'model.text_model.layers',
        'language_model.model.layers',
        'language_model.layers',
        'model.model.layers',
        'model.layers',
    ]
    failures = []
    for path in candidates:
        if not path:
            continue
        try:
            layers = resolve_module_path(model, path)
        except (AttributeError, KeyError) as exc:
            failures.append(f'{path}: {exc}')
            continue
        if isinstance(layers, torch.nn.ModuleList) and len(layers) >= 8:
            return path, layers
        failures.append(f'{path}: not a decoder ModuleList (type={type(layers).__name__})')
    available = [
        name for name, module in model.named_modules()
        if isinstance(module, torch.nn.ModuleList) and len(module) >= 8
    ]
    raise RuntimeError(
        'Could not locate language decoder layers. Tried:\n'
        + '\n'.join(failures)
        + '\nModuleList candidates: '
        + ', '.join(available)
    )

LAYER_PATH, DECODER_LAYERS = discover_decoder_layers(adapter.model, LAYER_PATH_OVERRIDE)
layer_indices = sorted(set(
    int(round(value))
    for value in torch.linspace(0, len(DECODER_LAYERS) - 1, steps=LAYER_COUNT).tolist()
))
normalized_depths = [(index + 1) / len(DECODER_LAYERS) for index in layer_indices]
assert len(layer_indices) == LAYER_COUNT, layer_indices
print('Model class:', type(adapter.model).__name__)
print('Layer path:', LAYER_PATH)
print('Decoder layer count:', len(DECODER_LAYERS))
print('Selected indices:', layer_indices)
print('Normalized depths:', [round(value, 4) for value in normalized_depths])
```

## 3. Residual-stream capture


```python
%%cell_progress 8/14 Define safe activation capture utilities
cell_stage('Defining physical targets and hook extraction')
import os
from datetime import datetime, timezone

from visualtorque_checks import sha256_file

def physical_targets(manifest_record):
    left = [stack for stack in manifest_record['stacks'] if stack['side'] == 'left']
    right = [stack for stack in manifest_record['stacks'] if stack['side'] == 'right']
    derived = manifest_record['derived']
    return {
        'left_units': int(sum(int(stack['units']) for stack in left)),
        'right_units': int(sum(int(stack['units']) for stack in right)),
        'left_distance': int(left[0]['distance']) if len(left) == 1 else None,
        'right_distance': int(right[0]['distance']) if len(right) == 1 else None,
        'left_torque': int(derived['left_torque']),
        'right_torque': int(derived['right_torque']),
        'net_torque': int(derived['net_torque']),
        'outcome': str(derived['outcome']),
    }

targets_by_scene = {scene_id: physical_targets(record) for scene_id, record in manifest_by_scene.items()}

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        tensor = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        tensor = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        tensor = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported decoder-layer output type: {type(output).__name__}')
    if tensor.ndim != 3:
        raise ValueError(f'Expected [batch, sequence, hidden], got {tuple(tensor.shape)}')
    return tensor

def capture_activation_batch(examples):
    if not examples:
        raise ValueError('capture batch must not be empty')
    if any(example.labels != examples[0].labels for example in examples):
        raise ValueError('all examples in a batch must use the same labels')
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('Activation discovery requires single-token A/B/C scoring for this model.')
    fixed_prefix, token_by_label = plan
    labels = list(examples[0].labels)
    conversations = [example.messages() for example in examples]
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(conversations, add_generation_prompt=True)),
        fixed_prefix,
    )
    captured = {}
    handles = []
    for index in layer_indices:
        def hook(_module, _inputs, output, layer_index=index):
            hidden = output_hidden_tensor(output)
            captured[layer_index] = hidden[:, -1, :].detach().to(
                device='cpu', dtype=torch.float16
            )
        handles.append(DECODER_LAYERS[index].register_forward_hook(hook))
    started = time.perf_counter()
    try:
        with torch.inference_mode():
            output = adapter.model(**inputs, use_cache=False)
            next_logprobs = torch.log_softmax(output.logits[:, -1, :].float(), dim=-1)
    finally:
        for handle in handles:
            handle.remove()
    missing = sorted(set(layer_indices) - set(captured))
    if missing:
        raise RuntimeError(f'Decoder hooks did not fire for layers: {missing}')
    activations = torch.stack([captured[index] for index in layer_indices], dim=1)
    if activations.shape[0] != len(examples):
        raise RuntimeError(f'Activation batch mismatch: {tuple(activations.shape)}')
    score_rows = torch.stack([
        next_logprobs[:, token_by_label[label]].detach().cpu() for label in labels
    ], dim=1)
    latency = (time.perf_counter() - started) / len(examples)
    records = []
    for row, example in enumerate(examples):
        scores = {label: float(score_rows[row, column]) for column, label in enumerate(labels)}
        predicted_label = max(labels, key=lambda label: scores[label])
        targets = targets_by_scene[example.scene_id]
        records.append({
            'schema_version': 'activation-discovery-1.0',
            'example_id': example.example_id,
            'question_id': example.question_id,
            'scene_id': example.scene_id,
            'split': example.split,
            'problem_type': example.problem_type,
            'variant': example.variant,
            'gold_semantic': str(example.gold_semantic),
            'gold_label': example.gold_label,
            'label_to_semantic': example.label_to_semantic,
            'predicted_label': predicted_label,
            'predicted_semantic': str(example.label_to_semantic[predicted_label]),
            'candidate_logprobs': scores,
            'latency_seconds': latency,
            **targets,
        })
    return {'activations': activations, 'records': records, 'labels': labels}

def torch_load_trusted(path):
    try:
        return torch.load(path, map_location='cpu', weights_only=False)
    except TypeError:
        return torch.load(path, map_location='cpu')

def capture_identity():
    return {
        'schema_version': 'activation-discovery-1.0',
        'model_id': MODEL_ID,
        'requested_revision': MODEL_REVISION,
        'resolved_revision': adapter.metadata.get('resolved_revision'),
        'dtype': MODEL_DTYPE,
        'dataset_fingerprint': example_config.dataset_fingerprint(),
        'activation_manifest_sha256': sha256_file(ACTIVATION_MANIFEST),
        'example_count': len(discovery_examples),
        'scene_count': len(manifest_scene_ids),
        'allowed_splits': ['probe_dev', 'probe_train'],
        'condition': 'image_rule',
        'question_kind': 'outcome',
        'permutations_per_scene': 6,
        'layer_path': LAYER_PATH,
        'decoder_layer_count': len(DECODER_LAYERS),
        'layer_indices': layer_indices,
        'normalized_depths': normalized_depths,
        'capture_position': 'final_token_after_generation_prompt_and_fixed_prefix',
        'activation_dtype': 'float16',
        'seed': SEED,
    }

def json_fallback(value):
    if isinstance(value, Path):
        return str(value)
    if hasattr(value, 'item'):
        return value.item()
    raise TypeError(f'Object of type {type(value).__name__} is not JSON serializable')

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False, default=json_fallback) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def run_resumable_capture():
    identity = capture_identity()
    config_path = CAPTURE_DIR / 'capture_config.json'
    if config_path.is_file():
        previous = json.loads(config_path.read_text(encoding='utf-8'))
        if previous.get('identity') != identity:
            raise ValueError(
                'Existing activation directory belongs to a different run. '
                'Choose a new CAPTURE_DIR instead of mixing activations.'
            )
    else:
        write_json_atomic(config_path, {
            'identity': identity,
            'adapter_metadata': adapter.metadata,
            'created_at_utc': datetime.now(timezone.utc).isoformat(),
        })

    shard_paths = sorted(CAPTURE_DIR.glob('shard_*.pt'))
    completed = set()
    for path in tqdm(shard_paths, desc='Reading completed shards', unit='shard'):
        payload = torch_load_trusted(path)
        for record in payload['records']:
            example_id = str(record['example_id'])
            if example_id in completed:
                raise RuntimeError(f'Duplicate captured example: {example_id}')
            completed.add(example_id)
    remaining = [example for example in discovery_examples if example.example_id not in completed]
    expected_ids = {example.example_id for example in discovery_examples}
    if not completed <= expected_ids:
        raise RuntimeError('Existing shards contain examples outside the frozen manifest.')
    print('Already captured:', len(completed), '| remaining:', len(remaining))

    next_shard = 0
    if shard_paths:
        next_shard = max(int(path.stem.split('_')[-1]) for path in shard_paths) + 1
    pending_activations = []
    pending_records = []

    def flush():
        nonlocal next_shard, pending_activations, pending_records
        if not pending_records:
            return
        payload = {
            'schema_version': 'activation-discovery-shard-1.0',
            'layer_indices': layer_indices,
            'normalized_depths': normalized_depths,
            'activations': torch.cat(pending_activations, dim=0),
            'records': pending_records,
        }
        destination = CAPTURE_DIR / f'shard_{next_shard:05d}.pt'
        temporary = CAPTURE_DIR / f'shard_{next_shard:05d}.pt.tmp'
        torch.save(payload, temporary)
        temporary.replace(destination)
        next_shard += 1
        pending_activations = []
        pending_records = []

    progress = tqdm(total=len(remaining), desc='Activation capture', unit='example', dynamic_ncols=True)
    for start in range(0, len(remaining), BATCH_SIZE):
        batch = remaining[start:start + BATCH_SIZE]
        captured = capture_activation_batch(batch)
        pending_activations.append(captured['activations'])
        pending_records.extend(captured['records'])
        progress.update(len(batch))
        progress.set_postfix_str(
            f"shard buffer {len(pending_records)}/{SHARD_EXAMPLES}", refresh=True
        )
        if len(pending_records) >= SHARD_EXAMPLES:
            flush()
    flush()
    progress.close()

    index_rows = []
    final_paths = sorted(CAPTURE_DIR.glob('shard_*.pt'))
    seen = set()
    for path in tqdm(final_paths, desc='Building capture index', unit='shard'):
        payload = torch_load_trusted(path)
        if list(payload['layer_indices']) != list(layer_indices):
            raise RuntimeError(f'Layer mismatch in {path.name}')
        for row, record in enumerate(payload['records']):
            if record['example_id'] in seen:
                raise RuntimeError(f"Duplicate example: {record['example_id']}")
            seen.add(record['example_id'])
            index_rows.append({'shard': path.name, 'row': row, **record})
    if seen != expected_ids:
        raise RuntimeError(f'Capture incomplete: expected {len(expected_ids)}, found {len(seen)}')
    index_path = CAPTURE_DIR / 'capture_index.jsonl'
    temporary_index = CAPTURE_DIR / 'capture_index.jsonl.tmp'
    with temporary_index.open('w', encoding='utf-8') as handle:
        for record in index_rows:
            handle.write(json.dumps(record, ensure_ascii=False, sort_keys=True) + '\n')
    temporary_index.replace(index_path)
    summary = {
        'expected_examples': len(expected_ids),
        'completed_examples': len(seen),
        'scene_count': len({record['scene_id'] for record in index_rows}),
        'shard_count': len(final_paths),
        'index_path': str(index_path),
    }
    write_json_atomic(CAPTURE_DIR / 'capture_summary.json', summary)
    return summary

print('Capture utilities ready.')
```

## 4. Train/dev discovery probes


```python
%%cell_progress 12/14 Fit discovery probes and invariance diagnostics
cell_stage('Importing probe utilities')
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.metrics import (
    balanced_accuracy_score, f1_score, mean_absolute_error, r2_score,
)
from sklearn.preprocessing import LabelEncoder, StandardScaler

RIDGE_ALPHAS = (0.01, 0.1, 1.0, 10.0, 100.0, 1000.0)
LOGISTIC_CS = (0.001, 0.01, 0.1, 1.0, 10.0)
CONTINUOUS_TARGETS = (
    'left_units', 'right_units', 'left_distance', 'right_distance',
    'left_torque', 'right_torque', 'net_torque',
)

def safe_pearson(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    if len(left) < 2 or not np.std(left) or not np.std(right):
        return None
    return float(np.corrcoef(left, right)[0, 1])

def safe_spearman(left, right):
    left = pd.Series(np.asarray(left, dtype=float)).rank(method='average')
    right = pd.Series(np.asarray(right, dtype=float)).rank(method='average')
    return safe_pearson(left.to_numpy(), right.to_numpy())

cell_stage('Fitting continuous scene-level probes')
continuous_rows = []
for target in tqdm(CONTINUOUS_TARGETS, desc='Continuous targets', unit='target'):
    values = pd.to_numeric(scene_metadata[target], errors='coerce').to_numpy(dtype=float)
    valid = np.isfinite(values)
    train_mask = valid & scene_metadata['split'].eq('probe_train').to_numpy()
    dev_mask = valid & scene_metadata['split'].eq('probe_dev').to_numpy()
    if train_mask.sum() < 10 or dev_mask.sum() < 3:
        raise RuntimeError(f'Insufficient train/dev examples for {target}: {train_mask.sum()}/{dev_mask.sum()}')
    y_train, y_dev = values[train_mask], values[dev_mask]
    layer_iterator = tqdm(
        list(enumerate(zip(layer_indices, normalized_depths))),
        desc=f'{target} layers', unit='layer', leave=False,
    )
    for layer_position, (layer_index, depth) in layer_iterator:
        scaler = StandardScaler()
        X_train = scaler.fit_transform(scene_activations[train_mask, layer_position, :])
        X_dev = scaler.transform(scene_activations[dev_mask, layer_position, :])
        candidates = []
        for alpha in RIDGE_ALPHAS:
            estimator = Ridge(alpha=alpha, solver='lsqr', max_iter=5000)
            estimator.fit(X_train, y_train)
            prediction = estimator.predict(X_dev)
            candidates.append((float(r2_score(y_dev, prediction)), alpha, prediction))
        dev_r2, best_alpha, prediction = max(candidates, key=lambda item: (item[0], -item[1]))
        rng = np.random.default_rng(SEED + 1000 * layer_position + sum(ord(c) for c in target))
        shuffled = rng.permutation(y_train)
        shuffled_estimator = Ridge(alpha=best_alpha, solver='lsqr', max_iter=5000)
        shuffled_estimator.fit(X_train, shuffled)
        shuffled_prediction = shuffled_estimator.predict(X_dev)
        baseline_prediction = np.full_like(y_dev, y_train.mean(), dtype=float)
        continuous_rows.append({
            'model_id': MODEL_ID,
            'target': target,
            'layer_index': layer_index,
            'normalized_depth': depth,
            'train_count': int(train_mask.sum()),
            'dev_count': int(dev_mask.sum()),
            'selected_alpha': best_alpha,
            'dev_r2': dev_r2,
            'dev_mae': float(mean_absolute_error(y_dev, prediction)),
            'dev_pearson': safe_pearson(y_dev, prediction),
            'dev_spearman': safe_spearman(y_dev, prediction),
            'mean_baseline_dev_r2': float(r2_score(y_dev, baseline_prediction)),
            'mean_baseline_dev_mae': float(mean_absolute_error(y_dev, baseline_prediction)),
            'shuffled_target_dev_r2': float(r2_score(y_dev, shuffled_prediction)),
        })
continuous_results = pd.DataFrame(continuous_rows)
continuous_results.to_csv(PROBE_DIR / 'continuous_probe_results.csv', index=False)

def fit_classification_curve(activations, metadata, target, *, representation):
    labels = metadata[target].astype(str).to_numpy()
    encoder = LabelEncoder().fit(labels)
    encoded = encoder.transform(labels)
    train_mask = metadata['split'].eq('probe_train').to_numpy()
    dev_mask = metadata['split'].eq('probe_dev').to_numpy()
    y_train, y_dev = encoded[train_mask], encoded[dev_mask]
    if len(np.unique(y_train)) < 2:
        raise RuntimeError(f'{target} has fewer than two train classes.')
    rows = []
    layer_iterator = tqdm(
        list(enumerate(zip(layer_indices, normalized_depths))),
        desc=f'{target} classification layers', unit='layer', leave=False,
    )
    for layer_position, (layer_index, depth) in layer_iterator:
        scaler = StandardScaler()
        X_train = scaler.fit_transform(activations[train_mask, layer_position, :])
        X_dev = scaler.transform(activations[dev_mask, layer_position, :])
        candidates = []
        for regularization in LOGISTIC_CS:
            estimator = LogisticRegression(
                C=regularization, solver='lbfgs', max_iter=2000, class_weight='balanced',
            )
            estimator.fit(X_train, y_train)
            prediction = estimator.predict(X_dev)
            score = float(balanced_accuracy_score(y_dev, prediction))
            candidates.append((score, regularization, prediction))
        balanced, best_c, prediction = max(candidates, key=lambda item: (item[0], -item[1]))
        majority = Counter(y_train).most_common(1)[0][0]
        baseline = np.full_like(y_dev, majority)
        rows.append({
            'model_id': MODEL_ID,
            'representation': representation,
            'target': target,
            'classes': '|'.join(str(value) for value in encoder.classes_),
            'layer_index': layer_index,
            'normalized_depth': depth,
            'train_count': int(train_mask.sum()),
            'dev_count': int(dev_mask.sum()),
            'selected_c': best_c,
            'dev_balanced_accuracy': balanced,
            'dev_macro_f1': float(f1_score(
                y_dev, prediction, labels=np.arange(len(encoder.classes_)),
                average='macro', zero_division=0,
            )),
            'majority_baseline_balanced_accuracy': float(balanced_accuracy_score(y_dev, baseline)),
        })
    return rows

cell_stage('Fitting semantic outcome and answer-label probes')
classification_rows = []
classification_rows.extend(fit_classification_curve(
    scene_activations, scene_metadata, 'outcome', representation='six_permutation_scene_mean'
))
classification_rows.extend(fit_classification_curve(
    variant_activations, variant_metadata, 'gold_label', representation='permutation_level'
))
classification_rows.extend(fit_classification_curve(
    variant_activations, variant_metadata, 'predicted_label', representation='permutation_level'
))
classification_results = pd.DataFrame(classification_rows)
classification_results.to_csv(PROBE_DIR / 'classification_probe_results.csv', index=False)

cell_stage('Measuring within-scene permutation dispersion')
dispersion_rows = []
variant_scene_ids = variant_metadata['scene_id'].astype(str).to_numpy()
for layer_position, (layer_index, depth) in enumerate(zip(layer_indices, normalized_depths)):
    scene_means = []
    within_energies = []
    for scene_id in scene_ids:
        group = variant_activations[variant_scene_ids == scene_id, layer_position, :]
        mean = group.mean(axis=0)
        scene_means.append(mean)
        within_energies.append(float(np.mean((group - mean) ** 2)))
    scene_means = np.stack(scene_means)
    global_mean = scene_means.mean(axis=0)
    between_energy = float(np.mean((scene_means - global_mean) ** 2))
    within_energy = float(np.mean(within_energies))
    dispersion_rows.append({
        'model_id': MODEL_ID,
        'layer_index': layer_index,
        'normalized_depth': depth,
        'within_scene_permutation_energy': within_energy,
        'between_scene_energy': between_energy,
        'within_to_between_ratio': within_energy / between_energy if between_energy else None,
    })
permutation_dispersion = pd.DataFrame(dispersion_rows)
permutation_dispersion.to_csv(PROBE_DIR / 'permutation_dispersion.csv', index=False)
print('Continuous probe rows:', len(continuous_results))
print('Classification probe rows:', len(classification_results))
display(continuous_results.sort_values(['target', 'dev_r2'], ascending=[True, False]).groupby('target').head(1).round(4))
display(classification_results.sort_values(['target', 'dev_balanced_accuracy'], ascending=[True, False]).groupby('target').head(1).round(4))
```
