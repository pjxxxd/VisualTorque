```python
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 12 tracked cells.')
```

```python
%%cell_progress 1/12 Locate local project
cell_stage('Locating local VisualTorque project')
from pathlib import Path
import os
import sys

def find_project_root():
    configured = os.environ.get('VISUALTORQUE_ROOT')
    candidates = []
    if configured:
        candidates.append(Path(configured).expanduser())
    for base in (Path.cwd(), *Path.cwd().parents):
        candidates.extend((base, base / 'VisualTorque'))
    checked = set()
    for candidate in candidates:
        candidate = candidate.resolve()
        if candidate in checked:
            continue
        checked.add(candidate)
        if (candidate / 'toolkit' / 'pyproject.toml').is_file():
            return candidate
    raise FileNotFoundError(
        'Could not locate toolkit/pyproject.toml. Start Jupyter inside the repository '
        'or set VISUALTORQUE_ROOT to the repository path.'
    )

PROJECT_ROOT = find_project_root()
TOOLKIT_ROOT = PROJECT_ROOT / 'toolkit'
DATA_ROOT = PROJECT_ROOT / 'data'
RESULTS_ROOT = PROJECT_ROOT / 'results'
SCRATCH_ROOT = PROJECT_ROOT / '.cache'
for directory in (DATA_ROOT, RESULTS_ROOT, SCRATCH_ROOT):
    directory.mkdir(parents=True, exist_ok=True)
print('Project:', PROJECT_ROOT)
```

## 2. Frozen probes and exact additive controls


```python
%%cell_progress 5/12 Define metrics, bootstrap, and scene features
cell_stage('Defining deterministic analysis utilities')
def torch_load_trusted(path):
    try:
        return torch.load(path, map_location='cpu', weights_only=False)
    except TypeError:
        return torch.load(path, map_location='cpu')

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False, default=json_default)
        + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def json_default(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, (np.integer, np.floating)):
        return value.item()
    if isinstance(value, np.ndarray):
        return value.tolist()
    raise TypeError(f'Cannot serialize {type(value).__name__}')

from visualtorque_checks import sha256_file

def safe_pearson(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    if len(left) < 2 or np.std(left) == 0 or np.std(right) == 0:
        return None
    return float(np.corrcoef(left, right)[0, 1])

def safe_spearman(left, right):
    left_rank = pd.Series(np.asarray(left)).rank(method='average').to_numpy()
    right_rank = pd.Series(np.asarray(right)).rank(method='average').to_numpy()
    return safe_pearson(left_rank, right_rank)

def regression_metrics(truth, prediction):
    truth = np.asarray(truth, dtype=float)
    prediction = np.asarray(prediction, dtype=float)
    return {
        'r2': float(r2_score(truth, prediction)),
        'mae': float(mean_absolute_error(truth, prediction)),
        'pearson': safe_pearson(truth, prediction),
        'spearman': safe_spearman(truth, prediction),
    }

def stratified_bootstrap_indices(labels, rng):
    labels = np.asarray(labels)
    parts = []
    for label in sorted(set(labels.tolist())):
        members = np.flatnonzero(labels == label)
        parts.append(rng.choice(members, size=len(members), replace=True))
    return np.concatenate(parts)

def percentile_interval(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return None, None
    low, high = np.quantile(values, [0.025, 0.975])
    return float(low), float(high)

def scene_feature_record(scene):
    left = [stack for stack in scene['stacks'] if stack['side'] == 'left']
    right = [stack for stack in scene['stacks'] if stack['side'] == 'right']
    return {
        'scene_id': str(scene['scene_id']),
        'split': str(scene['split']),
        'problem_type': str(scene['problem_type']),
        'outcome': str(scene['derived']['outcome']),
        'net_torque': float(scene['derived']['net_torque']),
        'left_units': float(sum(stack['units'] for stack in left)),
        'right_units': float(sum(stack['units'] for stack in right)),
        'left_distance_sum': float(sum(stack['distance'] for stack in left)),
        'right_distance_sum': float(sum(stack['distance'] for stack in right)),
        'left_stack_count': float(len(left)),
        'right_stack_count': float(len(right)),
        'left_units_multiset': ','.join(
            str(value) for value in sorted(stack['units'] for stack in left)
        ),
        'left_distance_multiset': ','.join(
            str(value) for value in sorted(stack['distance'] for stack in left)
        ),
        'right_units_multiset': ','.join(
            str(value) for value in sorted(stack['units'] for stack in right)
        ),
        'right_distance_multiset': ','.join(
            str(value) for value in sorted(stack['distance'] for stack in right)
        ),
        'left_torque': float(scene['derived']['left_torque']),
        'right_torque': float(scene['derived']['right_torque']),
    }

feature_table = pd.DataFrame([
    scene_feature_record(scene) for scene in scene_records
]).set_index('scene_id', drop=False)
assert feature_table.index.is_unique
print('Feature table:', feature_table.shape)
```

```python
%%cell_progress 6/12 Reconstruct and verify both frozen probes
cell_stage('Loading and aligning saved scene-mean activations')
model_data = {}
verification_rows = []
dataset_fingerprints = set()
for model_name, spec in MODEL_SPECS.items():
    paths = artifact_paths[model_name]
    selection = json.loads(
        paths['discovery_selection'].read_text(encoding='utf-8')
    )
    discovery_config = json.loads(
        paths['discovery_config'].read_text(encoding='utf-8')
    )
    confirmatory_config = json.loads(
        paths['confirmatory_config'].read_text(encoding='utf-8')
    )
    confirmatory_summary = json.loads(
        paths['confirmatory_summary'].read_text(encoding='utf-8')
    )
    choice = selection['candidate_net_torque_layer']
    layer_index = int(choice['layer_index'])
    alpha = float(choice['selected_alpha'])
    assert layer_index == spec['expected_layer']
    assert abs(alpha - spec['expected_alpha']) < 1e-12
    assert confirmatory_summary['test_used_for_selection'] is False
    assert confirmatory_summary['evaluation_split'] == 'probe_test'
    dataset_fingerprints.add(
        discovery_config['identity']['dataset_fingerprint']
    )
    dataset_fingerprints.add(
        confirmatory_config['identity']['dataset_fingerprint']
    )

    discovery = torch_load_trusted(paths['discovery_activations'])
    test = torch_load_trusted(paths['test_activations'])
    layer_indices = [int(value) for value in discovery['layer_indices']]
    layer_position = layer_indices.index(layer_index)
    X_fit = discovery['activations'][:, layer_position, :].float().numpy()
    fit_metadata = pd.DataFrame(discovery['metadata'])
    X_test = test['activations'].float().numpy()
    test_metadata = pd.DataFrame(test['metadata'])
    assert set(fit_metadata['split']) <= {'probe_train', 'probe_dev'}
    assert set(test_metadata['split']) == {'probe_test'}
    assert len(fit_metadata) == X_fit.shape[0]
    assert len(test_metadata) == X_test.shape[0]

    fit_order = np.argsort(fit_metadata['scene_id'].astype(str).to_numpy())
    test_order = np.argsort(test_metadata['scene_id'].astype(str).to_numpy())
    fit_metadata = fit_metadata.iloc[fit_order].reset_index(drop=True)
    test_metadata = test_metadata.iloc[test_order].reset_index(drop=True)
    X_fit = X_fit[fit_order]
    X_test = X_test[test_order]
    fit_ids = fit_metadata['scene_id'].astype(str).tolist()
    test_ids = test_metadata['scene_id'].astype(str).tolist()
    fit_features = feature_table.loc[fit_ids].reset_index(drop=True)
    test_features = feature_table.loc[test_ids].reset_index(drop=True)
    y_fit = fit_features['net_torque'].to_numpy(dtype=float)
    y_test = test_features['net_torque'].to_numpy(dtype=float)
    Z_fit = fit_features[ADDITIVE_COLUMNS].to_numpy(dtype=float)
    Z_test = test_features[ADDITIVE_COLUMNS].to_numpy(dtype=float)

    scaler = StandardScaler()
    frozen_probe = Ridge(alpha=alpha, solver='lsqr', max_iter=5000)
    frozen_probe.fit(scaler.fit_transform(X_fit), y_fit)
    primary_prediction = frozen_probe.predict(scaler.transform(X_test))
    recomputed = regression_metrics(y_test, primary_prediction)
    reported_r2 = float(confirmatory_summary['point_metrics']['test_r2'])
    assert abs(recomputed['r2'] - reported_r2) < 0.005, (
        model_name, recomputed['r2'], reported_r2
    )
    verification_rows.append({
        'model': model_name,
        'layer': layer_index,
        'alpha': alpha,
        'fit_scenes': len(y_fit),
        'test_scenes': len(y_test),
        'reported_r2': reported_r2,
        'reloaded_r2': recomputed['r2'],
    })
    model_data[model_name] = {
        'spec': spec,
        'paths': paths,
        'alpha': alpha,
        'layer_index': layer_index,
        'X_fit': X_fit,
        'X_test': X_test,
        'Z_fit': Z_fit,
        'Z_test': Z_test,
        'y_fit': y_fit,
        'y_test': y_test,
        'fit_metadata': fit_metadata,
        'test_metadata': test_metadata,
        'fit_features': fit_features,
        'test_features': test_features,
        'primary_prediction': primary_prediction,
        'confirmatory_summary': confirmatory_summary,
    }

assert len(dataset_fingerprints) == 1, dataset_fingerprints
assert (
    model_data['Qwen']['test_metadata']['scene_id'].astype(str).tolist()
    == model_data['Gemma']['test_metadata']['scene_id'].astype(str).tolist()
)
verification = pd.DataFrame(verification_rows)
display(verification.round(6))
print('Frozen-probe reconstruction passed for both models.')
```

## 3. Conditional information beyond additive primitives


```python
%%cell_progress 7/12 Run additive, activation, and stacked incremental tests
cell_stage('Defining cross-fitted prediction and paired bootstrap')
def cross_fitted_activation_predictions(X, y, alpha, seed):
    predictions = np.empty(len(y), dtype=float)
    splitter = KFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
    for train_indices, heldout_indices in splitter.split(X):
        scaler = StandardScaler()
        estimator = Ridge(alpha=alpha, solver='lsqr', max_iter=5000)
        estimator.fit(scaler.fit_transform(X[train_indices]), y[train_indices])
        predictions[heldout_indices] = estimator.predict(
            scaler.transform(X[heldout_indices])
        )
    return predictions

def bootstrap_prediction_difference(
    truth, first, second, labels, metric, seed
):
    rng = np.random.default_rng(seed)
    values = []
    for _ in range(BOOTSTRAP_REPLICATES):
        indices = stratified_bootstrap_indices(labels, rng)
        if metric == 'r2':
            first_value = r2_score(truth[indices], first[indices])
            second_value = r2_score(truth[indices], second[indices])
        elif metric == 'mae':
            first_value = mean_absolute_error(truth[indices], first[indices])
            second_value = mean_absolute_error(truth[indices], second[indices])
        else:
            raise ValueError(metric)
        values.append(float(second_value - first_value))
    return np.asarray(values)

incremental_rows = []
incremental_bootstrap_rows = []
for model_offset, (model_name, data) in enumerate(model_data.items()):
    cell_stage(f'Cross-fitting {model_name} activation predictions')
    activation_oof = cross_fitted_activation_predictions(
        data['X_fit'], data['y_fit'], data['alpha'], SEED + model_offset
    )
    additive_model = LinearRegression()
    additive_model.fit(data['Z_fit'], data['y_fit'])
    additive_prediction = additive_model.predict(data['Z_test'])

    stacker = LinearRegression()
    stacker.fit(
        np.column_stack([data['Z_fit'], activation_oof]),
        data['y_fit'],
    )
    stacked_prediction = stacker.predict(np.column_stack([
        data['Z_test'], data['primary_prediction']
    ]))
    data['activation_oof'] = activation_oof
    data['additive_prediction'] = additive_prediction
    data['stacked_prediction'] = stacked_prediction

    prediction_map = {
        'additive_primitives': additive_prediction,
        'activation_probe': data['primary_prediction'],
        'additive_plus_activation_oof_stacker': stacked_prediction,
    }
    for method, prediction in prediction_map.items():
        incremental_rows.append({
            'model': model_name,
            'method': method,
            **regression_metrics(data['y_test'], prediction),
        })

    labels = data['test_features']['outcome'].astype(str).to_numpy()
    delta_values = bootstrap_prediction_difference(
        data['y_test'], additive_prediction, stacked_prediction,
        labels, 'r2', SEED + 100 + model_offset,
    )
    low, high = percentile_interval(delta_values)
    point = (
        r2_score(data['y_test'], stacked_prediction)
        - r2_score(data['y_test'], additive_prediction)
    )
    incremental_bootstrap_rows.append({
        'model': model_name,
        'comparison': 'stacked_minus_additive',
        'delta_r2': float(point),
        'ci95_low': low,
        'ci95_high': high,
        'replicates': len(delta_values),
    })
    pd.DataFrame({
        'replicate': np.arange(len(delta_values)),
        'model': model_name,
        'delta_r2': delta_values,
    }).to_csv(
        OUTPUT_DIR / f'{model_name.lower()}_incremental_bootstrap.csv',
        index=False,
    )

incremental_metrics = pd.DataFrame(incremental_rows)
incremental_bootstrap = pd.DataFrame(incremental_bootstrap_rows)
incremental_metrics.to_csv(
    OUTPUT_DIR / 'incremental_value_metrics.csv', index=False
)
incremental_bootstrap.to_csv(
    OUTPUT_DIR / 'incremental_value_bootstrap_summary.csv', index=False
)
display(incremental_metrics.round(4))
display(incremental_bootstrap.round(4))
```

```python
%%cell_progress 8/12 Run cross-fitted partialling-out residual probes
cell_stage('Defining cross-fitted linear residualization')
def linear_fit_with_intercept(features, targets):
    design = np.column_stack([np.ones(len(features)), features])
    coefficients, *_ = np.linalg.lstsq(design, targets, rcond=None)
    return coefficients

def linear_predict_with_intercept(features, coefficients):
    design = np.column_stack([np.ones(len(features)), features])
    return design @ coefficients

def cross_fitted_partial_residuals(
    Z_fit, X_fit, y_fit, Z_test, X_test, y_test, seed
):
    residual_y_fit = np.empty_like(y_fit, dtype=float)
    residual_X_fit = np.empty_like(X_fit, dtype=np.float32)
    splitter = KFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
    for train_indices, heldout_indices in splitter.split(Z_fit):
        y_coefficients = linear_fit_with_intercept(
            Z_fit[train_indices], y_fit[train_indices]
        )
        X_coefficients = linear_fit_with_intercept(
            Z_fit[train_indices], X_fit[train_indices]
        )
        residual_y_fit[heldout_indices] = (
            y_fit[heldout_indices]
            - linear_predict_with_intercept(
                Z_fit[heldout_indices], y_coefficients
            )
        )
        residual_X_fit[heldout_indices] = (
            X_fit[heldout_indices]
            - linear_predict_with_intercept(
                Z_fit[heldout_indices], X_coefficients
            )
        )

    full_y_coefficients = linear_fit_with_intercept(Z_fit, y_fit)
    full_X_coefficients = linear_fit_with_intercept(Z_fit, X_fit)
    additive_test_prediction = linear_predict_with_intercept(
        Z_test, full_y_coefficients
    )
    residual_y_test = y_test - additive_test_prediction
    residual_X_test = (
        X_test
        - linear_predict_with_intercept(Z_test, full_X_coefficients)
    ).astype(np.float32)
    return {
        'residual_y_fit': residual_y_fit,
        'residual_X_fit': residual_X_fit,
        'residual_y_test': residual_y_test,
        'residual_X_test': residual_X_test,
        'additive_test_prediction': additive_test_prediction,
    }

residual_rows = []
residual_bootstrap_rows = []
for model_offset, (model_name, data) in enumerate(model_data.items()):
    cell_stage(f'Partialling out additive features for {model_name}')
    partial = cross_fitted_partial_residuals(
        data['Z_fit'], data['X_fit'], data['y_fit'],
        data['Z_test'], data['X_test'], data['y_test'],
        SEED + 300 + model_offset,
    )
    residual_scaler = StandardScaler()
    residual_probe = Ridge(
        alpha=data['alpha'], solver='lsqr', max_iter=5000
    )
    residual_probe.fit(
        residual_scaler.fit_transform(partial['residual_X_fit']),
        partial['residual_y_fit'],
    )
    residual_prediction = residual_probe.predict(
        residual_scaler.transform(partial['residual_X_test'])
    )
    corrected_prediction = (
        partial['additive_test_prediction'] + residual_prediction
    )
    data['partial'] = partial
    data['residual_prediction'] = residual_prediction
    data['residual_corrected_prediction'] = corrected_prediction
    residual_metrics = regression_metrics(
        partial['residual_y_test'], residual_prediction
    )
    baseline_r2 = r2_score(
        data['y_test'], partial['additive_test_prediction']
    )
    corrected_r2 = r2_score(data['y_test'], corrected_prediction)
    residual_rows.append({
        'model': model_name,
        'frozen_alpha': data['alpha'],
        'residual_r2': residual_metrics['r2'],
        'residual_mae': residual_metrics['mae'],
        'residual_pearson': residual_metrics['pearson'],
        'residual_spearman': residual_metrics['spearman'],
        'additive_r2': float(baseline_r2),
        'residual_corrected_r2': float(corrected_r2),
        'conditional_delta_r2': float(corrected_r2 - baseline_r2),
    })

    labels = data['test_features']['outcome'].astype(str).to_numpy()
    rng = np.random.default_rng(SEED + 500 + model_offset)
    delta_values = []
    residual_r2_values = []
    for _ in tqdm(
        range(BOOTSTRAP_REPLICATES),
        desc=f'{model_name} conditional bootstrap', unit='replicate',
    ):
        indices = stratified_bootstrap_indices(labels, rng)
        baseline_value = r2_score(
            data['y_test'][indices],
            partial['additive_test_prediction'][indices],
        )
        corrected_value = r2_score(
            data['y_test'][indices], corrected_prediction[indices]
        )
        residual_value = r2_score(
            partial['residual_y_test'][indices],
            residual_prediction[indices],
        )
        delta_values.append(float(corrected_value - baseline_value))
        residual_r2_values.append(float(residual_value))
    delta_low, delta_high = percentile_interval(delta_values)
    residual_low, residual_high = percentile_interval(residual_r2_values)
    residual_bootstrap_rows.append({
        'model': model_name,
        'conditional_delta_r2': float(corrected_r2 - baseline_r2),
        'delta_r2_ci95_low': delta_low,
        'delta_r2_ci95_high': delta_high,
        'residual_r2': residual_metrics['r2'],
        'residual_r2_ci95_low': residual_low,
        'residual_r2_ci95_high': residual_high,
        'replicates': BOOTSTRAP_REPLICATES,
    })
    pd.DataFrame({
        'replicate': np.arange(BOOTSTRAP_REPLICATES),
        'model': model_name,
        'conditional_delta_r2': delta_values,
        'residual_r2': residual_r2_values,
    }).to_csv(
        OUTPUT_DIR / f'{model_name.lower()}_conditional_bootstrap.csv',
        index=False,
    )

residual_metrics_table = pd.DataFrame(residual_rows)
residual_bootstrap_summary = pd.DataFrame(residual_bootstrap_rows)
residual_metrics_table.to_csv(
    OUTPUT_DIR / 'conditional_residual_metrics.csv', index=False
)
residual_bootstrap_summary.to_csv(
    OUTPUT_DIR / 'conditional_residual_bootstrap_summary.csv', index=False
)
display(residual_metrics_table.round(4))
display(residual_bootstrap_summary.round(4))
```

## 4. Exact additive matching and a strict binding subset


```python
%%cell_progress 9/12 Build exact matched groups and pairwise predictions
cell_stage('Finding exact additive-feature matches in probe_test')
reference = model_data['Qwen']
matched_frame = reference['test_features'].copy()
matched_frame['additive_key'] = matched_frame[ADDITIVE_COLUMNS].apply(
    lambda row: '|'.join(str(int(value)) for value in row), axis=1
)
group_statistics = matched_frame.groupby('additive_key').agg(
    scene_count=('scene_id', 'size'),
    torque_count=('net_torque', 'nunique'),
)
valid_keys = group_statistics[
    (group_statistics['scene_count'] >= 2)
    & (group_statistics['torque_count'] >= 2)
].index.tolist()
assert len(valid_keys) >= 20, (
    f'Unexpectedly few matched groups: {len(valid_keys)}'
)

matched_scene_rows = []
matched_pair_rows = []
prediction_by_model = {
    model_name: dict(zip(
        data['test_metadata']['scene_id'].astype(str),
        data['primary_prediction'],
    ))
    for model_name, data in model_data.items()
}
additive_by_model = {
    model_name: dict(zip(
        data['test_metadata']['scene_id'].astype(str),
        data['additive_prediction'],
    ))
    for model_name, data in model_data.items()
}
for group_number, key in enumerate(sorted(valid_keys)):
    group_id = f'matched_{group_number:03d}'
    group = matched_frame[matched_frame['additive_key'].eq(key)].copy()
    group = group.sort_values(['net_torque', 'scene_id']).reset_index(drop=True)
    for _, row in group.iterrows():
        matched_scene_rows.append({
            'matched_group_id': group_id,
            'additive_key': key,
            **row.to_dict(),
        })
    for left_index, right_index in itertools.combinations(range(len(group)), 2):
        left = group.iloc[left_index]
        right = group.iloc[right_index]
        if float(left['net_torque']) == float(right['net_torque']):
            continue
        low = left if left['net_torque'] < right['net_torque'] else right
        high = right if left['net_torque'] < right['net_torque'] else left
        common = {
            'matched_group_id': group_id,
            'additive_key': key,
            'low_scene_id': str(low['scene_id']),
            'high_scene_id': str(high['scene_id']),
            'low_torque': float(low['net_torque']),
            'high_torque': float(high['net_torque']),
            'true_delta': float(high['net_torque'] - low['net_torque']),
        }
        for model_name in MODEL_SPECS:
            low_id = str(low['scene_id'])
            high_id = str(high['scene_id'])
            predicted_delta = (
                prediction_by_model[model_name][high_id]
                - prediction_by_model[model_name][low_id]
            )
            additive_delta = (
                additive_by_model[model_name][high_id]
                - additive_by_model[model_name][low_id]
            )
            assert abs(additive_delta) < 1e-8, additive_delta
            matched_pair_rows.append({
                **common,
                'model': model_name,
                'predicted_delta': float(predicted_delta),
                'pair_order_correct': bool(predicted_delta > 0),
                'additive_predicted_delta': float(additive_delta),
            })

matched_scenes = pd.DataFrame(matched_scene_rows)
matched_pairs = pd.DataFrame(matched_pair_rows)
matched_scenes.to_csv(OUTPUT_DIR / 'additive_matched_scenes.csv', index=False)
matched_pairs.to_csv(OUTPUT_DIR / 'additive_matched_pairs.csv', index=False)

cell_stage('Building the strict primitive-multiset binding subset')
strict_frame = reference['test_features'].copy()
strict_frame['strict_binding_key'] = strict_frame[
    STRICT_BINDING_COLUMNS
].apply(lambda row: '||'.join(str(value) for value in row), axis=1)
strict_statistics = strict_frame.groupby('strict_binding_key').agg(
    scene_count=('scene_id', 'size'),
    torque_count=('net_torque', 'nunique'),
)
strict_keys = strict_statistics[
    (strict_statistics['scene_count'] >= 2)
    & (strict_statistics['torque_count'] >= 2)
].index.tolist()
assert len(strict_keys) >= 5, (
    f'Unexpectedly few strict binding groups: {len(strict_keys)}'
)
strict_pair_rows = []
for group_number, key in enumerate(sorted(strict_keys)):
    group_id = f'strict_binding_{group_number:03d}'
    group = strict_frame[
        strict_frame['strict_binding_key'].eq(key)
    ].sort_values(['net_torque', 'scene_id']).reset_index(drop=True)
    for left_index, right_index in itertools.combinations(range(len(group)), 2):
        left = group.iloc[left_index]
        right = group.iloc[right_index]
        if float(left['net_torque']) == float(right['net_torque']):
            continue
        low = left if left['net_torque'] < right['net_torque'] else right
        high = right if left['net_torque'] < right['net_torque'] else left
        low_id, high_id = str(low['scene_id']), str(high['scene_id'])
        for model_name in MODEL_SPECS:
            predicted_delta = (
                prediction_by_model[model_name][high_id]
                - prediction_by_model[model_name][low_id]
            )
            additive_delta = (
                additive_by_model[model_name][high_id]
                - additive_by_model[model_name][low_id]
            )
            assert abs(additive_delta) < 1e-8, additive_delta
            strict_pair_rows.append({
                'strict_binding_group_id': group_id,
                'strict_binding_key': key,
                'low_scene_id': low_id,
                'high_scene_id': high_id,
                'low_torque': float(low['net_torque']),
                'high_torque': float(high['net_torque']),
                'true_delta': float(high['net_torque'] - low['net_torque']),
                'model': model_name,
                'predicted_delta': float(predicted_delta),
                'pair_order_correct': bool(predicted_delta > 0),
                'additive_predicted_delta': float(additive_delta),
            })
strict_binding_pairs = pd.DataFrame(strict_pair_rows)
strict_binding_pairs.to_csv(
    OUTPUT_DIR / 'strict_binding_pairs.csv', index=False
)
coverage = pd.DataFrame([{
    'matched_groups': matched_scenes['matched_group_id'].nunique(),
    'matched_scenes': matched_scenes['scene_id'].nunique(),
    'non_tied_pairs': len(matched_pairs) // len(MODEL_SPECS),
    'strict_binding_groups': len(strict_keys),
    'strict_binding_pairs': len(strict_binding_pairs) // len(MODEL_SPECS),
    'problem_types': ','.join(sorted(matched_scenes['problem_type'].unique())),
}])
display(coverage)
print('Pair rows by model:', matched_pairs.groupby('model').size().to_dict())
```

```python
%%cell_progress 10/12 Run matched-group cluster bootstrap and permutation null
cell_stage('Computing group-balanced additive-matched metrics')
matched_summary_rows = []
matched_bootstrap_frames = []
matched_null_frames = []
for model_offset, model_name in enumerate(MODEL_SPECS):
    frame = matched_pairs[matched_pairs['model'].eq(model_name)].copy()
    group_accuracy = frame.groupby('matched_group_id')[
        'pair_order_correct'
    ].mean()
    point_group_balanced = float(group_accuracy.mean())
    point_pair_weighted = float(frame['pair_order_correct'].mean())
    point_delta_pearson = safe_pearson(
        frame['true_delta'], frame['predicted_delta']
    )

    rng = np.random.default_rng(SEED + 900 + model_offset)
    group_ids = group_accuracy.index.to_numpy()
    bootstrap_values = []
    for _ in tqdm(
        range(BOOTSTRAP_REPLICATES),
        desc=f'{model_name} matched cluster bootstrap', unit='replicate',
    ):
        sampled = rng.choice(group_ids, size=len(group_ids), replace=True)
        bootstrap_values.append(float(group_accuracy.loc[sampled].mean()))
    ci_low, ci_high = percentile_interval(bootstrap_values)
    matched_bootstrap_frames.append(pd.DataFrame({
        'replicate': np.arange(BOOTSTRAP_REPLICATES),
        'model': model_name,
        'group_balanced_pair_ordering_accuracy': bootstrap_values,
    }))

    cell_stage(f'Running within-group permutation null for {model_name}')
    scene_predictions = prediction_by_model[model_name]
    group_scene_ids = {
        group_id: group['scene_id'].astype(str).tolist()
        for group_id, group in matched_scenes.groupby('matched_group_id')
    }
    group_pair_truth = {
        group_id: group[['low_scene_id', 'high_scene_id']].copy()
        for group_id, group in frame.groupby('matched_group_id')
    }
    null_values = []
    for _ in tqdm(
        range(MATCHED_PERMUTATIONS),
        desc=f'{model_name} matched permutation', unit='permutation',
    ):
        accuracies = []
        for group_id, scene_ids in group_scene_ids.items():
            values = np.asarray([scene_predictions[s] for s in scene_ids])
            shuffled = rng.permutation(values)
            permuted = dict(zip(scene_ids, shuffled))
            pair_frame = group_pair_truth[group_id]
            correct = [
                permuted[high_id] > permuted[low_id]
                for low_id, high_id in zip(
                    pair_frame['low_scene_id'], pair_frame['high_scene_id']
                )
            ]
            accuracies.append(float(np.mean(correct)))
        null_values.append(float(np.mean(accuracies)))
    empirical_p = float(
        (1 + np.sum(np.asarray(null_values) >= point_group_balanced))
        / (MATCHED_PERMUTATIONS + 1)
    )
    matched_null_frames.append(pd.DataFrame({
        'permutation': np.arange(MATCHED_PERMUTATIONS),
        'model': model_name,
        'group_balanced_pair_ordering_accuracy': null_values,
    }))
    matched_summary_rows.append({
        'model': model_name,
        'matched_groups': len(group_ids),
        'matched_pairs': len(frame),
        'group_balanced_pair_ordering_accuracy': point_group_balanced,
        'cluster_bootstrap_ci95_low': ci_low,
        'cluster_bootstrap_ci95_high': ci_high,
        'pair_weighted_ordering_accuracy': point_pair_weighted,
        'pair_delta_pearson': point_delta_pearson,
        'permutation_p_one_sided': empirical_p,
        'permutations': MATCHED_PERMUTATIONS,
    })

matched_summary = pd.DataFrame(matched_summary_rows)
matched_bootstrap = pd.concat(matched_bootstrap_frames, ignore_index=True)
matched_null = pd.concat(matched_null_frames, ignore_index=True)
matched_summary.to_csv(
    OUTPUT_DIR / 'additive_matched_summary.csv', index=False
)
matched_bootstrap.to_csv(
    OUTPUT_DIR / 'additive_matched_cluster_bootstrap.csv', index=False
)
matched_null.to_csv(
    OUTPUT_DIR / 'additive_matched_permutation_null.csv', index=False
)
strict_summary_rows = []
for model_name in MODEL_SPECS:
    strict_frame = strict_binding_pairs[
        strict_binding_pairs['model'].eq(model_name)
    ]
    correct = int(strict_frame['pair_order_correct'].sum())
    pair_count = len(strict_frame)
    exact_binomial_p = float(
        sum(math.comb(pair_count, value) for value in range(correct, pair_count + 1))
        / (2 ** pair_count)
    )
    strict_summary_rows.append({
        'model': model_name,
        'strict_binding_groups': strict_frame[
            'strict_binding_group_id'
        ].nunique(),
        'strict_binding_pairs': pair_count,
        'correct_pairs': correct,
        'pair_ordering_accuracy': float(correct / pair_count),
        'one_sided_exact_binomial_p': exact_binomial_p,
        'analysis_role': 'descriptive_due_to_small_sample',
    })
strict_binding_summary = pd.DataFrame(strict_summary_rows)
strict_binding_summary.to_csv(
    OUTPUT_DIR / 'strict_binding_summary.csv', index=False
)
display(matched_summary.round(4))
display(strict_binding_summary.round(4))
```

```python
%%cell_progress 11/12 Run paired Qwen-Gemma comparison
cell_stage('Bootstrapping paired model differences on the same scenes')
qwen = model_data['Qwen']
gemma = model_data['Gemma']
assert np.array_equal(qwen['y_test'], gemma['y_test'])
truth = qwen['y_test']
labels = qwen['test_features']['outcome'].astype(str).to_numpy()
rng = np.random.default_rng(SEED + 1200)
paired_rows = []
for _ in tqdm(
    range(BOOTSTRAP_REPLICATES),
    desc='Paired Qwen-Gemma scene bootstrap', unit='replicate',
):
    indices = stratified_bootstrap_indices(labels, rng)
    qwen_r2 = r2_score(truth[indices], qwen['primary_prediction'][indices])
    gemma_r2 = r2_score(truth[indices], gemma['primary_prediction'][indices])
    qwen_mae = mean_absolute_error(
        truth[indices], qwen['primary_prediction'][indices]
    )
    gemma_mae = mean_absolute_error(
        truth[indices], gemma['primary_prediction'][indices]
    )
    paired_rows.append({
        'delta_r2_gemma_minus_qwen': float(gemma_r2 - qwen_r2),
        'mae_improvement_qwen_minus_gemma': float(qwen_mae - gemma_mae),
    })
paired_model_bootstrap = pd.DataFrame(paired_rows)
paired_model_bootstrap.insert(
    0, 'replicate', np.arange(BOOTSTRAP_REPLICATES)
)
paired_model_bootstrap.to_csv(
    OUTPUT_DIR / 'paired_model_bootstrap.csv', index=False
)
r2_low, r2_high = percentile_interval(
    paired_model_bootstrap['delta_r2_gemma_minus_qwen']
)
mae_low, mae_high = percentile_interval(
    paired_model_bootstrap['mae_improvement_qwen_minus_gemma']
)
paired_model_summary = pd.DataFrame([{
    'comparison': 'Gemma_minus_Qwen',
    'delta_r2': float(
        r2_score(truth, gemma['primary_prediction'])
        - r2_score(truth, qwen['primary_prediction'])
    ),
    'delta_r2_ci95_low': r2_low,
    'delta_r2_ci95_high': r2_high,
    'mae_improvement_qwen_minus_gemma': float(
        mean_absolute_error(truth, qwen['primary_prediction'])
        - mean_absolute_error(truth, gemma['primary_prediction'])
    ),
    'mae_improvement_ci95_low': mae_low,
    'mae_improvement_ci95_high': mae_high,
    'replicates': BOOTSTRAP_REPLICATES,
}])

qwen_group = matched_pairs[
    matched_pairs['model'].eq('Qwen')
].groupby('matched_group_id')['pair_order_correct'].mean()
gemma_group = matched_pairs[
    matched_pairs['model'].eq('Gemma')
].groupby('matched_group_id')['pair_order_correct'].mean()
assert qwen_group.index.equals(gemma_group.index)
matched_difference = gemma_group - qwen_group
matched_rng = np.random.default_rng(SEED + 1300)
matched_delta_values = []
for _ in range(BOOTSTRAP_REPLICATES):
    indices = matched_rng.integers(0, len(matched_difference), len(matched_difference))
    matched_delta_values.append(float(matched_difference.iloc[indices].mean()))
matched_low, matched_high = percentile_interval(matched_delta_values)
paired_matched_summary = pd.DataFrame([{
    'comparison': 'Gemma_minus_Qwen',
    'delta_group_balanced_matched_accuracy': float(matched_difference.mean()),
    'ci95_low': matched_low,
    'ci95_high': matched_high,
    'matched_groups': len(matched_difference),
}])
paired_model_summary.to_csv(
    OUTPUT_DIR / 'paired_model_summary.csv', index=False
)
paired_matched_summary.to_csv(
    OUTPUT_DIR / 'paired_matched_summary.csv', index=False
)
display(paired_model_summary.round(4))
display(paired_matched_summary.round(4))
```
