```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Freeze failed-update and correct-flip image pairs
cell_stage('Joining the model-blind geometry with frozen behavior predictions')
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

candidate_groups = read_jsonl(CANDIDATE_PATH)
screen_records = read_jsonl(SCREEN_RESULT_PATH)
screen_by_key = {row['request_key']: row for row in screen_records}
assert len(screen_by_key) == len(screen_records)

def mapping_variant_for(family, option):
    index = int(option['context_value']) + int(family['counterbalance_phase'])
    return ('forward', 'reverse')[index % 2]

def request_key(scene_id, mapping_variant):
    return f'{scene_id}|{mapping_variant}'

pair_candidates = {}
for family in candidate_groups:
    for option in family['context_options']:
        if option['transition'] != 'flip':
            continue
        variant = mapping_variant_for(family, option)
        target = option['target']
        hybrid = option['hybrid']
        target_result = screen_by_key[request_key(target['scene_id'], variant)]
        hybrid_result = screen_by_key[request_key(hybrid['scene_id'], variant)]
        if target_result['correct'] and not hybrid_result['correct']:
            cohort = 'failed_update_rescue'
            direction = 1.0
        elif target_result['correct'] and hybrid_result['correct']:
            cohort = 'correct_flip_induction'
            direction = -1.0
        else:
            continue
        pair_key = (
            str(target['scene_id']), str(hybrid['scene_id']), variant, cohort
        )
        row = {
            'pair_id': 'flip_' + stable_key(pair_key)[:20],
            'family_id': family['family_id'],
            'factor': family['factor'],
            'cohort': cohort,
            'intervention_direction': direction,
            'changed_side': family['changed_side'],
            'factor_base': family['factor_base'],
            'factor_source': family['factor_source'],
            'factor_delta': family['factor_delta'],
            'target_deficit': family['target_deficit'],
            'context_value': option['context_value'],
            'physical_dose': option['physical_dose'],
            'hybrid_clearance': option['hybrid_clearance'],
            'mapping_variant': variant,
            'target': target,
            'hybrid': hybrid,
            'screen_target_prediction': target_result['predicted_semantic'],
            'screen_hybrid_prediction': hybrid_result['predicted_semantic'],
            'screen_hybrid_error_is_target_answer': bool(
                hybrid_result['predicted_semantic'] == target['outcome']
            ),
        }
        if pair_key in pair_candidates:
            # The physical pair can appear in more than one candidate family.
            previous = pair_candidates[pair_key]
            if stable_key(row['family_id']) < stable_key(previous['family_id']):
                pair_candidates[pair_key] = row
        else:
            pair_candidates[pair_key] = row

available_pairs = sorted(
    pair_candidates.values(), key=lambda row: row['pair_id']
)
available_counts = Counter(
    (row['factor'], row['cohort']) for row in available_pairs
)

def select_pairs(factor, cohort, target_count, forbidden, attempt):
    groups = defaultdict(list)
    for row in available_pairs:
        if row['factor'] == factor and row['cohort'] == cohort:
            groups[(row['changed_side'], row['mapping_variant'])].append(row)
    for key, values in groups.items():
        values.sort(key=lambda row: stable_key((attempt, key, row['pair_id'])))
    keys = sorted(groups, key=lambda key: stable_key((attempt, 'stratum', key)))
    cursors = {key: 0 for key in keys}
    selected, used = [], set(forbidden)
    while len(selected) < target_count:
        changed = False
        for key in keys:
            values = groups[key]
            while cursors[key] < len(values):
                row = values[cursors[key]]
                cursors[key] += 1
                scene_ids = {
                    str(row['target']['scene_id']),
                    str(row['hybrid']['scene_id']),
                }
                if scene_ids & used:
                    continue
                selected.append(row)
                used.update(scene_ids)
                changed = True
                break
            if len(selected) == target_count:
                break
        if not changed:
            break
    return selected

best_selected = []
for attempt in range(512):
    selected, used = [], set()
    for factor in FACTORS:
        chosen = select_pairs(
            factor, 'failed_update_rescue',
            TARGET_RESCUE_PAIRS_PER_FACTOR, used, attempt,
        )
        selected.extend(chosen)
        used.update(
            str(row[role]['scene_id'])
            for row in chosen for role in ('target', 'hybrid')
        )
    for factor in FACTORS:
        chosen = select_pairs(
            factor, 'correct_flip_induction',
            TARGET_INDUCTION_PAIRS_PER_FACTOR, used, 1000 + attempt,
        )
        selected.extend(chosen)
        used.update(
            str(row[role]['scene_id'])
            for row in chosen for role in ('target', 'hybrid')
        )
    if len(selected) > len(best_selected):
        best_selected = selected
    rescue_counts = Counter(
        row['factor'] for row in selected
        if row['cohort'] == 'failed_update_rescue'
    )
    if all(
        rescue_counts.get(factor, 0) >= TARGET_RESCUE_PAIRS_PER_FACTOR
        for factor in FACTORS
    ):
        best_selected = selected
        break

cohort_manifest = sorted(best_selected, key=lambda row: row['pair_id'])
assert len({row['pair_id'] for row in cohort_manifest}) == len(cohort_manifest)
write_jsonl_atomic(COHORT_MANIFEST_PATH, cohort_manifest)
cohort_counts = Counter(
    (row['factor'], row['cohort']) for row in cohort_manifest
)
cohort_identity = {
    'schema_version': 'fliperror-cohort-manifest-1.0',
    'available_counts': {
        '|'.join(key): value for key, value in sorted(available_counts.items())
    },
    'selected_counts': {
        '|'.join(key): value for key, value in sorted(cohort_counts.items())
    },
    'selection_uses_behavior_only': True,
    'selection_uses_activation_data': False,
    'scene_disjoint_across_selected_pairs': True,
    'pair_ids': [row['pair_id'] for row in cohort_manifest],
}
if COHORT_LOCK_PATH.is_file():
    previous = json.loads(COHORT_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == cohort_identity
else:
    write_json_atomic(COHORT_LOCK_PATH, {
        'identity': cohort_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

rescue_counts = Counter(
    row['factor'] for row in cohort_manifest
    if row['cohort'] == 'failed_update_rescue'
)
RUN_PRIMARY_TEST = all(
    rescue_counts.get(factor, 0) >= MIN_RESCUE_PAIRS_PER_FACTOR
    for factor in FACTORS
)
print('Available:', available_counts)
print('Selected:', cohort_counts)
print('Primary rescue support sufficient:', RUN_PRIMARY_TEST)

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=OUTPUT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {}
for example in outcome_examples:
    outcome_by_scene.setdefault(str(example.scene_id), example)
selected_scene_ids = {
    str(row[role]['scene_id'])
    for row in cohort_manifest for role in ('target', 'hybrid')
}
assert selected_scene_ids <= set(outcome_by_scene)
```

```python
%%cell_progress 8/17 Define the frozen prompt, examples, and token masks
cell_stage('Reconstructing the exact readiness-screen prompt and mapping')
import math
from dataclasses import replace as dataclass_replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def mapping_for_variant(variant):
    semantics = (
        ('left', 'balanced', 'right')
        if variant == 'forward'
        else ('right', 'balanced', 'left')
    )
    return tuple(zip(('A', 'B', 'C'), semantics))

def custom_prompt(mapping):
    options = '\n'.join(
        f'{label}. {SEMANTIC_TEXT[semantic]}'
        for label, semantic in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT,
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, role_name):
    role = row[role_name]
    mapping = mapping_for_variant(row['mapping_variant'])
    semantics = tuple(value for _, value in mapping)
    labels = tuple(label for label, _ in mapping)
    outcome = str(role['outcome'])
    base = outcome_by_scene[str(role['scene_id'])]
    return dataclass_replace(
        base,
        example_id=f"fliperror:{row['pair_id']}:{role_name}",
        question_id=f"fliperror_{row['pair_id']}",
        question_kind='outcome',
        variant='matched_clearance_fliperror',
        prompt=custom_prompt(mapping),
        labels=labels,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=labels[semantics.index(outcome)],
        gold_semantic=outcome,
    )

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    assert len(sections) == len(names)
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans, cursor = [], 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key])
        masks = discover_masks(current)
        for scope in (UPSTREAM_SCOPE, SEMANTIC_SCOPE):
            assert torch.equal(reference_masks[scope], masks[scope])
    return reference_masks

print('Prompt, examples, and token masks ready.')
```

```python
%%cell_progress 9/17 Define capture, graded steering, and restoration helpers
cell_stage('Defining exact replay and equal-norm route controls')

def site_key(layer_index, component):
    return f'L{int(layer_index)}:{component}'

def score_records(prepared, logits):
    records = []
    for batch_index, example in enumerate(prepared['examples']):
        logprobs = torch.log_softmax(
            logits[batch_index, -1, :].float(), dim=-1
        )
        scores = [
            float(logprobs[prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_score(example):
    prepared = prepare_examples([example])
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    records = score_records(prepared, output.logits)
    assert len(records) == 1
    return records[0]

def semantic_margin(record, positive, negative):
    mapping = dict(zip(
        record['candidate_semantics'], record['candidate_logprobs']
    ))
    return float(mapping[positive] - mapping[negative])

def centered_error(first, second):
    first = np.asarray(first, dtype=float)
    second = np.asarray(second, dtype=float)
    return float(np.max(np.abs(
        (first - first.mean()) - (second - second.mean())
    )))

def forward_capture(prepared, upstream_delta=None):
    assert len(prepared['examples']) == 1
    masks = discover_masks(prepared)
    captured, handles = {}, []
    if upstream_delta is not None:
        selected = masks[UPSTREAM_SCOPE]
        def upstream_hook(_module, _inputs, output):
            hidden = output_hidden_tensor(output)
            current = selected.to(hidden.device)
            delta = upstream_delta.to(hidden.device, hidden.dtype)
            modified = hidden.clone()
            modified[0, current] = hidden[0, current] + delta[current]
            return replace_output_hidden(output, modified)
        handles.append(
            DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
        )
    def capture_l48(_module, _inputs, output):
        captured['L48:decoder'] = (
            output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
        )
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(capture_l48)
    )
    for (layer_index, component), module in COMPONENT_MODULES.items():
        key = site_key(layer_index, component)
        def capture(_module, _inputs, output, key=key):
            captured[key] = (
                output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
            )
        handles.append(module.register_forward_hook(capture))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    assert len(records) == 1
    return records[0], captured, masks

def forward_replace_l48(example, payload):
    prepared = prepare_examples([example])
    masks = discover_masks(prepared)
    selected = masks[UPSTREAM_SCOPE]
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = selected.to(hidden.device)
        modified = hidden.clone()
        modified[0, current] = payload.to(hidden.device, hidden.dtype)[current]
        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    return score_records(prepared, output.logits)[0]

def masked_removal(intervention_state, clean_state, mask):
    removal = torch.zeros_like(intervention_state, dtype=torch.bfloat16)
    selected = mask.bool()
    removal[selected] = (
        intervention_state[selected].float() - clean_state[selected].float()
    ).to(torch.bfloat16)
    return removal

def random_removal_like(removal, mask, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    selected = mask.bool()
    random_values = torch.randn(
        removal[selected].shape, generator=generator, dtype=torch.float32
    )
    target_norm = torch.linalg.vector_norm(removal[selected].float())
    random_norm = torch.linalg.vector_norm(random_values)
    if float(target_norm) > 0:
        random_values *= target_norm / random_norm
    else:
        random_values.zero_()
    output = torch.zeros_like(removal, dtype=torch.float32)
    output[selected] = random_values
    return output.to(torch.bfloat16)

def score_restorations(example, upstream_delta, conditions):
    prepared = prepare_examples([example] * len(conditions))
    single = prepare_examples([example])
    masks = discover_masks(single)
    upstream_mask = masks[UPSTREAM_SCOPE]
    handles = []
    measurements = [
        {'restored_norm': 0.0, 'random_norm': 0.0, 'restored_sites': 0}
        for _ in conditions
    ]
    def upstream_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        selected = upstream_mask.to(hidden.device)
        delta = upstream_delta.to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        for batch_index in range(len(conditions)):
            modified[batch_index, selected] = (
                hidden[batch_index, selected] + delta[selected]
            )
        return replace_output_hidden(output, modified)
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
    )
    union_keys = sorted({
        key for condition in conditions for key in condition['removals']
    })
    for key in union_keys:
        layer_text, component = key.split(':')
        module = COMPONENT_MODULES[(int(layer_text[1:]), component)]
        def restore_hook(_module, _inputs, output, key=key):
            hidden = output_hidden_tensor(output)
            modified, changed = hidden.clone(), False
            for batch_index, condition in enumerate(conditions):
                if key not in condition['removals']:
                    continue
                removal = condition['removals'][key].to(
                    hidden.device, hidden.dtype
                )
                modified[batch_index] = hidden[batch_index] - removal
                norm = float(torch.linalg.vector_norm(
                    removal.float()
                ).detach().cpu())
                measurements[batch_index]['restored_sites'] += 1
                field = (
                    'random_norm' if condition['random_control']
                    else 'restored_norm'
                )
                measurements[batch_index][field] += norm ** 2
                changed = True
            return replace_output_hidden(output, modified) if changed else None
        handles.append(module.register_forward_hook(restore_hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, measurement in zip(records, conditions, measurements):
        measurement['restored_norm'] **= 0.5
        measurement['random_norm'] **= 0.5
        record.update({'condition': condition['condition'], **measurement})
    return records

print('Capture, steering, and restoration helpers ready.')
```

```python
%%cell_progress 10/17 Run mandatory behavior, alignment, and parity smoke tests
cell_stage('Revalidating frozen cohorts with exact batch-one inference')
screen_cohort_manifest = list(cohort_manifest)
runtime_replay = []
runtime_manifest = []
replay_progress = tqdm(
    screen_cohort_manifest,
    desc='Batch-one behavior replay', unit='pair',
)
for row in replay_progress:
    target_score = forward_score(make_example(row, 'target'))
    hybrid_score = forward_score(make_example(row, 'hybrid'))
    target_correct = (
        target_score['predicted_semantic'] == str(row['target']['outcome'])
    )
    hybrid_correct = (
        hybrid_score['predicted_semantic'] == str(row['hybrid']['outcome'])
    )
    if row['cohort'] == 'failed_update_rescue':
        stable = target_correct and not hybrid_correct
    else:
        stable = target_correct and hybrid_correct
    replay_record = {
        'pair_id': row['pair_id'],
        'factor': row['factor'],
        'cohort': row['cohort'],
        'screen_target_prediction': row['screen_target_prediction'],
        'screen_hybrid_prediction': row['screen_hybrid_prediction'],
        'runtime_target_prediction': target_score['predicted_semantic'],
        'runtime_hybrid_prediction': hybrid_score['predicted_semantic'],
        'target_screen_match': bool(
            target_score['predicted_semantic']
            == row['screen_target_prediction']
        ),
        'hybrid_screen_match': bool(
            hybrid_score['predicted_semantic']
            == row['screen_hybrid_prediction']
        ),
        'runtime_cohort_stable': bool(stable),
    }
    runtime_replay.append(replay_record)
    if stable:
        retained = dict(row)
        retained['runtime_target_prediction'] = (
            target_score['predicted_semantic']
        )
        retained['runtime_hybrid_prediction'] = (
            hybrid_score['predicted_semantic']
        )
        runtime_manifest.append(retained)

write_jsonl_atomic(RUNTIME_REPLAY_PATH, runtime_replay)
write_jsonl_atomic(RUNTIME_MANIFEST_PATH, runtime_manifest)
cohort_manifest = runtime_manifest
cohort_counts = Counter(
    (row['factor'], row['cohort']) for row in cohort_manifest
)
rescue_counts = Counter(
    row['factor'] for row in cohort_manifest
    if row['cohort'] == 'failed_update_rescue'
)
RUN_PRIMARY_TEST = all(
    rescue_counts.get(factor, 0) >= MIN_RESCUE_PAIRS_PER_FACTOR
    for factor in FACTORS
)
excluded_replay = [
    row for row in runtime_replay if not row['runtime_cohort_stable']
]
assert cohort_manifest, 'No cohort pairs remain stable under batch-one replay.'
print('Screen-selected pairs:', len(screen_cohort_manifest))
print('Runtime-stable pairs:', len(cohort_manifest))
print('Excluded before interventions:', len(excluded_replay))
print('Runtime-stable cohorts:', cohort_counts)

cell_stage('Replaying one stable pair and equal-norm route removal')
smoke_row = cohort_manifest[0]
target_example = make_example(smoke_row, 'target')
hybrid_example = make_example(smoke_row, 'hybrid')
prepared = {
    'target': prepare_examples([target_example]),
    'hybrid': prepare_examples([hybrid_example]),
}
masks = assert_aligned(prepared)
target_score, target_hidden, _ = forward_capture(prepared['target'])
hybrid_score, hybrid_hidden, _ = forward_capture(prepared['hybrid'])
assert target_score['predicted_semantic'] == smoke_row['runtime_target_prediction']
assert hybrid_score['predicted_semantic'] == smoke_row['runtime_hybrid_prediction']

base_delta = torch.zeros_like(hybrid_hidden['L48:decoder'])
selected = masks[UPSTREAM_SCOPE]
base_delta[selected] = (
    hybrid_hidden['L48:decoder'][selected].float()
    - target_hidden['L48:decoder'][selected].float()
).to(torch.bfloat16)
signed_delta = float(smoke_row['intervention_direction']) * ALPHAS[0] * base_delta
direct_score, direct_hidden, _ = forward_capture(
    prepared['hybrid'], signed_delta
)
real_removals, random_removals = {}, {}
for rank, site in enumerate(FROZEN_SITES[smoke_row['factor']], start=1):
    key = site['site_key']
    mask = masks[site['scope']]
    removal = masked_removal(
        direct_hidden[key], hybrid_hidden[key], mask
    )
    real_removals[key] = removal
    random_removals[key] = random_removal_like(
        removal, mask,
        f"smoke:{smoke_row['pair_id']}:{rank}:{key}",
    )
restored = score_restorations(
    hybrid_example, signed_delta,
    [
        {'condition': 'joint_restore', 'removals': real_removals,
         'random_control': False},
        {'condition': 'joint_random', 'removals': random_removals,
         'random_control': True},
    ],
)
self_score = forward_replace_l48(
    hybrid_example, hybrid_hidden['L48:decoder']
)
self_error = centered_error(
    self_score['candidate_logprobs'], hybrid_score['candidate_logprobs']
)
norm_error = abs(
    restored[0]['restored_norm'] - restored[1]['random_norm']
) / max(restored[0]['restored_norm'], 1e-12)
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert norm_error < NORM_RELATIVE_TOLERANCE, norm_error
print('Smoke cohort / factor:', smoke_row['cohort'], smoke_row['factor'])
print('Batch-one replay:', target_score['predicted_semantic'],
      hybrid_score['predicted_semantic'])
print('Self replay / norm error:', self_error, norm_error)
```

```python
%%cell_progress 11/17 Run graded rescue, induction, and route restoration
cell_stage('Processing independent pairs with resumable per-pair shards')

capture_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'screen_cohort_sha256': sha256_file(COHORT_MANIFEST_PATH),
    'runtime_cohort_sha256': sha256_file(RUNTIME_MANIFEST_PATH),
    'runtime_replay_sha256': sha256_file(RUNTIME_REPLAY_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'frozen_sites': FROZEN_SITES,
    'alphas': list(ALPHAS),
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(
    cohort_manifest, desc='Targeted flip-error interventions', unit='pair'
)
for row in progress:
    shard_path = SHARD_DIR / f"{row['pair_id']}.json"
    if shard_path.is_file():
        existing = json.loads(shard_path.read_text(encoding='utf-8'))
        assert existing['capture_identity_sha256'] == CAPTURE_SHA256
        assert len(existing['records']) == len(ALPHAS) * len(CONDITIONS) + 1
        continue
    target_example = make_example(row, 'target')
    hybrid_example = make_example(row, 'hybrid')
    prepared = {
        'target': prepare_examples([target_example]),
        'hybrid': prepare_examples([hybrid_example]),
    }
    masks = assert_aligned(prepared)
    target_score, target_hidden, _ = forward_capture(prepared['target'])
    hybrid_score, hybrid_hidden, _ = forward_capture(prepared['hybrid'])
    assert target_score['predicted_semantic'] == row['runtime_target_prediction']
    assert hybrid_score['predicted_semantic'] == row['runtime_hybrid_prediction']
    selected = masks[UPSTREAM_SCOPE]
    natural_delta = torch.zeros_like(hybrid_hidden['L48:decoder'])
    natural_delta[selected] = (
        hybrid_hidden['L48:decoder'][selected].float()
        - target_hidden['L48:decoder'][selected].float()
    ).to(torch.bfloat16)
    positive = str(row['hybrid']['outcome'])
    negative = str(row['target']['outcome'])
    assert positive == row['changed_side'] and negative != positive
    clean_margin = semantic_margin(hybrid_score, positive, negative)
    pair_records = []
    for alpha in ALPHAS:
        upstream_delta = (
            float(row['intervention_direction']) * float(alpha) * natural_delta
        ).to(torch.bfloat16)
        direct_score, direct_hidden, _ = forward_capture(
            prepared['hybrid'], upstream_delta
        )
        random_upstream_delta = random_removal_like(
            upstream_delta, selected,
            f"upstream:{row['pair_id']}:{alpha}",
        )
        upstream_random_score, _, _ = forward_capture(
            prepared['hybrid'], random_upstream_delta
        )
        real_removals, random_removals = {}, {}
        for rank, site in enumerate(FROZEN_SITES[row['factor']], start=1):
            key = site['site_key']
            mask = masks[site['scope']]
            removal = masked_removal(
                direct_hidden[key], hybrid_hidden[key], mask
            )
            real_removals[key] = removal
            random_removals[key] = random_removal_like(
                removal, mask,
                f"fliperror:{row['pair_id']}:{alpha}:{rank}:{key}",
            )
        restored = score_restorations(
            hybrid_example, upstream_delta,
            [
                {'condition': 'joint_restore', 'removals': real_removals,
                 'random_control': False},
                {'condition': 'joint_random', 'removals': random_removals,
                 'random_control': True},
            ],
        )
        direct_score.update({
            'condition': 'direct', 'restored_norm': 0.0,
            'random_norm': 0.0, 'restored_sites': 0,
        })
        upstream_random_score.update({
            'condition': 'upstream_random', 'restored_norm': 0.0,
            'random_norm': 0.0, 'restored_sites': 0,
        })
        by_condition = {
            record['condition']: record
            for record in [direct_score, upstream_random_score, *restored]
        }
        assert set(by_condition) == set(CONDITIONS)
        for condition in CONDITIONS:
            record = by_condition[condition]
            margin = semantic_margin(record, positive, negative)
            directed_shift = float(row['intervention_direction']) * (
                margin - clean_margin
            )
            desired_semantic = (
                positive if row['cohort'] == 'failed_update_rescue'
                else negative
            )
            record.update({
                'schema_version': 'fliperror-intervention-record-1.0',
                'pair_id': row['pair_id'],
                'family_id': row['family_id'],
                'factor': row['factor'],
                'cohort': row['cohort'],
                'alpha': float(alpha),
                'condition': condition,
                'intervention_direction': row['intervention_direction'],
                'changed_side': row['changed_side'],
                'mapping_variant': row['mapping_variant'],
                'physical_dose': row['physical_dose'],
                'target_deficit': row['target_deficit'],
                'hybrid_clearance': row['hybrid_clearance'],
                'positive_semantic': positive,
                'negative_semantic': negative,
                'desired_semantic': desired_semantic,
                'clean_predicted_semantic': hybrid_score['predicted_semantic'],
                'clean_candidate_semantics': hybrid_score['candidate_semantics'],
                'clean_candidate_logprobs': hybrid_score['candidate_logprobs'],
                'clean_margin': clean_margin,
                'intervention_margin': margin,
                'directed_shift': directed_shift,
                'desired_success': float(
                    record['predicted_semantic'] == desired_semantic
                ),
                'natural_delta_norm': float(torch.linalg.vector_norm(
                    natural_delta[selected].float()
                )),
                'upstream_intervention_norm': float(torch.linalg.vector_norm(
                    upstream_delta[selected].float()
                )),
                'upstream_random_control_norm': float(torch.linalg.vector_norm(
                    random_upstream_delta[selected].float()
                )),
            })
            pair_records.append(record)
    self_score = forward_replace_l48(
        hybrid_example, hybrid_hidden['L48:decoder']
    )
    self_score.update({
        'schema_version': 'fliperror-intervention-record-1.0',
        'pair_id': row['pair_id'], 'family_id': row['family_id'],
        'factor': row['factor'], 'cohort': row['cohort'],
        'alpha': 0.0, 'condition': 'self_replay',
        'clean_candidate_logprobs': hybrid_score['candidate_logprobs'],
        'centered_replay_error': centered_error(
            self_score['candidate_logprobs'], hybrid_score['candidate_logprobs']
        ),
    })
    pair_records.append(self_score)
    write_json_atomic(shard_path, {
        'schema_version': 'fliperror-pair-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'pair': row,
        'records': pair_records,
    })
    del prepared, target_hidden, hybrid_hidden, natural_delta
    torch.cuda.empty_cache()
    progress.set_postfix_str(
        f"{row['factor']} | {row['cohort']} | {row['pair_id']}"
    )

shard_paths = sorted(SHARD_DIR.glob('*.json'))
assert len(shard_paths) == len(cohort_manifest)
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'capture_identity': capture_identity,
    'capture_identity_sha256': CAPTURE_SHA256,
    'pairs': len(cohort_manifest),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Complete pair shards:', len(shard_paths))
```

```python
%%cell_progress 13/17 Test graded rescue and induction dose response
cell_stage('Estimating within-pair slopes across the three frozen strengths')

def simple_slope(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    centered = x - x.mean()
    return float(np.sum(centered * (y - y.mean())) / np.sum(centered ** 2))

pair_rows = []
for (factor, cohort, pair_id), current in cell_frame.groupby([
    'factor', 'cohort', 'pair_id'
]):
    current = current.sort_values('alpha')
    row = {'factor': factor, 'cohort': cohort, 'pair_id': pair_id}
    for metric in (
        'direct_shift', 'upstream_random_shift',
        'joint_restore_shift', 'joint_random_shift',
        'route_attenuation', 'route_specificity',
        'upstream_specificity',
    ):
        row[f'{metric}_slope'] = simple_slope(current['alpha'], current[metric])
    maximum = current[current['alpha'] == max(ALPHAS)].iloc[0]
    for metric in (
        'direct_success', 'upstream_random_success',
        'joint_restore_success', 'joint_random_success'
    ):
        row[f'max_{metric}'] = float(maximum[metric])
    pair_rows.append(row)
pair_frame = pd.DataFrame(pair_rows)
pair_frame.to_csv(ANALYSIS_DIR / 'pair_dose_slopes.csv', index=False)
display(pair_frame.groupby(['factor', 'cohort']).mean(
    numeric_only=True
).round(5))
```

```python
%%cell_progress 14/17 Bootstrap targeted causal and route-specific effects
cell_stage('Resampling independent image pairs within factor and cohort')

def bootstrap_mean(values, key):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    # Correct-flip induction is explicitly secondary and can contain only one
    # pair after the behavior screen; its interval is then descriptive.
    assert len(values) >= 1, (key, len(values))
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    estimates = values[draws].mean(axis=1)
    return {
        'estimate': float(values.mean()),
        'ci_low': float(np.quantile(estimates, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(estimates, 1 - ALPHA_LEVEL / 2)),
        'n_pairs': int(len(values)),
    }

METRICS = (
    'direct_shift_slope', 'joint_restore_shift_slope',
    'joint_random_shift_slope', 'upstream_random_shift_slope',
    'route_attenuation_slope', 'upstream_specificity_slope',
    'route_specificity_slope', 'max_direct_success',
    'max_joint_restore_success', 'max_joint_random_success',
)
test_rows = []
for (factor, cohort), current in pair_frame.groupby(['factor', 'cohort']):
    for metric in METRICS:
        summary = bootstrap_mean(
            current[metric], f'{factor}:{cohort}:{metric}'
        )
        test_rows.append({
            'factor': factor, 'cohort': cohort, 'metric': metric, **summary
        })
test_frame = pd.DataFrame(test_rows)
test_frame.to_csv(ANALYSIS_DIR / 'targeted_causal_tests.csv', index=False)

def lookup(factor, cohort, metric):
    rows = test_frame[
        (test_frame['factor'] == factor)
        & (test_frame['cohort'] == cohort)
        & (test_frame['metric'] == metric)
    ]
    return rows.iloc[0] if len(rows) else None

factor_decisions = {}
for factor in FACTORS:
    rescue_n = cohort_counts.get((factor, 'failed_update_rescue'), 0)
    direct = lookup(
        factor, 'failed_update_rescue', 'direct_shift_slope'
    )
    attenuation = lookup(
        factor, 'failed_update_rescue', 'route_attenuation_slope'
    )
    specificity = lookup(
        factor, 'failed_update_rescue', 'route_specificity_slope'
    )
    upstream_specificity = lookup(
        factor, 'failed_update_rescue', 'upstream_specificity_slope'
    )
    success = lookup(
        factor, 'failed_update_rescue', 'max_direct_success'
    )
    gates = {
        'minimum_error_support': rescue_n >= MIN_RESCUE_PAIRS_PER_FACTOR,
        'graded_targeted_rescue': bool(
            direct is not None and direct['ci_low'] > 0
        ),
        'rescue_reaches_any_answers': bool(
            success is not None and success['estimate'] > 0
        ),
        'factor_direction_specific_over_random': bool(
            upstream_specificity is not None
            and upstream_specificity['ci_low'] > 0
        ),
        'frozen_route_attenuates_rescue': bool(
            attenuation is not None and attenuation['ci_low'] > 0
        ),
        'frozen_route_specific_over_random': bool(
            specificity is not None and specificity['ci_low'] > 0
        ),
    }
    factor_decisions[factor] = {
        'confirmed': bool(all(gates.values())),
        'gates': {key: bool(value) for key, value in gates.items()},
        'rescue_pairs': rescue_n,
        'induction_pairs': cohort_counts.get(
            (factor, 'correct_flip_induction'), 0
        ),
    }

confirmed = [
    factor for factor in FACTORS if factor_decisions[factor]['confirmed']
]
steerable = [
    factor for factor in FACTORS
    if factor_decisions[factor]['gates']['graded_targeted_rescue']
]
if len(confirmed) == 2:
    PRIMARY_DECISION = 'targeted_flip_error_route_confirmed_for_both_factors'
elif len(confirmed) == 1:
    PRIMARY_DECISION = 'targeted_flip_error_route_confirmed_for_one_factor'
elif steerable:
    PRIMARY_DECISION = 'errors_steerable_without_full_route_specificity'
elif not RUN_PRIMARY_TEST:
    PRIMARY_DECISION = 'insufficient_independent_failed_update_errors'
else:
    PRIMARY_DECISION = 'failed_update_errors_not_rescued_by_factor_delta'

decision_document = {
    'schema_version': 'fliperror-rescue-decision-1.0',
    'primary_decision': PRIMARY_DECISION,
    'factor_decisions': factor_decisions,
    'induction_is_secondary_due_to_behavioral_scarcity': True,
    'claim_boundary': (
        'The test is conditional on behavior-defined failed-update errors and '
        'does not estimate unconditional model competence.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'fliperror_rescue_decision.json', decision_document)
display(test_frame.round(6))
print('Primary decision:', PRIMARY_DECISION)
```

```python
%%cell_progress 15/17 Audit cohorts, replay, norms, and frozen provenance
cell_stage('Verifying behavior labels, independence, numerical parity, and controls')

violations = []
selected_scene_ids = []
mapping_counts = Counter()
runtime_replay_rows = read_jsonl(RUNTIME_REPLAY_PATH)
runtime_replay_by_pair = {
    row['pair_id']: row for row in runtime_replay_rows
}
assert len(runtime_replay_by_pair) == len(screen_cohort_manifest)
runtime_excluded_count = sum(
    not row['runtime_cohort_stable'] for row in runtime_replay_rows
)
screen_prediction_mismatch_count = sum(
    not row['target_screen_match'] or not row['hybrid_screen_match']
    for row in runtime_replay_rows
)
for row in cohort_manifest:
    target_key = f"{row['target']['scene_id']}|{row['mapping_variant']}"
    hybrid_key = f"{row['hybrid']['scene_id']}|{row['mapping_variant']}"
    target_result = screen_by_key[target_key]
    hybrid_result = screen_by_key[hybrid_key]
    if row['cohort'] == 'failed_update_rescue':
        if not target_result['correct'] or hybrid_result['correct']:
            violations.append((row['pair_id'], 'rescue_behavior'))
    else:
        if not target_result['correct'] or not hybrid_result['correct']:
            violations.append((row['pair_id'], 'induction_behavior'))
    replay = runtime_replay_by_pair[row['pair_id']]
    if not replay['runtime_cohort_stable']:
        violations.append((row['pair_id'], 'runtime_cohort_unstable'))
    if (
        replay['runtime_target_prediction']
        != row['runtime_target_prediction']
        or replay['runtime_hybrid_prediction']
        != row['runtime_hybrid_prediction']
    ):
        violations.append((row['pair_id'], 'runtime_manifest_mismatch'))
    if row['target']['outcome'] == row['hybrid']['outcome']:
        violations.append((row['pair_id'], 'not_flip'))
    selected_scene_ids.extend([
        str(row['target']['scene_id']), str(row['hybrid']['scene_id'])
    ])
    mapping_counts[(row['factor'], row['cohort'], row['mapping_variant'])] += 1

self_errors = self_frame['centered_replay_error'].to_numpy(dtype=float)
upstream_norm_errors = []
for _, row in record_frame[record_frame['condition'] == 'direct'].iterrows():
    upstream_norm_errors.append(abs(
        float(row['upstream_intervention_norm'])
        - float(row['upstream_random_control_norm'])
    ) / max(float(row['upstream_intervention_norm']), 1e-12))
norm_errors = []
for _, current in record_frame.groupby(['pair_id', 'alpha']):
    real = current[current['condition'] == 'joint_restore'].iloc[0]
    random = current[current['condition'] == 'joint_random'].iloc[0]
    norm_errors.append(abs(
        float(real['restored_norm']) - float(random['random_norm'])
    ) / max(float(real['restored_norm']), 1e-12))
nonfinite_count = int((~np.isfinite(
    cell_frame.select_dtypes(include=[np.number]).to_numpy()
)).sum())
assert not violations, violations[:5]
assert len(selected_scene_ids) == len(set(selected_scene_ids))
assert max(self_errors) < SMOKE_CENTERED_LOGPROB_TOLERANCE
assert max(upstream_norm_errors) < NORM_RELATIVE_TOLERANCE
assert max(norm_errors) < NORM_RELATIVE_TOLERANCE
assert nonfinite_count == 0

audit_document = {
    'schema_version': 'fliperror-rescue-audit-1.1',
    'screen_selected_pairs': len(screen_cohort_manifest),
    'pairs': len(cohort_manifest),
    'runtime_excluded_before_intervention': runtime_excluded_count,
    'screen_to_batch_one_prediction_mismatches': (
        screen_prediction_mismatch_count
    ),
    'cohort_counts': {
        '|'.join(key): value for key, value in sorted(cohort_counts.items())
    },
    'behavior_selection_only': True,
    'activation_selection_used': False,
    'runtime_replay_completed_before_interventions': True,
    'route_frozen_from_prior_work': True,
    'scene_disjoint_selected_pairs': True,
    'behavior_violations': len(violations),
    'maximum_self_replay_error': float(max(self_errors)),
    'maximum_upstream_equal_norm_relative_error': float(
        max(upstream_norm_errors)
    ),
    'maximum_equal_norm_relative_error': float(max(norm_errors)),
    'nonfinite_count': nonfinite_count,
    'mapping_counts': {
        '|'.join(key): value for key, value in sorted(mapping_counts.items())
    },
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_document)
display(pd.DataFrame([audit_document]).T.rename(columns={0: 'value'}))
```
