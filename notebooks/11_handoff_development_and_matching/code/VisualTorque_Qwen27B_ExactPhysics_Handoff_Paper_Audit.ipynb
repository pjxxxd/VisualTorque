```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 11 tracked cells.')
```

```python
%%cell_progress 5/11 Reconstruct the 29 matched difference-in-differences
cell_stage('Computing one independent contrast per exact-physics match')

def matched_contrasts(frame, condition, metric):
    current = frame[frame['condition'].eq(condition)]
    wide = current.pivot(
        index='match_id', columns=['cohort', 'side'], values=metric
    ).sort_index()
    required_cells = {
        ('both_correct', 'left'), ('both_correct', 'right'),
        ('right_error', 'left'), ('right_error', 'right'),
    }
    assert required_cells.issubset(set(wide.columns))
    both_gap = wide[('both_correct', 'right')] - wide[('both_correct', 'left')]
    error_gap = wide[('right_error', 'right')] - wide[('right_error', 'left')]
    return pd.DataFrame({
        'both_correct_side_gap': both_gap,
        'right_error_side_gap': error_gap,
        'right_error_minus_right_correct': (
            wide[('right_error', 'right')] - wide[('both_correct', 'right')]
        ),
        'left_control_difference': (
            wide[('right_error', 'left')] - wide[('both_correct', 'left')]
        ),
        'difference_in_differences': error_gap - both_gap,
    })

per_match_parts = []
for metric in METRICS:
    current = matched_contrasts(estimand_frame, PRIMARY_CONDITION, metric)
    current = current.reset_index()
    current['metric'] = metric
    per_match_parts.append(current)
per_match = pd.concat(per_match_parts, ignore_index=True)
assert per_match.groupby('metric')['match_id'].nunique().eq(29).all()
per_match.to_csv(ANALYSIS_DIR / 'per_match_contrasts.csv', index=False)

runtime_means = per_match.groupby('metric')['difference_in_differences'].mean()
frozen_means = confirmation_tests[
    confirmation_tests['condition'].eq(PRIMARY_CONDITION)
    & confirmation_tests['contrast'].eq('difference_in_differences')
].set_index('metric')['estimate']
for metric in METRICS:
    assert abs(float(runtime_means[metric]) - float(frozen_means[metric])) < 1e-8

display(per_match.groupby('metric').agg(
    estimate=('difference_in_differences', 'mean'),
    median=('difference_in_differences', 'median'),
    positive_fraction=('difference_in_differences', lambda x: float((x > 0).mean())),
).round(5))
```

## 3. Test amplification rather than a sharp serial break

The early and late coordinates have different natural scales, so their raw values cannot be compared directly.


```python
%%cell_progress 6/11 Quantify normalized late-stage amplification
cell_stage('Normalizing paired early and late effects by frozen direction norms')

def bootstrap_summary(values, key):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    assert len(values) > 0
    rng = np.random.default_rng(stable_seed(('bootstrap', key)))
    draws = rng.choice(
        values, size=(BOOTSTRAP_REPLICATES, len(values)), replace=True
    ).mean(axis=1)
    return {
        'estimate': float(values.mean()),
        'ci_low': float(np.quantile(draws, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA_LEVEL / 2)),
        'n_matched_units': int(len(values)),
    }

did_wide = per_match.pivot(
    index='match_id', columns='metric', values='difference_in_differences'
).sort_index()
normalized = pd.DataFrame(index=did_wide.index)
normalized['early_L49'] = (
    did_wide['l49_toward_exact'] / float(direction['semantic_norm'])
)
normalized['late_L55'] = (
    did_wide['l55_toward_exact_code'] / float(direction['code_norm'])
)
normalized['late_minus_early'] = (
    normalized['late_L55'] - normalized['early_L49']
)
normalized['answer_logodds'] = did_wide['margin_shift']
normalized.reset_index().to_csv(
    ANALYSIS_DIR / 'normalized_per_match_amplification.csv', index=False
)

amplification_rows = []
for name in ('early_L49', 'late_L55', 'late_minus_early', 'answer_logodds'):
    amplification_rows.append({
        'estimand': name,
        **bootstrap_summary(normalized[name], ('normalized', name)),
    })
amplification_frame = pd.DataFrame(amplification_rows)
early_mean = float(normalized['early_L49'].mean())
late_mean = float(normalized['late_L55'].mean())
descriptive_fold_amplification = float(late_mean / early_mean)
amplification_frame['interpretation'] = [
    'small early error-associated route difference',
    'larger late response-state difference',
    'paired scale-normalized amplification',
    'downstream answer consequence',
]
amplification_frame.to_csv(
    ANALYSIS_DIR / 'normalized_amplification_tests.csv', index=False
)

leave_one_match_out = []
for held_out in normalized.index:
    kept = normalized.drop(index=held_out)
    leave_one_match_out.append({
        'held_out_match_id': held_out,
        'early_L49': float(kept['early_L49'].mean()),
        'late_L55': float(kept['late_L55'].mean()),
        'late_minus_early': float(kept['late_minus_early'].mean()),
    })
leave_one_match_out_frame = pd.DataFrame(leave_one_match_out)
leave_one_match_out_frame.to_csv(
    ANALYSIS_DIR / 'leave_one_match_out_amplification.csv', index=False
)

display(amplification_frame.round(5))
print('Descriptive late/early fold amplification:', round(descriptive_fold_amplification, 3))
print('Leave-one-match-out late-minus-early range:', (
    round(leave_one_match_out_frame['late_minus_early'].min(), 5),
    round(leave_one_match_out_frame['late_minus_early'].max(), 5),
))
```

```python
%%cell_progress 7/11 Compare discovery with exact-physics confirmation
cell_stage('Extracting the same frozen interaction from both studies')

def extract_test(frame, metric):
    row = frame[
        frame['condition'].eq(PRIMARY_CONDITION)
        & frame['metric'].eq(metric)
        & frame['contrast'].eq('difference_in_differences')
    ]
    assert len(row) == 1, (metric, len(row))
    return row.iloc[0]

replication_rows = []
for stage, frame, matching in (
    ('Discovery', discovery_tests, 'approximate physical matching'),
    ('Exact-physics confirmation', confirmation_tests, 'exact physical matching'),
):
    for metric in METRICS:
        row = extract_test(frame, metric)
        replication_rows.append({
            'stage': stage,
            'physical_matching': matching,
            'metric': metric,
            'estimate': float(row['estimate']),
            'ci_low': float(row['ci_low']),
            'ci_high': float(row['ci_high']),
            'n_matched_units': int(row['n_matched_units']),
            'positive_interval': bool(float(row['ci_low']) > 0),
        })
replication_frame = pd.DataFrame(replication_rows)
replication_frame.to_csv(
    ANALYSIS_DIR / 'discovery_confirmation_replication.csv', index=False
)

sign_table = replication_frame.pivot(
    index='metric', columns='stage', values='estimate'
)
sign_consistency = {
    metric: bool(
        np.sign(sign_table.loc[metric, 'Discovery'])
        == np.sign(sign_table.loc[metric, 'Exact-physics confirmation'])
    ) for metric in METRICS
}
assert all(sign_consistency.values())
display(replication_frame.round(5))
print('Direction-of-effect replication:', sign_consistency)
```

## 4. Build a causal evidence triangle

A numerical state difference is more convincing when it predicts a directional behavioral intervention.


```python
%%cell_progress 8/11 Assemble rescue, induction, and specificity evidence
cell_stage('Connecting internal-state effects to directional behavior changes')

baseline_accuracy = baseline_frame.groupby(['cohort', 'side'])['correct'].mean()

def behavior_result(condition, outcome):
    row = behavior_frame[
        behavior_frame['condition'].eq(condition)
        & behavior_frame['outcome'].eq(outcome)
    ]
    assert len(row) == 1, (condition, outcome, len(row))
    return row.iloc[0]

triangle_specs = (
    ('Observed right-error accuracy', 'baseline', 'right_error:right', None, None),
    ('L48 matched rescue', 'upstream_matched', 'right_error_rescue', None, None),
    ('L48 random rescue', 'upstream_random_matched', 'right_error_rescue', None, None),
    ('L49 direct rescue', 'l49_direct', 'right_error_rescue', None, None),
    ('L55 direct rescue', 'l55_direct', 'right_error_rescue', None, None),
    ('L48 reverse induction', 'upstream_reverse_matched', 'right_correct_induction', None, None),
    ('L49 reverse induction', 'l49_reverse', 'right_correct_induction', None, None),
    ('L55 stale induction', 'l55_stale', 'right_correct_induction', None, None),
    ('Left-control induction', 'upstream_reverse_matched', 'left_correct_induction', None, None),
)
triangle_rows = []
for label, condition, outcome, _, _ in triangle_specs:
    if condition == 'baseline':
        value = float(baseline_accuracy.loc[('right_error', 'right')])
        triangle_rows.append({
            'evidence': label, 'condition': condition, 'outcome': outcome,
            'estimate': value, 'ci_low': np.nan, 'ci_high': np.nan,
        })
    else:
        row = behavior_result(condition, outcome)
        triangle_rows.append({
            'evidence': label, 'condition': condition, 'outcome': outcome,
            'estimate': float(row['estimate']),
            'ci_low': float(row['ci_low']),
            'ci_high': float(row['ci_high']),
        })
triangle_frame = pd.DataFrame(triangle_rows)
triangle_frame.to_csv(ANALYSIS_DIR / 'causal_evidence_triangle.csv', index=False)

specificity_paper = specificity_frame.copy()
specificity_paper['positive_interval'] = specificity_paper['ci_low'] > 0
specificity_paper.to_csv(ANALYSIS_DIR / 'direction_specificity_paper.csv', index=False)

display(triangle_frame.round(4))
display(specificity_paper.round(4))
```

```python
%%cell_progress 9/11 Audit robustness and forbidden overclaims
cell_stage('Testing style stability, single-match sensitivity, and claim boundaries')

def frozen_result(metric, contrast):
    row = confirmation_tests[
        confirmation_tests['condition'].eq(PRIMARY_CONDITION)
        & confirmation_tests['metric'].eq(metric)
        & confirmation_tests['contrast'].eq(contrast)
    ]
    assert len(row) == 1
    return row.iloc[0]

early_raw = frozen_result('l49_toward_exact', 'difference_in_differences')
late_raw = frozen_result('l55_toward_exact_code', 'difference_in_differences')
answer_raw = frozen_result('margin_shift', 'difference_in_differences')
left_early = frozen_result('l49_toward_exact', 'left_control_difference')
left_late = frozen_result('l55_toward_exact_code', 'left_control_difference')
left_answer = frozen_result('margin_shift', 'left_control_difference')
amplification_test = amplification_frame.set_index('estimand').loc['late_minus_early']

style_summary = style_frame.groupby('metric').agg(
    held_out_families=('held_out_style_family', 'nunique'),
    minimum_estimate=('estimate', 'min'),
    minimum_ci_low=('ci_low', 'min'),
    maximum_estimate=('estimate', 'max'),
).reset_index()
style_summary.to_csv(ANALYSIS_DIR / 'style_robustness_summary.csv', index=False)

random_specificity = specificity_frame[
    specificity_frame['control_condition'].eq('upstream_random_matched')
].set_index('metric')
reverse_answer_specificity = specificity_frame[
    specificity_frame['control_condition'].eq('upstream_reverse_matched')
    & specificity_frame['metric'].eq('margin_shift')
].iloc[0]

supported = {
    'exact_physical_matching': bool(
        confirmation_audit['maximum_absolute_physical_smd'] == 0.0
    ),
    'early_error_associated_difference_is_positive': bool(early_raw['ci_low'] > 0),
    'late_error_associated_difference_is_positive': bool(late_raw['ci_low'] > 0),
    'answer_interaction_is_positive': bool(answer_raw['ci_low'] > 0),
    'normalized_late_exceeds_early': bool(amplification_test['ci_low'] > 0),
    'direct_exceeds_random_at_L49': bool(
        random_specificity.loc['l49_toward_exact', 'ci_low'] > 0
    ),
    'direct_exceeds_random_at_L55': bool(
        random_specificity.loc['l55_toward_exact_code', 'ci_low'] > 0
    ),
    'direct_exceeds_random_at_answer': bool(
        random_specificity.loc['margin_shift', 'ci_low'] > 0
    ),
    'all_style_holdouts_preserve_positive_answer_interval': bool(
        (style_frame[style_frame['metric'].eq('margin_shift')]['ci_low'] > 0).all()
    ),
    'leave_one_match_out_preserves_positive_amplification': bool(
        (leave_one_match_out_frame['late_minus_early'] > 0).all()
    ),
}

forbidden_or_unsupported = {
    'claim_no_early_difference': bool(not supported['early_error_associated_difference_is_positive']),
    'claim_effect_is_exclusive_to_right_side': bool(
        abs(float(left_early['estimate'])) < 1e-12
        and abs(float(left_late['estimate'])) < 1e-12
        and abs(float(left_answer['estimate'])) < 1e-12
    ),
    'claim_direct_answer_effect_exceeds_reverse': bool(
        reverse_answer_specificity['ci_low'] > 0
    ),
    'claim_universal_across_models_or_tasks': False,
    'claim_single_sharp_serial_break': False,
}

claim_audit = {
    'paper_ready_wording': (
        'Under exact physical matching, stable right-side errors are associated '
        'with a small early route difference that is amplified into a larger '
        'late response-state difference and a directional answer effect. The '
        'same frozen route can rescue the error, while reversing or replacing '
        'it can induce the corresponding error in otherwise correct cases.'
    ),
    'supported': supported,
    'forbidden_or_unsupported_claims': forbidden_or_unsupported,
    'important_caveats': [
        'L49 is small but statistically nonzero; do not call it absent.',
        'Left-control differences are also nonzero; use the matched interaction, not a right-only claim.',
        'The reverse direction is strongly behaviorally active, so direct-greater-than-reverse is not required.',
        'The claim is conditional on Qwen-27B, this task, and the frozen behavior-defined cohort.',
    ],
    'descriptive_fold_amplification': descriptive_fold_amplification,
}
write_json_atomic(CLAIM_AUDIT_PATH, claim_audit)

claim_table = pd.DataFrame([
    {'claim': key, 'supported': value, 'category': 'supported'}
    for key, value in supported.items()
] + [
    {'claim': key, 'supported': value, 'category': 'unsupported stronger claim'}
    for key, value in forbidden_or_unsupported.items()
])
claim_table.to_csv(ANALYSIS_DIR / 'claim_boundary_table.csv', index=False)
display(style_summary.round(5))
display(claim_table)
```
