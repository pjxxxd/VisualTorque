```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Build model-blind cross-context source candidates
cell_stage('Matching distance-error targets to answer-neutral source pairs')
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

candidate_groups = read_jsonl(CANDIDATE_PATH)
prior_screen = read_jsonl(SCREEN_RESULT_PATH)
prior_screen_by_key = {row['request_key']: row for row in prior_screen}
current_rows = read_jsonl(CURRENT_RUNTIME_MANIFEST_PATH)
distance_targets = [
    row for row in current_rows
    if row['factor'] == 'distance'
    and row['cohort'] == 'failed_update_rescue'
]
assert len(distance_targets) >= MIN_BUNDLES, len(distance_targets)

def mapping_variant_for(family, option):
    index = int(option['context_value']) + int(family['counterbalance_phase'])
    return ('forward', 'reverse')[index % 2]

def request_key(scene_id, mapping_variant):
    return f'{scene_id}|{mapping_variant}'

def prior_pair_correct(family, option):
    variant = mapping_variant_for(family, option)
    target = prior_screen_by_key[
        request_key(option['target']['scene_id'], variant)
    ]
    hybrid = prior_screen_by_key[
        request_key(option['hybrid']['scene_id'], variant)
    ]
    return bool(target['correct'] and hybrid['correct'])

excluded_scene_ids = {
    str(row[role]['scene_id'])
    for row in current_rows for role in ('target', 'hybrid')
}

neutral_options = {'distance': [], 'weight': []}
for family in candidate_groups:
    factor = str(family['factor'])
    for option in family['context_options']:
        if option['transition'] != 'no_flip':
            continue
        if option['target']['outcome'] != option['hybrid']['outcome']:
            continue
        scene_ids = {
            str(option['target']['scene_id']),
            str(option['hybrid']['scene_id']),
        }
        if scene_ids & excluded_scene_ids or len(scene_ids) != 2:
            continue
        if not prior_pair_correct(family, option):
            continue
        neutral_options[factor].append({
            'family_id': family['family_id'],
            'factor': factor,
            'changed_side': family['changed_side'],
            'factor_base': int(family['factor_base']),
            'factor_source': int(family['factor_source']),
            'factor_delta': int(family['factor_delta']),
            'context_value': int(option['context_value']),
            'physical_dose': int(option['physical_dose']),
            'target_deficit': int(option['target_deficit']),
            'mapping_variant_at_prior_screen': mapping_variant_for(
                family, option
            ),
            'target': option['target'],
            'hybrid': option['hybrid'],
        })

model_blind_pool = []
for target in distance_targets:
    desired = str(target['hybrid']['outcome'])
    distance_candidates = []
    for donor in neutral_options['distance']:
        if donor['changed_side'] != target['changed_side']:
            continue
        if donor['factor_base'] != int(target['factor_base']):
            continue
        if donor['factor_source'] != int(target['factor_source']):
            continue
        if donor['context_value'] == int(target['context_value']):
            continue
        if str(donor['target']['outcome']) == desired:
            continue
        distance_candidates.append(donor)
    distance_candidates.sort(key=lambda donor: (
        abs(donor['context_value'] - int(target['context_value'])),
        stable_key((target['pair_id'], 'distance', donor['family_id'],
                    donor['target']['scene_id'])),
    ))

    weight_candidates = []
    for donor in neutral_options['weight']:
        if donor['changed_side'] != target['changed_side']:
            continue
        if str(donor['target']['outcome']) == desired:
            continue
        current = dict(donor)
        current['dose_match_error'] = abs(
            int(donor['physical_dose']) - int(target['physical_dose'])
        )
        weight_candidates.append(current)
    weight_candidates.sort(key=lambda donor: (
        donor['dose_match_error'],
        stable_key((target['pair_id'], 'weight', donor['family_id'],
                    donor['target']['scene_id'])),
    ))

    model_blind_pool.append({
        'pair_id': target['pair_id'],
        'target_row': target,
        'distance_candidates': distance_candidates[
            :DONOR_CANDIDATES_PER_TARGET
        ],
        'weight_candidates': weight_candidates[
            :DONOR_CANDIDATES_PER_TARGET
        ],
    })

assert model_blind_pool
write_jsonl_atomic(MODEL_BLIND_POOL_PATH, model_blind_pool)
pool_identity = {
    'schema_version': 'correctsource-distance-pool-1.0',
    'target_pair_ids': [row['pair_id'] for row in model_blind_pool],
    'distance_candidate_counts': {
        row['pair_id']: len(row['distance_candidates'])
        for row in model_blind_pool
    },
    'weight_candidate_counts': {
        row['pair_id']: len(row['weight_candidates'])
        for row in model_blind_pool
    },
    'excluded_prior_scene_count': len(excluded_scene_ids),
    'selection_uses_activation_data': False,
    'seed': SEED,
}
if POOL_LOCK_PATH.is_file():
    previous_pool = json.loads(POOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous_pool.get('identity') == pool_identity
else:
    write_json_atomic(POOL_LOCK_PATH, {
        'identity': pool_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

print('Distance-error targets:', len(distance_targets))
print('Neutral source options:', {
    factor: len(values) for factor, values in neutral_options.items()
})
print('Per-target distance candidates:', {
    row['pair_id']: len(row['distance_candidates'])
    for row in model_blind_pool
})
print('Per-target weight controls:', {
    row['pair_id']: len(row['weight_candidates'])
    for row in model_blind_pool
})

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=OUTPUT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {}
for example in outcome_examples:
    outcome_by_scene.setdefault(str(example.scene_id), example)
required_scene_ids = {
    str(role['scene_id'])
    for row in model_blind_pool
    for role in (
        row['target_row']['target'], row['target_row']['hybrid'],
        *[
            donor[role_name]
            for donor_type in ('distance_candidates', 'weight_candidates')
            for donor in row[donor_type]
            for role_name in ('target', 'hybrid')
        ],
    )
}
assert required_scene_ids <= set(outcome_by_scene)
```

```python
%%cell_progress 8/17 Define the frozen prompt, examples, and token masks
cell_stage('Reconstructing the exact readiness-screen prompt and mapping')
import math
from dataclasses import replace as dataclass_replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def mapping_for_variant(variant):
    semantics = (
        ('left', 'balanced', 'right')
        if variant == 'forward'
        else ('right', 'balanced', 'left')
    )
    return tuple(zip(('A', 'B', 'C'), semantics))

def custom_prompt(mapping):
    options = '\n'.join(
        f'{label}. {SEMANTIC_TEXT[semantic]}'
        for label, semantic in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT,
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, role_name):
    role = row[role_name]
    mapping = mapping_for_variant(row['mapping_variant'])
    semantics = tuple(value for _, value in mapping)
    labels = tuple(label for label, _ in mapping)
    outcome = str(role['outcome'])
    base = outcome_by_scene[str(role['scene_id'])]
    return dataclass_replace(
        base,
        example_id=f"correctsource:{row['pair_id']}:{role_name}",
        question_id=f"fliperror_{row['pair_id']}",
        question_kind='outcome',
        variant='correctsource_distance_rescue',
        prompt=custom_prompt(mapping),
        labels=labels,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=labels[semantics.index(outcome)],
        gold_semantic=outcome,
    )

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    assert len(sections) == len(names)
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans, cursor = [], 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key])
        masks = discover_masks(current)
        for scope in (UPSTREAM_SCOPE, SEMANTIC_SCOPE):
            assert torch.equal(reference_masks[scope], masks[scope])
    return reference_masks

print('Prompt, examples, and token masks ready.')
```

```python
%%cell_progress 9/17 Define capture, graded steering, and restoration helpers
cell_stage('Defining exact replay and equal-norm route controls')

def site_key(layer_index, component):
    return f'L{int(layer_index)}:{component}'

def score_records(prepared, logits):
    records = []
    for batch_index, example in enumerate(prepared['examples']):
        logprobs = torch.log_softmax(
            logits[batch_index, -1, :].float(), dim=-1
        )
        scores = [
            float(logprobs[prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_score(example):
    prepared = prepare_examples([example])
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    records = score_records(prepared, output.logits)
    assert len(records) == 1
    return records[0]

def semantic_margin(record, positive, negative):
    mapping = dict(zip(
        record['candidate_semantics'], record['candidate_logprobs']
    ))
    return float(mapping[positive] - mapping[negative])

def centered_error(first, second):
    first = np.asarray(first, dtype=float)
    second = np.asarray(second, dtype=float)
    return float(np.max(np.abs(
        (first - first.mean()) - (second - second.mean())
    )))

def forward_capture(prepared, upstream_delta=None):
    assert len(prepared['examples']) == 1
    masks = discover_masks(prepared)
    captured, handles = {}, []
    if upstream_delta is not None:
        selected = masks[UPSTREAM_SCOPE]
        def upstream_hook(_module, _inputs, output):
            hidden = output_hidden_tensor(output)
            current = selected.to(hidden.device)
            delta = upstream_delta.to(hidden.device, hidden.dtype)
            modified = hidden.clone()
            modified[0, current] = hidden[0, current] + delta[current]
            return replace_output_hidden(output, modified)
        handles.append(
            DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
        )
    def capture_l48(_module, _inputs, output):
        captured['L48:decoder'] = (
            output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
        )
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(capture_l48)
    )
    for (layer_index, component), module in COMPONENT_MODULES.items():
        key = site_key(layer_index, component)
        def capture(_module, _inputs, output, key=key):
            captured[key] = (
                output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
            )
        handles.append(module.register_forward_hook(capture))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    assert len(records) == 1
    return records[0], captured, masks

def forward_replace_l48(example, payload):
    prepared = prepare_examples([example])
    masks = discover_masks(prepared)
    selected = masks[UPSTREAM_SCOPE]
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = selected.to(hidden.device)
        modified = hidden.clone()
        modified[0, current] = payload.to(hidden.device, hidden.dtype)[current]
        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    return score_records(prepared, output.logits)[0]

def masked_removal(intervention_state, clean_state, mask):
    removal = torch.zeros_like(intervention_state, dtype=torch.bfloat16)
    selected = mask.bool()
    removal[selected] = (
        intervention_state[selected].float() - clean_state[selected].float()
    ).to(torch.bfloat16)
    return removal

def random_removal_like(removal, mask, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    selected = mask.bool()
    random_values = torch.randn(
        removal[selected].shape, generator=generator, dtype=torch.float32
    )
    target_norm = torch.linalg.vector_norm(removal[selected].float())
    random_norm = torch.linalg.vector_norm(random_values)
    if float(target_norm) > 0:
        random_values *= target_norm / random_norm
    else:
        random_values.zero_()
    output = torch.zeros_like(removal, dtype=torch.float32)
    output[selected] = random_values
    return output.to(torch.bfloat16)

def score_restorations(example, upstream_delta, conditions):
    prepared = prepare_examples([example] * len(conditions))
    single = prepare_examples([example])
    masks = discover_masks(single)
    upstream_mask = masks[UPSTREAM_SCOPE]
    handles = []
    measurements = [
        {'restored_norm': 0.0, 'random_norm': 0.0, 'restored_sites': 0}
        for _ in conditions
    ]
    def upstream_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        selected = upstream_mask.to(hidden.device)
        delta = upstream_delta.to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        for batch_index in range(len(conditions)):
            modified[batch_index, selected] = (
                hidden[batch_index, selected] + delta[selected]
            )
        return replace_output_hidden(output, modified)
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
    )
    union_keys = sorted({
        key for condition in conditions for key in condition['removals']
    })
    for key in union_keys:
        layer_text, component = key.split(':')
        module = COMPONENT_MODULES[(int(layer_text[1:]), component)]
        def restore_hook(_module, _inputs, output, key=key):
            hidden = output_hidden_tensor(output)
            modified, changed = hidden.clone(), False
            for batch_index, condition in enumerate(conditions):
                if key not in condition['removals']:
                    continue
                removal = condition['removals'][key].to(
                    hidden.device, hidden.dtype
                )
                modified[batch_index] = hidden[batch_index] - removal
                norm = float(torch.linalg.vector_norm(
                    removal.float()
                ).detach().cpu())
                measurements[batch_index]['restored_sites'] += 1
                field = (
                    'random_norm' if condition['random_control']
                    else 'restored_norm'
                )
                measurements[batch_index][field] += norm ** 2
                changed = True
            return replace_output_hidden(output, modified) if changed else None
        handles.append(module.register_forward_hook(restore_hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, measurement in zip(records, conditions, measurements):
        measurement['restored_norm'] **= 0.5
        measurement['random_norm'] **= 0.5
        record.update({'condition': condition['condition'], **measurement})
    return records

print('Capture, steering, and restoration helpers ready.')
```

```python
%%cell_progress 10/17 Freeze batch-one-stable targets and source pairs
cell_stage('Replaying every target and source endpoint before activation capture')

def generic_example(scene_role, mapping_variant, request_id):
    row = {
        'pair_id': request_id,
        'mapping_variant': mapping_variant,
        'request': scene_role,
    }
    return make_example(row, 'request')

request_by_key = {}
for pool_row in model_blind_pool:
    target_row = pool_row['target_row']
    variant = target_row['mapping_variant']
    roles = [target_row['target'], target_row['hybrid']]
    for donor_type in ('distance_candidates', 'weight_candidates'):
        for donor in pool_row[donor_type]:
            roles.extend([donor['target'], donor['hybrid']])
    for role in roles:
        key = request_key(role['scene_id'], variant)
        request = {
            'request_key': key,
            'scene_id': str(role['scene_id']),
            'mapping_variant': variant,
            'gold_semantic': str(role['outcome']),
            'role': role,
        }
        if key in request_by_key:
            assert request_by_key[key]['gold_semantic'] == request['gold_semantic']
        else:
            request_by_key[key] = request

behavior_requests = sorted(
    request_by_key.values(), key=lambda row: row['request_key']
)
write_jsonl_atomic(BEHAVIOR_REQUEST_PATH, behavior_requests)
behavior_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'pool_sha256': sha256_file(MODEL_BLIND_POOL_PATH),
    'request_sha256': sha256_file(BEHAVIOR_REQUEST_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'batch_size': 1,
}
BEHAVIOR_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(behavior_identity, sort_keys=True).encode('utf-8')
).hexdigest()

replay_progress = tqdm(
    behavior_requests, desc='Batch-one endpoint replay', unit='image-map'
)
for request in replay_progress:
    shard_path = BEHAVIOR_SHARD_DIR / (
        stable_key(request['request_key'])[:24] + '.json'
    )
    if shard_path.is_file():
        document = json.loads(shard_path.read_text(encoding='utf-8'))
        assert document['behavior_identity_sha256'] == BEHAVIOR_SHA256
        continue
    score = forward_score(generic_example(
        request['role'], request['mapping_variant'], request['request_key']
    ))
    score.update({
        'request_key': request['request_key'],
        'mapping_variant': request['mapping_variant'],
        'correct': bool(
            score['predicted_semantic'] == request['gold_semantic']
        ),
    })
    write_json_atomic(shard_path, {
        'schema_version': 'correctsource-behavior-shard-1.0',
        'behavior_identity_sha256': BEHAVIOR_SHA256,
        'record': score,
    })

behavior_records = []
for request in behavior_requests:
    shard_path = BEHAVIOR_SHARD_DIR / (
        stable_key(request['request_key'])[:24] + '.json'
    )
    document = json.loads(shard_path.read_text(encoding='utf-8'))
    assert document['behavior_identity_sha256'] == BEHAVIOR_SHA256
    behavior_records.append(document['record'])
write_jsonl_atomic(BEHAVIOR_REPLAY_PATH, behavior_records)
behavior_by_key = {row['request_key']: row for row in behavior_records}
assert len(behavior_by_key) == len(behavior_records)

eligible_targets = []
for pool_row in model_blind_pool:
    target_row = pool_row['target_row']
    variant = target_row['mapping_variant']
    target_result = behavior_by_key[
        request_key(target_row['target']['scene_id'], variant)
    ]
    hybrid_result = behavior_by_key[
        request_key(target_row['hybrid']['scene_id'], variant)
    ]
    stable_stale_error = bool(
        target_result['correct']
        and not hybrid_result['correct']
        and hybrid_result['predicted_semantic']
        == str(target_row['target']['outcome'])
    )
    if not stable_stale_error:
        continue

    valid_distance = []
    for donor in pool_row['distance_candidates']:
        first = behavior_by_key[
            request_key(donor['target']['scene_id'], variant)
        ]
        second = behavior_by_key[
            request_key(donor['hybrid']['scene_id'], variant)
        ]
        if first['correct'] and second['correct']:
            valid_distance.append(donor)
    valid_weight = []
    for donor in pool_row['weight_candidates']:
        first = behavior_by_key[
            request_key(donor['target']['scene_id'], variant)
        ]
        second = behavior_by_key[
            request_key(donor['hybrid']['scene_id'], variant)
        ]
        if first['correct'] and second['correct']:
            valid_weight.append(donor)
    if valid_distance and valid_weight:
        eligible_targets.append({
            'target_row': target_row,
            'target_result': target_result,
            'hybrid_result': hybrid_result,
            'distance_candidates': valid_distance,
            'weight_candidates': valid_weight,
        })

best_selected = []
for attempt in range(1024):
    ordered = sorted(eligible_targets, key=lambda row: (
        len(row['distance_candidates']) * len(row['weight_candidates']),
        stable_key((attempt, row['target_row']['pair_id'])),
    ))
    selected, used = [], set()
    for current in ordered:
        target_row = current['target_row']
        combinations = []
        for distance_donor in current['distance_candidates']:
            distance_scenes = {
                str(distance_donor[role]['scene_id'])
                for role in ('target', 'hybrid')
            }
            for weight_donor in current['weight_candidates']:
                weight_scenes = {
                    str(weight_donor[role]['scene_id'])
                    for role in ('target', 'hybrid')
                }
                if distance_scenes & weight_scenes:
                    continue
                combinations.append((distance_donor, weight_donor))
        combinations.sort(key=lambda pair: (
            int(pair[1].get('dose_match_error', 10**9)),
            stable_key((attempt, target_row['pair_id'],
                        pair[0]['target']['scene_id'],
                        pair[1]['target']['scene_id'])),
        ))
        chosen = None
        for distance_donor, weight_donor in combinations:
            donor_scenes = {
                str(donor[role]['scene_id'])
                for donor in (distance_donor, weight_donor)
                for role in ('target', 'hybrid')
            }
            if donor_scenes & used:
                continue
            chosen = (distance_donor, weight_donor, donor_scenes)
            break
        if chosen is None:
            continue
        distance_donor, weight_donor, donor_scenes = chosen
        selected.append({
            'pair_id': target_row['pair_id'],
            'family_id': target_row['family_id'],
            'mapping_variant': target_row['mapping_variant'],
            'changed_side': target_row['changed_side'],
            'factor_base': int(target_row['factor_base']),
            'factor_source': int(target_row['factor_source']),
            'target_context_value': int(target_row['context_value']),
            'target_physical_dose': int(target_row['physical_dose']),
            'target': target_row['target'],
            'hybrid': target_row['hybrid'],
            'distance_target': distance_donor['target'],
            'distance_hybrid': distance_donor['hybrid'],
            'distance_source_family_id': distance_donor['family_id'],
            'distance_context_value': int(distance_donor['context_value']),
            'distance_physical_dose': int(distance_donor['physical_dose']),
            'weight_target': weight_donor['target'],
            'weight_hybrid': weight_donor['hybrid'],
            'weight_source_family_id': weight_donor['family_id'],
            'weight_physical_dose': int(weight_donor['physical_dose']),
            'weight_dose_match_error': int(
                weight_donor.get('dose_match_error', 0)
            ),
            'runtime_target_prediction': current[
                'target_result'
            ]['predicted_semantic'],
            'runtime_hybrid_prediction': current[
                'hybrid_result'
            ]['predicted_semantic'],
        })
        used.update(donor_scenes)
        if len(selected) == TARGET_BUNDLES:
            break
    if len(selected) > len(best_selected):
        best_selected = selected
    if len(best_selected) == TARGET_BUNDLES:
        break

confirmation_manifest = sorted(
    best_selected, key=lambda row: row['pair_id']
)
assert confirmation_manifest, 'No behavior-stable source bundles are available.'
write_jsonl_atomic(CONFIRMATION_MANIFEST_PATH, confirmation_manifest)
confirmation_identity = {
    'schema_version': 'correctsource-distance-confirmation-1.0',
    'behavior_replay_sha256': sha256_file(BEHAVIOR_REPLAY_PATH),
    'bundle_count': len(confirmation_manifest),
    'pair_ids': [row['pair_id'] for row in confirmation_manifest],
    'source_scene_ids': sorted({
        str(row[role]['scene_id'])
        for row in confirmation_manifest
        for role in (
            'distance_target', 'distance_hybrid',
            'weight_target', 'weight_hybrid',
        )
    }),
    'selection_uses_behavior_only': True,
    'selection_uses_activation_data': False,
}
if CONFIRMATION_LOCK_PATH.is_file():
    previous_confirmation = json.loads(
        CONFIRMATION_LOCK_PATH.read_text(encoding='utf-8')
    )
    assert previous_confirmation.get('identity') == confirmation_identity
else:
    write_json_atomic(CONFIRMATION_LOCK_PATH, {
        'identity': confirmation_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

RUN_PRIMARY_TEST = len(confirmation_manifest) >= MIN_BUNDLES
print('Batch-one requests:', len(behavior_records))
print('Eligible target errors:', len(eligible_targets))
print('Frozen independent bundles:', len(confirmation_manifest))
print('Primary support sufficient:', RUN_PRIMARY_TEST)

def masked_delta(source_state, target_state, mask):
    output = torch.zeros_like(source_state, dtype=torch.bfloat16)
    selected = mask.bool()
    output[selected] = (
        source_state[selected].float() - target_state[selected].float()
    ).to(torch.bfloat16)
    return output

def rescale_delta(delta, mask, target_norm):
    selected = mask.bool()
    current_norm = torch.linalg.vector_norm(delta[selected].float())
    assert float(current_norm) > 0
    output = torch.zeros_like(delta, dtype=torch.float32)
    output[selected] = delta[selected].float() * (
        float(target_norm) / float(current_norm)
    )
    return output.to(torch.bfloat16)

# Override the inherited batched helper. Qwen-27B can change a near-tie
# prediction when the batch shape changes, so every confirmatory condition is
# scored as a single example.
def score_restorations(example, upstream_delta, conditions):
    records = []
    for condition in conditions:
        prepared = prepare_examples([example])
        masks = discover_masks(prepared)
        upstream_mask = masks[UPSTREAM_SCOPE]
        handles = []

        def upstream_hook(_module, _inputs, output):
            hidden = output_hidden_tensor(output)
            selected = upstream_mask.to(hidden.device)
            delta = upstream_delta.to(hidden.device, hidden.dtype)
            modified = hidden.clone()
            modified[0, selected] = hidden[0, selected] + delta[selected]
            return replace_output_hidden(output, modified)

        handles.append(
            DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(
                upstream_hook
            )
        )
        squared_norm = 0.0
        for key, removal in condition['removals'].items():
            layer_text, component = key.split(':')
            module = COMPONENT_MODULES[(int(layer_text[1:]), component)]

            def restore_hook(
                _module, _inputs, output, removal=removal
            ):
                hidden = output_hidden_tensor(output)
                current = removal.to(hidden.device, hidden.dtype)
                return replace_output_hidden(output, hidden - current)

            handles.append(module.register_forward_hook(restore_hook))
            squared_norm += float(torch.linalg.vector_norm(
                removal.float()
            )) ** 2
        try:
            with torch.inference_mode():
                output = adapter.model(**prepared['inputs'], use_cache=False)
        finally:
            for handle in handles:
                handle.remove()
        record = score_records(prepared, output.logits)[0]
        norm = squared_norm ** 0.5
        record.update({
            'condition': condition['condition'],
            'restored_norm': (
                0.0 if condition['random_control'] else norm
            ),
            'random_norm': (
                norm if condition['random_control'] else 0.0
            ),
            'restored_sites': len(condition['removals']),
        })
        records.append(record)
    return records

cell_stage('Checking alignment, self replay, and equal-norm controls')
smoke_row = confirmation_manifest[0]
smoke_roles = (
    'target', 'hybrid', 'distance_target', 'distance_hybrid',
    'weight_target', 'weight_hybrid',
)
smoke_examples = {
    role: make_example(smoke_row, role) for role in smoke_roles
}
smoke_prepared = {
    role: prepare_examples([example])
    for role, example in smoke_examples.items()
}
smoke_masks = assert_aligned(smoke_prepared)
smoke_scores, smoke_hidden = {}, {}
for role in smoke_roles:
    smoke_scores[role], smoke_hidden[role], _ = forward_capture(
        smoke_prepared[role]
    )
assert smoke_scores['target']['predicted_semantic'] == str(
    smoke_row['target']['outcome']
)
assert smoke_scores['hybrid']['predicted_semantic'] == str(
    smoke_row['target']['outcome']
)
for role in (
    'distance_target', 'distance_hybrid',
    'weight_target', 'weight_hybrid',
):
    assert smoke_scores[role]['predicted_semantic'] == str(
        smoke_row[role]['outcome']
    )

selected = smoke_masks[UPSTREAM_SCOPE]
own_delta = masked_delta(
    smoke_hidden['hybrid']['L48:decoder'],
    smoke_hidden['target']['L48:decoder'], selected,
)
reference_norm = torch.linalg.vector_norm(own_delta[selected].float())
distance_raw = masked_delta(
    smoke_hidden['distance_hybrid']['L48:decoder'],
    smoke_hidden['distance_target']['L48:decoder'], selected,
)
distance_delta = rescale_delta(distance_raw, selected, reference_norm)
random_delta = random_removal_like(
    own_delta, selected, f"smoke:{smoke_row['pair_id']}:random"
)
direct_score, direct_hidden, _ = forward_capture(
    smoke_prepared['hybrid'], ALPHAS[0] * distance_delta
)
real_removals, random_removals = {}, {}
for rank, site in enumerate(FROZEN_SITES, start=1):
    key = site['site_key']
    mask = smoke_masks[site['scope']]
    removal = masked_removal(
        direct_hidden[key], smoke_hidden['hybrid'][key], mask
    )
    real_removals[key] = removal
    random_removals[key] = random_removal_like(
        removal, mask,
        f"smoke:{smoke_row['pair_id']}:route:{rank}:{key}",
    )
restored = score_restorations(
    smoke_examples['hybrid'], ALPHAS[0] * distance_delta,
    [
        {'condition': 'distance_route_restore',
         'removals': real_removals, 'random_control': False},
        {'condition': 'distance_route_random',
         'removals': random_removals, 'random_control': True},
    ],
)
self_score = forward_replace_l48(
    smoke_examples['hybrid'], smoke_hidden['hybrid']['L48:decoder']
)
self_error = centered_error(
    self_score['candidate_logprobs'],
    smoke_scores['hybrid']['candidate_logprobs'],
)
upstream_norm_error = abs(
    float(torch.linalg.vector_norm(distance_delta[selected].float()))
    - float(torch.linalg.vector_norm(random_delta[selected].float()))
) / max(float(reference_norm), 1e-12)
route_norm_error = abs(
    restored[0]['restored_norm'] - restored[1]['random_norm']
) / max(restored[0]['restored_norm'], 1e-12)
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert upstream_norm_error < NORM_RELATIVE_TOLERANCE, upstream_norm_error
assert route_norm_error < NORM_RELATIVE_TOLERANCE, route_norm_error
print('Smoke pair:', smoke_row['pair_id'])
print('Self / upstream norm / route norm errors:',
      self_error, upstream_norm_error, route_norm_error)
```

```python
%%cell_progress 11/17 Run cross-context rescue and route restoration
cell_stage('Processing independent target/source bundles with resumable shards')

capture_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'confirmation_sha256': sha256_file(CONFIRMATION_MANIFEST_PATH),
    'behavior_replay_sha256': sha256_file(BEHAVIOR_REPLAY_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'frozen_sites': FROZEN_SITES,
    'alphas': list(ALPHAS),
    'conditions': list(CONDITIONS),
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(
    confirmation_manifest,
    desc='Correct-source distance rescue', unit='target'
)
for row in progress:
    shard_path = SHARD_DIR / f"{row['pair_id']}.json"
    if shard_path.is_file():
        existing = json.loads(shard_path.read_text(encoding='utf-8'))
        assert existing['capture_identity_sha256'] == CAPTURE_SHA256
        assert len(existing['records']) == len(ALPHAS) * len(CONDITIONS) + 1
        continue

    roles = (
        'target', 'hybrid', 'distance_target', 'distance_hybrid',
        'weight_target', 'weight_hybrid',
    )
    examples = {role: make_example(row, role) for role in roles}
    prepared = {
        role: prepare_examples([example]) for role, example in examples.items()
    }
    masks = assert_aligned(prepared)
    scores, hidden = {}, {}
    for role in roles:
        scores[role], hidden[role], _ = forward_capture(prepared[role])

    assert scores['target']['predicted_semantic'] == row[
        'runtime_target_prediction'
    ]
    assert scores['hybrid']['predicted_semantic'] == row[
        'runtime_hybrid_prediction'
    ]
    for role in (
        'distance_target', 'distance_hybrid',
        'weight_target', 'weight_hybrid',
    ):
        assert scores[role]['predicted_semantic'] == str(row[role]['outcome'])

    selected = masks[UPSTREAM_SCOPE]
    own_raw = masked_delta(
        hidden['hybrid']['L48:decoder'],
        hidden['target']['L48:decoder'], selected,
    )
    distance_raw = masked_delta(
        hidden['distance_hybrid']['L48:decoder'],
        hidden['distance_target']['L48:decoder'], selected,
    )
    weight_raw = masked_delta(
        hidden['weight_hybrid']['L48:decoder'],
        hidden['weight_target']['L48:decoder'], selected,
    )
    reference_norm = torch.linalg.vector_norm(own_raw[selected].float())
    assert float(reference_norm) > 0
    own_direction = rescale_delta(own_raw, selected, reference_norm)
    distance_direction = rescale_delta(
        distance_raw, selected, reference_norm
    )
    weight_direction = rescale_delta(weight_raw, selected, reference_norm)
    random_direction = random_removal_like(
        own_raw, selected, f"upstream:{row['pair_id']}:random"
    )

    positive = str(row['hybrid']['outcome'])
    negative = str(row['target']['outcome'])
    assert positive != negative
    clean_margin = semantic_margin(scores['hybrid'], positive, negative)
    pair_records = []
    for alpha in ALPHAS:
        deltas = {
            'own_failed': float(alpha) * own_direction,
            'correct_distance': float(alpha) * distance_direction,
            'reversed_distance': -float(alpha) * distance_direction,
            'weight_control': float(alpha) * weight_direction,
            'random_control': float(alpha) * random_direction,
        }
        direct_records = {}
        distance_hidden = None
        for condition, delta in deltas.items():
            score, current_hidden, _ = forward_capture(
                prepared['hybrid'], delta
            )
            score.update({
                'condition': condition,
                'restored_norm': 0.0,
                'random_norm': 0.0,
                'restored_sites': 0,
            })
            direct_records[condition] = score
            if condition == 'correct_distance':
                distance_hidden = current_hidden

        real_removals, random_removals = {}, {}
        for rank, site in enumerate(FROZEN_SITES, start=1):
            key = site['site_key']
            mask = masks[site['scope']]
            removal = masked_removal(
                distance_hidden[key], hidden['hybrid'][key], mask
            )
            real_removals[key] = removal
            random_removals[key] = random_removal_like(
                removal, mask,
                f"route:{row['pair_id']}:{alpha}:{rank}:{key}",
            )
        restored = score_restorations(
            examples['hybrid'], deltas['correct_distance'],
            [
                {'condition': 'distance_route_restore',
                 'removals': real_removals, 'random_control': False},
                {'condition': 'distance_route_random',
                 'removals': random_removals, 'random_control': True},
            ],
        )
        by_condition = {
            record['condition']: record
            for record in [*direct_records.values(), *restored]
        }
        assert set(by_condition) == set(CONDITIONS)

        for condition in CONDITIONS:
            record = by_condition[condition]
            margin = semantic_margin(record, positive, negative)
            direction_by_condition = {
                'own_failed': own_direction,
                'correct_distance': distance_direction,
                'reversed_distance': -distance_direction,
                'weight_control': weight_direction,
                'random_control': random_direction,
                'distance_route_restore': distance_direction,
                'distance_route_random': distance_direction,
            }
            intervention_norm = float(torch.linalg.vector_norm(
                (
                    float(alpha) * direction_by_condition[condition]
                )[selected].float()
            ))
            record.update({
                'schema_version': 'correctsource-distance-record-1.0',
                'pair_id': row['pair_id'],
                'family_id': row['family_id'],
                'alpha': float(alpha),
                'condition': condition,
                'mapping_variant': row['mapping_variant'],
                'changed_side': row['changed_side'],
                'target_context_value': row['target_context_value'],
                'distance_context_value': row['distance_context_value'],
                'target_physical_dose': row['target_physical_dose'],
                'distance_physical_dose': row['distance_physical_dose'],
                'weight_physical_dose': row['weight_physical_dose'],
                'weight_dose_match_error': row['weight_dose_match_error'],
                'positive_semantic': positive,
                'negative_semantic': negative,
                'clean_predicted_semantic': scores[
                    'hybrid'
                ]['predicted_semantic'],
                'clean_candidate_semantics': scores[
                    'hybrid'
                ]['candidate_semantics'],
                'clean_candidate_logprobs': scores[
                    'hybrid'
                ]['candidate_logprobs'],
                'clean_margin': clean_margin,
                'intervention_margin': margin,
                'margin_shift': margin - clean_margin,
                'rescue_success': float(
                    record['predicted_semantic'] == positive
                ),
                'reference_own_raw_norm': float(reference_norm),
                'distance_raw_norm': float(torch.linalg.vector_norm(
                    distance_raw[selected].float()
                )),
                'weight_raw_norm': float(torch.linalg.vector_norm(
                    weight_raw[selected].float()
                )),
                'upstream_intervention_norm': intervention_norm,
            })
            pair_records.append(record)

    self_score = forward_replace_l48(
        examples['hybrid'], hidden['hybrid']['L48:decoder']
    )
    self_score.update({
        'schema_version': 'correctsource-distance-record-1.0',
        'pair_id': row['pair_id'],
        'family_id': row['family_id'],
        'alpha': 0.0,
        'condition': 'self_replay',
        'clean_candidate_logprobs': scores['hybrid']['candidate_logprobs'],
        'centered_replay_error': centered_error(
            self_score['candidate_logprobs'],
            scores['hybrid']['candidate_logprobs'],
        ),
    })
    pair_records.append(self_score)
    write_json_atomic(shard_path, {
        'schema_version': 'correctsource-distance-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'bundle': row,
        'records': pair_records,
    })
    del prepared, hidden, own_raw, distance_raw, weight_raw
    torch.cuda.empty_cache()
    progress.set_postfix_str(row['pair_id'])

shard_paths = sorted(SHARD_DIR.glob('*.json'))
assert len(shard_paths) == len(confirmation_manifest)
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'capture_identity': capture_identity,
    'capture_identity_sha256': CAPTURE_SHA256,
    'bundles': len(confirmation_manifest),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Complete bundle shards:', len(shard_paths))
```

## 2. Cross-context causal rescue evidence


```python
%%cell_progress 12/17 Assemble cross-context rescue estimands
cell_stage('Releasing the model and assembling paired condition contrasts')
import gc
try:
    del adapter
    del DECODER_LAYERS
except NameError:
    pass
gc.collect()
torch.cuda.empty_cache()

records, self_records = [], []
for path in sorted(SHARD_DIR.glob('*.json')):
    document = json.loads(path.read_text(encoding='utf-8'))
    assert document['capture_identity_sha256'] == CAPTURE_SHA256
    for record in document['records']:
        if record['condition'] == 'self_replay':
            self_records.append(record)
        else:
            records.append(record)
record_frame = pd.DataFrame(records)
self_frame = pd.DataFrame(self_records)
assert len(record_frame) == (
    len(confirmation_manifest) * len(ALPHAS) * len(CONDITIONS)
)
assert len(self_frame) == len(confirmation_manifest)
record_frame.to_csv(
    ANALYSIS_DIR / 'cross_context_intervention_records.csv', index=False
)

cell_rows = []
for (pair_id, alpha), current in record_frame.groupby(['pair_id', 'alpha']):
    by_condition = {
        row['condition']: row for _, row in current.iterrows()
    }
    assert set(by_condition) == set(CONDITIONS)
    reference = by_condition['correct_distance']
    row = {
        'pair_id': pair_id,
        'family_id': reference['family_id'],
        'alpha': float(alpha),
        'mapping_variant': reference['mapping_variant'],
        'changed_side': reference['changed_side'],
        'target_context_value': reference['target_context_value'],
        'distance_context_value': reference['distance_context_value'],
        'target_physical_dose': reference['target_physical_dose'],
        'distance_physical_dose': reference['distance_physical_dose'],
        'weight_physical_dose': reference['weight_physical_dose'],
        'weight_dose_match_error': reference['weight_dose_match_error'],
    }
    for condition in CONDITIONS:
        row[f'{condition}_shift'] = float(
            by_condition[condition]['margin_shift']
        )
        row[f'{condition}_success'] = float(
            by_condition[condition]['rescue_success']
        )
    row['distance_vs_own'] = (
        row['correct_distance_shift'] - row['own_failed_shift']
    )
    row['distance_vs_reverse'] = (
        row['correct_distance_shift'] - row['reversed_distance_shift']
    )
    row['distance_vs_weight'] = (
        row['correct_distance_shift'] - row['weight_control_shift']
    )
    row['distance_vs_random'] = (
        row['correct_distance_shift'] - row['random_control_shift']
    )
    row['route_attenuation'] = (
        row['correct_distance_shift']
        - row['distance_route_restore_shift']
    )
    row['route_specificity'] = (
        row['distance_route_random_shift']
        - row['distance_route_restore_shift']
    )
    cell_rows.append(row)

cell_frame = pd.DataFrame(cell_rows)
cell_frame.to_csv(
    ANALYSIS_DIR / 'cross_context_rescue_estimands.csv', index=False
)
display(cell_frame.groupby('alpha')[[
    'own_failed_shift', 'correct_distance_shift',
    'reversed_distance_shift', 'weight_control_shift', 'random_control_shift',
    'distance_vs_own', 'distance_vs_reverse',
    'distance_vs_weight', 'distance_vs_random',
    'route_attenuation', 'route_specificity',
    'correct_distance_success', 'distance_route_restore_success',
]].mean().round(5))
```

```python
%%cell_progress 13/17 Estimate within-target graded rescue slopes
cell_stage('Reducing the three frozen strengths to one estimate per target')

def simple_slope(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    centered = x - x.mean()
    denominator = np.sum(centered ** 2)
    assert denominator > 0
    return float(np.sum(centered * (y - y.mean())) / denominator)

SLOPE_METRICS = (
    'own_failed_shift', 'correct_distance_shift',
    'reversed_distance_shift', 'weight_control_shift', 'random_control_shift',
    'distance_route_restore_shift', 'distance_route_random_shift',
    'distance_vs_own', 'distance_vs_reverse',
    'distance_vs_weight', 'distance_vs_random',
    'route_attenuation', 'route_specificity',
)
pair_rows = []
for pair_id, current in cell_frame.groupby('pair_id'):
    current = current.sort_values('alpha')
    row = {'pair_id': pair_id}
    for metric in SLOPE_METRICS:
        row[f'{metric}_slope'] = simple_slope(
            current['alpha'], current[metric]
        )
    maximum = current[current['alpha'] == max(ALPHAS)].iloc[0]
    for condition in CONDITIONS:
        row[f'max_{condition}_success'] = float(
            maximum[f'{condition}_success']
        )
    pair_rows.append(row)
pair_frame = pd.DataFrame(pair_rows)
pair_frame.to_csv(ANALYSIS_DIR / 'pair_rescue_slopes.csv', index=False)
display(pd.DataFrame([pair_frame.mean(numeric_only=True)]).round(5))
```

```python
%%cell_progress 14/17 Bootstrap the predeclared causal contrasts
cell_stage('Resampling independent failed target-hybrid pairs')

def bootstrap_mean(values, key):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    assert len(values) >= 1, (key, len(values))
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    estimates = values[draws].mean(axis=1)
    return {
        'estimate': float(values.mean()),
        'ci_low': float(np.quantile(estimates, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(estimates, 1 - ALPHA_LEVEL / 2)),
        'n_pairs': int(len(values)),
    }

METRICS = tuple(f'{metric}_slope' for metric in SLOPE_METRICS) + (
    'max_correct_distance_success',
    'max_own_failed_success',
    'max_weight_control_success',
    'max_random_control_success',
    'max_distance_route_restore_success',
    'max_distance_route_random_success',
)
test_rows = []
for metric in METRICS:
    summary = bootstrap_mean(pair_frame[metric], metric)
    test_rows.append({'metric': metric, **summary})
test_frame = pd.DataFrame(test_rows)
test_frame.to_csv(ANALYSIS_DIR / 'confirmatory_tests.csv', index=False)

def lookup(metric):
    rows = test_frame[test_frame['metric'] == metric]
    assert len(rows) == 1, metric
    return rows.iloc[0]

distance = lookup('correct_distance_shift_slope')
versus_own = lookup('distance_vs_own_slope')
versus_reverse = lookup('distance_vs_reverse_slope')
versus_weight = lookup('distance_vs_weight_slope')
versus_random = lookup('distance_vs_random_slope')
attenuation = lookup('route_attenuation_slope')
specificity = lookup('route_specificity_slope')
success = lookup('max_correct_distance_success')
gates = {
    'minimum_independent_support': bool(
        len(pair_frame) >= MIN_BUNDLES
    ),
    'correct_source_has_positive_graded_effect': bool(
        distance['ci_low'] > 0
    ),
    'correct_source_beats_failed_own_update': bool(
        versus_own['ci_low'] > 0
    ),
    'correct_source_beats_reversed_distance': bool(
        versus_reverse['ci_low'] > 0
    ),
    'correct_source_beats_weight_control': bool(
        versus_weight['ci_low'] > 0
    ),
    'correct_source_beats_random_control': bool(
        versus_random['ci_low'] > 0
    ),
    'correct_source_reaches_any_answers': bool(
        success['estimate'] > 0
    ),
    'frozen_route_attenuates_rescue': bool(
        attenuation['ci_low'] > 0
    ),
    'frozen_route_specific_over_random': bool(
        specificity['ci_low'] > 0
    ),
}
if all(gates.values()):
    PRIMARY_DECISION = 'missing_distance_update_confirmed'
elif not gates['minimum_independent_support']:
    PRIMARY_DECISION = 'insufficient_independent_cross_context_sources'
elif (
    gates['correct_source_has_positive_graded_effect']
    and gates['correct_source_reaches_any_answers']
):
    PRIMARY_DECISION = 'correct_source_rescues_without_full_specificity'
elif not gates['correct_source_beats_failed_own_update']:
    PRIMARY_DECISION = 'correct_source_not_better_than_failed_own_update'
else:
    PRIMARY_DECISION = 'distance_errors_not_rescued_by_correct_source_update'

decision_document = {
    'schema_version': 'correctsource-distance-decision-1.0',
    'primary_decision': PRIMARY_DECISION,
    'gates': {key: bool(value) for key, value in gates.items()},
    'bundle_count': len(pair_frame),
    'claim_boundary': (
        'The inference is conditional on stale-answer distance errors and '
        'answer-neutral correctly solved sources; it does not estimate '
        'unconditional physical competence.'
    ),
}
write_json_atomic(
    ANALYSIS_DIR / 'correctsource_distance_decision.json',
    decision_document,
)
display(test_frame.round(6))
print('Decision gates:', gates)
print('Primary decision:', PRIMARY_DECISION)
```

```python
%%cell_progress 15/17 Audit geometry, behavior, norms, and provenance
cell_stage('Verifying source meaning, independence, replay, and equal-norm controls')

violations = []
selected_scene_ids = []
mapping_counts = Counter()
for row in confirmation_manifest:
    desired = str(row['hybrid']['outcome'])
    if str(row['target']['outcome']) == desired:
        violations.append((row['pair_id'], 'target_not_outcome_flip'))
    if row['runtime_target_prediction'] != str(row['target']['outcome']):
        violations.append((row['pair_id'], 'target_not_correct'))
    if row['runtime_hybrid_prediction'] != str(row['target']['outcome']):
        violations.append((row['pair_id'], 'hybrid_not_stale_answer'))

    if (
        row['distance_target']['outcome']
        != row['distance_hybrid']['outcome']
        or str(row['distance_target']['outcome']) == desired
    ):
        violations.append((row['pair_id'], 'distance_source_answer_leak'))
    if (
        int(row['distance_target']['distance']) != int(row['factor_base'])
        or int(row['distance_hybrid']['distance'])
        != int(row['factor_source'])
        or int(row['distance_target']['units'])
        != int(row['distance_hybrid']['units'])
        or int(row['distance_context_value'])
        == int(row['target_context_value'])
    ):
        violations.append((row['pair_id'], 'distance_source_geometry'))

    if (
        row['weight_target']['outcome'] != row['weight_hybrid']['outcome']
        or str(row['weight_target']['outcome']) == desired
        or int(row['weight_target']['distance'])
        != int(row['weight_hybrid']['distance'])
        or int(row['weight_target']['units'])
        == int(row['weight_hybrid']['units'])
    ):
        violations.append((row['pair_id'], 'weight_control_geometry'))

    variant = row['mapping_variant']
    for role in (
        'distance_target', 'distance_hybrid',
        'weight_target', 'weight_hybrid',
    ):
        replay = behavior_by_key[request_key(row[role]['scene_id'], variant)]
        if not replay['correct']:
            violations.append((row['pair_id'], role + '_not_correct'))
    selected_scene_ids.extend(
        str(row[role]['scene_id'])
        for role in (
            'target', 'hybrid', 'distance_target', 'distance_hybrid',
            'weight_target', 'weight_hybrid',
        )
    )
    mapping_counts[variant] += 1

self_errors = self_frame['centered_replay_error'].to_numpy(dtype=float)
upstream_norm_errors = []
for _, current in record_frame.groupby(['pair_id', 'alpha']):
    norms = current.groupby('condition')[
        'upstream_intervention_norm'
    ].first()
    upstream_norm_errors.append(
        (float(norms.max()) - float(norms.min()))
        / max(float(norms.max()), 1e-12)
    )
route_norm_errors = []
for _, current in record_frame.groupby(['pair_id', 'alpha']):
    real = current[
        current['condition'] == 'distance_route_restore'
    ].iloc[0]
    random = current[
        current['condition'] == 'distance_route_random'
    ].iloc[0]
    route_norm_errors.append(abs(
        float(real['restored_norm']) - float(random['random_norm'])
    ) / max(float(real['restored_norm']), 1e-12))
nonfinite_count = int((~np.isfinite(
    cell_frame.select_dtypes(include=[np.number]).to_numpy()
)).sum())

assert not violations, violations[:10]
assert len(selected_scene_ids) == len(set(selected_scene_ids))
assert max(self_errors) < SMOKE_CENTERED_LOGPROB_TOLERANCE
assert max(upstream_norm_errors) < NORM_RELATIVE_TOLERANCE
assert max(route_norm_errors) < NORM_RELATIVE_TOLERANCE
assert nonfinite_count == 0

audit_document = {
    'schema_version': 'correctsource-distance-audit-1.0',
    'bundles': len(confirmation_manifest),
    'minimum_support_met': RUN_PRIMARY_TEST,
    'behavior_selection_only': True,
    'activation_selection_used': False,
    'sources_answer_neutral': True,
    'distance_change_exactly_matched': True,
    'distance_weight_context_changed': True,
    'source_answer_differs_from_desired_target_answer': True,
    'route_frozen_from_prior_work': True,
    'all_selected_scenes_unique': True,
    'behavior_or_geometry_violations': len(violations),
    'maximum_self_replay_error': float(max(self_errors)),
    'maximum_upstream_equal_norm_relative_error': float(
        max(upstream_norm_errors)
    ),
    'maximum_route_equal_norm_relative_error': float(
        max(route_norm_errors)
    ),
    'maximum_weight_dose_match_error': int(max(
        row['weight_dose_match_error'] for row in confirmation_manifest
    )),
    'nonfinite_count': nonfinite_count,
    'mapping_counts': dict(mapping_counts),
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_document)
display(pd.DataFrame([audit_document]).T.rename(columns={0: 'value'}))
```
