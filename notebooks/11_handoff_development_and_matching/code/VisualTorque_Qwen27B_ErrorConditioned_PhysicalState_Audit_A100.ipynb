```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Build the exhaustive single-stack support
cell_stage('Loading all 900 bounded single-stack configurations and questions')
import itertools
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')

def is_single_stack_scene(row):
    stacks = row.get('stacks', [])
    side_counts = Counter(
        str(stack.get('side')) for stack in stacks
    )
    metadata_family = str(row.get('metadata', {}).get('family', ''))
    return bool(
        metadata_family == 'single_stack'
        and str(row.get('scene_id', '')).startswith('single_')
        and len(stacks) == 2
        and side_counts == {'left': 1, 'right': 1}
    )

single_scenes = [
    row for row in scene_records if is_single_stack_scene(row)
]
assert len(single_scenes) == 900, {
    'single_stack_scenes': len(single_scenes),
    'total_scenes': len(scene_records),
    'metadata_families': dict(Counter(
        str(row.get('metadata', {}).get('family', ''))
        for row in scene_records
    )),
    'note': (
        'problem_type describes the physical cue pattern; it is not the '
        'single-stack versus multi-stack structural family.'
    ),
}
scene_by_id = {str(row['scene_id']): row for row in single_scenes}

def scene_features(scene):
    left = next(stack for stack in scene['stacks'] if stack['side'] == 'left')
    right = next(stack for stack in scene['stacks'] if stack['side'] == 'right')
    derived = scene['derived']
    left_units = int(left['units'])
    right_units = int(right['units'])
    left_distance = int(left['distance'])
    right_distance = int(right['distance'])
    left_torque = int(derived['left_torque'])
    right_torque = int(derived['right_torque'])
    net_torque = int(derived['net_torque'])
    def sign(value):
        return int(value > 0) - int(value < 0)
    weight_cue = sign(left_units - right_units)
    distance_cue = sign(left_distance - right_distance)
    torque_sign = sign(net_torque)
    if torque_sign != 0 and weight_cue != torque_sign and distance_cue == torque_sign:
        diagnostic_cell = 'weight_conflict'
    elif torque_sign != 0 and distance_cue != torque_sign and weight_cue == torque_sign:
        diagnostic_cell = 'distance_conflict'
    elif torque_sign != 0 and weight_cue == torque_sign and distance_cue == torque_sign:
        diagnostic_cell = 'both_cues_aligned'
    else:
        diagnostic_cell = 'mixed_or_balanced'
    return {
        'left_units': left_units,
        'right_units': right_units,
        'left_distance': left_distance,
        'right_distance': right_distance,
        'left_torque': left_torque,
        'right_torque': right_torque,
        'net_torque': net_torque,
        'outcome': str(derived['outcome']),
        'weight_cue': weight_cue,
        'distance_cue': distance_cue,
        'torque_sign': torque_sign,
        'diagnostic_cell': diagnostic_cell,
        'absolute_net_torque': abs(net_torque),
        'margin_bin': min(abs(net_torque) // 4, 5),
    }

physical_by_scene = {
    scene_id: scene_features(scene)
    for scene_id, scene in scene_by_id.items()
}

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=OUTPUT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=QUESTION_KINDS,
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1,
    generation_every=0,
    seed=SEED,
)
all_examples = tuple(load_behavior_examples(example_config))
all_examples = tuple(
    example for example in all_examples
    if str(example.scene_id) in scene_by_id
)

examples_by_scene_kind = defaultdict(list)
for example in all_examples:
    examples_by_scene_kind[(str(example.scene_id), example.question_kind)].append(
        example
    )

outcome_example_by_scene = {}
groups = defaultdict(list)
for scene in single_scenes:
    groups[(scene['split'], scene['derived']['outcome'])].append(scene)
for group_key, scenes in sorted(groups.items()):
    scenes.sort(key=lambda row: stable_key(('outcome-order', row['scene_id'])))
    for index, scene in enumerate(scenes):
        scene_id = str(scene['scene_id'])
        desired_variant = f'permutation_{index % 6}'
        candidates = examples_by_scene_kind[(scene_id, 'outcome')]
        selected = next(
            example for example in candidates
            if example.variant == desired_variant
        )
        outcome_example_by_scene[scene_id] = selected

task_example_by_scene_kind = {}
for scene_id in scene_by_id:
    for kind in QUESTION_KINDS:
        if kind == 'outcome':
            task_example_by_scene_kind[(scene_id, kind)] = (
                outcome_example_by_scene[scene_id]
            )
        else:
            values = examples_by_scene_kind[(scene_id, kind)]
            assert len(values) == 1, (scene_id, kind, len(values))
            task_example_by_scene_kind[(scene_id, kind)] = values[0]

outcome_manifest = []
for scene in sorted(single_scenes, key=lambda row: row['scene_id']):
    scene_id = str(scene['scene_id'])
    example = outcome_example_by_scene[scene_id]
    outcome_manifest.append({
        'scene_id': scene_id,
        'split': str(scene['split']),
        'outcome': str(scene['derived']['outcome']),
        'variant': str(example.variant),
        'gold_label': str(example.gold_label),
        **physical_by_scene[scene_id],
    })
write_jsonl_atomic(OUTCOME_MANIFEST_PATH, outcome_manifest)

mapping_counts = Counter(
    (row['split'], row['outcome'], row['variant'])
    for row in outcome_manifest
)
assert len(outcome_manifest) == 900
assert set(row['split'] for row in outcome_manifest) == {
    'probe_train', 'probe_dev', 'probe_test'
}
print('Scene splits:', Counter(row['split'] for row in outcome_manifest))
print('Outcome mappings balanced within one count:',
      max(mapping_counts.values()) - min(mapping_counts.values()) <= 1)
```

```python
%%cell_progress 7/17 Load pinned Qwen-27B and decoder layers
cell_stage('Loading Qwen-27B and resolving the frozen capture depths')
import gc
import itertools

if 'adapter' in globals():
    del adapter
    gc.collect()
    torch.cuda.empty_cache()

adapter = HuggingFaceMultimodalAdapter(
    MODEL_ID,
    revision=MODEL_REVISION,
    dtype=MODEL_DTYPE,
    device_map=DEVICE_MAP,
)
display(adapter.metadata)

def resolve_module_path(root, path):
    value = root
    for component in path.split('.'):
        value = getattr(value, component)
    return value

def discover_decoder_layers(model):
    candidates = (
        'model.language_model.layers', 'model.text_model.layers',
        'language_model.model.layers', 'model.model.layers', 'model.layers',
    )
    for path in candidates:
        try:
            layers = resolve_module_path(model, path)
        except AttributeError:
            continue
        if isinstance(layers, torch.nn.ModuleList) and len(layers) >= 8:
            return path, layers
    raise RuntimeError('Could not locate decoder layers.')

LAYER_PATH, DECODER_LAYERS = discover_decoder_layers(adapter.model)
assert len(DECODER_LAYERS) == EXPECTED_LAYER_COUNT
print('Layer path:', LAYER_PATH)
print('Decoder layers:', len(DECODER_LAYERS))
```

```python
%%cell_progress 8/17 Define exact scoring and hidden-state capture
cell_stage('Defining batch-one A/B/C scoring and final-position capture')

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def prepare_example(example):
    plan = adapter._candidate_plan(example)
    if plan is None:
        raise RuntimeError('This audit requires single-token A/B/C labels.')
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages()], add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'example': example,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def score_from_logits(prepared, logits):
    example = prepared['example']
    attention_mask = prepared['inputs']['attention_mask']
    final_index = int(attention_mask[0].sum().item()) - 1
    next_logprobs = torch.log_softmax(
        logits[0, final_index].float(), dim=-1
    )
    labels = tuple(example.labels)
    values = [
        float(next_logprobs[prepared['token_by_label'][label]].detach().cpu())
        for label in labels
    ]
    best_index = int(np.argmax(values))
    predicted_label = labels[best_index]
    predicted_semantic = example.label_to_semantic[predicted_label]
    return {
        'scene_id': str(example.scene_id),
        'question_kind': str(example.question_kind),
        'variant': str(example.variant),
        'split': str(example.split),
        'candidate_labels': list(labels),
        'candidate_semantics': list(example.option_semantics),
        'candidate_logprobs': values,
        'predicted_label': str(predicted_label),
        'predicted_semantic': predicted_semantic,
        'gold_label': str(example.gold_label),
        'gold_semantic': example.gold_semantic,
        'correct': bool(predicted_semantic == example.gold_semantic),
        'gold_logprob': float(values[labels.index(example.gold_label)]),
        'top_logprob': float(max(values)),
        'confidence_gap': float(
            sorted(values, reverse=True)[0] - sorted(values, reverse=True)[1]
        ),
    }

def forward_capture(example, capture_hidden=True):
    prepared = prepare_example(example)
    captured = {}
    handles = []
    if capture_hidden:
        for layer_index in CAPTURE_LAYERS:
            def hook(_module, _inputs, output, layer_index=layer_index):
                hidden = output_hidden_tensor(output)
                attention_mask = prepared['inputs']['attention_mask']
                final_index = int(attention_mask[0].sum().item()) - 1
                captured[layer_index] = (
                    hidden[0, final_index].detach().cpu().to(torch.bfloat16)
                )
            handles.append(
                DECODER_LAYERS[layer_index].register_forward_hook(hook)
            )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_from_logits(prepared, output.logits)
    if capture_hidden:
        assert set(captured) == set(CAPTURE_LAYERS)
    return record, captured

def save_torch_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    torch.save(value, temporary)
    temporary.replace(path)

def torch_load_owned(path):
    return torch.load(path, map_location='cpu', weights_only=False)

smoke_scene_id = outcome_manifest[0]['scene_id']
smoke_score, smoke_hidden = forward_capture(
    outcome_example_by_scene[smoke_scene_id]
)
assert smoke_score['scene_id'] == smoke_scene_id
assert set(smoke_hidden) == set(CAPTURE_LAYERS)
assert all(tensor.ndim == 1 for tensor in smoke_hidden.values())
print('Smoke score:', smoke_score['predicted_semantic'],
      '| gold:', smoke_score['gold_semantic'])
print('Hidden width:', smoke_hidden[CAPTURE_LAYERS[0]].numel())
```

```python
%%cell_progress 9/17 Score and capture all 900 outcome scenes
cell_stage('Running resumable exhaustive outcome capture')

capture_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'outcome_manifest_sha256': sha256_file(OUTCOME_MANIFEST_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'layers': list(CAPTURE_LAYERS),
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()

ordered_scene_ids = [row['scene_id'] for row in outcome_manifest]
outcome_chunks = [
    ordered_scene_ids[index:index + OUTCOME_SHARD_SIZE]
    for index in range(0, len(ordered_scene_ids), OUTCOME_SHARD_SIZE)
]
progress = tqdm(
    enumerate(outcome_chunks), total=len(outcome_chunks),
    desc='Exhaustive outcome capture', unit='shard',
)
for shard_index, scene_ids in progress:
    shard_path = OUTCOME_SHARD_DIR / f'outcome_{shard_index:03d}.pt'
    if shard_path.is_file():
        existing = torch_load_owned(shard_path)
        assert existing['capture_identity_sha256'] == CAPTURE_SHA256
        assert existing['scene_ids'] == scene_ids
        continue
    records = []
    layer_rows = {layer: [] for layer in CAPTURE_LAYERS}
    for scene_id in scene_ids:
        score, hidden = forward_capture(outcome_example_by_scene[scene_id])
        score.update(physical_by_scene[scene_id])
        records.append(score)
        for layer in CAPTURE_LAYERS:
            layer_rows[layer].append(hidden[layer])
    payload = {
        'schema_version': 'outcome-hidden-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'scene_ids': scene_ids,
        'records': records,
        'hidden': {
            str(layer): torch.stack(layer_rows[layer])
            for layer in CAPTURE_LAYERS
        },
    }
    save_torch_atomic(shard_path, payload)
    torch.cuda.empty_cache()
    progress.set_postfix_str(scene_ids[-1])

outcome_records = []
hidden_rows = {layer: [] for layer in CAPTURE_LAYERS}
captured_scene_ids = []
for shard_index in range(len(outcome_chunks)):
    payload = torch_load_owned(
        OUTCOME_SHARD_DIR / f'outcome_{shard_index:03d}.pt'
    )
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    outcome_records.extend(payload['records'])
    captured_scene_ids.extend(payload['scene_ids'])
    for layer in CAPTURE_LAYERS:
        hidden_rows[layer].append(payload['hidden'][str(layer)])

assert captured_scene_ids == ordered_scene_ids
assert len(outcome_records) == 900
OUTCOME_HIDDEN = {
    layer: torch.cat(hidden_rows[layer], dim=0).float().numpy()
    for layer in CAPTURE_LAYERS
}
outcome_frame = pd.DataFrame(outcome_records)
outcome_frame.to_csv(ANALYSIS_DIR / 'all_outcome_scores.csv', index=False)
print(outcome_frame.groupby('split')['correct'].agg(['count', 'mean']))
```

```python
%%cell_progress 10/17 Select physical readouts on train and dev only
cell_stage('Selecting layer and ridge strength without opening probe_test labels')
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

metadata = pd.DataFrame(outcome_manifest)
assert metadata['scene_id'].tolist() == captured_scene_ids
train_indices = np.flatnonzero(metadata['split'].to_numpy() == 'probe_train')
dev_indices = np.flatnonzero(metadata['split'].to_numpy() == 'probe_dev')
test_indices = np.flatnonzero(metadata['split'].to_numpy() == 'probe_test')
assert not (set(train_indices) & set(dev_indices))
assert not (set(train_indices) & set(test_indices))
assert not (set(dev_indices) & set(test_indices))

target_matrix = metadata[list(PROBE_TARGETS)].to_numpy(dtype=float)
train_target_mean = target_matrix[train_indices].mean(axis=0)
train_target_scale = target_matrix[train_indices].std(axis=0)
assert np.all(train_target_scale > 0)
standard_targets = (
    (target_matrix - train_target_mean[None, :])
    / train_target_scale[None, :]
)

selection_rows = []
for layer in tqdm(CAPTURE_LAYERS, desc='Probe layer selection', unit='layer'):
    X = OUTCOME_HIDDEN[layer]
    for alpha in RIDGE_ALPHAS:
        model = make_pipeline(
            StandardScaler(), Ridge(alpha=alpha, solver='lsqr')
        )
        model.fit(X[train_indices], standard_targets[train_indices])
        prediction = model.predict(X[dev_indices])
        for target_index, target in enumerate(PROBE_TARGETS):
            error = mean_absolute_error(
                standard_targets[dev_indices, target_index],
                prediction[:, target_index],
            )
            selection_rows.append({
                'target': target,
                'layer': int(layer),
                'alpha': float(alpha),
                'dev_standardized_mae': float(error),
            })

selection_frame = pd.DataFrame(selection_rows)
frozen_selection = {}
for target in PROBE_TARGETS:
    current = selection_frame[selection_frame['target'] == target].copy()
    current = current.sort_values(
        ['dev_standardized_mae', 'layer', 'alpha']
    )
    best = current.iloc[0]
    frozen_selection[target] = {
        'layer': int(best['layer']),
        'alpha': float(best['alpha']),
        'dev_standardized_mae': float(best['dev_standardized_mae']),
    }

probe_identity = {
    'schema_version': 'physical-readout-selection-1.0',
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'capture_identity_sha256': CAPTURE_SHA256,
    'fit_split': 'probe_train',
    'selection_split': 'probe_dev',
    'test_split_opened_for_selection': False,
    'selection': frozen_selection,
}
if PROBE_LOCK_PATH.is_file():
    previous = json.loads(PROBE_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == probe_identity
else:
    write_json_atomic(PROBE_LOCK_PATH, {
        'identity': probe_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
selection_frame.to_csv(ANALYSIS_DIR / 'probe_dev_selection.csv', index=False)
display(pd.DataFrame(frozen_selection).T)
```

```python
%%cell_progress 11/17 Freeze behavior-only error and matched-control scenes
cell_stage('Selecting probe_test errors before inspecting probe predictions or tasks')

test_behavior = outcome_frame[outcome_frame['split'] == 'probe_test'].copy()
test_behavior['scene_id'] = test_behavior['scene_id'].astype(str)

error_rows = test_behavior[~test_behavior['correct']].copy()
error_rows['selection_key'] = error_rows.apply(
    lambda row: stable_key((
        'error', row['outcome'], row['weight_cue'], row['distance_cue'],
        row['margin_bin'], row['scene_id'],
    )),
    axis=1,
)

error_strata = {}
for key, current in error_rows.groupby(
    ['outcome', 'weight_cue', 'distance_cue'], sort=True
):
    error_strata[key] = current.sort_values('selection_key').to_dict('records')
stratum_keys = sorted(
    error_strata,
    key=lambda key: stable_key(('error-stratum', key)),
)
stratum_cursors = {key: 0 for key in stratum_keys}
selected_errors = []
while len(selected_errors) < min(MAX_CONFIRMATION_ERRORS, len(error_rows)):
    changed = False
    for key in stratum_keys:
        cursor = stratum_cursors[key]
        values = error_strata[key]
        if cursor >= len(values):
            continue
        selected_errors.append(values[cursor])
        stratum_cursors[key] += 1
        changed = True
        if len(selected_errors) == min(MAX_CONFIRMATION_ERRORS, len(error_rows)):
            break
    if not changed:
        break

correct_pool = test_behavior[test_behavior['correct']].copy()
used_controls = set()
selected_controls = []

def control_distance(error, control):
    exact_penalty = 0.0
    exact_penalty += 100.0 * (str(error['outcome']) != str(control['outcome']))
    exact_penalty += 20.0 * (int(error['weight_cue']) != int(control['weight_cue']))
    exact_penalty += 20.0 * (int(error['distance_cue']) != int(control['distance_cue']))
    numerical = (
        abs(int(error['margin_bin']) - int(control['margin_bin']))
        + 0.25 * abs(int(error['left_units']) - int(control['left_units']))
        + 0.25 * abs(int(error['right_units']) - int(control['right_units']))
        + 0.15 * abs(int(error['left_distance']) - int(control['left_distance']))
        + 0.15 * abs(int(error['right_distance']) - int(control['right_distance']))
    )
    return exact_penalty + numerical

for error in selected_errors:
    candidates = correct_pool[
        ~correct_pool['scene_id'].isin(used_controls)
    ].copy()
    if candidates.empty:
        break
    candidates['match_distance'] = candidates.apply(
        lambda control: control_distance(error, control), axis=1
    )
    candidates['match_key'] = candidates['scene_id'].map(
        lambda scene_id: stable_key(('control', error['scene_id'], scene_id))
    )
    chosen = candidates.sort_values(
        ['match_distance', 'match_key']
    ).iloc[0].to_dict()
    used_controls.add(str(chosen['scene_id']))
    chosen['matched_error_scene_id'] = str(error['scene_id'])
    selected_controls.append(chosen)

paired_count = min(len(selected_errors), len(selected_controls))
selected_errors = selected_errors[:paired_count]
selected_controls = selected_controls[:paired_count]
assert paired_count > 0, 'No probe_test error/control pairs were available.'

confirmation_manifest = []
for row in selected_errors:
    confirmation_manifest.append({
        'scene_id': str(row['scene_id']),
        'cohort': 'outcome_error',
        'matched_scene_id': next(
            str(control['scene_id']) for control in selected_controls
            if control['matched_error_scene_id'] == str(row['scene_id'])
        ),
        'outcome_prediction': row['predicted_semantic'],
        'outcome_gold': row['gold_semantic'],
        'outcome_confidence_gap': float(row['confidence_gap']),
    })
for row in selected_controls:
    confirmation_manifest.append({
        'scene_id': str(row['scene_id']),
        'cohort': 'matched_correct',
        'matched_scene_id': str(row['matched_error_scene_id']),
        'outcome_prediction': row['predicted_semantic'],
        'outcome_gold': row['gold_semantic'],
        'outcome_confidence_gap': float(row['confidence_gap']),
        'match_distance': float(row['match_distance']),
    })
confirmation_manifest.sort(key=lambda row: (row['cohort'], row['scene_id']))
write_jsonl_atomic(CONFIRMATION_MANIFEST_PATH, confirmation_manifest)

confirmation_identity = {
    'schema_version': 'error-matched-control-scenes-1.0',
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'outcome_capture_sha256': CAPTURE_SHA256,
    'selection_uses_behavior_only': True,
    'selection_uses_hidden_states': False,
    'selection_uses_probe_predictions': False,
    'selection_uses_intermediate_tasks': False,
    'split': 'probe_test',
    'error_count': paired_count,
    'control_count': paired_count,
    'scene_ids': [row['scene_id'] for row in confirmation_manifest],
}
if CONFIRMATION_LOCK_PATH.is_file():
    previous = json.loads(CONFIRMATION_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == confirmation_identity
else:
    write_json_atomic(CONFIRMATION_LOCK_PATH, {
        'identity': confirmation_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

print('Probe-test outcome errors available:', len(error_rows))
print('Frozen error/control pairs:', paired_count)
print('Maximum control match distance:',
      max(float(row.get('match_distance', 0.0)) for row in selected_controls))
```

```python
%%cell_progress 12/17 Score intermediate physical questions on frozen scenes
cell_stage('Scoring weight, distance, side torque, and net torque')

confirmation_scene_ids = sorted({
    row['scene_id'] for row in confirmation_manifest
})
task_kinds = tuple(kind for kind in QUESTION_KINDS if kind != 'outcome')
task_chunks = [
    confirmation_scene_ids[index:index + TASK_SHARD_SIZE]
    for index in range(0, len(confirmation_scene_ids), TASK_SHARD_SIZE)
]
task_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'confirmation_manifest_sha256': sha256_file(CONFIRMATION_MANIFEST_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'task_kinds': list(task_kinds),
}
TASK_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(task_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(
    enumerate(task_chunks), total=len(task_chunks),
    desc='Intermediate physical questions', unit='shard',
)
for shard_index, scene_ids in progress:
    shard_path = TASK_SHARD_DIR / f'tasks_{shard_index:03d}.json'
    if shard_path.is_file():
        existing = json.loads(shard_path.read_text(encoding='utf-8'))
        assert existing['task_identity_sha256'] == TASK_SHA256
        assert existing['scene_ids'] == scene_ids
        continue
    records = []
    for scene_id in scene_ids:
        for kind in task_kinds:
            score, _ = forward_capture(
                task_example_by_scene_kind[(scene_id, kind)],
                capture_hidden=False,
            )
            records.append(score)
    write_json_atomic(shard_path, {
        'schema_version': 'intermediate-task-shard-1.0',
        'task_identity_sha256': TASK_SHA256,
        'scene_ids': scene_ids,
        'records': records,
    })
    progress.set_postfix_str(scene_ids[-1])

task_records = []
for shard_index in range(len(task_chunks)):
    payload = json.loads(
        (TASK_SHARD_DIR / f'tasks_{shard_index:03d}.json').read_text(
            encoding='utf-8'
        )
    )
    assert payload['task_identity_sha256'] == TASK_SHA256
    task_records.extend(payload['records'])
task_frame = pd.DataFrame(task_records)
expected_task_records = len(confirmation_scene_ids) * len(task_kinds)
assert len(task_frame) == expected_task_records
task_frame.to_csv(ANALYSIS_DIR / 'intermediate_task_scores.csv', index=False)
display(task_frame.groupby('question_kind')['correct'].agg(['count', 'mean']))
```

## 3. Error-conditioned physical-state map


```python
%%cell_progress 13/17 Apply frozen readouts to untouched probe_test
cell_stage('Refitting frozen readouts on train plus dev and opening probe_test once')

fit_indices = np.concatenate([train_indices, dev_indices])
probe_predictions = {}
probe_models = {}
probe_rows = []
for target_index, target in enumerate(PROBE_TARGETS):
    selection = frozen_selection[target]
    layer = int(selection['layer'])
    alpha = float(selection['alpha'])
    model = make_pipeline(
        StandardScaler(), Ridge(alpha=alpha, solver='lsqr')
    )
    model.fit(
        OUTCOME_HIDDEN[layer][fit_indices],
        target_matrix[fit_indices, target_index],
    )
    prediction = model.predict(OUTCOME_HIDDEN[layer][test_indices])
    probe_predictions[target] = prediction
    probe_models[target] = model
    gold = target_matrix[test_indices, target_index]
    if target == 'net_torque':
        exact_or_sign = float(np.mean(np.sign(prediction) == np.sign(gold)))
        accuracy_name = 'sign_accuracy'
    else:
        exact_or_sign = float(np.mean(np.abs(prediction - gold) < ROUND_TOLERANCE))
        accuracy_name = 'rounded_exact_accuracy'
    probe_rows.append({
        'target': target,
        'layer': layer,
        'alpha': alpha,
        'test_r2': float(r2_score(gold, prediction)),
        'test_mae': float(mean_absolute_error(gold, prediction)),
        accuracy_name: exact_or_sign,
    })

probe_test_frame = metadata.iloc[test_indices].copy().reset_index(drop=True)
for target in PROBE_TARGETS:
    probe_test_frame[f'probe_{target}'] = probe_predictions[target]
    gold = probe_test_frame[target].to_numpy(dtype=float)
    predicted = probe_test_frame[f'probe_{target}'].to_numpy(dtype=float)
    if target == 'net_torque':
        available = np.sign(predicted) == np.sign(gold)
    else:
        available = np.abs(predicted - gold) < ROUND_TOLERANCE
    probe_test_frame[f'probe_{target}_available'] = available

probe_result_frame = pd.DataFrame(probe_rows)
probe_result_frame.to_csv(ANALYSIS_DIR / 'frozen_probe_test_results.csv', index=False)
probe_test_frame.to_csv(ANALYSIS_DIR / 'probe_test_predictions.csv', index=False)
display(probe_result_frame.round(4))
```

```python
%%cell_progress 14/17 Build the same-scene physical fracture map
cell_stage('Joining outcome behavior, intermediate tasks, and frozen readouts')

cohort_by_scene = {
    row['scene_id']: row['cohort'] for row in confirmation_manifest
}
outcome_by_scene_frame = outcome_frame.set_index('scene_id')
probe_by_scene_frame = probe_test_frame.set_index('scene_id')
task_correct = task_frame.pivot(
    index='scene_id', columns='question_kind', values='correct'
)
task_prediction = task_frame.pivot(
    index='scene_id', columns='question_kind', values='predicted_semantic'
)

fracture_rows = []
for scene_id in confirmation_scene_ids:
    physical = physical_by_scene[scene_id]
    outcome_row = outcome_by_scene_frame.loc[scene_id]
    probe_row = probe_by_scene_frame.loc[scene_id]
    task_row = task_correct.loc[scene_id]

    weight_task_pass = bool(
        task_row['left_units'] and task_row['right_units']
    )
    distance_task_pass = bool(
        task_row['left_distance'] and task_row['right_distance']
    )
    factor_task_pass = bool(weight_task_pass and distance_task_pass)
    side_torque_pass = bool(
        task_row['left_torque'] and task_row['right_torque']
    )
    net_torque_pass = bool(task_row['net_torque'])
    outcome_pass = bool(outcome_row['correct'])

    weight_probe_pass = bool(
        probe_row['probe_left_units_available']
        and probe_row['probe_right_units_available']
    )
    distance_probe_pass = bool(
        probe_row['probe_left_distance_available']
        and probe_row['probe_right_distance_available']
    )
    factor_probe_pass = bool(weight_probe_pass and distance_probe_pass)
    side_torque_probe_pass = bool(all(
        probe_row[f'probe_{target}_available']
        for target in ('left_torque', 'right_torque')
    ))
    net_torque_probe_pass = bool(
        probe_row['probe_net_torque_available']
    )

    weight_available = bool(weight_task_pass or weight_probe_pass)
    distance_available = bool(distance_task_pass or distance_probe_pass)
    if not weight_available and distance_available:
        candidate_fracture = 'weight_reading_or_binding'
    elif not distance_available and weight_available:
        candidate_fracture = 'distance_reading_or_binding'
    elif not weight_available and not distance_available:
        candidate_fracture = 'factor_reading_or_binding'
    elif factor_task_pass and factor_probe_pass and not side_torque_pass:
        candidate_fracture = 'factor_to_side_torque'
    elif side_torque_pass and not net_torque_pass:
        candidate_fracture = 'side_torque_to_net_torque'
    elif net_torque_pass and net_torque_probe_pass and not outcome_pass:
        candidate_fracture = 'net_torque_to_outcome'
    else:
        candidate_fracture = 'ambiguous_or_mixed'

    fracture_rows.append({
        'scene_id': scene_id,
        'cohort': cohort_by_scene[scene_id],
        'outcome': physical['outcome'],
        'diagnostic_cell': physical['diagnostic_cell'],
        'weight_cue': physical['weight_cue'],
        'distance_cue': physical['distance_cue'],
        'net_torque': physical['net_torque'],
        'weight_task_pass': weight_task_pass,
        'weight_probe_pass': weight_probe_pass,
        'distance_task_pass': distance_task_pass,
        'distance_probe_pass': distance_probe_pass,
        'factor_task_pass': factor_task_pass,
        'factor_probe_pass': factor_probe_pass,
        'side_torque_task_pass': side_torque_pass,
        'side_torque_probe_pass': side_torque_probe_pass,
        'net_torque_task_pass': net_torque_pass,
        'net_torque_probe_pass': net_torque_probe_pass,
        'outcome_pass': outcome_pass,
        'candidate_fracture': candidate_fracture,
        'outcome_prediction': outcome_row['predicted_semantic'],
        'outcome_confidence_gap': float(outcome_row['confidence_gap']),
    })

fracture_frame = pd.DataFrame(fracture_rows)
fracture_frame.to_csv(ANALYSIS_DIR / 'scene_fracture_map.csv', index=False)

error_fractures = fracture_frame[
    fracture_frame['cohort'] == 'outcome_error'
].copy()
control_fractures = fracture_frame[
    fracture_frame['cohort'] == 'matched_correct'
].copy()
candidate_counts = error_fractures['candidate_fracture'].value_counts()

intervention_candidates = error_fractures[
    error_fractures['candidate_fracture'] != 'ambiguous_or_mixed'
].copy()
intervention_candidates['selection_key'] = intervention_candidates.apply(
    lambda row: stable_key((
        'future-intervention', row['candidate_fracture'], row['scene_id']
    )),
    axis=1,
)
intervention_candidates = intervention_candidates.sort_values(
    ['candidate_fracture', 'selection_key']
)
write_jsonl_atomic(
    ANALYSIS_DIR / 'candidate_stagewise_intervention_scenes.jsonl',
    intervention_candidates.to_dict('records'),
)

print('Candidate fractures among outcome errors:')
display(candidate_counts.rename('scenes').to_frame())
print('These are candidates, not final causal localizations.')
```

```python
%%cell_progress 15/17 Estimate error-conditioned stage support
cell_stage('Bootstrapping scene-level stage rates and matched differences')

STAGE_COLUMNS = (
    'weight_task_pass', 'weight_probe_pass',
    'distance_task_pass', 'distance_probe_pass',
    'factor_task_pass', 'factor_probe_pass',
    'side_torque_task_pass', 'side_torque_probe_pass',
    'net_torque_task_pass', 'net_torque_probe_pass',
)

def percentile_summary(draws):
    draws = np.asarray(draws, dtype=float)
    return {
        'estimate': float(np.mean(draws)),
        'ci_low': float(np.quantile(draws, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA_LEVEL / 2)),
    }

def bootstrap_rate(frame, column, key):
    values = frame[column].to_numpy(dtype=float)
    assert len(values) > 0
    generator = np.random.default_rng(
        int(stable_key(('bootstrap-rate', key))[:16], 16) % (2**63 - 1)
    )
    indices = generator.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    draws = values[indices].mean(axis=1)
    summary = percentile_summary(draws)
    summary['estimate'] = float(values.mean())
    summary['n_scenes'] = int(len(values))
    return summary

test_rows = []
for cohort, current in fracture_frame.groupby('cohort'):
    for column in STAGE_COLUMNS:
        test_rows.append({
            'cohort': cohort,
            'metric': column,
            **bootstrap_rate(current, column, (cohort, column)),
        })

error_scene_ids = [
    row['scene_id'] for row in confirmation_manifest
    if row['cohort'] == 'outcome_error'
]
control_by_error = {
    row['matched_error_scene_id']: row['scene_id']
    for row in selected_controls
}
fracture_by_scene = fracture_frame.set_index('scene_id')
for column in STAGE_COLUMNS:
    differences = np.asarray([
        float(fracture_by_scene.loc[error_id, column])
        - float(fracture_by_scene.loc[control_by_error[error_id], column])
        for error_id in error_scene_ids
    ])
    generator = np.random.default_rng(
        int(stable_key(('paired-bootstrap', column))[:16], 16) % (2**63 - 1)
    )
    indices = generator.integers(
        0, len(differences),
        size=(BOOTSTRAP_REPLICATES, len(differences)),
    )
    summary = percentile_summary(differences[indices].mean(axis=1))
    summary['estimate'] = float(differences.mean())
    summary['n_scenes'] = int(len(differences))
    test_rows.append({
        'cohort': 'error_minus_matched_correct',
        'metric': column,
        **summary,
    })

stage_test_frame = pd.DataFrame(test_rows)
stage_test_frame.to_csv(ANALYSIS_DIR / 'stage_support_tests.csv', index=False)

error_stage_support = {
    column: bootstrap_rate(
        error_fractures, column, ('error-primary', column)
    )
    for column in STAGE_COLUMNS
}
write_json_atomic(ANALYSIS_DIR / 'stage_support_summary.json', {
    'schema_version': 'error-conditioned-stage-support-1.0',
    'error_scenes': int(len(error_fractures)),
    'matched_correct_scenes': int(len(control_fractures)),
    'candidate_fracture_counts': {
        str(key): int(value) for key, value in candidate_counts.items()
    },
    'error_stage_support': error_stage_support,
    'interpretation': (
        'Availability on an error is evidence against total information '
        'absence, but it does not by itself establish causal use.'
    ),
})
display(stage_test_frame.round(4))
```
