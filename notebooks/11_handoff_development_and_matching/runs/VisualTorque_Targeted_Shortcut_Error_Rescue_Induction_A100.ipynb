```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. Independent same-answer shortcut-conflict manifest


```python
%%cell_progress 6/17 Build the independent same-answer conflict manifest
cell_stage('Enumerating conflict scenes and same-answer factor edits')
from collections import Counter, defaultdict
from itertools import product
import shutil

from visual_torque_toolkit.render import render_scene
from visual_torque_toolkit.world import Scene, Side, Stack

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def sign(value):
    return int(value > 0) - int(value < 0)

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

def shape_signature(shape):
    lu, ld, ru, rd = (int(value) for value in shape)
    return tuple(sorted((
        ('left', ld, lu), ('right', rd, ru),
    )))

def canonical_signature(stacks):
    return tuple(sorted(
        (str(stack['side']), int(stack['distance']), int(stack['units']))
        for stack in stacks
    ))

def stacks_from_shape(shape):
    lu, ld, ru, rd = (int(value) for value in shape)
    return [
        {'side': 'left', 'distance': ld, 'units': lu},
        {'side': 'right', 'distance': rd, 'units': ru},
    ]

def scene_features(shape):
    lu, ld, ru, rd = (int(value) for value in shape)
    left_torque = lu * ld
    right_torque = ru * rd
    net_torque = left_torque - right_torque
    weight_score = lu - ru
    distance_score = ld - rd
    return {
        'left_torque': int(left_torque),
        'right_torque': int(right_torque),
        'net_torque': int(net_torque),
        'total_torque': int(left_torque + right_torque),
        'normalized_margin': float(
            abs(net_torque) / max(1, left_torque + right_torque)
        ),
        'weight_score': int(weight_score),
        'distance_score': int(distance_score),
        'strategy_predictions': {
            'exact_torque': outcome(net_torque),
            'weight_only': outcome(weight_score),
            'distance_only': outcome(distance_score),
        },
        'outcome': outcome(net_torque),
    }

def diagnostic_stratum(features):
    predictions = features['strategy_predictions']
    exact = predictions['exact_torque']
    weight = predictions['weight_only']
    distance = predictions['distance_only']
    if exact == 'balanced' or weight == 'balanced' or distance == 'balanced':
        return None
    if weight != exact and distance == exact:
        diagnostic_cell = 'weight_conflict'
    elif distance != exact and weight == exact:
        diagnostic_cell = 'distance_conflict'
    else:
        return None
    margin = float(features['normalized_margin'])
    if 0.05 <= margin < 0.20:
        margin_bin = 'hard'
    elif 0.20 <= margin < 0.40:
        margin_bin = 'moderate'
    else:
        return None
    return diagnostic_cell, margin_bin, exact

def signatures_in(value):
    found = set()
    if isinstance(value, dict):
        stacks = value.get('stacks')
        if isinstance(stacks, list) and stacks and all(
            isinstance(stack, dict)
            and {'side', 'distance', 'units'} <= set(stack)
            for stack in stacks
        ):
            try:
                found.add(canonical_signature(stacks))
            except (TypeError, ValueError):
                pass
        for child in value.values():
            found.update(signatures_in(child))
    elif isinstance(value, list):
        for child in value:
            found.update(signatures_in(child))
    return found

dataset_rows = read_jsonl(DATASET_DIR / 'scenes.jsonl')
dataset_signatures = {
    canonical_signature(row['stacks']) for row in dataset_rows
}
prior_signatures = set()
prior_manifest_paths = []
results_root = PROJECT_ROOT / 'results'
if results_root.is_dir():
    prior_manifest_paths = sorted(
        path for path in results_root.rglob('*manifest*.jsonl')
        if SHARED_ROOT not in path.parents
    )
for path in tqdm(prior_manifest_paths, desc='Auditing prior geometries', unit='file'):
    try:
        rows = read_jsonl(path)
    except Exception:
        continue
    for row in rows:
        prior_signatures.update(signatures_in(row))
print('Dataset/prior signatures:', len(dataset_signatures), len(prior_signatures))

pool_by_shape = {}
for shape in product(range(1, 9), range(1, 7), range(1, 9), range(1, 7)):
    # Dataset v1 uses at most five blocks per stack; six-to-eight-block
    # candidates enlarge the independent pool while remaining fully visible.
    if max(shape[0], shape[2]) <= 5:
        continue
    signature = shape_signature(shape)
    if signature in dataset_signatures:
        continue
    features = scene_features(shape)
    pool_by_shape[shape] = {
        'shape': shape, 'signature': signature, 'features': features,
    }
pool = list(pool_by_shape.values())
by_stratum = defaultdict(list)
for record in pool:
    stratum = diagnostic_stratum(record['features'])
    if stratum is not None:
        by_stratum[stratum].append(record)
for records in by_stratum.values():
    records.sort(key=lambda row: stable_key(row['shape']))

def edit_distance(left, right):
    return sum(abs(int(a) - int(b)) for a, b in zip(left, right))

def role_candidates(base, role, used_signatures):
    base_shape = base['shape']
    lu, ld, ru, rd = base_shape
    base_features = base['features']
    exact = base_features['strategy_predictions']['exact_torque']
    shortcut_type = (
        'weight' if diagnostic_stratum(base_features)[0] == 'weight_conflict'
        else 'distance'
    )
    shortcut_prediction = base_features['strategy_predictions'][
        shortcut_type + '_only'
    ]
    base_score = int(base_features[shortcut_type + '_score'])
    candidates = []
    for record in pool:
        shape = record['shape']
        if shape == base_shape or record['signature'] in used_signatures:
            continue
        features = record['features']
        predictions = features['strategy_predictions']
        if features['outcome'] != exact:
            continue
        if shortcut_type == 'weight':
            factor_only = shape[1] == ld and shape[3] == rd
            irrelevant_only = shape[0] == lu and shape[2] == ru
        else:
            factor_only = shape[0] == lu and shape[2] == ru
            irrelevant_only = shape[1] == ld and shape[3] == rd
        current_score = int(features[shortcut_type + '_score'])
        if role == 'cue_aligned_source':
            valid = (
                factor_only
                and predictions[shortcut_type + '_only'] == exact
            )
        elif role == 'cue_amplified_source':
            valid = (
                factor_only
                and predictions[shortcut_type + '_only'] == shortcut_prediction
                and abs(current_score) > abs(base_score)
            )
        elif role == 'irrelevant_source':
            other = 'distance' if shortcut_type == 'weight' else 'weight'
            valid = (
                irrelevant_only
                and predictions[shortcut_type + '_only'] == shortcut_prediction
                and predictions[other + '_only'] == exact
            )
        else:
            raise KeyError(role)
        if valid:
            candidates.append(record)
    candidates.sort(key=lambda row: (
        int(row['signature'] in prior_signatures),
        edit_distance(base_shape, row['shape']),
        stable_key((role, base_shape, row['shape'])),
    ))
    return candidates

cell_stage('Selecting balanced target-disjoint four-scene families')
selected = []
used_signatures = set(dataset_signatures)
for diagnostic_cell in CONFLICT_CELLS:
    for margin_bin in MARGIN_BINS:
        for exact_outcome in OUTCOMES:
            stratum = (diagnostic_cell, margin_bin, exact_outcome)
            chosen = 0
            for base in by_stratum[stratum]:
                if (
                    base['signature'] in used_signatures
                    or base['signature'] in prior_signatures
                ):
                    continue
                provisional = used_signatures | {base['signature']}
                aligned = role_candidates(
                    base, 'cue_aligned_source', provisional
                )
                if not aligned:
                    continue
                aligned_record = aligned[0]
                provisional.add(aligned_record['signature'])
                amplified = role_candidates(
                    base, 'cue_amplified_source', provisional
                )
                if not amplified:
                    continue
                amplified_record = amplified[0]
                provisional.add(amplified_record['signature'])
                irrelevant = role_candidates(
                    base, 'irrelevant_source', provisional
                )
                if not irrelevant:
                    continue
                irrelevant_record = irrelevant[0]

                records = {
                    'base': base,
                    'cue_aligned_source': aligned_record,
                    'cue_amplified_source': amplified_record,
                    'irrelevant_source': irrelevant_record,
                }
                family_id = 'shortcut_' + stable_key((
                    diagnostic_cell, margin_bin, exact_outcome,
                    tuple(record['shape'] for record in records.values()),
                ))[:18]
                roles = {}
                for role, record in records.items():
                    scene_id = family_id + '__' + role
                    roles[role] = {
                        'scene_id': scene_id,
                        'image_filename': scene_id + '.png',
                        'stacks': stacks_from_shape(record['shape']),
                        'features': record['features'],
                        'prior_geometry_overlap': bool(
                            record['signature'] in prior_signatures
                        ),
                    }
                selected.append({
                    'family_id': family_id,
                    'split': 'confirmation',
                    'diagnostic_cell': diagnostic_cell,
                    'shortcut_type': (
                        'weight' if diagnostic_cell == 'weight_conflict'
                        else 'distance'
                    ),
                    'margin_bin': margin_bin,
                    'base_outcome': exact_outcome,
                    'roles': roles,
                })
                used_signatures.update(
                    record['signature'] for record in records.values()
                )
                chosen += 1
                if chosen == FAMILIES_PER_STRATUM:
                    break
            assert chosen == FAMILIES_PER_STRATUM, (
                stratum, chosen, len(by_stratum[stratum])
            )

selected.sort(key=lambda row: row['family_id'])
assert len(selected) == EXPECTED_FAMILIES
all_role_signatures = [
    canonical_signature(role['stacks'])
    for row in selected for role in row['roles'].values()
]
assert len(all_role_signatures) == len(set(all_role_signatures))
assert not any(
    canonical_signature(row['roles']['base']['stacks']) in prior_signatures
    for row in selected
)

counts = Counter((
    row['diagnostic_cell'], row['margin_bin'], row['base_outcome']
) for row in selected)
assert set(counts.values()) == {FAMILIES_PER_STRATUM}
manifest_identity = {
    'schema_version': 'targeted-shortcut-error-manifest-1.0',
    'dataset_scenes_sha256': sha256_file(DATASET_DIR / 'scenes.jsonl'),
    'selection_uses_model_outputs': False,
    'all_targets_disjoint_from_prior_manifests': True,
    'all_roles_disjoint_within_manifest': True,
    'families': len(selected),
    'role_geometries': len(all_role_signatures),
    'source_prior_overlap_count': int(sum(
        role['prior_geometry_overlap']
        for row in selected
        for name, role in row['roles'].items() if name != 'base'
    )),
    'stratum_counts': {
        '|'.join(key): int(value) for key, value in sorted(counts.items())
    },
    'selection_seed': SEED,
    'family_ids': [row['family_id'] for row in selected],
}
if MANIFEST_PATH.is_file() and MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity
    assert read_jsonl(MANIFEST_PATH) == selected
else:
    write_jsonl_atomic(MANIFEST_PATH, selected)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

shortcut_manifest = read_jsonl(MANIFEST_PATH)
cell_stage('Rendering or verifying same-answer scene quartets')
for row in tqdm(shortcut_manifest, desc='Rendering conflict families', unit='family'):
    exact = row['roles']['base']['features']['outcome']
    for role_name, role in row['roles'].items():
        assert role['features']['outcome'] == exact
        image_path = SCENE_DIR / role['image_filename']
        if image_path.is_file():
            continue
        stacks = tuple(
            Stack(
                stack_id=f'{role_name}_{index}',
                side=Side(str(stack['side'])),
                distance=int(stack['distance']),
                units=int(stack['units']),
            )
            for index, stack in enumerate(role['stacks'])
        )
        render_scene(Scene(role['scene_id'], stacks), image_path)
for row in shortcut_manifest:
    for role in row['roles'].values():
        assert (SCENE_DIR / role['image_filename']).is_file()

cell_stage('Caching rendered scenes on the Colab local SSD')
RUNTIME_SCENE_DIR = Path('/content/visualtorque_targeted_shortcut_scenes_v1')
RUNTIME_SCENE_DIR.mkdir(parents=True, exist_ok=True)
scene_files = [
    role['image_filename']
    for row in shortcut_manifest for role in row['roles'].values()
]
for filename in tqdm(scene_files, desc='Caching rendered scenes', unit='image'):
    source = SCENE_DIR / filename
    destination = RUNTIME_SCENE_DIR / filename
    if not destination.is_file() or destination.stat().st_size != source.stat().st_size:
        shutil.copy2(source, destination)
assert all((RUNTIME_SCENE_DIR / filename).is_file() for filename in scene_files)

manifest_frame = pd.DataFrame([{
    'family_id': row['family_id'],
    'diagnostic_cell': row['diagnostic_cell'],
    'shortcut_type': row['shortcut_type'],
    'margin_bin': row['margin_bin'],
    'base_outcome': row['base_outcome'],
    'source_prior_overlap_count': sum(
        role['prior_geometry_overlap']
        for name, role in row['roles'].items() if name != 'base'
    ),
} for row in shortcut_manifest])
manifest_frame.to_csv(SHARED_ROOT / 'shortcut_error_manifest.csv', index=False)
display(manifest_frame.groupby([
    'diagnostic_cell', 'margin_bin', 'base_outcome'
]).size().rename('n').reset_index())
print('Families / rendered scenes:',
      len(shortcut_manifest), len(all_role_signatures))
print('Target prior overlaps: 0 | source prior overlaps:',
      manifest_identity['source_prior_overlap_count'])
```

## 3. Factor-specific rescue, induction, and matched controls


```python
%%cell_progress 8/17 Define outcome prompts, bundles, and exact masks
cell_stage('Building identical semantic prompts for all four scene roles')
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Compare the two side totals.'
)
OUTCOME_TEXT = {
    'left': 'left side', 'balanced': 'balanced', 'right': 'right side',
}
LABELS = ('A', 'B', 'C')

def rotated_semantics(row):
    values = ['left', 'balanced', 'right']
    offset = int(stable_key(
        f"options:{row['family_id']}"
    )[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def custom_prompt(row, semantics):
    options = '\n'.join(
        f'{label}. {OUTCOME_TEXT[value]}'
        for label, value in zip(LABELS, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: Which side will rotate downward?',
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, role):
    semantics = rotated_semantics(row)
    role_record = row['roles'][role]
    features = role_record['features']
    gold = str(features['outcome'])
    prompt = custom_prompt(row, semantics)
    return BehaviorExample(
        example_id=f"shortcut:{row['family_id']}:{role}",
        question_id=f"shortcut_{row['family_id']}",
        scene_id=str(role_record['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind='outcome',
        variant='targeted_shortcut_same_answer',
        split='confirmation',
        problem_type=str(row['diagnostic_cell']),
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=prompt,
        image_path=(RUNTIME_SCENE_DIR / role_record['image_filename']).resolve(),
        labels=LABELS,
        option_text=tuple(OUTCOME_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def make_bundle(row):
    return {role: make_example(row, role) for role in SCENE_ROLES}

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples, 'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    return {**masks, 'tokens': tokens}

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key]), (
                name, key
            )
        masks = discover_masks(current)
        for _, scope, _ in FROZEN_SITES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Same-answer prompts and token masks ready.')
```

```python
%%cell_progress 9/17 Define capture and additive/replace intervention scorers
cell_stage('Defining candidate scoring and contrastive factor transport')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = (
                output_hidden_tensor(output).detach().cpu().clone()
            )
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def capture_named_examples(examples_by_name):
    names = list(examples_by_name)
    score_by_name = {}
    hidden_by_name = {name: {} for name in names}
    for start in range(0, len(names), CAPTURE_BATCH_SIZE):
        actual = names[start:start + CAPTURE_BATCH_SIZE]
        padded = list(actual)
        while len(padded) < CAPTURE_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepare_examples([examples_by_name[name] for name in padded])
        scores, hidden = forward_capture_batch(prepared)
        for index, name in enumerate(actual):
            score_by_name[name] = scores[index]
            for layer_index, tensor in hidden.items():
                hidden_by_name[name][layer_index] = tensor[index]
        del prepared, hidden
        torch.cuda.empty_cache()
    return score_by_name, hidden_by_name

def matched_random_delta(delta, mask, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(delta.shape, generator=generator, dtype=torch.float32)
    selected = mask.bool()
    output = torch.zeros_like(noise)
    delta_selected = delta.float()[selected]
    noise_selected = noise[selected]
    delta_norm = float(torch.linalg.vector_norm(delta_selected))
    noise_norm = float(torch.linalg.vector_norm(noise_selected))
    if delta_norm > 0 and noise_norm > 0:
        output[selected] = noise_selected * (delta_norm / noise_norm)
    return output.to(dtype=delta.dtype)

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert len(layer_indices) == 1
    layer_index = next(iter(layer_indices))
    payloads = torch.stack([row['payload_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    modes = [row['intervention_mode'] for row in conditions]
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payloads.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == payload.shape
        modified = hidden.clone()
        for index, mode in enumerate(modes):
            current_mask = mask[index]
            if mode == 'add':
                modified[index, current_mask] = (
                    hidden[index, current_mask] + payload[index, current_mask]
                )
            elif mode == 'replace':
                modified[index, current_mask] = payload[index, current_mask]
            else:
                raise ValueError(mode)
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('payload_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)

    def run_chunk(chunk):
        actual_size = len(chunk)
        padded = list(chunk)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepared_cache.get(len(padded))
        if prepared is None:
            prepared = prepare_examples([target_example] * len(padded))
            prepared_cache[len(padded)] = prepared
        try:
            return forward_intervention_chunk(prepared, padded)[:actual_size]
        except torch.cuda.OutOfMemoryError as exc:
            prepared_cache.pop(len(padded), None)
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Intervention batch ran out of memory. Restart on an A100; '
                'do not change the frozen batch size within a model run.'
            ) from exc

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

print('Targeted additive and replacement scorers ready.')
```

```python
%%cell_progress 10/17 Run physics, alignment, and parity smoke tests
cell_stage('Checking same-answer edits, exact masks, and additive parity')
smoke_row = shortcut_manifest[0]
roles = smoke_row['roles']
base_features = roles['base']['features']
aligned_features = roles['cue_aligned_source']['features']
amplified_features = roles['cue_amplified_source']['features']
irrelevant_features = roles['irrelevant_source']['features']
shortcut_type = smoke_row['shortcut_type']
shortcut_key = shortcut_type + '_only'
exact = base_features['outcome']
shortcut = base_features['strategy_predictions'][shortcut_key]
assert exact != shortcut
assert {role['features']['outcome'] for role in roles.values()} == {exact}
assert aligned_features['strategy_predictions'][shortcut_key] == exact
assert amplified_features['strategy_predictions'][shortcut_key] == shortcut
assert abs(amplified_features[shortcut_type + '_score']) > abs(
    base_features[shortcut_type + '_score']
)
assert irrelevant_features['strategy_predictions'][shortcut_key] == shortcut

smoke_bundle = make_bundle(smoke_row)
smoke_prepared = {
    name: prepare_examples([example])
    for name, example in smoke_bundle.items()
}
smoke_masks = assert_aligned(smoke_prepared)
for _, scope, _ in FROZEN_SITES:
    assert bool(smoke_masks[scope].any()), scope

smoke_scores, smoke_hidden = capture_named_examples(smoke_bundle)
layer_index = FROZEN_SITES[0][0]
base_hidden = smoke_hidden['base'][layer_index]
aligned_delta = (
    smoke_hidden['cue_aligned_source'][layer_index] - base_hidden
)
zero_delta = base_hidden - base_hidden
smoke_conditions = [
    {
        'condition': 'self_all_tokens',
        'intervention': 'self_replay',
        'intervention_mode': 'replace',
        'layer_index': layer_index,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': base_hidden,
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'condition': 'aligned_add_all_tokens',
        'intervention': 'cue_aligned',
        'intervention_mode': 'add',
        'layer_index': layer_index,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': aligned_delta,
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'condition': 'zero_add_all_tokens',
        'intervention': 'zero_add',
        'intervention_mode': 'add',
        'layer_index': layer_index,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': zero_delta,
        'patch_mask': smoke_masks['valid_tokens'],
    },
]
smoke_records = score_conditions(
    smoke_bundle['base'], {}, smoke_conditions
)
self_error = float(np.max(np.abs(
    centered_scores(smoke_records[0])
    - centered_scores(smoke_scores['base'])
)))
aligned_error = float(np.max(np.abs(
    centered_scores(smoke_records[1])
    - centered_scores(smoke_scores['cue_aligned_source'])
)))
zero_error = float(np.max(np.abs(
    centered_scores(smoke_records[2])
    - centered_scores(smoke_scores['base'])
)))
print('Centered self / aligned-source / zero-add parity:',
      self_error, aligned_error, zero_error)
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert aligned_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, aligned_error
assert zero_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, zero_error
print('Smoke family:', smoke_row['family_id'], shortcut_type)
print('All source outcomes:', {
    name: role['features']['outcome'] for name, role in roles.items()
})
print('Frozen token counts:', {
    f'{label}:L{layer}:{scope}': int(smoke_masks[scope].sum())
    for layer, scope, label in FROZEN_SITES
})
del smoke_prepared, smoke_hidden, aligned_delta, zero_delta
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/17 Capture or resume targeted rescue/induction interventions
cell_stage('Locking capture identity and checking completed family shards')

def norm_match_delta(reference_delta, control_delta, mask):
    selected = mask.bool()
    output = torch.zeros_like(control_delta)
    reference = reference_delta.float()[selected]
    control = control_delta.float()[selected]
    reference_norm = float(torch.linalg.vector_norm(reference))
    control_norm = float(torch.linalg.vector_norm(control))
    assert reference_norm > 0 and control_norm > 0
    output[selected] = (
        control * (reference_norm / control_norm)
    ).to(dtype=control_delta.dtype)
    return output

RECORDS_PER_FAMILY = len(FROZEN_SITES) * len(INTERVENTIONS)
EXPECTED_PATCH_RECORDS = len(shortcut_manifest) * RECORDS_PER_FAMILY
capture_identity = {
    'schema_version': 'targeted-shortcut-error-capture-1.0',
    'model_key': MODEL_KEY,
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'protocol_lock_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'frozen_sites': [list(site) for site in FROZEN_SITES],
    'scene_roles': list(SCENE_ROLES),
    'interventions': list(INTERVENTIONS),
    'capture_batch_size': CAPTURE_BATCH_SIZE,
    'patch_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'records_per_family': RECORDS_PER_FAMILY,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'layer_path': LAYER_PATH,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing capture differs. Use a new versioned output directory.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_family_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_family_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    family_id = str(payload['family']['family_id'])
    assert family_id not in completed_family_ids
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    completed_family_ids.add(family_id)
    completed_records += len(payload['patch_records'])

progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc=f'{MODEL_KEY} targeted error interventions', unit='condition',
    dynamic_ncols=True,
)
for family_index, row in enumerate(shortcut_manifest):
    family_id = str(row['family_id'])
    if family_id in completed_family_ids:
        continue
    progress.set_postfix_str(
        f"{row['diagnostic_cell']} | {row['margin_bin']} | {row['base_outcome']}"
    )
    bundle = make_bundle(row)
    prepared_single = {
        name: prepare_examples([example])
        for name, example in bundle.items()
    }
    masks = assert_aligned(prepared_single)
    scores, hidden = capture_named_examples(bundle)
    family_patch_records = []
    conditions = []
    for layer_index, scope, site_label in FROZEN_SITES:
        base_hidden = hidden['base'][layer_index]
        aligned_delta = (
            hidden['cue_aligned_source'][layer_index] - base_hidden
        )
        amplified_delta = (
            hidden['cue_amplified_source'][layer_index] - base_hidden
        )
        irrelevant_delta = (
            hidden['irrelevant_source'][layer_index] - base_hidden
        )
        wrong_aligned = norm_match_delta(
            aligned_delta, irrelevant_delta, masks[scope]
        )
        wrong_amplified = norm_match_delta(
            amplified_delta, irrelevant_delta, masks[scope]
        )
        random_aligned = matched_random_delta(
            aligned_delta, masks[scope],
            f'{family_id}:{site_label}:random_aligned',
        )
        random_amplified = matched_random_delta(
            amplified_delta, masks[scope],
            f'{family_id}:{site_label}:random_amplified',
        )
        payloads = {
            'self_replay': ('replace', base_hidden),
            'cue_aligned': ('add', aligned_delta),
            'cue_amplified': ('add', amplified_delta),
            'wrong_factor_aligned_norm': ('add', wrong_aligned),
            'wrong_factor_amplified_norm': ('add', wrong_amplified),
            'random_aligned_norm': ('add', random_aligned),
            'random_amplified_norm': ('add', random_amplified),
        }
        for intervention in INTERVENTIONS:
            mode, payload_hidden = payloads[intervention]
            conditions.append({
                'analysis_family': 'targeted_shortcut_error',
                'family_id': family_id,
                'diagnostic_cell': row['diagnostic_cell'],
                'shortcut_type': row['shortcut_type'],
                'margin_bin': row['margin_bin'],
                'intervention': intervention,
                'intervention_mode': mode,
                'site_label': site_label,
                'scope': scope,
                'layer_index': int(layer_index),
                'payload_hidden': payload_hidden,
                'patch_mask': masks[scope],
            })
    prepared_cache = {1: prepared_single['base']}
    family_patch_records.extend(
        score_conditions(bundle['base'], prepared_cache, conditions)
    )
    progress.update(len(family_patch_records))
    assert len(family_patch_records) == RECORDS_PER_FAMILY
    shard_path = SHARD_DIR / f'shard_family_{family_index:04d}.json'
    write_json_atomic(shard_path, {
        'schema_version': 'targeted-shortcut-error-family-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'family': row,
        'baselines': scores,
        'patch_records': family_patch_records,
    })
    del prepared_single, prepared_cache, scores, hidden, conditions
    torch.cuda.empty_cache()
progress.close()

completed = sorted(SHARD_DIR.glob('shard_family_*.json'))
assert len(completed) == len(shortcut_manifest)
observed_records = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    observed_records += len(payload['patch_records'])
assert observed_records == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'families': len(completed),
    'patch_records': observed_records,
    'capture_identity_sha256': CAPTURE_SHA256,
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Capture complete:', observed_records, 'patch records')
```

```python
%%cell_progress 12/17 Release GPU and assemble family-site estimands
cell_stage('Releasing model memory and loading completed family shards')
import gc

del adapter
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    assert semantic in semantics, (semantic, semantics)
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def semantic_probability(record, semantic):
    semantics = list(record['candidate_semantics'])
    assert semantic in semantics
    return float(record['candidate_probabilities'][semantics.index(semantic)])

def record_lookup(payload, site_label, intervention):
    matches = [
        record for record in payload['patch_records']
        if record['site_label'] == site_label
        and record['intervention'] == intervention
    ]
    assert len(matches) == 1, (site_label, intervention, len(matches))
    return matches[0]

estimand_rows = []
for shard_path in tqdm(
    sorted(SHARD_DIR.glob('shard_family_*.json')),
    desc='Assembling targeted-error estimands', unit='family',
):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    family = payload['family']
    base_features = family['roles']['base']['features']
    exact = str(base_features['outcome'])
    shortcut_key = family['shortcut_type'] + '_only'
    shortcut = str(base_features['strategy_predictions'][shortcut_key])
    assert exact != shortcut
    clean = payload['baselines']['base']
    clean_margin = (
        semantic_logprob(clean, exact)
        - semantic_logprob(clean, shortcut)
    )
    clean_prediction = str(clean['predicted_semantic'])
    baseline_group = (
        'shortcut_error' if clean_prediction == shortcut
        else 'exact_correct' if clean_prediction == exact
        else 'other_prediction'
    )
    for _, scope, site_label in FROZEN_SITES:
        self_record = record_lookup(payload, site_label, 'self_replay')
        self_margin = (
            semantic_logprob(self_record, exact)
            - semantic_logprob(self_record, shortcut)
        )
        self_shift = self_margin - clean_margin
        row = {
            'model_key': MODEL_KEY,
            'family_id': family['family_id'],
            'diagnostic_cell': family['diagnostic_cell'],
            'shortcut_type': family['shortcut_type'],
            'margin_bin': family['margin_bin'],
            'base_outcome': family['base_outcome'],
            'site_label': site_label,
            'scope': scope,
            'exact_semantic': exact,
            'shortcut_semantic': shortcut,
            'baseline_prediction': clean_prediction,
            'baseline_group': baseline_group,
            'clean_exact_shortcut_margin': clean_margin,
            'clean_exact_probability': semantic_probability(clean, exact),
            'self_raw_shift': self_shift,
            'self_prediction': str(self_record['predicted_semantic']),
            'aligned_source_prediction': str(
                payload['baselines']['cue_aligned_source']['predicted_semantic']
            ),
            'amplified_source_prediction': str(
                payload['baselines']['cue_amplified_source']['predicted_semantic']
            ),
            'source_prior_overlap_count': sum(
                role['prior_geometry_overlap']
                for name, role in family['roles'].items() if name != 'base'
            ),
        }
        for intervention in INTERVENTIONS:
            record = record_lookup(payload, site_label, intervention)
            patched_margin = (
                semantic_logprob(record, exact)
                - semantic_logprob(record, shortcut)
            )
            raw_shift = patched_margin - clean_margin
            row[f'{intervention}__raw_shift'] = raw_shift
            row[f'{intervention}__centered_shift'] = raw_shift - self_shift
            row[f'{intervention}__prediction'] = str(
                record['predicted_semantic']
            )
            row[f'{intervention}__exact_probability'] = (
                semantic_probability(record, exact)
            )
        estimand_rows.append(row)

estimand_frame = pd.DataFrame(estimand_rows)
assert len(estimand_frame) == len(shortcut_manifest) * len(FROZEN_SITES)
estimand_frame.to_csv(
    ANALYSIS_DIR / 'family_site_estimands.csv', index=False
)
print('Family-site estimands:', len(estimand_frame))
display(estimand_frame.groupby([
    'diagnostic_cell', 'baseline_group', 'site_label'
]).size().rename('n').reset_index())
```

```python
%%cell_progress 13/17 Run frozen targeted rescue and induction tests
cell_stage('Bootstrapping model-specific error and correct cohorts')

def bootstrap_summary(values, key, replicates=BOOTSTRAP_REPLICATES):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    result = {
        'n': int(len(array)),
        'mean': float(array.mean()) if len(array) else np.nan,
        'ci_low': np.nan, 'ci_high': np.nan,
        'p_directional': np.nan,
    }
    if len(array) < 2:
        return result
    rng = np.random.default_rng(
        int(stable_key(f'bootstrap:{MODEL_KEY}:{key}')[:16], 16)
        % (2**63 - 1)
    )
    draws = rng.choice(
        array, size=(replicates, len(array)), replace=True
    ).mean(axis=1)
    result.update({
        'ci_low': float(np.quantile(draws, ALPHA / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
        'p_directional': float(
            (1 + np.count_nonzero(draws <= 0)) / (replicates + 1)
        ),
    })
    return result

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    output = np.full(len(values), np.nan)
    finite = np.where(np.isfinite(values))[0]
    if not len(finite):
        return output
    order = finite[np.argsort(values[finite])]
    adjusted = np.empty(len(order), dtype=float)
    running = 1.0
    for rank_index in range(len(order) - 1, -1, -1):
        rank = rank_index + 1
        candidate = values[order[rank_index]] * len(order) / rank
        running = min(running, candidate)
        adjusted[rank_index] = min(1.0, running)
    output[order] = adjusted
    return output

test_rows = []
for site_label in (PRIMARY_SITE_LABEL, SECONDARY_SITE_LABEL):
    site = estimand_frame[estimand_frame['site_label'] == site_label]
    for diagnostic_cell in CONFLICT_CELLS:
        cell_rows = site[site['diagnostic_cell'] == diagnostic_cell]
        rescue = cell_rows[cell_rows['baseline_group'] == 'shortcut_error']
        induction = cell_rows[cell_rows['baseline_group'] == 'exact_correct']
        vectors = {
            ('rescue', 'main'): rescue['cue_aligned__centered_shift'],
            ('rescue', 'vs_wrong_factor'): (
                rescue['cue_aligned__centered_shift']
                - rescue['wrong_factor_aligned_norm__centered_shift']
            ),
            ('rescue', 'vs_random'): (
                rescue['cue_aligned__centered_shift']
                - rescue['random_aligned_norm__centered_shift']
            ),
            ('rescue', 'targeted_flip'): (
                (rescue['cue_aligned__prediction'] == rescue['exact_semantic']).astype(float)
                - (rescue['self_replay__prediction'] == rescue['exact_semantic']).astype(float)
            ),
            ('induction', 'main'): -induction['cue_amplified__centered_shift'],
            ('induction', 'vs_wrong_factor'): (
                induction['wrong_factor_amplified_norm__centered_shift']
                - induction['cue_amplified__centered_shift']
            ),
            ('induction', 'vs_random'): (
                induction['random_amplified_norm__centered_shift']
                - induction['cue_amplified__centered_shift']
            ),
            ('induction', 'targeted_flip'): (
                (induction['cue_amplified__prediction'] == induction['shortcut_semantic']).astype(float)
                - (induction['self_replay__prediction'] == induction['shortcut_semantic']).astype(float)
            ),
        }
        for (direction, contrast), values in vectors.items():
            summary = bootstrap_summary(
                values,
                f'{site_label}:{diagnostic_cell}:{direction}:{contrast}',
            )
            test_rows.append({
                'model_key': MODEL_KEY,
                'site_label': site_label,
                'diagnostic_cell': diagnostic_cell,
                'direction': direction,
                'contrast': contrast,
                'confirmatory_site': site_label == PRIMARY_SITE_LABEL,
                'predeclared_strongest_pair': (
                    diagnostic_cell in SPEC['predeclared_pairs']
                ),
                **summary,
            })

tests_frame = pd.DataFrame(test_rows)
confirmatory_logodds = (
    tests_frame['confirmatory_site']
    & tests_frame['contrast'].isin(('main', 'vs_wrong_factor', 'vs_random'))
)
tests_frame['q_value'] = np.nan
tests_frame.loc[confirmatory_logodds, 'q_value'] = bh_adjust(
    tests_frame.loc[confirmatory_logodds, 'p_directional']
)
tests_frame.to_csv(ANALYSIS_DIR / 'targeted_error_tests.csv', index=False)

decision_rows = []
for diagnostic_cell in CONFLICT_CELLS:
    direction_status = {}
    for direction in ('rescue', 'induction'):
        rows = tests_frame[
            (tests_frame['site_label'] == PRIMARY_SITE_LABEL)
            & (tests_frame['diagnostic_cell'] == diagnostic_cell)
            & (tests_frame['direction'] == direction)
        ].set_index('contrast')
        main = rows.loc['main']
        wrong = rows.loc['vs_wrong_factor']
        random = rows.loc['vs_random']
        powered = int(main['n']) >= MIN_PRIMARY_COHORT
        passed = bool(
            powered
            and float(main['ci_low']) > CAUSAL_SESOI_LOGODDS
            and float(wrong['ci_low']) > 0
            and float(random['ci_low']) > 0
            and float(main['q_value']) < 0.05
            and float(wrong['q_value']) < 0.05
            and float(random['q_value']) < 0.05
        )
        direction_status[direction] = {
            'powered': powered,
            'passed': passed,
            'n': int(main['n']),
            'main_mean': float(main['mean']),
            'main_ci_low': float(main['ci_low']),
            'main_ci_high': float(main['ci_high']),
        }
    if not all(value['powered'] for value in direction_status.values()):
        status = 'underpowered_for_bidirectional_claim'
    elif all(value['passed'] for value in direction_status.values()):
        status = 'bidirectional_specific_error_control'
    elif direction_status['rescue']['passed']:
        status = 'rescue_only'
    elif direction_status['induction']['passed']:
        status = 'induction_only'
    else:
        status = 'no_specific_direction_passes'
    decision_rows.append({
        'model_key': MODEL_KEY,
        'diagnostic_cell': diagnostic_cell,
        'status': status,
        'predeclared_strongest_pair': (
            diagnostic_cell in SPEC['predeclared_pairs']
        ),
        'rescue_powered': direction_status['rescue']['powered'],
        'rescue_passed': direction_status['rescue']['passed'],
        'rescue_n': direction_status['rescue']['n'],
        'induction_powered': direction_status['induction']['powered'],
        'induction_passed': direction_status['induction']['passed'],
        'induction_n': direction_status['induction']['n'],
    })

decision_frame = pd.DataFrame(decision_rows)
decision_frame.to_csv(
    ANALYSIS_DIR / 'targeted_error_decisions.csv', index=False
)
display(tests_frame.round(4))
display(decision_frame)
```

```python
%%cell_progress 14/17 Audit cohorts, sources, controls, and numerical replay
cell_stage('Checking cohort balance, same-answer sources, and self replay')

primary = estimand_frame[
    estimand_frame['site_label'] == PRIMARY_SITE_LABEL
].copy()
cohort_counts = primary.groupby([
    'diagnostic_cell', 'margin_bin', 'base_outcome', 'baseline_group'
]).size().rename('n').reset_index()
cohort_counts.to_csv(ANALYSIS_DIR / 'cohort_counts.csv', index=False)

self_audit = {
    'n': int(len(primary)),
    'max_abs_raw_shift': float(primary['self_raw_shift'].abs().max()),
    'p95_abs_raw_shift': float(primary['self_raw_shift'].abs().quantile(0.95)),
    'target_prior_overlap_count': 0,
    'source_prior_overlap_count': int(
        manifest_frame['source_prior_overlap_count'].sum()
    ),
}

source_behavior_rows = []
for diagnostic_cell in CONFLICT_CELLS:
    rows = primary[primary['diagnostic_cell'] == diagnostic_cell]
    source_behavior_rows.append({
        'diagnostic_cell': diagnostic_cell,
        'n': int(len(rows)),
        'aligned_source_exact_rate': float(np.mean(
            rows['aligned_source_prediction'] == rows['exact_semantic']
        )),
        'amplified_source_shortcut_rate': float(np.mean(
            rows['amplified_source_prediction'] == rows['shortcut_semantic']
        )),
        'clean_exact_rate': float(np.mean(
            rows['baseline_group'] == 'exact_correct'
        )),
        'clean_shortcut_error_rate': float(np.mean(
            rows['baseline_group'] == 'shortcut_error'
        )),
    })
source_behavior_frame = pd.DataFrame(source_behavior_rows)
source_behavior_frame.to_csv(
    ANALYSIS_DIR / 'source_behavior_audit.csv', index=False
)

stratum_rows = []
for diagnostic_cell in CONFLICT_CELLS:
    for margin_bin in MARGIN_BINS:
        rows = primary[
            (primary['diagnostic_cell'] == diagnostic_cell)
            & (primary['margin_bin'] == margin_bin)
        ]
        rescue = rows[rows['baseline_group'] == 'shortcut_error']
        induction = rows[rows['baseline_group'] == 'exact_correct']
        stratum_rows.append({
            'diagnostic_cell': diagnostic_cell,
            'margin_bin': margin_bin,
            'rescue_n': int(len(rescue)),
            'rescue_mean': float(
                rescue['cue_aligned__centered_shift'].mean()
            ) if len(rescue) else np.nan,
            'induction_n': int(len(induction)),
            'induction_mean': float(
                -induction['cue_amplified__centered_shift'].mean()
            ) if len(induction) else np.nan,
            'wrong_factor_rescue_mean': float(
                rescue['wrong_factor_aligned_norm__centered_shift'].mean()
            ) if len(rescue) else np.nan,
            'wrong_factor_induction_mean': float(
                -induction['wrong_factor_amplified_norm__centered_shift'].mean()
            ) if len(induction) else np.nan,
        })
stratum_frame = pd.DataFrame(stratum_rows)
stratum_frame.to_csv(ANALYSIS_DIR / 'stratum_robustness.csv', index=False)

loo_rows = []
for diagnostic_cell in CONFLICT_CELLS:
    rows = primary[primary['diagnostic_cell'] == diagnostic_cell]
    for direction, group, column, multiplier in (
        ('rescue', 'shortcut_error', 'cue_aligned__centered_shift', 1.0),
        ('induction', 'exact_correct', 'cue_amplified__centered_shift', -1.0),
    ):
        cohort = rows[rows['baseline_group'] == group]
        values = multiplier * cohort[column].to_numpy(dtype=float)
        if len(values) < 3:
            minimum = np.nan
            maximum = np.nan
        else:
            loo = [
                float(np.delete(values, index).mean())
                for index in range(len(values))
            ]
            minimum, maximum = min(loo), max(loo)
        loo_rows.append({
            'diagnostic_cell': diagnostic_cell,
            'direction': direction,
            'n': int(len(values)),
            'minimum_leave_one_family_out_mean': minimum,
            'maximum_leave_one_family_out_mean': maximum,
        })
loo_frame = pd.DataFrame(loo_rows)
loo_frame.to_csv(ANALYSIS_DIR / 'leave_one_family_out.csv', index=False)

print('Self-replay audit:', self_audit)
display(cohort_counts)
display(source_behavior_frame.round(3))
display(stratum_frame.round(3))
display(loo_frame.round(3))
```
