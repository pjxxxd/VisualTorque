```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 15 tracked cells.')
```

```python
%%cell_progress 4/15 Lock the readiness and competence protocol
cell_stage('Freezing model, exclusions, competence criterion, and sample targets')
import hashlib
import json
from datetime import datetime, timezone

MODEL_ID = 'Qwen/Qwen3.5-27B'
MODEL_SLUG = 'Qwen__Qwen3.5-27B__373612cb'
MODEL_REVISION = '373612cb4f9288fcf550b526d86d85febedd9801'
MODEL_DTYPE = 'bfloat16'
DEVICE_MAP = 'auto'
EXPECTED_LAYER_COUNT = 64
SEED = 20261015

FACTORS = ('weight', 'distance')
BACKGROUNDS = ('b1', 'b2', 'b3')
MAPPING_VARIANTS = ('forward', 'reverse')
SCREEN_BATCH_SIZE = 4
TARGET_FAMILIES_PER_FACTOR = 24
MIN_FAMILIES_PER_FACTOR = 12
CONTEXTS_PER_FAMILY = 3

PRIOR_ROUTE_ROOT = (
    PROJECT_ROOT / 'results' / 'qwen27b_factor_path_token_component_v2'
)
PRIOR_GRADED_ROOT = (
    PROJECT_ROOT / 'results' / 'qwen27b_graded_product_law_v1'
)
PRIOR_ROUTE_MANIFEST = PRIOR_ROUTE_ROOT / 'fresh_decision_neutral_manifest.jsonl'
PRIOR_GRADED_MANIFEST = PRIOR_GRADED_ROOT / 'graded_product_manifest.jsonl'
PRIOR_ROUTE_SELECTION = PRIOR_ROUTE_ROOT / 'analysis' / 'frozen_sites.json'
PRIOR_GRADED_RUN = PRIOR_GRADED_ROOT / 'RUN_COMPLETE.json'
for required in (
    PRIOR_ROUTE_MANIFEST, PRIOR_GRADED_MANIFEST,
    PRIOR_ROUTE_SELECTION, PRIOR_GRADED_RUN,
):
    assert required.is_file(), f'Missing prerequisite: {required}'

OUTPUT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'qwen27b_competence_matched_clearance_readiness_v1'
)
CANDIDATE_PATH = OUTPUT_ROOT / 'model_blind_candidate_groups.jsonl'
CANDIDATE_LOCK_PATH = OUTPUT_ROOT / 'candidate_lock.json'
REQUEST_PATH = OUTPUT_ROOT / 'screen_requests.jsonl'
SCREEN_SHARD_DIR = OUTPUT_ROOT / 'screen_shards'
SCREEN_RESULT_PATH = OUTPUT_ROOT / 'behavior_screen_results.jsonl'
SELECTED_MANIFEST_PATH = OUTPUT_ROOT / 'frozen_confirmation_manifest.jsonl'
ANALYSIS_DIR = OUTPUT_ROOT / 'analysis'
FIGURE_DIR = ANALYSIS_DIR / 'figures'
for path in (OUTPUT_ROOT, SCREEN_SHARD_DIR, ANALYSIS_DIR, FIGURE_DIR):
    path.mkdir(parents=True, exist_ok=True)

def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return 'sha256:' + digest.hexdigest()

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def write_jsonl_atomic(path, rows):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

protocol_identity = {
    'schema_version': 'competence-matched-clearance-readiness-1.0',
    'analysis_status': 'behavior-screened readiness; no activation analysis',
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'factors': list(FACTORS),
    'target_families_per_factor': TARGET_FAMILIES_PER_FACTOR,
    'minimum_families_per_factor': MIN_FAMILIES_PER_FACTOR,
    'contexts_per_family': CONTEXTS_PER_FAMILY,
    'competence_criterion': (
        'target and natural hybrid both exactly correct in every selected '
        'context; selected contexts include at least one no-flip and one flip'
    ),
    'clearance_match': (
        'all target contexts in a family have identical signed net-torque deficit'
    ),
    'mapping_assignment': 'deterministic forward/reverse by family and context',
    'selection_may_use': ['natural endpoint exact correctness'],
    'selection_may_not_use': [
        'activations', 'layers', 'token regions', 'intervention effects'
    ],
    'prior_route_manifest_sha256': sha256_file(PRIOR_ROUTE_MANIFEST),
    'prior_graded_manifest_sha256': sha256_file(PRIOR_GRADED_MANIFEST),
    'seed': SEED,
}
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_readiness_protocol.json'
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing protocol differs. Use a new versioned output directory.'
    )
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

print('Model:', MODEL_ID, '@', MODEL_REVISION)
print('Target / minimum families per factor:',
      TARGET_FAMILIES_PER_FACTOR, MIN_FAMILIES_PER_FACTOR)
print('Output:', OUTPUT_ROOT)
```

```python
%%cell_progress 6/15 Build the independent matched-clearance candidate pool
cell_stage('Enumerating exact-deficit weight and distance groups')
import itertools
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}

def single_shape(scene):
    if len(scene['stacks']) != 2:
        return None
    left = [stack for stack in scene['stacks'] if stack['side'] == 'left']
    right = [stack for stack in scene['stacks'] if stack['side'] == 'right']
    if len(left) != 1 or len(right) != 1:
        return None
    return (
        int(left[0]['units']), int(left[0]['distance']),
        int(right[0]['units']), int(right[0]['distance']),
    )

scene_by_shape = {}
for scene in scene_records:
    shape = single_shape(scene)
    if shape is not None:
        assert shape not in scene_by_shape
        scene_by_shape[shape] = str(scene['scene_id'])
assert len(scene_by_shape) == 900

def prior_scene_ids(path):
    values = set()
    for row in read_jsonl(path):
        if isinstance(row.get('roles'), dict):
            for role in row['roles'].values():
                if isinstance(role, dict) and 'scene_id' in role:
                    values.add(str(role['scene_id']))
        for key, value in row.items():
            if key.endswith('_scene_id') and isinstance(value, (str, int)):
                values.add(str(value))
            elif key.endswith('_scene_ids') and isinstance(value, list):
                values.update(str(item) for item in value)
    return values

excluded_scene_ids = (
    prior_scene_ids(PRIOR_ROUTE_MANIFEST)
    | prior_scene_ids(PRIOR_GRADED_MANIFEST)
)
assert excluded_scene_ids

def opposite_side(side):
    return 'right' if side == 'left' else 'left'

def shape_for(side, units, distance, opposite_units, opposite_distance):
    if side == 'left':
        return (units, distance, opposite_units, opposite_distance)
    return (opposite_units, opposite_distance, units, distance)

def role_record(side, units, distance, opposite_units, opposite_distance):
    shape = shape_for(
        side, units, distance, opposite_units, opposite_distance
    )
    scene_id = scene_by_shape.get(shape)
    if scene_id is None:
        return None
    scene = scene_by_id[scene_id]
    return {
        'scene_id': scene_id,
        'units': int(units),
        'distance': int(distance),
        'opposite_units': int(opposite_units),
        'opposite_distance': int(opposite_distance),
        'side_torque': int(scene['derived'][f'{side}_torque']),
        'net_torque': int(scene['derived']['net_torque']),
        'outcome': str(scene['derived']['outcome']),
    }

def product_pairs(total):
    return [
        (units, distance)
        for units in range(1, 6) for distance in range(1, 7)
        if units * distance == total
    ]

def choose_context_pair(side, base, source, context, deficit, factor):
    if factor == 'weight':
        target_units, target_distance = base, context
        hybrid_units, hybrid_distance = source, context
    else:
        target_units, target_distance = context, base
        hybrid_units, hybrid_distance = context, source
    opposite_torque = target_units * target_distance + deficit
    options = []
    for opposite_units, opposite_distance in product_pairs(opposite_torque):
        target = role_record(
            side, target_units, target_distance,
            opposite_units, opposite_distance,
        )
        hybrid = role_record(
            side, hybrid_units, hybrid_distance,
            opposite_units, opposite_distance,
        )
        if target is None or hybrid is None:
            continue
        scene_ids = {target['scene_id'], hybrid['scene_id']}
        if scene_ids & excluded_scene_ids or len(scene_ids) != 2:
            continue
        if target['outcome'] != opposite_side(side):
            continue
        dose = (
            (source - base) * context
            if factor == 'weight' else context * (source - base)
        )
        coordinate = dose - deficit
        expected = (
            opposite_side(side) if coordinate < 0
            else ('balanced' if coordinate == 0 else side)
        )
        if hybrid['outcome'] != expected:
            continue
        options.append({
            'context_value': int(context),
            'physical_dose': int(dose),
            'target_deficit': int(deficit),
            'hybrid_clearance': int(coordinate),
            'transition': (
                'no_flip' if coordinate < 0
                else ('balanced' if coordinate == 0 else 'flip')
            ),
            'target': target,
            'hybrid': hybrid,
        })
    if not options:
        return None
    return sorted(options, key=lambda row: stable_key((
        factor, side, base, source, context, deficit,
        row['target']['scene_id'], row['hybrid']['scene_id'],
    )))[0]

def choose_donor(side, base, source, factor, forbidden):
    candidates = []
    contexts = range(1, 7) if factor == 'weight' else range(1, 6)
    for context in contexts:
        if factor == 'weight':
            anchor_units, anchor_distance = base, context
            source_units, source_distance = source, context
        else:
            anchor_units, anchor_distance = context, base
            source_units, source_distance = context, source
        source_torque = source_units * source_distance
        for opposite_torque in range(source_torque + 1, 31):
            for opposite_units, opposite_distance in product_pairs(opposite_torque):
                anchor = role_record(
                    side, anchor_units, anchor_distance,
                    opposite_units, opposite_distance,
                )
                donor_source = role_record(
                    side, source_units, source_distance,
                    opposite_units, opposite_distance,
                )
                if anchor is None or donor_source is None:
                    continue
                scene_ids = {anchor['scene_id'], donor_source['scene_id']}
                if (
                    scene_ids & excluded_scene_ids
                    or scene_ids & forbidden
                    or len(scene_ids) != 2
                ):
                    continue
                if not all(
                    role['outcome'] == opposite_side(side)
                    for role in (anchor, donor_source)
                ):
                    continue
                candidates.append({
                    'source_context': int(context),
                    'anchor': anchor,
                    'source': donor_source,
                })
    if not candidates:
        return None
    return sorted(candidates, key=lambda row: stable_key((
        'donor', factor, side, base, source,
        row['source_context'], row['anchor']['scene_id'],
    )))[0]

if CANDIDATE_PATH.is_file() and CANDIDATE_LOCK_PATH.is_file():
    candidate_groups = read_jsonl(CANDIDATE_PATH)
    cell_stage('Using the existing locked candidate pool')
else:
    candidate_groups = []
    for factor in FACTORS:
        bases = range(1, 6) if factor == 'weight' else range(1, 7)
        contexts = range(1, 7) if factor == 'weight' else range(1, 6)
        for side in ('left', 'right'):
            for base, source in itertools.combinations(bases, 2):
                delta = source - base
                for deficit in range(1, 30):
                    context_options = []
                    for context in contexts:
                        option = choose_context_pair(
                            side, base, source, context, deficit, factor
                        )
                        if option is not None:
                            context_options.append(option)
                    transitions = {row['transition'] for row in context_options}
                    if len(context_options) < CONTEXTS_PER_FAMILY:
                        continue
                    if 'no_flip' not in transitions or 'flip' not in transitions:
                        continue
                    context_scene_ids = {
                        role['scene_id']
                        for option in context_options
                        for role in (option['target'], option['hybrid'])
                    }
                    donor = choose_donor(
                        side, base, source, factor, context_scene_ids
                    )
                    if donor is None:
                        continue
                    family_id = 'matched_' + factor + '_' + stable_key((
                        factor, side, base, source, deficit,
                        donor['anchor']['scene_id'],
                        tuple(option['target']['scene_id'] for option in context_options),
                    ))[:18]
                    candidate_groups.append({
                        'family_id': family_id,
                        'factor': factor,
                        'changed_side': side,
                        'factor_base': int(base),
                        'factor_source': int(source),
                        'factor_delta': int(delta),
                        'target_deficit': int(deficit),
                        'counterbalance_phase': int(
                            stable_key(('phase', family_id))[-1], 16
                        ) % 2,
                        'donor': donor,
                        'context_options': sorted(
                            context_options,
                            key=lambda row: row['physical_dose'],
                        ),
                    })
    candidate_groups = sorted(
        candidate_groups, key=lambda row: row['family_id']
    )
    assert candidate_groups
    write_jsonl_atomic(CANDIDATE_PATH, candidate_groups)
    candidate_identity = {
        'schema_version': 'matched-clearance-candidates-1.0',
        'candidate_count': len(candidate_groups),
        'counts_by_factor': dict(Counter(
            row['factor'] for row in candidate_groups
        )),
        'excluded_scene_count': len(excluded_scene_ids),
        'selection_uses_model_outputs': False,
        'exact_target_clearance_match': True,
        'family_ids': [row['family_id'] for row in candidate_groups],
        'seed': SEED,
    }
    write_json_atomic(CANDIDATE_LOCK_PATH, {
        'identity': candidate_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

assert all(
    len(row['context_options']) >= CONTEXTS_PER_FAMILY
    and {'no_flip', 'flip'} <= {
        option['transition'] for option in row['context_options']
    }
    for row in candidate_groups
)
print('Candidate groups:', len(candidate_groups))
print('By factor:', Counter(row['factor'] for row in candidate_groups))
print('Excluded prior scenes:', len(excluded_scene_ids))

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=OUTPUT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {}
for example in outcome_examples:
    outcome_by_scene.setdefault(str(example.scene_id), example)
print('Outcome examples:', len(outcome_by_scene))
```

## 2. Large natural-endpoint competence screen


```python
%%cell_progress 8/15 Define the frozen direct prompt and screen scorer
cell_stage('Defining deterministic label mappings and batched log-probability scoring')
import math
from dataclasses import replace as dataclass_replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def mapping_variant_for(family, option):
    index = int(option['context_value']) + int(family['counterbalance_phase'])
    return MAPPING_VARIANTS[index % len(MAPPING_VARIANTS)]

def mapping_for_variant(variant):
    semantics = (
        ('left', 'balanced', 'right')
        if variant == 'forward'
        else ('right', 'balanced', 'left')
    )
    return tuple(zip(('A', 'B', 'C'), semantics))

def custom_prompt(mapping):
    options = '\n'.join(
        f'{label}. {SEMANTIC_TEXT[semantic]}'
        for label, semantic in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT,
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def screen_key(scene_id, mapping_variant):
    return f'{scene_id}|{mapping_variant}'

def make_screen_example(request):
    mapping = mapping_for_variant(request['mapping_variant'])
    semantics = tuple(value for _, value in mapping)
    labels = tuple(label for label, _ in mapping)
    outcome = str(request['gold_semantic'])
    base = outcome_by_scene[str(request['scene_id'])]
    return dataclass_replace(
        base,
        example_id='readiness:' + request['request_key'],
        question_id='readiness_' + stable_key(request['request_key'])[:18],
        question_kind='outcome',
        variant='competence_matched_clearance_readiness',
        prompt=custom_prompt(mapping),
        labels=labels,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=labels[semantics.index(outcome)],
        gold_semantic=outcome,
    )

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def score_records(prepared, logits):
    output = []
    for batch_index, example in enumerate(prepared['examples']):
        logprobs = torch.log_softmax(
            logits[batch_index, -1, :].float(), dim=-1
        )
        scores = [
            float(logprobs[prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        predicted_index = int(np.argmax(scores))
        output.append({
            'scene_id': str(example.scene_id),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': bool(
                example.option_semantics[predicted_index] == example.gold_semantic
            ),
        })
    return output

print('Frozen prompt and scorer ready.')
```

```python
%%cell_progress 9/15 Build the deduplicated behavior-screen request set
cell_stage('Deduplicating scene and label-mapping requests across candidate groups')

request_by_key = {}
usage_count = Counter()
for family in candidate_groups:
    for option in family['context_options']:
        variant = mapping_variant_for(family, option)
        for role_name in ('target', 'hybrid'):
            role = option[role_name]
            key = screen_key(role['scene_id'], variant)
            request = {
                'request_key': key,
                'scene_id': str(role['scene_id']),
                'mapping_variant': variant,
                'gold_semantic': str(role['outcome']),
            }
            if key in request_by_key:
                assert request_by_key[key] == request
            else:
                request_by_key[key] = request
            usage_count[key] += 1

screen_requests = sorted(
    request_by_key.values(), key=lambda row: row['request_key']
)
assert screen_requests
assert len(screen_requests) <= 1800, len(screen_requests)
if REQUEST_PATH.is_file():
    previous_requests = read_jsonl(REQUEST_PATH)
    assert previous_requests == screen_requests
else:
    write_jsonl_atomic(REQUEST_PATH, screen_requests)

REQUEST_SHA256 = sha256_file(REQUEST_PATH)
print('Candidate usages / unique model requests:',
      sum(usage_count.values()), len(screen_requests))
print('Mapping counts:', Counter(
    row['mapping_variant'] for row in screen_requests
))
```

```python
%%cell_progress 10/15 Run the resumable natural endpoint screen
cell_stage('Scoring target and hybrid images without activation capture')

def chunks(values, size):
    for start in range(0, len(values), size):
        yield start // size, values[start:start + size]

screen_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'candidate_sha256': sha256_file(CANDIDATE_PATH),
    'request_sha256': REQUEST_SHA256,
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'batch_size': SCREEN_BATCH_SIZE,
}
SCREEN_IDENTITY_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(screen_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(
    total=len(screen_requests),
    desc='Natural endpoint behavior screen', unit='image-map',
)
for chunk_index, current_requests in chunks(screen_requests, SCREEN_BATCH_SIZE):
    shard_path = SCREEN_SHARD_DIR / f'chunk_{chunk_index:05d}.json'
    if shard_path.is_file():
        document = json.loads(shard_path.read_text(encoding='utf-8'))
        assert document['screen_identity_sha256'] == SCREEN_IDENTITY_SHA256
        assert len(document['records']) == len(current_requests)
        progress.update(len(current_requests))
        continue
    examples = [make_screen_example(row) for row in current_requests]
    prepared = prepare_examples(examples)
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    scored = score_records(prepared, output.logits)
    records = []
    for request, score in zip(current_requests, scored):
        assert score['scene_id'] == request['scene_id']
        score.update({
            'request_key': request['request_key'],
            'mapping_variant': request['mapping_variant'],
        })
        records.append(score)
    write_json_atomic(shard_path, {
        'schema_version': 'readiness-screen-shard-1.0',
        'screen_identity_sha256': SCREEN_IDENTITY_SHA256,
        'records': records,
    })
    progress.update(len(current_requests))
    progress.set_postfix_str(
        f'{chunk_index + 1} chunks | {len(current_requests)} current'
    )
    del prepared, output
    torch.cuda.empty_cache()
progress.close()

screen_records = []
for chunk_index, current_requests in chunks(screen_requests, SCREEN_BATCH_SIZE):
    shard_path = SCREEN_SHARD_DIR / f'chunk_{chunk_index:05d}.json'
    document = json.loads(shard_path.read_text(encoding='utf-8'))
    assert document['screen_identity_sha256'] == SCREEN_IDENTITY_SHA256
    screen_records.extend(document['records'])
assert len(screen_records) == len(screen_requests)
assert len({row['request_key'] for row in screen_records}) == len(screen_records)
write_jsonl_atomic(SCREEN_RESULT_PATH, screen_records)
print('Completed unique behavior requests:', len(screen_records))
```

```python
%%cell_progress 11/15 Find behavior-competent matched-clearance families
cell_stage('Requiring exact correctness at every selected target and hybrid endpoint')

screen_by_key = {row['request_key']: row for row in screen_records}
cell_rows = []
qualified_groups = []
for family in candidate_groups:
    competent_options = []
    for option in family['context_options']:
        variant = mapping_variant_for(family, option)
        target_key = screen_key(option['target']['scene_id'], variant)
        hybrid_key = screen_key(option['hybrid']['scene_id'], variant)
        target_result = screen_by_key[target_key]
        hybrid_result = screen_by_key[hybrid_key]
        endpoint_pair_correct = bool(
            target_result['correct'] and hybrid_result['correct']
        )
        current = {
            'family_id': family['family_id'],
            'factor': family['factor'],
            'changed_side': family['changed_side'],
            'factor_delta': family['factor_delta'],
            'target_deficit': family['target_deficit'],
            'context_value': option['context_value'],
            'physical_dose': option['physical_dose'],
            'transition': option['transition'],
            'mapping_variant': variant,
            'target_correct': bool(target_result['correct']),
            'hybrid_correct': bool(hybrid_result['correct']),
            'endpoint_pair_correct': endpoint_pair_correct,
        }
        cell_rows.append(current)
        if endpoint_pair_correct:
            competent = dict(option)
            competent['mapping_variant'] = variant
            competent_options.append(competent)

    valid_triples = []
    for triple in itertools.combinations(competent_options, CONTEXTS_PER_FAMILY):
        transitions = {row['transition'] for row in triple}
        if 'no_flip' not in transitions or 'flip' not in transitions:
            continue
        doses = [int(row['physical_dose']) for row in triple]
        if len(set(doses)) != CONTEXTS_PER_FAMILY:
            continue
        valid_triples.append(tuple(sorted(
            triple, key=lambda row: row['physical_dose']
        )))
    if not valid_triples:
        continue
    selected_triple = sorted(valid_triples, key=lambda triple: (
        -(triple[-1]['physical_dose'] - triple[0]['physical_dose']),
        -int(any(row['transition'] == 'balanced' for row in triple)),
        stable_key((family['family_id'], tuple(
            row['context_value'] for row in triple
        ))),
    ))[0]
    qualified = dict(family)
    qualified['selected_contexts'] = list(selected_triple)
    qualified['all_scene_ids'] = sorted({
        family['donor'][role]['scene_id']
        for role in ('anchor', 'source')
    } | {
        option[role]['scene_id']
        for option in selected_triple for role in ('target', 'hybrid')
    })
    assert len(qualified['all_scene_ids']) == 2 + 2 * CONTEXTS_PER_FAMILY
    qualified_groups.append(qualified)

screen_cell_frame = pd.DataFrame(cell_rows)
screen_cell_frame.to_csv(ANALYSIS_DIR / 'candidate_context_screen.csv', index=False)
qualified_summary = pd.DataFrame([{
    'family_id': row['family_id'],
    'factor': row['factor'],
    'changed_side': row['changed_side'],
    'factor_delta': row['factor_delta'],
    'target_deficit': row['target_deficit'],
    'dose_min': min(x['physical_dose'] for x in row['selected_contexts']),
    'dose_max': max(x['physical_dose'] for x in row['selected_contexts']),
    'mapping_pattern': '|'.join(
        x['mapping_variant'] for x in row['selected_contexts']
    ),
} for row in qualified_groups])
qualified_summary.to_csv(ANALYSIS_DIR / 'qualified_family_pool.csv', index=False)

print('Qualified groups:', len(qualified_groups))
print('By factor:', Counter(row['factor'] for row in qualified_groups))
display(screen_cell_frame.groupby(['factor', 'transition'])[[
    'target_correct', 'hybrid_correct', 'endpoint_pair_correct'
]].mean().round(4))
```

```python
%%cell_progress 12/15 Freeze an independent confirmation manifest if feasible
cell_stage('Selecting scene-disjoint, factor-balanced qualified families')

def select_factor_support(factor, forbidden, attempt, target_count):
    groups = defaultdict(list)
    for row in qualified_groups:
        if row['factor'] != factor:
            continue
        groups[(row['changed_side'], row['factor_delta'])].append(row)
    for key, values in groups.items():
        values.sort(key=lambda row: stable_key((attempt, key, row['family_id'])))
    keys = sorted(groups, key=lambda key: stable_key((attempt, 'stratum', key)))
    cursors = {key: 0 for key in keys}
    selected, used = [], set(forbidden)
    while len(selected) < target_count:
        changed = False
        for key in keys:
            values = groups[key]
            while cursors[key] < len(values):
                row = values[cursors[key]]
                cursors[key] += 1
                scene_ids = set(row['all_scene_ids'])
                if scene_ids & used:
                    continue
                selected.append(row)
                used.update(scene_ids)
                changed = True
                break
            if len(selected) == target_count:
                break
        if not changed:
            break
    return selected

selected_by_factor = None
selection_policy = None
for attempt in range(512):
    order = FACTORS if attempt % 2 else tuple(reversed(FACTORS))
    current, globally_used = {}, set()
    for factor in order:
        chosen = select_factor_support(
            factor, globally_used, attempt, TARGET_FAMILIES_PER_FACTOR
        )
        current[factor] = chosen
        globally_used.update(
            scene_id for row in chosen for scene_id in row['all_scene_ids']
        )
    if all(
        len(current[factor]) == TARGET_FAMILIES_PER_FACTOR
        for factor in FACTORS
    ):
        selected_by_factor = current
        selection_policy = 'global_scene_disjoint_target_reached'
        break

if selected_by_factor is None:
    selected_by_factor = {}
    for factor in FACTORS:
        best = []
        for attempt in range(512):
            chosen = select_factor_support(
                factor, set(), 1000 + attempt, TARGET_FAMILIES_PER_FACTOR
            )
            if len(chosen) > len(best):
                best = chosen
            if len(best) == TARGET_FAMILIES_PER_FACTOR:
                break
        selected_by_factor[factor] = best
    selection_policy = 'within_factor_scene_disjoint_maximum_support'

selected_manifest = []
for factor in FACTORS:
    for row in selected_by_factor[factor]:
        frozen = {
            key: value for key, value in row.items()
            if key not in ('context_options', 'selected_contexts')
        }
        frozen['backgrounds'] = {
            background: context
            for background, context in zip(BACKGROUNDS, row['selected_contexts'])
        }
        selected_manifest.append(frozen)
selected_manifest.sort(key=lambda row: row['family_id'])
write_jsonl_atomic(SELECTED_MANIFEST_PATH, selected_manifest)

selected_counts = Counter(row['factor'] for row in selected_manifest)
minimum_count = min(selected_counts.get(factor, 0) for factor in FACTORS)
if minimum_count >= TARGET_FAMILIES_PER_FACTOR:
    READINESS_DECISION = 'ready_for_24_family_confirmatory_mechanism'
elif minimum_count >= MIN_FAMILIES_PER_FACTOR:
    READINESS_DECISION = 'ready_for_reduced_power_confirmatory_mechanism'
else:
    READINESS_DECISION = 'behavioral_ceiling_prevents_confirmatory_mechanism'

print('Selection policy:', selection_policy)
print('Selected counts:', selected_counts)
print('Readiness decision:', READINESS_DECISION)
```

## 3. Readiness decision and frozen downstream support


```python
%%cell_progress 13/15 Audit readiness, independence, and effective sample size
cell_stage('Checking competence, exact clearance matching, and prior-scene exclusion')
import gc
try:
    del adapter
except NameError:
    pass
gc.collect()
torch.cuda.empty_cache()

violations = []
factor_scene_lists = defaultdict(list)
mapping_counts = Counter()
for row in selected_manifest:
    contexts = list(row['backgrounds'].values())
    deficits = {int(option['target_deficit']) for option in contexts}
    transitions = {option['transition'] for option in contexts}
    if len(deficits) != 1:
        violations.append((row['family_id'], 'clearance'))
    if not {'no_flip', 'flip'} <= transitions:
        violations.append((row['family_id'], 'transition_span'))
    if len({option['physical_dose'] for option in contexts}) != CONTEXTS_PER_FAMILY:
        violations.append((row['family_id'], 'dose'))
    scene_ids = set(row['all_scene_ids'])
    if scene_ids & excluded_scene_ids:
        violations.append((row['family_id'], 'prior_overlap'))
    if len(scene_ids) != 2 + 2 * CONTEXTS_PER_FAMILY:
        violations.append((row['family_id'], 'within_family_overlap'))
    factor_scene_lists[row['factor']].extend(scene_ids)
    for option in contexts:
        mapping_counts[(row['factor'], option['mapping_variant'])] += 1
        for role_name in ('target', 'hybrid'):
            role = option[role_name]
            result = screen_by_key[
                screen_key(role['scene_id'], option['mapping_variant'])
            ]
            if not result['correct']:
                violations.append((row['family_id'], 'competence'))

within_factor_disjoint = all(
    len(values) == len(set(values))
    for values in factor_scene_lists.values()
)
cross_factor_overlap = (
    set(factor_scene_lists.get('weight', []))
    & set(factor_scene_lists.get('distance', []))
)
assert not violations, violations[:5]
assert within_factor_disjoint

screen_request_frame = pd.DataFrame(screen_records)
screen_accuracy = float(screen_request_frame['correct'].mean())
qualified_counts = Counter(row['factor'] for row in qualified_groups)
audit_document = {
    'schema_version': 'competence-matched-clearance-readiness-audit-1.0',
    'candidate_groups': len(candidate_groups),
    'unique_behavior_requests': len(screen_records),
    'unconditional_request_accuracy': screen_accuracy,
    'qualified_counts': dict(qualified_counts),
    'selected_counts': dict(selected_counts),
    'selection_policy': selection_policy,
    'readiness_decision': READINESS_DECISION,
    'violations': len(violations),
    'exact_clearance_match': True,
    'competence_conditioned': True,
    'selection_uses_activation_data': False,
    'within_factor_scene_disjoint': within_factor_disjoint,
    'global_scene_disjoint': not cross_factor_overlap,
    'cross_factor_overlap_count': len(cross_factor_overlap),
    'prior_scene_overlap_count': len(
        set(scene_id for values in factor_scene_lists.values() for scene_id in values)
        & excluded_scene_ids
    ),
    'mapping_counts': {
        '|'.join(key): value for key, value in sorted(mapping_counts.items())
    },
}
write_json_atomic(ANALYSIS_DIR / 'readiness_audit.json', audit_document)
display(pd.DataFrame([audit_document]).T.rename(columns={0: 'value'}))
```
