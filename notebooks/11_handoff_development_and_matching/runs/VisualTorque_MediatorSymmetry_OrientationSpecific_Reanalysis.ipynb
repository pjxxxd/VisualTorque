```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 3/10 Lock the corrective orientation-specific analysis
cell_stage('Freezing parent files, pair unit, contrasts, and inherited margins')
import hashlib
import json
from datetime import datetime, timezone

SEED = 20261109
BOOTSTRAP_REPLICATES = 20000
ALPHA_LEVEL = 0.05

PARENT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'qwen27b_mediator_matched_handoff_symmetry_v1'
)
PARENT_RUN_PATH = PARENT_ROOT / 'RUN_COMPLETE.json'
PARENT_PROTOCOL_PATH = PARENT_ROOT / 'frozen_protocol.json'
PARENT_BASELINE_PATH = PARENT_ROOT / 'baseline_mapping_replay.jsonl'
PARENT_ESTIMANDS_PATH = PARENT_ROOT / 'analysis' / 'mapping_averaged_estimands.csv'
PARENT_RECORDS_PATH = PARENT_ROOT / 'analysis' / 'paired_intervention_records.csv'
PARENT_DECISION_PATH = PARENT_ROOT / 'analysis' / 'mediator_symmetry_decision.json'
PARENT_AUDIT_PATH = PARENT_ROOT / 'analysis' / 'mediator_symmetry_audit.json'

OUTPUT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'mediator_symmetry_orientation_reanalysis_v1'
)
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_reanalysis_protocol.json'
ANALYSIS_DIR = OUTPUT_ROOT / 'analysis'
FIGURE_DIR = ANALYSIS_DIR / 'figures'
for path in (OUTPUT_ROOT, ANALYSIS_DIR, FIGURE_DIR):
    path.mkdir(parents=True, exist_ok=True)

from visualtorque_checks import sha256_file

def json_default(value):
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(type(value).__name__)

def write_json_atomic(path, value):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    temporary.write_text(json.dumps(
        value, indent=2, ensure_ascii=False, default=json_default
    ) + '\n', encoding='utf-8')
    temporary.replace(path)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

required = (
    PARENT_RUN_PATH, PARENT_PROTOCOL_PATH, PARENT_BASELINE_PATH,
    PARENT_ESTIMANDS_PATH, PARENT_RECORDS_PATH,
    PARENT_DECISION_PATH, PARENT_AUDIT_PATH,
)
for path in required:
    assert path.is_file(), f'Missing parent artifact: {path}'

parent_protocol = json.loads(PARENT_PROTOCOL_PATH.read_text(encoding='utf-8'))
parent_identity = parent_protocol['identity']
ANSWER_EQUIVALENCE_MARGIN = float(parent_identity['answer_equivalence_margin'])
CODE_EQUIVALENCE_MARGIN = float(parent_identity['code_equivalence_margin'])
MEDIATOR_MATCH_FRACTION = 0.20
parent_audit = json.loads(PARENT_AUDIT_PATH.read_text(encoding='utf-8'))
SEMANTIC_NORM = float(
    parent_identity['gain_calibration']['target_l49_projection']
)
L49_EQUIVALENCE_MARGIN = MEDIATOR_MATCH_FRACTION * SEMANTIC_NORM

protocol_identity = {
    'schema_version': 'orientation-specific-mediator-reanalysis-1.0',
    'analysis_status': 'corrective post-hoc reanalysis; no new model outputs',
    'parent_run_sha256': sha256_file(PARENT_RUN_PATH),
    'parent_records_sha256': sha256_file(PARENT_RECORDS_PATH),
    'parent_estimands_sha256': sha256_file(PARENT_ESTIMANDS_PATH),
    'bootstrap_unit': 'original-mirror pair within orientation',
    'orientation_specific_equivalence_required': True,
    'pooled_cancellation_forbidden': True,
    'margins': {
        'l49_coordinate': L49_EQUIVALENCE_MARGIN,
        'l55_code': CODE_EQUIVALENCE_MARGIN,
        'answer_margin': ANSWER_EQUIVALENCE_MARGIN,
    },
    'seed': SEED,
}
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Corrective analysis only; no model will be loaded.')
print('Equivalence margins:', protocol_identity['margins'])
```

```python
%%cell_progress 4/10 Load and validate all parent scene-level records
cell_stage('Checking pair counts, mappings, conditions, and parent completion')

parent_run = json.loads(PARENT_RUN_PATH.read_text(encoding='utf-8'))
parent_decision = json.loads(PARENT_DECISION_PATH.read_text(encoding='utf-8'))
baseline_frame = pd.DataFrame(read_jsonl(PARENT_BASELINE_PATH))
estimand_frame = pd.read_csv(PARENT_ESTIMANDS_PATH)
record_frame = pd.read_csv(PARENT_RECORDS_PATH)

assert parent_run['status'] == 'complete'
assert estimand_frame['pair_id'].nunique() == 16
assert estimand_frame['scene_id'].nunique() == 32
assert set(estimand_frame['orientation']) == {'original', 'mirror'}
assert baseline_frame.groupby(['pair_id', 'orientation']).size().eq(2).all()
assert record_frame.groupby(
    ['scene_id', 'mapping_variant', 'condition']
).size().eq(1).all()
print('Parent automatic decision:', parent_decision['primary_decision'])
print('Pairs / scenes / intervention records:',
      estimand_frame['pair_id'].nunique(),
      estimand_frame['scene_id'].nunique(), len(record_frame))
```

```python
%%cell_progress 5/10 Quantify the orientation-correctness confound
cell_stage('Separating what the mirror design changes simultaneously')

baseline_summary = baseline_frame.groupby('orientation').agg(
    mapping_runs=('correct', 'size'),
    accuracy=('correct', 'mean'),
    shortcut_rate=('predicted_shortcut', 'mean'),
).reset_index()
baseline_pivot = baseline_summary.set_index('orientation')
BASELINE_PERFECTLY_CONFOUNDED = bool(
    baseline_pivot.loc['original', 'accuracy'] == 0
    and baseline_pivot.loc['mirror', 'accuracy'] == 1
)
display(baseline_summary.round(4))
print('Orientation perfectly confounded with baseline correctness:',
      BASELINE_PERFECTLY_CONFOUNDED)
```

## 2. Non-cancelling orientation-specific tests


```python
%%cell_progress 6/10 Run separate orientation-specific equivalence tests
cell_stage('Bootstrapping 16 pairs without pooling orientations')

rng = np.random.default_rng(SEED)

def bootstrap_summary(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    assert len(values) == 16, len(values)
    draws = rng.choice(
        values, size=(BOOTSTRAP_REPLICATES, len(values)), replace=True
    ).mean(axis=1)
    return {
        'estimate': float(values.mean()),
        'ci_low': float(np.quantile(draws, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA_LEVEL / 2)),
        'n_pairs': 16,
    }

def pivot(frame, metric):
    return frame.pivot(index='pair_id', columns='condition', values=metric)

orientation_tests = []
equivalence = {}
margin_by_metric = {
    'l49_toward_exact': L49_EQUIVALENCE_MARGIN,
    'l55_toward_exact_code': CODE_EQUIVALENCE_MARGIN,
    'margin_shift': ANSWER_EQUIVALENCE_MARGIN,
}
name_by_metric = {
    'l49_toward_exact': 'l49_coordinate',
    'l55_toward_exact_code': 'l55_code',
    'margin_shift': 'answer_margin',
}

for orientation in ('original', 'mirror'):
    current = estimand_frame[estimand_frame['orientation'].eq(orientation)]
    equivalence[orientation] = {}
    for metric, margin in margin_by_metric.items():
        values = pivot(current, metric)
        difference = values['upstream_matched'] - values['l49_direct']
        result = bootstrap_summary(difference)
        passed = bool(result['ci_low'] > -margin and result['ci_high'] < margin)
        name = name_by_metric[metric]
        equivalence[orientation][name] = {
            'margin': float(margin), 'passed': passed, **result,
        }
        orientation_tests.append({
            'family': 'orientation_equivalence',
            'orientation': orientation, 'metric': name,
            'equivalence_margin': float(margin),
            'equivalence_passed': passed, **result,
        })

    l49 = pivot(current, 'l49_toward_exact')
    for name, values in (
        ('matched_shift', l49['upstream_matched']),
        ('matched_minus_random',
         l49['upstream_matched'] - l49['upstream_random_matched']),
        ('matched_minus_reverse',
         l49['upstream_matched'] - l49['upstream_reverse_matched']),
    ):
        orientation_tests.append({
            'family': 'orientation_transport',
            'orientation': orientation, 'metric': name,
            **bootstrap_summary(values),
        })

test_frame = pd.DataFrame(orientation_tests)
test_frame.to_csv(ANALYSIS_DIR / 'orientation_specific_tests.csv', index=False)
display(test_frame.round(4))
```

```python
%%cell_progress 7/10 Test paired side symmetry without pooling downstream states
cell_stage('Comparing original and mirror effects within each physical pair')

side_tests = []
for metric in ('l49_toward_exact', 'l55_toward_exact_code', 'margin_shift'):
    current = estimand_frame[
        estimand_frame['condition'].eq('upstream_matched')
    ].pivot(index='pair_id', columns='orientation', values=metric)
    result = bootstrap_summary(current['original'] - current['mirror'])
    side_tests.append({
        'metric': name_by_metric[metric] + '_original_minus_mirror', **result,
    })
side_frame = pd.DataFrame(side_tests)
side_frame.to_csv(ANALYSIS_DIR / 'paired_side_differences.csv', index=False)
display(side_frame.round(4))
```

```python
%%cell_progress 8/10 Apply the corrected non-cancelling decision rule
cell_stage('Requiring every claimed equivalence within each orientation')

transport_frame = test_frame[test_frame['family'].eq('orientation_transport')]
transport_lookup = transport_frame.set_index(['orientation', 'metric'])
transport_gates = {}
for orientation in ('original', 'mirror'):
    transport_gates[orientation] = bool(
        transport_lookup.loc[(orientation, 'matched_shift'), 'ci_low'] > 0
        and transport_lookup.loc[(orientation, 'matched_minus_random'), 'ci_low'] > 0
        and transport_lookup.loc[(orientation, 'matched_minus_reverse'), 'ci_low'] > 0
    )

l49_match_both = all(
    equivalence[orientation]['l49_coordinate']['passed']
    for orientation in ('original', 'mirror')
)
l55_equivalent_both = all(
    equivalence[orientation]['l55_code']['passed']
    for orientation in ('original', 'mirror')
)
answer_equivalent_both = all(
    equivalence[orientation]['answer_margin']['passed']
    for orientation in ('original', 'mirror')
)

if all(transport_gates.values()) and l49_match_both:
    if l55_equivalent_both and answer_equivalent_both:
        CORRECTED_DECISION = 'l49_coordinate_is_orientation_specific_sufficient_mediator'
    else:
        CORRECTED_DECISION = (
            'symmetric_l48_to_l49_transport_but_l49_sufficiency_not_established'
        )
elif not transport_gates['mirror']:
    CORRECTED_DECISION = 'fixed_l48_route_does_not_transport_on_left_mirrors'
else:
    CORRECTED_DECISION = 'global_gain_does_not_match_l49_in_both_orientations'

decision = {
    'analysis_status': 'corrective post-hoc reanalysis',
    'parent_automatic_decision': parent_decision['primary_decision'],
    'corrected_decision': CORRECTED_DECISION,
    'transport_gates': transport_gates,
    'l49_matched_in_both_orientations': l49_match_both,
    'l55_equivalent_in_both_orientations': l55_equivalent_both,
    'answer_equivalent_in_both_orientations': answer_equivalent_both,
    'orientation_baseline_correctness_perfectly_confound': (
        BASELINE_PERFECTLY_CONFOUNDED
    ),
    'orientation_specific_equivalence': equivalence,
    'interpretation': (
        'The fixed route transports symmetrically to the measured L49 '
        'coordinate. Failure of orientation-specific downstream equivalence '
        'means that this coordinate is not established as a sufficient '
        'mediator. Side and baseline correctness remain inseparable here.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'corrected_mediator_decision.json', decision)
print('Parent automatic decision:', parent_decision['primary_decision'])
print('Corrected decision:', CORRECTED_DECISION)
print('Orientation equivalence:', {
    orientation: {
        metric: value['passed']
        for metric, value in equivalence[orientation].items()
    } for orientation in equivalence
})
```

## 3. Corrected result
