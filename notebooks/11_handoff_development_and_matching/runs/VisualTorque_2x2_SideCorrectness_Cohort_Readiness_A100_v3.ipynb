```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 14 tracked cells.')
```

```python
%%cell_progress 4/14 Lock the 2x2 readiness protocol
cell_stage('Freezing model, support exclusions, mappings, tiers, and gates')
import hashlib
import json
from datetime import datetime, timezone

MODEL_KEY = 'qwen27b'
MODEL_ID = 'Qwen/Qwen3.5-27B'
MODEL_REVISION = '373612cb4f9288fcf550b526d86d85febedd9801'
MODEL_DTYPE = 'bfloat16'
DEVICE_MAP = 'auto'

SEED = 20261112
LABELS = ('A', 'B', 'C')
MAPPING_VARIANTS = ('forward', 'reverse')
QUESTION_KINDS = ('outcome',)
MIN_PAIRS_PER_RECIPROCAL_PATTERN = 8
TARGET_PAIRS_PER_RECIPROCAL_PATTERN = 16

PARENT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'qwen27b_nettorque_semantic_handoff_mediation_v1'
)
PARENT_RUN_PATH = PARENT_ROOT / 'RUN_COMPLETE.json'
PARENT_SUPPORT_PATH = PARENT_ROOT / 'frozen_support.jsonl'
DEPTH_SUPPORT_PATH = (
    PROJECT_ROOT / 'results' / 'semantic_label_depth_localization_v1'
    / 'depth_support.jsonl'
)

OUTPUT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'side_correctness_2x2_readiness_v3'
)
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_protocol.json'
CANDIDATE_PATH = OUTPUT_ROOT / 'paired_candidates.jsonl'
MIRROR_DIR = OUTPUT_ROOT / 'mirrored_images'
SCREEN_SHARD_DIR = OUTPUT_ROOT / 'screen_shards'
SCREEN_RECORDS_PATH = OUTPUT_ROOT / 'screen_records.jsonl'
PAIR_TABLE_PATH = OUTPUT_ROOT / 'pair_readiness.csv'
SELECTED_MANIFEST_PATH = OUTPUT_ROOT / 'selected_2x2_manifest.jsonl'
ANALYSIS_DIR = OUTPUT_ROOT / 'analysis'
FIGURE_DIR = ANALYSIS_DIR / 'figures'
for path in (
    OUTPUT_ROOT, MIRROR_DIR, SCREEN_SHARD_DIR, ANALYSIS_DIR, FIGURE_DIR,
):
    path.mkdir(parents=True, exist_ok=True)

def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return 'sha256:' + digest.hexdigest()

def stable_key(value, seed=SEED):
    return hashlib.sha256(f'{seed}:{value!r}'.encode('utf-8')).hexdigest()

def json_default(value):
    if isinstance(value, np.generic):
        return value.item()
    if torch.is_tensor(value) and value.numel() == 1:
        return value.item()
    raise TypeError(type(value).__name__)

def write_json_atomic(path, value):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    temporary.write_text(json.dumps(
        value, indent=2, ensure_ascii=False, default=json_default
    ) + '\n', encoding='utf-8')
    temporary.replace(path)

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(
                row, sort_keys=True, default=json_default
            ) + '\n')
    temporary.replace(path)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

for required in (PARENT_RUN_PATH, PARENT_SUPPORT_PATH, DEPTH_SUPPORT_PATH):
    assert required.is_file(), f'Missing prerequisite: {required}'

protocol_identity = {
    'schema_version': 'side-correctness-2x2-readiness-1.2',
    'analysis_status': 'behavior-only support screening',
    'model_id': MODEL_ID, 'revision': MODEL_REVISION,
    'parent_support_sha256': sha256_file(PARENT_SUPPORT_PATH),
    'depth_support_sha256': sha256_file(DEPTH_SUPPORT_PATH),
    'primary_tier': (
        'both side-swapped configurations are probe_test; parent target and '
        'direction-support geometries excluded'
    ),
    'stable_correct': 'both mappings predict the exact physical side',
    'stable_error': 'both mappings predict the opposite physical side',
    'required_reciprocal_patterns': [
        'left_correct__right_error', 'left_error__right_correct',
    ],
    'reciprocal_matching': (
        'one-to-one within diagnostic_cell x absolute-net-torque margin_bin'
    ),
    'direction_geometry_parser': (
        'full-stack orientation-neutral signature v3; arbitrary stack count'
    ),
    'minimum_pairs_per_pattern': MIN_PAIRS_PER_RECIPROCAL_PATTERN,
    'target_pairs_per_pattern': TARGET_PAIRS_PER_RECIPROCAL_PATTERN,
    'selection_uses_hidden_states': False,
    'selection_uses_future_interventions': False,
    'seed': SEED,
}
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing protocol differs. Use a new versioned output directory.'
    )
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Primary minimum / target pairs per reciprocal pattern:',
      MIN_PAIRS_PER_RECIPROCAL_PATTERN,
      TARGET_PAIRS_PER_RECIPROCAL_PATTERN)
```

```python
%%cell_progress 6/14 Build strict and exploratory side-swapped pairs
cell_stage('Loading all bounded single-stack configurations and outcome prompts')
from collections import Counter, defaultdict
from dataclasses import replace as dataclass_replace
from PIL import Image, ImageOps

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')

def is_single_stack_scene(row):
    stacks = row.get('stacks', [])
    counts = Counter(str(stack.get('side')) for stack in stacks)
    return bool(
        str(row.get('metadata', {}).get('family', '')) == 'single_stack'
        and str(row.get('scene_id', '')).startswith('single_')
        and len(stacks) == 2 and counts == {'left': 1, 'right': 1}
    )

single_scenes = [row for row in scene_records if is_single_stack_scene(row)]
assert len(single_scenes) == 900, {
    'single_stack_scenes': len(single_scenes),
    'total_scenes': len(scene_records),
    'note': 'Single-stack is stored in metadata.family, not problem_type.',
}
scene_by_id = {str(row['scene_id']): row for row in single_scenes}

def scene_features(scene):
    left = next(stack for stack in scene['stacks'] if stack['side'] == 'left')
    right = next(stack for stack in scene['stacks'] if stack['side'] == 'right')
    derived = scene['derived']
    values = {
        'left_units': int(left['units']),
        'right_units': int(right['units']),
        'left_distance': int(left['distance']),
        'right_distance': int(right['distance']),
        'left_torque': int(derived['left_torque']),
        'right_torque': int(derived['right_torque']),
        'net_torque': int(derived['net_torque']),
        'outcome': str(derived['outcome']),
    }
    sign = lambda value: int(value > 0) - int(value < 0)
    weight_cue = sign(values['left_units'] - values['right_units'])
    distance_cue = sign(values['left_distance'] - values['right_distance'])
    torque_sign = sign(values['net_torque'])
    if torque_sign != 0 and weight_cue != torque_sign and distance_cue == torque_sign:
        diagnostic_cell = 'weight_conflict'
    elif torque_sign != 0 and distance_cue != torque_sign and weight_cue == torque_sign:
        diagnostic_cell = 'distance_conflict'
    elif torque_sign != 0 and weight_cue == torque_sign and distance_cue == torque_sign:
        diagnostic_cell = 'both_cues_aligned'
    else:
        diagnostic_cell = 'mixed_nonbalanced'
    values.update({
        'weight_cue': weight_cue, 'distance_cue': distance_cue,
        'torque_sign': torque_sign, 'diagnostic_cell': diagnostic_cell,
        'absolute_net_torque': abs(values['net_torque']),
        'margin_bin': min(abs(values['net_torque']) // 4, 5),
    })
    return values

def ordered_shape(features):
    return (
        (int(features['left_units']), int(features['left_distance'])),
        (int(features['right_units']), int(features['right_distance'])),
    )

def pair_geometry(features):
    return tuple(sorted(ordered_shape(features)))

physical_by_scene = {
    scene_id: scene_features(scene) for scene_id, scene in scene_by_id.items()
}

loader_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR, output_dir=OUTPUT_ROOT / '_unused_loader',
    model_id=MODEL_ID, model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,), question_kinds=QUESTION_KINDS,
    splits=('probe_train', 'probe_dev', 'probe_test'), batch_size=1,
    generation_every=0, seed=SEED,
)
all_examples = tuple(load_behavior_examples(loader_config))
examples_by_scene = defaultdict(list)
for example in all_examples:
    scene_id = str(example.scene_id)
    if scene_id in scene_by_id and example.question_kind == 'outcome':
        examples_by_scene[scene_id].append(example)

outcome_example_by_scene = {}
for scene_id, candidates in examples_by_scene.items():
    candidates = sorted(candidates, key=lambda example: str(example.variant))
    outcome_example_by_scene[scene_id] = candidates[
        int(stable_key(('prompt-variant', scene_id))[:8], 16) % len(candidates)
    ]
assert len(outcome_example_by_scene) == len(single_scenes)

parent_scene_ids = {
    str(row['scene_id']) for row in read_jsonl(PARENT_SUPPORT_PATH)
    if row.get('scene_id') is not None
}
direction_geometry_keys = set()
direction_role_count = 0
direction_stack_count_distribution = Counter()

def stack_geometry_key(stacks):
    records = tuple(sorted(
        (str(stack['side']), int(stack['units']), int(stack['distance']))
        for stack in stacks
    ))
    mirrored = tuple(sorted(
        ({'left': 'right', 'right': 'left'}[str(stack['side'])],
         int(stack['units']), int(stack['distance']))
        for stack in stacks
    ))
    return min(records, mirrored)

def role_geometry_key(role):
    stacks = role.get('stacks', [])
    if stacks:
        assert all(
            str(stack.get('side')) in ('left', 'right')
            and 'units' in stack and 'distance' in stack
            for stack in stacks
        ), role
        return stack_geometry_key(stacks)
    features = role.get('features', {})
    required = ('left_units', 'right_units', 'left_distance', 'right_distance')
    if all(key in features for key in required):
        return stack_geometry_key((
            {'side': 'left', 'units': features['left_units'],
             'distance': features['left_distance']},
            {'side': 'right', 'units': features['right_units'],
             'distance': features['right_distance']},
        ))
    return None

for family in read_jsonl(DEPTH_SUPPORT_PATH):
    for role in family.get('roles', {}).values():
        direction_role_count += 1
        direction_stack_count_distribution[len(role.get('stacks', []))] += 1
        geometry_key = role_geometry_key(role)
        assert geometry_key is not None, (
            'Direction-support role has neither parseable stacks nor legacy '
            f'features: {role.keys()}'
        )
        direction_geometry_keys.add(geometry_key)
assert direction_role_count > 0
assert direction_geometry_keys

by_geometry = defaultdict(list)
for scene in single_scenes:
    scene_id = str(scene['scene_id'])
    features = physical_by_scene[scene_id]
    if features['net_torque'] != 0:
        by_geometry[pair_geometry(features)].append(scene)

candidate_manifest = []
pair_rows = []
excluded_parent_pair_count = 0
excluded_direction_pair_count = 0
for geometry, members in sorted(by_geometry.items(), key=lambda item: repr(item[0])):
    member_by_shape = {ordered_shape(physical_by_scene[str(row['scene_id'])]): row
                       for row in members}
    if len(member_by_shape) != 2:
        continue
    ordered_members = sorted(
        member_by_shape.values(), key=lambda row: stable_key(('anchor', row['scene_id']))
    )
    anchor, counterpart = ordered_members
    anchor_id = str(anchor['scene_id'])
    counterpart_id = str(counterpart['scene_id'])
    if {anchor_id, counterpart_id} & parent_scene_ids:
        excluded_parent_pair_count += 1
        continue
    if stack_geometry_key(anchor['stacks']) in direction_geometry_keys:
        excluded_direction_pair_count += 1
        continue
    splits = {str(anchor['split']), str(counterpart['split'])}
    if splits == {'probe_test'}:
        tier = 'strict'
    elif 'probe_test' in splits:
        tier = 'exploratory'
    else:
        continue

    pair_id = 'side_pair_' + stable_key((geometry, anchor_id))[:20]
    base_features = physical_by_scene[anchor_id]
    mirror_features = {
        **base_features,
        'left_units': base_features['right_units'],
        'right_units': base_features['left_units'],
        'left_distance': base_features['right_distance'],
        'right_distance': base_features['left_distance'],
        'left_torque': base_features['right_torque'],
        'right_torque': base_features['left_torque'],
        'net_torque': -base_features['net_torque'],
        'outcome': {'left': 'right', 'right': 'left'}[base_features['outcome']],
        'weight_cue': -base_features['weight_cue'],
        'distance_cue': -base_features['distance_cue'],
        'torque_sign': -base_features['torque_sign'],
    }
    mirror_id = f'{anchor_id}__readiness_mirror'
    mirror_path = MIRROR_DIR / f'{mirror_id}.png'
    anchor_example = outcome_example_by_scene[anchor_id]
    if not mirror_path.is_file():
        with Image.open(anchor_example.image_path) as image:
            ImageOps.mirror(image.convert('RGB')).save(mirror_path)
    outcome_example_by_scene[mirror_id] = dataclass_replace(
        anchor_example, scene_id=mirror_id, image_path=mirror_path.resolve(),
        outcome=mirror_features['outcome'],
        net_torque=int(mirror_features['net_torque']),
    )
    rows = []
    for orientation, scene_id, features in (
        ('anchor', anchor_id, base_features),
        ('mirror', mirror_id, mirror_features),
    ):
        exact = str(features['outcome'])
        wrong = {'left': 'right', 'right': 'left'}[exact]
        row = {
            'pair_id': pair_id, 'scene_id': scene_id,
            'source_dataset_scene_id': anchor_id,
            'counterpart_dataset_scene_id': counterpart_id,
            'orientation': orientation, 'tier': tier,
            'dataset_splits': sorted(splits),
            'pair_geometry': [list(item) for item in geometry],
            'exact_semantic': exact, 'shortcut_semantic': wrong,
            **features,
        }
        rows.append(row)
        candidate_manifest.append(row)
    pair_rows.append({
        'pair_id': pair_id, 'tier': tier,
        'source_dataset_scene_id': anchor_id,
        'counterpart_dataset_scene_id': counterpart_id,
        'left_scene_id': next(row['scene_id'] for row in rows
                              if row['exact_semantic'] == 'left'),
        'right_scene_id': next(row['scene_id'] for row in rows
                               if row['exact_semantic'] == 'right'),
    })

assert candidate_manifest and len(candidate_manifest) == 2 * len(pair_rows)
assert all(
    Counter(row['exact_semantic'] for row in candidate_manifest
            if row['pair_id'] == pair['pair_id']) == {'left': 1, 'right': 1}
    for pair in pair_rows
)
write_jsonl_atomic(CANDIDATE_PATH, candidate_manifest)
print('Strict / exploratory pairs:', Counter(row['tier'] for row in pair_rows))
print('Parent support scene IDs audited / candidate pairs excluded:',
      len(parent_scene_ids), excluded_parent_pair_count)
print('Direction roles audited:', direction_role_count)
print('Direction role stack-count distribution:',
      dict(sorted(direction_stack_count_distribution.items())))
print('Unique full-stack direction geometries audited:',
      len(direction_geometry_keys))
print('Candidate pairs excluded by direction geometry:',
      excluded_direction_pair_count)
print('Candidate orientations:', len(candidate_manifest))
```

## 2. Two-mapping Qwen-27B endpoint scoring


```python
%%cell_progress 7/14 Load pinned Qwen-27B
cell_stage('Loading weights for behavior-only 2x2 screening')
import gc
if 'adapter' in globals():
    del adapter
    gc.collect()
    torch.cuda.empty_cache()

adapter = HuggingFaceMultimodalAdapter(
    MODEL_ID, revision=MODEL_REVISION,
    dtype=MODEL_DTYPE, device_map=DEVICE_MAP,
)
display(adapter.metadata)
assert int(adapter.metadata['parameter_count']) > 20_000_000_000
print('Behavior-only model ready. No decoder hook is registered.')
```

```python
%%cell_progress 8/14 Define counterbalanced physical-side prompts
cell_stage('Separating physical sides from A/B/C response codes')
import math

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {'left': 'left side', 'balanced': 'equal', 'right': 'right side'}

def mapping_by_label(semantics):
    assert set(semantics) == {'left', 'balanced', 'right'}
    return tuple({
        'position': index, 'label': LABELS[index], 'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def target_mappings(row):
    exact, wrong = row['exact_semantic'], row['shortcut_semantic']
    assert {exact, wrong} == {'left', 'right'}
    balanced_label = LABELS[int(stable_key(
        ('balanced-label', row['scene_id'])
    )[:8], 16) % 3]
    side_labels = [label for label in LABELS if label != balanced_label]
    result = {}
    for variant, exact_label, wrong_label in (
        ('forward', side_labels[0], side_labels[1]),
        ('reverse', side_labels[1], side_labels[0]),
    ):
        by_label = {
            exact_label: exact, wrong_label: wrong, balanced_label: 'balanced'
        }
        result[variant] = mapping_by_label(tuple(by_label[label] for label in LABELS))
    return result

def semantics_by_label(mapping):
    return tuple(next(
        item['semantic'] for item in mapping if item['label'] == label
    ) for label in LABELS)

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def custom_prompt(mapping):
    options = '\n'.join(
        f"{item['label']}. {SEMANTIC_TEXT[item['semantic']]}" for item in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT, 'Question: ' + QUESTION_TEXT, 'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option letter '
        'and nothing else. Valid replies are: A, B, C.',
    ])

def build_target_examples(row):
    mappings = target_mappings(row)
    base = outcome_example_by_scene[row['scene_id']]
    examples = {}
    for variant, mapping in mappings.items():
        semantics = semantics_by_label(mapping)
        examples[variant] = dataclass_replace(
            base,
            example_id=f"side_readiness:{row['scene_id']}:{variant}",
            question_id=f"side_readiness_{row['scene_id']}_{variant}",
            variant=f'side_readiness_{variant}', prompt=custom_prompt(mapping),
            labels=LABELS, option_text=tuple(SEMANTIC_TEXT[x] for x in semantics),
            option_semantics=semantics,
            gold_label=LABELS[semantics.index(row['exact_semantic'])],
            gold_semantic=row['exact_semantic'],
        )
    return examples, mappings

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples], add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

print('Two counterbalanced answer mappings are ready.')
```

```python
%%cell_progress 9/14 Define behavior-only scoring helpers
cell_stage('Defining candidate-restricted A/B/C scoring')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        values = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(values)
        weights = [math.exp(value - maximum) for value in values]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(values))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': values,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_score_prepared(prepared):
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    return score_records(prepared, output.logits)

def semantic_margin(record, positive_semantic, negative_semantic):
    semantics = list(record['candidate_semantics'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(
        values[semantics.index(positive_semantic)]
        - values[semantics.index(negative_semantic)]
    )

print('Behavior-only scorer ready; no hidden state is captured.')
```

```python
%%cell_progress 10/14 Run image, mapping, and scoring smoke tests
cell_stage('Checking one exact mirror pair under both answer mappings')
smoke_pair_id = candidate_manifest[0]['pair_id']
smoke_rows = [row for row in candidate_manifest if row['pair_id'] == smoke_pair_id]
assert len(smoke_rows) == 2
smoke_outputs = []
for row in smoke_rows:
    examples, mappings = build_target_examples(row)
    assert set(examples) == set(MAPPING_VARIANTS)
    assert label_for(mappings['forward'], row['exact_semantic']) != label_for(
        mappings['reverse'], row['exact_semantic']
    )
    prepared = prepare_examples([examples[key] for key in MAPPING_VARIANTS])
    scores = forward_score_prepared(prepared)
    for variant, score in zip(MAPPING_VARIANTS, scores):
        smoke_outputs.append({
            'physical_side': row['exact_semantic'],
            'mapping': variant,
            'predicted_semantic': score['predicted_semantic'],
            'exact_minus_opposite': semantic_margin(
                score, row['exact_semantic'], row['shortcut_semantic']
            ),
        })
assert {row['physical_side'] for row in smoke_outputs} == {'left', 'right'}
display(pd.DataFrame(smoke_outputs))
print('Smoke pair:', smoke_pair_id)
```

```python
%%cell_progress 11/14 Screen every orientation under two answer mappings
cell_stage('Running resumable behavior-only screening')

screen_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'candidate_sha256': sha256_file(CANDIDATE_PATH),
    'model_id': MODEL_ID, 'revision': MODEL_REVISION,
    'mapping_variants': list(MAPPING_VARIANTS),
}
SCREEN_IDENTITY_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(screen_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(candidate_manifest, desc='Two-mapping endpoint screen', unit='scene')
for row in progress:
    shard_name = stable_key(('screen', row['scene_id']))[:24] + '.json'
    shard_path = SCREEN_SHARD_DIR / shard_name
    if shard_path.is_file():
        existing = json.loads(shard_path.read_text(encoding='utf-8'))
        assert existing['screen_identity_sha256'] == SCREEN_IDENTITY_SHA256
        assert existing['scene_id'] == row['scene_id']
        assert len(existing['records']) == len(MAPPING_VARIANTS)
        continue
    examples, mappings = build_target_examples(row)
    prepared = prepare_examples([examples[key] for key in MAPPING_VARIANTS])
    scores = forward_score_prepared(prepared)
    records = []
    for variant, score in zip(MAPPING_VARIANTS, scores):
        score.update({
            'pair_id': row['pair_id'], 'scene_id': row['scene_id'],
            'tier': row['tier'], 'orientation': row['orientation'],
            'mapping_variant': variant,
            'exact_semantic': row['exact_semantic'],
            'opposite_semantic': row['shortcut_semantic'],
            'exact_minus_opposite': semantic_margin(
                score, row['exact_semantic'], row['shortcut_semantic']
            ),
            'is_exact': bool(score['predicted_semantic'] == row['exact_semantic']),
            'is_opposite': bool(
                score['predicted_semantic'] == row['shortcut_semantic']
            ),
        })
        records.append(score)
    write_json_atomic(shard_path, {
        'screen_identity_sha256': SCREEN_IDENTITY_SHA256,
        'scene_id': row['scene_id'], 'records': records,
    })
    progress.set_postfix_str(
        f"{row['tier']} | {row['exact_semantic']} | "
        + '/'.join(record['predicted_semantic'] for record in records)
    )
progress.close()

screen_records = []
for row in candidate_manifest:
    shard_path = SCREEN_SHARD_DIR / (
        stable_key(('screen', row['scene_id']))[:24] + '.json'
    )
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    assert payload['screen_identity_sha256'] == SCREEN_IDENTITY_SHA256
    screen_records.extend(payload['records'])
assert len(screen_records) == len(candidate_manifest) * len(MAPPING_VARIANTS)
write_jsonl_atomic(SCREEN_RECORDS_PATH, screen_records)
print('Saved mapping-level records:', len(screen_records))
```

## 3. Deconfounded 2×2 readiness decision


```python
%%cell_progress 12/14 Classify stable outcomes and reciprocal pair patterns
cell_stage('Separating physical direction from baseline correctness')

candidate_by_scene = {row['scene_id']: row for row in candidate_manifest}
records_by_scene = defaultdict(list)
for record in screen_records:
    records_by_scene[record['scene_id']].append(record)

scene_status = {}
for scene_id, records in records_by_scene.items():
    assert {row['mapping_variant'] for row in records} == set(MAPPING_VARIANTS)
    candidate = candidate_by_scene[scene_id]
    predictions = tuple(row['predicted_semantic'] for row in sorted(
        records, key=lambda row: MAPPING_VARIANTS.index(row['mapping_variant'])
    ))
    if all(value == candidate['exact_semantic'] for value in predictions):
        status = 'correct'
    elif all(value == candidate['shortcut_semantic'] for value in predictions):
        status = 'error'
    else:
        status = 'unstable_or_balanced'
    scene_status[scene_id] = {
        'status': status, 'predictions': list(predictions),
        'minimum_absolute_margin': min(
            abs(float(row['exact_minus_opposite'])) for row in records
        ),
        'mean_exact_minus_opposite': float(np.mean([
            row['exact_minus_opposite'] for row in records
        ])),
    }

pair_table_rows = []
for pair_id in sorted({row['pair_id'] for row in candidate_manifest}):
    rows = [row for row in candidate_manifest if row['pair_id'] == pair_id]
    left = next(row for row in rows if row['exact_semantic'] == 'left')
    right = next(row for row in rows if row['exact_semantic'] == 'right')
    left_state = scene_status[left['scene_id']]
    right_state = scene_status[right['scene_id']]
    pattern = f"left_{left_state['status']}__right_{right_state['status']}"
    pair_table_rows.append({
        'pair_id': pair_id, 'tier': left['tier'],
        'diagnostic_cell': left['diagnostic_cell'],
        'margin_bin': int(left['margin_bin']),
        'left_scene_id': left['scene_id'], 'right_scene_id': right['scene_id'],
        'left_status': left_state['status'],
        'right_status': right_state['status'],
        'pattern': pattern,
        'left_minimum_absolute_margin': left_state['minimum_absolute_margin'],
        'right_minimum_absolute_margin': right_state['minimum_absolute_margin'],
        'pair_minimum_absolute_margin': min(
            left_state['minimum_absolute_margin'],
            right_state['minimum_absolute_margin'],
        ),
    })

pair_frame = pd.DataFrame(pair_table_rows)
pair_frame.to_csv(PAIR_TABLE_PATH, index=False)
print('Strict patterns:')
display(pair_frame[pair_frame['tier'] == 'strict']['pattern'].value_counts().to_frame())
print('All eligible patterns (strict + exploratory):')
display(pair_frame['pattern'].value_counts().to_frame())
```

```python
%%cell_progress 13/14 Decide readiness and freeze a balanced proposed cohort
cell_stage('Applying precommitted minimum and target gates')

required_patterns = (
    'left_correct__right_error',
    'left_error__right_correct',
)
strict_frame = pair_frame[pair_frame['tier'] == 'strict'].copy()
strict_counts = {
    pattern: int((strict_frame['pattern'] == pattern).sum())
    for pattern in required_patterns
}
all_counts = {
    pattern: int((pair_frame['pattern'] == pattern).sum())
    for pattern in required_patterns
}

def matched_reciprocal_units(frame):
    matched = []
    left_pattern, right_pattern = required_patterns
    strata = sorted(set(
        zip(frame['diagnostic_cell'], frame['margin_bin'])
    ))
    for diagnostic_cell, margin_bin in strata:
        subset = frame[
            (frame['diagnostic_cell'] == diagnostic_cell)
            & (frame['margin_bin'] == margin_bin)
        ]
        choices = {}
        for pattern in required_patterns:
            current = subset[subset['pattern'] == pattern].copy()
            current['matching_key'] = current['pair_id'].map(
                lambda value, pattern=pattern: stable_key(
                    ('matched-unit', diagnostic_cell, margin_bin, pattern, value)
                )
            )
            choices[pattern] = current.sort_values('matching_key').to_dict('records')
        for index in range(min(len(choices[left_pattern]), len(choices[right_pattern]))):
            matched.append({
                'diagnostic_cell': diagnostic_cell, 'margin_bin': int(margin_bin),
                left_pattern: choices[left_pattern][index],
                right_pattern: choices[right_pattern][index],
                'unit_key': stable_key((
                    'unit-order', diagnostic_cell, margin_bin,
                    choices[left_pattern][index]['pair_id'],
                    choices[right_pattern][index]['pair_id'],
                )),
            })
    return sorted(matched, key=lambda row: row['unit_key'])

strict_matched_units = matched_reciprocal_units(strict_frame)
all_matched_units = matched_reciprocal_units(pair_frame)
strict_matched_available = len(strict_matched_units)
all_matched_available = len(all_matched_units)
strict_ready = (
    strict_matched_available >= MIN_PAIRS_PER_RECIPROCAL_PATTERN
)
strict_target_met = (
    strict_matched_available >= TARGET_PAIRS_PER_RECIPROCAL_PATTERN
)
exploratory_ready = (
    all_matched_available >= MIN_PAIRS_PER_RECIPROCAL_PATTERN
)

if strict_target_met:
    PRIMARY_DECISION = 'strict_2x2_support_reaches_preferred_target'
elif strict_ready:
    PRIMARY_DECISION = 'strict_2x2_support_is_ready_at_minimum'
elif exploratory_ready:
    PRIMARY_DECISION = 'strict_support_insufficient_exploratory_2x2_available'
else:
    PRIMARY_DECISION = 'side_by_correctness_2x2_support_insufficient'

if strict_ready:
    selection_units = strict_matched_units
    selection_tier = 'strict'
elif exploratory_ready:
    selection_units = all_matched_units
    selection_tier = 'exploratory_not_primary_confirmation'
else:
    selection_units = []
    selection_tier = 'none'

selected_pair_rows = []
if selection_units:
    selected_units = selection_units[:TARGET_PAIRS_PER_RECIPROCAL_PATTERN]
    for unit in selected_units:
        for pattern in required_patterns:
            selected_pair_rows.append(unit[pattern])

selected_pair_ids = {row['pair_id'] for row in selected_pair_rows}
selected_manifest = []
for row in candidate_manifest:
    if row['pair_id'] not in selected_pair_ids:
        continue
    state = scene_status[row['scene_id']]
    selected_manifest.append({
        **row,
        'baseline_status': state['status'],
        'factorial_cell': f"{row['exact_semantic']}_{state['status']}",
        'selection_tier': selection_tier,
        'mapping_predictions': state['predictions'],
        'minimum_absolute_margin': state['minimum_absolute_margin'],
    })
write_jsonl_atomic(SELECTED_MANIFEST_PATH, selected_manifest)

cell_counts = Counter(row['factorial_cell'] for row in selected_manifest)
if selected_manifest:
    assert len(cell_counts) == 4 and len(set(cell_counts.values())) == 1, cell_counts
decision = {
    'primary_decision': PRIMARY_DECISION,
    'strict_counts': strict_counts, 'all_eligible_counts': all_counts,
    'strict_matched_pairs_per_pattern': int(strict_matched_available),
    'all_eligible_matched_pairs_per_pattern': int(all_matched_available),
    'strict_ready_at_minimum': bool(strict_ready),
    'strict_preferred_target_met': bool(strict_target_met),
    'exploratory_ready_at_minimum': bool(exploratory_ready),
    'selected_tier': selection_tier,
    'selected_pairs': len(selected_pair_ids),
    'selected_factorial_cells': dict(cell_counts),
    'claim_boundary': (
        'Only a strict cohort can support the primary deconfounded causal test. '
        'Exploratory pairs are a feasibility diagnostic and cannot upgrade the '
        'strict decision.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'readiness_decision.json', decision)
display(decision)
```
