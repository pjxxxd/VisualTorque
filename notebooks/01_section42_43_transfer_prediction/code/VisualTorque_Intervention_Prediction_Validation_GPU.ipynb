```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')

IMPLEMENTATION_SHA256 = '160372af558929c42134517d3ffc09527278d018135dd39a46f32eb6987dde7d'
```

```python
%%cell_progress 4/14 Load protocol and prediction utilities
"""Pure CPU utilities embedded in the follow-up notebooks; no model execution."""
import hashlib
import json
from pathlib import Path
from collections import defaultdict, Counter
from itertools import combinations, product
import numpy as np
import pandas as pd


def plain(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(type(value).__name__)


def digest(value):
    return hashlib.sha256(json.dumps(value, sort_keys=True, default=plain, allow_nan=False).encode()).hexdigest()


from visualtorque_checks import file_hash


def read_json(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))


def read_jsonl(path):
    return [json.loads(s) for s in Path(path).read_text(encoding='utf-8').splitlines() if s.strip()]


def save_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + '.tmp')
    temp.write_text(json.dumps(value, ensure_ascii=False, indent=2, default=plain, allow_nan=False) + '\n', encoding='utf-8')
    temp.replace(path)


def freeze(path, identity):
    identity = json.loads(json.dumps(identity, default=plain, sort_keys=True, allow_nan=False))
    if Path(path).is_file():
        assert read_json(path) == identity, f'Configuration changed; preserve the old run and use a new RUN_ID: {path}'
    else:
        save_json(path, identity)
    return digest(identity)


def cached_json(directory, key, identity, compute):
    path = Path(directory) / (digest(str(key))[:24] + '.json')
    expected = digest(identity)
    if path.is_file():
        old = read_json(path)
        assert old['key'] == str(key) and old['identity'] == expected, f'Incompatible cache: {path}'
        return old['result']
    value = compute()
    save_json(path, dict(key=str(key), identity=expected, result=value))
    return value


def bootstrap(values, key='', replicates=10000):
    x = np.asarray(values, dtype=float)
    assert x.ndim == 1 and len(x) and np.isfinite(x).all()
    rng = np.random.default_rng(int(digest(str(key))[:16], 16))
    draws = np.concatenate([rng.choice(x, (min(250, replicates-i), len(x)), replace=True).mean(1)
                            for i in range(0, replicates, 250)])
    return dict(n=len(x), mean=float(x.mean()), lower=float(np.quantile(draws, .025)),
                upper=float(np.quantile(draws, .975)),
                p_value=float((1 + (draws <= 0).sum()) / (replicates + 1)))


def bh(values):
    p = np.asarray(values, dtype=float)
    assert np.isfinite(p).all() and ((p >= 0) & (p <= 1)).all()
    order = np.argsort(p)
    q = np.empty_like(p)
    previous = 1.
    for rank in range(len(p)-1, -1, -1):
        i = order[rank]
        previous = min(previous, p[i]*len(p)/(rank+1))
        q[i] = previous
    return q


def paired(left, right, id_col, value):
    assert left[id_col].is_unique and right[id_col].is_unique
    assert set(left[id_col]) == set(right[id_col]), 'Different paired support'
    out = left[[id_col, value]].merge(right[[id_col, value]], on=id_col, validate='one_to_one', suffixes=('_l', '_r'))
    return out[value+'_l'].to_numpy(float) - out[value+'_r'].to_numpy(float)


def semantic_scores(record):
    if 'semantic_logprobs' in record:
        return {str(k): float(v) for k, v in record['semantic_logprobs'].items()}
    values = record['candidate_logprobs']
    if isinstance(values, dict):
        return {str(record['label_to_semantic'][k]): float(v) for k, v in values.items()}
    return dict(zip(map(str, record['candidate_semantics']), map(float, values)))


def correct_margin(record):
    values = semantic_scores(record)
    gold = str(record['gold_semantic'])
    assert gold in values and len(values) >= 2
    return values[gold] - max(v for k, v in values.items() if k != gold)


def random_unit(shape, key):
    r = np.random.default_rng(int(digest(key)[:16], 16)).normal(size=shape)
    return r / np.linalg.norm(r)


def coordinate_clamp(pre, untreated, axis, cast):
    h, base, u = [np.asarray(x, dtype=np.float64) for x in (pre, untreated, axis)]
    u = u / np.linalg.norm(u)
    a = float(np.sum((h-base)*u))
    return np.asarray(cast(h-a*u), dtype=np.float64), a


def native_norm_edit(pre, direction, norm, cast, tolerance=.02, steps=48):
    h = np.asarray(pre, dtype=np.float64)
    r = np.asarray(direction, dtype=np.float64)
    assert np.isfinite(h).all() and np.isfinite(r).all() and norm >= 0 and np.linalg.norm(r) > 0
    r /= np.linalg.norm(r)
    if norm == 0:
        return h.copy(), dict(actual_norm=0., target_norm=0., relative_error=0., matched=True)
    lo, hi = 0., max(float(norm), 1e-12)
    for _ in range(32):
        if np.linalg.norm(np.asarray(cast(h+hi*r))-h) >= norm:
            break
        hi *= 2
    best, error = h.copy(), 1.
    for _ in range(steps):
        mid = (lo+hi)/2
        candidate = np.asarray(cast(h+mid*r), dtype=np.float64)
        actual = float(np.linalg.norm(candidate-h))
        current = abs(actual-norm)/norm
        if current < error:
            best, error = candidate.copy(), current
        if error <= tolerance:
            break
        if actual < norm:
            lo = mid
        else:
            hi = mid
    return best, dict(actual_norm=float(np.linalg.norm(best-h)), target_norm=float(norm),
                      relative_error=float(error), matched=bool(error <= tolerance))


def equal_removal(pre, untreated, axis, random_direction, cast, tolerance=.02):
    target, coefficient = coordinate_clamp(pre, untreated, axis, cast)
    target_norm = float(np.linalg.norm(target-np.asarray(pre)))
    sign = 1. if coefficient >= 0 else -1.
    post, audit = native_norm_edit(pre, -sign*np.asarray(random_direction), target_norm, cast, tolerance)
    audit.update(reference_coefficient=coefficient, reference_actual_norm=target_norm)
    return post, audit


def geometry_signature(stacks):
    a = tuple(sorted((str(s['side']), int(s['distance']), int(s['units'])) for s in stacks))
    b = tuple(sorted(('right' if side == 'left' else 'left', d, u) for side, d, u in a))
    return min(a, b)  # mirrors belong to the same physical family


def geometries_in(value):
    found = set()
    if isinstance(value, dict):
        stacks = value.get('stacks')
        if isinstance(stacks, list) and stacks and all(isinstance(s, dict) and {'side','distance','units'} <= set(s) for s in stacks):
            found.add(geometry_signature(stacks))
        # Some prior matched cohorts store geometry instead of stack records.
        geom = value.get('pair_geometry')
        if isinstance(geom, (list, tuple)) and len(geom) == 4 and all(isinstance(x, (int,float)) for x in geom):
            u1,d1,u2,d2 = map(int, geom)
            found.add(geometry_signature([dict(side='left',distance=d1,units=u1),dict(side='right',distance=d2,units=u2)]))
        if isinstance(geom, (list, tuple)) and len(geom)==2 and all(isinstance(x,(list,tuple)) and len(x)==2 for x in geom):
            (u1,d1),(u2,d2)=geom
            found.add(geometry_signature([dict(side='left',distance=int(d1),units=int(u1)),dict(side='right',distance=int(d2),units=int(u2))]))
        for child in value.values():
            found.update(geometries_in(child))
    elif isinstance(value, list):
        for child in value:
            found.update(geometries_in(child))
    return found


def physics(stacks):
    units = {s:sum(int(x['units']) for x in stacks if x['side']==s) for s in ('left','right')}
    dist = {s:sum(int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    torque = {s:sum(int(x['units'])*int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    out = lambda x:'left' if x>0 else 'right' if x<0 else 'balanced'
    net=torque['left']-torque['right']; w=units['left']-units['right']; d=dist['left']-dist['right']
    return dict(net_torque=net,outcome=out(net),weight_score=w,distance_score=d,
                normalized_margin=abs(net)/max(1,sum(torque.values())),
                left_units=units['left'],right_units=units['right'],left_torque=torque['left'],right_torque=torque['right'],
                strategy_predictions=dict(exact_torque=out(net),weight_only=out(w),distance_only=out(d)))


def exclusion_snapshot(project, followup_root, dataset_dir=None):
    lock=Path(followup_root)/'prior_geometry_snapshot.json'
    if lock.is_file():
        old=read_json(lock)
        for path, sha in old['files'].items():
            assert file_hash(Path(path))==sha, f'Prior input changed: {path}'
        return {tuple(tuple(x) for x in sig) for sig in old['signatures']}, old
    paths={Path(dataset_dir or project/'data/dataset_v1')/'scenes.jsonl'}
    for pattern in ('*manifest*.jsonl','*support*.jsonl','*geometries*.jsonl'):
        paths.update(p for p in (project/'results').rglob(pattern) if Path(followup_root) not in p.parents)
    excluded=set(); hashes={}
    for path in sorted(paths):
        assert path.is_file(), f'Missing required geometry input: {path}'
        excluded.update(geometries_in(read_jsonl(path)))
        hashes[str(path.resolve())]=file_hash(path)
    assert excluded, 'No prior physical geometry was indexed'
    old=dict(files=hashes,signatures=sorted(excluded),mirror_canonical=True)
    save_json(lock,old)
    return excluded,old


def inverse_families(n, seed, excluded):
    assert n%4==0
    pool=defaultdict(list)
    for dl in combinations(range(1,7),2):
        for dr in combinations(range(1,7),2):
            for units in product(range(1,6), repeat=4):
                stacks=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),dl+dr,units)]
                f=physics(stacks)
                pool[dl+dr].append((stacks,f))
    candidates=[]
    for distances,rows in pool.items():
        by_net=defaultdict(list)
        for stacks,f in rows:
            if f['outcome']==f['strategy_predictions']['weight_only'] and f['outcome']!='balanced':
                by_net[f['net_torque']].append((stacks,f))
        for stacks,f in rows:
            margin=f['normalized_margin']; difficulty='hard' if .08<=margin<.22 else 'moderate' if .22<=margin<.45 else None
            if (difficulty and f['outcome']!='balanced' and f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome'])
                    and f['strategy_predictions']['distance_only']==f['outcome'] and abs(f['distance_score'])>=2):
                sources=by_net.get(f['net_torque'],[])
                if sources:candidates.append((digest([seed,stacks]),difficulty,stacks,f,sources))
    candidates.sort(key=lambda r:r[0]); used=set(excluded); counts=Counter(); result=[]
    for _,difficulty,stacks,f,sources in candidates:
        stratum=(difficulty,f['outcome'])
        if counts[stratum]>=n//4 or geometry_signature(stacks) in used:continue
        valid=[(s,g) for s,g in sources if geometry_signature(s) not in used|{geometry_signature(stacks)}]
        if not valid:continue
        valid.sort(key=lambda x:(sum(abs(a['units']-b['units']) for a,b in zip(stacks,x[0])),digest([seed,x[0]])))
        src,sf=valid[0];fid='fixed_'+digest([seed,stacks,src])[:20]
        roles={name:dict(scene_id=fid+'__'+name,image_filename=fid+'__'+name+'.png',stacks=s,features=g)
               for name,s,g in [('base',stacks,f),('aligned_source',src,sf)]}
        result.append(dict(family_id=fid,remapping_split='new_fixed_confirmation',roles=roles,margin_bin=difficulty))
        used.update([geometry_signature(stacks),geometry_signature(src)]);counts[stratum]+=1
        if len(result)==n:break
    assert len(result)==n and set(counts.values())=={n//4}, (len(result),counts)
    return sorted(result,key=lambda r:r['family_id'])


def repair_families(n, seed, excluded):
    assert n%4==0
    candidates=defaultdict(list)
    for ul,dl,ur,dr in product(range(1,9),range(1,7),range(1,9),range(1,7)):
        s=[dict(side='left',units=ul,distance=dl),dict(side='right',units=ur,distance=dr)]
        f=physics(s)
        if f['net_torque']==0:continue
        if ul==ur and dl!=dr:kind='equal_weight_distance'
        elif f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome']):kind='weight_distance_conflict'
        elif f['strategy_predictions']['weight_only']==f['outcome'] and f['strategy_predictions']['distance_only']==f['outcome']:kind='aligned_cues'
        else:continue
        candidates[kind].append(s)
    for units in product(range(1,6),repeat=4):
        for distances in [((1,3),(2,5)),((2,5),(1,4)),((1,6),(3,5)),((2,4),(3,6))]:
            s=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),distances[0]+distances[1],units)]
            if physics(s)['net_torque']!=0:candidates['multi_stack_sum'].append(s)
    used=set(excluded); result=[]
    for kind in ('equal_weight_distance','weight_distance_conflict','multi_stack_sum','aligned_cues'):
        rows=sorted(candidates[kind],key=lambda s:digest([seed,kind,s])); count=0
        for stacks in rows:
            sig=geometry_signature(stacks)
            if sig in used:continue
            fid='repair_'+digest([seed,kind,sig])[:20];scenes=[]
            for mirror in (False,True):
                current=[dict(s,side=('right' if s['side']=='left' else 'left')) if mirror else dict(s) for s in stacks]
                for style in ('standard','alternate'):
                    sid=f'{fid}__m{int(mirror)}__{style}'
                    scenes.append(dict(scene_id=sid,image_filename=sid+'.png',stacks=current,features=physics(current),style=style,mirror=mirror))
            result.append(dict(family_id=fid,stratum=kind,scenes=scenes));used.add(sig);count+=1
            if count==n//4:break
        assert count==n//4,(kind,count,'Insufficient unused geometries; do not silently change task ranges')
    return sorted(result,key=lambda r:r['family_id'])

"""CPU-only protocol, support, prediction and analysis for prospective interventions.

Embedded with selected, unchanged followup_core utilities by the notebook builder.
No function in this module executes a language model.
"""
from datetime import datetime, timezone
import math

SEMANTICS = ('left', 'balanced', 'right')
LABELS = ('A', 'B', 'C')
SCHEMA = 'prospective-intervention-effects-v1'


def utc_now():
    return datetime.now(timezone.utc).isoformat()


def center(x):
    x = np.asarray(x, dtype=float)
    return x - x.mean(axis=-1, keepdims=True)


def bf16(x):
    """Round finite FP32 values to BF16, ties-to-even; return FP32 numpy values."""
    a = np.asarray(x, dtype=np.float32).copy()
    assert np.isfinite(a).all()
    u = a.view(np.uint32)
    u = (u + np.uint32(0x7fff) + ((u >> 16) & 1)) & np.uint32(0xffff0000)
    return u.view(np.float32)


def state_hash(x):
    return hashlib.sha256(np.asarray(x, dtype='<f4').tobytes()).hexdigest()


def inventory_geometries(project, output_root, extra_roots=()):
    """Index available historical geometry, including JSON supports, once per run.

    The scope is the files present at initial freezing, not all experiments ever run.
    Unreadable candidate files stop the inventory instead of silently disappearing.
    """
    lock = output_root / 'historical_geometry_snapshot.json'
    if lock.exists():
        obj = read_json(lock)
        for name, expected in obj['files'].items():
            assert file_hash(name) == expected, 'A frozen historical input changed: ' + name
        return {tuple(tuple(t) for t in s) for s in obj['signatures']}, obj
    paths = set()
    for root in (project / 'data', project / 'results', *extra_roots):
        root = Path(root)
        if not root.exists():
            continue
        for p in root.rglob('*'):
            if not p.is_file() or output_root.resolve() in p.resolve().parents:
                continue
            if p.suffix.lower() not in ('.json', '.jsonl'):
                continue
            if any(k in p.name.lower() for k in ('manifest', 'support', 'famil', 'scene', 'geometr')):
                paths.add(p)
    signatures, hashes = set(), {}
    for p in sorted(paths):
        obj = read_jsonl(p) if p.suffix == '.jsonl' else read_json(p)
        signatures.update(geometries_in(obj))
        hashes[str(p.resolve())] = file_hash(p)
    assert signatures, 'No historical scenes found. Restore the project data/results before generating new support.'
    obj = dict(files=hashes, signatures=sorted(signatures), created_at_utc=utc_now(),
               scope='available geometry-bearing JSON/JSONL files only; unimported archives are not indexed')
    save_json(lock, obj)
    return signatures, obj


def physical_families(n, seed, excluded):
    """Cross donor distance x recipient distance, with exact one-block changes.

    Baseline recipient net torque is equal across its two distances. Background
    stacks compensate for changing distance; this is a factorial family, not a
    claim that changing distance leaves every other visible primitive unchanged.
    """
    assert n % 4 == 0
    rng = np.random.default_rng(seed)
    used = set(excluded)
    families = []
    pools = {}

    def sample_pair(d, net, dw, mirror, local, target_units):
        key=(d,net,target_units)
        if key not in pools:
            choices=[]
            for u2 in range(1,8):
                for r1,r2 in combinations(range(1,7),2):
                    for v1 in range(1,8):
                        numerator=target_units*d + u2*6-v1*r1-net
                        if numerator%r2==0 and 1<=numerator//r2<=7:
                            choices.append((u2,r1,r2,v1,numerator//r2))
            pools[key]=choices
        for index in rng.permutation(len(pools[key])):
            u2,r1,r2,v1,v2=pools[key][int(index)]
            # The changed stack is always the innermost stack on its side.
            a = [dict(side=s, distance=int(x), units=int(w)) for s,x,w in
                 [('left',d,target_units),('left',6,u2),('right',r1,v1),('right',r2,v2)]]
            b = [dict(s) for s in a]
            b[0]['units'] += dw
            if mirror:
                for x in a+b:
                    x['side'] = 'right' if x['side'] == 'left' else 'left'
            sigs = {geometry_signature(a), geometry_signature(b)}
            if len(sigs) == 2 and not sigs & (used | local):
                return a, b, sigs
        raise RuntimeError(f'No unused matched geometry: d={d}, net={net}, dw={dw}, target_units={target_units}. '
                           'Do not relax support rules after looking at outputs.')

    for i in range(n):
        mirror = bool(i % 2)
        dw = 1 if (i // 2) % 2 == 0 else -1
        near = int(rng.choice([1, 2]))
        far = int(rng.choice([4, 5]))
        target_units = int(rng.integers(2,7))
        clearance = int(rng.integers(near+1, far))
        donor_sign = 1 if (i // 4) % 2 == 0 else -1
        donor_net = donor_sign * int(rng.integers(far+2, far+10))
        local, roles = set(), {}
        for owner in ('source', 'recipient'):
            for dname, d in (('near', near), ('far', far)):
                net = donor_net if owner == 'source' else -dw*clearance
                a, b, sigs = sample_pair(d, net, dw, mirror, local, target_units)
                local.update(sigs)
                for endpoint, stacks in (('before', a), ('after', b)):
                    roles[f'{owner}_{dname}_{endpoint}'] = dict(stacks=stacks, features=physics(stacks))
        fid = 'physical_' + digest([seed, i, roles])[:20]
        for name, role in roles.items():
            role.update(scene_id=fid+'__'+name, image_filename=fid+'__'+name+'.png')
        side = 'right' if mirror else 'left'
        families.append(dict(family_id=fid, module='physical', stratum=f'{side}_dw{dw}',
                             near=near, far=far, dw=dw, target_side=side, roles=roles))
        used.update(local)
    return families


def assign_splits(families, n_calibration, n_validation, seed):
    groups = defaultdict(list)
    for f in families:
        groups[f['stratum']].append(f)
    assert len(groups) == 4
    for key, rows in groups.items():
        rows.sort(key=lambda f: digest([seed, f['family_id']]))
        assert len(rows) == (n_calibration+n_validation)//4
        for i, f in enumerate(rows):
            f['split'] = 'calibration' if i < n_calibration//4 else 'validation'
    for phase, n in [('calibration', n_calibration), ('validation', n_validation)]:
        selected = sorted([f for f in families if f['split'] == phase], key=lambda f:f['family_id'])
        # A fixed quarter of families, balanced across the four design strata.
        for key in groups:
            subset = [f for f in selected if f['stratum'] == key]
            for j, f in enumerate(subset):
                f['random_control_family'] = j < len(subset)//4
    return families


def generate_support(n_calibration, n_validation, seed, excluded):
    assert n_calibration % 16 == n_validation % 16 == 0
    n = n_calibration+n_validation
    content = inverse_families(n, seed, excluded)
    for f in content:
        f.update(module='content', stratum=f["margin_bin"]+'_'+f['roles']['base']['features']['outcome'])
    physical = physical_families(n, seed+1, excluded | geometries_in(content))
    rows = assign_splits(content, n_calibration, n_validation, seed)
    rows += assign_splits(physical, n_calibration, n_validation, seed)
    signatures = [geometry_signature(r['stacks']) for f in rows for r in f['roles'].values()]
    assert len(signatures) == len(set(signatures))
    assert not set(signatures) & excluded
    return rows


def mapping_reference(fid):
    offset = int(digest(['mapping', fid])[:8], 16) % 3
    sem = list(SEMANTICS[offset:] + SEMANTICS[:offset])
    return [dict(position=i, label=LABELS[i], semantic=sem[i]) for i in range(3)]


def mapping_conflict(reference, semantic, reverse=False):
    pos = next(x['position'] for x in reference if x['semantic'] == semantic)
    label = reference[pos]['label']
    ps = [i for i in range(3) if i != pos]
    ss = [s for s in SEMANTICS if s != semantic]
    ls = [s for s in LABELS if s != label]
    if reverse:
        ss.reverse()
        ls.reverse()
    result = [None]*3
    result[pos] = dict(position=pos, label=ls[0], semantic=ss[0])
    result[ps[0]] = dict(position=ps[0], label=label, semantic=ss[1])
    result[ps[1]] = dict(position=ps[1], label=ls[1], semantic=semantic)
    assert len({next(x['label'] for x in result if x['semantic'] == semantic),label,result[pos]['label']}) == 3
    return result


def transport_vector(values, source_mapping, recipient_mapping, coordinate):
    values = np.asarray(values, dtype=float)  # Always A/B/C order.
    source_index = {x[coordinate]: LABELS.index(x['label']) for x in source_mapping}
    target = {x['label']: values[source_index[x[coordinate]]] for x in recipient_mapping}
    return center([target[label] for label in LABELS])


def predictor_features(row):
    """Explicit allowlist. No recipient gold, CF scores or post-intervention state."""
    s0 = np.asarray(row['s0'], float)
    source_delta = np.asarray(row['source_delta'], float)
    sm, rm = row['source_mapping'], row['recipient_mapping']
    features = {}
    for name, coord in [('semantic','semantic'),('letter','label'),('position','position')]:
        features['source_'+name] = transport_vector(source_delta, sm, rm, coord).tolist()
        # Source argmax endpoint changes; unchanging source decisions give zero.
        cat = np.eye(3)[row['source_after_argmax']] - np.eye(3)[row['source_before_argmax']]
        features['choice_'+name] = transport_vector(cat, sm, rm, coord).tolist()
        features['copy_'+name] = transport_vector(np.eye(3)[row['source_after_argmax']],sm,rm,coord).tolist()
    weights = np.exp(s0-s0.max()); probs = weights/weights.sum()
    gap = float(np.sort(s0)[-1]-np.sort(s0)[-2])
    conf = [*s0, gap, float(-np.sum(probs*np.log(probs+1e-12))),
            float(row['planned_norm']), float(row['relative_norm'])]
    features['confidence'] = conf
    if row['module'] == 'physical':
        axis = np.array([{'left':1., 'balanced':0., 'right':-1.}[
            next(x['semantic'] for x in rm if x['label'] == label)] for label in LABELS])
        sign = 1 if row['target_side'] == 'left' else -1
        signed_dw = sign*int(row['dw'])
        features['recipient_weight'] = (axis*signed_dw*row['recipient_distance']).tolist()
        features['source_torque'] = (axis*signed_dw*row['source_distance']).tolist()
        # Privileged structured-input baseline, explicitly labelled in the protocol.
        features['physical_difficulty'] = conf + list(map(float, row['baseline_physics']))
    return features


def model_names(module):
    names = ['null','confidence', 'source_semantic','source_letter','source_position',
             'choice_semantic','choice_letter','choice_position',
             'copy_semantic','copy_letter','copy_position']
    if module == 'physical':
        names += ['physical_difficulty','recipient_weight','source_torque',
                  'behavior_plus_weight','behavior_plus_source_torque','natural_response']
    return names


def ridge_inputs(row, name):
    f = predictor_features(row)
    if name in ('confidence', 'physical_difficulty'):
        return f[name]
    # Same answer/confidence baseline on both sides of the nested comparison.
    values = f['physical_difficulty'] + f['source_semantic'] + f['source_letter'] + f['source_position']
    if name == 'behavior_plus_weight':
        return values + f['recipient_weight']
    if name == 'behavior_plus_source_torque':
        return values + f['source_torque']
    if name == 'behavior':
        return values
    raise KeyError(name)


def all_model_names(module):
    names = model_names(module)
    return names + (['behavior'] if module == 'physical' else [])


def fit_predictor(rows, name):
    assert rows and all(r['split'] == 'calibration' for r in rows)
    y = np.array([center(r['actual_delta']) for r in rows])
    if name == 'null':
        return dict(kind='null')
    if name in ('confidence','physical_difficulty','behavior','behavior_plus_weight','behavior_plus_source_torque'):
        x = np.asarray([ridge_inputs(r,name) for r in rows], float)
        mean, scale = x.mean(0), x.std(0)
        scale[scale < 1e-8] = 1.
        z = np.column_stack([np.ones(len(x)), (x-mean)/scale])
        penalty = np.eye(z.shape[1]); penalty[0,0] = 0
        coef = np.linalg.solve(z.T@z/len(z) + penalty, z.T@y/len(z))
        return dict(kind='ridge', name=name, mean=mean, scale=scale, coef=coef, ridge_alpha=1.)
    x = np.asarray([r['natural_input_delta'] if name=='natural_response' else predictor_features(r)[name] for r in rows])
    beta = max(0., float(np.sum(x*y)/max(float(np.sum(x*x)), 1e-12)))
    return dict(kind='scalar', name=name, beta=beta)


def predict_one(row, fitted):
    if fitted['kind'] == 'null':
        return np.zeros(3)
    if fitted['kind'] == 'scalar':
        value = row['natural_input_delta'] if fitted['name']=='natural_response' else predictor_features(row)[fitted['name']]
        return center(np.asarray(value)*fitted['beta'])
    x = np.asarray(ridge_inputs(row, fitted['name']))
    z = np.r_[1., (x-np.asarray(fitted['mean']))/np.asarray(fitted['scale'])]
    return center(z@np.asarray(fitted['coef']))


def operation_key(row):
    return f"{row['module']}:L{row['layer']}"


def fit_all(rows):
    """Five family folds select a strong answer baseline; test scores never select."""
    output = {}
    for key in sorted({operation_key(r) for r in rows}):
        sub = [r for r in rows if operation_key(r) == key]
        fids = sorted({r['family_id'] for r in sub}, key=lambda s:digest(['fold',s]))
        folds = {fid:i%5 for i,fid in enumerate(fids)}
        names = all_model_names(sub[0]['module'])
        cv = {}
        for name in names:
            losses = defaultdict(list)
            for fold in range(5):
                train = [r for r in sub if folds[r['family_id']] != fold]
                valid = [r for r in sub if folds[r['family_id']] == fold]
                if not valid:
                    continue
                fitted = fit_predictor(train,name)
                for r in valid:
                    losses[r['family_id']].append(float(np.mean((predict_one(r,fitted)-r['actual_delta'])**2)))
            cv[name] = float(np.mean([np.mean(v) for v in losses.values()]))
        eligible = [n for n in names if n not in ('recipient_weight','source_torque','behavior_plus_weight',
                                                'behavior_plus_source_torque','natural_response')]
        best = min(eligible, key=lambda n:(cv[n],n))
        output[key] = dict(models={n:fit_predictor(sub,n) for n in names}, cv_family_mse=cv,
                           selected_baseline=best, calibration_families=fids, folds=folds,
                           baseline_gap_quartiles=np.quantile(
                               [np.sort(r['s0'])[-1]-np.sort(r['s0'])[-2] for r in sub],[.25,.5,.75]).tolist())
    return output


def seal_predictions(path, rows, fitted, identity):
    """Write once and verify on resume. Call before the first validation patch."""
    assert len({r['trial_id'] for r in rows}) == len(rows)
    records = []
    for r in rows:
        models = fitted[operation_key(r)]['models']
        records.append(dict(trial_id=r['trial_id'], family_id=r['family_id'],
                            input_sha256=digest({k:v for k,v in r.items() if k not in ('gold','cf_gold')}),
                            predicted_deltas={n:predict_one(r,m).tolist() for n,m in models.items()}))
    body = dict(identity=identity, records=records)
    if Path(path).exists():
        old = read_json(path)
        assert old['payload'] == body, 'Predictions or their inputs changed after sealing.'
    else:
        save_json(path, dict(created_at_utc=utc_now(), payload=body))
    return file_hash(path)


def paired_interval(values, key, replicates=10000):
    """Family bootstrap CI and two-sided paired sign-flip test."""
    x = np.asarray(values, float)
    assert len(x) and np.isfinite(x).all()
    rng = np.random.default_rng(int(digest(key)[:16],16))
    draws, null = [], []
    for start in range(0,replicates,250):
        n = min(250,replicates-start)
        draws.extend(rng.choice(x,(n,len(x)),replace=True).mean(1))
        null.extend((rng.choice([-1.,1.],(n,len(x)))*x).mean(1))
    lo, hi = np.quantile(draws,[.025,.975])
    p = (1+int(np.sum(np.abs(null) >= abs(x.mean())-1e-15)))/(replicates+1)
    return dict(n_families=len(x), mean=float(x.mean()), lower=float(lo), upper=float(hi), p_value=float(p))


def holm(values):
    p = np.asarray(values,float); order = np.argsort(p); q = np.empty(len(p)); previous = 0.
    for rank,i in enumerate(order):
        previous = max(previous,min(1.,float(p[i])*(len(p)-rank)))
        q[i] = previous
    return q


def transition(before, after, gold):
    if before != gold and after == gold: return 'repair'
    if before == gold and after != gold: return 'harm'
    if before != after: return 'wrong_to_other_wrong'
    return 'unchanged'


def evaluate_predictions(rows, fitted, sealed, outdir):
    """All target rows, then family means. No exclusions based on desired effects."""
    outdir = Path(outdir); outdir.mkdir(parents=True,exist_ok=True)
    prediction_map = {r['trial_id']:r for r in sealed['payload']['records']}
    assert set(prediction_map) == {r['trial_id'] for r in rows}
    records = []
    for r in rows:
        s0, actual = np.asarray(r['s0']), np.asarray(r['actual_delta'])
        before, after = int(np.argmax(s0)), int(np.argmax(s0+actual))
        gold = LABELS.index(r['gold'])
        gap = float(np.sort(s0)[-1]-np.sort(s0)[-2])
        gap_bin = int(np.searchsorted(fitted[operation_key(r)]['baseline_gap_quartiles'],gap,side='right'))
        for name, predicted in prediction_map[r['trial_id']]['predicted_deltas'].items():
            p = np.asarray(predicted); pred_after = int(np.argmax(s0+p))
            records.append(dict(trial_id=r['trial_id'], family_id=r['family_id'],
                operation=operation_key(r), predictor=name, mapping=r['mapping_name'],
                source_distance=r.get('source_distance'), recipient_distance=r.get('recipient_distance'),
                source_argmax_stable=r['source_before_argmax']==r['source_after_argmax'],
                baseline_gap=gap,baseline_gap_bin=gap_bin,
                mse=float(np.mean((p-actual)**2)),
                sign_agreement=float(np.mean(np.sign(p)==np.sign(actual))),
                cosine=(float(np.dot(p,actual)/(np.linalg.norm(p)*np.linalg.norm(actual)))
                        if np.linalg.norm(p)*np.linalg.norm(actual)>1e-10 else None),
                post_answer_match=int(pred_after==after), actual_change=int(after!=before),
                predicted_change=int(pred_after!=before),
                actual_transition=transition(before,after,gold),
                predicted_transition=transition(before,pred_after,gold)))
    df = pd.DataFrame(records); df.to_csv(outdir/'trial_prediction_metrics.csv',index=False)
    metrics = ['mse','post_answer_match','actual_change','predicted_change','sign_agreement','cosine']
    fam = df.groupby(['operation','predictor','family_id'])[metrics].mean().reset_index()
    fam.to_csv(outdir/'family_prediction_metrics.csv',index=False)
    summary = fam.groupby(['operation','predictor'])[metrics].mean().reset_index()
    summary.to_csv(outdir/'prediction_summary.csv',index=False)
    tests = []
    for op, sub in fam.groupby('operation'):
        pivot = sub.pivot(index='family_id',columns='predictor',values='mse')
        baseline = fitted[op]['selected_baseline']
        if op.startswith('content:'):
            comparisons = [('source_semantic','source_letter'),('source_letter','source_position'),
                           (baseline,'null'), (baseline,'confidence')]
        else:
            comparisons = [('recipient_weight','source_torque'), ('recipient_weight',baseline),
                           ('behavior_plus_weight','behavior'),('behavior_plus_weight','behavior_plus_source_torque'),
                           ('recipient_weight','natural_response')]
        for index,(a,b) in enumerate(comparisons):
            result = paired_interval((pivot[b]-pivot[a]).values, [op,index,a,b])
            tests.append(dict(operation=op, better_if_positive=a, comparator=b, **result))
    for t,q in zip(tests,holm([t['p_value'] for t in tests])):
        t['holm_p_within_model'] = float(q)
    pd.DataFrame(tests).to_csv(outdir/'primary_comparisons.csv',index=False)
    # Rare events get explicit denominators; undefined recall is blank, never 100%.
    transitions = []
    for (op,name),g in df.groupby(['operation','predictor']):
        for cls in ('repair','harm','wrong_to_other_wrong','unchanged'):
            true = g.actual_transition.eq(cls); pred = g.predicted_transition.eq(cls)
            tp = int((true&pred).sum()); nt = int(true.sum()); npred = int(pred.sum())
            transitions.append(dict(operation=op,predictor=name,event=cls,n_actual=nt,n_predicted=npred,
                n_true_positive=tp,recall=tp/nt if nt else None,
                precision=tp/npred if npred else None,
                f1=2*tp/(nt+npred) if nt+npred else None))
    event_frame=pd.DataFrame(transitions)
    event_frame.to_csv(outdir/'transition_events.csv',index=False)
    event_frame.groupby(['operation','predictor'])[['recall','f1']].mean().rename(
        columns={'recall':'macro_recall_over_observed_classes','f1':'macro_f1_over_nonempty_classes'}).to_csv(
            outdir/'transition_macro_metrics.csv')
    for column in ('mapping','source_argmax_stable','baseline_gap_bin'):
        per_family = df.groupby(['operation','predictor',column,'family_id'])[metrics].mean().reset_index()
        per_family.groupby(['operation','predictor',column])[metrics].mean().to_csv(outdir/f'by_{column}.csv')
    return summary, pd.DataFrame(tests)

IMPLEMENTATION_SHA256='14d29c1a79a01017fede9dfb8ae4972a1a32061c442957226fb841a0536515bf'
```

```python
%%cell_progress 6/14 Load model execution definitions
def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(scores))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def example_stamp(example):
    return dict(example_id=example.example_id,prompt=example.prompt,image=file_hash(example.image_path),
        labels=list(example.labels),semantics=list(example.option_semantics),gold=str(example.gold_semantic),
        torque=int(example.net_torque),split=example.split)

def current_runtime_identity():
    import transformers
    return dict(torch=torch.__version__,transformers=transformers.__version__,cuda=torch.version.cuda,
        dtype=MODEL_DTYPE,batch_size=1,checkpoint=MODEL_REVISION,layer_path=LAYER_PATH,
        implementation=IMPLEMENTATION_SHA256,
        toolkit={str(p.relative_to(TOOLKIT_ROOT)):file_hash(p) for p in sorted((TOOLKIT_ROOT/'src').rglob('*.py'))})
"""GPU execution, pre-intervention sealing and exports for the prospective study."""
import shutil
import zipfile
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample
from visual_torque_toolkit.render import RenderStyle, render_scene
from visual_torque_toolkit.world import Scene, Side, Stack

RULE_TEXT = ('Each square is one unit of weight. Distance is measured in tick '
             'intervals from the center pivot. Torque equals weight times distance. '
             'Total torque on a side is the sum over stacks on that side.')
QUESTION_TEXT = {'outcome':'Which way will the balance scale tip?',
                 'weight_more':'Which side has more total weight?',
                 'weight_less':'Which side has less total weight?'}
SEMANTIC_TEXT = {'left':'left side','balanced':'equal','right':'right side'}


def question_gold(role, task):
    f = role['features']
    value = f['net_torque'] if task == 'outcome' else f['weight_score']*(1 if task == 'weight_more' else -1)
    return 'left' if value>0 else 'right' if value<0 else 'balanced'


def make_predictive_example(family, role_name, task, mapping):
    role = family['roles'][role_name]
    sem = [next(x['semantic'] for x in mapping if x['label']==label) for label in LABELS]
    options = '\n'.join(f"{x['label']}. {SEMANTIC_TEXT[x['semantic']]}" for x in mapping)
    prompt = '\n\n'.join(['Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT, 'Question: '+QUESTION_TEXT[task], 'Options:\n'+options,
        'Do not explain your reasoning. Reply with exactly one option letter and nothing else. Valid replies are: A, B, C.'])
    gold = question_gold(role,task)
    eid = family['family_id']+'__'+role_name+'__'+task+'__'+digest(mapping)[:12]
    return BehaviorExample(example_id=eid,question_id=eid,scene_id=role['scene_id'],
        condition=PromptCondition.IMAGE_RULE,question_kind=task,variant=SCHEMA,
        split=family['split'],problem_type='controlled_four_stack_family',outcome=gold,
        net_torque=int(role['features']['net_torque']),prompt=prompt,
        image_path=IMAGE_DIR/role['image_filename'],labels=LABELS,
        option_text=tuple(SEMANTIC_TEXT[s] for s in sem),option_semantics=tuple(sem),
        gold_label=LABELS[sem.index(gold)],gold_semantic=gold)


def prepared_fingerprint(prepared):
    tensors = {}
    for key, value in prepared['inputs'].items():
        if torch.is_tensor(value):
            x = value.detach().cpu().contiguous()
            raw = x.view(torch.uint8).numpy().tobytes()
            tensors[key] = dict(shape=list(x.shape),dtype=str(x.dtype),sha256=hashlib.sha256(raw).hexdigest())
        else:
            tensors[key] = str(value)
    return digest(dict(inputs=tensors,candidate_tokens=prepared['token_by_label']))


def predictive_forward(example, layer=None, payload=None, random_key=None, expected_pre=None, expected_input=None):
    prepared = prepare_examples([example])
    fingerprint = prepared_fingerprint(prepared)
    if expected_input is not None:
        assert fingerprint == expected_input, 'Prepared prompt/image tensors changed.'
    mask = final_mask(prepared)
    captured, audits, handles = {}, {}, []
    active = CAPTURE_LAYERS if layer is None else (int(layer),)
    for index in active:
        def hook(_module,_inputs,output,index=index):
            hidden = output_hidden_tensor(output)
            m = mask.to(hidden.device)
            assert hidden.dtype == torch.bfloat16 and int(m.sum()) == 1
            pre = hidden[0,m]
            before = pre.detach().float().cpu().numpy().copy()
            if payload is not None:
                assert index == layer
                # Check the current untreated state against the state used to plan this write.
                assert expected_pre is not None
                max_error = float(np.max(np.abs(before-np.asarray(expected_pre))))
                assert max_error == 0., (
                    f'Untreated-state replay differs ({max_error}). Do not loosen the check; '
                    'inspect runtime/input identity and restart this run consistently.')
                delta = torch.as_tensor(np.asarray(payload).copy(),device=pre.device,dtype=pre.dtype).reshape(pre.shape)
                desired = float(((pre+delta).float()-pre.float()).norm())
                if random_key is None:
                    post = pre+delta
                    audit = dict(matched=True,relative_error=0.,target_norm=desired)
                else:
                    cast = lambda x:torch.as_tensor(np.asarray(x).copy(),dtype=torch.bfloat16).float().numpy()
                    post_np,audit = native_norm_edit(before,random_unit(before.shape,random_key),desired,cast)
                    post = torch.as_tensor(post_np,device=pre.device,dtype=pre.dtype)
                audit.update(actual_norm=float((post.float()-pre.float()).norm()),
                             pre_sha256=state_hash(before), planned_payload_sha256=state_hash(payload))
                audits[str(index)] = audit
                modified = hidden.clone(); modified[0,m] = post
                captured[index] = post.detach().float().cpu().numpy().copy()
                return replace_output_hidden(output,modified)
            captured[index] = before
            return None
        handles.append(DECODER_LAYERS[index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            out = adapter.model(**prepared['inputs'],use_cache=False)
        record = score_records(prepared,out.logits)[0]
    finally:
        for handle in handles:
            handle.remove()
    record.update(input_sha256=fingerprint,write_audits=audits)
    return record,captured


def trial_specs(family):
    source_mapping = mapping_reference(family['family_id'])
    if family['module'] == 'content':
        routes = [dict(source_before='base',source_after='aligned_source',recipient='base',
                       task=task,tag=task) for task in ('weight_more','weight_less')]
    else:
        routes = [dict(source_before=f'source_{ds}_before',source_after=f'source_{ds}_after',
                       recipient=f'recipient_{dr}_before',cf_role=f'recipient_{dr}_after',
                       task='outcome',tag=f'{ds}_{dr}',source_distance=family[ds],
                       recipient_distance=family[dr])
                  for ds in ('near','far') for dr in ('near','far')]
    for route in routes:
        if family['module'] == 'content':
            focus = question_gold(family['roles'][route['source_after']],route['task'])
        else:
            side_sign = 1 if family['target_side']=='left' else -1
            focus = 'left' if family['dw']*side_sign>0 else 'right'
        mappings = [('identity',source_mapping),
                    ('conflict_forward',mapping_conflict(source_mapping,focus)),
                    ('conflict_reverse',mapping_conflict(source_mapping,focus,True))]
        for mapping_name,mapping in mappings:
            for layer in SPEC[family['module']+'_layers']:
                tid = family['family_id']+f"__{route['tag']}__{mapping_name}__L{layer}"
                yield dict(route,trial_id=tid,layer=layer,source_mapping=source_mapping,
                           recipient_mapping=mapping,mapping_name=mapping_name)


def render_support(families):
    IMAGE_DIR.mkdir(parents=True,exist_ok=True)
    for f in tqdm(families,desc='Render frozen support'):
        for r in f['roles'].values():
            path = IMAGE_DIR/r['image_filename']
            if path.exists():
                continue
            scene = Scene(r['scene_id'],tuple(Stack(str(i),Side(s['side']),int(s['distance']),int(s['units']))
                                              for i,s in enumerate(r['stacks'])))
            render_scene(scene,path,RenderStyle(ticks_per_side=6))


def capture_family(family):
    """Donors, untreated recipients and a separately labelled extra-input benchmark.

    Natural CF images are queried before sealing, but enter ONLY natural_response.
    They are not hidden-state interventions or inputs to the other predictors.
    """
    folder = PHASE_ROOT/'pre_intervention'
    folder.mkdir(parents=True,exist_ok=True)
    path = folder/(family['family_id']+'.json')
    archive = folder/(family['family_id']+'.npz')
    identity = digest([RUNTIME_IDENTITY,family])
    if path.exists():
        obj = read_json(path)
        assert obj['identity'] == identity and file_hash(archive) == obj['arrays_sha256']
        return obj
    arrays, examples, records, rows = {}, {}, {}, []

    def capture(role,task,mapping):
        ex = make_predictive_example(family,role,task,mapping)
        key = digest(ex.example_id)[:20]
        if key not in examples:
            rec, hidden = predictive_forward(ex)
            examples[key] = dict(role=role,task=task,mapping=mapping,stamp=example_stamp(ex))
            records[key] = rec
            for layer,h in hidden.items():
                arrays[key+f'__L{layer}'] = h
        return key

    for spec in trial_specs(family):
        sb = capture(spec['source_before'],spec['task'],spec['source_mapping'])
        sa = capture(spec['source_after'],spec['task'],spec['source_mapping'])
        rb = capture(spec['recipient'],'outcome',spec['recipient_mapping'])
        layer = spec['layer']
        h0 = arrays[rb+f'__L{layer}']
        # Exact raw native-BF16 subtraction and unit dose; no layer-content label
        # is automatically inherited by this physical-difference operation.
        delta = bf16(arrays[sa+f'__L{layer}']-arrays[sb+f'__L{layer}'])
        realized = bf16(h0+delta)-h0
        norm = float(np.linalg.norm(realized))
        array_key = spec['trial_id']+'__delta'
        arrays[array_key] = delta
        gold = make_predictive_example(family,spec['recipient'],'outcome',spec['recipient_mapping']).gold_label
        row = dict(spec,family_id=family['family_id'],module=family['module'],split=family['split'],
            stratum=family['stratum'],gold=gold,source_key_before=sb,source_key_after=sa,
            recipient_key=rb,payload_key=array_key,pre_key=rb+f'__L{layer}',
            payload_sha256=state_hash(delta),pre_sha256=state_hash(h0),
            source_delta=center(np.array(records[sa]['candidate_logprobs'])-records[sb]['candidate_logprobs']).tolist(),
            source_before_argmax=int(np.argmax(records[sb]['candidate_logprobs'])),
            source_after_argmax=int(np.argmax(records[sa]['candidate_logprobs'])),
            s0=center(records[rb]['candidate_logprobs']).tolist(),
            planned_norm=norm,relative_norm=norm/max(float(np.linalg.norm(h0)),1e-12),
            random_control_family=family['random_control_family'])
        if family['module']=='physical':
            f = family['roles'][spec['recipient']]['features']
            row.update(dw=family['dw'],target_side=family['target_side'],
                       baseline_physics=[f['net_torque'],f['normalized_margin'],f['weight_score'],f['distance_score']])
            row['cf_gold'] = make_predictive_example(family,spec['cf_role'],'outcome',spec['recipient_mapping']).gold_label
            cf_key=capture(spec['cf_role'],'outcome',spec['recipient_mapping'])
            row['natural_input_key']=cf_key
            row['natural_input_delta']=(center(records[cf_key]['candidate_logprobs'])-np.asarray(row['s0'])).tolist()
        rows.append(row)
    # Numpy arrays are data, not pickled Python. Marker written only after the archive.
    temp = archive.with_suffix('.tmp.npz')
    np.savez_compressed(temp,**arrays); temp.replace(archive)
    obj = dict(identity=identity,arrays_sha256=file_hash(archive),family=family,examples=examples,
               baseline_records=records,rows=rows,created_at_utc=utc_now())
    save_json(path,obj)
    return obj


def pre_intervention_rows(families):
    rows, manifest = [], {}
    for family in tqdm(families,desc='Capture donor and untreated recipient states'):
        obj = capture_family(family)
        rows.extend(obj['rows'])
        p = PHASE_ROOT/'pre_intervention'/(family['family_id']+'.json')
        manifest[p.name] = file_hash(p)
    freeze(PHASE_ROOT/'pre_intervention_manifest.json',manifest)
    return rows


def verify_prediction_seal():
    if PHASE != 'validation':
        return None
    path = PHASE_ROOT/'predictions_before_intervention.json'
    assert path.is_file(), 'Save all predictions before running any validation interventions.'
    marker = read_json(PHASE_ROOT/'prediction_seal.json')
    assert file_hash(path) == marker['prediction_sha256']
    assert marker['frozen_models_sha256'] == file_hash(MODEL_ROOT/'frozen_predictors.json')
    assert marker['pre_manifest_sha256'] == file_hash(PHASE_ROOT/'pre_intervention_manifest.json')
    return marker


def execute_targets(families):
    seal = verify_prediction_seal()
    expected_seal = digest(seal) if seal is not None else 'calibration-only'
    folder = PHASE_ROOT/'interventions'; folder.mkdir(parents=True,exist_ok=True)
    all_rows, control_rows, natural_rows = [], [], []
    for family in tqdm(families,desc='Validate fixed interventions'):
        pre_path = PHASE_ROOT/'pre_intervention'/(family['family_id']+'.json')
        obj = read_json(pre_path)
        archive = pre_path.with_suffix('.npz')
        assert file_hash(archive)==obj['arrays_sha256']
        assert file_hash(pre_path)==read_json(PHASE_ROOT/'pre_intervention_manifest.json')[pre_path.name]
        with np.load(archive,allow_pickle=False) as arrays:
            for row in tqdm(obj['rows'],desc=family['family_id'],leave=False):
                path = folder/(row['trial_id']+'.json')
                identity = digest([RUNTIME_IDENTITY,row,expected_seal,obj['arrays_sha256']])
                if path.exists():
                    saved = read_json(path)
                    assert saved['identity']==identity, 'Intervention cache identity changed.'
                else:
                    ex = make_predictive_example(family,row['recipient'],'outcome',row['recipient_mapping'])
                    baseline = obj['baseline_records'][row['recipient_key']]
                    kwargs = dict(layer=row['layer'],payload=arrays[row['payload_key']],
                                  expected_pre=arrays[row['pre_key']],expected_input=baseline['input_sha256'])
                    assert state_hash(kwargs['payload'])==row['payload_sha256']
                    record,_ = predictive_forward(ex,**kwargs)
                    control = []
                    if family['random_control_family']:
                        for seed in RANDOM_SEEDS:
                            rec,_ = predictive_forward(ex,random_key=[SEED,row['trial_id'],seed],**kwargs)
                            control.append(dict(seed=seed,record=rec))
                    natural = None
                    if family['module']=='physical':
                        natural = obj['baseline_records'][row['natural_input_key']]
                    saved = dict(identity=identity,record=record,controls=control,natural_cf=natural,
                                 sealed_prediction_identity=expected_seal,created_at_utc=utc_now())
                    save_json(path,saved)
                record = saved['record']
                actual = center(record['candidate_logprobs'])-np.asarray(row['s0'])
                all_rows.append(dict(row,actual_delta=actual.tolist(),post_record=record))
                for control in saved['controls']:
                    rec = control['record']; audit = rec['write_audits'][str(row['layer'])]
                    control_rows.append(dict(trial_id=row['trial_id'],family_id=row['family_id'],
                        operation=operation_key(row),seed=control['seed'],**audit,
                        actual_delta=(center(rec['candidate_logprobs'])-np.asarray(row['s0'])).tolist()))
                if saved['natural_cf'] is not None:
                    cf = saved['natural_cf']
                    natural_rows.append(dict(trial_id=row['trial_id'],family_id=row['family_id'],
                        operation=operation_key(row),mapping=row['mapping_name'],
                        source_distance=row['source_distance'],recipient_distance=row['recipient_distance'],
                        baseline_gold=row['gold'],cf_gold=row['cf_gold'],
                        baseline_answer=LABELS[int(np.argmax(row['s0']))],
                        patched_answer=record['predicted_label'],natural_answer=cf['predicted_label'],
                        natural_delta=(center(cf['candidate_logprobs'])-np.asarray(row['s0'])).tolist(),
                        patch_delta=actual.tolist()))
    expected = sum(len(list(trial_specs(f))) for f in families)
    assert len(all_rows)==expected and len({r['trial_id'] for r in all_rows})==expected
    save_json(PHASE_ROOT/'target_rows.json',all_rows)
    save_json(PHASE_ROOT/'random_control_rows.json',control_rows)
    save_json(PHASE_ROOT/'natural_counterfactual_rows.json',natural_rows)
    return all_rows


def analyze_controls_and_counterfactuals():
    out = PHASE_ROOT/'analysis'; out.mkdir(parents=True,exist_ok=True)
    target = {r['trial_id']:r for r in read_json(PHASE_ROOT/'target_rows.json')}
    controls = read_json(PHASE_ROOT/'random_control_rows.json')
    pd.DataFrame(controls).to_csv(out/'random_native_norm_audit.csv',index=False)
    comparisons = []
    for c in controls:
        row = target[c['trial_id']]
        if row['module']=='physical':
            axis = np.asarray(predictor_features(row)['recipient_weight'])
        else:
            axis = np.asarray(predictor_features(row)['source_semantic'])
        norm = np.linalg.norm(axis)
        if norm:
            value = float(np.dot(np.asarray(row['actual_delta'])-c['actual_delta'],axis/norm))
        else:
            value = None
        comparisons.append(dict(operation=c['operation'],family_id=c['family_id'],seed=c['seed'],
                                matched=c['matched'],target_minus_random_projection=value))
    frame = pd.DataFrame(comparisons)
    frame.to_csv(out/'random_specificity_trials.csv',index=False)
    # Show all attempts; unmatched controls never silently count as valid equal-norm evidence.
    control_tests = []
    for op,g in frame.groupby('operation'):
        for scope in ('all_attempts_descriptive','all_controls_matched_families'):
            if scope.endswith('families'):
                valid = g.groupby('family_id').matched.all()
                sub = g[g.family_id.isin(valid[valid].index)]
            else: sub=g
            means = sub.groupby('family_id').target_minus_random_projection.mean().dropna()
            if len(means):
                control_tests.append(dict(operation=op,scope=scope,
                    **paired_interval(means.values,[op,scope,'random'])))
    pd.DataFrame(control_tests).to_csv(out/'random_specificity_secondary.csv',index=False)
    natural = read_json(PHASE_ROOT/'natural_counterfactual_rows.json')
    for r in natural:
        r['natural_cf_correct'] = int(r['natural_answer']==r['cf_gold'])
        r['patched_cf_correct'] = int(r['patched_answer']==r['cf_gold'])
        r['baseline_cf_correct'] = int(r['baseline_answer']==r['cf_gold'])
        r['baseline_original_correct'] = int(r['baseline_answer']==r['baseline_gold'])
        r['both_natural_endpoints_correct'] = r['natural_cf_correct']*r['baseline_original_correct']
        r['patch_matches_natural_answer'] = int(r['patched_answer']==r['natural_answer'])
        r['patch_natural_delta_mse'] = float(np.mean((np.asarray(r['patch_delta'])-r['natural_delta'])**2))
        r['cf_flip_by_physics'] = r['baseline_gold'] != r['cf_gold']
    df = pd.DataFrame(natural)
    df.to_csv(out/'natural_counterfactual_trials.csv',index=False)
    columns = ['natural_cf_correct','patched_cf_correct','baseline_cf_correct',
               'baseline_original_correct','both_natural_endpoints_correct',
               'patch_matches_natural_answer','patch_natural_delta_mse']
    if len(df):
        fam = df.groupby(['operation','family_id'])[columns].mean().reset_index()
        fam.to_csv(out/'natural_counterfactual_family_metrics.csv',index=False)
        fam.groupby('operation')[columns].mean().to_csv(out/'natural_counterfactual_summary.csv')
        df.groupby(['operation','source_distance','recipient_distance','family_id'])[columns].mean().groupby(
            ['operation','source_distance','recipient_distance']).mean().to_csv(out/'distance_factorial_secondary.csv')
    return dict(random_writes=len(controls),unmatched_random_writes=sum(not c['matched'] for c in controls))


def export_phase(status):
    """Include support, frozen models, sealed predictions, compact states and raw outputs."""
    completion=PHASE_ROOT/'completion.json'
    status=read_json(completion)['status'] if completion.is_file() else 'partial_capture_not_complete'
    manifest = {}
    candidates = [SUPPORT_PATH,RUN_ROOT/'protocol.json',RUN_ROOT/'historical_geometry_snapshot.json',
                  RUN_ROOT/'image_manifest.json',MODEL_ROOT/'runtime_identity.json',
                  MODEL_ROOT/'frozen_predictors.json']
    candidates += [p for p in PHASE_ROOT.rglob('*') if p.is_file() and not p.name.endswith('.tmp')]
    for p in sorted(set(candidates)):
        if p.is_file():
            manifest[str(p.relative_to(RUN_ROOT))] = file_hash(p)
    archive = RUN_ROOT/f'VisualTorque_Intervention_Prediction_{MODEL_KEY}_{PHASE}_Output.zip'
    with zipfile.ZipFile(archive,'w',zipfile.ZIP_DEFLATED) as z:
        for name in manifest:
            z.write(RUN_ROOT/name,name)
        z.writestr('export_manifest.json',json.dumps(dict(status=status,files=manifest,created_at_utc=utc_now()),indent=2))
    print('Output ZIP:',archive)
    print('Status:',status)
    try:
        from google.colab import files
        files.download(str(archive))
    except Exception:
        print('Automatic browser download did not start; the ZIP is saved at the Drive path above.')
    return archive

IO_IMPLEMENTATION_SHA256='7126756f1b170caaab12637e9d90f5e828e0d6b84a2f93b2ee7efc9d02375f65'
"""I/O-only acceleration. Same candidate files, hashes and geometry extraction.

Kept separate from the scientific implementation hash so an already frozen study
can resume. A separate I/O provenance file records this optimization.
"""
import os
import time
from collections import deque
from concurrent.futures import ThreadPoolExecutor, wait, FIRST_COMPLETED

HISTORY_IO_VERSION = 'scandir-single-read-v1'
HISTORY_IO_WORKERS = 4
_history_progress_time = 0.


def _history_progress(message, force=False):
    global _history_progress_time
    now=time.monotonic()
    if force or now-_history_progress_time >= 4.:
        print(message,flush=True)
        if callable(globals().get('cell_stage')):
            cell_stage(message)
        _history_progress_time=now


def _history_candidate(name):
    lower=name.lower()
    return (lower.endswith(('.json','.jsonl')) and
            any(word in lower for word in ('manifest','support','famil','scene','geometr')))


def _history_scan_directory(folder, excluded_root):
    directories, candidates = [], []
    count=0
    # DirEntry caches file type; noncandidate files never call is_file/resolve.
    with os.scandir(folder) as entries:
        for entry in entries:
            count+=1
            if entry.is_dir(follow_symlinks=False):
                if os.path.normcase(os.path.abspath(entry.path)) != excluded_root:
                    directories.append(entry.path)
            elif _history_candidate(entry.name) and entry.is_file():
                candidates.append(entry.path)
    return directories,candidates,count


def _history_discover(roots, output_root):
    excluded=os.path.normcase(os.path.abspath(output_root))
    queue=deque(str(root) for root in roots
                if root.exists() and os.path.normcase(os.path.abspath(root)) != excluded)
    seen, candidates=set(),set()
    n_dirs=n_entries=0
    _history_progress('Scanning history directories; no GPU inference is running.',True)
    with ThreadPoolExecutor(max_workers=HISTORY_IO_WORKERS) as pool:
        pending={}
        while queue or pending:
            while queue and len(pending)<HISTORY_IO_WORKERS:
                folder=queue.popleft()
                key=os.path.normcase(os.path.abspath(folder))
                if key in seen or key==excluded or key.startswith(excluded+os.sep):
                    continue
                seen.add(key)
                pending[pool.submit(_history_scan_directory,folder,excluded)]=folder
            if not pending:
                continue
            ready,_=wait(pending,timeout=1.,return_when=FIRST_COMPLETED)
            for future in ready:
                folder=pending.pop(future)
                try:
                    directories,files,count=future.result()
                except OSError as exc:
                    raise RuntimeError('Cannot list historical directory: '+folder) from exc
                queue.extend(directories); candidates.update(files)
                n_dirs+=1; n_entries+=count
            active=next(iter(pending.values()),queue[0] if queue else '-')
            _history_progress(f'History scan: {n_dirs} directories; {n_entries} entries; '
                              f'{len(candidates)} candidate JSON files. Current: {active}')
    # Resolve only candidate paths; preserve the old canonical-path scope.
    result=set()
    for name in sorted(candidates):
        path=Path(name).resolve()
        if output_root not in path.parents:
            result.add(str(path))
    _history_progress(f'Directory scan complete: {n_dirs} directories; {len(result)} JSON files to read.',True)
    return sorted(result)


def _history_read_once(name):
    """Parse and hash the SAME bytes, with one Drive read and streaming JSONL."""
    path=Path(name); hasher=hashlib.sha256(); signatures=set()
    if path.suffix == '.jsonl':
        with path.open('rb') as stream:
            for line in stream:
                hasher.update(line)
                if line.strip():
                    signatures.update(geometries_in(json.loads(line.decode('utf-8'))))
    else:
        with path.open('rb') as stream:
            payload=stream.read()
        hasher.update(payload)
        signatures.update(geometries_in(json.loads(payload.decode('utf-8'))))
    return name,hasher.hexdigest(),signatures


def _history_parallel_items(paths, work, description):
    iterator=iter(paths); total=len(paths); done=0
    with ThreadPoolExecutor(max_workers=HISTORY_IO_WORKERS) as pool:
        pending={}
        exhausted=False
        while pending or not exhausted:
            while len(pending)<HISTORY_IO_WORKERS and not exhausted:
                try:
                    name=next(iterator)
                except StopIteration:
                    exhausted=True; break
                pending[pool.submit(work,name)]=name
            if not pending:
                continue
            ready,_=wait(pending,timeout=1.,return_when=FIRST_COMPLETED)
            for future in ready:
                name=pending.pop(future)
                try:
                    result=future.result()
                except Exception as exc:
                    raise RuntimeError('Cannot read historical input: '+name) from exc
                done+=1
                yield result
            active=next(iter(pending.values()),'-')
            _history_progress(f'{description}: {done}/{total} files; current: {active}')
    _history_progress(f'{description} complete: {done}/{total} files.',True)


def fast_history_inventory(project, output_root, extra_roots=()):
    output_root=Path(output_root).resolve()
    lock=output_root/'historical_geometry_snapshot.json'
    if lock.exists():
        obj=read_json(lock)
        _history_progress(f"Verifying the frozen history snapshot ({len(obj['files'])} files); no directory rescan.",True)
        for name,actual in _history_parallel_items(
                sorted(obj['files']),lambda name:(name,file_hash(name)),'Verify frozen history'):
            assert actual==obj['files'][name], 'A frozen historical input changed: '+name
        return {tuple(tuple(t) for t in sig) for sig in obj['signatures']},obj
    roots=[(Path(project)/'data').resolve(),(Path(project)/'results').resolve(),
           *(Path(p).resolve() for p in extra_roots)]
    # A complete file list is saved before reading it. An interrupted read can
    # reuse this original inventory, exactly as a completed snapshot would.
    listing=output_root/'history_file_list_v1.json'
    root_identity=[str(p) for p in roots]
    if listing.exists():
        obj=read_json(listing)
        assert obj['roots']==root_identity, 'Historical roots changed after the file list was frozen.'
        paths=obj['files']
        _history_progress(f'Reusing the frozen file list: {len(paths)} files; directory rescan skipped.',True)
    else:
        paths=_history_discover(roots,output_root)
        save_json(listing,dict(roots=root_identity,files=paths,created_at_utc=utc_now(),
                              scope='candidate set fixed after completed directory enumeration'))
    signatures,hashes=set(),{}
    for name,sha,geometries in _history_parallel_items(paths,_history_read_once,'Read and hash history'):
        signatures.update(geometries); hashes[name]=sha
    assert signatures, 'No historical scenes found. Restore the project data/results before generating new support.'
    obj=dict(files=hashes,signatures=sorted(signatures),created_at_utc=utc_now(),
             scope='available geometry-bearing JSON/JSONL files only; unimported archives are not indexed')
    save_json(lock,obj)
    _history_progress(f'History ready: {len(signatures)} distinct mirror-canonical geometries excluded.',True)
    return signatures,obj


def enable_history_io_acceleration():
    required=('inventory_geometries','geometries_in','save_json','file_hash')
    assert all(name in globals() for name in required), (
        'Run the original utility/configuration cells first, then run this patch in the same notebook namespace.')
    globals()['inventory_geometries']=fast_history_inventory
    # Distinguish the three tasks formerly sharing one unchanged stage label.
    original=globals().get('generate_support')
    if original and not getattr(original,'_history_stage_wrapped',False):
        def generate_with_progress(*args,**kwargs):
            _history_progress('Generating new physical families on CPU; history scanning is complete.',True)
            result=original(*args,**kwargs)
            _history_progress(f'Generated {len(result)} families. Preparing images next.',True)
            return result
        generate_with_progress._history_stage_wrapped=True
        globals()['generate_support']=generate_with_progress
    render=globals().get('render_support')
    if render and not getattr(render,'_history_stage_wrapped',False):
        def render_with_progress(*args,**kwargs):
            _history_progress('Rendering / reusing support images; history scanning is complete.',True)
            result=render(*args,**kwargs)
            _history_progress('Checking image hashes and freezing image identity.',True)
            return result
        render_with_progress._history_stage_wrapped=True
        globals()['render_support']=render_with_progress
    if 'RUN_ROOT' in globals():
        receipt=dict(version=HISTORY_IO_VERSION,
            io_implementation_sha256=globals().get('IO_IMPLEMENTATION_SHA256'),
            scientific_implementation_sha256=globals().get('IMPLEMENTATION_SHA256'),
            workers=HISTORY_IO_WORKERS,enabled_at_utc=utc_now(),
            policy='same history filename filter and full content hashes; no experimental parameters changed')
        save_json(Path(RUN_ROOT)/'io_acceleration_v1.json',receipt)
        if 'PHASE_ROOT' in globals():
            # The existing exporter already includes all phase files.
            save_json(Path(PHASE_ROOT)/'io_acceleration_v1.json',receipt)
    print('Fast history I/O enabled. Keep the same RUN_ID and rerun the interrupted support cell.',flush=True)


enable_history_io_acceleration()
```
