```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. Frozen support and three-valued natural controls


```python
%%cell_progress 6/17 Audit support and lock three-valued task controls
cell_stage('Loading the frozen 29 triplets and scene metadata')
from collections import Counter

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}
source_rows = read_jsonl(SOURCE_MANIFEST_PATH)
assert len(source_rows) == EXPECTED_TRIPLETS
assert len({str(row['triplet_id']) for row in source_rows}) == EXPECTED_TRIPLETS
assert source_decision['manifest_sha256'] == sha256_file(SOURCE_MANIFEST_PATH)

prior_rows = read_jsonl(PRIOR_ORTHOGONAL_MANIFEST_PATH)
prior_scene_ids = {
    str(row[key]) for row in prior_rows
    for key in ('base_scene_id', 'weight_scene_id', 'distance_scene_id')
}
triplet_scene_ids = {
    str(row[key]) for row in source_rows
    for key in ('base_scene_id', 'weight_scene_id', 'distance_scene_id')
}
assert len(triplet_scene_ids) == 3 * EXPECTED_TRIPLETS
assert not (triplet_scene_ids & prior_scene_ids)

def side_torque(scene, side):
    return int(scene['derived'][f'{side}_torque'])

def total_units(scene):
    return int(scene['derived']['left_units']) + int(scene['derived']['right_units'])

def total_distance(scene):
    return sum(int(stack['distance']) for stack in scene['stacks'])

def choose_control(row, excluded):
    side = str(row['changed_side'])
    base = scene_by_id[str(row['base_scene_id'])]
    physical = scene_by_id[str(row['weight_scene_id'])]
    base_torque = side_torque(base, side)
    physical_torque = side_torque(physical, side)
    desired_outcome = str(physical['derived']['outcome'])
    physical_net = int(physical['derived']['net_torque'])
    old_control = str(row['outcome_matched_control_scene_id'])
    ranked = []
    for scene in scene_records:
        scene_id = str(scene['scene_id'])
        if scene_id in excluded or scene_id in prior_scene_ids:
            continue
        candidate_torque = side_torque(scene, side)
        if candidate_torque in {base_torque, physical_torque}:
            continue
        if str(scene['derived']['outcome']) != desired_outcome:
            continue
        if int(scene['derived']['net_torque']) == physical_net:
            continue
        ranked.append((
            -int(scene_id == old_control),
            -int(scene['problem_type'] == physical['problem_type']),
            -int(len(scene['stacks']) == len(physical['stacks'])),
            abs(total_units(scene) - total_units(physical)),
            abs(total_distance(scene) - total_distance(physical)),
            abs(candidate_torque - physical_torque),
            abs(int(scene['derived']['net_torque']) - physical_net),
            stable_key(f"control:{row['triplet_id']}:{scene_id}"),
            scene_id,
        ))
    assert ranked, f"No three-valued control for {row['triplet_id']}"
    ranked.sort()
    return ranked[0][-1]

cell_stage('Selecting globally unique controls without model information')
selected = []
used_controls = set()
for row in sorted(source_rows, key=lambda value: stable_key(value['triplet_id'])):
    control_id = choose_control(
        row, triplet_scene_ids | used_controls,
    )
    used_controls.add(control_id)
    control = scene_by_id[control_id]
    side = str(row['changed_side'])
    candidate = dict(row)
    candidate.update({
        'outcome_matched_control_scene_id': control_id,
        'outcome_matched_control_outcome': str(control['derived']['outcome']),
        'outcome_matched_control_net_torque': int(control['derived']['net_torque']),
        'outcome_matched_control_side_torque': side_torque(control, side),
        'control_reused_from_terminal_bridge': bool(
            control_id == str(row['outcome_matched_control_scene_id'])
        ),
    })
    base_torque = int(row['task_values']['side_torque']['base'])
    physical_torque = int(row['task_values']['side_torque']['weight_source'])
    assert len({
        base_torque, physical_torque,
        candidate['outcome_matched_control_side_torque'],
    }) == 3
    assert (
        candidate['outcome_matched_control_outcome']
        == str(row['task_values']['outcome']['weight_source'])
    )
    selected.append(candidate)

selected.sort(key=lambda row: str(row['triplet_id']))
assert len(selected) == EXPECTED_TRIPLETS
all_selected_scenes = triplet_scene_ids | used_controls
assert len(all_selected_scenes) == 4 * EXPECTED_TRIPLETS
assert not (used_controls & prior_scene_ids)

manifest_identity = {
    'schema_version': 'qwen27b-query-conditional-manifest-1.0',
    'source_manifest_sha256': sha256_file(SOURCE_MANIFEST_PATH),
    'selection_seed': SEED,
    'triplets': EXPECTED_TRIPLETS,
    'scene_disjoint_triplets_and_controls': True,
    'model_outputs_consulted': False,
    'three_distinct_side_torque_values': True,
    'triplet_ids': [row['triplet_id'] for row in selected],
    'control_scene_ids': [
        row['outcome_matched_control_scene_id'] for row in selected
    ],
}
if MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity
    assert read_jsonl(MANIFEST_PATH) == selected
else:
    write_jsonl_atomic(MANIFEST_PATH, selected)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

factorial_manifest = read_jsonl(MANIFEST_PATH)
manifest_frame = pd.DataFrame([{
    key: value for key, value in row.items() if key != 'task_values'
} for row in factorial_manifest])
manifest_frame.to_csv(EXPERIMENT_ROOT / 'query_conditional_manifest.csv', index=False)
print('Triplets:', len(factorial_manifest))
print('Unique controls:', len(used_controls))
print('Controls retained from terminal bridge:', int(
    manifest_frame['control_reused_from_terminal_bridge'].sum()
))
display(manifest_frame.groupby([
    'changed_side', 'factor_direction', 'outcome_transition'
]).size().rename('n').reset_index())

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=EXPERIMENT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)
print('Pinned image examples:', len(outcome_by_scene))
```

## 3. Identical-support dual-task interventions


```python
%%cell_progress 8/17 Define paired torque/outcome prompts and exact token masks
cell_stage('Building three-valued task prompts on identical scene support')
import math
from dataclasses import replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side and compare the two side totals.'
)

def display_value(task, value):
    if task == 'side_torque':
        return str(int(value))
    return {
        'left': 'left side',
        'balanced': 'balanced',
        'right': 'right side',
    }[str(value)]

def task_question(task, side):
    if task == 'side_torque':
        return f'What is the total {side}-side torque?'
    if task == 'outcome':
        return 'Which side will rotate downward?'
    raise KeyError(task)

def rotated(values, row, task):
    values = list(values)
    assert len(values) == 3 and len(set(values)) == 3
    offset = int(stable_key(
        f"options:{row['triplet_id']}:{task}"
    )[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def task_semantics(row, task):
    if task == 'side_torque':
        return rotated([
            int(row['task_values']['side_torque']['base']),
            int(row['task_values']['side_torque']['weight_source']),
            int(row['outcome_matched_control_side_torque']),
        ], row, task)
    return rotated(['left', 'balanced', 'right'], row, task)

def custom_prompt(task, side, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {display_value(task, value)}'
        for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + task_question(task, side),
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def task_values(row, task):
    if task == 'side_torque':
        return {
            'base': int(row['task_values']['side_torque']['base']),
            'weight_source': int(
                row['task_values']['side_torque']['weight_source']
            ),
            'distance_source': int(
                row['task_values']['side_torque']['distance_source']
            ),
            'outcome_matched_control': int(
                row['outcome_matched_control_side_torque']
            ),
        }
    return {
        'base': str(row['task_values']['outcome']['base']),
        'weight_source': str(row['task_values']['outcome']['weight_source']),
        'distance_source': str(row['task_values']['outcome']['distance_source']),
        'outcome_matched_control': str(
            row['outcome_matched_control_outcome']
        ),
    }

def make_task_bundle(row, task):
    side = str(row['changed_side'])
    semantics = task_semantics(row, task)
    values = task_values(row, task)
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, side, semantics)
    scene_ids = {
        'base': str(row['base_scene_id']),
        'weight_source': str(row['weight_scene_id']),
        'distance_source': str(row['distance_scene_id']),
        'outcome_matched_control': str(
            row['outcome_matched_control_scene_id']
        ),
    }
    examples = {}
    for name, scene_id in scene_ids.items():
        value = values[name]
        assert value in semantics, (task, name, value, semantics)
        gold_index = semantics.index(value)
        original = outcome_by_scene[scene_id]
        examples[name] = replace(
            original,
            example_id=(
                f"image_rule:query_{row['triplet_id']}_{task}_{name}"
            ),
            question_id=f"query_{row['triplet_id']}_{task}",
            question_kind=task,
            variant='qwen27b_query_conditional_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(display_value(task, value) for value in semantics),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    goal_semantic = values['weight_source']
    assert values['distance_source'] == goal_semantic
    assert values['base'] != goal_semantic
    if task == 'side_torque':
        assert values['outcome_matched_control'] not in {
            values['base'], goal_semantic,
        }
    else:
        assert values['outcome_matched_control'] == goal_semantic
    return {
        'task': task,
        'semantics': semantics,
        'examples': examples,
        'base_semantic': values['base'],
        'goal_semantic': goal_semantic,
        'control_semantic': values['outcome_matched_control'],
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    masks['pre_options_tokens'] = (
        masks['intro_tokens'] | masks['rule_tokens']
        | masks['question_tokens']
    )
    assert bool(masks['pre_options_tokens'].any())
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(
                reference['inputs'][key], current['inputs'][key]
            ), (name, key)
        masks = discover_masks(current)
        for scope in REQUIRED_SCOPES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Query-conditional prompts and masks ready.')
```

```python
%%cell_progress 9/17 Define frozen-site capture and patch scorers
cell_stage('Defining candidate scoring and direct decoder-output patching')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': example.option_semantics[predicted_index] == example.gold_semantic,
        })
    return records

def forward_capture(prepared):
    hidden = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden[layer_index] = (
                output_hidden_tensor(output)[0].detach().cpu().clone()
            )
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits)[0], hidden

def forward_patch(prepared, donor_hidden, layer_index, patch_mask):
    audit = {}
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = donor_hidden.unsqueeze(0).to(hidden.device, hidden.dtype)
        mask = patch_mask.unsqueeze(0).to(hidden.device)
        assert donor.shape == hidden.shape, (donor.shape, hidden.shape)
        modified = hidden.clone()
        modified[mask] = donor[mask]
        audit['patched_tokens'] = int(mask.sum().item())
        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[int(layer_index)].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    return score_records(prepared, output.logits)[0], audit['patched_tokens']

print('Frozen-site scorer ready.')
```

```python
%%cell_progress 10/17 Run mandatory dual-task alignment and parity tests
cell_stage('Checking three-valued semantics, masks, and exact patch ceilings')
smoke_row = factorial_manifest[0]

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

smoke_errors = []
task_mask_counts = {}
task_prompts = {}
for task in TASKS:
    bundle = make_task_bundle(smoke_row, task)
    prepared = {
        name: prepare_examples([example])
        for name, example in bundle['examples'].items()
    }
    masks = assert_aligned(prepared)
    for scope in REQUIRED_SCOPES:
        assert bool(masks[scope].any()), (task, scope)
    task_mask_counts[task] = {
        scope: int(masks[scope].sum()) for scope in REQUIRED_SCOPES
    }
    task_prompts[task] = bundle['examples']['base'].prompt
    assert len(set(bundle['semantics'])) == 3
    if task == 'side_torque':
        assert bundle['control_semantic'] not in {
            bundle['base_semantic'], bundle['goal_semantic'],
        }
    else:
        assert bundle['control_semantic'] == bundle['goal_semantic']

    base_score, base_hidden = forward_capture(prepared['base'])
    source_score, source_hidden = forward_capture(prepared['weight_source'])
    self_patch, _ = forward_patch(
        prepared['base'], base_hidden[48], 48, masks['valid_tokens']
    )
    source_patch, _ = forward_patch(
        prepared['base'], source_hidden[48], 48, masks['valid_tokens']
    )
    self_error = float(np.max(np.abs(
        centered_scores(self_patch) - centered_scores(base_score)
    )))
    source_error = float(np.max(np.abs(
        centered_scores(source_patch) - centered_scores(source_score)
    )))
    assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, (task, self_error)
    assert source_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, (task, source_error)
    smoke_errors.append((task, self_error, source_error))
    del base_hidden, source_hidden
    torch.cuda.empty_cache()

assert task_prompts['side_torque'] != task_prompts['outcome']
print('Centered self/source parity by task:', smoke_errors)
print('Token counts by task:', task_mask_counts)
print('Three-valued task/control semantics passed.')
```

```python
%%cell_progress 11/17 Capture or resume frozen-site dual-task interventions
cell_stage('Freezing capture identity and validating completed triplet shards')

RECORDS_PER_TASK = len(DIRECT_SITES) * len(DONOR_TYPES)
RECORDS_PER_TRIPLET = len(TASKS) * RECORDS_PER_TASK
EXPECTED_PATCH_RECORDS = EXPECTED_TRIPLETS * RECORDS_PER_TRIPLET
assert RECORDS_PER_TASK == 18 and RECORDS_PER_TRIPLET == 36

capture_identity = {
    'schema_version': 'qwen27b-query-conditional-capture-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'dataset_fingerprint': example_config.dataset_fingerprint(),
    'claim_registry_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'tasks': list(TASKS),
    'sites': [list(site) for site in DIRECT_SITES],
    'donor_types': list(DONOR_TYPES),
    'records_per_triplet': RECORDS_PER_TRIPLET,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'layer_path': LAYER_PATH,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing capture has a different identity. Use a new version.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_triplet_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    triplet_id = str(payload['triplet']['triplet_id'])
    assert triplet_id not in completed_ids
    assert len(payload['patch_records']) == RECORDS_PER_TRIPLET
    completed_ids.add(triplet_id)
    completed_records += len(payload['patch_records'])

def donor_scene_id(row, donor_type):
    return {
        'weight_source': str(row['weight_scene_id']),
        'distance_source': str(row['distance_scene_id']),
        'outcome_matched_control': str(row['outcome_matched_control_scene_id']),
    }[donor_type]

progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc='Qwen-27B query-conditional patches', unit='condition',
    dynamic_ncols=True,
)
names = ('base',) + DONOR_TYPES

for triplet_index, row in enumerate(factorial_manifest):
    triplet_id = str(row['triplet_id'])
    if triplet_id in completed_ids:
        continue
    triplet_baselines = {}
    triplet_records = []
    for task in TASKS:
        progress.set_postfix_str(f'{triplet_id[-8:]} | {task} baselines')
        bundle = make_task_bundle(row, task)
        prepared = {
            name: prepare_examples([bundle['examples'][name]])
            for name in names
        }
        masks = assert_aligned(prepared)
        clean_scores = {}
        clean_hidden = {}
        for name in names:
            score, hidden = forward_capture(prepared[name])
            clean_scores[name] = score
            clean_hidden[name] = hidden
        triplet_baselines[task] = {
            'task': task,
            'semantics': list(bundle['semantics']),
            'base_semantic': bundle['base_semantic'],
            'goal_semantic': bundle['goal_semantic'],
            'control_semantic': bundle['control_semantic'],
            'scores': clean_scores,
        }

        for layer_index, scope, site_label in DIRECT_SITES:
            for donor_type in DONOR_TYPES:
                score, patched_tokens = forward_patch(
                    prepared['base'], clean_hidden[donor_type][layer_index],
                    layer_index, masks[scope],
                )
                score.update({
                    'analysis_family': 'query_conditional_direct_patch',
                    'condition': f'{task}__{donor_type}__{site_label}',
                    'task': task,
                    'donor_type': donor_type,
                    'donor_scene_id': donor_scene_id(row, donor_type),
                    'site_label': site_label,
                    'layer_index': int(layer_index),
                    'scope': scope,
                    'patch_point': 'decoder_output',
                    'patched_tokens': patched_tokens,
                })
                triplet_records.append(score)
                progress.update(1)
        del clean_hidden, prepared
        torch.cuda.empty_cache()

    assert len(triplet_records) == RECORDS_PER_TRIPLET
    write_json_atomic(
        SHARD_DIR / f'shard_triplet_{triplet_index:04d}.json',
        {
            'schema_version': 'qwen27b-query-conditional-triplet-shard-1.0',
            'capture_identity_sha256': CAPTURE_SHA256,
            'triplet': row,
            'baselines': triplet_baselines,
            'patch_records': triplet_records,
        },
    )
progress.close()

completed = sorted(SHARD_DIR.glob('shard_triplet_*.json'))
assert len(completed) == EXPECTED_TRIPLETS
observed = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_TRIPLET
    observed += len(payload['patch_records'])
assert observed == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'triplets': len(completed),
    'patch_records': observed,
    'capture_identity_sha256': CAPTURE_SHA256,
    'peak_gpu_memory_gb': round(torch.cuda.max_memory_allocated() / 2**30, 3),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Capture complete:', observed, 'patch records')
```

## 4. Query interaction, robustness, and decision


```python
%%cell_progress 12/17 Release the GPU and assemble paired task metrics
cell_stage('Releasing model memory')
del adapter
gc.collect()
torch.cuda.empty_cache()

cell_stage('Computing goal-vs-base semantic-margin shifts')

def semantic_margin(record, goal_semantic, base_semantic):
    semantics = list(record['candidate_semantics'])
    scores = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(
        scores[semantics.index(goal_semantic)]
        - scores[semantics.index(base_semantic)]
    )

patch_rows = []
baseline_rows = []
for path in tqdm(
    sorted(SHARD_DIR.glob('shard_triplet_*.json')),
    desc='Assembling task metrics', unit='triplet',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    row = payload['triplet']
    patches = pd.DataFrame(payload['patch_records'])
    for task in TASKS:
        baseline = payload['baselines'][task]
        base_semantic = baseline['base_semantic']
        goal_semantic = baseline['goal_semantic']
        scores = baseline['scores']
        base_margin = semantic_margin(scores['base'], goal_semantic, base_semantic)
        natural = {
            donor: semantic_margin(
                scores[donor], goal_semantic, base_semantic
            ) - base_margin
            for donor in DONOR_TYPES
        }
        baseline_rows.append({
            'triplet_id': str(row['triplet_id']),
            'task': task,
            'changed_side': str(row['changed_side']),
            'factor_direction': str(row['factor_direction']),
            'outcome_transition': str(row['outcome_transition']),
            'weight_natural_margin': natural['weight_source'],
            'distance_natural_margin': natural['distance_source'],
            'control_goal_margin': natural['outcome_matched_control'],
            'base_correct': bool(scores['base']['correct']),
            'weight_correct': bool(scores['weight_source']['correct']),
            'distance_correct': bool(scores['distance_source']['correct']),
            'control_correct': bool(scores['outcome_matched_control']['correct']),
            'base_gold_label': str(scores['base']['gold_semantic']),
        })
        subset = patches[patches['task'] == task]
        assert len(subset) == RECORDS_PER_TASK
        for _, patch in subset.iterrows():
            record = patch.to_dict()
            patch_rows.append({
                'triplet_id': str(row['triplet_id']),
                'task': task,
                'changed_side': str(row['changed_side']),
                'factor_direction': str(row['factor_direction']),
                'outcome_transition': str(row['outcome_transition']),
                'donor_type': str(record['donor_type']),
                'site_label': str(record['site_label']),
                'layer_index': int(record['layer_index']),
                'scope': str(record['scope']),
                'margin_shift': (
                    semantic_margin(record, goal_semantic, base_semantic)
                    - base_margin
                ),
                'natural_margin': natural[str(record['donor_type'])],
                'patched_tokens': int(record['patched_tokens']),
            })

patch_frame = pd.DataFrame(patch_rows)
baseline_frame = pd.DataFrame(baseline_rows)
assert len(patch_frame) == EXPECTED_PATCH_RECORDS
assert len(baseline_frame) == EXPECTED_TRIPLETS * len(TASKS)

scale_rows = []
for task in TASKS:
    subset = baseline_frame[baseline_frame['task'] == task]
    per_triplet = subset[[
        'weight_natural_margin', 'distance_natural_margin'
    ]].abs().mean(axis=1)
    value = float(np.median(per_triplet))
    assert np.isfinite(value) and value > 0, (task, value)
    scale_rows.append({'task': task, 'task_natural_scale': value})
scale_frame = pd.DataFrame(scale_rows)

paired_rows = []
for (triplet_id, task, site_label), group in patch_frame.groupby(
    ['triplet_id', 'task', 'site_label'], sort=True
):
    assert len(group) == len(DONOR_TYPES)
    effects = group.set_index('donor_type')['margin_shift'].to_dict()
    weight = float(effects['weight_source'])
    distance = float(effects['distance_source'])
    control = float(effects['outcome_matched_control'])
    first = group.iloc[0]
    paired_rows.append({
        'triplet_id': triplet_id,
        'task': task,
        'changed_side': str(first['changed_side']),
        'factor_direction': str(first['factor_direction']),
        'outcome_transition': str(first['outcome_transition']),
        'site_label': site_label,
        'layer_index': int(first['layer_index']),
        'scope': str(first['scope']),
        'weight_effect': weight,
        'distance_effect': distance,
        'control_effect': control,
        'mean_physical_effect': (weight + distance) / 2,
        'donor_effect_difference': weight - distance,
        'weight_minus_control': weight - control,
        'distance_minus_control': distance - control,
        'mean_physical_minus_control': (weight + distance) / 2 - control,
        'min_physical_minus_control': min(weight, distance) - control,
    })

paired_frame = pd.DataFrame(paired_rows).merge(
    scale_frame, on='task', validate='many_to_one'
)
expected_paired = EXPECTED_TRIPLETS * len(TASKS) * len(DIRECT_SITES)
assert len(paired_frame) == expected_paired
assert not paired_frame.duplicated(['triplet_id', 'task', 'site_label']).any()
effect_columns = (
    'weight_effect', 'distance_effect', 'control_effect',
    'mean_physical_effect', 'donor_effect_difference',
    'weight_minus_control', 'distance_minus_control',
    'mean_physical_minus_control', 'min_physical_minus_control',
)
for column in effect_columns:
    paired_frame[f'std_{column}'] = (
        paired_frame[column] / paired_frame['task_natural_scale']
    )

patch_frame.to_csv(ANALYSIS_DIR / 'patch_level_metrics.csv', index=False)
baseline_frame.to_csv(ANALYSIS_DIR / 'clean_task_baselines.csv', index=False)
paired_frame.to_csv(ANALYSIS_DIR / 'paired_task_metrics.csv', index=False)
scale_frame.to_csv(ANALYSIS_DIR / 'task_natural_scales.csv', index=False)
print('Patch/baseline/paired rows:',
      len(patch_frame), len(baseline_frame), len(paired_frame))
display(scale_frame)
display(baseline_frame.groupby('task')[[
    'base_correct', 'weight_correct', 'distance_correct', 'control_correct'
]].mean())
```

```python
%%cell_progress 13/17 Test torque specificity and paired query interactions
cell_stage('Running paired bootstrap, sign-flip, FDR, and equivalence tests')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def summarize(values, label):
    values = pd.to_numeric(pd.Series(values), errors='coerce').dropna().to_numpy(float)
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    indices = rng.integers(0, len(values), size=(BOOTSTRAP_REPLICATES, len(values)))
    boot = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice((-1.0, 1.0), size=(SIGN_FLIP_REPLICATES, len(values)))
    null = (signs * values).mean(axis=1)
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
        'sign_flip_p': float(
            (1 + np.sum(np.abs(null) >= abs(observed)))
            / (SIGN_FLIP_REPLICATES + 1)
        ),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

TEST_METRICS = (
    'std_weight_effect', 'std_distance_effect', 'std_control_effect',
    'std_mean_physical_effect', 'std_donor_effect_difference',
    'std_weight_minus_control', 'std_distance_minus_control',
    'std_mean_physical_minus_control',
)
test_rows = []
for (task, site_label), subset in paired_frame.groupby(
    ['task', 'site_label'], sort=True
):
    for metric in TEST_METRICS:
        test_rows.append({
            'family': 'task_site', 'task': task,
            'site_label': site_label,
            'layer_index': int(subset.iloc[0]['layer_index']),
            'scope': str(subset.iloc[0]['scope']),
            'metric': metric,
            **summarize(subset[metric], f'task:{task}:{site_label}:{metric}'),
        })
task_tests = pd.DataFrame(test_rows)
task_tests['q_value'] = np.nan
positive_metrics = {
    'std_weight_effect', 'std_distance_effect',
    'std_weight_minus_control', 'std_distance_minus_control',
}
indices = task_tests.index[task_tests['metric'].isin(positive_metrics)]
task_tests.loc[indices, 'q_value'] = bh_adjust(
    task_tests.loc[indices, 'sign_flip_p'].to_numpy(float)
)

torque = paired_frame[paired_frame['task'] == 'side_torque'].copy()
outcome = paired_frame[paired_frame['task'] == 'outcome'].copy()
interaction_frame = torque.merge(
    outcome, on=['triplet_id', 'site_label'], validate='one_to_one',
    suffixes=('_torque', '_outcome'),
)
interaction_frame['interaction_mean_specificity'] = (
    interaction_frame['std_mean_physical_minus_control_torque']
    - interaction_frame['std_mean_physical_minus_control_outcome']
)
interaction_frame['interaction_weight_specificity'] = (
    interaction_frame['std_weight_minus_control_torque']
    - interaction_frame['std_weight_minus_control_outcome']
)
interaction_frame['interaction_distance_specificity'] = (
    interaction_frame['std_distance_minus_control_torque']
    - interaction_frame['std_distance_minus_control_outcome']
)
interaction_frame['interaction_control_effect'] = (
    interaction_frame['std_control_effect_outcome']
    - interaction_frame['std_control_effect_torque']
)

interaction_metrics = (
    'interaction_mean_specificity',
    'interaction_weight_specificity',
    'interaction_distance_specificity',
    'interaction_control_effect',
)
interaction_rows = []
for site_label, subset in interaction_frame.groupby('site_label', sort=True):
    for metric in interaction_metrics:
        interaction_rows.append({
            'family': 'query_interaction',
            'site_label': site_label,
            'layer_index': int(subset.iloc[0]['layer_index_torque']),
            'scope': str(subset.iloc[0]['scope_torque']),
            'metric': metric,
            **summarize(
                subset[metric], f'interaction:{site_label}:{metric}'
            ),
        })
interaction_tests = pd.DataFrame(interaction_rows)
interaction_tests['q_value'] = np.nan
primary_indices = interaction_tests.index[
    interaction_tests['metric'] == 'interaction_mean_specificity'
]
interaction_tests.loc[primary_indices, 'q_value'] = bh_adjust(
    interaction_tests.loc[
        primary_indices, 'sign_flip_p'
    ].to_numpy(float)
)

def task_test(task, site_label, metric):
    rows = task_tests[
        (task_tests['task'] == task)
        & (task_tests['site_label'] == site_label)
        & (task_tests['metric'] == metric)
    ]
    assert len(rows) == 1, (task, site_label, metric, len(rows))
    return rows.iloc[0]

def interaction_test(site_label, metric='interaction_mean_specificity'):
    rows = interaction_tests[
        (interaction_tests['site_label'] == site_label)
        & (interaction_tests['metric'] == metric)
    ]
    assert len(rows) == 1, (site_label, metric, len(rows))
    return rows.iloc[0]

def positive(row):
    return bool(row['ci_low'] > 0 and row['q_value'] < ALPHA)

def equivalent(row):
    return bool(
        row['ci_low'] > -EQUIVALENCE_FRACTION
        and row['ci_high'] < EQUIVALENCE_FRACTION
    )

site_gate_rows = []
for site_label in [site[2] for site in DIRECT_SITES]:
    weight = task_test('side_torque', site_label, 'std_weight_effect')
    distance = task_test('side_torque', site_label, 'std_distance_effect')
    difference = task_test(
        'side_torque', site_label, 'std_donor_effect_difference'
    )
    weight_specific = task_test(
        'side_torque', site_label, 'std_weight_minus_control'
    )
    distance_specific = task_test(
        'side_torque', site_label, 'std_distance_minus_control'
    )
    interaction = interaction_test(site_label)
    positive_donors = bool(positive(weight) and positive(distance))
    equivalent_donors = equivalent(difference)
    torque_specific = bool(
        positive(weight_specific) and positive(distance_specific)
    )
    interaction_gate = bool(
        interaction['ci_low'] > 0 and interaction['q_value'] < ALPHA
    )
    site_gate_rows.append({
        'site_label': site_label,
        'layer_index': int(weight['layer_index']),
        'scope': str(weight['scope']),
        'positive_torque_donors': positive_donors,
        'equivalent_torque_donors': equivalent_donors,
        'torque_specific_vs_control': torque_specific,
        'query_interaction_gate': interaction_gate,
        'query_conditional_gate': bool(
            positive_donors and equivalent_donors
            and torque_specific and interaction_gate
        ),
    })
site_gates = pd.DataFrame(site_gate_rows)

task_tests.to_csv(ANALYSIS_DIR / 'task_site_tests.csv', index=False)
interaction_frame.to_csv(
    ANALYSIS_DIR / 'paired_query_interactions.csv', index=False
)
interaction_tests.to_csv(
    ANALYSIS_DIR / 'query_interaction_tests.csv', index=False
)
site_gates.to_csv(ANALYSIS_DIR / 'site_gates.csv', index=False)
display(site_gates)
display(interaction_tests[
    interaction_tests['metric'] == 'interaction_mean_specificity'
].sort_values('layer_index'))
```

```python
%%cell_progress 14/17 Audit competence, option balance, and interaction stability
cell_stage('Running leave-one-out and metadata-stratified sensitivity checks')

robustness_rows = []
stratum_rows = []
for site_label, subset in interaction_frame.groupby('site_label', sort=True):
    values = subset['interaction_mean_specificity'].to_numpy(float)
    loo_means = np.asarray([
        np.delete(values, index).mean() for index in range(len(values))
    ])
    side_means = subset.groupby(
        'changed_side_torque'
    )['interaction_mean_specificity'].mean()
    direction_means = subset.groupby(
        'factor_direction_torque'
    )['interaction_mean_specificity'].mean()
    robustness_rows.append({
        'site_label': site_label,
        'n': len(values),
        'mean': float(values.mean()),
        'loo_min_mean': float(loo_means.min()),
        'loo_max_mean': float(loo_means.max()),
        'both_changed_sides_positive': bool(
            len(side_means) == 2 and (side_means > 0).all()
        ),
        'both_factor_directions_positive': bool(
            len(direction_means) == 2 and (direction_means > 0).all()
        ),
        'robustness_gate': bool(
            loo_means.min() > 0
            and len(side_means) == 2 and (side_means > 0).all()
            and len(direction_means) == 2 and (direction_means > 0).all()
        ),
    })
    for column in (
        'changed_side_torque', 'factor_direction_torque',
        'outcome_transition_torque',
    ):
        for value, group in subset.groupby(column):
            stratum_rows.append({
                'site_label': site_label,
                'stratum_variable': column.replace('_torque', ''),
                'stratum_value': str(value),
                'n': len(group),
                'interaction_mean': float(
                    group['interaction_mean_specificity'].mean()
                ),
            })

robustness_frame = pd.DataFrame(robustness_rows)
stratum_frame = pd.DataFrame(stratum_rows)

competence_rows = []
for task, subset in baseline_frame.groupby('task'):
    competence_rows.append({
        'task': task, 'n_triplets': subset['triplet_id'].nunique(),
        'base_accuracy': float(subset['base_correct'].mean()),
        'weight_accuracy': float(subset['weight_correct'].mean()),
        'distance_accuracy': float(subset['distance_correct'].mean()),
        'control_accuracy': float(subset['control_correct'].mean()),
        'both_physical_correct_rate': float(
            (subset['weight_correct'] & subset['distance_correct']).mean()
        ),
    })
competence_frame = pd.DataFrame(competence_rows)

label_rows = []
for row in factorial_manifest:
    for task in TASKS:
        semantics = task_semantics(row, task)
        values = task_values(row, task)
        for role in ('base', 'weight_source', 'outcome_matched_control'):
            label_rows.append({
                'task': task, 'role': role,
                'label': ('A', 'B', 'C')[semantics.index(values[role])],
            })
label_frame = pd.DataFrame(label_rows)
label_balance = label_frame.groupby(
    ['task', 'role', 'label']
).size().rename('n').reset_index()

robustness_frame.to_csv(
    ANALYSIS_DIR / 'interaction_robustness.csv', index=False
)
stratum_frame.to_csv(
    ANALYSIS_DIR / 'interaction_strata.csv', index=False
)
competence_frame.to_csv(
    ANALYSIS_DIR / 'clean_competence.csv', index=False
)
label_balance.to_csv(
    ANALYSIS_DIR / 'answer_label_balance.csv', index=False
)
display(competence_frame)
display(robustness_frame)
display(label_balance)
```

```python
%%cell_progress 16/17 Apply the frozen query-conditional decision tree
cell_stage('Combining causal, interaction, and stability gates')

decision_gates = site_gates.merge(
    robustness_frame[['site_label', 'robustness_gate', 'loo_min_mean']],
    on='site_label', validate='one_to_one',
)
decision_gates['stable_query_conditional_gate'] = (
    decision_gates['query_conditional_gate']
    & decision_gates['robustness_gate']
)

primary_rows = decision_gates[
    decision_gates['site_label'].isin(PRIMARY_SOURCE_SITES)
]
label_entangled_rows = decision_gates[
    decision_gates['site_label'].isin(LABEL_ENTANGLED_SOURCE_SITES)
]
terminal_rows = decision_gates[
    decision_gates['site_label'] == TERMINAL_SITE
]
assert len(primary_rows) == len(PRIMARY_SOURCE_SITES)
assert len(label_entangled_rows) == len(LABEL_ENTANGLED_SOURCE_SITES)
assert len(terminal_rows) == 1

source_supported = bool(primary_rows['stable_query_conditional_gate'].any())
label_entangled_supported = bool(
    label_entangled_rows['stable_query_conditional_gate'].any()
)
terminal_supported = bool(
    terminal_rows.iloc[0]['stable_query_conditional_gate']
)
any_interaction = bool(decision_gates['query_interaction_gate'].any())
any_torque_specificity = bool(
    decision_gates['torque_specific_vs_control'].any()
)

if source_supported and terminal_supported:
    decision = 'query_conditional_route_supported_on_reused_support'
elif source_supported:
    decision = 'query_conditional_l48_torque_use_supported_on_reused_support'
elif label_entangled_supported:
    decision = 'option_entangled_l48_task_switch_supported_only'
elif terminal_supported:
    decision = 'late_query_conditional_readout_supported_only'
elif any_interaction and not any_torque_specificity:
    decision = 'task_interaction_without_torque_specific_torque_query'
else:
    decision = 'query_conditional_torque_use_not_supported'

decision_payload = {
    'schema_version': 'qwen27b-query-conditional-decision-1.0',
    'decision': decision,
    'model_id': MODEL_ID,
    'model_revision': MODEL_REVISION,
    'triplets': EXPECTED_TRIPLETS,
    'support_status': 'reused_from_terminal_bridge_followup',
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'claim_registry_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'capture_config_sha256': CAPTURE_SHA256,
    'source_terminal_decision': source_decision['decision'],
    'source_supported': source_supported,
    'label_entangled_source_supported': label_entangled_supported,
    'terminal_supported': terminal_supported,
    'any_corrected_query_interaction': any_interaction,
    'any_torque_specific_site': any_torque_specificity,
    'site_gates': decision_gates.to_dict(orient='records'),
    'limitations': [
        'The scene support was reused after the terminal-bridge result motivated this test.',
        'The natural control is metadata-matched but can differ in residual visual properties.',
        'The test localizes grouped decoder-output transport, not a unique circuit.',
        'A positive result requires independent model or scene-support replication.',
    ],
}
DECISION_PATH = ANALYSIS_DIR / 'query_conditional_decision.json'
write_json_atomic(DECISION_PATH, decision_payload)
decision_gates.to_csv(ANALYSIS_DIR / 'decision_gates.csv', index=False)
print(json.dumps(decision_payload, indent=2))
```
