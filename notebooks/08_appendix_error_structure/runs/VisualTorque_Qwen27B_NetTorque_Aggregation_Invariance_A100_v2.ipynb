```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. Independent natural factorial quartets


```python
%%cell_progress 6/17 Build independent, scene-disjoint factorial quartets
cell_stage('Loading Dataset v1 geometries for strict exclusion')
from collections import Counter, defaultdict
from itertools import combinations

from visual_torque_toolkit.render import RenderStyle, render_scene
from visual_torque_toolkit.world import Scene, Side, Stack

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

def canonical_record_signature(scene):
    return tuple(sorted(
        (str(stack['side']), int(stack['distance']), int(stack['units']))
        for stack in scene['stacks']
    ))

dataset_scenes = read_jsonl(DATASET_DIR / 'scenes.jsonl')
dataset_signatures = {canonical_record_signature(scene) for scene in dataset_scenes}
assert len(dataset_signatures) == 2100

MAX_DISTANCE = 6
BASE_MAX_UNITS = 4
ENDPOINT_MAX_UNITS = 5
side_shapes = []
for distance in range(1, MAX_DISTANCE + 1):
    for units in range(1, BASE_MAX_UNITS + 1):
        side_shapes.append(((distance, units),))
for near, far in combinations(range(1, MAX_DISTANCE + 1), 2):
    for near_units in range(1, BASE_MAX_UNITS + 1):
        for far_units in range(1, BASE_MAX_UNITS + 1):
            side_shapes.append(((near, near_units), (far, far_units)))
side_shapes = tuple(side_shapes)

def side_torque(shape):
    return sum(int(distance) * int(units) for distance, units in shape)

def edit_shape(shape, stack_index, factor, step):
    edited = [list(value) for value in shape]
    distance, units = edited[stack_index]
    if factor == 'units':
        units += step
    elif factor == 'distance':
        distance += step
    else:
        raise KeyError(factor)
    if not (1 <= units <= ENDPOINT_MAX_UNITS):
        return None
    if not (1 <= distance <= MAX_DISTANCE):
        return None
    if any(
        index != stack_index and int(value[0]) == distance
        for index, value in enumerate(edited)
    ):
        return None
    edited[stack_index] = [distance, units]
    return tuple(sorted((int(d), int(u)) for d, u in edited))

def full_signature(left_shape, right_shape):
    return tuple(sorted(
        [('left', int(d), int(u)) for d, u in left_shape]
        + [('right', int(d), int(u)) for d, u in right_shape]
    ))

def scene_id(signature):
    digest = hashlib.sha256(repr(signature).encode('utf-8')).hexdigest()[:18]
    return f'netagg_{digest}'

def make_scene(signature):
    stacks = []
    side_counts = defaultdict(int)
    for side, distance, units in signature:
        index = side_counts[side]
        side_counts[side] += 1
        stacks.append(Stack(
            stack_id=f'{side[0]}{index}', side=Side(side),
            distance=int(distance), units=int(units),
        ))
    return Scene(scene_id(signature), tuple(stacks))

cell_stage('Enumerating balanced natural 2x2 factorial quartets')
by_torque = defaultdict(list)
for shape in side_shapes:
    by_torque[side_torque(shape)].append(shape)

candidates = []
seen_quartets = set()
for torque, shapes in tqdm(
    sorted(by_torque.items()), desc='Enumerating equal-torque bases', unit='torque'
):
    for left_shape in shapes:
        for right_shape in shapes:
            if left_shape == right_shape:
                continue
            for left_index in range(len(left_shape)):
                for right_index in range(len(right_shape)):
                    for left_factor, right_factor in (
                        ('units', 'units'), ('distance', 'distance'),
                        ('units', 'distance'), ('distance', 'units'),
                    ):
                        for step in (-1, 1):
                            left_edited = edit_shape(
                                left_shape, left_index, left_factor, step
                            )
                            right_edited = edit_shape(
                                right_shape, right_index, right_factor, step
                            )
                            if left_edited is None or right_edited is None:
                                continue
                            delta_left = side_torque(left_edited) - torque
                            delta_right = side_torque(right_edited) - torque
                            if delta_left == 0 or delta_left != delta_right:
                                continue
                            signatures = {
                                'base': full_signature(left_shape, right_shape),
                                'left_source': full_signature(left_edited, right_shape),
                                'right_source': full_signature(left_shape, right_edited),
                                'double_source': full_signature(left_edited, right_edited),
                            }
                            signature_tuple = tuple(signatures[role] for role in SCENE_ROLES)
                            if len(set(signature_tuple)) != 4:
                                continue
                            if any(value in dataset_signatures for value in signature_tuple):
                                continue
                            if signature_tuple in seen_quartets:
                                continue
                            seen_quartets.add(signature_tuple)
                            pattern = f'{left_factor}_{right_factor}'
                            structure = f'{len(left_shape)}x{len(right_shape)}'
                            candidates.append({
                                'signatures': signatures,
                                'edit_pattern': pattern,
                                'direction': 'increase' if step > 0 else 'decrease',
                                'structure': structure,
                                'delta_torque': int(delta_left),
                                'base_side_torque': int(torque),
                            })

assert len(candidates) >= TARGET_QUARTETS, len(candidates)
cell_stage('Selecting balanced strata with globally disjoint geometries')
for row in candidates:
    row['_key'] = stable_key((
        row['edit_pattern'], row['direction'], row['structure'],
        tuple(row['signatures'].values()),
    ))
strata = defaultdict(list)
for row in sorted(candidates, key=lambda value: value['_key']):
    strata[(row['edit_pattern'], row['direction'], row['structure'])].append(row)

selected = []
used_signatures = set()
ordered_strata = sorted(strata, key=stable_key)
while len(selected) < TARGET_QUARTETS:
    progressed = False
    for stratum in ordered_strata:
        choices = strata[stratum]
        chosen_index = None
        for index, row in enumerate(choices):
            values = set(row['signatures'].values())
            if not (values & used_signatures):
                chosen_index = index
                break
        if chosen_index is None:
            continue
        row = choices.pop(chosen_index)
        selected.append(row)
        used_signatures.update(row['signatures'].values())
        progressed = True
        if len(selected) == TARGET_QUARTETS:
            break
    if not progressed:
        break

assert len(selected) >= MIN_QUARTETS, (
    f'Only {len(selected)} scene-disjoint custom quartets; need {MIN_QUARTETS}.'
)
selected = selected[:TARGET_QUARTETS]

cell_stage('Validating physics and rendering custom images')
style = RenderStyle(ticks_per_side=MAX_DISTANCE)
manifest_rows = []
all_scene_ids = set()
for row in tqdm(selected, desc='Rendering factorial quartets', unit='quartet'):
    quartet_id = 'quartet_' + hashlib.sha256(
        repr(tuple(row['signatures'].values())).encode('utf-8')
    ).hexdigest()[:18]
    scenes_by_role = {
        role: make_scene(row['signatures'][role]) for role in SCENE_ROLES
    }
    base = scenes_by_role['base']
    left = scenes_by_role['left_source']
    right = scenes_by_role['right_source']
    double = scenes_by_role['double_source']
    delta = int(row['delta_torque'])
    assert base.net_torque == double.net_torque == 0
    assert left.net_torque == delta and right.net_torque == -delta
    assert base.outcome.value == double.outcome.value == 'balanced'
    role_ids = {role: scene.scene_id for role, scene in scenes_by_role.items()}
    assert not (set(role_ids.values()) & all_scene_ids)
    all_scene_ids.update(role_ids.values())
    image_paths = {}
    for role, scene in scenes_by_role.items():
        path = IMAGE_DIR / f'{scene.scene_id}.png'
        if not path.is_file():
            render_scene(scene, path, style)
        image_paths[role] = str(path.relative_to(CUSTOM_SCENE_ROOT).as_posix())
    task_values = {
        'left_torque': {
            role: int(scene.left_torque) for role, scene in scenes_by_role.items()
        },
        'right_torque': {
            role: int(scene.right_torque) for role, scene in scenes_by_role.items()
        },
        'net_torque': {
            role: int(scene.net_torque) for role, scene in scenes_by_role.items()
        },
        'outcome': {
            role: str(scene.outcome.value) for role, scene in scenes_by_role.items()
        },
    }
    manifest_rows.append({
        'quartet_id': quartet_id,
        'scene_ids': role_ids,
        'image_paths': image_paths,
        'signatures': {
            role: [list(value) for value in signature]
            for role, signature in row['signatures'].items()
        },
        'edit_pattern': row['edit_pattern'],
        'direction': row['direction'],
        'structure': row['structure'],
        'delta_torque': delta,
        'base_side_torque': int(row['base_side_torque']),
        'double_side_torque': int(double.left_torque),
        'task_values': task_values,
    })

manifest_rows.sort(key=lambda value: value['quartet_id'])
manifest_identity = {
    'schema_version': 'net-torque-factorial-quartet-manifest-1.0',
    'dataset_scenes_sha256': sha256_file(DATASET_DIR / 'scenes.jsonl'),
    'selection_seed': SEED,
    'quartets': len(manifest_rows),
    'scene_disjoint': True,
    'dataset_geometry_overlap': 0,
    'model_outputs_consulted': False,
    'quartet_ids': [row['quartet_id'] for row in manifest_rows],
}
if MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity
    assert read_jsonl(MANIFEST_PATH) == manifest_rows
else:
    write_jsonl_atomic(MANIFEST_PATH, manifest_rows)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

factorial_manifest = read_jsonl(MANIFEST_PATH)
manifest_frame = pd.DataFrame([{
    'quartet_id': row['quartet_id'],
    'edit_pattern': row['edit_pattern'],
    'direction': row['direction'],
    'structure': row['structure'],
    'delta_torque': row['delta_torque'],
    'base_side_torque': row['base_side_torque'],
    'double_side_torque': row['double_side_torque'],
} for row in factorial_manifest])
manifest_frame.to_csv(EXPERIMENT_ROOT / 'factorial_quartet_manifest.csv', index=False)
assert len({
    row['scene_ids'][role]
    for row in factorial_manifest for role in SCENE_ROLES
}) == 4 * len(factorial_manifest)
print('Candidate quartets:', len(candidates))
print('Selected quartets:', len(factorial_manifest))
print('Custom scenes:', 4 * len(factorial_manifest))
display(manifest_frame.groupby([
    'edit_pattern', 'direction', 'structure'
]).size().rename('n').reset_index())
```

## 3. Four-task fixed-site interventions


```python
%%cell_progress 8/17 Define four-task prompts and exact token masks
cell_stage('Building identical-option prompts for each quartet')
import math

from visual_torque_toolkit.behavior.data import BehaviorExample
from visual_torque_toolkit.behavior.schema import PromptCondition

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side, then compute left torque minus right torque.'
)

def display_value(task, value):
    if task == 'outcome':
        return {
            'left': 'left side', 'balanced': 'balanced', 'right': 'right side',
        }[str(value)]
    return str(int(value))

def task_question(task):
    return {
        'left_torque': 'What is the total left-side torque?',
        'right_torque': 'What is the total right-side torque?',
        'net_torque': 'What is left torque minus right torque?',
        'outcome': 'Which side will rotate downward?',
    }[task]

def third_numeric_value(base, changed):
    base, changed = int(base), int(changed)
    step = changed - base
    candidates = (
        changed + step, base - step, base + 2 * abs(step),
        max(0, base - 2 * abs(step)),
    )
    for value in candidates:
        if value >= 0 and value not in {base, changed}:
            return int(value)
    raise RuntimeError((base, changed))

def rotated(values, row, task):
    values = list(values)
    assert len(values) == 3 and len(set(values)) == 3, (task, values)
    offset = int(stable_key(
        f"options:{row['quartet_id']}:{task}"
    )[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def task_semantics(row, task):
    values = row['task_values'][task]
    if task == 'left_torque':
        base = int(values['base'])
        changed = int(values['left_source'])
        return rotated([base, changed, third_numeric_value(base, changed)], row, task)
    if task == 'right_torque':
        base = int(values['base'])
        changed = int(values['right_source'])
        return rotated([base, changed, third_numeric_value(base, changed)], row, task)
    if task == 'net_torque':
        return rotated([
            int(values['base']), int(values['left_source']),
            int(values['right_source']),
        ], row, task)
    return rotated(['left', 'balanced', 'right'], row, task)

def custom_prompt(task, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {display_value(task, value)}'
        for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + task_question(task),
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_task_bundle(row, task):
    semantics = task_semantics(row, task)
    values = {
        role: row['task_values'][task][role] for role in SCENE_ROLES
    }
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, semantics)
    examples = {}
    for role in SCENE_ROLES:
        value = values[role]
        if task != 'outcome':
            value = int(value)
        else:
            value = str(value)
        assert value in semantics, (task, role, value, semantics)
        gold_index = semantics.index(value)
        scene_id_value = str(row['scene_ids'][role])
        examples[role] = BehaviorExample(
            example_id=f"image_rule:{row['quartet_id']}:{task}:{role}",
            question_id=f"query_{row['quartet_id']}_{task}",
            scene_id=scene_id_value,
            condition=PromptCondition.IMAGE_RULE,
            question_kind=task,
            variant='net_torque_factorial_fixed_options',
            split='independent_custom_mechanism_test',
            problem_type='factorial_net_torque_invariance',
            outcome=str(row['task_values']['outcome'][role]),
            net_torque=int(row['task_values']['net_torque'][role]),
            prompt=prompt,
            image_path=(CUSTOM_SCENE_ROOT / row['image_paths'][role]).resolve(),
            labels=labels,
            option_text=tuple(display_value(task, value) for value in semantics),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    if task == 'left_torque':
        assert values['base'] == values['right_source']
        assert values['left_source'] == values['double_source']
    elif task == 'right_torque':
        assert values['base'] == values['left_source']
        assert values['right_source'] == values['double_source']
    else:
        assert values['base'] == values['double_source']
        assert len({values['base'], values['left_source'], values['right_source']}) == 3
    return {
        'task': task, 'semantics': semantics, 'values': values,
        'examples': examples, 'base_semantic': values['base'],
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = IMAGE_TOKEN_ID_OVERRIDE if isinstance(
            IMAGE_TOKEN_ID_OVERRIDE, (list, tuple)
        ) else [IMAGE_TOKEN_ID_OVERRIDE]
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()], dtype=torch.bool
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False, return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention, 'image_tokens': image,
        'all_text_except_final': all_text, 'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    masks['pre_options_tokens'] = (
        masks['intro_tokens'] | masks['rule_tokens'] | masks['question_tokens']
    )
    assert bool(masks['pre_options_tokens'].any())
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key]), (name, key)
        masks = discover_masks(current)
        for scope in REQUIRED_SCOPES:
            assert torch.equal(reference_masks[scope], masks[scope]), (name, scope)
    return reference_masks

print('Factorial prompts and token masks ready.')
```

```python
%%cell_progress 9/17 Define capture and batched factorial interventions
cell_stage('Defining candidate scoring and grouped decoder-output interventions')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': example.option_semantics[predicted_index] == example.gold_semantic,
        })
    return records

def forward_capture(prepared):
    hidden = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden[layer_index] = output_hidden_tensor(output)[0].detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits)[0], hidden

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(condition['layer_index']) for condition in conditions}
    assert len(layer_indices) == 1, layer_indices
    layer_index = next(iter(layer_indices))
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = torch.stack([
            condition['intervention_hidden'] for condition in conditions
        ]).to(hidden.device, hidden.dtype)
        masks = torch.stack([
            condition['patch_mask'] for condition in conditions
        ]).to(hidden.device)
        assert donor.shape == hidden.shape, (donor.shape, hidden.shape)
        assert masks.shape == hidden.shape[:2], (masks.shape, hidden.shape)
        modified = hidden.clone()
        modified[masks] = donor[masks]
        audit['patched_tokens'] = [int(mask.sum().item()) for mask in masks]
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    assert len(records) == len(conditions)
    for record, condition, count in zip(
        records, conditions, audit['patched_tokens']
    ):
        record.update({
            key: value for key, value in condition.items()
            if key not in {'intervention_hidden', 'patch_mask'}
        })
        record['patched_tokens'] = count
    return records

def centered_logprob_error(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    left -= left.mean()
    right -= right.mean()
    return float(np.max(np.abs(left - right)))

print('Capture and factorial intervention scorers ready.')
```

```python
%%cell_progress 10/17 Run alignment, physics, and parity smoke tests
cell_stage('Checking quartet physics, task semantics, and exact token alignment')

smoke_row = factorial_manifest[0]
assert smoke_row['task_values']['net_torque']['base'] == 0
assert smoke_row['task_values']['net_torque']['double_source'] == 0
assert (
    smoke_row['task_values']['net_torque']['left_source']
    == -smoke_row['task_values']['net_torque']['right_source']
)
assert smoke_row['task_values']['outcome']['base'] == 'balanced'
assert smoke_row['task_values']['outcome']['double_source'] == 'balanced'

smoke_bundle = make_task_bundle(smoke_row, 'net_torque')
smoke_prepared = {
    role: prepare_examples([example])
    for role, example in smoke_bundle['examples'].items()
}
smoke_masks = assert_aligned(smoke_prepared)
for _, scope, _ in DIRECT_SITES:
    assert bool(smoke_masks[scope].any()), scope

base_score, base_hidden = forward_capture(smoke_prepared['base'])
left_score, left_hidden = forward_capture(smoke_prepared['left_source'])
right_score, right_hidden = forward_capture(smoke_prepared['right_source'])
double_score, double_hidden = forward_capture(smoke_prepared['double_source'])

layer_index = 48
additive_hidden = (
    base_hidden[layer_index].float()
    + left_hidden[layer_index].float() - base_hidden[layer_index].float()
    + right_hidden[layer_index].float() - base_hidden[layer_index].float()
)
assert additive_hidden.shape == double_hidden[layer_index].shape

self_condition = {
    'condition': 'self_base', 'intervention': 'self_base',
    'layer_index': layer_index, 'site_label': 'smoke_valid',
    'scope': 'valid_tokens', 'intervention_hidden': base_hidden[layer_index],
    'patch_mask': smoke_masks['valid_tokens'],
}
smoke_batch = prepare_examples([smoke_bundle['examples']['base']])
smoke_records = forward_intervention_chunk(smoke_batch, [self_condition])
self_error = centered_logprob_error(
    smoke_records[0]['candidate_logprobs'], base_score['candidate_logprobs']
)
assert np.isfinite(self_error), self_error
assert smoke_records[0]['patched_tokens'] == int(smoke_masks['valid_tokens'].sum())
print('Centered self-replay audit:', self_error)
if self_error > SMOKE_MAX_SELF_REPLAY_ERROR:
    print(
        'Warning: self replay exceeds the audit threshold; continuing because '
        'all reported donor effects use the matched self replay as their zero.'
    )
print('All donor effects will be centered against a same-site self replay.')
print('Token counts:', {
    scope: int(smoke_masks[scope].sum()) for scope in REQUIRED_SCOPES
})
print('Smoke clean predictions:', {
    'base': base_score['predicted_semantic'],
    'left': left_score['predicted_semantic'],
    'right': right_score['predicted_semantic'],
    'double': double_score['predicted_semantic'],
})
```

```python
%%cell_progress 11/17 Capture or resume the factorial intervention grid
cell_stage('Running fixed-site quartet interventions; follow the inner bar')

EXPECTED_QUARTETS = len(factorial_manifest)
RECORDS_PER_TASK = len(DIRECT_SITES) * len(INTERVENTIONS)
EXPECTED_PATCH_RECORDS = EXPECTED_QUARTETS * len(TASKS) * RECORDS_PER_TASK
manifest_sha256 = sha256_file(MANIFEST_PATH)
protocol_sha256 = sha256_file(PROTOCOL_LOCK_PATH)

def shard_path(index):
    return SHARD_DIR / f'shard_quartet_{index:04d}.json'

completed = 0
for quartet_index, row in enumerate(tqdm(
    factorial_manifest, desc='Qwen-27B net-torque factorial audit', unit='quartet'
)):
    path = shard_path(quartet_index)
    if path.is_file():
        payload = json.loads(path.read_text(encoding='utf-8'))
        if (
            payload.get('quartet_id') == row['quartet_id']
            and payload.get('manifest_sha256') == manifest_sha256
            and payload.get('protocol_sha256') == protocol_sha256
            and len(payload.get('patch_records', []))
            == len(TASKS) * RECORDS_PER_TASK
        ):
            completed += 1
            continue

    payload = {
        'schema_version': 'net-torque-factorial-shard-2.0',
        'quartet_id': row['quartet_id'],
        'manifest_sha256': manifest_sha256,
        'protocol_sha256': protocol_sha256,
        'quartet': row,
        'baselines': {},
        'patch_records': [],
        'closure_records': [],
    }
    for task in TASKS:
        bundle = make_task_bundle(row, task)
        prepared_by_role = {
            role: prepare_examples([example])
            for role, example in bundle['examples'].items()
        }
        masks = assert_aligned(prepared_by_role)
        scores = {}
        hidden = {}
        for role in SCENE_ROLES:
            scores[role], hidden[role] = forward_capture(prepared_by_role[role])
        payload['baselines'][task] = {
            'semantics': list(bundle['semantics']),
            'values': bundle['values'],
            'scores': scores,
        }

        for layer_index in CAPTURE_LAYERS:
            additive = (
                hidden['base'][layer_index].float()
                + hidden['left_source'][layer_index].float()
                - hidden['base'][layer_index].float()
                + hidden['right_source'][layer_index].float()
                - hidden['base'][layer_index].float()
            )
            for site_layer, scope, site_label in DIRECT_SITES:
                if site_layer != layer_index:
                    continue
                mask = masks[scope]
                natural = hidden['double_source'][layer_index].float()
                base_state = hidden['base'][layer_index].float()
                denominator = (
                    torch.linalg.vector_norm(
                        (hidden['left_source'][layer_index].float() - base_state)[mask]
                    )
                    + torch.linalg.vector_norm(
                        (hidden['right_source'][layer_index].float() - base_state)[mask]
                    )
                ) / 2
                closure = torch.linalg.vector_norm((natural - additive)[mask])
                payload['closure_records'].append({
                    'quartet_id': row['quartet_id'], 'task': task,
                    'layer_index': layer_index, 'scope': scope,
                    'site_label': site_label,
                    'relative_closure_error': float(
                        (closure / denominator.clamp_min(1e-8)).cpu()
                    ),
                })

        conditions = []
        for layer_index, scope, site_label in DIRECT_SITES:
            intervention_states = {
                'self_base': hidden['base'][layer_index],
                'left_source': hidden['left_source'][layer_index],
                'right_source': hidden['right_source'][layer_index],
                'natural_double': hidden['double_source'][layer_index],
                'additive_double': (
                    hidden['base'][layer_index].float()
                    + hidden['left_source'][layer_index].float()
                    - hidden['base'][layer_index].float()
                    + hidden['right_source'][layer_index].float()
                    - hidden['base'][layer_index].float()
                ),
            }
            for intervention, intervention_hidden in intervention_states.items():
                conditions.append({
                    'quartet_id': row['quartet_id'], 'task': task,
                    'condition': f'{intervention}_{site_label}',
                    'intervention': intervention,
                    'layer_index': layer_index, 'scope': scope,
                    'site_label': site_label,
                    'intervention_hidden': intervention_hidden,
                    'patch_mask': masks[scope],
                })

        by_layer = defaultdict(list)
        for condition in conditions:
            by_layer[int(condition['layer_index'])].append(condition)
        for layer_index, layer_conditions in sorted(by_layer.items()):
            for start in range(0, len(layer_conditions), PATCH_CONDITION_BATCH_SIZE):
                chunk = layer_conditions[start:start + PATCH_CONDITION_BATCH_SIZE]
                prepared_batch = prepare_examples([
                    bundle['examples']['base'] for _ in chunk
                ])
                payload['patch_records'].extend(
                    forward_intervention_chunk(prepared_batch, chunk)
                )

    assert len(payload['patch_records']) == len(TASKS) * RECORDS_PER_TASK
    temporary = path.with_suffix('.json.tmp')
    temporary.write_text(
        json.dumps(payload, sort_keys=True) + '\n', encoding='utf-8'
    )
    temporary.replace(path)
    completed += 1

assert completed == EXPECTED_QUARTETS
actual_records = 0
for path in SHARD_DIR.glob('shard_quartet_*.json'):
    payload = json.loads(path.read_text(encoding='utf-8'))
    if payload.get('manifest_sha256') == manifest_sha256:
        actual_records += len(payload.get('patch_records', []))
assert actual_records == EXPECTED_PATCH_RECORDS, (actual_records, EXPECTED_PATCH_RECORDS)
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'quartets': EXPECTED_QUARTETS,
    'patch_records': actual_records,
    'manifest_sha256': manifest_sha256,
    'protocol_sha256': protocol_sha256,
    'peak_gpu_gb': round(torch.cuda.max_memory_allocated() / 2**30, 3),
})
print('Capture complete:', actual_records, 'patch records')
```

```python
%%cell_progress 12/17 Release GPU and assemble quartet-level estimands
cell_stage('Releasing model memory')
import gc
del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_score(record, semantic):
    semantics = list(record['candidate_semantics'])
    scores = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(scores[semantics.index(semantic)])

def semantic_contrast(record, goal, reference):
    return semantic_score(record, goal) - semantic_score(record, reference)

def gold_margin(record, semantic):
    semantics = list(record['candidate_semantics'])
    scores = np.asarray(record['candidate_logprobs'], dtype=float)
    index = semantics.index(semantic)
    return float(scores[index] - np.max(np.delete(scores, index)))

def centered_rmse(left_record, right_record):
    left = np.asarray(left_record['candidate_logprobs'], dtype=float)
    right = np.asarray(right_record['candidate_logprobs'], dtype=float)
    left -= left.mean()
    right -= right.mean()
    return float(np.sqrt(np.mean((left - right) ** 2)))

baseline_rows = []
raw_rows = []
closure_rows = []
for path in tqdm(
    sorted(SHARD_DIR.glob('shard_quartet_*.json')),
    desc='Assembling quartet metrics', unit='quartet',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    row = payload['quartet']
    closure_rows.extend(payload['closure_records'])
    patches = pd.DataFrame(payload['patch_records'])
    for task in TASKS:
        baseline = payload['baselines'][task]
        values = baseline['values']
        scores = baseline['scores']
        base_semantic = values['base']
        baseline_rows.append({
            'quartet_id': row['quartet_id'], 'task': task,
            'edit_pattern': row['edit_pattern'],
            'direction': row['direction'], 'structure': row['structure'],
            'delta_torque': int(row['delta_torque']),
            **{
                f'{role}_correct': bool(scores[role]['correct'])
                for role in SCENE_ROLES
            },
            **{
                f'{role}_gold_margin': gold_margin(scores[role], values[role])
                for role in SCENE_ROLES
            },
        })
        subset = patches[patches['task'] == task]
        assert len(subset) == RECORDS_PER_TASK
        for site_label, site_group in subset.groupby('site_label'):
            records = {
                str(record['intervention']): record.to_dict()
                for _, record in site_group.iterrows()
            }
            assert set(records) == set(INTERVENTIONS)
            base_score = scores['base']
            self_record = records['self_base']
            base_clean_gold = gold_margin(base_score, base_semantic)
            base_replay_gold = gold_margin(self_record, base_semantic)
            natural_double = records['natural_double']
            additive_double = records['additive_double']
            left_record = records['left_source']
            right_record = records['right_source']

            left_semantic = values['left_source']
            right_semantic = values['right_source']
            if left_semantic != base_semantic:
                left_shift = (
                    semantic_contrast(left_record, left_semantic, base_semantic)
                    - semantic_contrast(self_record, left_semantic, base_semantic)
                )
            else:
                left_shift = 0.0
            if right_semantic != base_semantic:
                right_shift = (
                    semantic_contrast(right_record, right_semantic, base_semantic)
                    - semantic_contrast(self_record, right_semantic, base_semantic)
                )
            else:
                right_shift = 0.0

            if task == 'left_torque':
                changed = values['left_source']
                relevant = left_shift
                irrelevant = (
                    semantic_contrast(right_record, changed, base_semantic)
                    - semantic_contrast(self_record, changed, base_semantic)
                )
            elif task == 'right_torque':
                changed = values['right_source']
                relevant = right_shift
                irrelevant = (
                    semantic_contrast(left_record, changed, base_semantic)
                    - semantic_contrast(self_record, changed, base_semantic)
                )
            else:
                relevant = np.nan
                irrelevant = np.nan

            raw_rows.append({
                'quartet_id': row['quartet_id'], 'task': task,
                'edit_pattern': row['edit_pattern'],
                'direction': row['direction'], 'structure': row['structure'],
                'delta_torque': int(row['delta_torque']),
                'site_label': site_label,
                'layer_index': int(site_group.iloc[0]['layer_index']),
                'scope': str(site_group.iloc[0]['scope']),
                'left_target_shift': float(left_shift),
                'right_target_shift': float(right_shift),
                'signed_donor_difference': float(left_shift - right_shift),
                'side_relevant_effect': float(relevant),
                'side_irrelevant_effect': float(irrelevant),
                'side_selectivity': float(relevant - irrelevant),
                'natural_double_base_change': float(
                    gold_margin(natural_double, base_semantic) - base_replay_gold
                ),
                'additive_double_base_change': float(
                    gold_margin(additive_double, base_semantic) - base_replay_gold
                ),
                'additive_minus_natural_base': float(
                    gold_margin(additive_double, base_semantic)
                    - gold_margin(natural_double, base_semantic)
                ),
                'additive_natural_centered_rmse': centered_rmse(
                    additive_double, natural_double
                ),
                'natural_double_clean_rmse': centered_rmse(
                    natural_double, scores['double_source']
                ),
                'self_replay_base_change': float(
                    base_replay_gold - base_clean_gold
                ),
                'base_clean_gold_margin': float(base_clean_gold),
                'base_replay_gold_margin': float(base_replay_gold),
            })

raw_frame = pd.DataFrame(raw_rows)
baseline_frame = pd.DataFrame(baseline_rows)
closure_frame = pd.DataFrame(closure_rows)
assert len(raw_frame) == EXPECTED_QUARTETS * len(TASKS) * len(DIRECT_SITES)

scale_rows = []
for task in TASKS:
    values = []
    for path in sorted(SHARD_DIR.glob('shard_quartet_*.json')):
        payload = json.loads(path.read_text(encoding='utf-8'))
        baseline = payload['baselines'][task]
        task_values = baseline['values']
        scores = baseline['scores']
        base_semantic = task_values['base']
        roles = (
            ('left_source',) if task == 'left_torque' else
            ('right_source',) if task == 'right_torque' else
            ('left_source', 'right_source')
        )
        for role in roles:
            goal = task_values[role]
            natural = (
                semantic_contrast(scores[role], goal, base_semantic)
                - semantic_contrast(scores['base'], goal, base_semantic)
            )
            values.append(abs(float(natural)))
    scale = float(np.median(values))
    assert np.isfinite(scale) and scale > 0, (task, scale)
    scale_rows.append({'task': task, 'task_natural_scale': scale})
scale_frame = pd.DataFrame(scale_rows)
paired_frame = raw_frame.merge(scale_frame, on='task', validate='many_to_one')
metric_columns = (
    'left_target_shift', 'right_target_shift', 'signed_donor_difference',
    'side_relevant_effect', 'side_irrelevant_effect', 'side_selectivity',
    'natural_double_base_change', 'additive_double_base_change',
    'additive_minus_natural_base', 'additive_natural_centered_rmse',
    'natural_double_clean_rmse', 'self_replay_base_change',
)
for column in metric_columns:
    paired_frame[f'std_{column}'] = (
        paired_frame[column] / paired_frame['task_natural_scale']
    )

baseline_frame.to_csv(ANALYSIS_DIR / 'clean_baselines.csv', index=False)
raw_frame.to_csv(ANALYSIS_DIR / 'raw_quartet_metrics.csv', index=False)
paired_frame.to_csv(ANALYSIS_DIR / 'standardized_quartet_metrics.csv', index=False)
closure_frame.to_csv(ANALYSIS_DIR / 'hidden_factorial_closure.csv', index=False)
scale_frame.to_csv(ANALYSIS_DIR / 'task_natural_scales.csv', index=False)
print('Baseline/paired/closure rows:',
      len(baseline_frame), len(paired_frame), len(closure_frame))
display(scale_frame)
display(baseline_frame.groupby('task')[[
    f'{role}_correct' for role in SCENE_ROLES
]].mean())
```

```python
%%cell_progress 13/17 Test side binding, signed aggregation, and invariance
cell_stage('Running paired bootstrap, sign-flip, FDR, and equivalence tests')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def summarize(values, label):
    values = pd.to_numeric(pd.Series(values), errors='coerce').dropna().to_numpy(float)
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    indices = rng.integers(0, len(values), size=(BOOTSTRAP_REPLICATES, len(values)))
    boot = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice((-1.0, 1.0), size=(SIGN_FLIP_REPLICATES, len(values)))
    null = (signs * values).mean(axis=1)
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
        'sign_flip_p': float(
            (1 + np.sum(np.abs(null) >= abs(observed)))
            / (SIGN_FLIP_REPLICATES + 1)
        ),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

TEST_METRICS = (
    'std_left_target_shift', 'std_right_target_shift',
    'std_signed_donor_difference', 'std_side_relevant_effect',
    'std_side_irrelevant_effect', 'std_side_selectivity',
    'std_natural_double_base_change', 'std_additive_double_base_change',
    'std_additive_minus_natural_base',
    'std_additive_natural_centered_rmse',
    'std_natural_double_clean_rmse', 'std_self_replay_base_change',
)
test_rows = []
for (task, site_label), subset in paired_frame.groupby(['task', 'site_label']):
    for metric in TEST_METRICS:
        test_rows.append({
            'task': task, 'site_label': site_label,
            'layer_index': int(subset.iloc[0]['layer_index']),
            'scope': str(subset.iloc[0]['scope']), 'metric': metric,
            **summarize(subset[metric], f'{task}:{site_label}:{metric}'),
        })
tests = pd.DataFrame(test_rows)
tests['q_value'] = np.nan
positive_mask = (
    (
        tests['task'].isin(('left_torque', 'right_torque'))
        & (tests['metric'] == 'std_side_selectivity')
    )
    | (
        tests['task'].isin(('net_torque', 'outcome'))
        & tests['metric'].isin((
            'std_left_target_shift', 'std_right_target_shift'
        ))
    )
)
for task, indices in tests[positive_mask].groupby('task').groups.items():
    tests.loc[indices, 'q_value'] = bh_adjust(
        tests.loc[indices, 'sign_flip_p'].to_numpy(float)
    )

def test_row(task, site_label, metric):
    rows = tests[
        (tests['task'] == task) & (tests['site_label'] == site_label)
        & (tests['metric'] == metric)
    ]
    assert len(rows) == 1, (task, site_label, metric, len(rows))
    return rows.iloc[0]

def positive(row):
    return bool(row['ci_low'] > 0 and row['q_value'] < ALPHA)

def equivalent(row):
    return bool(
        row['ci_low'] > -EQUIVALENCE_FRACTION
        and row['ci_high'] < EQUIVALENCE_FRACTION
    )

site_gate_rows = []
for _, _, site_label in DIRECT_SITES:
    left_binding = test_row('left_torque', site_label, 'std_side_selectivity')
    right_binding = test_row('right_torque', site_label, 'std_side_selectivity')
    net_left = test_row('net_torque', site_label, 'std_left_target_shift')
    net_right = test_row('net_torque', site_label, 'std_right_target_shift')
    net_difference = test_row(
        'net_torque', site_label, 'std_signed_donor_difference'
    )
    net_invariance = test_row(
        'net_torque', site_label, 'std_natural_double_base_change'
    )
    net_additive = test_row(
        'net_torque', site_label, 'std_additive_minus_natural_base'
    )
    outcome_left = test_row('outcome', site_label, 'std_left_target_shift')
    outcome_right = test_row('outcome', site_label, 'std_right_target_shift')
    outcome_difference = test_row(
        'outcome', site_label, 'std_signed_donor_difference'
    )
    outcome_invariance = test_row(
        'outcome', site_label, 'std_natural_double_base_change'
    )
    side_binding_gate = bool(positive(left_binding) and positive(right_binding))
    signed_net_gate = bool(
        positive(net_left) and positive(net_right) and equivalent(net_difference)
    )
    same_net_invariance_gate = equivalent(net_invariance)
    factorial_diagnostic_gate = equivalent(net_additive)
    terminal_outcome_gate = bool(
        positive(outcome_left) and positive(outcome_right)
        and equivalent(outcome_difference) and equivalent(outcome_invariance)
    )
    site_gate_rows.append({
        'site_label': site_label,
        'layer_index': int(net_left['layer_index']),
        'scope': str(net_left['scope']),
        'side_binding_gate': side_binding_gate,
        'signed_net_gate': signed_net_gate,
        'same_net_invariance_gate': same_net_invariance_gate,
        'factorial_diagnostic_gate': factorial_diagnostic_gate,
        'net_aggregation_gate': bool(
            side_binding_gate and signed_net_gate and same_net_invariance_gate
        ),
        'terminal_outcome_gate': terminal_outcome_gate,
    })
site_gates = pd.DataFrame(site_gate_rows)

tests.to_csv(ANALYSIS_DIR / 'bootstrap_tests.csv', index=False)
site_gates.to_csv(ANALYSIS_DIR / 'site_gates.csv', index=False)
display(site_gates)
display(tests[
    (tests['task'].isin(('net_torque', 'outcome')))
    & (tests['metric'].isin((
        'std_left_target_shift', 'std_right_target_shift',
        'std_signed_donor_difference', 'std_natural_double_base_change',
    )))
].sort_values(['task', 'layer_index', 'site_label', 'metric']))
```

```python
%%cell_progress 14/17 Audit competence, strata, labels, and leave-one-out stability
cell_stage('Running sensitivity checks on the frozen quartet support')

robustness_rows = []
stratum_rows = []
for task in ('net_torque', 'outcome'):
    for site_label, subset in paired_frame[
        paired_frame['task'] == task
    ].groupby('site_label'):
        metrics = {
            'left': subset['std_left_target_shift'].to_numpy(float),
            'right': subset['std_right_target_shift'].to_numpy(float),
            'invariance': subset['std_natural_double_base_change'].to_numpy(float),
        }
        left_loo = np.asarray([
            np.delete(metrics['left'], index).mean()
            for index in range(len(subset))
        ])
        right_loo = np.asarray([
            np.delete(metrics['right'], index).mean()
            for index in range(len(subset))
        ])
        direction = subset.groupby('direction')[
            ['std_left_target_shift', 'std_right_target_shift']
        ].mean()
        pattern = subset.groupby('edit_pattern')[
            ['std_left_target_shift', 'std_right_target_shift']
        ].mean()
        robustness_rows.append({
            'task': task, 'site_label': site_label, 'n': len(subset),
            'left_loo_min': float(left_loo.min()),
            'right_loo_min': float(right_loo.min()),
            'both_directions_positive': bool(
                len(direction) == 2 and (direction > 0).all().all()
            ),
            'all_patterns_positive': bool(
                len(pattern) >= 3 and (pattern > 0).all().all()
            ),
            'robustness_gate': bool(
                left_loo.min() > 0 and right_loo.min() > 0
                and len(direction) == 2 and (direction > 0).all().all()
                and len(pattern) >= 3 and (pattern > 0).all().all()
            ),
        })
        for column in ('direction', 'edit_pattern', 'structure'):
            for value, group in subset.groupby(column):
                stratum_rows.append({
                    'task': task, 'site_label': site_label,
                    'stratum_variable': column,
                    'stratum_value': str(value), 'n': len(group),
                    'left_mean': float(group['std_left_target_shift'].mean()),
                    'right_mean': float(group['std_right_target_shift'].mean()),
                    'invariance_mean': float(
                        group['std_natural_double_base_change'].mean()
                    ),
                })

robustness_frame = pd.DataFrame(robustness_rows)
stratum_frame = pd.DataFrame(stratum_rows)

competence_rows = []
for task, subset in baseline_frame.groupby('task'):
    competence_rows.append({
        'task': task, 'quartets': subset['quartet_id'].nunique(),
        **{
            f'{role}_accuracy': float(subset[f'{role}_correct'].mean())
            for role in SCENE_ROLES
        },
        'all_four_correct_rate': float(
            subset[[f'{role}_correct' for role in SCENE_ROLES]].all(axis=1).mean()
        ),
    })
competence_frame = pd.DataFrame(competence_rows)

label_rows = []
for row in factorial_manifest:
    for task in TASKS:
        semantics = task_semantics(row, task)
        for role in SCENE_ROLES:
            value = row['task_values'][task][role]
            if task != 'outcome':
                value = int(value)
            label_rows.append({
                'task': task, 'role': role,
                'label': ('A', 'B', 'C')[semantics.index(value)],
            })
label_balance = pd.DataFrame(label_rows).groupby(
    ['task', 'role', 'label']
).size().rename('n').reset_index()

robustness_wide = robustness_frame.pivot(
    index='site_label', columns='task', values='robustness_gate'
).rename(columns={
    'net_torque': 'net_robustness_gate',
    'outcome': 'outcome_robustness_gate',
}).reset_index()
decision_gates = site_gates.merge(
    robustness_wide, on='site_label', validate='one_to_one',
)
decision_gates['stable_net_aggregation_gate'] = (
    decision_gates['net_aggregation_gate']
    & decision_gates['net_robustness_gate']
)
decision_gates['stable_terminal_outcome_gate'] = (
    decision_gates['terminal_outcome_gate']
    & decision_gates['outcome_robustness_gate']
)

robustness_frame.to_csv(ANALYSIS_DIR / 'net_aggregation_robustness.csv', index=False)
stratum_frame.to_csv(ANALYSIS_DIR / 'net_aggregation_strata.csv', index=False)
competence_frame.to_csv(ANALYSIS_DIR / 'clean_competence.csv', index=False)
label_balance.to_csv(ANALYSIS_DIR / 'answer_label_balance.csv', index=False)
decision_gates.to_csv(ANALYSIS_DIR / 'decision_gates.csv', index=False)
display(competence_frame)
display(decision_gates)
display(label_balance)
```
