```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready.')

IMPLEMENTATION_SHA256 = '160372af558929c42134517d3ffc09527278d018135dd39a46f32eb6987dde7d'
IMPLEMENTATION_SHA256 = 'a35c2afb83df2ed082f34819d0f3f4562a2928e574bc670a3a32fb93b8eaa94c'
```

```python
%%cell_progress 3/8 Load shared validation utilities
"""Pure CPU utilities embedded in the follow-up notebooks; no model execution."""
import hashlib
import json
from pathlib import Path
from collections import defaultdict, Counter
from itertools import combinations, product
import numpy as np
import pandas as pd


def plain(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(type(value).__name__)


def digest(value):
    return hashlib.sha256(json.dumps(value, sort_keys=True, default=plain, allow_nan=False).encode()).hexdigest()


from visualtorque_checks import file_hash


def read_json(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))


def read_jsonl(path):
    return [json.loads(s) for s in Path(path).read_text(encoding='utf-8').splitlines() if s.strip()]


def save_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + '.tmp')
    temp.write_text(json.dumps(value, ensure_ascii=False, indent=2, default=plain, allow_nan=False) + '\n', encoding='utf-8')
    temp.replace(path)


def freeze(path, identity):
    identity = json.loads(json.dumps(identity, default=plain, sort_keys=True, allow_nan=False))
    if Path(path).is_file():
        assert read_json(path) == identity, f'Configuration changed; preserve the old run and use a new RUN_ID: {path}'
    else:
        save_json(path, identity)
    return digest(identity)


def cached_json(directory, key, identity, compute):
    path = Path(directory) / (digest(str(key))[:24] + '.json')
    expected = digest(identity)
    if path.is_file():
        old = read_json(path)
        assert old['key'] == str(key) and old['identity'] == expected, f'Incompatible cache: {path}'
        return old['result']
    value = compute()
    save_json(path, dict(key=str(key), identity=expected, result=value))
    return value


def bootstrap(values, key='', replicates=10000):
    x = np.asarray(values, dtype=float)
    assert x.ndim == 1 and len(x) and np.isfinite(x).all()
    rng = np.random.default_rng(int(digest(str(key))[:16], 16))
    draws = np.concatenate([rng.choice(x, (min(250, replicates-i), len(x)), replace=True).mean(1)
                            for i in range(0, replicates, 250)])
    return dict(n=len(x), mean=float(x.mean()), lower=float(np.quantile(draws, .025)),
                upper=float(np.quantile(draws, .975)),
                p_value=float((1 + (draws <= 0).sum()) / (replicates + 1)))


def bh(values):
    p = np.asarray(values, dtype=float)
    assert np.isfinite(p).all() and ((p >= 0) & (p <= 1)).all()
    order = np.argsort(p)
    q = np.empty_like(p)
    previous = 1.
    for rank in range(len(p)-1, -1, -1):
        i = order[rank]
        previous = min(previous, p[i]*len(p)/(rank+1))
        q[i] = previous
    return q


def paired(left, right, id_col, value):
    assert left[id_col].is_unique and right[id_col].is_unique
    assert set(left[id_col]) == set(right[id_col]), 'Different paired support'
    out = left[[id_col, value]].merge(right[[id_col, value]], on=id_col, validate='one_to_one', suffixes=('_l', '_r'))
    return out[value+'_l'].to_numpy(float) - out[value+'_r'].to_numpy(float)


def semantic_scores(record):
    if 'semantic_logprobs' in record:
        return {str(k): float(v) for k, v in record['semantic_logprobs'].items()}
    values = record['candidate_logprobs']
    if isinstance(values, dict):
        return {str(record['label_to_semantic'][k]): float(v) for k, v in values.items()}
    return dict(zip(map(str, record['candidate_semantics']), map(float, values)))


def correct_margin(record):
    values = semantic_scores(record)
    gold = str(record['gold_semantic'])
    assert gold in values and len(values) >= 2
    return values[gold] - max(v for k, v in values.items() if k != gold)


def random_unit(shape, key):
    r = np.random.default_rng(int(digest(key)[:16], 16)).normal(size=shape)
    return r / np.linalg.norm(r)


def coordinate_clamp(pre, untreated, axis, cast):
    h, base, u = [np.asarray(x, dtype=np.float64) for x in (pre, untreated, axis)]
    u = u / np.linalg.norm(u)
    a = float(np.sum((h-base)*u))
    return np.asarray(cast(h-a*u), dtype=np.float64), a


def native_norm_edit(pre, direction, norm, cast, tolerance=.02, steps=48):
    h = np.asarray(pre, dtype=np.float64)
    r = np.asarray(direction, dtype=np.float64)
    assert np.isfinite(h).all() and np.isfinite(r).all() and norm >= 0 and np.linalg.norm(r) > 0
    r /= np.linalg.norm(r)
    if norm == 0:
        return h.copy(), dict(actual_norm=0., target_norm=0., relative_error=0., matched=True)
    lo, hi = 0., max(float(norm), 1e-12)
    for _ in range(32):
        if np.linalg.norm(np.asarray(cast(h+hi*r))-h) >= norm:
            break
        hi *= 2
    best, error = h.copy(), 1.
    for _ in range(steps):
        mid = (lo+hi)/2
        candidate = np.asarray(cast(h+mid*r), dtype=np.float64)
        actual = float(np.linalg.norm(candidate-h))
        current = abs(actual-norm)/norm
        if current < error:
            best, error = candidate.copy(), current
        if error <= tolerance:
            break
        if actual < norm:
            lo = mid
        else:
            hi = mid
    return best, dict(actual_norm=float(np.linalg.norm(best-h)), target_norm=float(norm),
                      relative_error=float(error), matched=bool(error <= tolerance))


def equal_removal(pre, untreated, axis, random_direction, cast, tolerance=.02):
    target, coefficient = coordinate_clamp(pre, untreated, axis, cast)
    target_norm = float(np.linalg.norm(target-np.asarray(pre)))
    sign = 1. if coefficient >= 0 else -1.
    post, audit = native_norm_edit(pre, -sign*np.asarray(random_direction), target_norm, cast, tolerance)
    audit.update(reference_coefficient=coefficient, reference_actual_norm=target_norm)
    return post, audit


def geometry_signature(stacks):
    a = tuple(sorted((str(s['side']), int(s['distance']), int(s['units'])) for s in stacks))
    b = tuple(sorted(('right' if side == 'left' else 'left', d, u) for side, d, u in a))
    return min(a, b)  # mirrors belong to the same physical family


def geometries_in(value):
    found = set()
    if isinstance(value, dict):
        stacks = value.get('stacks')
        if isinstance(stacks, list) and stacks and all(isinstance(s, dict) and {'side','distance','units'} <= set(s) for s in stacks):
            found.add(geometry_signature(stacks))
        # Some prior matched cohorts store geometry instead of stack records.
        geom = value.get('pair_geometry')
        if isinstance(geom, (list, tuple)) and len(geom) == 4 and all(isinstance(x, (int,float)) for x in geom):
            u1,d1,u2,d2 = map(int, geom)
            found.add(geometry_signature([dict(side='left',distance=d1,units=u1),dict(side='right',distance=d2,units=u2)]))
        if isinstance(geom, (list, tuple)) and len(geom)==2 and all(isinstance(x,(list,tuple)) and len(x)==2 for x in geom):
            (u1,d1),(u2,d2)=geom
            found.add(geometry_signature([dict(side='left',distance=int(d1),units=int(u1)),dict(side='right',distance=int(d2),units=int(u2))]))
        for child in value.values():
            found.update(geometries_in(child))
    elif isinstance(value, list):
        for child in value:
            found.update(geometries_in(child))
    return found


def physics(stacks):
    units = {s:sum(int(x['units']) for x in stacks if x['side']==s) for s in ('left','right')}
    dist = {s:sum(int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    torque = {s:sum(int(x['units'])*int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    out = lambda x:'left' if x>0 else 'right' if x<0 else 'balanced'
    net=torque['left']-torque['right']; w=units['left']-units['right']; d=dist['left']-dist['right']
    return dict(net_torque=net,outcome=out(net),weight_score=w,distance_score=d,
                normalized_margin=abs(net)/max(1,sum(torque.values())),
                left_units=units['left'],right_units=units['right'],left_torque=torque['left'],right_torque=torque['right'],
                strategy_predictions=dict(exact_torque=out(net),weight_only=out(w),distance_only=out(d)))


def exclusion_snapshot(project, followup_root, dataset_dir=None):
    lock=Path(followup_root)/'prior_geometry_snapshot.json'
    if lock.is_file():
        old=read_json(lock)
        for path, sha in old['files'].items():
            assert file_hash(Path(path))==sha, f'Prior input changed: {path}'
        return {tuple(tuple(x) for x in sig) for sig in old['signatures']}, old
    paths={Path(dataset_dir or project/'data/dataset_v1')/'scenes.jsonl'}
    for pattern in ('*manifest*.jsonl','*support*.jsonl','*geometries*.jsonl'):
        paths.update(p for p in (project/'results').rglob(pattern) if Path(followup_root) not in p.parents)
    excluded=set(); hashes={}
    for path in sorted(paths):
        assert path.is_file(), f'Missing required geometry input: {path}'
        excluded.update(geometries_in(read_jsonl(path)))
        hashes[str(path.resolve())]=file_hash(path)
    assert excluded, 'No prior physical geometry was indexed'
    old=dict(files=hashes,signatures=sorted(excluded),mirror_canonical=True)
    save_json(lock,old)
    return excluded,old


def inverse_families(n, seed, excluded):
    assert n%4==0
    pool=defaultdict(list)
    for dl in combinations(range(1,7),2):
        for dr in combinations(range(1,7),2):
            for units in product(range(1,6), repeat=4):
                stacks=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),dl+dr,units)]
                f=physics(stacks)
                pool[dl+dr].append((stacks,f))
    candidates=[]
    for distances,rows in pool.items():
        by_net=defaultdict(list)
        for stacks,f in rows:
            if f['outcome']==f['strategy_predictions']['weight_only'] and f['outcome']!='balanced':
                by_net[f['net_torque']].append((stacks,f))
        for stacks,f in rows:
            margin=f['normalized_margin']; difficulty='hard' if .08<=margin<.22 else 'moderate' if .22<=margin<.45 else None
            if (difficulty and f['outcome']!='balanced' and f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome'])
                    and f['strategy_predictions']['distance_only']==f['outcome'] and abs(f['distance_score'])>=2):
                sources=by_net.get(f['net_torque'],[])
                if sources:candidates.append((digest([seed,stacks]),difficulty,stacks,f,sources))
    candidates.sort(key=lambda r:r[0]); used=set(excluded); counts=Counter(); result=[]
    for _,difficulty,stacks,f,sources in candidates:
        stratum=(difficulty,f['outcome'])
        if counts[stratum]>=n//4 or geometry_signature(stacks) in used:continue
        valid=[(s,g) for s,g in sources if geometry_signature(s) not in used|{geometry_signature(stacks)}]
        if not valid:continue
        valid.sort(key=lambda x:(sum(abs(a['units']-b['units']) for a,b in zip(stacks,x[0])),digest([seed,x[0]])))
        src,sf=valid[0];fid='fixed_'+digest([seed,stacks,src])[:20]
        roles={name:dict(scene_id=fid+'__'+name,image_filename=fid+'__'+name+'.png',stacks=s,features=g)
               for name,s,g in [('base',stacks,f),('aligned_source',src,sf)]}
        result.append(dict(family_id=fid,remapping_split='new_fixed_confirmation',roles=roles,margin_bin=difficulty))
        used.update([geometry_signature(stacks),geometry_signature(src)]);counts[stratum]+=1
        if len(result)==n:break
    assert len(result)==n and set(counts.values())=={n//4}, (len(result),counts)
    return sorted(result,key=lambda r:r['family_id'])


def repair_families(n, seed, excluded):
    assert n%4==0
    candidates=defaultdict(list)
    for ul,dl,ur,dr in product(range(1,9),range(1,7),range(1,9),range(1,7)):
        s=[dict(side='left',units=ul,distance=dl),dict(side='right',units=ur,distance=dr)]
        f=physics(s)
        if f['net_torque']==0:continue
        if ul==ur and dl!=dr:kind='equal_weight_distance'
        elif f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome']):kind='weight_distance_conflict'
        elif f['strategy_predictions']['weight_only']==f['outcome'] and f['strategy_predictions']['distance_only']==f['outcome']:kind='aligned_cues'
        else:continue
        candidates[kind].append(s)
    for units in product(range(1,6),repeat=4):
        for distances in [((1,3),(2,5)),((2,5),(1,4)),((1,6),(3,5)),((2,4),(3,6))]:
            s=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),distances[0]+distances[1],units)]
            if physics(s)['net_torque']!=0:candidates['multi_stack_sum'].append(s)
    used=set(excluded); result=[]
    for kind in ('equal_weight_distance','weight_distance_conflict','multi_stack_sum','aligned_cues'):
        rows=sorted(candidates[kind],key=lambda s:digest([seed,kind,s])); count=0
        for stacks in rows:
            sig=geometry_signature(stacks)
            if sig in used:continue
            fid='repair_'+digest([seed,kind,sig])[:20];scenes=[]
            for mirror in (False,True):
                current=[dict(s,side=('right' if s['side']=='left' else 'left')) if mirror else dict(s) for s in stacks]
                for style in ('standard','alternate'):
                    sid=f'{fid}__m{int(mirror)}__{style}'
                    scenes.append(dict(scene_id=sid,image_filename=sid+'.png',stacks=current,features=physics(current),style=style,mirror=mirror))
            result.append(dict(family_id=fid,stratum=kind,scenes=scenes));used.add(sig);count+=1
            if count==n//4:break
        assert count==n//4,(kind,count,'Insufficient unused geometries; do not silently change task ranges')
    return sorted(result,key=lambda r:r['family_id'])


def intervention_conditions(model, sign, upstream, semantic, code, untreated, key, seeds=(0, 1, 2, 3)):
    """Condition topology is fixed before observing the recipient's answer."""
    if model == 'gemma12b':
        add = {31: dict(action='add', payload=sign * upstream)}
        return [('zero', -1, {}), ('targeted', -1, add)] + [
            ('random', s, {31: dict(action='matched_add', reference=sign * upstream, key=key + ['random', s])})
            for s in seeds]
    assert model == 'qwen27b' and set(untreated) >= {49, 55}
    u = {48: dict(action='add', payload=sign * upstream)}
    a = {49: dict(action='add', payload=sign * semantic)}
    b = {55: dict(action='add', payload=code)}
    blocks = {49: dict(action='clamp', untreated=untreated[49], axis=sign * semantic),
              55: dict(action='clamp', untreated=untreated[55], axis=code)}
    rows = [('zero', -1, {}), ('u', -1, u), ('a', -1, a), ('b', -1, b),
            ('bypass', -1, {**u, 49: blocks[49], **b})]
    for name, layer, payload in [('u', 48, sign * upstream), ('a', 49, sign * semantic)]:
        for s in seeds:
            rows.append((name + '_random', s, {layer: dict(action='matched_add', reference=payload, key=key + [name, s])}))
    for layer in (49, 55):
        rows += [(f'self_{layer}', -1, {layer: blocks[layer]}),
                 (f'self_random_{layer}', 0, {layer: dict(blocks[layer], action='random_coordinate', key=key + ['self', layer, 0])})]
    for name, start, layer in [('ua', u, 49), ('ub', u, 55), ('ab', a, 55)]:
        rows.append((name + '_block', -1, {**start, layer: blocks[layer]}))
        rows.append((name + '_coordinate', 0, {**start, layer: dict(blocks[layer], action='random_coordinate', key=key + [name, 'coordinate', 0])}))
        for s in seeds:
            rows.append((name + '_equal', s, {**start, layer: dict(blocks[layer], action='equal_remove', key=key + [name, 'equal', s])}))
    assert len(rows) == 35 and len({(n, s) for n, s, _ in rows}) == len(rows)
    return rows

def validate_trial_grid(frame, families, model):
    expected = [('zero', -1), ('targeted', -1)] + [('random', s) for s in range(4)] if model == 'gemma12b' else None
    if expected is None:
        mock = intervention_conditions('qwen27b', 1, np.ones(2), np.ones(2), np.ones(2), {49: np.zeros(2), 55: np.zeros(2)}, [])
        expected = [(n, s) for n, s, _ in mock]
    assert not frame.duplicated(['family_id', 'scene_id', 'variant', 'condition', 'random_seed']).any()
    assert set(frame.family_id) == {f['family_id'] for f in families}
    for family in families:
        sub = frame[frame.family_id.eq(family['family_id'])]
        assert set(sub.scene_id) == {s['scene_id'] for s in family['scenes']}
        assert len(sub) == 4 * 2 * len(expected)
        for _, unit in sub.groupby(['scene_id', 'variant']):
            assert set(zip(unit.condition, unit.random_seed)) == set(expected)
        assert set(sub.variant) == {'forward', 'reverse'}

def bool_value(value):
    if isinstance(value, (bool, np.bool_)):
        return bool(value)
    if isinstance(value, str) and value.lower() in ('true', 'false'):
        return value.lower() == 'true'
    if isinstance(value, (int, float, np.number)) and value in (0, 1):
        return bool(value)
    raise ValueError(f'Missing or invalid boolean: {value!r}')

def tracked_json(path, inputs):
    inputs[str(path)] = file_hash(path)
    return read_json(path)

def tracked_csv(path, inputs):
    inputs[str(path)] = file_hash(path)
    try:
        return pd.read_csv(path)
    except pd.errors.EmptyDataError:
        return pd.DataFrame()

def load_science_result(root, model, inputs):
    root = Path(root)
    paths = {'completion': root/'analysis/RUN_COMPLETE.json', 'protocol': root/'protocol.json',
             'capture': root/'capture_identity.json', 'support': root/'cohort/support.json',
             'records': root/'analysis/intervention_records.json', 'primary': root/'analysis/primary_tests.csv',
             'vectors': root/'frozen_directions.npz', 'reference': root/'reference.json',
             'images': root/'cohort/images.json'}
    missing = [str(p) for p in paths.values() if not p.is_file()]
    if missing:
        return dict(model=model, status='missing_inputs', issues=missing)
    result = tracked_json(paths['completion'], inputs)
    protocol = tracked_json(paths['protocol'], inputs)
    capture = tracked_json(paths['capture'], inputs)
    assert protocol['schema'] == ('qwen_three_path-1.0' if model == 'qwen27b' else 'gemma_l31_candidate-1.0')
    assert protocol['layers'] == ([48, 49, 55] if model == 'qwen27b' else [31])
    assert protocol['random_seeds'] == [0, 1, 2, 3] and protocol['norm_tolerance'] == .02
    for key in ('vectors', 'reference', 'images'):
        inputs[str(paths[key])] = file_hash(paths[key])
    assert result['status'] == 'complete' and result['model'] == protocol['model'] == model
    for key, path_key in [('protocol_sha256', 'protocol'), ('capture_sha256', 'capture'),
                          ('intervention_records_sha256', 'records')]:
        assert result[key] == file_hash(paths[path_key]), f'Completion hash mismatch: {paths[path_key]}'
    assert capture['protocol'] == protocol
    for key, path_key in [('support', 'support'), ('vectors', 'vectors'), ('reference', 'reference'), ('images', 'images')]:
        assert capture[key] == file_hash(paths[path_key]), f'Capture input changed: {paths[path_key]}'
    families = tracked_json(paths['support'], inputs)['families']
    records = tracked_json(paths['records'], inputs)
    primary = tracked_csv(paths['primary'], inputs)
    frame = pd.DataFrame([{k: v for k, v in r.items() if k not in ('record', 'clean_record')} for r in records])
    validate_trial_grid(frame, families, model)
    assert len(records) == result['condition_records'] and len(families) == result['families'] == protocol['n_families']
    assert len(primary) == protocol['primary_tests'] == (14 if model == 'qwen27b' else 4)
    assert primary.n.eq(len(families)).all() and primary.scope.eq('all_attempted').all()
    assert frame.valid.map(bool_value).all() == bool_value(result['all_controls_valid'])
    assert int(frame.groupby('family_id').valid.all().sum()) == result['fully_valid_families']
    optional = {}
    for name in ('technical_failures', 'valid_family_sensitivity', 'family_metrics', 'subgroup_descriptive',
                 'repair_and_harm_counts', 'secondary_bypass'):
        path = root/'analysis'/f'{name}.csv'
        if path.is_file():
            optional[name] = tracked_csv(path, inputs)
    return dict(model=model, status='validated_complete', issues=[], root=str(root), completion=result,
                protocol=protocol, capture=capture, families=families, records=records, frame=frame,
                primary=primary, optional=optional)

def verify_record_metrics(bundle):
    """Reconstruct baseline pairing and behavioral metrics from saved candidate scores."""
    rows = bundle['records']
    zeros = {(r['family_id'], r['scene_id'], r['variant']): r['record'] for r in rows if r['condition'] == 'zero'}
    for row in rows:
        clean = zeros[row['family_id'], row['scene_id'], row['variant']]
        assert digest(row['clean_record']) == digest(clean), 'Mixed current untreated baseline'
        out = row['record']
        assert clean['gold_semantic'] == out['gold_semantic'] == row['physical_outcome']
        correct = out['predicted_semantic'] == out['gold_semantic']
        baseline = clean['predicted_semantic'] == clean['gold_semantic']
        assert bool_value(row['correct']) == correct and bool_value(row['baseline_correct']) == baseline
        expected = {'accuracy_change': float(correct)-float(baseline),
                    'margin_change': correct_margin(out)-correct_margin(clean),
                    'repaired': not baseline and correct, 'harmed': baseline and not correct,
                    'answer_changed': out['predicted_semantic'] != clean['predicted_semantic']}
        for key, value in expected.items():
            assert abs(float(row[key])-float(value)) < 1e-10, f'Record metric mismatch: {key}'

def csv_markdown(frame, columns=None, limit=40):
    if frame.empty:
        return '无记录。'
    shown = frame if columns is None else frame[columns]
    shown = shown.head(limit)
    escape = lambda x: str(x).replace('|', '/').replace('\n', ' ')
    lines = ['| '+' | '.join(map(escape, shown.columns))+' |', '| '+' | '.join(['---']*len(shown.columns))+' |']
    for row in shown.itertuples(index=False, name=None):
        lines.append('| '+' | '.join(escape(f'{v:.6g}' if isinstance(v, float) else v) for v in row)+' |')
    if len(frame) > limit:
        lines += ['', f'此处显示前 {limit} 行，完整 {len(frame)} 行保存在 CSV。']
    return '\n'.join(lines)

"""Small registered-output packet; never include stale files from another run."""


class DiagnosticPacket:
    def __init__(self, root, identity):
        self.root = Path(root)
        self.root.mkdir(parents=True, exist_ok=True)
        freeze(self.root/'protocol.json', identity)
        self.outputs = {self.root/'protocol.json'}
        self.inputs = {}

    def track(self, path):
        path = Path(path)
        self.inputs[str(path)] = file_hash(path)
        return path

    def csv(self, name, frame):
        path = self.root/name
        frame.to_csv(path, index=False, encoding='utf-8-sig')
        self.outputs.add(path)
        return path

    def json(self, name, value):
        path = self.root/name
        save_json(path, value)
        self.outputs.add(path)
        return path

    def text(self, name, value):
        path = self.root/name
        path.write_text(value, encoding='utf-8')
        self.outputs.add(path)
        return path

    def finish(self, summary, report, archive_name, auto_download=True):
        import zipfile
        changed = [p for p, h in self.inputs.items() if not Path(p).is_file() or file_hash(p) != h]
        summary = dict(summary, source_files_unchanged=not changed, changed_inputs=changed)
        if changed:
            summary['status'] = 'inputs_changed_results_not_comparable'
        self.json('input_manifest.json', self.inputs)
        self.json('RUN_COMPLETE.json', summary)
        self.text('REPORT_ZH.md', report)
        archive = self.root/archive_name
        temporary = archive.with_suffix('.zip.tmp')
        with zipfile.ZipFile(temporary, 'w', compression=zipfile.ZIP_DEFLATED) as bundle:
            for path in sorted(self.outputs):
                bundle.write(path, str(path.relative_to(self.root)))
        temporary.replace(archive)
        print(json.dumps(summary, ensure_ascii=False, indent=2))
        print('完整结果包：', archive)
        if auto_download:
            try:
                from google.colab import files
                files.download(str(archive))
            except Exception as error:
                print('自动下载未触发，请从以上 Drive 路径手动下载：', type(error).__name__, str(error))
        return archive


def diagnostic_show(title, frame, limit=30):
    print('\n'+title)
    if frame.empty:
        print('无记录或此项输入缺失；请查看 coverage。')
        return
    with pd.option_context('display.max_rows', limit+2, 'display.max_columns', None, 'display.max_colwidth', 90):
        display(frame.head(limit))
    if len(frame) > limit:
        print(f'这里显示前 {limit} 行；结果 ZIP 保存全部 {len(frame)} 行。')


"""Saved-answer diagnostics. Embedded after the original CPU validation helpers."""


def record_semantics(record):
    if 'label_to_semantic' in record:
        mapping = {str(k): str(v) for k, v in record['label_to_semantic'].items()}
    else:
        labels, values = record['candidate_labels'], record['candidate_semantics']
        assert len(labels) == len(values) == 3
        mapping = dict(zip(map(str, labels), map(str, values)))
    assert set(mapping) == {'A', 'B', 'C'} and set(mapping.values()) == {'left', 'right', 'balanced'}
    scores = record['candidate_logprobs']
    labels = list(record.get('candidate_labels', mapping))
    scores = {k: float(scores[k]) for k in labels} if isinstance(scores, dict) else dict(zip(labels, map(float, scores)))
    assert set(scores) == set(mapping) and np.isfinite(list(scores.values())).all()
    pred_label = str(record.get('predicted_label', max(labels, key=scores.get)))
    assert pred_label in mapping and scores[pred_label] >= max(scores.values()) - 1e-12
    assert mapping[pred_label] == str(record['predicted_semantic'])
    if 'gold_label' in record:
        assert mapping[str(record['gold_label'])] == str(record['gold_semantic'])
    return mapping, pred_label


def swap_side(value):
    return {'left': 'right', 'right': 'left', 'balanced': 'balanced'}[value]


def stack_signature(stacks, mirror=False):
    return sorted((swap_side(s['side']) if mirror else s['side'], int(s['units']), int(s['distance'])) for s in stacks)


def behavior_rows(bundle):
    verify_record_metrics(bundle)
    lookup = {}
    for family in bundle['families']:
        assert len(family['scenes']) == 4
        for style in ('standard', 'alternate'):
            sub = {bool_value(s['mirror']): s for s in family['scenes'] if s['style'] == style}
            assert set(sub) == {False, True}
            assert stack_signature(sub[False]['stacks'], True) == stack_signature(sub[True]['stacks'])
        for scene in family['scenes']:
            assert scene['scene_id'] not in lookup
            features = physics(scene['stacks'])
            assert features == scene['features'], 'Saved geometry/features differ'
            lookup[scene['scene_id']] = (family, scene, features)
    out = []
    for row in bundle['records']:
        family, scene, features = lookup[row['scene_id']]
        record, clean = row['record'], row['clean_record']
        mapping, pred_label = record_semantics(record)
        clean_mapping, baseline_label = record_semantics(clean)
        assert mapping == clean_mapping
        assert family['family_id'] == row['family_id'] and family['stratum'] == row['stratum']
        assert bool_value(row['mirror']) == bool_value(scene['mirror']) and row['style'] == scene['style']
        pred, gold, base = str(record['predicted_semantic']), features['outcome'], str(clean['predicted_semantic'])
        assert gold == record['gold_semantic'] == clean['gold_semantic']
        assert int(record['net_torque']) == int(clean['net_torque']) == features['net_torque']
        w, d = features['strategy_predictions']['weight_only'], features['strategy_predictions']['distance_only']
        mistake = ('correct' if pred == gold else 'both_heuristics' if pred == w == d else
                   'weight_only_prediction' if pred == w else 'distance_only_prediction' if pred == d else
                   'other_balanced' if pred == 'balanced' else 'other_direction')
        out.append(dict(model=bundle['model'], family_id=row['family_id'], scene_id=row['scene_id'],
            stratum=row['stratum'], style=row['style'], mirror=bool_value(row['mirror']), variant=row['variant'],
            condition=row['condition'], random_seed=int(row['random_seed']), valid=bool_value(row['valid']),
            gold=gold, predicted=pred, baseline_predicted=base, predicted_label=pred_label, baseline_label=baseline_label,
            label_left=next(k for k, v in mapping.items() if v == 'left'),
            label_right=next(k for k, v in mapping.items() if v == 'right'),
            label_balanced=next(k for k, v in mapping.items() if v == 'balanced'),
            net_torque=features['net_torque'], left_units=features['left_units'], right_units=features['right_units'],
            left_torque=features['left_torque'], right_torque=features['right_torque'],
            weight_prediction=w, distance_prediction=d, correct=pred == gold, baseline_correct=base == gold,
            repaired=base != gold and pred == gold, harmed=base == gold and pred != gold,
            weight_disagrees_physics=w != gold, directional_weight_conflict=w != gold and w != 'balanced',
            follows_weight=pred == w, follows_distance=pred == d, mistake_category=mistake,
            margin_change=float(row['margin_change'])))
    return pd.DataFrame(out)


def family_ratio(frame, numerator, denominator, key, unit_columns=None, replicates=5000):
    """Seed means first; ratios of family totals with a family-cluster bootstrap.

    Report the pooled ratio and equally weighted within-family ratios separately.
    Undefined denominators stay missing; no confirmatory p values are produced.
    """
    unit_columns = unit_columns or ['family_id', 'scene_id', 'variant']
    x = frame[unit_columns].copy()
    x['num'] = np.asarray(numerator, float)
    x['den'] = np.asarray(denominator, float)
    assert np.isfinite(x[['num', 'den']]).all().all()
    assert (x.num >= 0).all() and (x.den >= x.num).all()
    units = x.groupby(unit_columns, sort=True)[['num', 'den']].mean()
    families = units.groupby(level='family_id').sum()
    total = float(families.den.sum())
    if total == 0:
        return dict(families=len(families), eligible_families=0, numerator_seed_average=0., denominator_seed_average=0.,
                    pooled_fraction=None, lower=None, upper=None, family_mean_fraction=None)
    rng = np.random.default_rng(int(digest(['error_structure', key])[:16], 16))
    counts = families[['num', 'den']].to_numpy(float)
    estimates = []
    for start in range(0, replicates, 250):
        sampled = counts[rng.integers(0, len(counts), (min(250, replicates-start), len(counts)))].sum(1)
        estimates.extend((sampled[:, 0][sampled[:, 1] > 0] / sampled[:, 1][sampled[:, 1] > 0]).tolist())
    return dict(families=len(families), eligible_families=int(families.den.gt(0).sum()),
        numerator_seed_average=float(families.num.sum()), denominator_seed_average=total,
        pooled_fraction=float(families.num.sum()/total), lower=float(np.quantile(estimates, .025)),
        upper=float(np.quantile(estimates, .975)),
        family_mean_fraction=float((families.num/families.den.replace(0, np.nan)).mean()))


def error_rate_tables(frame):
    rows, confusion = [], []
    for condition, block in frame.groupby('condition', sort=True):
        groups = [('all', 'all', block)]
        groups += [(column, str(value), sub) for column in ('stratum', 'gold', 'style', 'variant')
                   for value, sub in block.groupby(column, sort=True)]
        for group, value, sub in groups:
            all_rows = np.ones(len(sub), bool)
            specs = [('accuracy', sub.correct, all_rows), ('repair_of_original_errors', sub.repaired, ~sub.baseline_correct),
                ('harm_of_original_correct', sub.harmed, sub.baseline_correct),
                ('weight_choice_when_weight_disagrees', sub.follows_weight & sub.weight_disagrees_physics, sub.weight_disagrees_physics),
                ('weight_choice_on_directional_conflicts', sub.follows_weight & sub.directional_weight_conflict, sub.directional_weight_conflict),
                ('weight_agreement_among_errors', sub.follows_weight & ~sub.correct, ~sub.correct)]
            specs += [('predicted_'+s, sub.predicted.eq(s), all_rows) for s in ('left', 'right', 'balanced')]
            for metric, num, den in specs:
                rows.append(dict(model=sub.model.iloc[0], condition=condition, group=group, value=value, metric=metric,
                    **family_ratio(sub, num, den, [sub.model.iloc[0], condition, group, value, metric]),
                    interpretation='posthoc_descriptive_no_new_confirmation'))
        units = block.groupby(['family_id','scene_id','variant','gold','predicted','predicted_label','mistake_category']).size().rename('draws').reset_index()
        seeds = block.groupby(['family_id','scene_id','variant']).size().rename('seeds').reset_index()
        units = units.merge(seeds, on=['family_id','scene_id','variant'], validate='many_to_one')
        units['seed_average_units'] = units.draws/units.seeds
        for (gold, predicted, letter, category), sub in units.groupby(['gold','predicted','predicted_label','mistake_category']):
            confusion.append(dict(model=block.model.iloc[0], condition=condition, gold=gold, predicted=predicted,
                predicted_label=letter, mistake_category=category, seed_average_units=float(sub.seed_average_units.sum())))
    return pd.DataFrame(rows), pd.DataFrame(confusion)


def paired_behavior(frame, pairing):
    """Match the same family/render/mapping; never pair records by row order."""
    assert pairing in ('mirror', 'mapping')
    varying = 'mirror' if pairing == 'mirror' else 'variant'
    keep = ['model','family_id','style','condition','random_seed'] + (['variant'] if pairing == 'mirror' else ['scene_id','mirror'])
    rows = []
    for key, sub in frame.groupby(keep, sort=True):
        assert len(sub) == 2, (pairing, key, len(sub))
        ordered = sub.set_index(varying)
        assert set(ordered.index) == ({False, True} if pairing == 'mirror' else {'forward','reverse'})
        a, b = (ordered.loc[False], ordered.loc[True]) if pairing == 'mirror' else (ordered.loc['forward'], ordered.loc['reverse'])
        if pairing == 'mirror':
            assert b.gold == swap_side(a.gold)
            assert all(a[c] == b[c] for c in ('label_left','label_right','label_balanced'))
        else:
            assert a.gold == b.gold and a.label_balanced == b.label_balanced
            assert a.label_left == b.label_right and a.label_right == b.label_left
        rows.append(dict(zip(keep, key)) | dict(pairing=pairing, pair_id=digest([pairing, list(key)])[:24],
            scene_a=str(a.get('scene_id', key[-2] if pairing == 'mapping' else '')),
            gold_a=a.gold, gold_b=b.gold, prediction_a=a.predicted, prediction_b=b.predicted,
            label_a=a.predicted_label, label_b=b.predicted_label,
            semantic_consistent=b.predicted == (swap_side(a.predicted) if pairing == 'mirror' else a.predicted),
            both_correct=bool(a.correct and b.correct), both_wrong=bool(not a.correct and not b.correct),
            same_direction=bool(a.predicted == b.predicted and a.predicted != 'balanced'),
            same_letter=bool(a.predicted_label == b.predicted_label),
            same_letter_different_semantics=bool(a.predicted_label == b.predicted_label and a.predicted != b.predicted),
            original_error_count=int(not a.baseline_correct)+int(not b.baseline_correct),
            repaired_count=int(a.repaired)+int(b.repaired), harmed_count=int(a.harmed)+int(b.harmed)))
    return pd.DataFrame(rows)


def pair_summaries(pairs):
    rows = []
    for (model, pairing, condition), block in pairs.groupby(['model','pairing','condition'], sort=True):
        units = ['family_id','style','variant'] if pairing == 'mirror' else ['family_id','scene_id']
        for metric in ('semantic_consistent','both_correct','both_wrong','same_direction','same_letter','same_letter_different_semantics'):
            rows.append(dict(model=model, pairing=pairing, condition=condition, metric=metric,
                **family_ratio(block, block[metric], np.ones(len(block)), [model,pairing,condition,metric], units)))
    return pd.DataFrame(rows)


def paired_consistency_changes(pairs):
    rows = []
    for (model, pairing), block in pairs.groupby(['model','pairing'], sort=True):
        units = ['family_id','style','variant'] if pairing == 'mirror' else ['family_id','scene_id']
        for metric in ('semantic_consistent','both_correct'):
            means = block.groupby(units+['condition'])[metric].mean().reset_index()
            zero = means[means.condition.eq('zero')].drop(columns='condition').rename(columns={metric:'baseline'})
            for condition, sub in means[means.condition.ne('zero')].groupby('condition'):
                paired = sub.merge(zero, on=units, validate='one_to_one')
                assert len(paired) == len(zero)
                paired['change'] = paired[metric]-paired.baseline
                family = paired.groupby('family_id').change.mean().to_numpy(float)
                ci = bootstrap(family, ['paired_consistency',model,pairing,condition,metric], replicates=5000)
                rows.append(dict(model=model,pairing=pairing,condition=condition,metric=metric,
                    **{k:ci[k] for k in ('n','mean','lower','upper')},interpretation='paired_posthoc_difference_not_new_confirmation'))
    return pd.DataFrame(rows)
```

```python
%%cell_progress 4/8 Configure one CPU run for both completed models
RUN_ID='main'
SOURCE_RUN_ID='main'
AUTO_DOWNLOAD=True
BASE=PROJECT_ROOT/'results/cross_model_independent_confirmation'
OUTPUT_ROOT=BASE/'error_structure'/RUN_ID
packet=DiagnosticPacket(OUTPUT_ROOT,dict(schema='error-structure-1.0',implementation=IMPLEMENTATION_SHA256,
    source_run=SOURCE_RUN_ID,models=['qwen27b','gemma12b'],new_scientific_confirmation=False,
    bootstrap='5000 family-cluster draws; seed-average numerator/denominator; descriptive intervals only',
    changes_to_existing_inputs=False,gpu_forwards=0))
roots={'qwen27b':BASE/'qwen_three_path'/SOURCE_RUN_ID/'Qwen__Qwen3.5-27B__373612cb',
       'gemma12b':BASE/'gemma_l31_candidate'/SOURCE_RUN_ID/'google__gemma-4-12B-it'}
selected={'qwen27b':['zero','u','u_random','a','a_random','b','bypass'], 'gemma12b':['zero','targeted','random']}
bundles={};coverage=[];sections=[]
print('CPU 一次读取 Qwen-27B 和 Gemma，无需切换模型。输出：',OUTPUT_ROOT)
```

```python
%%cell_progress 5/8 Read actual answers and independently check the physical labels
for model,root in tqdm(roots.items(),desc='Reading saved answers'):
    try:
        bundle=load_science_result(root,model,packet.inputs)
        if bundle['status']=='validated_complete':
            answers=behavior_rows(bundle)
            packet.csv(model+'_all_saved_answer_details.csv',answers)
            packet.json(model+'_physical_families.json',bundle['families'])
            packet.json(model+'_original_completion.json',bundle['completion'])
            subset=answers[answers.condition.isin(selected[model])].copy()
            assert subset.valid.all(),'Selected repair conditions include an invalid technical control.'
            bundles[model]=(bundle,subset)
        coverage.append(dict(model=model,status=bundle['status'],issues='; '.join(bundle['issues'])))
    except (AssertionError,ValueError,KeyError,TypeError,OSError) as error:
        coverage.append(dict(model=model,status='input_integrity_error',issues=type(error).__name__+': '+str(error)))
packet.csv('coverage.csv',pd.DataFrame(coverage))
diagnostic_show('输入覆盖情况',pd.DataFrame(coverage))
```

```python
%%cell_progress 6/8 Count errors, heuristic choices and repairs with explicit denominators
for model,(bundle,answers) in bundles.items():
    rates,confusion=error_rate_tables(answers)
    packet.csv(model+'_error_rates_all_groups.csv',rates)
    packet.csv(model+'_answer_confusion_counts.csv',confusion)
    native=rates[(rates.condition=='zero') & rates['group'].isin(['stratum','gold']) &
        rates.metric.isin(['accuracy','weight_choice_on_directional_conflicts','weight_choice_when_weight_disagrees'])]
    shown=native[['group','value','metric','numerator_seed_average','denominator_seed_average','pooled_fraction','lower','upper']]
    diagnostic_show(model+'：原模型准确率与重量预测一致率（分母明确列出）',shown,40)
    repair=rates[(rates['group']=='gold') & rates.metric.isin(['repair_of_original_errors','harm_of_original_correct'])]
    diagnostic_show(model+'：左右各自的修复率与新增错误率',repair[['condition','value','metric','numerator_seed_average','denominator_seed_average','pooled_fraction']],35)
    diagnostic_show(model+'：原模型错误实际选向哪里',confusion[confusion.condition=='zero'],30)
    sections.append('## '+model+'：错误类型与左右表现\n\n'+csv_markdown(shown))
```

```python
%%cell_progress 7/8 Compare mirrored scenes and option-letter swaps within each family
for model,(bundle,answers) in bundles.items():
    mirror=paired_behavior(answers,'mirror')
    mapping=paired_behavior(answers,'mapping')
    pair_table=pd.concat([mirror,mapping],ignore_index=True)
    summary=pair_summaries(pair_table)
    changes=paired_consistency_changes(pair_table)
    packet.csv(model+'_mirror_pair_details.csv',mirror)
    packet.csv(model+'_mapping_pair_details.csv',mapping)
    packet.csv(model+'_pair_consistency_summary.csv',summary)
    packet.csv(model+'_paired_consistency_change_vs_zero.csv',changes)
    important=summary[summary.metric.isin(['semantic_consistent','both_correct','same_letter_different_semantics'])]
    diagnostic_show(model+'：镜像应交换左右；换选项字母应保持物理答案',important[['pairing','condition','metric','numerator_seed_average','denominator_seed_average','pooled_fraction','lower','upper']],40)
    sections.append('## '+model+'：镜像和答案映射\n\n'+csv_markdown(important))
    diagnostic_show(model+'：相对未修改答案的一致性变化（同一家族内配对）',changes,25)
```
