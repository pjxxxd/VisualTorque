The previous audit found that confidence across different physical families was associated with stronger downstream route attenuation and specificity, but it did not predict the total weight effect on unseen families.

```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_TRACKED_CELLS = 14
_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print(f'Cell progress helper ready: {_TRACKED_CELLS} tracked cells.')
```

```python
%%cell_progress 3/14 Lock the within-context analysis
cell_stage('Freezing units, predictors, outcomes, and decision gates')

SEED = 20261114
BOOTSTRAP_REPLICATES = 10000
ROBUSTNESS_BOOTSTRAP_REPLICATES = 4000
ALPHA = 0.05
MINIMUM_FAMILIES = 20
MINIMUM_DISCORDANT_BUNDLES = 12

PRIMARY_METRIC = 'route_specificity'
CONFIRMATORY_METRIC = 'route_attenuation'
BOUNDARY_METRIC = 'shared_over_random'
OUTCOME_METRICS = (
    PRIMARY_METRIC, CONFIRMATORY_METRIC, BOUNDARY_METRIC,
)
PRIMARY_COMPETENCE_RAW = 'min_endpoint_gold_margin'
SECONDARY_COMPETENCE_RAW = 'log_transition_norm'
FIXED_COLUMNS = ('side_code', 'mapping_code', 'side_mapping_code')

PREVIOUS_RUN_PATH = PREVIOUS_ROOT / 'RUN_COMPLETE.json'
PREVIOUS_PROTOCOL_PATH = PREVIOUS_ROOT / 'frozen_analysis_protocol.json'
PREVIOUS_DECISION_PATH = (
    PREVIOUS_ROOT / 'analysis' / 'competence_gating_decision.json'
)
INPUT_ESTIMAND_PATH = (
    PREVIOUS_ROOT / 'analysis' / 'target_level_competence_estimands.csv'
)
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_withincontext_protocol.json'

def json_default(value):
    if isinstance(value, np.generic):
        return value.item()
    if isinstance(value, Path):
        return str(value)
    raise TypeError(type(value).__name__)

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False,
                   default=json_default) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return 'sha256:' + digest.hexdigest()

def stable_seed(value):
    text = f'{SEED}:{value!r}'.encode('utf-8')
    return int(hashlib.sha256(text).hexdigest()[:16], 16) % (2**63 - 1)

for required in (
    PREVIOUS_RUN_PATH, PREVIOUS_PROTOCOL_PATH,
    PREVIOUS_DECISION_PATH, INPUT_ESTIMAND_PATH,
):
    assert required.is_file(), f'Missing prerequisite: {required}'

input_hashes = {
    'previous_run': sha256_file(PREVIOUS_RUN_PATH),
    'previous_protocol': sha256_file(PREVIOUS_PROTOCOL_PATH),
    'previous_decision': sha256_file(PREVIOUS_DECISION_PATH),
    'target_estimands': sha256_file(INPUT_ESTIMAND_PATH),
}
protocol_identity = {
    'schema_version': 'weight-route-withincontext-gate-audit-1.0',
    'analysis_only': True,
    'input_hashes': input_hashes,
    'bundle': 'family_id x target_key; four paired side-by-mapping cells',
    'transformation': (
        'outcomes and continuous predictors centered within each bundle; '
        'side and mapping use orthogonal effect coding'
    ),
    'primary_outcome': PRIMARY_METRIC,
    'confirmatory_outcome': CONFIRMATORY_METRIC,
    'boundary_outcome': BOUNDARY_METRIC,
    'primary_predictor': PRIMARY_COMPETENCE_RAW,
    'secondary_predictor': SECONDARY_COMPETENCE_RAW,
    'strata': 'natural edit changes versus preserves the gold outcome',
    'matched_check': (
        'within-bundle both-endpoints-correct versus not-both-correct cells'
    ),
    'heldout_unit': 'complete physical family',
    'bootstrap_unit': 'complete physical family',
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'seed': SEED,
}
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous['identity'] == protocol_identity, (
        'Existing protocol differs; use a new versioned output directory.'
    )
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

print('Primary outcome:', PRIMARY_METRIC)
print('Bundle:', protocol_identity['bundle'])
```

```python
%%cell_progress 4/14 Load and validate the paired target-level data
cell_stage('Checking complete four-cell bundles')

previous_run = json.loads(PREVIOUS_RUN_PATH.read_text(encoding='utf-8'))
previous_decision = json.loads(
    PREVIOUS_DECISION_PATH.read_text(encoding='utf-8')
)
frame = pd.read_csv(INPUT_ESTIMAND_PATH)
required_columns = {
    'family_id', 'target_key', 'orientation', 'changed_side',
    'mapping_variant', 'clean_correct', 'natural_correct',
    'gold_outcome_changed', 'min_endpoint_gold_margin',
    'mean_endpoint_gold_margin', 'clean_gold_margin',
    'natural_gold_margin', 'log_transition_norm',
    'route_specificity', 'route_attenuation', 'shared_over_random',
}
assert required_columns.issubset(frame.columns), sorted(
    required_columns - set(frame.columns)
)
for column in (
    'clean_correct', 'natural_correct', 'gold_outcome_changed'
):
    if frame[column].dtype != bool:
        frame[column] = frame[column].astype(str).str.lower().map({
            'true': True, 'false': False,
        })
    assert frame[column].notna().all(), column

frame['bundle_id'] = (
    frame['family_id'].astype(str) + '::' + frame['target_key'].astype(str)
)
assert frame['family_id'].nunique() >= MINIMUM_FAMILIES
assert frame.groupby('bundle_id').size().eq(4).all()
assert frame.groupby('bundle_id')['family_id'].nunique().eq(1).all()
assert frame.groupby('bundle_id')['gold_outcome_changed'].nunique().eq(1).all()
assert set(zip(frame['changed_side'], frame['mapping_variant'])) == {
    ('left', 'forward'), ('left', 'reverse'),
    ('right', 'forward'), ('right', 'reverse'),
}
assert np.isfinite(frame[list(OUTCOME_METRICS)].to_numpy(dtype=float)).all()

print('Previous decision:', previous_decision['primary_decision'])
print('Families / bundles / rows:',
      frame['family_id'].nunique(), frame['bundle_id'].nunique(), len(frame))
```

## 2. Remove physical-family and target-context difficulty


```python
%%cell_progress 5/14 Construct within-bundle outcomes and predictors
cell_stage('Centering every measure inside the same four-cell bundle')

frame['side_code'] = np.where(frame['changed_side'] == 'right', 0.5, -0.5)
frame['mapping_code'] = np.where(
    frame['mapping_variant'] == 'reverse', 0.5, -0.5
)
frame['side_mapping_code'] = frame['side_code'] * frame['mapping_code']
frame['both_endpoints_correct'] = (
    frame['clean_correct'] & frame['natural_correct']
)

continuous_predictors = (
    'min_endpoint_gold_margin', 'mean_endpoint_gold_margin',
    'clean_gold_margin', 'natural_gold_margin', 'log_transition_norm',
)
for column in continuous_predictors:
    within = frame[column] - frame.groupby('bundle_id')[column].transform('mean')
    scale = float(within.std(ddof=0))
    assert scale > 0, column
    frame[column + '_within_raw'] = within
    frame[column + '_within_z'] = within / scale

for metric in OUTCOME_METRICS:
    frame[metric + '_within'] = (
        frame[metric]
        - frame.groupby('bundle_id')[metric].transform('mean')
    )
    assert np.allclose(
        frame.groupby('bundle_id')[metric + '_within'].mean().to_numpy(),
        0.0, atol=1e-12,
    )

PRIMARY_COMPETENCE = PRIMARY_COMPETENCE_RAW + '_within_z'
SECONDARY_COMPETENCE = SECONDARY_COMPETENCE_RAW + '_within_z'
WITHIN_OUTCOMES = tuple(metric + '_within' for metric in OUTCOME_METRICS)

frame.to_csv(ANALYSIS_DIR / 'withincontext_estimands.csv', index=False)
print('Primary within-context predictor:', PRIMARY_COMPETENCE)
print('Primary within-context outcome:', PRIMARY_METRIC + '_within')
```

```python
%%cell_progress 6/14 Audit outcome-flip and discordant matched support
cell_stage('Counting informative comparisons before fitting models')

bundle_summary = frame.groupby('bundle_id').agg(
    family_id=('family_id', 'first'),
    gold_outcome_changed=('gold_outcome_changed', 'first'),
    correct_cells=('both_endpoints_correct', 'sum'),
    margin_range=('min_endpoint_gold_margin',
                  lambda values: float(values.max() - values.min())),
).reset_index()
bundle_summary['discordant_correctness'] = bundle_summary[
    'correct_cells'
].between(1, 3)

support = bundle_summary.groupby('gold_outcome_changed').agg(
    bundles=('bundle_id', 'size'),
    families=('family_id', 'nunique'),
    discordant_bundles=('discordant_correctness', 'sum'),
    mean_margin_range=('margin_range', 'mean'),
).reset_index()
support['stratum'] = support['gold_outcome_changed'].map({
    False: 'outcome preserved', True: 'outcome changed',
})
assert bundle_summary['discordant_correctness'].sum() >= MINIMUM_DISCORDANT_BUNDLES
assert (support['discordant_bundles'] > 0).all(), support.to_dict(orient='records')
assert (bundle_summary['margin_range'] > 0).all()

support.to_csv(ANALYSIS_DIR / 'withincontext_support.csv', index=False)
bundle_summary.to_csv(ANALYSIS_DIR / 'bundle_support.csv', index=False)
display(support[[
    'stratum', 'bundles', 'families',
    'discordant_bundles', 'mean_margin_range',
]].round(4))
print('Total discordant bundles:', int(bundle_summary['discordant_correctness'].sum()))
```

```python
%%cell_progress 7/14 Describe paired correctness differences
cell_stage('Computing descriptive within-bundle contrasts')

matched_rows = []
for bundle_id, current in frame.groupby('bundle_id'):
    correct = current[current['both_endpoints_correct']]
    other = current[~current['both_endpoints_correct']]
    if correct.empty or other.empty:
        continue
    row = {
        'bundle_id': bundle_id,
        'family_id': current['family_id'].iloc[0],
        'gold_outcome_changed': bool(current['gold_outcome_changed'].iloc[0]),
        'correct_cells': int(len(correct)),
        'other_cells': int(len(other)),
        'margin_difference': float(
            correct['min_endpoint_gold_margin'].mean()
            - other['min_endpoint_gold_margin'].mean()
        ),
    }
    for metric in OUTCOME_METRICS:
        row[metric + '_correct_minus_other'] = float(
            correct[metric].mean() - other[metric].mean()
        )
    matched_rows.append(row)

matched_frame = pd.DataFrame(matched_rows)
assert len(matched_frame) >= MINIMUM_DISCORDANT_BUNDLES
matched_frame.to_csv(
    ANALYSIS_DIR / 'discordant_bundle_differences.csv', index=False
)
descriptive = matched_frame.groupby('gold_outcome_changed').agg(
    bundles=('bundle_id', 'size'),
    families=('family_id', 'nunique'),
    margin_difference=('margin_difference', 'mean'),
    route_specificity=(
        'route_specificity_correct_minus_other', 'mean'
    ),
    route_attenuation=(
        'route_attenuation_correct_minus_other', 'mean'
    ),
    shared_over_random=(
        'shared_over_random_correct_minus_other', 'mean'
    ),
).reset_index()
descriptive['stratum'] = descriptive['gold_outcome_changed'].map({
    False: 'outcome preserved', True: 'outcome changed',
})
display(descriptive.round(5))
```

## 3. Within-context inference and prediction


```python
%%cell_progress 8/14 Estimate within-context competence slopes
cell_stage('Bootstrapping complete physical families')

MODEL_COLUMNS = {
    'factorial': FIXED_COLUMNS,
    'within_competence': FIXED_COLUMNS + (
        PRIMARY_COMPETENCE, SECONDARY_COMPETENCE,
    ),
}

def fit_no_intercept(current, outcome, columns):
    matrix = current[list(columns)].to_numpy(dtype=float)
    values = current[outcome].to_numpy(dtype=float)
    coefficients = np.linalg.lstsq(matrix, values, rcond=None)[0]
    return dict(zip(columns, coefficients))

def clustered_coefficient_draws(current, outcome, columns, key, replicates,
                                family_order=None, draw_counts=None):
    columns = tuple(columns)
    if family_order is None:
        family_order = current['family_id'].drop_duplicates().to_numpy()
    else:
        family_order = np.asarray(family_order)
    xtx_by_family = []
    xty_by_family = []
    present = []
    for family in family_order:
        subset = current[current['family_id'] == family]
        if subset.empty:
            xtx_by_family.append(np.zeros((len(columns), len(columns))))
            xty_by_family.append(np.zeros(len(columns)))
            present.append(False)
            continue
        matrix = subset[list(columns)].to_numpy(dtype=float)
        values = subset[outcome].to_numpy(dtype=float)
        xtx_by_family.append(matrix.T @ matrix)
        xty_by_family.append(matrix.T @ values)
        present.append(True)
    xtx_by_family = np.asarray(xtx_by_family, dtype=float)
    xty_by_family = np.asarray(xty_by_family, dtype=float)
    present = np.asarray(present, dtype=bool)
    if draw_counts is None:
        generator = np.random.default_rng(stable_seed(key))
        draw_counts = generator.multinomial(
            len(family_order),
            np.full(len(family_order), 1 / len(family_order)),
            size=replicates,
        )
    xtx = np.einsum('rf,fij->rij', draw_counts, xtx_by_family)
    xty = np.einsum('rf,fj->rj', draw_counts, xty_by_family)
    inverse = np.linalg.pinv(xtx, rcond=1e-12)
    draws = np.einsum('rij,rj->ri', inverse, xty)
    valid = (draw_counts[:, present].sum(axis=1) > 0)
    return list(columns), draws, draw_counts, valid

def clustered_summary(current, outcome, model_name, key, replicates):
    columns = MODEL_COLUMNS[model_name]
    point = fit_no_intercept(current, outcome, columns)
    families = current['family_id'].drop_duplicates().to_numpy()
    names, draw_matrix, _, valid = clustered_coefficient_draws(
        current, outcome, columns, key, replicates,
    )
    assert valid.mean() > 0.99
    rows = []
    for column_index, name in enumerate(names):
        estimate = point[name]
        values = draw_matrix[valid, column_index]
        rows.append({
            'analysis': 'all_bundles', 'metric': outcome,
            'model': model_name, 'term': name,
            'estimate': float(estimate),
            'ci_low': float(np.quantile(values, ALPHA / 2)),
            'ci_high': float(np.quantile(values, 1 - ALPHA / 2)),
            'n_families': int(len(families)),
            'n_bundles': int(current['bundle_id'].nunique()),
        })
    return rows

coefficient_rows = []
for metric in OUTCOME_METRICS:
    outcome = metric + '_within'
    for model_name in MODEL_COLUMNS:
        coefficient_rows.extend(clustered_summary(
            frame, outcome, model_name,
            ('all', metric, model_name), BOOTSTRAP_REPLICATES,
        ))
coefficient_frame = pd.DataFrame(coefficient_rows)
coefficient_frame.to_csv(
    ANALYSIS_DIR / 'withincontext_clustered_coefficients.csv', index=False
)
display(coefficient_frame[
    (coefficient_frame['model'] == 'within_competence')
    & (coefficient_frame['term'].isin((
        PRIMARY_COMPETENCE, SECONDARY_COMPETENCE,
    )))
].round(6))
```

```python
%%cell_progress 9/14 Localize slopes to outcome-changing edits
cell_stage('Separating edits that require versus preserve the answer')

stratum_rows = []
for changed, current in frame.groupby('gold_outcome_changed'):
    label = 'outcome_changed' if changed else 'outcome_preserved'
    for metric in OUTCOME_METRICS:
        outcome = metric + '_within'
        rows = clustered_summary(
            current, outcome, 'within_competence',
            ('stratum', label, metric), BOOTSTRAP_REPLICATES,
        )
        for row in rows:
            row['analysis'] = label
        stratum_rows.extend(rows)
stratum_frame = pd.DataFrame(stratum_rows)
stratum_frame.to_csv(
    ANALYSIS_DIR / 'outcome_flip_stratified_coefficients.csv', index=False
)

primary_strata = stratum_frame[
    (stratum_frame['term'] == PRIMARY_COMPETENCE)
    & (stratum_frame['metric'].isin(tuple(
        metric + '_within' for metric in OUTCOME_METRICS
    )))
]
display(primary_strata.round(6))

# Directly test whether the competence slope differs between edits that change
# versus preserve the correct answer. Separate confidence intervals alone do
# not test this difference.
stratum_contrast_rows = []
all_families = frame['family_id'].drop_duplicates().to_numpy()
for metric in OUTCOME_METRICS:
    outcome = metric + '_within'
    changed = frame[frame['gold_outcome_changed']]
    preserved = frame[~frame['gold_outcome_changed']]
    point_changed = fit_no_intercept(
        changed, outcome, MODEL_COLUMNS['within_competence']
    )[PRIMARY_COMPETENCE]
    point_preserved = fit_no_intercept(
        preserved, outcome, MODEL_COLUMNS['within_competence']
    )[PRIMARY_COMPETENCE]
    estimate = float(point_changed - point_preserved)
    generator = np.random.default_rng(stable_seed((
        'flip-minus-preserved', metric
    )))
    draw_counts = generator.multinomial(
        len(all_families),
        np.full(len(all_families), 1 / len(all_families)),
        size=BOOTSTRAP_REPLICATES,
    )
    names, changed_draws, _, changed_valid = clustered_coefficient_draws(
        changed, outcome, MODEL_COLUMNS['within_competence'],
        ('paired-changed', metric), BOOTSTRAP_REPLICATES,
        family_order=all_families, draw_counts=draw_counts,
    )
    _, preserved_draws, _, preserved_valid = clustered_coefficient_draws(
        preserved, outcome, MODEL_COLUMNS['within_competence'],
        ('paired-preserved', metric), BOOTSTRAP_REPLICATES,
        family_order=all_families, draw_counts=draw_counts,
    )
    competence_index = names.index(PRIMARY_COMPETENCE)
    valid = changed_valid & preserved_valid
    draws = (
        changed_draws[valid, competence_index]
        - preserved_draws[valid, competence_index]
    )
    assert len(draws) >= int(0.99 * BOOTSTRAP_REPLICATES)
    stratum_contrast_rows.append({
        'metric': metric,
        'contrast': 'outcome_changed_minus_preserved_competence_slope',
        'estimate': estimate,
        'ci_low': float(np.quantile(draws, ALPHA / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
        'valid_bootstrap_draws': int(len(draws)),
    })
stratum_contrast_frame = pd.DataFrame(stratum_contrast_rows)
stratum_contrast_frame.to_csv(
    ANALYSIS_DIR / 'outcome_flip_slope_contrasts.csv', index=False
)
display(stratum_contrast_frame.round(6))
```

```python
%%cell_progress 10/14 Test discordant within-bundle correctness contrasts
cell_stage('Bootstrapping matched correct-versus-other differences by family')

matched_test_rows = []
for stratum_name, current in [
    ('all', matched_frame),
    ('outcome_changed', matched_frame[matched_frame['gold_outcome_changed']]),
    ('outcome_preserved', matched_frame[~matched_frame['gold_outcome_changed']]),
]:
    if current.empty:
        continue
    families = current['family_id'].drop_duplicates().to_numpy()
    for metric in OUTCOME_METRICS:
        column = metric + '_correct_minus_other'
        estimate = float(current[column].mean())
        generator = np.random.default_rng(stable_seed((
            'matched', stratum_name, metric
        )))
        family_statistics = current.groupby('family_id')[column].agg(
            ['sum', 'count']
        ).reindex(families)
        draw_counts = generator.multinomial(
            len(families), np.full(len(families), 1 / len(families)),
            size=BOOTSTRAP_REPLICATES,
        )
        numerators = draw_counts @ family_statistics['sum'].to_numpy(dtype=float)
        denominators = draw_counts @ family_statistics['count'].to_numpy(dtype=float)
        draws = numerators / denominators
        matched_test_rows.append({
            'stratum': stratum_name, 'metric': metric,
            'estimate': estimate,
            'ci_low': float(np.quantile(draws, ALPHA / 2)),
            'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
            'n_families': int(len(families)),
            'n_bundles': int(len(current)),
        })

matched_test_frame = pd.DataFrame(matched_test_rows)
matched_test_frame.to_csv(
    ANALYSIS_DIR / 'discordant_correctness_tests.csv', index=False
)
display(matched_test_frame.round(6))
```

```python
%%cell_progress 11/14 Predict within-bundle patterns in held-out families
cell_stage('Leaving out every physical family in turn')

PREDICTION_MODELS = {
    'bundle_mean': (),
    'factorial': FIXED_COLUMNS,
    'within_competence': FIXED_COLUMNS + (
        PRIMARY_COMPETENCE_RAW + '_within_raw',
        SECONDARY_COMPETENCE_RAW + '_within_raw',
    ),
}
CONTINUOUS_RAW = {
    PRIMARY_COMPETENCE_RAW + '_within_raw',
    SECONDARY_COMPETENCE_RAW + '_within_raw',
}

def train_test_matrices(train, test, model_name):
    columns = PREDICTION_MODELS[model_name]
    if not columns:
        return None, None
    train_columns = []
    test_columns = []
    for column in columns:
        train_values = train[column].to_numpy(dtype=float)
        test_values = test[column].to_numpy(dtype=float)
        if column in CONTINUOUS_RAW:
            scale = max(float(train_values.std(ddof=0)), 1e-12)
            train_values = train_values / scale
            test_values = test_values / scale
        train_columns.append(train_values)
        test_columns.append(test_values)
    return np.column_stack(train_columns), np.column_stack(test_columns)

prediction_rows = []
families = frame['family_id'].drop_duplicates().tolist()
for metric in OUTCOME_METRICS:
    outcome = metric + '_within'
    for held_family in tqdm(
        families, desc=f'Within-family holdout: {metric}', unit='family'
    ):
        train = frame[frame['family_id'] != held_family]
        test = frame[frame['family_id'] == held_family]
        observed = test[outcome].to_numpy(dtype=float)
        for model_name in PREDICTION_MODELS:
            train_matrix, test_matrix = train_test_matrices(
                train, test, model_name
            )
            if model_name == 'bundle_mean':
                predicted = np.zeros(len(test), dtype=float)
            else:
                coefficients = np.linalg.lstsq(
                    train_matrix, train[outcome].to_numpy(dtype=float),
                    rcond=None,
                )[0]
                predicted = test_matrix @ coefficients
            for row_index, prediction, observation in zip(
                test.index, predicted, observed
            ):
                prediction_rows.append({
                    'metric': metric, 'model': model_name,
                    'family_id': held_family,
                    'bundle_id': test.loc[row_index, 'bundle_id'],
                    'gold_outcome_changed': bool(
                        test.loc[row_index, 'gold_outcome_changed']
                    ),
                    'row_index': int(row_index),
                    'observed_within': float(observation),
                    'predicted_within': float(prediction),
                    'squared_error': float((observation - prediction) ** 2),
                    'absolute_error': float(abs(observation - prediction)),
                })

prediction_frame = pd.DataFrame(prediction_rows)
prediction_frame.to_csv(
    ANALYSIS_DIR / 'heldout_family_within_predictions.csv', index=False
)
comparison_rows = []
gain_rows = []
for stratum_name, subset in [
    ('all', prediction_frame),
    ('outcome_changed', prediction_frame[prediction_frame['gold_outcome_changed']]),
    ('outcome_preserved', prediction_frame[~prediction_frame['gold_outcome_changed']]),
]:
    for (metric, model_name), current in subset.groupby(['metric', 'model']):
        observed = current['observed_within'].to_numpy(dtype=float)
        predicted = current['predicted_within'].to_numpy(dtype=float)
        denominator = float((observed ** 2).sum())
        comparison_rows.append({
            'stratum': stratum_name, 'metric': metric, 'model': model_name,
            'rmse': float(np.sqrt(np.mean((observed - predicted) ** 2))),
            'mae': float(np.mean(np.abs(observed - predicted))),
            'within_r2': float(
                1 - ((observed - predicted) ** 2).sum() / denominator
            ),
            'correlation': float(np.corrcoef(observed, predicted)[0, 1])
            if float(np.std(predicted)) > 0 else float('nan'),
        })

    family_error = subset.groupby([
        'metric', 'model', 'family_id'
    ])['squared_error'].mean().reset_index()
    for metric in OUTCOME_METRICS:
        current = family_error[family_error['metric'] == metric].pivot(
            index='family_id', columns='model', values='squared_error'
        )
        values = (
            current['factorial'] - current['within_competence']
        ).to_numpy(dtype=float)
        generator = np.random.default_rng(stable_seed((
            'prediction-gain', stratum_name, metric
        )))
        indices = generator.integers(
            0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
        )
        draws = values[indices].mean(axis=1)
        gain_rows.append({
            'stratum': stratum_name, 'metric': metric,
            'mse_reduction_over_factorial': float(values.mean()),
            'ci_low': float(np.quantile(draws, ALPHA / 2)),
            'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
            'families_improved_fraction': float((values > 0).mean()),
            'n_families': int(len(values)),
        })

comparison_frame = pd.DataFrame(comparison_rows)
gain_frame = pd.DataFrame(gain_rows)
comparison_frame.to_csv(
    ANALYSIS_DIR / 'within_prediction_model_comparison.csv', index=False
)
gain_frame.to_csv(
    ANALYSIS_DIR / 'within_prediction_gain.csv', index=False
)
display(comparison_frame[comparison_frame['stratum'] == 'all'].round(6))
display(gain_frame.round(6))
```

## 4. Robustness, visualization, and decision


```python
%%cell_progress 12/14 Run within-context influence and predictor robustness
cell_stage('Checking whether one family or one margin definition drives results')

ALTERNATIVE_PREDICTORS = (
    'clean_gold_margin_within_z',
    'natural_gold_margin_within_z',
    'min_endpoint_gold_margin_within_z',
    'mean_endpoint_gold_margin_within_z',
)

def single_predictor_slope(current, metric, predictor):
    columns = FIXED_COLUMNS + (predictor,)
    return fit_no_intercept(
        current, metric + '_within', columns
    )[predictor]

robustness_rows = []
families_array = frame['family_id'].drop_duplicates().to_numpy()
for metric in OUTCOME_METRICS:
    for predictor in ALTERNATIVE_PREDICTORS:
        estimate = single_predictor_slope(frame, metric, predictor)
        columns = FIXED_COLUMNS + (predictor,)
        names, draw_matrix, _, valid = clustered_coefficient_draws(
            frame, metric + '_within', columns,
            ('robustness', metric, predictor),
            ROBUSTNESS_BOOTSTRAP_REPLICATES,
        )
        draws = draw_matrix[valid, names.index(predictor)]
        robustness_rows.append({
            'metric': metric, 'predictor': predictor,
            'estimate': float(estimate),
            'ci_low': float(np.quantile(draws, ALPHA / 2)),
            'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
        })

influence_rows = []
for metric in OUTCOME_METRICS:
    for family in families_array:
        current = frame[frame['family_id'] != family]
        influence_rows.append({
            'metric': metric, 'omitted_family': family,
            'slope': single_predictor_slope(
                current, metric, PRIMARY_COMPETENCE
            ),
        })
robustness_frame = pd.DataFrame(robustness_rows)
influence_frame = pd.DataFrame(influence_rows)
robustness_frame.to_csv(
    ANALYSIS_DIR / 'within_predictor_robustness.csv', index=False
)
influence_frame.to_csv(
    ANALYSIS_DIR / 'within_leave_one_family_out_influence.csv', index=False
)
display(robustness_frame.round(6))
display(influence_frame.groupby('metric').agg(
    minimum=('slope', 'min'), median=('slope', 'median'),
    maximum=('slope', 'max'),
    positive_fraction=('slope', lambda values: (values > 0).mean()),
).round(6))
```

```python
%%cell_progress 13/14 Save the within-context paper figure and decision
cell_stage('Rendering matched, stratified, and predictive results')

def coefficient_lookup(source, analysis, metric, term):
    current = source[
        (source['analysis'] == analysis)
        & (source['metric'] == metric + '_within')
        & (source['term'] == term)
    ]
    assert len(current) == 1, (analysis, metric, term, len(current))
    return current.iloc[0]

primary_row = coefficient_lookup(
    coefficient_frame, 'all_bundles', PRIMARY_METRIC, PRIMARY_COMPETENCE
)
confirmatory_row = coefficient_lookup(
    coefficient_frame, 'all_bundles', CONFIRMATORY_METRIC, PRIMARY_COMPETENCE
)
boundary_row = coefficient_lookup(
    coefficient_frame, 'all_bundles', BOUNDARY_METRIC, PRIMARY_COMPETENCE
)
matched_primary = matched_test_frame[
    (matched_test_frame['stratum'] == 'all')
    & (matched_test_frame['metric'] == PRIMARY_METRIC)
].iloc[0]
predictive_primary = gain_frame[
    (gain_frame['stratum'] == 'all')
    & (gain_frame['metric'] == PRIMARY_METRIC)
].iloc[0]

gates = {
    'within_route_specificity_slope_positive': bool(primary_row['ci_low'] > 0),
    'within_route_attenuation_slope_positive': bool(
        confirmatory_row['ci_low'] > 0
    ),
    'boundary_total_effect_not_required': True,
    'discordant_correctness_difference_positive': bool(
        matched_primary['ci_low'] > 0
    ),
    'heldout_within_pattern_gain_positive': bool(
        predictive_primary['ci_low'] > 0
    ),
}
if (
    gates['within_route_specificity_slope_positive']
    and gates['heldout_within_pattern_gain_positive']
    and (
        gates['within_route_attenuation_slope_positive']
        or gates['discordant_correctness_difference_positive']
    )
):
    PRIMARY_DECISION = 'withincontext_route_gating_supported'
elif (
    gates['within_route_specificity_slope_positive']
    or gates['discordant_correctness_difference_positive']
):
    PRIMARY_DECISION = 'withincontext_route_gating_partially_supported'
else:
    PRIMARY_DECISION = 'betweenfamily_difficulty_explains_prior_gate_association'

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
labels = ['Outcome preserved', 'Outcome changed']
matched_plot = matched_test_frame[
    matched_test_frame['stratum'].isin((
        'outcome_preserved', 'outcome_changed'
    ))
    & (matched_test_frame['metric'] == PRIMARY_METRIC)
].set_index('stratum').loc[['outcome_preserved', 'outcome_changed']]
x = np.arange(2)
y = matched_plot['estimate'].to_numpy()
yerr = np.vstack([
    y - matched_plot['ci_low'].to_numpy(),
    matched_plot['ci_high'].to_numpy() - y,
])
axes[0, 0].errorbar(x, y, yerr=yerr, fmt='o', capsize=5,
                    color='#2563EB', ecolor='#64748B')
axes[0, 0].axhline(0, color='black', linewidth=0.8)
axes[0, 0].set_xticks(x, labels)
axes[0, 0].set_ylabel('Route specificity: correct minus other cell')
axes[0, 0].set_title('A. Discordant cells of the same physical bundle')

colors = {False: '#D97706', True: '#2563EB'}
for changed, current in frame.groupby('gold_outcome_changed'):
    label = 'Outcome changed' if changed else 'Outcome preserved'
    axes[0, 1].scatter(
        current[PRIMARY_COMPETENCE],
        current[PRIMARY_METRIC + '_within'],
        s=20, alpha=0.40, color=colors[bool(changed)], label=label,
    )
    x_values = current[PRIMARY_COMPETENCE].to_numpy(dtype=float)
    y_values = current[PRIMARY_METRIC + '_within'].to_numpy(dtype=float)
    slope, intercept = np.polyfit(x_values, y_values, 1)
    grid = np.linspace(x_values.min(), x_values.max(), 100)
    axes[0, 1].plot(grid, intercept + slope * grid,
                    color=colors[bool(changed)], linewidth=2)
axes[0, 1].axhline(0, color='black', linewidth=0.8)
axes[0, 1].set_xlabel('Within-bundle endpoint confidence')
axes[0, 1].set_ylabel('Within-bundle route specificity')
axes[0, 1].set_title('B. Same-context competence association')
axes[0, 1].legend(frameon=False)

rows = [primary_row, confirmatory_row, boundary_row]
names = ['Route specificity', 'Route attenuation', 'Total weight effect']
x = np.arange(3)
y = np.asarray([row['estimate'] for row in rows])
yerr = np.vstack([
    y - np.asarray([row['ci_low'] for row in rows]),
    np.asarray([row['ci_high'] for row in rows]) - y,
])
axes[1, 0].errorbar(x, y, yerr=yerr, fmt='o', capsize=5,
                    color='#7C3AED', ecolor='#64748B')
axes[1, 0].axhline(0, color='black', linewidth=0.8)
axes[1, 0].set_xticks(x, names, rotation=10)
axes[1, 0].set_ylabel('Within-bundle slope per 1 SD')
axes[1, 0].set_title('C. Family-clustered within-context slopes')

primary_comparison = comparison_frame[
    (comparison_frame['stratum'] == 'all')
    & (comparison_frame['metric'] == PRIMARY_METRIC)
].set_index('model').loc[
    ['bundle_mean', 'factorial', 'within_competence']
]
axes[1, 1].bar(
    np.arange(3), primary_comparison['rmse'],
    color=['#9CA3AF', '#D97706', '#2563EB'],
)
axes[1, 1].set_xticks(
    np.arange(3), ['Bundle mean', 'Side + mapping', '+ within competence']
)
axes[1, 1].set_ylabel('Held-out-family within-pattern RMSE')
axes[1, 1].set_title('D. Prediction of unseen physical families')

fig.suptitle('Does route organization track competence within the same scene?',
             fontsize=16)
fig.tight_layout(rect=(0, 0, 1, 0.96))
figure_path = FIGURE_DIR / 'withincontext_route_gating.png'
fig.savefig(figure_path, dpi=220, bbox_inches='tight')
fig.savefig(figure_path.with_suffix('.pdf'), bbox_inches='tight')
plt.show()

decision = {
    'schema_version': 'weight-route-withincontext-gate-decision-1.0',
    'primary_decision': PRIMARY_DECISION,
    'gates': gates,
    'primary_within_slope': primary_row.to_dict(),
    'confirmatory_within_slope': confirmatory_row.to_dict(),
    'boundary_within_slope': boundary_row.to_dict(),
    'matched_primary_difference': matched_primary.to_dict(),
    'heldout_predictive_gain': predictive_primary.to_dict(),
    'claim_boundary': (
        'A positive result means route organization covaries with untouched '
        'endpoint competence across orientation and answer mappings of the '
        'same physical bundle, and generalizes as a within-bundle pattern to '
        'held-out families. It does not make competence a causal mechanism. '
        'A null result means the earlier cross-family association was likely '
        'scene difficulty rather than a within-scene gate.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'withincontext_gate_decision.json', decision)
print('Primary decision:', PRIMARY_DECISION)
print('Gates:', gates)
print('Saved figure:', figure_path)
```
