```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Build the graded model-blind factorial manifest
cell_stage('Enumerating weight-by-distance and distance-by-weight dose families')
import itertools
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}

def single_shape(scene):
    if len(scene['stacks']) != 2:
        return None
    left = [stack for stack in scene['stacks'] if stack['side'] == 'left']
    right = [stack for stack in scene['stacks'] if stack['side'] == 'right']
    if len(left) != 1 or len(right) != 1:
        return None
    return (
        int(left[0]['units']), int(left[0]['distance']),
        int(right[0]['units']), int(right[0]['distance']),
    )

scene_by_shape = {}
for scene in scene_records:
    shape = single_shape(scene)
    if shape is not None:
        assert shape not in scene_by_shape
        scene_by_shape[shape] = str(scene['scene_id'])
assert len(scene_by_shape) == 900, len(scene_by_shape)

prior_rows = read_jsonl(PRIOR_MANIFEST_PATH)
excluded_scene_ids = {
    str(role['scene_id'])
    for row in prior_rows for role in row['roles'].values()
}
assert len(excluded_scene_ids) >= 1

all_prior_scene_ids = set(excluded_scene_ids)
results_root = PROJECT_ROOT / 'results'
if results_root.is_dir():
    prior_paths = sorted(
        path for path in results_root.rglob('*manifest*.jsonl')
        if SHARED_ROOT not in path.parents
    )
    for path in tqdm(prior_paths, desc='Auditing historical scene support', unit='file'):
        try:
            rows = read_jsonl(path)
        except Exception:
            continue
        for row in rows:
            for key, value in row.items():
                if key.endswith('_scene_id') and isinstance(value, (str, int)):
                    all_prior_scene_ids.add(str(value))
                elif key.endswith('_scene_ids') and isinstance(value, (list, tuple)):
                    all_prior_scene_ids.update(str(item) for item in value)
            if isinstance(row.get('roles'), dict):
                for role in row['roles'].values():
                    if isinstance(role, dict) and 'scene_id' in role:
                        all_prior_scene_ids.add(str(role['scene_id']))

def opposite_side(side):
    return 'right' if side == 'left' else 'left'

def shape_for(side, units, distance, fixed_units, fixed_distance):
    if side == 'left':
        return (units, distance, fixed_units, fixed_distance)
    return (fixed_units, fixed_distance, units, distance)

def role_record(side, units, distance, fixed_units, fixed_distance):
    shape = shape_for(side, units, distance, fixed_units, fixed_distance)
    scene_id = scene_by_shape.get(shape)
    if scene_id is None:
        return None
    scene = scene_by_id[scene_id]
    return {
        'scene_id': scene_id,
        'units': int(units),
        'distance': int(distance),
        'side_torque': int(scene['derived'][f'{side}_torque']),
        'net_torque': int(scene['derived']['net_torque']),
        'outcome': str(scene['derived']['outcome']),
    }

def graded_valid(side, roles):
    base_outcome = opposite_side(side)
    if any(value is None for value in roles.values()):
        return False
    if any(
        roles[name]['outcome'] != base_outcome
        for name in ('anchor', 'source', 'target_b1', 'target_b2', 'target_b3')
    ):
        return False
    hybrid_outcomes = {
        roles[f'hybrid_{background}']['outcome'] for background in BACKGROUNDS
    }
    return base_outcome in hybrid_outcomes and side in hybrid_outcomes

if MANIFEST_PATH.is_file() and MANIFEST_LOCK_PATH.is_file():
    factor_manifest = read_jsonl(MANIFEST_PATH)
    lock = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert lock['identity']['selected_count'] == len(factor_manifest)
    candidates = None
    cell_stage('Using the existing locked graded manifest')
else:
    candidates = []
    for side in ('left', 'right'):
        for fixed_units in range(1, 6):
            for fixed_distance in range(1, 7):
                for w0, w1 in itertools.combinations(range(1, 6), 2):
                    delta = w1 - w0
                    for d0 in range(1, 7):
                        for contexts in itertools.combinations(
                            [value for value in range(1, 7) if value != d0], 3
                        ):
                            values = {
                                'anchor': (w0, d0), 'source': (w1, d0),
                            }
                            for background, distance in zip(BACKGROUNDS, contexts):
                                values[f'target_{background}'] = (w0, distance)
                                values[f'hybrid_{background}'] = (w1, distance)
                            roles = {
                                name: role_record(
                                    side, units, distance,
                                    fixed_units, fixed_distance,
                                )
                                for name, (units, distance) in values.items()
                            }
                            if not graded_valid(side, roles):
                                continue
                            doses = {
                                background: int(delta * distance)
                                for background, distance in zip(BACKGROUNDS, contexts)
                            }
                            family_id = 'graded_weight_' + stable_key((
                                side, fixed_units, fixed_distance,
                                w0, w1, d0, contexts,
                            ))[:18]
                            candidates.append({
                                'family_id': family_id,
                                'factor': 'weight',
                                'changed_side': side,
                                'fixed_opposite_units': fixed_units,
                                'fixed_opposite_distance': fixed_distance,
                                'factor_base': w0,
                                'factor_source': w1,
                                'factor_delta': delta,
                                'source_context': d0,
                                'background_contexts': dict(zip(BACKGROUNDS, contexts)),
                                'background_doses': doses,
                                'roles': roles,
                            })

                for d0, d1 in itertools.combinations(range(1, 7), 2):
                    delta = d1 - d0
                    for w0 in range(1, 6):
                        for contexts in itertools.combinations(
                            [value for value in range(1, 6) if value != w0], 3
                        ):
                            values = {
                                'anchor': (w0, d0), 'source': (w0, d1),
                            }
                            for background, units in zip(BACKGROUNDS, contexts):
                                values[f'target_{background}'] = (units, d0)
                                values[f'hybrid_{background}'] = (units, d1)
                            roles = {
                                name: role_record(
                                    side, units, distance,
                                    fixed_units, fixed_distance,
                                )
                                for name, (units, distance) in values.items()
                            }
                            if not graded_valid(side, roles):
                                continue
                            doses = {
                                background: int(units * delta)
                                for background, units in zip(BACKGROUNDS, contexts)
                            }
                            family_id = 'graded_distance_' + stable_key((
                                side, fixed_units, fixed_distance,
                                d0, d1, w0, contexts,
                            ))[:18]
                            candidates.append({
                                'family_id': family_id,
                                'factor': 'distance',
                                'changed_side': side,
                                'fixed_opposite_units': fixed_units,
                                'fixed_opposite_distance': fixed_distance,
                                'factor_base': d0,
                                'factor_source': d1,
                                'factor_delta': delta,
                                'source_context': w0,
                                'background_contexts': dict(zip(BACKGROUNDS, contexts)),
                                'background_doses': doses,
                                'roles': roles,
                            })

    assert candidates
    for row in candidates:
        row['immediate_prior_overlap_count'] = sum(
            role['scene_id'] in excluded_scene_ids
            for role in row['roles'].values()
        )
        row['historical_overlap_count'] = sum(
            role['scene_id'] in all_prior_scene_ids
            for role in row['roles'].values()
        )
        row['hybrid_pattern'] = '|'.join(
            row['roles'][f'hybrid_{background}']['outcome']
            for background in BACKGROUNDS
        )

    def select_factor_support(factor, forbidden, attempt, strict_exclusion):
        groups = defaultdict(list)
        for row in candidates:
            if row['factor'] != factor:
                continue
            scene_ids = {role['scene_id'] for role in row['roles'].values()}
            if strict_exclusion and scene_ids & excluded_scene_ids:
                continue
            groups[(row['changed_side'], row['factor_delta'])].append(row)
        for key, rows in groups.items():
            rows.sort(key=lambda row: (
                row['immediate_prior_overlap_count'],
                row['historical_overlap_count'],
                stable_key((attempt, key, row['family_id'])),
            ))
        group_keys = sorted(
            groups, key=lambda key: stable_key((attempt, 'group', key))
        )
        cursors = {key: 0 for key in group_keys}
        chosen, used = [], set(forbidden)
        while len(chosen) < FAMILIES_PER_FACTOR:
            changed = False
            for key in group_keys:
                rows = groups[key]
                while cursors[key] < len(rows):
                    row = rows[cursors[key]]
                    cursors[key] += 1
                    scene_ids = {role['scene_id'] for role in row['roles'].values()}
                    if scene_ids & used:
                        continue
                    chosen.append(row)
                    used.update(scene_ids)
                    changed = True
                    break
                if len(chosen) == FAMILIES_PER_FACTOR:
                    break
            if not changed:
                break
        return chosen

    selected_by_factor = None
    selection_policy = None
    selected_attempt = None
    for attempt in range(512):
        order = FACTORS if attempt % 2 else tuple(reversed(FACTORS))
        current, globally_used = {}, set()
        for factor in order:
            chosen = select_factor_support(
                factor, globally_used, attempt, strict_exclusion=True
            )
            if len(chosen) != FAMILIES_PER_FACTOR:
                break
            current[factor] = chosen
            globally_used.update(
                role['scene_id']
                for row in chosen for role in row['roles'].values()
            )
        if set(current) == set(FACTORS):
            selected_by_factor = current
            selection_policy = 'global_disjoint_immediate_prior_excluded'
            selected_attempt = attempt
            break

    if selected_by_factor is None:
        current = {}
        for factor in FACTORS:
            best = []
            for attempt in range(512):
                chosen = select_factor_support(
                    factor, set(), 1000 + attempt, strict_exclusion=True
                )
                if len(chosen) > len(best):
                    best = chosen
                if len(chosen) == FAMILIES_PER_FACTOR:
                    break
            if len(best) != FAMILIES_PER_FACTOR:
                current = None
                break
            current[factor] = best
        if current is not None:
            selected_by_factor = current
            selection_policy = 'within_factor_disjoint_immediate_prior_excluded'
            selected_attempt = None

    if selected_by_factor is None:
        selected_by_factor = {}
        for factor in FACTORS:
            best = []
            for attempt in range(512):
                chosen = select_factor_support(
                    factor, set(), 100000 + attempt, strict_exclusion=False
                )
                if len(chosen) > len(best):
                    best = chosen
                if len(chosen) == FAMILIES_PER_FACTOR:
                    break
            assert len(best) == FAMILIES_PER_FACTOR, (
                factor, len(best), Counter(row['factor'] for row in candidates)
            )
            selected_by_factor[factor] = best
        selection_policy = 'within_factor_disjoint_prior_minimized_fallback'
        selected_attempt = None

    selected = []
    for factor in FACTORS:
        ordered = sorted(
            selected_by_factor[factor], key=lambda row: stable_key(row['family_id'])
        )
        for index, original in enumerate(ordered):
            row = dict(original)
            row['counterbalance_phase'] = index % 2
            selected.append(row)
    factor_manifest = sorted(selected, key=lambda row: row['family_id'])

    factor_scene_lists = {
        factor: [
            role['scene_id']
            for row in factor_manifest if row['factor'] == factor
            for role in row['roles'].values()
        ]
        for factor in FACTORS
    }
    assert all(
        len(values) == len(set(values))
        for values in factor_scene_lists.values()
    )
    cross_factor_overlap = set.intersection(*(
        set(values) for values in factor_scene_lists.values()
    ))
    immediate_overlap = {
        scene_id for values in factor_scene_lists.values() for scene_id in values
        if scene_id in excluded_scene_ids
    }
    temporary = MANIFEST_PATH.with_suffix(MANIFEST_PATH.suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in factor_manifest:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(MANIFEST_PATH)
    manifest_identity = {
        'schema_version': 'graded-product-manifest-1.0',
        'selected_count': len(factor_manifest),
        'counts_by_factor': dict(Counter(
            row['factor'] for row in factor_manifest
        )),
        'selection_policy': selection_policy,
        'selection_attempt': selected_attempt,
        'within_factor_scene_disjoint': True,
        'global_scene_disjoint': not cross_factor_overlap,
        'cross_factor_scene_overlap_count': len(cross_factor_overlap),
        'immediate_prior_scene_overlap_count': len(immediate_overlap),
        'selection_uses_model_outputs': False,
        'backgrounds_per_family': len(BACKGROUNDS),
        'families_span_no_flip_and_flip': True,
        'family_ids': [row['family_id'] for row in factor_manifest],
        'seed': SEED,
    }
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

assert len(factor_manifest) == EXPECTED_FAMILIES
assert Counter(row['factor'] for row in factor_manifest) == {
    factor: FAMILIES_PER_FACTOR for factor in FACTORS
}
for row in factor_manifest:
    doses = [row['background_doses'][background] for background in BACKGROUNDS]
    assert len(set(doses)) == len(BACKGROUNDS)
    assert all(value > 0 for value in doses)
    assert graded_valid(row['changed_side'], row['roles'])

manifest_frame = pd.DataFrame([{
    'family_id': row['family_id'],
    'factor': row['factor'],
    'changed_side': row['changed_side'],
    'factor_delta': row['factor_delta'],
    'source_context': row['source_context'],
    'hybrid_pattern': row['hybrid_pattern'],
    'immediate_prior_overlap_count': row['immediate_prior_overlap_count'],
    'historical_overlap_count': row['historical_overlap_count'],
    'counterbalance_phase': row['counterbalance_phase'],
} for row in factor_manifest])
manifest_frame.to_csv(ANALYSIS_DIR / 'manifest_summary.csv', index=False)
display(manifest_frame.groupby([
    'factor', 'changed_side', 'factor_delta', 'hybrid_pattern'
]).size().rename('n').reset_index())
print('Candidate / selected families:',
      len(candidates) if candidates is not None else 'locked',
      len(factor_manifest))
print('Manifest policy:',
      json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))['identity'])

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=SHARED_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {}
for example in outcome_examples:
    outcome_by_scene.setdefault(str(example.scene_id), example)
selected_scene_ids = {
    str(role['scene_id'])
    for row in factor_manifest for role in row['roles'].values()
}
missing = selected_scene_ids - set(outcome_by_scene)
assert not missing, sorted(missing)[:5]
print('Selected unique scenes:', len(selected_scene_ids))
```

```python
%%cell_progress 8/17 Define counterbalanced prompts and graded token masks
cell_stage('Defining label-swapped outcome bundles')
import math
from dataclasses import replace as dataclass_replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def mapping_for(row, background_index):
    variant = MAPPING_VARIANTS[
        (int(row['counterbalance_phase']) + background_index) % 2
    ]
    semantics = (
        ('left', 'balanced', 'right')
        if variant == 'forward'
        else ('right', 'balanced', 'left')
    )
    return variant, tuple(zip(('A', 'B', 'C'), semantics))

def custom_prompt(mapping):
    options = '\n'.join(
        f'{label}. {SEMANTIC_TEXT[semantic]}'
        for label, semantic in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT,
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, role_name, mapping, background, mapping_variant):
    role = row['roles'][role_name]
    scene_id = str(role['scene_id'])
    semantics = tuple(value for _, value in mapping)
    labels = tuple(label for label, _ in mapping)
    outcome = str(role['outcome'])
    base = outcome_by_scene[scene_id]
    return dataclass_replace(
        base,
        example_id=(
            f"decision_neutral:{row['family_id']}:{background}:"
            f"{mapping_variant}:{role_name}"
        ),
        question_id=(
            f"decision_neutral_{row['family_id']}_{background}_"
            f"{mapping_variant}"
        ),
        question_kind='outcome',
        variant='decision_neutral_factor_semantic',
        prompt=custom_prompt(mapping),
        labels=labels,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=labels[semantics.index(outcome)],
        gold_semantic=outcome,
    )

def make_bundle(row, background_index):
    background = BACKGROUNDS[background_index]
    target_role = f'target_{background}'
    hybrid_role = f'hybrid_{background}'
    mapping_variant, mapping = mapping_for(row, background_index)
    role_names = ('anchor', 'source', target_role, hybrid_role)
    examples = {
        role: make_example(
            row, role, mapping, background, mapping_variant
        )
        for role in role_names
    }
    target_semantic = row['roles'][target_role]['outcome']
    hybrid_semantic = row['roles'][hybrid_role]['outcome']
    assert target_semantic == row['roles']['source']['outcome']
    assert target_semantic != hybrid_semantic
    return {
        'background': background,
        'mapping_variant': mapping_variant,
        'mapping': mapping,
        'examples': examples,
        'anchor_role': 'anchor',
        'source_role': 'source',
        'target_role': target_role,
        'hybrid_role': hybrid_role,
        'target_semantic': target_semantic,
        'hybrid_semantic': hybrid_semantic,
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key]), (
                name, key
            )
        masks = discover_masks(current)
        for scope in (UPSTREAM_SCOPE, SEMANTIC_SCOPE):
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Counterbalanced prompts and token regions ready.')

# Override the binary-flip bundle with a graded, continuously scored bundle.
def make_bundle(row, background_index):
    background = BACKGROUNDS[background_index]
    target_role = f'target_{background}'
    hybrid_role = f'hybrid_{background}'
    mapping_variant, mapping = mapping_for(row, background_index)
    role_names = ('anchor', 'source', target_role, hybrid_role)
    examples = {
        role: make_example(
            row, role, mapping, background, mapping_variant
        )
        for role in role_names
    }
    positive = str(row['changed_side'])
    negative = opposite_side(positive)
    assert row['roles']['anchor']['outcome'] == negative
    assert row['roles']['source']['outcome'] == negative
    assert row['roles'][target_role]['outcome'] == negative
    return {
        'background': background,
        'mapping_variant': mapping_variant,
        'mapping': mapping,
        'examples': examples,
        'anchor_role': 'anchor',
        'source_role': 'source',
        'target_role': target_role,
        'hybrid_role': hybrid_role,
        'positive_semantic': positive,
        'negative_semantic': negative,
        'target_semantic': row['roles'][target_role]['outcome'],
        'hybrid_semantic': row['roles'][hybrid_role]['outcome'],
        'physical_dose': float(row['background_doses'][background]),
        'context_value': float(row['background_contexts'][background]),
        'source_context': float(row['source_context']),
        'factor_delta': float(row['factor_delta']),
    }

print('Graded prompt bundles ready.')
```

```python
%%cell_progress 9/17 Define frozen-route capture and restoration helpers
cell_stage('Defining exact clean restoration and equal-norm random controls')
import math

def site_key(layer_index, component):
    return f'L{int(layer_index)}:{component}'

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def score_records(prepared, logits):
    records = []
    for batch_index, example in enumerate(prepared['examples']):
        logprobs = torch.log_softmax(
            logits[batch_index, -1, :].float(), dim=-1
        )
        scores = [
            float(logprobs[prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': [value / total for value in weights],
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def semantic_margin(record, positive, negative):
    mapping = dict(zip(
        record['candidate_semantics'], record['candidate_logprobs']
    ))
    return float(mapping[positive] - mapping[negative])

def centered_error(first, second):
    first = np.asarray(first, dtype=float)
    second = np.asarray(second, dtype=float)
    return float(np.max(np.abs(
        (first - first.mean()) - (second - second.mean())
    )))

def forward_capture_components(prepared, upstream_payload=None, upstream_mode=None):
    assert len(prepared['examples']) == 1
    masks = discover_masks(prepared)
    upstream_mask = masks[UPSTREAM_SCOPE]
    captured = {}
    handles = []
    if upstream_payload is not None:
        def upstream_hook(_module, _inputs, output):
            hidden = output_hidden_tensor(output)
            selected = upstream_mask.to(hidden.device)
            payload = upstream_payload.to(hidden.device, hidden.dtype)
            modified = hidden.clone()
            if upstream_mode == 'add':
                modified[0, selected] = hidden[0, selected] + payload[selected]
            elif upstream_mode == 'replace':
                modified[0, selected] = payload[selected]
            else:
                raise ValueError(upstream_mode)
            return replace_output_hidden(output, modified)
        handles.append(
            DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
        )

    def capture_l48(_module, _inputs, output):
        captured['L48:decoder'] = (
            output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
        )
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(capture_l48)
    )
    for (layer_index, component), module in COMPONENT_MODULES.items():
        key = site_key(layer_index, component)
        def capture(_module, _inputs, output, key=key):
            captured[key] = (
                output_hidden_tensor(output)[0].detach().cpu().to(torch.bfloat16)
            )
        handles.append(module.register_forward_hook(capture))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    assert len(records) == 1
    assert len(captured) == 1 + len(COMPONENT_MODULES)
    return records[0], captured, masks

def forward_score_upstream(example, upstream_payload, upstream_mode):
    prepared = prepare_examples([example])
    masks = discover_masks(prepared)
    selected = masks[UPSTREAM_SCOPE]
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = selected.to(hidden.device)
        payload = upstream_payload.to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        if upstream_mode == 'add':
            modified[0, current] = hidden[0, current] + payload[current]
        elif upstream_mode == 'replace':
            modified[0, current] = payload[current]
        else:
            raise ValueError(upstream_mode)
        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    return score_records(prepared, output.logits)[0]

def forward_score(example):
    prepared = prepare_examples([example])
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    return score_records(prepared, output.logits)[0]

def masked_removal(additive_state, clean_state, mask):
    removal = torch.zeros_like(additive_state, dtype=torch.bfloat16)
    selected = mask.bool()
    removal[selected] = (
        additive_state[selected].float() - clean_state[selected].float()
    ).to(torch.bfloat16)
    return removal

def random_removal_like(removal, mask, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.zeros_like(removal, dtype=torch.float32)
    selected = mask.bool()
    random_values = torch.randn(
        removal[selected].shape, generator=generator, dtype=torch.float32
    )
    target_norm = torch.linalg.vector_norm(removal[selected].float())
    random_norm = torch.linalg.vector_norm(random_values)
    if float(target_norm) > 0:
        random_values = random_values * (target_norm / random_norm)
    else:
        random_values.zero_()
    noise[selected] = random_values
    return noise.to(torch.bfloat16)

def score_restoration_conditions(example, upstream_delta, conditions):
    assert 1 <= len(conditions) <= SITE_BATCH_SIZE
    prepared = prepare_examples([example] * len(conditions))
    single = prepare_examples([example])
    masks = discover_masks(single)
    upstream_mask = masks[UPSTREAM_SCOPE]
    handles = []
    measurements = [
        {'restored_norm': 0.0, 'random_norm': 0.0, 'restored_sites': 0}
        for _ in conditions
    ]

    def upstream_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        selected = upstream_mask.to(hidden.device)
        payload = upstream_delta.to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        for batch_index in range(len(conditions)):
            modified[batch_index, selected] = (
                hidden[batch_index, selected] + payload[selected]
            )
        return replace_output_hidden(output, modified)
    handles.append(
        DECODER_LAYERS[UPSTREAM_LAYER].register_forward_hook(upstream_hook)
    )

    union_keys = sorted({
        key for condition in conditions for key in condition['removals']
    })
    for key in union_keys:
        layer_text, component = key.split(':')
        layer_index = int(layer_text[1:])
        module = COMPONENT_MODULES[(layer_index, component)]
        def restore_hook(_module, _inputs, output, key=key):
            hidden = output_hidden_tensor(output)
            modified = hidden.clone()
            changed = False
            for batch_index, condition in enumerate(conditions):
                if key not in condition['removals']:
                    continue
                removal = condition['removals'][key].to(
                    hidden.device, hidden.dtype
                )
                modified[batch_index] = hidden[batch_index] - removal
                norm = float(torch.linalg.vector_norm(
                    removal.float()
                ).detach().cpu())
                measurements[batch_index]['restored_sites'] += 1
                if condition.get('random_control', False):
                    measurements[batch_index]['random_norm'] += norm ** 2
                else:
                    measurements[batch_index]['restored_norm'] += norm ** 2
                changed = True
            return replace_output_hidden(output, modified) if changed else None
        handles.append(module.register_forward_hook(restore_hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, measurement in zip(records, conditions, measurements):
        measurement['restored_norm'] = measurement['restored_norm'] ** 0.5
        measurement['random_norm'] = measurement['random_norm'] ** 0.5
        record.update({'condition': condition['condition'], **measurement})
    return records

def chunks(values, size):
    for start in range(0, len(values), size):
        yield values[start:start + size]

print('Component capture and batched restoration helpers ready.')
```

```python
%%cell_progress 10/17 Run physics, alignment, and route-parity smoke tests
cell_stage('Checking graded doses and one full intervention path')

smoke_row = factor_manifest[0]
smoke_doses = [
    smoke_row['background_doses'][background] for background in BACKGROUNDS
]
assert len(set(smoke_doses)) == len(BACKGROUNDS)
assert min(smoke_doses) > 0
smoke_hybrid_outcomes = {
    smoke_row['roles'][f'hybrid_{background}']['outcome']
    for background in BACKGROUNDS
}
assert opposite_side(smoke_row['changed_side']) in smoke_hybrid_outcomes
assert smoke_row['changed_side'] in smoke_hybrid_outcomes

smoke_bundle = make_bundle(smoke_row, 0)
role_names = (
    smoke_bundle['anchor_role'], smoke_bundle['source_role'],
    smoke_bundle['target_role'],
)
smoke_prepared = {
    name: prepare_examples([smoke_bundle['examples'][name]])
    for name in role_names
}
smoke_masks = assert_aligned(smoke_prepared)
smoke_scores, smoke_hidden = {}, {}
for name in role_names:
    smoke_scores[name], smoke_hidden[name], _ = forward_capture_components(
        smoke_prepared[name]
    )
anchor, source, target = role_names
upstream_delta = torch.zeros_like(smoke_hidden[source]['L48:decoder'])
upstream_mask = smoke_masks[UPSTREAM_SCOPE]
upstream_delta[upstream_mask] = (
    smoke_hidden[source]['L48:decoder'][upstream_mask].float()
    - smoke_hidden[anchor]['L48:decoder'][upstream_mask].float()
).to(torch.bfloat16)
additive_score, additive_hidden, _ = forward_capture_components(
    smoke_prepared[target], upstream_delta, 'add'
)

selected_removals, random_removals = {}, {}
for rank, site in enumerate(FROZEN_SITES[smoke_row['factor']], start=1):
    key = site['site_key']
    mask = smoke_masks[site['scope']]
    removal = masked_removal(
        additive_hidden[key], smoke_hidden[target][key], mask
    )
    random_removal = random_removal_like(
        removal, mask,
        f"smoke:{smoke_row['family_id']}:{rank}:{key}:{site['scope']}",
    )
    selected_removals[key] = removal
    random_removals[key] = random_removal
smoke_records = score_restoration_conditions(
    smoke_bundle['examples'][target], upstream_delta,
    [
        {
            'condition': 'joint_restore',
            'removals': selected_removals,
            'random_control': False,
        },
        {
            'condition': 'joint_random',
            'removals': random_removals,
            'random_control': True,
        },
    ],
)
self_score = forward_score_upstream(
    smoke_bundle['examples'][target],
    smoke_hidden[target]['L48:decoder'], 'replace',
)
self_error = centered_error(
    self_score['candidate_logprobs'],
    smoke_scores[target]['candidate_logprobs'],
)
real_norm = smoke_records[0]['restored_norm']
random_norm = smoke_records[1]['random_norm']
norm_error = abs(real_norm - random_norm) / max(real_norm, 1e-12)
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert norm_error < NORM_RELATIVE_TOLERANCE, norm_error
print('Smoke factor / doses / hybrid outcomes:',
      smoke_row['factor'], smoke_doses, smoke_hybrid_outcomes)
print('Self replay / joint norm error:', self_error, norm_error)
```

```python
%%cell_progress 11/17 Run frozen graded interventions
cell_stage('Capturing three target contexts per family with resumable shards')

def oriented_margin(record, positive, negative):
    return semantic_margin(record, positive, negative)

def add_metadata(record, row, bundle, condition):
    record.update({
        'schema_version': 'graded-product-record-1.0',
        'family_id': row['family_id'],
        'factor': row['factor'],
        'background': bundle['background'],
        'mapping_variant': bundle['mapping_variant'],
        'condition': condition,
        'changed_side': row['changed_side'],
        'positive_semantic': bundle['positive_semantic'],
        'negative_semantic': bundle['negative_semantic'],
        'target_semantic': bundle['target_semantic'],
        'hybrid_semantic': bundle['hybrid_semantic'],
        'physical_dose': bundle['physical_dose'],
        'context_value': bundle['context_value'],
        'source_context': bundle['source_context'],
        'source_dose': bundle['factor_delta'] * bundle['source_context'],
        'factor_delta': bundle['factor_delta'],
    })
    return record

capture_identity = {
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'frozen_sites': FROZEN_SITES,
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()

progress = tqdm(
    total=len(factor_manifest) * len(BACKGROUNDS),
    desc='Graded product-law families', unit='background',
)
for row in factor_manifest:
    family_path = SHARD_DIR / f"{row['family_id']}.json"
    if family_path.is_file():
        existing = json.loads(family_path.read_text(encoding='utf-8'))
        assert existing['capture_identity_sha256'] == CAPTURE_SHA256
        assert len(existing['records']) == len(BACKGROUNDS) * len(CONDITIONS)
        progress.update(len(BACKGROUNDS))
        continue
    family_records = []
    for background_index, _ in enumerate(BACKGROUNDS):
        bundle = make_bundle(row, background_index)
        role_names = (
            bundle['anchor_role'], bundle['source_role'], bundle['target_role']
        )
        prepared = {
            name: prepare_examples([bundle['examples'][name]])
            for name in role_names
        }
        masks = assert_aligned(prepared)
        scores, hidden = {}, {}
        for name in role_names:
            scores[name], hidden[name], _ = forward_capture_components(
                prepared[name]
            )
            torch.cuda.empty_cache()
        anchor, source, target = role_names
        upstream_mask = masks[UPSTREAM_SCOPE]
        upstream_delta = torch.zeros_like(hidden[source]['L48:decoder'])
        upstream_delta[upstream_mask] = (
            hidden[source]['L48:decoder'][upstream_mask].float()
            - hidden[anchor]['L48:decoder'][upstream_mask].float()
        ).to(torch.bfloat16)
        additive_score, additive_hidden, _ = forward_capture_components(
            prepared[target], upstream_delta, 'add'
        )

        selected_removals, random_removals = {}, {}
        for rank, site in enumerate(FROZEN_SITES[row['factor']], start=1):
            key = site['site_key']
            mask = masks[site['scope']]
            removal = masked_removal(
                additive_hidden[key], hidden[target][key], mask
            )
            random_removal = random_removal_like(
                removal, mask,
                f"graded:{row['factor']}:{row['family_id']}:"
                f"{bundle['background']}:{rank}:{key}:{site['scope']}",
            )
            selected_removals[key] = removal
            random_removals[key] = random_removal
        restored = score_restoration_conditions(
            bundle['examples'][target], upstream_delta,
            [
                {
                    'condition': 'joint_restore',
                    'removals': dict(selected_removals),
                    'random_control': False,
                },
                {
                    'condition': 'joint_random',
                    'removals': dict(random_removals),
                    'random_control': True,
                },
            ],
        )
        restored_by_condition = {
            record['condition']: record for record in restored
        }
        whole_donor = forward_score_upstream(
            bundle['examples'][target], hidden[source]['L48:decoder'], 'replace'
        )
        natural_hybrid = forward_score(
            bundle['examples'][bundle['hybrid_role']]
        )
        self_replay = forward_score_upstream(
            bundle['examples'][target], hidden[target]['L48:decoder'], 'replace'
        )
        plain = {
            'additive_total': additive_score,
            'whole_donor': whole_donor,
            'natural_hybrid': natural_hybrid,
            'self_replay': self_replay,
        }
        for record in plain.values():
            record.update({
                'restored_norm': 0.0, 'random_norm': 0.0,
                'restored_sites': 0,
            })
        cell_records = {**plain, **restored_by_condition}
        assert set(cell_records) == set(CONDITIONS)
        clean = scores[target]
        clean_margin = oriented_margin(
            clean, bundle['positive_semantic'], bundle['negative_semantic']
        )
        for condition in CONDITIONS:
            record = cell_records[condition]
            add_metadata(record, row, bundle, condition)
            record.update({
                'clean_candidate_semantics': clean['candidate_semantics'],
                'clean_candidate_logprobs': clean['candidate_logprobs'],
                'clean_predicted_semantic': clean['predicted_semantic'],
                'clean_oriented_margin': clean_margin,
                'intervention_oriented_margin': oriented_margin(
                    record, bundle['positive_semantic'], bundle['negative_semantic']
                ),
                'upstream_delta_norm': float(torch.linalg.vector_norm(
                    upstream_delta[upstream_mask].float()
                )),
            })
            record['answer_effect'] = (
                record['intervention_oriented_margin'] - clean_margin
            )
            family_records.append(record)
        del prepared, scores, hidden, additive_hidden, upstream_delta
        torch.cuda.empty_cache()
        progress.update(1)
        progress.set_postfix_str(
            f"{row['factor']} | {row['family_id']} | {bundle['background']}"
        )
    assert len(family_records) == len(BACKGROUNDS) * len(CONDITIONS)
    write_json_atomic(family_path, {
        'schema_version': 'graded-product-family-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'family': row,
        'records': family_records,
    })
progress.close()

shard_paths = sorted(SHARD_DIR.glob('*.json'))
assert len(shard_paths) == EXPECTED_FAMILIES
write_json_atomic(SHARED_ROOT / 'capture_complete.json', {
    'capture_identity': capture_identity,
    'capture_identity_sha256': CAPTURE_SHA256,
    'families': EXPECTED_FAMILIES,
    'records': EXPECTED_FAMILIES * len(BACKGROUNDS) * len(CONDITIONS),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Complete family shards:', len(shard_paths))
```

```python
%%cell_progress 12/17 Assemble dose-response estimands
cell_stage('Releasing the model and constructing family-level slopes')
import gc
try:
    del adapter
    del DECODER_LAYERS
except NameError:
    pass
gc.collect()
torch.cuda.empty_cache()

records = []
for path in sorted(SHARD_DIR.glob('*.json')):
    document = json.loads(path.read_text(encoding='utf-8'))
    assert document['capture_identity_sha256'] == CAPTURE_SHA256
    records.extend(document['records'])
record_frame = pd.DataFrame(records)
assert len(record_frame) == (
    EXPECTED_FAMILIES * len(BACKGROUNDS) * len(CONDITIONS)
)
record_frame.to_csv(ANALYSIS_DIR / 'graded_records.csv', index=False)

index_columns = [
    'factor', 'family_id', 'background', 'mapping_variant',
    'changed_side', 'positive_semantic', 'negative_semantic',
    'target_semantic', 'hybrid_semantic', 'physical_dose',
    'context_value', 'source_context', 'source_dose', 'factor_delta',
    'clean_oriented_margin',
]
cell_frame = record_frame.pivot_table(
    index=index_columns, columns='condition',
    values='answer_effect', aggfunc='first',
).reset_index()
cell_frame['route_attenuation'] = (
    cell_frame['additive_total'] - cell_frame['joint_restore']
)
cell_frame['random_attenuation'] = (
    cell_frame['additive_total'] - cell_frame['joint_random']
)
cell_frame['route_specificity'] = (
    cell_frame['joint_random'] - cell_frame['joint_restore']
)
cell_frame.to_csv(ANALYSIS_DIR / 'graded_cell_estimands.csv', index=False)

def simple_slope(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    centered = x - x.mean()
    denominator = float(np.sum(centered ** 2))
    assert denominator > 0
    return float(np.sum(centered * (y - y.mean())) / denominator)

slope_rows = []
for (factor, family_id), current in cell_frame.groupby(['factor', 'family_id']):
    current = current.sort_values('physical_dose')
    row = {
        'factor': factor,
        'family_id': family_id,
        'dose_min': float(current['physical_dose'].min()),
        'dose_max': float(current['physical_dose'].max()),
        'dose_range': float(current['physical_dose'].max() - current['physical_dose'].min()),
    }
    for condition in CONDITIONS:
        row[f'{condition}_slope'] = simple_slope(
            current['physical_dose'], current[condition]
        )
    row['route_slope_attenuation'] = (
        row['additive_total_slope'] - row['joint_restore_slope']
    )
    row['random_slope_attenuation'] = (
        row['additive_total_slope'] - row['joint_random_slope']
    )
    row['route_slope_specificity'] = (
        row['joint_random_slope'] - row['joint_restore_slope']
    )
    effects = current['additive_total'].to_numpy(dtype=float)
    row['additive_monotonic'] = float(np.all(np.diff(effects) > 0))
    slope_rows.append(row)
family_slopes = pd.DataFrame(slope_rows)
assert len(family_slopes) == EXPECTED_FAMILIES
family_slopes.to_csv(ANALYSIS_DIR / 'family_product_slopes.csv', index=False)
display(family_slopes.groupby('factor').mean(numeric_only=True).round(5))
```

```python
%%cell_progress 13/17 Test held-background product-law prediction
cell_stage('Comparing product dose with context-free and raw-context models')

def ridge_fit_predict(train, test, features, family_levels, ridge=1e-6):
    train_families = train['family_id'].astype(str).to_numpy()
    test_families = test['family_id'].astype(str).to_numpy()
    family_to_index = {value: index for index, value in enumerate(family_levels)}
    def design(frame, names, family_values):
        one_hot = np.zeros((len(frame), len(family_levels)), dtype=float)
        for row_index, family in enumerate(family_values):
            one_hot[row_index, family_to_index[family]] = 1.0
        continuous = frame[names].to_numpy(dtype=float) if names else np.empty((len(frame), 0))
        return np.column_stack([one_hot, continuous])
    X_train = design(train, features, train_families)
    X_test = design(test, features, test_families)
    y_train = train['additive_total'].to_numpy(dtype=float)
    penalty = ridge * np.eye(X_train.shape[1])
    coefficients = np.linalg.solve(
        X_train.T @ X_train + penalty,
        X_train.T @ y_train,
    )
    return X_test @ coefficients

MODEL_FEATURES = {
    'context_free': ['clean_oriented_margin'],
    'raw_context': ['clean_oriented_margin', 'context_value'],
    'donor_carryover': ['clean_oriented_margin', 'source_dose'],
    'product_law': ['clean_oriented_margin', 'physical_dose'],
}
prediction_rows = []
for factor in FACTORS:
    current = cell_frame[cell_frame['factor'] == factor].copy()
    family_levels = sorted(current['family_id'].astype(str).unique())
    for held_background in BACKGROUNDS:
        train = current[current['background'] != held_background]
        test = current[current['background'] == held_background].copy()
        assert set(test['family_id']) == set(family_levels)
        predictions = {
            model: ridge_fit_predict(
                train, test, features, family_levels
            )
            for model, features in MODEL_FEATURES.items()
        }
        truth = test['additive_total'].to_numpy(dtype=float)
        for row_index, (_, row) in enumerate(test.iterrows()):
            output = {
                'factor': factor,
                'family_id': row['family_id'],
                'background': held_background,
                'truth': float(truth[row_index]),
            }
            for model in MODEL_FEATURES:
                output[f'{model}_prediction'] = float(predictions[model][row_index])
                output[f'{model}_squared_error'] = float(
                    (truth[row_index] - predictions[model][row_index]) ** 2
                )
            prediction_rows.append(output)
prediction_frame = pd.DataFrame(prediction_rows)
prediction_frame.to_csv(ANALYSIS_DIR / 'held_background_predictions.csv', index=False)

family_cv = prediction_frame.groupby(['factor', 'family_id'], as_index=False)[[
    f'{model}_squared_error' for model in MODEL_FEATURES
]].mean()
family_cv['product_vs_context_free'] = (
    family_cv['context_free_squared_error']
    - family_cv['product_law_squared_error']
)
family_cv['product_vs_raw_context'] = (
    family_cv['raw_context_squared_error']
    - family_cv['product_law_squared_error']
)
family_cv['product_vs_donor_carryover'] = (
    family_cv['donor_carryover_squared_error']
    - family_cv['product_law_squared_error']
)
family_cv.to_csv(ANALYSIS_DIR / 'family_cv_metrics.csv', index=False)
display(family_cv.groupby('factor').mean(numeric_only=True).round(6))
```

```python
%%cell_progress 14/17 Run frozen product and route-mediation gates
cell_stage('Bootstrapping factorial families without behavior filtering')

def bootstrap_mean(values, key):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    assert len(values) == FAMILIES_PER_FACTOR, (key, len(values))
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    estimates = values[draws].mean(axis=1)
    return {
        'estimate': float(values.mean()),
        'ci_low': float(np.quantile(estimates, ALPHA / 2)),
        'ci_high': float(np.quantile(estimates, 1 - ALPHA / 2)),
        'n_families': int(len(values)),
    }

SLOPE_METRICS = (
    'additive_total_slope', 'joint_restore_slope', 'joint_random_slope',
    'natural_hybrid_slope', 'whole_donor_slope',
    'route_slope_attenuation', 'random_slope_attenuation',
    'route_slope_specificity', 'additive_monotonic',
)
CV_METRICS = (
    'product_vs_context_free', 'product_vs_raw_context',
    'product_vs_donor_carryover',
)
test_rows = []
decisions = {}
for factor in FACTORS:
    slope_current = family_slopes[family_slopes['factor'] == factor]
    cv_current = family_cv[family_cv['factor'] == factor]
    summaries = {}
    for metric in SLOPE_METRICS:
        summary = bootstrap_mean(
            slope_current[metric].to_numpy(), f'{factor}:{metric}'
        )
        summaries[metric] = summary
        test_rows.append({'factor': factor, 'metric': metric, **summary})
    for metric in CV_METRICS:
        summary = bootstrap_mean(
            cv_current[metric].to_numpy(), f'{factor}:{metric}'
        )
        summaries[metric] = summary
        test_rows.append({'factor': factor, 'metric': metric, **summary})
    gates = {
        'positive_product_slope': (
            summaries['additive_total_slope']['ci_low'] > PRODUCT_SLOPE_SESOI
        ),
        'beats_context_free': (
            summaries['product_vs_context_free']['ci_low'] > 0
        ),
        'beats_raw_context': (
            summaries['product_vs_raw_context']['ci_low'] > 0
        ),
        'beats_donor_carryover': (
            summaries['product_vs_donor_carryover']['ci_low'] > 0
        ),
        'natural_hybrid_dose_response': (
            summaries['natural_hybrid_slope']['ci_low'] > 0
        ),
        'route_attenuates_product_slope': (
            summaries['route_slope_attenuation']['ci_low']
            > MEDIATION_SLOPE_SESOI
        ),
        'route_specific_over_random': (
            summaries['route_slope_specificity']['ci_low'] > 0
        ),
    }
    decisions[factor] = {
        'confirmed': bool(all(gates.values())),
        'gates': {key: bool(value) for key, value in gates.items()},
        'summaries': summaries,
    }

confirmed = [factor for factor in FACTORS if decisions[factor]['confirmed']]
positive_slopes = [
    factor for factor in FACTORS
    if decisions[factor]['gates']['positive_product_slope']
]
if len(confirmed) == 2:
    PRIMARY_DECISION = 'graded_product_law_route_confirmed_for_both_factors'
elif len(confirmed) == 1:
    PRIMARY_DECISION = 'graded_product_law_route_confirmed_for_one_factor'
elif len(positive_slopes) >= 1:
    PRIMARY_DECISION = 'dose_response_without_full_product_route_confirmation'
else:
    PRIMARY_DECISION = 'context_invariant_or_unreliable_factor_effect'

test_frame = pd.DataFrame(test_rows)
test_frame.to_csv(ANALYSIS_DIR / 'product_route_tests.csv', index=False)
decision_document = {
    'schema_version': 'graded-product-law-decision-1.0',
    'primary_decision': PRIMARY_DECISION,
    'factor_decisions': decisions,
    'interpretation_guardrail': (
        'A positive result supports functional quantitative recombination and '
        'route mediation; it does not establish isolated weight or distance neurons.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'graded_product_decision.json', decision_document)
display(test_frame.round(6))
print('Primary decision:', PRIMARY_DECISION)
```

```python
%%cell_progress 15/17 Audit support, physics, mappings, and parity
cell_stage('Verifying model-blind selection and numerical controls')

manifest_lock = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
physics_violations = []
factor_scene_lists = defaultdict(list)
mapping_counts = Counter()
for row in factor_manifest:
    if not graded_valid(row['changed_side'], row['roles']):
        physics_violations.append((row['family_id'], 'graded_valid'))
    doses = [row['background_doses'][background] for background in BACKGROUNDS]
    if len(set(doses)) != len(BACKGROUNDS) or min(doses) <= 0:
        physics_violations.append((row['family_id'], 'dose'))
    factor_scene_lists[row['factor']].extend(
        str(role['scene_id']) for role in row['roles'].values()
    )
    for background_index, _ in enumerate(BACKGROUNDS):
        variant, _ = mapping_for(row, background_index)
        mapping_counts[(row['factor'], variant)] += 1

within_factor_disjoint = all(
    len(values) == len(set(values))
    for values in factor_scene_lists.values()
)
cross_factor_overlap = set.intersection(*(
    set(values) for values in factor_scene_lists.values()
))
immediate_prior_overlap = {
    scene_id for values in factor_scene_lists.values() for scene_id in values
    if scene_id in excluded_scene_ids
}
self_errors = []
norm_errors = []
for _, current in record_frame.groupby(['factor', 'family_id', 'background']):
    clean = current.iloc[0]['clean_candidate_logprobs']
    self_record = current[current['condition'] == 'self_replay'].iloc[0]
    self_errors.append(centered_error(
        self_record['candidate_logprobs'], clean
    ))
    real = current[current['condition'] == 'joint_restore'].iloc[0]
    random = current[current['condition'] == 'joint_random'].iloc[0]
    denominator = max(float(real['restored_norm']), 1e-12)
    norm_errors.append(abs(
        float(real['restored_norm']) - float(random['random_norm'])
    ) / denominator)

nonfinite_count = int((~np.isfinite(
    cell_frame.select_dtypes(include=[np.number]).to_numpy()
)).sum())
assert not physics_violations, physics_violations[:3]
assert within_factor_disjoint
assert max(self_errors) < SMOKE_CENTERED_LOGPROB_TOLERANCE
assert max(norm_errors) < NORM_RELATIVE_TOLERANCE
assert nonfinite_count == 0
assert all(
    mapping_counts[(factor, 'forward')]
    == mapping_counts[(factor, 'reverse')]
    for factor in FACTORS
)
audit_document = {
    'schema_version': 'graded-product-law-audit-1.0',
    'manifest_selection_policy': manifest_lock['identity']['selection_policy'],
    'selection_uses_model_outputs': False,
    'route_selected_from_current_data': False,
    'families': len(factor_manifest),
    'backgrounds_per_family': len(BACKGROUNDS),
    'records': len(record_frame),
    'physics_violations': len(physics_violations),
    'within_factor_scene_disjoint': within_factor_disjoint,
    'global_scene_disjoint': not cross_factor_overlap,
    'cross_factor_scene_overlap_count': len(cross_factor_overlap),
    'immediate_prior_scene_overlap_count': len(immediate_prior_overlap),
    'maximum_self_replay_error': float(max(self_errors)),
    'maximum_equal_norm_relative_error': float(max(norm_errors)),
    'nonfinite_count': nonfinite_count,
    'mapping_counts': {
        '|'.join(key): value for key, value in sorted(mapping_counts.items())
    },
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_document)
display(pd.DataFrame([audit_document]).T.rename(columns={0: 'value'}))
```
