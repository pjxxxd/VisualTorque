# Cross-Context Factor Recombination

Tests whether a weight or distance change is a transportable internal factor rather than a copy of the donor scene's completed torque.

[实验设置与运行说明](../../../docs/experiment_notes/10_exploratory_composition_and_gating/158ff50445dfc023.md)


```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. Model-blind 3×3 cross-context manifest


```python
%%cell_progress 6/17 Build or verify the model-independent 3x3 manifest
cell_stage('Loading Dataset v1 scenes and prior scene support')
import itertools
from collections import defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}

def single_shape(scene):
    if len(scene['stacks']) != 2:
        return None
    left = [stack for stack in scene['stacks'] if stack['side'] == 'left']
    right = [stack for stack in scene['stacks'] if stack['side'] == 'right']
    if len(left) != 1 or len(right) != 1:
        return None
    return (
        int(left[0]['units']), int(left[0]['distance']),
        int(right[0]['units']), int(right[0]['distance']),
    )

scene_by_shape = {}
for scene in scene_records:
    shape = single_shape(scene)
    if shape is not None:
        assert shape not in scene_by_shape
        scene_by_shape[shape] = str(scene['scene_id'])
assert len(scene_by_shape) == 900, len(scene_by_shape)

prior_scene_ids = set()
results_root = PROJECT_ROOT / 'results'
prior_manifest_paths = []
if results_root.is_dir():
    prior_manifest_paths = sorted(
        path for path in results_root.rglob('*manifest*.jsonl')
        if SHARED_ROOT not in path.parents
    )
for path in tqdm(prior_manifest_paths, desc='Reading prior manifests', unit='file'):
    try:
        rows = read_jsonl(path)
    except Exception:
        continue
    for row in rows:
        for key, value in row.items():
            if key.endswith('_scene_id') and isinstance(value, (str, int)):
                prior_scene_ids.add(str(value))
            elif key.endswith('_scene_ids') and isinstance(value, (list, tuple)):
                prior_scene_ids.update(str(item) for item in value)
        if isinstance(row.get('roles'), dict):
            for role in row['roles'].values():
                if isinstance(role, dict) and 'scene_id' in role:
                    prior_scene_ids.add(str(role['scene_id']))
print('Prior scene IDs available for overlap audit:', len(prior_scene_ids))

def shape_for(side, units, distance, fixed_units, fixed_distance):
    if side == 'left':
        return (units, distance, fixed_units, fixed_distance)
    return (fixed_units, fixed_distance, units, distance)

def transport_roles():
    return {
        'weight_to_d1': {
            'factor': 'weight', 'anchor': 's00', 'source': 's10',
            'target': 's01', 'hybrid': 's11', 'background': 'd1',
        },
        'weight_to_d2': {
            'factor': 'weight', 'anchor': 's00', 'source': 's10',
            'target': 's02', 'hybrid': 's12', 'background': 'd2',
        },
        'distance_to_w1': {
            'factor': 'distance', 'anchor': 's00', 'source': 's01',
            'target': 's10', 'hybrid': 's11', 'background': 'w1',
        },
        'distance_to_w2': {
            'factor': 'distance', 'anchor': 's00', 'source': 's01',
            'target': 's20', 'hybrid': 's21', 'background': 'w2',
        },
    }

TRANSPORT_ROLE_MAP = transport_roles()

if MANIFEST_PATH.is_file() and MANIFEST_LOCK_PATH.is_file():
    grid_manifest = read_jsonl(MANIFEST_PATH)
    lock = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert lock['identity']['selected_count'] == len(grid_manifest)
    cell_stage('Using the existing locked shared manifest')
else:
    cell_stage('Enumerating model-blind 3x3 factorial grids')
    candidates = []
    for side in ('left', 'right'):
        for fixed_units in range(1, 6):
            for fixed_distance in range(1, 7):
                for weight_combo in itertools.combinations(range(1, 6), 3):
                    for distance_combo in itertools.combinations(range(1, 7), 3):
                        for weight_sign in (-1, 1):
                            weights = (
                                tuple(weight_combo) if weight_sign > 0
                                else tuple(reversed(weight_combo))
                            )
                            for distance_sign in (-1, 1):
                                distances = (
                                    tuple(distance_combo) if distance_sign > 0
                                    else tuple(reversed(distance_combo))
                                )
                                scene_ids = {}
                                for i, units in enumerate(weights):
                                    for j, distance in enumerate(distances):
                                        shape = shape_for(
                                            side, units, distance,
                                            fixed_units, fixed_distance,
                                        )
                                        scene_id = scene_by_shape.get(shape)
                                        if scene_id is None:
                                            break
                                        scene_ids[f's{i}{j}'] = scene_id
                                if len(scene_ids) != 9:
                                    continue
                                torques = {
                                    f's{i}{j}': int(weights[i] * distances[j])
                                    for i in range(3) for j in range(3)
                                }
                                valid = True
                                for spec in TRANSPORT_ROLE_MAP.values():
                                    triad = {
                                        torques[spec['hybrid']],
                                        torques[spec['target']],
                                        torques[spec['source']],
                                    }
                                    if len(triad) != 3:
                                        valid = False
                                        break
                                if not valid:
                                    continue
                                family_id = 'crossgrid_' + stable_key('|'.join([
                                    side, str(fixed_units), str(fixed_distance),
                                    repr(weights), repr(distances),
                                ]))[:18]
                                candidates.append({
                                    'family_id': family_id,
                                    'changed_side': side,
                                    'fixed_opposite_units': fixed_units,
                                    'fixed_opposite_distance': fixed_distance,
                                    'weights': list(weights),
                                    'distances': list(distances),
                                    'weight_sign': weight_sign,
                                    'distance_sign': distance_sign,
                                    'scene_ids': scene_ids,
                                    'prior_overlap_count': sum(
                                        scene_id in prior_scene_ids
                                        for scene_id in scene_ids.values()
                                    ),
                                })

    assert len(candidates) >= TOTAL_FAMILIES, len(candidates)
    groups = defaultdict(list)
    for row in candidates:
        groups[(
            row['changed_side'], row['weight_sign'], row['distance_sign'],
        )].append(row)
    for rows in groups.values():
        rows.sort(key=lambda row: (
            row['prior_overlap_count'], stable_key(row['family_id'])
        ))
    group_keys = sorted(groups, key=lambda value: stable_key(repr(value)))
    cursors = {key: 0 for key in group_keys}
    selected = []
    used_scenes = set()
    while len(selected) < TOTAL_FAMILIES:
        changed = False
        for key in group_keys:
            rows = groups[key]
            while cursors[key] < len(rows):
                row = rows[cursors[key]]
                cursors[key] += 1
                current = set(row['scene_ids'].values())
                if current & used_scenes:
                    continue
                selected.append(row)
                used_scenes.update(current)
                changed = True
                break
            if len(selected) == TOTAL_FAMILIES:
                break
        if not changed:
            break
    assert len(selected) >= MIN_FAMILIES, (
        f'Only {len(selected)} scene-disjoint grids; need {MIN_FAMILIES}.'
    )
    selected = selected[:TOTAL_FAMILIES]

    def role_record(row, i, j):
        role = f's{i}{j}'
        scene_id = row['scene_ids'][role]
        scene = scene_by_id[scene_id]
        side = row['changed_side']
        stack = next(
            stack for stack in scene['stacks'] if stack['side'] == side
        )
        return {
            'scene_id': scene_id,
            'units': int(stack['units']),
            'distance': int(stack['distance']),
            'side_torque': int(scene['derived'][f'{side}_torque']),
            'net_torque': int(scene['derived']['net_torque']),
            'outcome': str(scene['derived']['outcome']),
        }

    grid_manifest = []
    for index, row in enumerate(selected):
        roles = {
            f's{i}{j}': role_record(row, i, j)
            for i in range(3) for j in range(3)
        }
        grid_manifest.append({
            **{key: value for key, value in row.items() if key != 'scene_ids'},
            'split': 'pilot' if index < PILOT_FAMILIES else 'confirmation',
            'roles': roles,
            'transports': TRANSPORT_ROLE_MAP,
        })
    grid_manifest.sort(key=lambda row: str(row['family_id']))
    write_jsonl_atomic(MANIFEST_PATH, grid_manifest)
    manifest_identity = {
        'schema_version': 'cross-context-grid-manifest-1.0',
        'dataset_scene_count': len(scene_records),
        'candidate_count': len(candidates),
        'selected_count': len(grid_manifest),
        'pilot_count': sum(row['split'] == 'pilot' for row in grid_manifest),
        'confirmation_count': sum(
            row['split'] == 'confirmation' for row in grid_manifest
        ),
        'scene_disjoint': True,
        'selection_uses_model_information': False,
        'selection_seed': SEED,
        'family_ids': [row['family_id'] for row in grid_manifest],
    }
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

assert len(grid_manifest) >= MIN_FAMILIES
all_new_scene_ids = [
    role['scene_id'] for row in grid_manifest for role in row['roles'].values()
]
assert len(all_new_scene_ids) == len(set(all_new_scene_ids))
manifest_frame = pd.DataFrame([{
    'family_id': row['family_id'],
    'split': row['split'],
    'changed_side': row['changed_side'],
    'weight_sign': row['weight_sign'],
    'distance_sign': row['distance_sign'],
    'prior_overlap_count': row['prior_overlap_count'],
} for row in grid_manifest])
manifest_frame.to_csv(SHARED_ROOT / 'cross_context_grid_manifest.csv', index=False)
display(manifest_frame.groupby([
    'split', 'changed_side', 'weight_sign', 'distance_sign'
]).size().rename('n').reset_index())
print('Families / scenes:', len(grid_manifest), len(all_new_scene_ids))
print('Families with prior-scene overlap:', int(
    (manifest_frame['prior_overlap_count'] > 0).sum()
))

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=SHARED_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example
    for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)
print('Pinned revision:', MODEL_REVISION)
```

## 3. Contrastive factor transport and diagnostic triads


```python
%%cell_progress 8/17 Define cross-context prompts, bundles, and exact masks
cell_stage('Defining factor, torque, and net-torque diagnostic prompts')
import math
from dataclasses import replace as dataclass_replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side and compare the two side totals.'
)

def task_question(task, side):
    if task == 'side_units':
        return f'How many blocks are on the {side} side?'
    if task == 'side_distance':
        return f'How many tick intervals from the pivot is the {side} stack?'
    if task == 'side_torque':
        return f'What is the total {side}-side torque?'
    if task == 'net_torque':
        return 'What is left torque minus right torque?'
    raise KeyError(task)

def role_value(row, role, task):
    values = row['roles'][role]
    if task == 'side_units':
        return int(values['units'])
    if task == 'side_distance':
        return int(values['distance'])
    if task == 'side_torque':
        return int(values['side_torque'])
    if task == 'net_torque':
        return int(values['net_torque'])
    raise KeyError(task)

def option_semantics(row, transport_name, task):
    spec = row['transports'][transport_name]
    hybrid = role_value(row, spec['hybrid'], task)
    target = role_value(row, spec['target'], task)
    donor = role_value(row, spec['source'], task)
    if task in ('side_torque', 'net_torque'):
        semantics = [hybrid, target, donor]
        assert len(set(semantics)) == 3, (
            row['family_id'], transport_name, task, semantics
        )
    else:
        semantics = []
        for value in (hybrid, target, donor):
            if value not in semantics:
                semantics.append(value)
        anchor = target
        for value in (
            anchor + 1, anchor - 1, 0, 1,
            anchor + 2, anchor - 2, anchor + 3, anchor - 3,
        ):
            if len(semantics) == 3:
                break
            if value not in semantics:
                semantics.append(value)
        assert len(semantics) == 3
    offset = int(stable_key(
        f"options:{row['family_id']}:{transport_name}:{task}"
    )[:8], 16) % 3
    return tuple(semantics[offset:] + semantics[:offset])

def custom_prompt(task, side, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {value}' for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + task_question(task, side),
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def goal_and_control(row, transport_name, task):
    spec = row['transports'][transport_name]
    factor = spec['factor']
    target = role_value(row, spec['target'], task)
    hybrid = role_value(row, spec['hybrid'], task)
    donor = role_value(row, spec['source'], task)
    relevant = (
        task in ('side_torque', 'net_torque')
        or (task == 'side_units' and factor == 'weight')
        or (task == 'side_distance' and factor == 'distance')
    )
    goal = hybrid if relevant else target
    alternatives = [value for value in (target, donor, hybrid) if value != goal]
    control = alternatives[0] if alternatives else goal
    return goal, control, relevant

def make_task_bundle(row, transport_name, task):
    spec = row['transports'][transport_name]
    semantics = option_semantics(row, transport_name, task)
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, str(row['changed_side']), semantics)
    role_names = tuple(dict.fromkeys((
        spec['anchor'], spec['source'], spec['target'], spec['hybrid'],
    )))
    examples = {}
    for role in role_names:
        scene_id = str(row['roles'][role]['scene_id'])
        value = role_value(row, role, task)
        gold_index = semantics.index(value) if value in semantics else 0
        base = outcome_by_scene[scene_id]
        examples[role] = dataclass_replace(
            base,
            example_id=(
                f"image_rule:cross_context_{row['family_id']}_"
                f"{transport_name}_{task}_{role}"
            ),
            question_id=(
                f"cross_context_{row['family_id']}_{transport_name}_{task}"
            ),
            question_kind=task,
            variant='cross_context_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(str(value) for value in semantics),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    goal, control, relevant = goal_and_control(row, transport_name, task)
    return {
        'transport': transport_name,
        'factor': spec['factor'],
        'background': spec['background'],
        'task': task,
        'semantics': semantics,
        'examples': examples,
        'anchor_role': spec['anchor'],
        'source_role': spec['source'],
        'target_role': spec['target'],
        'hybrid_role': spec['hybrid'],
        'anchor_semantic': role_value(row, spec['anchor'], task),
        'donor_semantic': role_value(row, spec['source'], task),
        'target_semantic': role_value(row, spec['target'], task),
        'hybrid_semantic': role_value(row, spec['hybrid'], task),
        'goal_semantic': goal,
        'control_semantic': control,
        'task_relevant': relevant,
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key]), (
                name, key
            )
        masks = discover_masks(current)
        for _, scope, _ in FROZEN_SITES:
            assert torch.equal(reference_masks[scope], masks[scope]), (name, scope)
    return reference_masks

print('Cross-context prompts and masks ready.')
```

```python
%%cell_progress 9/17 Define capture and additive/replace intervention scorers
cell_stage('Defining candidate scoring and contrastive factor transport')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = (
                output_hidden_tensor(output).detach().cpu().clone()
            )
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def capture_named_examples(examples_by_name):
    names = list(examples_by_name)
    score_by_name = {}
    hidden_by_name = {name: {} for name in names}
    for start in range(0, len(names), CAPTURE_BATCH_SIZE):
        actual = names[start:start + CAPTURE_BATCH_SIZE]
        padded = list(actual)
        while len(padded) < CAPTURE_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepare_examples([examples_by_name[name] for name in padded])
        scores, hidden = forward_capture_batch(prepared)
        for index, name in enumerate(actual):
            score_by_name[name] = scores[index]
            for layer_index, tensor in hidden.items():
                hidden_by_name[name][layer_index] = tensor[index]
        del prepared, hidden
        torch.cuda.empty_cache()
    return score_by_name, hidden_by_name

def matched_random_delta(delta, mask, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(delta.shape, generator=generator, dtype=torch.float32)
    selected = mask.bool()
    output = torch.zeros_like(noise)
    delta_selected = delta.float()[selected]
    noise_selected = noise[selected]
    delta_norm = float(torch.linalg.vector_norm(delta_selected))
    noise_norm = float(torch.linalg.vector_norm(noise_selected))
    if delta_norm > 0 and noise_norm > 0:
        output[selected] = noise_selected * (delta_norm / noise_norm)
    return output.to(dtype=delta.dtype)

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert len(layer_indices) == 1
    layer_index = next(iter(layer_indices))
    payloads = torch.stack([row['payload_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    modes = [row['intervention_mode'] for row in conditions]
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payloads.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == payload.shape
        modified = hidden.clone()
        for index, mode in enumerate(modes):
            current_mask = mask[index]
            if mode == 'add':
                modified[index, current_mask] = (
                    hidden[index, current_mask] + payload[index, current_mask]
                )
            elif mode == 'replace':
                modified[index, current_mask] = payload[index, current_mask]
            else:
                raise ValueError(mode)
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('payload_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)

    def run_chunk(chunk):
        actual_size = len(chunk)
        padded = list(chunk)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepared_cache.get(len(padded))
        if prepared is None:
            prepared = prepare_examples([target_example] * len(padded))
            prepared_cache[len(padded)] = prepared
        try:
            return forward_intervention_chunk(prepared, padded)[:actual_size]
        except torch.cuda.OutOfMemoryError as exc:
            prepared_cache.pop(len(padded), None)
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Intervention batch ran out of memory. Restart on an A100; '
                'do not change the frozen batch size within a model run.'
            ) from exc

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

print('Contrastive additive and replacement scorers ready.')
```

```python
%%cell_progress 10/17 Run physics, alignment, and parity smoke tests
cell_stage('Checking 3x3 physics, fanout roles, masks, and intervention parity')
smoke_row = next(row for row in grid_manifest if row['split'] == 'pilot')
smoke_bundle = make_task_bundle(smoke_row, 'weight_to_d1', 'side_torque')
assert len({
    smoke_bundle['hybrid_semantic'], smoke_bundle['target_semantic'],
    smoke_bundle['donor_semantic'],
}) == 3
assert smoke_bundle['hybrid_semantic'] == (
    smoke_row['weights'][1] * smoke_row['distances'][1]
)

smoke_prepared = {
    name: prepare_examples([example])
    for name, example in smoke_bundle['examples'].items()
}
smoke_masks = assert_aligned(smoke_prepared)
for _, scope, _ in FROZEN_SITES:
    assert bool(smoke_masks[scope].any()), scope

smoke_scores, smoke_hidden = capture_named_examples(
    smoke_bundle['examples']
)
target_role = smoke_bundle['target_role']
source_role = smoke_bundle['source_role']
anchor_role = smoke_bundle['anchor_role']
hybrid_role = smoke_bundle['hybrid_role']
smoke_layer = FROZEN_SITES[0][0]
zero_delta = (
    smoke_hidden[target_role][smoke_layer]
    - smoke_hidden[target_role][smoke_layer]
)
factor_delta = (
    smoke_hidden[source_role][smoke_layer]
    - smoke_hidden[anchor_role][smoke_layer]
)
assert float(torch.linalg.vector_norm(factor_delta.float())) > 0
smoke_conditions = [
    {
        'condition': 'self_all_tokens',
        'intervention': 'self_replay',
        'intervention_mode': 'replace',
        'layer_index': smoke_layer,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': smoke_hidden[target_role][smoke_layer],
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'condition': 'hybrid_all_tokens',
        'intervention': 'natural_hybrid',
        'intervention_mode': 'replace',
        'layer_index': smoke_layer,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': smoke_hidden[hybrid_role][smoke_layer],
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'condition': 'zero_add_all_tokens',
        'intervention': 'zero_add',
        'intervention_mode': 'add',
        'layer_index': smoke_layer,
        'scope': 'valid_tokens',
        'site_label': 'smoke_all_tokens',
        'payload_hidden': zero_delta,
        'patch_mask': smoke_masks['valid_tokens'],
    },
]
smoke_records = score_conditions(
    smoke_bundle['examples'][target_role], {}, smoke_conditions
)
self_error = float(np.max(np.abs(
    centered_scores(smoke_records[0])
    - centered_scores(smoke_scores[target_role])
)))
hybrid_error = float(np.max(np.abs(
    centered_scores(smoke_records[1])
    - centered_scores(smoke_scores[hybrid_role])
)))
zero_error = float(np.max(np.abs(
    centered_scores(smoke_records[2])
    - centered_scores(smoke_scores[target_role])
)))
print('Centered self / hybrid / zero-add parity:',
      self_error, hybrid_error, zero_error)
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert hybrid_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, hybrid_error
assert zero_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, zero_error
print('Smoke family:', smoke_row['family_id'])
print('Diagnostic triad H/T/D:',
      smoke_bundle['hybrid_semantic'], smoke_bundle['target_semantic'],
      smoke_bundle['donor_semantic'])
print('Frozen token counts:', {
    f'{label}:L{layer}:{scope}': int(smoke_masks[scope].sum())
    for layer, scope, label in FROZEN_SITES
})
del smoke_prepared, smoke_hidden, factor_delta, zero_delta
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/17 Capture or resume all cross-context interventions
cell_stage('Locking capture identity and checking completed family shards')

def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return 'sha256:' + digest.hexdigest()

CONDITIONS_PER_BUNDLE = len(FROZEN_SITES) * len(INTERVENTIONS)
RECORDS_PER_FAMILY = len(TRANSPORTS) * len(TASKS) * CONDITIONS_PER_BUNDLE
EXPECTED_PATCH_RECORDS = len(grid_manifest) * RECORDS_PER_FAMILY
capture_identity = {
    'schema_version': 'cross-context-recombination-capture-1.0',
    'model_key': MODEL_KEY,
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'dataset_fingerprint': example_config.dataset_fingerprint(),
    'protocol_lock_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'frozen_sites': [list(site) for site in FROZEN_SITES],
    'transports': list(TRANSPORTS),
    'tasks': list(TASKS),
    'interventions': list(INTERVENTIONS),
    'capture_batch_size': CAPTURE_BATCH_SIZE,
    'patch_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'records_per_family': RECORDS_PER_FAMILY,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'layer_path': LAYER_PATH,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing capture differs. Use a new versioned output directory.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_family_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_family_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    family_id = str(payload['family']['family_id'])
    assert family_id not in completed_family_ids
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    completed_family_ids.add(family_id)
    completed_records += len(payload['patch_records'])

progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc=f'{MODEL_KEY} cross-context patches', unit='condition',
    dynamic_ncols=True,
)
for family_index, row in enumerate(grid_manifest):
    family_id = str(row['family_id'])
    if family_id in completed_family_ids:
        continue
    progress.set_postfix_str(
        f"{row['split']} | {row['changed_side']} | W{row['weight_sign']:+d} D{row['distance_sign']:+d}"
    )
    family_baselines = {}
    family_patch_records = []
    for transport_name in TRANSPORTS:
        for task in TASKS:
            bundle = make_task_bundle(row, transport_name, task)
            examples = bundle['examples']
            prepared_single = {
                name: prepare_examples([example])
                for name, example in examples.items()
            }
            masks = assert_aligned(prepared_single)
            scores, hidden = capture_named_examples(examples)
            target_role = bundle['target_role']
            source_role = bundle['source_role']
            anchor_role = bundle['anchor_role']
            hybrid_role = bundle['hybrid_role']
            bundle_key = f'{transport_name}__{task}'
            family_baselines[bundle_key] = {
                key: value for key, value in bundle.items()
                if key != 'examples'
            }
            family_baselines[bundle_key]['scores'] = scores

            conditions = []
            for layer_index, scope, site_label in FROZEN_SITES:
                delta = (
                    hidden[source_role][layer_index]
                    - hidden[anchor_role][layer_index]
                )
                random_delta = matched_random_delta(
                    delta, masks[scope],
                    f'{family_id}:{transport_name}:{task}:{site_label}',
                )
                payloads = {
                    'additive_delta': ('add', delta),
                    'whole_donor': ('replace', hidden[source_role][layer_index]),
                    'natural_hybrid': ('replace', hidden[hybrid_role][layer_index]),
                    'self_replay': ('replace', hidden[target_role][layer_index]),
                    'random_matched': ('add', random_delta),
                }
                for intervention in INTERVENTIONS:
                    mode, payload_hidden = payloads[intervention]
                    conditions.append({
                        'analysis_family': 'cross_context_recombination',
                        'family_id': family_id,
                        'split': row['split'],
                        'transport': transport_name,
                        'factor': bundle['factor'],
                        'background': bundle['background'],
                        'task': task,
                        'task_relevant': bool(bundle['task_relevant']),
                        'intervention': intervention,
                        'intervention_mode': mode,
                        'site_label': site_label,
                        'scope': scope,
                        'layer_index': int(layer_index),
                        'anchor_scene_id': row['roles'][anchor_role]['scene_id'],
                        'donor_scene_id': row['roles'][source_role]['scene_id'],
                        'target_scene_id': row['roles'][target_role]['scene_id'],
                        'hybrid_scene_id': row['roles'][hybrid_role]['scene_id'],
                        'payload_hidden': payload_hidden,
                        'patch_mask': masks[scope],
                    })
            prepared_cache = {1: prepared_single[target_role]}
            records = score_conditions(
                examples[target_role], prepared_cache, conditions
            )
            family_patch_records.extend(records)
            progress.update(len(records))
            del prepared_single, scores, hidden, prepared_cache, conditions
            torch.cuda.empty_cache()

    assert len(family_patch_records) == RECORDS_PER_FAMILY
    shard_path = SHARD_DIR / f'shard_family_{family_index:04d}.json'
    write_json_atomic(shard_path, {
        'schema_version': 'cross-context-family-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'family': row,
        'baselines': family_baselines,
        'patch_records': family_patch_records,
    })
progress.close()

completed = sorted(SHARD_DIR.glob('shard_family_*.json'))
assert len(completed) == len(grid_manifest)
observed_records = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    observed_records += len(payload['patch_records'])
assert observed_records == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'families': len(completed),
    'patch_records': observed_records,
    'capture_identity_sha256': CAPTURE_SHA256,
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Capture complete:', observed_records, 'patch records')
```

```python
%%cell_progress 12/17 Release GPU and assemble recombination estimands
cell_stage('Releasing model memory and loading completed family shards')
import gc

del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    assert semantic in semantics, (semantic, semantics)
    return float(record['candidate_logprobs'][semantics.index(semantic)])

long_rows = []
for shard_path in tqdm(
    sorted(SHARD_DIR.glob('shard_family_*.json')),
    desc='Assembling intervention rows', unit='family',
):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    family = payload['family']
    baselines = payload['baselines']
    for record in payload['patch_records']:
        bundle_key = f"{record['transport']}__{record['task']}"
        meta = baselines[bundle_key]
        clean_target = meta['scores'][meta['target_role']]
        hybrid = meta['hybrid_semantic']
        target = meta['target_semantic']
        donor = meta['donor_semantic']
        goal = meta['goal_semantic']
        control = meta['control_semantic']

        patched_ht = semantic_logprob(record, hybrid) - semantic_logprob(record, target)
        clean_ht = (
            semantic_logprob(clean_target, hybrid)
            - semantic_logprob(clean_target, target)
        )
        patched_hd = semantic_logprob(record, hybrid) - semantic_logprob(record, donor)
        clean_hd = (
            semantic_logprob(clean_target, hybrid)
            - semantic_logprob(clean_target, donor)
        )
        patched_goal = semantic_logprob(record, goal) - semantic_logprob(record, control)
        clean_goal = (
            semantic_logprob(clean_target, goal)
            - semantic_logprob(clean_target, control)
        )
        predicted = record['predicted_semantic']
        if len({hybrid, target, donor}) == 3:
            if predicted == hybrid:
                winner_class = 'hybrid'
            elif predicted == target:
                winner_class = 'target'
            elif predicted == donor:
                winner_class = 'donor'
            else:
                winner_class = 'other'
        else:
            winner_class = 'factor_task'

        long_rows.append({
            'family_id': family['family_id'],
            'split': family['split'],
            'changed_side': family['changed_side'],
            'weight_sign': int(family['weight_sign']),
            'distance_sign': int(family['distance_sign']),
            'prior_overlap_count': int(family['prior_overlap_count']),
            'transport': record['transport'],
            'factor': record['factor'],
            'background': record['background'],
            'task': record['task'],
            'task_relevant': bool(record['task_relevant']),
            'site_label': record['site_label'],
            'layer_index': int(record['layer_index']),
            'scope': record['scope'],
            'intervention': record['intervention'],
            'patched_tokens': int(record['patched_tokens']),
            'hybrid_semantic': hybrid,
            'target_semantic': target,
            'donor_semantic': donor,
            'goal_semantic': goal,
            'control_semantic': control,
            'winner_class': winner_class,
            'hybrid_target_margin': patched_ht,
            'hybrid_donor_margin': patched_hd,
            'goal_control_margin': patched_goal,
            'raw_hybrid_target_shift': patched_ht - clean_ht,
            'raw_hybrid_donor_shift': patched_hd - clean_hd,
            'raw_goal_shift': patched_goal - clean_goal,
        })

long_frame = pd.DataFrame(long_rows)
key_columns = [
    'family_id', 'transport', 'task', 'site_label',
]
self_frame = long_frame[
    long_frame['intervention'] == 'self_replay'
][key_columns + [
    'raw_hybrid_target_shift', 'raw_hybrid_donor_shift', 'raw_goal_shift',
]].rename(columns={
    'raw_hybrid_target_shift': 'self_hybrid_target_shift',
    'raw_hybrid_donor_shift': 'self_hybrid_donor_shift',
    'raw_goal_shift': 'self_goal_shift',
})
assert not self_frame.duplicated(key_columns).any()
long_frame = long_frame.merge(self_frame, on=key_columns, how='left', validate='many_to_one')
long_frame['centered_hybrid_target_shift'] = (
    long_frame['raw_hybrid_target_shift']
    - long_frame['self_hybrid_target_shift']
)
long_frame['centered_hybrid_donor_shift'] = (
    long_frame['raw_hybrid_donor_shift']
    - long_frame['self_hybrid_donor_shift']
)
long_frame['centered_goal_shift'] = (
    long_frame['raw_goal_shift'] - long_frame['self_goal_shift']
)
long_frame.to_csv(ANALYSIS_DIR / 'intervention_records.csv', index=False)

index_columns = [
    'family_id', 'split', 'changed_side', 'weight_sign', 'distance_sign',
    'prior_overlap_count', 'transport', 'factor', 'background', 'task',
    'task_relevant', 'site_label', 'layer_index', 'scope',
]
value_columns = [
    'centered_hybrid_target_shift', 'centered_hybrid_donor_shift',
    'centered_goal_shift', 'hybrid_donor_margin',
]
wide_frame = long_frame.pivot_table(
    index=index_columns, columns='intervention', values=value_columns,
    aggfunc='first', observed=True,
).reset_index()
wide_frame.columns = [
    column if isinstance(column, str) else (
        column[0] if not column[1] else f'{column[1]}__{column[0]}'
    )
    for column in wide_frame.columns
]
required_intervention_columns = [
    f'{intervention}__{metric}'
    for intervention in INTERVENTIONS
    for metric in value_columns
]
missing_columns = [
    column for column in required_intervention_columns
    if column not in wide_frame.columns
]
assert not missing_columns, missing_columns
wide_frame['recombination_vs_copy'] = (
    wide_frame['additive_delta__centered_hybrid_donor_shift']
    - wide_frame['whole_donor__centered_hybrid_donor_shift']
)
wide_frame['additive_vs_random'] = (
    wide_frame['additive_delta__centered_hybrid_target_shift']
    - wide_frame['random_matched__centered_hybrid_target_shift']
)
wide_frame['ceiling_minus_additive'] = (
    wide_frame['natural_hybrid__centered_hybrid_target_shift']
    - wide_frame['additive_delta__centered_hybrid_target_shift']
)
wide_frame.to_csv(ANALYSIS_DIR / 'recombination_estimands.csv', index=False)
print('Long intervention records:', len(long_frame))
print('Family-task-site estimands:', len(wide_frame))
display(wide_frame.head())
```

```python
%%cell_progress 13/17 Run frozen held-out recombination and factor-specificity tests
cell_stage('Bootstrapping confirmation families without filtering by correctness')

def bootstrap_mean(values, key, replicates=BOOTSTRAP_REPLICATES):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    assert len(array) > 1, (key, len(array))
    rng = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = rng.choice(array, size=(replicates, len(array)), replace=True).mean(axis=1)
    return {
        'n': int(len(array)),
        'mean': float(array.mean()),
        'ci_low': float(np.quantile(draws, ALPHA / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
    }

confirmation = wide_frame[wide_frame['split'] == 'confirmation'].copy()
primary = confirmation[
    (confirmation['task'] == 'side_torque')
    & (confirmation['site_label'] == 'factor_composition')
].copy()
assert set(primary['factor']) == {'weight', 'distance'}

test_rows = []
factor_decisions = {}
for factor in ('weight', 'distance'):
    factor_rows = primary[primary['factor'] == factor]
    summaries = {}
    for metric in (
        'additive_delta__centered_hybrid_target_shift',
        'recombination_vs_copy', 'additive_vs_random',
        'natural_hybrid__centered_hybrid_target_shift',
    ):
        summary = bootstrap_mean(
            factor_rows[metric], f'{MODEL_KEY}:{factor}:{metric}'
        )
        summaries[metric] = summary
        test_rows.append({
            'model_key': MODEL_KEY, 'factor': factor,
            'test_name': metric, **summary,
        })

    fanout = factor_rows.groupby('family_id', observed=True)[
        'additive_delta__centered_hybrid_target_shift'
    ].min()
    fanout_summary = bootstrap_mean(
        fanout, f'{MODEL_KEY}:{factor}:fanout_minimum'
    )
    summaries['fanout_minimum'] = fanout_summary
    test_rows.append({
        'model_key': MODEL_KEY, 'factor': factor,
        'test_name': 'fanout_minimum', **fanout_summary,
    })
    passed = (
        summaries['additive_delta__centered_hybrid_target_shift']['ci_low']
        > PRIMARY_RAW_SESOI
        and summaries['recombination_vs_copy']['ci_low'] > 0
        and summaries['additive_vs_random']['ci_low'] > 0
        and summaries['fanout_minimum']['ci_low'] > 0
        and summaries['natural_hybrid__centered_hybrid_target_shift']['ci_low'] > 0
    )
    factor_decisions[factor] = {
        'passed': bool(passed),
        'summaries': summaries,
    }

factor_control_rows = []
for factor in ('weight', 'distance'):
    relevant_task = 'side_units' if factor == 'weight' else 'side_distance'
    irrelevant_task = 'side_distance' if factor == 'weight' else 'side_units'
    relevant = confirmation[
        (confirmation['factor'] == factor)
        & (confirmation['task'] == relevant_task)
        & (confirmation['site_label'] == 'factor_composition')
    ]
    irrelevant = confirmation[
        (confirmation['factor'] == factor)
        & (confirmation['task'] == irrelevant_task)
        & (confirmation['site_label'] == 'factor_composition')
    ]
    relevant_summary = bootstrap_mean(
        relevant['additive_delta__centered_goal_shift'],
        f'{MODEL_KEY}:{factor}:relevant_factor_task',
    )
    irrelevant_summary = bootstrap_mean(
        irrelevant['additive_delta__centered_goal_shift'],
        f'{MODEL_KEY}:{factor}:irrelevant_factor_task',
    )
    factor_control_rows.extend([
        {'model_key': MODEL_KEY, 'factor': factor, 'role': 'relevant', **relevant_summary},
        {'model_key': MODEL_KEY, 'factor': factor, 'role': 'irrelevant', **irrelevant_summary},
    ])

net_rows = confirmation[
    (confirmation['task'] == 'net_torque')
    & (confirmation['site_label'] == 'factor_composition')
]
for factor in ('weight', 'distance'):
    summary = bootstrap_mean(
        net_rows[net_rows['factor'] == factor][
            'additive_delta__centered_hybrid_target_shift'
        ],
        f'{MODEL_KEY}:{factor}:net_torque_transport',
    )
    test_rows.append({
        'model_key': MODEL_KEY, 'factor': factor,
        'test_name': 'net_torque_transport', **summary,
    })

tests_frame = pd.DataFrame(test_rows)
factor_controls_frame = pd.DataFrame(factor_control_rows)
tests_frame.to_csv(ANALYSIS_DIR / 'confirmatory_tests.csv', index=False)
factor_controls_frame.to_csv(
    ANALYSIS_DIR / 'factor_specificity_controls.csv', index=False
)
PRIMARY_DECISION = (
    'both_factors_recombine'
    if all(value['passed'] for value in factor_decisions.values())
    else 'one_factor_recombines'
    if any(value['passed'] for value in factor_decisions.values())
    else 'no_factor_passes_frozen_gate'
)
print('Frozen primary decision:', PRIMARY_DECISION)
display(tests_frame)
display(factor_controls_frame)
```

```python
%%cell_progress 14/17 Audit fanout, orientation, overlap, site, and replay robustness
cell_stage('Checking that effects are not confined to one background or orientation')

robustness_rows = []
robust_subset = wide_frame[wide_frame['task'].isin(['side_torque', 'net_torque'])]
groupings = {
    'background': ['factor', 'background', 'task', 'site_label'],
    'orientation': [
        'factor', 'changed_side', 'weight_sign', 'distance_sign',
        'task', 'site_label',
    ],
    'overlap': ['factor', 'task', 'site_label', 'prior_overlap_count'],
    'split': ['factor', 'task', 'site_label', 'split'],
}
for audit_name, columns in groupings.items():
    grouped = robust_subset.groupby(columns, observed=True)
    for keys, group in grouped:
        keys = keys if isinstance(keys, tuple) else (keys,)
        robustness_rows.append({
            'audit': audit_name,
            **dict(zip(columns, keys)),
            'n': int(len(group)),
            'mean_hybrid_shift': float(group[
                'additive_delta__centered_hybrid_target_shift'
            ].mean()),
            'mean_vs_copy': float(group['recombination_vs_copy'].mean()),
            'mean_vs_random': float(group['additive_vs_random'].mean()),
        })

robustness_frame = pd.DataFrame(robustness_rows)
robustness_frame.to_csv(ANALYSIS_DIR / 'robustness_cells.csv', index=False)

self_rows = long_frame[long_frame['intervention'] == 'self_replay']
self_audit = {
    'n': int(len(self_rows)),
    'max_abs_raw_hybrid_target_shift': float(
        self_rows['raw_hybrid_target_shift'].abs().max()
    ),
    'p95_abs_raw_hybrid_target_shift': float(
        self_rows['raw_hybrid_target_shift'].abs().quantile(0.95)
    ),
    'max_abs_raw_goal_shift': float(self_rows['raw_goal_shift'].abs().max()),
}

primary_values = primary.groupby('family_id', observed=True)[
    'additive_delta__centered_hybrid_target_shift'
].mean()
leave_one_out = []
for family_id in primary_values.index:
    retained = primary_values.drop(index=family_id)
    leave_one_out.append(float(retained.mean()))
loo_audit = {
    'n_families': int(len(primary_values)),
    'minimum_leave_one_family_out_mean': float(min(leave_one_out)),
    'maximum_leave_one_family_out_mean': float(max(leave_one_out)),
}
write_json_atomic(ANALYSIS_DIR / 'numerical_robustness.json', {
    'self_replay': self_audit,
    'leave_one_family_out': loo_audit,
})
print('Self-replay numerical audit:', self_audit)
print('Leave-one-family-out audit:', loo_audit)
display(robustness_frame.head(20))
```

```python
%%cell_progress 15/17 Save the paper-facing cross-context figure
cell_stage('Rendering transport, donor-copy, fanout, and site panels')
import matplotlib.pyplot as plt

def mean_ci(values, key):
    summary = bootstrap_mean(values, key, replicates=5000)
    return summary['mean'], summary['ci_low'], summary['ci_high']

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
colors = {'weight': '#D97706', 'distance': '#2563EB'}

metric_labels = [
    ('additive_delta__centered_hybrid_target_shift', 'Additive factor delta'),
    ('whole_donor__centered_hybrid_target_shift', 'Whole donor'),
    ('random_matched__centered_hybrid_target_shift', 'Matched random'),
    ('natural_hybrid__centered_hybrid_target_shift', 'Natural hybrid ceiling'),
]
x = np.arange(len(metric_labels))
width = 0.35
for factor_index, factor in enumerate(('weight', 'distance')):
    subset = primary[primary['factor'] == factor]
    means, lower, upper = [], [], []
    for metric, _ in metric_labels:
        mean, low, high = mean_ci(subset[metric], f'figure:{factor}:{metric}')
        means.append(mean); lower.append(mean - low); upper.append(high - mean)
    axes[0, 0].bar(
        x + (factor_index - 0.5) * width, means, width,
        yerr=np.vstack([lower, upper]), capsize=3,
        color=colors[factor], label=factor,
    )
axes[0, 0].axhline(0, color='black', linewidth=1)
axes[0, 0].set_xticks(x, [label for _, label in metric_labels], rotation=15)
axes[0, 0].set_ylabel('Centered hybrid-vs-target shift')
axes[0, 0].set_title('A. Does the edit produce the new hybrid value?')
axes[0, 0].legend(frameon=False)

for factor_index, factor in enumerate(('weight', 'distance')):
    subset = primary[primary['factor'] == factor]
    values = []
    errors = [[], []]
    for metric in ('recombination_vs_copy', 'additive_vs_random'):
        mean, low, high = mean_ci(subset[metric], f'figure2:{factor}:{metric}')
        values.append(mean); errors[0].append(mean - low); errors[1].append(high - mean)
    positions = np.arange(2) + (factor_index - 0.5) * width
    axes[0, 1].bar(
        positions, values, width, yerr=np.asarray(errors), capsize=3,
        color=colors[factor], label=factor,
    )
axes[0, 1].axhline(0, color='black', linewidth=1)
axes[0, 1].set_xticks(np.arange(2), ['Versus whole donor', 'Versus random'])
axes[0, 1].set_ylabel('Specific recombination advantage')
axes[0, 1].set_title('B. Is the effect more than copying or corruption?')

fanout_plot = primary.groupby(
    ['factor', 'background'], observed=True
)['additive_delta__centered_hybrid_target_shift'].agg(['mean', 'sem']).reset_index()
for factor in ('weight', 'distance'):
    subset = fanout_plot[fanout_plot['factor'] == factor]
    axes[1, 0].errorbar(
        subset['background'], subset['mean'], yerr=1.96 * subset['sem'],
        marker='o', capsize=4, color=colors[factor], label=factor,
    )
axes[1, 0].axhline(0, color='black', linewidth=1)
axes[1, 0].set_ylabel('Hybrid-vs-target shift')
axes[1, 0].set_title('C. The same delta in two target backgrounds')
axes[1, 0].legend(frameon=False)

site_plot = confirmation[
    confirmation['task'] == 'side_torque'
].groupby(['factor', 'site_label'], observed=True)[
    'additive_delta__centered_hybrid_target_shift'
].agg(['mean', 'sem']).reset_index()
site_labels = list(dict.fromkeys(site_plot['site_label']))
for factor_index, factor in enumerate(('weight', 'distance')):
    subset = site_plot[site_plot['factor'] == factor].set_index('site_label')
    means = [subset.loc[label, 'mean'] for label in site_labels]
    sems = [subset.loc[label, 'sem'] for label in site_labels]
    axes[1, 1].bar(
        np.arange(len(site_labels)) + (factor_index - 0.5) * width,
        means, width, yerr=np.asarray(sems) * 1.96, capsize=3,
        color=colors[factor], label=factor,
    )
axes[1, 1].axhline(0, color='black', linewidth=1)
axes[1, 1].set_xticks(np.arange(len(site_labels)), site_labels)
axes[1, 1].set_ylabel('Hybrid-vs-target shift')
axes[1, 1].set_title('D. Composition site versus late answer site')

fig.suptitle(f'VisualTorque cross-context recombination — {MODEL_KEY}', fontsize=16)
fig.tight_layout()
figure_path = FIGURE_DIR / 'cross_context_recombination.png'
fig.savefig(figure_path, dpi=220, bbox_inches='tight')
plt.show()
print('Saved:', figure_path)
```

```python
%%cell_progress 16/17 Apply the frozen gate and write the model report
cell_stage('Writing the model-specific claim card and failure interpretation')

if PRIMARY_DECISION == 'both_factors_recombine':
    interpretation = (
        'Both weight and distance deltas transported to two new backgrounds, '
        'favored the unseen hybrid torque, and exceeded whole-donor and '
        'matched-random controls at the frozen factor-composition site.'
    )
elif PRIMARY_DECISION == 'one_factor_recombines':
    passed = [factor for factor, value in factor_decisions.items() if value['passed']]
    failed = [factor for factor, value in factor_decisions.items() if not value['passed']]
    interpretation = (
        f"Transportable recombination is factor-specific: {passed} passed and "
        f"{failed} did not pass the frozen gate. Do not claim a complete "
        'weight-distance program.'
    )
else:
    copy_mean = float(primary['recombination_vs_copy'].mean())
    if copy_mean <= 0:
        interpretation = (
            'The additive edit did not outperform whole-donor copying. Prior '
            'patch effects should be described as context-bound or holistic, '
            'not as independent factor transport.'
        )
    else:
        interpretation = (
            'The frozen intervention did not provide confirmatory evidence of '
            'fanout recombination. This does not prove the information is absent; '
            'it limits the causal compositionality claim at these sites.'
        )

decision_payload = {
    'schema_version': 'cross-context-recombination-decision-1.0',
    'model_key': MODEL_KEY,
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'capture_identity_sha256': CAPTURE_SHA256,
    'primary_decision': PRIMARY_DECISION,
    'factor_decisions': factor_decisions,
    'interpretation': interpretation,
    'family_counts': {
        str(key): int(value)
        for key, value in manifest_frame['split'].value_counts().items()
    },
    'self_replay_audit': self_audit,
    'leave_one_family_out': loo_audit,
    'limitations': [
        'The claim is local to frozen decoder-output sites and token scopes.',
        'An additive activation delta is a causal test, not proof of a literal multiplication neuron.',
        'Final-answer error rescue and induction require a separate notebook.',
        'Model-specific failures must not be averaged away in a pooled score.',
    ],
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
}
write_json_atomic(ANALYSIS_DIR / 'recombination_decision.json', decision_payload)

report_lines = [
    f'# Cross-context factor recombination: {MODEL_KEY}',
    '',
    f'Primary decision: **{PRIMARY_DECISION}**',
    '',
    interpretation,
    '',
    '## Factor decisions',
    '',
]
for factor, value in factor_decisions.items():
    report_lines.append(f"- {factor}: {'PASS' if value['passed'] else 'NOT CONFIRMED'}")
report_lines.extend([
    '',
    '## Interpretation boundary',
    '',
    'The decisive comparison is hybrid versus target and donor. A change toward '
    'the donor value is source-result copying; no change is non-use; a change '
    'toward the unseen hybrid value supports cross-context recombination.',
])
(ANALYSIS_DIR / 'REPORT.md').write_text(
    '\n'.join(report_lines) + '\n', encoding='utf-8'
)
display(decision_payload)
print('Report:', ANALYSIS_DIR / 'REPORT.md')
```
