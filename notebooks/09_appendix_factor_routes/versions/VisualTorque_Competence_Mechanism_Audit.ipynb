```python
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = (
            int(match.group(1)), int(match.group(2)), match.group(3)
        )
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s",
                refresh=True,
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(
                f"complete | elapsed {elapsed:.0f}s", refresh=True
            )
        else:
            bar.set_postfix_str(
                f"failed | elapsed {elapsed:.0f}s", refresh=True
            )
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 12 tracked cells.')
```

```python
%%cell_progress 3/12 Freeze the post-confirmatory audit protocol
cell_stage('Locking hypotheses, families, and decision rules')
SEED = 20260826
BOOTSTRAP_REPLICATES = 5000
PERMUTATION_REPLICATES = 10000
ALPHA = 0.05
FOCAL_SITES = (
    (16, 'all_text_except_final'),
    (16, 'options_tokens'),
)
STRATA_COLUMNS = ('edit_type', 'relevant_task')
MATCH_STRATA = ('edit_type', 'task_family')
AUDIT_LOCK_PATH = AUDIT_DIR / 'audit_protocol.json'

audit_identity = {
    'schema_version': 'competence-mechanism-audit-1.0',
    'analysis_status': (
        'post-confirmatory robustness audit motivated by observed '
        'custom-query competence heterogeneity'
    ),
    'source_files': {
        key: sha256_file(path) for key, path in REQUIRED_PATHS.items()
    },
    'focal_sites': [list(site) for site in FOCAL_SITES],
    'formal_success_test': (
        'one-sided pair sign-flip and pair bootstrap of '
        'value-minus-outcome recovery among actual query-both-correct pairs; '
        'BH across two focal sites'
    ),
    'formal_interaction_test': (
        'success-minus-failure value effect, blocked by edit_type and '
        'relevant_task; fixed harmonic block weights; one-sided within-block '
        'label permutation and stratified bootstrap; BH across two sites'
    ),
    'alternative_directional_metric': (
        'same-value minus outcome-only donor difference in source-answer '
        'candidate logit contrast; no recovery denominator'
    ),
    'matching': (
        'deterministic no-replacement nearest matching within edit_type and '
        'task_family on log natural-effect norm and absolute oracle value gap'
    ),
    'sensitivity': (
        '2.5/97.5 percent winsorization and natural-effect norm thresholds '
        'at the 0, 10, 25, and 50 percent quantiles'
    ),
    'stability': (
        'edit/task-family strata and leave-one-edit-out successful effects'
    ),
    'multiplicity': 'BH separately across the two frozen focal sites',
    'bootstrap_unit': 'counterfactual pair',
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'permutation_replicates': PERMUTATION_REPLICATES,
    'alpha': ALPHA,
    'decision_rule': {
        'robust': (
            'at least one focal site passes corrected successful recovery, '
            'corrected blocked interaction, corrected alternative directional '
            'metric, corrected matched interaction, and all leave-one-edit-out '
            'successful means are positive'
        ),
        'heterogeneous': (
            'a site passes successful recovery and alternative metric plus '
            'either blocked or matched interaction, but leave-one-edit-out '
            'stability fails'
        ),
        'metric_only': (
            'successful recovery passes but the alternative directional metric '
            'does not'
        ),
    },
    'interpretation_limit': (
        'This audit is not independent confirmation; conditioning on model '
        'success may select easier or higher-margin computations, matching '
        'cannot remove unmeasured confounding, and grouped patching does not '
        'identify a unique circuit or algorithm.'
    ),
    'seed': SEED,
}
if AUDIT_LOCK_PATH.is_file():
    previous = json.loads(AUDIT_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == audit_identity, (
        'Existing audit uses different source files or rules. Use a new '
        'audit version rather than overwriting it.'
    )
else:
    write_json_atomic(AUDIT_LOCK_PATH, {
        'identity': audit_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Focal sites:', FOCAL_SITES)
print('Audit status: post-confirmatory, not independent confirmation')
```

## 2. Reconstruct pair-level directional effects


```python
%%cell_progress 4/12 Build the pair-level competence audit table
cell_stage('Joining pair metadata to recovery effects')
manifest_columns = [
    'pair_id', 'relevant_task', 'control_task',
    'relevant_source_value', 'relevant_target_value',
    'competence_cohort', 'edit_type',
]
assert set(manifest_columns) <= set(pair_manifest.columns)
metadata = pair_manifest[manifest_columns].copy()
metadata['absolute_value_gap'] = (
    metadata['relevant_source_value'].astype(float)
    - metadata['relevant_target_value'].astype(float)
).abs()

def task_family(task):
    task = str(task)
    if 'units' in task:
        return 'units'
    if 'torque' in task:
        return 'torque'
    if 'stack' in task:
        return 'structure'
    return 'other'

metadata['task_family'] = metadata['relevant_task'].map(task_family)
gap_frame = pair_gaps.merge(
    metadata.drop(columns=['competence_cohort', 'edit_type']),
    on='pair_id', how='left', validate='many_to_one',
)
assert gap_frame['relevant_task'].notna().all()

cell_stage('Reconstructing denominator-free source-answer effects')
relevant_patch = patch_frame[
    (patch_frame['query_role'] == 'relevant_changed')
    & patch_frame['source_answer_contrast_gain'].notna()
].copy()
margin_pivot = relevant_patch.pivot_table(
    index=['pair_id', 'layer_index', 'scope'],
    columns='donor_type',
    values='source_answer_contrast_gain',
    aggfunc='first',
).reset_index()
assert {
    'matched_source', 'same_task_value_different_layout',
    'same_outcome_different_task_value',
} <= set(margin_pivot.columns)
margin_pivot['answer_margin_value_minus_outcome'] = (
    margin_pivot['same_task_value_different_layout']
    - margin_pivot['same_outcome_different_task_value']
)

natural = relevant_patch.groupby(
    ['pair_id', 'layer_index', 'scope'], as_index=False
).agg(
    natural_effect_norm=('natural_effect_norm', 'first'),
    valid_recovery=('valid_recovery', 'first'),
)
site_frame = (
    gap_frame.merge(
        margin_pivot[[
            'pair_id', 'layer_index', 'scope',
            'answer_margin_value_minus_outcome',
        ]],
        on=['pair_id', 'layer_index', 'scope'],
        how='left', validate='one_to_one',
    )
    .merge(
        natural,
        on=['pair_id', 'layer_index', 'scope'],
        how='left', validate='one_to_one',
    )
)
site_frame['query_both_correct'] = as_bool(
    site_frame['query_both_correct']
).astype(bool)
site_frame['log_natural_effect_norm'] = np.log1p(
    site_frame['natural_effect_norm'].astype(float)
)
site_frame['recovery_value_minus_outcome'] = (
    site_frame['value_minus_outcome'].astype(float)
)
assert not site_frame.duplicated(
    ['pair_id', 'layer_index', 'scope']
).any()
assert site_frame['answer_margin_value_minus_outcome'].notna().all()
focal_frame = site_frame[
    site_frame.apply(
        lambda row: (int(row['layer_index']), row['scope'])
        in set(FOCAL_SITES), axis=1,
    )
].copy()
assert focal_frame['pair_id'].nunique() >= 180
focal_frame.to_csv(AUDIT_DIR / 'pair_level_audit_table.csv', index=False)
print('Auditable pairs:', focal_frame['pair_id'].nunique())
print('Successful pairs:', focal_frame.loc[
    focal_frame['query_both_correct'], 'pair_id'
].nunique())
display(focal_frame.groupby(
    ['layer_index', 'scope', 'query_both_correct']
).size().rename('n').reset_index())
```

```python
%%cell_progress 5/12 Define and self-test pair-level statistical utilities
cell_stage('Defining deterministic bootstrap and permutation tests')

def stable_seed(label):
    digest = hashlib.sha256(f'{SEED}:{label}'.encode('utf-8')).hexdigest()
    return int(digest[:8], 16)

def bh_adjust(p_values):
    values = np.asarray(p_values, dtype=float)
    adjusted = np.full(values.shape, np.nan, dtype=float)
    valid = np.isfinite(values)
    if not valid.any():
        return adjusted
    observed = values[valid]
    order = np.argsort(observed)
    ranked = observed[order]
    scaled = ranked * len(ranked) / np.arange(1, len(ranked) + 1)
    scaled = np.minimum.accumulate(scaled[::-1])[::-1]
    restored = np.empty_like(scaled)
    restored[order] = np.minimum(scaled, 1.0)
    adjusted[valid] = restored
    return adjusted

def mean_test(values, label):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'p_value': np.nan,
            'standardized_mean': np.nan,
        }
    rng = np.random.default_rng(stable_seed(label))
    counts = rng.multinomial(
        len(values), np.full(len(values), 1 / len(values)),
        size=BOOTSTRAP_REPLICATES,
    )
    bootstrap = counts @ values / len(values)
    signs = rng.choice(
        np.asarray([-1.0, 1.0]),
        size=(PERMUTATION_REPLICATES, len(values)),
    )
    permuted = signs @ values / len(values)
    observed = float(values.mean())
    scale = float(values.std(ddof=1)) if len(values) > 1 else np.nan
    return {
        'n': int(len(values)),
        'mean': observed,
        'ci_low': float(np.quantile(bootstrap, 0.025)),
        'ci_high': float(np.quantile(bootstrap, 0.975)),
        'p_value': float(
            (1 + np.sum(permuted >= observed))
            / (PERMUTATION_REPLICATES + 1)
        ),
        'standardized_mean': (
            observed / scale if np.isfinite(scale) and scale > 0 else np.nan
        ),
    }

def stratified_interaction(
    frame, value_column, label,
    strata_columns=STRATA_COLUMNS,
):
    working = frame[
        [*strata_columns, 'query_both_correct', value_column]
    ].dropna().copy()
    blocks = []
    for keys, group in working.groupby(list(strata_columns), sort=True):
        success = group.loc[
            group['query_both_correct'], value_column
        ].to_numpy(dtype=float)
        failure = group.loc[
            ~group['query_both_correct'], value_column
        ].to_numpy(dtype=float)
        if len(success) and len(failure):
            weight = len(success) * len(failure) / (
                len(success) + len(failure)
            )
            blocks.append((keys, success, failure, weight))
    if not blocks:
        return {
            'success_n': 0, 'failure_n': 0, 'block_count': 0,
            'effect': np.nan, 'ci_low': np.nan, 'ci_high': np.nan,
            'p_value': np.nan,
        }
    weights = np.asarray([block[3] for block in blocks], dtype=float)
    weights /= weights.sum()
    observed_parts = np.asarray([
        success.mean() - failure.mean()
        for _, success, failure, _ in blocks
    ])
    observed = float(weights @ observed_parts)
    rng = np.random.default_rng(stable_seed(label))
    bootstrap = np.zeros(BOOTSTRAP_REPLICATES, dtype=float)
    permutation = np.zeros(PERMUTATION_REPLICATES, dtype=float)
    for weight, (_, success, failure, _) in zip(weights, blocks):
        success_counts = rng.multinomial(
            len(success), np.full(len(success), 1 / len(success)),
            size=BOOTSTRAP_REPLICATES,
        )
        failure_counts = rng.multinomial(
            len(failure), np.full(len(failure), 1 / len(failure)),
            size=BOOTSTRAP_REPLICATES,
        )
        bootstrap += weight * (
            success_counts @ success / len(success)
            - failure_counts @ failure / len(failure)
        )
        pooled = np.concatenate([success, failure])
        success_n = len(success)
        random_order = np.argsort(
            rng.random((PERMUTATION_REPLICATES, len(pooled))), axis=1
        )
        chosen = random_order[:, :success_n]
        chosen_sum = pooled[chosen].sum(axis=1)
        total = pooled.sum()
        permutation += weight * (
            chosen_sum / success_n
            - (total - chosen_sum) / len(failure)
        )
    return {
        'success_n': int(sum(len(block[1]) for block in blocks)),
        'failure_n': int(sum(len(block[2]) for block in blocks)),
        'block_count': int(len(blocks)),
        'effect': observed,
        'ci_low': float(np.quantile(bootstrap, 0.025)),
        'ci_high': float(np.quantile(bootstrap, 0.975)),
        'p_value': float(
            (1 + np.sum(permutation >= observed))
            / (PERMUTATION_REPLICATES + 1)
        ),
    }

cell_stage('Running deterministic utility smoke tests')
positive_test = mean_test(
    np.linspace(0.2, 1.0, 24), 'utility-positive'
)
assert positive_test['mean'] > 0 and positive_test['ci_low'] > 0
synthetic = pd.DataFrame({
    'edit_type': ['a'] * 20 + ['b'] * 20,
    'relevant_task': ['x'] * 10 + ['y'] * 10
                     + ['x'] * 10 + ['y'] * 10,
    'query_both_correct': ([True] * 5 + [False] * 5) * 4,
    'value': ([1.0] * 5 + [0.0] * 5) * 4,
})
interaction_test = stratified_interaction(
    synthetic, 'value', 'utility-interaction'
)
assert interaction_test['effect'] > 0.9
print('Statistical utilities passed deterministic smoke tests.')
```

## 3. Formal competence tests and metric triangulation


```python
%%cell_progress 6/12 Test successful effects and blocked competence interactions
cell_stage('Testing the two frozen layer-16 sites')
formal_rows = []
for layer_index, scope in FOCAL_SITES:
    site = focal_frame[
        (focal_frame['layer_index'] == layer_index)
        & (focal_frame['scope'] == scope)
    ]
    success = site[site['query_both_correct']]
    failure = site[~site['query_both_correct']]
    formal_rows.append({
        'layer_index': int(layer_index), 'scope': scope,
        **{f'success_{key}': value for key, value in mean_test(
            success['recovery_value_minus_outcome'],
            f'formal-success:L{layer_index}:{scope}',
        ).items()},
        **{f'failure_{key}': value for key, value in mean_test(
            failure['recovery_value_minus_outcome'],
            f'formal-failure:L{layer_index}:{scope}',
        ).items()},
        **{f'interaction_{key}': value for key, value in
           stratified_interaction(
               site, 'recovery_value_minus_outcome',
               f'formal-interaction:L{layer_index}:{scope}',
           ).items()},
    })
formal_tests = pd.DataFrame(formal_rows)
formal_tests['success_q_bh'] = bh_adjust(
    formal_tests['success_p_value']
)
formal_tests['interaction_q_bh'] = bh_adjust(
    formal_tests['interaction_p_value']
)
formal_tests['success_pass'] = (
    (formal_tests['success_mean'] > 0)
    & (formal_tests['success_ci_low'] > 0)
    & (formal_tests['success_q_bh'] < ALPHA)
)
formal_tests['interaction_pass'] = (
    (formal_tests['interaction_effect'] > 0)
    & (formal_tests['interaction_ci_low'] > 0)
    & (formal_tests['interaction_q_bh'] < ALPHA)
)
formal_tests.to_csv(
    AUDIT_DIR / 'formal_competence_tests.csv', index=False
)
display(formal_tests[[
    'layer_index', 'scope',
    'success_n', 'success_mean', 'success_ci_low', 'success_ci_high',
    'success_q_bh', 'success_pass',
    'failure_n', 'failure_mean',
    'interaction_effect', 'interaction_ci_low', 'interaction_ci_high',
    'interaction_q_bh', 'interaction_pass',
]])
```

```python
%%cell_progress 7/12 Audit alternative metrics and recovery sensitivity
cell_stage('Testing the denominator-free source-answer margin')
alternative_rows = []
sensitivity_rows = []
for layer_index, scope in FOCAL_SITES:
    site = focal_frame[
        (focal_frame['layer_index'] == layer_index)
        & (focal_frame['scope'] == scope)
    ].copy()
    success = site[site['query_both_correct']]
    margin_success = mean_test(
        success['answer_margin_value_minus_outcome'],
        f'margin-success:L{layer_index}:{scope}',
    )
    margin_interaction = stratified_interaction(
        site, 'answer_margin_value_minus_outcome',
        f'margin-interaction:L{layer_index}:{scope}',
    )

    lower, upper = site['recovery_value_minus_outcome'].quantile(
        [0.025, 0.975]
    )
    site['winsorized_recovery'] = site[
        'recovery_value_minus_outcome'
    ].clip(lower, upper)
    winsor_success = mean_test(
        site.loc[site['query_both_correct'], 'winsorized_recovery'],
        f'winsor-success:L{layer_index}:{scope}',
    )
    winsor_interaction = stratified_interaction(
        site, 'winsorized_recovery',
        f'winsor-interaction:L{layer_index}:{scope}',
    )
    alternative_rows.append({
        'layer_index': int(layer_index), 'scope': scope,
        **{f'margin_success_{key}': value
           for key, value in margin_success.items()},
        **{f'margin_interaction_{key}': value
           for key, value in margin_interaction.items()},
        **{f'winsor_success_{key}': value
           for key, value in winsor_success.items()},
        **{f'winsor_interaction_{key}': value
           for key, value in winsor_interaction.items()},
        'winsor_lower': float(lower),
        'winsor_upper': float(upper),
    })

    quantiles = site['natural_effect_norm'].quantile(
        [0.0, 0.1, 0.25, 0.5]
    )
    for quantile, threshold in quantiles.items():
        retained = site[site['natural_effect_norm'] >= threshold]
        retained_success = retained[retained['query_both_correct']]
        sensitivity_rows.append({
            'layer_index': int(layer_index), 'scope': scope,
            'natural_norm_quantile': float(quantile),
            'natural_norm_threshold': float(threshold),
            'retained_pairs': int(retained['pair_id'].nunique()),
            **{f'success_{key}': value for key, value in mean_test(
                retained_success['recovery_value_minus_outcome'],
                f'threshold-success:L{layer_index}:{scope}:{quantile}',
            ).items()},
            **{f'interaction_{key}': value for key, value in
               stratified_interaction(
                   retained, 'recovery_value_minus_outcome',
                   f'threshold-interaction:L{layer_index}:{scope}:{quantile}',
               ).items()},
        })

alternative_tests = pd.DataFrame(alternative_rows)
for family in (
    'margin_success', 'margin_interaction',
    'winsor_success', 'winsor_interaction',
):
    p_column = (
        f'{family}_p_value'
        if family in ('margin_success', 'winsor_success')
        else f'{family}_p_value'
    )
    alternative_tests[f'{family}_q_bh'] = bh_adjust(
        alternative_tests[p_column]
    )
alternative_tests['margin_success_pass'] = (
    (alternative_tests['margin_success_mean'] > 0)
    & (alternative_tests['margin_success_ci_low'] > 0)
    & (alternative_tests['margin_success_q_bh'] < ALPHA)
)
alternative_tests['margin_interaction_pass'] = (
    (alternative_tests['margin_interaction_effect'] > 0)
    & (alternative_tests['margin_interaction_ci_low'] > 0)
    & (alternative_tests['margin_interaction_q_bh'] < ALPHA)
)
alternative_tests['winsor_success_pass'] = (
    (alternative_tests['winsor_success_mean'] > 0)
    & (alternative_tests['winsor_success_ci_low'] > 0)
    & (alternative_tests['winsor_success_q_bh'] < ALPHA)
)
alternative_tests['winsor_interaction_pass'] = (
    (alternative_tests['winsor_interaction_effect'] > 0)
    & (alternative_tests['winsor_interaction_ci_low'] > 0)
    & (alternative_tests['winsor_interaction_q_bh'] < ALPHA)
)
sensitivity_frame = pd.DataFrame(sensitivity_rows)
alternative_tests.to_csv(
    AUDIT_DIR / 'alternative_metric_tests.csv', index=False
)
sensitivity_frame.to_csv(
    AUDIT_DIR / 'natural_effect_sensitivity.csv', index=False
)
display(alternative_tests[[
    'layer_index', 'scope',
    'margin_success_mean', 'margin_success_ci_low',
    'margin_success_ci_high', 'margin_success_q_bh',
    'margin_success_pass',
    'margin_interaction_effect', 'margin_interaction_q_bh',
    'winsor_success_mean', 'winsor_success_ci_low',
    'winsor_success_q_bh', 'winsor_success_pass',
]])
print('Natural-effect sensitivity:')
display(sensitivity_frame[[
    'layer_index', 'scope', 'natural_norm_quantile',
    'retained_pairs', 'success_n', 'success_mean',
    'success_ci_low', 'interaction_effect', 'interaction_ci_low',
]])
```

## 4. Matched competence comparison and structural stability


```python
%%cell_progress 8/12 Match successful and unsuccessful computations
cell_stage('Constructing site-independent pair covariates')
pair_covariates = focal_frame.groupby('pair_id', as_index=False).agg(
    query_both_correct=('query_both_correct', 'first'),
    edit_type=('edit_type', 'first'),
    relevant_task=('relevant_task', 'first'),
    task_family=('task_family', 'first'),
    absolute_value_gap=('absolute_value_gap', 'first'),
    log_natural_effect_norm=('log_natural_effect_norm', 'mean'),
)
assert not pair_covariates.duplicated('pair_id').any()

def standardized_mean_difference(success, failure):
    success = np.asarray(success, dtype=float)
    failure = np.asarray(failure, dtype=float)
    pooled = np.sqrt(
        (success.var(ddof=1) + failure.var(ddof=1)) / 2
    )
    return (
        float((success.mean() - failure.mean()) / pooled)
        if np.isfinite(pooled) and pooled > 0 else np.nan
    )

match_rows = []
for keys, group in pair_covariates.groupby(list(MATCH_STRATA), sort=True):
    success = group[group['query_both_correct']].copy()
    failure = group[~group['query_both_correct']].copy()
    if success.empty or failure.empty:
        continue
    covariates = ['absolute_value_gap', 'log_natural_effect_norm']
    center = group[covariates].mean()
    scale = group[covariates].std(ddof=0).replace(0, 1).fillna(1)
    success_z = (success[covariates] - center) / scale
    failure_z = (failure[covariates] - center) / scale
    success_order = sorted(
        range(len(success)),
        key=lambda index: hashlib.sha256(
            f"{SEED}:match:{success.iloc[index]['pair_id']}".encode()
        ).hexdigest(),
    )
    available = set(range(len(failure)))
    for success_index in success_order:
        if not available:
            break
        distances = [
            (
                float(np.linalg.norm(
                    success_z.iloc[success_index].to_numpy(dtype=float)
                    - failure_z.iloc[failure_index].to_numpy(dtype=float)
                )),
                hashlib.sha256(
                    f"{SEED}:failure:{failure.iloc[failure_index]['pair_id']}"
                    .encode()
                ).hexdigest(),
                failure_index,
            )
            for failure_index in available
        ]
        distance, _, failure_index = min(distances)
        available.remove(failure_index)
        match_rows.append({
            'edit_type': keys[0], 'task_family': keys[1],
            'success_pair_id': success.iloc[success_index]['pair_id'],
            'failure_pair_id': failure.iloc[failure_index]['pair_id'],
            'match_distance': distance,
        })
match_frame = pd.DataFrame(match_rows)
assert not match_frame.empty
assert not match_frame['success_pair_id'].duplicated().any()
assert not match_frame['failure_pair_id'].duplicated().any()
match_frame.to_csv(AUDIT_DIR / 'matched_pair_map.csv', index=False)

cell_stage('Testing paired success-minus-failure effects')
matched_rows = []
for layer_index, scope in FOCAL_SITES:
    site = focal_frame[
        (focal_frame['layer_index'] == layer_index)
        & (focal_frame['scope'] == scope)
    ][[
        'pair_id', 'recovery_value_minus_outcome',
        'answer_margin_value_minus_outcome',
    ]]
    joined = (
        match_frame.merge(
            site.add_prefix('success_'),
            left_on='success_pair_id', right_on='success_pair_id',
            validate='one_to_one',
        )
        .merge(
            site.add_prefix('failure_'),
            left_on='failure_pair_id', right_on='failure_pair_id',
            validate='one_to_one',
        )
    )
    joined['recovery_difference'] = (
        joined['success_recovery_value_minus_outcome']
        - joined['failure_recovery_value_minus_outcome']
    )
    joined['margin_difference'] = (
        joined['success_answer_margin_value_minus_outcome']
        - joined['failure_answer_margin_value_minus_outcome']
    )
    matched_rows.append({
        'layer_index': int(layer_index), 'scope': scope,
        **{f'recovery_{key}': value for key, value in mean_test(
            joined['recovery_difference'],
            f'matched-recovery:L{layer_index}:{scope}',
        ).items()},
        **{f'margin_{key}': value for key, value in mean_test(
            joined['margin_difference'],
            f'matched-margin:L{layer_index}:{scope}',
        ).items()},
    })
matched_tests = pd.DataFrame(matched_rows)
matched_tests['recovery_q_bh'] = bh_adjust(
    matched_tests['recovery_p_value']
)
matched_tests['margin_q_bh'] = bh_adjust(
    matched_tests['margin_p_value']
)
matched_tests['recovery_pass'] = (
    (matched_tests['recovery_mean'] > 0)
    & (matched_tests['recovery_ci_low'] > 0)
    & (matched_tests['recovery_q_bh'] < ALPHA)
)
matched_tests['margin_pass'] = (
    (matched_tests['margin_mean'] > 0)
    & (matched_tests['margin_ci_low'] > 0)
    & (matched_tests['margin_q_bh'] < ALPHA)
)
matched_tests.to_csv(
    AUDIT_DIR / 'matched_competence_tests.csv', index=False
)

matched_success = pair_covariates.merge(
    match_frame[['success_pair_id']].rename(
        columns={'success_pair_id': 'pair_id'}
    ), on='pair_id', validate='one_to_one',
)
matched_failure = pair_covariates.merge(
    match_frame[['failure_pair_id']].rename(
        columns={'failure_pair_id': 'pair_id'}
    ), on='pair_id', validate='one_to_one',
)
balance_rows = []
for covariate in ('absolute_value_gap', 'log_natural_effect_norm'):
    raw_success = pair_covariates.loc[
        pair_covariates['query_both_correct'], covariate
    ]
    raw_failure = pair_covariates.loc[
        ~pair_covariates['query_both_correct'], covariate
    ]
    balance_rows.append({
        'covariate': covariate,
        'raw_smd': standardized_mean_difference(
            raw_success, raw_failure
        ),
        'matched_smd': standardized_mean_difference(
            matched_success[covariate], matched_failure[covariate]
        ),
    })
balance_frame = pd.DataFrame(balance_rows)
balance_frame.to_csv(AUDIT_DIR / 'matching_balance.csv', index=False)
print('Matched pairs:', len(match_frame))
display(matched_tests)
display(balance_frame)
```

```python
%%cell_progress 9/12 Audit edit, task-family, and leave-one-edit-out stability
cell_stage('Estimating edit and task-family heterogeneity')
heterogeneity_rows = []
for layer_index, scope in FOCAL_SITES:
    site = focal_frame[
        (focal_frame['layer_index'] == layer_index)
        & (focal_frame['scope'] == scope)
    ]
    for group_type, group_column in (
        ('edit_type', 'edit_type'),
        ('task_family', 'task_family'),
    ):
        for group_name, group in site.groupby(group_column):
            success = group[group['query_both_correct']]
            strata = (
                ('relevant_task',)
                if group_column == 'edit_type' else ('edit_type',)
            )
            heterogeneity_rows.append({
                'layer_index': int(layer_index), 'scope': scope,
                'group_type': group_type,
                'group_name': str(group_name),
                **{f'success_{key}': value for key, value in mean_test(
                    success['recovery_value_minus_outcome'],
                    f'heterogeneity-success:L{layer_index}:{scope}:'
                    f'{group_type}:{group_name}',
                ).items()},
                **{f'interaction_{key}': value for key, value in
                   stratified_interaction(
                       group, 'recovery_value_minus_outcome',
                       f'heterogeneity-interaction:L{layer_index}:{scope}:'
                       f'{group_type}:{group_name}',
                       strata_columns=strata,
                   ).items()},
            })
heterogeneity_frame = pd.DataFrame(heterogeneity_rows)
heterogeneity_frame.to_csv(
    AUDIT_DIR / 'edit_task_heterogeneity.csv', index=False
)

cell_stage('Running leave-one-edit-out sensitivity')
leaveout_rows = []
edits = sorted(focal_frame['edit_type'].unique())
for layer_index, scope in FOCAL_SITES:
    site = focal_frame[
        (focal_frame['layer_index'] == layer_index)
        & (focal_frame['scope'] == scope)
    ]
    for excluded_edit in edits:
        retained = site[site['edit_type'] != excluded_edit]
        success = retained[retained['query_both_correct']]
        leaveout_rows.append({
            'layer_index': int(layer_index), 'scope': scope,
            'excluded_edit': excluded_edit,
            **{f'success_{key}': value for key, value in mean_test(
                success['recovery_value_minus_outcome'],
                f'leaveout-success:L{layer_index}:{scope}:{excluded_edit}',
            ).items()},
            **{f'interaction_{key}': value for key, value in
               stratified_interaction(
                   retained, 'recovery_value_minus_outcome',
                   f'leaveout-interaction:L{layer_index}:{scope}:{excluded_edit}',
               ).items()},
        })
leaveout_frame = pd.DataFrame(leaveout_rows)
leaveout_frame.to_csv(
    AUDIT_DIR / 'leave_one_edit_out.csv', index=False
)
print('Edit/task-family heterogeneity:')
display(heterogeneity_frame[[
    'layer_index', 'scope', 'group_type', 'group_name',
    'success_n', 'success_mean', 'success_ci_low', 'success_ci_high',
    'interaction_effect', 'interaction_ci_low',
]])
print('Leave-one-edit-out:')
display(leaveout_frame[[
    'layer_index', 'scope', 'excluded_edit',
    'success_n', 'success_mean', 'success_ci_low',
    'interaction_effect', 'interaction_ci_low',
]])
```

```python
%%cell_progress 11/12 Apply the frozen audit decision rule and write report
cell_stage('Combining corrected, alternative, matched, and stability gates')
decision_rows = []
for layer_index, scope in FOCAL_SITES:
    formal = formal_tests[
        (formal_tests['layer_index'] == layer_index)
        & (formal_tests['scope'] == scope)
    ].iloc[0]
    alternative = alternative_tests[
        (alternative_tests['layer_index'] == layer_index)
        & (alternative_tests['scope'] == scope)
    ].iloc[0]
    matched = matched_tests[
        (matched_tests['layer_index'] == layer_index)
        & (matched_tests['scope'] == scope)
    ].iloc[0]
    leaveout = leaveout_frame[
        (leaveout_frame['layer_index'] == layer_index)
        & (leaveout_frame['scope'] == scope)
    ]
    leaveout_all_positive = bool(
        (leaveout['success_mean'] > 0).all()
    )
    full_pass = bool(
        formal['success_pass']
        and formal['interaction_pass']
        and alternative['margin_success_pass']
        and matched['recovery_pass']
        and leaveout_all_positive
    )
    core_signal = bool(
        formal['success_pass']
        and alternative['margin_success_pass']
        and (
            formal['interaction_pass']
            or matched['recovery_pass']
        )
    )
    decision_rows.append({
        'layer_index': int(layer_index), 'scope': scope,
        'successful_recovery_pass': bool(formal['success_pass']),
        'blocked_interaction_pass': bool(formal['interaction_pass']),
        'alternative_margin_pass': bool(
            alternative['margin_success_pass']
        ),
        'matched_interaction_pass': bool(matched['recovery_pass']),
        'leave_one_edit_out_all_positive': leaveout_all_positive,
        'core_signal': core_signal,
        'full_robustness_pass': full_pass,
    })
decision_frame = pd.DataFrame(decision_rows)
decision_frame.to_csv(
    AUDIT_DIR / 'competence_mechanism_gate.csv', index=False
)

any_full = bool(decision_frame['full_robustness_pass'].any())
any_core = bool(decision_frame['core_signal'].any())
any_recovery = bool(
    decision_frame['successful_recovery_pass'].any()
)
any_margin = bool(decision_frame['alternative_margin_pass'].any())
if any_full:
    decision_code = 'robust_competence_conditional_route'
    recommended_next = (
        'Run a new competence-first confirmatory experiment on untouched '
        'pairs. Pre-screen the exact custom physical queries, freeze '
        'successful and matched-error cohorts, and retain only the passing '
        'layer-16 grouped sites plus negative-control sites.'
    )
elif any_core:
    decision_code = 'competence_signal_but_edit_heterogeneous'
    recommended_next = (
        'Do not claim a general torque route. Design the next independent '
        'confirmation around the stable edit/task families identified here, '
        'with a separately powered transfer test to the failing families.'
    )
elif any_recovery and not any_margin:
    decision_code = 'recovery_metric_only_signal'
    recommended_next = (
        'Treat the successful-subset result as recovery-normalization '
        'sensitivity rather than a physical-value mechanism. Do not run '
        'component decomposition.'
    )
else:
    decision_code = 'no_robust_competence_mechanism'
    recommended_next = (
        'Freeze the Qwen grouped-route result as a representation/use '
        'dissociation and replicate the measurement protocol on Gemma '
        'before investing in finer localization.'
    )

decision = {
    'schema_version': 'competence-mechanism-audit-decision-1.0',
    'decision': decision_code,
    'site_gates': decision_rows,
    'matched_pairs': int(len(match_frame)),
    'recommended_next': recommended_next,
    'interpretation_limit': audit_identity['interpretation_limit'],
}
write_json_atomic(AUDIT_DIR / 'audit_decision.json', decision)

report_lines = [
    '# Competence–mechanism robustness audit', '',
    '- Status: **post-confirmatory robustness audit**; not an independent '
    'confirmation.',
    f"- Source pairs: {pair_manifest['pair_id'].nunique()}.",
    f"- Auditable recovery pairs: {focal_frame['pair_id'].nunique()}.",
    f'- Deterministically matched success/error pairs: {len(match_frame)}.',
    '', '## Corrected focal-site tests', '',
]
for _, row in formal_tests.iterrows():
    report_lines.append(
        f"- L{int(row['layer_index'])} `{row['scope']}`: successful "
        f"recovery {row['success_mean']:.3f} "
        f"[{row['success_ci_low']:.3f}, {row['success_ci_high']:.3f}], "
        f"q={row['success_q_bh']:.4f}; blocked success−failure "
        f"interaction {row['interaction_effect']:.3f} "
        f"[{row['interaction_ci_low']:.3f}, "
        f"{row['interaction_ci_high']:.3f}], "
        f"q={row['interaction_q_bh']:.4f}."
    )
report_lines.extend([
    '', '## Directional and matched robustness', '',
])
for _, row in alternative_tests.iterrows():
    matched = matched_tests[
        (matched_tests['layer_index'] == row['layer_index'])
        & (matched_tests['scope'] == row['scope'])
    ].iloc[0]
    report_lines.append(
        f"- L{int(row['layer_index'])} `{row['scope']}`: denominator-free "
        f"source-answer effect {row['margin_success_mean']:.3f} "
        f"[{row['margin_success_ci_low']:.3f}, "
        f"{row['margin_success_ci_high']:.3f}], "
        f"q={row['margin_success_q_bh']:.4f}; matched recovery interaction "
        f"{matched['recovery_mean']:.3f} "
        f"[{matched['recovery_ci_low']:.3f}, "
        f"{matched['recovery_ci_high']:.3f}], "
        f"q={matched['recovery_q_bh']:.4f}."
    )
report_lines.extend([
    '', '## Frozen audit gate', '',
])
for row in decision_rows:
    report_lines.append(
        f"- L{row['layer_index']} `{row['scope']}`: successful recovery="
        f"{row['successful_recovery_pass']}; blocked interaction="
        f"{row['blocked_interaction_pass']}; alternative margin="
        f"{row['alternative_margin_pass']}; matched interaction="
        f"{row['matched_interaction_pass']}; all leave-one-edit-out positive="
        f"{row['leave_one_edit_out_all_positive']}; full pass="
        f"{row['full_robustness_pass']}."
    )
report_lines.extend([
    '', '## Decision', '',
    f'- **`{decision_code}`**',
    f'- Next: {recommended_next}',
    '', '## Interpretation limit', '',
    audit_identity['interpretation_limit'], '',
])
report_path = AUDIT_DIR / 'competence_mechanism_audit_report.md'
report_path.write_text(
    '\n'.join(report_lines) + '\n', encoding='utf-8'
)
print(report_path.read_text(encoding='utf-8'))
```
