```python
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 16 tracked cells.')
```

```python
%%cell_progress 6/16 Select 64 pairs and load fixed-variant examples
cell_stage('Loading confirmatory manifest and Dataset v1 records')
from collections import Counter, defaultdict
from dataclasses import replace

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

confirmatory_manifest_path = CONFIRMATORY_ROOT / 'pair_manifest.jsonl'
confirmatory_decision_path = (
    CONFIRMATORY_ROOT / 'analysis' / 'confirmatory_decision_summary.json'
)
discovery_config_path = DISCOVERY_ROOT / 'capture_config.json'
discovery_selection_path = (
    DISCOVERY_ROOT / 'probe_results' / 'discovery_selection.json'
)
for required in (
    confirmatory_manifest_path, confirmatory_decision_path,
    discovery_config_path, discovery_selection_path,
):
    assert required.is_file(), f'Missing prerequisite: {required}'
confirmatory_decision = json.loads(
    confirmatory_decision_path.read_text(encoding='utf-8')
)
assert confirmatory_decision['three_stage_route_supported'] is True

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}
confirmatory_pairs = read_jsonl(confirmatory_manifest_path)
discovery_config = json.loads(
    discovery_config_path.read_text(encoding='utf-8')
)
discovery_identity = discovery_config['identity']
MODEL_REVISION = str(discovery_identity['resolved_revision'])
discovery_selection = json.loads(
    discovery_selection_path.read_text(encoding='utf-8')
)
SELECTED_LAYER_INDEX = int(
    discovery_selection['candidate_net_torque_layer']['layer_index']
)
assert SELECTED_LAYER_INDEX == EXPECTED_SELECTED_LAYER

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

def balanced_manifest_take(rows, count):
    groups = defaultdict(list)
    for row in rows:
        groups[str(row['strategy_group'])].append(row)
    for values in groups.values():
        values.sort(key=lambda row: stable_key(row['pair_id']))
    keys = sorted(groups, key=lambda key: stable_key('group:' + key))
    cursor = {key: 0 for key in keys}
    selected = []
    while len(selected) < min(count, len(rows)):
        changed = False
        for key in keys:
            index = cursor[key]
            if index < len(groups[key]) and len(selected) < count:
                selected.append(groups[key][index])
                cursor[key] += 1
                changed = True
        if not changed:
            break
    return selected

pair_manifest = []
for edit_type in EDIT_TYPES:
    candidates = [
        row for row in confirmatory_pairs if row['edit_type'] == edit_type
    ]
    selected = balanced_manifest_take(candidates, PAIRS_PER_EDIT)
    assert len(selected) == PAIRS_PER_EDIT
    pair_manifest.extend(selected)
pair_manifest.sort(key=lambda row: row['pair_id'])

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=(SCRATCH_ROOT / 'unused_task_token_component'),
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1,
    generation_every=0,
    seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example
    for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)

manifest_path = OUTPUT_ROOT / 'pair_manifest.jsonl'
temporary_manifest = manifest_path.with_suffix('.jsonl.tmp')
with temporary_manifest.open('w', encoding='utf-8') as handle:
    for row in pair_manifest:
        handle.write(json.dumps(row, sort_keys=True) + '\n')
temporary_manifest.replace(manifest_path)
selection = pd.DataFrame(pair_manifest)
print('Selected pairs:', len(selection))
display(pd.crosstab(selection['edit_type'], selection['strategy_group']))
print('Outcome transitions:')
display(pd.crosstab(selection['edit_type'], selection['outcome_transition']))
```

```python
%%cell_progress 7/16 Load Qwen and resolve decoder components
import gc
if 'adapter' in globals():
    del adapter
    gc.collect()
    torch.cuda.empty_cache()
cell_stage('Loading pinned Qwen weights')
adapter = HuggingFaceMultimodalAdapter(
    MODEL_ID,
    revision=MODEL_REVISION,
    dtype=MODEL_DTYPE,
    device_map=DEVICE_MAP,
)
display(adapter.metadata)

def resolve_module_path(root, path):
    value = root
    for component in path.split('.'):
        value = getattr(value, component)
    return value

def discover_decoder_layers(model):
    candidates = [
        discovery_identity.get('layer_path'),
        'model.language_model.layers', 'model.text_model.layers',
        'language_model.model.layers', 'model.model.layers', 'model.layers',
    ]
    for path in candidates:
        if not path:
            continue
        try:
            layers = resolve_module_path(model, path)
        except (AttributeError, KeyError):
            continue
        if isinstance(layers, torch.nn.ModuleList) and len(layers) >= 8:
            return path, layers
    raise RuntimeError('Could not locate decoder layers.')

LAYER_PATH, DECODER_LAYERS = discover_decoder_layers(adapter.model)
assert len(DECODER_LAYERS) == 32
ALL_DECODER_CAPTURE_LAYERS = tuple(sorted(set(
    TASK_LAYERS + TOKEN_SCAN_LAYERS + TOKEN_CONTROL_LAYERS
)))
ATTENTION_MODULES = {}
ATTENTION_KINDS = {}
MLP_MODULES = {}

def resolve_token_mixer(layer, layer_index):
    preferred = ('self_attn', 'linear_attn', 'attention', 'mixer')
    for name in preferred:
        module = getattr(layer, name, None)
        if isinstance(module, torch.nn.Module):
            return name, module
    discovered = [
        (name, module)
        for name, module in layer.named_children()
        if isinstance(module, torch.nn.Module)
        and any(marker in name.lower() for marker in (
            'attn', 'attention', 'mixer', 'delta'
        ))
    ]
    if len(discovered) == 1:
        return discovered[0]
    child_names = [name for name, _ in layer.named_children()]
    raise RuntimeError(
        f'Could not uniquely resolve token mixer at layer {layer_index}; '
        f'layer={type(layer).__name__}, children={child_names}, '
        f'candidates={[name for name, _ in discovered]}'
    )

def resolve_mlp(layer, layer_index):
    for name in ('mlp', 'feed_forward', 'ffn'):
        module = getattr(layer, name, None)
        if isinstance(module, torch.nn.Module):
            return name, module
    child_names = [name for name, _ in layer.named_children()]
    raise RuntimeError(
        f'Could not resolve MLP at layer {layer_index}; '
        f'layer={type(layer).__name__}, children={child_names}'
    )

for layer_index in COMPONENT_LAYERS:
    layer = DECODER_LAYERS[layer_index]
    attention_kind, attention = resolve_token_mixer(layer, layer_index)
    _mlp_kind, mlp = resolve_mlp(layer, layer_index)
    ATTENTION_MODULES[layer_index] = attention
    ATTENTION_KINDS[layer_index] = attention_kind
    MLP_MODULES[layer_index] = mlp
print('Layer path:', LAYER_PATH)
print('Decoder capture layers:', ALL_DECODER_CAPTURE_LAYERS)
print('Token mixers by component layer:', ATTENTION_KINDS)
print('MLP modules resolved:', tuple(MLP_MODULES))
```

## 3. Aligned query construction, token masks, and patch scorers


```python
%%cell_progress 8/16 Define latent queries and exact token masks
cell_stage('Defining pair-aligned latent-variable questions')
import math

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick intervals '
    "from the central pivot. A stack's turning effect is its number of squares "
    'multiplied by its distance. Add the turning effects on each side and compare '
    'the two side totals.'
)
TASK_QUESTIONS = {
    'left_units': 'How many blocks are on the left side?',
    'right_units': 'How many blocks are on the right side?',
    'left_distance': 'How many tick intervals from the pivot is the left stack?',
    'right_distance': 'How many tick intervals from the pivot is the right stack?',
    'left_torque': 'What is the total left-side torque in block-distance units?',
    'right_torque': 'What is the total right-side torque in block-distance units?',
    'net_torque': 'What is left torque minus right torque?',
}

def scene_task_value(scene, task):
    derived = scene['derived']
    if task in (
        'left_units', 'right_units', 'left_torque',
        'right_torque', 'net_torque',
    ):
        return int(derived[task])
    if task in ('left_distance', 'right_distance'):
        side = task.split('_')[0]
        stacks = [stack for stack in scene['stacks'] if stack['side'] == side]
        if len(stacks) != 1:
            return None
        return int(stacks[0]['distance'])
    if task == 'outcome':
        return str(derived['outcome'])
    raise KeyError(task)

def numeric_option_semantics(pair_id, task, source_value, target_value):
    values = []
    for value in (source_value, target_value):
        if value not in values:
            values.append(int(value))
    distractors = (
        target_value + 1, target_value - 1,
        source_value + 1, source_value - 1,
        0, 1, -1,
    )
    for value in distractors:
        if int(value) not in values:
            values.append(int(value))
        if len(values) == 3:
            break
    assert len(values) == 3
    offset = int(stable_key(f'options:{pair_id}:{task}')[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def custom_prompt(question, semantics):
    labels = ('A', 'B', 'C')
    option_lines = [
        f'{label}. {value}' for label, value in zip(labels, semantics)
    ]
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + question,
        'Options:\n' + '\n'.join(option_lines),
        'Do not explain your reasoning. Reply with exactly one option letter and '
        'nothing else. Valid replies are: A, B, C.',
    ])

def build_task_examples(pair, task):
    source_scene = scene_by_id[pair['source_scene_id']]
    target_scene = scene_by_id[pair['target_scene_id']]
    source_base = outcome_by_scene[pair['source_scene_id']]
    target_base = outcome_by_scene[pair['target_scene_id']]
    if task == 'outcome':
        return source_base, target_base, (
            pair['source_outcome'] != pair['target_outcome']
        )
    source_value = scene_task_value(source_scene, task)
    target_value = scene_task_value(target_scene, task)
    if source_value is None or target_value is None:
        return None
    semantics = numeric_option_semantics(
        pair['pair_id'], task, source_value, target_value
    )
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(TASK_QUESTIONS[task], semantics)

    def convert(base, value, role):
        gold_index = semantics.index(value)
        return replace(
            base,
            example_id=(
                f"image_rule:route_{pair['pair_id']}_{task}_{role}"
            ),
            question_id=f"route_{pair['pair_id']}_{task}",
            question_kind=task,
            variant='route_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(str(value) for value in semantics),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    return (
        convert(source_base, source_value, 'source'),
        convert(target_base, target_value, 'target'),
        source_value != target_value,
    )

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples, 'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    candidate_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        candidate_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            candidate_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in candidate_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any())
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if ids.tolist()[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (
                max(0, min(token_end, end) - max(token_start, start)),
                name,
            )
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(source_prepared, target_prepared):
    for key in ('input_ids', 'attention_mask'):
        assert torch.equal(
            source_prepared['inputs'][key], target_prepared['inputs'][key]
        ), key
    source_masks = discover_masks(source_prepared)
    target_masks = discover_masks(target_prepared)
    for scope in TASK_SCOPES + TOKEN_SCAN_GROUPS:
        assert torch.equal(source_masks[scope], target_masks[scope]), scope
    return target_masks

print('Latent-query factory and exact masks ready.')
```

```python
%%cell_progress 9/16 Define decoder and component patch scorers
cell_stage('Defining general candidate-space scoring')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': (
                example.option_semantics[predicted_index] == example.gold_semantic
            ),
        })
    return records

def forward_capture(prepared, decoder_layers, capture_components=False):
    assert len(prepared['examples']) == 1
    decoder_hidden = {}
    component_hidden = {}
    handles = []
    for layer_index in decoder_layers:
        def layer_hook(_module, _inputs, output, layer_index=layer_index):
            decoder_hidden[layer_index] = (
                output_hidden_tensor(output)[0].detach().cpu().clone()
            )
        handles.append(
            DECODER_LAYERS[layer_index].register_forward_hook(layer_hook)
        )
    if capture_components:
        for layer_index in COMPONENT_LAYERS:
            def attention_hook(
                _module, _inputs, output, layer_index=layer_index,
            ):
                component_hidden[(layer_index, 'attention_update')] = (
                    output_hidden_tensor(output)[0].detach().cpu().clone()
                )
            def mlp_hook(
                _module, _inputs, output, layer_index=layer_index,
            ):
                component_hidden[(layer_index, 'mlp_update')] = (
                    output_hidden_tensor(output)[0].detach().cpu().clone()
                )
            handles.append(
                ATTENTION_MODULES[layer_index].register_forward_hook(attention_hook)
            )
            handles.append(
                MLP_MODULES[layer_index].register_forward_hook(mlp_hook)
            )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return (
        score_records(prepared, output.logits)[0],
        decoder_hidden, component_hidden,
    )

def patch_module_for_condition(condition):
    layer_index = int(condition['layer_index'])
    patch_point = condition['patch_point']
    if patch_point == 'decoder_output':
        return DECODER_LAYERS[layer_index]
    if patch_point == 'attention_update':
        return ATTENTION_MODULES[layer_index]
    if patch_point == 'mlp_update':
        return MLP_MODULES[layer_index]
    raise KeyError(patch_point)

def forward_patch_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    modules = {id(patch_module_for_condition(row)) for row in conditions}
    assert len(modules) == 1
    module = patch_module_for_condition(conditions[0])
    donors = torch.stack([row['donor_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = donors.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == donor.shape
        modified = hidden.clone()
        modified[mask] = donor[mask]
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = module.register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('donor_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        key = (
            int(condition['layer_index']), condition['patch_point']
        )
        grouped[key].append(condition)

    def run_chunk(chunk):
        size = len(chunk)
        try:
            prepared = prepared_cache.get(size)
            if prepared is None:
                prepared = prepare_examples([target_example] * size)
                prepared_cache[size] = prepared
            return forward_patch_chunk(prepared, chunk)
        except torch.cuda.OutOfMemoryError:
            prepared_cache.pop(size, None)
            try:
                del prepared
            except UnboundLocalError:
                pass
            torch.cuda.empty_cache()
            if size == 1:
                raise
            midpoint = size // 2
            return run_chunk(chunk[:midpoint]) + run_chunk(chunk[midpoint:])

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

print('Decoder/component capture and batched patching ready.')
```

```python
%%cell_progress 10/16 Run alignment, component, and batch smoke tests
cell_stage('Checking fixed outcome token layout')
smoke_pair = pair_manifest[0]
source_example = outcome_by_scene[smoke_pair['source_scene_id']]
target_example = outcome_by_scene[smoke_pair['target_scene_id']]
control_example = outcome_by_scene[
    smoke_pair['same_source_outcome_control_scene_id']
]
source_prepared = prepare_examples([source_example])
target_prepared = prepare_examples([target_example])
control_prepared = prepare_examples([control_example])
masks = assert_aligned(source_prepared, target_prepared)
assert_aligned(control_prepared, target_prepared)
source_score, source_hidden, source_components = forward_capture(
    source_prepared, ALL_DECODER_CAPTURE_LAYERS, capture_components=True
)
target_score, target_hidden, target_components = forward_capture(
    target_prepared, ALL_DECODER_CAPTURE_LAYERS,
    capture_components=True,
)
control_score, control_hidden, control_components = forward_capture(
    control_prepared, ALL_DECODER_CAPTURE_LAYERS, capture_components=True
)
smoke_conditions = [
    {
        'analysis_family': 'smoke', 'condition': 'self_all_tokens',
        'donor_type': 'self_target', 'scope': 'valid_tokens',
        'layer_index': 16, 'patch_point': 'decoder_output',
        'donor_scene_id': smoke_pair['target_scene_id'],
        'donor_hidden': target_hidden[16],
        'patch_mask': masks['valid_tokens'],
    },
    {
        'analysis_family': 'smoke', 'condition': 'source_all_tokens',
        'donor_type': 'matched_source', 'scope': 'valid_tokens',
        'layer_index': 16, 'patch_point': 'decoder_output',
        'donor_scene_id': smoke_pair['source_scene_id'],
        'donor_hidden': source_hidden[16],
        'patch_mask': masks['valid_tokens'],
    },
]
smoke_batch = prepare_examples([target_example] * 2)
smoke_records = forward_patch_chunk(smoke_batch, smoke_conditions)
self_error = max(
    abs(a - b) for a, b in zip(
        smoke_records[0]['candidate_logprobs'],
        target_score['candidate_logprobs'],
    )
)
source_error = max(
    abs(a - b) for a, b in zip(
        smoke_records[1]['candidate_logprobs'],
        source_score['candidate_logprobs'],
    )
)
assert self_error < SMOKE_LOGPROB_TOLERANCE
assert source_error < SMOKE_LOGPROB_TOLERANCE
for layer_index in COMPONENT_LAYERS:
    for component_type in COMPONENT_TYPES:
        tensor = source_components[(layer_index, component_type)]
        assert tensor.shape == source_hidden[layer_index].shape

component_smoke_conditions = []
for component_type, scope in (
    ('attention_update', 'options_tokens'),
    ('mlp_update', 'separator_chat_tokens'),
):
    component_smoke_conditions.append({
        'analysis_family': 'smoke',
        'condition': f'self_{component_type}_{scope}',
        'donor_type': 'self_target', 'scope': scope,
        'layer_index': 16, 'patch_point': component_type,
        'donor_scene_id': smoke_pair['target_scene_id'],
        'donor_hidden': target_components[(16, component_type)],
        'patch_mask': masks[scope],
    })
component_smoke_records = score_conditions(
    target_example, {1: target_prepared}, component_smoke_conditions
)
component_self_error = max(
    abs(value - target_value)
    for record in component_smoke_records
    for value, target_value in zip(
        record['candidate_logprobs'],
        target_score['candidate_logprobs'],
    )
)
assert component_self_error < SMOKE_LOGPROB_TOLERANCE

token_positions = []
for group_name in TOKEN_SCAN_GROUPS:
    for group_rank, token_index in enumerate(
        torch.where(masks[group_name])[0].tolist()
    ):
        token_positions.append({
            'token_group': group_name,
            'group_rank': group_rank,
            'token_index': int(token_index),
            'token_string': str(masks['tokens'][token_index]),
        })
assert len(token_positions) == 40, len(token_positions)

cell_stage('Checking all numeric query alignments')
applicable_tasks = []
for task in LATENT_TASKS:
    examples = build_task_examples(smoke_pair, task)
    if examples is None:
        continue
    task_source, task_target, value_changed = examples
    task_source_prepared = prepare_examples([task_source])
    task_target_prepared = prepare_examples([task_target])
    assert_aligned(task_source_prepared, task_target_prepared)
    applicable_tasks.append((task, value_changed))
print('Applicable smoke tasks:', applicable_tasks)
print('Token scan positions:', len(token_positions))
print('First token descriptors:', token_positions[:8])
print('Self/source ceiling errors:', self_error, source_error)
print('Component self-patch error:', component_self_error)
print('Component shapes and query alignments passed.')
del source_hidden, target_hidden, control_hidden
del source_components, target_components, control_components
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/16 Capture or resume task, token, and component patches
cell_stage('Freezing capture identity and expected pair jobs')

def sha256_file(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return 'sha256:' + digest.hexdigest()

applicable_task_counts = {}
for pair in pair_manifest:
    count = sum(
        build_task_examples(pair, task) is not None for task in LATENT_TASKS
    )
    applicable_task_counts[pair['pair_id']] = count
token_count = len(token_positions)
per_pair_expected = {}
for pair in pair_manifest:
    task_records = (
        applicable_task_counts[pair['pair_id']]
        * len(TASK_LAYERS) * len(TASK_SCOPES)
    )
    token_records = token_count * (
        len(TOKEN_SCAN_LAYERS) + len(TOKEN_CONTROL_LAYERS)
    )
    component_records = (
        len(COMPONENT_LAYERS) * len(COMPONENT_TYPES)
        * len(COMPONENT_SCOPES) * len(DONOR_TYPES)
    )
    per_pair_expected[pair['pair_id']] = (
        task_records + token_records + component_records
    )
EXPECTED_PATCH_RECORDS = sum(per_pair_expected.values())
capture_identity = {
    'schema_version': 'task-token-component-capture-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'dataset_fingerprint': example_config.dataset_fingerprint(),
    'protocol_lock_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'pair_manifest_sha256': sha256_file(manifest_path),
    'confirmatory_decision_sha256': sha256_file(confirmatory_decision_path),
    'discovery_selection_sha256': sha256_file(discovery_selection_path),
    'token_positions': token_positions,
    'expected_pairs': len(pair_manifest),
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'per_pair_expected_records': per_pair_expected,
    'layer_path': LAYER_PATH,
    'token_mixer_kinds': ATTENTION_KINDS,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing capture differs. Use a new output version.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_pair_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_pair_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    pair_id = payload['pair']['pair_id']
    assert pair_id not in completed_pair_ids
    assert len(payload['patch_records']) == per_pair_expected[pair_id]
    completed_pair_ids.add(pair_id)
    completed_records += len(payload['patch_records'])
progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc='Task/token/component conditions', unit='condition',
    dynamic_ncols=True,
)

for pair_index, pair in enumerate(pair_manifest):
    pair_id = pair['pair_id']
    if pair_id in completed_pair_ids:
        continue
    progress.set_postfix_str(f"{pair['edit_type']} | {pair_id}")
    outcome_source = outcome_by_scene[pair['source_scene_id']]
    outcome_target = outcome_by_scene[pair['target_scene_id']]
    outcome_control = outcome_by_scene[
        pair['same_source_outcome_control_scene_id']
    ]
    source_prepared = prepare_examples([outcome_source])
    target_prepared = prepare_examples([outcome_target])
    control_prepared = prepare_examples([outcome_control])
    masks = assert_aligned(source_prepared, target_prepared)
    assert_aligned(control_prepared, target_prepared)
    source_score, source_hidden, source_components = forward_capture(
        source_prepared, ALL_DECODER_CAPTURE_LAYERS,
        capture_components=True,
    )
    target_score, _, _ = forward_capture(
        target_prepared, TASK_LAYERS,
    )
    control_score, control_hidden, control_components = forward_capture(
        control_prepared, ALL_DECODER_CAPTURE_LAYERS,
        capture_components=True,
    )
    baselines = {'outcome': {'source': source_score, 'target': target_score}}
    prepared_cache = {1: target_prepared}
    patch_records = []

    for task in LATENT_TASKS:
        task_examples = build_task_examples(pair, task)
        if task_examples is None:
            continue
        task_source, task_target, value_changed = task_examples
        if task == 'outcome':
            task_source_score = source_score
            task_target_score = target_score
            task_source_hidden = source_hidden
            task_masks = masks
            task_cache = prepared_cache
        else:
            task_source_prepared = prepare_examples([task_source])
            task_target_prepared = prepare_examples([task_target])
            task_masks = assert_aligned(
                task_source_prepared, task_target_prepared
            )
            task_source_score, task_source_hidden, _ = forward_capture(
                task_source_prepared, TASK_LAYERS
            )
            task_target_score, _, _ = forward_capture(
                task_target_prepared, ()
            )
            task_cache = {1: task_target_prepared}
        baselines[task] = {
            'source': task_source_score,
            'target': task_target_score,
            'value_changed': bool(value_changed),
        }
        conditions = []
        for layer_index in TASK_LAYERS:
            for scope in TASK_SCOPES:
                conditions.append({
                    'analysis_family': 'latent_task_route',
                    'condition': f'{task}__matched_source__{scope}',
                    'task': task,
                    'task_value_changed': bool(value_changed),
                    'donor_type': 'matched_source',
                    'scope': scope,
                    'layer_index': int(layer_index),
                    'patch_point': 'decoder_output',
                    'donor_scene_id': pair['source_scene_id'],
                    'donor_hidden': task_source_hidden[layer_index],
                    'patch_mask': task_masks[scope],
                })
        records = score_conditions(task_target, task_cache, conditions)
        patch_records.extend(records)
        progress.update(len(records))
        if task != 'outcome':
            del task_source_hidden, task_cache

    token_conditions = []
    for layer_index in TOKEN_SCAN_LAYERS:
        for descriptor in token_positions:
            mask = torch.zeros_like(masks['valid_tokens'])
            mask[descriptor['token_index']] = True
            token_conditions.append({
                'analysis_family': 'individual_token_scan',
                'condition': (
                    f"matched__L{layer_index}__T{descriptor['token_index']}"
                ),
                'task': 'outcome',
                'task_value_changed': (
                    pair['source_outcome'] != pair['target_outcome']
                ),
                'donor_type': 'matched_source',
                'scope': descriptor['token_group'],
                **descriptor,
                'layer_index': int(layer_index),
                'patch_point': 'decoder_output',
                'donor_scene_id': pair['source_scene_id'],
                'donor_hidden': source_hidden[layer_index],
                'patch_mask': mask,
            })
    for layer_index in TOKEN_CONTROL_LAYERS:
        for descriptor in token_positions:
            mask = torch.zeros_like(masks['valid_tokens'])
            mask[descriptor['token_index']] = True
            token_conditions.append({
                'analysis_family': 'individual_token_scan',
                'condition': (
                    f"control__L{layer_index}__T{descriptor['token_index']}"
                ),
                'task': 'outcome',
                'task_value_changed': (
                    pair['source_outcome'] != pair['target_outcome']
                ),
                'donor_type': 'unrelated_same_source_outcome',
                'scope': descriptor['token_group'],
                **descriptor,
                'layer_index': int(layer_index),
                'patch_point': 'decoder_output',
                'donor_scene_id': pair[
                    'same_source_outcome_control_scene_id'
                ],
                'donor_hidden': control_hidden[layer_index],
                'patch_mask': mask,
            })
    records = score_conditions(
        outcome_target, prepared_cache, token_conditions
    )
    patch_records.extend(records)
    progress.update(len(records))

    component_conditions = []
    for layer_index in COMPONENT_LAYERS:
        for component_type in COMPONENT_TYPES:
            for scope in COMPONENT_SCOPES:
                for donor_type in DONOR_TYPES:
                    donor_components = (
                        source_components
                        if donor_type == 'matched_source'
                        else control_components
                    )
                    donor_scene_id = (
                        pair['source_scene_id']
                        if donor_type == 'matched_source'
                        else pair['same_source_outcome_control_scene_id']
                    )
                    component_conditions.append({
                        'analysis_family': 'component_patch',
                        'condition': (
                            f'{component_type}__{donor_type}__{scope}'
                        ),
                        'task': 'outcome',
                        'task_value_changed': (
                            pair['source_outcome'] != pair['target_outcome']
                        ),
                        'donor_type': donor_type,
                        'scope': scope,
                        'component_type': component_type,
                        'component_module_kind': (
                            ATTENTION_KINDS[layer_index]
                            if component_type == 'attention_update'
                            else 'mlp'
                        ),
                        'layer_index': int(layer_index),
                        'patch_point': component_type,
                        'donor_scene_id': donor_scene_id,
                        'donor_hidden': donor_components[
                            (layer_index, component_type)
                        ],
                        'patch_mask': masks[scope],
                    })
    records = score_conditions(
        outcome_target, prepared_cache, component_conditions
    )
    patch_records.extend(records)
    progress.update(len(records))

    assert len(patch_records) == per_pair_expected[pair_id], (
        pair_id, len(patch_records), per_pair_expected[pair_id]
    )
    payload = {
        'schema_version': 'task-token-component-pair-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'pair': pair,
        'baselines': baselines,
        'outcome_control_baseline': control_score,
        'patch_records': patch_records,
    }
    destination = SHARD_DIR / f'shard_{pair_id}.json'
    write_json_atomic(destination, payload)
    completed_pair_ids.add(pair_id)
    del source_hidden, control_hidden, source_components, control_components
    del prepared_cache, source_prepared, target_prepared, control_prepared
    if (pair_index + 1) % 4 == 0:
        torch.cuda.empty_cache()

progress.close()
assert len(completed_pair_ids) == len(pair_manifest)
actual_records = sum(
    len(json.loads(path.read_text(encoding='utf-8'))['patch_records'])
    for path in SHARD_DIR.glob('shard_pair_*.json')
)
assert actual_records == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_summary.json', {
    'pairs': len(pair_manifest),
    'patch_records': actual_records,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'token_positions': token_count,
})
print('Capture complete:', actual_records, 'patch records')
```

```python
%%cell_progress 12/16 Assemble task-relative recovery metrics
cell_stage('Releasing model and reading pair shards')
del adapter
import gc
gc.collect()
torch.cuda.empty_cache()

patch_rows = []
baseline_rows = []
for path in tqdm(
    sorted(SHARD_DIR.glob('shard_pair_*.json')),
    desc='Reading route shards', unit='pair',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    pair = payload['pair']
    baseline_by_task = payload['baselines']
    for task, values in baseline_by_task.items():
        source = values['source']
        target = values['target']
        baseline_rows.append({
            **pair, 'task': task,
            'task_value_changed': bool(values.get('value_changed',
                pair['source_outcome'] != pair['target_outcome']
            )),
            'source_correct': bool(source['correct']),
            'target_correct': bool(target['correct']),
            'both_correct': bool(source['correct'] and target['correct']),
            'source_predicted_semantic': source['predicted_semantic'],
            'target_predicted_semantic': target['predicted_semantic'],
        })
    for record in payload['patch_records']:
        task = record['task']
        source = baseline_by_task[task]['source']
        target = baseline_by_task[task]['target']
        source_vector = np.asarray(
            source['candidate_logprobs'], dtype=np.float64
        )
        target_vector = np.asarray(
            target['candidate_logprobs'], dtype=np.float64
        )
        patched_vector = np.asarray(
            record['candidate_logprobs'], dtype=np.float64
        )
        source_vector -= source_vector.mean()
        target_vector -= target_vector.mean()
        patched_vector -= patched_vector.mean()
        natural_delta = source_vector - target_vector
        denominator = float(np.dot(natural_delta, natural_delta))
        natural_norm = float(np.linalg.norm(natural_delta))
        patch_delta = patched_vector - target_vector
        recovery = None
        if natural_norm >= NATURAL_EFFECT_EPS:
            recovery = float(
                np.dot(patch_delta, natural_delta) / denominator
            )
        patch_rows.append({
            **pair, **record,
            'natural_effect_norm': natural_norm,
            'valid_recovery': recovery is not None,
            'recovery': recovery,
            'patch_effect_norm': float(np.linalg.norm(patch_delta)),
            'source_correct': bool(source['correct']),
            'target_correct': bool(target['correct']),
            'both_correct': bool(source['correct'] and target['correct']),
        })
patch_frame = pd.DataFrame(patch_rows)
baseline_frame = pd.DataFrame(baseline_rows)
assert len(patch_frame) == EXPECTED_PATCH_RECORDS
patch_frame.to_csv(ANALYSIS_DIR / 'patch_records.csv', index=False)
baseline_frame.to_csv(ANALYSIS_DIR / 'task_baselines.csv', index=False)
print('Patch records:', len(patch_frame))
print('Valid recovery by family:')
display(
    patch_frame.groupby('analysis_family')['valid_recovery'].mean()
)
print('Task applicability and both-correct rates:')
display(baseline_frame.groupby('task').agg(
    pairs=('pair_id', 'nunique'),
    value_changed=('task_value_changed', 'mean'),
    both_correct=('both_correct', 'mean'),
))
```

```python
%%cell_progress 13/16 Analyze latent-task routing and relevance
cell_stage('Bootstrapping task-specific route curves')
rng = np.random.default_rng(SEED + 1000)

def bootstrap_values(values, replicates=BOOTSTRAP_REPLICATES):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {'n': 0, 'mean': np.nan, 'ci_low': np.nan, 'ci_high': np.nan}
    draws = rng.choice(
        values, size=(replicates, len(values)), replace=True
    ).mean(axis=1)
    return {
        'n': int(len(values)), 'mean': float(values.mean()),
        'ci_low': float(np.quantile(draws, 0.025)),
        'ci_high': float(np.quantile(draws, 0.975)),
    }

task_frame = patch_frame[
    (patch_frame['analysis_family'] == 'latent_task_route')
    & patch_frame['valid_recovery']
].copy()
task_summary_rows = []
for keys, group in task_frame.groupby(
    ['task', 'task_value_changed', 'scope', 'layer_index'], sort=True
):
    values = group.groupby('pair_id')['recovery'].mean().to_numpy()
    task_summary_rows.append({
        'task': keys[0], 'task_value_changed': bool(keys[1]),
        'scope': keys[2], 'layer_index': int(keys[3]),
        **bootstrap_values(values),
    })
task_summary = pd.DataFrame(task_summary_rows)
task_summary.to_csv(ANALYSIS_DIR / 'task_route_summary.csv', index=False)

relevance_rows = []
for keys, group in task_frame.groupby(
    ['task', 'scope', 'layer_index'], sort=True
):
    changed = group[group['task_value_changed']].groupby(
        'pair_id'
    )['recovery'].mean()
    unchanged = group[~group['task_value_changed']].groupby(
        'pair_id'
    )['recovery'].mean()
    stats = bootstrap_values(changed.to_numpy())
    relevance_rows.append({
        'task': keys[0], 'scope': keys[1],
        'layer_index': int(keys[2]),
        'changed_n': int(len(changed)),
        'unchanged_n': int(len(unchanged)),
        'changed_mean': stats['mean'],
        'changed_ci_low': stats['ci_low'],
        'changed_ci_high': stats['ci_high'],
        'unchanged_mean': float(unchanged.mean()) if len(unchanged) else np.nan,
        'descriptive_changed_minus_unchanged': (
            float(changed.mean() - unchanged.mean())
            if len(changed) and len(unchanged) else np.nan
        ),
    })
relevance_frame = pd.DataFrame(relevance_rows)
relevance_frame.to_csv(ANALYSIS_DIR / 'task_relevance.csv', index=False)
display(task_summary.head(20))
```

```python
%%cell_progress 14/16 Analyze individual tokens and attention-versus-MLP updates
cell_stage('Finding matched and pair-specific token positions')
token_frame = patch_frame[
    (patch_frame['analysis_family'] == 'individual_token_scan')
    & patch_frame['valid_recovery']
].copy()
token_rows = []
for keys, group in token_frame.groupby(
    [
        'donor_type', 'layer_index', 'token_group',
        'group_rank', 'token_index', 'token_string',
    ],
    sort=True,
):
    values = group.groupby('pair_id')['recovery'].mean().to_numpy()
    token_rows.append({
        'donor_type': keys[0], 'layer_index': int(keys[1]),
        'token_group': keys[2], 'group_rank': int(keys[3]),
        'token_index': int(keys[4]), 'token_string': keys[5],
        **bootstrap_values(values),
    })
token_summary = pd.DataFrame(token_rows)
token_summary.to_csv(ANALYSIS_DIR / 'individual_token_summary.csv', index=False)

token_gap_rows = []
for layer_index in TOKEN_CONTROL_LAYERS:
    subset = token_frame[token_frame['layer_index'] == layer_index]
    pivot = subset.pivot_table(
        index=['pair_id', 'token_index'], columns='donor_type',
        values='recovery', aggfunc='first',
    )
    assert set(DONOR_TYPES) <= set(pivot.columns)
    difference = (
        pivot['matched_source']
        - pivot['unrelated_same_source_outcome']
    )
    for token_index, values in difference.groupby('token_index'):
        descriptor = next(
            row for row in token_positions
            if row['token_index'] == int(token_index)
        )
        token_gap_rows.append({
            'layer_index': int(layer_index), **descriptor,
            **bootstrap_values(values.to_numpy()),
        })
token_gap_frame = pd.DataFrame(token_gap_rows)
token_gap_frame.to_csv(
    ANALYSIS_DIR / 'individual_token_control_gaps.csv', index=False
)

component_frame = patch_frame[
    (patch_frame['analysis_family'] == 'component_patch')
    & patch_frame['valid_recovery']
].copy()
component_rows = []
for keys, group in component_frame.groupby(
    ['donor_type', 'component_type', 'scope', 'layer_index'], sort=True
):
    values = group.groupby('pair_id')['recovery'].mean().to_numpy()
    component_rows.append({
        'donor_type': keys[0], 'component_type': keys[1],
        'scope': keys[2], 'layer_index': int(keys[3]),
        **bootstrap_values(values),
    })
component_summary = pd.DataFrame(component_rows)
component_summary.to_csv(
    ANALYSIS_DIR / 'component_summary.csv', index=False
)

component_difference_rows = []
matched_components = component_frame[
    component_frame['donor_type'] == 'matched_source'
]
for keys, group in matched_components.groupby(['scope', 'layer_index']):
    pivot = group.pivot_table(
        index='pair_id', columns='component_type',
        values='recovery', aggfunc='first',
    )
    difference = (
        pivot['attention_update'] - pivot['mlp_update']
    ).dropna()
    component_difference_rows.append({
        'scope': keys[0], 'layer_index': int(keys[1]),
        'contrast': 'attention_minus_mlp',
        **bootstrap_values(difference.to_numpy()),
    })
component_difference = pd.DataFrame(component_difference_rows)
component_difference.to_csv(
    ANALYSIS_DIR / 'attention_minus_mlp.csv', index=False
)

component_control_rows = []
for keys, group in component_frame.groupby(
    ['component_type', 'scope', 'layer_index'], sort=True
):
    pivot = group.pivot_table(
        index='pair_id', columns='donor_type',
        values='recovery', aggfunc='first',
    )
    assert set(DONOR_TYPES) <= set(pivot.columns)
    difference = (
        pivot['matched_source']
        - pivot['unrelated_same_source_outcome']
    ).dropna()
    component_control_rows.append({
        'component_type': keys[0], 'scope': keys[1],
        'layer_index': int(keys[2]),
        'contrast': 'matched_minus_unrelated_same_outcome',
        **bootstrap_values(difference.to_numpy()),
    })
component_control_gap = pd.DataFrame(component_control_rows)
component_control_gap.to_csv(
    ANALYSIS_DIR / 'component_control_gaps.csv', index=False
)

correctness_rows = []
for keys, group in patch_frame[
    patch_frame['analysis_family'].isin([
        'individual_token_scan', 'component_patch'
    ])
    & patch_frame['valid_recovery']
    & (patch_frame['donor_type'] == 'matched_source')
].groupby(
    ['analysis_family', 'both_correct', 'scope', 'layer_index'],
    sort=True,
):
    values = group.groupby('pair_id')['recovery'].mean().to_numpy()
    correctness_rows.append({
        'analysis_family': keys[0], 'both_correct': bool(keys[1]),
        'scope': keys[2], 'layer_index': int(keys[3]),
        **bootstrap_values(values),
    })
correctness_summary = pd.DataFrame(correctness_rows)
correctness_summary.to_csv(
    ANALYSIS_DIR / 'correctness_stratified_summary.csv', index=False
)
print('Top pair-specific individual tokens:')
display(token_gap_frame.sort_values('ci_low', ascending=False).head(15))
print('Attention minus MLP contrasts:')
display(component_difference)
print('Pair-specific component effects:')
display(component_control_gap)
```

```python
%%cell_progress 16/16 Write route-decomposition report and next decision
cell_stage('Summarizing what is task-specific and which component to inspect')
report_lines = [
    '# Task, token, and component routing report',
    '',
    f'- Model: `{MODEL_ID}`',
    f'- Revision: `{MODEL_REVISION}`',
    f'- Pairs: {len(pair_manifest)} ({PAIRS_PER_EDIT} per edit)',
    f'- Fixed outcome variant: `{FIXED_VARIANT}`',
    '',
    '## Latent-query routing',
    '',
]
for task in LATENT_TASKS:
    rows = task_summary[
        (task_summary['task'] == task)
        & task_summary['task_value_changed']
    ]
    if rows.empty:
        report_lines.append(f'- {task}: no changed-value pairs available.')
        continue
    best = rows.loc[rows['mean'].idxmax()]
    report_lines.append(
        f"- {task}: peak changed-value recovery {best['mean']:.3f} "
        f"[{best['ci_low']:.3f}, {best['ci_high']:.3f}] at layer "
        f"{int(best['layer_index'])}, scope `{best['scope']}`, n={int(best['n'])}."
    )

report_lines.extend([
    '',
    'A route that appears with similar timing for units, distances, side torque, net '
    'torque, and outcome is a generic multimodal answer-selection scaffold. Selective '
    'recovery only when a task-relevant latent changes is evidence that physical '
    'content, rather than arbitrary scene identity, is being transported.',
    '',
    '## Individual-token localization',
    '',
])
for _, row in token_gap_frame.sort_values(
    'ci_low', ascending=False
).head(10).iterrows():
    report_lines.append(
        f"- layer {int(row['layer_index'])}, token {int(row['token_index'])} "
        f"`{row['token_string']}` ({row['token_group']} rank "
        f"{int(row['group_rank'])}): matched-control gap {row['mean']:.3f} "
        f"[{row['ci_low']:.3f}, {row['ci_high']:.3f}]."
    )

report_lines.extend([
    '',
    '## Attention versus MLP',
    '',
])
for _, row in component_difference.iterrows():
    winner = (
        'attention' if row['ci_low'] > 0
        else ('MLP' if row['ci_high'] < 0 else 'inconclusive')
    )
    report_lines.append(
        f"- layer {int(row['layer_index'])}, `{row['scope']}`: "
        f"attention−MLP {row['mean']:.3f} "
        f"[{row['ci_low']:.3f}, {row['ci_high']:.3f}] → {winner}."
    )

attention_candidates = component_difference[
    component_difference['ci_low'] > 0
].merge(
    component_control_gap[
        (component_control_gap['component_type'] == 'attention_update')
        & (component_control_gap['ci_low'] > 0)
    ][['scope', 'layer_index', 'mean', 'ci_low', 'ci_high']],
    on=['scope', 'layer_index'], how='inner',
    suffixes=('_attention_minus_mlp', '_matched_control'),
)
mlp_candidates = component_difference[
    component_difference['ci_high'] < 0
].merge(
    component_control_gap[
        (component_control_gap['component_type'] == 'mlp_update')
        & (component_control_gap['ci_low'] > 0)
    ][['scope', 'layer_index', 'mean', 'ci_low', 'ci_high']],
    on=['scope', 'layer_index'], how='inner',
    suffixes=('_attention_minus_mlp', '_matched_control'),
)
significant_attention = attention_candidates
significant_mlp = mlp_candidates
significant_attention.to_csv(
    ANALYSIS_DIR / 'head_search_candidates.csv', index=False
)
significant_mlp.to_csv(
    ANALYSIS_DIR / 'mlp_search_candidates.csv', index=False
)
pair_specific_tokens = token_gap_frame[token_gap_frame['ci_low'] > 0]
report_lines.extend([
    '',
    '## Next experiment decision',
    '',
    f'- Pair-specific individual token positions: {len(pair_specific_tokens)}.',
    f'- Attention-dominant layer/scope tests: {len(significant_attention)}.',
    f'- MLP-dominant layer/scope tests: {len(significant_mlp)}.',
    '',
])
if len(significant_attention):
    report_lines.append(
        'Proceed to attention-head path patching only at the attention-dominant '
        'layer/scope combinations and the pair-specific token positions listed above.'
    )
elif len(significant_mlp):
    report_lines.append(
        'Prioritize MLP-neuron/subspace analysis at the MLP-dominant layers; a broad '
        'head search is not justified by these results.'
    )
else:
    report_lines.append(
        'Do not search heads or neurons yet. The causal update is distributed or the '
        'component intervention lacks power; first inspect correctness strata and '
        'matched-control token effects.'
    )
report_lines.extend([
    '',
    '## Limits',
    '',
    protocol_identity['interpretation_limit'],
    '',
])
report_path = ANALYSIS_DIR / 'task_token_component_report.md'
report_path.write_text('\n'.join(report_lines), encoding='utf-8')
decision = {
    'model_id': MODEL_ID,
    'pairs': len(pair_manifest),
    'pair_specific_token_count': int(len(pair_specific_tokens)),
    'attention_dominant_tests': int(len(significant_attention)),
    'mlp_dominant_tests': int(len(significant_mlp)),
    'report_path': str(report_path),
    'recommended_next': (
        'attention_head_path_patching'
        if len(significant_attention) else
        ('mlp_subspace_analysis' if len(significant_mlp)
         else 'correctness_stratified_route_diagnosis')
    ),
}
write_json_atomic(
    ANALYSIS_DIR / 'task_token_component_decision.json', decision
)
print(report_path.read_text(encoding='utf-8'))
print('Completed task/token/component routing.')
```
