```python
stage('Loading and hashing frozen source artifacts')
import hashlib
import json
import math
from datetime import datetime, timezone

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

SOURCE_PATHS = {
    'orthogonal_manifest': ORTHOGONAL_ROOT / 'orthogonal_triplet_manifest.jsonl',
    'orthogonal_manifest_lock': ORTHOGONAL_ROOT / 'manifest_lock.json',
    'orthogonal_cross_synthesis': (
        ORTHOGONAL_ROOT / 'cross_model_analysis' / 'cross_model_synthesis.json'
    ),
    'factorial_manifest': FACTORIAL_ROOT / 'factorial_triplet_manifest.jsonl',
}
for model in MODEL_SPECS:
    SOURCE_PATHS[f'orthogonal_{model}_paired'] = (
        ORTHOGONAL_ANALYSIS[model] / 'paired_triplet_metrics.csv'
    )
    SOURCE_PATHS[f'orthogonal_{model}_patch'] = (
        ORTHOGONAL_ANALYSIS[model] / 'patch_level_metrics.csv'
    )
    SOURCE_PATHS[f'orthogonal_{model}_decision'] = (
        ORTHOGONAL_ANALYSIS[model] / 'orthogonal_value_binding_decision.json'
    )
    SOURCE_PATHS[f'factorial_{model}_paired'] = (
        FACTORIAL_ANALYSIS[model] / 'paired_triplet_metrics.csv'
    )
missing = [str(path) for path in SOURCE_PATHS.values() if not path.is_file()]
assert not missing, 'Missing required artifacts:\n' + '\n'.join(missing)

from visualtorque_checks import sha256_file

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def as_bool(series):
    if series.dtype == bool:
        return series.astype(bool)
    mapped = series.astype(str).str.lower().map({
        'true': True, 'false': False, '1': True, '0': False,
    })
    assert mapped.notna().all(), sorted(series.astype(str).unique())
    return mapped.astype(bool)

SEED = 20260908
BOOTSTRAP_REPLICATES = 10000
SIGN_FLIP_REPLICATES = 20000
ALPHA = 0.05
FOCAL_SITES = ('middle_text', 'middle_options')
PRIMARY_EQUIVALENCE_FRACTION = 0.25
EQUIVALENCE_FRACTIONS = (0.10, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40)

source_hashes = {key: sha256_file(path) for key, path in SOURCE_PATHS.items()}
protocol_identity = {
    'schema_version': 'orthogonal-binding-robustness-audit-1.0',
    'status': (
        'post-confirmatory robustness audit; the hypotheses were informed by '
        'earlier results and are not an independent preregistration'
    ),
    'source_hashes': source_hashes,
    'population': '48 scalar-orthogonal, previous-factorial-scene-excluded triplets',
    'physical_strata': [
        'changed side', 'factor direction', 'outcome change',
        'small versus large torque change', 'other-prior-scene overlap',
    ],
    'correctness_boundaries': [
        'base correct', 'relevant source correct',
        'either source correct', 'both sources correct',
        'common cross-model base correct',
    ],
    'equivalence_fractions': list(EQUIVALENCE_FRACTIONS),
    'standardization': 'divide each model/task effect by median absolute natural margin',
    'replication': (
        'compare effect signs and intervals with the scene-disjoint preceding '
        'same-manifest factorial experiment'
    ),
    'bootstrap_unit': 'triplet',
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'sign_flip_replicates': SIGN_FLIP_REPLICATES,
    'interpretation_limit': (
        'grouped decoder-output transport under fixed multiple-choice prompts; '
        'not architecture attribution or a unique circuit'
    ),
    'seed': SEED,
}
PROTOCOL_PATH = OUTPUT_DIR / 'audit_protocol.json'
if PROTOCOL_PATH.is_file():
    previous = json.loads(PROTOCOL_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing audit uses different sources or rules. Use a new output version.'
    )
else:
    write_json_atomic(PROTOCOL_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Protocol:', PROTOCOL_PATH)
```

## 2. Exact support and derived metrics


```python
stage('Validating exact support and deriving normalized metrics')

manifest_rows = read_jsonl(SOURCE_PATHS['orthogonal_manifest'])
manifest = pd.DataFrame([
    {key: value for key, value in row.items() if key != 'task_values'}
    for row in manifest_rows
])
manifest['triplet_id'] = manifest['triplet_id'].astype(str)
for column in (
    'outcome_changed', 'previous_factorial_scene_overlap',
    'any_prior_scene_overlap',
):
    manifest[column] = as_bool(manifest[column])
assert len(manifest) == manifest['triplet_id'].nunique() == 48
assert not manifest['previous_factorial_scene_overlap'].any()
manifest['delta_magnitude'] = manifest['delta_side_torque'].abs().astype(int)
manifest['delta_magnitude_group'] = np.where(
    manifest['delta_magnitude'] <= 4, 'small_3_4', 'large_6_12'
)

paired_frames = []
patch_frames = []
decisions = {}
for model in MODEL_SPECS:
    paired_model = pd.read_csv(SOURCE_PATHS[f'orthogonal_{model}_paired'])
    patch_model = pd.read_csv(SOURCE_PATHS[f'orthogonal_{model}_patch'])
    paired_model['model'] = model
    patch_model['model'] = model
    for frame in (paired_model, patch_model):
        frame['triplet_id'] = frame['triplet_id'].astype(str)
        for column in (
            'estimable', 'outcome_changed', 'any_prior_scene_overlap',
            'base_correct', 'weight_source_correct',
            'distance_source_correct', 'all_three_correct',
        ):
            if column in frame:
                frame[column] = as_bool(frame[column])
    assert set(paired_model['triplet_id']) == set(manifest['triplet_id'])
    assert set(patch_model['triplet_id']) == set(manifest['triplet_id'])
    assert len(paired_model) == 960 and len(patch_model) == 1920
    assert not paired_model.duplicated([
        'triplet_id', 'task', 'site_label'
    ]).any()
    assert not patch_model.duplicated([
        'triplet_id', 'task', 'site_label', 'donor_type'
    ]).any()
    decisions[model] = json.loads(
        SOURCE_PATHS[f'orthogonal_{model}_decision'].read_text(encoding='utf-8')
    )
    assert decisions[model]['manifest_sha256'] == source_hashes['orthogonal_manifest']
    paired_frames.append(paired_model)
    patch_frames.append(patch_model)

paired = pd.concat(paired_frames, ignore_index=True)
patch = pd.concat(patch_frames, ignore_index=True)
manifest_columns = [
    'triplet_id', 'base_units', 'changed_units',
    'base_distance', 'changed_distance', 'unit_step', 'distance_step',
    'delta_magnitude', 'delta_magnitude_group',
    'previous_factorial_scene_overlap',
]
paired = paired.merge(
    manifest[manifest_columns], on='triplet_id', how='left', validate='many_to_one'
)
patch = patch.merge(
    manifest[manifest_columns], on='triplet_id', how='left', validate='many_to_one'
)
assert paired.groupby(['model', 'task', 'site_label']).size().eq(48).all()

scale = (
    paired[paired['estimable']]
    .groupby(['model', 'task'])['natural_margin']
    .apply(lambda values: float(np.median(np.abs(values))))
    .rename('task_natural_scale')
    .reset_index()
)
assert (scale['task_natural_scale'] > 0).all()
paired = paired.merge(scale, on=['model', 'task'], validate='many_to_one')
effect_columns = (
    'weight_effect', 'distance_effect', 'mean_donor_effect',
    'donor_effect_difference', 'factor_selectivity',
    'relevant_route_effect', 'irrelevant_route_effect',
    'cross_value_leakage', 'abs_donor_gap',
)
for column in effect_columns:
    paired[f'std_{column}'] = paired[column] / paired['task_natural_scale']
stabilizer = PRIMARY_EQUIVALENCE_FRACTION * paired['task_natural_scale']
paired['donor_dominance_index'] = (
    paired['donor_effect_difference']
    / (
        paired['weight_effect'].abs()
        + paired['distance_effect'].abs()
        + stabilizer
    )
)

paired['both_sources_correct'] = (
    paired['weight_source_correct'] & paired['distance_source_correct']
)
paired['either_source_correct'] = (
    paired['weight_source_correct'] | paired['distance_source_correct']
)
paired['relevant_source_correct'] = np.where(
    paired['task'] == 'side_distance',
    paired['distance_source_correct'],
    paired['weight_source_correct'],
).astype(bool)

cross_synthesis = json.loads(
    SOURCE_PATHS['orthogonal_cross_synthesis'].read_text(encoding='utf-8')
)
assert cross_synthesis['exact_shared_triplets'] == 48
assert cross_synthesis['support_status'] == 'exact_same_stimulus_support_complete'

paired.to_csv(TABLE_DIR / 'orthogonal_derived_metrics.csv', index=False)
print('Exact support: 48 triplets × 5 tasks × 4 sites × 2 models')
display(manifest.groupby([
    'changed_side', 'factor_direction', 'outcome_changed',
    'delta_magnitude_group',
]).size().rename('n').reset_index())
display(scale)
```

```python
stage('Defining deterministic paired inference')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def paired_summary(values, label):
    values = pd.to_numeric(pd.Series(values), errors='coerce').to_numpy(float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    indices = rng.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    bootstrap = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice(
        (-1.0, 1.0), size=(SIGN_FLIP_REPLICATES, len(values))
    )
    null = (signs * values).mean(axis=1)
    p_value = float(
        (1 + np.sum(np.abs(null) >= abs(observed)))
        / (SIGN_FLIP_REPLICATES + 1)
    )
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(bootstrap, ALPHA / 2)),
        'ci_high': float(np.quantile(bootstrap, 1 - ALPHA / 2)),
        'sign_flip_p': p_value,
    }

def independent_difference(left, right, label):
    left = pd.to_numeric(pd.Series(left), errors='coerce').dropna().to_numpy(float)
    right = pd.to_numeric(pd.Series(right), errors='coerce').dropna().to_numpy(float)
    if not len(left) or not len(right):
        return {
            'n_left': len(left), 'n_right': len(right),
            'mean': np.nan, 'ci_low': np.nan, 'ci_high': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    li = rng.integers(0, len(left), size=(BOOTSTRAP_REPLICATES, len(left)))
    ri = rng.integers(0, len(right), size=(BOOTSTRAP_REPLICATES, len(right)))
    boot = left[li].mean(axis=1) - right[ri].mean(axis=1)
    return {
        'n_left': int(len(left)), 'n_right': int(len(right)),
        'mean': float(left.mean() - right.mean()),
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

def exact_mcnemar(left, right):
    left = np.asarray(left, dtype=bool)
    right = np.asarray(right, dtype=bool)
    left_only = int(np.sum(left & ~right))
    right_only = int(np.sum(~left & right))
    discordant = left_only + right_only
    if discordant == 0:
        return left_only, right_only, 1.0
    tail = sum(
        math.comb(discordant, value)
        for value in range(0, min(left_only, right_only) + 1)
    ) / (2 ** discordant)
    return left_only, right_only, min(1.0, 2 * tail)

print('Inference utilities ready.')
```

## 3. Within-model and exact cross-model robustness


```python
stage('Testing within-model binding, leakage, and torque routes')

within_plan = []
for model in MODEL_SPECS:
    for site in FOCAL_SITES:
        for task in ('side_units', 'side_distance'):
            within_plan.extend([
                (model, 'factor_binding', task, site, 'factor_selectivity'),
                (model, 'factor_binding', task, site, 'relevant_route_effect'),
                (model, 'factor_binding', task, site, 'irrelevant_route_effect'),
                (model, 'cross_value_leakage', task, site, 'cross_value_leakage'),
            ])
        for metric in (
            'weight_effect', 'distance_effect', 'mean_donor_effect',
            'donor_effect_difference', 'displacement_cosine',
        ):
            within_plan.append(
                (model, 'side_torque_route', 'side_torque', site, metric)
            )
        within_plan.append(
            (model, 'downstream_boundary', 'net_torque', site, 'mean_donor_effect')
        )
    within_plan.append(
        (model, 'downstream_boundary', 'outcome', 'late_answer', 'mean_donor_effect')
    )

within_rows = []
for model, family, task, site, metric in within_plan:
    subset = paired[
        (paired['model'] == model)
        & (paired['task'] == task)
        & (paired['site_label'] == site)
        & paired['estimable']
    ]
    within_rows.append({
        'model': model, 'family': family, 'task': task,
        'site_label': site, 'metric': metric,
        **paired_summary(subset[metric], f'within:{model}:{task}:{site}:{metric}'),
    })
within_summary = pd.DataFrame(within_rows)
within_summary['q_value'] = np.nan
for _, indices in within_summary.groupby('family').groups.items():
    within_summary.loc[indices, 'q_value'] = bh_adjust(
        within_summary.loc[indices, 'sign_flip_p']
    )

within_summary['equivalence_bound'] = np.nan
within_summary['equivalent'] = False
for model in MODEL_SPECS:
    for task in ('side_torque', 'net_torque', 'outcome'):
        sites = FOCAL_SITES if task != 'outcome' else ('late_answer',)
        for site in sites:
            subset = paired[
                (paired['model'] == model) & (paired['task'] == task)
                & (paired['site_label'] == site) & paired['estimable']
            ]
            if not len(subset):
                continue
            bound = PRIMARY_EQUIVALENCE_FRACTION * float(
                subset['task_natural_scale'].iloc[0]
            )
            difference = paired_summary(
                subset['donor_effect_difference'],
                f'equivalence:{model}:{task}:{site}',
            )
            mask = (
                (within_summary['model'] == model)
                & (within_summary['task'] == task)
                & (within_summary['site_label'] == site)
                & (within_summary['metric'] == 'donor_effect_difference')
            )
            if mask.any():
                within_summary.loc[mask, 'equivalence_bound'] = bound
                within_summary.loc[mask, 'equivalent'] = (
                    difference['ci_low'] > -bound
                    and difference['ci_high'] < bound
                )

standardized_within_rows = []
for model, family, task, site, metric in within_plan:
    std_metric = f'std_{metric}'
    if std_metric not in paired.columns:
        continue
    subset = paired[
        (paired['model'] == model) & (paired['task'] == task)
        & (paired['site_label'] == site) & paired['estimable']
    ]
    standardized_within_rows.append({
        'model': model, 'family': family, 'task': task,
        'site_label': site, 'metric': std_metric,
        **paired_summary(
            subset[std_metric], f'within_std:{model}:{task}:{site}:{metric}'
        ),
    })
standardized_within = pd.DataFrame(standardized_within_rows)
within_summary.to_csv(TABLE_DIR / 'within_model_primary.csv', index=False)
standardized_within.to_csv(
    TABLE_DIR / 'within_model_standardized.csv', index=False
)
display(within_summary[
    within_summary['site_label'].isin(FOCAL_SITES)
].sort_values(['model', 'task', 'site_label', 'metric']))
```

```python
stage('Testing exact-pair raw and standardized cross-model contrasts')

join_keys = ['triplet_id', 'task', 'site_label']
qwen = paired[paired['model'] == 'qwen'].copy()
gemma = paired[paired['model'] == 'gemma'].copy()
value_columns = [
    'factor_selectivity', 'cross_value_leakage',
    'donor_effect_difference', 'mean_donor_effect',
    'std_factor_selectivity', 'std_cross_value_leakage',
    'std_donor_effect_difference', 'std_mean_donor_effect',
    'donor_dominance_index',
    'changed_side', 'factor_direction', 'outcome_changed',
    'delta_magnitude_group', 'any_prior_scene_overlap',
    'base_correct', 'weight_source_correct', 'distance_source_correct',
    'either_source_correct', 'both_sources_correct',
]
exact = qwen[join_keys + value_columns].merge(
    gemma[join_keys + value_columns],
    on=join_keys, how='inner', validate='one_to_one',
    suffixes=('_qwen', '_gemma'),
)
assert len(exact) == 960
for column in (
    'changed_side', 'factor_direction', 'outcome_changed',
    'delta_magnitude_group', 'any_prior_scene_overlap',
):
    assert (exact[f'{column}_qwen'] == exact[f'{column}_gemma']).all()
    exact[column] = exact[f'{column}_qwen']

cross_plan = []
for site in FOCAL_SITES:
    cross_plan.extend([
        ('side_units', site, 'factor_selectivity'),
        ('side_units', site, 'cross_value_leakage'),
        ('side_distance', site, 'factor_selectivity'),
        ('side_distance', site, 'cross_value_leakage'),
        ('side_torque', site, 'donor_effect_difference'),
        ('side_torque', site, 'mean_donor_effect'),
    ])

cross_rows = []
standardized_rows = []
for task, site, metric in cross_plan:
    subset = exact[
        (exact['task'] == task) & (exact['site_label'] == site)
    ]
    raw_difference = subset[f'{metric}_qwen'] - subset[f'{metric}_gemma']
    cross_rows.append({
        'task': task, 'site_label': site, 'metric': metric,
        'contrast': 'qwen_minus_gemma',
        **paired_summary(raw_difference, f'cross:{task}:{site}:{metric}'),
    })
    std_metric = f'std_{metric}'
    standardized_difference = (
        subset[f'{std_metric}_qwen'] - subset[f'{std_metric}_gemma']
    )
    standardized_rows.append({
        'task': task, 'site_label': site, 'metric': std_metric,
        'contrast': 'qwen_minus_gemma',
        **paired_summary(
            standardized_difference, f'cross_std:{task}:{site}:{metric}'
        ),
    })

for site in FOCAL_SITES:
    subset = exact[
        (exact['task'] == 'side_torque') & (exact['site_label'] == site)
    ]
    difference = (
        subset['donor_dominance_index_qwen']
        - subset['donor_dominance_index_gemma']
    )
    standardized_rows.append({
        'task': 'side_torque', 'site_label': site,
        'metric': 'donor_dominance_index',
        'contrast': 'qwen_minus_gemma',
        **paired_summary(difference, f'cross_dominance:{site}'),
    })

cross_summary = pd.DataFrame(cross_rows)
cross_summary['q_value'] = bh_adjust(cross_summary['sign_flip_p'])
standardized_cross = pd.DataFrame(standardized_rows)
standardized_cross['q_value'] = bh_adjust(
    standardized_cross['sign_flip_p']
)
cross_summary.to_csv(TABLE_DIR / 'cross_model_primary.csv', index=False)
standardized_cross.to_csv(
    TABLE_DIR / 'cross_model_standardized.csv', index=False
)
exact.to_csv(TABLE_DIR / 'exact_cross_model_rows.csv', index=False)
display(cross_summary.sort_values(['task', 'site_label', 'metric']))
display(standardized_cross.sort_values(['task', 'site_label', 'metric']))
```

## 4. Physical, competence, equivalence, and replication boundaries


```python
stage('Auditing physical strata and heterogeneity')

stratum_specs = [
    ('changed_side', ('left', 'right')),
    ('factor_direction', ('decrease', 'increase')),
    ('outcome_changed', (False, True)),
    ('delta_magnitude_group', ('small_3_4', 'large_6_12')),
]
if manifest['any_prior_scene_overlap'].nunique() == 2:
    stratum_specs.append(('any_prior_scene_overlap', (False, True)))
within_stratum_specs = (
    ('qwen', 'side_torque', 'donor_effect_difference'),
    ('gemma', 'side_torque', 'donor_effect_difference'),
    ('qwen', 'side_distance', 'cross_value_leakage'),
    ('gemma', 'side_units', 'cross_value_leakage'),
)
stratum_rows = []
for model, task, metric in within_stratum_specs:
    for site in FOCAL_SITES:
        base = paired[
            (paired['model'] == model) & (paired['task'] == task)
            & (paired['site_label'] == site) & paired['estimable']
        ]
        for column, levels in stratum_specs:
            values_by_level = {}
            for level in levels:
                subset = base[base[column] == level]
                values_by_level[level] = subset[metric]
                stratum_rows.append({
                    'analysis': 'within', 'model_or_contrast': model,
                    'task': task, 'site_label': site, 'metric': metric,
                    'stratum': column, 'level': str(level),
                    'row_type': 'level',
                    **paired_summary(
                        subset[metric],
                        f'stratum:{model}:{task}:{site}:{metric}:{column}:{level}',
                    ),
                })
            difference = independent_difference(
                values_by_level[levels[0]], values_by_level[levels[1]],
                f'heterogeneity:{model}:{task}:{site}:{metric}:{column}',
            )
            stratum_rows.append({
                'analysis': 'within', 'model_or_contrast': model,
                'task': task, 'site_label': site, 'metric': metric,
                'stratum': column,
                'level': f'{levels[0]}_minus_{levels[1]}',
                'row_type': 'heterogeneity',
                'n': difference['n_left'] + difference['n_right'],
                'mean': difference['mean'],
                'ci_low': difference['ci_low'],
                'ci_high': difference['ci_high'],
                'sign_flip_p': np.nan,
            })

for site in FOCAL_SITES:
    base = exact[
        (exact['task'] == 'side_torque') & (exact['site_label'] == site)
    ].copy()
    base['cross_route_difference'] = (
        base['donor_effect_difference_qwen']
        - base['donor_effect_difference_gemma']
    )
    for column, levels in stratum_specs:
        values_by_level = {}
        for level in levels:
            subset = base[base[column] == level]
            values_by_level[level] = subset['cross_route_difference']
            stratum_rows.append({
                'analysis': 'cross', 'model_or_contrast': 'qwen_minus_gemma',
                'task': 'side_torque', 'site_label': site,
                'metric': 'donor_effect_difference',
                'stratum': column, 'level': str(level),
                'row_type': 'level',
                **paired_summary(
                    subset['cross_route_difference'],
                    f'cross_stratum:{site}:{column}:{level}',
                ),
            })
        difference = independent_difference(
            values_by_level[levels[0]], values_by_level[levels[1]],
            f'cross_heterogeneity:{site}:{column}',
        )
        stratum_rows.append({
            'analysis': 'cross', 'model_or_contrast': 'qwen_minus_gemma',
            'task': 'side_torque', 'site_label': site,
            'metric': 'donor_effect_difference',
            'stratum': column,
            'level': f'{levels[0]}_minus_{levels[1]}',
            'row_type': 'heterogeneity',
            'n': difference['n_left'] + difference['n_right'],
            'mean': difference['mean'],
            'ci_low': difference['ci_low'],
            'ci_high': difference['ci_high'],
            'sign_flip_p': np.nan,
        })

stratum_audit = pd.DataFrame(stratum_rows)
stratum_audit.to_csv(TABLE_DIR / 'physical_strata.csv', index=False)
display(stratum_audit[
    stratum_audit['row_type'] == 'heterogeneity'
].sort_values([
    'analysis', 'model_or_contrast', 'site_label', 'stratum'
]))
```

```python
stage('Auditing baseline competence and correctness sensitivity')

baseline = patch[[
    'model', 'triplet_id', 'task', 'base_correct',
    'weight_source_correct', 'distance_source_correct',
]].drop_duplicates(['model', 'triplet_id', 'task'])
assert len(baseline) == 2 * 48 * 5
for column in (
    'base_correct', 'weight_source_correct', 'distance_source_correct',
):
    baseline[column] = as_bool(baseline[column])

accuracy = baseline.groupby(['model', 'task']).agg(
    n=('triplet_id', 'nunique'),
    base_accuracy=('base_correct', 'mean'),
    weight_source_accuracy=('weight_source_correct', 'mean'),
    distance_source_accuracy=('distance_source_correct', 'mean'),
).reset_index()
mcnemar_rows = []
for model in MODEL_SPECS:
    for task in sorted(baseline['task'].unique()):
        subset = baseline[
            (baseline['model'] == model) & (baseline['task'] == task)
        ]
        weight_only, distance_only, p_value = exact_mcnemar(
            subset['weight_source_correct'], subset['distance_source_correct']
        )
        mcnemar_rows.append({
            'model': model, 'task': task,
            'weight_only_correct': weight_only,
            'distance_only_correct': distance_only,
            'mcnemar_exact_p': p_value,
        })
mcnemar = pd.DataFrame(mcnemar_rows)

correctness_populations = {
    'all': lambda frame: np.ones(len(frame), dtype=bool),
    'base_correct': lambda frame: frame['base_correct'].to_numpy(bool),
    'relevant_source_correct': lambda frame: (
        frame['relevant_source_correct'].to_numpy(bool)
    ),
    'either_source_correct': lambda frame: (
        frame['either_source_correct'].to_numpy(bool)
    ),
    'both_sources_correct': lambda frame: (
        frame['both_sources_correct'].to_numpy(bool)
    ),
}
competence_specs = (
    ('qwen', 'side_torque', 'donor_effect_difference'),
    ('gemma', 'side_torque', 'donor_effect_difference'),
    ('qwen', 'side_distance', 'cross_value_leakage'),
    ('gemma', 'side_units', 'cross_value_leakage'),
)
competence_rows = []
for model, task, metric in competence_specs:
    for site in FOCAL_SITES:
        base = paired[
            (paired['model'] == model) & (paired['task'] == task)
            & (paired['site_label'] == site) & paired['estimable']
        ]
        for population, selector in correctness_populations.items():
            subset = base[selector(base)]
            competence_rows.append({
                'analysis': 'within', 'population': population,
                'model_or_contrast': model, 'task': task,
                'site_label': site, 'metric': metric,
                **paired_summary(
                    subset[metric],
                    f'competence:{population}:{model}:{task}:{site}:{metric}',
                ),
            })

for site in FOCAL_SITES:
    base = exact[
        (exact['task'] == 'side_torque') & (exact['site_label'] == site)
    ].copy()
    base['cross_route_difference'] = (
        base['donor_effect_difference_qwen']
        - base['donor_effect_difference_gemma']
    )
    common_masks = {
        'all': np.ones(len(base), dtype=bool),
        'common_base_correct': (
            base['base_correct_qwen'] & base['base_correct_gemma']
        ).to_numpy(bool),
        'common_either_source_correct': (
            (base['weight_source_correct_qwen'] | base['distance_source_correct_qwen'])
            & (base['weight_source_correct_gemma'] | base['distance_source_correct_gemma'])
        ).to_numpy(bool),
    }
    for population, mask in common_masks.items():
        subset = base[mask]
        competence_rows.append({
            'analysis': 'cross', 'population': population,
            'model_or_contrast': 'qwen_minus_gemma',
            'task': 'side_torque', 'site_label': site,
            'metric': 'donor_effect_difference',
            **paired_summary(
                subset['cross_route_difference'],
                f'cross_competence:{population}:{site}',
            ),
        })

competence_sensitivity = pd.DataFrame(competence_rows)
behavioral_binding = paired[
    paired['task'].isin(('side_units', 'side_distance'))
][[
    'model', 'triplet_id', 'task', 'cross_value_behavior_error'
]].drop_duplicates()
behavioral_binding_summary = behavioral_binding.groupby([
    'model', 'task'
]).agg(
    n=('triplet_id', 'nunique'),
    cross_value_error_rate=('cross_value_behavior_error', 'mean'),
    cross_value_error_count=('cross_value_behavior_error', 'sum'),
).reset_index()

accuracy.to_csv(TABLE_DIR / 'baseline_accuracy.csv', index=False)
mcnemar.to_csv(TABLE_DIR / 'baseline_mcnemar.csv', index=False)
competence_sensitivity.to_csv(
    TABLE_DIR / 'correctness_sensitivity.csv', index=False
)
behavioral_binding_summary.to_csv(
    TABLE_DIR / 'behavioral_cross_value_errors.csv', index=False
)
display(accuracy)
display(mcnemar)
display(competence_sensitivity.sort_values([
    'analysis', 'model_or_contrast', 'site_label', 'population'
]))
display(behavioral_binding_summary)
```

```python
stage('Auditing equivalence-bound sensitivity and full-chain limits')

equivalence_rows = []
for model in MODEL_SPECS:
    for task in ('side_torque', 'net_torque', 'outcome'):
        sites = FOCAL_SITES if task != 'outcome' else ('late_answer',)
        for site in sites:
            subset = paired[
                (paired['model'] == model) & (paired['task'] == task)
                & (paired['site_label'] == site) & paired['estimable']
            ]
            if not len(subset):
                continue
            weight = paired_summary(
                subset['weight_effect'], f'eq_weight:{model}:{task}:{site}'
            )
            distance = paired_summary(
                subset['distance_effect'], f'eq_distance:{model}:{task}:{site}'
            )
            difference = paired_summary(
                subset['donor_effect_difference'],
                f'eq_difference:{model}:{task}:{site}',
            )
            natural_scale = float(subset['task_natural_scale'].iloc[0])
            for fraction in EQUIVALENCE_FRACTIONS:
                bound = fraction * natural_scale
                equivalent = bool(
                    difference['ci_low'] > -bound
                    and difference['ci_high'] < bound
                )
                positive_donors = bool(
                    weight['ci_low'] > 0 and distance['ci_low'] > 0
                )
                equivalence_rows.append({
                    'model': model, 'task': task, 'site_label': site,
                    'fraction': fraction, 'natural_scale': natural_scale,
                    'bound': bound,
                    'difference_mean': difference['mean'],
                    'difference_ci_low': difference['ci_low'],
                    'difference_ci_high': difference['ci_high'],
                    'weight_ci_low': weight['ci_low'],
                    'distance_ci_low': distance['ci_low'],
                    'equivalent': equivalent,
                    'positive_donors': positive_donors,
                    'composition_gate': bool(equivalent and positive_donors),
                })
equivalence_sensitivity = pd.DataFrame(equivalence_rows)

threshold_rows = []
for keys, group in equivalence_sensitivity.groupby([
    'model', 'task', 'site_label'
]):
    passing = group[group['composition_gate']]
    threshold_rows.append({
        'model': keys[0], 'task': keys[1], 'site_label': keys[2],
        'minimum_tested_fraction': (
            float(passing['fraction'].min()) if len(passing) else np.nan
        ),
        'passes_at_primary_0_25': bool(
            group.loc[np.isclose(group['fraction'], 0.25), 'composition_gate'].iloc[0]
        ),
    })
equivalence_thresholds = pd.DataFrame(threshold_rows)

chain_rows = []
for model in MODEL_SPECS:
    factor_binding = bool(decisions[model]['factor_binding_supported'])
    side_torque = bool(decisions[model]['side_torque_equivalence_supported'])
    late_outcome = bool(decisions[model]['late_outcome_use_supported'])
    chain_rows.append({
        'model': model, 'factor_binding': factor_binding,
        'side_torque_equivalence': side_torque,
        'late_outcome_use': late_outcome,
        'full_chain': bool(factor_binding and side_torque and late_outcome),
    })
chain_boundary = pd.DataFrame(chain_rows)
equivalence_sensitivity.to_csv(
    TABLE_DIR / 'equivalence_sensitivity.csv', index=False
)
equivalence_thresholds.to_csv(
    TABLE_DIR / 'equivalence_thresholds.csv', index=False
)
chain_boundary.to_csv(TABLE_DIR / 'full_chain_boundary.csv', index=False)
display(equivalence_sensitivity[
    (equivalence_sensitivity['task'] == 'side_torque')
    & equivalence_sensitivity['fraction'].isin((0.15, 0.20, 0.25, 0.30, 0.35))
])
display(equivalence_thresholds)
display(chain_boundary)
```

```python
stage('Testing independent-image replication against the prior factorial run')

previous_manifest_rows = read_jsonl(SOURCE_PATHS['factorial_manifest'])
previous_scene_ids = {
    str(row[key])
    for row in previous_manifest_rows
    for key in ('base_scene_id', 'weight_scene_id', 'distance_scene_id')
}
orthogonal_scene_ids = {
    str(row[key])
    for row in manifest_rows
    for key in ('base_scene_id', 'weight_scene_id', 'distance_scene_id')
}
assert len(previous_scene_ids) == len(orthogonal_scene_ids) == 144
assert not (previous_scene_ids & orthogonal_scene_ids)

previous_frames = []
for model in MODEL_SPECS:
    frame = pd.read_csv(SOURCE_PATHS[f'factorial_{model}_paired'])
    frame['model'] = model
    frame['experiment'] = 'previous_factorial'
    frame['triplet_id'] = frame['triplet_id'].astype(str)
    previous_frames.append(frame)
previous_paired = pd.concat(previous_frames, ignore_index=True)
assert len(previous_paired) == 1920

current_for_replication = paired.copy()
current_for_replication['experiment'] = 'orthogonal_control'
replication_plan = (
    ('side_units', 'factor_selectivity'),
    ('side_distance', 'factor_selectivity'),
    ('side_torque', 'donor_effect_difference'),
    ('side_torque', 'mean_donor_effect'),
)
replication_rows = []
for model in MODEL_SPECS:
    for task, metric in replication_plan:
        for site in FOCAL_SITES:
            experiment_results = {}
            for experiment, frame in (
                ('previous_factorial', previous_paired),
                ('orthogonal_control', current_for_replication),
            ):
                subset = frame[
                    (frame['model'] == model) & (frame['task'] == task)
                    & (frame['site_label'] == site)
                ]
                result = paired_summary(
                    subset[metric],
                    f'replication:{experiment}:{model}:{task}:{site}:{metric}',
                )
                experiment_results[experiment] = result
                replication_rows.append({
                    'row_type': 'experiment', 'experiment': experiment,
                    'model': model, 'task': task, 'site_label': site,
                    'metric': metric, **result,
                })
            old = experiment_results['previous_factorial']
            new = experiment_results['orthogonal_control']
            replication_rows.append({
                'row_type': 'sign_replication',
                'experiment': 'orthogonal_vs_previous',
                'model': model, 'task': task, 'site_label': site,
                'metric': metric, 'n': old['n'] + new['n'],
                'mean': new['mean'] - old['mean'],
                'ci_low': np.nan, 'ci_high': np.nan,
                'sign_flip_p': np.nan,
                'same_mean_sign': bool(np.sign(old['mean']) == np.sign(new['mean'])),
                'both_ci_positive': bool(old['ci_low'] > 0 and new['ci_low'] > 0),
                'both_ci_negative': bool(old['ci_high'] < 0 and new['ci_high'] < 0),
            })
replication = pd.DataFrame(replication_rows)

cross_replication_rows = []
for experiment, frame in (
    ('previous_factorial', previous_paired),
    ('orthogonal_control', current_for_replication),
):
    left = frame[frame['model'] == 'qwen']
    right = frame[frame['model'] == 'gemma']
    merged = left[[
        'triplet_id', 'task', 'site_label', 'donor_effect_difference'
    ]].merge(
        right[[
            'triplet_id', 'task', 'site_label', 'donor_effect_difference'
        ]],
        on=['triplet_id', 'task', 'site_label'],
        validate='one_to_one', suffixes=('_qwen', '_gemma'),
    )
    for site in FOCAL_SITES:
        subset = merged[
            (merged['task'] == 'side_torque')
            & (merged['site_label'] == site)
        ]
        difference = (
            subset['donor_effect_difference_qwen']
            - subset['donor_effect_difference_gemma']
        )
        cross_replication_rows.append({
            'experiment': experiment, 'site_label': site,
            'metric': 'side_torque_route_difference_qwen_minus_gemma',
            **paired_summary(
                difference, f'cross_replication:{experiment}:{site}'
            ),
        })
cross_replication = pd.DataFrame(cross_replication_rows)

replication.to_csv(TABLE_DIR / 'independent_scene_replication.csv', index=False)
cross_replication.to_csv(
    TABLE_DIR / 'cross_model_route_replication.csv', index=False
)
display(replication.sort_values([
    'model', 'task', 'site_label', 'metric', 'row_type'
]))
display(cross_replication)
print('Scene overlap between experiments:', len(
    previous_scene_ids & orthogonal_scene_ids
))
```

```python
stage('Applying conservative claim gates and finalizing artifacts')

def within_record(model, task, site, metric):
    rows = within_summary[
        (within_summary['model'] == model)
        & (within_summary['task'] == task)
        & (within_summary['site_label'] == site)
        & (within_summary['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def cross_record(task, site, metric, standardized=False):
    frame = standardized_cross if standardized else cross_summary
    rows = frame[
        (frame['task'] == task) & (frame['site_label'] == site)
        & (frame['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

shared_factor_binding = all(
    bool(decisions[model]['factor_site_gates'][site])
    for model in MODEL_SPECS for site in FOCAL_SITES
)
qwen_weight_dominance = all(
    bool(decisions['qwen']['weight_dominance_site_gates'][site])
    for site in FOCAL_SITES
)
qwen_weight_to_distance_leakage = all(
    bool(decisions['qwen']['leakage_site_flags'][site]['side_distance'])
    for site in FOCAL_SITES
)
gemma_option_gate = equivalence_thresholds[
    (equivalence_thresholds['model'] == 'gemma')
    & (equivalence_thresholds['task'] == 'side_torque')
    & (equivalence_thresholds['site_label'] == 'middle_options')
].iloc[0]
gemma_partial_convergence = bool(gemma_option_gate['passes_at_primary_0_25'])
gemma_convergence_at_20pct = bool(
    gemma_option_gate['minimum_tested_fraction'] <= 0.20
)

cross_route_raw = all(
    cross_record('side_torque', site,
                 'donor_effect_difference')['ci_low'] > 0
    and cross_record('side_torque', site,
                     'donor_effect_difference')['q_value'] < ALPHA
    for site in FOCAL_SITES
)
cross_route_standardized = all(
    cross_record('side_torque', site,
                 'std_donor_effect_difference', True)['ci_low'] > 0
    and cross_record('side_torque', site,
                     'std_donor_effect_difference', True)['q_value'] < ALPHA
    for site in FOCAL_SITES
)
gemma_relative_distance_to_units_leakage = all(
    cross_record('side_units', site,
                 'cross_value_leakage')['ci_high'] < 0
    and cross_record('side_units', site,
                     'cross_value_leakage')['q_value'] < ALPHA
    for site in FOCAL_SITES
)

def replicated_positive(model, task, metric):
    rows = replication[
        (replication['row_type'] == 'sign_replication')
        & (replication['model'] == model)
        & (replication['task'] == task)
        & (replication['metric'] == metric)
        & replication['site_label'].isin(FOCAL_SITES)
    ]
    return bool(len(rows) == 2 and rows['both_ci_positive'].fillna(False).all())

qwen_route_replication = replicated_positive(
    'qwen', 'side_torque', 'donor_effect_difference'
)
cross_route_replication = all(
    cross_replication[
        cross_replication['site_label'] == site
    ].sort_values('experiment')['ci_low'].gt(0).all()
    for site in FOCAL_SITES
)
complete_chain_absent = bool((~chain_boundary['full_chain']).all())
low_complete_torque_competence = all(
    paired[
        (paired['model'] == model)
        & (paired['task'].isin(('side_torque', 'net_torque')))
    ][['triplet_id', 'task', 'all_three_correct']]
    .drop_duplicates()['all_three_correct'].mean() < 0.10
    for model in MODEL_SPECS
)

core_confirmed = all((
    shared_factor_binding,
    qwen_weight_dominance,
    cross_route_raw,
    cross_route_standardized,
    qwen_route_replication,
    cross_route_replication,
    complete_chain_absent,
))
decision = (
    'orthogonal_control_confirms_model_specific_composition_routes'
    if core_confirmed else
    'orthogonal_route_story_requires_targeted_followup'
)

claim_card = {
    'schema_version': 'orthogonal-binding-robustness-claim-card-1.0',
    'decision': decision,
    'claims': {
        'shared_factor_specific_binding_under_scalar_orthogonality': (
            shared_factor_binding
        ),
        'qwen_weight_dominance_survives_scalar_orthogonality': (
            qwen_weight_dominance
        ),
        'qwen_weight_dominance_replicates_on_independent_images': (
            qwen_route_replication
        ),
        'gemma_option_level_partial_torque_convergence_at_0_25_bound': (
            gemma_partial_convergence
        ),
        'gemma_convergence_survives_0_20_bound': gemma_convergence_at_20pct,
        'cross_model_route_difference_raw': cross_route_raw,
        'cross_model_route_difference_natural_scale_standardized': (
            cross_route_standardized
        ),
        'cross_model_route_difference_replicates_previous_experiment': (
            cross_route_replication
        ),
        'qwen_weight_to_distance_cross_value_leakage': (
            qwen_weight_to_distance_leakage
        ),
        'gemma_has_more_distance_to_units_leakage_than_qwen': (
            gemma_relative_distance_to_units_leakage
        ),
        'complete_factor_to_torque_to_outcome_chain_absent': (
            complete_chain_absent
        ),
        'correct_torque_chain_competence_is_low': (
            low_complete_torque_competence
        ),
    },
    'sample': {
        'orthogonal_triplets': 48,
        'orthogonal_scenes': 144,
        'previous_factorial_scenes': 144,
        'scene_overlap_between_experiments': 0,
        'exact_cross_model_rows': 960,
    },
    'paper_language': (
        'Scalar-orthogonal, independent-image controls confirm that both '
        'models causally separate units and distance but allocate these '
        'factors differently when forming torque judgments. Qwen retains a '
        'replicable weight-dominant route with weight-to-distance leakage; '
        'Gemma shows option-level partial convergence. The cross-model route '
        'difference survives natural-scale standardization, while neither '
        'model supports a complete factor-to-torque-to-outcome chain.'
    ),
    'nonclaims': [
        'The audit is an independent preregistration.',
        'Architecture alone caused the model difference.',
        'Either model implements literal multiplication.',
        'The reported route is a unique circuit.',
        'Low torque accuracy is evidence that no torque information exists.',
    ],
    'source_hashes': source_hashes,
}
write_json_atomic(OUTPUT_DIR / 'claim_card.json', claim_card)

report = [
    '# Orthogonal binding robustness audit', '',
    f'Decision: `{decision}`', '', '## Claim card', '',
]
for claim, value in claim_card['claims'].items():
    report.append(f'- {claim}: **{value}**')
report.extend([
    '', '## Conservative paper language', '',
    claim_card['paper_language'], '',
    'The orthogonal scenes are independent of the previous factorial images, '
    'but the hypotheses and audit were informed by earlier results.', '',
    '## Nonclaims', '',
])
report.extend(f'- {value}' for value in claim_card['nonclaims'])
report_path = OUTPUT_DIR / 'orthogonal_binding_robustness_report.md'
report_path.write_text('\n'.join(report) + '\n', encoding='utf-8')

artifact_paths = sorted(
    path for path in OUTPUT_DIR.rglob('*') if path.is_file()
)
artifact_manifest = {
    str(path.relative_to(OUTPUT_DIR)): sha256_file(path)
    for path in artifact_paths
    if path.name != 'artifact_manifest.json'
}
write_json_atomic(OUTPUT_DIR / 'artifact_manifest.json', artifact_manifest)
write_json_atomic(OUTPUT_DIR / 'audit_complete.json', {
    'decision': decision,
    'artifact_count': len(artifact_manifest),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
_progress.n = TOTAL_STAGES
_progress.set_postfix_str('complete')
_progress.close()
print(json.dumps(claim_card, indent=2))
print('Report:', report_path)
print('Tables:', TABLE_DIR)
print('Figures:', FIGURE_DIR)
```
