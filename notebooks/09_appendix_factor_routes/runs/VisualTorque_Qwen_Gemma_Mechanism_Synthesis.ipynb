```python
from pathlib import Path
import os
import sys

from tqdm.auto import tqdm
from IPython.display import display

TOTAL_STAGES = 10
_progress = tqdm(total=TOTAL_STAGES, desc='Qwen–Gemma synthesis', unit='stage')

def stage(label):
    if _progress.n < TOTAL_STAGES:
        _progress.update(1)
    _progress.set_postfix_str(label)
    print(f'[{_progress.n}/{TOTAL_STAGES}] {label}')
```

## 1. Locate the completed experiment artifacts


```python
stage('Locating project and completed runs')

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')

def find_project_root():
    configured = os.environ.get('VISUALTORQUE_ROOT')
    candidates = []
    if configured:
        candidates.append(Path(configured).expanduser())
    if IN_COLAB:
        candidates.extend((
            Path('/content/drive/MyDrive/VisualTorque'),
            Path('/content/drive/MyDrive/VisualTorque/VisualTorque'),
        ))
    for base in (Path.cwd(), *Path.cwd().parents):
        candidates.extend((base, base / 'VisualTorque'))
    checked = set()
    for candidate in candidates:
        candidate = candidate.resolve()
        if candidate in checked:
            continue
        checked.add(candidate)
        if (candidate / 'results').is_dir() and (
            (candidate / 'toolkit' / 'pyproject.toml').is_file()
            or (candidate / 'notebooks').is_dir()
        ):
            return candidate
    raise FileNotFoundError(
        'Could not locate VisualTorque. Start Jupyter inside the repository, '
        'mount Drive, or set VISUALTORQUE_ROOT.'
    )

PROJECT_ROOT = find_project_root()
RESULTS_ROOT = PROJECT_ROOT / 'results'

def resolve_analysis_dir(exact_relative, model_tokens):
    exact = RESULTS_ROOT / exact_relative / 'analysis'
    if (exact / 'patch_records.csv').is_file():
        return exact
    candidates = []
    for path in RESULTS_ROOT.rglob('patch_records.csv'):
        if path.parent.name != 'analysis':
            continue
        lowered = str(path).lower()
        if all(token.lower() in lowered for token in model_tokens):
            candidates.append(path.parent)
    if len(candidates) == 1:
        return candidates[0]
    raise FileNotFoundError(
        f'Could not uniquely locate analysis for {model_tokens}. '
        f'Expected {exact}; candidates={candidates}'
    )

QWEN_ANALYSIS = resolve_analysis_dir(
    Path('Qwen__Qwen3.5-9B') / 'competence_grouped_route_confirmatory_v1',
    ('qwen', 'competence_grouped_route_confirmatory_v1'),
)
GEMMA_ANALYSIS = resolve_analysis_dir(
    Path('google__gemma-4-12B-it')
    / 'competence_grouped_route_gemma_replication_v3',
    ('gemma', 'competence_grouped_route_gemma_replication_v3'),
)
OUTPUT_DIR = RESULTS_ROOT / 'cross_model' / 'qwen_gemma_mechanism_synthesis_v1'
FIGURE_DIR = OUTPUT_DIR / 'figures'
FIGURE_DIR.mkdir(parents=True, exist_ok=True)
print('Project:', PROJECT_ROOT)
print('Qwen:', QWEN_ANALYSIS)
print('Gemma:', GEMMA_ANALYSIS)
print('Output:', OUTPUT_DIR)
```

```python
stage('Loading and validating source tables')

import hashlib
import json
from datetime import datetime, timezone

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

MODEL_SPECS = {
    'Qwen': {
        'analysis': QWEN_ANALYSIS,
        'layer': 16,
        'color': '#2563eb',
    },
    'Gemma': {
        'analysis': GEMMA_ANALYSIS,
        'layer': 25,
        'color': '#f97316',
    },
}
SITE_SCOPES = ('all_text_except_final', 'options_tokens')
DONORS = (
    'matched_source',
    'same_task_value_different_layout',
    'same_outcome_different_task_value',
)

from visualtorque_checks import sha256_file

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def as_bool(series):
    if series.dtype == bool:
        return series
    mapped = series.astype(str).str.lower().map({
        'true': True, 'false': False, '1': True, '0': False,
    })
    assert mapped.notna().all(), sorted(series.astype(str).unique())
    return mapped.astype(bool)

def task_family(task):
    task = str(task)
    if 'units' in task:
        return 'units'
    if 'torque' in task:
        return 'torque'
    if 'stack' in task:
        return 'structure'
    return 'other'

source_tables = {}
source_hashes = {}
for model, spec in MODEL_SPECS.items():
    paths = {
        'patch_records': spec['analysis'] / 'patch_records.csv',
        'query_baselines': spec['analysis'] / 'query_baselines.csv',
        'pair_manifest': spec['analysis'] / 'pair_manifest.csv',
        'decision_summary': spec['analysis'] / 'decision_summary.json',
    }
    missing = [str(path) for path in paths.values() if not path.is_file()]
    assert not missing, f'{model} missing artifacts:\n' + '\n'.join(missing)
    patch = pd.read_csv(paths['patch_records'])
    baseline = pd.read_csv(paths['query_baselines'])
    manifest = pd.read_csv(paths['pair_manifest'])
    required_patch = {
        'pair_id', 'query_role', 'layer_index', 'scope', 'donor_type',
        'recovery', 'valid_recovery', 'source_answer_contrast_gain',
        'natural_effect_norm', 'query_both_correct', 'competence_cohort',
        'edit_type', 'source_value', 'target_value',
    }
    assert required_patch <= set(patch.columns), (
        model, sorted(required_patch - set(patch.columns))
    )
    assert set(DONORS) <= set(patch['donor_type'].unique()), model
    for frame in (patch, baseline):
        for column in ('query_both_correct', 'valid_recovery'):
            if column in frame:
                frame[column] = as_bool(frame[column])
    assert patch['pair_id'].nunique() == 192, (model, patch['pair_id'].nunique())
    source_tables[model] = {
        'patch': patch, 'baseline': baseline, 'manifest': manifest,
        'decision': json.loads(paths['decision_summary'].read_text('utf-8')),
    }
    source_hashes[model] = {
        name: sha256_file(path) for name, path in paths.items()
    }
    print(model, 'pairs=', patch['pair_id'].nunique(),
          'records=', len(patch), 'decision=',
          source_tables[model]['decision'].get('decision'))
```

```python
stage('Freezing the cross-model analysis protocol')

SEED = 20260820
BOOTSTRAP_REPLICATES = 5000
PERMUTATION_REPLICATES = 10000
ALPHA = 0.05
REPLICATION_FRACTION = 0.50
STRATA = ('edit_type', 'task_family')
PROTOCOL_PATH = OUTPUT_DIR / 'synthesis_protocol.json'

protocol_identity = {
    'schema_version': 'qwen-gemma-mechanism-synthesis-1.0',
    'status': 'cross-model synthesis of completed confirmatory experiments',
    'source_hashes': source_hashes,
    'mapped_focal_sites': {
        'Qwen': {'layer': 16, 'scopes': list(SITE_SCOPES)},
        'Gemma': {'layer': 25, 'scopes': list(SITE_SCOPES)},
    },
    'primary_population': 'actual custom-query source and target both correct',
    'metrics': {
        'recovery': (
            'same-task-value minus same-outcome-different-value recovery'
        ),
        'margin': (
            'same-task-value minus same-outcome-different-value '
            'source-answer contrast gain; denominator-free'
        ),
    },
    'formal_cross_model_tests': (
        'Qwen minus Gemma successful-query effect and the difference of '
        'competence interactions, blocked by edit type and task family'
    ),
    'sensitivity': (
        '2.5/97.5% winsorization, natural-effect thresholds at '
        '0/10/25/50%, deterministic success-error matching, '
        'and leave-one-edit-out effects'
    ),
    'replication_fraction': REPLICATION_FRACTION,
    'multiplicity': 'BH across the two mapped scopes within each test family',
    'bootstrap_unit': 'counterfactual pair',
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'permutation_replicates': PERMUTATION_REPLICATES,
    'alpha': ALPHA,
    'seed': SEED,
    'interpretation_limit': (
        'A negative mapped-site result excludes a shared route under the '
        'prespecified depth mapping; it does not exclude Gemma-specific, '
        'distributed, or differently timed physical computation.'
    ),
}
if PROTOCOL_PATH.is_file():
    previous = json.loads(PROTOCOL_PATH.read_text('utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing synthesis used different inputs or rules. Use a new '
        'versioned output directory instead of overwriting it.'
    )
else:
    write_json_atomic(PROTOCOL_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Protocol:', PROTOCOL_PATH)
```

## 2. Pair-level statistics


```python
stage('Defining deterministic statistical utilities')

def stable_seed(label):
    digest = hashlib.sha256(f'{SEED}:{label}'.encode()).hexdigest()
    return int(digest[:8], 16)

def bh_adjust(p_values):
    values = np.asarray(p_values, dtype=float)
    result = np.full(values.shape, np.nan)
    valid = np.isfinite(values)
    if not valid.any():
        return result
    observed = values[valid]
    order = np.argsort(observed)
    ranked = observed[order]
    adjusted = ranked * len(ranked) / np.arange(1, len(ranked) + 1)
    adjusted = np.minimum.accumulate(adjusted[::-1])[::-1]
    restored = np.empty_like(adjusted)
    restored[order] = np.minimum(adjusted, 1.0)
    result[valid] = restored
    return result

def mean_test(values, label):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if not len(values):
        return dict(n=0, mean=np.nan, ci_low=np.nan, ci_high=np.nan,
                    p_value=np.nan, standardized_mean=np.nan)
    rng = np.random.default_rng(stable_seed(label))
    counts = rng.multinomial(
        len(values), np.full(len(values), 1 / len(values)),
        size=BOOTSTRAP_REPLICATES,
    )
    bootstrap = counts @ values / len(values)
    signs = rng.choice(
        np.asarray([-1.0, 1.0]),
        size=(PERMUTATION_REPLICATES, len(values)),
    )
    permutation = signs @ values / len(values)
    observed = float(values.mean())
    standard_deviation = (
        float(values.std(ddof=1)) if len(values) > 1 else np.nan
    )
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(bootstrap, 0.025)),
        'ci_high': float(np.quantile(bootstrap, 0.975)),
        'p_value': float(
            (1 + np.sum(permutation >= observed))
            / (PERMUTATION_REPLICATES + 1)
        ),
        'standardized_mean': (
            observed / standard_deviation
            if np.isfinite(standard_deviation) and standard_deviation > 0
            else np.nan
        ),
    }

def blocked_success_failure(frame, value_column, label):
    blocks = []
    for keys, group in frame.groupby(list(STRATA), sort=True):
        success = group.loc[group['query_both_correct'], value_column]
        failure = group.loc[~group['query_both_correct'], value_column]
        success = success.dropna().to_numpy(dtype=float)
        failure = failure.dropna().to_numpy(dtype=float)
        if len(success) and len(failure):
            weight = len(success) * len(failure) / (len(success) + len(failure))
            blocks.append((keys, success, failure, weight))
    if not blocks:
        return dict(success_n=0, failure_n=0, block_count=0,
                    effect=np.nan, ci_low=np.nan, ci_high=np.nan,
                    p_value=np.nan)
    weights = np.asarray([row[3] for row in blocks], dtype=float)
    weights /= weights.sum()
    observed = float(weights @ np.asarray([
        success.mean() - failure.mean()
        for _, success, failure, _ in blocks
    ]))
    rng = np.random.default_rng(stable_seed(label))
    bootstrap = np.zeros(BOOTSTRAP_REPLICATES)
    permutation = np.zeros(PERMUTATION_REPLICATES)
    for weight, (_, success, failure, _) in zip(weights, blocks):
        success_counts = rng.multinomial(
            len(success), np.full(len(success), 1 / len(success)),
            size=BOOTSTRAP_REPLICATES,
        )
        failure_counts = rng.multinomial(
            len(failure), np.full(len(failure), 1 / len(failure)),
            size=BOOTSTRAP_REPLICATES,
        )
        bootstrap += weight * (
            success_counts @ success / len(success)
            - failure_counts @ failure / len(failure)
        )
        pooled = np.concatenate([success, failure])
        order = np.argsort(
            rng.random((PERMUTATION_REPLICATES, len(pooled))), axis=1
        )
        selected = pooled[order[:, :len(success)]].sum(axis=1)
        permutation += weight * (
            selected / len(success)
            - (pooled.sum() - selected) / len(failure)
        )
    return {
        'success_n': int(sum(len(row[1]) for row in blocks)),
        'failure_n': int(sum(len(row[2]) for row in blocks)),
        'block_count': int(len(blocks)), 'effect': observed,
        'ci_low': float(np.quantile(bootstrap, 0.025)),
        'ci_high': float(np.quantile(bootstrap, 0.975)),
        'p_value': float(
            (1 + np.sum(permutation >= observed))
            / (PERMUTATION_REPLICATES + 1)
        ),
    }

def blocked_cross_model(frame, value_column, label, interaction=False):
    blocks = []
    for keys, group in frame.groupby(list(STRATA), sort=True):
        cells = {}
        competence_values = (True, False) if interaction else (True,)
        for competence in competence_values:
            for model in ('Qwen', 'Gemma'):
                values = group.loc[
                    (group['model'] == model)
                    & (group['query_both_correct'] == competence),
                    value_column,
                ].dropna().to_numpy(dtype=float)
                cells[(model, competence)] = values
        if all(len(values) for values in cells.values()):
            sizes = np.asarray([len(values) for values in cells.values()])
            weight = len(sizes) / np.sum(1 / sizes)
            blocks.append((keys, cells, weight))
    if not blocks:
        return dict(block_count=0, effect=np.nan, ci_low=np.nan,
                    ci_high=np.nan, p_value=np.nan)
    weights = np.asarray([row[2] for row in blocks], dtype=float)
    weights /= weights.sum()

    def cell_contrast(cells):
        success = (
            cells[('Qwen', True)].mean()
            - cells[('Gemma', True)].mean()
        )
        if not interaction:
            return success
        failure = (
            cells[('Qwen', False)].mean()
            - cells[('Gemma', False)].mean()
        )
        return success - failure

    observed = float(weights @ np.asarray([
        cell_contrast(cells) for _, cells, _ in blocks
    ]))
    rng = np.random.default_rng(stable_seed(label))
    bootstrap = np.zeros(BOOTSTRAP_REPLICATES)
    permutation = np.zeros(PERMUTATION_REPLICATES)
    for weight, (_, cells, _) in zip(weights, blocks):
        bootstrap_cells = {}
        for key, values in cells.items():
            counts = rng.multinomial(
                len(values), np.full(len(values), 1 / len(values)),
                size=BOOTSTRAP_REPLICATES,
            )
            bootstrap_cells[key] = counts @ values / len(values)
        success_boot = (
            bootstrap_cells[('Qwen', True)]
            - bootstrap_cells[('Gemma', True)]
        )
        bootstrap += weight * success_boot
        if interaction:
            bootstrap += weight * (
                -bootstrap_cells[('Qwen', False)]
                + bootstrap_cells[('Gemma', False)]
            )

        permuted_by_competence = {}
        competence_values = (True, False) if interaction else (True,)
        for competence in competence_values:
            qwen = cells[('Qwen', competence)]
            gemma = cells[('Gemma', competence)]
            pooled = np.concatenate([qwen, gemma])
            order = np.argsort(
                rng.random((PERMUTATION_REPLICATES, len(pooled))), axis=1
            )
            selected = pooled[order[:, :len(qwen)]].sum(axis=1)
            permuted_by_competence[competence] = (
                selected / len(qwen)
                - (pooled.sum() - selected) / len(gemma)
            )
        permutation += weight * permuted_by_competence[True]
        if interaction:
            permutation -= weight * permuted_by_competence[False]
    return {
        'block_count': int(len(blocks)), 'effect': observed,
        'ci_low': float(np.quantile(bootstrap, 0.025)),
        'ci_high': float(np.quantile(bootstrap, 0.975)),
        'p_value': float(
            (1 + np.sum(permutation >= observed))
            / (PERMUTATION_REPLICATES + 1)
        ),
    }

positive_smoke = mean_test(np.linspace(0.2, 1.0, 24), 'positive-smoke')
assert positive_smoke['mean'] > 0 and positive_smoke['ci_low'] > 0
print('Statistical utility self-tests passed.')
```

```python
stage('Reconstructing unified pair-level effects')

pair_frames = []
competence_rows = []
for model, spec in MODEL_SPECS.items():
    patch = source_tables[model]['patch'].copy()
    baseline = source_tables[model]['baseline'].copy()
    relevant = patch[
        (patch['query_role'] == 'relevant_changed')
        & (patch['layer_index'].astype(int) == spec['layer'])
        & patch['scope'].isin(SITE_SCOPES)
    ].copy()
    relevant['query_both_correct'] = as_bool(relevant['query_both_correct'])
    index_columns = [
        'pair_id', 'layer_index', 'scope', 'competence_cohort',
        'query_both_correct', 'edit_type',
    ]
    recovery = relevant.pivot_table(
        index=index_columns, columns='donor_type', values='recovery',
        aggfunc='first',
    ).reset_index()
    margin = relevant.pivot_table(
        index=index_columns, columns='donor_type',
        values='source_answer_contrast_gain', aggfunc='first',
    ).reset_index()
    assert set(DONORS) <= set(recovery.columns), model
    assert set(DONORS) <= set(margin.columns), model
    recovery['recovery_value_minus_outcome'] = (
        recovery['same_task_value_different_layout']
        - recovery['same_outcome_different_task_value']
    )
    margin['margin_value_minus_outcome'] = (
        margin['same_task_value_different_layout']
        - margin['same_outcome_different_task_value']
    )
    natural = relevant.groupby(
        ['pair_id', 'layer_index', 'scope'], as_index=False
    ).agg(
        natural_effect_norm=('natural_effect_norm', 'first'),
        source_value=('source_value', 'first'),
        target_value=('target_value', 'first'),
    )
    manifest = source_tables[model]['manifest'].copy()
    metadata_columns = ['pair_id']
    if 'relevant_task' in manifest:
        metadata_columns.append('relevant_task')
    metadata = manifest[metadata_columns].drop_duplicates('pair_id')
    unified = (
        recovery[index_columns + ['recovery_value_minus_outcome']]
        .merge(
            margin[index_columns + ['margin_value_minus_outcome']],
            on=index_columns, how='outer', validate='one_to_one',
        )
        .merge(
            natural, on=['pair_id', 'layer_index', 'scope'],
            how='left', validate='one_to_one',
        )
        .merge(metadata, on='pair_id', how='left', validate='many_to_one')
    )
    if 'relevant_task' not in unified or unified['relevant_task'].isna().all():
        unified['relevant_task'] = 'unknown'
    unified['task_family'] = unified['relevant_task'].map(task_family)
    unified['absolute_value_gap'] = (
        unified['source_value'].astype(float)
        - unified['target_value'].astype(float)
    ).abs()
    unified['log_natural_effect_norm'] = np.log1p(
        unified['natural_effect_norm'].astype(float)
    )
    unified['model'] = model
    unified['mapped_site'] = unified['scope'].map({
        'all_text_except_final': 'all_text',
        'options_tokens': 'options',
    })
    assert not unified.duplicated(['pair_id', 'mapped_site']).any(), model
    assert unified['pair_id'].nunique() == 192, (model, unified['pair_id'].nunique())
    pair_frames.append(unified)

    relevant_baseline = baseline[baseline['query_role'] == 'relevant_changed'].copy()
    relevant_baseline['query_both_correct'] = as_bool(
        relevant_baseline['query_both_correct']
    )
    competence_rows.append({
        'model': model,
        'pairs': int(relevant_baseline['pair_id'].nunique()),
        'query_both_correct_n': int(relevant_baseline.loc[
            relevant_baseline['query_both_correct'], 'pair_id'
        ].nunique()),
        'query_both_correct_rate': float(
            relevant_baseline['query_both_correct'].mean()
        ),
    })

pair_frame = pd.concat(pair_frames, ignore_index=True)
competence_summary = pd.DataFrame(competence_rows)
pair_frame.to_csv(OUTPUT_DIR / 'unified_pair_level.csv', index=False)
competence_summary.to_csv(OUTPUT_DIR / 'competence_summary.csv', index=False)
display(competence_summary)
display(pair_frame.groupby(
    ['model', 'mapped_site', 'query_both_correct']
).size().rename('n').reset_index())
```

## 3. Primary within-model and cross-model tests


```python
stage('Running primary and cross-model tests')

METRICS = {
    'recovery': 'recovery_value_minus_outcome',
    'margin': 'margin_value_minus_outcome',
}
within_rows = []
cross_rows = []
for metric, value_column in METRICS.items():
    for mapped_site in ('all_text', 'options'):
        site = pair_frame[pair_frame['mapped_site'] == mapped_site]
        for model in ('Qwen', 'Gemma'):
            model_site = site[site['model'] == model]
            success = model_site[model_site['query_both_correct']]
            failure = model_site[~model_site['query_both_correct']]
            within_rows.append({
                'model': model, 'mapped_site': mapped_site,
                'metric': metric,
                **{f'success_{key}': value for key, value in mean_test(
                    success[value_column],
                    f'primary:{model}:{mapped_site}:{metric}:success',
                ).items()},
                **{f'failure_{key}': value for key, value in mean_test(
                    failure[value_column],
                    f'primary:{model}:{mapped_site}:{metric}:failure',
                ).items()},
                **{f'competence_{key}': value for key, value in
                   blocked_success_failure(
                       model_site, value_column,
                       f'competence:{model}:{mapped_site}:{metric}',
                   ).items()},
            })
        cross_rows.append({
            'mapped_site': mapped_site, 'metric': metric,
            'test': 'qwen_minus_gemma_success',
            **blocked_cross_model(
                site, value_column,
                f'cross-success:{mapped_site}:{metric}', interaction=False,
            ),
        })
        cross_rows.append({
            'mapped_site': mapped_site, 'metric': metric,
            'test': 'model_by_competence_interaction',
            **blocked_cross_model(
                site, value_column,
                f'cross-interaction:{mapped_site}:{metric}', interaction=True,
            ),
        })

within_tests = pd.DataFrame(within_rows)
cross_tests = pd.DataFrame(cross_rows)
within_tests['success_q_bh'] = np.nan
within_tests['competence_q_bh'] = np.nan
for keys, indices in within_tests.groupby(['model', 'metric']).groups.items():
    within_tests.loc[indices, 'success_q_bh'] = bh_adjust(
        within_tests.loc[indices, 'success_p_value']
    )
    within_tests.loc[indices, 'competence_q_bh'] = bh_adjust(
        within_tests.loc[indices, 'competence_p_value']
    )
within_tests['success_pass'] = (
    (within_tests['success_mean'] > 0)
    & (within_tests['success_ci_low'] > 0)
    & (within_tests['success_q_bh'] < ALPHA)
)
within_tests['competence_pass'] = (
    (within_tests['competence_effect'] > 0)
    & (within_tests['competence_ci_low'] > 0)
    & (within_tests['competence_q_bh'] < ALPHA)
)
cross_tests['q_bh'] = np.nan
for keys, indices in cross_tests.groupby(['metric', 'test']).groups.items():
    cross_tests.loc[indices, 'q_bh'] = bh_adjust(
        cross_tests.loc[indices, 'p_value']
    )
cross_tests['passes_fdr'] = (
    (cross_tests['effect'] > 0)
    & (cross_tests['ci_low'] > 0)
    & (cross_tests['q_bh'] < ALPHA)
)
within_tests.to_csv(OUTPUT_DIR / 'within_model_tests.csv', index=False)
cross_tests.to_csv(OUTPUT_DIR / 'cross_model_tests.csv', index=False)
display(within_tests[[
    'model', 'mapped_site', 'metric', 'success_n', 'success_mean',
    'success_ci_low', 'success_ci_high', 'success_q_bh', 'success_pass',
    'competence_effect', 'competence_ci_low', 'competence_ci_high',
    'competence_q_bh', 'competence_pass',
]])
display(cross_tests)
```

```python
stage('Running sensitivity, matching, and replication-fraction tests')

def winsorized(series):
    values = series.astype(float)
    lower, upper = values.quantile([0.025, 0.975])
    return values.clip(lower, upper)

def deterministic_matches(frame):
    rows = []
    for keys, group in frame.groupby(list(STRATA), sort=True):
        success = group[group['query_both_correct']].copy()
        failure = group[~group['query_both_correct']].copy()
        if success.empty or failure.empty:
            continue
        features = ['log_natural_effect_norm', 'absolute_value_gap']
        pooled = group[features]
        scale = pooled.std(ddof=0).replace(0, 1).fillna(1)
        success = success.sort_values('pair_id')
        available = set(failure.index)
        for success_index, success_row in success.iterrows():
            if not available:
                break
            candidate_indices = sorted(available)
            candidate = failure.loc[candidate_indices]
            distances = (((candidate[features] - success_row[features]) / scale) ** 2).sum(axis=1)
            failure_index = distances.sort_values(kind='mergesort').index[0]
            rows.append({
                'success_index': int(success_index),
                'failure_index': int(failure_index),
                'edit_type': keys[0], 'task_family': keys[1],
                'distance': float(distances.loc[failure_index]),
            })
            available.remove(failure_index)
    return pd.DataFrame(rows)

sensitivity_rows = []
matching_rows = []
replication_rows = []
for metric, value_column in METRICS.items():
    for mapped_site in ('all_text', 'options'):
        site = pair_frame[pair_frame['mapped_site'] == mapped_site].copy()
        for model in ('Qwen', 'Gemma'):
            model_site = site[site['model'] == model].copy()
            model_site['winsorized_value'] = winsorized(model_site[value_column])
            sensitivity_rows.append({
                'analysis': 'winsorized_success', 'model': model,
                'mapped_site': mapped_site, 'metric': metric,
                'threshold_quantile': np.nan,
                **mean_test(
                    model_site.loc[
                        model_site['query_both_correct'], 'winsorized_value'
                    ], f'winsor:{model}:{mapped_site}:{metric}',
                ),
            })
            if metric == 'recovery':
                for quantile in (0.0, 0.10, 0.25, 0.50):
                    threshold = model_site['natural_effect_norm'].quantile(quantile)
                    selected = model_site[
                        model_site['query_both_correct']
                        & (model_site['natural_effect_norm'] >= threshold)
                    ]
                    sensitivity_rows.append({
                        'analysis': 'natural_effect_threshold',
                        'model': model, 'mapped_site': mapped_site,
                        'metric': metric, 'threshold_quantile': quantile,
                        'threshold_value': float(threshold),
                        **mean_test(
                            selected[value_column],
                            f'threshold:{model}:{mapped_site}:{quantile}',
                        ),
                    })
            for held_out in sorted(model_site['edit_type'].dropna().unique()):
                selected = model_site[
                    model_site['query_both_correct']
                    & (model_site['edit_type'] != held_out)
                ]
                sensitivity_rows.append({
                    'analysis': f'leave_out:{held_out}', 'model': model,
                    'mapped_site': mapped_site, 'metric': metric,
                    'threshold_quantile': np.nan,
                    **mean_test(
                        selected[value_column],
                        f'leaveout:{model}:{mapped_site}:{metric}:{held_out}',
                    ),
                })

            matchable = model_site.dropna(subset=[value_column]).copy()
            matches = deterministic_matches(matchable)
            if len(matches):
                differences = (
                    matchable.loc[matches['success_index'], value_column].to_numpy()
                    - matchable.loc[matches['failure_index'], value_column].to_numpy()
                )
                match_test = mean_test(
                    differences,
                    f'matched:{model}:{mapped_site}:{metric}',
                )
                matching_rows.append({
                    'model': model, 'mapped_site': mapped_site,
                    'metric': metric, 'pairs': len(matches),
                    'mean_match_distance': float(matches['distance'].mean()),
                    **match_test,
                })

        qwen = site[
            (site['model'] == 'Qwen') & site['query_both_correct']
        ][value_column].dropna().to_numpy(dtype=float)
        gemma = site[
            (site['model'] == 'Gemma') & site['query_both_correct']
        ][value_column].dropna().to_numpy(dtype=float)
        rng = np.random.default_rng(stable_seed(
            f'replication-fraction:{mapped_site}:{metric}'
        ))
        qwen_draws = qwen[rng.integers(
            0, len(qwen), size=(BOOTSTRAP_REPLICATES, len(qwen))
        )].mean(axis=1)
        gemma_draws = gemma[rng.integers(
            0, len(gemma), size=(BOOTSTRAP_REPLICATES, len(gemma))
        )].mean(axis=1)
        fraction_contrast = gemma_draws - REPLICATION_FRACTION * qwen_draws
        replication_rows.append({
            'mapped_site': mapped_site, 'metric': metric,
            'qwen_mean': float(qwen.mean()),
            'gemma_mean': float(gemma.mean()),
            'replication_fraction': REPLICATION_FRACTION,
            'contrast_gemma_minus_fraction_qwen': float(
                gemma.mean() - REPLICATION_FRACTION * qwen.mean()
            ),
            'ci_low': float(np.quantile(fraction_contrast, 0.025)),
            'ci_high': float(np.quantile(fraction_contrast, 0.975)),
            'half_replication_excluded': bool(
                np.quantile(fraction_contrast, 0.975) < 0
            ),
        })

sensitivity = pd.DataFrame(sensitivity_rows)
matching = pd.DataFrame(matching_rows)
replication_tests = pd.DataFrame(replication_rows)
matching['q_bh'] = np.nan
for keys, indices in matching.groupby(['model', 'metric']).groups.items():
    matching.loc[indices, 'q_bh'] = bh_adjust(
        matching.loc[indices, 'p_value']
    )
matching['passes_fdr'] = (
    (matching['mean'] > 0)
    & (matching['ci_low'] > 0)
    & (matching['q_bh'] < ALPHA)
)
sensitivity.to_csv(OUTPUT_DIR / 'sensitivity_tests.csv', index=False)
matching.to_csv(OUTPUT_DIR / 'matched_competence_tests.csv', index=False)
replication_tests.to_csv(
    OUTPUT_DIR / 'replication_fraction_tests.csv', index=False
)
print('Matched competence comparisons:')
display(matching)
print('Does Gemma reach at least 50% of Qwen effect?')
display(replication_tests)
```

```python
stage('Applying claim gates and writing the synthesis report')

def within_row(model, site, metric):
    rows = within_tests[
        (within_tests['model'] == model)
        & (within_tests['mapped_site'] == site)
        & (within_tests['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def cross_row(site, metric, test):
    rows = cross_tests[
        (cross_tests['mapped_site'] == site)
        & (cross_tests['metric'] == metric)
        & (cross_tests['test'] == test)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def matched_row(model, site, metric):
    rows = matching[
        (matching['model'] == model)
        & (matching['mapped_site'] == site)
        & (matching['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def sensitivity_stable(model, site, metric):
    rows = sensitivity[
        (sensitivity['model'] == model)
        & (sensitivity['mapped_site'] == site)
        & (sensitivity['metric'] == metric)
    ]
    winsor = rows[rows['analysis'] == 'winsorized_success']
    leave_out = rows[rows['analysis'].str.startswith('leave_out:')]
    return bool(
        len(winsor) == 1
        and winsor.iloc[0]['ci_low'] > 0
        and len(leave_out) == 4
        and (leave_out['mean'] > 0).all()
    )

site_decisions = []
for site in ('all_text', 'options'):
    qwen_recovery = within_row('Qwen', site, 'recovery')
    qwen_margin = within_row('Qwen', site, 'margin')
    gemma_recovery = within_row('Gemma', site, 'recovery')
    gemma_margin = within_row('Gemma', site, 'margin')
    cross_recovery = cross_row(
        site, 'recovery', 'qwen_minus_gemma_success'
    )
    cross_margin = cross_row(
        site, 'margin', 'qwen_minus_gemma_success'
    )
    qwen_signal = bool(
        qwen_recovery['success_pass'] and qwen_margin['success_pass']
        and sensitivity_stable('Qwen', site, 'recovery')
        and sensitivity_stable('Qwen', site, 'margin')
    )
    gemma_signal = bool(
        gemma_recovery['success_pass'] and gemma_margin['success_pass']
        and sensitivity_stable('Gemma', site, 'recovery')
        and sensitivity_stable('Gemma', site, 'margin')
    )
    model_difference = bool(
        cross_recovery['passes_fdr'] and cross_margin['passes_fdr']
    )
    qwen_competence = bool(
        qwen_recovery['competence_pass'] and qwen_margin['competence_pass']
        and matched_row('Qwen', site, 'recovery')['passes_fdr']
        and matched_row('Qwen', site, 'margin')['passes_fdr']
        and sensitivity_stable('Qwen', site, 'recovery')
        and sensitivity_stable('Qwen', site, 'margin')
    )
    gemma_competence = bool(
        gemma_recovery['competence_pass'] and gemma_margin['competence_pass']
        and matched_row('Gemma', site, 'recovery')['passes_fdr']
        and matched_row('Gemma', site, 'margin')['passes_fdr']
        and sensitivity_stable('Gemma', site, 'recovery')
        and sensitivity_stable('Gemma', site, 'margin')
    )
    site_decisions.append({
        'mapped_site': site,
        'qwen_successful_value_signal': qwen_signal,
        'gemma_successful_value_signal': gemma_signal,
        'corrected_cross_model_difference': model_difference,
        'qwen_competence_interaction': qwen_competence,
        'gemma_competence_interaction': gemma_competence,
    })
site_decision_frame = pd.DataFrame(site_decisions)

shared_route = bool((
    site_decision_frame['qwen_successful_value_signal']
    & site_decision_frame['gemma_successful_value_signal']
).any())
model_specific = bool((
    site_decision_frame['qwen_successful_value_signal']
    & ~site_decision_frame['gemma_successful_value_signal']
    & site_decision_frame['corrected_cross_model_difference']
).any())
competence_mechanism = bool((
    site_decision_frame['qwen_competence_interaction']
    | site_decision_frame['gemma_competence_interaction']
).any())

if shared_route:
    route_decision = 'shared_mapped_route_supported'
elif model_specific:
    route_decision = 'qwen_specific_mapped_route_supported'
elif site_decision_frame['qwen_successful_value_signal'].any():
    route_decision = 'gemma_replication_not_supported_model_difference_inconclusive'
else:
    route_decision = 'no_stable_mapped_route_across_models'

competence_decision = (
    'competence_mechanism_supported'
    if competence_mechanism
    else 'competence_mechanism_not_robustly_supported'
)

claim_rows = [
    {
        'claim': 'Physical-value transport is detectable in successful Qwen computations.',
        'status': (
            'supported' if site_decision_frame[
                'qwen_successful_value_signal'
            ].any() else 'not supported'
        ),
        'required_evidence': 'Corrected positive recovery and margin at a mapped site.',
    },
    {
        'claim': 'The same mapped route replicates in Gemma.',
        'status': 'supported' if shared_route else 'not supported',
        'required_evidence': 'Corrected positive recovery and margin in Gemma.',
    },
    {
        'claim': 'The mapped physical-value route differs between models.',
        'status': 'supported' if model_specific else 'inconclusive',
        'required_evidence': 'Qwen signal plus corrected Qwen−Gemma contrasts on both metrics.',
    },
    {
        'claim': 'The route explains success versus failure.',
        'status': 'supported' if competence_mechanism else 'not supported',
        'required_evidence': (
            'Corrected blocked and matched competence interactions on both '
            'metrics, with winsorized and edit leave-out stability.'
        ),
    },
    {
        'claim': 'A universal VLM torque circuit has been identified.',
        'status': 'not supported',
        'required_evidence': 'Would require cross-model replication and finer causal localization.',
    },
]
claim_table = pd.DataFrame(claim_rows)
site_decision_frame.to_csv(OUTPUT_DIR / 'site_decisions.csv', index=False)
claim_table.to_csv(OUTPUT_DIR / 'claim_evidence_table.csv', index=False)

decision = {
    'schema_version': 'qwen-gemma-mechanism-synthesis-decision-1.0',
    'route_decision': route_decision,
    'competence_decision': competence_decision,
    'site_decisions': site_decisions,
    'interpretation_limit': protocol_identity['interpretation_limit'],
    'recommended_next': (
        'If the mapped model difference is supported, freeze it as the '
        'cross-model result and do not search Gemma heads post hoc. If it is '
        'inconclusive, report the failed replication and use a separately '
        'versioned Gemma-wide localization only as exploratory follow-up.'
    ),
}
write_json_atomic(OUTPUT_DIR / 'decision_summary.json', decision)

report_lines = [
    '# Qwen–Gemma grouped-route synthesis', '',
    f'- Qwen analysis: `{QWEN_ANALYSIS}`',
    f'- Gemma analysis: `{GEMMA_ANALYSIS}`',
    f'- Route decision: **`{route_decision}`**',
    f'- Competence decision: **`{competence_decision}`**',
    '', '## Custom-query competence', '',
]
for _, row in competence_summary.iterrows():
    report_lines.append(
        f"- {row['model']}: {int(row['query_both_correct_n'])}/"
        f"{int(row['pairs'])} ({row['query_both_correct_rate']:.1%}) "
        'relevant-query pairs both correct.'
    )
report_lines.extend(['', '## Successful-query effects', ''])
for _, row in within_tests.iterrows():
    report_lines.append(
        f"- {row['model']} / {row['mapped_site']} / {row['metric']}: "
        f"{row['success_mean']:.3f} [{row['success_ci_low']:.3f}, "
        f"{row['success_ci_high']:.3f}], q={row['success_q_bh']:.4f}, "
        f"n={int(row['success_n'])}, pass={bool(row['success_pass'])}."
    )
report_lines.extend(['', '## Cross-model tests', ''])
for _, row in cross_tests.iterrows():
    report_lines.append(
        f"- {row['mapped_site']} / {row['metric']} / {row['test']}: "
        f"{row['effect']:.3f} [{row['ci_low']:.3f}, "
        f"{row['ci_high']:.3f}], q={row['q_bh']:.4f}, "
        f"pass={bool(row['passes_fdr'])}."
    )
report_lines.extend(['', '## Claim–evidence table', ''])
for _, row in claim_table.iterrows():
    report_lines.append(
        f"- **{row['status']}** — {row['claim']} "
        f"Criterion: {row['required_evidence']}"
    )
report_lines.extend([
    '', '## Interpretation limit', '',
    protocol_identity['interpretation_limit'], '',
    '## Recommended next step', '',
    decision['recommended_next'], '',
])
report_path = OUTPUT_DIR / 'qwen_gemma_synthesis_report.md'
report_path.write_text('\n'.join(report_lines) + '\n', encoding='utf-8')
display(site_decision_frame)
display(claim_table)
print(report_path.read_text('utf-8'))
```
