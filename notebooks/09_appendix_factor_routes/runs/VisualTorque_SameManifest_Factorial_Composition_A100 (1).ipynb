```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. Model-independent factorial manifest

Selection uses only exact physical scene values and counterfactual metadata.


```python
%%cell_progress 6/17 Build or verify the model-independent factorial manifest
cell_stage('Loading scenes, counterfactual pairs, and prior pair IDs')
from collections import defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
pair_records = read_jsonl(DATASET_DIR / 'counterfactual_pairs.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}

prior_manifest_paths = (
    PROJECT_ROOT / 'results' / 'Qwen__Qwen3.5-9B'
    / 'natural_counterfactual_causal_tracing_v1' / 'pair_manifest.jsonl',
    PROJECT_ROOT / 'results' / 'Qwen__Qwen3.5-9B'
    / 'counterfactual_causal_tracing_confirmatory_v1' / 'pair_manifest.jsonl',
    PROJECT_ROOT / 'results' / 'Qwen__Qwen3.5-9B'
    / 'task_token_component_routing_v1' / 'pair_manifest.jsonl',
    PROJECT_ROOT / 'results' / 'Qwen__Qwen3.5-9B'
    / 'competence_grouped_route_confirmatory_v1' / 'pair_manifest.jsonl',
    PROJECT_ROOT / 'results' / 'google__gemma-4-12B-it'
    / 'competence_grouped_route_gemma_replication_v3' / 'pair_manifest.jsonl',
)
prior_pair_ids = set()
for path in prior_manifest_paths:
    if path.is_file():
        prior_pair_ids.update(
            str(row['pair_id']) for row in read_jsonl(path)
        )

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

def stacks_on_side(scene, side):
    return [stack for stack in scene['stacks'] if stack['side'] == side]

def single_stack(scene, side):
    values = stacks_on_side(scene, side)
    return values[0] if len(values) == 1 else None

def atomic_pair_change(pair):
    source = scene_by_id[str(pair['source_scene_id'])]
    target = scene_by_id[str(pair['target_scene_id'])]
    changed = []
    for side in ('left', 'right'):
        source_stack = single_stack(source, side)
        target_stack = single_stack(target, side)
        if source_stack is None or target_stack is None:
            return None
        difference = (
            int(source_stack['units']) != int(target_stack['units'])
            or int(source_stack['distance']) != int(target_stack['distance'])
        )
        if difference:
            changed.append((side, source_stack, target_stack))
    if len(changed) != 1:
        return None
    side, source_stack, target_stack = changed[0]
    delta_units = int(source_stack['units']) - int(target_stack['units'])
    delta_distance = (
        int(source_stack['distance']) - int(target_stack['distance'])
    )
    delta_side_torque = (
        int(source_stack['torque']) - int(target_stack['torque'])
    )
    observed_delta = (
        int(source['derived']['net_torque'])
        - int(target['derived']['net_torque'])
    )
    expected_delta = (
        delta_side_torque if side == 'left' else -delta_side_torque
    )
    if observed_delta != expected_delta:
        return None
    if int(pair['delta_net_torque']) != observed_delta:
        return None
    if pair['edit_type'] == 'add_unit':
        if delta_units == 0 or delta_distance != 0:
            return None
    elif pair['edit_type'] == 'move_stack':
        if delta_units != 0 or delta_distance == 0:
            return None
    else:
        return None
    return {
        'changed_side': side,
        'delta_units': delta_units,
        'delta_distance': delta_distance,
        'delta_side_torque': delta_side_torque,
        'delta_net_torque_verified': observed_delta,
    }

def task_value(scene, task, side):
    derived = scene['derived']
    if task == 'side_units':
        return int(derived[f'{side}_units'])
    if task == 'side_distance':
        stack = single_stack(scene, side)
        assert stack is not None
        return int(stack['distance'])
    if task == 'side_torque':
        return int(derived[f'{side}_torque'])
    if task == 'net_torque':
        return int(derived['net_torque'])
    if task == 'outcome':
        return str(derived['outcome'])
    raise KeyError(task)

cell_stage('Enumerating same-base, same-delta weight/distance triplets')
grouped = defaultdict(lambda: defaultdict(list))
for pair in pair_records:
    if pair.get('split') != 'intervention_test':
        continue
    if pair.get('edit_type') not in ('add_unit', 'move_stack'):
        continue
    change = atomic_pair_change(pair)
    if change is None:
        continue
    key = (
        str(pair['target_scene_id']),
        int(pair['delta_net_torque']),
        change['changed_side'],
    )
    grouped[key][str(pair['edit_type'])].append((pair, change))

candidates = []
for (base_scene_id, delta_net_torque, side), by_edit in grouped.items():
    if not by_edit['add_unit'] or not by_edit['move_stack']:
        continue
    weight_pair, weight_change = sorted(
        by_edit['add_unit'], key=lambda item: stable_key(item[0]['pair_id'])
    )[0]
    distance_pair, distance_change = sorted(
        by_edit['move_stack'], key=lambda item: stable_key(item[0]['pair_id'])
    )[0]
    base = scene_by_id[base_scene_id]
    weight = scene_by_id[str(weight_pair['source_scene_id'])]
    distance = scene_by_id[str(distance_pair['source_scene_id'])]
    if len({base['scene_id'], weight['scene_id'], distance['scene_id']}) != 3:
        continue

    values = {
        task: {
            'base': task_value(base, task, side),
            'weight_source': task_value(weight, task, side),
            'distance_source': task_value(distance, task, side),
        }
        for task in TASKS
    }
    valid = (
        values['side_units']['weight_source']
        != values['side_units']['base']
        and values['side_units']['distance_source']
        == values['side_units']['base']
        and values['side_distance']['distance_source']
        != values['side_distance']['base']
        and values['side_distance']['weight_source']
        == values['side_distance']['base']
        and values['side_torque']['weight_source']
        == values['side_torque']['distance_source']
        != values['side_torque']['base']
        and values['net_torque']['weight_source']
        == values['net_torque']['distance_source']
        != values['net_torque']['base']
        and values['outcome']['weight_source']
        == values['outcome']['distance_source']
    )
    if not valid:
        continue
    pair_ids = (
        str(weight_pair['pair_id']), str(distance_pair['pair_id'])
    )
    triplet_id = 'factorial_' + hashlib.sha256(
        '|'.join((base_scene_id, *pair_ids)).encode('utf-8')
    ).hexdigest()[:18]
    candidates.append({
        'triplet_id': triplet_id,
        'base_scene_id': base_scene_id,
        'weight_scene_id': str(weight['scene_id']),
        'distance_scene_id': str(distance['scene_id']),
        'weight_pair_id': pair_ids[0],
        'distance_pair_id': pair_ids[1],
        'changed_side': side,
        'delta_net_torque': int(delta_net_torque),
        'delta_side_torque': int(weight_change['delta_side_torque']),
        'outcome_changed': (
            values['outcome']['base']
            != values['outcome']['weight_source']
        ),
        'prior_pair_overlap': any(
            pair_id in prior_pair_ids for pair_id in pair_ids
        ),
        'base_problem_type': str(base['problem_type']),
        'weight_problem_type': str(weight['problem_type']),
        'distance_problem_type': str(distance['problem_type']),
        'task_values': values,
    })

assert len(candidates) >= MIN_TRIPLETS, (
    f'Only {len(candidates)} valid factorial triplets; need {MIN_TRIPLETS}.'
)

def stratum(row):
    return (
        bool(row['prior_pair_overlap']),
        str(row['changed_side']),
        int(np.sign(row['delta_net_torque'])),
        int(bool(row['outcome_changed'])),
        min(abs(int(row['delta_net_torque'])), 4),
    )

groups = defaultdict(list)
for row in candidates:
    groups[stratum(row)].append(row)
for rows in groups.values():
    rows.sort(key=lambda row: stable_key(row['triplet_id']))
ordered_keys = sorted(
    groups,
    key=lambda key: (
        key[0], stable_key('stratum:' + repr(key))
    ),
)
selected = []
used_scenes = set()
cursor = {key: 0 for key in ordered_keys}
while len(selected) < MAX_TRIPLETS:
    changed = False
    for key in ordered_keys:
        rows = groups[key]
        while cursor[key] < len(rows):
            row = rows[cursor[key]]
            cursor[key] += 1
            scene_ids = {
                row['base_scene_id'], row['weight_scene_id'],
                row['distance_scene_id'],
            }
            if scene_ids & used_scenes:
                continue
            selected.append(row)
            used_scenes.update(scene_ids)
            changed = True
            break
        if len(selected) == MAX_TRIPLETS:
            break
    if not changed:
        break
assert len(selected) >= MIN_TRIPLETS, (
    f'Only {len(selected)} scene-disjoint triplets; need {MIN_TRIPLETS}.'
)
selected.sort(key=lambda row: str(row['triplet_id']))

manifest_identity = {
    'schema_version': 'visualtorque-factorial-triplet-manifest-1.0',
    'dataset_scene_count': len(scene_records),
    'dataset_pair_count': len(pair_records),
    'selection_seed': SEED,
    'max_triplets': MAX_TRIPLETS,
    'min_triplets': MIN_TRIPLETS,
    'candidate_count': len(candidates),
    'selected_count': len(selected),
    'selection_uses_model_information': False,
    'scene_disjoint_across_triplets': True,
    'triplet_ids': [row['triplet_id'] for row in selected],
}
if MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity, (
        'The existing shared manifest was built from a different dataset or '
        'selection protocol. Use a new output version.'
    )
    existing = read_jsonl(MANIFEST_PATH)
    assert existing == selected, 'Shared manifest content drifted.'
else:
    write_jsonl_atomic(MANIFEST_PATH, selected)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
factorial_manifest = read_jsonl(MANIFEST_PATH)
manifest_frame = pd.DataFrame([
    {key: value for key, value in row.items() if key != 'task_values'}
    for row in factorial_manifest
])
manifest_frame.to_csv(SHARED_ROOT / 'factorial_triplet_manifest.csv', index=False)
print('Candidate triplets:', len(candidates))
print('Selected scene-disjoint triplets:', len(factorial_manifest))
print('Prior-pair overlap:', int(manifest_frame['prior_pair_overlap'].sum()))
display(manifest_frame.groupby(
    ['changed_side', 'outcome_changed', 'prior_pair_overlap']
).size().rename('n').reset_index())

discovery_config_path = (
    PROJECT_ROOT / 'results' / MODEL_SLUG
    / 'activation_discovery_v1' / 'capture_config.json'
)
assert discovery_config_path.is_file(), (
    f'Missing prior activation discovery artifact: {discovery_config_path}'
)
discovery_config = json.loads(
    discovery_config_path.read_text(encoding='utf-8')
)
MODEL_REVISION = str(
    discovery_config['identity']['resolved_revision']
)

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=SHARED_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example
    for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)
print('Pinned revision:', MODEL_REVISION)
```

## 3. Causal transport on factor, torque, and outcome queries

The two source scenes share the same resulting torque but reach it through different factors.


```python
%%cell_progress 8/17 Define factorial tasks, aligned prompts, and exact masks
cell_stage('Defining factor, torque, and outcome queries')
import math
from dataclasses import replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side and compare the two side totals.'
)

def task_question(task, side):
    if task == 'side_units':
        return f'How many blocks are on the {side} side?'
    if task == 'side_distance':
        return (
            f'How many tick intervals from the pivot is the stack on the '
            f'{side} side?'
        )
    if task == 'side_torque':
        return f'What is the total {side}-side torque?'
    if task == 'net_torque':
        return 'What is left torque minus right torque?'
    if task == 'outcome':
        return 'Which side will rotate downward?'
    raise KeyError(task)

def option_semantics(row, task):
    values = row['task_values'][task]
    if task == 'outcome':
        semantics = ['left', 'balanced', 'right']
    else:
        semantics = []
        for name in ('base', 'weight_source', 'distance_source'):
            value = int(values[name])
            if value not in semantics:
                semantics.append(value)
        anchor = semantics[0]
        for value in (
            anchor + 1, anchor - 1, 0, 1, -1,
            anchor + 2, anchor - 2, anchor + 3, anchor - 3,
        ):
            if len(semantics) == 3:
                break
            if value not in semantics:
                semantics.append(value)
        assert len(semantics) == 3
    offset = int(stable_key(
        f"options:{row['triplet_id']}:{task}"
    )[:8], 16) % 3
    return tuple(semantics[offset:] + semantics[:offset])

def display_value(task, value):
    if task != 'outcome':
        return str(value)
    return {
        'left': 'left side',
        'balanced': 'balanced',
        'right': 'right side',
    }[str(value)]

def custom_prompt(task, side, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {display_value(task, value)}'
        for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + task_question(task, side),
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_task_bundle(row, task):
    side = str(row['changed_side'])
    semantics = option_semantics(row, task)
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, side, semantics)
    scene_ids = {
        'base': str(row['base_scene_id']),
        'weight_source': str(row['weight_scene_id']),
        'distance_source': str(row['distance_scene_id']),
    }
    examples = {}
    for name, scene_id in scene_ids.items():
        value = row['task_values'][task][name]
        if task != 'outcome':
            value = int(value)
        else:
            value = str(value)
        assert value in semantics, (task, name, value, semantics)
        base = outcome_by_scene[scene_id]
        gold_index = semantics.index(value)
        examples[name] = replace(
            base,
            example_id=(
                f"image_rule:factorial_{row['triplet_id']}_{task}_{name}"
            ),
            question_id=f"factorial_{row['triplet_id']}_{task}",
            question_kind=task,
            variant='same_manifest_factorial_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(
                display_value(task, value) for value in semantics
            ),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    goal_name = {
        'side_units': 'weight_source',
        'side_distance': 'distance_source',
        'side_torque': 'weight_source',
        'net_torque': 'weight_source',
        'outcome': 'weight_source',
    }[task]
    return {
        'task': task,
        'semantics': semantics,
        'examples': examples,
        'base_semantic': examples['base'].gold_semantic,
        'goal_semantic': examples[goal_name].gold_semantic,
        'goal_name': goal_name,
        'estimable': (
            examples['base'].gold_semantic
            != examples[goal_name].gold_semantic
        ),
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(
                reference['inputs'][key], current['inputs'][key]
            ), (name, key)
        masks = discover_masks(current)
        for _, scope, _ in FROZEN_SITES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Factorial prompts and exact masks ready.')
```

```python
%%cell_progress 9/17 Define decoder capture and batched patch scorers
cell_stage('Defining candidate-space scoring and decoder patching')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': (
                example.option_semantics[predicted_index]
                == example.gold_semantic
            ),
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = (
                output_hidden_tensor(output).detach().cpu().clone()
            )
        handles.append(
            DECODER_LAYERS[layer_index].register_forward_hook(hook)
        )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def forward_patch_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert len(layer_indices) == 1
    layer_index = next(iter(layer_indices))
    donors = torch.stack([row['donor_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = donors.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == donor.shape
        modified = hidden.clone()
        modified[mask] = donor[mask]
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('donor_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)

    def run_chunk(chunk):
        actual_size = len(chunk)
        assert 1 <= actual_size <= PATCH_CONDITION_BATCH_SIZE
        padded = list(chunk)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        size = len(padded)
        prepared = prepared_cache.get(size)
        if prepared is None:
            prepared = prepare_examples([target_example] * size)
            prepared_cache[size] = prepared
        try:
            return forward_patch_chunk(prepared, padded)[:actual_size]
        except torch.cuda.OutOfMemoryError as exc:
            prepared_cache.pop(size, None)
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Fixed batch-size patching ran out of memory. Restart with '
                'an A100; changing batch size changes numerical parity.'
            ) from exc

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

print('Capture and patch scorers ready.')
```

## 4. Confirmatory analysis and exact cross-model synthesis

The primary metric is a denominator-free changed-vs-base candidate-margin shift.


```python
%%cell_progress 12/17 Release GPU and assemble denominator-free metrics
cell_stage('Releasing model memory')
del adapter
gc.collect()
torch.cuda.empty_cache()

cell_stage('Reading triplet shards and computing pair-level effects')
long_rows = []
paired_rows = []

def centered_vector(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def semantic_margin(record, goal_semantic, base_semantic):
    semantics = list(record['candidate_semantics'])
    goal_index = semantics.index(goal_semantic)
    base_index = semantics.index(base_semantic)
    scores = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(scores[goal_index] - scores[base_index])

def cosine(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    denominator = float(np.linalg.norm(left) * np.linalg.norm(right))
    if denominator <= 1e-12:
        return np.nan
    return float(np.dot(left, right) / denominator)

for path in tqdm(
    sorted(SHARD_DIR.glob('shard_triplet_*.json')),
    desc='Assembling triplet metrics', unit='triplet',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    row = payload['triplet']
    patches = pd.DataFrame(payload['patch_records'])
    for task in TASKS:
        baseline = payload['baselines'][task]
        base_record = baseline['scores']['base']
        weight_record = baseline['scores']['weight_source']
        distance_record = baseline['scores']['distance_source']
        base_semantic = baseline['base_semantic']
        goal_semantic = baseline['goal_semantic']
        estimable = bool(baseline['estimable'])
        base_vector = centered_vector(base_record)
        goal_record = baseline['scores'][baseline['goal_name']]
        goal_vector = centered_vector(goal_record)
        natural_vector = goal_vector - base_vector
        if estimable:
            base_margin = semantic_margin(
                base_record, goal_semantic, base_semantic
            )
            goal_margin = semantic_margin(
                goal_record, goal_semantic, base_semantic
            )
            natural_margin = goal_margin - base_margin
        else:
            base_margin = np.nan
            goal_margin = np.nan
            natural_margin = np.nan

        task_patches = patches[patches['task'] == task]
        for _, patch in task_patches.iterrows():
            patch_record = patch.to_dict()
            patch_vector = centered_vector(patch_record)
            displacement = patch_vector - base_vector
            if estimable:
                patch_margin = semantic_margin(
                    patch_record, goal_semantic, base_semantic
                )
                margin_shift = patch_margin - base_margin
                recovery = (
                    margin_shift / natural_margin
                    if abs(natural_margin) >= NATURAL_EFFECT_EPS
                    else np.nan
                )
            else:
                patch_margin = np.nan
                margin_shift = np.nan
                recovery = np.nan
            long_rows.append({
                'model_key': MODEL_KEY,
                'model_id': MODEL_ID,
                'triplet_id': row['triplet_id'],
                'changed_side': row['changed_side'],
                'delta_net_torque': row['delta_net_torque'],
                'outcome_changed': bool(row['outcome_changed']),
                'prior_pair_overlap': bool(row['prior_pair_overlap']),
                'task': task,
                'estimable': estimable,
                'donor_type': str(patch['donor_type']),
                'site_label': str(patch['site_label']),
                'layer_index': int(patch['layer_index']),
                'scope': str(patch['scope']),
                'base_semantic': base_semantic,
                'goal_semantic': goal_semantic,
                'margin_shift': margin_shift,
                'natural_margin': natural_margin,
                'recovery': recovery,
                'vector_alignment_to_natural': cosine(
                    displacement, natural_vector
                ),
                'displacement_0': float(displacement[0]),
                'displacement_1': float(displacement[1]),
                'displacement_2': float(displacement[2]),
                'base_correct': bool(base_record['correct']),
                'weight_source_correct': bool(weight_record['correct']),
                'distance_source_correct': bool(distance_record['correct']),
                'all_three_correct': bool(
                    base_record['correct']
                    and weight_record['correct']
                    and distance_record['correct']
                ),
                'patched_tokens': int(patch['patched_tokens']),
            })

        for _, _, site_label in FROZEN_SITES:
            site_records = [
                value for value in long_rows
                if value['triplet_id'] == row['triplet_id']
                and value['task'] == task
                and value['site_label'] == site_label
            ]
            assert len(site_records) == 2, (
                row['triplet_id'], task, site_label, len(site_records)
            )
            by_donor = {
                value['donor_type']: value for value in site_records
            }
            weight_effect = by_donor['weight_source']['margin_shift']
            distance_effect = by_donor['distance_source']['margin_shift']
            weight_disp = np.asarray([
                by_donor['weight_source'][f'displacement_{index}']
                for index in range(3)
            ])
            distance_disp = np.asarray([
                by_donor['distance_source'][f'displacement_{index}']
                for index in range(3)
            ])
            if task == 'side_units':
                factor_selectivity = weight_effect - distance_effect
            elif task == 'side_distance':
                factor_selectivity = distance_effect - weight_effect
            else:
                factor_selectivity = np.nan
            paired_rows.append({
                'model_key': MODEL_KEY,
                'model_id': MODEL_ID,
                'triplet_id': row['triplet_id'],
                'changed_side': row['changed_side'],
                'delta_net_torque': row['delta_net_torque'],
                'outcome_changed': bool(row['outcome_changed']),
                'prior_pair_overlap': bool(row['prior_pair_overlap']),
                'task': task,
                'estimable': estimable,
                'site_label': site_label,
                'weight_effect': weight_effect,
                'distance_effect': distance_effect,
                'mean_donor_effect': (
                    np.nanmean([weight_effect, distance_effect])
                    if estimable else np.nan
                ),
                'donor_effect_difference': (
                    weight_effect - distance_effect
                    if estimable else np.nan
                ),
                'abs_donor_gap': (
                    abs(weight_effect - distance_effect)
                    if estimable else np.nan
                ),
                'factor_selectivity': factor_selectivity,
                'displacement_cosine': cosine(weight_disp, distance_disp),
                'natural_margin': natural_margin,
                'all_three_correct': by_donor[
                    'weight_source'
                ]['all_three_correct'],
            })

long_frame = pd.DataFrame(long_rows)
paired_frame = pd.DataFrame(paired_rows)
expected_long = (
    len(factorial_manifest) * len(TASKS)
    * len(FROZEN_SITES) * len(DONOR_TYPES)
)
expected_paired = (
    len(factorial_manifest) * len(TASKS) * len(FROZEN_SITES)
)
assert len(long_frame) == expected_long
assert len(paired_frame) == expected_paired
assert not long_frame.duplicated([
    'triplet_id', 'task', 'site_label', 'donor_type'
]).any()
long_frame.to_csv(ANALYSIS_DIR / 'patch_level_metrics.csv', index=False)
paired_frame.to_csv(ANALYSIS_DIR / 'paired_triplet_metrics.csv', index=False)
baseline_accuracy = (
    long_frame.groupby(['task'])[
        ['base_correct', 'weight_source_correct', 'distance_source_correct']
    ].mean().reset_index()
)
baseline_accuracy.to_csv(
    ANALYSIS_DIR / 'baseline_task_accuracy.csv', index=False
)
display(baseline_accuracy)
print('Patch rows:', len(long_frame), '| Paired rows:', len(paired_frame))
```

```python
%%cell_progress 13/17 Run frozen factor-separation and torque-convergence tests
cell_stage('Defining deterministic paired bootstrap and sign-flip tests')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def paired_summary(frame, column, family, task, site_label, population):
    values = pd.to_numeric(frame[column], errors='coerce').to_numpy(float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {
            'family': family, 'metric': column, 'task': task,
            'site_label': site_label, 'population': population,
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(
        MODEL_KEY, family, task, site_label, population, column
    ))
    indices = rng.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    boot = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice(
        (-1.0, 1.0),
        size=(SIGN_FLIP_REPLICATES, len(values)),
    )
    null = (signs * values).mean(axis=1)
    p_value = float(
        (1 + np.sum(np.abs(null) >= abs(observed)))
        / (SIGN_FLIP_REPLICATES + 1)
    )
    return {
        'family': family, 'metric': column, 'task': task,
        'site_label': site_label, 'population': population,
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
        'sign_flip_p': p_value,
    }

summary_rows = []
populations = {
    'all': paired_frame,
    'all_three_correct_secondary': paired_frame[
        paired_frame['all_three_correct']
    ],
}
metric_plan = {
    'side_units': ('factor_selectivity',),
    'side_distance': ('factor_selectivity',),
    'side_torque': (
        'weight_effect', 'distance_effect', 'mean_donor_effect',
        'donor_effect_difference', 'displacement_cosine',
    ),
    'net_torque': (
        'weight_effect', 'distance_effect', 'mean_donor_effect',
        'donor_effect_difference', 'displacement_cosine',
    ),
    'outcome': (
        'weight_effect', 'distance_effect', 'mean_donor_effect',
        'donor_effect_difference', 'displacement_cosine',
    ),
}
for population, population_frame in populations.items():
    for task, metrics in metric_plan.items():
        task_frame = population_frame[
            (population_frame['task'] == task)
            & population_frame['estimable']
        ]
        for _, _, site_label in FROZEN_SITES:
            site_frame = task_frame[
                task_frame['site_label'] == site_label
            ]
            for metric in metrics:
                summary_rows.append(paired_summary(
                    site_frame, metric,
                    'factor' if task.startswith('side_') and task != 'side_torque'
                    else 'torque' if task in ('side_torque', 'net_torque')
                    else 'outcome',
                    task, site_label, population,
                ))
summary_frame = pd.DataFrame(summary_rows)

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

summary_frame['q_value'] = np.nan
for keys, indices in summary_frame.groupby(
    ['population', 'family', 'metric']
).groups.items():
    summary_frame.loc[indices, 'q_value'] = bh_adjust(
        summary_frame.loc[indices, 'sign_flip_p'].to_numpy(float)
    )

summary_frame['equivalence_bound'] = np.nan
summary_frame['equivalent'] = False
all_population = paired_frame[paired_frame['estimable']]
for task in ('side_torque', 'net_torque', 'outcome'):
    for _, _, site_label in FROZEN_SITES:
        subset = all_population[
            (all_population['task'] == task)
            & (all_population['site_label'] == site_label)
        ]
        scale_values = np.abs(
            pd.to_numeric(subset['natural_margin'], errors='coerce')
        )
        scale_values = scale_values[np.isfinite(scale_values)]
        bound = (
            EQUIVALENCE_FRACTION * float(np.median(scale_values))
            if len(scale_values) else np.nan
        )
        mask = (
            (summary_frame['population'] == 'all')
            & (summary_frame['task'] == task)
            & (summary_frame['site_label'] == site_label)
            & (summary_frame['metric'] == 'donor_effect_difference')
        )
        summary_frame.loc[mask, 'equivalence_bound'] = bound
        summary_frame.loc[mask, 'equivalent'] = (
            (summary_frame.loc[mask, 'ci_low'] > -bound)
            & (summary_frame.loc[mask, 'ci_high'] < bound)
        )

summary_frame.to_csv(
    ANALYSIS_DIR / 'confirmatory_summary.csv', index=False
)
display(summary_frame[
    (summary_frame['population'] == 'all')
    & summary_frame['site_label'].isin(('middle_text', 'middle_options'))
    & summary_frame['metric'].isin((
        'factor_selectivity', 'weight_effect', 'distance_effect',
        'donor_effect_difference',
    ))
].sort_values(['site_label', 'task', 'metric']))
```

```python
%%cell_progress 14/17 Test layerwise convergence and competence boundaries
cell_stage('Comparing early image and middle text transport')
convergence_rows = []
for task in ('side_torque', 'net_torque'):
    task_frame = paired_frame[
        (paired_frame['task'] == task) & paired_frame['estimable']
    ]
    for metric, direction in (
        ('displacement_cosine', 1.0),
        ('abs_donor_gap', -1.0),
    ):
        pivot = task_frame.pivot(
            index='triplet_id', columns='site_label', values=metric
        )
        pivot = pivot.dropna(subset=['early_image', 'middle_text'])
        pivot['convergence_gain'] = direction * (
            pivot['middle_text'] - pivot['early_image']
        )
        convergence_rows.append(paired_summary(
            pivot.reset_index(), 'convergence_gain',
            'layerwise_convergence', task,
            'middle_text_minus_early_image', 'all',
        ))
convergence_frame = pd.DataFrame(convergence_rows)
convergence_frame.to_csv(
    ANALYSIS_DIR / 'layerwise_convergence.csv', index=False
)
display(convergence_frame)

cell_stage('Auditing correctness without changing the primary population')
competence_base = paired_frame[
    ['triplet_id', 'task', 'all_three_correct']
].drop_duplicates()
competence_counts = (
    competence_base.groupby(['task'])
    .agg(
        triplets=('triplet_id', 'nunique'),
        all_three_correct=('all_three_correct', 'sum'),
    )
    .reset_index()
)
competence_counts['all_three_correct_rate'] = (
    competence_counts['all_three_correct']
    / competence_counts['triplets']
)
competence_counts.to_csv(
    ANALYSIS_DIR / 'competence_counts.csv', index=False
)
display(competence_counts)

secondary = summary_frame[
    (summary_frame['population'] == 'all_three_correct_secondary')
    & summary_frame['site_label'].isin(('middle_text', 'middle_options'))
    & summary_frame['metric'].isin((
        'factor_selectivity', 'weight_effect', 'distance_effect',
        'donor_effect_difference',
    ))
].copy()
secondary.to_csv(
    ANALYSIS_DIR / 'competence_secondary_summary.csv', index=False
)
print(
    'Primary inference retains every triplet. The table above is a '
    'predeclared secondary boundary analysis only.'
)
```

```python
%%cell_progress 16/17 Apply the frozen mechanistic gate and write the report
cell_stage('Evaluating factor, combination, and use gates')

def summary_record(task, site_label, metric):
    rows = summary_frame[
        (summary_frame['population'] == 'all')
        & (summary_frame['task'] == task)
        & (summary_frame['site_label'] == site_label)
        & (summary_frame['metric'] == metric)
    ]
    assert len(rows) == 1, (task, site_label, metric, len(rows))
    return rows.iloc[0]

focal_sites = ('middle_text', 'middle_options')
factor_site_gates = {}
torque_site_gates = {}
for site_label in focal_sites:
    units = summary_record('side_units', site_label, 'factor_selectivity')
    distance = summary_record(
        'side_distance', site_label, 'factor_selectivity'
    )
    factor_site_gates[site_label] = bool(
        units['ci_low'] > 0 and units['q_value'] < ALPHA
        and distance['ci_low'] > 0 and distance['q_value'] < ALPHA
    )

    task_gates = {}
    for task in ('side_torque', 'net_torque'):
        weight = summary_record(task, site_label, 'weight_effect')
        distance_effect = summary_record(
            task, site_label, 'distance_effect'
        )
        difference = summary_record(
            task, site_label, 'donor_effect_difference'
        )
        task_gates[task] = bool(
            weight['ci_low'] > 0 and weight['q_value'] < ALPHA
            and distance_effect['ci_low'] > 0
            and distance_effect['q_value'] < ALPHA
            and bool(difference['equivalent'])
        )
    torque_site_gates[site_label] = bool(all(task_gates.values()))

outcome_weight = summary_record('outcome', 'late_answer', 'weight_effect')
outcome_distance = summary_record(
    'outcome', 'late_answer', 'distance_effect'
)
outcome_difference = summary_record(
    'outcome', 'late_answer', 'donor_effect_difference'
)
late_outcome_use = bool(
    outcome_weight['n'] >= 12
    and outcome_weight['ci_low'] > 0
    and outcome_weight['q_value'] < ALPHA
    and outcome_distance['ci_low'] > 0
    and outcome_distance['q_value'] < ALPHA
    and bool(outcome_difference['equivalent'])
)
factor_separation = bool(any(factor_site_gates.values()))
torque_convergence = bool(any(torque_site_gates.values()))
full_chain = bool(
    any(
        factor_site_gates[site] and torque_site_gates[site]
        for site in focal_sites
    )
    and late_outcome_use
)

if full_chain:
    decision = 'full_factor_to_torque_to_outcome_chain_supported'
elif torque_convergence and factor_separation:
    decision = 'factor_to_torque_composition_supported_outcome_use_incomplete'
elif torque_convergence:
    decision = 'torque_equivalent_transport_supported_factor_chain_incomplete'
elif factor_separation:
    decision = 'factor_specific_transport_supported_torque_convergence_incomplete'
else:
    decision = 'factorial_composition_chain_not_supported_at_frozen_sites'

decision_payload = {
    'schema_version': 'factorial-composition-decision-1.0',
    'model_key': MODEL_KEY,
    'model_id': MODEL_ID,
    'decision': decision,
    'factor_separation_supported': factor_separation,
    'torque_convergence_supported': torque_convergence,
    'late_outcome_use_supported': late_outcome_use,
    'full_composition_chain_supported': full_chain,
    'factor_site_gates': factor_site_gates,
    'torque_site_gates': torque_site_gates,
    'triplets': len(factorial_manifest),
    'outcome_changed_triplets': int(
        pd.Series([row['outcome_changed'] for row in factorial_manifest]).sum()
    ),
    'primary_population': 'all triplets',
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'capture_config_sha256': CAPTURE_SHA256,
    'limitations': [
        'Mapped grouped decoder outputs are tested, not individual heads or neurons.',
        'Positive equivalence supports a shared transported task value, not a unique multiplication algorithm.',
        'A negative gate is local to the frozen prompts, sites, and patch operation.',
    ],
}
write_json_atomic(ANALYSIS_DIR / 'factorial_composition_decision.json', decision_payload)

report_lines = [
    f'# Same-manifest factorial composition — {MODEL_KEY}',
    '',
    f'- Decision: `{decision}`',
    f'- Triplets: {len(factorial_manifest)} (all enter primary inference)',
    f'- Factor separation: {factor_separation}',
    f'- Torque convergence: {torque_convergence}',
    f'- Late outcome use: {late_outcome_use}',
    f'- Full factor → torque → outcome chain: {full_chain}',
    '',
    '## Interpretation',
    '',
    (
        'The experiment holds the base image and resulting torque change fixed '
        'while changing whether that torque arose from weight or distance. '
        'Factor selectivity tests whether the donor routes retain their distinct '
        'physical content; torque equivalence tests whether those routes converge '
        'on the same task-relevant causal effect.'
    ),
    '',
    'Correctness-conditioned results are secondary and do not define the sample.',
    '',
    '## Limits',
    '',
    '- This is evidence about causal transport at mapped grouped sites.',
    '- It does not by itself identify literal multiplication or a unique circuit.',
    '- Failure does not imply that the model lacks torque information elsewhere.',
]
report_path = ANALYSIS_DIR / 'factorial_composition_report.md'
report_path.write_text('\n'.join(report_lines) + '\n', encoding='utf-8')
print(json.dumps(decision_payload, indent=2))
print('Report:', report_path)
```

```python
%%cell_progress 17/17 Run the optional exact-pair cross-model synthesis
cell_stage('Checking whether both model runs are complete')
model_metric_paths = {
    key: (
        SHARED_ROOT / 'models' / spec['model_slug']
        / 'analysis' / 'paired_triplet_metrics.csv'
    )
    for key, spec in MODEL_SPECS.items()
}
model_decision_paths = {
    key: (
        SHARED_ROOT / 'models' / spec['model_slug']
        / 'analysis' / 'factorial_composition_decision.json'
    )
    for key, spec in MODEL_SPECS.items()
}
if not all(path.is_file() for path in model_metric_paths.values()):
    missing = [
        key for key, path in model_metric_paths.items() if not path.is_file()
    ]
    print(
        'Cross-model synthesis is waiting for:', missing,
        '\nRun this same notebook in a fresh runtime with MODEL_KEY set to '
        'the missing model. All current model artifacts are complete.'
    )
else:
    cell_stage('Merging exact triplet, task, and site rows')
    frames = {
        key: pd.read_csv(path) for key, path in model_metric_paths.items()
    }
    join_keys = ['triplet_id', 'task', 'site_label']
    for key, frame in frames.items():
        assert not frame.duplicated(join_keys).any(), key
    qwen_ids = set(frames['qwen']['triplet_id'])
    gemma_ids = set(frames['gemma']['triplet_id'])
    assert qwen_ids == gemma_ids == {
        row['triplet_id'] for row in factorial_manifest
    }
    metrics = (
        'factor_selectivity', 'weight_effect', 'distance_effect',
        'mean_donor_effect', 'donor_effect_difference',
        'displacement_cosine',
    )
    merged = frames['qwen'][join_keys + list(metrics)].merge(
        frames['gemma'][join_keys + list(metrics)],
        on=join_keys, how='inner', validate='one_to_one',
        suffixes=('_qwen', '_gemma'),
    )
    assert len(merged) == len(frames['qwen']) == len(frames['gemma'])

    cross_rows = []
    for task in TASKS:
        for site_label in (
            'early_image', 'middle_text', 'middle_options', 'late_answer'
        ):
            subset = merged[
                (merged['task'] == task)
                & (merged['site_label'] == site_label)
            ].copy()
            for metric in metrics:
                difference_column = f'{metric}_qwen_minus_gemma'
                subset[difference_column] = (
                    pd.to_numeric(subset[f'{metric}_qwen'], errors='coerce')
                    - pd.to_numeric(subset[f'{metric}_gemma'], errors='coerce')
                )
                cross_rows.append(paired_summary(
                    subset, difference_column,
                    'exact_same_manifest_cross_model',
                    task, site_label, 'all',
                ))
    cross_frame = pd.DataFrame(cross_rows)
    cross_frame.to_csv(
        CROSS_MODEL_DIR / 'exact_pair_cross_model_summary.csv', index=False
    )
    merged.to_csv(
        CROSS_MODEL_DIR / 'exact_pair_merged_metrics.csv', index=False
    )

    decisions = {
        key: json.loads(path.read_text(encoding='utf-8'))
        for key, path in model_decision_paths.items()
    }
    synthesis = {
        'schema_version': 'same-manifest-cross-model-synthesis-1.0',
        'manifest_sha256': sha256_file(MANIFEST_PATH),
        'exact_shared_triplets': len(qwen_ids),
        'exact_shared_rows': len(merged),
        'qwen_decision': decisions['qwen']['decision'],
        'gemma_decision': decisions['gemma']['decision'],
        'primary_cross_model_estimand': 'paired Qwen minus Gemma on identical triplets',
        'support_status': 'exact_same_stimulus_support_complete',
        'interpretation': (
            'Differences can be attributed to the two frozen model/site '
            'systems on this manifest without nearest-neighbor support '
            'extrapolation. They still do not identify architecture alone '
            'as the cause.'
        ),
    }
    write_json_atomic(
        CROSS_MODEL_DIR / 'cross_model_synthesis.json', synthesis
    )

    display(cross_frame[
        cross_frame['site_label'].isin(('middle_text', 'middle_options'))
        & cross_frame['task'].isin((
            'side_units', 'side_distance', 'side_torque', 'net_torque'
        ))
        & cross_frame['metric'].isin((
            'factor_selectivity_qwen_minus_gemma',
            'mean_donor_effect_qwen_minus_gemma',
            'donor_effect_difference_qwen_minus_gemma',
        ))
    ].sort_values(['task', 'site_label', 'metric']))

    plot_rows = cross_frame[
        (cross_frame['site_label'] == 'middle_text')
        & cross_frame['task'].isin((
            'side_units', 'side_distance', 'side_torque', 'net_torque'
        ))
        & cross_frame['metric'].isin((
            'factor_selectivity_qwen_minus_gemma',
            'mean_donor_effect_qwen_minus_gemma',
        ))
    ].copy()
    fig, ax = plt.subplots(figsize=(9, 4.5))
    labels = [
        f"{row.task}\n{row.metric.replace('_qwen_minus_gemma', '')}"
        for row in plot_rows.itertuples()
    ]
    x = np.arange(len(plot_rows))
    y = plot_rows['mean'].to_numpy(float)
    ax.errorbar(
        x, y,
        yerr=np.vstack([
            y - plot_rows['ci_low'].to_numpy(float),
            plot_rows['ci_high'].to_numpy(float) - y,
        ]),
        fmt='o', capsize=4, color='#4338ca',
    )
    ax.axhline(0, color='black', linewidth=0.8)
    ax.set_xticks(x, labels, rotation=20, ha='right')
    ax.set_ylabel('paired Qwen − Gemma effect')
    ax.set_title('Exact same-manifest cross-model comparison')
    fig.tight_layout()
    cross_figure = CROSS_MODEL_DIR / 'exact_pair_cross_model.png'
    fig.savefig(cross_figure, dpi=180, bbox_inches='tight')
    plt.show()
    print(json.dumps(synthesis, indent=2))
    print('Cross-model outputs:', CROSS_MODEL_DIR)
```
