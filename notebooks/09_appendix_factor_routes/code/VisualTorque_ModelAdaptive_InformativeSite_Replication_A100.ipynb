```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

## 2. New strategy-diagnostic confirmation support


```python
%%cell_progress 6/17 Build the independent model-adaptive replication manifest
cell_stage('Enumerating new single-stack and multi-stack geometries')
from collections import Counter, defaultdict
from itertools import combinations, product

from visual_torque_toolkit.render import render_scene
from visual_torque_toolkit.world import Scene, Side, Stack

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

def sign(value):
    return int(value > 0) - int(value < 0)

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

def canonical_signature(stacks):
    return tuple(sorted(
        (str(stack['side']), int(stack['distance']), int(stack['units']))
        for stack in stacks
    ))

def mirror_stacks(stacks):
    return [
        {
            'side': 'right' if stack['side'] == 'left' else 'left',
            'distance': int(stack['distance']),
            'units': int(stack['units']),
        }
        for stack in stacks
    ]

def scene_features(stacks):
    left = [stack for stack in stacks if stack['side'] == 'left']
    right = [stack for stack in stacks if stack['side'] == 'right']
    left_torque = sum(int(s['units']) * int(s['distance']) for s in left)
    right_torque = sum(int(s['units']) * int(s['distance']) for s in right)
    delta_torque = left_torque - right_torque
    delta_weight = sum(int(s['units']) for s in left) - sum(
        int(s['units']) for s in right
    )
    delta_distance = sum(int(s['distance']) for s in left) - sum(
        int(s['distance']) for s in right
    )
    max_left = max(int(s['units']) * int(s['distance']) for s in left)
    max_right = max(int(s['units']) * int(s['distance']) for s in right)
    delta_max = max_left - max_right
    lexical_score = (
        delta_weight if abs(delta_weight) > 1 else delta_distance
    )
    scores = {
        'exact_torque': int(delta_torque),
        'weight_only': int(delta_weight),
        'distance_only': int(delta_distance),
        'max_contribution': int(delta_max),
        'lexicographic': int(lexical_score),
    }
    predictions = {name: outcome(value) for name, value in scores.items()}
    return {
        'left_torque': int(left_torque),
        'right_torque': int(right_torque),
        'net_torque': int(delta_torque),
        'total_torque': int(left_torque + right_torque),
        'normalized_margin': float(
            abs(delta_torque) / max(1, left_torque + right_torque)
        ),
        'strategy_scores': scores,
        'strategy_predictions': predictions,
        'outcome': predictions['exact_torque'],
        'stack_count': len(stacks),
    }

def classify_diagnostic(features):
    prediction = features['strategy_predictions']
    exact = prediction['exact_torque']
    if exact == 'balanced':
        return None
    if (
        prediction['weight_only'] != exact
        and prediction['distance_only'] == exact
        and prediction['weight_only'] != 'balanced'
    ):
        return 'weight_conflict'
    if (
        prediction['distance_only'] != exact
        and prediction['weight_only'] == exact
        and prediction['distance_only'] != 'balanced'
    ):
        return 'distance_conflict'
    if (
        features['stack_count'] > 2
        and prediction['max_contribution'] != exact
        and prediction['max_contribution'] != 'balanced'
    ):
        return 'max_vs_sum'
    if (
        prediction['lexicographic'] != exact
        and prediction['lexicographic'] != 'balanced'
    ):
        return 'lexicographic_conflict'
    nonzero = [prediction[name] for name in CANDIDATE_STRATEGIES]
    if (
        len(set(nonzero)) == 1
        and exact != 'balanced'
        and features['normalized_margin'] >= 0.35
    ):
        return 'congruent_high'
    return None

dataset_rows = read_jsonl(DATASET_DIR / 'scenes.jsonl')
dataset_signatures = {
    canonical_signature(row['stacks']) for row in dataset_rows
}
ancestry_paths = (EARLIER_MANIFEST_PATH, LATE_MANIFEST_PATH)
for path in ancestry_paths:
    assert path.is_file(), f'Missing prerequisite manifest: {path}'
ancestry_manifests = {
    str(path): read_jsonl(path) for path in ancestry_paths
}
prior_signatures = {
    canonical_signature(row['roles'][role]['stacks'])
    for rows in ancestry_manifests.values()
    for row in rows for role in SCENE_ROLES
}
expected_prior = 3 * sum(len(rows) for rows in ancestry_manifests.values())
assert len(prior_signatures) == expected_prior, (
    len(prior_signatures), expected_prior
)
dataset_signatures |= prior_signatures
print('Excluded all prior geometries:', len(prior_signatures))

pool_by_signature = {}
for lu, ld, ru, rd in product(range(1, 7), repeat=4):
    stacks = [
        {'side': 'left', 'distance': ld, 'units': lu},
        {'side': 'right', 'distance': rd, 'units': ru},
    ]
    signature = canonical_signature(stacks)
    if signature in dataset_signatures:
        continue
    features = scene_features(stacks)
    if features['outcome'] != 'balanced':
        pool_by_signature[signature] = {
            'stacks': stacks, 'signature': signature, **features,
        }

for left_distances in combinations(range(1, 7), 2):
    for right_distances in combinations(range(1, 7), 2):
        for units in product(range(1, 5), repeat=4):
            stacks = [
                {'side': 'left', 'distance': left_distances[0], 'units': units[0]},
                {'side': 'left', 'distance': left_distances[1], 'units': units[1]},
                {'side': 'right', 'distance': right_distances[0], 'units': units[2]},
                {'side': 'right', 'distance': right_distances[1], 'units': units[3]},
            ]
            signature = canonical_signature(stacks)
            if signature in dataset_signatures:
                continue
            features = scene_features(stacks)
            if features['outcome'] != 'balanced':
                pool_by_signature[signature] = {
                    'stacks': stacks, 'signature': signature, **features,
                }

pool = list(pool_by_signature.values())
by_net = defaultdict(list)
by_cell = defaultdict(list)
for record in pool:
    by_net[int(record['net_torque'])].append(record)
    diagnostic = classify_diagnostic(record)
    if diagnostic is not None:
        by_cell[diagnostic].append(record)
for records in by_net.values():
    records.sort(key=lambda row: stable_key(row['signature']))
for records in by_cell.values():
    records.sort(key=lambda row: stable_key(row['signature']))

def different_predictions(left, right):
    return [
        name for name in CANDIDATE_STRATEGIES
        if left['strategy_predictions'][name]
        != right['strategy_predictions'][name]
    ]

def invariant_candidates(base, used_signatures, require_two=False):
    candidates = []
    for donor in by_net[int(base['net_torque'])]:
        signature = donor['signature']
        mirror_signature = canonical_signature(mirror_stacks(donor['stacks']))
        if signature in used_signatures or mirror_signature in used_signatures:
            continue
        if signature == base['signature']:
            continue
        differences = different_predictions(base, donor)
        required = 2 if require_two else 1
        if len(differences) < required:
            continue
        geometry_distance = sum(
            abs(a[1] - b[1]) + abs(a[2] - b[2])
            for a, b in zip(base['signature'], donor['signature'])
        ) if len(base['signature']) == len(donor['signature']) else 99
        candidates.append((
            -len(differences), geometry_distance,
            stable_key((base['signature'], signature)), donor,
        ))
    candidates.sort(key=lambda value: value[:-1])
    return [value[-1] for value in candidates]

cell_stage('Selecting balanced, scene-disjoint diagnostic families')
selected = []
used_signatures = set()

def append_family(cell_name, base, donor, cell_index):
    mirror = {
        'stacks': mirror_stacks(base['stacks']),
    }
    mirror.update(scene_features(mirror['stacks']))
    mirror['signature'] = canonical_signature(mirror['stacks'])
    assert mirror['net_torque'] == -base['net_torque']
    assert donor['net_torque'] == base['net_torque']
    differences = different_predictions(base, donor)
    diagnostic_strategy = (
        TARGET_STRATEGY[cell_name]
        if cell_name in TARGET_STRATEGY
        else differences[0] if differences else 'weight_only'
    )
    family_id = 'abstraction_' + hashlib.sha256(
        repr((cell_name, base['signature'], donor['signature'])).encode('utf-8')
    ).hexdigest()[:18]
    split = (
        'calibration' if cell_index < CALIBRATION_PER_CELL
        else 'confirmation'
    )
    roles = {}
    for role, scene_record in (
        ('base', base), ('mirror_source', mirror),
        ('invariant_source', donor),
    ):
        scene_id = family_id + '__' + role
        roles[role] = {
            'scene_id': scene_id,
            'image_filename': scene_id + '.png',
            'stacks': scene_record['stacks'],
            'features': {
                key: value for key, value in scene_record.items()
                if key not in {'stacks', 'signature'}
            },
        }
    selected.append({
        'family_id': family_id,
        'diagnostic_cell': cell_name,
        'split': split,
        'prompt_style': (
            'semantic' if (cell_index // 2) % 2 == 0 else 'nonce'
        ),
        'diagnostic_strategy': diagnostic_strategy,
        'base_outcome': base['outcome'],
        'invariant_strategy_differences': differences,
        'roles': roles,
    })
    used_signatures.update({
        base['signature'], mirror['signature'], donor['signature'],
    })

for cell_name in (
    'weight_conflict', 'distance_conflict', 'max_vs_sum',
    'lexicographic_conflict', 'congruent_high',
):
    chosen = 0
    for candidate in by_cell[cell_name]:
        desired = 'left' if chosen % 2 == 0 else 'right'
        base = candidate
        if candidate['outcome'] != desired:
            mirrored_signature = canonical_signature(mirror_stacks(candidate['stacks']))
            base = pool_by_signature.get(mirrored_signature)
            if base is None:
                continue
        base_mirror = canonical_signature(mirror_stacks(base['stacks']))
        if base['signature'] in used_signatures or base_mirror in used_signatures:
            continue
        donors = invariant_candidates(base, used_signatures | {
            base['signature'], base_mirror,
        })
        if not donors:
            continue
        append_family(cell_name, base, donors[0], chosen)
        chosen += 1
        if chosen == FAMILIES_PER_CELL:
            break
    assert chosen == FAMILIES_PER_CELL, (cell_name, chosen)

cell_stage('Selecting equal-torque recomposition families')
chosen = 0
recomposition_bases = sorted(
    [row for row in pool if row['stack_count'] > 2],
    key=lambda row: stable_key(('recomposition', row['signature'])),
)
for candidate in recomposition_bases:
    desired = 'left' if chosen % 2 == 0 else 'right'
    base = candidate
    if candidate['outcome'] != desired:
        mirrored_signature = canonical_signature(mirror_stacks(candidate['stacks']))
        base = pool_by_signature.get(mirrored_signature)
        if base is None:
            continue
    base_mirror = canonical_signature(mirror_stacks(base['stacks']))
    if base['signature'] in used_signatures or base_mirror in used_signatures:
        continue
    donors = invariant_candidates(
        base, used_signatures | {base['signature'], base_mirror},
        require_two=True,
    )
    if not donors:
        continue
    append_family('equal_torque_recomposition', base, donors[0], chosen)
    chosen += 1
    if chosen == FAMILIES_PER_CELL:
        break
assert chosen == FAMILIES_PER_CELL, ('equal_torque_recomposition', chosen)

selected.sort(key=lambda row: row['family_id'])
assert len(selected) == EXPECTED_FAMILIES
assert len(used_signatures) == EXPECTED_FAMILIES * 3
counts = Counter((row['diagnostic_cell'], row['split']) for row in selected)
for cell_name in DIAGNOSTIC_CELLS:
    assert counts[(cell_name, 'calibration')] == CALIBRATION_PER_CELL
    assert counts[(cell_name, 'confirmation')] == CONFIRMATION_PER_CELL
    outcomes = Counter(
        row['base_outcome'] for row in selected
        if row['diagnostic_cell'] == cell_name
    )
    assert outcomes == {
        'left': FAMILIES_PER_CELL // 2,
        'right': FAMILIES_PER_CELL // 2,
    }, (cell_name, outcomes)
    prompt_counts = Counter(
        (row['prompt_style'], row['base_outcome'])
        for row in selected if row['diagnostic_cell'] == cell_name
    )
    assert (
        len(prompt_counts) == 4
        and set(prompt_counts.values()) == {FAMILIES_PER_CELL // 4}
    ), (
        cell_name, prompt_counts
    )

manifest_identity = {
    'schema_version': 'model-adaptive-strategy-family-manifest-1.0',
    'dataset_scenes_sha256': sha256_file(DATASET_DIR / 'scenes.jsonl'),
    'dataset_geometries_excluded': True,
    'ancestry_sha256': {
        str(path.name): sha256_file(path) for path in ancestry_paths
    },
    'prior_geometries_excluded': len(prior_signatures),
    'model_outputs_consulted': False,
    'selection_seed': SEED,
    'pool_size': len(pool),
    'families': EXPECTED_FAMILIES,
    'unique_geometries': len(used_signatures),
    'cell_counts': {
        f'{cell}:{split}': int(count)
        for (cell, split), count in sorted(counts.items())
    },
    'family_ids': [row['family_id'] for row in selected],
}
if MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity
    assert read_jsonl(MANIFEST_PATH) == selected
else:
    write_jsonl_atomic(MANIFEST_PATH, selected)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

strategy_manifest = read_jsonl(MANIFEST_PATH)
cell_stage('Rendering or verifying 360 persistent diagnostic scenes')
for row in tqdm(strategy_manifest, desc='Rendering strategy scenes', unit='scene-family'):
    for role in SCENE_ROLES:
        role_record = row['roles'][role]
        image_path = SCENE_DIR / role_record['image_filename']
        if image_path.is_file():
            continue
        stacks = tuple(
            Stack(
                stack_id=f'{role}_{index}',
                side=Side(str(stack['side'])),
                distance=int(stack['distance']),
                units=int(stack['units']),
            )
            for index, stack in enumerate(role_record['stacks'])
        )
        render_scene(Scene(role_record['scene_id'], stacks), image_path)
for row in strategy_manifest:
    for role in SCENE_ROLES:
        assert (SCENE_DIR / row['roles'][role]['image_filename']).is_file()

manifest_frame = pd.DataFrame([{
    'family_id': row['family_id'],
    'diagnostic_cell': row['diagnostic_cell'],
    'split': row['split'],
    'prompt_style': row['prompt_style'],
    'diagnostic_strategy': row['diagnostic_strategy'],
    'base_outcome': row['base_outcome'],
    'invariant_strategy_differences': ','.join(
        row['invariant_strategy_differences']
    ),
} for row in strategy_manifest])
manifest_frame.to_csv(SHARED_ROOT / 'adaptive_site_family_manifest.csv', index=False)
identifiability_rows = []
for cell_name in DIAGNOSTIC_CELLS:
    cell_rows = [
        row for row in strategy_manifest
        if row['diagnostic_cell'] == cell_name
    ]
    for left_index, left_strategy in enumerate(CANDIDATE_STRATEGIES):
        for right_strategy in CANDIDATE_STRATEGIES[left_index + 1:]:
            disagreements = [
                row['roles']['base']['features']['strategy_predictions'][left_strategy]
                != row['roles']['base']['features']['strategy_predictions'][right_strategy]
                for row in cell_rows
            ]
            identifiability_rows.append({
                'diagnostic_cell': cell_name,
                'left_strategy': left_strategy,
                'right_strategy': right_strategy,
                'base_prediction_disagreement_rate': float(np.mean(disagreements)),
            })
identifiability_frame = pd.DataFrame(identifiability_rows)
identifiability_frame.to_csv(
    SHARED_ROOT / 'strategy_identifiability.csv', index=False
)
display(manifest_frame.groupby([
    'diagnostic_cell', 'split', 'prompt_style', 'base_outcome'
]).size().rename('n').reset_index())
print('Pool:', len(pool), '| selected families:', len(strategy_manifest))
print('New rendered geometries:', len(used_signatures))
```

```python
%%cell_progress 8/17 Define balanced prompts and aligned adaptive-site masks
cell_stage('Building role-aligned semantic and nonce-codebook prompts')
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add the turning effects '
    'on each side and compare the two totals.'
)
OUTCOME_TEXT = {
    'left': 'left side', 'balanced': 'balanced', 'right': 'right side',
}
LABELS = ('A', 'B', 'C')
NONCE_WORDS = ('dax', 'wug', 'zorp')

def rotated_semantics(row):
    values = ['left', 'balanced', 'right']
    offset = int(stable_key(
        f"options:{row['family_id']}"
    )[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def custom_prompt(row, prompt_style, semantics):
    if prompt_style == 'semantic':
        introduction = 'Solve this balance-scale question using only the supplied scene.'
        displayed = [OUTCOME_TEXT[value] for value in semantics]
    elif prompt_style == 'nonce':
        nonce_by_semantic = {
            semantic: NONCE_WORDS[index]
            for index, semantic in enumerate(rotated_semantics({
                **row, 'family_id': row['family_id'] + ':nonce'
            }))
        }
        codebook = '; '.join(
            f"{nonce_by_semantic[value]} means {OUTCOME_TEXT[value]}"
            for value in ('left', 'balanced', 'right')
        )
        introduction = (
            'Solve this balance-scale question using only the supplied scene. '
            'Use this family-specific codebook: ' + codebook + '.'
        )
        displayed = [nonce_by_semantic[value] for value in semantics]
    else:
        raise ValueError(prompt_style)
    options = '\n'.join(
        f'{label}. {text}' for label, text in zip(LABELS, displayed)
    )
    return '\n\n'.join([
        introduction,
        RULE_TEXT,
        'Question: Which side will rotate downward?',
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, role, prompt_style):
    semantics = rotated_semantics(row)
    role_record = row['roles'][role]
    features = role_record['features']
    gold = str(features['outcome'])
    assert gold in semantics
    prompt = custom_prompt(row, prompt_style, semantics)
    return BehaviorExample(
        example_id=(
            f"abstraction:{row['family_id']}:{role}:{prompt_style}"
        ),
        question_id=f"abstraction_{row['family_id']}_{prompt_style}",
        scene_id=str(role_record['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind='outcome',
        variant=f'competing_abstraction_{prompt_style}',
        split=str(row['split']),
        problem_type=str(row['diagnostic_cell']),
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=prompt,
        image_path=(SCENE_DIR / role_record['image_filename']).resolve(),
        labels=LABELS,
        option_text=tuple(OUTCOME_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def make_bundle(row, prompt_style='semantic'):
    return {
        role: make_example(row, role, prompt_style)
        for role in SCENE_ROLES
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples, 'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False, return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention, 'image_tokens': image,
        'all_text_except_final': all_text, 'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    masks['pre_options_tokens'] = (
        masks['intro_tokens'] | masks['rule_tokens'] | masks['question_tokens']
    )
    assert bool(masks['pre_options_tokens'].any())
    assert bool(masks['options_tokens'].any())
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(reference['inputs'][key], current['inputs'][key]), (
                name, key
            )
        masks = discover_masks(current)
        for scope in REQUIRED_SCOPES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Balanced adaptive-site prompts and masks ready.')
```

```python
%%cell_progress 9/17 Define clean scoring and adaptive-site interventions
cell_stage('Defining candidate scoring and grouped decoder-output interventions')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': example.option_semantics[predicted_index] == example.gold_semantic,
        })
    return records

def forward_capture(prepared):
    hidden = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden[layer_index] = output_hidden_tensor(output)[0].detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits)[0], hidden

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(condition['layer_index']) for condition in conditions}
    assert len(layer_indices) == 1, layer_indices
    layer_index = next(iter(layer_indices))
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = torch.stack([
            condition['intervention_hidden'] for condition in conditions
        ]).to(hidden.device, hidden.dtype)
        masks = torch.stack([
            condition['patch_mask'] for condition in conditions
        ]).to(hidden.device)
        assert donor.shape == hidden.shape, (donor.shape, hidden.shape)
        assert masks.shape == hidden.shape[:2], (masks.shape, hidden.shape)
        modified = hidden.clone()
        modified[masks] = donor[masks]
        audit['patched_tokens'] = [int(mask.sum().item()) for mask in masks]
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    assert len(records) == len(conditions)
    for record, condition, count in zip(
        records, conditions, audit['patched_tokens']
    ):
        record.update({
            key: value for key, value in condition.items()
            if key not in {'intervention_hidden', 'patch_mask'}
        })
        record['patched_tokens'] = count
    return records

def centered_logprob_error(left, right):
    left = np.asarray(left, dtype=float)
    right = np.asarray(right, dtype=float)
    left -= left.mean()
    right -= right.mean()
    return float(np.max(np.abs(left - right)))

def forward_score(prepared):
    with torch.inference_mode():
        output = adapter.model(**prepared['inputs'], use_cache=False)
    return score_records(prepared, output.logits)[0]

print('Adaptive-site scorers ready.')
```

```python
%%cell_progress 12/17 Release GPU and assemble adaptive-site causal metrics
cell_stage('Releasing model memory before analysis')
del adapter
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

cell_stage('Loading completed family shards')

def record_lookup(payload, site_role, intervention):
    matches = [
        record for record in payload['patch_records']
        if record['site_role'] == site_role
        and record['intervention'] == intervention
    ]
    assert len(matches) == 1, (site_role, intervention, len(matches))
    return matches[0]

def log_probability(score, semantic):
    index = score['candidate_semantics'].index(semantic)
    return float(score['candidate_logprobs'][index])

def probability_vector(score):
    values = np.asarray(score['candidate_logprobs'], dtype=float)
    values -= values.max()
    values = np.exp(values)
    return values / values.sum()

def semantic_margin(score, positive, negative):
    assert positive != negative
    return log_probability(score, positive) - log_probability(score, negative)

def centered_logprob_error(left, right):
    left_values = np.asarray(left['candidate_logprobs'], dtype=float)
    right_values = np.asarray(right['candidate_logprobs'], dtype=float)
    left_values -= left_values.mean()
    right_values -= right_values.mean()
    return float(np.max(np.abs(left_values - right_values)))

payloads = []
for family_index, row in enumerate(strategy_manifest):
    payload = json.loads(shard_path(family_index).read_text(encoding='utf-8'))
    assert payload['family_id'] == row['family_id']
    payloads.append(payload)

clean_rows = []
behavior_strategy_rows = []
behavior_effect_rows = []
late_effect_rows = []
congruent_rows = []
invariance_rows = []
replay_rows = []

for row, payload in tqdm(
    zip(strategy_manifest, payloads), total=len(payloads),
    desc='Assembling adaptive-site metrics', unit='family',
):
    base_score = payload['baselines']['base']
    base_features = row['roles']['base']['features']
    base_semantic = base_features['outcome']
    base_prediction = base_score['candidate_semantics'][
        int(np.argmax(base_score['candidate_logprobs']))
    ]
    clean_rows.append({
        'model_key': MODEL_KEY,
        'family_id': row['family_id'],
        'diagnostic_cell': row['diagnostic_cell'],
        'prompt_style': row['prompt_style'],
        'orientation': row['base_outcome'],
        'correct_semantic': base_semantic,
        'predicted_semantic': base_prediction,
        'exact_correct': int(base_prediction == base_semantic),
        'exact_logprob': log_probability(base_score, base_semantic),
    })
    for strategy in CANDIDATE_STRATEGIES:
        semantic = base_features['strategy_predictions'][strategy]
        behavior_strategy_rows.append({
            'model_key': MODEL_KEY,
            'family_id': row['family_id'],
            'diagnostic_cell': row['diagnostic_cell'],
            'prompt_style': row['prompt_style'],
            'orientation': row['base_outcome'],
            'strategy': strategy,
            'semantic': semantic,
            'logprob': log_probability(base_score, semantic),
        })
    cell = row['diagnostic_cell']
    if cell in TARGET_STRATEGY:
        target_strategy = TARGET_STRATEGY[cell]
        target_semantic = base_features['strategy_predictions'][target_strategy]
        exact_semantic = base_features['strategy_predictions']['exact_torque']
        assert target_semantic != exact_semantic
        behavior_effect_rows.append({
            'model_key': MODEL_KEY,
            'family_id': row['family_id'],
            'diagnostic_cell': cell,
            'prompt_style': row['prompt_style'],
            'orientation': row['base_outcome'],
            'target_strategy': target_strategy,
            'exact_minus_target_logprob': (
                log_probability(base_score, exact_semantic)
                - log_probability(base_score, target_semantic)
            ),
        })

    for _, _, site_role in FROZEN_SITES:
        self_records = [
            record_lookup(payload, site_role, f'self_replay_{index}')
            for index in range(SELF_REPLAY_REPLICATES)
        ]
        mirror_record = record_lookup(payload, site_role, 'mirror_source')
        invariant_record = record_lookup(payload, site_role, 'invariant_source')
        for replicate, record in enumerate(self_records):
            replay_rows.append({
                'model_key': MODEL_KEY,
                'family_id': row['family_id'],
                'diagnostic_cell': cell,
                'prompt_style': row['prompt_style'],
                'orientation': row['base_outcome'],
                'site_role': site_role,
                'replicate': replicate,
                'centered_logprob_error': centered_logprob_error(
                    record, base_score
                ),
            })

        if cell in TARGET_STRATEGY:
            target_strategy = TARGET_STRATEGY[cell]
            source_features = row['roles']['mirror_source']['features']
            exact_source = source_features['strategy_predictions']['exact_torque']
            exact_base = base_features['strategy_predictions']['exact_torque']
            target_source = source_features['strategy_predictions'][target_strategy]
            target_base = base_features['strategy_predictions'][target_strategy]
            assert exact_source != exact_base
            assert target_source != target_base

            def direct_index(score):
                exact_transport = semantic_margin(
                    score, exact_source, exact_base
                )
                target_transport = semantic_margin(
                    score, target_source, target_base
                )
                return exact_transport - target_transport

            self_indices = np.asarray([
                direct_index(record) for record in self_records
            ], dtype=float)
            mirror_index = direct_index(mirror_record)
            late_effect_rows.append({
                'model_key': MODEL_KEY,
                'family_id': row['family_id'],
                'diagnostic_cell': cell,
                'prompt_style': row['prompt_style'],
                'orientation': row['base_outcome'],
                'site_role': site_role,
                'target_strategy': target_strategy,
                'mirror_direct_index': mirror_index,
                'self_direct_index_mean': float(self_indices.mean()),
                'direct_causal_effect': float(
                    mirror_index - self_indices.mean()
                ),
                'self_replay_max_deviation': float(
                    np.max(np.abs(self_indices - self_indices.mean()))
                ),
            })

        if cell == 'congruent_high':
            source_semantic = row['roles']['mirror_source']['features'][
                'strategy_predictions'
            ]['exact_torque']
            base_exact = base_features['strategy_predictions']['exact_torque']
            assert source_semantic != base_exact
            self_margins = np.asarray([
                semantic_margin(record, source_semantic, base_exact)
                for record in self_records
            ], dtype=float)
            mirror_margin = semantic_margin(
                mirror_record, source_semantic, base_exact
            )
            congruent_rows.append({
                'model_key': MODEL_KEY,
                'family_id': row['family_id'],
                'prompt_style': row['prompt_style'],
                'orientation': row['base_outcome'],
                'site_role': site_role,
                'exact_transport_effect': float(
                    mirror_margin - self_margins.mean()
                ),
                'self_replay_max_deviation': float(
                    np.max(np.abs(self_margins - self_margins.mean()))
                ),
            })

        if cell == 'equal_torque_recomposition':
            self_probability = np.mean([
                probability_vector(record) for record in self_records
            ], axis=0)
            invariant_probability = probability_vector(invariant_record)
            invariance_rows.append({
                'model_key': MODEL_KEY,
                'family_id': row['family_id'],
                'prompt_style': row['prompt_style'],
                'orientation': row['base_outcome'],
                'site_role': site_role,
                'invariant_tvd': float(
                    0.5 * np.abs(
                        invariant_probability - self_probability
                    ).sum()
                ),
            })

clean_frame = pd.DataFrame(clean_rows)
behavior_strategy_frame = pd.DataFrame(behavior_strategy_rows)
behavior_effect_frame = pd.DataFrame(behavior_effect_rows)
late_effect_frame = pd.DataFrame(late_effect_rows)
congruent_frame = pd.DataFrame(congruent_rows)
invariance_frame = pd.DataFrame(invariance_rows)
replay_frame = pd.DataFrame(replay_rows)

for frame, name in (
    (clean_frame, 'clean_behavior.csv'),
    (behavior_strategy_frame, 'behavior_strategy_scores.csv'),
    (behavior_effect_frame, 'family_behavior_effects.csv'),
    (late_effect_frame, 'family_late_site_effects.csv'),
    (congruent_frame, 'family_congruent_transport.csv'),
    (invariance_frame, 'family_invariance_tvd.csv'),
    (replay_frame, 'self_replay_audit_long.csv'),
):
    frame.to_csv(OUTPUT_ROOT / name, index=False)

assert len(clean_frame) == EXPECTED_FAMILIES
assert len(late_effect_frame) == 4 * FAMILIES_PER_CELL * len(FROZEN_SITES)
assert len(congruent_frame) == FAMILIES_PER_CELL * len(FROZEN_SITES)
assert len(invariance_frame) == FAMILIES_PER_CELL * len(FROZEN_SITES)
assert len(replay_frame) == (
    EXPECTED_FAMILIES * len(FROZEN_SITES) * SELF_REPLAY_REPLICATES
)
display(clean_frame.groupby([
    'diagnostic_cell', 'prompt_style'
])['exact_correct'].mean().round(3))
display(late_effect_frame.groupby([
    'diagnostic_cell', 'site_role'
])['direct_causal_effect'].agg(['mean', 'std']).round(3))
```

```python
%%cell_progress 13/17 Test raw, efficacy-normalized, relocation, and invariance hypotheses
cell_stage('Bootstrapping family-level raw and normalized effects')

def bootstrap_mean(values, seed, draws=BOOTSTRAP_REPLICATES):
    values = np.asarray(values, dtype=float)
    assert values.ndim == 1 and len(values) >= 8
    generator = np.random.default_rng(seed)
    indices = generator.integers(0, len(values), size=(draws, len(values)))
    distribution = values[indices].mean(axis=1)
    low, high = np.quantile(distribution, [0.025, 0.975])
    p_two_sided = min(
        1.0,
        2.0 * min(
            float(np.mean(distribution <= 0.0)),
            float(np.mean(distribution >= 0.0)),
        ),
    )
    return {
        'n': int(len(values)),
        'estimate': float(values.mean()),
        'ci_low': float(low),
        'ci_high': float(high),
        'p_two_sided': float(p_two_sided),
    }

def bootstrap_ratio(numerator, denominator, seed, draws=BOOTSTRAP_REPLICATES):
    numerator = np.asarray(numerator, dtype=float)
    denominator = np.asarray(denominator, dtype=float)
    assert len(numerator) >= 8 and len(denominator) >= 8
    generator = np.random.default_rng(seed)
    numerator_index = generator.integers(
        0, len(numerator), size=(draws, len(numerator))
    )
    denominator_index = generator.integers(
        0, len(denominator), size=(draws, len(denominator))
    )
    denominator_means = denominator[denominator_index].mean(axis=1)
    denominator_stable = bool(np.quantile(denominator_means, 0.01) > 0)
    if not denominator_stable:
        return {
            'n_diagnostic': int(len(numerator)),
            'n_control': int(len(denominator)),
            'denominator_stable': False,
            'estimate': np.nan,
            'ci_low': np.nan,
            'ci_high': np.nan,
            'p_two_sided': 1.0,
        }
    distribution = (
        numerator[numerator_index].mean(axis=1) / denominator_means
    )
    low, high = np.quantile(distribution, [0.025, 0.975])
    p_two_sided = min(
        1.0,
        2.0 * min(
            float(np.mean(distribution <= 0.0)),
            float(np.mean(distribution >= 0.0)),
        ),
    )
    return {
        'n_diagnostic': int(len(numerator)),
        'n_control': int(len(denominator)),
        'denominator_stable': True,
        'estimate': float(numerator.mean() / denominator.mean()),
        'ci_low': float(low),
        'ci_high': float(high),
        'p_two_sided': float(p_two_sided),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    order = np.argsort(values)
    ranked = values[order]
    adjusted = ranked * len(values) / np.arange(1, len(values) + 1)
    adjusted = np.minimum.accumulate(adjusted[::-1])[::-1]
    output = np.empty_like(adjusted)
    output[order] = np.minimum(adjusted, 1.0)
    return output

site_replay_q95 = replay_frame.groupby('site_role')[
    'centered_logprob_error'
].quantile(.95).to_dict()

control_rows = []
for site_role, group in congruent_frame.groupby('site_role'):
    result = bootstrap_mean(
        group['exact_transport_effect'],
        SEED + 100 + sum(map(ord, site_role)),
    )
    effective_bound = max(
        RAW_SESOI_LOGODDS, float(site_replay_q95[site_role])
    )
    control_rows.append({
        'model_key': MODEL_KEY,
        'site_role': site_role,
        'site_priority': (
            'primary' if site_role == PRIMARY_SITE_ROLE else 'alternate'
        ),
        'replay_q95_centered_error': float(site_replay_q95[site_role]),
        'effective_raw_bound': effective_bound,
        'site_informative': bool(result['ci_low'] > effective_bound),
        **result,
    })
control_test_frame = pd.DataFrame(control_rows)
control_test_frame.to_csv(
    OUTPUT_ROOT / 'congruent_efficacy_tests.csv', index=False
)

raw_rows = []
normalized_rows = []
for (cell, site_role), group in late_effect_frame.groupby([
    'diagnostic_cell', 'site_role'
], sort=True):
    control = congruent_frame[congruent_frame['site_role'] == site_role]
    raw = bootstrap_mean(
        group['direct_causal_effect'],
        SEED + 200 + sum(map(ord, cell + site_role)),
    )
    normalized = bootstrap_ratio(
        group['direct_causal_effect'], control['exact_transport_effect'],
        SEED + 300 + sum(map(ord, cell + site_role)),
    )
    effective_bound = float(
        control_test_frame.set_index('site_role').loc[
            site_role, 'effective_raw_bound'
        ]
    )
    direction = EXPECTED_DIRECTION[cell]
    if direction == 'negative':
        raw_pass = raw['ci_high'] < -effective_bound
        normalized_pass = normalized['ci_high'] < -NORMALIZED_SESOI
        raw_equivalent = normalized_equivalent = False
    elif direction == 'positive':
        raw_pass = raw['ci_low'] > effective_bound
        normalized_pass = normalized['ci_low'] > NORMALIZED_SESOI
        raw_equivalent = normalized_equivalent = False
    else:
        raw_pass = normalized_pass = False
        raw_equivalent = (
            raw['ci_low'] > -effective_bound
            and raw['ci_high'] < effective_bound
        )
        normalized_equivalent = (
            normalized['ci_low'] > -NORMALIZED_SESOI
            and normalized['ci_high'] < NORMALIZED_SESOI
        )
    common = {
        'model_key': MODEL_KEY,
        'diagnostic_cell': cell,
        'site_role': site_role,
        'site_priority': (
            'primary' if site_role == PRIMARY_SITE_ROLE else 'alternate'
        ),
        'expected_direction': direction,
    }
    raw_rows.append({
        **common,
        'effective_bound': effective_bound,
        'directional_pass': bool(raw_pass),
        'equivalence_pass': bool(raw_equivalent),
        **raw,
    })
    normalized_rows.append({
        **common,
        'normalized_bound': NORMALIZED_SESOI,
        'directional_pass': bool(normalized_pass),
        'equivalence_pass': bool(normalized_equivalent),
        **normalized,
    })

raw_test_frame = pd.DataFrame(raw_rows)
normalized_test_frame = pd.DataFrame(normalized_rows)
raw_test_frame['q_value'] = np.nan
normalized_test_frame['q_value'] = np.nan
for frame in (raw_test_frame, normalized_test_frame):
    mask = (
        (frame['site_priority'] == 'primary')
        & frame['diagnostic_cell'].isin([
            'weight_conflict', 'distance_conflict', 'max_vs_sum'
        ])
    )
    frame.loc[mask, 'q_value'] = bh_adjust(
        frame.loc[mask, 'p_two_sided']
    )
raw_test_frame.to_csv(OUTPUT_ROOT / 'raw_site_tests.csv', index=False)
normalized_test_frame.to_csv(
    OUTPUT_ROOT / 'normalized_site_tests.csv', index=False
)

cell_gate_rows = []
control_lookup = control_test_frame.set_index('site_role')
for cell in TARGET_STRATEGY:
    raw = raw_test_frame[
        (raw_test_frame['diagnostic_cell'] == cell)
        & (raw_test_frame['site_role'] == PRIMARY_SITE_ROLE)
    ].iloc[0]
    normalized = normalized_test_frame[
        (normalized_test_frame['diagnostic_cell'] == cell)
        & (normalized_test_frame['site_role'] == PRIMARY_SITE_ROLE)
    ].iloc[0]
    site_ok = bool(control_lookup.loc[PRIMARY_SITE_ROLE, 'site_informative'])
    if EXPECTED_DIRECTION[cell] == 'equivalent':
        pass_claim = (
            site_ok and bool(raw['equivalence_pass'])
            and bool(normalized['equivalence_pass'])
        )
    else:
        pass_claim = (
            site_ok and bool(raw['directional_pass'])
            and bool(normalized['directional_pass'])
            and float(raw['q_value']) < .05
            and float(normalized['q_value']) < .05
        )
    cell_gate_rows.append({
        'model_key': MODEL_KEY,
        'diagnostic_cell': cell,
        'primary_site_role': PRIMARY_SITE_ROLE,
        'site_informative': site_ok,
        'raw_estimate': float(raw['estimate']),
        'raw_ci_low': float(raw['ci_low']),
        'raw_ci_high': float(raw['ci_high']),
        'normalized_estimate': float(normalized['estimate']),
        'normalized_ci_low': float(normalized['ci_low']),
        'normalized_ci_high': float(normalized['ci_high']),
        'normalization_denominator_stable': bool(
            normalized['denominator_stable']
        ),
        'claim_gate': bool(pass_claim),
    })
cell_gate_frame = pd.DataFrame(cell_gate_rows)
cell_gate_frame.to_csv(OUTPUT_ROOT / 'primary_cell_gates.csv', index=False)

congruent_pivot = congruent_frame.pivot(
    index='family_id', columns='site_role',
    values='exact_transport_effect',
)
relocation_values = (
    congruent_pivot[PRIMARY_SITE_ROLE]
    - congruent_pivot[ALTERNATE_SITE_ROLE]
)
relocation_result = bootstrap_mean(
    relocation_values, SEED + 400
)
relocation_test = {
    'model_key': MODEL_KEY,
    'primary_site_role': PRIMARY_SITE_ROLE,
    'alternate_site_role': ALTERNATE_SITE_ROLE,
    'sesoi': RELOCATION_SESOI_LOGODDS,
    'relocation_pass': bool(
        relocation_result['ci_low'] > RELOCATION_SESOI_LOGODDS
    ),
    **relocation_result,
}
write_json_atomic(OUTPUT_ROOT / 'route_relocation_test.json', relocation_test)

invariance_rows = []
for site_role, group in invariance_frame.groupby('site_role'):
    result = bootstrap_mean(
        group['invariant_tvd'],
        SEED + 500 + sum(map(ord, site_role)),
    )
    invariance_rows.append({
        'model_key': MODEL_KEY,
        'site_role': site_role,
        'site_priority': (
            'primary' if site_role == PRIMARY_SITE_ROLE else 'alternate'
        ),
        'equivalent': bool(result['ci_high'] < INVARIANCE_SESOI_TVD),
        'violated': bool(result['ci_low'] > INVARIANCE_SESOI_TVD),
        **result,
    })
invariance_test_frame = pd.DataFrame(invariance_rows)
invariance_test_frame.to_csv(
    OUTPUT_ROOT / 'equal_net_invariance_tests.csv', index=False
)
invariance_pivot = invariance_frame.pivot(
    index='family_id', columns='site_role', values='invariant_tvd'
)
stage_delta_result = bootstrap_mean(
    invariance_pivot['answer'] - invariance_pivot['options'],
    SEED + 600,
)
stage_delta_test = {
    'model_key': MODEL_KEY,
    'contrast': 'answer_minus_options_equal_net_tvd',
    'sesoi': STAGE_DELTA_SESOI_TVD,
    'answer_expansion_pass': bool(
        stage_delta_result['ci_low'] > STAGE_DELTA_SESOI_TVD
    ),
    **stage_delta_result,
}
write_json_atomic(OUTPUT_ROOT / 'equal_net_stage_delta_test.json', stage_delta_test)

display(control_test_frame.round(4))
display(cell_gate_frame.round(4))
display(relocation_test)
display(invariance_test_frame.round(4))
display(stage_delta_test)
```

```python
%%cell_progress 14/17 Audit prompt, orientation, replay, and leave-one-out robustness
cell_stage('Checking normalized effects across nuisance subgroups')

def expected_sign(cell, value):
    direction = EXPECTED_DIRECTION[cell]
    if direction == 'negative':
        return value < 0
    if direction == 'positive':
        return value > 0
    return abs(value) < NORMALIZED_SESOI

subgroup_rows = []
for grouping in ('prompt_style', 'orientation'):
    diagnostic_groups = late_effect_frame[
        late_effect_frame['site_role'] == PRIMARY_SITE_ROLE
    ].groupby(['diagnostic_cell', grouping], sort=True)
    for (cell, level), group in diagnostic_groups:
        controls = congruent_frame[
            (congruent_frame['site_role'] == PRIMARY_SITE_ROLE)
            & (congruent_frame[grouping] == level)
        ]
        result = bootstrap_ratio(
            group['direct_causal_effect'],
            controls['exact_transport_effect'],
            SEED + 700 + sum(map(ord, cell + grouping + str(level))),
        )
        subgroup_rows.append({
            'model_key': MODEL_KEY,
            'grouping': grouping,
            'level': level,
            'diagnostic_cell': cell,
            'site_role': PRIMARY_SITE_ROLE,
            **result,
        })
subgroup_frame = pd.DataFrame(subgroup_rows)
subgroup_frame.to_csv(
    OUTPUT_ROOT / 'normalized_nuisance_subgroups.csv', index=False
)

prompt_consistency_rows = []
for cell, group in subgroup_frame[
    subgroup_frame['grouping'] == 'prompt_style'
].groupby('diagnostic_cell'):
    prompt_consistency_rows.append({
        'model_key': MODEL_KEY,
        'diagnostic_cell': cell,
        'primary_site_role': PRIMARY_SITE_ROLE,
        'all_prompt_point_estimates_expected': bool(all(
            expected_sign(cell, value) for value in group['estimate']
        )),
        'prompt_estimate_range': float(
            group['estimate'].max() - group['estimate'].min()
        ),
    })
prompt_consistency_frame = pd.DataFrame(prompt_consistency_rows)
prompt_consistency_frame.to_csv(
    OUTPUT_ROOT / 'prompt_consistency.csv', index=False
)

replay_summary = replay_frame.groupby('site_role').agg(
    n=('centered_logprob_error', 'size'),
    mean_centered_error=('centered_logprob_error', 'mean'),
    q95_centered_error=(
        'centered_logprob_error', lambda values: np.quantile(values, .95)
    ),
    max_centered_error=('centered_logprob_error', 'max'),
).reset_index()
replay_summary.insert(0, 'model_key', MODEL_KEY)
replay_summary.to_csv(OUTPUT_ROOT / 'self_replay_summary.csv', index=False)

primary_control_mean = float(
    congruent_frame[
        congruent_frame['site_role'] == PRIMARY_SITE_ROLE
    ]['exact_transport_effect'].mean()
)
loo_rows = []
primary_effects = late_effect_frame[
    late_effect_frame['site_role'] == PRIMARY_SITE_ROLE
]
for cell, group in primary_effects.groupby('diagnostic_cell'):
    values = group['direct_causal_effect'].to_numpy(dtype=float)
    leave_one_out = (
        (values.sum() - values) / (len(values) - 1)
        / primary_control_mean
    )
    loo_rows.append({
        'model_key': MODEL_KEY,
        'diagnostic_cell': cell,
        'minimum_loo_normalized_mean': float(leave_one_out.min()),
        'maximum_loo_normalized_mean': float(leave_one_out.max()),
        'all_loo_expected': bool(all(
            expected_sign(cell, value) for value in leave_one_out
        )),
    })
loo_frame = pd.DataFrame(loo_rows)
loo_frame.to_csv(
    OUTPUT_ROOT / 'leave_one_family_out_normalized.csv', index=False
)

accuracy_frame = clean_frame.groupby([
    'diagnostic_cell', 'prompt_style'
])['exact_correct'].agg(['count', 'mean']).reset_index()
accuracy_frame.insert(0, 'model_key', MODEL_KEY)
accuracy_frame.to_csv(
    OUTPUT_ROOT / 'clean_accuracy_by_prompt.csv', index=False
)

display(prompt_consistency_frame)
display(replay_summary.round(4))
display(loo_frame.round(4))
display(accuracy_frame.round(3))
```

```python
%%cell_progress 16/17 Apply the frozen model-adaptive decision tree
cell_stage('Combining efficacy, normalization, prompt, and relocation gates')

primary_gates = cell_gate_frame.set_index('diagnostic_cell')
prompt_lookup = prompt_consistency_frame.set_index('diagnostic_cell')
cell_decisions = []
for cell in TARGET_STRATEGY:
    gate = primary_gates.loc[cell]
    prompt_ok = bool(
        prompt_lookup.loc[cell, 'all_prompt_point_estimates_expected']
    )
    if not bool(gate['site_informative']):
        status = 'site_uninformative'
    elif bool(gate['claim_gate']) and prompt_ok:
        if EXPECTED_DIRECTION[cell] == 'negative':
            status = 'fallback_route_confirmed'
        elif EXPECTED_DIRECTION[cell] == 'positive':
            status = 'exact_route_confirmed'
        else:
            status = 'route_equivalent_within_bounds'
    elif (
        EXPECTED_DIRECTION[cell] == 'positive'
        and gate['normalized_ci_high'] < -NORMALIZED_SESOI
    ) or (
        EXPECTED_DIRECTION[cell] == 'negative'
        and gate['normalized_ci_low'] > NORMALIZED_SESOI
    ):
        status = 'wrong_direction'
    else:
        status = 'unresolved'
    cell_decisions.append({
        'diagnostic_cell': cell,
        'status': status,
        'prompt_direction_consistent': prompt_ok,
        'raw_estimate': float(gate['raw_estimate']),
        'raw_ci_low': float(gate['raw_ci_low']),
        'raw_ci_high': float(gate['raw_ci_high']),
        'normalized_estimate': float(gate['normalized_estimate']),
        'normalized_ci_low': float(gate['normalized_ci_low']),
        'normalized_ci_high': float(gate['normalized_ci_high']),
    })

distance_confirmed = next(
    row['status'] == 'exact_route_confirmed'
    for row in cell_decisions
    if row['diagnostic_cell'] == 'distance_conflict'
)
max_confirmed = next(
    row['status'] == 'exact_route_confirmed'
    for row in cell_decisions
    if row['diagnostic_cell'] == 'max_vs_sum'
)
relocation_confirmed = bool(relocation_test['relocation_pass'])
if distance_confirmed and relocation_confirmed and max_confirmed:
    model_status = 'adaptive_distance_and_sum_route_confirmed'
elif distance_confirmed and relocation_confirmed:
    model_status = 'adaptive_distance_route_confirmed'
elif distance_confirmed:
    model_status = 'distance_route_confirmed_without_relocation'
else:
    model_status = 'adaptive_route_not_confirmed'

primary_invariance = invariance_test_frame[
    invariance_test_frame['site_role'] == PRIMARY_SITE_ROLE
].iloc[0]
if bool(primary_invariance['equivalent']):
    invariance_status = 'primary_equal_net_invariance_supported'
elif bool(primary_invariance['violated']):
    invariance_status = 'primary_equal_net_invariance_violated'
else:
    invariance_status = 'primary_equal_net_invariance_unresolved'

decision = {
    'schema_version': 'model-adaptive-route-decision-1.0',
    'model_key': MODEL_KEY,
    'model_id': MODEL_ID,
    'manifest_sha256': manifest_sha256,
    'protocol_sha256': protocol_sha256,
    'primary_site_role': PRIMARY_SITE_ROLE,
    'alternate_site_role': ALTERNATE_SITE_ROLE,
    'model_status': model_status,
    'distance_confirmed': distance_confirmed,
    'max_sum_confirmed': max_confirmed,
    'relocation_confirmed': relocation_confirmed,
    'cell_decisions': cell_decisions,
    'invariance_status': invariance_status,
    'answer_expansion_pass': bool(
        stage_delta_test['answer_expansion_pass']
    ),
    'claim_boundary': (
        'The result identifies a model-specific causally sufficient route. '
        'It does not imply a unique serial physical algorithm or a human-like '
        'calculation process.'
    ),
}
write_json_atomic(OUTPUT_ROOT / 'adaptive_route_decision.json', decision)
display(pd.DataFrame(cell_decisions))
display(decision)
```

```python
%%cell_progress 17/17 Export model report and optional cross-model synthesis
cell_stage('Writing the model-adaptive evidence report')

report_lines = [
    '# Model-Adaptive Informative-Site Replication',
    '',
    f'- Model: `{MODEL_ID}` (`{MODEL_KEY}`)',
    f'- Primary site: `{PRIMARY_SITE_ROLE}`',
    f'- Alternate site: `{ALTERNATE_SITE_ROLE}`',
    f'- Model decision: **{decision["model_status"]}**',
    f'- Route relocation: **{decision["relocation_confirmed"]}**',
    f'- Equal-net result: **{decision["invariance_status"]}**',
    '',
    '## Frozen primary-site decisions',
    '',
]
for row in cell_decisions:
    report_lines.append(
        f'- `{row["diagnostic_cell"]}`: **{row["status"]}**; '
        f'raw={row["raw_estimate"]:.3f} '
        f'[{row["raw_ci_low"]:.3f}, {row["raw_ci_high"]:.3f}]; '
        f'normalized={row["normalized_estimate"]:.3f} '
        f'[{row["normalized_ci_low"]:.3f}, '
        f'{row["normalized_ci_high"]:.3f}].'
    )
report_lines.extend([
    '',
    '## Interpretation boundary',
    '',
    decision['claim_boundary'],
])
(OUTPUT_ROOT / 'MODEL_REPORT.md').write_text(
    '\n'.join(report_lines) + '\n', encoding='utf-8'
)

cell_stage('Checking for all three same-manifest model runs')
model_roots = {
    key: SHARED_ROOT / 'models' / specification['model_slug']
    for key, specification in MODEL_SPECS.items()
}
required_names = (
    'family_late_site_effects.csv', 'family_congruent_transport.csv',
    'family_invariance_tvd.csv', 'primary_cell_gates.csv',
    'route_relocation_test.json', 'adaptive_route_decision.json',
    'capture_config.json',
)
available_models = [
    key for key, root in model_roots.items()
    if all((root / name).is_file() for name in required_names)
]
print('Available same-manifest models:', available_models)

def summarize_distribution(distribution, estimate):
    distribution = np.asarray(distribution, dtype=float)
    distribution = distribution[np.isfinite(distribution)]
    if len(distribution) == 0 or not np.isfinite(estimate):
        return {
            'estimate': np.nan, 'ci_low': np.nan, 'ci_high': np.nan,
            'p_two_sided': 1.0,
        }
    low, high = np.quantile(distribution, [0.025, 0.975])
    p_two_sided = min(
        1.0,
        2.0 * min(
            float(np.mean(distribution <= 0)),
            float(np.mean(distribution >= 0)),
        ),
    )
    return {
        'estimate': float(estimate),
        'ci_low': float(low),
        'ci_high': float(high),
        'p_two_sided': float(p_two_sided),
    }

if len(available_models) == len(MODEL_SPECS):
    reference_hash = sha256_file(MANIFEST_PATH)
    model_data = {}
    for key, root in model_roots.items():
        capture_config = json.loads(
            (root / 'capture_config.json').read_text(encoding='utf-8')
        )
        assert capture_config['identity']['manifest_sha256'] == reference_hash
        model_data[key] = {
            'effects': pd.read_csv(root / 'family_late_site_effects.csv'),
            'controls': pd.read_csv(root / 'family_congruent_transport.csv'),
            'invariance': pd.read_csv(root / 'family_invariance_tvd.csv'),
            'gates': pd.read_csv(root / 'primary_cell_gates.csv'),
            'decision': json.loads(
                (root / 'adaptive_route_decision.json').read_text(
                    encoding='utf-8'
                )
            ),
        }

    generator = np.random.default_rng(SEED + 900)
    draws = BOOTSTRAP_REPLICATES

    def ratio_distribution(key, cell):
        site = MODEL_SPECS[key]['primary_site_role']
        numerator = model_data[key]['effects']
        numerator = numerator[
            (numerator['diagnostic_cell'] == cell)
            & (numerator['site_role'] == site)
        ]['direct_causal_effect'].to_numpy(dtype=float)
        denominator = model_data[key]['controls']
        denominator = denominator[
            denominator['site_role'] == site
        ]['exact_transport_effect'].to_numpy(dtype=float)
        numerator_index = generator.integers(
            0, len(numerator), size=(draws, len(numerator))
        )
        denominator_index = generator.integers(
            0, len(denominator), size=(draws, len(denominator))
        )
        denominator_mean = denominator[denominator_index].mean(axis=1)
        if np.quantile(denominator_mean, .01) <= 0:
            return np.full(draws, np.nan), np.nan
        distribution = (
            numerator[numerator_index].mean(axis=1) / denominator_mean
        )
        return distribution, float(numerator.mean() / denominator.mean())

    ratio_cache = {
        (key, cell): ratio_distribution(key, cell)
        for key in MODEL_SPECS
        for cell in ('distance_conflict', 'max_vs_sum')
    }
    cross_rows = []

    distance_distribution = np.mean([
        ratio_cache[(key, 'distance_conflict')][0]
        for key in MODEL_SPECS
    ], axis=0)
    distance_estimates = {
        key: ratio_cache[(key, 'distance_conflict')][1]
        for key in MODEL_SPECS
    }
    individual_distance_pass = all(
        bool(model_data[key]['gates'].set_index('diagnostic_cell').loc[
            'distance_conflict', 'claim_gate'
        ])
        for key in MODEL_SPECS
    )
    distance_result = summarize_distribution(
        distance_distribution,
        np.mean(list(distance_estimates.values())),
    )
    cross_rows.append({
        'hypothesis': 'shared_adaptive_distance_route',
        'bound': NORMALIZED_SESOI,
        'all_individual_model_gates_pass': individual_distance_pass,
        'effect_pass': bool(
            distance_result['ci_low'] > NORMALIZED_SESOI
            and individual_distance_pass
        ),
        'model_estimates': json.dumps(
            distance_estimates, sort_keys=True
        ),
        **distance_result,
    })

    large_keys = ('gemma12b', 'qwen27b')
    large_max_distribution = np.mean([
        ratio_cache[(key, 'max_vs_sum')][0] for key in large_keys
    ], axis=0)
    max_estimates = {
        key: ratio_cache[(key, 'max_vs_sum')][1]
        for key in MODEL_SPECS
    }
    large_max_result = summarize_distribution(
        large_max_distribution,
        np.mean([max_estimates[key] for key in large_keys]),
    )
    cross_rows.append({
        'hypothesis': 'large_model_normalized_summation_route',
        'bound': NORMALIZED_SESOI,
        'all_individual_model_gates_pass': None,
        'effect_pass': bool(
            large_max_result['ci_low'] > NORMALIZED_SESOI
        ),
        'model_estimates': json.dumps(max_estimates, sort_keys=True),
        **large_max_result,
    })

    max_interaction_distribution = (
        large_max_distribution
        - ratio_cache[('qwen9b', 'max_vs_sum')][0]
    )
    max_interaction_estimate = (
        np.mean([max_estimates[key] for key in large_keys])
        - max_estimates['qwen9b']
    )
    max_interaction_result = summarize_distribution(
        max_interaction_distribution, max_interaction_estimate
    )
    cross_rows.append({
        'hypothesis': 'large_minus_qwen9b_summation_route',
        'bound': NORMALIZED_SESOI,
        'all_individual_model_gates_pass': None,
        'effect_pass': bool(
            max_interaction_result['ci_low'] > NORMALIZED_SESOI
        ),
        'model_estimates': json.dumps(max_estimates, sort_keys=True),
        **max_interaction_result,
    })

    relocation_distributions = []
    relocation_estimates = {}
    for key in MODEL_SPECS:
        primary = MODEL_SPECS[key]['primary_site_role']
        alternate = 'answer' if primary == 'options' else 'options'
        pivot = model_data[key]['controls'].pivot(
            index='family_id', columns='site_role',
            values='exact_transport_effect',
        )
        primary_values = pivot[primary].to_numpy(dtype=float)
        alternate_values = pivot[alternate].to_numpy(dtype=float)
        indices = generator.integers(
            0, len(pivot), size=(draws, len(pivot))
        )
        sampled_primary = primary_values[indices].mean(axis=1)
        sampled_alternate = alternate_values[indices].mean(axis=1)
        if np.quantile(sampled_primary, .01) <= 0:
            relocation_distributions.append(np.full(draws, np.nan))
            relocation_estimates[key] = np.nan
        else:
            relocation_distributions.append(
                (sampled_primary - sampled_alternate) / sampled_primary
            )
            relocation_estimates[key] = float(
                (primary_values.mean() - alternate_values.mean())
                / primary_values.mean()
            )
    relocation_distribution = np.mean(relocation_distributions, axis=0)
    relocation_result = summarize_distribution(
        relocation_distribution,
        np.mean(list(relocation_estimates.values())),
    )
    all_relocation_pass = all(
        model_data[key]['decision']['relocation_confirmed']
        for key in MODEL_SPECS
    )
    cross_rows.append({
        'hypothesis': 'model_specific_route_relocation',
        'bound': NORMALIZED_SESOI,
        'all_individual_model_gates_pass': all_relocation_pass,
        'effect_pass': bool(
            relocation_result['ci_low'] > NORMALIZED_SESOI
            and all_relocation_pass
        ),
        'model_estimates': json.dumps(
            relocation_estimates, sort_keys=True
        ),
        **relocation_result,
    })

    expansion_distributions = []
    expansion_estimates = {}
    for key in large_keys:
        pivot = model_data[key]['invariance'].pivot(
            index='family_id', columns='site_role', values='invariant_tvd'
        )
        delta = (
            pivot['answer'] - pivot['options']
        ).to_numpy(dtype=float)
        indices = generator.integers(
            0, len(delta), size=(draws, len(delta))
        )
        expansion_distributions.append(delta[indices].mean(axis=1))
        expansion_estimates[key] = float(delta.mean())
    expansion_distribution = np.mean(expansion_distributions, axis=0)
    expansion_result = summarize_distribution(
        expansion_distribution,
        np.mean(list(expansion_estimates.values())),
    )
    cross_rows.append({
        'hypothesis': 'large_model_answer_factorization_expansion',
        'bound': STAGE_DELTA_SESOI_TVD,
        'all_individual_model_gates_pass': None,
        'effect_pass': bool(
            expansion_result['ci_low'] > STAGE_DELTA_SESOI_TVD
        ),
        'model_estimates': json.dumps(
            expansion_estimates, sort_keys=True
        ),
        **expansion_result,
    })

    cross_frame = pd.DataFrame(cross_rows)
    cross_frame['q_value'] = bh_adjust(cross_frame['p_two_sided'])
    cross_frame['confirmatory_pass'] = (
        cross_frame['effect_pass'] & (cross_frame['q_value'] < .05)
    )
    cross_frame.to_csv(
        CROSS_MODEL_DIR / 'model_adaptive_cross_model_tests.csv',
        index=False,
    )
    passed = dict(zip(
        cross_frame['hypothesis'], cross_frame['confirmatory_pass']
    ))
    if (
        passed['shared_adaptive_distance_route']
        and passed['model_specific_route_relocation']
        and passed['large_model_normalized_summation_route']
        and passed['large_minus_qwen9b_summation_route']
    ):
        cross_status = 'model_adaptive_conditional_routes_confirmed'
    elif (
        passed['shared_adaptive_distance_route']
        and passed['model_specific_route_relocation']
    ):
        cross_status = 'shared_distance_route_with_relocation_confirmed'
    else:
        cross_status = 'model_adaptive_route_hypothesis_not_confirmed'
    cross_decision = {
        'schema_version': 'model-adaptive-cross-model-decision-1.0',
        'manifest_sha256': reference_hash,
        'available_models': available_models,
        'status': cross_status,
        'tests': cross_frame.to_dict(orient='records'),
        'interpretation': (
            'Model-specific informative sites are compared after normalization '
            'by causal efficacy. The result does not assume a universal serial '
            'or human-like physical calculation path.'
        ),
    }
    write_json_atomic(
        CROSS_MODEL_DIR / 'model_adaptive_cross_model_decision.json',
        cross_decision,
    )
    display(cross_frame.round(4))
    display(cross_decision)
else:
    print(
        'Cross-model synthesis deferred. Run the unchanged notebook with '
        "MODEL_KEY='qwen9b', 'gemma12b', and 'qwen27b'."
    )

print('Model report:', OUTPUT_ROOT / 'MODEL_REPORT.md')
print('Decision:', OUTPUT_ROOT / 'adaptive_route_decision.json')
print('Figure:', FIGURE_DIR / 'model_adaptive_route_replication.png')
```
