```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Lock the metadata-only discovery/confirmation split
cell_stage('Verifying the exact orthogonal manifest and prior 27B run')
from collections import defaultdict
from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

def split_key(value):
    return hashlib.sha256(f'{SPLIT_SEED}:{value}'.encode('utf-8')).hexdigest()

raw_manifest = read_jsonl(MANIFEST_PATH)
assert len(raw_manifest) == EXPECTED_TRIPLETS
assert len({str(row['triplet_id']) for row in raw_manifest}) == EXPECTED_TRIPLETS
for row in raw_manifest:
    assert len(set(map(int, row['scalar_values']))) == 4
    assert set(TASKS) <= set(row['task_values'])

prior_decision = json.loads(PRIOR_DECISION_PATH.read_text(encoding='utf-8'))
assert prior_decision['manifest_sha256'] == sha256_file(MANIFEST_PATH)
prior_paired = pd.read_csv(PRIOR_PAIRED_PATH)
prior_paired['triplet_id'] = prior_paired['triplet_id'].astype(str)
assert set(prior_paired['triplet_id']) == {
    str(row['triplet_id']) for row in raw_manifest
}

cell_stage('Creating a deterministic metadata-stratified split')
strata = defaultdict(list)
for row in raw_manifest:
    key = (
        str(row['changed_side']), str(row['factor_direction']),
        bool(row['outcome_changed']),
    )
    strata[key].append(row)

discovery_ids = set()
odd_middle_candidates = []
for key in sorted(strata, key=str):
    rows = sorted(strata[key], key=lambda row: split_key(row['triplet_id']))
    half = len(rows) // 2
    discovery_ids.update(str(row['triplet_id']) for row in rows[:half])
    if len(rows) % 2:
        odd_middle_candidates.append(rows[half])
needed = DISCOVERY_TRIPLETS - len(discovery_ids)
assert 0 <= needed <= len(odd_middle_candidates)
extras = sorted(
    odd_middle_candidates, key=lambda row: split_key('extra:' + row['triplet_id'])
)[:needed]
discovery_ids.update(str(row['triplet_id']) for row in extras)
confirmation_ids = {
    str(row['triplet_id']) for row in raw_manifest
} - discovery_ids
assert len(discovery_ids) == DISCOVERY_TRIPLETS
assert len(confirmation_ids) == CONFIRMATION_TRIPLETS
assert not (discovery_ids & confirmation_ids)

factorial_manifest = [
    {
        **row,
        'analysis_split': (
            'discovery' if str(row['triplet_id']) in discovery_ids
            else 'confirmation'
        ),
    }
    for row in raw_manifest
]
split_identity = {
    'schema_version': 'qwen27b-depth-audit-split-1.0',
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'split_seed': SPLIT_SEED,
    'discovery_ids': sorted(discovery_ids),
    'confirmation_ids': sorted(confirmation_ids),
}
SPLIT_LOCK_PATH = EXPERIMENT_ROOT / 'split_lock.json'
if SPLIT_LOCK_PATH.is_file():
    previous = json.loads(SPLIT_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == split_identity
else:
    write_json_atomic(SPLIT_LOCK_PATH, {
        'identity': split_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}
example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=EXPERIMENT_ROOT / '_unused_behavior_loader',
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example
    for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)

manifest_frame = pd.DataFrame([
    {key: value for key, value in row.items() if key != 'task_values'}
    for row in factorial_manifest
])
manifest_frame.to_csv(EXPERIMENT_ROOT / 'depth_audit_manifest.csv', index=False)
display(manifest_frame.groupby([
    'analysis_split', 'changed_side', 'factor_direction', 'outcome_changed'
]).size().rename('n').reset_index())
print('Discovery/confirmation:', len(discovery_ids), len(confirmation_ids))
```

## 3. Fixed-grid causal transport


```python
%%cell_progress 8/17 Define orthogonal tasks, diagnostic options, and exact masks
cell_stage('Defining factor, cross-value, torque, and outcome queries')
import math
from dataclasses import replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side and compare the two side totals.'
)

def task_question(task, side):
    if task == 'side_units':
        return f'How many blocks are on the {side} side?'
    if task == 'side_distance':
        return (
            f'How many tick intervals from the pivot is the stack on the '
            f'{side} side?'
        )
    if task == 'side_torque':
        return f'What is the total {side}-side torque?'
    if task == 'net_torque':
        return 'What is left torque minus right torque?'
    if task == 'outcome':
        return 'Which side will rotate downward?'
    raise KeyError(task)

def option_semantics(row, task):
    values = row['task_values'][task]
    if task == 'outcome':
        semantics = ['left', 'balanced', 'right']
    elif task == 'side_units':
        semantics = [
            int(values['base']),
            int(values['weight_source']),
            int(row['task_values']['side_distance']['distance_source']),
        ]
    elif task == 'side_distance':
        semantics = [
            int(values['base']),
            int(values['distance_source']),
            int(row['task_values']['side_units']['weight_source']),
        ]
    else:
        semantics = []
        for name in ('base', 'weight_source', 'distance_source'):
            value = int(values[name])
            if value not in semantics:
                semantics.append(value)
        anchor = semantics[0]
        for value in (
            anchor + 1, anchor - 1, 0, 1, -1,
            anchor + 2, anchor - 2, anchor + 3, anchor - 3,
        ):
            if len(semantics) == 3:
                break
            if value not in semantics:
                semantics.append(value)
    assert len(semantics) == 3 and len(set(semantics)) == 3, (task, semantics)
    offset = int(stable_key(
        f"options:{row['triplet_id']}:{task}"
    )[:8], 16) % 3
    return tuple(semantics[offset:] + semantics[:offset])

def display_value(task, value):
    if task != 'outcome':
        return str(value)
    return {
        'left': 'left side',
        'balanced': 'balanced',
        'right': 'right side',
    }[str(value)]

def custom_prompt(task, side, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {display_value(task, value)}'
        for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + task_question(task, side),
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_task_bundle(row, task):
    side = str(row['changed_side'])
    semantics = option_semantics(row, task)
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, side, semantics)
    scene_ids = {
        'base': str(row['base_scene_id']),
        'weight_source': str(row['weight_scene_id']),
        'distance_source': str(row['distance_scene_id']),
    }
    examples = {}
    for name, scene_id in scene_ids.items():
        value = row['task_values'][task][name]
        value = str(value) if task == 'outcome' else int(value)
        assert value in semantics, (task, name, value, semantics)
        base = outcome_by_scene[scene_id]
        gold_index = semantics.index(value)
        examples[name] = replace(
            base,
            example_id=(
                f"image_rule:depth_{row['triplet_id']}_{task}_{name}"
            ),
            question_id=f"depth_{row['triplet_id']}_{task}",
            question_kind=task,
            variant='qwen27b_depth_relocalization_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(
                display_value(task, value) for value in semantics
            ),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    goal_name = {
        'side_units': 'weight_source',
        'side_distance': 'distance_source',
        'side_torque': 'weight_source',
        'net_torque': 'weight_source',
        'outcome': 'weight_source',
    }[task]
    cross_semantic = None
    if task == 'side_units':
        cross_semantic = int(
            row['task_values']['side_distance']['distance_source']
        )
    elif task == 'side_distance':
        cross_semantic = int(
            row['task_values']['side_units']['weight_source']
        )
    if cross_semantic is not None:
        assert cross_semantic in semantics
        assert cross_semantic not in {
            examples['base'].gold_semantic,
            examples[goal_name].gold_semantic,
        }
    return {
        'task': task,
        'semantics': semantics,
        'examples': examples,
        'base_semantic': examples['base'].gold_semantic,
        'goal_semantic': examples[goal_name].gold_semantic,
        'cross_semantic': cross_semantic,
        'goal_name': goal_name,
        'estimable': (
            examples['base'].gold_semantic
            != examples[goal_name].gold_semantic
        ),
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(
                reference['inputs'][key], current['inputs'][key]
            ), (name, key)
        masks = discover_masks(current)
        for scope in REQUIRED_SCOPES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Depth-audit prompts and exact masks ready.')
```

```python
%%cell_progress 9/17 Define decoder capture and batched patch scorers
cell_stage('Defining candidate-space scoring and decoder patching')
from collections import defaultdict

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': (
                example.option_semantics[predicted_index]
                == example.gold_semantic
            ),
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = (
                output_hidden_tensor(output).detach().cpu().clone()
            )
        handles.append(
            DECODER_LAYERS[layer_index].register_forward_hook(hook)
        )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def forward_patch_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert len(layer_indices) == 1
    layer_index = next(iter(layer_indices))
    donors = torch.stack([row['donor_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = donors.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == donor.shape
        modified = hidden.clone()
        modified[mask] = donor[mask]
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('donor_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)

    def run_chunk(chunk):
        actual_size = len(chunk)
        assert 1 <= actual_size <= PATCH_CONDITION_BATCH_SIZE
        padded = list(chunk)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        size = len(padded)
        prepared = prepared_cache.get(size)
        if prepared is None:
            prepared = prepare_examples([target_example] * size)
            prepared_cache[size] = prepared
        try:
            return forward_patch_chunk(prepared, padded)[:actual_size]
        except torch.cuda.OutOfMemoryError as exc:
            prepared_cache.pop(size, None)
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Fixed batch-size patching ran out of memory. Restart with '
                'an A100; changing batch size changes numerical parity.'
            ) from exc

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

print('Capture and patch scorers ready.')
```

```python
%%cell_progress 10/17 Run mandatory depth-grid smoke tests
cell_stage('Checking alignment, grid coverage, scopes, and patch parity')
smoke_row = factorial_manifest[0]
assert len(set(smoke_row['scalar_values'])) == 4
smoke_bundle = make_task_bundle(smoke_row, 'side_torque')
smoke_prepared = {
    name: prepare_examples([example])
    for name, example in smoke_bundle['examples'].items()
}
smoke_masks = assert_aligned(smoke_prepared)
for scope in REQUIRED_SCOPES:
    assert bool(smoke_masks[scope].any()), scope

target_prepared = prepare_examples([smoke_bundle['examples']['base']])
source_prepared = prepare_examples([smoke_bundle['examples']['weight_source']])
target_scores, target_hidden = forward_capture_batch(target_prepared)
source_scores, source_hidden = forward_capture_batch(source_prepared)
smoke_layer = MAPPED_MIDDLE_LAYER
smoke_conditions = [
    {
        'condition': 'self_all_tokens', 'task': 'side_torque',
        'donor_type': 'self_base', 'site_label': 'smoke_self',
        'scope': 'valid_tokens', 'layer_index': smoke_layer,
        'donor_scene_id': smoke_row['base_scene_id'],
        'donor_hidden': target_hidden[smoke_layer][0],
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'condition': 'source_all_tokens', 'task': 'side_torque',
        'donor_type': 'weight_source', 'site_label': 'smoke_source',
        'scope': 'valid_tokens', 'layer_index': smoke_layer,
        'donor_scene_id': smoke_row['weight_scene_id'],
        'donor_hidden': source_hidden[smoke_layer][0],
        'patch_mask': smoke_masks['valid_tokens'],
    },
]
smoke_records = score_conditions(
    smoke_bundle['examples']['base'], {}, smoke_conditions
)

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

self_error = float(np.max(np.abs(
    centered_scores(smoke_records[0]) - centered_scores(target_scores[0])
)))
source_error = float(np.max(np.abs(
    centered_scores(smoke_records[1]) - centered_scores(source_scores[0])
)))
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert source_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, source_error
print('Centered self/source parity:', self_error, source_error)
print('Grid layers:', CAPTURE_LAYERS)
print('Token counts:', {
    scope: int(smoke_masks[scope].sum()) for scope in REQUIRED_SCOPES
})
del target_hidden, source_hidden, target_prepared, source_prepared
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/17 Capture or resume the fixed depth grid
cell_stage('Freezing capture identity and checking completed shards')

SITES_PER_TASK = {task: len(sites) for task, sites in TASK_SITES.items()}
RECORDS_PER_TRIPLET = sum(SITES_PER_TASK.values()) * len(DONOR_TYPES)
EXPECTED_PATCH_RECORDS = EXPECTED_TRIPLETS * RECORDS_PER_TRIPLET
capture_identity = {
    'schema_version': 'qwen27b-depth-relocalization-capture-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'dataset_fingerprint': example_config.dataset_fingerprint(),
    'claim_registry_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'split_lock_sha256': sha256_file(SPLIT_LOCK_PATH),
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'grid_layers': list(GRID_LAYERS),
    'task_sites': {
        task: [list(site) for site in sites]
        for task, sites in TASK_SITES.items()
    },
    'records_per_triplet': RECORDS_PER_TRIPLET,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'fixed_patch_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'layer_path': LAYER_PATH,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing depth capture has a different identity.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_triplet_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    triplet_id = str(payload['triplet']['triplet_id'])
    assert triplet_id not in completed_ids
    assert len(payload['patch_records']) == RECORDS_PER_TRIPLET
    completed_ids.add(triplet_id)
    completed_records += len(payload['patch_records'])

progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc='Qwen-27B fixed depth grid', unit='condition', dynamic_ncols=True,
)
for triplet_index, row in enumerate(factorial_manifest):
    triplet_id = str(row['triplet_id'])
    if triplet_id in completed_ids:
        continue
    progress.set_postfix_str(f"{row['analysis_split']} | {triplet_id[-8:]}")
    triplet_baselines = {}
    triplet_patch_records = []
    for task in TASKS:
        bundle = make_task_bundle(row, task)
        examples = bundle['examples']
        prepared = {
            name: prepare_examples([example])
            for name, example in examples.items()
        }
        masks = assert_aligned(prepared)
        scores = {}
        hidden = {}
        for capture_name in ('base', 'weight_source', 'distance_source'):
            capture_prepared = prepare_examples([examples[capture_name]])
            capture_scores, capture_hidden = forward_capture_batch(capture_prepared)
            scores[capture_name] = capture_scores[0]
            hidden[capture_name] = {
                layer_index: value[0]
                for layer_index, value in capture_hidden.items()
            }
            del capture_prepared, capture_scores, capture_hidden
            torch.cuda.empty_cache()
        triplet_baselines[task] = {
            'task': task,
            'semantics': list(bundle['semantics']),
            'base_semantic': bundle['base_semantic'],
            'goal_semantic': bundle['goal_semantic'],
            'cross_semantic': bundle['cross_semantic'],
            'goal_name': bundle['goal_name'],
            'estimable': bool(bundle['estimable']),
            'scores': scores,
        }

        conditions = []
        for layer_index, scope, site_label in TASK_SITES[task]:
            for donor_type in DONOR_TYPES:
                conditions.append({
                    'analysis_family': 'qwen27b_depth_relocalization',
                    'condition': f'{task}__{donor_type}__{site_label}',
                    'task': task,
                    'donor_type': donor_type,
                    'site_label': site_label,
                    'scope': scope,
                    'layer_index': int(layer_index),
                    'donor_scene_id': str(
                        row[f'{donor_type.split("_")[0]}_scene_id']
                    ),
                    'donor_hidden': hidden[donor_type][layer_index],
                    'patch_mask': masks[scope],
                })
        prepared_cache = {1: prepared['base']}
        records = score_conditions(examples['base'], prepared_cache, conditions)
        triplet_patch_records.extend(records)
        progress.update(len(records))
        del prepared, hidden, prepared_cache
        torch.cuda.empty_cache()

    assert len(triplet_patch_records) == RECORDS_PER_TRIPLET
    write_json_atomic(
        SHARD_DIR / f'shard_triplet_{triplet_index:04d}.json',
        {
            'schema_version': 'qwen27b-depth-relocalization-triplet-shard-1.0',
            'capture_identity_sha256': CAPTURE_SHA256,
            'triplet': row,
            'baselines': triplet_baselines,
            'patch_records': triplet_patch_records,
        },
    )
progress.close()

completed = sorted(SHARD_DIR.glob('shard_triplet_*.json'))
assert len(completed) == EXPECTED_TRIPLETS
observed = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_TRIPLET
    observed += len(payload['patch_records'])
assert observed == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'triplets': len(completed),
    'patch_records': observed,
    'capture_identity_sha256': CAPTURE_SHA256,
    'peak_gpu_memory_gb': round(torch.cuda.max_memory_allocated() / 2**30, 3),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Capture complete:', observed, 'patch records')
```

## 4. Discovery selection and held-out confirmation


```python
%%cell_progress 12/17 Release GPU and assemble depth-profile metrics
cell_stage('Releasing model memory')
del adapter
gc.collect()
torch.cuda.empty_cache()

cell_stage('Reading depth shards and computing paired effects')

def centered_vector(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def semantic_margin(record, goal_semantic, base_semantic):
    semantics = list(record['candidate_semantics'])
    scores = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(
        scores[semantics.index(goal_semantic)]
        - scores[semantics.index(base_semantic)]
    )

long_rows = []
paired_rows = []
for path in tqdm(
    sorted(SHARD_DIR.glob('shard_triplet_*.json')),
    desc='Assembling depth metrics', unit='triplet',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    row = payload['triplet']
    patches = pd.DataFrame(payload['patch_records'])
    for task in TASKS:
        baseline = payload['baselines'][task]
        base_record = baseline['scores']['base']
        weight_record = baseline['scores']['weight_source']
        distance_record = baseline['scores']['distance_source']
        base_semantic = baseline['base_semantic']
        goal_semantic = baseline['goal_semantic']
        cross_semantic = baseline['cross_semantic']
        estimable = bool(baseline['estimable'])
        goal_record = baseline['scores'][baseline['goal_name']]
        if estimable:
            base_margin = semantic_margin(base_record, goal_semantic, base_semantic)
            goal_margin = semantic_margin(goal_record, goal_semantic, base_semantic)
            natural_margin = goal_margin - base_margin
        else:
            base_margin = natural_margin = np.nan
        base_cross_margin = (
            semantic_margin(base_record, cross_semantic, base_semantic)
            if cross_semantic is not None else np.nan
        )

        task_records = []
        for _, patch in patches[patches['task'] == task].iterrows():
            patch_record = patch.to_dict()
            if estimable:
                margin_shift = (
                    semantic_margin(patch_record, goal_semantic, base_semantic)
                    - base_margin
                )
            else:
                margin_shift = np.nan
            cross_value_shift = (
                semantic_margin(patch_record, cross_semantic, base_semantic)
                - base_cross_margin
                if cross_semantic is not None else np.nan
            )
            record = {
                'triplet_id': str(row['triplet_id']),
                'analysis_split': str(row['analysis_split']),
                'changed_side': str(row['changed_side']),
                'factor_direction': str(row['factor_direction']),
                'outcome_changed': bool(row['outcome_changed']),
                'delta_side_torque': int(row['delta_side_torque']),
                'task': task,
                'estimable': estimable,
                'donor_type': str(patch['donor_type']),
                'site_label': str(patch['site_label']),
                'layer_index': int(patch['layer_index']),
                'scope': str(patch['scope']),
                'margin_shift': margin_shift,
                'cross_value_shift': cross_value_shift,
                'natural_margin': natural_margin,
                'base_correct': bool(base_record['correct']),
                'weight_source_correct': bool(weight_record['correct']),
                'distance_source_correct': bool(distance_record['correct']),
                'all_three_correct': bool(
                    base_record['correct'] and weight_record['correct']
                    and distance_record['correct']
                ),
                'patched_tokens': int(patch['patched_tokens']),
            }
            long_rows.append(record)
            task_records.append(record)

        for layer_index, scope, site_label in TASK_SITES[task]:
            site_records = [
                value for value in task_records
                if value['site_label'] == site_label
            ]
            assert len(site_records) == 2
            by_donor = {value['donor_type']: value for value in site_records}
            weight_effect = by_donor['weight_source']['margin_shift']
            distance_effect = by_donor['distance_source']['margin_shift']
            if task == 'side_units':
                factor_selectivity = weight_effect - distance_effect
                cross_value_leakage = by_donor['distance_source']['cross_value_shift']
            elif task == 'side_distance':
                factor_selectivity = distance_effect - weight_effect
                cross_value_leakage = by_donor['weight_source']['cross_value_shift']
            else:
                factor_selectivity = cross_value_leakage = np.nan
            paired_rows.append({
                'triplet_id': str(row['triplet_id']),
                'analysis_split': str(row['analysis_split']),
                'changed_side': str(row['changed_side']),
                'factor_direction': str(row['factor_direction']),
                'outcome_changed': bool(row['outcome_changed']),
                'delta_side_torque': int(row['delta_side_torque']),
                'task': task,
                'estimable': estimable,
                'site_label': site_label,
                'layer_index': int(layer_index),
                'scope': scope,
                'weight_effect': weight_effect,
                'distance_effect': distance_effect,
                'mean_donor_effect': (
                    np.nanmean([weight_effect, distance_effect])
                    if estimable else np.nan
                ),
                'donor_effect_difference': (
                    weight_effect - distance_effect if estimable else np.nan
                ),
                'factor_selectivity': factor_selectivity,
                'cross_value_leakage': cross_value_leakage,
                'natural_margin': natural_margin,
                'base_correct': bool(base_record['correct']),
                'weight_source_correct': bool(weight_record['correct']),
                'distance_source_correct': bool(distance_record['correct']),
                'all_three_correct': bool(
                    base_record['correct'] and weight_record['correct']
                    and distance_record['correct']
                ),
            })

long_frame = pd.DataFrame(long_rows)
paired_frame = pd.DataFrame(paired_rows)
expected_paired = EXPECTED_TRIPLETS * sum(SITES_PER_TASK.values())
assert len(long_frame) == EXPECTED_PATCH_RECORDS
assert len(paired_frame) == expected_paired
assert not paired_frame.duplicated([
    'triplet_id', 'task', 'site_label'
]).any()

scale = (
    paired_frame[paired_frame['estimable']]
    .groupby('task')['natural_margin']
    .apply(lambda values: float(np.median(np.abs(values))))
    .rename('task_natural_scale').reset_index()
)
assert (scale['task_natural_scale'] > 0).all()
paired_frame = paired_frame.merge(scale, on='task', validate='many_to_one')
for column in (
    'weight_effect', 'distance_effect', 'mean_donor_effect',
    'donor_effect_difference', 'factor_selectivity',
    'cross_value_leakage',
):
    paired_frame[f'std_{column}'] = (
        paired_frame[column] / paired_frame['task_natural_scale']
    )

long_frame.to_csv(ANALYSIS_DIR / 'patch_level_metrics.csv', index=False)
paired_frame.to_csv(ANALYSIS_DIR / 'paired_depth_metrics.csv', index=False)
scale.to_csv(ANALYSIS_DIR / 'natural_scales.csv', index=False)
print('Patch/paired rows:', len(long_frame), len(paired_frame))
display(scale)
```

```python
%%cell_progress 13/17 Select candidate depths on discovery only
cell_stage('Defining paired inference utilities')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def summarize(values, label):
    values = pd.to_numeric(pd.Series(values), errors='coerce').dropna().to_numpy(float)
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    indices = rng.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    boot = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice((-1.0, 1.0), size=(SIGN_FLIP_REPLICATES, len(values)))
    null = (signs * values).mean(axis=1)
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
        'sign_flip_p': float(
            (1 + np.sum(np.abs(null) >= abs(observed)))
            / (SIGN_FLIP_REPLICATES + 1)
        ),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

cell_stage('Building discovery profiles without touching confirmation rows')
discovery = paired_frame[
    (paired_frame['analysis_split'] == 'discovery')
    & paired_frame['estimable']
].copy()
confirmation = paired_frame[
    (paired_frame['analysis_split'] == 'confirmation')
    & paired_frame['estimable']
].copy()
assert discovery['triplet_id'].nunique() == DISCOVERY_TRIPLETS
assert confirmation['triplet_id'].nunique() == CONFIRMATION_TRIPLETS

profile_rows = []
profile_plan = []
for layer in GRID_LAYERS:
    profile_plan.extend([
        ('side_units', f'text_L{layer:02d}', 'std_factor_selectivity'),
        ('side_distance', f'text_L{layer:02d}', 'std_factor_selectivity'),
        ('side_torque', f'text_L{layer:02d}', 'std_weight_effect'),
        ('side_torque', f'text_L{layer:02d}', 'std_distance_effect'),
        ('side_torque', f'options_L{layer:02d}', 'std_weight_effect'),
        ('side_torque', f'options_L{layer:02d}', 'std_distance_effect'),
        ('outcome', f'answer_L{layer:02d}', 'std_weight_effect'),
        ('outcome', f'answer_L{layer:02d}', 'std_distance_effect'),
    ])
for task, site_label, metric in profile_plan:
    subset = discovery[
        (discovery['task'] == task) & (discovery['site_label'] == site_label)
    ]
    profile_rows.append({
        'task': task, 'site_label': site_label,
        'layer_index': int(subset['layer_index'].iloc[0]),
        'metric': metric,
        **summarize(subset[metric], f'discovery:{task}:{site_label}:{metric}'),
    })
discovery_profile = pd.DataFrame(profile_rows)

def discovery_mean(task, site_label, metric):
    row = discovery_profile[
        (discovery_profile['task'] == task)
        & (discovery_profile['site_label'] == site_label)
        & (discovery_profile['metric'] == metric)
    ]
    assert len(row) == 1
    return float(row.iloc[0]['mean'])

factor_scores = []
for layer in GRID_LAYERS:
    values = [
        discovery_mean(task, f'text_L{layer:02d}', 'std_factor_selectivity')
        for task in ('side_units', 'side_distance')
    ]
    factor_scores.append((min(values), layer, values))
factor_candidate = sorted(factor_scores, key=lambda row: (-row[0], row[1]))[0][1]

def select_min_donor_candidate(task, prefix):
    values = []
    for layer in GRID_LAYERS:
        site = f'{prefix}_L{layer:02d}'
        weight = discovery_mean(task, site, 'std_weight_effect')
        distance = discovery_mean(task, site, 'std_distance_effect')
        values.append((min(weight, distance), layer, weight, distance))
    return sorted(values, key=lambda row: (-row[0], row[1]))[0][1], values

torque_text_candidate, torque_text_scores = select_min_donor_candidate(
    'side_torque', 'text'
)
torque_options_candidate, torque_options_scores = select_min_donor_candidate(
    'side_torque', 'options'
)
outcome_candidate, outcome_scores = select_min_donor_candidate(
    'outcome', 'answer'
)

selection_identity = {
    'schema_version': 'qwen27b-depth-candidate-selection-1.0',
    'split_lock_sha256': sha256_file(SPLIT_LOCK_PATH),
    'paired_metrics_sha256': sha256_file(
        ANALYSIS_DIR / 'paired_depth_metrics.csv'
    ),
    'factor_layer': int(factor_candidate),
    'torque_text_layer': int(torque_text_candidate),
    'torque_options_layer': int(torque_options_candidate),
    'outcome_answer_layer': int(outcome_candidate),
}
SELECTION_PATH = ANALYSIS_DIR / 'discovery_selection.json'
if SELECTION_PATH.is_file():
    previous = json.loads(SELECTION_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == selection_identity
else:
    write_json_atomic(SELECTION_PATH, {
        'identity': selection_identity,
        'selected_at_utc': datetime.now(timezone.utc).isoformat(),
    })
discovery_profile.to_csv(
    ANALYSIS_DIR / 'discovery_depth_profile.csv', index=False
)
print('Discovery-selected layers:', selection_identity)
display(discovery_profile.sort_values(['task', 'metric', 'layer_index']))
```

```python
%%cell_progress 14/17 Test selected depths on untouched confirmation triplets
cell_stage('Running confirmation-only causal and relocation tests')

test_rows = []

def add_test(family, stage, task, site_label, metric, values):
    test_rows.append({
        'family': family, 'stage': stage, 'task': task,
        'site_label': site_label, 'metric': metric,
        **summarize(values, f'confirm:{family}:{stage}:{task}:{site_label}:{metric}'),
    })

def values_at(task, site_label, metric):
    rows = confirmation[
        (confirmation['task'] == task)
        & (confirmation['site_label'] == site_label)
    ].sort_values('triplet_id')
    return rows[['triplet_id', metric]].dropna()

def paired_site_difference(task, candidate_site, mapped_site, metric):
    candidate = values_at(task, candidate_site, metric)
    mapped = values_at(task, mapped_site, metric)
    exact = candidate.merge(
        mapped, on='triplet_id', validate='one_to_one',
        suffixes=('_candidate', '_mapped'),
    )
    return exact[f'{metric}_candidate'] - exact[f'{metric}_mapped']

for task in ('side_units', 'side_distance'):
    candidate_site = f'text_L{factor_candidate:02d}'
    mapped_site = f'text_L{MAPPED_MIDDLE_LAYER:02d}'
    add_test(
        'factor_binding', 'candidate_effect', task, candidate_site,
        'std_factor_selectivity',
        values_at(task, candidate_site, 'std_factor_selectivity')[
            'std_factor_selectivity'
        ],
    )
    add_test(
        'factor_binding', 'candidate_minus_mapped', task, candidate_site,
        'std_factor_selectivity_shift',
        paired_site_difference(
            task, candidate_site, mapped_site, 'std_factor_selectivity'
        ),
    )

torque_candidates = {
    'torque_text': (
        f'text_L{torque_text_candidate:02d}',
        f'text_L{MAPPED_MIDDLE_LAYER:02d}',
    ),
    'torque_options': (
        f'options_L{torque_options_candidate:02d}',
        f'options_L{MAPPED_MIDDLE_LAYER:02d}',
    ),
}
for family, (candidate_site, mapped_site) in torque_candidates.items():
    for metric in (
        'std_weight_effect', 'std_distance_effect',
        'std_mean_donor_effect', 'std_donor_effect_difference',
    ):
        add_test(
            family, 'candidate_effect', 'side_torque', candidate_site,
            metric, values_at('side_torque', candidate_site, metric)[metric],
        )
    add_test(
        family, 'candidate_minus_mapped', 'side_torque', candidate_site,
        'std_mean_donor_effect_shift',
        paired_site_difference(
            'side_torque', candidate_site, mapped_site,
            'std_mean_donor_effect',
        ),
    )

outcome_candidate_site = f'answer_L{outcome_candidate:02d}'
outcome_mapped_site = f'answer_L{MAPPED_LATE_LAYER:02d}'
for metric in (
    'std_weight_effect', 'std_distance_effect',
    'std_mean_donor_effect', 'std_donor_effect_difference',
):
    add_test(
        'outcome_answer', 'candidate_effect', 'outcome',
        outcome_candidate_site, metric,
        values_at('outcome', outcome_candidate_site, metric)[metric],
    )
add_test(
    'outcome_answer', 'candidate_minus_mapped', 'outcome',
    outcome_candidate_site, 'std_mean_donor_effect_shift',
    paired_site_difference(
        'outcome', outcome_candidate_site, outcome_mapped_site,
        'std_mean_donor_effect',
    ),
)

confirmation_tests = pd.DataFrame(test_rows)
confirmation_tests['q_value'] = np.nan
for _, indices in confirmation_tests.groupby('family').groups.items():
    confirmation_tests.loc[indices, 'q_value'] = bh_adjust(
        confirmation_tests.loc[indices, 'sign_flip_p'].to_numpy(float)
    )

def test_record(family, stage, metric, task=None):
    rows = confirmation_tests[
        (confirmation_tests['family'] == family)
        & (confirmation_tests['stage'] == stage)
        & (confirmation_tests['metric'] == metric)
    ]
    if task is not None:
        rows = rows[rows['task'] == task]
    assert len(rows) == 1, (family, stage, metric, task, len(rows))
    return rows.iloc[0]

factor_binding_gate = all(
    test_record(
        'factor_binding', 'candidate_effect',
        'std_factor_selectivity', task,
    )['ci_low'] > 0
    and test_record(
        'factor_binding', 'candidate_effect',
        'std_factor_selectivity', task,
    )['q_value'] < ALPHA
    for task in ('side_units', 'side_distance')
)
factor_relocation_gate = bool(
    factor_candidate != MAPPED_MIDDLE_LAYER
    and any(
        test_record(
            'factor_binding', 'candidate_minus_mapped',
            'std_factor_selectivity_shift', task,
        )['ci_low'] > 0
        and test_record(
            'factor_binding', 'candidate_minus_mapped',
            'std_factor_selectivity_shift', task,
        )['q_value'] < ALPHA
        for task in ('side_units', 'side_distance')
    )
)

torque_gates = {}
for family, candidate_layer in (
    ('torque_text', torque_text_candidate),
    ('torque_options', torque_options_candidate),
):
    weight = test_record(family, 'candidate_effect', 'std_weight_effect')
    distance = test_record(family, 'candidate_effect', 'std_distance_effect')
    difference = test_record(
        family, 'candidate_effect', 'std_donor_effect_difference'
    )
    shift = test_record(
        family, 'candidate_minus_mapped', 'std_mean_donor_effect_shift'
    )
    positive_donors = bool(
        weight['ci_low'] > 0 and weight['q_value'] < ALPHA
        and distance['ci_low'] > 0 and distance['q_value'] < ALPHA
    )
    equivalent_donors = bool(
        difference['ci_low'] > -EQUIVALENCE_FRACTION
        and difference['ci_high'] < EQUIVALENCE_FRACTION
    )
    torque_gates[family] = {
        'candidate_layer': int(candidate_layer),
        'positive_donors': positive_donors,
        'equivalent_donors': equivalent_donors,
        'composition_gate': bool(positive_donors and equivalent_donors),
        'relocation_gate': bool(
            candidate_layer != MAPPED_MIDDLE_LAYER
            and shift['ci_low'] > 0 and shift['q_value'] < ALPHA
        ),
    }

outcome_weight = test_record(
    'outcome_answer', 'candidate_effect', 'std_weight_effect'
)
outcome_distance = test_record(
    'outcome_answer', 'candidate_effect', 'std_distance_effect'
)
outcome_difference = test_record(
    'outcome_answer', 'candidate_effect', 'std_donor_effect_difference'
)
outcome_shift = test_record(
    'outcome_answer', 'candidate_minus_mapped',
    'std_mean_donor_effect_shift',
)
outcome_positive = bool(
    outcome_weight['ci_low'] > 0 and outcome_weight['q_value'] < ALPHA
    and outcome_distance['ci_low'] > 0
    and outcome_distance['q_value'] < ALPHA
)
outcome_equivalent = bool(
    outcome_difference['ci_low'] > -EQUIVALENCE_FRACTION
    and outcome_difference['ci_high'] < EQUIVALENCE_FRACTION
)
outcome_use_gate = bool(outcome_positive and outcome_equivalent)
outcome_relocation_gate = bool(
    outcome_candidate != MAPPED_LATE_LAYER
    and outcome_shift['ci_low'] > 0 and outcome_shift['q_value'] < ALPHA
)

any_torque_composition = any(
    value['composition_gate'] for value in torque_gates.values()
)
any_torque_relocation = any(
    value['relocation_gate'] for value in torque_gates.values()
)
complete_chain_gate = bool(
    factor_binding_gate and any_torque_composition and outcome_use_gate
)
any_confirmed_relocation = bool(
    factor_relocation_gate or any_torque_relocation or outcome_relocation_gate
)

confirmation_tests.to_csv(
    ANALYSIS_DIR / 'confirmation_tests.csv', index=False
)
display(confirmation_tests.sort_values(['family', 'stage', 'metric']))
print('Factor binding/relocation:', factor_binding_gate, factor_relocation_gate)
print('Torque gates:', torque_gates)
print('Outcome use/relocation:', outcome_use_gate, outcome_relocation_gate)
print('Complete chain:', complete_chain_gate)
```

```python
%%cell_progress 15/17 Save discovery and confirmation depth profiles
cell_stage('Rendering held-out depth profiles')

profile_rows = []
plot_plan = []
for layer in GRID_LAYERS:
    plot_plan.extend([
        ('side_units', f'text_L{layer:02d}', 'std_factor_selectivity', 'factor_units'),
        ('side_distance', f'text_L{layer:02d}', 'std_factor_selectivity', 'factor_distance'),
        ('side_torque', f'text_L{layer:02d}', 'std_mean_donor_effect', 'torque_text'),
        ('side_torque', f'options_L{layer:02d}', 'std_mean_donor_effect', 'torque_options'),
        ('outcome', f'answer_L{layer:02d}', 'std_mean_donor_effect', 'outcome_answer'),
    ])
for task, site_label, metric, panel in plot_plan:
    subset = confirmation[
        (confirmation['task'] == task)
        & (confirmation['site_label'] == site_label)
    ]
    profile_rows.append({
        'panel': panel, 'task': task, 'site_label': site_label,
        'layer_index': int(subset['layer_index'].iloc[0]),
        'metric': metric,
        **summarize(subset[metric], f'confirm_profile:{panel}:{site_label}'),
    })
confirmation_profile = pd.DataFrame(profile_rows)
confirmation_profile.to_csv(
    ANALYSIS_DIR / 'confirmation_depth_profile.csv', index=False
)

fig, axes = plt.subplots(2, 2, figsize=(13, 9))

ax = axes[0, 0]
for panel, label, color in (
    ('factor_units', 'units', '#2563eb'),
    ('factor_distance', 'distance', '#f97316'),
):
    rows = confirmation_profile[
        confirmation_profile['panel'] == panel
    ].sort_values('layer_index')
    ax.errorbar(rows['layer_index'], rows['mean'], yerr=np.vstack([
        rows['mean'] - rows['ci_low'], rows['ci_high'] - rows['mean'],
    ]), marker='o', capsize=3, label=label, color=color)
ax.axvline(MAPPED_MIDDLE_LAYER, color='black', linestyle='--', linewidth=1)
ax.axvline(factor_candidate, color='#dc2626', linestyle=':', linewidth=1)
ax.axhline(0, color='black', linewidth=0.7)
ax.set_title('A. Factor binding — confirmation')
ax.set_xlabel('decoder layer')
ax.set_ylabel('standardized selectivity')
ax.legend(frameon=False)

for ax, panel, title, candidate_layer in (
    (axes[0, 1], 'torque_text', 'B. Torque text route', torque_text_candidate),
    (axes[1, 0], 'torque_options', 'C. Torque option route', torque_options_candidate),
    (axes[1, 1], 'outcome_answer', 'D. Late answer route', outcome_candidate),
):
    rows = confirmation_profile[
        confirmation_profile['panel'] == panel
    ].sort_values('layer_index')
    ax.errorbar(rows['layer_index'], rows['mean'], yerr=np.vstack([
        rows['mean'] - rows['ci_low'], rows['ci_high'] - rows['mean'],
    ]), marker='o', capsize=3, color='#4338ca')
    mapped = MAPPED_LATE_LAYER if panel == 'outcome_answer' else MAPPED_MIDDLE_LAYER
    ax.axvline(mapped, color='black', linestyle='--', linewidth=1, label='mapped')
    ax.axvline(candidate_layer, color='#dc2626', linestyle=':', linewidth=1, label='selected')
    ax.axhline(0, color='black', linewidth=0.7)
    ax.set_title(title + ' — confirmation')
    ax.set_xlabel('decoder layer')
    ax.set_ylabel('standardized mean donor effect')
    ax.legend(frameon=False)

fig.suptitle('Qwen3.5-27B depth relocalization audit')
fig.tight_layout()
figure_path = FIGURE_DIR / 'qwen27b_depth_relocalization.png'
fig.savefig(figure_path, dpi=180, bbox_inches='tight')
plt.show()
print('Saved:', figure_path)
```

```python
%%cell_progress 16/17 Apply the frozen depth-audit decision tree
cell_stage('Classifying relocation, broad routing, attenuation, or recovery')

candidate_mean_positive = any(
    test_record(family, 'candidate_effect', 'std_mean_donor_effect')['ci_low'] > 0
    and test_record(family, 'candidate_effect', 'std_mean_donor_effect')['q_value'] < ALPHA
    for family in ('torque_text', 'torque_options', 'outcome_answer')
)

if complete_chain_gate and any_confirmed_relocation:
    decision = 'depth_relocalization_recovers_complete_chain'
elif any_torque_relocation and any_torque_composition:
    decision = 'depth_relocalization_recovers_balanced_composition_route'
elif any_confirmed_relocation:
    decision = 'depth_relocalization_recovers_partial_route'
elif candidate_mean_positive:
    decision = 'broad_route_without_confirmed_peak_shift'
else:
    decision = 'mapped_route_attenuation_not_recovered_on_grid'

decision_payload = {
    'schema_version': 'qwen27b-depth-relocalization-decision-1.0',
    'decision': decision,
    'model_id': MODEL_ID,
    'model_revision': MODEL_REVISION,
    'manifest_sha256': sha256_file(MANIFEST_PATH),
    'split_lock_sha256': sha256_file(SPLIT_LOCK_PATH),
    'claim_registry_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'capture_config_sha256': CAPTURE_SHA256,
    'selection': selection_identity,
    'factor_binding_confirmed': bool(factor_binding_gate),
    'factor_relocation_confirmed': bool(factor_relocation_gate),
    'torque_gates': torque_gates,
    'outcome_use_confirmed': bool(outcome_use_gate),
    'outcome_relocation_confirmed': bool(outcome_relocation_gate),
    'any_relocation_confirmed': bool(any_confirmed_relocation),
    'complete_chain_confirmed': bool(complete_chain_gate),
    'prior_mapped_site_decision': prior_decision['decision'],
    'limitations': [
        'The audit covers eight frozen decoder depths, not every layer.',
        'Grouped decoder-output patching does not identify a unique component circuit.',
        'Each confirmatory half contains 24 triplets, so small shifts may remain unresolved.',
        'Failure to recover a route does not exclude unsampled depths or distributed computation.',
    ],
}
DECISION_PATH = ANALYSIS_DIR / 'depth_relocalization_decision.json'
write_json_atomic(DECISION_PATH, decision_payload)
print(json.dumps(decision_payload, indent=2))
```

```python
%%cell_progress 17/17 Verify mapped-site replication and export the report
cell_stage('Checking new mapped-site rows against the prior frozen 27B run')

site_map = {
    ('side_units', 'middle_text'): f'text_L{MAPPED_MIDDLE_LAYER:02d}',
    ('side_distance', 'middle_text'): f'text_L{MAPPED_MIDDLE_LAYER:02d}',
    ('side_torque', 'middle_text'): f'text_L{MAPPED_MIDDLE_LAYER:02d}',
    ('side_torque', 'middle_options'): f'options_L{MAPPED_MIDDLE_LAYER:02d}',
    ('outcome', 'late_answer'): f'answer_L{MAPPED_LATE_LAYER:02d}',
}
parity_rows = []
for (task, prior_site), audit_site in site_map.items():
    old = prior_paired[
        (prior_paired['task'] == task)
        & (prior_paired['site_label'] == prior_site)
    ].copy()
    new = paired_frame[
        (paired_frame['task'] == task)
        & (paired_frame['site_label'] == audit_site)
    ].copy()
    exact = new.merge(
        old, on='triplet_id', validate='one_to_one', suffixes=('_new', '_old')
    )
    for metric in (
        'weight_effect', 'distance_effect', 'mean_donor_effect',
        'donor_effect_difference', 'factor_selectivity',
    ):
        differences = pd.to_numeric(
            exact[f'{metric}_new'], errors='coerce'
        ) - pd.to_numeric(exact[f'{metric}_old'], errors='coerce')
        finite = differences[np.isfinite(differences)]
        parity_rows.append({
            'task': task, 'prior_site': prior_site,
            'audit_site': audit_site, 'metric': metric,
            'n': len(finite),
            'max_abs_difference': (
                float(np.max(np.abs(finite))) if len(finite) else np.nan
            ),
        })
parity_frame = pd.DataFrame(parity_rows)
parity_frame.to_csv(ANALYSIS_DIR / 'mapped_site_parity.csv', index=False)
max_parity_error = float(parity_frame['max_abs_difference'].max())
assert max_parity_error < 1e-5, max_parity_error

interpretations = {
    'depth_relocalization_recovers_complete_chain': (
        'A discovery-selected depth reproduced on held-out scenes and restored '
        'the full factor-to-torque-to-answer chain.'
    ),
    'depth_relocalization_recovers_balanced_composition_route': (
        'A held-out depth shift recovered balanced torque composition, but the '
        'complete downstream chain is still incomplete.'
    ),
    'depth_relocalization_recovers_partial_route': (
        'At least one factor, torque, or answer route moved to a different '
        'depth, but the complete chain was not recovered.'
    ),
    'broad_route_without_confirmed_peak_shift': (
        'The held-out profile contains causal signal, but no discovery-selected '
        'depth is reliably stronger than the mapped reference depth.'
    ),
    'mapped_route_attenuation_not_recovered_on_grid': (
        'The fixed grid did not recover a stronger route; attenuation or a '
        'distributed/component-specific computation remains the supported boundary.'
    ),
}
report_lines = [
    '# Qwen-27B depth relocalization audit', '',
    f"- Decision: `{decision_payload['decision']}`",
    f'- Discovery/confirmation triplets: {DISCOVERY_TRIPLETS}/{CONFIRMATION_TRIPLETS}',
    f'- Selected factor layer: {factor_candidate}',
    f'- Selected torque text layer: {torque_text_candidate}',
    f'- Selected torque option layer: {torque_options_candidate}',
    f'- Selected outcome layer: {outcome_candidate}',
    f'- Factor binding confirmed: {factor_binding_gate}',
    f'- Any relocation confirmed: {any_confirmed_relocation}',
    f'- Complete chain confirmed: {complete_chain_gate}',
    f'- Mapped-site parity max error: {max_parity_error:.3g}', '',
    '## Interpretation', '',
    interpretations[decision_payload['decision']], '',
    'Selection used discovery triplets only. Every causal gate and candidate-versus-'
    'mapped comparison used the untouched confirmation split.',
]
REPORT_PATH = ANALYSIS_DIR / 'depth_relocalization_report.md'
REPORT_PATH.write_text('\n'.join(report_lines) + '\n', encoding='utf-8')
write_json_atomic(ANALYSIS_DIR / 'paper_claim_card.json', {
    'decision': decision_payload['decision'],
    'interpretation': interpretations[decision_payload['decision']],
    'selection': selection_identity,
    'factor_binding': bool(factor_binding_gate),
    'torque_gates': torque_gates,
    'outcome_use': bool(outcome_use_gate),
    'complete_chain': bool(complete_chain_gate),
    'mapped_site_parity_max_error': max_parity_error,
})
display(parity_frame)
print(REPORT_PATH.read_text(encoding='utf-8'))
print('All outputs:', ANALYSIS_DIR)
```
