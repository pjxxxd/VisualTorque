```python
# Run this helper once. Every later code cell uses %%cell_progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 16 tracked cells.')
```

```python
%%cell_progress 6/16 Select independent competence cohorts and task-value donors
cell_stage('Loading scenes, pairs, prior manifests, and behavior predictions')
from collections import Counter, defaultdict

from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

scene_records = read_jsonl(DATASET_DIR / 'scenes.jsonl')
pair_records = read_jsonl(DATASET_DIR / 'counterfactual_pairs.jsonl')
scene_by_id = {str(row['scene_id']): row for row in scene_records}

behavior_path = (
    PROJECT_ROOT / 'results' / 'behavior_comparison_v3'
    / 'scene_level_predictions.jsonl'
)
robustness_decision_path = ROBUSTNESS_ROOT / 'robustness_decision.json'
discovery_config_path = DISCOVERY_ROOT / 'capture_config.json'
discovery_selection_path = (
    DISCOVERY_ROOT / 'probe_results' / 'discovery_selection.json'
)
for required in (
    behavior_path, robustness_decision_path,
    discovery_config_path, discovery_selection_path,
):
    assert required.is_file(), f'Missing prerequisite: {required}'
robustness_decision = json.loads(
    robustness_decision_path.read_text(encoding='utf-8')
)
assert robustness_decision['decision'] == 'no_targeted_component_search_yet'
discovery_config = json.loads(
    discovery_config_path.read_text(encoding='utf-8')
)
MODEL_REVISION = str(
    discovery_config['identity']['resolved_revision']
)
discovery_selection = json.loads(
    discovery_selection_path.read_text(encoding='utf-8')
)
SELECTED_LAYER_INDEX = int(
    discovery_selection['candidate_net_torque_layer']['layer_index']
)
assert SELECTED_LAYER_INDEX == EXPECTED_SELECTED_LAYER

behavior = pd.read_json(behavior_path, lines=True)
behavior = behavior[behavior['model_name'] == 'Gemma4-12B'].copy()
assert behavior['scene_id'].nunique() == len(scene_records)
behavior['scene_id'] = behavior['scene_id'].astype(str)
correct_by_scene = behavior.set_index('scene_id')['aggregated_correct'].map(
    bool
).to_dict()

prior_manifest_paths = (
    QWEN_REFERENCE_ROOT / 'natural_counterfactual_causal_tracing_v1'
    / 'pair_manifest.jsonl',
    QWEN_REFERENCE_ROOT / 'counterfactual_causal_tracing_confirmatory_v1'
    / 'pair_manifest.jsonl',
    QWEN_REFERENCE_ROOT / 'task_token_component_routing_v1'
    / 'pair_manifest.jsonl',
    QWEN_REFERENCE_ROOT / 'competence_grouped_route_confirmatory_v1'
    / 'pair_manifest.jsonl',
)
excluded_pair_ids = set()
reference_scene_ids = set()
for path in prior_manifest_paths:
    if not path.is_file():
        continue
    for row in read_jsonl(path):
        excluded_pair_ids.add(str(row['pair_id']))
        for key, value in row.items():
            if key.endswith('_scene_id') and value is not None:
                reference_scene_ids.add(str(value))
excluded_scene_ids = set()
print('Excluded prior pair IDs:', len(excluded_pair_ids))
print('Reference endpoint scenes retained for cross-model replication:',
      len(reference_scene_ids))

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value}'.encode('utf-8')).hexdigest()

scenes_by_outcome = defaultdict(list)
for scene in scene_records:
    scenes_by_outcome[str(scene['derived']['outcome'])].append(scene)

def task_value(scene, task):
    derived = scene['derived']
    if task in (
        'left_units', 'right_units', 'left_torque',
        'right_torque', 'net_torque', 'outcome',
    ):
        value = derived[task]
        return str(value) if task == 'outcome' else int(value)
    if task == 'total_units':
        return int(derived['left_units']) + int(derived['right_units'])
    if task == 'torque_magnitude':
        return abs(int(derived['net_torque']))
    if task == 'stack_count':
        return len(scene['stacks'])
    if task == 'max_single_torque':
        return max(
            int(stack['units']) * int(stack['distance'])
            for stack in scene['stacks']
        )
    raise KeyError(task)

RELEVANT_PRIORITIES = {
    'add_unit': (
        'left_units', 'right_units',
        'left_torque', 'right_torque', 'net_torque',
    ),
    'move_stack': (
        'left_torque', 'right_torque', 'net_torque',
        'max_single_torque',
    ),
    'redecompose': (
        'stack_count', 'max_single_torque',
        'left_units', 'right_units',
        'left_torque', 'right_torque',
    ),
    'mirror': (
        'net_torque', 'left_torque', 'right_torque',
        'left_units', 'right_units',
    ),
}
CONTROL_PRIORITIES = {
    'add_unit': (
        'right_units', 'left_units', 'torque_magnitude', 'stack_count',
    ),
    'move_stack': (
        'total_units', 'left_units', 'right_units', 'stack_count',
    ),
    'redecompose': (
        'total_units', 'left_units', 'right_units', 'net_torque',
    ),
    'mirror': (
        'total_units', 'stack_count', 'max_single_torque',
        'torque_magnitude',
    ),
}

def make_task_plan(pair):
    source = scene_by_id[str(pair['source_scene_id'])]
    target = scene_by_id[str(pair['target_scene_id'])]
    values = {}
    for task in set(
        RELEVANT_PRIORITIES[pair['edit_type']]
        + CONTROL_PRIORITIES[pair['edit_type']]
    ):
        values[task] = (task_value(source, task), task_value(target, task))
    current_endpoints = {
        str(pair['source_scene_id']), str(pair['target_scene_id'])
    }
    desired_outcome = str(source['derived']['outcome'])
    for relevant in RELEVANT_PRIORITIES[pair['edit_type']]:
        source_value, target_value = values[relevant]
        if source_value == target_value:
            continue
        for control in CONTROL_PRIORITIES[pair['edit_type']]:
            control_source, control_target = values[control]
            if control == relevant or control_source != control_target:
                continue
            donor_candidates = [
                scene for scene in scenes_by_outcome[desired_outcome]
                if str(scene['scene_id']) not in current_endpoints
                and str(scene['scene_id']) not in excluded_scene_ids
                and task_value(scene, control) == control_source
            ]
            has_value = any(
                task_value(scene, relevant) == source_value
                for scene in donor_candidates
            )
            has_outcome_control = any(
                task_value(scene, relevant)
                not in {source_value, target_value}
                for scene in donor_candidates
            )
            if has_value and has_outcome_control:
                return {
                    'relevant_task': relevant,
                    'relevant_source_value': source_value,
                    'relevant_target_value': target_value,
                    'control_task': control,
                    'control_source_value': control_source,
                    'control_target_value': control_target,
                }
    return None

def strategy_group(problem_type):
    name = str(problem_type)
    if 'conflict' in name:
        return 'conflict'
    if 'compatible' in name:
        return 'compatible'
    if name == 'multi_stack':
        return 'multi_stack'
    if name in ('weight', 'distance'):
        return 'single_cue'
    if name == 'balance':
        return 'balance'
    return 'other'

def pair_cohort(pair):
    source_correct = bool(correct_by_scene[str(pair['source_scene_id'])])
    target_correct = bool(correct_by_scene[str(pair['target_scene_id'])])
    return 'both_correct' if source_correct and target_correct else 'error'

def acceptable_pair(pair, edit_type, cohort):
    if str(pair['pair_id']) in excluded_pair_ids:
        return False
    if pair.get('split') != 'intervention_test':
        return False
    if pair.get('edit_type') != edit_type:
        return False
    endpoints = {
        str(pair['source_scene_id']), str(pair['target_scene_id'])
    }
    if endpoints & excluded_scene_ids:
        return False
    if edit_type == 'redecompose':
        if not (
            pair['source_outcome'] == pair['target_outcome']
            and int(pair['delta_net_torque']) == 0
        ):
            return False
    return pair_cohort(pair) == cohort and make_task_plan(pair) is not None

def round_robin_order(rows):
    groups = defaultdict(list)
    for pair in rows:
        target = scene_by_id[str(pair['target_scene_id'])]
        key = (
            strategy_group(target['problem_type']),
            str(target['problem_type']),
            f"{pair['target_outcome']}->{pair['source_outcome']}",
        )
        groups[key].append(pair)
    for values in groups.values():
        values.sort(key=lambda row: stable_key(row['pair_id']))
    keys = sorted(groups, key=lambda key: stable_key('stratum:' + repr(key)))
    ordered = []
    cursor = {key: 0 for key in keys}
    while True:
        changed = False
        for key in keys:
            index = cursor[key]
            if index < len(groups[key]):
                ordered.append(groups[key][index])
                cursor[key] += 1
                changed = True
        if not changed:
            break
    return ordered

selected_pairs = []
selected_endpoint_ids = set()
availability_rows = []
for edit_type in EDIT_TYPES:
    for cohort in COMPETENCE_COHORTS:
        candidates = [
            pair for pair in pair_records
            if acceptable_pair(pair, edit_type, cohort)
        ]
        availability_rows.append({
            'edit_type': edit_type, 'cohort': cohort,
            'available_before_endpoint_deduplication': len(candidates),
        })
        chosen = []
        for pair in round_robin_order(candidates):
            endpoints = {
                str(pair['source_scene_id']), str(pair['target_scene_id'])
            }
            if endpoints & selected_endpoint_ids:
                continue
            plan = make_task_plan(pair)
            row = {
                **pair, **plan, 'competence_cohort': cohort,
                'source_behavior_correct': bool(
                    correct_by_scene[str(pair['source_scene_id'])]
                ),
                'target_behavior_correct': bool(
                    correct_by_scene[str(pair['target_scene_id'])]
                ),
                'strategy_group': strategy_group(
                    scene_by_id[str(pair['target_scene_id'])]['problem_type']
                ),
            }
            chosen.append(row)
            selected_endpoint_ids.update(endpoints)
            if len(chosen) == PAIRS_PER_EDIT_COHORT:
                break
        assert len(chosen) == PAIRS_PER_EDIT_COHORT, (
            edit_type, cohort, len(chosen), len(candidates)
        )
        selected_pairs.extend(chosen)

cell_stage('Selecting value-matched and outcome-only natural donors')
endpoint_ids = {
    str(row[key]) for row in selected_pairs
    for key in ('source_scene_id', 'target_scene_id')
}
# Pair endpoints remain unique, but they may serve as a capped donor for a
# different pair. Excluding all 384 endpoints makes rare physical values
# impossible to match and is not required for pair-level independence.
donor_exclusions = excluded_scene_ids
donor_reuse = Counter()

def nuisance_features(scene):
    stacks = scene['stacks']
    derived = scene['derived']
    return np.asarray([
        int(derived['left_units']) + int(derived['right_units']),
        len(stacks),
        max(int(stack['distance']) for stack in stacks),
        abs(int(derived['net_torque'])),
        max(int(stack['units']) * int(stack['distance']) for stack in stacks),
    ], dtype=float)

feature_scale = np.std(
    np.stack([nuisance_features(scene) for scene in scene_records]),
    axis=0,
)
feature_scale[feature_scale == 0] = 1

def nuisance_distance(left, right):
    return float(np.linalg.norm(
        (nuisance_features(left) - nuisance_features(right))
        / feature_scale
    ))

def select_donor_pair(row, salt):
    source = scene_by_id[str(row['source_scene_id'])]
    desired_outcome = str(source['derived']['outcome'])
    relevant_task = str(row['relevant_task'])
    control_task = str(row['control_task'])
    source_value = row['relevant_source_value']
    target_value = row['relevant_target_value']
    control_value = row['control_source_value']
    assert control_value == row['control_target_value']
    current_endpoints = {
        str(row['source_scene_id']), str(row['target_scene_id'])
    }
    candidates = [
        scene for scene in scenes_by_outcome[desired_outcome]
        if str(scene['scene_id']) not in donor_exclusions
        and str(scene['scene_id']) not in current_endpoints
        and donor_reuse[str(scene['scene_id'])] < MAX_DONOR_REUSE
        and task_value(scene, control_task) == control_value
    ]
    value_candidates = [
        scene for scene in candidates
        if task_value(scene, relevant_task) == source_value
    ]
    outcome_candidates = [
        scene for scene in candidates
        if task_value(scene, relevant_task)
        not in {source_value, target_value}
    ]
    assert value_candidates and outcome_candidates, (
        row['pair_id'], relevant_task, control_task,
        len(value_candidates), len(outcome_candidates)
    )
    value_candidates.sort(key=lambda scene: (
        nuisance_distance(source, scene),
        donor_reuse[str(scene['scene_id'])],
        stable_key(f'{salt}:value:{scene["scene_id"]}'),
    ))
    outcome_candidates.sort(key=lambda scene: (
        nuisance_distance(source, scene),
        donor_reuse[str(scene['scene_id'])],
        stable_key(f'{salt}:outcome:{scene["scene_id"]}'),
    ))
    best = None
    for value_scene in value_candidates[:24]:
        for outcome_scene in outcome_candidates[:24]:
            if value_scene['scene_id'] == outcome_scene['scene_id']:
                continue
            value_distance = nuisance_distance(source, value_scene)
            outcome_distance = nuisance_distance(source, outcome_scene)
            score = (
                abs(value_distance - outcome_distance)
                + 0.20 * nuisance_distance(value_scene, outcome_scene)
                + 0.25 * (
                    donor_reuse[str(value_scene['scene_id'])]
                    + donor_reuse[str(outcome_scene['scene_id'])]
                )
            )
            key = (score, stable_key(
                f'{salt}:{value_scene["scene_id"]}:{outcome_scene["scene_id"]}'
            ))
            if best is None or key < best[0]:
                best = (key, value_scene, outcome_scene,
                        value_distance, outcome_distance)
    assert best is not None
    _, value_scene, outcome_scene, value_distance, outcome_distance = best
    donor_reuse[str(value_scene['scene_id'])] += 1
    donor_reuse[str(outcome_scene['scene_id'])] += 1
    return {
        'same_task_value_scene_id': str(value_scene['scene_id']),
        'same_task_value_relevant_value': task_value(
            value_scene, relevant_task
        ),
        'same_task_value_control_value': task_value(
            value_scene, control_task
        ),
        'same_task_value_nuisance_distance': value_distance,
        'outcome_only_scene_id': str(outcome_scene['scene_id']),
        'outcome_only_relevant_value': task_value(
            outcome_scene, relevant_task
        ),
        'outcome_only_control_value': task_value(
            outcome_scene, control_task
        ),
        'outcome_only_nuisance_distance': outcome_distance,
    }

pair_manifest = []
for row in tqdm(
    selected_pairs, desc='Selecting natural donor triplets', unit='pair'
):
    donors = select_donor_pair(row, str(row['pair_id']))
    pair_manifest.append({
        **row, **donors,
    })
pair_manifest.sort(key=lambda row: str(row['pair_id']))

manifest_path = OUTPUT_ROOT / 'pair_manifest.jsonl'
temporary = manifest_path.with_suffix('.jsonl.tmp')
with temporary.open('w', encoding='utf-8') as handle:
    for row in pair_manifest:
        handle.write(json.dumps(row, sort_keys=True) + '\n')
temporary.replace(manifest_path)
pd.DataFrame(availability_rows).to_csv(
    ANALYSIS_DIR / 'pair_pool_availability.csv', index=False
)
selection_frame = pd.DataFrame(pair_manifest)
selection_frame.to_csv(ANALYSIS_DIR / 'pair_manifest.csv', index=False)

example_config = BehaviorRunConfig(
    dataset_dir=DATASET_DIR,
    output_dir=Path('/content/unused_grouped_route'),
    model_id=MODEL_ID,
    model_revision=MODEL_REVISION,
    conditions=(PromptCondition.IMAGE_RULE,),
    question_kinds=('outcome',),
    splits=('probe_train', 'probe_dev', 'probe_test'),
    batch_size=1, generation_every=0, seed=SEED,
)
outcome_examples = tuple(load_behavior_examples(example_config))
outcome_by_scene = {
    str(example.scene_id): example
    for example in outcome_examples
    if example.variant == FIXED_VARIANT
}
assert len(outcome_by_scene) == len(scene_records)
assert len(pair_manifest) == (
    len(EDIT_TYPES) * len(COMPETENCE_COHORTS)
    * PAIRS_PER_EDIT_COHORT
)
assert max(donor_reuse.values()) <= MAX_DONOR_REUSE
print('Selected pairs:', len(pair_manifest))
display(selection_frame.groupby(
    ['edit_type', 'competence_cohort']
).size().unstack(fill_value=0))
print('Relevant tasks:')
display(selection_frame.groupby(
    ['edit_type', 'relevant_task']
).size().rename('n').reset_index())
print('Unchanged control tasks:')
display(selection_frame.groupby(
    ['edit_type', 'control_task']
).size().rename('n').reset_index())
print('Maximum donor reuse:', max(donor_reuse.values()))
```

## 3. Pair-aligned task prompts and grouped token masks

Source, target, and both unrelated donors receive exactly the same pair-specific prompt and A/B/C option mapping.


```python
%%cell_progress 8/16 Define pair-aligned tasks, options, and exact masks
cell_stage('Defining custom latent-variable queries')
import math
from dataclasses import replace

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    "intervals from the central pivot. A stack's turning effect is its "
    'number of squares multiplied by its distance. Add turning effects '
    'within each side and compare the two side totals.'
)
TASK_QUESTIONS = {
    'left_units': 'How many blocks are on the left side?',
    'right_units': 'How many blocks are on the right side?',
    'left_torque': 'What is the total left-side torque?',
    'right_torque': 'What is the total right-side torque?',
    'net_torque': 'What is left torque minus right torque?',
    'total_units': 'How many blocks are present in total?',
    'torque_magnitude': 'What is the absolute value of net torque?',
    'stack_count': 'How many distinct block stacks are shown?',
    'max_single_torque': (
        'What is the largest torque contributed by any single stack?'
    ),
}

def option_semantics(row, query_role):
    prefix = 'relevant' if query_role == 'relevant_changed' else 'control'
    values = []
    for key in (
        f'{prefix}_source_value', f'{prefix}_target_value',
        f'outcome_only_{prefix}_value',
    ):
        value = int(row[key])
        if value not in values:
            values.append(value)
    anchor = values[0]
    for value in (
        anchor + 1, anchor - 1, 0, 1, -1,
        anchor + 2, anchor - 2,
    ):
        if len(values) >= 3:
            break
        if value not in values:
            values.append(value)
    assert len(values) == 3
    offset = int(stable_key(
        f"options:{row['pair_id']}:{query_role}"
    )[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def custom_prompt(task, semantics):
    labels = ('A', 'B', 'C')
    options = '\n'.join(
        f'{label}. {value}'
        for label, value in zip(labels, semantics)
    )
    return '\n\n'.join([
        'Solve this balance-scale question using only the supplied scene.',
        RULE_TEXT,
        'Question: ' + TASK_QUESTIONS[task],
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_query_bundle(row, query_role):
    prefix = 'relevant' if query_role == 'relevant_changed' else 'control'
    task = str(row[f'{prefix}_task'])
    semantics = option_semantics(row, query_role)
    labels = ('A', 'B', 'C')
    prompt = custom_prompt(task, semantics)
    scene_ids = {
        'source': str(row['source_scene_id']),
        'target': str(row['target_scene_id']),
        'same_task_value_different_layout': str(
            row['same_task_value_scene_id']
        ),
        'same_outcome_different_task_value': str(
            row['outcome_only_scene_id']
        ),
    }

    def value_for(name):
        if name == 'source':
            return int(row[f'{prefix}_source_value'])
        if name == 'target':
            return int(row[f'{prefix}_target_value'])
        if name == 'same_task_value_different_layout':
            return int(row[f'same_task_value_{prefix}_value'])
        return int(row[f'outcome_only_{prefix}_value'])

    examples = {}
    for name, scene_id in scene_ids.items():
        value = value_for(name)
        assert value in semantics, (task, name, value, semantics)
        base = outcome_by_scene[scene_id]
        gold_index = semantics.index(value)
        examples[name] = replace(
            base,
            example_id=(
                f"image_rule:grouped_{row['pair_id']}_"
                f'{query_role}_{name}'
            ),
            question_id=(
                f"grouped_{row['pair_id']}_{query_role}"
            ),
            question_kind=task,
            variant='grouped_route_fixed_options',
            prompt=prompt,
            labels=labels,
            option_text=tuple(str(value) for value in semantics),
            option_semantics=semantics,
            gold_label=labels[gold_index],
            gold_semantic=value,
        )
    return {
        'task': task, 'semantics': semantics,
        'examples': examples,
        'source_value': int(row[f'{prefix}_source_value']),
        'target_value': int(row[f'{prefix}_target_value']),
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(
        output.last_hidden_state
    ):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {
        'examples': examples, 'inputs': inputs,
        'token_by_label': token_by_label,
    }

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = (
            IMAGE_TOKEN_ID_OVERRIDE
            if isinstance(IMAGE_TOKEN_ID_OVERRIDE, (list, tuple))
            else [IMAGE_TOKEN_ID_OVERRIDE]
        )
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any())
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False,
        return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if ids.tolist()[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention,
        'image_tokens': image,
        'all_text_except_final': all_text,
        'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    covered = torch.zeros_like(attention)
    for name in names:
        covered |= masks[name]
    masks['separator_chat_tokens'] = all_text & ~covered
    assert torch.equal(
        masks['options_tokens'] | masks['intro_tokens']
        | masks['rule_tokens'] | masks['question_tokens']
        | masks['answer_instruction_tokens']
        | masks['separator_chat_tokens'],
        all_text,
    )
    masks['tokens'] = tokens
    return masks

def assert_aligned(prepared_by_name):
    names = list(prepared_by_name)
    reference = prepared_by_name[names[0]]
    reference_masks = discover_masks(reference)
    for name in names[1:]:
        current = prepared_by_name[name]
        for key in ('input_ids', 'attention_mask'):
            assert torch.equal(
                reference['inputs'][key], current['inputs'][key]
            ), (name, key)
        masks = discover_masks(current)
        for _, scope in FROZEN_SITES:
            assert torch.equal(reference_masks[scope], masks[scope]), (
                name, scope
            )
    return reference_masks

print('Grouped-query and exact-mask utilities ready.')
```

```python
%%cell_progress 9/16 Define grouped decoder capture and patch scorers
cell_stage('Defining candidate-space scoring and grouped patching')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[
                row_index, prepared['token_by_label'][label]
            ].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        total = sum(weights)
        probabilities = [value / total for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
            'correct': (
                example.option_semantics[predicted_index]
                == example.gold_semantic
            ),
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = (
                output_hidden_tensor(output).detach().cpu().clone()
            )
        handles.append(
            DECODER_LAYERS[layer_index].register_forward_hook(hook)
        )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def forward_capture(prepared):
    assert len(prepared['examples']) == 1
    records, hidden = forward_capture_batch(prepared)
    return records[0], {
        layer_index: value[0]
        for layer_index, value in hidden.items()
    }

def forward_patch_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert len(layer_indices) == 1
    layer_index = next(iter(layer_indices))
    donors = torch.stack([row['donor_hidden'] for row in conditions])
    masks = torch.stack([row['patch_mask'] for row in conditions])
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        donor = donors.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        assert hidden.shape == donor.shape
        modified = hidden.clone()
        modified[mask] = donor[mask]
        audit['counts'] = mask.sum(dim=1).detach().cpu().tolist()
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[layer_index].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('donor_hidden', 'patch_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)

    def run_chunk(chunk):
        actual_size = len(chunk)
        assert 1 <= actual_size <= PATCH_CONDITION_BATCH_SIZE
        padded = list(chunk)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        size = len(padded)
        try:
            prepared = prepared_cache.get(size)
            if prepared is None:
                prepared = prepare_examples([target_example] * size)
                prepared_cache[size] = prepared
            return forward_patch_chunk(prepared, padded)[:actual_size]
        except torch.cuda.OutOfMemoryError:
            prepared_cache.pop(size, None)
            try:
                del prepared
            except UnboundLocalError:
                pass
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Fixed batch-size-4 patching ran out of memory. Reducing '
                'the batch would invalidate the batch-matched baseline; '
                'use a larger GPU or lower the image/runtime footprint.'
            )

    for group in grouped.values():
        for start in range(0, len(group), PATCH_CONDITION_BATCH_SIZE):
            records.extend(run_chunk(
                group[start:start + PATCH_CONDITION_BATCH_SIZE]
            ))
    return records

print('Grouped decoder capture and patching ready.')
```

```python
%%cell_progress 10/16 Run mandatory grouped-route smoke tests
cell_stage('Checking task changes, donor semantics, masks, and patch parity')
smoke_pair = pair_manifest[0]
smoke_bundle = make_query_bundle(smoke_pair, 'relevant_changed')
assert smoke_bundle['source_value'] != smoke_bundle['target_value']
control_bundle = make_query_bundle(smoke_pair, 'unchanged_control')
assert control_bundle['source_value'] == control_bundle['target_value']

smoke_prepared = {
    name: prepare_examples([example])
    for name, example in smoke_bundle['examples'].items()
}
smoke_masks = assert_aligned(smoke_prepared)
for _, scope in FROZEN_SITES:
    assert bool(smoke_masks[scope].any()), scope
smoke_target_batch = prepare_examples([
    smoke_bundle['examples']['target']
] * PATCH_CONDITION_BATCH_SIZE)
smoke_source_batch = prepare_examples([
    smoke_bundle['examples']['source']
] * PATCH_CONDITION_BATCH_SIZE)
smoke_target_scores, smoke_target_hidden = forward_capture_batch(
    smoke_target_batch
)
smoke_source_scores, smoke_source_hidden = forward_capture_batch(
    smoke_source_batch
)

SMOKE_LAYER = FOCAL_SITES[0][0]
smoke_conditions = [
    {
        'analysis_family': 'smoke',
        'condition': 'self_all_tokens',
        'query_role': 'relevant_changed',
        'donor_type': 'self_target',
        'scope': 'valid_tokens',
        'layer_index': SMOKE_LAYER,
        'donor_scene_id': smoke_pair['target_scene_id'],
        'donor_hidden': smoke_target_hidden[SMOKE_LAYER][0],
        'patch_mask': smoke_masks['valid_tokens'],
    },
    {
        'analysis_family': 'smoke',
        'condition': 'source_all_tokens',
        'query_role': 'relevant_changed',
        'donor_type': 'matched_source',
        'scope': 'valid_tokens',
        'layer_index': SMOKE_LAYER,
        'donor_scene_id': smoke_pair['source_scene_id'],
        'donor_hidden': smoke_source_hidden[SMOKE_LAYER][0],
        'patch_mask': smoke_masks['valid_tokens'],
    },
]
smoke_records = score_conditions(
    smoke_bundle['examples']['target'], {}, smoke_conditions
)
def centered_candidate_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

self_raw_error = max(
    abs(a - b) for a, b in zip(
        smoke_records[0]['candidate_logprobs'],
        smoke_target_scores[0]['candidate_logprobs'],
    )
)
source_raw_error = max(
    abs(a - b) for a, b in zip(
        smoke_records[1]['candidate_logprobs'],
        smoke_source_scores[0]['candidate_logprobs'],
    )
)
self_error = max(
    np.abs(
        centered_candidate_scores(smoke_records[0])
        - centered_candidate_scores(smoke_target_scores[0])
    )
)
source_error = max(
    np.abs(
        centered_candidate_scores(smoke_records[1])
        - centered_candidate_scores(smoke_source_scores[0])
    )
)
print(
    'Raw self/source logprob errors:',
    self_raw_error, source_raw_error,
)
print(
    'Centered self/source candidate errors:',
    self_error, source_error,
)
assert self_error < SMOKE_LOGPROB_TOLERANCE, self_error
assert source_error < SMOKE_LOGPROB_TOLERANCE, source_error
print('Smoke relevant/control tasks:',
      smoke_bundle['task'], control_bundle['task'])
print('Relevant semantics:', smoke_bundle['semantics'])
print('Frozen-site token counts:', {
    f'L{layer}:{scope}': int(smoke_masks[scope].sum())
    for layer, scope in FROZEN_SITES
})
del (
    smoke_prepared, smoke_target_batch, smoke_source_batch,
    smoke_target_hidden, smoke_source_hidden,
)
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/16 Capture or resume grouped-route interventions
cell_stage('Freezing capture identity and expected pair shards')

from visualtorque_checks import sha256_file

CONDITIONS_PER_QUERY = len(FROZEN_SITES) * len(DONOR_TYPES)
RECORDS_PER_PAIR = len(QUERY_ROLES) * CONDITIONS_PER_QUERY
EXPECTED_PATCH_RECORDS = len(pair_manifest) * RECORDS_PER_PAIR
capture_identity = {
    'schema_version': 'competence-grouped-route-capture-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'dtype': MODEL_DTYPE,
    'dataset_fingerprint': example_config.dataset_fingerprint(),
    'protocol_lock_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'pair_manifest_sha256': sha256_file(manifest_path),
    'frozen_sites': [list(site) for site in FROZEN_SITES],
    'donor_types': list(DONOR_TYPES),
    'query_roles': list(QUERY_ROLES),
    'expected_pairs': len(pair_manifest),
    'records_per_pair': RECORDS_PER_PAIR,
    'expected_patch_records': EXPECTED_PATCH_RECORDS,
    'fixed_patch_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'partial_batches_are_padded': True,
    'layer_path': LAYER_PATH,
    'seed': SEED,
}
capture_config_path = OUTPUT_ROOT / 'capture_config.json'
if capture_config_path.is_file():
    previous = json.loads(capture_config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing grouped-route capture differs. Use a new output version.'
    )
else:
    write_json_atomic(capture_config_path, {
        'identity': capture_identity,
        'adapter_metadata': adapter.metadata,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
CAPTURE_SHA256 = sha256_file(capture_config_path)

completed_pair_ids = set()
completed_records = 0
for path in sorted(SHARD_DIR.glob('shard_pair_*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    pair_id = str(payload['pair']['pair_id'])
    assert pair_id not in completed_pair_ids
    assert len(payload['patch_records']) == RECORDS_PER_PAIR
    completed_pair_ids.add(pair_id)
    completed_records += len(payload['patch_records'])
progress = tqdm(
    total=EXPECTED_PATCH_RECORDS, initial=completed_records,
    desc='Grouped-route conditions', unit='condition',
    dynamic_ncols=True,
)

for pair_index, pair in enumerate(pair_manifest):
    pair_id = str(pair['pair_id'])
    if pair_id in completed_pair_ids:
        continue
    progress.set_postfix_str(
        f"{pair['edit_type']} | {pair['competence_cohort']}"
    )
    pair_baselines = {}
    pair_patch_records = []
    for query_role in QUERY_ROLES:
        bundle = make_query_bundle(pair, query_role)
        examples = bundle['examples']
        prepared = {
            name: prepare_examples([example])
            for name, example in examples.items()
        }
        masks = assert_aligned(prepared)
        scores = {}
        hidden = {}
        capture_names = (
            'source', 'target',
            'same_task_value_different_layout',
            'same_outcome_different_task_value',
        )
        capture_prepared = prepare_examples([
            examples[name] for name in capture_names
        ])
        capture_scores, capture_hidden = forward_capture_batch(
            capture_prepared
        )
        for row_index, name in enumerate(capture_names):
            scores[name] = capture_scores[row_index]
            hidden[name] = {
                layer_index: value[row_index]
                for layer_index, value in capture_hidden.items()
            }
        pair_baselines[query_role] = {
            'task': bundle['task'],
            'source_value': bundle['source_value'],
            'target_value': bundle['target_value'],
            'semantics': list(bundle['semantics']),
            'scores': scores,
        }

        donor_hidden = {
            'matched_source': hidden['source'],
            'same_task_value_different_layout': hidden[
                'same_task_value_different_layout'
            ],
            'same_outcome_different_task_value': hidden[
                'same_outcome_different_task_value'
            ],
        }
        donor_scene_ids = {
            'matched_source': str(pair['source_scene_id']),
            'same_task_value_different_layout': str(
                pair['same_task_value_scene_id']
            ),
            'same_outcome_different_task_value': str(
                pair['outcome_only_scene_id']
            ),
        }
        conditions = []
        for layer_index, scope in FROZEN_SITES:
            for donor_type in DONOR_TYPES:
                conditions.append({
                    'analysis_family': 'grouped_route',
                    'condition': (
                        f'{query_role}__{donor_type}__'
                        f'L{layer_index}__{scope}'
                    ),
                    'query_role': query_role,
                    'task': bundle['task'],
                    'task_value_changed': bool(
                        bundle['source_value'] != bundle['target_value']
                    ),
                    'donor_type': donor_type,
                    'scope': scope,
                    'layer_index': int(layer_index),
                    'donor_scene_id': donor_scene_ids[donor_type],
                    'donor_hidden': donor_hidden[donor_type][layer_index],
                    'patch_mask': masks[scope],
                })
        prepared_cache = {1: prepared['target']}
        records = score_conditions(
            examples['target'], prepared_cache, conditions
        )
        pair_patch_records.extend(records)
        progress.update(len(records))
        del (
            prepared, hidden, donor_hidden, prepared_cache,
            capture_prepared, capture_hidden,
        )
        torch.cuda.empty_cache()

    assert len(pair_patch_records) == RECORDS_PER_PAIR
    payload = {
        'schema_version': 'competence-grouped-route-pair-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'pair': pair,
        'baselines': pair_baselines,
        'patch_records': pair_patch_records,
    }
    shard_path = SHARD_DIR / f'shard_pair_{pair_index:04d}.json'
    write_json_atomic(shard_path, payload)
progress.close()

completed = sorted(SHARD_DIR.glob('shard_pair_*.json'))
assert len(completed) == len(pair_manifest)
observed_records = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_PAIR
    observed_records += len(payload['patch_records'])
assert observed_records == EXPECTED_PATCH_RECORDS
write_json_atomic(OUTPUT_ROOT / 'capture_complete.json', {
    'pairs': len(completed),
    'patch_records': observed_records,
    'capture_identity_sha256': CAPTURE_SHA256,
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
print('Capture complete:', observed_records, 'patch records')
```

```python
%%cell_progress 12/16 Release GPU and assemble grouped-route metrics
cell_stage('Releasing model memory')
del adapter
gc.collect()
torch.cuda.empty_cache()

cell_stage('Reading atomic pair shards')
patch_rows = []
baseline_rows = []
for path in tqdm(
    sorted(SHARD_DIR.glob('shard_pair_*.json')),
    desc='Reading grouped-route shards', unit='pair',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    pair = payload['pair']
    baselines = payload['baselines']
    for query_role, baseline in baselines.items():
        scores = baseline['scores']
        baseline_rows.append({
            **pair,
            'query_role': query_role,
            'task': baseline['task'],
            'task_value_changed': bool(
                baseline['source_value'] != baseline['target_value']
            ),
            'source_value': baseline['source_value'],
            'target_value': baseline['target_value'],
            'source_correct': bool(scores['source']['correct']),
            'target_correct': bool(scores['target']['correct']),
            'query_both_correct': bool(
                scores['source']['correct']
                and scores['target']['correct']
            ),
            'value_donor_correct': bool(
                scores['same_task_value_different_layout']['correct']
            ),
            'outcome_donor_correct': bool(
                scores['same_outcome_different_task_value']['correct']
            ),
        })
    for record in payload['patch_records']:
        baseline = baselines[record['query_role']]
        source = baseline['scores']['source']
        target = baseline['scores']['target']
        source_vector = np.asarray(
            source['candidate_logprobs'], dtype=np.float64
        )
        target_vector = np.asarray(
            target['candidate_logprobs'], dtype=np.float64
        )
        patched_vector = np.asarray(
            record['candidate_logprobs'], dtype=np.float64
        )
        source_vector -= source_vector.mean()
        target_vector -= target_vector.mean()
        patched_vector -= patched_vector.mean()
        natural_delta = source_vector - target_vector
        natural_norm = float(np.linalg.norm(natural_delta))
        denominator = float(np.dot(natural_delta, natural_delta))
        patch_delta = patched_vector - target_vector
        recovery = None
        if natural_norm >= NATURAL_EFFECT_EPS:
            recovery = float(
                np.dot(patch_delta, natural_delta) / denominator
            )
        source_answer_gain = None
        if baseline['source_value'] != baseline['target_value']:
            semantics = list(baseline['semantics'])
            source_index = semantics.index(baseline['source_value'])
            target_index = semantics.index(baseline['target_value'])
            source_answer_gain = float(
                patch_delta[source_index] - patch_delta[target_index]
            )
        patch_rows.append({
            **pair, **record,
            'source_value': baseline['source_value'],
            'target_value': baseline['target_value'],
            'natural_effect_norm': natural_norm,
            'valid_recovery': recovery is not None,
            'recovery': recovery,
            'patch_effect_norm': float(np.linalg.norm(patch_delta)),
            'patched_centered_0': float(patched_vector[0]),
            'patched_centered_1': float(patched_vector[1]),
            'patched_centered_2': float(patched_vector[2]),
            'source_answer_contrast_gain': source_answer_gain,
            'source_correct': bool(source['correct']),
            'target_correct': bool(target['correct']),
            'query_both_correct': bool(
                source['correct'] and target['correct']
            ),
        })
patch_frame = pd.DataFrame(patch_rows)
baseline_frame = pd.DataFrame(baseline_rows)
assert len(patch_frame) == EXPECTED_PATCH_RECORDS
assert patch_frame['pair_id'].nunique() == len(pair_manifest)
patch_frame.to_csv(ANALYSIS_DIR / 'patch_records.csv', index=False)
baseline_frame.to_csv(ANALYSIS_DIR / 'query_baselines.csv', index=False)
print('Patch records:', len(patch_frame))
print('Valid recovery by query role:')
display(patch_frame.groupby('query_role')['valid_recovery'].mean())
print('Actual custom-query competence:')
display(baseline_frame.groupby(
    ['competence_cohort', 'query_role']
).agg(
    pairs=('pair_id', 'nunique'),
    both_correct=('query_both_correct', 'mean'),
    source_accuracy=('source_correct', 'mean'),
    target_accuracy=('target_correct', 'mean'),
))
```

```python
%%cell_progress 13/16 Test physical-value specificity and query selectivity
cell_stage('Building pair-level donor gaps at six frozen sites')
rng = np.random.default_rng(SEED + 1000)

def bh_adjust(p_values):
    values = np.asarray(p_values, dtype=float)
    result = np.full(values.shape, np.nan)
    valid = np.isfinite(values)
    if not valid.any():
        return result
    observed = values[valid]
    order = np.argsort(observed)
    ranked = observed[order]
    adjusted = ranked * len(ranked) / np.arange(1, len(ranked) + 1)
    adjusted = np.minimum.accumulate(adjusted[::-1])[::-1]
    restored = np.empty_like(adjusted)
    restored[order] = np.minimum(adjusted, 1.0)
    result[valid] = restored
    return result

def paired_statistics(values, label):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'p_value': np.nan,
        }
    local_rng = np.random.default_rng(int(
        hashlib.sha256(f'{SEED}:{label}'.encode()).hexdigest()[:8], 16
    ))
    counts = local_rng.multinomial(
        len(values), np.full(len(values), 1 / len(values)),
        size=BOOTSTRAP_REPLICATES,
    )
    draws = counts @ values / len(values)
    signs = local_rng.choice(
        np.asarray([-1.0, 1.0]),
        size=(SIGN_FLIP_REPLICATES, len(values)),
    )
    permuted = signs @ values / len(values)
    observed = float(values.mean())
    return {
        'n': int(len(values)), 'mean': observed,
        'ci_low': float(np.quantile(draws, 0.025)),
        'ci_high': float(np.quantile(draws, 0.975)),
        'p_value': float(
            (1 + np.sum(permuted >= observed))
            / (SIGN_FLIP_REPLICATES + 1)
        ),
    }

valid = patch_frame[
    patch_frame['valid_recovery']
    & (patch_frame['query_role'] == 'relevant_changed')
].copy()
recovery_pivot = valid.pivot_table(
    index=[
        'pair_id', 'layer_index', 'scope',
        'competence_cohort', 'query_both_correct', 'edit_type',
    ],
    columns='donor_type', values='recovery', aggfunc='first',
).reset_index()
assert set(DONOR_TYPES) <= set(recovery_pivot.columns)
recovery_pivot['value_minus_outcome'] = (
    recovery_pivot['same_task_value_different_layout']
    - recovery_pivot['same_outcome_different_task_value']
)
recovery_pivot['matched_minus_outcome'] = (
    recovery_pivot['matched_source']
    - recovery_pivot['same_outcome_different_task_value']
)
recovery_pivot.to_csv(
    ANALYSIS_DIR / 'pair_level_donor_gaps.csv', index=False
)

cell_stage('Measuring prompt-gated separation of the same donor pair')
separation_rows = []
vector_columns = (
    'patched_centered_0', 'patched_centered_1',
    'patched_centered_2',
)
separation_keys = [
    'pair_id', 'query_role', 'layer_index', 'scope',
    'competence_cohort', 'query_both_correct', 'edit_type',
]
for keys, group in patch_frame.groupby(separation_keys, sort=False):
    donors = group.set_index('donor_type')
    assert set(DONOR_TYPES) <= set(donors.index)
    value_vector = donors.loc[
        'same_task_value_different_layout', list(vector_columns)
    ].to_numpy(dtype=float)
    outcome_vector = donors.loc[
        'same_outcome_different_task_value', list(vector_columns)
    ].to_numpy(dtype=float)
    matched_vector = donors.loc[
        'matched_source', list(vector_columns)
    ].to_numpy(dtype=float)
    separation_rows.append({
        **dict(zip(separation_keys, keys)),
        'value_outcome_separation': float(np.linalg.norm(
            value_vector - outcome_vector
        )),
        'matched_outcome_separation': float(np.linalg.norm(
            matched_vector - outcome_vector
        )),
    })
separation_frame = pd.DataFrame(separation_rows)
assert len(separation_frame) == (
    len(pair_manifest) * len(QUERY_ROLES) * len(FROZEN_SITES)
)
separation_frame.to_csv(
    ANALYSIS_DIR / 'pair_level_donor_separation.csv', index=False
)

populations = {
    'all_pairs': lambda frame: pd.Series(True, index=frame.index),
    'outcome_both_correct_cohort': lambda frame: (
        frame['competence_cohort'] == 'both_correct'
    ),
    'actual_query_both_correct': lambda frame: frame['query_both_correct'],
    'outcome_error_cohort': lambda frame: (
        frame['competence_cohort'] == 'error'
    ),
}
specificity_rows = []
for population, selector in populations.items():
    subset = recovery_pivot[selector(recovery_pivot)]
    for layer_index, scope in FROZEN_SITES:
        site = subset[
            (subset['layer_index'] == layer_index)
            & (subset['scope'] == scope)
        ]
        for contrast in (
            'value_minus_outcome', 'matched_minus_outcome'
        ):
            specificity_rows.append({
                'population': population,
                'layer_index': int(layer_index),
                'scope': scope,
                'contrast': contrast,
                **paired_statistics(
                    site[contrast].to_numpy(),
                    f'{population}:L{layer_index}:'
                    f'{scope}:{contrast}',
                ),
            })
specificity_frame = pd.DataFrame(specificity_rows)
specificity_frame['q_bh'] = np.nan
for keys, indices in specificity_frame.groupby(
    ['population', 'contrast']
).groups.items():
    specificity_frame.loc[indices, 'q_bh'] = bh_adjust(
        specificity_frame.loc[indices, 'p_value']
    )
specificity_frame['passes_fdr'] = (
    (specificity_frame['mean'] > 0)
    & (specificity_frame['q_bh'] < ALPHA)
)
specificity_frame.to_csv(
    ANALYSIS_DIR / 'donor_value_specificity.csv', index=False
)

cell_stage('Computing within-pair relevant-minus-unchanged selectivity')
selectivity_rows = []
for population, selector in populations.items():
    subset = separation_frame[selector(separation_frame)]
    for layer_index, scope in FROZEN_SITES:
        site = subset[
            (subset['layer_index'] == layer_index)
            & (subset['scope'] == scope)
        ]
        pivot = site.pivot_table(
            index='pair_id', columns='query_role',
            values='value_outcome_separation', aggfunc='first',
        )
        if set(QUERY_ROLES) <= set(pivot.columns):
            difference = (
                pivot['relevant_changed']
                - pivot['unchanged_control']
            ).dropna()
        else:
            difference = pd.Series(dtype=float)
        selectivity_rows.append({
            'population': population,
            'layer_index': int(layer_index),
            'scope': scope,
            'contrast': 'relevant_minus_unchanged_donor_separation',
            **paired_statistics(
                difference.to_numpy(),
                f'selectivity:{population}:L{layer_index}:{scope}',
            ),
        })
selectivity_frame = pd.DataFrame(selectivity_rows)
selectivity_frame['q_bh'] = np.nan
for population, indices in selectivity_frame.groupby('population').groups.items():
    selectivity_frame.loc[indices, 'q_bh'] = bh_adjust(
        selectivity_frame.loc[indices, 'p_value']
    )
selectivity_frame['passes_fdr'] = (
    (selectivity_frame['mean'] > 0)
    & (selectivity_frame['q_bh'] < ALPHA)
)
selectivity_frame.to_csv(
    ANALYSIS_DIR / 'within_pair_query_selectivity.csv', index=False
)

primary_specificity = specificity_frame[
    (specificity_frame['population'] == 'outcome_both_correct_cohort')
    & (specificity_frame['contrast'] == 'value_minus_outcome')
]
primary_selectivity = selectivity_frame[
    selectivity_frame['population'] == 'outcome_both_correct_cohort'
]
print('Primary physical-value specificity:')
display(primary_specificity)
print('Primary within-pair query selectivity:')
display(primary_selectivity)
```

```python
%%cell_progress 14/16 Audit competence, edit-family stability, and donor matching
cell_stage('Stratifying focal-site effects by success and edit family')
focal_rows = []
for layer_index, scope in FOCAL_SITES:
    site = recovery_pivot[
        (recovery_pivot['layer_index'] == layer_index)
        & (recovery_pivot['scope'] == scope)
    ]
    groups = [
        ('all_pairs', site),
        ('outcome_both_correct', site[
            site['competence_cohort'] == 'both_correct'
        ]),
        ('outcome_error', site[
            site['competence_cohort'] == 'error'
        ]),
        ('actual_query_both_correct', site[site['query_both_correct']]),
        ('actual_query_not_both_correct', site[~site['query_both_correct']]),
    ]
    groups.extend(
        (f'edit:{edit_type}', group)
        for edit_type, group in site.groupby('edit_type')
    )
    for group_name, group in groups:
        focal_rows.append({
            'layer_index': int(layer_index), 'scope': scope,
            'group': group_name,
            **paired_statistics(
                group['value_minus_outcome'].to_numpy(),
                f'focal:{layer_index}:{scope}:{group_name}',
            ),
        })
focal_stability = pd.DataFrame(focal_rows)
focal_stability.to_csv(
    ANALYSIS_DIR / 'focal_site_stability.csv', index=False
)

donor_balance_rows = []
for row in pair_manifest:
    donor_balance_rows.append({
        'pair_id': row['pair_id'],
        'edit_type': row['edit_type'],
        'competence_cohort': row['competence_cohort'],
        'value_donor_distance': row[
            'same_task_value_nuisance_distance'
        ],
        'outcome_donor_distance': row[
            'outcome_only_nuisance_distance'
        ],
    })
donor_balance = pd.DataFrame(donor_balance_rows)
donor_balance['distance_difference'] = (
    donor_balance['value_donor_distance']
    - donor_balance['outcome_donor_distance']
)
donor_balance.to_csv(
    ANALYSIS_DIR / 'donor_nuisance_balance.csv', index=False
)
print('Focal-site competence and edit stability:')
display(focal_stability)
print('Donor nuisance-distance balance:')
display(donor_balance.agg(
    n=('pair_id', 'nunique'),
    value_distance=('value_donor_distance', 'mean'),
    outcome_distance=('outcome_donor_distance', 'mean'),
    mean_difference=('distance_difference', 'mean'),
))
```

```python
%%cell_progress 16/16 Apply the frozen gate and write the mechanistic report
cell_stage('Evaluating focal sites without post-hoc site selection')

def get_site(frame, layer_index, scope):
    rows = frame[
        (frame['layer_index'] == layer_index)
        & (frame['scope'] == scope)
    ]
    assert len(rows) == 1, (layer_index, scope, len(rows))
    return rows.iloc[0]

successful_specificity = specificity_frame[
    (specificity_frame['population'] == 'actual_query_both_correct')
    & (specificity_frame['contrast'] == 'value_minus_outcome')
]
focal_decisions = []
for layer_index, scope in FOCAL_SITES:
    value_row = get_site(primary_specificity, layer_index, scope)
    selectivity_row = get_site(primary_selectivity, layer_index, scope)
    success_row = get_site(successful_specificity, layer_index, scope)
    passed = bool(
        value_row['passes_fdr']
        and selectivity_row['passes_fdr']
        and success_row['n'] >= 20
        and success_row['ci_low'] > 0
    )
    focal_decisions.append({
        'layer_index': int(layer_index), 'scope': scope,
        'value_specificity_mean': float(value_row['mean']),
        'value_specificity_q': float(value_row['q_bh']),
        'query_selectivity_mean': float(selectivity_row['mean']),
        'query_selectivity_q': float(selectivity_row['q_bh']),
        'successful_query_n': int(success_row['n']),
        'successful_query_mean': float(success_row['mean']),
        'successful_query_ci_low': float(success_row['ci_low']),
        'successful_query_ci_high': float(success_row['ci_high']),
        'passed_full_gate': passed,
    })
focal_decision_frame = pd.DataFrame(focal_decisions)
focal_decision_frame.to_csv(
    ANALYSIS_DIR / 'focal_site_decisions.csv', index=False
)

any_full_pass = bool(focal_decision_frame['passed_full_gate'].any())
any_value_pass = bool(primary_specificity[
    primary_specificity.apply(
        lambda row: (int(row['layer_index']), row['scope'])
        in set(FOCAL_SITES), axis=1
    )
]['passes_fdr'].any())
sufficient_success_n = bool(
    successful_specificity[
        successful_specificity.apply(
            lambda row: (int(row['layer_index']), row['scope'])
            in set(FOCAL_SITES), axis=1
        )
    ]['n'].max() >= 20
)

if any_full_pass:
    decision_code = 'gemma_cross_model_route_replication_supported'
    recommended_next = (
        'Run the frozen stratified Qwen-Gemma comparison. Do not localize heads '
        'until the cross-model effect sizes, competence strata, and negative '
        'controls have been compared.'
    )
elif not sufficient_success_n:
    decision_code = 'gemma_replication_underpowered_successful_query_subset'
    recommended_next = (
        'Report the prespecified power failure. Do not change sites or query '
        'definitions; enlarge only the untouched Gemma pair pool in a new '
        'versioned replication.'
    )
elif any_value_pass:
    decision_code = 'gemma_value_signal_without_query_selectivity'
    recommended_next = (
        'Treat the effect as a shared value/answer scaffold and compare it '
        'directly with Qwen; do not run a Gemma head search.'
    )
else:
    decision_code = 'gemma_cross_model_route_replication_not_supported'
    recommended_next = (
        'Freeze the cross-model dissociation: the Qwen grouped route does '
        'not transfer to Gemma under the prespecified mapping. Compare '
        'representations and behavioral competence before new localization.'
    )

decision = {
    'schema_version': 'gemma-competence-grouped-route-decision-3.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'pairs': len(pair_manifest),
    'patch_records': len(patch_frame),
    'focal_site_decisions': focal_decisions,
    'decision': decision_code,
    'recommended_next': recommended_next,
    'interpretation_limit': protocol_identity['interpretation_limit'],
}
write_json_atomic(ANALYSIS_DIR / 'decision_summary.json', decision)

report_lines = [
    '# Gemma cross-model grouped-route replication report', '',
    f'- Model: `{MODEL_ID}`',
    f'- Revision: `{MODEL_REVISION}`',
    f'- New pairs: {len(pair_manifest)} '
    f'({PAIRS_PER_EDIT_COHORT} per edit × competence cohort).',
    f'- Patch records: {len(patch_frame)}.',
    f'- Frozen sites: {list(FROZEN_SITES)}.',
    '', '## Actual custom-query competence', '',
]
competence_report = baseline_frame.groupby(
    ['competence_cohort', 'query_role']
).agg(
    n=('pair_id', 'nunique'),
    both_correct=('query_both_correct', 'mean'),
).reset_index()
for _, row in competence_report.iterrows():
    report_lines.append(
        f"- {row['competence_cohort']} / {row['query_role']}: "
        f"{int(row['n'])} pairs, task-both-correct "
        f"{row['both_correct']:.1%}."
    )
report_lines.extend([
    '', '## Primary physical-value specificity', '',
    'Positive values mean that a different-layout donor with the same task '
    'value outperforms a same-outcome donor with a different task value.',
    '',
])
for _, row in primary_specificity.iterrows():
    report_lines.append(
        f"- L{int(row['layer_index'])} `{row['scope']}`: "
        f"{row['mean']:.3f} [{row['ci_low']:.3f}, "
        f"{row['ci_high']:.3f}], q={row['q_bh']:.4f}, "
        f"n={int(row['n'])}, pass={bool(row['passes_fdr'])}."
    )
report_lines.extend([
    '', '## Within-pair prompt-gated donor separation', '',
    'Positive values mean that the same donor pair is more separable when '
    'its differing physical value is queried than when the donor-matched '
    'unchanged control value is queried.',
    '',
])
for _, row in primary_selectivity.iterrows():
    report_lines.append(
        f"- L{int(row['layer_index'])} `{row['scope']}`: "
        f"{row['mean']:.3f} [{row['ci_low']:.3f}, "
        f"{row['ci_high']:.3f}], q={row['q_bh']:.4f}, "
        f"n={int(row['n'])}, pass={bool(row['passes_fdr'])}."
    )
report_lines.extend([
    '', '## Frozen focal-site gate', '',
])
for row in focal_decisions:
    report_lines.append(
        f"- L{row['layer_index']} `{row['scope']}`: value q="
        f"{row['value_specificity_q']:.4f}; selectivity q="
        f"{row['query_selectivity_q']:.4f}; successful-query "
        f"effect {row['successful_query_mean']:.3f} "
        f"[{row['successful_query_ci_low']:.3f}, "
        f"{row['successful_query_ci_high']:.3f}], "
        f"n={row['successful_query_n']}; full pass="
        f"{row['passed_full_gate']}."
    )
report_lines.extend([
    '', '## Decision', '',
    f'- **`{decision_code}`**',
    f'- Next: {recommended_next}',
    '', '## Limits', '',
    protocol_identity['interpretation_limit'], '',
])
report_path = ANALYSIS_DIR / 'grouped_route_confirmatory_report.md'
report_path.write_text('\n'.join(report_lines) + '\n', encoding='utf-8')

expected_outputs = (
    'pair_pool_availability.csv', 'pair_manifest.csv',
    'patch_records.csv', 'query_baselines.csv',
    'pair_level_donor_gaps.csv', 'pair_level_donor_separation.csv',
    'donor_value_specificity.csv',
    'within_pair_query_selectivity.csv', 'focal_site_stability.csv',
    'donor_nuisance_balance.csv', 'focal_site_decisions.csv',
    'decision_summary.json', 'grouped_route_confirmatory_report.md',
)
for name in expected_outputs:
    assert (ANALYSIS_DIR / name).is_file(), name
manifest = {
    'schema_version': 'competence-grouped-route-artifacts-1.0',
    'protocol_sha256': sha256_file(PROTOCOL_LOCK_PATH),
    'capture_config_sha256': sha256_file(capture_config_path),
    'pair_manifest_sha256': sha256_file(manifest_path),
    'outputs': {
        name: sha256_file(ANALYSIS_DIR / name)
        for name in expected_outputs
    },
    'figures': {
        path.name: sha256_file(path)
        for path in sorted(FIGURE_DIR.glob('*.png'))
    },
    'decision': decision,
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
}
write_json_atomic(ANALYSIS_DIR / 'run_manifest.json', manifest)
print(report_path.read_text(encoding='utf-8'))
print('Decision:', decision_code)
print('Completed Gemma grouped-route cross-model replication.')
```
