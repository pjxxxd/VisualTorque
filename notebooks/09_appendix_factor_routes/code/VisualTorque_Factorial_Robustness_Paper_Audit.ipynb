```python
from pathlib import Path
import os
import sys

from tqdm.auto import tqdm
from IPython.display import display

TOTAL_STAGES = 12
_progress = tqdm(total=TOTAL_STAGES, desc='Factorial robustness audit', unit='stage')

def stage(label):
    if _progress.n < TOTAL_STAGES:
        _progress.update(1)
    _progress.set_postfix_str(label)
    print(f'[{_progress.n}/{TOTAL_STAGES}] {label}')
```

## 1. Locate and freeze the post-confirmatory audit


```python
stage('Locating completed same-manifest artifacts')

IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import drive
    drive.mount('/content/drive')

def find_project_root():
    configured = os.environ.get('VISUALTORQUE_ROOT')
    candidates = []
    if configured:
        candidates.append(Path(configured).expanduser())
    if IN_COLAB:
        candidates.extend((
            Path('/content/drive/MyDrive/VisualTorque'),
            Path('/content/drive/MyDrive/VisualTorque/VisualTorque'),
        ))
    for base in (Path.cwd(), *Path.cwd().parents):
        candidates.extend((base, base / 'VisualTorque'))
    checked = set()
    for candidate in candidates:
        candidate = candidate.resolve()
        if candidate in checked:
            continue
        checked.add(candidate)
        marker = (
            candidate / 'results'
            / 'same_manifest_factorial_composition_v1'
            / 'factorial_triplet_manifest.jsonl'
        )
        if marker.is_file():
            return candidate
    raise FileNotFoundError(
        'Could not locate completed factorial results. Mount Drive, start '
        'Jupyter inside the repository, or set VISUALTORQUE_ROOT.'
    )

PROJECT_ROOT = find_project_root()
FACTORIAL_ROOT = (
    PROJECT_ROOT / 'results' / 'same_manifest_factorial_composition_v1'
)
MODEL_SPECS = {
    'qwen': 'Qwen__Qwen3.5-9B',
    'gemma': 'google__gemma-4-12B-it',
}
MODEL_ANALYSIS = {
    key: FACTORIAL_ROOT / 'models' / slug / 'analysis'
    for key, slug in MODEL_SPECS.items()
}
OUTPUT_DIR = FACTORIAL_ROOT / 'robustness_paper_audit_v1'
FIGURE_DIR = OUTPUT_DIR / 'figures'
TABLE_DIR = OUTPUT_DIR / 'tables'
for path in (OUTPUT_DIR, FIGURE_DIR, TABLE_DIR):
    path.mkdir(parents=True, exist_ok=True)
print('Project:', PROJECT_ROOT)
print('Factorial results:', FACTORIAL_ROOT)
print('Audit output:', OUTPUT_DIR)
```

```python
stage('Loading and hashing frozen source artifacts')
import hashlib
import json
import math
from datetime import datetime, timezone

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

SOURCE_PATHS = {
    'manifest': FACTORIAL_ROOT / 'factorial_triplet_manifest.jsonl',
    'manifest_lock': FACTORIAL_ROOT / 'manifest_lock.json',
}
for model, analysis_dir in MODEL_ANALYSIS.items():
    SOURCE_PATHS[f'{model}_paired'] = analysis_dir / 'paired_triplet_metrics.csv'
    SOURCE_PATHS[f'{model}_patch'] = analysis_dir / 'patch_level_metrics.csv'
    SOURCE_PATHS[f'{model}_decision'] = (
        analysis_dir / 'factorial_composition_decision.json'
    )
missing = [str(path) for path in SOURCE_PATHS.values() if not path.is_file()]
assert not missing, 'Missing required artifacts:\n' + '\n'.join(missing)

from visualtorque_checks import sha256_file

def write_json_atomic(path, value):
    path = Path(path)
    temporary = path.with_suffix(path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def as_bool(series):
    if series.dtype == bool:
        return series.astype(bool)
    mapped = series.astype(str).str.lower().map({
        'true': True, 'false': False, '1': True, '0': False,
    })
    assert mapped.notna().all(), sorted(series.astype(str).unique())
    return mapped.astype(bool)

SEED = 20260831
BOOTSTRAP_REPLICATES = 10000
SIGN_FLIP_REPLICATES = 20000
ALPHA = 0.05
EQUIVALENCE_FRACTION = 0.25
FOCAL_SITES = ('middle_text', 'middle_options')
PROTOCOL_PATH = OUTPUT_DIR / 'audit_protocol.json'
source_hashes = {
    key: sha256_file(path) for key, path in SOURCE_PATHS.items()
}
protocol_identity = {
    'schema_version': 'factorial-robustness-paper-audit-1.0',
    'status': (
        'post-confirmatory robustness audit; contrast selection was '
        'informed by the completed factorial experiment and is not an '
        'independent preregistered confirmation'
    ),
    'source_hashes': source_hashes,
    'primary_population': 'all 48 exact same-manifest triplets',
    'robustness_population': '27 triplets with prior_pair_overlap false',
    'overlap_population': '21 triplets with prior_pair_overlap true',
    'primary_within_model_families': {
        'factor_binding': [
            'side_units factor_selectivity at two focal sites',
            'side_distance factor_selectivity at two focal sites',
        ],
        'side_torque_transport': [
            'weight effect, distance effect, and donor difference at two focal sites',
        ],
        'downstream_boundary': [
            'net_torque mean donor effect at two focal sites',
            'outcome mean donor effect at late_answer',
        ],
    },
    'primary_cross_model_contrasts': [
        'side_units factor_selectivity Qwen minus Gemma at two focal sites',
        'side_distance factor_selectivity Qwen minus Gemma at two focal sites',
        'side_torque donor difference Qwen minus Gemma at two focal sites',
        'side_torque mean donor effect Qwen minus Gemma at two focal sites',
    ],
    'multiplicity': (
        'BH within each within-model family and across the eight raw '
        'primary cross-model contrasts'
    ),
    'equivalence_bound': (
        f'{EQUIVALENCE_FRACTION:.2f} times the model/task median '
        'absolute natural margin'
    ),
    'bootstrap_unit': 'factorial triplet',
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'sign_flip_replicates': SIGN_FLIP_REPLICATES,
    'interpretation_limit': (
        'This audit tests robustness and paper claims. It cannot turn '
        'post-observation contrasts into independent confirmation, and '
        'it does not attribute differences to architecture alone.'
    ),
    'seed': SEED,
}
if PROTOCOL_PATH.is_file():
    previous = json.loads(PROTOCOL_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing audit uses different sources or rules. Use a new output version.'
    )
else:
    write_json_atomic(PROTOCOL_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Protocol:', PROTOCOL_PATH)
```

## 2. Exact support and derived route metrics


```python
stage('Validating exact support and constructing common metrics')
manifest_rows = read_jsonl(SOURCE_PATHS['manifest'])
manifest = pd.DataFrame([
    {key: value for key, value in row.items() if key != 'task_values'}
    for row in manifest_rows
])
manifest['triplet_id'] = manifest['triplet_id'].astype(str)
manifest['prior_pair_overlap'] = as_bool(manifest['prior_pair_overlap'])
manifest['outcome_changed'] = as_bool(manifest['outcome_changed'])
assert len(manifest) == manifest['triplet_id'].nunique() == 48
assert (~manifest['prior_pair_overlap']).sum() == 27
assert manifest['prior_pair_overlap'].sum() == 21

paired_frames = []
patch_frames = []
decisions = {}
for model in MODEL_SPECS:
    paired = pd.read_csv(SOURCE_PATHS[f'{model}_paired'])
    patch = pd.read_csv(SOURCE_PATHS[f'{model}_patch'])
    paired['model'] = model
    patch['model'] = model
    for frame in (paired, patch):
        frame['triplet_id'] = frame['triplet_id'].astype(str)
        for column in (
            'estimable', 'outcome_changed', 'prior_pair_overlap',
            'all_three_correct',
        ):
            if column in frame:
                frame[column] = as_bool(frame[column])
    assert set(paired['triplet_id']) == set(manifest['triplet_id'])
    assert set(patch['triplet_id']) == set(manifest['triplet_id'])
    assert len(paired) == 960 and len(patch) == 1920
    assert not paired.duplicated(['triplet_id', 'task', 'site_label']).any()
    assert not patch.duplicated([
        'triplet_id', 'task', 'site_label', 'donor_type'
    ]).any()
    decisions[model] = json.loads(
        SOURCE_PATHS[f'{model}_decision'].read_text(encoding='utf-8')
    )
    paired_frames.append(paired)
    patch_frames.append(patch)

paired = pd.concat(paired_frames, ignore_index=True)
patch = pd.concat(patch_frames, ignore_index=True)
assert paired.groupby(['model', 'task', 'site_label']).size().eq(48).all()
assert patch.groupby([
    'model', 'task', 'site_label', 'donor_type'
]).size().eq(48).all()
for model, decision in decisions.items():
    assert decision['manifest_sha256'] == source_hashes['manifest']
    assert decision['decision'] == (
        'factorial_composition_chain_not_supported_at_frozen_sites'
    )

scale = (
    paired[paired['estimable']]
    .groupby(['model', 'task'])['natural_margin']
    .apply(lambda values: float(np.median(np.abs(values))))
    .rename('task_natural_scale')
    .reset_index()
)
assert (scale['task_natural_scale'] > 0).all()
paired = paired.merge(scale, on=['model', 'task'], validate='many_to_one')
for column in (
    'weight_effect', 'distance_effect', 'mean_donor_effect',
    'donor_effect_difference', 'factor_selectivity', 'abs_donor_gap',
):
    paired[f'std_{column}'] = (
        paired[column] / paired['task_natural_scale']
    )
stabilizer = EQUIVALENCE_FRACTION * paired['task_natural_scale']
paired['donor_dominance_index'] = (
    paired['donor_effect_difference']
    / (
        paired['weight_effect'].abs()
        + paired['distance_effect'].abs()
        + stabilizer
    )
)
paired.to_csv(TABLE_DIR / 'exact_paired_derived_metrics.csv', index=False)
print('Exact support: 48 triplets × 5 tasks × 4 sites × 2 models')
display(manifest.groupby([
    'prior_pair_overlap', 'changed_side', 'outcome_changed'
]).size().rename('n').reset_index())
display(scale)
```

```python
stage('Defining deterministic paired inference')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def paired_summary(values, label):
    values = pd.to_numeric(pd.Series(values), errors='coerce').to_numpy(float)
    values = values[np.isfinite(values)]
    if len(values) == 0:
        return {
            'n': 0, 'mean': np.nan, 'ci_low': np.nan,
            'ci_high': np.nan, 'sign_flip_p': np.nan,
        }
    rng = np.random.default_rng(seed_for(label))
    indices = rng.integers(
        0, len(values), size=(BOOTSTRAP_REPLICATES, len(values))
    )
    bootstrap = values[indices].mean(axis=1)
    observed = float(values.mean())
    signs = rng.choice(
        (-1.0, 1.0),
        size=(SIGN_FLIP_REPLICATES, len(values)),
    )
    null = (signs * values).mean(axis=1)
    p_value = float(
        (1 + np.sum(np.abs(null) >= abs(observed)))
        / (SIGN_FLIP_REPLICATES + 1)
    )
    return {
        'n': int(len(values)),
        'mean': observed,
        'ci_low': float(np.quantile(bootstrap, ALPHA / 2)),
        'ci_high': float(np.quantile(bootstrap, 1 - ALPHA / 2)),
        'sign_flip_p': p_value,
    }

def independent_difference(left, right, label):
    left = pd.to_numeric(pd.Series(left), errors='coerce').dropna().to_numpy(float)
    right = pd.to_numeric(pd.Series(right), errors='coerce').dropna().to_numpy(float)
    if not len(left) or not len(right):
        return {'n_left': len(left), 'n_right': len(right), 'mean': np.nan,
                'ci_low': np.nan, 'ci_high': np.nan}
    rng = np.random.default_rng(seed_for(label))
    left_index = rng.integers(
        0, len(left), size=(BOOTSTRAP_REPLICATES, len(left))
    )
    right_index = rng.integers(
        0, len(right), size=(BOOTSTRAP_REPLICATES, len(right))
    )
    boot = left[left_index].mean(axis=1) - right[right_index].mean(axis=1)
    return {
        'n_left': int(len(left)), 'n_right': int(len(right)),
        'mean': float(left.mean() - right.mean()),
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    result = np.full(len(values), np.nan)
    valid = np.where(np.isfinite(values))[0]
    if len(valid) == 0:
        return result
    order = valid[np.argsort(values[valid])]
    ranked = values[order] * len(order) / np.arange(1, len(order) + 1)
    ranked = np.minimum.accumulate(ranked[::-1])[::-1]
    result[order] = np.minimum(ranked, 1.0)
    return result

def population_frame(frame, population):
    if population == 'all':
        return frame
    if population == 'unseen':
        return frame[~frame['prior_pair_overlap']]
    if population == 'overlap':
        return frame[frame['prior_pair_overlap']]
    raise KeyError(population)

print('Inference utilities ready.')
```

## 3. Within-model and exact cross-model claims


```python
stage('Testing within-model factor binding and torque transport')
within_plan = []
for model in MODEL_SPECS:
    for site in FOCAL_SITES:
        within_plan.extend([
            (model, 'factor_binding', 'side_units', site, 'factor_selectivity'),
            (model, 'factor_binding', 'side_distance', site, 'factor_selectivity'),
            (model, 'side_torque_transport', 'side_torque', site, 'weight_effect'),
            (model, 'side_torque_transport', 'side_torque', site, 'distance_effect'),
            (model, 'side_torque_transport', 'side_torque', site, 'donor_effect_difference'),
            (model, 'downstream_boundary', 'net_torque', site, 'mean_donor_effect'),
        ])
    within_plan.append((
        model, 'downstream_boundary', 'outcome',
        'late_answer', 'mean_donor_effect',
    ))

within_rows = []
for population in ('all', 'unseen', 'overlap'):
    current = population_frame(paired, population)
    for model, family, task, site, metric in within_plan:
        subset = current[
            (current['model'] == model)
            & (current['task'] == task)
            & (current['site_label'] == site)
            & current['estimable']
        ]
        result = paired_summary(
            subset[metric],
            f'within:{population}:{model}:{task}:{site}:{metric}',
        )
        within_rows.append({
            'population': population, 'model': model,
            'family': family, 'task': task, 'site_label': site,
            'metric': metric, **result,
        })
within_summary = pd.DataFrame(within_rows)
within_summary['q_value'] = np.nan
all_mask = within_summary['population'] == 'all'
for keys, indices in within_summary[all_mask].groupby(
    'family'
).groups.items():
    within_summary.loc[indices, 'q_value'] = bh_adjust(
        within_summary.loc[indices, 'sign_flip_p']
    )

within_summary['equivalence_bound'] = np.nan
within_summary['equivalent'] = False
for population in ('all', 'unseen', 'overlap'):
    current = population_frame(paired, population)
    for model in MODEL_SPECS:
        for task in ('side_torque', 'net_torque', 'outcome'):
            for site in (
                FOCAL_SITES if task != 'outcome' else ('late_answer',)
            ):
                subset = current[
                    (current['model'] == model)
                    & (current['task'] == task)
                    & (current['site_label'] == site)
                    & current['estimable']
                ]
                bound = EQUIVALENCE_FRACTION * float(
                    subset['task_natural_scale'].iloc[0]
                ) if len(subset) else np.nan
                difference = paired_summary(
                    subset['donor_effect_difference'],
                    f'equivalence:{population}:{model}:{task}:{site}',
                )
                mask = (
                    (within_summary['population'] == population)
                    & (within_summary['model'] == model)
                    & (within_summary['task'] == task)
                    & (within_summary['site_label'] == site)
                    & (within_summary['metric'] == 'donor_effect_difference')
                )
                if mask.any():
                    within_summary.loc[mask, 'equivalence_bound'] = bound
                    within_summary.loc[mask, 'equivalent'] = (
                        difference['ci_low'] > -bound
                        and difference['ci_high'] < bound
                    )
within_summary.to_csv(TABLE_DIR / 'within_model_audit.csv', index=False)
display(within_summary[
    (within_summary['population'].isin(('all', 'unseen')))
    & (within_summary['family'].isin((
        'factor_binding', 'side_torque_transport'
    )))
].sort_values(['population', 'model', 'site_label', 'task', 'metric']))
```

```python
stage('Auditing side and outcome-change strata')
stratum_rows = []
stratum_specs = (
    ('changed_side', ('left', 'right')),
    ('outcome_changed', (False, True)),
)
metric_specs = (
    ('qwen', 'side_units', 'factor_selectivity'),
    ('qwen', 'side_torque', 'donor_effect_difference'),
    ('gemma', 'side_units', 'factor_selectivity'),
    ('gemma', 'side_distance', 'factor_selectivity'),
    ('gemma', 'side_torque', 'donor_effect_difference'),
)
for model, task, metric in metric_specs:
    for site in FOCAL_SITES:
        base = paired[
            (paired['model'] == model)
            & (paired['task'] == task)
            & (paired['site_label'] == site)
            & paired['estimable']
        ]
        for column, levels in stratum_specs:
            level_values = {}
            for level in levels:
                subset = base[base[column] == level]
                result = paired_summary(
                    subset[metric],
                    f'stratum:{model}:{task}:{site}:{metric}:{column}:{level}',
                )
                level_values[level] = subset[metric]
                stratum_rows.append({
                    'model': model, 'task': task, 'site_label': site,
                    'metric': metric, 'stratum': column,
                    'level': str(level), 'row_type': 'level', **result,
                })
            difference = independent_difference(
                level_values[levels[0]], level_values[levels[1]],
                f'heterogeneity:{model}:{task}:{site}:{metric}:{column}',
            )
            stratum_rows.append({
                'model': model, 'task': task, 'site_label': site,
                'metric': metric, 'stratum': column,
                'level': f'{levels[0]}_minus_{levels[1]}',
                'row_type': 'heterogeneity',
                'n': difference['n_left'] + difference['n_right'],
                'mean': difference['mean'],
                'ci_low': difference['ci_low'],
                'ci_high': difference['ci_high'],
                'sign_flip_p': np.nan,
            })
stratum_audit = pd.DataFrame(stratum_rows)
stratum_audit.to_csv(TABLE_DIR / 'physical_strata.csv', index=False)
display(stratum_audit[
    stratum_audit['row_type'] == 'heterogeneity'
].sort_values(['model', 'task', 'site_label', 'stratum']))
```

```python
stage('Auditing paired baseline competence')
baseline = patch[[
    'model', 'triplet_id', 'task', 'base_correct',
    'weight_source_correct', 'distance_source_correct',
    'prior_pair_overlap',
]].drop_duplicates(['model', 'triplet_id', 'task'])
assert len(baseline) == 2 * 48 * 5
for column in (
    'base_correct', 'weight_source_correct',
    'distance_source_correct', 'prior_pair_overlap',
):
    baseline[column] = as_bool(baseline[column])

accuracy_rows = []
for population in ('all', 'unseen'):
    current = population_frame(baseline, population)
    for model in MODEL_SPECS:
        for task in sorted(current['task'].unique()):
            subset = current[
                (current['model'] == model) & (current['task'] == task)
            ]
            for role, column in (
                ('base', 'base_correct'),
                ('weight_source', 'weight_source_correct'),
                ('distance_source', 'distance_source_correct'),
            ):
                accuracy_rows.append({
                    'population': population, 'model': model,
                    'task': task, 'role': role, 'n': len(subset),
                    'accuracy': float(subset[column].mean()),
                })
accuracy = pd.DataFrame(accuracy_rows)

def exact_mcnemar(left, right):
    left = np.asarray(left, dtype=bool)
    right = np.asarray(right, dtype=bool)
    left_only = int(np.sum(left & ~right))
    right_only = int(np.sum(~left & right))
    discordant = left_only + right_only
    if discordant == 0:
        return left_only, right_only, 1.0
    tail = sum(
        math.comb(discordant, value)
        for value in range(0, min(left_only, right_only) + 1)
    ) / (2 ** discordant)
    return left_only, right_only, min(1.0, 2 * tail)

mcnemar_rows = []
for population in ('all', 'unseen'):
    current = population_frame(baseline, population)
    for model in MODEL_SPECS:
        for task in sorted(current['task'].unique()):
            subset = current[
                (current['model'] == model) & (current['task'] == task)
            ]
            left_only, right_only, p_value = exact_mcnemar(
                subset['weight_source_correct'],
                subset['distance_source_correct'],
            )
            mcnemar_rows.append({
                'population': population, 'model': model,
                'task': task, 'weight_only_correct': left_only,
                'distance_only_correct': right_only,
                'mcnemar_exact_p': p_value,
            })
mcnemar = pd.DataFrame(mcnemar_rows)
accuracy.to_csv(TABLE_DIR / 'baseline_accuracy.csv', index=False)
mcnemar.to_csv(TABLE_DIR / 'baseline_weight_distance_mcnemar.csv', index=False)
display(accuracy[accuracy['population'] == 'all'].pivot_table(
    index=['model', 'task'], columns='role', values='accuracy'
).reset_index())
display(mcnemar)
```

```python
stage('Recomputing clearly labeled convergence metrics')
convergence_rows = []
for population in ('all', 'unseen'):
    current = population_frame(paired, population)
    for model in MODEL_SPECS:
        for task in ('side_torque', 'net_torque'):
            subset = current[
                (current['model'] == model)
                & (current['task'] == task)
                & current['estimable']
            ]
            for destination in ('middle_text', 'middle_options'):
                cosine = subset.pivot(
                    index='triplet_id', columns='site_label',
                    values='displacement_cosine',
                ).dropna(subset=['early_image', destination])
                cosine_gain = cosine[destination] - cosine['early_image']
                convergence_rows.append({
                    'population': population, 'model': model,
                    'task': task, 'destination': destination,
                    'metric': 'cosine_gain',
                    **paired_summary(
                        cosine_gain,
                        f'convergence:cosine:{population}:{model}:{task}:{destination}',
                    ),
                })
                gap = subset.pivot(
                    index='triplet_id', columns='site_label',
                    values='abs_donor_gap',
                ).dropna(subset=['early_image', destination])
                gap_reduction = gap['early_image'] - gap[destination]
                convergence_rows.append({
                    'population': population, 'model': model,
                    'task': task, 'destination': destination,
                    'metric': 'gap_reduction',
                    **paired_summary(
                        gap_reduction,
                        f'convergence:gap:{population}:{model}:{task}:{destination}',
                    ),
                })
convergence = pd.DataFrame(convergence_rows)
convergence.to_csv(TABLE_DIR / 'labeled_convergence.csv', index=False)
display(convergence.sort_values([
    'population', 'model', 'task', 'destination', 'metric'
]))
```

```python
stage('Applying conservative claim gates and finalizing artifacts')

def within_record(population, model, task, site, metric):
    rows = within_summary[
        (within_summary['population'] == population)
        & (within_summary['model'] == model)
        & (within_summary['task'] == task)
        & (within_summary['site_label'] == site)
        & (within_summary['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def cross_record(population, task, site, metric):
    rows = cross_summary[
        (cross_summary['population'] == population)
        & (cross_summary['task'] == task)
        & (cross_summary['site_label'] == site)
        & (cross_summary['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

shared_unit_route = all(
    within_record(population, model, 'side_units', site,
                  'factor_selectivity')['ci_low'] > 0
    for population in ('all', 'unseen')
    for model in MODEL_SPECS
    for site in FOCAL_SITES
) and all(
    within_record('all', model, 'side_units', site,
                  'factor_selectivity')['q_value'] < ALPHA
    for model in MODEL_SPECS
    for site in FOCAL_SITES
)
gemma_distance_misbinding = all(
    within_record(population, 'gemma', 'side_distance', site,
                  'factor_selectivity')['ci_high'] < 0
    for population in ('all', 'unseen')
    for site in FOCAL_SITES
) and all(
    within_record('all', 'gemma', 'side_distance', site,
                  'factor_selectivity')['q_value'] < ALPHA
    for site in FOCAL_SITES
)
qwen_weight_dominant_torque = all(
    within_record(population, 'qwen', 'side_torque', site,
                  'donor_effect_difference')['ci_low'] > 0
    for population in ('all', 'unseen')
    for site in FOCAL_SITES
) and all(
    within_record('all', 'qwen', 'side_torque', site,
                  'donor_effect_difference')['q_value'] < ALPHA
    for site in FOCAL_SITES
)
exact_cross_model_route_difference = all(
    cross_record(population, 'side_torque', site,
                 'donor_effect_difference')['ci_low'] > 0
    for population in ('all', 'unseen')
    for site in FOCAL_SITES
) and all(
    cross_record('all', 'side_torque', site,
                 'donor_effect_difference')['q_value'] < ALPHA
    for site in FOCAL_SITES
)
exact_cross_distance_binding_difference = all(
    cross_record(population, 'side_distance', site,
                 'factor_selectivity')['ci_low'] > 0
    for population in ('all', 'unseen')
    for site in FOCAL_SITES
) and all(
    cross_record('all', 'side_distance', site,
                 'factor_selectivity')['q_value'] < ALPHA
    for site in FOCAL_SITES
)
gemma_partial_convergence_checks = []
for population in ('all', 'unseen'):
    weight = within_record(
        population, 'gemma', 'side_torque',
        'middle_options', 'weight_effect',
    )
    distance = within_record(
        population, 'gemma', 'side_torque',
        'middle_options', 'distance_effect',
    )
    difference = within_record(
        population, 'gemma', 'side_torque',
        'middle_options', 'donor_effect_difference',
    )
    gemma_partial_convergence_checks.append(bool(
        weight['ci_low'] > 0
        and distance['ci_low'] > 0
        and bool(difference['equivalent'])
        and (
            population != 'all'
            or (
                weight['q_value'] < ALPHA
                and distance['q_value'] < ALPHA
            )
        )
    ))
gemma_partial_side_torque_convergence = all(
    gemma_partial_convergence_checks
)

full_chain_not_supported = all(
    not decisions[model]['full_composition_chain_supported']
    for model in MODEL_SPECS
)
if (
    shared_unit_route
    and qwen_weight_dominant_torque
    and exact_cross_model_route_difference
):
    decision = 'robust_model_specific_factorization_routes_full_chain_absent'
else:
    decision = 'factorial_route_story_requires_additional_confirmation'

claim_card = {
    'schema_version': 'factorial-robustness-claim-card-1.0',
    'decision': decision,
    'claims': {
        'shared_unit_route_robust_on_unseen_triplets': shared_unit_route,
        'gemma_distance_misbinding_robust_on_unseen_triplets': (
            gemma_distance_misbinding
        ),
        'qwen_weight_dominant_side_torque_robust': (
            qwen_weight_dominant_torque
        ),
        'gemma_partial_side_torque_convergence_robust': (
            gemma_partial_side_torque_convergence
        ),
        'exact_cross_model_route_difference_robust': (
            exact_cross_model_route_difference
        ),
        'exact_cross_model_distance_binding_difference_robust': (
            exact_cross_distance_binding_difference
        ),
        'full_factor_to_torque_to_outcome_chain_not_supported': (
            full_chain_not_supported
        ),
    },
    'sample': {
        'all_triplets': 48,
        'unseen_triplets': 27,
        'prior_overlap_triplets': 21,
    },
    'scope': (
        'Frozen prompts, grouped decoder-output sites, and the exact '
        'same-manifest factorial intervention.'
    ),
    'paper_language': (
        'Both models transport unit information, but they do not share '
        'a complete factorization-invariant physical computation. Qwen '
        'is strongly weight-dominant; Gemma is less asymmetric at the '
        'side-torque stage but shows unstable distance binding. Neither '
        'model supports the full net-torque-to-outcome causal chain.'
    ),
    'nonclaims': [
        'Architecture alone caused the model difference.',
        'Either model lacks all distance or torque information.',
        'The intervention identifies literal multiplication or a unique circuit.',
        'The post-confirmatory robustness audit is an independent preregistration.',
    ],
    'source_hashes': source_hashes,
}
write_json_atomic(OUTPUT_DIR / 'claim_card.json', claim_card)

report = [
    '# Factorial robustness and paper audit', '',
    f"Decision: `{decision}`", '',
    '## Claim card', '',
]
for claim, value in claim_card['claims'].items():
    report.append(f'- {claim}: **{value}**')
report.extend([
    '', '## Conservative interpretation', '',
    claim_card['paper_language'], '',
    'The all-triplet analysis is the original support. The 27-triplet '
    'non-overlap subset is a robustness check, not an independently '
    'selected confirmatory experiment.', '',
    '## Nonclaims', '',
])
report.extend(f'- {value}' for value in claim_card['nonclaims'])
report_path = OUTPUT_DIR / 'factorial_robustness_report.md'
report_path.write_text('\n'.join(report) + '\n', encoding='utf-8')

artifact_paths = sorted(
    [path for path in OUTPUT_DIR.rglob('*') if path.is_file()]
)
artifact_manifest = {
    str(path.relative_to(OUTPUT_DIR)): sha256_file(path)
    for path in artifact_paths
    if path.name != 'artifact_manifest.json'
}
write_json_atomic(OUTPUT_DIR / 'artifact_manifest.json', artifact_manifest)
write_json_atomic(OUTPUT_DIR / 'audit_complete.json', {
    'decision': decision,
    'artifact_count': len(artifact_manifest),
    'completed_at_utc': datetime.now(timezone.utc).isoformat(),
})
_progress.n = TOTAL_STAGES
_progress.set_postfix_str('complete')
_progress.close()
print(json.dumps(claim_card, indent=2))
print('Report:', report_path)
print('Tables:', TABLE_DIR)
print('Figures:', FIGURE_DIR)
```
