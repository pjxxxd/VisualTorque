```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 5/17 Resolve every registered source artifact
cell_stage('Resolving exact paths with unique recursive fallback')

from visualtorque_checks import sha256_file

def resolve_artifact(specification):
    exact = RESULTS_ROOT / specification['relative']
    if exact.is_file():
        return exact, 'exact', []
    tokens = tuple(str(value).lower() for value in specification['tokens'])
    candidates = [
        path for path in RESULTS_ROOT.rglob(specification['filename'])
        if all(token in str(path).lower() for token in tokens)
    ]
    candidates = sorted(set(candidates))
    if len(candidates) == 1:
        return candidates[0], 'fallback_unique', []
    status = 'missing' if not candidates else 'ambiguous'
    return None, status, [str(path) for path in candidates]

resolved_sources = {}
inventory_rows = []
for specification in SOURCE_REGISTRY:
    path, resolution, candidates = resolve_artifact(specification)
    evidence_id = specification['evidence_id']
    if path is not None:
        resolved_sources[evidence_id] = path
    inventory_rows.append({
        **specification,
        'resolved_path': str(path) if path is not None else '',
        'resolution': resolution,
        'candidate_count': len(candidates),
        'candidates': ' | '.join(candidates),
        'sha256': sha256_file(path) if path is not None else '',
    })

artifact_inventory = pd.DataFrame(inventory_rows)
artifact_inventory.to_csv(TABLE_DIR / 'source_artifact_inventory.csv', index=False)
display(artifact_inventory[[
    'evidence_id', 'stage', 'tier', 'paper_role', 'resolution'
]])
print('Resolved:', len(resolved_sources), '/', len(SOURCE_REGISTRY))
```

## 2. Source decisions and computation-chain evidence


```python
%%cell_progress 6/17 Parse and normalize the frozen decisions
cell_stage('Reading heterogeneous decision and claim-card schemas')

decision_payloads = {}
decision_rows = []
for specification in SOURCE_REGISTRY:
    evidence_id = specification['evidence_id']
    path = resolved_sources.get(evidence_id)
    if path is None:
        actual = None
        payload = None
    else:
        payload = json.loads(path.read_text(encoding='utf-8'))
        actual = payload.get('decision')
        assert actual is None or isinstance(actual, str), evidence_id
        decision_payloads[evidence_id] = payload
    expected = specification['expected_decision']
    if path is None:
        audit_status = 'missing_source'
    elif actual == expected:
        audit_status = 'matched'
    elif actual is None:
        audit_status = 'missing_top_level_decision'
    else:
        audit_status = 'unexpected_decision'
    decision_rows.append({
        'evidence_id': evidence_id,
        'stage': specification['stage'],
        'tier': specification['tier'],
        'paper_role': specification['paper_role'],
        'required': specification['required'],
        'expected_decision': expected,
        'actual_decision': actual or '',
        'audit_status': audit_status,
        'source_path': str(path) if path is not None else '',
    })

decision_audit = pd.DataFrame(decision_rows)
decision_audit.to_csv(TABLE_DIR / 'decision_audit.csv', index=False)
display(decision_audit)
```

```python
%%cell_progress 7/17 Apply source completeness and reconciliation gates
cell_stage('Checking missing, ambiguous, and unexpected evidence')

required_rows = decision_audit[decision_audit['required'].astype(bool)]
missing_required = required_rows[
    required_rows['audit_status'] == 'missing_source'
]['evidence_id'].tolist()
unreconciled_required = required_rows[
    required_rows['audit_status'].isin([
        'missing_top_level_decision', 'unexpected_decision',
    ])
]['evidence_id'].tolist()
all_required_resolved = not missing_required
all_required_reconciled = not unreconciled_required

assert decision_audit['evidence_id'].is_unique
assert set(decision_audit['evidence_id']) == set(EXPECTED_DECISIONS)
assert artifact_inventory['evidence_id'].is_unique
print('All required sources resolved:', all_required_resolved)
print('All required decisions reconciled:', all_required_reconciled)
if missing_required:
    print('Missing:', missing_required)
if unreconciled_required:
    print('Needs manual reconciliation:', unreconciled_required)
```
