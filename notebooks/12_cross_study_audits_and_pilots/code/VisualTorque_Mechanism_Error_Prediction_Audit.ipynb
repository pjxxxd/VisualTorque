```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 5/17 Load and validate the frozen phase-map outputs
cell_stage('Reading manifest, baselines, causal metrics, and saved scores')

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

manifest_rows = read_jsonl(MANIFEST_PATH)
manifest_frame = pd.DataFrame(manifest_rows)
baseline_frame = pd.read_csv(BASELINE_PATH)
route_frame = pd.read_csv(ROUTE_PATH)
phase_decision = json.loads(PHASE_DECISION_PATH.read_text(encoding='utf-8'))

assert len(manifest_frame) == 80
assert len(baseline_frame) == 320
assert len(route_frame) == 960
assert phase_decision['families'] == 80
assert set(PRIMARY_TASKS).issubset(set(baseline_frame['task']))
assert set(FROZEN_SITES) == set(route_frame['site_label'])

score_rows = []
for path in tqdm(
    sorted(PHASE_SHARD_DIR.glob('shard_family_*.json')),
    desc='Recovering non-gold clean confidence', unit='family',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    family_id = str(payload['family_id'])
    for task in PRIMARY_TASKS:
        score = payload['baselines'][task]['scores']['base']
        probabilities = np.asarray(score['candidate_probabilities'], dtype=float)
        assert len(probabilities) == 3 and np.isclose(probabilities.sum(), 1.0)
        ordered = np.sort(probabilities)
        top1_gap = float(ordered[-1] - ordered[-2])
        entropy = float(
            -np.sum(probabilities * np.log(probabilities.clip(1e-12)))
            / np.log(len(probabilities))
        )
        score_rows.append({
            'family_id': family_id, 'task': task,
            'clean_top1_gap': top1_gap,
            'clean_normalized_entropy': entropy,
            'shard_base_correct': bool(score['correct']),
            'shard_base_prediction': str(score['predicted_semantic']),
        })
score_frame = pd.DataFrame(score_rows)
assert len(score_frame) == 160
print('Manifest / baseline / route:',
      len(manifest_frame), len(baseline_frame), len(route_frame))
print('Recovered confidence rows:', len(score_frame))
```

## 2. No-leakage family × task prediction table


```python
%%cell_progress 6/17 Build the no-leakage family × task prediction table
cell_stage('Joining targets, confidence, matched mechanisms, and wrong-task controls')

base = baseline_frame[
    baseline_frame['task'].isin(PRIMARY_TASKS)
].copy()
assert len(base) == 160
base['family_id'] = base['family_id'].astype(str)
base['same_net_instability'] = (~base['clean_cue_invariant'].astype(bool)).astype(int)
base['base_error'] = (~base['base_correct'].astype(bool)).astype(int)
base['mirror_error'] = (~base['mirror_correct'].astype(bool)).astype(int)
base['design_cell'] = base['cue_class'].astype(str) + '__' + base['margin_bin'].astype(str)

assert (base['base_semantic'].astype(str) == base['cue_semantic'].astype(str)).all()
base = base.merge(
    score_frame, on=['family_id', 'task'], how='left', validate='one_to_one'
)
assert base['clean_top1_gap'].notna().all()
assert (
    base['base_correct'].astype(bool)
    == base['shard_base_correct'].astype(bool)
).all()

route = route_frame[route_frame['task'].isin(PRIMARY_TASKS)].copy()
route['family_id'] = route['family_id'].astype(str)
route['abs_std_cue_leakage'] = route['std_cue_base_margin_change'].abs()
mechanism_wide = route.pivot(
    index=['family_id', 'task'],
    columns='site_label',
    values=list(MECHANISM_COMPONENTS),
)
mechanism_wide.columns = [
    f'mech__{component}__{site}'
    for component, site in mechanism_wide.columns
]
mechanism_wide = mechanism_wide.reset_index()
MECHANISM_FEATURES = [
    column for column in mechanism_wide.columns if column.startswith('mech__')
]
assert len(MECHANISM_FEATURES) == len(MECHANISM_COMPONENTS) * len(FROZEN_SITES)

prediction_frame = base.merge(
    mechanism_wide, on=['family_id', 'task'], how='left', validate='one_to_one'
)
wrong = mechanism_wide.copy()
wrong['task'] = wrong['task'].map({
    'net_torque': 'outcome', 'outcome': 'net_torque',
})
wrong = wrong.rename(columns={
    column: column.replace('mech__', 'wrong__', 1)
    for column in MECHANISM_FEATURES
})
WRONG_TASK_FEATURES = [
    column.replace('mech__', 'wrong__', 1)
    for column in MECHANISM_FEATURES
]
prediction_frame = prediction_frame.merge(
    wrong[['family_id', 'task'] + WRONG_TASK_FEATURES],
    on=['family_id', 'task'], how='left', validate='one_to_one',
)

METADATA_NUMERIC = ['base_margin', 'cue_source_margin_gap']
METADATA_CATEGORICAL = [
    'task', 'cue_class', 'margin_bin', 'cue_type', 'base_outcome',
]
BEHAVIOR_NUMERIC = [
    'clean_top1_gap', 'clean_normalized_entropy',
]
FEATURE_SPEC = {
    'metadata': {
        'numeric': METADATA_NUMERIC,
        'categorical': METADATA_CATEGORICAL,
    },
    'behavior': {
        'numeric': METADATA_NUMERIC + BEHAVIOR_NUMERIC,
        'categorical': METADATA_CATEGORICAL,
    },
    'wrong_task_mechanism': {
        'numeric': METADATA_NUMERIC + BEHAVIOR_NUMERIC + WRONG_TASK_FEATURES,
        'categorical': METADATA_CATEGORICAL,
    },
    'matched_mechanism': {
        'numeric': METADATA_NUMERIC + BEHAVIOR_NUMERIC + MECHANISM_FEATURES,
        'categorical': METADATA_CATEGORICAL,
    },
}
required_columns = sorted(set(
    TARGETS + tuple(METADATA_NUMERIC) + tuple(METADATA_CATEGORICAL)
    + tuple(BEHAVIOR_NUMERIC) + tuple(MECHANISM_FEATURES)
    + tuple(WRONG_TASK_FEATURES)
))
assert prediction_frame[required_columns].notna().all().all()
assert len(prediction_frame) == 160
assert prediction_frame['family_id'].nunique() == 80

prediction_frame.to_csv(ANALYSIS_DIR / 'prediction_table.csv', index=False)
print('Prediction rows / families:', len(prediction_frame),
      prediction_frame['family_id'].nunique())
print('Mechanism features:', MECHANISM_FEATURES)
```

```python
%%cell_progress 7/17 Validate targets, balance, and feature falsifications
cell_stage('Checking prevalence, design cells, task pairing, and target leakage')

prevalence_rows = []
for target in TARGETS:
    for task, subset in [('all', prediction_frame)] + [
        (task, prediction_frame[prediction_frame['task'] == task])
        for task in PRIMARY_TASKS
    ]:
        prevalence_rows.append({
            'target': target, 'task': task,
            'n': int(len(subset)),
            'positives': int(subset[target].sum()),
            'prevalence': float(subset[target].mean()),
        })
prevalence_frame = pd.DataFrame(prevalence_rows)
display(prevalence_frame)

assert prediction_frame.groupby('family_id')['task'].nunique().eq(2).all()
assert set(prediction_frame['design_cell']) == {
    'congruent__high', 'congruent__low',
    'conflict__high', 'conflict__low',
}
assert prediction_frame.groupby('design_cell')['family_id'].nunique().eq(20).all()

# The clean behavior baseline uses only the unmodified base prediction's
# uncertainty. Cue and mirror correctness, gold margin, and intervention
# outcomes are never included as predictor columns.
for feature_set, specification in FEATURE_SPEC.items():
    columns = specification['numeric'] + specification['categorical']
    assert not any(
        forbidden in column
        for column in columns
        for forbidden in (
            'correct', 'instability', 'error', 'gold_margin',
            'cue_prediction', 'mirror_prediction',
        )
    ), (feature_set, columns)
print('All feature sets pass target-leakage checks.')
```

## 3. Frozen predictors and out-of-family evaluation


```python
%%cell_progress 8/17 Define frozen probabilistic predictors and metrics
cell_stage('Defining preprocessing, logistic regression, and proper scores')
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    brier_score_loss, log_loss, roc_auc_score,
)
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

def make_predictor(feature_set):
    specification = FEATURE_SPEC[feature_set]
    numeric = specification['numeric']
    categorical = specification['categorical']
    preprocessor = ColumnTransformer([
        ('numeric', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scale', StandardScaler()),
        ]), numeric),
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical),
    ])
    return Pipeline([
        ('preprocess', preprocessor),
        ('model', LogisticRegression(
            C=LOGISTIC_C, penalty='l2', solver='liblinear',
            max_iter=5000, random_state=SEED,
        )),
    ])

def probability_metrics(y_true, probability):
    y_true = np.asarray(y_true, dtype=int)
    probability = np.asarray(probability, dtype=float).clip(1e-6, 1 - 1e-6)
    return {
        'n': int(len(y_true)),
        'positives': int(y_true.sum()),
        'prevalence': float(y_true.mean()),
        'log_loss': float(log_loss(y_true, probability, labels=[0, 1])),
        'brier': float(brier_score_loss(y_true, probability)),
        'auroc': (
            float(roc_auc_score(y_true, probability))
            if len(np.unique(y_true)) == 2 else np.nan
        ),
    }

print('Frozen L2 logistic predictors ready.')
```

```python
%%cell_progress 9/17 Run mandatory group-split and pipeline smoke tests
cell_stage('Checking family isolation and probability outputs')

target = PRIMARY_TARGET
splitter = StratifiedGroupKFold(
    n_splits=CV_SPLITS, shuffle=True, random_state=SEED,
)
train_index, test_index = next(splitter.split(
    prediction_frame,
    prediction_frame[target],
    groups=prediction_frame['family_id'],
))
train_families = set(prediction_frame.iloc[train_index]['family_id'])
test_families = set(prediction_frame.iloc[test_index]['family_id'])
assert train_families.isdisjoint(test_families)
assert len(train_families | test_families) == 80

for feature_set in FEATURE_SETS:
    model = make_predictor(feature_set)
    model.fit(
        prediction_frame.iloc[train_index],
        prediction_frame.iloc[train_index][target],
    )
    probability = model.predict_proba(
        prediction_frame.iloc[test_index]
    )[:, 1]
    assert len(probability) == len(test_index)
    assert np.isfinite(probability).all()
    assert ((probability > 0) & (probability < 1)).all()
print('Smoke train/test families:', len(train_families), len(test_families))
print('All four predictors produce finite held-out probabilities.')
```

```python
%%cell_progress 10/17 Run repeated family-grouped cross-fitting
cell_stage('Generating out-of-family probabilities; follow the repeat bar')

crossfit_records = []
for target in tqdm(TARGETS, desc='Prediction targets', unit='target'):
    for feature_set in FEATURE_SETS:
        probability_sum = np.zeros(len(prediction_frame), dtype=float)
        probability_count = np.zeros(len(prediction_frame), dtype=int)
        for repeat in tqdm(
            range(CV_REPEATS),
            desc=f'{target} | {feature_set}', unit='repeat', leave=False,
        ):
            splitter = StratifiedGroupKFold(
                n_splits=CV_SPLITS, shuffle=True,
                random_state=SEED + 1009 * repeat,
            )
            for fold, (train_index, test_index) in enumerate(splitter.split(
                prediction_frame,
                prediction_frame[target],
                groups=prediction_frame['family_id'],
            )):
                train_groups = set(
                    prediction_frame.iloc[train_index]['family_id']
                )
                test_groups = set(
                    prediction_frame.iloc[test_index]['family_id']
                )
                assert train_groups.isdisjoint(test_groups)
                model = make_predictor(feature_set)
                model.fit(
                    prediction_frame.iloc[train_index],
                    prediction_frame.iloc[train_index][target],
                )
                probability = model.predict_proba(
                    prediction_frame.iloc[test_index]
                )[:, 1]
                probability_sum[test_index] += probability
                probability_count[test_index] += 1
        assert np.all(probability_count == CV_REPEATS), (
            target, feature_set, np.unique(probability_count)
        )
        averaged = probability_sum / probability_count
        for row_index, probability in enumerate(averaged):
            row = prediction_frame.iloc[row_index]
            crossfit_records.append({
                'row_index': int(row_index),
                'family_id': str(row['family_id']),
                'task': str(row['task']),
                'design_cell': str(row['design_cell']),
                'target': target,
                'feature_set': feature_set,
                'label': int(row[target]),
                'probability': float(probability),
            })

crossfit_frame = pd.DataFrame(crossfit_records)
assert len(crossfit_frame) == (
    len(prediction_frame) * len(TARGETS) * len(FEATURE_SETS)
)
crossfit_frame.to_csv(
    ANALYSIS_DIR / 'repeated_group_crossfit_predictions.csv', index=False
)
metric_rows = []
for (target, feature_set), subset in crossfit_frame.groupby([
    'target', 'feature_set'
]):
    metric_rows.append({
        'target': target, 'feature_set': feature_set,
        **probability_metrics(subset['label'], subset['probability']),
    })
crossfit_metrics = pd.DataFrame(metric_rows)
crossfit_metrics.to_csv(
    ANALYSIS_DIR / 'repeated_group_crossfit_metrics.csv', index=False
)
display(crossfit_metrics.sort_values(['target', 'log_loss']))
```

```python
%%cell_progress 11/17 Run leave-one-design-cell-out transfer
cell_stage('Holding out each cue-congruence × margin cell in turn')

loco_records = []
design_cells = sorted(prediction_frame['design_cell'].unique())
for target in tqdm(TARGETS, desc='LOCO targets', unit='target'):
    for held_out_cell in design_cells:
        test_mask = prediction_frame['design_cell'] == held_out_cell
        train_mask = ~test_mask
        assert prediction_frame.loc[test_mask, 'family_id'].nunique() == 20
        assert set(prediction_frame.loc[test_mask, 'family_id']).isdisjoint(
            set(prediction_frame.loc[train_mask, 'family_id'])
        )
        y_train = prediction_frame.loc[train_mask, target]
        assert y_train.nunique() == 2, (target, held_out_cell)
        for feature_set in FEATURE_SETS:
            model = make_predictor(feature_set)
            model.fit(prediction_frame.loc[train_mask], y_train)
            probability = model.predict_proba(
                prediction_frame.loc[test_mask]
            )[:, 1]
            for row_index, value in zip(
                prediction_frame.index[test_mask], probability
            ):
                row = prediction_frame.loc[row_index]
                loco_records.append({
                    'row_index': int(row_index),
                    'family_id': str(row['family_id']),
                    'task': str(row['task']),
                    'held_out_cell': held_out_cell,
                    'target': target, 'feature_set': feature_set,
                    'label': int(row[target]),
                    'probability': float(value),
                })

loco_frame = pd.DataFrame(loco_records)
assert len(loco_frame) == (
    len(prediction_frame) * len(TARGETS) * len(FEATURE_SETS)
)
loco_frame.to_csv(
    ANALYSIS_DIR / 'leave_one_cell_out_predictions.csv', index=False
)
loco_metric_rows = []
for (target, feature_set), subset in loco_frame.groupby([
    'target', 'feature_set'
]):
    loco_metric_rows.append({
        'target': target, 'feature_set': feature_set,
        **probability_metrics(subset['label'], subset['probability']),
    })
loco_metrics = pd.DataFrame(loco_metric_rows)
loco_metrics.to_csv(
    ANALYSIS_DIR / 'leave_one_cell_out_metrics.csv', index=False
)
display(loco_metrics.sort_values(['target', 'log_loss']))
```

## 4. Family inference, robustness, and paper decision


```python
%%cell_progress 12/17 Estimate family-level predictive improvements
cell_stage('Running cluster bootstrap and family sign-flip inference')

def seed_for(*values):
    digest = hashlib.sha256(
        (str(SEED) + '|' + '|'.join(map(str, values))).encode('utf-8')
    ).hexdigest()
    return int(digest[:8], 16)

def add_row_losses(frame):
    result = frame.copy()
    probability = result['probability'].to_numpy(float).clip(1e-6, 1 - 1e-6)
    label = result['label'].to_numpy(int)
    result['row_log_loss'] = -(
        label * np.log(probability)
        + (1 - label) * np.log(1 - probability)
    )
    result['row_brier'] = (probability - label) ** 2
    return result

def family_comparison(frame, target, comparator, metric, label):
    subset = add_row_losses(frame[frame['target'] == target])
    family = subset.groupby([
        'family_id', 'feature_set'
    ])[metric].mean().unstack()
    assert 'matched_mechanism' in family and comparator in family
    improvement = (
        family[comparator] - family['matched_mechanism']
    ).dropna().to_numpy(float)
    assert len(improvement) == 80
    rng = np.random.default_rng(seed_for(label, metric))
    indices = rng.integers(
        0, len(improvement),
        size=(BOOTSTRAP_REPLICATES, len(improvement)),
    )
    boot = improvement[indices].mean(axis=1)
    observed = float(improvement.mean())
    signs = rng.choice(
        (-1.0, 1.0),
        size=(PERMUTATION_REPLICATES, len(improvement)),
    )
    null = signs @ improvement / len(improvement)
    p_value = float(
        (1 + np.sum(null >= observed))
        / (PERMUTATION_REPLICATES + 1)
    )
    return {
        'target': target, 'comparator': comparator, 'metric': metric,
        'families': int(len(improvement)), 'estimate': observed,
        'ci_low': float(np.quantile(boot, ALPHA / 2)),
        'ci_high': float(np.quantile(boot, 1 - ALPHA / 2)),
        'sign_flip_p': p_value,
    }

comparison_rows = []
for target in TARGETS:
    for comparator in ('behavior', 'wrong_task_mechanism'):
        for metric in ('row_log_loss', 'row_brier'):
            comparison_rows.append(family_comparison(
                crossfit_frame, target, comparator, metric,
                f'crossfit:{target}:{comparator}',
            ))
comparison_frame = pd.DataFrame(comparison_rows)

loco_comparison_rows = []
for target in TARGETS:
    for comparator in ('behavior', 'wrong_task_mechanism'):
        for metric in ('row_log_loss', 'row_brier'):
            loco_comparison_rows.append(family_comparison(
                loco_frame, target, comparator, metric,
                f'loco:{target}:{comparator}',
            ))
loco_comparison_frame = pd.DataFrame(loco_comparison_rows)

comparison_frame.to_csv(
    ANALYSIS_DIR / 'crossfit_family_improvements.csv', index=False
)
loco_comparison_frame.to_csv(
    ANALYSIS_DIR / 'loco_family_improvements.csv', index=False
)
display(comparison_frame)
display(loco_comparison_frame)
```

```python
%%cell_progress 13/17 Audit task, design-cell, and calibration robustness
cell_stage('Checking whether predictive gains generalize across frozen strata')

loss_frame = add_row_losses(
    crossfit_frame[crossfit_frame['target'] == PRIMARY_TARGET]
)
loss_wide = loss_frame.pivot_table(
    index=['family_id', 'task', 'design_cell'],
    columns='feature_set', values='row_log_loss', aggfunc='mean',
).reset_index()
loss_wide['mechanism_gain_vs_behavior'] = (
    loss_wide['behavior'] - loss_wide['matched_mechanism']
)
loss_wide['mechanism_gain_vs_wrong'] = (
    loss_wide['wrong_task_mechanism'] - loss_wide['matched_mechanism']
)

stratum_rows = []
for group_name, group_columns in (
    ('task', ['task']), ('design_cell', ['design_cell']),
):
    for key, subset in loss_wide.groupby(group_columns):
        if isinstance(key, tuple):
            key = '|'.join(map(str, key))
        stratum_rows.append({
            'group': group_name, 'stratum': str(key),
            'n': int(len(subset)),
            'gain_vs_behavior': float(
                subset['mechanism_gain_vs_behavior'].mean()
            ),
            'gain_vs_wrong': float(
                subset['mechanism_gain_vs_wrong'].mean()
            ),
        })
stratum_frame = pd.DataFrame(stratum_rows)

primary_predictions = crossfit_frame[
    crossfit_frame['target'] == PRIMARY_TARGET
].copy()
calibration_rows = []
for feature_set, subset in primary_predictions.groupby('feature_set'):
    ranked = subset.copy()
    ranked['bin'] = pd.qcut(
        ranked['probability'], q=5, duplicates='drop', labels=False,
    )
    for bin_index, group in ranked.groupby('bin'):
        calibration_rows.append({
            'feature_set': feature_set, 'bin': int(bin_index),
            'n': int(len(group)),
            'mean_probability': float(group['probability'].mean()),
            'event_rate': float(group['label'].mean()),
        })
calibration_frame = pd.DataFrame(calibration_rows)

stratum_frame.to_csv(
    ANALYSIS_DIR / 'primary_stratum_robustness.csv', index=False
)
calibration_frame.to_csv(
    ANALYSIS_DIR / 'primary_calibration.csv', index=False
)
display(stratum_frame)
display(calibration_frame)
```

```python
%%cell_progress 15/17 Apply the frozen predictive-validity decision tree
cell_stage('Combining prevalence, specificity, proper scores, and transfer')

primary_prevalence = prevalence_frame[
    (prevalence_frame['target'] == PRIMARY_TARGET)
    & (prevalence_frame['task'] == 'all')
].iloc[0]
prevalence_gate = bool(
    int(primary_prevalence['positives']) >= MIN_POSITIVES
    and MIN_PREVALENCE <= float(primary_prevalence['prevalence']) <= MAX_PREVALENCE
)

def comparison_row(frame, comparator, metric):
    subset = frame[
        (frame['target'] == PRIMARY_TARGET)
        & (frame['comparator'] == comparator)
        & (frame['metric'] == metric)
    ]
    assert len(subset) == 1
    return subset.iloc[0]

log_vs_behavior = comparison_row(
    comparison_frame, 'behavior', 'row_log_loss'
)
log_vs_wrong = comparison_row(
    comparison_frame, 'wrong_task_mechanism', 'row_log_loss'
)
brier_vs_behavior = comparison_row(
    comparison_frame, 'behavior', 'row_brier'
)
loco_log_vs_behavior = comparison_row(
    loco_comparison_frame, 'behavior', 'row_log_loss'
)

predictive_increment = bool(
    log_vs_behavior['estimate'] > 0
    and log_vs_behavior['ci_low'] > 0
    and log_vs_behavior['sign_flip_p'] < ALPHA
)
task_specificity = bool(
    log_vs_wrong['estimate'] > 0
    and log_vs_wrong['ci_low'] > 0
    and log_vs_wrong['sign_flip_p'] < ALPHA
)
brier_gate = bool(
    brier_vs_behavior['estimate'] > 0
    and brier_vs_behavior['ci_low'] > 0
)
cell_transfer_gate = bool(loco_log_vs_behavior['estimate'] > 0)
cell_rows = stratum_frame[stratum_frame['group'] == 'design_cell']
positive_design_cells = int((cell_rows['gain_vs_behavior'] > 0).sum())
stratum_gate = positive_design_cells >= 3

if not prevalence_gate:
    decision = 'primary_target_prevalence_blocks_claim'
elif (
    predictive_increment and task_specificity and brier_gate
    and cell_transfer_gate and stratum_gate
):
    decision = 'mechanism_predicts_counterfactual_instability'
elif predictive_increment and not task_specificity:
    decision = 'predictive_signal_not_task_specific'
elif predictive_increment and not cell_transfer_gate:
    decision = 'mechanism_predicts_random_folds_but_not_condition_transfer'
else:
    decision = 'mechanism_prediction_not_supported'

decision_payload = {
    'schema_version': 'mechanism-error-prediction-decision-1.0',
    'decision': decision,
    'model_id': MODEL_ID, 'model_revision': MODEL_REVISION,
    'families': 80, 'rows': len(prediction_frame),
    'primary_target': PRIMARY_TARGET,
    'primary_prevalence': float(primary_prevalence['prevalence']),
    'primary_positives': int(primary_prevalence['positives']),
    'prevalence_gate': prevalence_gate,
    'predictive_increment': predictive_increment,
    'task_specificity': task_specificity,
    'brier_gate': brier_gate,
    'cell_transfer_gate': cell_transfer_gate,
    'positive_design_cells': positive_design_cells,
    'stratum_gate': stratum_gate,
    'log_loss_vs_behavior': log_vs_behavior.to_dict(),
    'log_loss_vs_wrong_task': log_vs_wrong.to_dict(),
    'brier_vs_behavior': brier_vs_behavior.to_dict(),
    'loco_log_loss_vs_behavior': loco_log_vs_behavior.to_dict(),
    'crossfit_metrics': crossfit_metrics.to_dict(orient='records'),
    'loco_metrics': loco_metrics.to_dict(orient='records'),
    'limitations': [
        'Predictive validity is evaluated on one model and one controlled task family.',
        'The mechanism features use controlled donor interventions and are not deployment-time features.',
        'Repeated cross-fitting reduces leakage but does not create independent models or checkpoints.',
        'Wrong-task features are a specificity control, not a random-feature null distribution.',
        'A null does not show that every internal mechanism measurement lacks predictive value.',
    ],
}
DECISION_PATH = ANALYSIS_DIR / 'mechanism_error_prediction_decision.json'
write_json_atomic(DECISION_PATH, decision_payload)
print(json.dumps(decision_payload, indent=2))
```

```python
%%cell_progress 16/17 Export the claim card and readable report
cell_stage('Writing the decision-aware predictive-validity report')

interpretations = {
    'primary_target_prevalence_blocks_claim': (
        'The primary event is too rare or too common for the frozen '
        'probabilistic comparison to support a strong predictive claim.'
    ),
    'mechanism_predicts_counterfactual_instability': (
        'Task-matched causal mechanism measurements predict whether a '
        'same-net visual counterfactual changes the model answer, beyond scene '
        'metadata, clean confidence, and wrong-task mechanism features, and '
        'the advantage transfers across held-out design cells.'
    ),
    'predictive_signal_not_task_specific': (
        'Mechanism features improve prediction over behavior, but matched-task '
        'features do not beat the wrong-task control. The signal cannot be '
        'attributed to the claimed task-specific physical route.'
    ),
    'mechanism_predicts_random_folds_but_not_condition_transfer': (
        'Mechanism features help within repeated family folds but do not retain '
        'an advantage when a full cue-by-margin condition is held out.'
    ),
    'mechanism_prediction_not_supported': (
        'The frozen causal mechanism features do not add reliable held-out '
        'prediction of same-net instability beyond behavior and controls.'
    ),
}

report_lines = [
    '# VisualTorque mechanism-to-error prediction audit', '',
    f"- Decision: `{decision}`",
    '- Families: 80',
    f'- Prediction rows: {len(prediction_frame)}',
    f'- Primary target prevalence: {primary_prevalence["prevalence"]:.3f}',
    f'- Primary positives: {int(primary_prevalence["positives"])}',
    f'- Predictive increment gate: {predictive_increment}',
    f'- Task-specificity gate: {task_specificity}',
    f'- Leave-one-cell transfer gate: {cell_transfer_gate}', '',
    '## Interpretation', '', interpretations[decision], '',
    '## Evidence boundary', '',
    'Every reported probability is out-of-family. The main comparison uses '
    'proper probabilistic loss rather than in-sample accuracy. The mechanism '
    'model must beat both clean behavior and wrong-task mechanism controls; '
    'secondary error targets do not replace the primary same-net test.', '',
    '## Recommended next step', '',
]
if decision == 'mechanism_predicts_counterfactual_instability':
    report_lines.extend([
        'Freeze this feature definition and replicate predictive validity on '
        'Gemma-4-12B or on an independently generated shortcut-reversal support. '
        'The paper can then claim that causal mechanism measurements predict '
        'where behavior will fail, not only describe activations.', '',
    ])
elif decision == 'predictive_signal_not_task_specific':
    report_lines.extend([
        'Do not claim a torque-specific mechanism predictor. Test whether the '
        'signal is generic intervention sensitivity or confidence not captured '
        'by the clean baseline.', '',
    ])
elif decision == 'mechanism_predicts_random_folds_but_not_condition_transfer':
    report_lines.extend([
        'Treat the signal as in-distribution only. Do not use it to motivate '
        'OOD reliability or mechanism-guided repair without new evidence.', '',
    ])
else:
    report_lines.extend([
        'Retain the causal results as an explanatory boundary rather than a '
        'predictive metric. The next paper step should consolidate the existing '
        'mechanism story instead of adding more post-hoc predictors.', '',
    ])

REPORT_PATH = ANALYSIS_DIR / 'mechanism_error_prediction_report.md'
REPORT_PATH.write_text('\n'.join(report_lines) + '\n', encoding='utf-8')
write_json_atomic(ANALYSIS_DIR / 'paper_claim_card.json', {
    'decision': decision,
    'interpretation': interpretations[decision],
    'primary_target': PRIMARY_TARGET,
    'predictive_increment': predictive_increment,
    'task_specificity': task_specificity,
    'cell_transfer_gate': cell_transfer_gate,
    'nonclaim': (
        'This analysis does not establish a universal reliability metric or '
        'identify a unique physical circuit.'
    ),
})
print(REPORT_PATH.read_text(encoding='utf-8'))
print('All outputs:', ANALYSIS_DIR)
```
