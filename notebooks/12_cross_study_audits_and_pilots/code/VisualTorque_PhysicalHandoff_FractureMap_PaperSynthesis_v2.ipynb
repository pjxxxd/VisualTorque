```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_TRACKED_CELLS = 13
_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()
    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )
    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set(); worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close(); _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1); _SETUP_PROGRESS.close()
print(f'Cell progress helper ready: {_TRACKED_CELLS} tracked cells.')
```

```python
%%cell_progress 4/13 Resolve and audit evidence decisions
cell_stage('Loading decision artifacts without searching alternate outcomes')

from visualtorque_checks import sha256_file

def extract_decision(payload):
    for key in (
        'primary_decision', 'robustness_decision', 'route_decision',
        'decision', 'status', 'final_decision',
    ):
        value = payload.get(key)
        if isinstance(value, str):
            return value
    raise KeyError('No top-level decision field found.')

decision_payloads = {}
rows = []
for specification in tqdm(SOURCE_REGISTRY, desc='Evidence decisions', unit='source'):
    path = RESULTS_ROOT / specification['relative_path']
    if not path.is_file():
        rows.append({
            **specification, 'path': str(path), 'resolution': 'missing',
            'actual_decision': None, 'decision_matches_registry': False,
            'sha256': None,
        })
        continue
    payload = json.loads(path.read_text(encoding='utf-8'))
    actual = extract_decision(payload)
    decision_payloads[specification['evidence_id']] = payload
    rows.append({
        **specification, 'path': str(path), 'resolution': 'resolved',
        'actual_decision': actual,
        'decision_matches_registry': actual in specification['expected'],
        'sha256': sha256_file(path),
    })

decision_frame = pd.DataFrame(rows)
required_failures = decision_frame[
    decision_frame['required']
    & (
        (decision_frame['resolution'] != 'resolved')
        | (~decision_frame['decision_matches_registry'])
    )
]
assert required_failures.empty, required_failures[[
    'evidence_id', 'resolution', 'actual_decision', 'expected', 'path'
]].to_dict('records')
decision_frame.to_csv(TABLE_DIR / 'evidence_decision_inventory.csv', index=False)
display(decision_frame[[
    'evidence_id', 'stage', 'role', 'resolution', 'actual_decision'
]])
```

```python
%%cell_progress 5/13 Interpret the newest outcome-flip evidence boundary
cell_stage('Separating absence of support from inadequate support')

latest = decision_frame[
    decision_frame['evidence_id'] == 'E10_outcomeflip_confirmation'
].iloc[0]
if latest['resolution'] == 'missing':
    OUTCOME_FLIP_STATUS = 'pending'
    OUTCOME_FLIP_NOTE = (
        'The A100 outcome-flip confirmation has not completed. No conclusion '
        'is inferred from the earlier outcome-preserving audit.'
    )
else:
    payload = decision_payloads['E10_outcomeflip_confirmation']
    gates = payload.get('gates', {})
    adequate = bool(gates.get('adequate_discordant_support', False))
    if not adequate:
        OUTCOME_FLIP_STATUS = 'inconclusive_support_shortfall'
        OUTCOME_FLIP_NOTE = (
            'The analysis completed, but the preregistered discordant-support '
            'gate was not met. Treat this as underpowered, not as evidence that '
            'outcome-changing route gating is absent.'
        )
    elif latest['actual_decision'] == 'outcomeflip_withincontext_route_gating_confirmed':
        OUTCOME_FLIP_STATUS = 'supported'
        OUTCOME_FLIP_NOTE = 'Outcome-changing within-context route gating is confirmed.'
    elif latest['actual_decision'] == 'outcomeflip_route_gating_partially_supported':
        OUTCOME_FLIP_STATUS = 'partial'
        OUTCOME_FLIP_NOTE = 'Only a subset of the frozen outcome-flip gates passed.'
    else:
        OUTCOME_FLIP_STATUS = 'not_supported_under_frozen_test'
        OUTCOME_FLIP_NOTE = (
            'The adequately powered frozen outcome-flip test did not confirm '
            'route gating under its prespecified estimands.'
        )
print('Outcome-flip status:', OUTCOME_FLIP_STATUS)
print(OUTCOME_FLIP_NOTE)
```

```python
%%cell_progress 10/13 Export a numeric inventory for reviewer audit
cell_stage('Flattening decision payloads without selecting favorable metrics')

def flatten_numeric(value, prefix=''):
    rows = []
    if isinstance(value, dict):
        for key, child in value.items():
            child_prefix = f'{prefix}.{key}' if prefix else str(key)
            rows.extend(flatten_numeric(child, child_prefix))
    elif isinstance(value, list):
        for index, child in enumerate(value):
            rows.extend(flatten_numeric(child, f'{prefix}[{index}]'))
    elif isinstance(value, bool):
        rows.append((prefix, int(value), 'bool'))
    elif isinstance(value, (int, float)) and np.isfinite(value):
        rows.append((prefix, float(value), 'number'))
    return rows

numeric_rows = []
for evidence_id, payload in decision_payloads.items():
    for key, value, value_type in flatten_numeric(payload):
        numeric_rows.append({
            'evidence_id': evidence_id, 'field': key,
            'value': value, 'value_type': value_type,
        })
numeric_frame = pd.DataFrame(numeric_rows)
numeric_frame.to_csv(TABLE_DIR / 'decision_numeric_inventory.csv', index=False)
print('Numeric fields exported:', len(numeric_frame))
```
