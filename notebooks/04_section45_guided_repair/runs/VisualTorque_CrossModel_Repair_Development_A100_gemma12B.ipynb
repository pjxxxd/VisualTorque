```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')

IMPLEMENTATION_SHA256 = '160372af558929c42134517d3ffc09527278d018135dd39a46f32eb6987dde7d'
IMPLEMENTATION_SHA256 = 'fc1a7b1b66a4e089a8ac85bbb28bc6aeb55ed8bb1c8ab138ac1e8678ecea9e4c'
```

```python
%%cell_progress 4/13 Load shared validation utilities
"""Pure CPU utilities embedded in the follow-up notebooks; no model execution."""
import hashlib
import json
from pathlib import Path
from collections import defaultdict, Counter
from itertools import combinations, product
import numpy as np
import pandas as pd


def plain(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(type(value).__name__)


def digest(value):
    return hashlib.sha256(json.dumps(value, sort_keys=True, default=plain, allow_nan=False).encode()).hexdigest()


def file_hash(path):
    h = hashlib.sha256()
    with Path(path).open('rb') as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b''):
            h.update(chunk)
    return h.hexdigest()


def read_json(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))


def read_jsonl(path):
    return [json.loads(s) for s in Path(path).read_text(encoding='utf-8').splitlines() if s.strip()]


def save_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + '.tmp')
    temp.write_text(json.dumps(value, ensure_ascii=False, indent=2, default=plain, allow_nan=False) + '\n', encoding='utf-8')
    temp.replace(path)


def freeze(path, identity):
    identity = json.loads(json.dumps(identity, default=plain, sort_keys=True, allow_nan=False))
    if Path(path).is_file():
        assert read_json(path) == identity, f'Configuration changed; preserve the old run and use a new RUN_ID: {path}'
    else:
        save_json(path, identity)
    return digest(identity)


def cached_json(directory, key, identity, compute):
    path = Path(directory) / (digest(str(key))[:24] + '.json')
    expected = digest(identity)
    if path.is_file():
        old = read_json(path)
        assert old['key'] == str(key) and old['identity'] == expected, f'Incompatible cache: {path}'
        return old['result']
    value = compute()
    save_json(path, dict(key=str(key), identity=expected, result=value))
    return value


def bootstrap(values, key='', replicates=10000):
    x = np.asarray(values, dtype=float)
    assert x.ndim == 1 and len(x) and np.isfinite(x).all()
    rng = np.random.default_rng(int(digest(str(key))[:16], 16))
    draws = np.concatenate([rng.choice(x, (min(250, replicates-i), len(x)), replace=True).mean(1)
                            for i in range(0, replicates, 250)])
    return dict(n=len(x), mean=float(x.mean()), lower=float(np.quantile(draws, .025)),
                upper=float(np.quantile(draws, .975)),
                p_value=float((1 + (draws <= 0).sum()) / (replicates + 1)))


def bh(values):
    p = np.asarray(values, dtype=float)
    assert np.isfinite(p).all() and ((p >= 0) & (p <= 1)).all()
    order = np.argsort(p)
    q = np.empty_like(p)
    previous = 1.
    for rank in range(len(p)-1, -1, -1):
        i = order[rank]
        previous = min(previous, p[i]*len(p)/(rank+1))
        q[i] = previous
    return q


def paired(left, right, id_col, value):
    assert left[id_col].is_unique and right[id_col].is_unique
    assert set(left[id_col]) == set(right[id_col]), 'Different paired support'
    out = left[[id_col, value]].merge(right[[id_col, value]], on=id_col, validate='one_to_one', suffixes=('_l', '_r'))
    return out[value+'_l'].to_numpy(float) - out[value+'_r'].to_numpy(float)


def semantic_scores(record):
    if 'semantic_logprobs' in record:
        return {str(k): float(v) for k, v in record['semantic_logprobs'].items()}
    values = record['candidate_logprobs']
    if isinstance(values, dict):
        return {str(record['label_to_semantic'][k]): float(v) for k, v in values.items()}
    return dict(zip(map(str, record['candidate_semantics']), map(float, values)))


def correct_margin(record):
    values = semantic_scores(record)
    gold = str(record['gold_semantic'])
    assert gold in values and len(values) >= 2
    return values[gold] - max(v for k, v in values.items() if k != gold)


def random_unit(shape, key):
    r = np.random.default_rng(int(digest(key)[:16], 16)).normal(size=shape)
    return r / np.linalg.norm(r)


def coordinate_clamp(pre, untreated, axis, cast):
    h, base, u = [np.asarray(x, dtype=np.float64) for x in (pre, untreated, axis)]
    u = u / np.linalg.norm(u)
    a = float(np.sum((h-base)*u))
    return np.asarray(cast(h-a*u), dtype=np.float64), a


def native_norm_edit(pre, direction, norm, cast, tolerance=.02, steps=48):
    h = np.asarray(pre, dtype=np.float64)
    r = np.asarray(direction, dtype=np.float64)
    assert np.isfinite(h).all() and np.isfinite(r).all() and norm >= 0 and np.linalg.norm(r) > 0
    r /= np.linalg.norm(r)
    if norm == 0:
        return h.copy(), dict(actual_norm=0., target_norm=0., relative_error=0., matched=True)
    lo, hi = 0., max(float(norm), 1e-12)
    for _ in range(32):
        if np.linalg.norm(np.asarray(cast(h+hi*r))-h) >= norm:
            break
        hi *= 2
    best, error = h.copy(), 1.
    for _ in range(steps):
        mid = (lo+hi)/2
        candidate = np.asarray(cast(h+mid*r), dtype=np.float64)
        actual = float(np.linalg.norm(candidate-h))
        current = abs(actual-norm)/norm
        if current < error:
            best, error = candidate.copy(), current
        if error <= tolerance:
            break
        if actual < norm:
            lo = mid
        else:
            hi = mid
    return best, dict(actual_norm=float(np.linalg.norm(best-h)), target_norm=float(norm),
                      relative_error=float(error), matched=bool(error <= tolerance))


def equal_removal(pre, untreated, axis, random_direction, cast, tolerance=.02):
    target, coefficient = coordinate_clamp(pre, untreated, axis, cast)
    target_norm = float(np.linalg.norm(target-np.asarray(pre)))
    sign = 1. if coefficient >= 0 else -1.
    post, audit = native_norm_edit(pre, -sign*np.asarray(random_direction), target_norm, cast, tolerance)
    audit.update(reference_coefficient=coefficient, reference_actual_norm=target_norm)
    return post, audit


def geometry_signature(stacks):
    a = tuple(sorted((str(s['side']), int(s['distance']), int(s['units'])) for s in stacks))
    b = tuple(sorted(('right' if side == 'left' else 'left', d, u) for side, d, u in a))
    return min(a, b)  # mirrors belong to the same physical family


def geometries_in(value):
    found = set()
    if isinstance(value, dict):
        stacks = value.get('stacks')
        if isinstance(stacks, list) and stacks and all(isinstance(s, dict) and {'side','distance','units'} <= set(s) for s in stacks):
            found.add(geometry_signature(stacks))
        # Some prior matched cohorts store geometry instead of stack records.
        geom = value.get('pair_geometry')
        if isinstance(geom, (list, tuple)) and len(geom) == 4 and all(isinstance(x, (int,float)) for x in geom):
            u1,d1,u2,d2 = map(int, geom)
            found.add(geometry_signature([dict(side='left',distance=d1,units=u1),dict(side='right',distance=d2,units=u2)]))
        if isinstance(geom, (list, tuple)) and len(geom)==2 and all(isinstance(x,(list,tuple)) and len(x)==2 for x in geom):
            (u1,d1),(u2,d2)=geom
            found.add(geometry_signature([dict(side='left',distance=int(d1),units=int(u1)),dict(side='right',distance=int(d2),units=int(u2))]))
        for child in value.values():
            found.update(geometries_in(child))
    elif isinstance(value, list):
        for child in value:
            found.update(geometries_in(child))
    return found


def physics(stacks):
    units = {s:sum(int(x['units']) for x in stacks if x['side']==s) for s in ('left','right')}
    dist = {s:sum(int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    torque = {s:sum(int(x['units'])*int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    out = lambda x:'left' if x>0 else 'right' if x<0 else 'balanced'
    net=torque['left']-torque['right']; w=units['left']-units['right']; d=dist['left']-dist['right']
    return dict(net_torque=net,outcome=out(net),weight_score=w,distance_score=d,
                normalized_margin=abs(net)/max(1,sum(torque.values())),
                left_units=units['left'],right_units=units['right'],left_torque=torque['left'],right_torque=torque['right'],
                strategy_predictions=dict(exact_torque=out(net),weight_only=out(w),distance_only=out(d)))


def exclusion_snapshot(project, followup_root, dataset_dir=None):
    lock=Path(followup_root)/'prior_geometry_snapshot.json'
    if lock.is_file():
        old=read_json(lock)
        for path, sha in old['files'].items():
            assert file_hash(Path(path))==sha, f'Prior input changed: {path}'
        return {tuple(tuple(x) for x in sig) for sig in old['signatures']}, old
    paths={Path(dataset_dir or project/'data/dataset_v1')/'scenes.jsonl'}
    for pattern in ('*manifest*.jsonl','*support*.jsonl','*geometries*.jsonl'):
        paths.update(p for p in (project/'results').rglob(pattern) if Path(followup_root) not in p.parents)
    excluded=set(); hashes={}
    for path in sorted(paths):
        assert path.is_file(), f'Missing required geometry input: {path}'
        excluded.update(geometries_in(read_jsonl(path)))
        hashes[str(path.resolve())]=file_hash(path)
    assert excluded, 'No prior physical geometry was indexed'
    old=dict(files=hashes,signatures=sorted(excluded),mirror_canonical=True)
    save_json(lock,old)
    return excluded,old


def inverse_families(n, seed, excluded):
    assert n%4==0
    pool=defaultdict(list)
    for dl in combinations(range(1,7),2):
        for dr in combinations(range(1,7),2):
            for units in product(range(1,6), repeat=4):
                stacks=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),dl+dr,units)]
                f=physics(stacks)
                pool[dl+dr].append((stacks,f))
    candidates=[]
    for distances,rows in pool.items():
        by_net=defaultdict(list)
        for stacks,f in rows:
            if f['outcome']==f['strategy_predictions']['weight_only'] and f['outcome']!='balanced':
                by_net[f['net_torque']].append((stacks,f))
        for stacks,f in rows:
            margin=f['normalized_margin']; difficulty='hard' if .08<=margin<.22 else 'moderate' if .22<=margin<.45 else None
            if (difficulty and f['outcome']!='balanced' and f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome'])
                    and f['strategy_predictions']['distance_only']==f['outcome'] and abs(f['distance_score'])>=2):
                sources=by_net.get(f['net_torque'],[])
                if sources:candidates.append((digest([seed,stacks]),difficulty,stacks,f,sources))
    candidates.sort(key=lambda r:r[0]); used=set(excluded); counts=Counter(); result=[]
    for _,difficulty,stacks,f,sources in candidates:
        stratum=(difficulty,f['outcome'])
        if counts[stratum]>=n//4 or geometry_signature(stacks) in used:continue
        valid=[(s,g) for s,g in sources if geometry_signature(s) not in used|{geometry_signature(stacks)}]
        if not valid:continue
        valid.sort(key=lambda x:(sum(abs(a['units']-b['units']) for a,b in zip(stacks,x[0])),digest([seed,x[0]])))
        src,sf=valid[0];fid='fixed_'+digest([seed,stacks,src])[:20]
        roles={name:dict(scene_id=fid+'__'+name,image_filename=fid+'__'+name+'.png',stacks=s,features=g)
               for name,s,g in [('base',stacks,f),('aligned_source',src,sf)]}
        result.append(dict(family_id=fid,remapping_split='new_fixed_confirmation',roles=roles,margin_bin=difficulty))
        used.update([geometry_signature(stacks),geometry_signature(src)]);counts[stratum]+=1
        if len(result)==n:break
    assert len(result)==n and set(counts.values())=={n//4}, (len(result),counts)
    return sorted(result,key=lambda r:r['family_id'])


def repair_families(n, seed, excluded):
    assert n%4==0
    candidates=defaultdict(list)
    for ul,dl,ur,dr in product(range(1,9),range(1,7),range(1,9),range(1,7)):
        s=[dict(side='left',units=ul,distance=dl),dict(side='right',units=ur,distance=dr)]
        f=physics(s)
        if f['net_torque']==0:continue
        if ul==ur and dl!=dr:kind='equal_weight_distance'
        elif f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome']):kind='weight_distance_conflict'
        elif f['strategy_predictions']['weight_only']==f['outcome'] and f['strategy_predictions']['distance_only']==f['outcome']:kind='aligned_cues'
        else:continue
        candidates[kind].append(s)
    for units in product(range(1,6),repeat=4):
        for distances in [((1,3),(2,5)),((2,5),(1,4)),((1,6),(3,5)),((2,4),(3,6))]:
            s=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),distances[0]+distances[1],units)]
            if physics(s)['net_torque']!=0:candidates['multi_stack_sum'].append(s)
    used=set(excluded); result=[]
    for kind in ('equal_weight_distance','weight_distance_conflict','multi_stack_sum','aligned_cues'):
        rows=sorted(candidates[kind],key=lambda s:digest([seed,kind,s])); count=0
        for stacks in rows:
            sig=geometry_signature(stacks)
            if sig in used:continue
            fid='repair_'+digest([seed,kind,sig])[:20];scenes=[]
            for mirror in (False,True):
                current=[dict(s,side=('right' if s['side']=='left' else 'left')) if mirror else dict(s) for s in stacks]
                for style in ('standard','alternate'):
                    sid=f'{fid}__m{int(mirror)}__{style}'
                    scenes.append(dict(scene_id=sid,image_filename=sid+'.png',stacks=current,features=physics(current),style=style,mirror=mirror))
            result.append(dict(family_id=fid,stratum=kind,scenes=scenes));used.add(sig);count+=1
            if count==n//4:break
        assert count==n//4,(kind,count,'Insufficient unused geometries; do not silently change task ranges')
    return sorted(result,key=lambda r:r['family_id'])
```

```python
%%cell_progress 7/13 Load pinned model
cell_stage(f'Loading {MODEL_KEY} weights for depth localization')
import gc
if 'adapter' in globals():
    del adapter
    gc.collect()
    torch.cuda.empty_cache()

adapter = HuggingFaceMultimodalAdapter(
    MODEL_ID, revision=MODEL_REVISION,
    dtype=MODEL_DTYPE, device_map=DEVICE_MAP,
)
display(adapter.metadata)
assert adapter.metadata['resolved_revision'] == MODEL_REVISION

def resolve_module_path(root, path):
    value = root
    for component in path.split('.'):
        value = getattr(value, component)
    return value

def discover_decoder_layers(model):
    candidates = (
        'model.language_model.layers', 'model.text_model.layers',
        'language_model.model.layers', 'model.model.layers',
        'model.layers', 'language_model.layers',
    )
    for path in candidates:
        try:
            layers = resolve_module_path(model, path)
        except AttributeError:
            continue
        if isinstance(layers, torch.nn.ModuleList) and len(layers) >= 8:
            return path, layers
    raise RuntimeError('Could not locate decoder layers.')

LAYER_PATH, DECODER_LAYERS = discover_decoder_layers(adapter.model)
assert len(DECODER_LAYERS) == EXPECTED_LAYER_COUNT
assert all(0 <= layer < len(DECODER_LAYERS) for layer in CAPTURE_LAYERS)
print('Layer path:', LAYER_PATH)
print('Capture layers:', CAPTURE_LAYERS)
assert not any(v == 'disk' for v in getattr(adapter.model,'hf_device_map',{}).values()), 'Use a GPU with enough memory for this model.'
```

```python
%%cell_progress 8/13 Define final-token capture and native edits
def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(scores))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def label_score(record, label):
    labels = list(record['candidate_labels'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    index = labels.index(label)
    other = np.delete(values, index)
    return float(values[index] - np.logaddexp.reduce(other))

"""Final-token GPU helpers embedded after the proven adapter/scoring definitions."""

def final_forward(example, capture_layers=(), operations=None):
    operations=operations or {}
    prepared=prepare_examples([example]);mask=final_mask(prepared)
    captured={};audits={};handles=[]
    for layer in sorted(set(capture_layers)|set(operations)):
        def hook(_module,_inputs,output,layer=layer):
            hidden=output_hidden_tensor(output);m=mask.to(hidden.device)
            assert int(m.sum())==1 and hidden.dtype==torch.bfloat16
            pre=hidden[0,m];post=pre;op=operations.get(layer)
            if op:
                cast=lambda x:torch.as_tensor(np.asarray(x).copy(),dtype=pre.dtype).float().numpy()
                h=pre.detach().float().cpu().numpy()
                if op['action']=='add':
                    delta=torch.as_tensor(op['payload'],device=pre.device,dtype=pre.dtype).reshape(pre.shape)
                    post=pre+delta
                    audits[layer]=dict(action='add',requested_norm=float(delta.float().norm()))
                elif op['action']=='matched_add':
                    target=torch.as_tensor(op['reference'],device=pre.device,dtype=pre.dtype).reshape(pre.shape)
                    desired=float(((pre+target).float()-pre.float()).norm())
                    r=random_unit(h.shape,op['key'])
                    value,audit=native_norm_edit(h,r,desired,cast)
                    post=torch.as_tensor(value,device=pre.device,dtype=pre.dtype)
                    audits[layer]=dict(action='matched_add',**audit)
                elif op['action'] in ('clamp','random_coordinate','equal_remove'):
                    base=torch.as_tensor(op['untreated'],device=pre.device,dtype=pre.dtype).reshape(pre.shape)
                    axis=torch.as_tensor(op['axis'],device=pre.device,dtype=torch.float32).reshape(pre.shape)
                    axis=axis/axis.norm()
                    if op['action']=='random_coordinate':
                        axis=torch.as_tensor(random_unit(h.shape,op['key']),device=pre.device,dtype=torch.float32)
                    coefficient=((pre.float()-base.float())*axis).sum(-1,keepdim=True)
                    reference=(pre.float()-coefficient*axis).to(pre.dtype)
                    target_norm=float((reference.float()-pre.float()).norm())
                    if op['action']=='equal_remove':
                        sign=1. if float(coefficient.item())>=0 else -1.
                        value,audit=native_norm_edit(h,-sign*random_unit(h.shape,op['key']),target_norm,cast)
                        post=torch.as_tensor(value,device=pre.device,dtype=pre.dtype)
                    else:
                        post=reference;audit=dict(target_norm=target_norm,matched=True,relative_error=0.)
                    audits[layer]=dict(action=op['action'],coefficient=float(coefficient.item()),**audit)
                else:raise ValueError(op['action'])
                audits[layer]['actual_norm']=float((post.float()-pre.float()).norm())
                audits[layer]['hidden_norm']=float(pre.float().norm())
                modified=hidden.clone();modified[0,m]=post
            captured[layer]=post.detach().float().cpu().numpy().copy()
            return replace_output_hidden(output,modified) if op else None
        handles.append(DECODER_LAYERS[layer].register_forward_hook(hook))
    try:
        with torch.inference_mode():out=adapter.model(**prepared['inputs'],use_cache=False)
    finally:
        for handle in handles:handle.remove()
    record=score_records(prepared,out.logits)[0]
    record.update(example_id=example.example_id,scene_id=str(example.scene_id),variant=example.variant,
        split=example.split,net_torque=int(example.net_torque),correct=record['predicted_semantic']==str(example.gold_semantic),
        write_audits={str(k):v for k,v in audits.items()})
    assert set(captured)==set(capture_layers)|set(operations)
    return record,captured


def example_stamp(example):
    return dict(example_id=example.example_id,prompt=example.prompt,image=file_hash(example.image_path),
        labels=list(example.labels),semantics=list(example.option_semantics),gold=str(example.gold_semantic),
        torque=int(example.net_torque),split=example.split)


def current_runtime_identity():
    import transformers
    return dict(torch=torch.__version__,transformers=transformers.__version__,cuda=torch.version.cuda,
        dtype=MODEL_DTYPE,batch_size=1,checkpoint=MODEL_REVISION,layer_path=LAYER_PATH,
        implementation=IMPLEMENTATION_SHA256,
        toolkit={str(p.relative_to(TOOLKIT_ROOT)):file_hash(p) for p in sorted((TOOLKIT_ROOT/'src').rglob('*.py'))})
```

```python
%%cell_progress 9/13 Load only historical training/development and response-direction inputs
from sklearn.preprocessing import StandardScaler
from sklearn.cross_decomposition import PLSRegression
from visual_torque_toolkit.behavior import BehaviorRunConfig,PromptCondition,load_behavior_examples
from visual_torque_toolkit.behavior.data import BehaviorExample
from dataclasses import replace as dataclass_replace
import shutil
ACTIVATION_MANIFEST=PROJECT_ROOT/'results/behavior_comparison_v3/activation_pilot_manifest.jsonl'
DIRECTION_SUPPORT_PATH=PROJECT_ROOT/'results/semantic_label_depth_localization_v1/depth_support.jsonl'
DIRECTION_SCENE_DIR=PROJECT_ROOT/'results/inverse_query_weight_route_specificity_v1/rendered_scenes'
for path in (ACTIVATION_MANIFEST,DIRECTION_SUPPORT_PATH):assert path.is_file(),path
development_manifest=read_jsonl(ACTIVATION_MANIFEST)
assert len(development_manifest)==254 and {r['split'] for r in development_manifest}<={'probe_train','probe_dev'}
direction_support=read_jsonl(DIRECTION_SUPPORT_PATH)
assert len(direction_support)==32
RULE_TEXT=('Each square is one unit of weight. Distance is measured in tick intervals from the center pivot. '
           'Torque equals weight times distance. Total torque on a side is the sum over stacks on that side.')
LABELS=('A','B','C');SEMANTIC_TEXT={'left':'left side','right':'right side','balanced':'equal'}
def repair_prompt(task,semantics):
    question={'outcome':'Which way will the balance scale tip?','weight_more':'Which side has more total weight?',
              'weight_less':'Which side has less total weight?'}[task]
    return '\n\n'.join(['Answer this question using only the supplied balance-scale scene.',RULE_TEXT,'Question: '+question,
        'Options:\n'+'\n'.join(f'{l}. {SEMANTIC_TEXT[s]}' for l,s in zip(LABELS,semantics)),
        'Do not explain your reasoning. Reply with exactly one option letter and nothing else. Valid replies are: A, B, C.'])
def new_example(sid,path,features,semantics,variant,split,task='outcome'):
    val=features['net_torque'] if task=='outcome' else features['weight_score']*(1 if task=='weight_more' else -1)
    gold='left' if val>0 else 'right' if val<0 else 'balanced'
    return BehaviorExample(example_id=f'followup:{sid}:{task}:{variant}',question_id=f'followup_{sid}_{task}_{variant}',
        scene_id=sid,condition=PromptCondition.IMAGE_RULE,question_kind=task,variant=variant,split=split,
        problem_type='fixed_physics_repair',outcome=gold,net_torque=int(features['net_torque']),
        prompt=repair_prompt(task,semantics),image_path=Path(path),labels=LABELS,
        option_text=tuple(SEMANTIC_TEXT[s] for s in semantics),option_semantics=tuple(semantics),
        gold_label=LABELS[semantics.index(gold)],gold_semantic=gold)
def two_mappings(scene, family_id):
    # Rotation is family-fixed; mirror and render variants stay in the same family.
    offset=int(digest(['repair_mapping',family_id])[:8],16)%3
    base=['left','balanced','right'];base=base[offset:]+base[:offset]
    reverse=[{'left':'right','right':'left','balanced':'balanced'}[x] for x in base]
    return {'forward':tuple(base),'reverse':tuple(reverse)}
def code_key(wrong_label,correct_label):return str(wrong_label)+'_'+str(correct_label)
```

```python
%%cell_progress 11/13 Capture final-token states and fit supervised PLS4 prototype directions
train_states={};raw_baselines={}
for sid in tqdm(sorted(by_scene),desc='Capturing training and development scenes'):
    examples=sorted(by_scene[sid],key=lambda e:e.variant)
    def compute(examples=examples):
        states=defaultdict(list);scores=[]
        for ex in examples:
            record,h=final_forward(ex,CANDIDATE_LAYERS);scores.append(record)
            for layer in CANDIDATE_LAYERS:states[str(layer)].append(h[layer].ravel())
        return dict(scores=scores,mean_hidden={k:np.mean(v,axis=0).tolist() for k,v in states.items()})
    cached=cached_json(OUTPUT_ROOT/'capture',sid,train_identity,compute)
    train_states[sid]=cached['mean_hidden'];raw_baselines[sid]=cached['scores']
U_DIRECTIONS={};U_BASE_DOSES={}
y=np.array([int(by_scene[s][0].net_torque) for s in TRAIN_IDS],float)
assert (y>0).any() and (y<0).any()
for layer in CANDIDATE_LAYERS:
    X=np.array([train_states[s][str(layer)] for s in TRAIN_IDS],float)
    scaler=StandardScaler().fit(X);Z=scaler.transform(X)
    pls=PLSRegression(n_components=4,scale=False,max_iter=1000).fit(Z,y.reshape(-1,1))
    scores=pls.transform(Z);difference=scores[y>0].mean(0)-scores[y<0].mean(0)
    decoder=np.linalg.lstsq(scores-scores.mean(0),Z-Z.mean(0),rcond=1e-5)[0]
    direction=(difference@decoder)*scaler.scale_/2
    assert np.isfinite(direction).all() and np.linalg.norm(direction)>0
    U_DIRECTIONS[layer]=direction.astype('float32')
    # The new-model reference dose is defined using training states only.
    # It is NOT the legacy Qwen-27B mediator-matched alpha.
    U_BASE_DOSES[layer]=float(.01*np.median(np.linalg.norm(X,axis=1))/np.linalg.norm(direction))
print('Frozen base dose: 1% of median training hidden-state norm; development compares factors 0.5 and 1.')
```

```python
%%cell_progress 12/13 Build later answer directions from the disjoint historical response support
code_vectors=defaultdict(list);code_norms=[]
direction_hashes={}
for family in tqdm(direction_support,desc='Building later answer directions'):
    # Exact prior rotating-label seed and family identity; no confirmation outcomes.
    reference_key='options:'+family['family_id']
    offset=int(hashlib.sha256(f'20260901:{reference_key!r}'.encode()).hexdigest()[:8],16)%3
    semantics=['left','balanced','right'];semantics=semantics[offset:]+semantics[:offset]
    for task in ('weight_more','weight_less'):
        captured={};gold={}
        for role in ('base','aligned_source'):
            r=family['roles'][role];path=DIRECTION_SCENE_DIR/r['image_filename'];assert path.is_file(),path
            direction_hashes[str(path)]=file_hash(path)
            ex=new_example(r['scene_id'],path,r['features'],semantics,task,'historical_direction_fit',task)
            def compute(ex=ex):
                record,h=final_forward(ex,(DOWNSTREAM_LAYER,));return dict(record=record,state=h[DOWNSTREAM_LAYER].ravel().tolist())
            value=cached_json(OUTPUT_ROOT/'code_capture',ex.example_id,dict(runtime=runtime,input=example_stamp(ex),layer=DOWNSTREAM_LAYER),compute)
            captured[role]=np.asarray(value['state'],float);gold[role]=ex.gold_label
        delta=captured['aligned_source']-captured['base'];norm=float(np.linalg.norm(delta))
        if norm==0:continue
        code_vectors[code_key(gold['base'],gold['aligned_source'])].append(delta/norm);code_norms.append(norm)
CODE_DIRECTIONS={k:np.mean(v,axis=0) for k,v in code_vectors.items()}
assert len(CODE_DIRECTIONS)==6,'Historical support must span all six label transitions'
code_size=float(np.median(code_norms))
for key,value in CODE_DIRECTIONS.items():
    assert np.linalg.norm(value)>0
    CODE_DIRECTIONS[key]=(value/np.linalg.norm(value)*code_size).astype('float32')
freeze(OUTPUT_ROOT/'code_inputs.json',direction_hashes)
```

```python
%%cell_progress 13/13 Evaluate the bounded layer-dose grid against four matched random directions
dev_rows=[]
for sid in tqdm(DEV_IDS,desc='Development layer/dose comparison'):
    for ex in sorted(by_scene[sid],key=lambda e:e.variant):
        def evaluate(ex=ex):
            base,_=final_forward(ex);rows=[];sign=1. if ex.net_torque>0 else -1. if ex.net_torque<0 else 0.
            for layer in CANDIDATE_LAYERS:
                for factor in (.5,1.):
                    payload=sign*U_DIRECTIONS[layer]*U_BASE_DOSES[layer]*factor
                    for condition,seed in [('targeted',-1)]+[('random',s) for s in RANDOM_SEEDS]:
                        op=dict(action='add',payload=payload) if condition=='targeted' else dict(action='matched_add',reference=payload,key=[SEED,sid,ex.variant,layer,factor,seed])
                        record,_=final_forward(ex,operations={layer:op})
                        rows.append(dict(scene_id=sid,variant=ex.variant,layer=layer,factor=factor,condition=condition,random_seed=seed,
                            baseline_correct=base['correct'],accuracy_change=float(record['correct'])-float(base['correct']),
                            margin_change=correct_margin(record)-correct_margin(base),norm_valid=all(a.get('matched',True) for a in record['write_audits'].values()),record=record))
            return rows
        dev_rows.extend(cached_json(OUTPUT_ROOT/'development_trials',ex.example_id,
            dict(training=digest(train_identity),directions=digest(U_DIRECTIONS),doses=U_BASE_DOSES,input=example_stamp(ex)),evaluate))
dev=pd.DataFrame([{k:v for k,v in r.items() if k!='record'} for r in dev_rows]);dev.to_csv(ANALYSIS_DIR/'development_trials.csv',index=False)
save_json(ANALYSIS_DIR/'development_records.json',dev_rows)
rows=[]
for (layer,factor),sub in dev.groupby(['layer','factor']):
    target=sub[sub.condition.eq('targeted')].groupby('scene_id')[['accuracy_change','margin_change']].mean().reset_index()
    random=sub[sub.condition.eq('random')].groupby('scene_id')[['accuracy_change','margin_change']].mean().reset_index()
    direct=bootstrap(target.accuracy_change.to_numpy(float),[layer,factor,'accuracy'])
    specificity=bootstrap(paired(target,random,'scene_id','accuracy_change'),[layer,factor,'specificity'])
    valid=bool(sub.norm_valid.all())
    rows.append(dict(layer=int(layer),factor=float(factor),accuracy_change=direct['mean'],accuracy_lower=direct['lower'],
        specificity=specificity['mean'],specificity_lower=specificity['lower'],all_norms_valid=valid,
        eligible=bool(valid and direct['lower']>0 and specificity['lower']>0)))
selection_table=pd.DataFrame(rows);selection_table.to_csv(ANALYSIS_DIR/'development_selection.csv',index=False);display(selection_table)
eligible=selection_table[selection_table.eligible].sort_values(['specificity','layer','factor'],ascending=[False,True,True])
selected=eligible.iloc[0] if len(eligible) else None
selection=dict(model=MODEL_KEY,model_id=MODEL_ID,revision=MODEL_REVISION,downstream_layer=DOWNSTREAM_LAYER,
    eligible=selected is not None,method='pls4_prototype_train_only_six_map_mean',
    dose_rule='training median residual norm 1%, times frozen factor 0.5 or 1',
    training_identity=digest(train_identity),runtime=runtime,implementation=IMPLEMENTATION_SHA256,
    development_scenes=DEV_IDS,fit_scenes=TRAIN_IDS,direction_support_sha256=file_hash(DIRECTION_SUPPORT_PATH),
    conclusion='selected_for_new_confirmation' if selected is not None else 'no_upstream_repair_established_in_fixed_grid')
if selected is not None:
    layer=int(selected.layer);dose=U_BASE_DOSES[layer]*float(selected.factor)
    arrays={'upstream':U_DIRECTIONS[layer],**{'code_'+k:v for k,v in CODE_DIRECTIONS.items()}}
    vector_path=OUTPUT_ROOT/'frozen_directions.npz'
    if vector_path.is_file():
        with np.load(vector_path,allow_pickle=False) as old:
            assert set(old.files)==set(arrays) and all(np.array_equal(old[k],v) for k,v in arrays.items())
    else:np.savez_compressed(vector_path,**arrays)
    selection.update(upstream_layer=layer,dose=dose,directions_sha256=file_hash(OUTPUT_ROOT/'frozen_directions.npz'))
freeze(OUTPUT_ROOT/'selected_route.json',selection)
save_json(ANALYSIS_DIR/'RUN_COMPLETE.json',dict(status='development_complete',selection=selection))
(ANALYSIS_DIR/'REPORT_ZH.md').write_text('# 修正位置开发\n\n'+('已冻结候选，接下来只在新家族上确认。' if selected is not None else '固定候选范围内未建立有效上游修正；不开展依赖解释。')+'\n',encoding='utf-8')
print('Development finished:',OUTPUT_ROOT,'| eligible:',selection['eligible'])
```
