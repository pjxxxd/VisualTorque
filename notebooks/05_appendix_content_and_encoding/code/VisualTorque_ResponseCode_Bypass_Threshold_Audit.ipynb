```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_TRACKED_CELLS = 11
_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()
    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )
    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set(); worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close(); _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1); _SETUP_PROGRESS.close()
print(f'Cell progress helper ready: {_TRACKED_CELLS} tracked cells.')
```

```python
%%cell_progress 2/11 Verify the preinstalled analysis environment
cell_stage('Importing Colab-provided numerical libraries without upgrading them')
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

def one_sided_binomial_p(positive, total):
    positive, total = int(positive), int(total)
    assert 0 <= positive <= total
    if total == 0:
        return 1.0
    return float(sum(
        math.comb(total, value) for value in range(positive, total + 1)
    ) / (2 ** total))

print('numpy:', np.__version__)
print('pandas:', pd.__version__)
print('matplotlib:', plt.matplotlib.__version__)
print('No pip upgrade is performed in this notebook.')
```

```python
%%cell_progress 3/11 Freeze the response-code bypass curve audit
cell_stage('Locking parent identities, curve metrics, and interpretation gates')
import hashlib
import json
from datetime import datetime, timezone

SEED = 20261208
DOSES = (0.0, 0.5, 1.0, 1.5, 2.0, 3.0, 4.0)
LOW_DOSES = (0.0, 0.5, 1.0, 1.5, 2.0)
HIGH_DOSES = (2.0, 3.0, 4.0)
ROUTES = ('direct', 'sequential')
CURVE_MODES = ('recoded', 'target_code_only')
BOOTSTRAP_REPLICATES = 20000
ALPHA_LEVEL = 0.05
THRESHOLD_SENTINEL = max(DOSES) + 1.0

PARENT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'qwen27b_recipient_recode_dose_response_confirmation_v1'
)
PARENT_RUN_PATH = PARENT_ROOT / 'RUN_COMPLETE.json'
PARENT_PROTOCOL_PATH = PARENT_ROOT / 'frozen_protocol.json'
PARENT_PAIR_PATH = PARENT_ROOT / 'frozen_pair_manifest.jsonl'
PARENT_RECORDS_PATH = PARENT_ROOT / 'analysis' / 'dose_condition_records.csv'
PARENT_CURVES_PATH = PARENT_ROOT / 'analysis' / 'dose_curves.csv'
PARENT_DECISION_PATH = PARENT_ROOT / 'analysis' / 'dose_confirmation_decision.json'
PARENT_AUDIT_PATH = PARENT_ROOT / 'analysis' / 'audit.json'

OUTPUT_ROOT = (
    PROJECT_ROOT / 'results'
    / 'qwen27b_response_code_bypass_threshold_audit_v1'
)
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_protocol.json'
ANALYSIS_DIR = OUTPUT_ROOT / 'analysis'
FIGURE_DIR = ANALYSIS_DIR / 'figures'
for directory in (OUTPUT_ROOT, ANALYSIS_DIR, FIGURE_DIR):
    directory.mkdir(parents=True, exist_ok=True)

from visualtorque_checks import sha256_file_by_name as sha256_file

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def json_default(value):
    if isinstance(value, np.generic): return value.item()
    if isinstance(value, Path): return str(value)
    raise TypeError(type(value).__name__)

def write_json_atomic(file_path, value):
    file_path = Path(file_path)
    temporary = file_path.with_suffix(file_path.suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False,
                   default=json_default) + '\n', encoding='utf-8'
    )
    temporary.replace(file_path)

required = (
    PARENT_RUN_PATH, PARENT_PROTOCOL_PATH, PARENT_PAIR_PATH,
    PARENT_RECORDS_PATH, PARENT_CURVES_PATH,
    PARENT_DECISION_PATH, PARENT_AUDIT_PATH,
)
for file_path in required:
    assert Path(file_path).is_file(), f'Missing prerequisite: {file_path}'

protocol_identity = {
    'schema_version': 'response-code-bypass-threshold-audit-1.0',
    'analysis_status': 'post_hoc_curve_audit_without_new_model_forwards',
    'parent_run_sha256': sha256_file(PARENT_RUN_PATH),
    'parent_protocol_sha256': sha256_file(PARENT_PROTOCOL_PATH),
    'parent_pair_sha256': sha256_file(PARENT_PAIR_PATH),
    'parent_records_sha256': sha256_file(PARENT_RECORDS_PATH),
    'doses': list(DOSES),
    'low_doses': list(LOW_DOSES), 'high_doses': list(HIGH_DOSES),
    'curve_comparison': 'recoded minus target_code_only within pair and dose',
    'threshold_metric': (
        'minimum tested alpha producing the target label; alpha 5 is a '
        'right-censoring sentinel if no tested dose succeeds'
    ),
    'curve_metrics': [
        'normalized full-range AUC advantage',
        'low-dose AUC advantage', 'high-dose AUC advantage',
        'code-only threshold minus recoded threshold',
        'linear dose-by-residual interaction slope',
    ],
    'bootstrap_unit': 'complete physical family',
    'interpretation_limit': (
        'This reanalysis can refine the bypass boundary but cannot convert '
        'reused support into an independent replication.'
    ),
    'seed': SEED,
}
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity, (
        'Existing protocol differs. Use a new output directory.'
    )
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Output:', OUTPUT_ROOT)
```

## Dosewise and threshold-sensitive residual tests


```python
%%cell_progress 5/11 Build pairwise residual-advantage curves and flip thresholds
cell_stage('Pairing recoded and target-code-only records at every alpha')

comparison = dose_frame[
    dose_frame['mode'].isin(CURVE_MODES)
].pivot(
    index=['pair_id', 'family_id', 'split', 'stale_error', 'route', 'alpha'],
    columns='mode',
    values=['desired_vs_source_margin', 'reaches_desired_label'],
).reset_index()
comparison.columns = [
    '__'.join(value for value in column if value)
    if isinstance(column, tuple) else column
    for column in comparison.columns
]
comparison = comparison.rename(columns={
    'desired_vs_source_margin__recoded': 'recoded_margin',
    'desired_vs_source_margin__target_code_only': 'code_only_margin',
    'reaches_desired_label__recoded': 'recoded_success',
    'reaches_desired_label__target_code_only': 'code_only_success',
})
comparison['margin_advantage'] = (
    comparison['recoded_margin'] - comparison['code_only_margin']
)
comparison['success_advantage'] = (
    comparison['recoded_success'].astype(float)
    - comparison['code_only_success'].astype(float)
)
assert len(comparison) == len(pair_frame) * len(ROUTES) * len(DOSES)

def normalized_auc(alpha, values):
    alpha = np.asarray(alpha, dtype=float)
    values = np.asarray(values, dtype=float)
    order = np.argsort(alpha)
    alpha, values = alpha[order], values[order]
    denominator = float(alpha[-1] - alpha[0])
    assert denominator > 0
    return float(np.trapezoid(values, alpha) / denominator)

def minimum_success_dose(current, success_column):
    successful = current[current[success_column].astype(bool)]
    return (
        float(successful['alpha'].min())
        if not successful.empty else float(THRESHOLD_SENTINEL)
    )

pair_rows = []
for keys, current in comparison.groupby(
    ['pair_id', 'family_id', 'split', 'stale_error', 'route'], sort=True
):
    current = current.sort_values('alpha')
    assert tuple(current['alpha'].astype(float)) == DOSES
    low = current[current['alpha'].isin(LOW_DOSES)]
    high = current[current['alpha'].isin(HIGH_DOSES)]
    recoded_threshold = minimum_success_dose(current, 'recoded_success')
    code_only_threshold = minimum_success_dose(current, 'code_only_success')
    pair_rows.append({
        **dict(zip(
            ['pair_id', 'family_id', 'split', 'stale_error', 'route'], keys
        )),
        'full_auc_advantage': normalized_auc(
            current['alpha'], current['margin_advantage']
        ),
        'low_auc_advantage': normalized_auc(
            low['alpha'], low['margin_advantage']
        ),
        'high_auc_advantage': normalized_auc(
            high['alpha'], high['margin_advantage']
        ),
        'interaction_slope': float(np.polyfit(
            current['alpha'].to_numpy(dtype=float),
            current['margin_advantage'].to_numpy(dtype=float), 1,
        )[0]),
        'recoded_threshold': recoded_threshold,
        'code_only_threshold': code_only_threshold,
        'threshold_reduction': code_only_threshold - recoded_threshold,
        'recoded_censored': bool(recoded_threshold == THRESHOLD_SENTINEL),
        'code_only_censored': bool(code_only_threshold == THRESHOLD_SENTINEL),
        'alpha0_margin_advantage': float(
            current[np.isclose(current['alpha'], 0)]['margin_advantage'].iloc[0]
        ),
        'alpha3_margin_advantage': float(
            current[np.isclose(current['alpha'], 3)]['margin_advantage'].iloc[0]
        ),
    })

pair_estimands = pd.DataFrame(pair_rows)
comparison.to_csv(ANALYSIS_DIR / 'dosewise_residual_advantage.csv', index=False)
pair_estimands.to_csv(ANALYSIS_DIR / 'pair_curve_estimands.csv', index=False)
display(pair_estimands.groupby(['split', 'route']).agg(
    full_auc=('full_auc_advantage', 'mean'),
    low_auc=('low_auc_advantage', 'mean'),
    high_auc=('high_auc_advantage', 'mean'),
    threshold_reduction=('threshold_reduction', 'mean'),
    interaction_slope=('interaction_slope', 'mean'),
).round(4))
```

```python
%%cell_progress 6/11 Bootstrap the residual contribution at each dose
cell_stage('Estimating family-clustered dosewise intervals')

def family_interval(frame, value_column, key):
    family_values = frame.groupby('family_id')[value_column].mean().to_numpy(dtype=float)
    estimate = float(family_values.mean())
    generator = np.random.default_rng(
        int(stable_key(('bootstrap', key))[:16], 16) % (2**63 - 1)
    )
    draws = np.empty(BOOTSTRAP_REPLICATES, dtype=float)
    for index in range(BOOTSTRAP_REPLICATES):
        draws[index] = generator.choice(
            family_values, size=len(family_values), replace=True
        ).mean()
    return {
        'estimate': estimate,
        'ci_low': float(np.quantile(draws, ALPHA_LEVEL / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA_LEVEL / 2)),
        'pairs': int(len(frame)), 'families': int(len(family_values)),
    }

dose_test_rows = []
for split in ('all', 'discovery', 'confirmation'):
    split_frame = comparison if split == 'all' else comparison[
        comparison['split'] == split
    ]
    for route in ROUTES:
        for alpha in DOSES:
            current = split_frame[
                (split_frame['route'] == route)
                & np.isclose(split_frame['alpha'], alpha)
            ]
            dose_test_rows.append({
                'split': split, 'route': route, 'alpha': alpha,
                'metric': 'margin_advantage',
                **family_interval(
                    current, 'margin_advantage',
                    ('dose', split, route, alpha, 'margin'),
                ),
            })
            dose_test_rows.append({
                'split': split, 'route': route, 'alpha': alpha,
                'metric': 'success_advantage',
                **family_interval(
                    current, 'success_advantage',
                    ('dose', split, route, alpha, 'success'),
                ),
            })

dose_tests = pd.DataFrame(dose_test_rows)
dose_tests.to_csv(ANALYSIS_DIR / 'dosewise_family_bootstrap.csv', index=False)
display(dose_tests[
    (dose_tests['split'] == 'confirmation')
    & (dose_tests['metric'] == 'margin_advantage')
].round(4))
```

```python
%%cell_progress 7/11 Test AUC, threshold, and dose-by-residual interaction
cell_stage('Testing whether physical residuals reduce the steering requirement')

metrics = (
    'full_auc_advantage', 'low_auc_advantage', 'high_auc_advantage',
    'threshold_reduction', 'interaction_slope',
    'alpha0_margin_advantage', 'alpha3_margin_advantage',
)
curve_test_rows = []
for split in ('all', 'discovery', 'confirmation'):
    split_frame = pair_estimands if split == 'all' else pair_estimands[
        pair_estimands['split'] == split
    ]
    for route in ROUTES:
        current = split_frame[split_frame['route'] == route]
        for metric in metrics:
            curve_test_rows.append({
                'split': split, 'route': route, 'metric': metric,
                **family_interval(
                    current, metric, ('curve', split, route, metric)
                ),
            })

curve_tests = pd.DataFrame(curve_test_rows)
curve_tests.to_csv(ANALYSIS_DIR / 'curve_level_family_bootstrap.csv', index=False)

sign_rows = []
for split in ('all', 'discovery', 'confirmation'):
    split_frame = pair_estimands if split == 'all' else pair_estimands[
        pair_estimands['split'] == split
    ]
    for route in ROUTES:
        current = split_frame[split_frame['route'] == route]
        family_values = current.groupby('family_id')['threshold_reduction'].mean()
        nonzero = family_values[family_values != 0]
        if len(nonzero):
            positive = int((nonzero > 0).sum())
            sign_p = one_sided_binomial_p(positive, len(nonzero))
        else:
            positive, sign_p = 0, 1.0
        sign_rows.append({
            'split': split, 'route': route,
            'families': len(family_values),
            'nonzero_families': len(nonzero),
            'positive_threshold_reductions': positive,
            'one_sided_sign_p': sign_p,
        })
sign_frame = pd.DataFrame(sign_rows)
sign_frame.to_csv(ANALYSIS_DIR / 'threshold_sign_tests.csv', index=False)

display(curve_tests[curve_tests['split'] == 'confirmation'].round(4))
display(sign_frame)

def lookup(route, metric, split='confirmation'):
    return curve_tests[
        (curve_tests['split'] == split)
        & (curve_tests['route'] == route)
        & (curve_tests['metric'] == metric)
    ].iloc[0]

route_gates = {}
for route in ROUTES:
    threshold_gate = bool(
        lookup(route, 'threshold_reduction')['ci_low'] > 0
    )
    low_dose_gate = bool(
        lookup(route, 'low_auc_advantage')['ci_low'] > 0
    )
    full_curve_gate = bool(
        lookup(route, 'full_auc_advantage')['ci_low'] > 0
    )
    declining_gate = bool(
        lookup(route, 'interaction_slope')['ci_high'] < 0
    )
    high_dose_advantage = bool(
        lookup(route, 'high_auc_advantage')['ci_low'] > 0
    )
    route_gates[route] = {
        'threshold_reduction_gate': threshold_gate,
        'low_dose_residual_gate': low_dose_gate,
        'full_curve_residual_gate': full_curve_gate,
        'residual_declines_with_dose': declining_gate,
        'high_dose_residual_advantage': high_dose_advantage,
    }

if all(g['threshold_reduction_gate'] for g in route_gates.values()):
    PRIMARY_DECISION = 'physical_residual_reduces_required_response_code_dose'
elif any(
    g['low_dose_residual_gate'] and g['residual_declines_with_dose']
    for g in route_gates.values()
):
    PRIMARY_DECISION = 'weak_low_dose_physical_support_overridden_by_output_code'
elif not any(
    g['low_dose_residual_gate'] or g['full_curve_residual_gate']
    for g in route_gates.values()
):
    PRIMARY_DECISION = 'pure_response_code_bypass_on_current_support'
else:
    PRIMARY_DECISION = 'mixed_residual_contribution_inconclusive'

decision = {
    'primary_decision': PRIMARY_DECISION,
    'route_gates': route_gates,
    'analysis_status': protocol_identity['analysis_status'],
    'interpretation': (
        'Threshold and curve metrics test whether the transported residual '
        'reduces the amount of oracle response-code steering needed for repair.'
    ),
}
write_json_atomic(ANALYSIS_DIR / 'bypass_threshold_decision.json', decision)
display(decision)
```

```python
%%cell_progress 8/11 Audit monotonicity, splits, censoring, and leave-one-family-out stability
cell_stage('Checking curve completeness and sensitivity to individual families')

monotonic_rows = []
for keys, current in comparison.groupby(['pair_id', 'route']):
    current = current.sort_values('alpha')
    for prefix in ('recoded', 'code_only'):
        values = current[f'{prefix}_margin'].to_numpy(dtype=float)
        monotonic_rows.append({
            'pair_id': keys[0], 'route': keys[1], 'mode': prefix,
            'nondecreasing': bool(np.all(np.diff(values) >= -0.25)),
            'minimum_step': float(np.diff(values).min()),
        })
monotonic_frame = pd.DataFrame(monotonic_rows)

lofo_rows = []
for route in ROUTES:
    current = pair_estimands[
        (pair_estimands['split'] == 'confirmation')
        & (pair_estimands['route'] == route)
    ]
    for held_out in sorted(current['family_id'].unique()):
        remaining = current[current['family_id'] != held_out]
        lofo_rows.append({
            'route': route, 'held_out_family': held_out,
            'low_auc_advantage': float(remaining['low_auc_advantage'].mean()),
            'full_auc_advantage': float(remaining['full_auc_advantage'].mean()),
            'threshold_reduction': float(remaining['threshold_reduction'].mean()),
            'interaction_slope': float(remaining['interaction_slope'].mean()),
        })
lofo_frame = pd.DataFrame(lofo_rows)
lofo_frame.to_csv(ANALYSIS_DIR / 'leave_one_family_out.csv', index=False)

audit = {
    'pairs': len(pair_frame), 'families': pair_frame['family_id'].nunique(),
    'discovery_confirmation_family_overlap': len(
        set(pair_frame[pair_frame['split'] == 'discovery']['family_id'])
        & set(pair_frame[pair_frame['split'] == 'confirmation']['family_id'])
    ),
    'complete_pair_route_dose_grid': bool(
        len(comparison) == len(pair_frame) * len(ROUTES) * len(DOSES)
    ),
    'recoded_censored_pairs': int(pair_estimands['recoded_censored'].sum()),
    'code_only_censored_pairs': int(pair_estimands['code_only_censored'].sum()),
    'monotonic_curve_rate': float(monotonic_frame['nondecreasing'].mean()),
    'minimum_lofo_low_auc': {
        route: float(lofo_frame[lofo_frame['route'] == route][
            'low_auc_advantage'
        ].min()) for route in ROUTES
    },
    'minimum_lofo_threshold_reduction': {
        route: float(lofo_frame[lofo_frame['route'] == route][
            'threshold_reduction'
        ].min()) for route in ROUTES
    },
    'new_model_forwards': 0,
    'post_hoc_analysis': True,
    'confirmation_support_reused_from_parent': True,
}
assert audit['discovery_confirmation_family_overlap'] == 0
assert audit['complete_pair_route_dose_grid']
assert audit['new_model_forwards'] == 0
write_json_atomic(ANALYSIS_DIR / 'audit.json', audit)
display(audit)
```

```python
%%cell_progress 9/11 Save the response-code bypass threshold figure
cell_stage('Rendering dosewise residual advantage and threshold comparisons')

fig, axes = plt.subplots(2, 2, figsize=(15, 11))
colors = {'direct': '#2563EB', 'sequential': '#D97706'}

for route in ROUTES:
    current = dose_tests[
        (dose_tests['split'] == 'confirmation')
        & (dose_tests['route'] == route)
        & (dose_tests['metric'] == 'margin_advantage')
    ].sort_values('alpha')
    axes[0, 0].plot(
        current['alpha'], current['estimate'], marker='o',
        color=colors[route], label=route.title(),
    )
    axes[0, 0].fill_between(
        current['alpha'].to_numpy(dtype=float),
        current['ci_low'].to_numpy(dtype=float),
        current['ci_high'].to_numpy(dtype=float),
        color=colors[route], alpha=0.18,
    )
axes[0, 0].axhline(0, color='black', linewidth=1)
axes[0, 0].set_xlabel('Target-code dose')
axes[0, 0].set_ylabel('Recoded minus target-code-only margin')
axes[0, 0].set_title('A. Does the physical residual add value at each dose?')
axes[0, 0].legend(frameon=False)

for route in ROUTES:
    current = comparison[
        (comparison['split'] == 'confirmation')
        & (comparison['route'] == route)
    ].groupby('alpha', as_index=False).agg(
        recoded=('recoded_success', 'mean'),
        code_only=('code_only_success', 'mean'),
    )
    axes[0, 1].plot(
        current['alpha'], current['recoded'], marker='o',
        color=colors[route], label=f'{route}: recoded',
    )
    axes[0, 1].plot(
        current['alpha'], current['code_only'], marker='x', linestyle='--',
        color=colors[route], label=f'{route}: code only',
    )
axes[0, 1].set_ylim(-0.03, 1.03)
axes[0, 1].set_xlabel('Target-code dose')
axes[0, 1].set_ylabel('Target-answer rate')
axes[0, 1].set_title('B. Behavioral flip thresholds')
axes[0, 1].legend(frameon=False, fontsize=8)

confirmation_estimands = pair_estimands[
    pair_estimands['split'] == 'confirmation'
].copy()
for index, route in enumerate(ROUTES):
    current = confirmation_estimands[
        confirmation_estimands['route'] == route
    ]
    axes[1, 0].scatter(
        np.full(len(current), index) - 0.12,
        current['recoded_threshold'], color=colors[route],
        marker='o', alpha=0.8,
    )
    axes[1, 0].scatter(
        np.full(len(current), index) + 0.12,
        current['code_only_threshold'], color=colors[route],
        marker='x', alpha=0.8,
    )
    for _, row in current.iterrows():
        axes[1, 0].plot(
            [index - 0.12, index + 0.12],
            [row['recoded_threshold'], row['code_only_threshold']],
            color='#9CA3AF', linewidth=0.8,
        )
axes[1, 0].set_xticks(range(len(ROUTES)), ['Direct', 'Sequential'])
axes[1, 0].set_ylabel('Minimum successful dose')
axes[1, 0].set_title('C. Pairwise dose needed for an answer flip')

forest_metrics = ('low_auc_advantage', 'high_auc_advantage', 'threshold_reduction')
forest = curve_tests[
    (curve_tests['split'] == 'confirmation')
    & (curve_tests['metric'].isin(forest_metrics))
].copy()
forest['label'] = forest['route'] + ': ' + forest['metric'].str.replace('_', ' ')
y = np.arange(len(forest))
axes[1, 1].errorbar(
    forest['estimate'], y,
    xerr=np.vstack([
        forest['estimate'] - forest['ci_low'],
        forest['ci_high'] - forest['estimate'],
    ]), fmt='o', color='#111827', capsize=4,
)
axes[1, 1].axvline(0, color='#9CA3AF', linestyle='--')
axes[1, 1].set_yticks(y, forest['label'])
axes[1, 1].set_title('D. Confirmation curve-level contrasts')

fig.suptitle('Physical residual contribution versus response-code bypass', fontsize=16)
fig.tight_layout()
figure_path = FIGURE_DIR / 'response_code_bypass_threshold_audit.png'
fig.savefig(figure_path, dpi=220, bbox_inches='tight')
plt.show()
print('Saved:', figure_path)
```

```python
%%cell_progress 10/11 Write the paper-facing bypass interpretation
cell_stage('Writing a result that separates behavioral repair from mechanism restoration')

def interval(route, metric, split='confirmation'):
    row = lookup(route, metric, split)
    return f"{row['estimate']:.4f} [{row['ci_low']:.4f}, {row['ci_high']:.4f}]"

lines = [
    '# Response-Code Bypass Threshold Audit', '',
    f'Primary decision: {PRIMARY_DECISION}', '',
    '## Question', '',
    (
        'At the discovery-selected dose, recoded and target-code-only '
        'conditions both reached ceiling accuracy. This audit asks whether '
        'the transported physical residual nevertheless reduces the dose '
        'needed for a flip or improves the response curve below ceiling.'
    ), '',
    '## Confirmation curve results', '',
]
for route in ROUTES:
    lines.extend([
        f'### {route.title()}', '',
        f'- Low-dose AUC advantage: {interval(route, "low_auc_advantage")}',
        f'- High-dose AUC advantage: {interval(route, "high_auc_advantage")}',
        f'- Full-range AUC advantage: {interval(route, "full_auc_advantage")}',
        f'- Code-only minus recoded flip threshold: {interval(route, "threshold_reduction")}',
        f'- Dose-by-residual interaction slope: {interval(route, "interaction_slope")}',
        '',
    ])
lines.extend([
    '## Interpretation boundary', '',
    (
        'A positive threshold reduction would mean that the physical residual '
        'lowers the amount of oracle output steering required. A positive '
        'low-dose effect that disappears at high doses would instead indicate '
        'weak subthreshold physical support that is eventually overwhelmed by '
        'the response-code intervention. No advantage across the curve supports '
        'a cleaner response-code bypass interpretation.'
    ),
    (
        'This is a post-hoc analysis of an existing dose grid. It adds no new '
        'model evaluations and is not an independent replication.'
    ), '',
])
report_path = ANALYSIS_DIR / 'RESPONSE_CODE_BYPASS_THRESHOLD_REPORT.md'
report_path.write_text('\n'.join(lines), encoding='utf-8')
print(report_path.read_text(encoding='utf-8'))
```
