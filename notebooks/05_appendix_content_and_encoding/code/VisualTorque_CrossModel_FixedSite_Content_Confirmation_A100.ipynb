```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')

IMPLEMENTATION_SHA256 = '160372af558929c42134517d3ffc09527278d018135dd39a46f32eb6987dde7d'
IMPLEMENTATION_SHA256 = '6631da39a2e1a6c885408e1d338df998a82bde9be59bd7b692df54d2a0b3708f'
```

```python
%%cell_progress 4/15 Load shared validation utilities
"""Pure CPU utilities embedded in the follow-up notebooks; no model execution."""
import hashlib
import json
from pathlib import Path
from collections import defaultdict, Counter
from itertools import combinations, product
import numpy as np
import pandas as pd


def plain(value):
    if isinstance(value, Path):
        return str(value)
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, np.generic):
        return value.item()
    raise TypeError(type(value).__name__)


def digest(value):
    return hashlib.sha256(json.dumps(value, sort_keys=True, default=plain, allow_nan=False).encode()).hexdigest()


from visualtorque_checks import file_hash


def read_json(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))


def read_jsonl(path):
    return [json.loads(s) for s in Path(path).read_text(encoding='utf-8').splitlines() if s.strip()]


def save_json(path, value):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    temp = path.with_suffix(path.suffix + '.tmp')
    temp.write_text(json.dumps(value, ensure_ascii=False, indent=2, default=plain, allow_nan=False) + '\n', encoding='utf-8')
    temp.replace(path)


def freeze(path, identity):
    identity = json.loads(json.dumps(identity, default=plain, sort_keys=True, allow_nan=False))
    if Path(path).is_file():
        assert read_json(path) == identity, f'Configuration changed; preserve the old run and use a new RUN_ID: {path}'
    else:
        save_json(path, identity)
    return digest(identity)


def cached_json(directory, key, identity, compute):
    path = Path(directory) / (digest(str(key))[:24] + '.json')
    expected = digest(identity)
    if path.is_file():
        old = read_json(path)
        assert old['key'] == str(key) and old['identity'] == expected, f'Incompatible cache: {path}'
        return old['result']
    value = compute()
    save_json(path, dict(key=str(key), identity=expected, result=value))
    return value


def bootstrap(values, key='', replicates=10000):
    x = np.asarray(values, dtype=float)
    assert x.ndim == 1 and len(x) and np.isfinite(x).all()
    rng = np.random.default_rng(int(digest(str(key))[:16], 16))
    draws = np.concatenate([rng.choice(x, (min(250, replicates-i), len(x)), replace=True).mean(1)
                            for i in range(0, replicates, 250)])
    return dict(n=len(x), mean=float(x.mean()), lower=float(np.quantile(draws, .025)),
                upper=float(np.quantile(draws, .975)),
                p_value=float((1 + (draws <= 0).sum()) / (replicates + 1)))


def bh(values):
    p = np.asarray(values, dtype=float)
    assert np.isfinite(p).all() and ((p >= 0) & (p <= 1)).all()
    order = np.argsort(p)
    q = np.empty_like(p)
    previous = 1.
    for rank in range(len(p)-1, -1, -1):
        i = order[rank]
        previous = min(previous, p[i]*len(p)/(rank+1))
        q[i] = previous
    return q


def paired(left, right, id_col, value):
    assert left[id_col].is_unique and right[id_col].is_unique
    assert set(left[id_col]) == set(right[id_col]), 'Different paired support'
    out = left[[id_col, value]].merge(right[[id_col, value]], on=id_col, validate='one_to_one', suffixes=('_l', '_r'))
    return out[value+'_l'].to_numpy(float) - out[value+'_r'].to_numpy(float)


def semantic_scores(record):
    if 'semantic_logprobs' in record:
        return {str(k): float(v) for k, v in record['semantic_logprobs'].items()}
    values = record['candidate_logprobs']
    if isinstance(values, dict):
        return {str(record['label_to_semantic'][k]): float(v) for k, v in values.items()}
    return dict(zip(map(str, record['candidate_semantics']), map(float, values)))


def correct_margin(record):
    values = semantic_scores(record)
    gold = str(record['gold_semantic'])
    assert gold in values and len(values) >= 2
    return values[gold] - max(v for k, v in values.items() if k != gold)


def random_unit(shape, key):
    r = np.random.default_rng(int(digest(key)[:16], 16)).normal(size=shape)
    return r / np.linalg.norm(r)


def coordinate_clamp(pre, untreated, axis, cast):
    h, base, u = [np.asarray(x, dtype=np.float64) for x in (pre, untreated, axis)]
    u = u / np.linalg.norm(u)
    a = float(np.sum((h-base)*u))
    return np.asarray(cast(h-a*u), dtype=np.float64), a


def native_norm_edit(pre, direction, norm, cast, tolerance=.02, steps=48):
    h = np.asarray(pre, dtype=np.float64)
    r = np.asarray(direction, dtype=np.float64)
    assert np.isfinite(h).all() and np.isfinite(r).all() and norm >= 0 and np.linalg.norm(r) > 0
    r /= np.linalg.norm(r)
    if norm == 0:
        return h.copy(), dict(actual_norm=0., target_norm=0., relative_error=0., matched=True)
    lo, hi = 0., max(float(norm), 1e-12)
    for _ in range(32):
        if np.linalg.norm(np.asarray(cast(h+hi*r))-h) >= norm:
            break
        hi *= 2
    best, error = h.copy(), 1.
    for _ in range(steps):
        mid = (lo+hi)/2
        candidate = np.asarray(cast(h+mid*r), dtype=np.float64)
        actual = float(np.linalg.norm(candidate-h))
        current = abs(actual-norm)/norm
        if current < error:
            best, error = candidate.copy(), current
        if error <= tolerance:
            break
        if actual < norm:
            lo = mid
        else:
            hi = mid
    return best, dict(actual_norm=float(np.linalg.norm(best-h)), target_norm=float(norm),
                      relative_error=float(error), matched=bool(error <= tolerance))


def equal_removal(pre, untreated, axis, random_direction, cast, tolerance=.02):
    target, coefficient = coordinate_clamp(pre, untreated, axis, cast)
    target_norm = float(np.linalg.norm(target-np.asarray(pre)))
    sign = 1. if coefficient >= 0 else -1.
    post, audit = native_norm_edit(pre, -sign*np.asarray(random_direction), target_norm, cast, tolerance)
    audit.update(reference_coefficient=coefficient, reference_actual_norm=target_norm)
    return post, audit


def geometry_signature(stacks):
    a = tuple(sorted((str(s['side']), int(s['distance']), int(s['units'])) for s in stacks))
    b = tuple(sorted(('right' if side == 'left' else 'left', d, u) for side, d, u in a))
    return min(a, b)  # mirrors belong to the same physical family


def geometries_in(value):
    found = set()
    if isinstance(value, dict):
        stacks = value.get('stacks')
        if isinstance(stacks, list) and stacks and all(isinstance(s, dict) and {'side','distance','units'} <= set(s) for s in stacks):
            found.add(geometry_signature(stacks))
        # Some prior matched cohorts store geometry instead of stack records.
        geom = value.get('pair_geometry')
        if isinstance(geom, (list, tuple)) and len(geom) == 4 and all(isinstance(x, (int,float)) for x in geom):
            u1,d1,u2,d2 = map(int, geom)
            found.add(geometry_signature([dict(side='left',distance=d1,units=u1),dict(side='right',distance=d2,units=u2)]))
        if isinstance(geom, (list, tuple)) and len(geom)==2 and all(isinstance(x,(list,tuple)) and len(x)==2 for x in geom):
            (u1,d1),(u2,d2)=geom
            found.add(geometry_signature([dict(side='left',distance=int(d1),units=int(u1)),dict(side='right',distance=int(d2),units=int(u2))]))
        for child in value.values():
            found.update(geometries_in(child))
    elif isinstance(value, list):
        for child in value:
            found.update(geometries_in(child))
    return found


def physics(stacks):
    units = {s:sum(int(x['units']) for x in stacks if x['side']==s) for s in ('left','right')}
    dist = {s:sum(int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    torque = {s:sum(int(x['units'])*int(x['distance']) for x in stacks if x['side']==s) for s in ('left','right')}
    out = lambda x:'left' if x>0 else 'right' if x<0 else 'balanced'
    net=torque['left']-torque['right']; w=units['left']-units['right']; d=dist['left']-dist['right']
    return dict(net_torque=net,outcome=out(net),weight_score=w,distance_score=d,
                normalized_margin=abs(net)/max(1,sum(torque.values())),
                left_units=units['left'],right_units=units['right'],left_torque=torque['left'],right_torque=torque['right'],
                strategy_predictions=dict(exact_torque=out(net),weight_only=out(w),distance_only=out(d)))


def exclusion_snapshot(project, followup_root, dataset_dir=None):
    lock=Path(followup_root)/'prior_geometry_snapshot.json'
    if lock.is_file():
        old=read_json(lock)
        for path, sha in old['files'].items():
            assert file_hash(Path(path))==sha, f'Prior input changed: {path}'
        return {tuple(tuple(x) for x in sig) for sig in old['signatures']}, old
    paths={Path(dataset_dir or project/'data/dataset_v1')/'scenes.jsonl'}
    for pattern in ('*manifest*.jsonl','*support*.jsonl','*geometries*.jsonl'):
        paths.update(p for p in (project/'results').rglob(pattern) if Path(followup_root) not in p.parents)
    excluded=set(); hashes={}
    for path in sorted(paths):
        assert path.is_file(), f'Missing required geometry input: {path}'
        excluded.update(geometries_in(read_jsonl(path)))
        hashes[str(path.resolve())]=file_hash(path)
    assert excluded, 'No prior physical geometry was indexed'
    old=dict(files=hashes,signatures=sorted(excluded),mirror_canonical=True)
    save_json(lock,old)
    return excluded,old


def inverse_families(n, seed, excluded):
    assert n%4==0
    pool=defaultdict(list)
    for dl in combinations(range(1,7),2):
        for dr in combinations(range(1,7),2):
            for units in product(range(1,6), repeat=4):
                stacks=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),dl+dr,units)]
                f=physics(stacks)
                pool[dl+dr].append((stacks,f))
    candidates=[]
    for distances,rows in pool.items():
        by_net=defaultdict(list)
        for stacks,f in rows:
            if f['outcome']==f['strategy_predictions']['weight_only'] and f['outcome']!='balanced':
                by_net[f['net_torque']].append((stacks,f))
        for stacks,f in rows:
            margin=f['normalized_margin']; difficulty='hard' if .08<=margin<.22 else 'moderate' if .22<=margin<.45 else None
            if (difficulty and f['outcome']!='balanced' and f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome'])
                    and f['strategy_predictions']['distance_only']==f['outcome'] and abs(f['distance_score'])>=2):
                sources=by_net.get(f['net_torque'],[])
                if sources:candidates.append((digest([seed,stacks]),difficulty,stacks,f,sources))
    candidates.sort(key=lambda r:r[0]); used=set(excluded); counts=Counter(); result=[]
    for _,difficulty,stacks,f,sources in candidates:
        stratum=(difficulty,f['outcome'])
        if counts[stratum]>=n//4 or geometry_signature(stacks) in used:continue
        valid=[(s,g) for s,g in sources if geometry_signature(s) not in used|{geometry_signature(stacks)}]
        if not valid:continue
        valid.sort(key=lambda x:(sum(abs(a['units']-b['units']) for a,b in zip(stacks,x[0])),digest([seed,x[0]])))
        src,sf=valid[0];fid='fixed_'+digest([seed,stacks,src])[:20]
        roles={name:dict(scene_id=fid+'__'+name,image_filename=fid+'__'+name+'.png',stacks=s,features=g)
               for name,s,g in [('base',stacks,f),('aligned_source',src,sf)]}
        result.append(dict(family_id=fid,remapping_split='new_fixed_confirmation',roles=roles,margin_bin=difficulty))
        used.update([geometry_signature(stacks),geometry_signature(src)]);counts[stratum]+=1
        if len(result)==n:break
    assert len(result)==n and set(counts.values())=={n//4}, (len(result),counts)
    return sorted(result,key=lambda r:r['family_id'])


def repair_families(n, seed, excluded):
    assert n%4==0
    candidates=defaultdict(list)
    for ul,dl,ur,dr in product(range(1,9),range(1,7),range(1,9),range(1,7)):
        s=[dict(side='left',units=ul,distance=dl),dict(side='right',units=ur,distance=dr)]
        f=physics(s)
        if f['net_torque']==0:continue
        if ul==ur and dl!=dr:kind='equal_weight_distance'
        elif f['strategy_predictions']['weight_only'] not in ('balanced',f['outcome']):kind='weight_distance_conflict'
        elif f['strategy_predictions']['weight_only']==f['outcome'] and f['strategy_predictions']['distance_only']==f['outcome']:kind='aligned_cues'
        else:continue
        candidates[kind].append(s)
    for units in product(range(1,6),repeat=4):
        for distances in [((1,3),(2,5)),((2,5),(1,4)),((1,6),(3,5)),((2,4),(3,6))]:
            s=[dict(side=side,distance=d,units=u) for side,d,u in zip(('left','left','right','right'),distances[0]+distances[1],units)]
            if physics(s)['net_torque']!=0:candidates['multi_stack_sum'].append(s)
    used=set(excluded); result=[]
    for kind in ('equal_weight_distance','weight_distance_conflict','multi_stack_sum','aligned_cues'):
        rows=sorted(candidates[kind],key=lambda s:digest([seed,kind,s])); count=0
        for stacks in rows:
            sig=geometry_signature(stacks)
            if sig in used:continue
            fid='repair_'+digest([seed,kind,sig])[:20];scenes=[]
            for mirror in (False,True):
                current=[dict(s,side=('right' if s['side']=='left' else 'left')) if mirror else dict(s) for s in stacks]
                for style in ('standard','alternate'):
                    sid=f'{fid}__m{int(mirror)}__{style}'
                    scenes.append(dict(scene_id=sid,image_filename=sid+'.png',stacks=current,features=physics(current),style=style,mirror=mirror))
            result.append(dict(family_id=fid,stratum=kind,scenes=scenes));used.add(sig);count+=1
            if count==n//4:break
        assert count==n//4,(kind,count,'Insufficient unused geometries; do not silently change task ranges')
    return sorted(result,key=lambda r:r['family_id'])
```

```python
%%cell_progress 9/15 Define original remapping prompts
cell_stage('Constructing mappings that separate semantic, label, and position')
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = {
    'outcome': 'Which way will the balance scale tip?',
    'weight_more': 'Which side has more total weight?',
    'weight_less': 'Which side has less total weight?',
}
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def task_semantic(features, task):
    if task == 'outcome':
        value = int(features['net_torque'])
    elif task == 'weight_more':
        value = int(features['weight_score'])
    elif task == 'weight_less':
        value = -int(features['weight_score'])
    else:
        raise KeyError(task)
    return outcome(value)

def reference_mapping(row):
    semantics = ['left', 'balanced', 'right']
    offset = int(stable_key(f"options:{row['family_id']}")[:8], 16) % 3
    semantics = semantics[offset:] + semantics[:offset]
    return tuple({
        'position': index, 'label': LABELS[index], 'semantic': semantics[index]
    } for index in range(3))

def diagnostic_mapping(row, donor_semantic, variant):
    reference = reference_mapping(row)
    donor_position = next(
        item['position'] for item in reference
        if item['semantic'] == donor_semantic
    )
    donor_label = reference[donor_position]['label']
    other_positions = [value for value in range(3) if value != donor_position]
    other_semantics = [
        value for value in ('left', 'balanced', 'right')
        if value != donor_semantic
    ]
    other_labels = [value for value in LABELS if value != donor_label]
    if variant == 'reverse':
        other_semantics.reverse()
        other_labels.reverse()
    elif variant != 'forward':
        raise KeyError(variant)
    label_position, semantic_position = other_positions
    mapping = [None, None, None]
    mapping[donor_position] = {
        'position': donor_position,
        'label': other_labels[0],
        'semantic': other_semantics[0],
    }
    mapping[label_position] = {
        'position': label_position,
        'label': donor_label,
        'semantic': other_semantics[1],
    }
    mapping[semantic_position] = {
        'position': semantic_position,
        'label': other_labels[1],
        'semantic': donor_semantic,
    }
    assert {item['label'] for item in mapping} == set(LABELS)
    assert {item['semantic'] for item in mapping} == {'left', 'balanced', 'right'}
    semantic_candidate = next(
        item['label'] for item in mapping if item['semantic'] == donor_semantic
    )
    label_candidate = donor_label
    position_candidate = mapping[donor_position]['label']
    assert len({semantic_candidate, label_candidate, position_candidate}) == 3
    diagnostics = {
        'donor_semantic': donor_semantic,
        'donor_label': donor_label,
        'donor_position': donor_position,
        'semantic_candidate': semantic_candidate,
        'label_candidate': label_candidate,
        'position_candidate': position_candidate,
    }
    return tuple(mapping), diagnostics

def semantics_by_label(mapping):
    return tuple(
        next(item['semantic'] for item in mapping if item['label'] == label)
        for label in LABELS
    )

def custom_prompt(task, mapping):
    options = '\n'.join(
        f"{item['label']}. {SEMANTIC_TEXT[item['semantic']]}"
        for item in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT[task],
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, task, role, mapping, name):
    role_record = row['roles'][role]
    features = role_record['features']
    semantics = semantics_by_label(mapping)
    gold = task_semantic(features, task)
    return BehaviorExample(
        example_id=f"remap:{row['family_id']}:{name}:{role}",
        question_id=f"remap_{row['family_id']}_{name}",
        scene_id=str(role_record['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind=task,
        variant='response_state_remapping',
        split=str(row['remapping_split']),
        problem_type='semantic_label_position_separation',
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=custom_prompt(task, mapping),
        image_path=(RUNTIME_SCENE_DIR / role_record['image_filename']).resolve(),
        labels=LABELS,
        option_text=tuple(
            SEMANTIC_TEXT[value] for value in semantics
        ),
        option_semantics=semantics,
        gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def donor_semantic_for(row, donor_task):
    return task_semantic(row['roles']['aligned_source']['features'], donor_task)

def build_reference_examples(row):
    mapping = reference_mapping(row)
    examples = {}
    for donor_task in DONOR_TASKS:
        for role in ('base', 'aligned_source'):
            key = f'ref__{donor_task}__{role}'
            examples[key] = make_example(
                row, donor_task, role, mapping, f'ref_{donor_task}'
            )
    return examples

def build_diagnostic_examples(row):
    examples = {}
    diagnostics = {}
    for donor_task in DONOR_TASKS:
        donor_semantic = donor_semantic_for(row, donor_task)
        for target_task in TARGET_TASKS:
            for variant in DIAGNOSTIC_VARIANTS:
                mapping, info = diagnostic_mapping(
                    row, donor_semantic, variant
                )
                key = f'diag__{donor_task}__{target_task}__{variant}'
                examples[key] = make_example(
                    row, target_task, 'base', mapping, key
                )
                diagnostics[key] = info
    return examples, diagnostics

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

print('Depth-localization mappings ready.')
```

```python
%%cell_progress 10/15 Define native matched remapping interventions
cell_stage('Defining variable-layer final-token transport')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(scores))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_capture(prepared, layers):
    hidden_by_layer = {}
    handles = []
    for layer in layers:
        def hook(_module, _inputs, output, layer=layer):
            hidden_by_layer[layer] = output_hidden_tensor(
                output
            ).detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def capture_examples(examples, layers):
    scores = {}
    hidden = {name: {} for name in examples}
    masks = {}
    for name, example in examples.items():
        prepared = prepare_examples([example])
        masks[name] = final_mask(prepared)
        current_scores, current_hidden = forward_capture(prepared, layers)
        scores[name] = current_scores[0]
        for layer, tensor in current_hidden.items():
            hidden[name][layer] = tensor[0]
        del prepared, current_hidden
    return scores, hidden, masks

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def matched_random(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    return (noise * (selected_norm(tensor) / selected_norm(noise))).to(tensor.dtype)

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layers = {int(row['layer_index']) for row in conditions}
    assert len(layers) == 1
    layer = next(iter(layers))
    payloads = torch.stack([row['payload_selected'] for row in conditions])
    masks = torch.stack([row['target_mask'] for row in conditions])
    modes = [row['intervention_mode'] for row in conditions]
    counts = []
    displacement_audits = []
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payloads.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        modified = hidden.clone()
        displacement_audits.clear()
        counts.clear()
        for index, mode in enumerate(modes):
            current = mask[index]
            assert int(current.sum()) == int(payload[index].shape[0]) == 1
            if mode == 'add':
                modified[index, current] = hidden[index, current] + payload[index]
            elif mode == 'replace':
                modified[index, current] = payload[index]
            else:
                raise ValueError(mode)
            if str(conditions[index].get('condition','')).endswith('_random'):
                reference=conditions[index]['reference_payload_selected'].to(device=hidden.device,dtype=hidden.dtype)
                pre_native=hidden[index,current]
                target_norm=float(((pre_native+reference).float()-pre_native.float()).norm())
                cast=lambda x:torch.as_tensor(np.asarray(x).copy(),dtype=hidden.dtype).float().numpy()
                raw_direction=payload[index].float().cpu().numpy()
                if np.linalg.norm(raw_direction)==0:raw_direction=np.ones_like(raw_direction)
                post,audit=native_norm_edit(pre_native.float().cpu().numpy(),raw_direction,target_norm,cast)
                modified[index,current]=torch.as_tensor(post,device=hidden.device,dtype=hidden.dtype)
                conditions[index]['native_norm_matched']=audit['matched']
                conditions[index]['native_norm_relative_error']=audit['relative_error']
                conditions[index]['reference_actual_delta_norm']=target_norm
            counts.append(int(current.sum()))
            pre = hidden[index, current].float()
            realized = modified[index, current].float() - pre
            requested = (payload[index].float() if mode == 'add' else payload[index].float() - pre)
            displacement_audits.append({
                'actual_delta_norm': float(realized.norm()),
                'requested_delta_norm': float(requested.norm()),
                'cast_error_norm': float((realized - requested).norm()),
            })

        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[layer].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, counts):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('payload_selected', 'target_mask', 'reference_payload_selected')
        })
        record['patched_tokens'] = count
    for record, audit_row in zip(records, displacement_audits):
        record.update(audit_row)
    return records


def score_conditions(example, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)
    for layer_conditions in grouped.values():
        for start in range(0, len(layer_conditions), PATCH_CONDITION_BATCH_SIZE):
            actual = layer_conditions[start:start + PATCH_CONDITION_BATCH_SIZE]
            padded = list(actual)
            while len(padded) < PATCH_CONDITION_BATCH_SIZE:
                padded.append(padded[-1])
            prepared = prepare_examples([example] * len(padded))
            records.extend(forward_intervention_chunk(
                prepared, padded
            )[:len(actual)])
            del prepared
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def label_score(record, label):
    labels = list(record['candidate_labels'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    index = labels.index(label)
    other = np.delete(values, index)
    return float(values[index] - np.logaddexp.reduce(other))

def entropy(record):
    probabilities = np.asarray(record['candidate_probabilities'], dtype=float)
    return float(-np.sum(probabilities * np.log(np.clip(probabilities, 1e-12, 1))))

print('Multi-layer final-token scorers ready.')
```

```python
%%cell_progress 11/15 Verify one unchanged critical path
cell_stage('Checking all fixed depths and three mutually exclusive candidates')
smoke_row = depth_support[0]
reference_examples = build_reference_examples(smoke_row)
diagnostic_examples, diagnostic_info = build_diagnostic_examples(smoke_row)
for key, info in diagnostic_info.items():
    assert len({
        info['semantic_candidate'], info['label_candidate'],
        info['position_candidate'],
    }) == 3, key

smoke_examples = {**reference_examples, **diagnostic_examples}
smoke_scores, smoke_hidden, smoke_masks = capture_examples(
    smoke_examples, SCAN_LAYERS
)
token_ids = []
for name, example in smoke_examples.items():
    prepared = prepare_examples([example])
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    token_ids.append(ids[final_mask(prepared)].tolist())
assert all(value == token_ids[0] for value in token_ids)

target_key = 'diag__weight_more__outcome__forward'
smoke_errors = {}
for layer in SCAN_LAYERS:
    self_record = score_conditions(smoke_examples[target_key], [{
        'family_id': smoke_row['family_id'], 'condition': 'self_replay',
        'layer_index': layer, 'intervention_mode': 'replace',
        'payload_selected': smoke_hidden[target_key][layer][smoke_masks[target_key]],
        'target_mask': smoke_masks[target_key],
    }])[0]
    error = float(np.max(np.abs(
        centered_scores(self_record) - centered_scores(smoke_scores[target_key])
    )))
    assert error < SMOKE_CENTERED_LOGPROB_TOLERANCE, (layer, error)
    assert self_record['patched_tokens'] == 1
    smoke_errors[layer] = error
print('Three diagnostic candidates:', diagnostic_info[target_key])
print('Final token ID:', token_ids[0])
print('Self parity by layer:', smoke_errors)
del smoke_hidden
torch.cuda.empty_cache()
```

```python
%%cell_progress 12/15 Capture or resume new families
cell_stage('Applying donor and equal-norm random directions at every depth')


# Include code, environment, exact pixels and prompt construction in resumption identity.
import inspect
import platform
import transformers
def fingerprint_sources():
    files = sorted((TOOLKIT_ROOT / 'src' / 'visual_torque_toolkit').rglob('*.py'))
    return {str(p.relative_to(TOOLKIT_ROOT)): sha256_file(p) for p in files}

image_hashes = {name: sha256_file(PRIOR_SCENE_DIR / name)
                for name in sorted(set(image_filenames))}
for name, expected_hash in image_hashes.items():
    assert sha256_file(RUNTIME_SCENE_DIR / name) == expected_hash
runtime_identity = {
    'implementation_sha256': IMPLEMENTATION_SHA256,
    'torch': torch.__version__, 'transformers': transformers.__version__,
    'dtype': MODEL_DTYPE,
    'capture_batch_size': globals().get('CAPTURE_BATCH_SIZE', 1),
    'patch_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'toolkit_files': fingerprint_sources(), 'image_hashes': image_hashes,
}

capture_identity = {
    'runtime_identity': runtime_identity,
    'schema_version': 'semantic-label-depth-capture-1.0',
    'protocol_identity': canonical_protocol_identity,
    'support_identity': support_identity,
    'model_id': MODEL_ID,
    'resolved_revision': adapter.metadata['resolved_revision'],
    'scan_layers': list(SCAN_LAYERS),
    'normalized_depths': list(NORMALIZED_DEPTHS),
    'scope': SITE_SCOPE,
    'records_per_family': RECORDS_PER_FAMILY,
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()
config_path = OUTPUT_ROOT / 'capture_config.json'
if config_path.is_file():
    previous = json.loads(config_path.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity
else:
    write_json_atomic(config_path, {
        'identity': capture_identity,
        'capture_identity_sha256': CAPTURE_SHA256,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })

completed_ids = set()
for path in sorted(SHARD_DIR.glob('*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    assert payload['family_id'] not in completed_ids, 'Duplicate family shard'
    completed_ids.add(payload['family_id'])
assert completed_ids <= {str(r['family_id']) for r in depth_support}
print('Completed / total:', len(completed_ids), len(depth_support))

remaining_families = sum(
    str(row['family_id']) not in completed_ids for row in depth_support
)
layer_progress = tqdm(
    total=remaining_families * len(SCAN_LAYERS),
    desc=f'{MODEL_KEY} intervention layers', unit='layer-family',
)
for row in tqdm(depth_support, desc=f'{MODEL_KEY} depth scan', unit='family'):
    family_id = str(row['family_id'])
    if family_id in completed_ids:
        continue
    reference_examples = build_reference_examples(row)
    diagnostic_examples, diagnostic_info = build_diagnostic_examples(row)
    all_examples = {**reference_examples, **diagnostic_examples}
    scores, hidden, masks = capture_examples(all_examples, SCAN_LAYERS)
    donor_deltas = {layer: {} for layer in SCAN_LAYERS}
    random_deltas = {layer: {} for layer in SCAN_LAYERS}
    donor_norms = {str(layer): {} for layer in SCAN_LAYERS}
    for layer in SCAN_LAYERS:
        for donor_task in DONOR_TASKS:
            base_key = f'ref__{donor_task}__base'
            source_key = f'ref__{donor_task}__aligned_source'
            delta = (
                hidden[source_key][layer][masks[source_key]]
                - hidden[base_key][layer][masks[base_key]]
            )
            donor_deltas[layer][donor_task] = delta
            donor_norms[str(layer)][donor_task] = selected_norm(delta)
            random_deltas[layer][donor_task] = matched_random(
                delta, f'depth:{family_id}:{layer}:{donor_task}'
            )

    patch_records = []
    for layer in SCAN_LAYERS:
        normalized_depth = (layer + 1) / EXPECTED_LAYER_COUNT
        for donor_task in DONOR_TASKS:
            reference_key = f'ref__{donor_task}__base'
            donor_norm = donor_norms[str(layer)][donor_task]
            identity_conditions = []
            for condition, payload in (
                ('identity_donor', donor_deltas[layer][donor_task]),
                ('identity_random', random_deltas[layer][donor_task]),
            ):
                identity_conditions.append({
                    'family_id': family_id, 'mapping_type': 'identity',
                    'variant': 'reference', 'donor_task': donor_task,
                    'target_task': donor_task, 'condition': condition,
                    'layer_index': layer,
                    'normalized_depth': normalized_depth,
                    'intervention_mode': 'add',
                    'payload_selected': payload,
                    'reference_payload_selected': donor_deltas[layer][donor_task],
                    'target_mask': masks[reference_key],
                    'donor_norm': donor_norm,
                    'applied_norm': selected_norm(payload),
                })
            patch_records.extend(score_conditions(
                reference_examples[reference_key], identity_conditions
            ))
            for variant in DIAGNOSTIC_VARIANTS:
                key = f'diag__{donor_task}__outcome__{variant}'
                conditions = []
                for condition, payload in (
                    ('diagnostic_donor', donor_deltas[layer][donor_task]),
                    ('diagnostic_random', random_deltas[layer][donor_task]),
                ):
                    conditions.append({
                        'family_id': family_id,
                        'mapping_type': 'diagnostic', 'variant': variant,
                        'donor_task': donor_task, 'target_task': 'outcome',
                        'condition': condition, 'layer_index': layer,
                        'normalized_depth': normalized_depth,
                        'intervention_mode': 'add',
                        'payload_selected': payload,
                    'reference_payload_selected': donor_deltas[layer][donor_task],
                        'target_mask': masks[key],
                        'donor_norm': donor_norm,
                        'applied_norm': selected_norm(payload),
                    })
                current = score_conditions(diagnostic_examples[key], conditions)
                for record in current:
                    record['diagnostics'] = diagnostic_info[key]
                    record['target_key'] = key
                patch_records.extend(current)
        layer_progress.update(1)
    assert len(patch_records) == RECORDS_PER_FAMILY
    write_json_atomic(SHARD_DIR / f'{family_id}.json', {
        'schema_version': 'semantic-label-depth-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'family_id': family_id,
        'baseline_scores': scores,
        'diagnostic_info': diagnostic_info,
        'donor_norms': donor_norms,
        'patch_records': patch_records,
    })
    del hidden
    torch.cuda.empty_cache()
layer_progress.close()

shards = sorted(SHARD_DIR.glob('*.json'))
assert len(shards) == EXPECTED_FAMILIES
print('Depth capture complete:', len(shards), 'families')
```

```python
%%cell_progress 13/15 Assemble original family estimands
cell_stage('Converting every layer intervention into family-level effects')
import gc
del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_score(record, semantic):
    semantics = list(record['candidate_semantics'])
    label = record['candidate_labels'][semantics.index(semantic)]
    return label_score(record, label)

def find_record(records, layer, donor_task, target_task, variant, condition):
    matches = [
        record for record in records
        if int(record['layer_index']) == int(layer)
        and record['donor_task'] == donor_task
        and record['target_task'] == target_task
        and record['variant'] == variant
        and record['condition'] == condition
    ]
    assert len(matches) == 1, (
        layer, donor_task, target_task, variant, condition, len(matches)
    )
    return matches[0]

estimand_rows = []
identity_rows = []
behavior_rows = []
for shard_path in tqdm(sorted(SHARD_DIR.glob('*.json')), desc='Estimands', unit='family'):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    family_id = payload['family_id']
    records = payload['patch_records']
    row = next(item for item in depth_support if item['family_id'] == family_id)
    for layer in SCAN_LAYERS:
        normalized_depth = (layer + 1) / EXPECTED_LAYER_COUNT
        for donor_task in DONOR_TASKS:
            base_key = f'ref__{donor_task}__base'
            base = payload['baseline_scores'][base_key]
            base_semantic = task_semantic(
                row['roles']['base']['features'], donor_task
            )
            source_semantic = donor_semantic_for(row, donor_task)
            base_margin = (
                semantic_score(base, source_semantic)
                - semantic_score(base, base_semantic)
            )
            donor_record = find_record(
                records, layer, donor_task, donor_task,
                'reference', 'identity_donor'
            )
            random_record = find_record(
                records, layer, donor_task, donor_task,
                'reference', 'identity_random'
            )
            identity_rows.append({
                'family_id': family_id, 'layer_index': layer,
                'normalized_depth': normalized_depth,
                'donor_task': donor_task,
                'effect': float(
                    (semantic_score(donor_record, source_semantic)
                     - semantic_score(donor_record, base_semantic) - base_margin)
                    - (semantic_score(random_record, source_semantic)
                       - semantic_score(random_record, base_semantic) - base_margin)
                ),
            })
            for variant in DIAGNOSTIC_VARIANTS:
                key = f'diag__{donor_task}__outcome__{variant}'
                baseline = payload['baseline_scores'][key]
                info = payload['diagnostic_info'][key]
                donor = find_record(
                    records, layer, donor_task, 'outcome',
                    variant, 'diagnostic_donor'
                )
                random = find_record(
                    records, layer, donor_task, 'outcome',
                    variant, 'diagnostic_random'
                )
                effects = {}
                for state_class, candidate_key in (
                    ('semantic', 'semantic_candidate'),
                    ('label', 'label_candidate'),
                    ('position', 'position_candidate'),
                ):
                    label = info[candidate_key]
                    effects[state_class] = float(
                        (label_score(donor, label) - label_score(baseline, label))
                        - (label_score(random, label) - label_score(baseline, label))
                    )
                estimand_rows.append({
                    'family_id': family_id, 'layer_index': layer,
                    'normalized_depth': normalized_depth,
                    'donor_task': donor_task, 'target_task': 'outcome',
                    'variant': variant,
                    'semantic_effect': effects['semantic'],
                    'label_effect': effects['label'],
                    'position_effect': effects['position'],
                    **info,
                })
                if layer == ENDPOINT_LAYER:
                    behavior_rows.append({
                        'family_id': family_id, 'donor_task': donor_task,
                        'variant': variant,
                        'correct': (
                            baseline['predicted_semantic']
                            == baseline['gold_semantic']
                        ),
                    })

estimand_frame = pd.DataFrame(estimand_rows)
identity_frame = pd.DataFrame(identity_rows)
behavior_frame = pd.DataFrame(behavior_rows)
assert len(estimand_frame) == (
    EXPECTED_FAMILIES * len(SCAN_LAYERS)
    * len(DONOR_TASKS) * len(DIAGNOSTIC_VARIANTS)
)
assert len(identity_frame) == (
    EXPECTED_FAMILIES * len(SCAN_LAYERS) * len(DONOR_TASKS)
)

family_rows = []
for (family_id, layer), frame in estimand_frame.groupby([
    'family_id', 'layer_index'
]):
    semantic = float(frame['semantic_effect'].mean())
    label = float(frame['label_effect'].mean())
    position = float(frame['position_effect'].mean())
    identity = float(identity_frame[
        (identity_frame['family_id'] == family_id)
        & (identity_frame['layer_index'] == layer)
    ]['effect'].mean())
    family_rows.append({
        'family_id': family_id, 'layer_index': int(layer),
        'normalized_depth': (int(layer) + 1) / EXPECTED_LAYER_COUNT,
        'identity_carrier': identity,
        'semantic_effect': semantic,
        'label_effect': label,
        'position_effect': position,
        'semantic_dominance': semantic - max(label, position),
        'label_dominance': label - max(semantic, position),
        'position_dominance': position - max(semantic, label),
        'label_minus_semantic': label - semantic,
    })
family_frame = pd.DataFrame(family_rows)
estimand_frame.to_csv(ANALYSIS_DIR / 'depth_estimands.csv', index=False)
family_frame.to_csv(ANALYSIS_DIR / 'family_depth_metrics.csv', index=False)
behavior_frame.to_csv(ANALYSIS_DIR / 'endpoint_behavior.csv', index=False)
print('Estimands / family-depth rows:', len(estimand_frame), len(family_frame))
```

```python
%%cell_progress 15/15 Global 37-test confirmation and fixed-endpoint pairwise comparisons
PRIMARY_METRICS=('identity_carrier','semantic_effect','label_effect','position_effect',
                 'semantic_dominance','label_dominance','position_dominance')
all_frames={};coverage=[];valid_by_model={}
for key,spec in MODEL_SPECS.items():
    folder=SHARED_ROOT/'models'/spec['model_slug']/'analysis'
    complete=folder/'completion.json';table=folder/'family_depth_metrics.csv'
    if not complete.is_file() or not table.is_file():
        coverage.append(dict(model=key,status='not_run'));continue
    marker=read_json(complete)
    assert marker['protocol']==digest(canonical_protocol_identity)
    assert marker['support']==support_identity and marker['implementation']==IMPLEMENTATION_SHA256
    assert marker['family_table_sha256']==file_hash(table)
    data=pd.read_csv(table)
    assert set(data.layer_index)==set(spec['scan_layers']) and not data.duplicated(['family_id','layer_index']).any()
    for layer,sub in data.groupby('layer_index'):assert set(sub.family_id)==set(support_identity['family_ids'])
    all_frames[key]=data;valid_by_model[key]=bool(marker['all_native_controls_valid'])
    coverage.append(dict(model=key,status='complete',all_native_controls_valid=valid_by_model[key]))
display(pd.DataFrame(coverage));pd.DataFrame(coverage).to_csv(CROSS_MODEL_DIR/'coverage.csv',index=False)
tests=[]
for key,data in all_frames.items():
    for layer,sub in data.groupby('layer_index'):
        for metric in PRIMARY_METRICS:
            tests.append(dict(model=key,layer=int(layer),metric=metric,**bootstrap(sub[metric].to_numpy(float),[key,layer,metric])))
    if len(MODEL_SPECS[key]['scan_layers'])==2:
        earlier,later=MODEL_SPECS[key]['scan_layers']
        pivot=data.assign(relative=data.label_effect-data.semantic_effect).pivot(index='family_id',columns='layer_index',values='relative')
        tests.append(dict(model=key,layer=-1,metric='later_minus_earlier_label_relative_to_semantic',
                          **bootstrap((pivot[later]-pivot[earlier]).to_numpy(float),[key,'two_site_difference'])))
test_frame=pd.DataFrame(tests);test_frame['q_value']=np.nan
complete=len(all_frames)==3
decisions=[]
if complete:
    assert len(test_frame)==37
    test_frame['q_value']=bh(test_frame.p_value)
    for key,spec in MODEL_SPECS.items():
        for layer in spec['scan_layers']:
            sub=test_frame[(test_frame.model==key)&(test_frame.layer==layer)].set_index('metric')
            passes=lambda metric:bool(sub.loc[metric,'lower']>.1 and sub.loc[metric,'q_value']<.05)
            classes=[c for c in ('semantic','label','position') if passes('identity_carrier') and passes(c+'_effect') and passes(c+'_dominance')]
            classification=('technical_control_mismatch' if not valid_by_model[key] else classes[0] if len(classes)==1 else 'unresolved')
            decisions.append(dict(model=key,layer=layer,classification=classification,
                scope='fixed_site_only_not_a_continuous_transition'))
    pairwise=[]
    for left,right in combinations(MODEL_SPECS,2):
        a=all_frames[left][all_frames[left].layer_index.eq(MODEL_SPECS[left]['endpoint_layer'])]
        b=all_frames[right][all_frames[right].layer_index.eq(MODEL_SPECS[right]['endpoint_layer'])]
        result=bootstrap(paired(a,b,'family_id','label_effect'),[left,right,'endpoint'])
        result['p_value']=min(1.,2*min(result['p_value'],1-result['p_value']+1/10001))
        pairwise.append(dict(left_model=left,right_model=right,**result))
    p=pd.DataFrame(pairwise);p['q_value']=bh(p.p_value);p.to_csv(CROSS_MODEL_DIR/'endpoint_pairwise_secondary.csv',index=False)
test_frame.to_csv(CROSS_MODEL_DIR/'fixed_37_tests.csv',index=False)
pd.DataFrame(decisions).to_csv(CROSS_MODEL_DIR/'fixed_site_classification.csv',index=False)
display(test_frame.round(4));display(pd.DataFrame(decisions))
save_json(CROSS_MODEL_DIR/'status.json',dict(complete=complete,expected_tests=37,available_tests=len(tests),
    classification=decisions,interpretation='Two fixed sites can show a content difference; this does not establish a continuous conversion chain.'))
lines=['# 固定位置的新家族确认','',f'已完成模型：{", ".join(all_frames)}。',
       '全部三模型完成后才统一校正预定的 37 个主指标；未完成时不提前给出阶段判定。',
       '单项效应、完整类别判据、两个位置的配对差异分别展示。',
       '样本是新家族；层位来自此前探索。两个位置不能代表完整连续转换轨迹。']
(CROSS_MODEL_DIR/'REPORT_ZH.md').write_text('\n'.join(lines)+'\n',encoding='utf-8')
print('Results:',CROSS_MODEL_DIR)
```
