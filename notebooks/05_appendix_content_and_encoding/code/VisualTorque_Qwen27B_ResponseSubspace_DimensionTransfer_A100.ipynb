```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 19 tracked cells.')
```

```python
%%cell_progress 8/19 Define held-out target prompts and counterbalancing
cell_stage('Defining the unchanged three-codebook target factorial')
import itertools
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = 'Which way will the balance scale tip?'
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}
ALL_SEMANTIC_PERMUTATIONS = tuple(itertools.permutations(SEMANTICS))

def mapping_by_codebook(codebook_key, semantics):
    codes = CODEBOOKS[codebook_key]
    return tuple({
        'ordinal': index + 1,
        'label': codes[index],
        'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def semantics_by_code(mapping, codebook_key):
    return tuple(
        next(item['semantic'] for item in mapping if item['label'] == code)
        for code in CODEBOOKS[codebook_key]
    )

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def ordinal_label(codebook_key, label):
    return f"L{CODEBOOKS[codebook_key].index(label) + 1}"

def ordinal_transition(codebook_key, source_label, desired_label):
    return (
        f'{ordinal_label(codebook_key, source_label)}'
        f'->{ordinal_label(codebook_key, desired_label)}'
    )

def ordinal_pair_from_transition(transition):
    source_label, desired_label = transition.split('->')
    return ''.join(sorted((source_label, desired_label)))

def line_order_for_positions(codes, source_label, desired_label,
                             source_position, desired_position):
    order = [None, None, None]
    order[source_position - 1] = source_label
    order[desired_position - 1] = desired_label
    remaining = next(
        label for label in codes
        if label not in (source_label, desired_label)
    )
    order[next(index for index, value in enumerate(order) if value is None)] = remaining
    return tuple(order)

def custom_prompt(mapping, line_order, codebook_key):
    item_by_label = {item['label']: item for item in mapping}
    options = '\n'.join(
        f"{label}. {SEMANTIC_TEXT[item_by_label[label]['semantic']]}"
        for label in line_order
    )
    valid = ', '.join(CODEBOOKS[codebook_key])
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT,
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one response '
        f'code and nothing else. Valid reply codes are: {valid}.',
    ])

def make_example(row, mapping, line_order, codebook_key, mapping_index):
    role = row['family']['roles']['base']
    features = role['features']
    codes = CODEBOOKS[codebook_key]
    semantics = semantics_by_code(mapping, codebook_key)
    gold = str(features['outcome'])
    return BehaviorExample(
        example_id=(
            f"dimension-transfer:{codebook_key}:{row['family_id']}:m{mapping_index}"
        ),
        question_id=(
            f"dimension_transfer_{codebook_key}_{row['family_id']}_m{mapping_index}"
        ),
        scene_id=str(role['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind='outcome',
        variant='response_subspace_dimension_transfer',
        split='heldout_target_confirmation',
        problem_type='response_subspace_dimension_transfer',
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=custom_prompt(mapping, line_order, codebook_key),
        image_path=(TARGET_RUNTIME_DIR / role['image_filename']).resolve(),
        labels=codes,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=codes[semantics.index(gold)],
        gold_semantic=gold,
    )

def target_metadata(row, codebook_key, mapping_index):
    family = row['family']
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions'][
        family['shortcut_type'] + '_only'
    ])
    desired = exact if row['cohort'] == 'rescue' else shortcut
    source_semantic = shortcut if row['cohort'] == 'rescue' else exact
    assert desired in ('left', 'right')
    assert source_semantic in ('left', 'right')
    assert desired != source_semantic
    mapping = mapping_by_codebook(
        codebook_key, ALL_SEMANTIC_PERMUTATIONS[mapping_index]
    )
    source_label = label_for(mapping, source_semantic)
    desired_label = label_for(mapping, desired)
    transition = ordinal_transition(codebook_key, source_label, desired_label)
    transition_index = ORDINAL_TRANSITIONS.index(transition)
    position_transition = POSITION_TRANSITIONS[
        (transition_index + row['counterbalance_phase']) % 6
    ]
    source_position, desired_position = (
        int(value[1:]) for value in position_transition.split('->')
    )
    line_order = line_order_for_positions(
        CODEBOOKS[codebook_key], source_label, desired_label,
        source_position, desired_position,
    )
    return {
        'mapping': mapping,
        'line_order': line_order,
        'source_semantic': source_semantic,
        'desired_semantic': desired,
        'ordinal_transition': transition,
        'ordinal_pair': ordinal_pair_from_transition(transition),
        'position_transition': position_transition,
    }

preflight_rows = []
for codebook_key in CODEBOOKS:
    metadata = target_metadata(target_manifest[0], codebook_key, 0)
    example = make_example(
        target_manifest[0], metadata['mapping'], metadata['line_order'],
        codebook_key, 0,
    )
    plan = adapter._candidate_plan(example)
    assert plan is not None, codebook_key
    _, token_by_label = plan
    token_ids = [int(token_by_label[label]) for label in CODEBOOKS[codebook_key]]
    assert len(set(token_ids)) == 3
    preflight_rows.append({
        'codebook': codebook_key,
        'labels': list(CODEBOOKS[codebook_key]),
        'token_ids': token_ids,
    })
display(pd.DataFrame(preflight_rows))
```

```python
%%cell_progress 9/19 Define capture and transfer-intervention helpers
cell_stage('Defining final-token capture, projection, and matched removal')

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(type(output).__name__)
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('Response codes must be single-token candidates.')
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

def score_record(prepared, logits):
    example = prepared['examples'][0]
    logprobs = torch.log_softmax(logits[0, -1, :].float(), dim=-1)
    scores = [
        float(logprobs[prepared['token_by_label'][label]].detach().cpu())
        for label in example.labels
    ]
    maximum = max(scores)
    weights = [math.exp(value - maximum) for value in scores]
    probabilities = [value / sum(weights) for value in weights]
    predicted = int(np.argmax(scores))
    return {
        'candidate_labels': list(example.labels),
        'candidate_semantics': list(example.option_semantics),
        'candidate_logprobs': scores,
        'candidate_probabilities': probabilities,
        'predicted_label': example.labels[predicted],
        'predicted_semantic': example.option_semantics[predicted],
        'gold_label': example.gold_label,
        'gold_semantic': example.gold_semantic,
    }

def capture_selected_hidden(example, layers):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    captured = {}
    handles = []
    for layer_index in layers:
        def hook(_module, _inputs, output, layer_index=layer_index):
            captured[layer_index] = output_hidden_tensor(
                output
            ).detach().cpu()[0][mask].float().clone()
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_record(prepared, output.logits), captured

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def unit_direction(tensor):
    norm = selected_norm(tensor)
    assert norm > 0
    return tensor.float() / norm

def orthonormal_basis(matrix, dimension):
    matrix = matrix.float().reshape(matrix.shape[0], -1)
    _, _, vh = torch.linalg.svd(matrix, full_matrices=False)
    basis = vh[:dimension].contiguous()
    identity = basis @ basis.T
    assert float(torch.max(torch.abs(
        identity - torch.eye(dimension)
    ))) < ORTHONORMAL_TOLERANCE
    return basis

def random_basis(reference_basis, key):
    dimension, hidden = reference_basis.shape
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    matrix = torch.randn(hidden, dimension, generator=generator)
    q, _ = torch.linalg.qr(matrix, mode='reduced')
    return q.T.contiguous()

def projection_component(delta, basis):
    basis = basis.float()
    return (delta.float() @ basis.T) @ basis

def captured_energy(delta, basis):
    numerator = torch.sum(projection_component(delta, basis) ** 2)
    denominator = torch.sum(delta.float() ** 2)
    assert float(denominator) > 0
    return float(numerator / denominator)

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def score_transfer_condition(example, clean_l55, condition):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    measurements = {}

    def l49_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        payload = condition['l49_payload'].to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        modified[0, current] = hidden[0, current] + payload
        return replace_output_hidden(output, modified)

    def l55_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        clean = clean_l55.to(hidden.device, torch.float32)
        basis_1d = condition['basis_1d'].to(hidden.device, torch.float32)
        basis_within = condition['basis_within'].to(hidden.device, torch.float32)
        basis_other = condition['basis_other'].to(hidden.device, torch.float32)
        random_within = condition['random_within'].to(hidden.device, torch.float32)
        random_other = condition['random_other'].to(hidden.device, torch.float32)
        delta = hidden[0, current].float() - clean
        component_1d = projection_component(delta, basis_1d)
        component_within = projection_component(delta, basis_within)
        component_other = projection_component(delta, basis_other)
        raw_random_within = projection_component(delta, random_within)
        raw_random_other = projection_component(delta, random_other)

        def norm_match(component, reference):
            return component * (
                torch.linalg.vector_norm(reference)
                / torch.clamp(torch.linalg.vector_norm(component), min=1e-12)
            )

        component_random_within = norm_match(
            raw_random_within, component_within
        )
        component_random_other = norm_match(raw_random_other, component_other)
        mode = condition['mode']
        if mode == 'total':
            selected = hidden[0, current].float()
        elif mode == 'remove_1d':
            selected = hidden[0, current].float() - component_1d
        elif mode == 'remove_within':
            selected = hidden[0, current].float() - component_within
        elif mode == 'remove_other':
            selected = hidden[0, current].float() - component_other
        elif mode == 'remove_random_within':
            selected = hidden[0, current].float() - component_random_within
        elif mode == 'remove_random_other':
            selected = hidden[0, current].float() - component_random_other
        elif mode == 'clean_clamp':
            selected = clean
        else:
            raise ValueError(mode)
        modified = hidden.clone()
        modified[0, current] = selected.to(hidden.dtype)
        measurements.update({
            'l55_delta_norm': selected_norm(delta),
            'component_1d_norm': selected_norm(component_1d),
            'component_within_norm': selected_norm(component_within),
            'component_other_norm': selected_norm(component_other),
            'component_random_within_norm': selected_norm(component_random_within),
            'component_random_other_norm': selected_norm(component_random_other),
        })
        return replace_output_hidden(output, modified)

    handles = [
        DECODER_LAYERS[SEMANTIC_LAYER].register_forward_hook(l49_hook),
        DECODER_LAYERS[LABEL_LAYER].register_forward_hook(l55_hook),
    ]
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_record(prepared, output.logits)
    record.update({
        key: value for key, value in condition.items()
        if key not in (
            'l49_payload', 'basis_1d', 'basis_within', 'basis_other',
            'random_within', 'random_other',
        )
    })
    record.update(measurements)
    record['patched_tokens'] = int(mask.sum())
    return record

print('Capture and transfer-intervention helpers ready.')
```

```python
%%cell_progress 10/19 Fit expanded within- and cross-alphabet basis libraries
cell_stage('Fitting d1 through d32 without target outputs')

def bootstrap_summary(values, key, replicates=BOOTSTRAP_REPLICATES):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    assert len(array)
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.choice(
        array, size=(replicates, len(array)), replace=True
    ).mean(axis=1)
    return {
        'n': int(len(array)),
        'mean': float(array.mean()),
        'ci_low': float(np.quantile(draws, ALPHA / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
    }

WITHIN_BASES_BY_DIM = {key: {} for key in CODEBOOKS}
OTHER_BASES_BY_DIM = {key: {} for key in CODEBOOKS}
geometry_rows = []
family_energy_rows = []

basis_jobs = [
    (target_codebook, pair)
    for target_codebook in CODEBOOKS for pair in ORDINAL_PAIRS
]
for target_codebook, pair in tqdm(
    basis_jobs, desc='Fitting expanded response bases', unit='cell'
):
    other_codebooks = tuple(
        key for key in CODEBOOKS if key != target_codebook
    )
    calibration = direction_groups[
        ('calibration', target_codebook, pair)
    ]
    confirmation = direction_groups[
        ('confirmation', target_codebook, pair)
    ]
    other_calibration = [
        record for key in other_codebooks
        for record in direction_groups[('calibration', key, pair)]
    ]
    other_confirmation = [
        record for key in other_codebooks
        for record in direction_groups[('confirmation', key, pair)]
    ]
    calibration_matrix = torch.cat([
        unit_direction(record['l55_delta']) for record in calibration
    ], dim=0)
    confirmation_matrix = torch.cat([
        unit_direction(record['l55_delta']) for record in confirmation
    ], dim=0)
    other_calibration_matrix = torch.cat([
        unit_direction(record['l55_delta']) for record in other_calibration
    ], dim=0)
    other_confirmation_matrix = torch.cat([
        unit_direction(record['l55_delta']) for record in other_confirmation
    ], dim=0)
    within_full = orthonormal_basis(
        calibration_matrix, max(CANDIDATE_DIMS)
    )
    within_confirmation_full = orthonormal_basis(
        confirmation_matrix, max(CANDIDATE_DIMS)
    )
    other_full = orthonormal_basis(
        other_calibration_matrix, max(CANDIDATE_DIMS)
    )
    other_confirmation_full = orthonormal_basis(
        other_confirmation_matrix, max(CANDIDATE_DIMS)
    )
    WITHIN_BASES_BY_DIM[target_codebook][pair] = {}
    OTHER_BASES_BY_DIM[target_codebook][pair] = {}
    for dimension in CANDIDATE_DIMS:
        within_basis = within_full[:dimension].contiguous()
        within_confirmation_basis = (
            within_confirmation_full[:dimension].contiguous()
        )
        other_basis = other_full[:dimension].contiguous()
        other_confirmation_basis = (
            other_confirmation_full[:dimension].contiguous()
        )
        WITHIN_BASES_BY_DIM[target_codebook][pair][dimension] = within_basis
        OTHER_BASES_BY_DIM[target_codebook][pair][dimension] = other_basis
        within_random = random_basis(
            within_basis,
            f'geometry-within:{target_codebook}:{pair}:d{dimension}',
        )
        other_random = random_basis(
            other_basis,
            f'geometry-other:{target_codebook}:{pair}:d{dimension}',
        )
        within_singular = torch.linalg.svdvals(
            within_basis @ within_confirmation_basis.T
        )
        other_singular = torch.linalg.svdvals(
            other_basis @ other_confirmation_basis.T
        )
        cross_singular = torch.linalg.svdvals(
            within_basis @ other_basis.T
        )
        by_family = defaultdict(list)
        for record in confirmation:
            delta = unit_direction(record['l55_delta'])
            by_family[str(record['family_id'])].append({
                'within': captured_energy(delta, within_basis),
                'other': captured_energy(delta, other_basis),
                'random_within': captured_energy(delta, within_random),
                'random_other': captured_energy(delta, other_random),
            })
        for family_id, values in by_family.items():
            row = {
                'family_id': family_id,
                'codebook': target_codebook,
                'pair': pair,
                'dimension': dimension,
            }
            for metric in (
                'within', 'other', 'random_within', 'random_other'
            ):
                row[metric + '_energy'] = float(np.mean([
                    value[metric] for value in values
                ]))
            family_energy_rows.append(row)
        current_family_rows = [
            row for row in family_energy_rows
            if row['codebook'] == target_codebook
            and row['pair'] == pair
            and row['dimension'] == dimension
        ]
        within_gains = [
            row['within_energy'] - row['random_within_energy']
            for row in current_family_rows
        ]
        other_gains = [
            row['other_energy'] - row['random_other_energy']
            for row in current_family_rows
        ]
        within_summary = bootstrap_summary(
            within_gains,
            f'within-geometry:{target_codebook}:{pair}:d{dimension}',
        )
        other_summary = bootstrap_summary(
            other_gains,
            f'other-geometry:{target_codebook}:{pair}:d{dimension}',
        )
        geometry_rows.append({
            'codebook': target_codebook,
            'pair': pair,
            'dimension': dimension,
            'within_energy': float(np.mean([
                row['within_energy'] for row in current_family_rows
            ])),
            'other_energy': float(np.mean([
                row['other_energy'] for row in current_family_rows
            ])),
            'random_within_energy': float(np.mean([
                row['random_within_energy'] for row in current_family_rows
            ])),
            'random_other_energy': float(np.mean([
                row['random_other_energy'] for row in current_family_rows
            ])),
            'within_gain': within_summary['mean'],
            'within_gain_ci_low': within_summary['ci_low'],
            'within_gain_ci_high': within_summary['ci_high'],
            'other_gain': other_summary['mean'],
            'other_gain_ci_low': other_summary['ci_low'],
            'other_gain_ci_high': other_summary['ci_high'],
            'within_stability': float(torch.mean(within_singular ** 2)),
            'other_stability': float(torch.mean(other_singular ** 2)),
            'within_other_overlap': float(torch.mean(cross_singular ** 2)),
        })

geometry_frame = pd.DataFrame(geometry_rows)
family_energy_frame = pd.DataFrame(family_energy_rows)
assert len(geometry_frame) == (
    len(CODEBOOKS) * len(ORDINAL_PAIRS) * len(CANDIDATE_DIMS)
)
print('Expanded geometry rows:', len(geometry_frame))
```

```python
%%cell_progress 11/19 Select dimensions and freeze cross-alphabet geometry
cell_stage('Selecting dimensions and measuring alphabet-excluded transfer')

SELECTED_DIM = {key: {} for key in CODEBOOKS}
SELECTION_PASS = {key: {} for key in CODEBOOKS}
selected_rows = []
transfer_rows = []
for (codebook_key, pair), current in geometry_frame.groupby(['codebook', 'pair']):
    current = current.sort_values('dimension').copy()
    ceiling_energy = float(
        current[current['dimension'] == max(CANDIDATE_DIMS)][
            'within_energy'
        ].iloc[0]
    )
    current['relative_to_ceiling'] = (
        current['within_energy'] / max(ceiling_energy, 1e-12)
    )
    eligible = current[
        (current['relative_to_ceiling'] >= RELATIVE_ENERGY_GATE)
        & (current['within_gain_ci_low'] > ENERGY_GAIN_CI_GATE)
        & (current['within_stability'] > SUBSPACE_STABILITY_GATE)
    ]
    if len(eligible):
        selected_dimension = int(eligible['dimension'].min())
        selection_pass = True
    else:
        selected_dimension = max(CANDIDATE_DIMS)
        selection_pass = False
    SELECTED_DIM[codebook_key][pair] = selected_dimension
    SELECTION_PASS[codebook_key][pair] = selection_pass
    for _, row in current.iterrows():
        selected_rows.append({
            **row.to_dict(),
            'selected': int(row['dimension']) == selected_dimension,
            'selection_pass': selection_pass,
        })
    selected = current[current['dimension'] == selected_dimension].iloc[0]
    family_selected = family_energy_frame[
        (family_energy_frame['codebook'] == codebook_key)
        & (family_energy_frame['pair'] == pair)
        & (family_energy_frame['dimension'] == selected_dimension)
    ]
    other_gains = (
        family_selected['other_energy']
        - family_selected['random_other_energy']
    )
    within_advantages = (
        family_selected['within_energy'] - family_selected['other_energy']
    )
    other_summary = bootstrap_summary(
        other_gains,
        f'selected-other:{codebook_key}:{pair}',
    )
    advantage_summary = bootstrap_summary(
        within_advantages,
        f'selected-within-advantage:{codebook_key}:{pair}',
    )
    transfer_rows.append({
        'codebook': codebook_key,
        'pair': pair,
        'selected_dimension': selected_dimension,
        'selection_pass': selection_pass,
        'within_energy': float(selected['within_energy']),
        'other_energy': float(selected['other_energy']),
        'other_gain': other_summary['mean'],
        'other_gain_ci_low': other_summary['ci_low'],
        'other_gain_ci_high': other_summary['ci_high'],
        'within_advantage': advantage_summary['mean'],
        'within_advantage_ci_low': advantage_summary['ci_low'],
        'within_advantage_ci_high': advantage_summary['ci_high'],
        'transfer_ratio': float(
            selected['other_energy'] / max(selected['within_energy'], 1e-12)
        ),
        'within_other_overlap': float(selected['within_other_overlap']),
        'geometry_transfer_pass': bool(
            other_summary['ci_low'] > GEOMETRY_TRANSFER_CI_GATE
        ),
    })

geometry_frame = pd.DataFrame(selected_rows)
transfer_geometry_frame = pd.DataFrame(transfer_rows)
selected_values = [
    SELECTED_DIM[key][pair] for key in CODEBOOKS for pair in ORDINAL_PAIRS
]
if all(value < max(CANDIDATE_DIMS) for value in selected_values):
    dimension_status = 'dimension_saturation_resolved_below_d32'
elif all(value == max(CANDIDATE_DIMS) for value in selected_values):
    dimension_status = 'dimension_ceiling_unresolved_at_d32'
else:
    dimension_status = 'dimension_saturation_partially_resolved'

subspace_payload = {
    'schema_version': 'response-subspace-expanded-bases-1.0',
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'candidate_dimensions': list(CANDIDATE_DIMS),
    'within_bases': WITHIN_BASES_BY_DIM,
    'other_bases': OTHER_BASES_BY_DIM,
    'selected_dim': SELECTED_DIM,
    'selection_pass': SELECTION_PASS,
}
selection_identity = {
    'schema_version': 'response-subspace-expanded-selection-identity-1.0',
    'protocol_identity': canonical_protocol_identity,
    'support_identity': support_identity,
    'selected_dim': SELECTED_DIM,
    'selection_pass': SELECTION_PASS,
    'dimension_status': dimension_status,
    'selected_without_target_outputs': True,
}
if SUBSPACE_TENSOR_PATH.is_file() and SUBSPACE_SELECTION_PATH.is_file():
    selection_payload = json.loads(
        SUBSPACE_SELECTION_PATH.read_text(encoding='utf-8')
    )
    assert selection_payload['identity'] == selection_identity
    SUBSPACE_TENSOR_SHA256 = sha256_file(SUBSPACE_TENSOR_PATH)
    assert selection_payload['subspace_tensor_sha256'] == SUBSPACE_TENSOR_SHA256
    print('Reusing the frozen expanded subspace library.')
else:
    assert not SUBSPACE_TENSOR_PATH.exists()
    assert not SUBSPACE_SELECTION_PATH.exists()
    temporary = SUBSPACE_TENSOR_PATH.with_suffix('.pt.tmp')
    torch.save(subspace_payload, temporary)
    temporary.replace(SUBSPACE_TENSOR_PATH)
    SUBSPACE_TENSOR_SHA256 = sha256_file(SUBSPACE_TENSOR_PATH)
    selection_payload = {
        'schema_version': 'response-subspace-expanded-selection-1.0',
        'identity': selection_identity,
        'subspace_tensor_sha256': SUBSPACE_TENSOR_SHA256,
        'geometry': geometry_frame.to_dict(orient='records'),
        'transfer_geometry': transfer_geometry_frame.to_dict(orient='records'),
    }
    write_json_atomic(SUBSPACE_SELECTION_PATH, selection_payload)
geometry_frame.to_csv(ANALYSIS_DIR / 'expanded_geometry.csv', index=False)
family_energy_frame.to_csv(
    ANALYSIS_DIR / 'expanded_family_energy.csv', index=False
)
transfer_geometry_frame.to_csv(
    ANALYSIS_DIR / 'cross_alphabet_geometry.csv', index=False
)
display(pd.DataFrame(SELECTED_DIM).T.rename_axis('codebook'))
display(transfer_geometry_frame.round(4))
print('Dimension status:', dimension_status)
```

```python
%%cell_progress 15/19 Assemble cell, family, and pair estimands
cell_stage('Releasing the model and calculating matched causal contrasts')
del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def lookup(records, mode):
    matches = [record for record in records if record['mode'] == mode]
    assert len(matches) == 1
    return matches[0]

def cycle_gap(frame, value):
    return float(
        frame[frame['position_transition'].isin(POSITION_FAVORED)][value].mean()
        - frame[frame['position_transition'].isin(POSITION_REVERSE)][value].mean()
    )

cell_rows = []
for path in tqdm(
    sorted(TARGET_SHARD_DIR.glob('*.json')),
    desc='Target estimands', unit='family',
):
    payload = json.loads(path.read_text(encoding='utf-8'))
    family = payload['family']
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions'][
        family['shortcut_type'] + '_only'
    ])
    multiplier = 1.0 if payload['cohort'] == 'rescue' else -1.0
    for cell in payload['cells']:
        clean_margin = (
            semantic_logprob(cell['baseline'], exact)
            - semantic_logprob(cell['baseline'], shortcut)
        )
        records = {mode: lookup(cell['records'], mode) for mode in TARGET_MODES}
        effects = {}
        for mode, record in records.items():
            margin = (
                semantic_logprob(record, exact)
                - semantic_logprob(record, shortcut)
            )
            effects[mode] = multiplier * (margin - clean_margin)
        attenuation = {
            mode: effects['total'] - effects[mode]
            for mode in TARGET_MODES if mode != 'total'
        }
        cell_rows.append({
            'family_id': payload['family_id'],
            'cohort': payload['cohort'],
            'diagnostic_cell': family['diagnostic_cell'],
            'codebook': cell['codebook'],
            'pair': cell['pair'],
            'transition': cell['transition'],
            'position_transition': cell['position_transition'],
            'selected_dimension': cell['selected_dimension'],
            'selection_pass': cell['selection_pass'],
            'effect_total': effects['total'],
            'attenuation_1d': attenuation['remove_1d'],
            'attenuation_within': attenuation['remove_within'],
            'attenuation_other': attenuation['remove_other'],
            'attenuation_random_within': attenuation['remove_random_within'],
            'attenuation_random_other': attenuation['remove_random_other'],
            'specificity_within': (
                attenuation['remove_within']
                - attenuation['remove_random_within']
            ),
            'specificity_other': (
                attenuation['remove_other']
                - attenuation['remove_random_other']
            ),
            'within_advantage': (
                attenuation['remove_within'] - attenuation['remove_other']
            ),
        })

cell_frame = pd.DataFrame(cell_rows)
assert len(cell_frame) == EXPECTED_TARGET_CELLS
cell_frame.to_csv(ANALYSIS_DIR / 'target_cell_estimands.csv', index=False)

family_rows = []
pair_rows = []
metric_columns = (
    'attenuation_1d', 'attenuation_within', 'attenuation_other',
    'attenuation_random_within', 'attenuation_random_other',
    'specificity_within', 'specificity_other', 'within_advantage',
)
for (family_id, codebook_key), current in cell_frame.groupby([
    'family_id', 'codebook'
]):
    assert len(current) == TARGET_MAPPINGS
    row = {
        'family_id': family_id,
        'codebook': codebook_key,
        'cohort': current['cohort'].iloc[0],
        'diagnostic_cell': current['diagnostic_cell'].iloc[0],
        'mean_total_effect': float(current['effect_total'].mean()),
        'position_cycle_other': cycle_gap(current, 'attenuation_other'),
    }
    row.update({
        'mean_' + metric: float(current[metric].mean())
        for metric in metric_columns
    })
    family_rows.append(row)
    for pair, pair_current in current.groupby('pair'):
        assert len(pair_current) == 2
        pair_row = {
            'family_id': family_id,
            'codebook': codebook_key,
            'pair': pair,
            'selected_dimension': int(
                pair_current['selected_dimension'].iloc[0]
            ),
            'selection_pass': bool(pair_current['selection_pass'].iloc[0]),
        }
        pair_row.update({
            metric: float(pair_current[metric].mean())
            for metric in metric_columns
        })
        pair_rows.append(pair_row)

family_frame = pd.DataFrame(family_rows)
pair_frame = pd.DataFrame(pair_rows)
assert len(family_frame) == EXPECTED_TARGET_FAMILIES * len(CODEBOOKS)
assert len(pair_frame) == (
    EXPECTED_TARGET_FAMILIES * len(CODEBOOKS) * len(ORDINAL_PAIRS)
)
family_frame.to_csv(ANALYSIS_DIR / 'target_family_estimands.csv', index=False)
pair_frame.to_csv(ANALYSIS_DIR / 'target_pair_estimands.csv', index=False)
display(cell_frame.groupby(['codebook', 'pair'])[[
    'attenuation_1d', 'attenuation_within', 'attenuation_other',
    'specificity_within', 'specificity_other', 'within_advantage',
]].mean().round(4))
```

```python
%%cell_progress 16/19 Test dimension saturation and cross-alphabet transfer
cell_stage('Running family bootstrap, sign-flip, and equivalence tests')

def signflip_p(values, key, alternative='greater'):
    array = np.asarray(values, dtype=float)
    observed = float(array.mean())
    generator = np.random.default_rng(
        int(stable_key(f'signflip:{key}')[:16], 16) % (2**63 - 1)
    )
    exceed = 0
    remaining = SIGNFLIP_REPLICATES
    while remaining:
        count = min(10000, remaining)
        signs = generator.choice((-1.0, 1.0), size=(count, len(array)))
        draws = (signs * array).mean(axis=1)
        if alternative == 'greater':
            exceed += int(np.count_nonzero(draws >= observed))
        else:
            exceed += int(np.count_nonzero(np.abs(draws) >= abs(observed)))
        remaining -= count
    return float((1 + exceed) / (SIGNFLIP_REPLICATES + 1))

test_rows = []
macro_metrics = (
    'mean_total_effect', 'mean_attenuation_1d',
    'mean_attenuation_within', 'mean_attenuation_other',
    'mean_specificity_within', 'mean_specificity_other',
    'mean_within_advantage', 'position_cycle_other',
)
for codebook_key in CODEBOOKS:
    current = family_frame[family_frame['codebook'] == codebook_key]
    for metric in macro_metrics:
        alternative = 'two-sided' if metric == 'position_cycle_other' else 'greater'
        summary = bootstrap_summary(
            current[metric], f'macro:{codebook_key}:{metric}'
        )
        test_rows.append({
            'test_family': 'codebook',
            'codebook': codebook_key,
            'pair': 'all',
            'metric': metric,
            **summary,
            'p_signflip': signflip_p(
                current[metric], f'macro:{codebook_key}:{metric}', alternative
            ),
        })

pair_metrics = (
    'attenuation_1d', 'attenuation_within', 'attenuation_other',
    'specificity_within', 'specificity_other', 'within_advantage',
)
for (codebook_key, pair), current in pair_frame.groupby(['codebook', 'pair']):
    for metric in pair_metrics:
        summary = bootstrap_summary(
            current[metric], f'pair:{codebook_key}:{pair}:{metric}'
        )
        test_rows.append({
            'test_family': 'pair',
            'codebook': codebook_key,
            'pair': pair,
            'metric': metric,
            **summary,
            'p_signflip': signflip_p(
                current[metric], f'pair:{codebook_key}:{pair}:{metric}'
            ),
        })
tests_frame = pd.DataFrame(test_rows)
tests_frame.to_csv(ANALYSIS_DIR / 'dimension_transfer_tests.csv', index=False)

def pair_test(codebook_key, pair, metric):
    row = tests_frame[
        (tests_frame['test_family'] == 'pair')
        & (tests_frame['codebook'] == codebook_key)
        & (tests_frame['pair'] == pair)
        & (tests_frame['metric'] == metric)
    ]
    assert len(row) == 1
    return row.iloc[0]

all_pairs = [
    (codebook_key, pair)
    for codebook_key in CODEBOOKS for pair in ORDINAL_PAIRS
]
positive_control_pass = all(
    pair_test(codebook_key, 'L1L2', 'attenuation_1d')['ci_low'] > 0
    for codebook_key in CODEBOOKS
)
within_specific_pairs = [
    (codebook_key, pair) for codebook_key, pair in all_pairs
    if pair_test(codebook_key, pair, 'specificity_within')['ci_low']
    > CAUSAL_SPECIFICITY_SESOI
]
cross_transfer_pairs = [
    (codebook_key, pair) for codebook_key, pair in all_pairs
    if pair_test(codebook_key, pair, 'specificity_other')['ci_low']
    > CAUSAL_SPECIFICITY_SESOI
]
codebook_residual_pairs = [
    (codebook_key, pair) for codebook_key, pair in all_pairs
    if pair_test(codebook_key, pair, 'within_advantage')['ci_low']
    > OWN_ADVANTAGE_SESOI
]
equivalent_pairs = [
    (codebook_key, pair) for codebook_key, pair in all_pairs
    if pair_test(codebook_key, pair, 'within_advantage')['ci_low']
    > -OWN_ADVANTAGE_EQUIVALENCE_BOUND
    and pair_test(codebook_key, pair, 'within_advantage')['ci_high']
    < OWN_ADVANTAGE_EQUIVALENCE_BOUND
]

if not positive_control_pass:
    primary_decision = 'l1l2_positive_control_failed'
elif set(within_specific_pairs) != set(all_pairs):
    primary_decision = 'expanded_within_subspaces_not_uniformly_causal'
elif set(cross_transfer_pairs) == set(all_pairs):
    if codebook_residual_pairs:
        primary_decision = 'shared_core_with_codebook_specific_residual'
    elif set(equivalent_pairs) == set(all_pairs):
        primary_decision = 'predominantly_alphabet_invariant_response_subspace'
    else:
        primary_decision = 'shared_core_with_mixed_specificity'
elif cross_transfer_pairs:
    primary_decision = 'pair_specific_cross_alphabet_transfer'
else:
    primary_decision = 'predominantly_codebook_specific_response_subspaces'

decision_payload = {
    'schema_version': 'response-subspace-dimension-transfer-decision-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'primary_decision': primary_decision,
    'dimension_status': dimension_status,
    'positive_control_pass': positive_control_pass,
    'selected_dim': SELECTED_DIM,
    'selection_pass': SELECTION_PASS,
    'within_specific_pairs': [list(value) for value in within_specific_pairs],
    'cross_transfer_pairs': [list(value) for value in cross_transfer_pairs],
    'codebook_residual_pairs': [list(value) for value in codebook_residual_pairs],
    'equivalent_pairs': [list(value) for value in equivalent_pairs],
}
display(tests_frame.round(4))
display(pd.DataFrame([decision_payload]).drop(columns=[
    'selected_dim', 'selection_pass', 'within_specific_pairs',
    'cross_transfer_pairs', 'codebook_residual_pairs', 'equivalent_pairs',
]))
```

```python
%%cell_progress 17/19 Audit supports, bases, shards, norms, and replay
cell_stage('Verifying all structural and numerical invariants')

assert len(direction_records) == EXPECTED_DIRECTION_RECORDS
direction_nonfinite = sum(
    not bool(torch.isfinite(record['l55_delta']).all())
    for record in direction_records
)
assert direction_nonfinite == 0

maximum_orthonormal_error = 0.0
for codebook_key in CODEBOOKS:
    for pair in ORDINAL_PAIRS:
        for dimension in CANDIDATE_DIMS:
            for basis in (
                WITHIN_BASES_BY_DIM[codebook_key][pair][dimension],
                OTHER_BASES_BY_DIM[codebook_key][pair][dimension],
            ):
                maximum_orthonormal_error = max(
                    maximum_orthonormal_error,
                    float(torch.max(torch.abs(
                        basis @ basis.T - torch.eye(dimension)
                    ))),
                )
assert maximum_orthonormal_error < ORTHONORMAL_TOLERANCE

target_cell_count = 0
target_record_count = 0
mode_counts = Counter()
target_nonfinite = 0
maximum_norm_relative_error = 0.0
for path in sorted(TARGET_SHARD_DIR.glob('*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == TARGET_CAPTURE_SHA256
    assert len(payload['cells']) == TARGET_CELLS_PER_FAMILY
    for cell in payload['cells']:
        target_cell_count += 1
        assert len(cell['records']) == len(TARGET_MODES)
        for record in cell['records']:
            target_record_count += 1
            mode_counts[record['mode']] += 1
            values = [
                *record['candidate_logprobs'], record['l55_delta_norm'],
                record['component_1d_norm'], record['component_within_norm'],
                record['component_other_norm'],
                record['component_random_within_norm'],
                record['component_random_other_norm'],
            ]
            if not np.isfinite(values).all():
                target_nonfinite += 1
            assert record['patched_tokens'] == 1
            for observed, reference in (
                (
                    record['component_random_within_norm'],
                    record['component_within_norm'],
                ),
                (
                    record['component_random_other_norm'],
                    record['component_other_norm'],
                ),
            ):
                relative = abs(observed - reference) / max(reference, 1e-12)
                maximum_norm_relative_error = max(
                    maximum_norm_relative_error, relative
                )
assert target_cell_count == EXPECTED_TARGET_CELLS
assert target_record_count == EXPECTED_TARGET_RECORDS
assert mode_counts == Counter({
    mode: EXPECTED_TARGET_CELLS for mode in TARGET_MODES
})
assert target_nonfinite == 0
assert maximum_norm_relative_error < NORM_MATCH_TOLERANCE

audit_payload = {
    'schema_version': 'response-subspace-dimension-transfer-audit-1.0',
    'support_identity': support_identity,
    'subspace_tensor_sha256': SUBSPACE_TENSOR_SHA256,
    'target_capture_sha256': TARGET_CAPTURE_SHA256,
    'direction_records_reused': len(direction_records),
    'target_families': EXPECTED_TARGET_FAMILIES,
    'target_cells': target_cell_count,
    'target_records': target_record_count,
    'mode_counts': dict(mode_counts),
    'clean_clamp_error': smoke_clamp_error,
    'maximum_orthonormal_error': maximum_orthonormal_error,
    'maximum_norm_relative_error': maximum_norm_relative_error,
    'direction_nonfinite': direction_nonfinite,
    'target_nonfinite': target_nonfinite,
    'direction_target_geometry_overlap': 0,
    'selected_without_target_outputs': True,
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_payload)
display(pd.DataFrame([audit_payload]))
```
