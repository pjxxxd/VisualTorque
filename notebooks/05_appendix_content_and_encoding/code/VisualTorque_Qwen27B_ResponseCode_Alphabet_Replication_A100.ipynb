```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 19 tracked cells.')
```

```python
%%cell_progress 6/19 Freeze disjoint supports, target split, and six-phase counterbalance
cell_stage('Selecting 42 targets and caching all required images')
import itertools
import math
import shutil

def canonical_signature(stacks):
    return tuple(sorted(
        (str(stack['side']), int(stack['distance']), int(stack['units']))
        for stack in stacks
    ))

direction_signatures = {
    canonical_signature(role['stacks'])
    for row in direction_support for role in row['roles'].values()
}
target_signatures = {
    canonical_signature(row['family']['roles']['base']['stacks'])
    for row in stable_manifest
}
assert not (direction_signatures & target_signatures)

def stratum(row):
    return (row['cohort'], str(row['family']['diagnostic_cell']))

by_stratum = defaultdict(list)
for row in stable_manifest:
    by_stratum[stratum(row)].append(row)
largest_stratum = sorted(
    by_stratum,
    key=lambda key: (-len(by_stratum[key]), stable_key(f'largest:{key}')),
)[0]
excluded_row = sorted(
    by_stratum[largest_stratum],
    key=lambda row: stable_key(f"exclude:{row['family_id']}"),
)[-1]
selected = [
    row for row in stable_manifest
    if row['family_id'] != excluded_row['family_id']
]
assert len(selected) == EXPECTED_TARGET_FAMILIES

def proportional_allocation(counts, total):
    population = sum(counts.values())
    raw = {key: total * value / population for key, value in counts.items()}
    allocated = {key: int(math.floor(value)) for key, value in raw.items()}
    remaining = total - sum(allocated.values())
    order = sorted(
        counts,
        key=lambda key: (-(raw[key] - allocated[key]), stable_key(key)),
    )
    for key in order[:remaining]:
        allocated[key] += 1
    assert sum(allocated.values()) == total
    assert all(0 <= allocated[key] <= counts[key] for key in counts)
    return allocated

selected_counts = Counter(stratum(row) for row in selected)
discovery_allocation = proportional_allocation(
    selected_counts, DISCOVERY_FAMILIES
)
target_manifest = []
for key in sorted(selected_counts):
    rows = sorted(
        [row for row in selected if stratum(row) == key],
        key=lambda row: stable_key(f"target-split:{row['family_id']}"),
    )
    boundary = discovery_allocation[key]
    for index, row in enumerate(rows):
        target_manifest.append({
            **row,
            'analysis_split': 'discovery' if index < boundary else 'confirmation',
        })
assert Counter(row['analysis_split'] for row in target_manifest) == {
    'discovery': DISCOVERY_FAMILIES,
    'confirmation': CONFIRMATION_FAMILIES,
}
for split in ('discovery', 'confirmation'):
    rows = sorted(
        [row for row in target_manifest if row['analysis_split'] == split],
        key=lambda row: stable_key(f"phase:{split}:{row['family_id']}"),
    )
    assert len(rows) % 6 == 0
    for index, row in enumerate(rows):
        row['counterbalance_phase'] = index % 6
    assert Counter(row['counterbalance_phase'] for row in rows) == {
        phase: len(rows) // 6 for phase in range(6)
    }
target_manifest.sort(key=lambda row: row['family_id'])

ALL_SEMANTIC_PERMUTATIONS = tuple(itertools.permutations(SEMANTICS))
ordered_direction_rows = sorted(
    direction_support,
    key=lambda row: stable_key(f"direction-map:{row['family_id']}"),
)
DIRECTION_MAPPING_INDEX = {
    row['family_id']: index % 6
    for index, row in enumerate(ordered_direction_rows)
}

def pair_for_semantics(semantics):
    left = semantics.index('left') + 1
    right = semantics.index('right') + 1
    return ''.join(sorted((f'L{left}', f'L{right}')))

direction_by_pair = defaultdict(list)
for row in direction_support:
    semantics = ALL_SEMANTIC_PERMUTATIONS[
        DIRECTION_MAPPING_INDEX[row['family_id']]
    ]
    direction_by_pair[pair_for_semantics(semantics)].append(row)
assert set(direction_by_pair) == set(ORDINAL_PAIRS)
calibration_allocation = proportional_allocation(
    {key: len(value) for key, value in direction_by_pair.items()}, 16
)
calibration_ids = set()
confirmation_ids = set()
for pair, rows in direction_by_pair.items():
    ordered = sorted(
        rows, key=lambda row: stable_key(f"direction-split:{row['family_id']}")
    )
    boundary = calibration_allocation[pair]
    calibration_ids.update(row['family_id'] for row in ordered[:boundary])
    confirmation_ids.update(row['family_id'] for row in ordered[boundary:])
assert len(calibration_ids) == len(confirmation_ids) == 16
assert not (calibration_ids & confirmation_ids)

DIRECTION_LINE_ORDER_INDEX = {}
for split_name, split_ids in (
    ('calibration', calibration_ids),
    ('confirmation', confirmation_ids),
):
    for pair in ORDINAL_PAIRS:
        rows = sorted(
            [
                row for row in direction_by_pair[pair]
                if row['family_id'] in split_ids
            ],
            key=lambda row: stable_key(
                f"direction-order:{split_name}:{pair}:{row['family_id']}"
            ),
        )
        for index, row in enumerate(rows):
            DIRECTION_LINE_ORDER_INDEX[row['family_id']] = index % 6
assert len(DIRECTION_LINE_ORDER_INDEX) == EXPECTED_DIRECTION_FAMILIES

direction_split_identity = {
    'schema_version': 'alphabet-direction-split-1.0',
    'calibration_ids': sorted(calibration_ids),
    'confirmation_ids': sorted(confirmation_ids),
    'mapping_index': dict(sorted(DIRECTION_MAPPING_INDEX.items())),
    'line_order_index': dict(sorted(DIRECTION_LINE_ORDER_INDEX.items())),
    'calibration_pair_counts': dict(Counter(
        pair_for_semantics(ALL_SEMANTIC_PERMUTATIONS[
            DIRECTION_MAPPING_INDEX[row['family_id']]
        ]) for row in direction_support if row['family_id'] in calibration_ids
    )),
    'confirmation_pair_counts': dict(Counter(
        pair_for_semantics(ALL_SEMANTIC_PERMUTATIONS[
            DIRECTION_MAPPING_INDEX[row['family_id']]
        ]) for row in direction_support if row['family_id'] in confirmation_ids
    )),
}
if DIRECTION_SPLIT_PATH.is_file():
    assert json.loads(DIRECTION_SPLIT_PATH.read_text(encoding='utf-8')) == direction_split_identity
else:
    write_json_atomic(DIRECTION_SPLIT_PATH, direction_split_identity)

support_identity = {
    'schema_version': 'response-code-alphabet-support-1.0',
    'prior_tensor_sha256': sha256_file(PRIOR_TENSOR_PATH),
    'position_decision_sha256': sha256_file(POSITION_DECISION_PATH),
    'depth_support_sha256': sha256_file(DEPTH_SUPPORT_PATH),
    'excluded_family_id': excluded_row['family_id'],
    'target_rows': [{
        'family_id': row['family_id'],
        'cohort': row['cohort'],
        'diagnostic_cell': row['family']['diagnostic_cell'],
        'analysis_split': row['analysis_split'],
        'counterbalance_phase': row['counterbalance_phase'],
    } for row in target_manifest],
    'direction_split': direction_split_identity,
    'geometrically_disjoint': True,
}
if SUPPORT_LOCK_PATH.is_file():
    previous = json.loads(SUPPORT_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == support_identity
else:
    write_json_atomic(SUPPORT_LOCK_PATH, {
        'identity': support_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

DIRECTION_RUNTIME_DIR = Path('/content/visualtorque_alphabet_direction_images_v1')
TARGET_RUNTIME_DIR = Path('/content/visualtorque_alphabet_target_images_v1')
DIRECTION_RUNTIME_DIR.mkdir(parents=True, exist_ok=True)
TARGET_RUNTIME_DIR.mkdir(parents=True, exist_ok=True)
direction_files = sorted({
    role['image_filename']
    for row in direction_support for role in row['roles'].values()
})
target_files = sorted({
    row['family']['roles']['base']['image_filename']
    for row in target_manifest
})
for filename in tqdm(direction_files, desc='Caching direction images', unit='image'):
    source = INVERSE_SCENE_DIR / filename
    destination = DIRECTION_RUNTIME_DIR / filename
    assert source.is_file(), source
    if not destination.is_file() or destination.stat().st_size != source.stat().st_size:
        shutil.copy2(source, destination)
for filename in tqdm(target_files, desc='Caching target images', unit='image'):
    source = TARGET_SCENE_DIR / filename
    destination = TARGET_RUNTIME_DIR / filename
    assert source.is_file(), source
    if not destination.is_file() or destination.stat().st_size != source.stat().st_size:
        shutil.copy2(source, destination)

print('Target split:', Counter(row['analysis_split'] for row in target_manifest))
print('Target strata:', Counter(
    (row['analysis_split'], row['cohort'], row['family']['diagnostic_cell'])
    for row in target_manifest
))
print('Direction pair split:', direction_split_identity['calibration_pair_counts'],
      direction_split_identity['confirmation_pair_counts'])
```

```python
%%cell_progress 8/19 Define prompts and verify every response code is one token
cell_stage('Building codebook prompts and checking candidate tokenization')
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = {
    'outcome': 'Which way will the balance scale tip?',
    'weight_more': 'Which side has more total weight?',
    'weight_less': 'Which side has less total weight?',
}
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

def task_semantic(features, task):
    if task == 'outcome':
        value = int(features['net_torque'])
    elif task == 'weight_more':
        value = int(features['weight_score'])
    elif task == 'weight_less':
        value = -int(features['weight_score'])
    else:
        raise KeyError(task)
    return outcome(value)

def mapping_by_codebook(codebook_key, semantics):
    codes = CODEBOOKS[codebook_key]
    assert set(semantics) == set(SEMANTICS)
    return tuple({
        'ordinal': index + 1,
        'label': codes[index],
        'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def semantics_by_code(mapping, codebook_key):
    return tuple(
        next(item['semantic'] for item in mapping if item['label'] == code)
        for code in CODEBOOKS[codebook_key]
    )

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def mapping_key(mapping):
    abbreviation = {'left': 'L', 'balanced': 'B', 'right': 'R'}
    return ''.join(abbreviation[item['semantic']] for item in mapping)

def custom_prompt(task, mapping, line_order, codebook_key):
    item_by_label = {item['label']: item for item in mapping}
    options = '\n'.join(
        f"{label}. {SEMANTIC_TEXT[item_by_label[label]['semantic']]}"
        for label in line_order
    )
    valid = ', '.join(CODEBOOKS[codebook_key])
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT[task],
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one response '
        f'code and nothing else. Valid reply codes are: {valid}.',
    ])

def make_example(row, role, task, mapping, line_order, codebook_key,
                 name, image_root, split):
    features = role['features']
    codes = CODEBOOKS[codebook_key]
    semantics = semantics_by_code(mapping, codebook_key)
    gold = task_semantic(features, task)
    return BehaviorExample(
        example_id=f"alphabet:{codebook_key}:{row['family_id']}:{name}",
        question_id=f"alphabet_{codebook_key}_{row['family_id']}_{name}",
        scene_id=str(role['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind=task,
        variant='response_code_alphabet_replication',
        split=split,
        problem_type='response_code_alphabet',
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=custom_prompt(task, mapping, line_order, codebook_key),
        image_path=(image_root / role['image_filename']).resolve(),
        labels=codes,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=codes[semantics.index(gold)],
        gold_semantic=gold,
    )

def direction_reference_mapping(row, codebook_key):
    semantics = ALL_SEMANTIC_PERMUTATIONS[
        DIRECTION_MAPPING_INDEX[row['family_id']]
    ]
    return mapping_by_codebook(codebook_key, semantics)

def build_direction_examples(row, codebook_key):
    mapping = direction_reference_mapping(row, codebook_key)
    codes = CODEBOOKS[codebook_key]
    line_order = tuple(itertools.permutations(codes))[
        DIRECTION_LINE_ORDER_INDEX[row['family_id']]
    ]
    examples = {}
    for donor_task in DONOR_TASKS:
        for role_name in ('base', 'aligned_source'):
            key = f'{donor_task}__{role_name}'
            examples[key] = make_example(
                row, row['roles'][role_name], donor_task,
                mapping, line_order, codebook_key, key,
                DIRECTION_RUNTIME_DIR, 'direction_support',
            )
    return examples, mapping

token_rows = []
preflight_row = direction_support[0]
for codebook_key in CODEBOOKS:
    examples, _ = build_direction_examples(preflight_row, codebook_key)
    example = next(iter(examples.values()))
    plan = adapter._candidate_plan(example)
    assert plan is not None, (
        f'{codebook_key} codes are not compatible with single-token scoring.'
    )
    fixed_prefix, token_by_label = plan
    token_ids = [int(token_by_label[label]) for label in example.labels]
    assert len(set(token_ids)) == 3, (codebook_key, token_ids)
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    pieces = [str(tokenizer.convert_ids_to_tokens(value)) for value in token_ids]
    for label, token_id, piece in zip(example.labels, token_ids, pieces):
        token_rows.append({
            'codebook': codebook_key, 'label': label,
            'token_id': token_id, 'token_piece': piece,
            'fixed_prefix': repr(fixed_prefix),
        })
token_frame = pd.DataFrame(token_rows)
display(token_frame)
print('All three codebooks passed single-token preflight.')
```

```python
%%cell_progress 9/19 Define clean capture and L49-to-L55 interventions
cell_stage('Defining capture, scoring, and axis-removal hooks')

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('Response codes must be single-token candidates.')
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {
        'examples': examples,
        'inputs': inputs,
        'token_by_label': token_by_label,
    }

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

def score_record(prepared, logits):
    example = prepared['examples'][0]
    logprobs = torch.log_softmax(logits[0, -1, :].float(), dim=-1)
    scores = [
        float(logprobs[prepared['token_by_label'][label]].detach().cpu())
        for label in example.labels
    ]
    maximum = max(scores)
    weights = [math.exp(value - maximum) for value in scores]
    probabilities = [value / sum(weights) for value in weights]
    predicted = int(np.argmax(scores))
    return {
        'candidate_labels': list(example.labels),
        'candidate_semantics': list(example.option_semantics),
        'candidate_logprobs': scores,
        'candidate_probabilities': probabilities,
        'predicted_label': example.labels[predicted],
        'predicted_semantic': example.option_semantics[predicted],
        'gold_label': example.gold_label,
        'gold_semantic': example.gold_semantic,
    }

def forward_capture(prepared, layers):
    hidden_by_layer = {}
    handles = []
    for layer_index in layers:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = output_hidden_tensor(
                output
            ).detach().cpu().clone()
        handles.append(
            DECODER_LAYERS[layer_index].register_forward_hook(hook)
        )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_record(prepared, output.logits), hidden_by_layer

def capture_examples(examples, layers):
    scores = {}
    hidden = {name: {} for name in examples}
    masks = {}
    for name, example in examples.items():
        prepared = prepare_examples([example])
        masks[name] = final_mask(prepared)
        current_score, current_hidden = forward_capture(prepared, layers)
        scores[name] = current_score
        for layer_index, tensor in current_hidden.items():
            hidden[name][layer_index] = tensor[0]
        del prepared, current_hidden
    return scores, hidden, masks

def capture_clean(example):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    score, hidden = forward_capture(prepared, (LABEL_LAYER,))
    return score, hidden[LABEL_LAYER][0][mask]

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def unit_direction(tensor):
    norm = selected_norm(tensor)
    assert norm > 0
    return tensor.float() / norm

def cosine_similarity(left, right):
    left = left.float().reshape(-1)
    right = right.float().reshape(-1)
    denominator = torch.linalg.vector_norm(left) * torch.linalg.vector_norm(right)
    assert float(denominator) > 0
    return float(torch.dot(left, right) / denominator)

def matched_random(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    return unit_direction(noise)

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def score_interface_condition(example, clean_l55, condition):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    measurements = {}

    def l49_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        payload = condition['l49_payload'].to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        modified[0, current] = hidden[0, current] + payload
        return replace_output_hidden(output, modified)

    def l55_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        clean = clean_l55.to(hidden.device, torch.float32)
        axis = condition['axis'].to(hidden.device, torch.float32)
        random_axis = condition['random_axis'].to(hidden.device, torch.float32)
        delta = hidden[0, current].float() - clean
        coefficient = torch.sum(delta * axis)
        component = coefficient * axis
        random_component = coefficient * random_axis
        mode = condition['mode']
        if mode == 'total':
            selected = hidden[0, current].float()
        elif mode == 'remove_axis':
            selected = hidden[0, current].float() - component
        elif mode == 'remove_random':
            selected = hidden[0, current].float() - random_component
        elif mode == 'clean_clamp':
            selected = clean
        else:
            raise ValueError(mode)
        modified = hidden.clone()
        modified[0, current] = selected.to(hidden.dtype)
        measurements.update({
            'axis_coefficient': float(coefficient.detach().cpu()),
            'axis_component_norm': float(
                torch.linalg.vector_norm(component).detach().cpu()
            ),
            'l55_delta_norm': float(
                torch.linalg.vector_norm(delta).detach().cpu()
            ),
        })
        return replace_output_hidden(output, modified)

    handles = [
        DECODER_LAYERS[SEMANTIC_LAYER].register_forward_hook(l49_hook),
        DECODER_LAYERS[LABEL_LAYER].register_forward_hook(l55_hook),
    ]
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_record(prepared, output.logits)
    record.update({
        key: value for key, value in condition.items()
        if key not in ('l49_payload', 'axis', 'random_axis')
    })
    record.update(measurements)
    record['patched_tokens'] = int(mask.sum())
    return record

print('Capture and intervention functions ready.')
```

```python
%%cell_progress 10/19 Calibrate and confirm one L55 axis library per codebook
cell_stage('Building or loading disjoint codebook-specific axis libraries')

def ordinal_label(codebook_key, label):
    return f"L{CODEBOOKS[codebook_key].index(label) + 1}"

def ordinal_transition(codebook_key, source_label, desired_label):
    return f'{ordinal_label(codebook_key, source_label)}->{ordinal_label(codebook_key, desired_label)}'

def ordinal_pair(transition):
    left, right = transition.split('->')
    return ''.join(sorted((left, right)))

direction_identity = {
    'schema_version': 'alphabet-direction-library-1.0',
    'model_id': MODEL_ID,
    'revision': MODEL_REVISION,
    'codebooks': {key: list(value) for key, value in CODEBOOKS.items()},
    'direction_split': direction_split_identity,
    'semantic_direction_sha256': sha256_file(PRIOR_TENSOR_PATH),
    'axis_rule': 'unit(mean_unit_forward - mean_unit_reverse)',
}

if DIRECTION_LIBRARY_PATH.is_file() and DIRECTION_TENSOR_PATH.is_file():
    direction_metadata = json.loads(
        DIRECTION_LIBRARY_PATH.read_text(encoding='utf-8')
    )
    assert direction_metadata['identity'] == direction_identity
    assert direction_metadata['tensor_sha256'] == sha256_file(DIRECTION_TENSOR_PATH)
    payload = torch.load(
        DIRECTION_TENSOR_PATH, map_location='cpu', weights_only=False
    )
    AXES = {
        codebook_key: {
            pair: tensor.float()
            for pair, tensor in payload['axes'][codebook_key].items()
        }
        for codebook_key in CODEBOOKS
    }
    geometry_frame = pd.DataFrame(direction_metadata['geometry'])
    print('Loaded cached alphabet direction library.')
else:
    FROZEN_SEMANTIC_UNIT = unit_direction(SEMANTIC_DIRECTION)
    calibration_units = {
        key: defaultdict(list) for key in CODEBOOKS
    }
    confirmation_units = {
        key: defaultdict(list) for key in CODEBOOKS
    }
    semantic_units = {
        key: {'calibration': [], 'confirmation': []}
        for key in CODEBOOKS
    }
    progress = tqdm(
        total=len(direction_support) * len(CODEBOOKS),
        desc='Codebook axis calibration', unit='family-codebook',
    )
    for row in direction_support:
        split = (
            'calibration' if row['family_id'] in calibration_ids
            else 'confirmation'
        )
        for codebook_key in CODEBOOKS:
            examples, mapping = build_direction_examples(row, codebook_key)
            _, hidden, masks = capture_examples(examples, CAPTURE_LAYERS)
            destination = (
                calibration_units[codebook_key]
                if split == 'calibration'
                else confirmation_units[codebook_key]
            )
            for donor_task in DONOR_TASKS:
                base_key = f'{donor_task}__base'
                source_key = f'{donor_task}__aligned_source'
                base_semantic = task_semantic(
                    row['roles']['base']['features'], donor_task
                )
                source_semantic = task_semantic(
                    row['roles']['aligned_source']['features'], donor_task
                )
                assert {base_semantic, source_semantic} == {'left', 'right'}
                base_label = label_for(mapping, base_semantic)
                source_label = label_for(mapping, source_semantic)
                transition = ordinal_transition(
                    codebook_key, base_label, source_label
                )
                code_delta = (
                    hidden[source_key][LABEL_LAYER][masks[source_key]]
                    - hidden[base_key][LABEL_LAYER][masks[base_key]]
                )
                destination[transition].append(unit_direction(code_delta))
                semantic_delta = (
                    hidden[source_key][SEMANTIC_LAYER][masks[source_key]]
                    - hidden[base_key][SEMANTIC_LAYER][masks[base_key]]
                )
                left_delta = (
                    semantic_delta if source_semantic == 'left'
                    else -semantic_delta
                )
                semantic_units[codebook_key][split].append(
                    unit_direction(left_delta)
                )
            del hidden
            torch.cuda.empty_cache()
            progress.update(1)
            progress.set_postfix_str(f"{row['family_id']} | {codebook_key}")
    progress.close()

    AXES = {key: {} for key in CODEBOOKS}
    confirmation_axes = {key: {} for key in CODEBOOKS}
    geometry_rows = []
    expected = set(ORDINAL_TRANSITIONS)
    for codebook_key in CODEBOOKS:
        assert set(calibration_units[codebook_key]) == expected
        assert set(confirmation_units[codebook_key]) == expected
        assert min(len(value) for value in calibration_units[codebook_key].values()) >= 5
        assert min(len(value) for value in confirmation_units[codebook_key].values()) >= 5
        raw_calibration = {
            transition: unit_direction(torch.stack(values).mean(dim=0))
            for transition, values in calibration_units[codebook_key].items()
        }
        raw_confirmation = {
            transition: unit_direction(torch.stack(values).mean(dim=0))
            for transition, values in confirmation_units[codebook_key].items()
        }
        for pair in ORDINAL_PAIRS:
            low, high = pair[:2], pair[2:]
            forward = f'{low}->{high}'
            reverse = f'{high}->{low}'
            axis = unit_direction(
                raw_calibration[forward] - raw_calibration[reverse]
            )
            confirmation_axis = unit_direction(
                raw_confirmation[forward] - raw_confirmation[reverse]
            )
            AXES[codebook_key][pair] = axis
            confirmation_axes[codebook_key][pair] = confirmation_axis
            geometry_rows.append({
                'codebook': codebook_key,
                'pair': pair,
                'calibration_opposition_cosine': cosine_similarity(
                    raw_calibration[forward], -raw_calibration[reverse]
                ),
                'confirmation_opposition_cosine': cosine_similarity(
                    raw_confirmation[forward], -raw_confirmation[reverse]
                ),
                'split_axis_cosine': cosine_similarity(
                    axis, confirmation_axis
                ),
                'calibration_forward_count': len(
                    calibration_units[codebook_key][forward]
                ),
                'calibration_reverse_count': len(
                    calibration_units[codebook_key][reverse]
                ),
                'confirmation_forward_count': len(
                    confirmation_units[codebook_key][forward]
                ),
                'confirmation_reverse_count': len(
                    confirmation_units[codebook_key][reverse]
                ),
                'semantic_alignment_calibration': cosine_similarity(
                    unit_direction(torch.stack(
                        semantic_units[codebook_key]['calibration']
                    ).mean(dim=0)), FROZEN_SEMANTIC_UNIT,
                ),
                'semantic_alignment_confirmation': cosine_similarity(
                    unit_direction(torch.stack(
                        semantic_units[codebook_key]['confirmation']
                    ).mean(dim=0)), FROZEN_SEMANTIC_UNIT,
                ),
            })
    geometry_frame = pd.DataFrame(geometry_rows)
    tensor_payload = {
        'schema_version': 'alphabet-direction-tensors-1.0',
        'model_id': MODEL_ID,
        'revision': MODEL_REVISION,
        'axes': {
            key: {pair: value.cpu() for pair, value in AXES[key].items()}
            for key in CODEBOOKS
        },
        'confirmation_axes': {
            key: {
                pair: value.cpu()
                for pair, value in confirmation_axes[key].items()
            }
            for key in CODEBOOKS
        },
    }
    temporary_tensor = DIRECTION_TENSOR_PATH.with_suffix('.pt.tmp')
    torch.save(tensor_payload, temporary_tensor)
    temporary_tensor.replace(DIRECTION_TENSOR_PATH)
    direction_metadata = {
        'identity': direction_identity,
        'tensor_sha256': sha256_file(DIRECTION_TENSOR_PATH),
        'geometry': geometry_frame.to_dict(orient='records'),
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    }
    write_json_atomic(DIRECTION_LIBRARY_PATH, direction_metadata)
    print('Saved new alphabet direction library.')

assert len(geometry_frame) == len(CODEBOOKS) * len(ORDINAL_PAIRS)
for codebook_key in CODEBOOKS:
    assert set(AXES[codebook_key]) == set(ORDINAL_PAIRS)
    assert all(
        abs(selected_norm(axis) - 1.0) < NORM_TOLERANCE
        for axis in AXES[codebook_key].values()
    )
DIRECTION_TENSOR_SHA256 = sha256_file(DIRECTION_TENSOR_PATH)
geometry_frame.to_csv(ANALYSIS_DIR / 'alphabet_axis_geometry.csv', index=False)
display(geometry_frame.round(4))
```

```python
%%cell_progress 11/19 Build the reduced code-by-position target factorial
cell_stage('Assigning every code transition to every position transition by phase')

def line_order_for_positions(codes, source_label, desired_label,
                             source_position, desired_position):
    assert source_position != desired_position
    order = [None, None, None]
    order[source_position - 1] = source_label
    order[desired_position - 1] = desired_label
    remaining_label = next(
        label for label in codes
        if label not in (source_label, desired_label)
    )
    remaining_position = next(index for index, value in enumerate(order) if value is None)
    order[remaining_position] = remaining_label
    assert set(order) == set(codes)
    return tuple(order)

def target_cell_metadata(row, codebook_key, mapping_index):
    family = row['family']
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions'][
        family['shortcut_type'] + '_only'
    ])
    assert {exact, shortcut} == {'left', 'right'}
    desired = exact if row['cohort'] == 'rescue' else shortcut
    source = shortcut if row['cohort'] == 'rescue' else exact
    mapping = mapping_by_codebook(
        codebook_key, ALL_SEMANTIC_PERMUTATIONS[mapping_index]
    )
    source_label = label_for(mapping, source)
    desired_label = label_for(mapping, desired)
    transition = ordinal_transition(
        codebook_key, source_label, desired_label
    )
    transition_index = ORDINAL_TRANSITIONS.index(transition)
    position_transition = POSITION_TRANSITIONS[
        (transition_index + int(row['counterbalance_phase'])) % 6
    ]
    source_position, desired_position = (
        int(value[1:]) for value in position_transition.split('->')
    )
    line_order = line_order_for_positions(
        CODEBOOKS[codebook_key], source_label, desired_label,
        source_position, desired_position,
    )
    assert line_order.index(source_label) + 1 == source_position
    assert line_order.index(desired_label) + 1 == desired_position
    return {
        'mapping': mapping,
        'mapping_key': mapping_key(mapping),
        'line_order': line_order,
        'line_order_key': '-'.join(line_order),
        'source_semantic': source,
        'desired_semantic': desired,
        'source_label': source_label,
        'desired_label': desired_label,
        'ordinal_transition': transition,
        'ordinal_pair': ordinal_pair(transition),
        'position_transition': position_transition,
        'source_position': source_position,
        'desired_position': desired_position,
    }

def make_target_example(row, codebook_key, mapping_index, metadata):
    return make_example(
        row, row['family']['roles']['base'], 'outcome',
        metadata['mapping'], metadata['line_order'], codebook_key,
        f"target__m{mapping_index}__p{row['counterbalance_phase']}",
        TARGET_RUNTIME_DIR, row['analysis_split'],
    )

design_rows = []
for row in target_manifest:
    for codebook_key in CODEBOOKS:
        for mapping_index in range(6):
            metadata = target_cell_metadata(
                row, codebook_key, mapping_index
            )
            design_rows.append({
                'family_id': row['family_id'],
                'analysis_split': row['analysis_split'],
                'counterbalance_phase': row['counterbalance_phase'],
                'codebook': codebook_key,
                'mapping_index': mapping_index,
                'ordinal_transition': metadata['ordinal_transition'],
                'position_transition': metadata['position_transition'],
            })
design_frame = pd.DataFrame(design_rows)
assert len(design_frame) == EXPECTED_TARGET_CELLS
for (split, codebook_key), current in design_frame.groupby([
    'analysis_split', 'codebook'
]):
    expected_repeats = (
        DISCOVERY_FAMILIES if split == 'discovery'
        else CONFIRMATION_FAMILIES
    ) // 6
    counts = current.groupby([
        'ordinal_transition', 'position_transition'
    ]).size()
    assert len(counts) == 36
    assert set(counts) == {expected_repeats}
design_frame.to_csv(ANALYSIS_DIR / 'target_factorial_design.csv', index=False)
display(design_frame.groupby([
    'analysis_split', 'codebook',
    'ordinal_transition', 'position_transition'
]).size().rename('n').reset_index().head(12))
print('Reduced factorial is exactly balanced in both splits.')
```

```python
%%cell_progress 14/19 Assemble cell-level and physical-family estimands
cell_stage('Releasing the model and computing matched attenuation contrasts')
import gc
del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def lookup(records, mode):
    matches = [record for record in records if record['mode'] == mode]
    assert len(matches) == 1
    return matches[0]

def cycle_gap(frame, column, favored, reverse, value):
    return float(
        frame[frame[column].isin(favored)][value].mean()
        - frame[frame[column].isin(reverse)][value].mean()
    )

estimand_rows = []
for path in tqdm(sorted(SHARD_DIR.glob('*.json')), desc='Alphabet estimands', unit='family'):
    payload = json.loads(path.read_text(encoding='utf-8'))
    family = payload['family']
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions'][
        family['shortcut_type'] + '_only'
    ])
    multiplier = 1.0 if payload['cohort'] == 'rescue' else -1.0
    for cell in payload['cells']:
        clean_margin = (
            semantic_logprob(cell['baseline'], exact)
            - semantic_logprob(cell['baseline'], shortcut)
        )
        records = {
            mode: lookup(cell['records'], mode)
            for mode in INTERVENTION_MODES
        }
        effects = {}
        for mode, record in records.items():
            margin = (
                semantic_logprob(record, exact)
                - semantic_logprob(record, shortcut)
            )
            effects[mode] = multiplier * (margin - clean_margin)
        source = cell['source_semantic']
        desired = cell['desired_semantic']
        estimand_rows.append({
            'family_id': payload['family_id'],
            'analysis_split': payload['analysis_split'],
            'counterbalance_phase': payload['counterbalance_phase'],
            'cohort': payload['cohort'],
            'diagnostic_cell': family['diagnostic_cell'],
            'codebook': cell['codebook'],
            'mapping_index': cell['mapping_index'],
            'mapping_key': cell['mapping_key'],
            'line_order_key': cell['line_order_key'],
            'ordinal_transition': cell['ordinal_transition'],
            'position_transition': cell['position_transition'],
            'clean_margin': clean_margin,
            'effect_total': effects['total'],
            'effect_remove_axis': effects['remove_axis'],
            'effect_remove_random': effects['remove_random'],
            'axis_attenuation': effects['total'] - effects['remove_axis'],
            'axis_specificity': effects['remove_random'] - effects['remove_axis'],
            'baseline_prediction': cell['baseline']['predicted_semantic'],
            'total_prediction': records['total']['predicted_semantic'],
            'removed_prediction': records['remove_axis']['predicted_semantic'],
            'actual_desired_flip': float(
                cell['baseline']['predicted_semantic'] == source
                and records['total']['predicted_semantic'] == desired
            ),
            'blocked_desired_flip': float(
                cell['baseline']['predicted_semantic'] == source
                and records['total']['predicted_semantic'] == desired
                and records['remove_axis']['predicted_semantic'] != desired
            ),
        })

estimand_frame = pd.DataFrame(estimand_rows)
assert len(estimand_frame) == EXPECTED_TARGET_CELLS
estimand_frame.to_csv(ANALYSIS_DIR / 'alphabet_cell_estimands.csv', index=False)

family_rows = []
for (family_id, codebook_key), current in estimand_frame.groupby([
    'family_id', 'codebook'
]):
    assert len(current) == MAPPINGS_PER_FAMILY
    assert set(current['ordinal_transition']) == set(ORDINAL_TRANSITIONS)
    assert set(current['position_transition']) == set(POSITION_TRANSITIONS)
    family_rows.append({
        'family_id': family_id,
        'analysis_split': current['analysis_split'].iloc[0],
        'counterbalance_phase': int(current['counterbalance_phase'].iloc[0]),
        'cohort': current['cohort'].iloc[0],
        'diagnostic_cell': current['diagnostic_cell'].iloc[0],
        'codebook': codebook_key,
        'mean_total_effect': float(current['effect_total'].mean()),
        'mean_axis_attenuation': float(current['axis_attenuation'].mean()),
        'mean_axis_specificity': float(current['axis_specificity'].mean()),
        'ordinal_cycle_gap': cycle_gap(
            current, 'ordinal_transition',
            PRIOR_ORDINAL_FAVORED, PRIOR_ORDINAL_REVERSE,
            'axis_attenuation',
        ),
        'position_cycle_gap': cycle_gap(
            current, 'position_transition',
            POSITION_FAVORED, POSITION_REVERSE,
            'axis_attenuation',
        ),
        'actual_flip_rate': float(current['actual_desired_flip'].mean()),
        'blocked_flip_rate': float(current['blocked_desired_flip'].mean()),
    })
family_frame = pd.DataFrame(family_rows)
assert len(family_frame) == EXPECTED_TARGET_FAMILIES * len(CODEBOOKS)
family_frame.to_csv(ANALYSIS_DIR / 'alphabet_family_estimands.csv', index=False)
display(estimand_frame.groupby([
    'analysis_split', 'codebook', 'ordinal_transition'
])[['effect_total', 'axis_attenuation', 'axis_specificity']].mean().round(4))
print('Physical-family estimands:', len(family_frame))
```

```python
%%cell_progress 16/19 Test held-out replication, reversal, and disappearance
cell_stage('Running family bootstrap and sign-flip tests on confirmation only')

def stable_key_local(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def bootstrap_summary(values, key, alpha=ALPHA):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    result = {
        'n': int(len(array)),
        'mean': float(array.mean()),
        'ci_low': np.nan,
        'ci_high': np.nan,
        'alpha': float(alpha),
    }
    if len(array) < 2:
        return result
    generator = np.random.default_rng(
        int(stable_key_local(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.choice(
        array, size=(BOOTSTRAP_REPLICATES, len(array)), replace=True
    ).mean(axis=1)
    result.update({
        'ci_low': float(np.quantile(draws, alpha / 2)),
        'ci_high': float(np.quantile(draws, 1 - alpha / 2)),
    })
    return result

def signflip_p(values, key, alternative='greater'):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    assert len(array)
    observed = float(array.mean())
    generator = np.random.default_rng(
        int(stable_key_local(f'signflip:{key}')[:16], 16) % (2**63 - 1)
    )
    exceed = 0
    remaining = SIGNFLIP_REPLICATES
    while remaining:
        count = min(10000, remaining)
        signs = generator.choice((-1.0, 1.0), size=(count, len(array)))
        draws = (signs * array).mean(axis=1)
        if alternative == 'greater':
            exceed += int(np.count_nonzero(draws >= observed))
        elif alternative == 'two-sided':
            exceed += int(np.count_nonzero(np.abs(draws) >= abs(observed)))
        else:
            raise ValueError(alternative)
        remaining -= count
    return float((1 + exceed) / (SIGNFLIP_REPLICATES + 1))

metric_alternatives = {
    'mean_total_effect': 'greater',
    'mean_axis_attenuation': 'greater',
    'mean_axis_specificity': 'greater',
    'ordinal_cycle_gap': 'two-sided',
    'native_cycle_gap': 'greater',
    'position_cycle_gap': 'two-sided',
}
test_rows = []
for split in ('discovery', 'confirmation'):
    for codebook_key in CODEBOOKS:
        current = family_frame[
            (family_frame['analysis_split'] == split)
            & (family_frame['codebook'] == codebook_key)
        ]
        for metric, alternative in metric_alternatives.items():
            values = current[metric]
            test_rows.append({
                'test_family': 'codebook',
                'analysis_split': split,
                'codebook': codebook_key,
                'metric': metric,
                **bootstrap_summary(values, f'{split}:{codebook_key}:{metric}'),
                'p_signflip': signflip_p(
                    values, f'{split}:{codebook_key}:{metric}', alternative
                ),
            })

confirmation_pivot = family_frame[
    family_frame['analysis_split'] == 'confirmation'
].pivot(index='family_id', columns='codebook', values='ordinal_cycle_gap')
assert not confirmation_pivot.isna().any().any()
for codebook_key in ('xyz', 'digits'):
    difference = confirmation_pivot[codebook_key] - confirmation_pivot['abc']
    test_rows.append({
        'test_family': 'paired_codebook',
        'analysis_split': 'confirmation',
        'codebook': codebook_key,
        'metric': 'ordinal_cycle_gap_minus_abc',
        **bootstrap_summary(difference, f'paired:{codebook_key}:minus_abc'),
        'p_signflip': signflip_p(
            difference, f'paired:{codebook_key}:minus_abc', 'two-sided'
        ),
    })

tests_frame = pd.DataFrame(test_rows)
tests_frame.to_csv(ANALYSIS_DIR / 'alphabet_tests.csv', index=False)

def confirmation_test(codebook_key, metric):
    row = tests_frame[
        (tests_frame['test_family'] == 'codebook')
        & (tests_frame['analysis_split'] == 'confirmation')
        & (tests_frame['codebook'] == codebook_key)
        & (tests_frame['metric'] == metric)
    ]
    assert len(row) == 1
    return row.iloc[0]

codebook_status = {}
codebook_flags = {}
for codebook_key in CODEBOOKS:
    ordinal = confirmation_test(codebook_key, 'ordinal_cycle_gap')
    native = confirmation_test(codebook_key, 'native_cycle_gap')
    position = confirmation_test(codebook_key, 'position_cycle_gap')
    specificity = confirmation_test(codebook_key, 'mean_axis_specificity')
    geometry = geometry_frame[geometry_frame['codebook'] == codebook_key]
    axis_reliable = bool(
        (geometry['split_axis_cosine'] > AXIS_SPLIT_COSINE_GATE).all()
        and (geometry['semantic_alignment_calibration'] > SEMANTIC_ALIGNMENT_GATE).all()
        and (geometry['semantic_alignment_confirmation'] > SEMANTIC_ALIGNMENT_GATE).all()
    )
    specific = bool(
        specificity['ci_low'] > 0
        and specificity['p_signflip'] < ALPHA
    )
    position_equivalent = bool(
        position['ci_low'] > -EQUIVALENCE_BOUND
        and position['ci_high'] < EQUIVALENCE_BOUND
    )
    if ordinal['ci_low'] > CYCLE_SESOI:
        status = 'same_ordinal_cycle'
    elif ordinal['ci_high'] < -CYCLE_SESOI:
        status = 'reversed_ordinal_cycle'
    elif (
        ordinal['ci_low'] > -EQUIVALENCE_BOUND
        and ordinal['ci_high'] < EQUIVALENCE_BOUND
    ):
        status = 'cycle_equivalent_to_zero'
    elif native['ci_low'] > CYCLE_SESOI:
        status = 'discovery_selected_cycle_replicates'
    else:
        status = 'cycle_inconclusive'
    codebook_status[codebook_key] = status
    codebook_flags[codebook_key] = {
        'axis_reliable': axis_reliable,
        'specific_component': specific,
        'position_cycle_equivalent': position_equivalent,
        'discovery_orientation': orientation_by_codebook[codebook_key],
        'status': status,
    }

all_reliable = all(
    value['axis_reliable'] and value['specific_component']
    for value in codebook_flags.values()
)
if not all_reliable:
    primary_decision = 'one_or_more_codebook_axes_not_reliable'
elif codebook_status['abc'] != 'same_ordinal_cycle':
    primary_decision = 'abc_cycle_not_replicated_under_new_design'
elif all(
    codebook_status[key] == 'same_ordinal_cycle'
    for key in ('xyz', 'digits')
):
    primary_decision = 'ordinal_cycle_generalizes_across_code_alphabets'
elif all(
    codebook_status[key] == 'cycle_equivalent_to_zero'
    for key in ('xyz', 'digits')
):
    primary_decision = 'abc_specific_response_code_geometry'
elif all(
    codebook_status[key] in (
        'same_ordinal_cycle', 'reversed_ordinal_cycle',
        'discovery_selected_cycle_replicates',
    ) for key in ('xyz', 'digits')
):
    primary_decision = 'codebook_adaptive_directed_geometry'
elif any(
    codebook_status[key] in (
        'same_ordinal_cycle', 'reversed_ordinal_cycle',
        'discovery_selected_cycle_replicates',
    ) for key in ('xyz', 'digits')
):
    primary_decision = 'alphabet_dependent_mixed_geometry'
else:
    primary_decision = 'new_alphabet_cycle_inconclusive'

decision_payload = {
    'schema_version': 'response-code-alphabet-decision-1.0',
    'model_id': MODEL_ID,
    'resolved_revision': MODEL_REVISION,
    'primary_decision': primary_decision,
    'codebook_status': codebook_status,
    'codebook_flags': codebook_flags,
    'confirmation_tests': {
        codebook_key: {
            metric: json.loads(
                confirmation_test(codebook_key, metric).to_json()
            )
            for metric in metric_alternatives
        }
        for codebook_key in CODEBOOKS
    },
}
display(tests_frame[tests_frame['analysis_split'] == 'confirmation'].round(4))
display(pd.DataFrame.from_dict(codebook_flags, orient='index'))
print('Primary decision:', primary_decision)
```

```python
%%cell_progress 17/19 Audit tokenization, supports, balance, geometry, and replay
cell_stage('Verifying every frozen structural and numerical invariant')

record_count = 0
cell_count = 0
mode_counts = Counter()
combination_counts = Counter()
transition_counts = Counter()
position_counts = Counter()
nonfinite = []
for path in sorted(SHARD_DIR.glob('*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['cells']) == CELLS_PER_FAMILY
    local = Counter()
    for cell in payload['cells']:
        cell_count += 1
        split = payload['analysis_split']
        codebook_key = cell['codebook']
        transition = cell['ordinal_transition']
        position = cell['position_transition']
        transition_counts[(split, codebook_key, transition)] += 1
        position_counts[(split, codebook_key, position)] += 1
        combination_counts[(split, codebook_key, transition, position)] += 1
        local[(codebook_key, transition)] += 1
        assert len(cell['records']) == RECORDS_PER_CELL
        for record in cell['records']:
            record_count += 1
            mode_counts[record['mode']] += 1
            values = [
                *record['candidate_logprobs'],
                record['axis_coefficient'],
                record['axis_component_norm'],
                record['l55_delta_norm'],
            ]
            if not np.isfinite(values).all():
                nonfinite.append(record['condition'])
            assert record['patched_tokens'] == 1
    assert all(
        local[(codebook_key, transition)] == 1
        for codebook_key in CODEBOOKS
        for transition in ORDINAL_TRANSITIONS
    )

assert cell_count == EXPECTED_TARGET_CELLS
assert record_count == EXPECTED_RECORDS
assert mode_counts == Counter({
    mode: EXPECTED_TARGET_CELLS for mode in INTERVENTION_MODES
})
assert not nonfinite
for split, family_count in (
    ('discovery', DISCOVERY_FAMILIES),
    ('confirmation', CONFIRMATION_FAMILIES),
):
    repeat = family_count // 6
    for codebook_key in CODEBOOKS:
        for transition in ORDINAL_TRANSITIONS:
            assert transition_counts[(split, codebook_key, transition)] == family_count
        for position in POSITION_TRANSITIONS:
            assert position_counts[(split, codebook_key, position)] == family_count
        for transition in ORDINAL_TRANSITIONS:
            for position in POSITION_TRANSITIONS:
                assert combination_counts[(
                    split, codebook_key, transition, position
                )] == repeat

maximum_axis_norm_error = max(
    abs(selected_norm(axis) - 1.0)
    for codebook_axes in AXES.values()
    for axis in codebook_axes.values()
)
audit_payload = {
    'schema_version': 'response-code-alphabet-audit-1.0',
    'support_identity': support_identity,
    'direction_tensor_sha256': DIRECTION_TENSOR_SHA256,
    'families': EXPECTED_TARGET_FAMILIES,
    'discovery_families': DISCOVERY_FAMILIES,
    'confirmation_families': CONFIRMATION_FAMILIES,
    'factorial_cells': cell_count,
    'intervention_records': record_count,
    'mode_counts': dict(mode_counts),
    'combination_count_discovery': DISCOVERY_FAMILIES // 6,
    'combination_count_confirmation': CONFIRMATION_FAMILIES // 6,
    'single_token_table': token_frame.to_dict(orient='records'),
    'clean_clamp_error': smoke_clamp_error,
    'maximum_axis_norm_error': maximum_axis_norm_error,
    'nonfinite_measurements': len(nonfinite),
    'direction_target_geometry_overlap': 0,
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_payload)
display(pd.DataFrame([audit_payload]))
```
