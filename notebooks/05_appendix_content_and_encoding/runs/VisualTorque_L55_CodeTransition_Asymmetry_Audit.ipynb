```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 10 tracked cells.')
```

```python
%%cell_progress 3/10 Lock source files and the post-hoc audit protocol
cell_stage('Freezing inputs, transition definitions, and diagnostic gates')

SEED = 20260904
BOOTSTRAP_REPLICATES = 20000
SIGNFLIP_REPLICATES = 100000
ALPHA = 0.05
LABELS = ('A', 'B', 'C')
MAPPING_VARIANTS = ('forward', 'reverse')
DIRECTIONS = ('exact', 'shortcut')
MODES = ('total', 'remove_code', 'remove_stale', 'remove_random', 'code_only')
EXPECTED_FAMILIES = 48
EXPECTED_CURRENT_RECORDS = 960

PRIOR_ROOT = PROJECT_ROOT / 'results' / 'qwen27b_semantic_error_mediation_v1'
PRIOR_SHARD_DIR = PRIOR_ROOT / 'target_shards'
PRIOR_DECISION_PATH = PRIOR_ROOT / 'analysis' / 'semantic_error_decision.json'
CURRENT_ROOT = PROJECT_ROOT / 'results' / 'qwen27b_l55_code_subspace_mediation_v1'
CURRENT_SHARD_DIR = CURRENT_ROOT / 'family_shards'
CURRENT_DECISION_PATH = CURRENT_ROOT / 'analysis' / 'code_subspace_decision.json'
CURRENT_AUDIT_PATH = CURRENT_ROOT / 'analysis' / 'robustness_audit.json'

OUTPUT_ROOT = (
    PROJECT_ROOT / 'results' / 'qwen27b_l55_code_transition_asymmetry_audit_v1'
)
ANALYSIS_DIR = OUTPUT_ROOT / 'analysis'
FIGURE_DIR = ANALYSIS_DIR / 'figures'
for path in (OUTPUT_ROOT, ANALYSIS_DIR, FIGURE_DIR):
    path.mkdir(parents=True, exist_ok=True)
PROTOCOL_LOCK_PATH = OUTPUT_ROOT / 'frozen_protocol.json'

from visualtorque_checks import sha256_file

def write_json_atomic(path, value):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    temporary.write_text(
        json.dumps(value, indent=2, ensure_ascii=False) + '\n',
        encoding='utf-8',
    )
    temporary.replace(path)

for required in (
    PRIOR_SHARD_DIR, PRIOR_DECISION_PATH,
    CURRENT_SHARD_DIR, CURRENT_DECISION_PATH, CURRENT_AUDIT_PATH,
):
    assert Path(required).exists(), f'Missing prerequisite: {required}'

protocol_identity = {
    'schema_version': 'l55-code-transition-asymmetry-audit-1.0',
    'analysis_status': 'post_hoc_explanatory_audit',
    'prior_decision_sha256': sha256_file(PRIOR_DECISION_PATH),
    'current_decision_sha256': sha256_file(CURRENT_DECISION_PATH),
    'current_audit_sha256': sha256_file(CURRENT_AUDIT_PATH),
    'transition_order': list(LABELS),
    'orientation_rule': 'descending iff target label index < source label index',
    'scientific_cohort': (
        'intersection: same stable exact/shortcut group in both completed runs'
    ),
    'primary_diagnostic': (
        'family-paired descending minus ascending current-code attenuation'
    ),
    'claim_limit': (
        'behavioral transition asymmetry only; saved shards do not contain '
        'direction tensors, so estimator geometry and model geometry remain '
        'unresolved'
    ),
    'bootstrap_replicates': BOOTSTRAP_REPLICATES,
    'signflip_replicates': SIGNFLIP_REPLICATES,
    'seed': SEED,
}
if PROTOCOL_LOCK_PATH.is_file():
    previous = json.loads(PROTOCOL_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == protocol_identity
else:
    write_json_atomic(PROTOCOL_LOCK_PATH, {
        'identity': protocol_identity,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })
print('Output:', OUTPUT_ROOT)
```

```python
%%cell_progress 4/10 Load and verify both completed experiments
cell_stage('Reading 96 locked family shards and completed decisions')

def read_shards(directory):
    payloads = {}
    for path in tqdm(sorted(Path(directory).glob('*.json')), desc=Path(directory).name):
        payload = json.loads(path.read_text(encoding='utf-8'))
        family_id = str(payload['family_id'])
        assert family_id not in payloads
        payloads[family_id] = payload
    return payloads

prior_by_id = read_shards(PRIOR_SHARD_DIR)
current_by_id = read_shards(CURRENT_SHARD_DIR)
assert len(prior_by_id) == len(current_by_id) == EXPECTED_FAMILIES
assert set(prior_by_id) == set(current_by_id)

prior_decision = json.loads(PRIOR_DECISION_PATH.read_text(encoding='utf-8'))
current_decision = json.loads(CURRENT_DECISION_PATH.read_text(encoding='utf-8'))
current_audit = json.loads(CURRENT_AUDIT_PATH.read_text(encoding='utf-8'))
assert prior_decision['primary_decision'] == (
    'bidirectional_semantic_error_control_with_code_mediation'
)
assert current_decision['primary_decision'] == (
    'no_specific_l55_code_subspace_mediation'
)
assert int(current_audit['records']) == EXPECTED_CURRENT_RECORDS
assert current_audit['full_clean_clamp_error'] == 0
print('Prior decision:', prior_decision['primary_decision'])
print('Current decision:', current_decision['primary_decision'])
print('Common families:', len(current_by_id))
```

```python
%%cell_progress 5/10 Reconstruct cohorts, margins, transitions, and flips
cell_stage('Building one row per family, mapping, and physical direction')

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def group_for(payload, exact, shortcut):
    predictions = tuple(
        str(payload['baselines'][variant]['predicted_semantic'])
        for variant in MAPPING_VARIANTS
    )
    if all(value == shortcut for value in predictions):
        return 'stable_shortcut_error'
    if all(value == exact for value in predictions):
        return 'stable_exact_correct'
    return 'mapping_sensitive_or_other'

def current_lookup(payload, variant, direction, mode):
    matches = [
        row for row in payload['records']
        if row['mapping_variant'] == variant
        and row['direction'] == direction and row['mode'] == mode
    ]
    assert len(matches) == 1
    return matches[0]

def prior_lookup(payload, variant, intervention):
    matches = [
        row for row in payload['patch_records']
        if row['mapping_variant'] == variant
        and row['intervention'] == intervention
    ]
    assert len(matches) == 1
    return matches[0]

rows = []
family_status_rows = []
for family_id in tqdm(sorted(current_by_id), desc='Reconstructing audit rows', unit='family'):
    current = current_by_id[family_id]
    prior = prior_by_id[family_id]
    family = current['family']
    exact = str(family['roles']['base']['features']['outcome'])
    shortcut = str(family['roles']['base']['features']['strategy_predictions'][
        family['shortcut_type'] + '_only'
    ])
    prior_group = group_for(prior, exact, shortcut)
    current_group = group_for(current, exact, shortcut)
    family_status_rows.append({
        'family_id': family_id,
        'diagnostic_cell': family['diagnostic_cell'],
        'prior_group': prior_group, 'current_group': current_group,
        'group_stable': prior_group == current_group,
    })
    for variant in MAPPING_VARIANTS:
        current_clean = current['baselines'][variant]
        prior_clean = prior['baselines'][variant]
        current_clean_margin = (
            semantic_logprob(current_clean, exact)
            - semantic_logprob(current_clean, shortcut)
        )
        prior_clean_margin = (
            semantic_logprob(prior_clean, exact)
            - semantic_logprob(prior_clean, shortcut)
        )
        for direction, desired, multiplier, prior_names in (
            ('exact', exact, 1.0, ('l49_semantic_exact', 'l49_random_exact')),
            ('shortcut', shortcut, -1.0, ('l49_semantic_shortcut', 'l49_random_shortcut')),
        ):
            records = {
                mode: current_lookup(current, variant, direction, mode)
                for mode in MODES
            }
            transition = tuple(records['total']['current_transition'])
            assert len(transition) == 2 and transition[0] != transition[1]
            source_index = LABELS.index(transition[0])
            target_index = LABELS.index(transition[1])
            orientation = 'descending' if target_index < source_index else 'ascending'
            effects = {}
            for mode, record in records.items():
                margin = (
                    semantic_logprob(record, exact)
                    - semantic_logprob(record, shortcut)
                )
                effects[mode] = multiplier * (margin - current_clean_margin)
            prior_semantic = prior_lookup(prior, variant, prior_names[0])
            prior_random = prior_lookup(prior, variant, prior_names[1])
            expected_group = (
                'stable_shortcut_error' if direction == 'exact'
                else 'stable_exact_correct'
            )
            rows.append({
                'family_id': family_id,
                'diagnostic_cell': family['diagnostic_cell'],
                'mapping_variant': variant, 'direction': direction,
                'prior_group': prior_group, 'current_group': current_group,
                'intersection_eligible': bool(
                    prior_group == current_group == expected_group
                ),
                'transition': '→'.join(transition),
                'unordered_pair': ''.join(sorted(transition)),
                'orientation': orientation,
                'prior_clean_margin': prior_clean_margin,
                'current_clean_margin': current_clean_margin,
                'clean_margin_change': current_clean_margin - prior_clean_margin,
                **{f'effect_{mode}': value for mode, value in effects.items()},
                'attenuation_code': effects['total'] - effects['remove_code'],
                'specificity_random': effects['remove_random'] - effects['remove_code'],
                'specificity_stale': effects['remove_stale'] - effects['remove_code'],
                'current_total_flip': float(records['total']['predicted_semantic'] == desired),
                'current_removed_flip': float(records['remove_code']['predicted_semantic'] == desired),
                'prior_semantic_flip': float(prior_semantic['predicted_semantic'] == desired),
                'prior_random_flip': float(prior_random['predicted_semantic'] == desired),
                'code_component_norm': float(records['total']['code_component_norm']),
            })

audit_frame = pd.DataFrame(rows)
family_status_frame = pd.DataFrame(family_status_rows)
assert len(audit_frame) == EXPECTED_FAMILIES * 4
audit_frame.to_csv(ANALYSIS_DIR / 'transition_level_estimands.csv', index=False)
family_status_frame.to_csv(ANALYSIS_DIR / 'cohort_stability.csv', index=False)
print('Transition rows:', len(audit_frame))
print('Intersection-eligible rows:', int(audit_frame['intersection_eligible'].sum()))
```

```python
%%cell_progress 6/10 Audit cross-run cohort stability
cell_stage('Identifying switched families and decision-boundary sensitivity')

confusion = pd.crosstab(
    family_status_frame['prior_group'], family_status_frame['current_group']
)
switched = family_status_frame[~family_status_frame['group_stable']].copy()
margin_summary = audit_frame.groupby('family_id').agg(
    minimum_prior_abs_margin=('prior_clean_margin', lambda x: float(np.min(np.abs(x)))),
    minimum_current_abs_margin=('current_clean_margin', lambda x: float(np.min(np.abs(x)))),
    maximum_abs_margin_change=('clean_margin_change', lambda x: float(np.max(np.abs(x)))),
).reset_index()
switched = switched.merge(margin_summary, on='family_id', how='left')
switched.to_csv(ANALYSIS_DIR / 'switched_families.csv', index=False)

intersection_counts = audit_frame[
    audit_frame['intersection_eligible']
].groupby(['direction', 'diagnostic_cell'])['family_id'].nunique().rename('n').reset_index()
display(confusion)
display(switched.round(4))
display(intersection_counts)
print('Stable families:', int(family_status_frame['group_stable'].sum()), '/', EXPECTED_FAMILIES)
```

```python
%%cell_progress 7/10 Test ordered-label transition asymmetry
cell_stage('Bootstrapping families and running paired sign-flip tests')

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def bootstrap_summary(values, key):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    result = {'n': int(len(array)), 'mean': float(array.mean()) if len(array) else np.nan,
              'ci_low': np.nan, 'ci_high': np.nan}
    if len(array) < 2:
        return result
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.choice(
        array, size=(BOOTSTRAP_REPLICATES, len(array)), replace=True
    ).mean(axis=1)
    result.update({
        'ci_low': float(np.quantile(draws, ALPHA / 2)),
        'ci_high': float(np.quantile(draws, 1 - ALPHA / 2)),
    })
    return result

def signflip_p(values, key):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    if not len(array):
        return np.nan
    observed = float(array.mean())
    generator = np.random.default_rng(
        int(stable_key(f'signflip:{key}')[:16], 16) % (2**63 - 1)
    )
    exceed = 0
    remaining = SIGNFLIP_REPLICATES
    while remaining:
        count = min(10000, remaining)
        signs = generator.choice((-1.0, 1.0), size=(count, len(array)))
        exceed += int(np.count_nonzero((signs * array).mean(axis=1) >= observed))
        remaining -= count
    return float((1 + exceed) / (SIGNFLIP_REPLICATES + 1))

scientific = audit_frame[audit_frame['intersection_eligible']].copy()
orientation_family = scientific.groupby([
    'family_id', 'direction', 'orientation'
]).agg(
    attenuation_code=('attenuation_code', 'mean'),
    effect_total=('effect_total', 'mean'),
    effect_code_only=('effect_code_only', 'mean'),
    specificity_random=('specificity_random', 'mean'),
    literal_flip=('prior_semantic_flip', 'mean'),
    random_flip=('prior_random_flip', 'mean'),
).reset_index()

gap_rows = []
for direction in DIRECTIONS:
    current = orientation_family[orientation_family['direction'] == direction]
    pivot = current.pivot(index='family_id', columns='orientation', values='attenuation_code')
    assert {'ascending', 'descending'} <= set(pivot.columns)
    for family_id, row in pivot.dropna().iterrows():
        gap_rows.append({
            'family_id': family_id, 'direction': direction,
            'descending_attenuation': float(row['descending']),
            'ascending_attenuation': float(row['ascending']),
            'descending_minus_ascending': float(row['descending'] - row['ascending']),
        })
gap_frame = pd.DataFrame(gap_rows)
gap_frame.to_csv(ANALYSIS_DIR / 'paired_orientation_gaps.csv', index=False)

test_rows = []
for direction in DIRECTIONS:
    current = gap_frame[gap_frame['direction'] == direction]
    for metric in ('descending_attenuation', 'ascending_attenuation', 'descending_minus_ascending'):
        values = current[metric]
        test_rows.append({
            'test_family': 'orientation', 'direction': direction,
            'metric': metric,
            **bootstrap_summary(values, f'{direction}:{metric}'),
            'p_signflip': signflip_p(values, f'{direction}:{metric}'),
        })

transition_rows = []
for transition, current in scientific.groupby('transition'):
    family_values = current.groupby('family_id')['attenuation_code'].mean()
    transition_rows.append({
        'transition': transition,
        'orientation': current['orientation'].iloc[0],
        **bootstrap_summary(family_values, f'transition:{transition}'),
        'p_signflip': signflip_p(family_values, f'transition:{transition}'),
    })
transition_frame = pd.DataFrame(transition_rows).sort_values('transition')
tests_frame = pd.DataFrame(test_rows)
tests_frame.to_csv(ANALYSIS_DIR / 'orientation_tests.csv', index=False)
transition_frame.to_csv(ANALYSIS_DIR / 'six_transition_tests.csv', index=False)

def gap_test(direction):
    row = tests_frame[
        (tests_frame['direction'] == direction)
        & (tests_frame['metric'] == 'descending_minus_ascending')
    ]
    assert len(row) == 1
    return row.iloc[0]

rescue_gap = gap_test('exact')
induction_gap = gap_test('shortcut')
if (
    rescue_gap['n'] >= 8 and induction_gap['n'] >= 8
    and rescue_gap['ci_low'] > 0.05 and induction_gap['ci_low'] > 0.05
    and rescue_gap['p_signflip'] < ALPHA and induction_gap['p_signflip'] < ALPHA
):
    audit_decision = 'robust_ordered_label_transition_asymmetry'
elif rescue_gap['ci_low'] > 0 or induction_gap['ci_low'] > 0:
    audit_decision = 'directional_or_partial_transition_asymmetry'
else:
    audit_decision = 'no_robust_transition_asymmetry_on_intersection'

display(tests_frame.round(4))
display(transition_frame.round(4))
print('Audit decision:', audit_decision)
```

```python
%%cell_progress 8/10 Quantify literal flips by transition orientation
cell_stage('Separating score movement from actual answer changes')

flip_rows = []
for direction in DIRECTIONS:
    current = orientation_family[orientation_family['direction'] == direction]
    for orientation in ('ascending', 'descending'):
        values = current[current['orientation'] == orientation]
        semantic = values.groupby('family_id')['literal_flip'].mean()
        random = values.groupby('family_id')['random_flip'].mean()
        difference = semantic - random
        flip_rows.append({
            'direction': direction, 'orientation': orientation,
            'families': int(len(semantic)),
            'semantic_flip_rate': float(semantic.mean()),
            'random_flip_rate': float(random.mean()),
            **{
                'difference_mean': bootstrap_summary(
                    difference, f'flip:{direction}:{orientation}'
                )['mean'],
                'difference_ci_low': bootstrap_summary(
                    difference, f'flip:{direction}:{orientation}'
                )['ci_low'],
                'difference_ci_high': bootstrap_summary(
                    difference, f'flip:{direction}:{orientation}'
                )['ci_high'],
                'p_signflip': signflip_p(
                    difference, f'flip:{direction}:{orientation}'
                ),
            },
        })
flip_frame = pd.DataFrame(flip_rows)
flip_frame.to_csv(ANALYSIS_DIR / 'literal_flips_by_orientation.csv', index=False)
display(flip_frame.round(4))
```
