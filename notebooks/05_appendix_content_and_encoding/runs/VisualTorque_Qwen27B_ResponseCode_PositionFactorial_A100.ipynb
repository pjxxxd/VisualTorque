```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 7/17 Define the complete token-by-position prompt factorial
cell_stage('Crossing semantic-to-label mappings with independent line orders')
import itertools
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}

def mapping_by_label(semantics):
    assert set(semantics) == set(SEMANTICS)
    return tuple({
        'label': LABELS[index], 'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def semantics_by_label(mapping):
    return tuple(
        next(item['semantic'] for item in mapping if item['label'] == label)
        for label in LABELS
    )

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def mapping_key(mapping):
    abbreviation = {'left': 'L', 'balanced': 'B', 'right': 'R'}
    return ''.join(abbreviation[value] for value in semantics_by_label(mapping))

MAPPINGS = {
    mapping_key(mapping_by_label(semantics)): mapping_by_label(semantics)
    for semantics in itertools.permutations(SEMANTICS)
}
LINE_ORDER_MAP = {
    ''.join(order): tuple(order) for order in itertools.permutations(LABELS)
}
assert len(MAPPINGS) == LABEL_MAPPINGS
assert len(LINE_ORDER_MAP) == LINE_ORDERS

def custom_prompt(mapping, line_order):
    item_by_label = {item['label']: item for item in mapping}
    options = '\n'.join(
        f"{label}. {SEMANTIC_TEXT[item_by_label[label]['semantic']]}"
        for label in line_order
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: Which way will the balance scale tip?',
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_factorial_example(row, mapping, line_order, map_key, order_key):
    role = row['family']['roles']['base']
    features = role['features']
    semantics = semantics_by_label(mapping)
    gold = str(features['outcome'])
    return BehaviorExample(
        example_id=f"interface:{row['family_id']}:{map_key}:{order_key}",
        question_id=f"interface_{row['family_id']}_{map_key}_{order_key}",
        scene_id=str(role['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind='outcome', variant='response_code_position_factorial',
        split='prospective_prompt_factorial',
        problem_type=str(row['family']['diagnostic_cell']),
        outcome=gold, net_torque=int(features['net_torque']),
        prompt=custom_prompt(mapping, line_order),
        image_path=(TARGET_RUNTIME_DIR / role['image_filename']).resolve(),
        labels=LABELS,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def transition_metadata(mapping, line_order, source_semantic, desired_semantic):
    source_label = label_for(mapping, source_semantic)
    desired_label = label_for(mapping, desired_semantic)
    label_transition = f'{source_label}->{desired_label}'
    source_position = line_order.index(source_label) + 1
    desired_position = line_order.index(desired_label) + 1
    position_transition = f'P{source_position}->P{desired_position}'
    assert label_transition in LABEL_TRANSITIONS
    assert position_transition in POSITION_TRANSITIONS
    return {
        'source_label': source_label, 'desired_label': desired_label,
        'label_transition': label_transition,
        'label_pair': ''.join(sorted((source_label, desired_label))),
        'source_position': source_position,
        'desired_position': desired_position,
        'position_transition': position_transition,
    }

print('Mappings:', tuple(MAPPINGS))
print('Independent option orders:', tuple(LINE_ORDER_MAP))
```

## 3. Causal capture across token identity and line position


```python
%%cell_progress 8/17 Define clean capture and dual-hook interface interventions
cell_stage('Defining final-token L49 steering and L55 component removal')

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

def score_record(prepared, logits):
    example = prepared['examples'][0]
    logprobs = torch.log_softmax(logits[0, -1, :].float(), dim=-1)
    scores = [
        float(logprobs[prepared['token_by_label'][label]].detach().cpu())
        for label in example.labels
    ]
    maximum = max(scores)
    weights = [math.exp(value - maximum) for value in scores]
    probabilities = [value / sum(weights) for value in weights]
    predicted = int(np.argmax(scores))
    return {
        'candidate_labels': list(example.labels),
        'candidate_semantics': list(example.option_semantics),
        'candidate_logprobs': scores,
        'candidate_probabilities': probabilities,
        'predicted_label': example.labels[predicted],
        'predicted_semantic': example.option_semantics[predicted],
        'gold_label': example.gold_label, 'gold_semantic': example.gold_semantic,
    }

def capture_clean(example):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    captured = {}
    def hook(_module, _inputs, output):
        captured['hidden'] = output_hidden_tensor(output).detach().cpu().clone()
    handle = DECODER_LAYERS[LABEL_LAYER].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    return score_record(prepared, output.logits), captured['hidden'][0][mask]

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def unit_direction(tensor):
    norm = selected_norm(tensor)
    assert norm > 0
    return tensor.float() / norm

def matched_random(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    return unit_direction(noise)

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def score_interface_condition(example, clean_l55, condition):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    measurements = {}

    def l49_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        payload = condition['l49_payload'].to(hidden.device, hidden.dtype)
        modified = hidden.clone()
        modified[0, current] = hidden[0, current] + payload
        return replace_output_hidden(output, modified)

    def l55_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        current = mask.to(hidden.device)
        clean = clean_l55.to(hidden.device, torch.float32)
        axis = condition['symmetric_axis'].to(hidden.device, torch.float32)
        random_axis = condition['random_axis'].to(hidden.device, torch.float32)
        delta = hidden[0, current].float() - clean
        coefficient = torch.sum(delta * axis)
        component = coefficient * axis
        random_component = coefficient * random_axis
        mode = condition['mode']
        if mode == 'total':
            selected = hidden[0, current].float()
        elif mode == 'remove_symmetric':
            selected = hidden[0, current].float() - component
        elif mode == 'remove_random':
            selected = hidden[0, current].float() - random_component
        elif mode == 'clean_clamp':
            selected = clean
        else:
            raise ValueError(mode)
        modified = hidden.clone()
        modified[0, current] = selected.to(hidden.dtype)
        measurements.update({
            'axis_coefficient': float(coefficient.detach().cpu()),
            'axis_component_norm': float(torch.linalg.vector_norm(component).detach().cpu()),
            'l55_delta_norm': float(torch.linalg.vector_norm(delta).detach().cpu()),
        })
        return replace_output_hidden(output, modified)

    handles = [
        DECODER_LAYERS[SEMANTIC_LAYER].register_forward_hook(l49_hook),
        DECODER_LAYERS[LABEL_LAYER].register_forward_hook(l55_hook),
    ]
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_record(prepared, output.logits)
    record.update({
        key: value for key, value in condition.items()
        if key not in ('l49_payload', 'symmetric_axis', 'random_axis')
    })
    record.update(measurements)
    record['patched_tokens'] = int(mask.sum())
    return record

print('Interface scorer ready.')
```

## 4. Factorial localization and frozen decision


```python
%%cell_progress 11/17 Assemble the fully crossed family estimands
cell_stage('Releasing the model and computing matched label/position contrasts')
import gc
del adapter
gc.collect()
torch.cuda.empty_cache()

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def lookup(records, mode):
    matches = [record for record in records if record['mode'] == mode]
    assert len(matches) == 1
    return matches[0]

estimand_rows = []
for path in tqdm(sorted(SHARD_DIR.glob('*.json')), desc='Factorial estimands', unit='family'):
    payload = json.loads(path.read_text(encoding='utf-8'))
    family = payload['family']
    cohort = payload['cohort']
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions'][family['shortcut_type'] + '_only'])
    desired = exact if cohort == 'rescue' else shortcut
    source_semantic = shortcut if cohort == 'rescue' else exact
    multiplier = 1.0 if cohort == 'rescue' else -1.0
    for cell in payload['cells']:
        baseline = cell['baseline']
        clean_margin = semantic_logprob(baseline, exact) - semantic_logprob(baseline, shortcut)
        records = {mode: lookup(cell['records'], mode) for mode in INTERVENTION_MODES}
        effects = {}
        for mode, record in records.items():
            margin = semantic_logprob(record, exact) - semantic_logprob(record, shortcut)
            effects[mode] = multiplier * (margin - clean_margin)
        estimand_rows.append({
            'family_id': payload['family_id'], 'cohort': cohort,
            'diagnostic_cell': family['diagnostic_cell'],
            'mapping_key': cell['mapping_key'],
            'line_order_key': cell['line_order_key'],
            'label_transition': cell['label_transition'],
            'position_transition': cell['position_transition'],
            'label_pair': cell['label_pair'],
            'clean_margin': clean_margin,
            'effect_total': effects['total'],
            'effect_remove_symmetric': effects['remove_symmetric'],
            'effect_remove_random': effects['remove_random'],
            'symmetric_attenuation': (
                effects['total'] - effects['remove_symmetric']
            ),
            'symmetric_specificity': (
                effects['remove_random'] - effects['remove_symmetric']
            ),
            'baseline_prediction': baseline['predicted_semantic'],
            'total_prediction': records['total']['predicted_semantic'],
            'removed_prediction': records['remove_symmetric']['predicted_semantic'],
            'actual_desired_flip': float(
                baseline['predicted_semantic'] == source_semantic
                and records['total']['predicted_semantic'] == desired
            ),
            'blocked_desired_flip': float(
                baseline['predicted_semantic'] == source_semantic
                and records['total']['predicted_semantic'] == desired
                and records['remove_symmetric']['predicted_semantic'] != desired
            ),
        })

estimand_frame = pd.DataFrame(estimand_rows)
assert len(estimand_frame) == EXPECTED_FACTORIAL_CELLS
for family_id, current in estimand_frame.groupby('family_id'):
    assert len(current) == CELLS_PER_FAMILY
    assert len(current[['label_transition', 'position_transition']].drop_duplicates()) == 36
estimand_frame.to_csv(ANALYSIS_DIR / 'factorial_cell_estimands.csv', index=False)

def cycle_gap(frame, column, favored, reverse, value):
    return float(
        frame[frame[column].isin(favored)][value].mean()
        - frame[frame[column].isin(reverse)][value].mean()
    )

def decomposition(frame, value):
    matrix = frame.pivot(
        index='label_transition', columns='position_transition', values=value
    ).reindex(index=LABEL_TRANSITIONS, columns=POSITION_TRANSITIONS).to_numpy(float)
    assert matrix.shape == (6, 6) and np.isfinite(matrix).all()
    grand = float(matrix.mean())
    label_main = matrix.mean(axis=1) - grand
    position_main = matrix.mean(axis=0) - grand
    interaction = matrix - grand - label_main[:, None] - position_main[None, :]
    ss_label = float(6 * np.sum(label_main ** 2))
    ss_position = float(6 * np.sum(position_main ** 2))
    ss_interaction = float(np.sum(interaction ** 2))
    total = ss_label + ss_position + ss_interaction
    if total <= 1e-12:
        return 0.0, 0.0, 0.0
    return ss_label / total, ss_position / total, ss_interaction / total

family_rows = []
position_stratum_rows = []
for family_id, current in estimand_frame.groupby('family_id'):
    label_fraction, position_fraction, interaction_fraction = decomposition(
        current, 'symmetric_attenuation'
    )
    family_rows.append({
        'family_id': family_id,
        'cohort': current['cohort'].iloc[0],
        'diagnostic_cell': current['diagnostic_cell'].iloc[0],
        'label_cycle_gap': cycle_gap(
            current, 'label_transition', LABEL_FAVORED, LABEL_REVERSE,
            'symmetric_attenuation'
        ),
        'position_cycle_gap': cycle_gap(
            current, 'position_transition', POSITION_FAVORED, POSITION_REVERSE,
            'symmetric_attenuation'
        ),
        'total_label_cycle_gap': cycle_gap(
            current, 'label_transition', LABEL_FAVORED, LABEL_REVERSE,
            'effect_total'
        ),
        'total_position_cycle_gap': cycle_gap(
            current, 'position_transition', POSITION_FAVORED, POSITION_REVERSE,
            'effect_total'
        ),
        'symmetric_specificity': float(current['symmetric_specificity'].mean()),
        'label_fraction': label_fraction,
        'position_fraction': position_fraction,
        'interaction_fraction': interaction_fraction,
    })
    for position_transition in POSITION_TRANSITIONS:
        subset = current[current['position_transition'] == position_transition]
        assert len(subset) == len(LABEL_TRANSITIONS)
        position_stratum_rows.append({
            'family_id': family_id,
            'position_transition': position_transition,
            'label_cycle_gap': cycle_gap(
                subset, 'label_transition', LABEL_FAVORED, LABEL_REVERSE,
                'symmetric_attenuation'
            ),
        })

family_frame = pd.DataFrame(family_rows)
family_frame['label_minus_position_cycle'] = (
    family_frame['label_cycle_gap'] - family_frame['position_cycle_gap']
)
position_stratum_frame = pd.DataFrame(position_stratum_rows)
family_frame.to_csv(ANALYSIS_DIR / 'family_factorial_estimands.csv', index=False)
position_stratum_frame.to_csv(
    ANALYSIS_DIR / 'position_stratified_label_cycles.csv', index=False
)
display(estimand_frame.groupby('label_transition')[[
    'effect_total', 'symmetric_attenuation', 'symmetric_specificity'
]].mean().round(4))
display(estimand_frame.groupby('position_transition')[[
    'effect_total', 'symmetric_attenuation', 'symmetric_specificity'
]].mean().round(4))
print('Family estimands:', len(family_frame))
```

```python
%%cell_progress 12/17 Test token identity against option-position routing
cell_stage('Running family bootstrap, sign-flip, and equivalence tests')

def stable_key_local(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def bootstrap_summary(values, key, alpha=ALPHA):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    result = {
        'n': int(len(array)),
        'mean': float(array.mean()) if len(array) else np.nan,
        'ci_low': np.nan, 'ci_high': np.nan, 'alpha': float(alpha),
    }
    if len(array) < 2:
        return result
    generator = np.random.default_rng(
        int(stable_key_local(f'bootstrap:{key}')[:16], 16) % (2**63 - 1)
    )
    draws = generator.choice(
        array, size=(BOOTSTRAP_REPLICATES, len(array)), replace=True
    ).mean(axis=1)
    result.update({
        'ci_low': float(np.quantile(draws, alpha / 2)),
        'ci_high': float(np.quantile(draws, 1 - alpha / 2)),
    })
    return result

def signflip_p(values, key, alternative='greater'):
    array = np.asarray(values, dtype=float)
    array = array[np.isfinite(array)]
    assert len(array)
    observed = float(array.mean())
    generator = np.random.default_rng(
        int(stable_key_local(f'signflip:{key}')[:16], 16) % (2**63 - 1)
    )
    exceed = 0
    remaining = SIGNFLIP_REPLICATES
    while remaining:
        count = min(10000, remaining)
        signs = generator.choice((-1.0, 1.0), size=(count, len(array)))
        draws = (signs * array).mean(axis=1)
        if alternative == 'greater':
            exceed += int(np.count_nonzero(draws >= observed))
        elif alternative == 'two-sided':
            exceed += int(np.count_nonzero(np.abs(draws) >= abs(observed)))
        else:
            raise ValueError(alternative)
        remaining -= count
    return float((1 + exceed) / (SIGNFLIP_REPLICATES + 1))

macro_metrics = (
    'label_cycle_gap', 'position_cycle_gap', 'label_minus_position_cycle',
    'total_label_cycle_gap', 'total_position_cycle_gap',
    'symmetric_specificity', 'label_fraction', 'position_fraction',
    'interaction_fraction',
)
test_rows = []
for metric in macro_metrics:
    values = family_frame[metric]
    test_rows.append({
        'test_family': 'macro', 'stratum': 'all', 'metric': metric,
        **bootstrap_summary(values, f'macro:{metric}'),
        'p_signflip': signflip_p(
            values, f'macro:{metric}',
            'two-sided' if metric == 'position_cycle_gap' else 'greater'
        ),
    })
for position_transition in POSITION_TRANSITIONS:
    values = position_stratum_frame[
        position_stratum_frame['position_transition'] == position_transition
    ]['label_cycle_gap']
    test_rows.append({
        'test_family': 'position_stratified',
        'stratum': position_transition, 'metric': 'label_cycle_gap',
        **bootstrap_summary(
            values, f'position:{position_transition}', STRATIFIED_ALPHA
        ),
        'p_signflip': signflip_p(
            values, f'position:{position_transition}', 'greater'
        ),
    })
tests_frame = pd.DataFrame(test_rows)
tests_frame.to_csv(ANALYSIS_DIR / 'factorial_tests.csv', index=False)

def macro(metric):
    row = tests_frame[
        (tests_frame['test_family'] == 'macro')
        & (tests_frame['metric'] == metric)
    ]
    assert len(row) == 1
    return row.iloc[0]

label_cycle_replicated = bool(
    macro('label_cycle_gap')['ci_low'] > CYCLE_SESOI
    and macro('label_cycle_gap')['p_signflip'] < ALPHA
)
position_cycle_equivalent = bool(
    macro('position_cycle_gap')['ci_low'] > -EQUIVALENCE_BOUND
    and macro('position_cycle_gap')['ci_high'] < EQUIVALENCE_BOUND
)
position_cycle_present = bool(
    macro('position_cycle_gap')['ci_low'] > STRATUM_SESOI
    or macro('position_cycle_gap')['ci_high'] < -STRATUM_SESOI
)
label_dominates_position = bool(
    macro('label_minus_position_cycle')['ci_low'] > STRATUM_SESOI
)
specific_component = bool(
    macro('symmetric_specificity')['ci_low'] > 0
    and macro('symmetric_specificity')['p_signflip'] < ALPHA
)
stratified = tests_frame[
    tests_frame['test_family'] == 'position_stratified'
]
label_cycle_position_invariant = bool(
    len(stratified) == len(POSITION_TRANSITIONS)
    and (stratified['ci_low'] > STRATUM_SESOI).all()
)

if not specific_component:
    primary_decision = 'factorial_effect_not_specific_to_l55_axis'
elif not label_cycle_replicated:
    primary_decision = 'prior_label_cycle_not_replicated'
elif label_cycle_position_invariant and position_cycle_equivalent and label_dominates_position:
    primary_decision = 'response_token_cycle_survives_option_position_control'
elif label_cycle_position_invariant and position_cycle_present:
    primary_decision = 'response_token_cycle_with_position_modulation'
elif position_cycle_present and not label_cycle_position_invariant:
    primary_decision = 'option_position_or_interaction_dominates'
else:
    primary_decision = 'mixed_or_inconclusive_interface_gating'

decision_payload = {
    'schema_version': 'response-code-position-decision-1.0',
    'model_id': MODEL_ID, 'resolved_revision': MODEL_REVISION,
    'primary_decision': primary_decision,
    'label_cycle_replicated': label_cycle_replicated,
    'label_cycle_position_invariant': label_cycle_position_invariant,
    'position_cycle_equivalent': position_cycle_equivalent,
    'position_cycle_present': position_cycle_present,
    'label_dominates_position': label_dominates_position,
    'specific_component': specific_component,
    'macro_tests': {
        metric: json.loads(macro(metric).to_json()) for metric in macro_metrics
    },
    'position_stratified_tests': stratified.to_dict(orient='records'),
}
write_json_atomic(ANALYSIS_DIR / 'factorial_decision.json', decision_payload)
display(tests_frame.round(4))
display(pd.DataFrame([decision_payload]).drop(
    columns=['macro_tests', 'position_stratified_tests']
))
```

```python
%%cell_progress 13/17 Audit factorial balance, shards, norms, and replay
cell_stage('Verifying every frozen combinatorial and numerical invariant')

record_count = 0
cell_count = 0
mode_counts = Counter()
label_counts = Counter()
position_counts = Counter()
combination_counts = Counter()
nonfinite = []
for path in sorted(SHARD_DIR.glob('*.json')):
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['cells']) == CELLS_PER_FAMILY
    local_combinations = set()
    for cell in payload['cells']:
        cell_count += 1
        label_counts[cell['label_transition']] += 1
        position_counts[cell['position_transition']] += 1
        local_combinations.add((cell['label_transition'], cell['position_transition']))
        combination_counts[(cell['label_transition'], cell['position_transition'])] += 1
        assert len(cell['records']) == RECORDS_PER_CELL
        for record in cell['records']:
            record_count += 1
            mode_counts[record['mode']] += 1
            if not np.isfinite([
                *record['candidate_logprobs'],
                record['axis_coefficient'], record['axis_component_norm'],
                record['l55_delta_norm'],
            ]).all():
                nonfinite.append(record['condition'])
            assert record['patched_tokens'] == 1
    assert len(local_combinations) == CELLS_PER_FAMILY

assert cell_count == EXPECTED_FACTORIAL_CELLS
assert record_count == EXPECTED_RECORDS
assert mode_counts == Counter({
    mode: EXPECTED_FACTORIAL_CELLS for mode in INTERVENTION_MODES
})
assert label_counts == Counter({
    transition: EXPECTED_FAMILIES * LINE_ORDERS
    for transition in LABEL_TRANSITIONS
})
assert position_counts == Counter({
    transition: EXPECTED_FAMILIES * LABEL_MAPPINGS
    for transition in POSITION_TRANSITIONS
})
assert all(value == EXPECTED_FAMILIES for value in combination_counts.values())
assert len(combination_counts) == 36
assert not nonfinite

audit_payload = {
    'schema_version': 'response-code-position-audit-1.0',
    'support_identity': support_identity,
    'families': EXPECTED_FAMILIES,
    'factorial_cells': cell_count,
    'intervention_records': record_count,
    'mode_counts': dict(mode_counts),
    'label_transition_counts': dict(label_counts),
    'position_transition_counts': dict(position_counts),
    'combination_count_min': min(combination_counts.values()),
    'combination_count_max': max(combination_counts.values()),
    'clean_clamp_error': smoke_clamp_error,
    'nonfinite_measurements': len(nonfinite),
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', audit_payload)
display(pd.DataFrame([audit_payload]))
```
