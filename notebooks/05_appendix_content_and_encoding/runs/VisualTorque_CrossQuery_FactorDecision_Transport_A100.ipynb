```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 6/17 Reuse and verify the locked inverse-query manifest
cell_stage('Loading the exact prior manifest and verifying image support')
from collections import Counter, defaultdict
import shutil

def read_jsonl(path):
    with Path(path).open('r', encoding='utf-8') as handle:
        return [json.loads(line) for line in handle if line.strip()]

def write_jsonl_atomic(path, rows):
    temporary = Path(path).with_suffix(Path(path).suffix + '.tmp')
    with temporary.open('w', encoding='utf-8') as handle:
        for row in rows:
            handle.write(json.dumps(row, sort_keys=True) + '\n')
    temporary.replace(path)

def stable_key(value):
    return hashlib.sha256(f'{SEED}:{value!r}'.encode('utf-8')).hexdigest()

def sign(value):
    return int(value > 0) - int(value < 0)

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

assert PRIOR_MANIFEST_PATH.is_file(), (
    'Run VisualTorque_InverseQuery_WeightRoute_Specificity_A100 first: '
    f'{PRIOR_MANIFEST_PATH}'
)
prior_manifest_sha256 = sha256_file(PRIOR_MANIFEST_PATH)
locked_rows = read_jsonl(PRIOR_MANIFEST_PATH)
assert len(locked_rows) == EXPECTED_FAMILIES
assert len({row['family_id'] for row in locked_rows}) == EXPECTED_FAMILIES

physics_violations = []
image_filenames = []
for row in locked_rows:
    base = row['roles']['base']
    source = row['roles']['aligned_source']
    base_positions = tuple(
        (stack['side'], int(stack['distance'])) for stack in base['stacks']
    )
    source_positions = tuple(
        (stack['side'], int(stack['distance'])) for stack in source['stacks']
    )
    base_units = tuple(int(stack['units']) for stack in base['stacks'])
    source_units = tuple(int(stack['units']) for stack in source['stacks'])
    if base_positions != source_positions:
        physics_violations.append((row['family_id'], 'positions'))
    if base_units == source_units:
        physics_violations.append((row['family_id'], 'unchanged_units'))
    if int(base['features']['net_torque']) != int(source['features']['net_torque']):
        physics_violations.append((row['family_id'], 'net_torque'))
    if str(base['features']['outcome']) != str(source['features']['outcome']):
        physics_violations.append((row['family_id'], 'outcome'))
    image_filenames.extend(
        role['image_filename'] for role in row['roles'].values()
    )
assert not physics_violations, physics_violations[:5]
assert len(image_filenames) == 2 * EXPECTED_FAMILIES
assert all((PRIOR_SCENE_DIR / name).is_file() for name in image_filenames)

manifest_identity = {
    'schema_version': 'cross-query-manifest-lock-1.0',
    'prior_manifest_sha256': prior_manifest_sha256,
    'family_ids': [row['family_id'] for row in locked_rows],
    'families': len(locked_rows),
    'physical_images': len(image_filenames),
    'selection_uses_current_model_outputs': False,
    'exact_prior_support_reused': True,
}
if MANIFEST_PATH.is_file() and MANIFEST_LOCK_PATH.is_file():
    previous = json.loads(MANIFEST_LOCK_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == manifest_identity
    assert read_jsonl(MANIFEST_PATH) == locked_rows
else:
    write_jsonl_atomic(MANIFEST_PATH, locked_rows)
    write_json_atomic(MANIFEST_LOCK_PATH, {
        'identity': manifest_identity,
        'locked_at_utc': datetime.now(timezone.utc).isoformat(),
    })

transport_manifest = read_jsonl(MANIFEST_PATH)
RUNTIME_SCENE_DIR = Path('/content/visualtorque_cross_query_scenes_v1')
RUNTIME_SCENE_DIR.mkdir(parents=True, exist_ok=True)
for filename in tqdm(image_filenames, desc='Caching locked image pairs', unit='image'):
    source = PRIOR_SCENE_DIR / filename
    destination = RUNTIME_SCENE_DIR / filename
    if not destination.is_file() or destination.stat().st_size != source.stat().st_size:
        shutil.copy2(source, destination)
assert all((RUNTIME_SCENE_DIR / name).is_file() for name in image_filenames)
print('Prior manifest SHA256:', prior_manifest_sha256)
print('Families / images:', len(transport_manifest), len(image_filenames))
```

```python
%%cell_progress 8/17 Define fixed-option cross-query prompts and scopes
cell_stage('Building prompts with one option mapping shared across all tasks')
import math
from visual_torque_toolkit.behavior import PromptCondition
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = {
    'outcome': 'Which way will the balance scale tip?',
    'weight_more': 'Which side has more total weight?',
    'weight_less': 'Which side has less total weight?',
    'distance_more': 'Which side has the greater sum of stack distances?',
    'distance_less': 'Which side has the smaller sum of stack distances?',
}
SEMANTIC_TEXT = {
    'left': 'left side', 'balanced': 'equal', 'right': 'right side',
}
LABELS = ('A', 'B', 'C')

def task_semantic(features, task):
    if task == 'outcome':
        value = int(features['net_torque'])
    elif task == 'weight_more':
        value = int(features['weight_score'])
    elif task == 'weight_less':
        value = -int(features['weight_score'])
    elif task == 'distance_more':
        value = int(features['distance_score'])
    elif task == 'distance_less':
        value = -int(features['distance_score'])
    else:
        raise KeyError(task)
    return outcome(value)

def fixed_semantics(row):
    values = ['left', 'balanced', 'right']
    offset = int(stable_key(f"options:{row['family_id']}")[:8], 16) % 3
    return tuple(values[offset:] + values[:offset])

def custom_prompt(row, task, semantics):
    options = '\n'.join(
        f'{label}. {SEMANTIC_TEXT[value]}'
        for label, value in zip(LABELS, semantics)
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT,
        'Question: ' + QUESTION_TEXT[task],
        'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option '
        'letter and nothing else. Valid replies are: A, B, C.',
    ])

def make_example(row, task, role):
    semantics = fixed_semantics(row)
    role_record = row['roles'][role]
    features = role_record['features']
    gold = task_semantic(features, task)
    return BehaviorExample(
        example_id=f"crossquery:{row['family_id']}:{task}:{role}",
        question_id=f"crossquery_{row['family_id']}_{task}",
        scene_id=str(role_record['scene_id']),
        condition=PromptCondition.IMAGE_RULE,
        question_kind=task,
        variant='cross_query_factor_decision_transport',
        split='followup_confirmation',
        problem_type='factor_vs_decision_transport',
        outcome=gold,
        net_torque=int(features['net_torque']),
        prompt=custom_prompt(row, task, semantics),
        image_path=(RUNTIME_SCENE_DIR / role_record['image_filename']).resolve(),
        labels=LABELS,
        option_text=tuple(SEMANTIC_TEXT[value] for value in semantics),
        option_semantics=semantics,
        gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def make_bundle(row):
    return {
        f'{task}__{role}': make_example(row, task, role)
        for task in TARGET_TASKS for role in SCENE_ROLES
    }

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples],
            add_generation_prompt=True,
        )),
        fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def discover_masks(prepared):
    assert len(prepared['examples']) == 1
    example = prepared['examples'][0]
    ids = prepared['inputs']['input_ids'][0].detach().cpu()
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    tokenizer = getattr(adapter.processor, 'tokenizer', adapter.processor)
    tokens = tokenizer.convert_ids_to_tokens(ids.tolist())
    image_ids = set()
    if IMAGE_TOKEN_ID_OVERRIDE is not None:
        values = IMAGE_TOKEN_ID_OVERRIDE if isinstance(
            IMAGE_TOKEN_ID_OVERRIDE, (list, tuple)
        ) else [IMAGE_TOKEN_ID_OVERRIDE]
        image_ids.update(int(value) for value in values)
    for token_id, token_text in zip(ids.tolist(), tokens):
        lower = str(token_text).lower()
        if any(marker in lower for marker in ('image', 'vision', '<img', 'img>')):
            image_ids.add(int(token_id))
    image = torch.tensor(
        [int(token_id) in image_ids for token_id in ids.tolist()],
        dtype=torch.bool,
    ) & attention
    assert bool(image.any()), 'No image tokens detected.'
    final = torch.zeros_like(attention)
    final[int(torch.where(attention)[0][-1])] = True
    all_text = attention & ~image & ~final

    encoding = tokenizer(
        example.prompt, add_special_tokens=False, return_offsets_mapping=True,
    )
    prompt_ids = list(encoding['input_ids'])
    offsets = list(encoding['offset_mapping'])
    id_list = ids.tolist()
    starts = [
        start for start in range(len(ids) - len(prompt_ids) + 1)
        if id_list[start:start + len(prompt_ids)] == prompt_ids
    ]
    assert len(starts) == 1, len(starts)
    prompt_start = starts[0]
    sections = example.prompt.split('\n\n')
    assert len(sections) == 5
    names = (
        'intro_tokens', 'rule_tokens', 'question_tokens',
        'options_tokens', 'answer_instruction_tokens',
    )
    masks = {
        'valid_tokens': attention, 'image_tokens': image,
        'all_text_except_final': all_text, 'final_prompt_token': final,
    }
    spans = []
    cursor = 0
    for name, section in zip(names, sections):
        start = example.prompt.find(section, cursor)
        end = start + len(section)
        cursor = end
        spans.append((name, start, end))
        masks[name] = torch.zeros_like(attention)
    for offset_index, (token_start, token_end) in enumerate(offsets):
        overlaps = [
            (max(0, min(token_end, end) - max(token_start, start)), name)
            for name, start, end in spans
        ]
        overlap, name = max(overlaps)
        full_index = prompt_start + offset_index
        if overlap > 0 and bool(all_text[full_index]):
            masks[name][full_index] = True
    return {**masks, 'tokens': tokens, 'input_ids': ids}

def assert_within_task_aligned(prepared_by_name, task):
    base = prepared_by_name[f'{task}__base']
    source = prepared_by_name[f'{task}__aligned_source']
    for key in ('input_ids', 'attention_mask'):
        assert torch.equal(base['inputs'][key], source['inputs'][key]), (task, key)
    base_masks = discover_masks(base)
    source_masks = discover_masks(source)
    assert torch.equal(base_masks[SITE_SCOPE], source_masks[SITE_SCOPE])
    return base_masks

def assert_cross_task_scope(prepared_by_name, masks_by_task):
    reference = None
    for task in TARGET_TASKS:
        prepared = prepared_by_name[f'{task}__base']
        ids = prepared['inputs']['input_ids'][0].detach().cpu()
        selected_ids = ids[masks_by_task[task][SITE_SCOPE]].tolist()
        assert selected_ids, (task, SITE_SCOPE)
        if reference is None:
            reference = selected_ids
        else:
            assert selected_ids == reference, (task, selected_ids, reference)
    return reference

print('Cross-query prompts and transport scopes ready.')
```

```python
%%cell_progress 9/17 Define selected-state capture and transport scorers
cell_stage('Defining cross-query selected-token transport')

def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for row_index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[row_index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted_index = int(np.argmax(scores))
        records.append({
            'scene_id': str(example.scene_id),
            'example_id': str(example.example_id),
            'task': str(example.question_kind),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_semantic': example.option_semantics[predicted_index],
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_capture_batch(prepared):
    hidden_by_layer = {}
    handles = []
    for layer_index in CAPTURE_LAYERS:
        def hook(_module, _inputs, output, layer_index=layer_index):
            hidden_by_layer[layer_index] = output_hidden_tensor(
                output
            ).detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer_index].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def capture_named_examples(examples_by_name):
    names = list(examples_by_name)
    score_by_name = {}
    hidden_by_name = {name: {} for name in names}
    for start in range(0, len(names), CAPTURE_BATCH_SIZE):
        actual = names[start:start + CAPTURE_BATCH_SIZE]
        padded = list(actual)
        while len(padded) < CAPTURE_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepare_examples([examples_by_name[name] for name in padded])
        scores, hidden = forward_capture_batch(prepared)
        for index, name in enumerate(actual):
            score_by_name[name] = scores[index]
            for layer_index, tensor in hidden.items():
                hidden_by_name[name][layer_index] = tensor[index]
        del prepared, hidden
        torch.cuda.empty_cache()
    return score_by_name, hidden_by_name

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def scale_selected_to_norm(tensor, target_norm):
    current = selected_norm(tensor)
    assert current > 0 and target_norm > 0
    return tensor * (float(target_norm) / current)

def matched_random_selected(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    noise_norm = selected_norm(noise)
    target_norm = selected_norm(tensor)
    assert noise_norm > 0 and target_norm > 0
    return (noise * (target_norm / noise_norm)).to(dtype=tensor.dtype)

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layer_indices = {int(row['layer_index']) for row in conditions}
    assert layer_indices == {SITE_LAYER}
    payloads = torch.stack([row['payload_selected'] for row in conditions])
    masks = torch.stack([row['target_mask'] for row in conditions])
    modes = [row['intervention_mode'] for row in conditions]
    audit = {}

    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payloads.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        modified = hidden.clone()
        counts = []
        for index, mode in enumerate(modes):
            current_mask = mask[index]
            assert int(current_mask.sum()) == int(payload[index].shape[0])
            if mode == 'add':
                modified[index, current_mask] = (
                    hidden[index, current_mask] + payload[index]
                )
            elif mode == 'replace':
                modified[index, current_mask] = payload[index]
            else:
                raise ValueError(mode)
            counts.append(int(current_mask.sum()))
        audit['counts'] = counts
        return replace_output_hidden(output, modified)

    handle = DECODER_LAYERS[SITE_LAYER].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, audit['counts']):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('payload_selected', 'target_mask')
        })
        record['patched_tokens'] = int(count)
    return records

def score_conditions(target_example, prepared_cache, conditions):
    records = []
    for start in range(0, len(conditions), PATCH_CONDITION_BATCH_SIZE):
        actual = conditions[start:start + PATCH_CONDITION_BATCH_SIZE]
        padded = list(actual)
        while len(padded) < PATCH_CONDITION_BATCH_SIZE:
            padded.append(padded[-1])
        prepared = prepared_cache.get(len(padded))
        if prepared is None:
            prepared = prepare_examples([target_example] * len(padded))
            prepared_cache[len(padded)] = prepared
        try:
            records.extend(forward_intervention_chunk(
                prepared, padded
            )[:len(actual)])
        except torch.cuda.OutOfMemoryError as exc:
            prepared_cache.pop(len(padded), None)
            torch.cuda.empty_cache()
            raise RuntimeError(
                'Intervention batch ran out of memory. Restart on an A100; '
                'do not change the frozen batch size within a model run.'
            ) from exc
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

print('Selected-state cross-query scorers ready.')
```

```python
%%cell_progress 10/17 Run semantics, scope, and parity smoke tests
cell_stage('Checking inverse semantics, cross-task tokens, and self replay')
smoke_row = transport_manifest[0]
base_features = smoke_row['roles']['base']['features']
source_features = smoke_row['roles']['aligned_source']['features']
exact = str(base_features['outcome'])
shortcut = str(base_features['strategy_predictions']['weight_only'])
assert exact != shortcut
expected = {
    'outcome': (exact, exact),
    'weight_more': (shortcut, exact),
    'weight_less': (exact, shortcut),
    'distance_more': (exact, exact),
    'distance_less': (shortcut, shortcut),
}
for task, (base_value, source_value) in expected.items():
    assert task_semantic(base_features, task) == base_value
    assert task_semantic(source_features, task) == source_value

smoke_bundle = make_bundle(smoke_row)
smoke_prepared = {
    name: prepare_examples([example]) for name, example in smoke_bundle.items()
}
smoke_masks = {
    task: assert_within_task_aligned(smoke_prepared, task)
    for task in TARGET_TASKS
}
scope_token_ids = assert_cross_task_scope(smoke_prepared, smoke_masks)
smoke_scores, smoke_hidden = capture_named_examples(smoke_bundle)

target_task = 'outcome'
target_key = f'{target_task}__base'
target_mask = smoke_masks[target_task][SITE_SCOPE]
self_payload = smoke_hidden[target_key][SITE_LAYER][target_mask]
self_condition = {
    'analysis_family': 'smoke', 'family_id': smoke_row['family_id'],
    'donor_task': 'self', 'target_task': target_task,
    'intervention': 'self_replay', 'intervention_mode': 'replace',
    'site_label': SITE_LABEL, 'scope': SITE_SCOPE,
    'layer_index': SITE_LAYER, 'payload_selected': self_payload,
    'target_mask': target_mask,
}
self_record = score_conditions(
    smoke_bundle[target_key], {1: smoke_prepared[target_key]}, [self_condition]
)[0]
self_error = float(np.max(np.abs(
    centered_scores(self_record) - centered_scores(smoke_scores[target_key])
)))
assert self_error < SMOKE_CENTERED_LOGPROB_TOLERANCE, self_error
assert self_record['patched_tokens'] == len(scope_token_ids)

donor_task = 'weight_more'
donor_mask = smoke_masks[donor_task][SITE_SCOPE]
donor_delta = (
    smoke_hidden[f'{donor_task}__aligned_source'][SITE_LAYER][donor_mask]
    - smoke_hidden[f'{donor_task}__base'][SITE_LAYER][donor_mask]
)
cross_target = 'weight_less'
cross_condition = {
    'analysis_family': 'smoke', 'family_id': smoke_row['family_id'],
    'donor_task': donor_task, 'target_task': cross_target,
    'intervention': 'donor_delta_norm', 'intervention_mode': 'add',
    'site_label': SITE_LABEL, 'scope': SITE_SCOPE,
    'layer_index': SITE_LAYER, 'payload_selected': donor_delta,
    'target_mask': smoke_masks[cross_target][SITE_SCOPE],
}
cross_record = score_conditions(
    smoke_bundle[f'{cross_target}__base'], {}, [cross_condition]
)[0]
assert cross_record['patched_tokens'] == len(scope_token_ids)
print('Centered self parity:', self_error)
print('Transport scope / token count:', SITE_SCOPE, len(scope_token_ids))
print('Smoke task values:', expected)
print('Smoke cross-query prediction:', cross_record['predicted_semantic'])
del smoke_prepared, smoke_hidden
torch.cuda.empty_cache()
```

```python
%%cell_progress 11/17 Capture or resume the 2 x 5 transport matrix
cell_stage('Locking capture identity and checking family shards')

capture_identity = {
    'schema_version': 'cross-query-factor-decision-capture-1.0',
    'protocol_identity': protocol_identity,
    'prior_manifest_sha256': prior_manifest_sha256,
    'model_id': MODEL_ID,
    'requested_revision': MODEL_REVISION,
    'resolved_revision': adapter.metadata['resolved_revision'],
    'site': list(SPEC['site']),
    'donor_tasks': list(DONOR_TASKS),
    'target_tasks': list(TARGET_TASKS),
    'records_per_family': RECORDS_PER_FAMILY,
    'capture_batch_size': CAPTURE_BATCH_SIZE,
    'patch_condition_batch_size': PATCH_CONDITION_BATCH_SIZE,
    'seed': SEED,
}
CAPTURE_SHA256 = 'sha256:' + hashlib.sha256(
    json.dumps(capture_identity, sort_keys=True).encode('utf-8')
).hexdigest()
CAPTURE_CONFIG_PATH = OUTPUT_ROOT / 'capture_config.json'
if CAPTURE_CONFIG_PATH.is_file():
    previous = json.loads(CAPTURE_CONFIG_PATH.read_text(encoding='utf-8'))
    assert previous.get('identity') == capture_identity, (
        'Existing model output differs from this capture identity.'
    )
else:
    write_json_atomic(CAPTURE_CONFIG_PATH, {
        'identity': capture_identity,
        'capture_identity_sha256': CAPTURE_SHA256,
        'created_at_utc': datetime.now(timezone.utc).isoformat(),
    })

completed_family_ids = set()
for shard_path in sorted(SHARD_DIR.glob('*.json')):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    completed_family_ids.add(payload['family_id'])
print('Completed / total families:', len(completed_family_ids), len(transport_manifest))

progress = tqdm(
    transport_manifest, desc=f'{MODEL_KEY} cross-query transport', unit='family'
)
for family in progress:
    family_id = str(family['family_id'])
    if family_id in completed_family_ids:
        continue
    examples = make_bundle(family)
    prepared_single = {
        name: prepare_examples([example]) for name, example in examples.items()
    }
    masks = {
        task: assert_within_task_aligned(prepared_single, task)
        for task in TARGET_TASKS
    }
    assert_cross_task_scope(prepared_single, masks)
    baseline_scores, hidden = capture_named_examples(examples)

    raw_deltas = {}
    raw_norms = {}
    for donor_task in DONOR_TASKS:
        donor_mask = masks[donor_task][SITE_SCOPE]
        raw = (
            hidden[f'{donor_task}__aligned_source'][SITE_LAYER][donor_mask]
            - hidden[f'{donor_task}__base'][SITE_LAYER][donor_mask]
        )
        raw_deltas[donor_task] = raw
        raw_norms[donor_task] = selected_norm(raw)
    matched_norm = min(raw_norms.values())
    assert matched_norm > 0
    donor_deltas = {
        task: scale_selected_to_norm(raw_deltas[task], matched_norm)
        for task in DONOR_TASKS
    }
    random_deltas = {
        task: matched_random_selected(
            donor_deltas[task], f'{family_id}:{task}:random'
        )
        for task in DONOR_TASKS
    }
    applied_norms = {
        task: {
            'donor_delta_norm': selected_norm(donor_deltas[task]),
            'random_norm': selected_norm(random_deltas[task]),
        }
        for task in DONOR_TASKS
    }
    for task in DONOR_TASKS:
        for value in applied_norms[task].values():
            assert abs(value - matched_norm) / matched_norm < NORM_RELATIVE_TOLERANCE

    patch_records = []
    for target_task in TARGET_TASKS:
        target_key = f'{target_task}__base'
        target_mask = masks[target_task][SITE_SCOPE]
        conditions = [{
            'analysis_family': 'cross_query_transport',
            'family_id': family_id,
            'donor_task': 'self', 'target_task': target_task,
            'intervention': 'self_replay', 'intervention_mode': 'replace',
            'site_label': SITE_LABEL, 'scope': SITE_SCOPE,
            'layer_index': SITE_LAYER,
            'donor_raw_norm': None, 'matched_norm': None,
            'payload_selected': hidden[target_key][SITE_LAYER][target_mask],
            'target_mask': target_mask,
        }]
        for donor_task in DONOR_TASKS:
            for intervention, payload in (
                ('donor_delta_norm', donor_deltas[donor_task]),
                ('random_norm', random_deltas[donor_task]),
            ):
                conditions.append({
                    'analysis_family': 'cross_query_transport',
                    'family_id': family_id,
                    'donor_task': donor_task,
                    'target_task': target_task,
                    'intervention': intervention,
                    'intervention_mode': 'add',
                    'site_label': SITE_LABEL, 'scope': SITE_SCOPE,
                    'layer_index': SITE_LAYER,
                    'donor_raw_norm': float(raw_norms[donor_task]),
                    'matched_norm': float(matched_norm),
                    'payload_selected': payload,
                    'target_mask': target_mask,
                })
        patch_records.extend(score_conditions(
            examples[target_key], {1: prepared_single[target_key]}, conditions
        ))
    assert len(patch_records) == RECORDS_PER_FAMILY
    write_json_atomic(SHARD_DIR / f'{family_id}.json', {
        'schema_version': 'cross-query-family-shard-1.0',
        'capture_identity_sha256': CAPTURE_SHA256,
        'family_id': family_id,
        'baseline_scores': baseline_scores,
        'donor_raw_norms': raw_norms,
        'matched_norm': float(matched_norm),
        'applied_norms': applied_norms,
        'patch_records': patch_records,
    })
    del prepared_single, hidden, raw_deltas, donor_deltas, random_deltas
    torch.cuda.empty_cache()

completed = sorted(SHARD_DIR.glob('*.json'))
assert len(completed) == len(transport_manifest)
observed_records = 0
for path in completed:
    payload = json.loads(path.read_text(encoding='utf-8'))
    assert payload['capture_identity_sha256'] == CAPTURE_SHA256
    assert len(payload['patch_records']) == RECORDS_PER_FAMILY
    observed_records += len(payload['patch_records'])
assert observed_records == EXPECTED_PATCH_RECORDS
print('Capture complete:', len(completed), 'families /', observed_records, 'records')
```

## 4. Competing abstraction tests and cross-model synthesis


```python
%%cell_progress 12/17 Release GPU and assemble transport estimands
cell_stage('Releasing model memory and constructing the 2 x 5 matrix')
import gc
del adapter
gc.collect()
torch.cuda.empty_cache()

family_by_id = {str(row['family_id']): row for row in transport_manifest}

def semantic_logprob(record, semantic):
    semantics = list(record['candidate_semantics'])
    assert semantic in semantics
    return float(record['candidate_logprobs'][semantics.index(semantic)])

def exact_shortcut_logodds(record, exact, shortcut):
    return semantic_logprob(record, exact) - semantic_logprob(record, shortcut)

def find_record(records, donor_task, target_task, intervention):
    matches = [
        row for row in records
        if row['donor_task'] == donor_task
        and row['target_task'] == target_task
        and row['intervention'] == intervention
    ]
    assert len(matches) == 1, (donor_task, target_task, intervention, len(matches))
    return matches[0]

target_sign = {
    'outcome': 1.0, 'weight_more': 1.0, 'weight_less': -1.0,
    'distance_more': 1.0, 'distance_less': -1.0,
}
donor_sign = {'weight_more': 1.0, 'weight_less': -1.0}
estimand_rows = []
baseline_rows = []
self_rows = []
for shard_path in tqdm(
    sorted(SHARD_DIR.glob('*.json')), desc='Assembling transport estimands', unit='family'
):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    family_id = str(payload['family_id'])
    family = family_by_id[family_id]
    features = family['roles']['base']['features']
    exact = str(features['outcome'])
    shortcut = str(features['strategy_predictions']['weight_only'])
    assert exact != shortcut
    records = payload['patch_records']
    for target_task in TARGET_TASKS:
        base_record = payload['baseline_scores'][f'{target_task}__base']
        base_logodds = exact_shortcut_logodds(base_record, exact, shortcut)
        baseline_rows.append({
            'family_id': family_id, 'target_task': target_task,
            'predicted_semantic': base_record['predicted_semantic'],
            'gold_semantic': base_record['gold_semantic'],
            'correct': base_record['predicted_semantic'] == base_record['gold_semantic'],
        })
        self_record = find_record(records, 'self', target_task, 'self_replay')
        self_rows.append({
            'family_id': family_id, 'target_task': target_task,
            'max_centered_error': float(np.max(np.abs(
                centered_scores(self_record) - centered_scores(base_record)
            ))),
            'patched_tokens': int(self_record['patched_tokens']),
        })
        for donor_task in DONOR_TASKS:
            aligned = find_record(records, donor_task, target_task, 'donor_delta_norm')
            random = find_record(records, donor_task, target_task, 'random_norm')
            aligned_shift = exact_shortcut_logodds(
                aligned, exact, shortcut
            ) - base_logodds
            random_shift = exact_shortcut_logodds(
                random, exact, shortcut
            ) - base_logodds
            specific_raw = aligned_shift - random_shift
            estimand_rows.append({
                'family_id': family_id,
                'model_key': MODEL_KEY,
                'site_label': SITE_LABEL,
                'layer_index': SITE_LAYER,
                'scope': SITE_SCOPE,
                'donor_task': donor_task,
                'target_task': target_task,
                'aligned_raw_effect': float(aligned_shift),
                'random_raw_effect': float(random_shift),
                'specific_raw_effect': float(specific_raw),
                'target_aligned_effect': float(
                    target_sign[target_task] * specific_raw
                ),
                'donor_aligned_effect': float(
                    donor_sign[donor_task] * specific_raw
                ),
                'donor_raw_norm': float(aligned['donor_raw_norm']),
                'matched_norm': float(aligned['matched_norm']),
                'patched_tokens': int(aligned['patched_tokens']),
            })

estimand_frame = pd.DataFrame(estimand_rows)
baseline_frame = pd.DataFrame(baseline_rows)
self_frame = pd.DataFrame(self_rows)
assert len(estimand_frame) == (
    EXPECTED_FAMILIES * len(DONOR_TASKS) * len(TARGET_TASKS)
)
assert len(baseline_frame) == EXPECTED_FAMILIES * len(TARGET_TASKS)
estimand_frame.to_csv(ANALYSIS_DIR / 'cross_query_estimands.csv', index=False)
baseline_frame.to_csv(ANALYSIS_DIR / 'baseline_behavior.csv', index=False)
self_frame.to_csv(ANALYSIS_DIR / 'self_replay.csv', index=False)
display(estimand_frame.groupby([
    'donor_task', 'target_task'
])['target_aligned_effect'].agg(['count', 'mean']).reset_index())
```

```python
%%cell_progress 13/17 Test factor, target-decision, and donor-answer patterns
cell_stage('Bootstrapping cells and preregistered abstraction contrasts')

def bootstrap_summary(values, key, directional=False):
    array = np.asarray(values, dtype=float)
    assert len(array) >= MIN_FAMILIES and np.isfinite(array).all()
    generator = np.random.default_rng(
        int(stable_key(f'bootstrap:{MODEL_KEY}:{key}')[:16], 16)
    )
    draws = generator.choice(
        array, size=(BOOTSTRAP_REPLICATES, len(array)), replace=True
    ).mean(axis=1)
    mean = float(array.mean())
    ci_low, ci_high = np.quantile(draws, [ALPHA / 2, 1 - ALPHA / 2])
    if directional:
        p_value = float((1 + np.sum(draws <= 0)) / (BOOTSTRAP_REPLICATES + 1))
    else:
        p_value = float(min(
            1.0,
            2 * min(
                (1 + np.sum(draws <= 0)) / (BOOTSTRAP_REPLICATES + 1),
                (1 + np.sum(draws >= 0)) / (BOOTSTRAP_REPLICATES + 1),
            ),
        ))
    return {
        'n': int(len(array)), 'mean': mean,
        'ci_low': float(ci_low), 'ci_high': float(ci_high),
        'p_value': p_value,
    }

def bh_adjust(values):
    values = np.asarray(values, dtype=float)
    order = np.argsort(values)
    adjusted = np.empty_like(values)
    running = 1.0
    for reverse_rank in range(len(values) - 1, -1, -1):
        index = order[reverse_rank]
        rank = reverse_rank + 1
        running = min(running, values[index] * len(values) / rank)
        adjusted[index] = running
    return adjusted

test_rows = []
for donor_task in DONOR_TASKS:
    for target_task in TARGET_TASKS:
        values = estimand_frame[
            (estimand_frame['donor_task'] == donor_task)
            & (estimand_frame['target_task'] == target_task)
        ]['target_aligned_effect'].to_numpy(float)
        test_rows.append({
            'test_family': 'transport_cell',
            'donor_task': donor_task, 'target_task': target_task,
            'metric': 'target_aligned_effect',
            'primary_test': bool(
                donor_task != target_task
                and target_task in ('weight_more', 'weight_less')
            ),
            **bootstrap_summary(
                values, f'cell:{donor_task}:{target_task}', directional=True
            ),
        })

aggregate_rows = []
for family_id, frame in estimand_frame.groupby('family_id', sort=True):
    lookup = {
        (row.donor_task, row.target_task): row
        for row in frame.itertuples(index=False)
    }
    def aligned(donor, target):
        return float(lookup[(donor, target)].target_aligned_effect)
    def raw(donor, target):
        return float(lookup[(donor, target)].specific_raw_effect)
    cross_weight = np.mean([
        aligned('weight_more', 'weight_less'),
        aligned('weight_less', 'weight_more'),
    ])
    within_weight = np.mean([
        aligned('weight_more', 'weight_more'),
        aligned('weight_less', 'weight_less'),
    ])
    weight_all = np.mean([
        aligned(donor, target)
        for donor in DONOR_TASKS
        for target in ('weight_more', 'weight_less')
    ])
    distance_more = np.mean([
        aligned(donor, 'distance_more') for donor in DONOR_TASKS
    ])
    distance_less = np.mean([
        aligned(donor, 'distance_less') for donor in DONOR_TASKS
    ])
    distance_all = np.mean([distance_more, distance_less])
    distance_leakage = max(
        abs(raw(donor, target))
        for donor in DONOR_TASKS
        for target in ('distance_more', 'distance_less')
    )
    target_polarity = np.mean([
        aligned(donor, target)
        for donor in DONOR_TASKS for target in TARGET_TASKS
    ])
    donor_carryover = np.mean([
        float(donor_sign[donor]) * raw(donor, target)
        for donor in DONOR_TASKS for target in TARGET_TASKS
    ])
    donor_row_disagreement = np.mean([
        abs(raw('weight_more', target) - raw('weight_less', target))
        for target in TARGET_TASKS
    ])
    aggregate_rows.append({
        'family_id': family_id,
        'cross_query_weight_transfer': float(cross_weight),
        'within_query_weight_transfer': float(within_weight),
        'all_weight_transfer': float(weight_all),
        'distance_more_transfer': float(distance_more),
        'distance_less_transfer': float(distance_less),
        'distance_target_transfer': float(distance_all),
        'distance_leakage_max': float(distance_leakage),
        'factor_specificity': float(weight_all - distance_leakage),
        'target_polarity_transport': float(target_polarity),
        'donor_answer_carryover': float(donor_carryover),
        'donor_row_disagreement': float(donor_row_disagreement),
        'outcome_transfer': float(np.mean([
            aligned(donor, 'outcome') for donor in DONOR_TASKS
        ])),
    })
aggregate_frame = pd.DataFrame(aggregate_rows)

aggregate_metrics = (
    'cross_query_weight_transfer', 'within_query_weight_transfer',
    'all_weight_transfer', 'distance_more_transfer',
    'distance_less_transfer', 'distance_target_transfer',
    'distance_leakage_max', 'factor_specificity',
    'target_polarity_transport', 'donor_answer_carryover',
    'donor_row_disagreement', 'outcome_transfer',
)
for metric in aggregate_metrics:
    directional = metric not in (
        'factor_specificity', 'donor_row_disagreement'
    )
    test_rows.append({
        'test_family': 'aggregate', 'donor_task': 'both',
        'target_task': 'all', 'metric': metric,
        'primary_test': metric in (
            'cross_query_weight_transfer', 'factor_specificity',
            'target_polarity_transport', 'donor_answer_carryover',
        ),
        **bootstrap_summary(
            aggregate_frame[metric].to_numpy(float),
            f'aggregate:{metric}', directional=directional,
        ),
    })

tests_frame = pd.DataFrame(test_rows)
primary_indices = tests_frame.index[tests_frame['primary_test']].tolist()
tests_frame['q_value'] = np.nan
tests_frame.loc[primary_indices, 'q_value'] = bh_adjust(
    tests_frame.loc[primary_indices, 'p_value'].to_numpy(float)
)

def aggregate_test(metric):
    rows = tests_frame[
        (tests_frame['test_family'] == 'aggregate')
        & (tests_frame['metric'] == metric)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

def cell_test(donor, target):
    rows = tests_frame[
        (tests_frame['test_family'] == 'transport_cell')
        & (tests_frame['donor_task'] == donor)
        & (tests_frame['target_task'] == target)
    ]
    assert len(rows) == 1
    return rows.iloc[0]

cross_test = aggregate_test('cross_query_weight_transfer')
weight_test = aggregate_test('all_weight_transfer')
leakage_test = aggregate_test('distance_leakage_max')
specificity_test = aggregate_test('factor_specificity')
target_test = aggregate_test('target_polarity_transport')
donor_test = aggregate_test('donor_answer_carryover')
distance_test = aggregate_test('distance_target_transfer')
distance_more_test = aggregate_test('distance_more_transfer')
distance_less_test = aggregate_test('distance_less_transfer')
cross_cells = (
    cell_test('weight_more', 'weight_less'),
    cell_test('weight_less', 'weight_more'),
)
cross_query_passed = bool(
    cross_test['n'] >= MIN_FAMILIES
    and cross_test['ci_low'] > ROUTE_SESOI_LOGODDS
    and cross_test['q_value'] < ALPHA
    and all(row['ci_low'] > CELL_SESOI_LOGODDS for row in cross_cells)
)
distance_equivalence_passed = bool(
    leakage_test['ci_high'] < MAX_DISTANCE_LEAKAGE_LOGODDS
)
factor_specificity_passed = bool(
    specificity_test['ci_low'] > SPECIFICITY_SESOI_LOGODDS
    and specificity_test['q_value'] < ALPHA
)
distance_target_pattern_passed = bool(
    distance_test['ci_low'] > ROUTE_SESOI_LOGODDS
    and distance_more_test['ci_low'] > CELL_SESOI_LOGODDS
    and distance_less_test['ci_low'] > CELL_SESOI_LOGODDS
)
target_general_pattern_passed = bool(
    cross_query_passed
    and distance_target_pattern_passed
    and target_test['ci_low'] > ROUTE_SESOI_LOGODDS
    and target_test['q_value'] < ALPHA
)
donor_answer_pattern_passed = bool(
    donor_test['ci_low'] > ROUTE_SESOI_LOGODDS
    and donor_test['q_value'] < ALPHA
)

if cross_query_passed and distance_equivalence_passed and factor_specificity_passed:
    primary_decision = 'factor_specific_cross_query_weight_transport'
elif target_general_pattern_passed:
    primary_decision = 'target_conditioned_general_comparison_transport'
elif donor_answer_pattern_passed:
    primary_decision = 'donor_answer_carryover_transport'
elif cross_query_passed:
    primary_decision = 'cross_query_transport_without_factor_specificity'
else:
    primary_decision = 'no_stable_cross_query_transport'

decision_frame = pd.DataFrame([{
    'model_key': MODEL_KEY, 'site_label': SITE_LABEL,
    'primary_decision': primary_decision,
    'cross_query_weight_passed': cross_query_passed,
    'distance_equivalence_passed': distance_equivalence_passed,
    'factor_specificity_passed': factor_specificity_passed,
    'distance_target_pattern_passed': distance_target_pattern_passed,
    'target_general_pattern_passed': target_general_pattern_passed,
    'donor_answer_pattern_passed': donor_answer_pattern_passed,
    'cross_query_weight_effect': float(cross_test['mean']),
    'distance_target_effect': float(distance_test['mean']),
    'distance_leakage': float(leakage_test['mean']),
    'factor_specificity': float(specificity_test['mean']),
    'target_polarity_effect': float(target_test['mean']),
    'donor_answer_effect': float(donor_test['mean']),
}])
tests_frame.to_csv(ANALYSIS_DIR / 'cross_query_tests.csv', index=False)
aggregate_frame.to_csv(ANALYSIS_DIR / 'family_aggregate_metrics.csv', index=False)
decision_frame.to_csv(ANALYSIS_DIR / 'cross_query_decision.csv', index=False)
display(tests_frame.round(4))
display(decision_frame)
```

```python
%%cell_progress 14/17 Audit support, scope identity, norms, and replay
cell_stage('Auditing locked support and every intervention invariant')

task_semantic_violations = []
for family in transport_manifest:
    base_features = family['roles']['base']['features']
    source_features = family['roles']['aligned_source']['features']
    exact = str(base_features['outcome'])
    shortcut = str(base_features['strategy_predictions']['weight_only'])
    expected = {
        'outcome': (exact, exact),
        'weight_more': (shortcut, exact),
        'weight_less': (exact, shortcut),
        'distance_more': (exact, exact),
        'distance_less': (shortcut, shortcut),
    }
    for task, values in expected.items():
        observed = (
            task_semantic(base_features, task),
            task_semantic(source_features, task),
        )
        if observed != values:
            task_semantic_violations.append((family['family_id'], task, observed, values))
assert not task_semantic_violations, task_semantic_violations[:5]

self_audit = {
    'max_abs_logodds_shift': float(self_frame['max_centered_error'].max()),
    'mean_abs_logodds_shift': float(self_frame['max_centered_error'].mean()),
    'tolerance': SMOKE_CENTERED_LOGPROB_TOLERANCE,
}
assert self_audit['max_abs_logodds_shift'] < SMOKE_CENTERED_LOGPROB_TOLERANCE

norm_rows = []
for shard_path in sorted(SHARD_DIR.glob('*.json')):
    payload = json.loads(shard_path.read_text(encoding='utf-8'))
    raw_norms = payload['donor_raw_norms']
    matched = float(payload['matched_norm'])
    assert matched > 0
    applied = [
        float(value)
        for task_values in payload['applied_norms'].values()
        for value in task_values.values()
    ]
    relative_error = max(abs(value - matched) for value in applied) / matched
    norm_rows.append({
        'family_id': payload['family_id'],
        'weight_more_raw_norm': float(raw_norms['weight_more']),
        'weight_less_raw_norm': float(raw_norms['weight_less']),
        'matched_norm': matched,
        'matched_relative_error': float(relative_error),
    })
norm_frame = pd.DataFrame(norm_rows)
maximum_norm_relative_error = float(norm_frame['matched_relative_error'].max())
assert maximum_norm_relative_error < NORM_RELATIVE_TOLERANCE

patched_counts = estimand_frame['patched_tokens'].unique().tolist()
assert len(patched_counts) == 1 and patched_counts[0] > 0
behavior_accuracy = baseline_frame.groupby('target_task')['correct'].mean()
robustness_payload = {
    'schema_version': 'cross-query-factor-decision-audit-1.0',
    'model_key': MODEL_KEY,
    'prior_manifest_sha256': prior_manifest_sha256,
    'families': len(transport_manifest),
    'same_locked_manifest_as_inverse_query': True,
    'physics_violations': len(physics_violations),
    'task_semantic_violations': len(task_semantic_violations),
    'fixed_option_mapping_across_tasks': True,
    'cross_task_scope_token_ids_identical_in_smoke': True,
    'transport_scope': SITE_SCOPE,
    'patched_token_count': int(patched_counts[0]),
    'behavior_accuracy_by_task': {
        str(key): float(value) for key, value in behavior_accuracy.items()
    },
    'self_replay': self_audit,
    'donor_directions_norm_matched_within_family': True,
    'maximum_norm_relative_error': maximum_norm_relative_error,
    'norm_relative_tolerance': NORM_RELATIVE_TOLERANCE,
}
write_json_atomic(ANALYSIS_DIR / 'robustness_audit.json', robustness_payload)
norm_frame.to_csv(ANALYSIS_DIR / 'norm_matching_audit.csv', index=False)
display(pd.DataFrame([robustness_payload]))
```

```python
%%cell_progress 17/17 Run the exact-manifest cross-model synthesis
cell_stage('Comparing model-adaptive abstractions only when both runs exist')

decision_paths = {
    key: SHARED_ROOT / 'models' / spec['model_slug']
    / 'analysis' / 'cross_query_decision.json'
    for key, spec in MODEL_SPECS.items()
}
available = {
    key: json.loads(path.read_text(encoding='utf-8'))
    for key, path in decision_paths.items() if path.is_file()
}
print('Available model decisions:', sorted(available))
missing = [key for key in PRIMARY_MODEL_KEYS if key not in available]
if missing:
    print('Cross-model synthesis deferred. Still run:', missing)
else:
    hashes = {available[key]['prior_manifest_sha256'] for key in PRIMARY_MODEL_KEYS}
    assert len(hashes) == 1
    rows = []
    for key in PRIMARY_MODEL_KEYS:
        payload = available[key]
        metrics = payload['decision_metrics']
        rows.append({
            'model_key': key,
            'site_label': payload['frozen_site'][2],
            'layer_index': payload['frozen_site'][0],
            'primary_decision': payload['primary_decision'],
            'cross_query_weight_effect': metrics['cross_query_weight_effect'],
            'distance_target_effect': metrics['distance_target_effect'],
            'factor_specificity': metrics['factor_specificity'],
            'target_polarity_effect': metrics['target_polarity_effect'],
            'donor_answer_effect': metrics['donor_answer_effect'],
        })
    cross_model_frame = pd.DataFrame(rows)
    same_class = cross_model_frame['primary_decision'].nunique() == 1
    if same_class:
        interpretation = (
            'The two Qwen scales select the same abstraction class on the '
            'identical locked support, despite their different adaptive depths.'
        )
    else:
        interpretation = (
            'The two Qwen scales select different abstraction classes on the '
            'same support. Treat this as model-specific organization, not a '
            'monotonic scaling law.'
        )
    CROSS_MODEL_DIR.mkdir(parents=True, exist_ok=True)
    cross_model_frame.to_csv(
        CROSS_MODEL_DIR / 'cross_model_abstraction_summary.csv', index=False
    )
    write_json_atomic(CROSS_MODEL_DIR / 'cross_model_decision.json', {
        'schema_version': 'cross-query-cross-model-synthesis-1.0',
        'prior_manifest_sha256': next(iter(hashes)),
        'same_abstraction_class': bool(same_class),
        'model_decisions': {
            key: available[key]['primary_decision'] for key in PRIMARY_MODEL_KEYS
        },
        'interpretation': interpretation,
    })
    display(cross_model_frame)
    print(interpretation)
    print('Cross-model outputs:', CROSS_MODEL_DIR)
```
