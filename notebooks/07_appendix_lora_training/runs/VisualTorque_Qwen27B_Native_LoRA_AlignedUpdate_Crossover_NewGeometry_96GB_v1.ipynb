```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 4/17 Audit exposure and render 24 new geometry families
cell_stage('Audit exposure and render 24 new geometry families')
def physical_key(values):
    x=tuple(map(int,values));return min(x,x[2:]+x[:2])

def outcome(values):
    u,d,v,e=values;t=u*d-v*e
    return 'left' if t>0 else 'right' if t<0 else 'balanced'

def collect_geometries(obj):
    found=set()
    def visit(x):
        if isinstance(x,dict):
            keys=('left_units','left_distance','right_units','right_distance')
            if all(k in x for k in keys):found.add(physical_key([x[k] for k in keys]))
            vals=x.get('values')
            if isinstance(vals,(list,tuple)) and len(vals)==4 and all(isinstance(v,(int,float)) for v in vals):found.add(physical_key(vals))
            stacks=x.get('stacks')
            if isinstance(stacks,list) and len(stacks)==2 and all(isinstance(s,dict) and set(('units','distance','side'))<=set(s) for s in stacks):
                by={s['side']:s for s in stacks}
                if set(by)=={'left','right'}:found.add(physical_key([by['left']['units'],by['left']['distance'],by['right']['units'],by['right']['distance']]))
            for v in x.values():visit(v)
        elif isinstance(x,list):
            for v in x:visit(v)
    visit(obj);return found

def family_geometries(family):return {physical_key(v) for v in family['values'].values()}

def select_families(excluded,n):
    # Solely physical enumeration: all before/after truths and disjointness are known before inference.
    neutral={u:[] for u in range(2,MAX_UNITS-DELTA_UNITS+1)};targets=[]
    for u in range(2,MAX_UNITS-DELTA_UNITS+1):
        for d,v,e in itertools.product(range(1,MAX_DISTANCE+1),range(1,MAX_UNITS+1),range(1,MAX_DISTANCE+1)):
            before=(u,d,v,e);after=(u+DELTA_UNITS,d,v,e)
            if any(physical_key(x) in excluded for x in (before,after)):continue
            if outcome(before)==outcome(after)=='right':neutral[u].append((before,after))
        for v,e,df,ds in itertools.product(range(1,MAX_UNITS+1),range(1,MAX_DISTANCE+1),range(2,MAX_DISTANCE+1),range(1,MAX_DISTANCE)):
            if ds>=df:continue
            vals={'flip_before':(u,df,v,e),'flip_after':(u+DELTA_UNITS,df,v,e),
                  'stable_before':(u,ds,v,e),'stable_after':(u+DELTA_UNITS,ds,v,e)}
            if [outcome(vals[k]) for k in ('flip_before','flip_after','stable_before','stable_after')]!=['right','left','right','right']:continue
            keys={physical_key(x) for x in vals.values()}
            if len(keys)==4 and not keys&excluded:targets.append(vals)
    targets.sort(key=stable_key)
    # Preserve scarce recipient configurations when assigning neutral donors; this is a geometry-only packing rule.
    target_uses=Counter(physical_key(v) for vals in targets for v in vals.values())
    used=set(excluded);selected=[]
    for vals in targets:
        keys={physical_key(x) for x in vals.values()}
        if keys&used:continue
        u=vals['flip_before'][0];source=[];reserved=keys.copy()
        pool=sorted(neutral[u],key=lambda x:(sum(target_uses[physical_key(v)] for v in x),stable_key([vals,x])))
        for pair in pool:
            new={physical_key(x) for x in pair}
            if new&(used|reserved):continue
            source.append(pair);reserved|=new
            if len(source)==2:break
        if len(source)!=2:continue
        full=dict(vals)
        for i,pair in enumerate(source):
            full[f'source{i}_before'],full[f'source{i}_after']=pair
        fid='physical_update_'+stable_key(full)[:18]
        selected.append({'family_id':fid,'values':full,'before_units':u,'delta_units':DELTA_UNITS})
        used|=reserved
        if len(selected)==n:break
    require(len(selected)==n,f'Only {len(selected)} disjoint families available, requested {n}. No automatic relaxation. Inspect the exposure census and predeclare a new protocol if needed.')
    return selected,{'eligible_receiver_contexts':len(targets),'neutral_pairs_by_units':{str(k):len(v) for k,v in neutral.items()},'excluded_geometries':len(excluded)}
def read_dataset_metadata():
    names=('dataset_config.json','scenes.jsonl','counterfactual_pairs.jsonl','questions.jsonl','splits.json','statistics.json')
    digest=hashlib.sha256()
    directory = PROJECT_ROOT/'data/dataset_v1'
    if (directory/'scenes.jsonl').is_file():
        for name in names:
            assert (directory/name).is_file(), directory/name
            SOURCE_PATHS['dataset_'+name] = directory/name
            digest.update(name.encode());digest.update(b'\0');digest.update((directory/name).read_bytes());digest.update(b'\0')
        assert 'sha256:'+digest.hexdigest()==adapter_manifest['dataset_fingerprint'], 'Dataset differs from the SFT dataset.'
        return read_jsonl(directory/'scenes.jsonl')
    archive = PROJECT_ROOT/'data/dataset_v1.zip'
    assert archive.is_file(), 'Need the original Dataset v1 scenes.jsonl, or data/dataset_v1.zip, to audit SFT geometry.'
    SOURCE_PATHS['dataset_archive'] = archive
    with ZipFile(archive) as z:
        found = [n for n in z.namelist() if n.rstrip('/').split('/')[-1]=='scenes.jsonl']
        assert len(found)==1, f'Ambiguous Dataset v1 archive: {found}'
        prefix=found[0][:-len('scenes.jsonl')]
        for name in names:
            data=z.read(prefix+name)
            digest.update(name.encode());digest.update(b'\0');digest.update(data);digest.update(b'\0')
        assert 'sha256:'+digest.hexdigest()==adapter_manifest['dataset_fingerprint'], 'Dataset differs from the SFT dataset.'
        return [json.loads(s) for s in z.read(found[0]).decode('utf-8').splitlines() if s.strip()]

dataset_scenes=read_dataset_metadata()
dataset_by_id={str(r['scene_id']):r for r in dataset_scenes}
require(len(dataset_by_id)==len(dataset_scenes),'Duplicate dataset scene IDs.')
train_ids={str(r['scene_id']) for r in read_jsonl(SOURCE_PATHS['sft_train'])}
dev_ids={str(r['scene_id']) for r in read_jsonl(SOURCE_PATHS['sft_dev'])}
require(train_ids and dev_ids and train_ids.isdisjoint(dev_ids),'Invalid SFT train/dev split.')
require((train_ids|dev_ids)<=set(dataset_by_id),'Missing SFT geometry metadata.')
exposure=collect_geometries(dataset_scenes)
require(exposure,'Dataset single-stack census is empty.')
exposure_audit=[{'group':'all_dataset_v1_single_stack','geometries':len(exposure)},
    {'group':'sft_train','scenes':len(train_ids),'single_stack_geometries':len(collect_geometries([dataset_by_id[s] for s in train_ids]))},
    {'group':'sft_dev','scenes':len(dev_ids),'single_stack_geometries':len(collect_geometries([dataset_by_id[s] for s in dev_ids]))}]
for path in EXPOSURE_MANIFESTS:
    obj=read_jsonl(path) if path.suffix=='.jsonl' else read_json(path)
    require(bool(obj),'Empty historical manifest: '+str(path))
    keys=collect_geometries(obj)
    if path.name not in ('depth_support.jsonl','frozen_scene_manifest.jsonl'):require(keys,'No single-stack geometry parsed: '+str(path))
    exposure|=keys;exposure_audit.append({'path':str(path),'sha256':sha(path),'single_stack_geometries':len(keys)})
EXPOSURE_HASHES={str(p):sha(p) for p in [*EXPOSURE_MANIFESTS,*SOURCE_PATHS.values()]}
families,pool_audit=select_families(exposure,N_FAMILIES)
require(not set().union(*(family_geometries(f) for f in families))&exposure,'Historical overlap.')
freeze(SHARED_ROOT/'geometry_selection.json',{'families':families,'pool':pool_audit,'exposure_audit':exposure_audit,
    'excluded_keys':sorted(exposure),'domain':'units<=8; distances<=8; +2 update','selection':'geometry only; no model outcomes'})
scene_rows=[]
for f in tqdm(families,desc='Render eight scenes and their mirrors',unit='family'):
    for role,values in f['values'].items():
        u,d,v,e=values;sid=f['family_id']+'__'+role
        ap=SHARED_ROOT/'selected_images'/(sid+'__anchor.png');mp=SHARED_ROOT/'selected_images'/(sid+'__mirror.png')
        # Render into a new temporary file; validate immutable outputs before replacing anything.
        temp=ap.with_suffix('.render.png')
        render_scene(Scene(sid,(Stack('L',Side.LEFT,d,u),Stack('R',Side.RIGHT,e,v))),temp,RenderStyle(**STYLE))
        if ap.is_file():require(sha(ap)==sha(temp),'Frozen image changed: '+ap.name);temp.unlink()
        else:temp.replace(ap)
        temp=mp.with_suffix('.render.png')
        with Image.open(ap) as im:ImageOps.mirror(im.convert('RGB')).save(temp)
        if mp.is_file():require(sha(mp)==sha(temp),'Frozen mirror changed.');temp.unlink()
        else:temp.replace(mp)
        for orientation,path,vs in [('anchor',ap,values),('mirror',mp,values[2:]+values[:2])]:
            scene_rows.append({'family_id':f['family_id'],'scene_id':sid+'__'+orientation,'role':role,'orientation':orientation,
                'values':list(vs),'outcome':outcome(vs),'changed_side':'left' if orientation=='anchor' else 'right',
                'image_name':path.name,'image_sha256':sha(path)})
require(len(scene_rows)==N_FAMILIES*16,'Wrong geometry grid.')
require(len(set().union(*(family_geometries(f) for f in families)))==N_FAMILIES*8,'Cross-family geometry overlap.')
freeze(SHARED_ROOT/'scene_manifest.json',scene_rows)
SCENES={r['scene_id']:r for r in scene_rows}
SCENE_LOOKUP={(r['family_id'],r['orientation'],r['role']):r for r in scene_rows}
print('Frozen geometry:',N_FAMILIES,'families,',len(scene_rows),'rendered scenes; pool:',pool_audit)
display(pd.DataFrame([{'family':f['family_id'],'source0':f['values']['source0_before'],'source1':f['values']['source1_before'],
                      'flip_before':f['values']['flip_before'],'stable_before':f['values']['stable_before']} for f in families]))
```

```python
%%cell_progress 6/17 Define scoring, model modes and exact vector audits
cell_stage('Define scoring, model modes and exact vector audits')
INJECTION_AUDIT_SCHEMA='bf16-injection-vector-audit-2.0'
NUMERICAL_AUDIT_PATCH=IMPLEMENTATION
def audit_tensor_hash(t):
    t=t.detach().cpu().contiguous()
    return digest({'shape':list(t.shape),'dtype':str(t.dtype),
        'bytes_sha256':hashlib.sha256(t.view(torch.uint8).numpy().tobytes()).hexdigest()})

def audit_norm64(t):
    return float(torch.linalg.vector_norm(t.detach().cpu().double()))

def audit_injection(result,clean,payload,dose,input_key,condition_id):
    # Retain the actual pre-hook state: the post-hook state alone cannot distinguish
    # a correct addition from replay drift relative to the cached clean state.
    before=result['injection_before'].detach().cpu()
    cached=clean['states'][INJECTION_LAYER].detach().cpu()
    after=result['states'][INJECTION_LAYER].detach().cpu()
    cast=payload.detach().to(device='cpu',dtype=torch.bfloat16)
    require(before.shape==cached.shape==after.shape==cast.shape,'Injection audit: tensor shape mismatch.')
    require(before.dtype==cached.dtype==after.dtype==cast.dtype==torch.bfloat16,'Injection audit: expected BF16 tensors.')
    require(all(torch.isfinite(t).all() for t in (before,cached,after,cast)),'Injection audit: nonfinite tensor.')
    planned_after=cached+cast
    live_expected_after=before+cast
    actual_delta=after.float()-before.float()
    expected_delta=planned_after.float()-cached.float()
    actual_norm=norm(actual_delta)  # same CPU FP32 reporting convention as frozen v1 doses
    frozen_norm=float(dose['expected_actual_added_norm'])
    device_norm=float(result['edit']['actual_added_norm_device_fp32'])
    checks={'prestate_equal':torch.equal(before,cached),
        'hook_addition_equal':torch.equal(after,live_expected_after),
        'effective_delta_equal':torch.equal(actual_delta,expected_delta),
        'frozen_cpu_norm_equal':abs(norm(expected_delta)-frozen_norm)<1e-5}
    audit={'schema':INJECTION_AUDIT_SCHEMA,'patch_code_digest':NUMERICAL_AUDIT_PATCH,
        'passed':all(checks.values()),**checks,
        'cached_before_sha256':audit_tensor_hash(cached),'live_before_sha256':audit_tensor_hash(before),
        'expected_after_sha256':audit_tensor_hash(planned_after),'actual_after_sha256':audit_tensor_hash(after),
        'expected_delta_sha256':audit_tensor_hash(expected_delta),'actual_delta_sha256':audit_tensor_hash(actual_delta),
        'prestate_max_abs_error':float((before.float()-cached.float()).abs().max()),
        'hook_max_abs_error':float((after.float()-live_expected_after.float()).abs().max()),
        'delta_max_abs_error':float((actual_delta-expected_delta).abs().max()),
        'actual_norm_cpu_fp32':actual_norm,'expected_norm_cpu_fp32':norm(expected_delta),
        'actual_norm_cpu_fp64':audit_norm64(actual_delta),'expected_norm_cpu_fp64':audit_norm64(expected_delta),
        'frozen_expected_norm':frozen_norm,'device_norm_fp32':device_norm,
        'device_vs_frozen_abs_error':abs(device_norm-frozen_norm),
        'legacy_scalar_gate_would_pass':abs(device_norm-frozen_norm)<1e-5}
    if not audit['passed']:
        folder=OUTPUT_ROOT/'analysis/injection_diagnostics';folder.mkdir(parents=True,exist_ok=True)
        stem=input_key+'__'+condition_id
        tensor_path=folder/(stem+'.pt');json_path=folder/(stem+'.json')
        torch_atomic(tensor_path,{'cached_before':cached,'live_before':before,'actual_after':after,
            'payload_fp32':payload.detach().cpu(),'cast_payload':cast,'planned_after':planned_after})
        failed=[k for k,v in checks.items() if not v]
        write_json(json_path,{'run_id':RUN_ID,'input_key':input_key,'condition_id':condition_id,
            'failed_checks':failed,'audit':audit,'tensor_sha256':sha(tensor_path)})
        raise ValueError('Injection vector audit failed for '+input_key+' / '+condition_id+
            ': '+', '.join(failed)+'. Diagnostics: '+str(json_path)+
            '. Frozen doses and completed records were kept. Do not loosen the gate or delete locks.')
    result['edit']['actual_added_norm']=actual_norm
    return audit

def validate_injection_record(record):
    actual=float(record['edit']['actual_added_norm']);expected=float(record['dose']['expected_actual_added_norm'])
    require(math.isfinite(actual) and actual>=0 and math.isfinite(expected) and expected>=0,'Invalid saved injection norm.')
    require(abs(actual-expected)<1e-5,'Saved injection does not pass its frozen CPU/v1 norm gate.')
    audit=record.get('injection_audit')
    require(audit is not None,'New diagnostic records require the full vector audit.')
    require(audit['schema']==INJECTION_AUDIT_SCHEMA and audit['patch_code_digest']==NUMERICAL_AUDIT_PATCH,'Unknown injection audit version.')
    require(all(audit[k] is True for k in ('passed','prestate_equal','hook_addition_equal','effective_delta_equal','frozen_cpu_norm_equal')),'Failed saved vector checks.')
    for a,b in (('cached_before_sha256','live_before_sha256'),('expected_after_sha256','actual_after_sha256'),('expected_delta_sha256','actual_delta_sha256')):
        require(audit[a]==audit[b],'Saved injection vector hashes disagree.')
    require(all(audit[k]==0 for k in ('prestate_max_abs_error','hook_max_abs_error','delta_max_abs_error')),'Nonzero saved vector error.')
    require(audit['actual_norm_cpu_fp32']==actual and audit['expected_norm_cpu_fp32']==actual and audit['frozen_expected_norm']==expected,'Saved audit norm mismatch.')
    require(audit['actual_norm_cpu_fp64']==audit['expected_norm_cpu_fp64'] and math.isfinite(audit['actual_norm_cpu_fp64']),'Saved FP64 vector norm mismatch.')
    device=float(record['edit']['actual_added_norm_device_fp32'])
    require(math.isfinite(device) and device>=0 and audit['device_norm_fp32']==device,'Invalid device norm diagnostic.')
    require(audit['device_vs_frozen_abs_error']==abs(device-expected) and audit['legacy_scalar_gate_would_pass']==(abs(device-expected)<1e-5),'Device diagnostic mismatch.')
    return 'v2_exact_vectors'

def prepare(example):
    plan=adapter._candidate_plan(example)
    require(plan is not None,'A/B/C must be single-token candidates.')
    prefix,token_by_label=plan
    inputs=adapter._append_text_prefix(adapter._to_device(adapter._template([example.messages()],add_generation_prompt=True)),prefix)
    require(inputs['input_ids'].shape[0]==1,'All forwards use batch size one.')
    require(inputs['attention_mask'][0,-1].item()==1,'Final token must be unpadded.')
    fingerprint=digest({'ids':inputs['input_ids'].detach().cpu().tolist(),'attention':inputs['attention_mask'].detach().cpu().tolist(),
                       'token_by_label':{k:int(v) for k,v in token_by_label.items()}})
    return inputs,token_by_label,fingerprint
def hidden_tensor(output):
    if torch.is_tensor(output):return output
    if isinstance(output,(tuple,list)) and torch.is_tensor(output[0]):return output[0]
    raise TypeError('Unsupported decoder output: '+str(type(output)))
def replace_hidden(output,value):
    if torch.is_tensor(output):return value
    if isinstance(output,tuple):return (value,)+output[1:]
    if isinstance(output,list):return [value]+output[1:]
    raise TypeError(type(output))
def centered_score_error(a,b):
    a=np.asarray(a,dtype=float);b=np.asarray(b,dtype=float)
    return float(np.max(np.abs((a-a.mean())-(b-b.mean()))))
def forward(example,payload=None,expected=None):
    assert_native_precision();inputs,token_by_label,fingerprint=prepare(example)
    if expected is not None:require(fingerprint==expected,'Prepared input or candidate token identity changed.')
    states={};edit={};handles=[];injection_before={}
    def make_hook(layer):
        def hook(module,args,output):
            h=hidden_tensor(output);require(h.dtype==torch.bfloat16 and h.ndim==3,'Expected BF16 decoder state.')
            before=h[:, -1, :].detach().clone()
            if layer==INJECTION_LAYER and payload is not None:
                add=payload.to(device=h.device,dtype=h.dtype);require(add.shape==before.shape,'Payload shape mismatch.')
                after=before+add;changed=h.clone();changed[:,-1,:]=after
                injection_before['tensor']=before.cpu()
                edit.update(bf16_payload_norm=norm(add),actual_added_norm=norm((after.float()-before.float()).cpu()),
                            actual_added_norm_device_fp32=norm(after.float()-before.float()),
                            requested_fp32_norm=norm(payload),actual_cosine_to_requested=cosine(after.float()-before.float(),add.float()))
                states[layer]=after.cpu();return replace_hidden(output,changed)
            states[layer]=before.cpu()
        return hook
    try:
        for layer in CAPTURE_LAYERS:handles.append(DECODER_LAYERS[layer].register_forward_hook(make_hook(layer)))
        with torch.inference_mode():out=adapter.model(**inputs,use_cache=False)
        scores=torch.log_softmax(out.logits[0,-1].float(),dim=-1)
        values=[float(scores[token_by_label[k]].cpu()) for k in LABELS]
    finally:
        for h in handles:h.remove()
    require(set(states)==set(CAPTURE_LAYERS) and all(torch.isfinite(t).all() for t in states.values()),'Missing/nonfinite captured states.')
    require(np.isfinite(values).all(),'Nonfinite candidate scores.')
    idx=int(np.argmax(values));record={'candidate_labels':list(LABELS),'candidate_semantics':list(example.option_semantics),
        'candidate_logprobs':values,'predicted_label':LABELS[idx],'predicted_semantic':example.option_semantics[idx]}
    result={'score':record,'states':states,'fingerprint':fingerprint,'edit':edit}
    if payload is not None:
        require('tensor' in injection_before,'Injection hook was not called.')
        result['injection_before']=injection_before['tensor']
    return result
def clean_paths(key):return OUTPUT_ROOT/'clean_cache'/(key+'.pt'),OUTPUT_ROOT/'clean_records'/(key+'.json')
def load_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key);meta=read_json(jp)
    require(meta['run_id']==RUN_ID and meta['capture_key']==key and sha(tp)==meta['tensor_sha256'],'Clean capture identity/hash mismatch: '+key)
    data=torch.load(tp,map_location='cpu',weights_only=True)
    require(data['run_id']==RUN_ID and data['capture_key']==key,'Wrong tensor identity.')
    require(set(data['states'])==set(CAPTURE_LAYERS) and all(t.dtype==torch.bfloat16 and torch.isfinite(t).all() and norm(t)>0 for t in data['states'].values()),'Invalid clean tensor.')
    require(data['score']==meta['score'] and data['fingerprint']==meta['fingerprint'],'Clean metadata mismatch.')
    return data
def get_or_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key)
    if jp.is_file():return load_capture(row,mid)
    result=forward(make_example(row,mid));result.update(run_id=RUN_ID,capture_key=key)
    torch_atomic(tp,result)
    write_json(jp,{k:result[k] for k in ('run_id','capture_key','score','fingerprint')}|{'tensor_sha256':sha(tp)})
    return result
def semantic_scores(record):return dict(zip(record['candidate_semantics'],record['candidate_logprobs']))

# Use the active receiver for every forward, including smoke checks and restarts.
_forward_unscoped=forward
@contextmanager
def model_mode(mode):
    require(mode in MODES,'Unknown receiver mode.')
    if mode=='native' and hasattr(adapter.model,'disable_adapter'):
        with adapter.model.disable_adapter():yield
    elif mode=='native':yield  # pre-PEFT parity only
    else:
        require(hasattr(adapter.model,'disable_adapter'),'LoRA receiver needs an attached adapter.')
        yield
def forward(example,payload=None,expected=None):
    with model_mode(MODEL_MODE):return _forward_unscoped(example,payload,expected)

def activate(mode):
    global MODEL_MODE,OUTPUT_ROOT,RUN_ID
    require(mode in MODES,'Invalid active receiver.')
    MODEL_MODE=mode;OUTPUT_ROOT=SHARED_ROOT/mode;RUN_ID=RUN_IDS[mode]

@lru_cache(maxsize=4096)
def capture_for(mode,key):
    tp=SHARED_ROOT/mode/'clean_cache'/(key+'.pt');jp=SHARED_ROOT/mode/'clean_records'/(key+'.json')
    meta=read_json(jp);require(meta['run_id']==RUN_IDS[mode] and meta['capture_key']==key and sha(tp)==meta['tensor_sha256'],'Cross-model clean identity/hash mismatch.')
    data=torch.load(tp,map_location='cpu',weights_only=True)
    require(data['run_id']==RUN_IDS[mode] and data['capture_key']==key and data['score']==meta['score'] and data['fingerprint']==meta['fingerprint'],'Cross-model clean cache mismatch.')
    require(set(data['states'])==set(CAPTURE_LAYERS) and all(t.dtype==torch.bfloat16 and torch.isfinite(t).all() and norm(t)>0 for t in data['states'].values()),'Invalid cross-model clean state.')
    return data
def cc(row,mid):return capture_for(MODEL_MODE,capture_key(row,mid))
```

```python
%%cell_progress 7/17 Load base plus adapter and verify disabled-base parity
cell_stage('Load base plus adapter and verify disabled-base parity')
assert_native_precision()
# Release layer references too: they otherwise keep the previous 27B decoder on GPU when Run all is repeated.
for stale_name in ('DECODER_LAYERS','adapter'):
    globals().pop(stale_name,None)
gc.collect();torch.cuda.empty_cache()
adapter=HuggingFaceMultimodalAdapter(MODEL_ID,revision=MODEL_REVISION,dtype='bfloat16',device_map='cuda:0')
adapter.model.eval()
def resolve_module(root,path):
    for key in path.split('.'):root=getattr(root,key)
    return root
def discover_layers(model):
    for path in ('model.language_model.layers','model.text_model.layers','language_model.model.layers','model.model.layers','model.layers'):
        try:layers=resolve_module(model,path)
        except AttributeError:continue
        if isinstance(layers,torch.nn.ModuleList) and len(layers)==64:return path,layers
    raise RuntimeError('Expected 64-layer Qwen3.5 decoder not found.')
LAYER_PATH,DECODER_LAYERS=discover_layers(adapter.model)
require(all(p.dtype==torch.bfloat16 for p in adapter.model.parameters() if p.ndim>=2),'Base weights are not all BF16; no quantization is allowed.')

MODEL_MODE='native'
parity_examples=[make_example(*TARGET_INPUTS[i]) for i in (0,-1)]
bare=[forward(e) for e in parity_examples]
adapter.model=PeftModel.from_pretrained(adapter.model,str(ADAPTER_DIR),is_trainable=False)
adapter.model.eval()
require(any('lora_' in n for n,_ in adapter.model.named_parameters()),'Adapter missing.')
require(any(torch.count_nonzero(p).item()>0 for n,p in adapter.model.named_parameters() if 'lora_B' in n),'Adapter B weights are all zero.')
parity=[]
for e,bare_result in zip(parity_examples,bare):
    disabled=forward(e)
    same=all(torch.equal(bare_result['states'][l],disabled['states'][l]) for l in CAPTURE_LAYERS)
    error=centered_score_error(bare_result['score']['candidate_logprobs'],disabled['score']['candidate_logprobs'])
    require(same and error<=SCORE_TOLERANCE and bare_result['fingerprint']==disabled['fingerprint'],'Disabled adapter differs from bare base; do not proceed.')
    parity.append({'example_id':e.example_id,'exact_layer_states':same,'centered_score_error':error})
freeze(SHARED_ROOT/'disabled_base_parity.json',{'passed':True,'records':parity})
del bare
implementations=sorted({f'{rule.__module__}.{rule.__name__}' for module in adapter.model.modules()
    for rule in [getattr(module,'chunk_gated_delta_rule',None)] if rule is not None})
paths={inspect.getsourcefile(type(m)) for m in adapter.model.modules() if 'qwen3_5' in type(m).__module__}
runtime={'torch':torch.__version__,'transformers':importlib.metadata.version('transformers'),'peft':importlib.metadata.version('peft'),
    'numpy':np.__version__,'pandas':pd.__version__,'pillow':importlib.metadata.version('pillow'),'cuda':torch.version.cuda,
    'gpu':torch.cuda.get_device_name(0),'precision':assert_native_precision(),'decoder_path':LAYER_PATH,'batch_size':1,
    'linear_attention':implementations,'model_source_hashes':{Path(p).name:sha(p) for p in paths if p and Path(p).is_file()},
    'lora_parameter_dtypes':sorted({str(p.dtype) for n,p in adapter.model.named_parameters() if 'lora_' in n}),
    'synthetic_only':SYNTHETIC_ONLY}
RUNTIME_DIGEST=freeze(SHARED_ROOT/'runtime_lock.json',runtime)
RUN_IDS={mode:digest({'protocol':PROTOCOL_DIGEST,'runtime':RUNTIME_DIGEST,'mode':mode}) for mode in MODES}
for mode in MODES:freeze(SHARED_ROOT/mode/'run_identity.json',{'run_id':RUN_IDS[mode]})
activate('native')
print('One base + adapter loaded. Native parity passed; model modes share exact images and candidate tokens.')
```

```python
%%cell_progress 9/17 Freeze both component libraries and receiver dose matching
cell_stage('Freeze both component libraries and receiver dose matching')
def norm64(t):return float(torch.linalg.vector_norm(t.detach().cpu().double()))

def weight_payload(source_before,source_after,receiver_before):
    # No target after state, label, prediction or outcome is accepted here.
    require(source_before.shape==source_after.shape==receiver_before.shape,'Donor/receiver dimensions differ.')
    require(norm(source_before)>AXIS_EPS,'Zero donor baseline norm.')
    return (source_after.float()-source_before.float())*(norm(receiver_before)/norm(source_before))

def split_update(full,natural):
    v,g=full.double(),natural.double();gn=norm64(g)
    defined=gn>AXIS_EPS
    coefficient=float((v*g).sum()/(g*g).sum()) if defined else 0.
    a=coefficient*g if defined else torch.zeros_like(v);r=v-a
    scale=max(norm64(v),AXIS_EPS)
    require(norm64(v-a-r)<=1e-10*scale,'FP64 component closure failed.')
    require(not defined or abs(float((r*g).sum()))<=1e-10*max(norm64(r)*gn,1.),'FP64 residual orthogonality failed.')
    return a.float(),r.float(),{'axis_defined':defined,'signed_coefficient':coefficient,'full_norm64':norm64(v),
        'aligned_norm64':norm64(a),'residual_norm64':norm64(r),'natural_norm64':gn,
        'fp32_closure_relative_error':norm64(full-a.float()-r.float())/scale,
        'fp32_residual_projection':float((r.float().double()*g).sum()/gn) if defined else None}

def random_matched(shape,magnitude,seed,orthogonal_to=None):
    if magnitude<=AXIS_EPS:return torch.zeros(shape,dtype=torch.float32)
    generator=torch.Generator(device='cpu').manual_seed(seed)
    x=torch.randn(shape,generator=generator,dtype=torch.float64)
    if orthogonal_to is not None and norm64(orthogonal_to)>AXIS_EPS:
        q=orthogonal_to.double();x=x-(x*q).sum()/(q*q).sum()*q
    require(norm64(x)>AXIS_EPS,'Degenerate random control; no silent redraw.')
    return (x*(magnitude/norm64(x))).float()

def endpoint_flags(before,after,truth):
    a,b=before['score'],after['score']
    stable=a['predicted_semantic']==b['predicted_semantic']
    unique=all(np.sum(np.asarray(s['candidate_logprobs'])==max(s['candidate_logprobs']))==1 for s in (a,b))
    return {'prediction_stable':bool(stable),'stable_unique_top':bool(stable and unique),
        'source_both_correct':bool(a['predicted_semantic']==b['predicted_semantic']==truth),
        'source_confidence_change':float(sorted(b['candidate_logprobs'])[-1]-sorted(b['candidate_logprobs'])[-2]
                                      -sorted(a['candidate_logprobs'])[-1]+sorted(a['candidate_logprobs'])[-2])}

def displacement_geometry(delta,natural):
    d,g=delta.double(),natural.double();dn,gn=norm64(d),norm64(g)
    defined=gn>AXIS_EPS
    projection=float((d*g).sum()/gn) if defined else None
    return {'induced_norm':dn,'natural_cf_norm':gn,
        'projection_on_natural_cf':projection,'cosine_to_natural_cf':projection/dn if defined and dn>AXIS_EPS else None,
        'cf_aligned_fraction':projection/gn if defined else None,
        'relative_distance_to_cf':norm64(d-g)/gn if defined else None,
        'closeness_gain_over_clean':1-norm64(d-g)/gn if defined else None}

def tensor_fingerprint(t):return 'sha256:'+hashlib.sha256(t.contiguous().numpy().tobytes()).hexdigest()

DONORS={}
for library,f,ori,context,slot in itertools.product(MODES,families,('anchor','mirror'),range(2),range(2)):
    fid=f['family_id'];mid=source_mappings(fid,ori,context)[slot]
    rb=SCENE_LOOKUP[(fid,ori,f'source{context}_before')];ra=SCENE_LOOKUP[(fid,ori,f'source{context}_after')]
    before=capture_for(library,capture_key(rb,mid));after=capture_for(library,capture_key(ra,mid))
    require(rb['outcome']==ra['outcome'] and before['score']['candidate_semantics']==after['score']['candidate_semantics'],'Source neutrality/mapping failed.')
    DONORS[(library,fid,ori,context,slot)]={'before':before['states'][48],'after':after['states'][48],
        'source_semantic':rb['outcome'],'source_label':LABELS[PERMUTATIONS[mid].index(rb['outcome'])],
        'source_mapping':mid,'source_capture_keys':[capture_key(rb,mid),capture_key(ra,mid)],**endpoint_flags(before,after,rb['outcome'])}

@lru_cache(maxsize=8192)
def local_components(library,key,context,slot):
    row,mid=CAPTURE_LOOKUP[key];own=capture_for(library,key)['states'][48]
    after=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    g=capture_for(library,capture_key(after,mid))['states'][48].float()-own.float()
    donor=DONORS[(library,row['family_id'],row['orientation'],context,slot)]
    v=weight_payload(donor['before'],donor['after'],own);a,r,meta=split_update(v,g)
    source={k:donor[k] for k in ('source_semantic','source_label','source_mapping','source_capture_keys','prediction_stable','stable_unique_top','source_both_correct','source_confidence_change')}
    return v,a,r,g,{**source,**meta,'library_receiver_clean_norm':norm64(own),'zero_full_update':norm64(v)<=AXIS_EPS}

def make_payload(row,mid,plan,clean):
    own=clean['states'][48];key=capture_key(row,mid);kind=plan['kind']
    if plan['context']<0:
        after=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
        g=cc(after,mid)['states'][48].float()-own.float()
        p=torch.zeros_like(own,dtype=torch.float32) if kind=='zero' else g*(.5 if kind=='cf_half' else 1.)
        return p,{'axis_defined':norm64(g)>AXIS_EPS,'signed_coefficient':None,'prediction_stable':None,'stable_unique_top':None,
            'source_both_correct':None,'source_mapping':None,'source_semantic':None,'source_label':None,'source_capture_keys':[],
            'library':'shared','receiver':MODEL_MODE,'receiver_rescale':1.,'library_receiver_clean_norm':norm64(own)}
    library=plan['library'];v,a,r,g,meta=local_components(library,key,plan['context'],plan['source_mapping_slot'])
    # Identical Gaussian seed across library/model cells; match within each library.
    seedbase=[key,plan['context'],plan['source_mapping_slot']]
    if kind=='full':p=v
    elif kind=='half':p=.5*v
    elif kind=='aligned':p=a
    elif kind=='residual':p=r
    elif kind=='aligned_half_residual':p=a+.5*r
    elif kind.startswith('random_aligned_'):
        p=random_matched(a.shape,norm64(a),int(stable_key(['random_A',*seedbase,int(kind[-1])])[:15],16))
    elif kind.startswith('rotated_residual_'):
        p=a+random_matched(r.shape,norm64(r),int(stable_key(['rotate_R',*seedbase,int(kind[-1])])[:15],16),g)
    else:raise ValueError(kind)
    factor=norm64(own)/meta['library_receiver_clean_norm']
    return (p*factor).contiguous(),{**meta,'library':library,'receiver':MODEL_MODE,'receiver_rescale':factor}

def freeze_doses(mode):
    activate(mode);doses=[];components=[];matches=[]
    for row,mid in tqdm(TARGET_INPUTS,desc=mode+': freeze crossed payloads',unit='input'):
        key=capture_key(row,mid);clean=cc(row,mid);own=clean['states'][48]
        after=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
        g=cc(after,mid)['states'][48].float()-own.float();items={}
        for plan in PLAN:
            p,meta=make_payload(row,mid,plan,clean);cast=p.to(torch.bfloat16)
            require(torch.isfinite(p).all() and torch.isfinite(cast).all(),'Nonfinite payload.')
            effective=(own+cast).float()-own.float()
            d={'input_key':key,'condition_id':plan['condition_id'],'payload_sha256':tensor_fingerprint(p),
                'requested_norm':norm(p),'bf16_norm':norm(cast),'relative_norm':norm(cast)/norm(own),
                'requested_relative_norm64':norm64(p)/norm64(own),'effective_relative_norm64':norm64(effective)/norm64(own),
                'expected_actual_added_norm':norm(effective),'quantization_error_norm':norm64(effective-p),**meta,
                **{'requested_'+k:v for k,v in displacement_geometry(p,g).items()},
                **{'effective_'+k:v for k,v in displacement_geometry(effective,g).items()}}
            doses.append(d);items[plan['condition_id']]=(p,effective,d)
        for library,context,slot in itertools.product(MODES,range(2),range(2)):
            def item(kind):return items[f'{library}__{kind}__s{context}__m{slot}']
            v,a,r,axis,meta=local_components(library,key,context,slot)
            closure=item('full')[1]-item('aligned')[1]-item('residual')[1]
            pair={'receiver':mode,'library':library,'family_id':row['family_id'],'input_key':key,'context':context,'source_mapping_slot':slot,
                'receiver_context':row['role'].split('_')[0]}
            components.append({**pair,**meta,'effective_component_closure_relative':norm64(closure)/max(norm64(item('full')[1]),AXIS_EPS),
                'effective_A_norm':norm64(item('aligned')[1]),'clean_receiver_norm':norm64(own),
                'natural_axis_cross_cosine':cosine(axis,g)})
            for kind,reference in [(f'random_aligned_{s}','aligned') for s in range(2)]+[(f'rotated_residual_{s}','full') for s in range(2)]:
                p,eff,d=item(kind);ref,refeff,rd=item(reference)
                requested=abs(norm64(p)-norm64(ref))/max(norm64(ref),AXIS_EPS)
                require(requested<2e-6,'Requested control norm match failed.')
                # Rotation preserves the LIBRARY axis, not necessarily the receiver axis.
                axis_error=abs(float(((p-ref).double()*axis.double()).sum()))/max(norm64(ref)*norm64(axis),AXIS_EPS)
                if kind.startswith('rotated'):require(axis_error<2e-6,'Rotated R changed the library-axis projection.')
                matches.append({**pair,'kind':kind,'reference':reference,'requested_norm_relative_error':requested,
                    'effective_norm_relative_error':abs(norm64(eff)-norm64(refeff))/max(norm64(refeff),AXIS_EPS),
                    'library_projection_error_relative':axis_error if kind.startswith('rotated') else None})
    dose_digest=freeze(OUTPUT_ROOT/'frozen_doses.json',doses)
    pd.DataFrame(doses).to_csv(OUTPUT_ROOT/'analysis/dose_inventory.csv',index=False)
    pd.DataFrame(components).to_csv(OUTPUT_ROOT/'analysis/component_inventory.csv',index=False)
    pd.DataFrame(matches).to_csv(OUTPUT_ROOT/'analysis/control_matching_audit.csv',index=False)
    require(len(doses)==len(TARGET_INPUTS)*75 and len(components)==len(TARGET_INPUTS)*8,'Incomplete dose grid.')
    return dose_digest,{(r['input_key'],r['condition_id']):r for r in doses}

DOSE_DIGESTS={};DOSES_BY_MODE={}
for mode in MODES:DOSE_DIGESTS[mode],DOSES_BY_MODE[mode]=freeze_doses(mode)
crossdose=[]
for key,d in DOSES_BY_MODE['native'].items():
    if d['library']=='shared':continue
    e=DOSES_BY_MODE['lora'][key];a,b=d['requested_relative_norm64'],e['requested_relative_norm64']
    err=abs(a-b)/max(a,AXIS_EPS)
    require(err<2e-6,'Requested relative dose differs across receiver models.')
    crossdose.append({'input_key':key[0],'condition_id':key[1],'library':d['library'],'requested_error':err,
        'effective_error':abs(d['effective_relative_norm64']-e['effective_relative_norm64'])/max(d['effective_relative_norm64'],AXIS_EPS)})
pd.DataFrame(crossdose).to_csv(SHARED_ROOT/'analysis/cross_receiver_dose_matching.csv',index=False)
print('All 2x2 dose inventories frozen. Actual BF16 matching tails retained for sensitivity analysis.')
```

```python
%%cell_progress 10/17 Define resumable crossed intervention runner
cell_stage('Define resumable crossed intervention runner')
def shard_path(row,mid):return OUTPUT_ROOT/'intervention_shards'/(capture_key(row,mid)+'.json')
def shard_identity(row,mid):return {'run_id':RUN_ID,'all_clean_digest':ALL_CLEAN_DIGEST,'dose_digest':DOSE_DIGEST,'input_key':capture_key(row,mid)}
def validate_shard(shard,row,mid):
    require(shard['identity']==shard_identity(row,mid),'Shard identity mismatch.')
    ids=[r['condition']['condition_id'] for r in shard['records']]
    require(len(ids)==len(set(ids)) and set(ids)<={p['condition_id'] for p in PLAN},'Duplicate/unknown conditions.')
    plan_lookup={p['condition_id']:p for p in PLAN}
    for r in shard['records']:
        body={k:v for k,v in r.items() if k!='record_digest'}
        require(digest(body)==r['record_digest'],'Record checksum failed.')
        p=r['condition'];require(p==plan_lookup[p['condition_id']],'Condition metadata changed.')
        require(r['dose']==DOSES[(capture_key(row,mid),p['condition_id'])],'Frozen dose mismatch.')
        s=r['score'];require(s['candidate_labels']==list(LABELS) and s['candidate_semantics']==list(PERMUTATIONS[mid]),'Candidate map mismatch.')
        values=np.array(s['candidate_logprobs']);require(values.shape==(3,) and np.isfinite(values).all(),'Invalid candidate scores.')
        idx=int(values.argmax());require(s['predicted_label']==LABELS[idx] and s['predicted_semantic']==PERMUTATIONS[mid][idx],'Score/prediction mismatch.')
        require(set(r['traces'])==set(map(str,CAPTURE_LAYERS)),'Missing traces.')
        validate_injection_record(r)
    return shard

def run_one(row,mid):
    path=shard_path(row,mid)
    shard=validate_shard(read_json(path),row,mid) if path.is_file() else {'identity':shard_identity(row,mid),'records':[]}
    done={r['condition']['condition_id'] for r in shard['records']}
    if len(done)==len(PLAN):return
    clean=cc(row,mid);after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    natural=cc(after_row,mid);e=make_example(row,mid)
    for p in PLAN:
        if p['condition_id'] in done:continue
        payload,_=make_payload(row,mid,p,clean);dose=DOSES[(capture_key(row,mid),p['condition_id'])]
        require('sha256:'+hashlib.sha256(payload.contiguous().numpy().tobytes()).hexdigest()==dose['payload_sha256'],'Payload changed after freezing.')
        result=forward(e,payload,expected=clean['fingerprint'])
        injection_audit=audit_injection(result,clean,payload,dose,capture_key(row,mid),p['condition_id'])
        traces={}
        for layer in CAPTURE_LAYERS:
            delta=result['states'][layer].float()-clean['states'][layer].float()
            natural_delta=natural['states'][layer].float()-clean['states'][layer].float()
            traces[str(layer)]=displacement_geometry(delta,natural_delta)
        record={'condition':p,'score':result['score'],'dose':dose,'edit':result['edit'],'traces':traces,'injection_audit':injection_audit}
        record['record_digest']=digest(record);shard['records'].append(record);write_json(path,shard)
    validate_shard(shard,row,mid)

def run_all():
    start=time.monotonic();new_done=0
    for i,(row,mid) in enumerate(tqdm(TARGET_INPUTS,desc=MODEL_MODE+': frozen transfer conditions',unit='input')):
        path=shard_path(row,mid);complete=path.is_file() and len(read_json(path).get('records',[]))==len(PLAN)
        run_one(row,mid)
        if not complete:
            new_done+=1
            if new_done in (1,8):print('Remaining ETA from completed inputs:',round((time.monotonic()-start)/new_done*(len(TARGET_INPUTS)-i-1)/3600,2),'hours')

def run_receiver(mode):
    global DOSES,DOSE_DIGEST
    activate(mode);DOSES=DOSES_BY_MODE[mode];DOSE_DIGEST=DOSE_DIGESTS[mode]
    technical=[]
    for row,mid in (TARGET_INPUTS[0],TARGET_INPUTS[-1]):
        clean=cc(row,mid);e=make_example(row,mid);replay=forward(e,expected=clean['fingerprint'])
        require(all(torch.equal(replay['states'][l],clean['states'][l]) for l in CAPTURE_LAYERS),'Clean state replay changed.')
        require(centered_score_error(replay['score']['candidate_logprobs'],clean['score']['candidate_logprobs'])<=SCORE_TOLERANCE,'Clean score replay changed.')
        for library,kind in [('shared','zero'),('native','aligned'),('lora','aligned'),('native','rotated_residual_0'),('lora','rotated_residual_0')]:
            plan=next(p for p in PLAN if p['kind']==kind and p['library']==library)
            payload,_=make_payload(row,mid,plan,clean);dose=DOSES[(capture_key(row,mid),plan['condition_id'])]
            result=forward(e,payload,expected=clean['fingerprint'])
            audit=audit_injection(result,clean,payload,dose,capture_key(row,mid),plan['condition_id'])
            technical.append({'input_key':capture_key(row,mid),'condition_id':plan['condition_id'],'passed':audit['passed']})
    freeze(OUTPUT_ROOT/'technical_controls.json',technical)
    run_all()
```

```python
%%cell_progress 13/17 Verify both complete grids and state geometry
cell_stage('Verify both complete grids and state geometry')
FRAMES=[];TRACES=[]
for mode in MODES:
    activate(mode);DOSES=DOSES_BY_MODE[mode];DOSE_DIGEST=DOSE_DIGESTS[mode]

    records=[];trace_rows=[];shard_hashes={}
    for row,mid in tqdm(TARGET_INPUTS,desc='Verify completed grid',unit='input'):
        path=shard_path(row,mid);shard=validate_shard(read_json(path),row,mid)
        require(len(shard['records'])==len(PLAN),'Incomplete condition grid.')
        shard_hashes[path.name]=sha(path);clean=cc(row,mid)
        cfrow=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
        natural=cc(cfrow,mid);baseline=semantic_scores(clean['score']);ns=semantic_scores(natural['score'])
        changed=row['changed_side'];opposite='right' if changed=='left' else 'left';cfsem=cfrow['outcome']
        for record in shard['records']:
            s=record['score'];ss=semantic_scores(s);plan=record['condition'];dose=record['dose']
            base={'family_id':row['family_id'],'input_key':capture_key(row,mid),'scene_id':row['scene_id'],
                'orientation':row['orientation'],'receiver_context':row['role'].split('_')[0],'receiver_mapping':mid,**plan,
                'predicted_semantic':s['predicted_semantic'],'predicted_label':s['predicted_label'],
                'margin_shift':ss[changed]-ss[opposite]-baseline[changed]+baseline[opposite],
                'natural_margin_shift':ns[changed]-ns[opposite]-baseline[changed]+baseline[opposite],
                'cf_accuracy_gain':float(s['predicted_semantic']==cfsem)-float(clean['score']['predicted_semantic']==cfsem),
                'cf_success':s['predicted_semantic']==cfsem,'clean_cf_success':clean['score']['predicted_semantic']==cfsem,
                'natural_cf_correct':natural['score']['predicted_semantic']==cfsem,'clean_correct':clean['score']['predicted_semantic']==row['outcome'],
                'cf_best_other_gain':(ss[cfsem]-max(v for k,v in ss.items() if k!=cfsem))-(baseline[cfsem]-max(v for k,v in baseline.items() if k!=cfsem)),
                'axis_defined':dose['axis_defined'],'prediction_stable':dose['prediction_stable'],'stable_unique_top':dose['stable_unique_top'],
                'source_both_correct':dose['source_both_correct'],'signed_coefficient':dose['signed_coefficient'],
                'relative_payload_norm':dose['relative_norm'],'actual_added_norm':record['edit']['actual_added_norm'],
                'injection_audit_level':validate_injection_record(record),**{'logprob_'+k:v for k,v in zip(LABELS,s['candidate_logprobs'])}}
            records.append(base)
            for layer,t in record['traces'].items():
                # Direct distance is captured from FP64 vectors; also reconstruct it from scalar traces.
                n,m=t['induced_norm'],t['natural_cf_norm'];p=t['projection_on_natural_cf']
                require(math.isfinite(n) and math.isfinite(m) and n>=0 and m>=0,'Invalid trace norm.')
                if m>AXIS_EPS:
                    sq=n*n+m*m-2*p*m
                    require(sq>=-1e-9*max(n*n+m*m,1),'Impossible trace geometry.')
                    require(math.isclose(math.sqrt(max(0,sq))/m,t['relative_distance_to_cf'],rel_tol=1e-7,abs_tol=1e-7),'Distance reconstruction failed.')
                trace_rows.append({k:base[k] for k in ('family_id','input_key','condition_id','kind','context','source_mapping_slot','receiver_context','orientation','receiver_mapping')}|{'layer':int(layer),**t})
    frame=pd.DataFrame(records);traces=pd.DataFrame(trace_rows)
    require(len(frame)==len(TARGET_INPUTS)*len(PLAN) and not frame.duplicated(['input_key','condition_id']).any(),'Incomplete/duplicate intervention rows.')
    require(len(traces)==3*len(frame),'Trace count mismatch.')
    require(np.isfinite(frame[['margin_shift','cf_accuracy_gain','cf_best_other_gain','actual_added_norm']].to_numpy()).all(),'Nonfinite scores.')
    zero=frame[frame.kind=='zero']
    require(zero.margin_shift.abs().max()<=SCORE_TOLERANCE and zero.actual_added_norm.max()==0,'Zero control failed.')
    require(set(frame.injection_audit_level)=={'v2_exact_vectors'},'Every new record must have the vector audit.')
    frame.to_csv(OUTPUT_ROOT/'analysis/intervention_records.csv',index=False)
    traces.to_csv(OUTPUT_ROOT/'analysis/state_response_traces.csv',index=False)
    freeze(OUTPUT_ROOT/'shard_inventory.json',shard_hashes)
    print('Complete:',len(frame),'interventions;',len(traces),'traces; all actual BF16 additions verified.')

    frame['receiver']=mode;traces['receiver']=mode
    libmap={p['condition_id']:p['library'] for p in PLAN};traces['library']=traces.condition_id.map(libmap)
    FRAMES.append(frame);TRACES.append(traces)
frame=pd.concat(FRAMES,ignore_index=True);traces=pd.concat(TRACES,ignore_index=True)
require(len(frame)==N_FAMILIES*24*75*2 and len(traces)==3*len(frame),'Crossed total mismatch.')
require(not frame.duplicated(['receiver','input_key','condition_id']).any(),'Duplicate crossed records.')
frame.to_csv(SHARED_ROOT/'analysis/intervention_records.csv',index=False)
traces.to_csv(SHARED_ROOT/'analysis/state_response_traces.csv',index=False)
print('Both completed grids assembled:',len(frame),'interventions;',len(traces),'state traces.')
```

```python
%%cell_progress 14/17 Run six paired family tests and behavior comparisons
cell_stage('Run six paired family tests and behavior comparisons')
def estimate(values,key,test=False):
    values=pd.Series(values,dtype=float).dropna().sort_index()
    require(values.index.is_unique,'Statistics require one value per family.')
    if len(values)<2:return {'n_families':len(values),'estimate':float(values.mean()) if len(values) else None,'ci_low':None,'ci_high':None,'p_two_sided':None}
    x=values.to_numpy();require(np.isfinite(x).all(),'Invalid family effects.')
    rng=np.random.default_rng(int(stable_key(['stats',key])[:15],16));means=[]
    for start in range(0,BOOTSTRAPS,256):
        count=min(256,BOOTSTRAPS-start);means.extend(x[rng.integers(len(x),size=(count,len(x)))].mean(1))
    out={'n_families':len(x),'estimate':float(x.mean()),'ci_low':float(np.quantile(means,.025)),'ci_high':float(np.quantile(means,.975))}
    if test:
        exceed=0
        for start in range(0,SIGNFLIPS,1024):
            count=min(1024,SIGNFLIPS-start);signs=rng.integers(0,2,size=(count,len(x)))*2-1
            exceed+=int((np.abs((x*signs).mean(1))>=abs(x.mean())-1e-12).sum())
        out['p_two_sided']=(exceed+1)/(SIGNFLIPS+1)
    return out

def holm(pvalues):
    out=np.zeros(len(pvalues));running=0.
    for rank,i in enumerate(np.argsort(pvalues)):
        running=max(running,(len(pvalues)-rank)*pvalues[i]);out[i]=min(1.,running)
    return out

PAIR=['family_id','input_key','context','source_mapping_slot']
def paired_values(receiver,library,context,metric):
    sub=frame[(frame.receiver==receiver)&(frame.receiver_context==context)]
    w=sub[(sub.library==library)&(sub.context>=0)].pivot(index=PAIR,columns='kind',values=metric).sort_index()
    require(set(w)==set(PAIR_KINDS) and not w.isna().any().any(),'Incomplete crossed pairs: '+metric)
    zero=sub[sub.kind=='zero'].set_index('input_key')[metric]
    require(zero.index.is_unique,'Duplicate shared zero.')
    w['zero']=[zero.loc[k] for k in w.index.get_level_values('input_key')]
    w['random_A']=(w.random_aligned_0+w.random_aligned_1)/2
    w['rotated_R']=(w.rotated_residual_0+w.rotated_residual_1)/2
    return w
def effects(w):
    return pd.DataFrame({'A_zero':w.aligned-w.zero,'A_specific':w.aligned-w.random_A,'R_zero':w.residual-w.zero,
        'full_zero':w.full-w.zero,'full_minus_A':w.full-w.aligned,'full_minus_rotated':w.full-w.rotated_R,
        'interaction':w.full-w.aligned-w.residual+w.zero,'half_minus_full':w.half-w.full,
        'AhalfR_minus_full':w.aligned_half_residual-w.full,'AhalfR_minus_half':w.aligned_half_residual-w.half})
def core_primary(cell):
    nn,nl,ln,ll=[cell[k] for k in [('native','native'),('native','lora'),('lora','native'),('lora','lora')]]
    return {'native_own_A_zero':nn.A_zero,'native_own_A_random':nn.A_specific,
        'own_A_specific_model_delta':ll.A_specific-nn.A_specific,
        'fixed_native_library_receiver_delta':ln.A_specific-nn.A_specific,
        'fixed_lora_library_receiver_delta':ll.A_specific-nl.A_specific,
        'own_full_minus_A_model_delta':ll.full_minus_A-nn.full_minus_A}

CELL_EFFECTS={(r,s):effects(paired_values(r,s,'flip','margin_shift')) for r,s in itertools.product(MODES,MODES)}
primary=[];family_rows=[]
for name,values in core_primary(CELL_EFFECTS).items():
    require(not values.isna().any(),'Unpaired primary model difference.')
    fam=values.groupby('family_id').mean();require(len(fam)==N_FAMILIES,'Primary family missing.')
    primary.append({'test':name,**estimate(fam,name,True)})
    family_rows.extend({'test':name,'family_id':f,'value':float(v)} for f,v in fam.items())
primary=pd.DataFrame(primary);require(list(primary.test)==PRIMARY_TESTS,'Primary registry changed.')
primary['p_holm']=holm(primary.p_two_sided.to_numpy())
primary['evidence']=np.where(primary.p_holm<.05,np.where(primary.estimate>0,'positive','negative'),'unresolved')
primary.to_csv(SHARED_ROOT/'analysis/primary_six_tests.csv',index=False)
pd.DataFrame(family_rows).to_csv(SHARED_ROOT/'analysis/primary_family_values.csv',index=False)
secondary=[];robust=[];boundaries=[]
for context,metric in itertools.product(('flip','stable'),('margin_shift','cf_accuracy_gain','cf_best_other_gain')):
    cells={(r,s):effects(paired_values(r,s,context,metric)) for r,s in itertools.product(MODES,MODES)}
    for (receiver,library),c in cells.items():
        for name,values in c.items():
            fam=values.groupby('family_id').mean()
            secondary.append({'receiver_context':context,'metric':metric,'comparison':'cell','receiver':receiver,'library':library,'contrast':name,**estimate(fam,['cell',context,metric,receiver,library,name])})
            if metric=='margin_shift' and name in ('A_zero','A_specific','full_minus_A'):
                loo=[fam.drop(f).mean() for f in fam.index]
                robust.append({'receiver_context':context,'receiver':receiver,'library':library,'contrast':name,'positive_families':int((fam>0).sum()),'negative_families':int((fam<0).sum()),'loo_min':min(loo),'loo_max':max(loo)})
                metadata=frame[(frame.receiver==receiver)&(frame.library==library)&(frame.kind=='full')].set_index(PAIR)
                for mapping in range(6):
                    ids=metadata.index[metadata.receiver_mapping.eq(mapping)].intersection(values.index)
                    boundaries.append({'receiver_context':context,'receiver':receiver,'library':library,'contrast':name,'mapping':mapping,
                        **estimate(values.loc[ids].groupby('family_id').mean(),['mapping',context,receiver,library,name,mapping])})
    for name,vals in core_primary(cells).items():
        if name.startswith('native_own'):continue
        secondary.append({'receiver_context':context,'metric':metric,'comparison':'model_delta','contrast':name,**estimate(vals.groupby('family_id').mean(),['secondary',context,metric,name])})
    interaction=(cells[('lora','lora')].A_specific-cells[('lora','native')].A_specific)-(cells[('native','lora')].A_specific-cells[('native','native')].A_specific)
    secondary.append({'receiver_context':context,'metric':metric,'comparison':'library_x_receiver','contrast':'A_specific_interaction',**estimate(interaction.groupby('family_id').mean(),['interaction',context,metric])})
secondary=pd.DataFrame(secondary);robust=pd.DataFrame(robust);boundaries=pd.DataFrame(boundaries)
for name,table in [('secondary_effects',secondary),('family_robustness',robust),('mapping_boundaries',boundaries)]:table.to_csv(SHARED_ROOT/'analysis'/(name+'.csv'),index=False)

# Shared natural references are paired once per receiver input, not expanded over donors/libraries.
natural_rows=[];clean_rows=[]
for context in ('flip','stable'):
    z=frame[(frame.receiver_context==context)&(frame.library=='shared')]
    for mode in MODES:
        w=z[z.receiver==mode].pivot(index=['family_id','input_key'],columns='kind',values='cf_success').astype(float)
        for new,ref in [('cf_half','zero'),('cf_delta','zero'),('cf_half','cf_delta')]:
            natural_rows.append({'receiver':mode,'receiver_context':context,'contrast':new+'_minus_'+ref,'paired_inputs':len(w),**estimate((w[new]-w[ref]).groupby('family_id').mean(),['natural',mode,context,new,ref])})
    for role in (context+'_before',context+'_after'):
        table=pd.concat([pd.read_csv(SHARED_ROOT/m/'analysis/behavior_records.csv').assign(receiver=m) for m in MODES])
        sub=table[table.role==role];w=sub.pivot(index=['family_id','capture_key'],columns='receiver',values='correct').astype(float)
        require(not w.isna().any().any(),'Natural behavior model pairing failed.')
        clean_rows.append({'role':role,'native_accuracy':w.native.mean(),'lora_accuracy':w.lora.mean(),**estimate((w.lora-w.native).groupby('family_id').mean(),['clean',role])})
natural_stats=pd.DataFrame(natural_rows);clean_stats=pd.DataFrame(clean_rows)
natural_stats.to_csv(SHARED_ROOT/'analysis/paired_natural_reference_accuracy.csv',index=False)
clean_stats.to_csv(SHARED_ROOT/'analysis/paired_clean_accuracy.csv',index=False)
display(primary);display(clean_stats)
print('Six confirmatory tests complete. All other intervals are descriptive family bootstrap intervals.')
```

```python
%%cell_progress 15/17 Compare downstream distances and joint event coverage
cell_stage('Compare downstream distances and joint event coverage')
geometry_rows=[];joint_rows=[];GEOM_VALUES={}
for context,layer,receiver,library in itertools.product(('flip','stable'),(49,55),MODES,MODES):
    sub=traces[(traces.receiver==receiver)&(traces.receiver_context==context)&(traces.layer==layer)]
    w=sub[sub.library==library].pivot(index=PAIR,columns='kind',values='relative_distance_to_cf').sort_index()
    zero=sub[sub.kind=='zero'].set_index('input_key').relative_distance_to_cf
    w['zero']=[zero.loc[k] for k in w.index.get_level_values('input_key')]
    scores=paired_values(receiver,library,context,'margin_shift')
    for new,ref in [('aligned','zero'),('aligned','full'),('aligned_half_residual','full'),('half','full')]:
        vals=w[ref]-w[new];valid=vals.notna();label=new+'_minus_'+ref
        GEOM_VALUES[(context,layer,receiver,library,label)]=vals
        geometry_rows.append({'receiver_context':context,'layer':layer,'receiver':receiver,'library':library,'comparison':'cell','contrast':label,
            'valid_pairs':int(valid.sum()),'eligible_pairs':len(vals),**estimate(vals.groupby('family_id').mean(),['distance',context,layer,receiver,library,label])})
        event=((vals>1e-9)&((scores[new]-scores[ref])>1e-12)).astype(float).where(valid)
        joint_rows.append({'receiver_context':context,'layer':layer,'receiver':receiver,'library':library,'contrast':label,'valid_pairs':int(valid.sum()),
            **estimate(event.groupby('family_id').mean(),['joint',context,layer,receiver,library,label])})
for context,layer,label in itertools.product(('flip','stable'),(49,55),('aligned_minus_zero','aligned_minus_full')):
    for source in ('own','native','lora'):
        n=GEOM_VALUES[(context,layer,'native','native' if source=='own' else source,label)]
        l=GEOM_VALUES[(context,layer,'lora','lora' if source=='own' else source,label)]
        diff=l-n  # subtract on shared pairs FIRST, then average; NA retained.
        geometry_rows.append({'receiver_context':context,'layer':layer,'library':source,'comparison':'lora_minus_native','contrast':label,
            'valid_pairs':int(diff.notna().sum()),'eligible_pairs':len(diff),**estimate(diff.groupby('family_id').mean(),['distance_delta',context,layer,source,label])})
geometry_stats=pd.DataFrame(geometry_rows);joint_stats=pd.DataFrame(joint_rows)
geometry_stats.to_csv(SHARED_ROOT/'analysis/paired_downstream_distances.csv',index=False)
joint_stats.to_csv(SHARED_ROOT/'analysis/joint_score_distance_coverage.csv',index=False)
display(geometry_stats[(geometry_stats.receiver_context=='flip')&(geometry_stats.comparison=='lora_minus_native')])
print('Positive distance reduction = closer to receiver-specific natural CF; not accuracy or mediated proportion.')
```

```python
%%cell_progress 16/17 Audit BF16 and prediction-stability common masks
cell_stage('Audit BF16 and prediction-stability common masks')
quality=[]
for mode in MODES:
    c=pd.read_csv(SHARED_ROOT/mode/'analysis/component_inventory.csv',float_precision='round_trip')
    m=pd.read_csv(SHARED_ROOT/mode/'analysis/control_matching_audit.csv',float_precision='round_trip')
    index=['receiver','library',*PAIR]
    require(not c.duplicated(index).any(),'Duplicate component rows.')
    maxerr=m.groupby(index).effective_norm_relative_error.max().rename('control_norm_error')
    c=c.set_index(index).join(maxerr,validate='one_to_one')
    require(c.control_norm_error.notna().all(),'Missing matched controls.')
    for flag in ('stable_unique_top','axis_defined','zero_full_update'):
        require(c[flag].isin([True,False]).all(),'Invalid boolean in quality inventory.')
    quality.append(c.reset_index())
quality=pd.concat(quality,ignore_index=True)
pair_quality=quality.groupby(PAIR).agg(control_norm_error=('control_norm_error','max'),closure_error=('effective_component_closure_relative','max'),
    both_sources_stable=('stable_unique_top','all'),all_axes_defined=('axis_defined','all'))
cross=pd.read_csv(SHARED_ROOT/'analysis/cross_receiver_dose_matching.csv',float_precision='round_trip')
planmeta=pd.DataFrame(PLAN)[['condition_id','kind','context','source_mapping_slot']]
cross=cross.merge(planmeta,on='condition_id',validate='many_to_one')
inputmeta=pd.DataFrame([{'input_key':capture_key(r,m),'family_id':r['family_id'],'receiver_context':r['role'].split('_')[0]} for r,m in TARGET_INPUTS])
cross=cross.merge(inputmeta,on='input_key',validate='many_to_one')
crossmax=cross[cross.kind.isin(['aligned','random_aligned_0','random_aligned_1','full'])].groupby(PAIR).effective_error.max().rename('cross_receiver_error')
pair_quality=pair_quality.join(crossmax,validate='one_to_one')
require(pair_quality.cross_receiver_error.notna().all(),'Missing crossed matching quality.')
masks={'all':pd.Series(True,index=pair_quality.index),'both_sources_stable':pair_quality.both_sources_stable.astype(bool)}
for limit in (.01,.02,.05):
    masks[f'norm_{limit:.2f}']=pair_quality.control_norm_error.le(limit)&pair_quality.cross_receiver_error.le(limit)
    masks[f'norm_and_closure_{limit:.2f}']=masks[f'norm_{limit:.2f}']&pair_quality.closure_error.le(limit)
sensitivity_rows=[];supports=[]
for context in ('flip','stable'):
    cells={(r,s):effects(paired_values(r,s,context,'margin_shift')) for r,s in itertools.product(MODES,MODES)}
    estimates=core_primary(cells);idx=cells[('native','native')].index
    for name,globalmask in masks.items():
        mask=globalmask.reindex(idx);require(mask.notna().all(),'Mask index mismatch.')
        ids=idx[mask];families_kept=set(ids.get_level_values('family_id'))
        supports.append({'receiver_context':context,'subset':name,'pairs':len(ids),'total_pairs':len(idx),
            'families':len(families_kept),'distinct_inputs':len(set(ids.get_level_values('input_key')))})
        for test,vals in estimates.items():
            for population,selected in [('retained',vals.loc[ids]),('unfiltered_same_families',vals[vals.index.get_level_values('family_id').isin(families_kept)])]:
                sensitivity_rows.append({'receiver_context':context,'subset':name,'test':test,'population':population,
                    **estimate(selected.groupby('family_id').mean(),['sensitivity',context,name,test,population])})
for name,mask in masks.items():pair_quality[name]=mask
pair_quality.reset_index().to_csv(SHARED_ROOT/'analysis/common_pair_quality_masks.csv',index=False)
sensitivity=pd.DataFrame(sensitivity_rows);supports=pd.DataFrame(supports)
sensitivity.to_csv(SHARED_ROOT/'analysis/common_mask_sensitivity.csv',index=False)
supports.to_csv(SHARED_ROOT/'analysis/common_mask_support.csv',index=False)
display(supports[supports.receiver_context=='flip'])
print('Same pairs retained in every model/library cell; no outcome-based or model-specific exclusion.')
```

```python
%%cell_progress 17/17 Display report and export the complete review bundle
cell_stage('Display report and export the complete review bundle')
def md_table(table):
    def fmt(x):
        if x is None or (isinstance(x,(float,np.floating)) and not np.isfinite(x)):return 'NA'
        return f'{x:.5f}' if isinstance(x,(float,np.floating)) else str(x).replace('|','/')
    return '\n'.join(['| '+' | '.join(table.columns)+' |','| '+' | '.join(['---']*len(table.columns))+' |']+
                     ['| '+' | '.join(fmt(x) for x in row)+' |' for row in table.itertuples(index=False,name=None)])

FIGURES=[]
def finish(fig,name):
    if SYNTHETIC_ONLY:fig.suptitle('SYNTHETIC VALIDATION ONLY - NOT MODEL RESULTS',fontsize=10,color='crimson')
    path=SHARED_ROOT/'analysis/figures'/(name+'.png');fig.tight_layout();fig.savefig(path,dpi=160,bbox_inches='tight');display(fig);plt.close(fig);FIGURES.append(path)
def interval_plot(ax,table,label,title,xlabel,scale=1):
    table=table.reset_index(drop=True);y=np.arange(len(table))
    x=table.estimate.to_numpy(float)*scale;lo=table.ci_low.to_numpy(float)*scale;hi=table.ci_high.to_numpy(float)*scale
    ax.errorbar(x,y,xerr=np.maximum(0,np.vstack([x-lo,hi-x])),fmt='o',capsize=3)
    ax.axvline(0,color='gray',lw=1);ax.set_yticks(y,table[label]);ax.invert_yaxis();ax.set_title(title);ax.set_xlabel(xlabel)
fig,ax=plt.subplots(figsize=(11,4));interval_plot(ax,primary,'test','Six frozen primary contrasts','Signed log-probability margin');finish(fig,'primary_tests')
fig,axes=plt.subplots(1,2,figsize=(13,4))
for ax,metric in zip(axes,('margin_shift','cf_accuracy_gain')):
    t=secondary[(secondary.receiver_context=='flip')&(secondary.comparison=='cell')&(secondary.contrast=='A_zero')&(secondary.metric==metric)].copy()
    t['label']=t.receiver+' <- '+t.library
    interval_plot(ax,t,'label','A minus zero: '+metric,'Margin' if metric=='margin_shift' else 'Accuracy change (percentage points)',1 if metric=='margin_shift' else 100)
finish(fig,'crossed_cells_behavior')
fig,axes=plt.subplots(1,2,figsize=(13,4))
for ax,layer in zip(axes,(49,55)):
    t=geometry_stats[(geometry_stats.receiver_context=='flip')&(geometry_stats.comparison=='cell')&(geometry_stats.layer==layer)&(geometry_stats.contrast=='aligned_minus_full')].copy()
    t['label']=t.receiver+' <- '+t.library;interval_plot(ax,t,'label',f'L{layer}: A versus full','Normalized distance reduction (>0 = closer)')
finish(fig,'downstream_geometry')
fig,axes=plt.subplots(1,2,figsize=(14,4))
for ax,test in zip(axes,('fixed_native_library_receiver_delta','fixed_lora_library_receiver_delta')):
    t=sensitivity[(sensitivity.receiver_context=='flip')&(sensitivity.test==test)&(sensitivity.population=='retained')&sensitivity.subset.isin(['all','norm_0.01','norm_0.02','norm_0.05'])]
    interval_plot(ax,t,'subset',test,'LoRA - Native A-specific margin')
finish(fig,'bf16_sensitivity')
marker='SYNTHETIC ONLY — NOT MODEL EVIDENCE\n\n' if SYNTHETIC_ONLY else ''
accuracy=secondary[(secondary.metric=='cf_accuracy_gain')&((secondary.comparison!='cell')|secondary.contrast.isin(['A_zero','full_zero','full_minus_A']))].copy()
for field in ('estimate','ci_low','ci_high'):accuracy[field]=accuracy[field]*100
report_text=marker+'# Native／LoRA 对齐物理更新交叉确认\n\n'
report_text+=f'完整运行：{N_FAMILIES} 家族、{len(capture_inputs)*2} clean 捕获、{len(frame)} 干预、{len(traces)} 状态测量。\n\n'
report_text+='距离 1–8 格，刻度 8；与旧 1–6 格相比扩展范围。所有场景在新推理前冻结并排除已列历史、Dataset v1 和 SFT 几何。不对整个训练分布或未知历史暴露作泛化声明。\n\n'
report_text+='## 六项主检验\n\n'+md_table(primary)+'\n\n'
report_text+='## 真实端点的配对行为\n\n'+md_table(clean_stats)+'\n\n'
report_text+='## 干预的直接配对准确率变化（百分点）\n\n'+md_table(accuracy[['receiver_context','comparison','receiver','library','contrast','estimate','ci_low','ci_high','n_families']])+'\n\n'
report_text+='stable 行衡量原物理答案保留，负数表示损伤；flip 行以 after 的新物理答案评分。完整剂量对比见 secondary_effects.csv。\n\n'
report_text+='## 下游状态距离\n\n'+md_table(geometry_stats[(geometry_stats.receiver_context=='flip')&(geometry_stats.comparison=='lora_minus_native')])+'\n\n'
report_text+='## 自身库同时改善分数与距离的覆盖比例\n\n'+md_table(joint_stats[(joint_stats.receiver_context=='flip')&joint_stats.receiver.eq(joint_stats.library)&joint_stats.contrast.isin(['aligned_minus_zero','aligned_minus_full'])])+'\n\n'
report_text+='## BF16 共同掩码支持数\n\n'+md_table(supports[supports.receiver_context=='flip'])+'\n\n'
report_text+='## 分家族稳健性\n\n'+md_table(robust[robust.receiver_context=='flip'])+'\n\n'
report_text+='## 解释与停止标准\n\n'
report_text+='固定库模型差控制请求方向和相对范数，支持的只是这些干预下的接收响应变化；不能直接证明自然通路减弱或所有微调收益的中介。对角线差混合本模型更新和接收响应；库间范数不等，因此库交互也不是纯轴角度效应。A/R 使用目标 after，g 不是纯物理轴。L48 的正交与加和是构造；L49/L55 距离是相对接收模型自身反事实的标量，不是跨模型隐藏向量的绝对距离。\n\n'
report_text+='干预准确率以 after 场景真实答案评分，衡量 before 图片上的反事实更新实现程度，不是原始 before 任务普通准确率。真实自然端点准确率另报。margin 改善不等于准确率改善，状态更接近不等于每个输入更好。非显著不证明等效、R 无用或网络线性。half-full 与 A+0.5R 是剂量／残余边界，不确认 A 的最优剂量。所有次级区间与筛选子集都是描述性，不能升级为新的独立复现。主分析保留所有家族；零自然轴保留并标注，几何 NA 不作零。\n\n'
report_text+='六项检验无论正、负还是未确定都结束本轮，不按新输出重选层、方向、剂量或家族。单个 adapter 不代表跨训练种子的一般规律。完整结果与共同阈值支持数均已导出。\n'
report=SHARED_ROOT/'ALIGNED_UPDATE_CROSSOVER_REPORT_ZH.md';report.write_text(report_text,encoding='utf-8');display(Markdown(report_text))
require(tree_hashes(ADAPTER_DIR)==ADAPTER_HASHES,'Adapter changed during the experiment.')
require(tree_hashes(DRIVE_TOOLKIT_ROOT/'src')==TOOLKIT_HASHES and tree_hashes(TOOLKIT_ROOT/'src')==TOOLKIT_HASHES,'Toolkit changed during run.')
for path,h in EXPOSURE_HASHES.items():require(sha(path)==h,'Exposure/training input changed: '+path)
for row in scene_rows:require(sha(SHARED_ROOT/'selected_images'/row['image_name'])==row['image_sha256'],'Frozen rendered image changed: '+row['image_name'])
for mode in MODES:
    clean_lock=read_json(SHARED_ROOT/mode/'clean_inventory.json')
    require(clean_lock['digest']==CLEAN_DIGESTS[mode] and digest(clean_lock['identity'])==CLEAN_DIGESTS[mode],'Clean inventory lock changed.')
    for key,h in clean_lock['identity'].items():
        jp=SHARED_ROOT/mode/'clean_records'/(key+'.json')
        require(sha(jp)==h,'Clean record changed after dose freezing.')
        require(sha(SHARED_ROOT/mode/'clean_cache'/(key+'.pt'))==read_json(jp)['tensor_sha256'],'Clean tensor changed after dose freezing.')
decision={'primary_evidence':dict(zip(primary.test,primary.evidence)),'synthetic_only':SYNTHETIC_ONLY,'domain_expanded':True,
    'new_geometry_cohort':True,'target_informed':True,'independent_training_seed_replication':False,
    'all_families_retained':True,'natural_mediation_proven':False,'accuracy_must_be_separate':True}
write_json(SHARED_ROOT/'decision.json',decision)
# Include only known result scopes; clean tensors are small final-token states, not model weights.
paths=[]
for p in sorted(SHARED_ROOT.rglob('*')):
    if not p.is_file() or p.name in ('review_bundle.zip','RUN_COMPLETE.json') or p.suffix=='.tmp':continue
    if 'injection_diagnostics' in p.parts:continue
    paths.append(p)
hashes={p.relative_to(SHARED_ROOT).as_posix():sha(p) for p in paths}
complete={'status':'complete','schema':'aligned-update-model-crossover-1.0','protocol_digest':PROTOCOL_DIGEST,'runtime_digest':RUNTIME_DIGEST,
    'run_ids':RUN_IDS,'synthetic_only':SYNTHETIC_ONLY,'families':N_FAMILIES,'clean':len(capture_inputs)*2,'interventions':len(frame),'traces':len(traces),'decision':decision,'outputs':hashes}
write_json(SHARED_ROOT/'RUN_COMPLETE.json',complete)
zip_path=SHARED_ROOT/'review_bundle.zip';tmp=SHARED_ROOT/'review_bundle.tmp'
with ZipFile(tmp,'w',ZIP_DEFLATED) as z:
    for p in [*paths,SHARED_ROOT/'RUN_COMPLETE.json']:z.write(p,p.relative_to(SHARED_ROOT).as_posix())
with ZipFile(tmp) as z:
    require(z.testzip() is None,'ZIP integrity failure.')
    for name,h in hashes.items():require('sha256:'+hashlib.sha256(z.read(name)).hexdigest()==h,'ZIP content mismatch: '+name)
tmp.replace(zip_path)
print('Finished. Report:',report,'\nReview bundle:',zip_path,'\nZIP SHA256:',sha(zip_path))
```
