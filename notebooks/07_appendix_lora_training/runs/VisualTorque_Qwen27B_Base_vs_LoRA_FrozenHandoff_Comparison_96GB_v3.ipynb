```python
# Run once. Every later code cell reports progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(
        total=total, initial=initial, desc=label, unit='cell',
        dynamic_ncols=True, leave=True,
    )
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 2/17 Install dependencies and lock Native numerical precision
cell_stage('Install dependencies and lock Native numerical precision')
import importlib
import importlib.metadata
import json
import os
import subprocess
import sys

cell_stage('Configuring caches before importing Torch')
os.environ.setdefault('PYTORCH_CUDA_ALLOC_CONF', 'expandable_segments:True')
os.environ.setdefault('HF_HOME', '/content/hf_cache')
os.environ.setdefault('TOKENIZERS_PARALLELISM', 'false')

cell_stage('Removing incompatible optional Torch packages')
subprocess.run(
    [sys.executable, '-m', 'pip', 'uninstall', '-y', 'torchaudio', 'torchao'],
    check=False,
)
importlib.invalidate_caches()

cell_stage('Installing pinned inference and analysis dependencies')
subprocess.check_call([
    sys.executable, '-m', 'pip', 'install', '-q', '-U',
    '--upgrade-strategy', 'only-if-needed',
    'transformers==5.14.0', 'peft==0.20.0', 'accelerate>=1.10',
    'sentencepiece', 'protobuf', 'huggingface_hub>=0.34',
    'safetensors', 'tqdm>=4.66', 'scikit-learn>=1.5',
    'scipy>=1.13', 'pandas>=2.2', 'matplotlib>=3.8',
    'joblib>=1.4', 'pillow>=10.4',
])

cell_stage('Importing the immutable toolkit copy')
SRC_ROOT = TOOLKIT_ROOT / 'src'
PACKAGE_ROOT = SRC_ROOT / 'visual_torque_toolkit'
source_text = str(SRC_ROOT)
sys.path[:] = [entry for entry in sys.path if entry != source_text]
sys.path.insert(0, source_text)
for module_name in tuple(sys.modules):
    if module_name == 'visual_torque_toolkit' or module_name.startswith('visual_torque_toolkit.'):
        del sys.modules[module_name]
importlib.invalidate_caches()

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import visual_torque_toolkit
from IPython.display import display
from peft import PeftModel
from scipy.stats import binomtest
from visual_torque_toolkit.behavior import (
    BehaviorRunConfig, PromptCondition, load_behavior_examples,
)
from visual_torque_toolkit.behavior.adapter import HuggingFaceMultimodalAdapter
from visual_torque_toolkit.generate import (
    GenerationConfig, generate_multi_stack_scenes,
)
from visual_torque_toolkit.prompts import QuestionKind, generate_questions
from visual_torque_toolkit.render import RenderStyle, render_scene

# Defensive Qwen3.5 compatibility patch for exact candidate scoring.
def append_text_prefix_aligned(self, inputs, prefix):
    if not prefix:
        return inputs
    batch_size = inputs['input_ids'].shape[0]
    prefix_tensor = torch.tensor(
        prefix,
        dtype=inputs['input_ids'].dtype,
        device=inputs['input_ids'].device,
    ).unsqueeze(0).expand(batch_size, -1)
    inputs['input_ids'] = torch.cat((inputs['input_ids'], prefix_tensor), dim=1)
    attention_extension = torch.ones(
        (batch_size, len(prefix)),
        dtype=inputs['attention_mask'].dtype,
        device=inputs['attention_mask'].device,
    )
    inputs['attention_mask'] = torch.cat(
        (inputs['attention_mask'], attention_extension), dim=1
    )
    for key in ('token_type_ids', 'mm_token_type_ids'):
        if key not in inputs:
            continue
        extension = torch.zeros(
            (batch_size, len(prefix)),
            dtype=inputs[key].dtype,
            device=inputs[key].device,
        )
        inputs[key] = torch.cat((inputs[key], extension), dim=1)
    inputs.pop('position_ids', None)
    return inputs

HuggingFaceMultimodalAdapter._append_text_prefix = append_text_prefix_aligned

NATIVE_NUMERICS_POLICY = {
    'version': 'native-fp32-matmul-1.0',
    'float32_matmul_precision': 'highest',
    'tf32_matmul': False,
}

def native_precision_state():
    return {'float32_matmul_precision': torch.get_float32_matmul_precision(),
            'tf32_matmul': bool(torch.backends.cuda.matmul.allow_tf32)}

def assert_native_precision():
    expected = {key: NATIVE_NUMERICS_POLICY[key] for key in native_precision_state()}
    actual = native_precision_state()
    if actual != expected:
        raise RuntimeError(
            f'Native numerical policy changed: expected {expected}, got {actual}. '
            'Restart and Run all in this v3 notebook; do not reuse v2 captures '
            'or enable TF32 during a frozen comparison.'
        )
    return actual

# Explicitly undo flags left by a previous notebook in a reused Colab process.
torch.set_float32_matmul_precision('highest')
torch.backends.cuda.matmul.allow_tf32 = False
print('Frozen Native precision:', assert_native_precision())

toolkit_path = Path(visual_torque_toolkit.__file__).resolve()
assert PACKAGE_ROOT.resolve() in toolkit_path.parents
try:
    torchao_version = importlib.metadata.version('torchao')
except importlib.metadata.PackageNotFoundError:
    torchao_version = None
assert torchao_version is None
print('torch:', torch.__version__, '| CUDA:', torch.version.cuda)
print('transformers:', importlib.metadata.version('transformers'))
print('peft:', importlib.metadata.version('peft'))
print('Toolkit:', toolkit_path)
```

```python
%%cell_progress 5/17 Freeze exactly matching Native prompts
cell_stage('Freeze exactly matching Native prompts')
import math
from dataclasses import replace as dataclass_replace
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = {
    'outcome': 'Which way will the balance scale tip?',
    'weight_more': 'Which side has more total weight?',
    'weight_less': 'Which side has less total weight?',
}
SEMANTIC_TEXT = {'left': 'left side', 'balanced': 'equal', 'right': 'right side'}

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

def task_semantic(features, task):
    if task == 'outcome':
        value = int(features['net_torque'])
    elif task == 'weight_more':
        value = int(features['weight_score'])
    elif task == 'weight_less':
        value = -int(features['weight_score'])
    else:
        raise KeyError(task)
    return outcome(value)

def mapping_by_label(semantics):
    assert set(semantics) == {'left', 'balanced', 'right'}
    return tuple({
        'position': index, 'label': LABELS[index], 'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def direction_reference_mapping(row):
    semantics = ['left', 'balanced', 'right']
    offset = int(stable_key(
        f"options:{row['family_id']}", seed=DIRECTION_SEED
    )[:8], 16) % 3
    return mapping_by_label(semantics[offset:] + semantics[:offset])

def target_mappings(row):
    exact, wrong = row['exact_semantic'], row['shortcut_semantic']
    assert {exact, wrong} == {'left', 'right'}
    balanced_label = LABELS[int(stable_key(
        ('balanced-label', row['pair_instance_id']), seed=EXPANSION_SEED
    )[:8], 16) % 3]
    side_labels = [label for label in LABELS if label != balanced_label]
    result = {}
    for variant, exact_label, wrong_label in (
        ('forward', side_labels[0], side_labels[1]),
        ('reverse', side_labels[1], side_labels[0]),
    ):
        by_label = {
            exact_label: exact, wrong_label: wrong, balanced_label: 'balanced'
        }
        result[variant] = mapping_by_label(tuple(by_label[label] for label in LABELS))
    return result

def semantics_by_label(mapping):
    return tuple(next(
        item['semantic'] for item in mapping if item['label'] == label
    ) for label in LABELS)

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def custom_prompt(task, mapping):
    options = '\n'.join(
        f"{item['label']}. {SEMANTIC_TEXT[item['semantic']]}" for item in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT, 'Question: ' + QUESTION_TEXT[task], 'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option letter '
        'and nothing else. Valid replies are: A, B, C.',
    ])

def make_direction_example(row, role_record, task, mapping, name):
    features = role_record['features']
    semantics = semantics_by_label(mapping)
    gold = task_semantic(features, task)
    return BehaviorExample(
        example_id=f"exactphysics_direction:{row['family_id']}:{name}",
        question_id=f"exactphysics_direction_{row['family_id']}_{name}",
        scene_id=str(role_record['scene_id']), condition=PromptCondition.IMAGE_RULE,
        question_kind=task, variant='exactphysics_direction', split='direction_support',
        problem_type='semantic_direction', outcome=gold,
        net_torque=int(features['net_torque']), prompt=custom_prompt(task, mapping),
        image_path=(DIRECTION_RUNTIME_DIR / role_record['image_filename']).resolve(),
        labels=LABELS, option_text=tuple(SEMANTIC_TEXT[x] for x in semantics),
        option_semantics=semantics, gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def build_direction_examples(row):
    mapping = direction_reference_mapping(row)
    examples = {}
    for task in DONOR_TASKS:
        for role in ('base', 'aligned_source'):
            key = f'{task}__{role}'
            examples[key] = make_direction_example(
                row, row['roles'][role], task, mapping, key
            )
    return examples, mapping

def build_target_examples(row):
    mappings = target_mappings(row)
    base = outcome_example_by_scene[row['scene_id']]
    examples = {}
    for variant, mapping in mappings.items():
        semantics = semantics_by_label(mapping)
        examples[variant] = dataclass_replace(
            base,
            example_id=f"independent_exactphysics_target:{row['scene_id']}:{variant}",
            question_id=f"independent_exactphysics_target_{row['scene_id']}_{variant}",
            variant=f'independent_exactphysics_{variant}', prompt=custom_prompt('outcome', mapping),
            labels=LABELS, option_text=tuple(SEMANTIC_TEXT[x] for x in semantics),
            option_semantics=semantics,
            gold_label=LABELS[semantics.index(row['exact_semantic'])],
            gold_semantic=row['exact_semantic'],
        )
    return examples, mappings

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples], add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

print('Prompt builders ready.')

outcome_example_by_scene = {}
for row in scene_manifest:
    outcome_example_by_scene[row['scene_id']] = BehaviorExample(
        example_id=row['scene_id'],question_id=row['scene_id'],scene_id=row['scene_id'],
        condition=PromptCondition.IMAGE_RULE,question_kind='outcome',variant='frozen_bridge',
        split='native_selected_replay',problem_type='single_stack',outcome=row['exact_semantic'],
        net_torque=int(row['net_torque']),prompt='',image_path=Path(row['runtime_image_path']),
        labels=LABELS,option_text=('left side','equal','right side'),
        option_semantics=('left','balanced','right'),gold_label='A' if row['exact_semantic']=='left' else 'C',
        gold_semantic=row['exact_semantic'])
prompt_inventory=[]
for row in scene_manifest:
    examples,_=build_target_examples(row)
    for variant,example in examples.items():
        prompt_inventory.append({'scene_id':row['scene_id'],'mapping_variant':variant,
            'prompt':example.prompt,'labels':example.labels,'semantics':example.option_semantics,
            'image_sha256':image_hashes[row['scene_id']]})
PROMPT_DIGEST=freeze_json(OUTPUT_ROOT/'prompt_inventory.json',prompt_inventory)
assert len(prompt_inventory)==len(scene_manifest)*2
print('Frozen target inputs:',len(prompt_inventory))
```

```python
%%cell_progress 7/17 Define the original intervention helpers
cell_stage('Define the original intervention helpers')
def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(scores))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def forward_capture(prepared, layers):
    hidden_by_layer = {}
    handles = []
    for layer in layers:
        def hook(_module, _inputs, output, layer=layer):
            hidden_by_layer[layer] = output_hidden_tensor(
                output
            ).detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer].register_forward_hook(hook))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    return score_records(prepared, output.logits), hidden_by_layer

def capture_examples(examples, layers):
    scores = {}
    hidden = {name: {} for name in examples}
    masks = {}
    for name, example in examples.items():
        prepared = prepare_examples([example])
        masks[name] = final_mask(prepared)
        current_scores, current_hidden = forward_capture(prepared, layers)
        scores[name] = current_scores[0]
        for layer, tensor in current_hidden.items():
            hidden[name][layer] = tensor[0]
        del prepared, current_hidden
    return scores, hidden, masks

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def matched_random(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    return (noise * (selected_norm(tensor) / selected_norm(noise))).to(tensor.dtype)

def forward_intervention_chunk(prepared, conditions):
    assert len(prepared['examples']) == len(conditions)
    layers = {int(row['layer_index']) for row in conditions}
    assert len(layers) == 1
    layer = next(iter(layers))
    payloads = torch.stack([row['payload_selected'] for row in conditions])
    masks = torch.stack([row['target_mask'] for row in conditions])
    modes = [row['intervention_mode'] for row in conditions]
    counts = []
    def hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payloads.to(device=hidden.device, dtype=hidden.dtype)
        mask = masks.to(device=hidden.device)
        modified = hidden.clone()
        counts.clear()
        for index, mode in enumerate(modes):
            current = mask[index]
            assert int(current.sum()) == int(payload[index].shape[0]) == 1
            if mode == 'add':
                modified[index, current] = hidden[index, current] + payload[index]
            elif mode == 'replace':
                modified[index, current] = payload[index]
            else:
                raise ValueError(mode)
            counts.append(int(current.sum()))
        return replace_output_hidden(output, modified)
    handle = DECODER_LAYERS[layer].register_forward_hook(hook)
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        handle.remove()
    records = score_records(prepared, output.logits)
    for record, condition, count in zip(records, conditions, counts):
        record.update({
            key: value for key, value in condition.items()
            if key not in ('payload_selected', 'target_mask')
        })
        record['patched_tokens'] = count
    return records

def score_conditions(example, conditions):
    records = []
    grouped = defaultdict(list)
    for condition in conditions:
        grouped[int(condition['layer_index'])].append(condition)
    for layer_conditions in grouped.values():
        for start in range(0, len(layer_conditions), PATCH_CONDITION_BATCH_SIZE):
            actual = layer_conditions[start:start + PATCH_CONDITION_BATCH_SIZE]
            padded = list(actual)
            while len(padded) < PATCH_CONDITION_BATCH_SIZE:
                padded.append(padded[-1])
            prepared = prepare_examples([example] * len(padded))
            records.extend(forward_intervention_chunk(
                prepared, padded
            )[:len(actual)])
            del prepared
    return records

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def label_score(record, label):
    labels = list(record['candidate_labels'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    index = labels.index(label)
    other = np.delete(values, index)
    return float(values[index] - np.logaddexp.reduce(other))

def entropy(record):
    probabilities = np.asarray(record['candidate_probabilities'], dtype=float)
    return float(-np.sum(probabilities * np.log(np.clip(probabilities, 1e-12, 1))))

def cosine_similarity(left, right):
    left = left.float().reshape(-1)
    right = right.float().reshape(-1)
    denominator = torch.linalg.vector_norm(left) * torch.linalg.vector_norm(right)
    assert float(denominator) > 0
    return float(torch.dot(left, right) / denominator)

def rescale_direction(tensor, target_norm):
    norm = selected_norm(tensor)
    assert norm > 0 and target_norm > 0
    return (tensor.float() * (float(target_norm) / norm)).to(tensor.dtype)

def unit_direction(tensor):
    return rescale_direction(tensor, 1.0).float()

def score_l49_and_capture_l55(example, payload_selected, clean_l55, metadata):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    captured = {}
    def semantic_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payload_selected.to(device=hidden.device, dtype=hidden.dtype)
        current = mask.to(hidden.device)
        modified = hidden.clone()
        modified[0, current] = hidden[0, current] + payload
        return replace_output_hidden(output, modified)
    def label_capture_hook(_module, _inputs, output):
        captured['hidden'] = output_hidden_tensor(output).detach().cpu().clone()
    semantic_handle = DECODER_LAYERS[SEMANTIC_LAYER].register_forward_hook(
        semantic_hook
    )
    label_handle = DECODER_LAYERS[LABEL_LAYER].register_forward_hook(
        label_capture_hook
    )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        semantic_handle.remove()
        label_handle.remove()
    record = score_records(prepared, output.logits)[0]
    downstream = captured['hidden'][0][mask] - clean_l55
    record.update(metadata)
    record['patched_tokens'] = int(mask.sum())
    record['downstream_l55_norm'] = selected_norm(downstream)
    return record, downstream

def semantic_margin(record, exact_semantic, wrong_semantic):
    semantics = list(record['candidate_semantics'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(
        values[semantics.index(exact_semantic)]
        - values[semantics.index(wrong_semantic)]
    )

def vector_projection(delta, direction):
    unit = unit_direction(direction).reshape(-1)
    return float(torch.dot(delta.float().reshape(-1), unit))

def forward_add_and_capture(example, layer, payload_selected, capture_layers):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    captured = {}
    handles = []
    def intervention_hook(_module, _inputs, output):
        hidden = output_hidden_tensor(output)
        payload = payload_selected.to(device=hidden.device, dtype=hidden.dtype)
        current = mask.to(hidden.device)
        assert int(current.sum()) == int(payload.shape[0]) == 1
        modified = hidden.clone()
        modified[0, current] = hidden[0, current] + payload
        return replace_output_hidden(output, modified)
    handles.append(DECODER_LAYERS[layer].register_forward_hook(intervention_hook))
    for capture_layer in capture_layers:
        assert capture_layer > layer, (layer, capture_layer)
        def capture_hook(_module, _inputs, output, capture_layer=capture_layer):
            captured[capture_layer] = output_hidden_tensor(
                output
            ).detach().cpu().clone()[0]
        handles.append(
            DECODER_LAYERS[capture_layer].register_forward_hook(capture_hook)
        )
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_records(prepared, output.logits)[0]
    return record, {
        capture_layer: tensor[mask] for capture_layer, tensor in captured.items()
    }, int(mask.sum())

def clamp_selected_coordinate(selected, untreated, axis):
    untreated = untreated.to(device=selected.device, dtype=selected.dtype)
    axis_float = axis.to(device=selected.device, dtype=torch.float32)
    axis_float = axis_float / torch.linalg.vector_norm(axis_float)
    displacement = selected.float() - untreated.float()
    coefficient = torch.sum(displacement * axis_float, dim=-1, keepdim=True)
    clamped = selected.float() - coefficient * axis_float
    return clamped.to(selected.dtype), float(coefficient.detach().cpu().item())

def forward_path_condition(example, operations, untreated_selected, metadata):
    prepared = prepare_examples([example])
    mask = final_mask(prepared)
    captured, removed = {}, {}
    handles = []

    def make_hook(layer):
        operation = operations.get(layer)
        def hook(_module, _inputs, output):
            hidden = output_hidden_tensor(output)
            current = mask.to(hidden.device)
            selected = hidden[0, current]
            changed = False
            if operation is not None:
                action = operation['action']
                if action == 'add':
                    payload = operation['payload'].to(
                        device=hidden.device, dtype=hidden.dtype
                    )
                    selected = selected + payload
                    changed = True
                elif action == 'clamp_axis':
                    selected, coefficient = clamp_selected_coordinate(
                        selected, operation['untreated'], operation['axis']
                    )
                    removed[layer] = coefficient
                    changed = True
                else:
                    raise ValueError(action)
            captured[layer] = selected.detach().cpu().clone()
            if not changed:
                return None
            modified = hidden.clone()
            modified[0, current] = selected
            return replace_output_hidden(output, modified)
        return hook

    active_layers = sorted(set(operations) | {SEMANTIC_LAYER, LABEL_LAYER})
    for layer in active_layers:
        handles.append(DECODER_LAYERS[layer].register_forward_hook(make_hook(layer)))
    try:
        with torch.inference_mode():
            output = adapter.model(**prepared['inputs'], use_cache=False)
    finally:
        for handle in handles:
            handle.remove()
    record = score_records(prepared, output.logits)[0]
    record.update(metadata)
    record['patched_tokens'] = int(mask.sum())
    record['removed_projection_l49'] = float(removed.get(SEMANTIC_LAYER, 0.0))
    record['removed_projection_l55'] = float(removed.get(LABEL_LAYER, 0.0))
    record['l48_payload_norm'] = float(
        selected_norm(operations[UPSTREAM_LAYER]['payload'])
        if UPSTREAM_LAYER in operations else 0.0
    )
    record['l49_axis_norm'] = float(
        selected_norm(operations[SEMANTIC_LAYER]['axis'])
        if (SEMANTIC_LAYER in operations
            and operations[SEMANTIC_LAYER]['action'] == 'clamp_axis') else 0.0
    )
    record['l55_axis_norm'] = float(
        selected_norm(operations[LABEL_LAYER]['axis'])
        if (LABEL_LAYER in operations
            and operations[LABEL_LAYER]['action'] == 'clamp_axis') else 0.0
    )
    return record, captured

def build_path_operations(row, variant, mapping, untreated_selected):
    exact, wrong = row['exact_semantic'], row['shortcut_semantic']
    toward_exact_sign = 1.0 if exact == 'left' else -1.0
    exact_label = label_for(mapping, exact)
    wrong_label = label_for(mapping, wrong)
    upstream_payload = (
        toward_exact_sign * UPSTREAM_DIRECTION * MATCHED_UPSTREAM_ALPHA
    )
    semantic_payload = toward_exact_sign * SEMANTIC_DIRECTION
    code_payload = CODE_DIRECTIONS[(wrong_label, exact_label)]
    random_semantic = matched_random(
        semantic_payload,
        ('sequential-block', row['scene_id'], variant, 'l49'),
    )
    random_code = matched_random(
        code_payload,
        ('sequential-block', row['scene_id'], variant, 'l55'),
    )

    add_upstream = {'action': 'add', 'payload': upstream_payload}
    add_semantic = {'action': 'add', 'payload': semantic_payload}
    add_code = {'action': 'add', 'payload': code_payload}
    block_l49 = {
        'action': 'clamp_axis',
        'untreated': untreated_selected[SEMANTIC_LAYER],
        'axis': semantic_payload,
    }
    random_block_l49 = {
        'action': 'clamp_axis',
        'untreated': untreated_selected[SEMANTIC_LAYER],
        'axis': random_semantic,
    }
    block_l55 = {
        'action': 'clamp_axis',
        'untreated': untreated_selected[LABEL_LAYER],
        'axis': code_payload,
    }
    random_block_l55 = {
        'action': 'clamp_axis',
        'untreated': untreated_selected[LABEL_LAYER],
        'axis': random_code,
    }
    operations = {
        'self_block_l49': {SEMANTIC_LAYER: block_l49},
        'self_random_block_l49': {SEMANTIC_LAYER: random_block_l49},
        'self_block_l55': {LABEL_LAYER: block_l55},
        'self_random_block_l55': {LABEL_LAYER: random_block_l55},
        'upstream_repair': {UPSTREAM_LAYER: add_upstream},
        'upstream_repair_block_l49': {
            UPSTREAM_LAYER: add_upstream, SEMANTIC_LAYER: block_l49,
        },
        'upstream_repair_random_block_l49': {
            UPSTREAM_LAYER: add_upstream, SEMANTIC_LAYER: random_block_l49,
        },
        'upstream_repair_block_l55': {
            UPSTREAM_LAYER: add_upstream, LABEL_LAYER: block_l55,
        },
        'upstream_repair_random_block_l55': {
            UPSTREAM_LAYER: add_upstream, LABEL_LAYER: random_block_l55,
        },
        'l49_repair': {SEMANTIC_LAYER: add_semantic},
        'l49_repair_block_l55': {
            SEMANTIC_LAYER: add_semantic, LABEL_LAYER: block_l55,
        },
        'l49_repair_random_block_l55': {
            SEMANTIC_LAYER: add_semantic, LABEL_LAYER: random_block_l55,
        },
        'l55_repair': {LABEL_LAYER: add_code},
        'l55_bypass_after_l49_block': {
            UPSTREAM_LAYER: add_upstream,
            SEMANTIC_LAYER: block_l49,
            LABEL_LAYER: add_code,
        },
    }
    axes = {
        'semantic': semantic_payload, 'semantic_random': random_semantic,
        'code': code_payload, 'code_random': random_code,
    }
    return operations, axes
```

```python
%%cell_progress 8/17 Attach LoRA and verify disabled-base parity
cell_stage('Attach LoRA and verify disabled-base parity')
def compare_disabled_capture(bare_score, disabled_score, bare_selected, disabled_selected):
    score_error=float(np.max(np.abs(centered_scores(disabled_score)-centered_scores(bare_score))))
    assert set(bare_selected)==set(disabled_selected)==set(CAPTURE_LAYERS)
    layers={}
    for layer in CAPTURE_LAYERS:
        x,y=np.asarray(bare_selected[layer]),np.asarray(disabled_selected[layer])
        assert x.shape==y.shape and x.ndim==2 and x.shape[0]==1
        assert np.isfinite(x).all() and np.isfinite(y).all(), 'Nonfinite disabled-base hidden state.'
        layers[str(layer)]={'exactly_equal':bool(np.array_equal(x,y)),
            'max_absolute_error':float(np.max(np.abs(x-y))),
            'relative_l2_error':float(np.linalg.norm(x-y)/max(float(np.linalg.norm(x)),1e-12))}
    return {'centered_score_error':score_error,'layers':layers,
        'passed':bool(np.isfinite(score_error) and score_error<SMOKE_TOLERANCE and
                      all(value['exactly_equal'] for value in layers.values()))}

def capture_parity_example(example):
    prepared=prepare_examples([example])
    mask=final_mask(prepared)
    scores,hidden=forward_capture(prepared,CAPTURE_LAYERS)
    selected={layer:value[0,mask].float().numpy().copy() for layer,value in hidden.items()}
    return scores[0],selected

# Compare both scores and final-token coordinates in the SAME runtime. Equal
# answer scores alone cannot establish that a disabled adapter preserves an axis.
assert_native_precision()
smoke_rows=[next(r for r in scene_manifest if (r['cohort'],r['side'])==c)
            for c in itertools.product(COHORTS,SIDES)]
bare_captures={}
for row in smoke_rows:
    for v,e in build_target_examples(row)[0].items():
        bare_captures[(row['scene_id'],v)]=capture_parity_example(e)
adapter.model=PeftModel.from_pretrained(adapter.model,str(ADAPTER_DIR),is_trainable=False)
adapter.model.eval()
for p in adapter.model.parameters(): p.requires_grad_(False)
assert not any(p.requires_grad for p in adapter.model.parameters())
assert not getattr(adapter.model,'is_loaded_in_4bit',False)
assert not getattr(adapter.model,'is_loaded_in_8bit',False)

@contextmanager
def model_mode(mode):
    assert_native_precision()
    if mode=='native':
        with adapter.model.disable_adapter(): yield
    elif mode=='lora': yield
    else: raise ValueError(mode)

disabled_parity_rows=[]
with model_mode('native'):
    for row in smoke_rows:
        for v,e in build_target_examples(row)[0].items():
            score,selected=capture_parity_example(e)
            bare_score,bare_selected=bare_captures[(row['scene_id'],v)]
            disabled_parity_rows.append({'scene_id':row['scene_id'],'mapping_variant':v,
                **compare_disabled_capture(bare_score,score,bare_selected,selected)})
disabled_parity_report={'runtime':RUNTIME_DIGEST,'precision':assert_native_precision(),
    'passed':all(row['passed'] for row in disabled_parity_rows),'rows':disabled_parity_rows}
write_json(OUTPUT_ROOT/'disabled_base_parity.json',disabled_parity_report)
if not disabled_parity_report['passed']:
    raise RuntimeError('Bare/disabled Native hidden states or scores differ. '
        'See disabled_base_parity.json. No direction reconstruction or target intervention has run.')
disable_errors=[row['centered_score_error'] for row in disabled_parity_rows]
del bare_captures,bare_selected,selected
# Verify eval mode and that PEFT really changes a nonzero trained parameter.
lora_parameters=[p for n,p in adapter.model.named_parameters() if 'lora_B' in n]
assert lora_parameters and any(torch.count_nonzero(p).item()>0 for p in lora_parameters)
assert not adapter.model.training
print('Bare/disabled Native parity max error:',max(disable_errors))
print('L48/L49/L55 final-token coordinates: exactly equal for all 8 bare/disabled pairs.')
```

```python
%%cell_progress 9/17 Rebuild and freeze the Native directions
cell_stage('Rebuild and freeze the Native directions')
def compare_direction_calibration(observations, reference):
    # Norms cannot prove vector identity. Check a full calibration profile plus
    # the exact label/semantic alignment, then require target Native replay later.
    expected={r['family_id']:r for r in reference['rows']}
    actual={r['family_id']:r for r in observations}
    assert len(expected)==len(reference['rows']) and len(actual)==len(observations), 'Duplicate calibration family.'
    assert set(actual)==set(expected), 'Missing or unexpected direction calibration families.'
    norm_rows,score_rows=[],[]
    for family_id in sorted(expected):
        old,new=expected[family_id],actual[family_id]
        assert set(new['baseline_scores'])==set(old['baseline_scores']), ('Calibration prompt keys',family_id)
        for key,prior in old['baseline_scores'].items():
            current=new['baseline_scores'][key]
            assert current['candidate_labels']==prior['candidate_labels'], ('Direction answer labels changed',family_id,key)
            assert current['candidate_semantics']==prior['candidate_semantics'], ('Direction answer mapping changed',family_id,key)
            x=np.asarray(current['candidate_logprobs'],dtype=float)
            y=np.asarray(prior['candidate_logprobs'],dtype=float)
            assert x.shape==y.shape==(3,) and np.isfinite(x).all() and np.isfinite(y).all(), 'Invalid direction score.'
            error=float(np.max(np.abs((x-x.mean())-(y-y.mean()))))
            score_rows.append({'family_id':family_id,'prompt':key,'centered_logprob_error':error,
                'prediction_matches':current['predicted_semantic']==prior['predicted_semantic']})
        for layer in ('49','55'):
            for task in DONOR_TASKS:
                a=float(new['donor_norms'][layer][task]);b=float(old['donor_norms'][layer][task])
                assert math.isfinite(a) and math.isfinite(b) and a>0 and b>0, 'Non-finite or zero direction norm.'
                norm_rows.append({'family_id':family_id,'layer':int(layer),'task':task,
                    'current_norm':a,'historical_norm':b,'absolute_error':abs(a-b),'relative_error':abs(a-b)/b})
    norm_frame,score_frame=pd.DataFrame(norm_rows),pd.DataFrame(score_rows)
    layers={}
    for layer,frame in norm_frame.groupby('layer'):
        current=float(frame.current_norm.median());historical=float(frame.historical_norm.median())
        layers[str(layer)]={'current_median':current,'historical_median':historical,
            'absolute_median_error':abs(current-historical),'relative_median_error':abs(current-historical)/historical,
            'max_family_relative_error':float(frame.relative_error.max())}
    score_p95=float(score_frame.centered_logprob_error.quantile(.95))
    score_max=float(score_frame.centered_logprob_error.max())
    agreement=float(score_frame.prediction_matches.mean())
    gates={
        'median_norms_compatible':all(r['relative_median_error']<=DIRECTION_MEDIAN_RTOL for r in layers.values()),
        'family_norm_profile_compatible':all(r['max_family_relative_error']<=DIRECTION_PROFILE_MAX_RTOL for r in layers.values()),
        'calibration_scores_compatible':score_p95<=DIRECTION_SCORE_P95_ATOL and score_max<=DIRECTION_SCORE_MAX_ATOL,
        'calibration_predictions_compatible':agreement>=DIRECTION_PREDICTION_MIN_AGREEMENT,
    }
    return {'schema_version':'direction-rebuild-calibration-2.0','families':len(actual),'layers':layers,
        'score_p95_error':score_p95,'score_max_error':score_max,'prediction_agreement':agreement,
        'gates':gates,'passed':all(gates.values()),'bitwise_axis_identity_proven':False,
        'boundary':'Numerically compatible reconstruction, not equality of historical vector coordinates. Native target replay remains mandatory.'},norm_frame,score_frame

# Capture each original support prompt once; persist final-token states for restarts.
uncached_capture_examples = capture_examples
direction_capture_observations=[]
def capture_direction_examples(examples,layers):
    assert_native_precision()  # Applies to cache hits as well as fresh forwards.
    key=digest_object({k:{'id':e.example_id,'prompt':e.prompt,'image':sha256_file(e.image_path)} for k,e in examples.items()})
    path=DIRECTION_CACHE_DIR/(key.split(':')[1]+'.pt')
    identity={'protocol':PROTOCOL_DIGEST,'runtime':RUNTIME_DIGEST,'key':key,'layers':list(layers),'mode':'native'}
    if path.is_file():
        payload=torch.load(path,map_location='cpu',weights_only=True)
        assert payload['identity']==identity
    else:
        with model_mode('native'):
            scores,hidden,masks=uncached_capture_examples(examples,layers)
        selected={k:{l:t[masks[k]].clone() for l,t in h.items()} for k,h in hidden.items()}
        payload={'identity':identity,'scores':scores,'selected':selected}
        tmp=path.with_suffix('.tmp'); torch.save(payload,tmp); tmp.replace(path)
    family_ids={e.example_id.split(':')[1] for e in examples.values()}
    assert len(family_ids)==1
    norms={str(layer):{} for layer in layers}
    for layer in layers:
        for task in DONOR_TASKS:
            base=payload['selected'][task+'__base'][layer]
            source=payload['selected'][task+'__aligned_source'][layer]
            assert base.ndim==2 and base.shape[0]==1 and base.shape==source.shape, 'Invalid cached direction shape.'
            norms[str(layer)][task]=selected_norm(source-base)
    direction_capture_observations.append({'family_id':next(iter(family_ids)),
        'baseline_scores':payload['scores'],'donor_norms':norms})
    return payload['scores'],payload['selected'],{k:torch.tensor([True]) for k in examples}
cell_stage('Reconstructing the parent route without new scene outcomes')
from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import StandardScaler

left_units, left_raw_norms = [], []
code_units, code_raw_norms = defaultdict(list), []
for row in tqdm(direction_support, desc='L49/L55 directions', unit='family'):
    examples, mapping = build_direction_examples(row)
    scores, hidden, masks = capture_direction_examples(
        examples, (SEMANTIC_LAYER, LABEL_LAYER)
    )
    for task in DONOR_TASKS:
        base_key, source_key = f'{task}__base', f'{task}__aligned_source'
        base_semantic = task_semantic(row['roles']['base']['features'], task)
        source_semantic = task_semantic(
            row['roles']['aligned_source']['features'], task
        )
        semantic_delta = (
            hidden[source_key][SEMANTIC_LAYER][masks[source_key]]
            - hidden[base_key][SEMANTIC_LAYER][masks[base_key]]
        )
        left_delta = semantic_delta if source_semantic == 'left' else -semantic_delta
        left_units.append(unit_direction(left_delta))
        left_raw_norms.append(selected_norm(semantic_delta))
        code_delta = (
            hidden[source_key][LABEL_LAYER][masks[source_key]]
            - hidden[base_key][LABEL_LAYER][masks[base_key]]
        )
        transition = (
            label_for(mapping, base_semantic), label_for(mapping, source_semantic)
        )
        code_units[transition].append(unit_direction(code_delta))
        code_raw_norms.append(selected_norm(code_delta))
    del hidden
    torch.cuda.empty_cache()

SEMANTIC_STEERING_NORM = float(np.median(left_raw_norms))
CODE_STEERING_NORM = float(np.median(code_raw_norms))
SEMANTIC_DIRECTION = rescale_direction(
    torch.stack(left_units).mean(dim=0), SEMANTIC_STEERING_NORM
)
CODE_DIRECTIONS = {
    transition: rescale_direction(
        torch.stack(values).mean(dim=0), CODE_STEERING_NORM
    ) for transition, values in code_units.items()
}

audit_manifest = pd.DataFrame(read_jsonl(AUDIT_OUTCOME_MANIFEST_PATH))
hidden_rows, captured_ids = [], []
for path in tqdm(
    sorted(AUDIT_OUTCOME_SHARD_DIR.glob('outcome_*.pt')),
    desc='Loading prior L48 states', unit='shard',
):
    payload = torch.load(path, map_location='cpu', weights_only=False)
    hidden_rows.append(payload['hidden'][str(UPSTREAM_LAYER)].float())
    captured_ids.extend(str(value) for value in payload['scene_ids'])
assert audit_manifest['scene_id'].astype(str).tolist() == captured_ids
X = torch.cat(hidden_rows, dim=0).numpy()
y = audit_manifest['net_torque'].to_numpy(dtype=float)
fit = audit_manifest['split'].isin(['probe_train', 'probe_dev']).to_numpy()
scaler = StandardScaler().fit(X[fit])
X_fit, y_fit = scaler.transform(X[fit]), y[fit]
pls = PLSRegression(n_components=4, scale=False, max_iter=1000)
pls.fit(X_fit, y_fit.reshape(-1, 1))
scores = pls.transform(X_fit)
latent_delta = scores[y_fit > 0].mean(axis=0) - scores[y_fit < 0].mean(axis=0)
score_centered = scores - scores.mean(axis=0, keepdims=True)
X_centered = X_fit - X_fit.mean(axis=0, keepdims=True)
decoder = np.linalg.lstsq(score_centered, X_centered, rcond=1e-5)[0]
raw_delta = (latent_delta @ decoder) * scaler.scale_
UPSTREAM_DIRECTION = torch.from_numpy(
    (raw_delta / 2).astype('float32')
).reshape(1, -1)

parent_upstream = next(
    row for row in parent_direction['upstream']
    if int(row['layer']) == UPSTREAM_LAYER and row['method'] == UPSTREAM_METHOD
)

direction_calibration,norm_diagnostics,score_diagnostics=compare_direction_calibration(
    direction_capture_observations,HISTORICAL_CALIBRATION)
direction_calibration.update(runtime=runtime,upstream_current_norm=selected_norm(UPSTREAM_DIRECTION),
    upstream_historical_norm=float(parent_upstream['norm']))
direction_calibration['gates']['cached_L48_reconstruction_matches']=abs(selected_norm(UPSTREAM_DIRECTION)-float(parent_upstream['norm']))<1e-4
for layer,field in (('49','semantic_norm'),('55','code_norm')):
    historical=direction_calibration['layers'][layer]['historical_median']
    direction_calibration['gates'][f'historical_L{layer}_metadata_consistent']=bool(
        abs(historical-float(parent_direction[field]))<1e-4 and
        abs(historical-float(native_direction_metadata[field]))<1e-4)
direction_calibration['passed']=all(direction_calibration['gates'].values())
write_json(OUTPUT_ROOT/'direction_rebuild_diagnostics.json',direction_calibration)
write_json(OUTPUT_ROOT/'direction_rebuild_observations.json',direction_capture_observations)
norm_diagnostics.to_csv(ANALYSIS_DIR/'direction_norm_calibration.csv',index=False)
score_diagnostics.to_csv(ANALYSIS_DIR/'direction_score_calibration.csv',index=False)
display(pd.DataFrame(direction_calibration['layers']).T)
print('Calibration gates:',direction_calibration['gates'])
print('Diagnostics saved:',OUTPUT_ROOT/'direction_rebuild_diagnostics.json')
if not direction_calibration['passed']:
    raise RuntimeError('Native direction reconstruction is incompatible with the historical calibration. '
        'The diagnostic JSON/CSV above were saved before stopping. Do not delete assertions or adjust thresholds. '
        'Keep this runtime and provide direction_rebuild_diagnostics.json for diagnosis. '
        'No target intervention has run.')
# Keep the prescribed intervention amplitude, not a new amplitude inferred from
# a changed runtime. Only compatible reconstructions reach this step.
REBUILT_SEMANTIC_RAW_NORM,REBUILT_CODE_RAW_NORM=SEMANTIC_STEERING_NORM,CODE_STEERING_NORM
SEMANTIC_STEERING_NORM=float(native_direction_metadata['semantic_norm'])
CODE_STEERING_NORM=float(native_direction_metadata['code_norm'])
if REBUILT_SEMANTIC_RAW_NORM!=SEMANTIC_STEERING_NORM:
    SEMANTIC_DIRECTION=rescale_direction(SEMANTIC_DIRECTION,SEMANTIC_STEERING_NORM)
if REBUILT_CODE_RAW_NORM!=CODE_STEERING_NORM:
    CODE_DIRECTIONS={key:rescale_direction(value,CODE_STEERING_NORM) for key,value in CODE_DIRECTIONS.items()}

direction_metadata = {
    'schema_version': 'fixed-mediator-matched-directions-1.0',
    'upstream_layer': UPSTREAM_LAYER, 'upstream_method': UPSTREAM_METHOD,
    'upstream_norm': selected_norm(UPSTREAM_DIRECTION),
    'semantic_norm': SEMANTIC_STEERING_NORM,
    'code_norm': CODE_STEERING_NORM,
    'matched_upstream_alpha': MATCHED_UPSTREAM_ALPHA,
    'parent_direction_sha256': sha256_file(PARENT_DIRECTION_PATH),
}
direction_metadata.update(rebuild_calibration_sha256=digest_object(direction_calibration), raw_rebuilt_semantic_norm=REBUILT_SEMANTIC_RAW_NORM, raw_rebuilt_code_norm=REBUILT_CODE_RAW_NORM, coordinate_provenance='numerically calibrated reconstruction from original Native support')
write_json_atomic(DIRECTION_LIBRARY_PATH, direction_metadata)
print(direction_metadata)

native_direction_metadata=read_json(SOURCE_PATHS['native_directions'])
for key in ('upstream_norm','semantic_norm','code_norm','matched_upstream_alpha'):
    assert abs(float(direction_metadata[key])-float(native_direction_metadata[key]))<1e-4, ('Native direction parity',key)
assert set(CODE_DIRECTIONS)==set(itertools.permutations(LABELS,2))
frozen_axes={'upstream':UPSTREAM_DIRECTION,'semantic':SEMANTIC_DIRECTION,
             'code':{a+'>'+b:t for (a,b),t in CODE_DIRECTIONS.items()}}
axis_tensor_path=OUTPUT_ROOT/'frozen_native_axes.pt'
if axis_tensor_path.is_file():
    saved=torch.load(axis_tensor_path,map_location='cpu',weights_only=True)
    assert torch.equal(saved['upstream'],UPSTREAM_DIRECTION) and torch.equal(saved['semantic'],SEMANTIC_DIRECTION)
    assert all(torch.equal(saved['code'][k],v) for k,v in frozen_axes['code'].items())
else:
    tmp=axis_tensor_path.with_suffix('.tmp');torch.save(frozen_axes,tmp);tmp.replace(axis_tensor_path)
AXIS_DIGEST=sha256_file(axis_tensor_path)
CAPTURE_DIGEST=digest_object({'protocol':PROTOCOL_DIGEST,'runtime':RUNTIME_DIGEST,'prompts':PROMPT_DIGEST,'axes':AXIS_DIGEST})
write_json(OUTPUT_ROOT/'capture_identity.json',{'digest':CAPTURE_DIGEST})
del hidden_rows, X, X_fit, decoder, audit_manifest
gc.collect();torch.cuda.empty_cache()
print('Calibrated Native directions frozen at historical doses. Same tensors will be used in both models.')
```

```python
%%cell_progress 11/17 Define validated resumable input shards
cell_stage('Define validated resumable input shards')
def expected_conditions(row): return ALL_CONDITIONS if row['side']=='right' else CORE_CONDITIONS
def shard_path_for(mode,row,variant):
    return SHARD_DIR/f"{mode}__{row['scene_id']}__{variant}.json"
def validate_shard(payload,mode,row,variant):
    assert_native_precision()
    assert payload['capture_identity']==CAPTURE_DIGEST
    assert (payload['model'],payload['scene_id'],payload['mapping_variant'])==(mode,row['scene_id'],variant)
    assert Counter(r['condition'] for r in payload['records'])==Counter(expected_conditions(row))
    assert all(r['patched_tokens']==1 and math.isfinite(r['margin_shift']) for r in payload['records'])
    assert all(math.isfinite(v) for r in [payload['clean'],*payload['records']] for v in r['candidate_logprobs'])
    assert payload['payload_digest']==digest_object({k:payload[k] for k in ('clean','records')})

def run_mode(mode):
    assert_native_precision()
    jobs=[(row,v) for row in scene_manifest for v in MAPPING_VARIANTS]
    bar=tqdm(jobs,desc=f'{mode}: frozen handoff',unit='input')
    with model_mode(mode):
        for row,variant in bar:
            assert_native_precision()
            path=shard_path_for(mode,row,variant)
            if path.is_file():
                validate_shard(read_json(path),mode,row,variant);continue
            started=time.perf_counter()
            examples,mappings=build_target_examples(row);e=examples[variant]
            scores,hidden,masks=capture_examples({variant:e},CAPTURE_LAYERS)
            clean=scores[variant];exact,wrong=row['exact_semantic'],row['shortcut_semantic']
            clean_margin=semantic_margin(clean,exact,wrong)
            untreated={l:hidden[variant][l][masks[variant]] for l in (49,55)}
            operations,axes=build_path_operations(row,variant,mappings[variant],untreated)
            meta={k:row[k] for k in ('match_id','scene_id','cohort','side','style_family','style_id','diagnostic_cell',
                  'train_geometry_seen','dev_geometry_seen','unseen_train_dev_geometry')}
            meta.update(model=mode,mapping_variant=variant)
            clean.update(meta,clean_margin=clean_margin,correct=clean['predicted_semantic']==exact,
                opposite=clean['predicted_semantic']==wrong,
                clean_l48_norm=selected_norm(hidden[variant][48][masks[variant]]),
                clean_l49_norm=selected_norm(untreated[49]),clean_l55_norm=selected_norm(untreated[55]),
                clean_l49_coordinate=vector_projection(untreated[49],axes['semantic']),
                clean_l55_coordinate=vector_projection(untreated[55],axes['code']))
            records=[]
            for condition in expected_conditions(row):
                bar.set_postfix_str(row['cohort']+' | '+row['side']+' | '+condition)
                r,h=forward_path_condition(e,operations[condition],untreated,{'condition':condition})
                margin=semantic_margin(r,exact,wrong)
                r.update(meta,clean_margin=clean_margin,margin_shift=margin-clean_margin,
                    patched_correct=r['predicted_semantic']==exact,
                    newly_correct=(not clean['correct']) and r['predicted_semantic']==exact,
                    newly_wrong=clean['correct'] and r['predicted_semantic']!=exact,
                    l49_toward_exact=vector_projection(h[49]-untreated[49],axes['semantic']),
                    l55_toward_exact_code=vector_projection(h[55]-untreated[55],axes['code']))
                records.append(r)
            payload={'capture_identity':CAPTURE_DIGEST,'model':mode,'scene_id':row['scene_id'],
                     'mapping_variant':variant,'clean':clean,'records':records,
                     'elapsed_seconds':time.perf_counter()-started}
            payload['payload_digest']=digest_object({k:payload[k] for k in ('clean','records')})
            validate_shard(payload,mode,row,variant);write_json(path,payload)
            del hidden;torch.cuda.empty_cache()
    print(mode,'complete:',len(jobs),'inputs')
```

```python
%%cell_progress 15/17 Assemble paired behavior and mechanism effects
cell_stage('Assemble paired behavior and mechanism effects')
def derive_path_metrics(w):
    s=w.copy()
    s['upstream_total']=w['upstream_repair']
    for layer in (49,55):
        s[f'upstream_l{layer}_block_loss']=w['upstream_repair']-w[f'upstream_repair_block_l{layer}']+w[f'self_block_l{layer}']
        s[f'upstream_l{layer}_random_loss']=w['upstream_repair']-w[f'upstream_repair_random_block_l{layer}']+w[f'self_random_block_l{layer}']
        s[f'upstream_l{layer}_specific_loss']=s[f'upstream_l{layer}_block_loss']-s[f'upstream_l{layer}_random_loss']
    s['l49_total']=w['l49_repair']
    s['l49_l55_block_loss']=w['l49_repair']-w['l49_repair_block_l55']+w['self_block_l55']
    s['l49_l55_random_loss']=w['l49_repair']-w['l49_repair_random_block_l55']+w['self_random_block_l55']
    s['l49_l55_specific_loss']=s['l49_l55_block_loss']-s['l49_l55_random_loss']
    s['l55_direct']=w['l55_repair']
    s['l55_bypass_gain']=w['l55_bypass_after_l49_block']-w['upstream_repair_block_l49']
    return s

clean_records,patch_records=[],[]
for mode in MODES:
    for row in scene_manifest:
        for variant in MAPPING_VARIANTS:
            p=read_json(shard_path_for(mode,row,variant));validate_shard(p,mode,row,variant)
            clean_records.append(p['clean']);patch_records.extend(p['records'])
clean_frame,patch_frame=pd.DataFrame(clean_records),pd.DataFrame(patch_records)
assert len(clean_frame)==16*MATCHED_UNITS and len(patch_frame)==184*MATCHED_UNITS
clean_frame.to_csv(ANALYSIS_DIR/'clean_records.csv',index=False)
patch_frame.to_csv(ANALYSIS_DIR/'patch_records.csv',index=False)
scale_frame=clean_frame[['model','match_id','scene_id','cohort','side','mapping_variant','clean_l48_norm','clean_l49_norm','clean_l55_norm']].copy()
for layer,norm in ((48,selected_norm(UPSTREAM_DIRECTION)*MATCHED_UPSTREAM_ALPHA),
                   (49,selected_norm(SEMANTIC_DIRECTION)),(55,CODE_STEERING_NORM)):
    scale_frame[f'l{layer}_payload_over_clean_norm']=norm/scale_frame[f'clean_l{layer}_norm']
scale_frame.to_csv(ANALYSIS_DIR/'intervention_relative_scale.csv',index=False)
INDEX=['model','match_id','scene_id','cohort','side','style_family','style_id','diagnostic_cell','unseen_train_dev_geometry']
wide=patch_frame.pivot(index=INDEX+['mapping_variant'],columns='condition',values='margin_shift').reset_index()
mapping_paths=derive_path_metrics(wide)
PATH_METRICS=['upstream_total','upstream_l49_block_loss','upstream_l49_random_loss','upstream_l49_specific_loss',
    'upstream_l55_block_loss','upstream_l55_random_loss','upstream_l55_specific_loss','l49_total',
    'l49_l55_block_loss','l49_l55_random_loss','l49_l55_specific_loss','l55_direct','l55_bypass_gain']
mapping_paths.to_csv(ANALYSIS_DIR/'mapping_path_metrics.csv',index=False)
scene_paths=mapping_paths.groupby(INDEX,as_index=False)[PATH_METRICS].mean()
scene_paths.to_csv(ANALYSIS_DIR/'scene_path_metrics.csv',index=False)
behavior=clean_frame.pivot(index=['match_id','scene_id','cohort','side','mapping_variant','unseen_train_dev_geometry'],
    columns='model',values=['correct','clean_margin']).reset_index()
behavior.columns=['_'.join(str(x) for x in c if x) if isinstance(c,tuple) else c for c in behavior.columns]
behavior['transition']=np.select([
    ~behavior.correct_native.astype(bool)&behavior.correct_lora.astype(bool),
    behavior.correct_native.astype(bool)&~behavior.correct_lora.astype(bool),
    behavior.correct_native.astype(bool)&behavior.correct_lora.astype(bool)],
    ['repaired','regressed','both_correct'],default='both_wrong')
behavior.to_csv(ANALYSIS_DIR/'behavior_transitions_by_input.csv',index=False)
# No majority vote: two mappings can tie. Report both correct explicitly.
pair_success=clean_frame.groupby(['model','match_id','cohort','mapping_variant']).correct.all().reset_index(name='both_mirrors_correct')
pair_success.to_csv(ANALYSIS_DIR/'both_mirrors_correct.csv',index=False)
behavior_summary=clean_frame.groupby(['model','cohort','side']).agg(accuracy=('correct','mean'),margin=('clean_margin','mean'))
behavior_summary.to_csv(ANALYSIS_DIR/'behavior_summary.csv')
display(behavior_summary)
display(pd.crosstab([behavior.cohort,behavior.side],behavior.transition))
```

## 直接比较前后效应

记 M 为正确物理侧相对相反侧的 log-probability margin，ΔM 为干预后相对本模型 clean 的变化。


```python
%%cell_progress 16/17 Estimate direct LoRA-minus-Native differences
cell_stage('Estimate direct LoRA-minus-Native differences')
def mean_ci(values,key):
    values=np.asarray(values,dtype=float)
    assert np.isfinite(values).all()
    n=len(values)
    if n==0:return {'estimate':None,'ci_low':None,'ci_high':None,'n_families':0}
    if n==1:return {'estimate':float(values.mean()),'ci_low':None,'ci_high':None,'n_families':1}
    rng=np.random.default_rng(int(stable_key(key)[:16],16))
    means=values[rng.integers(0,n,size=(BOOTSTRAP_REPLICATES,n))].mean(axis=1)
    lo,hi=np.quantile(means,[.025,.975])
    return {'estimate':float(values.mean()),'ci_low':float(lo),'ci_high':float(hi),'n_families':n}

def paired_signflip(values,key):
    values=np.asarray(values,dtype=float);assert np.isfinite(values).all()
    if not len(values):return None
    observed=abs(values.mean())
    if observed==0:return 1.0
    if len(values)<=16:
        signs=np.asarray(list(itertools.product((-1,1),repeat=len(values))),dtype=float)
        return float((np.abs(signs@values/len(values))>=observed-1e-12).mean())
    rng=np.random.default_rng(int(stable_key(key)[:16],16))
    exceed=0
    for offset in range(0,SIGNFLIP_REPLICATES,2048):
        count=min(2048,SIGNFLIP_REPLICATES-offset)
        signs=rng.choice([-1.,1.],size=(count,len(values)))
        exceed+=int((np.abs(signs@values/len(values))>=observed-1e-12).sum())
    return (exceed+1)/(SIGNFLIP_REPLICATES+1)

def holm(pvalues):
    p=np.asarray(pvalues,dtype=float);order=np.argsort(p);result=np.empty_like(p)
    result[order]=np.minimum(1,np.maximum.accumulate((len(p)-np.arange(len(p)))*p[order]))
    return result.tolist()

def paired_metric(frame,metric):
    p=frame.pivot(index='match_id',columns='model',values=metric).sort_index()
    assert set(p.columns)==set(MODES) and p.notna().all().all()
    return p,p['lora'].to_numpy()-p['native'].to_numpy()

target=scene_paths[(scene_paths.cohort=='right_error')&(scene_paths.side=='right')]
primary=[]
for metric in PRIMARY_METRICS:
    p,d=paired_metric(target,metric);assert len(d)==MATCHED_UNITS
    primary.append({'metric':metric,**mean_ci(d,('primary',metric)),
        'native_mean':float(p.native.mean()),'lora_mean':float(p.lora.mean()),
        'p_two_sided':paired_signflip(d,('signflip',metric))})
for r,p in zip(primary,holm([r['p_two_sided'] for r in primary])):
    r['p_holm']=p
    r['evidence_of_change']=bool(p<.05 and NATIVE_PARITY_GATE and preflight['primary_test_resolution_sufficient'])
primary_frame=pd.DataFrame(primary);primary_frame.to_csv(ANALYSIS_DIR/'primary_paired_changes.csv',index=False)

# Secondary estimates are descriptive, with pointwise CIs and no further significance gates.
secondary=[]
for (cohort,side),group in scene_paths.groupby(['cohort','side']):
    for metric in PATH_METRICS:
        if group[metric].isna().any():continue  # Right-only interventions are structurally absent on the left.
        p,d=paired_metric(group,metric)
        secondary.append({'scope':cohort+'/'+side,'metric':metric,**mean_ci(d,('secondary',cohort,side,metric))})
unseen_target=target[target.match_id.isin(unseen_ids)]
for metric in PRIMARY_METRICS:
    if len(unseen_target):_,d=paired_metric(unseen_target,metric)
    else:d=[]
    secondary.append({'scope':'unseen_train_dev/right_error/right','metric':metric,**mean_ci(d,('unseen',metric))})
for variant,group in mapping_paths.groupby('mapping_variant'):
    group=group[(group.cohort=='right_error')&(group.side=='right')]
    for metric in PRIMARY_METRICS:
        _,d=paired_metric(group,metric)
        secondary.append({'scope':'mapping/'+variant,'metric':metric,**mean_ci(d,('mapping',variant,metric))})
for metric in PRIMARY_METRICS:
    # LoRA−Native change in the error-vs-correct right-side contrast.
    p=scene_paths[scene_paths.side=='right'].pivot(index='match_id',columns=['model','cohort'],values=metric)
    d=(p[('lora','right_error')]-p[('lora','both_correct')])-(p[('native','right_error')]-p[('native','both_correct')])
    secondary.append({'scope':'right_error_minus_right_correct/model_difference','metric':metric,**mean_ci(d,('interaction',metric))})
pd.DataFrame(secondary).to_csv(ANALYSIS_DIR/'secondary_paired_changes.csv',index=False)

behavior_effects=[]
for (cohort,side),g in clean_frame.groupby(['cohort','side']):
    for metric in ('correct','clean_margin'):
        family=g.groupby(['match_id','model'],as_index=False)[metric].mean()
        _,d=paired_metric(family,metric)
        behavior_effects.append({'cohort':cohort,'side':side,'metric':metric,**mean_ci(d,('behavior',cohort,side,metric))})
pd.DataFrame(behavior_effects).to_csv(ANALYSIS_DIR/'behavior_paired_changes.csv',index=False)
# Pair behavior to mechanism without testing post-treatment-selected groups as primary hypotheses.
joined=patch_frame.merge(behavior[['scene_id','mapping_variant','transition']],on=['scene_id','mapping_variant'],validate='many_to_one')
joined.groupby(['model','transition','condition']).agg(
    n_inputs=('scene_id','size'),n_families=('match_id','nunique'),margin_shift=('margin_shift','mean'),
    patched_accuracy=('patched_correct','mean'),newly_correct=('newly_correct','mean'),newly_wrong=('newly_wrong','mean')
    ).to_csv(ANALYSIS_DIR/'transition_conditioned_descriptive.csv')
display(primary_frame)
print('Positive estimate = larger frozen-axis transmission effect after LoRA, not necessarily better physics.')
print('Sign-flip tests assume symmetric family-level differences under the null; they do not randomize model training.')
```
