```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 17 tracked cells.')
```

```python
%%cell_progress 4/17 Exclude known geometries and freeze new families
cell_stage('Exclude known geometries and freeze new families')
def geometry_from_stacks(stacks, mirror_invariant=True):
    sides = tuple(tuple(sorted((int(s['units']), int(s['distance'])) for s in stacks if s['side']==side))
                  for side in ('left','right'))
    assert sum(map(len,sides)) == len(stacks) and len(stacks)>0
    return min(sides, sides[::-1]) if mirror_invariant else sides

def target_stacks(row):
    return [{'side':s,'units':int(row[f'{s}_units']),'distance':int(row[f'{s}_distance'])}
            for s in ('left','right')]

def read_dataset_metadata():
    names=('dataset_config.json','scenes.jsonl','counterfactual_pairs.jsonl','questions.jsonl','splits.json','statistics.json')
    digest=hashlib.sha256()
    directory = PROJECT_ROOT/'data/dataset_v1'
    if (directory/'scenes.jsonl').is_file():
        for name in names:
            assert (directory/name).is_file(), directory/name
            SOURCE_PATHS['dataset_'+name] = directory/name
            digest.update(name.encode());digest.update(b'\0');digest.update((directory/name).read_bytes());digest.update(b'\0')
        assert 'sha256:'+digest.hexdigest()==adapter_manifest['dataset_fingerprint'], 'Dataset differs from the SFT dataset.'
        return read_jsonl(directory/'scenes.jsonl')
    archive = PROJECT_ROOT/'data/dataset_v1.zip'
    assert archive.is_file(), 'Need the original Dataset v1 scenes.jsonl, or data/dataset_v1.zip, to audit SFT geometry.'
    SOURCE_PATHS['dataset_archive'] = archive
    with ZipFile(archive) as z:
        found = [n for n in z.namelist() if n.rstrip('/').split('/')[-1]=='scenes.jsonl']
        assert len(found)==1, f'Ambiguous Dataset v1 archive: {found}'
        prefix=found[0][:-len('scenes.jsonl')]
        for name in names:
            data=z.read(prefix+name)
            digest.update(name.encode());digest.update(b'\0');digest.update(data);digest.update(b'\0')
        assert 'sha256:'+digest.hexdigest()==adapter_manifest['dataset_fingerprint'], 'Dataset differs from the SFT dataset.'
        return [json.loads(s) for s in z.read(found[0]).decode('utf-8').splitlines() if s.strip()]
def physics_cell(units_left,distance_left,units_right,distance_right):
    torque=units_left*distance_left-units_right*distance_right
    sign=lambda x:int(x>0)-int(x<0)
    w,d,t=sign(units_left-units_right),sign(distance_left-distance_right),sign(torque)
    if t and w!=t and d==t: return 'weight_conflict'
    if t and d!=t and w==t: return 'distance_conflict'
    if t and w==t and d==t: return 'both_cues_aligned'
    return 'mixed_nonbalanced'

def choose_new_families(excluded):
    eligible=[]
    for pair in itertools.combinations_with_replacement(itertools.product(range(1,8),range(1,7)),2):
        (u,d),(v,e)=pair
        maximum=max(u,v)
        if maximum not in MAX_UNITS_STRATA or u*d==v*e: continue
        stacks=[{'side':'left','units':u,'distance':d},{'side':'right','units':v,'distance':e}]
        key=geometry_from_stacks(stacks)
        if key in excluded: continue
        eligible.append({'family_id':'axisdose_'+stable_key(pair)[:18],'pair':pair,'max_units':maximum,
            'geometry_key':key,'diagnostic_cell':physics_cell(u,d,v,e),'margin_bin':min(abs(u*d-v*e)//4,5)})
    chosen=[]
    for maximum in MAX_UNITS_STRATA:
        queues=defaultdict(list)
        for row in eligible:
            if row['max_units']==maximum: queues[(row['diagnostic_cell'],row['margin_bin'])].append(row)
        for key in queues: queues[key].sort(key=lambda r:stable_key(('order',r['family_id'])))
        keys=sorted(queues,key=lambda k:stable_key(('stratum',maximum,k)))
        require(sum(map(len,queues.values()))>=FAMILIES_PER_MAX_UNITS,
            f'Not enough unseen max-units={maximum} families; no silent replacement or sample reduction.')
        current=[]
        while len(current)<FAMILIES_PER_MAX_UNITS:
            for key in keys:
                if queues[key] and len(current)<FAMILIES_PER_MAX_UNITS: current.append(queues[key].pop(0))
        chosen.extend(current)
    require(len({r['geometry_key'] for r in chosen})==N_FAMILIES,'Duplicate mirror-equivalent geometry.')
    return chosen,eligible

dataset_scenes=read_dataset_metadata()
dataset_by_id={str(r['scene_id']):r for r in dataset_scenes}
require(len(dataset_by_id)==len(dataset_scenes),'Duplicate dataset scene IDs.')
train_ids={str(r['scene_id']) for r in read_jsonl(SOURCE_PATHS['sft_train'])}
dev_ids={str(r['scene_id']) for r in read_jsonl(SOURCE_PATHS['sft_dev'])}
require(train_ids and dev_ids and train_ids.isdisjoint(dev_ids),'Invalid SFT split.')
require((train_ids|dev_ids)<=set(dataset_by_id),'Missing SFT geometry metadata.')
direction_support=read_jsonl(DEPTH_SUPPORT_PATH)
audit_rows=read_jsonl(AUDIT_OUTCOME_MANIFEST_PATH)
old_scenes=read_jsonl(PRIOR_SCENES_PATH)
exclusions={
    'sft_train':{geometry_from_stacks(dataset_by_id[s]['stacks']) for s in train_ids},
    'sft_dev':{geometry_from_stacks(dataset_by_id[s]['stacks']) for s in dev_ids},
    'all_old_physical_audit':{geometry_from_stacks(target_stacks(r)) for r in audit_rows},
    'old_paired_cohort':{geometry_from_stacks(target_stacks(r)) for r in old_scenes},
    'direction_support':{geometry_from_stacks(role['stacks']) for r in direction_support for role in r['roles'].values()},
}
excluded=set().union(*exclusions.values())
families,eligible=choose_new_families(excluded)
freeze_json(OUTPUT_ROOT/'geometry_selection.json',{'families':families,'eligible':eligible,
    'excluded':{k:sorted(v) for k,v in exclusions.items()},'method':'physics_only_stratified_no_outcome_screening'})
from visual_torque_toolkit.world import Scene, Side, Stack
scene_manifest=[]
image_hashes={}
for family in tqdm(families,desc='Render new geometry mirrors',unit='family'):
    (u,d),(v,e)=family['pair']
    anchor=Scene(family['family_id'],(Stack('left',Side.LEFT,d,u),Stack('right',Side.RIGHT,e,v)))
    # Regenerate deterministically and compare hashes with any frozen images on resume.
    anchor_path=RUNTIME_ROOT/(family['family_id']+'__anchor.png')
    mirror_path=RUNTIME_ROOT/(family['family_id']+'__mirror.png')
    render_scene(anchor,anchor_path,RenderStyle(**STYLE_SPEC))
    with Image.open(anchor_path) as im: ImageOps.mirror(im.convert('RGB')).save(mirror_path)
    for orientation,path,values in [('anchor',anchor_path,(u,d,v,e)),('mirror',mirror_path,(v,e,u,d))]:
        lu,ld,ru,rd=values; torque=lu*ld-ru*rd; exact='left' if torque>0 else 'right'
        sid=family['family_id']+'__'+orientation
        image_hashes[sid]=sha256_file(path)
        persistent=OUTPUT_ROOT/'selected_images'/path.name; persistent.parent.mkdir(exist_ok=True)
        if persistent.exists(): require(sha256_file(persistent)==image_hashes[sid],'Rendered image changed on resume.')
        else: shutil.copy2(path,persistent)
        scene_manifest.append({'scene_id':sid,'family_id':family['family_id'],'match_id':family['family_id'],
            'pair_instance_id':family['family_id'],'orientation':orientation,'max_units':family['max_units'],
            'diagnostic_cell':family['diagnostic_cell'],'margin_bin':family['margin_bin'],
            'left_units':lu,'left_distance':ld,'right_units':ru,'right_distance':rd,'net_torque':torque,
            'exact_semantic':exact,'shortcut_semantic':'right' if exact=='left' else 'left',
            'side':exact,'runtime_image_path':str(path),'image_sha256':image_hashes[sid]})
require(len(scene_manifest)==2*N_FAMILIES,'Wrong rendered cohort size.')
require(all(geometry_from_stacks(target_stacks(r)) not in excluded for r in scene_manifest),'Geometry overlap.')
freeze_json(OUTPUT_ROOT/'scene_manifest.json', [{k:v for k,v in r.items() if k!='runtime_image_path'} for r in scene_manifest])
direction_image_hashes={}
for name in sorted({role['image_filename'] for r in direction_support for role in r['roles'].values()}):
    source=INVERSE_SCENE_DIR/name
    require(source.is_file(),f'Missing direction image: {source}')
    target=DIRECTION_RUNTIME_DIR/name;target.parent.mkdir(parents=True,exist_ok=True);shutil.copy2(source,target)
    direction_image_hashes[name]=sha256_file(source)
    require(sha256_file(target)==direction_image_hashes[name],'Direction image copy mismatch.')
    SOURCE_PATHS['direction_image/'+name]=source
# Deterministic old anchors are technical checks only. Keep their old mapping seed.
replay_rows=sorted(old_scenes,key=lambda r:stable_key(('replay',r['scene_id'])))[:4]
for row in replay_rows:
    name=Path(row['image_path']).name
    source=PRIOR_SCENES_PATH.parent/'selected_images'/name
    destination=RUNTIME_ROOT/('old_'+name);shutil.copy2(source,destination)
    row['runtime_image_path']=str(destination)
    SOURCE_PATHS['replay_image/'+row['scene_id']]=source
    SOURCE_PATHS['replay_shard/'+row['scene_id']]=SOURCE_ROOT/'paired_shards'/f"native__{row['scene_id']}__forward.json"
    require(SOURCE_PATHS['replay_shard/'+row['scene_id']].is_file(),'Missing old Native replay shard.')
print('New families:',len(families),'| max-units:',Counter(r['max_units'] for r in families))
print('Exclusion groups:',{k:len(v) for k,v in exclusions.items()})
display(pd.DataFrame(families).groupby(['max_units','diagnostic_cell']).size().rename('families').reset_index())
```

```python
%%cell_progress 5/17 Freeze prompts and scientific protocol
cell_stage('Freeze prompts and scientific protocol')
import math
from dataclasses import replace as dataclass_replace
from visual_torque_toolkit.behavior.data import BehaviorExample

RULE_TEXT = (
    'Each square is one unit of weight. Distance is measured in tick '
    'intervals from the center pivot. Torque equals weight times distance. '
    'Total torque on a side is the sum over stacks on that side.'
)
QUESTION_TEXT = {
    'outcome': 'Which way will the balance scale tip?',
    'weight_more': 'Which side has more total weight?',
    'weight_less': 'Which side has less total weight?',
}
SEMANTIC_TEXT = {'left': 'left side', 'balanced': 'equal', 'right': 'right side'}

def outcome(value):
    return 'left' if value > 0 else 'right' if value < 0 else 'balanced'

def task_semantic(features, task):
    if task == 'outcome':
        value = int(features['net_torque'])
    elif task == 'weight_more':
        value = int(features['weight_score'])
    elif task == 'weight_less':
        value = -int(features['weight_score'])
    else:
        raise KeyError(task)
    return outcome(value)

def mapping_by_label(semantics):
    assert set(semantics) == {'left', 'balanced', 'right'}
    return tuple({
        'position': index, 'label': LABELS[index], 'semantic': semantic,
    } for index, semantic in enumerate(semantics))

def direction_reference_mapping(row):
    semantics = ['left', 'balanced', 'right']
    offset = int(stable_key(
        f"options:{row['family_id']}", seed=DIRECTION_SEED
    )[:8], 16) % 3
    return mapping_by_label(semantics[offset:] + semantics[:offset])

def target_mappings(row):
    exact, wrong = row['exact_semantic'], row['shortcut_semantic']
    assert {exact, wrong} == {'left', 'right'}
    balanced_label = LABELS[int(stable_key(
        ('balanced-label', row['pair_instance_id']), seed=EXPANSION_SEED
    )[:8], 16) % 3]
    side_labels = [label for label in LABELS if label != balanced_label]
    result = {}
    for variant, exact_label, wrong_label in (
        ('forward', side_labels[0], side_labels[1]),
        ('reverse', side_labels[1], side_labels[0]),
    ):
        by_label = {
            exact_label: exact, wrong_label: wrong, balanced_label: 'balanced'
        }
        result[variant] = mapping_by_label(tuple(by_label[label] for label in LABELS))
    return result

def semantics_by_label(mapping):
    return tuple(next(
        item['semantic'] for item in mapping if item['label'] == label
    ) for label in LABELS)

def label_for(mapping, semantic):
    return next(item['label'] for item in mapping if item['semantic'] == semantic)

def custom_prompt(task, mapping):
    options = '\n'.join(
        f"{item['label']}. {SEMANTIC_TEXT[item['semantic']]}" for item in mapping
    )
    return '\n\n'.join([
        'Answer this question using only the supplied balance-scale scene.',
        RULE_TEXT, 'Question: ' + QUESTION_TEXT[task], 'Options:\n' + options,
        'Do not explain your reasoning. Reply with exactly one option letter '
        'and nothing else. Valid replies are: A, B, C.',
    ])

def make_direction_example(row, role_record, task, mapping, name):
    features = role_record['features']
    semantics = semantics_by_label(mapping)
    gold = task_semantic(features, task)
    return BehaviorExample(
        example_id=f"exactphysics_direction:{row['family_id']}:{name}",
        question_id=f"exactphysics_direction_{row['family_id']}_{name}",
        scene_id=str(role_record['scene_id']), condition=PromptCondition.IMAGE_RULE,
        question_kind=task, variant='exactphysics_direction', split='direction_support',
        problem_type='semantic_direction', outcome=gold,
        net_torque=int(features['net_torque']), prompt=custom_prompt(task, mapping),
        image_path=(DIRECTION_RUNTIME_DIR / role_record['image_filename']).resolve(),
        labels=LABELS, option_text=tuple(SEMANTIC_TEXT[x] for x in semantics),
        option_semantics=semantics, gold_label=LABELS[semantics.index(gold)],
        gold_semantic=gold,
    )

def build_direction_examples(row):
    mapping = direction_reference_mapping(row)
    examples = {}
    for task in DONOR_TASKS:
        for role in ('base', 'aligned_source'):
            key = f'{task}__{role}'
            examples[key] = make_direction_example(
                row, row['roles'][role], task, mapping, key
            )
    return examples, mapping

def build_target_examples(row):
    mappings = target_mappings(row)
    base = outcome_example_by_scene[row['scene_id']]
    examples = {}
    for variant, mapping in mappings.items():
        semantics = semantics_by_label(mapping)
        examples[variant] = dataclass_replace(
            base,
            example_id=f"independent_exactphysics_target:{row['scene_id']}:{variant}",
            question_id=f"independent_exactphysics_target_{row['scene_id']}_{variant}",
            variant=f'independent_exactphysics_{variant}', prompt=custom_prompt('outcome', mapping),
            labels=LABELS, option_text=tuple(SEMANTIC_TEXT[x] for x in semantics),
            option_semantics=semantics,
            gold_label=LABELS[semantics.index(row['exact_semantic'])],
            gold_semantic=row['exact_semantic'],
        )
    return examples, mappings

def output_hidden_tensor(output):
    if torch.is_tensor(output):
        hidden = output
    elif isinstance(output, (tuple, list)) and output and torch.is_tensor(output[0]):
        hidden = output[0]
    elif hasattr(output, 'last_hidden_state') and torch.is_tensor(output.last_hidden_state):
        hidden = output.last_hidden_state
    else:
        raise TypeError(f'Unsupported output: {type(output).__name__}')
    assert hidden.ndim == 3
    return hidden

def replace_output_hidden(output, modified):
    if torch.is_tensor(output):
        return modified
    if isinstance(output, tuple):
        return (modified,) + output[1:]
    if isinstance(output, list):
        return [modified] + output[1:]
    raise TypeError(type(output).__name__)

def prepare_examples(examples):
    examples = tuple(examples)
    plan = adapter._candidate_plan(examples[0])
    if plan is None:
        raise RuntimeError('A/B/C labels must be single-token candidates.')
    for example in examples[1:]:
        assert adapter._candidate_plan(example) == plan
    fixed_prefix, token_by_label = plan
    inputs = adapter._append_text_prefix(
        adapter._to_device(adapter._template(
            [example.messages() for example in examples], add_generation_prompt=True,
        )), fixed_prefix,
    )
    return {'examples': examples, 'inputs': inputs, 'token_by_label': token_by_label}

def final_mask(prepared):
    assert len(prepared['examples']) == 1
    attention = prepared['inputs']['attention_mask'][0].detach().cpu().bool()
    mask = torch.zeros_like(attention)
    mask[int(torch.where(attention)[0][-1])] = True
    return mask

print('Prompt builders ready.')

outcome_example_by_scene={}
for row in [*scene_manifest,*replay_rows]:
    outcome_example_by_scene[row['scene_id']]=BehaviorExample(
        example_id=row['scene_id'],question_id=row['scene_id'],scene_id=row['scene_id'],
        condition=PromptCondition.IMAGE_RULE,question_kind='outcome',variant='axis_dose',split='confirmation',
        problem_type='single_stack',outcome=row['exact_semantic'],net_torque=int(row['net_torque']),
        prompt='',image_path=Path(row['runtime_image_path']),labels=LABELS,
        option_text=('left side','equal','right side'),option_semantics=('left','balanced','right'),
        gold_label='A' if row['exact_semantic']=='left' else 'C',gold_semantic=row['exact_semantic'])
prompt_inventory=[]
for row in scene_manifest:
    for variant,e in build_target_examples(row)[0].items():
        prompt_inventory.append({'scene_id':row['scene_id'],'mapping':variant,'prompt':e.prompt,
            'labels':e.labels,'semantics':e.option_semantics,'image_sha256':row['image_sha256']})
PROMPT_DIGEST=freeze_json(OUTPUT_ROOT/'prompt_inventory.json',prompt_inventory)
protocol={'implementation':IMPLEMENTATION_SHA256,'model':MODEL_ID,'revision':MODEL_REVISION,
    'dtype':MODEL_DTYPE,'source_hashes':{k:sha256_file(p) for k,p in SOURCE_PATHS.items()},
    'toolkit_files':hash_tree(DRIVE_TOOLKIT_ROOT/'src'),'parent_capture':old_capture,
    'images':image_hashes,'prompts':PROMPT_DIGEST,'direction_images':direction_image_hashes,
    'geometry':digest_object(families),'n_families':N_FAMILIES,'scope':'new max-units 6/7 geometry; no outcome filtering',
    'fit_source':'original 32 depth-support families, same recipe in Native/LoRA',
    'l48':'fixed historical Native physical input; LoRA L48 axis not fitted',
    'scales':SCALES,'doses':DOSES,'clamp_dose':1.0,'random_axes':'one shared deterministic axis per input and layer',
    'relative_rule':'historical_native_norm * recipient_clean_norm / paired_native_clean_norm',
    'primary_scope':'all new families; average mirrors and maps; relative scale; dose +1; Native injection',
    'primary_tests':'3 route recipient-model differences with Native clamps + 3 model x clamp-source interactions; Holm over 6',
    'secondary':'input-axis crossover, dose curve, absolute/relative, signed projection, max-units and maps',
    'independent_unit':'mirror-invariant physical family','bootstrap':BOOTSTRAP_REPLICATES,
    'signflip':SIGNFLIP_REPLICATES,'seed':SEED,'direction_seed':DIRECTION_SEED,'mapping_seed':EXPANSION_SEED,
    'native_numerics_policy':NATIVE_NUMERICS_POLICY,'style':STYLE_SPEC}
PROTOCOL_DIGEST=freeze_json(OUTPUT_ROOT/'frozen_protocol.json',protocol)
print('Frozen before loading base / inspecting confirmation scores:',PROTOCOL_DIGEST)
```

```python
%%cell_progress 7/17 Define compact captures and verify disabled adapter
cell_stage('Define compact captures and verify disabled adapter')
def score_records(prepared, logits):
    logprobs = torch.log_softmax(logits[:, -1, :].float(), dim=-1)
    records = []
    for index, example in enumerate(prepared['examples']):
        scores = [
            float(logprobs[index, prepared['token_by_label'][label]].detach().cpu())
            for label in example.labels
        ]
        maximum = max(scores)
        weights = [math.exp(value - maximum) for value in scores]
        probabilities = [value / sum(weights) for value in weights]
        predicted = int(np.argmax(scores))
        records.append({
            'candidate_labels': list(example.labels),
            'candidate_semantics': list(example.option_semantics),
            'candidate_logprobs': scores,
            'candidate_probabilities': probabilities,
            'predicted_label': example.labels[predicted],
            'predicted_semantic': example.option_semantics[predicted],
            'gold_label': example.gold_label,
            'gold_semantic': example.gold_semantic,
        })
    return records

def selected_norm(tensor):
    return float(torch.linalg.vector_norm(tensor.float()))

def matched_random(tensor, key):
    generator = torch.Generator(device='cpu')
    generator.manual_seed(int(stable_key(key)[:16], 16) % (2**63 - 1))
    noise = torch.randn(tensor.shape, generator=generator, dtype=torch.float32)
    return (noise * (selected_norm(tensor) / selected_norm(noise))).to(tensor.dtype)

def centered_scores(record):
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return values - values.mean()

def cosine_similarity(left, right):
    left = left.float().reshape(-1)
    right = right.float().reshape(-1)
    denominator = torch.linalg.vector_norm(left) * torch.linalg.vector_norm(right)
    assert float(denominator) > 0
    return float(torch.dot(left, right) / denominator)

def rescale_direction(tensor, target_norm):
    norm = selected_norm(tensor)
    assert norm > 0 and target_norm > 0
    return (tensor.float() * (float(target_norm) / norm)).to(tensor.dtype)

def unit_direction(tensor):
    return rescale_direction(tensor, 1.0).float()

def semantic_margin(record, exact_semantic, wrong_semantic):
    semantics = list(record['candidate_semantics'])
    values = np.asarray(record['candidate_logprobs'], dtype=float)
    return float(
        values[semantics.index(exact_semantic)]
        - values[semantics.index(wrong_semantic)]
    )

def vector_projection(delta, direction):
    unit = unit_direction(direction).reshape(-1)
    return float(torch.dot(delta.float().reshape(-1), unit))

def clamp_selected_coordinate(selected, untreated, axis):
    untreated = untreated.to(device=selected.device, dtype=selected.dtype)
    axis_float = axis.to(device=selected.device, dtype=torch.float32)
    axis_float = axis_float / torch.linalg.vector_norm(axis_float)
    displacement = selected.float() - untreated.float()
    coefficient = torch.sum(displacement * axis_float, dim=-1, keepdim=True)
    clamped = selected.float() - coefficient * axis_float
    return clamped.to(selected.dtype), float(coefficient.detach().cpu().item())
def torch_save_atomic(path,payload):
    path=Path(path);tmp=path.with_suffix(path.suffix+'.tmp');torch.save(payload,tmp);tmp.replace(path)

def capture_one(example,layers=CAPTURE_LAYERS):
    assert_native_precision()
    prepared=prepare_examples([example]); mask=final_mask(prepared)
    require(int(mask.sum())==1,'Expected exactly one final token.')
    selected={};handles=[]
    for layer in layers:
        def hook(_module,_inputs,output,layer=layer):
            h=output_hidden_tensor(output)
            selected[layer]=h[0,mask.to(h.device)].detach().cpu().clone()
        handles.append(DECODER_LAYERS[layer].register_forward_hook(hook))
    try:
        with torch.inference_mode(): result=adapter.model(**prepared['inputs'],use_cache=False)
        score=score_records(prepared,result.logits)[0]
    finally:
        for handle in handles:handle.remove()
    require(set(selected)==set(layers),'Some requested layers did not execute.')
    require(all(t.ndim==2 and t.shape[0]==1 and torch.isfinite(t).all() for t in selected.values()),'Invalid final state.')
    return score,selected

def capture_parity_example(example):
    score,selected=capture_one(example)
    return score,{k:v.float().numpy().copy() for k,v in selected.items()}
def compare_disabled_capture(bare_score, disabled_score, bare_selected, disabled_selected):
    score_error=float(np.max(np.abs(centered_scores(disabled_score)-centered_scores(bare_score))))
    assert set(bare_selected)==set(disabled_selected)==set(CAPTURE_LAYERS)
    layers={}
    for layer in CAPTURE_LAYERS:
        x,y=np.asarray(bare_selected[layer]),np.asarray(disabled_selected[layer])
        assert x.shape==y.shape and x.ndim==2 and x.shape[0]==1
        assert np.isfinite(x).all() and np.isfinite(y).all(), 'Nonfinite disabled-base hidden state.'
        layers[str(layer)]={'exactly_equal':bool(np.array_equal(x,y)),
            'max_absolute_error':float(np.max(np.abs(x-y))),
            'relative_l2_error':float(np.linalg.norm(x-y)/max(float(np.linalg.norm(x)),1e-12))}
    return {'centered_score_error':score_error,'layers':layers,
        'passed':bool(np.isfinite(score_error) and score_error<SMOKE_TOLERANCE and
                      all(value['exactly_equal'] for value in layers.values()))}
# Technical parity uses development prompts only; no confirmation outcomes yet.
dev_smoke=[e for f in direction_support[:2] for e in build_direction_examples(f)[0].values()]
bare_captures=[capture_parity_example(e) for e in dev_smoke]
adapter.model=PeftModel.from_pretrained(adapter.model,str(ADAPTER_DIR),is_trainable=False)
adapter.model.eval()
for p in adapter.model.parameters():p.requires_grad_(False)
require(not getattr(adapter.model,'is_loaded_in_4bit',False) and not getattr(adapter.model,'is_loaded_in_8bit',False),'Quantized model.')
require(any(torch.count_nonzero(p).item()>0 for n,p in adapter.model.named_parameters() if 'lora_B' in n),'Adapter has no trained LoRA B.')

@contextmanager
def model_mode(mode):
    assert_native_precision()
    if mode=='native':
        with adapter.model.disable_adapter():yield
    elif mode=='lora':yield
    else:raise ValueError(mode)

disabled_rows=[]
with model_mode('native'):
    for e,(old_score,old_states) in zip(dev_smoke,bare_captures):
        score,states=capture_parity_example(e)
        disabled_rows.append({'example':e.example_id,**compare_disabled_capture(old_score,score,old_states,states)})
write_json(OUTPUT_ROOT/'disabled_base_parity.json',{'runtime':RUNTIME_DIGEST,'rows':disabled_rows,'passed':all(r['passed'] for r in disabled_rows)})
require(all(r['passed'] for r in disabled_rows),'Bare/disabled parity failed; diagnostic saved.')
del bare_captures
print('Bare and disabled Native final-token states are exactly equal on all development parity inputs.')
```

```python
%%cell_progress 9/17 Define independent injection and clamp crossover
cell_stage('Define independent injection and clamp crossover')
def intervention_plan():
    plans=[]
    for layer in (49,55):
        for clamp in ('native','lora','random'):
            plans.append(dict(kind='self',layer=layer,axis='none',scale='none',dose=0.0,mediator=layer,clamp=clamp))
    for scale in SCALES:
        for layer in CAPTURE_LAYERS:
            for source in (('native',) if layer==48 else MODES):
                for dose in DOSES:
                    plans.append(dict(kind='add',layer=layer,axis=source,scale=scale,dose=dose,mediator=0,clamp='none'))
                for mediator in ((49,55) if layer==48 else (55,) if layer==49 else ()):
                    for clamp in ('native','lora','random'):
                        plans.append(dict(kind='block',layer=layer,axis=source,scale=scale,dose=1.0,mediator=mediator,clamp=clamp))
    for p in plans:p['condition_id']=digest_object(p).split(':')[1][:24]
    require(len({p['condition_id'] for p in plans})==len(plans)==70,'Wrong or duplicate condition design.')
    return plans

PLAN=intervention_plan()
PLAN_BY_ID={p['condition_id']:p for p in PLAN}
freeze_json(OUTPUT_ROOT/'condition_plan.json',PLAN)

def axis_for(layer,source,row,mapping):
    sign=1.0 if row['exact_semantic']=='left' else -1.0
    if layer==48:
        require(source=='native','L48 is the fixed Native anchor.')
        return UPSTREAM_DIRECTION*sign
    if layer==49:return AXES[source]['semantic']*sign
    transition=label_for(mapping,row['shortcut_semantic'])+'>'+label_for(mapping,row['exact_semantic'])
    return AXES[source]['code'][transition]

def dose_target(reference_norm,recipient_clean_norm,native_clean_norm,scale,dose):
    require(all(math.isfinite(x) and x>0 for x in (reference_norm,recipient_clean_norm,native_clean_norm)),'Invalid dose anchor.')
    require(scale in SCALES and dose in DOSES,'Unknown dose condition.')
    return abs(dose)*reference_norm*(recipient_clean_norm/native_clean_norm if scale=='relative' else 1.0)

def make_payload(plan,row,mapping,own,native):
    layer=plan['layer'];axis=axis_for(layer,plan['axis'],row,mapping)
    reference=selected_norm(axis_for(layer,'native',row,mapping))*(MATCHED_UPSTREAM_ALPHA if layer==48 else 1.0)
    desired=dose_target(reference,selected_norm(own[layer]),selected_norm(native[layer]),plan['scale'],plan['dose'])
    fp32=axis.float()*(desired/selected_norm(axis))*(1.0 if plan['dose']>0 else -1.0)
    cast=fp32.to(dtype=own[layer].dtype)
    cast_norm=selected_norm(cast)
    require(abs(cast_norm-desired)/desired<.005,'Payload BF16 norm differs unexpectedly from prescribed dose.')
    effective=(own[layer]+cast).float()-own[layer].float()
    meta={'historical_native_reference_norm':reference,'recipient_clean_norm':selected_norm(own[layer]),
        'paired_native_clean_norm':selected_norm(native[layer]),'requested_payload_norm':desired,
        'fp32_payload_norm':selected_norm(fp32),'bf16_payload_norm':cast_norm,
        'bf16_payload_over_clean_norm':cast_norm/selected_norm(own[layer]),
        'effective_added_displacement_norm':selected_norm(effective)}
    return cast,meta

def random_axis(layer,row,variant,mapping):
    return matched_random(unit_direction(axis_for(layer,'native',row,mapping)),('axisdose_shared',row['scene_id'],variant,layer))

def make_operations(plan,row,variant,mapping,own,native):
    operations={};dose_meta={}
    if plan['kind']!='self':
        payload,dose_meta=make_payload(plan,row,mapping,own,native)
        operations[plan['layer']]={'action':'add','payload':payload}
    if plan['mediator']:
        layer=plan['mediator']
        axis=random_axis(layer,row,variant,mapping) if plan['clamp']=='random' else axis_for(layer,plan['clamp'],row,mapping)
        operations[layer]={'action':'clamp_axis','axis':axis,'untreated':own[layer]}
    return operations,dose_meta

def evaluate_operations(example,operations,own,row,mapping):
    assert_native_precision()
    prepared=prepare_examples([example]);mask=final_mask(prepared)
    require(int(mask.sum())==1,'Intervention must affect one final token.')
    traces={};handles=[]
    for layer in CAPTURE_LAYERS:
        def hook(_module,_inputs,output,layer=layer):
            hidden=output_hidden_tensor(output);indices=mask.to(hidden.device)
            before=hidden[0,indices];after=before;operation=operations.get(layer);trace={}
            if operation is not None:
                if operation['action']=='add':after=before+operation['payload'].to(device=hidden.device,dtype=hidden.dtype)
                elif operation['action']=='clamp_axis':
                    after,coefficient=clamp_selected_coordinate(before,operation['untreated'],operation['axis'])
                    trace['removed_projection_signed']=coefficient
                    trace['removed_projection_abs']=abs(coefficient)
                    trace['clamp_axis_unit_norm']=selected_norm(unit_direction(operation['axis']))
                    baseline=operation['untreated'].to(device=after.device)
                    trace['clamp_residual_signed']=vector_projection(after.float()-baseline.float(),operation['axis'].to(after.device))
                else:raise ValueError(operation['action'])
                trace['actual_local_edit_norm']=selected_norm(after.float()-before.float())
            baseline=own[layer].to(device=after.device)
            delta=after.float()-baseline.float()
            trace['displacement_from_own_clean_norm']=selected_norm(delta)
            for source in (('native',) if layer==48 else MODES):
                trace['projection_on_'+source]=vector_projection(delta,axis_for(layer,source,row,mapping).to(after.device))
            traces[str(layer)]=trace
            if operation is None:return None
            modified=hidden.clone();modified[0,indices]=after
            return replace_output_hidden(output,modified)
        handles.append(DECODER_LAYERS[layer].register_forward_hook(hook))
    try:
        with torch.inference_mode():result=adapter.model(**prepared['inputs'],use_cache=False)
        score=score_records(prepared,result.logits)[0]
    finally:
        for handle in handles:handle.remove()
    require(set(traces)=={'48','49','55'},'Missing intervention trace.')
    return score,traces

def validate_scores(score,example):
    require(score['candidate_labels']==list(example.labels) and score['candidate_semantics']==list(example.option_semantics),'Candidate mapping mismatch.')
    values=np.asarray(score['candidate_logprobs'],dtype=float)
    require(values.shape==(3,) and np.isfinite(values).all(),'Invalid candidate scores.')
    require(score['predicted_semantic']==example.option_semantics[int(np.argmax(values))],'Saved prediction disagrees with scores.')
```

```python
%%cell_progress 12/17 Define resumable per-condition runner
cell_stage('Define resumable per-condition runner')
def validate_intervention_record(record,plan,row,variant,example):
    require(all(record[k]==v for k,v in plan.items()),'Saved condition metadata differs.')
    validate_scores(record['score'],example)
    require(math.isfinite(record['margin_shift']),'Nonfinite margin shift.')
    require(record['patched_tokens']==1,'Wrong intervention token count.')
    require(set(record['traces'])=={'48','49','55'},'Missing trace layer.')
    require(all(math.isfinite(float(v)) for t in record['traces'].values() for v in t.values()),'Nonfinite trace.')
    if plan['kind']!='self':
        expected=record['dose_metadata']['effective_added_displacement_norm']
        actual=record['traces'][str(plan['layer'])]['actual_local_edit_norm']
        require(math.isclose(actual,expected,abs_tol=1e-4,rel_tol=1e-5),
            f'Actual hook displacement {actual} differs from frozen BF16 dose {expected}.')

def validate_partial(payload,mode,row,variant,example,clean):
    require(payload['run_identity']==RUN_DIGEST,'Intervention cache identity changed.')
    require((payload['model'],payload['scene_id'],payload['mapping'])==(mode,row['scene_id'],variant),'Wrong cached input.')
    require(payload['clean_sha256']==CLEAN_HASHES[clean_path(mode,row,variant).name],'Intervention clean reference differs.')
    records=payload['records'];ids=[r['condition_id'] for r in records]
    require(ids==[p['condition_id'] for p in PLAN[:len(ids)]] and len(ids)<=len(PLAN),'Incomplete, duplicate or out-of-order condition cache.')
    require(payload['records_digest']==digest_object(records),'Intervention record hash mismatch.')
    for record,plan in zip(records,PLAN):
        validate_intervention_record(record,plan,row,variant,example)
        shift=semantic_margin(record['score'],row['exact_semantic'],row['shortcut_semantic'])-semantic_margin(clean['score'],row['exact_semantic'],row['shortcut_semantic'])
        require(abs(shift-record['margin_shift'])<1e-9,'Saved margin shift cannot be reproduced.')

def run_mode(mode):
    assert_native_precision()
    jobs=list(itertools.product(scene_manifest,MAPPING_VARIANTS));durations=[]
    with model_mode(mode):
        for job_index,(row,variant) in enumerate(tqdm(jobs,desc=f'{mode}: crossed interventions',unit='input')):
            e=build_target_examples(row)[0][variant];mapping=target_mappings(row)[variant]
            clean=load_clean(mode,row,variant);native=load_clean('native',row,variant)['states']
            path=SHARD_DIR/f"{mode}__{row['scene_id']}__{variant}.json"
            if path.exists():payload=read_json(path);validate_partial(payload,mode,row,variant,e,clean)
            else:payload={'run_identity':RUN_DIGEST,'model':mode,'scene_id':row['scene_id'],'mapping':variant,
                'clean_sha256':CLEAN_HASHES[clean_path(mode,row,variant).name],'records':[],'records_digest':digest_object([])}
            if len(payload['records'])==len(PLAN):continue
            started=time.perf_counter();remaining=PLAN[len(payload['records']):]
            for plan in tqdm(remaining,desc=f"{mode} {row['scene_id']} {variant}",unit='condition',leave=False):
                operations,dose_meta=make_operations(plan,row,variant,mapping,clean['states'],native)
                score,traces=evaluate_operations(e,operations,clean['states'],row,mapping)
                record={**plan,'score':score,'traces':traces,'dose_metadata':dose_meta,'patched_tokens':1,
                    'centered_score_change':float(np.max(np.abs(centered_scores(score)-centered_scores(clean['score'])))),
                    'margin_shift':semantic_margin(score,row['exact_semantic'],row['shortcut_semantic'])-semantic_margin(clean['score'],row['exact_semantic'],row['shortcut_semantic'])}
                try:validate_intervention_record(record,plan,row,variant,e)
                except Exception:
                    write_json(OUTPUT_ROOT/'failed_intervention_diagnostic.json',{'model':mode,'scene':row,'mapping':variant,'record':record})
                    raise
                payload['records'].append(record);payload['records_digest']=digest_object(payload['records'])
                write_json(path,payload)  # atomic after every condition; interrupted inputs resume too
            durations.append((time.perf_counter()-started)/len(remaining))
            if len(durations)==1 or (job_index+1)%16==0:
                hours=(len(jobs)-job_index-1)*len(PLAN)*float(np.mean(durations))/3600
                print(f'{mode}: measured mean {np.mean(durations):.2f} s/condition; remaining in this mode ~{hours:.2f} h (uncached upper estimate).')
            del clean,native;gc.collect()
    print(mode,'complete; all planned inputs retained, no behavior filtering.')
```

```python
%%cell_progress 15/17 Validate records doses and paired block losses
cell_stage('Validate records doses and paired block losses')
def specific_loss_from_records(records,layer,source,scale,mediator,clamp_source):
    def pick(kind,clamp,mediator_value):
        hits=[r for r in records if r['kind']==kind and r['clamp']==clamp and r['mediator']==mediator_value
            and (kind=='self' or (r['layer']==layer and r['axis']==source and r['scale']==scale and r['dose']==1.0))]
        require(len(hits)==1,'Missing or ambiguous condition while assembling loss.')
        return hits[0]['margin_shift']
    total=pick('add','none',0)
    self_target=pick('self',clamp_source,mediator);self_random=pick('self','random',mediator)
    target_response=pick('block',clamp_source,mediator);random_response=pick('block','random',mediator)
    target_loss=total-target_response+self_target
    random_loss=total-random_response+self_random
    return {'total_effect':total,'target_block_loss':target_loss,'random_block_loss':random_loss,
        'specific_loss':target_loss-random_loss,'self_target':self_target,'self_random':self_random,
        'target_clamped_response':target_response,'random_clamped_response':random_response}

clean_rows=[];patch_rows=[];trace_rows=[];loss_rows=[];source_shard_hashes={}
for mode in MODES:
    for row in scene_manifest:
        for variant,e in build_target_examples(row)[0].items():
            clean=load_clean(mode,row,variant)
            path=SHARD_DIR/f"{mode}__{row['scene_id']}__{variant}.json"
            payload=read_json(path);validate_partial(payload,mode,row,variant,e,clean)
            require(len(payload['records'])==len(PLAN),'Run all intervention stages before analysis.')
            source_shard_hashes[path.name]=sha256_file(path)
            meta={k:row[k] for k in ('family_id','scene_id','max_units','side','diagnostic_cell','margin_bin')}
            meta.update(model=mode,mapping=variant)
            clean_rows.append({**meta,'clean_margin':semantic_margin(clean['score'],row['exact_semantic'],row['shortcut_semantic']),
                'correct':clean['score']['predicted_semantic']==row['exact_semantic'],
                'predicted_semantic':clean['score']['predicted_semantic'],
                **{f'clean_l{l}_norm':selected_norm(t) for l,t in clean['states'].items()}})
            for r in payload['records']:
                patch_rows.append({**meta,**{k:r[k] for k in PLAN[0]},'margin_shift':r['margin_shift'],
                    'centered_score_change':r['centered_score_change'],
                    'correct':r['score']['predicted_semantic']==row['exact_semantic'],**r['dose_metadata']})
                for layer,trace in r['traces'].items():
                    trace_rows.append({**meta,**{k:r[k] for k in PLAN[0]},'trace_layer':int(layer),**trace})
            for scale in SCALES:
                for layer,mediator in ((48,49),(48,55),(49,55)):
                    for source in (('native',) if layer==48 else MODES):
                        for clamp in MODES:
                            loss_rows.append({**meta,'route':f'L{layer}_via_L{mediator}','axis':source,'scale':scale,'clamp':clamp,
                                **specific_loss_from_records(payload['records'],layer,source,scale,mediator,clamp)})
clean_frame,patch_frame,trace_frame,loss_frame=map(pd.DataFrame,(clean_rows,patch_rows,trace_rows,loss_rows))
require(len(clean_frame)==8*N_FAMILIES and len(patch_frame)==70*8*N_FAMILIES,'Final count mismatch.')
require(not clean_frame.duplicated(['model','scene_id','mapping']).any(),'Duplicate clean input.')
require(not patch_frame.duplicated(['model','scene_id','mapping','condition_id']).any(),'Duplicate intervention.')
# Exact join keys; no dependence on pandas suffixes or optional exported columns.
dose_export=pd.DataFrame(dose_inventory)
check=patch_frame[patch_frame.kind=='add'].merge(dose_export,
    on=['model','scene_id','mapping','condition_id'],suffixes=('_actual','_planned'),validate='one_to_one')
for field in ('requested_payload_norm','bf16_payload_norm','bf16_payload_over_clean_norm','effective_added_displacement_norm'):
    require(np.allclose(check[field+'_actual'],check[field+'_planned'],atol=1e-10,rtol=1e-9),f'Realized dose mismatch: {field}')
paired_doses=dose_export.pivot(index=['scene_id','mapping','layer','axis','dose','scale'],columns='model',values='bf16_payload_over_clean_norm')
relative=paired_doses.reset_index().query("scale == 'relative'")
require(np.allclose(relative.native,relative.lora,atol=0,rtol=.01),'Relative BF16 dose matching failed.')
self_max=float(patch_frame.loc[patch_frame.kind=='self','margin_shift'].abs().max())
self_score_max=float(patch_frame.loc[patch_frame.kind=='self','centered_score_change'].max())
clamp_residual_max=float(trace_frame.clamp_residual_signed.abs().max())
write_json(ANALYSIS_DIR/'confirmation_control_diagnostics.json',{'self_max_margin_change':self_max,
    'self_max_centered_score_change':self_score_max,'clamp_max_abs_residual':clamp_residual_max})
require(self_max<=SMOKE_TOLERANCE and self_score_max<=SMOKE_TOLERANCE,'Confirmation self-clamp control failed; inspect saved diagnostics.')
require(clamp_residual_max<=CLAMP_TOLERANCE,'Confirmation clamp residual exceeded fixed tolerance; inspect saved diagnostics.')
# Native absolute and relative payloads are mathematically identical; verify repeat forwards.
native_repeat=patch_frame.query("model == 'native' and kind != 'self'").pivot(
    index=['scene_id','mapping','kind','layer','axis','dose','mediator','clamp'],columns='scale',values='margin_shift')
repeat_error=float((native_repeat.absolute-native_repeat.relative).abs().max())
require(repeat_error<=SMOKE_TOLERANCE,'Native identical-dose repeat mismatch.')
for name,frame in [('clean_records',clean_frame),('intervention_records',patch_frame),('signed_state_traces',trace_frame),('specific_block_losses',loss_frame)]:
    frame.to_csv(ANALYSIS_DIR/(name+'.csv'),index=False)
write_json(OUTPUT_ROOT/'source_shard_hashes.json',source_shard_hashes)
write_json(ANALYSIS_DIR/'final_integrity.json',{'passed':True,'clean':len(clean_frame),'interventions':len(patch_frame),
    'self_clamp_max_margin_shift':self_max,'native_repeat_max_error':repeat_error,'relative_payload_norm_rtol':.01})
print('Complete raw records, signed projections and separately verified dose columns saved.')
```

```python
%%cell_progress 16/17 Estimate family-paired changes and interactions
cell_stage('Estimate family-paired changes and interactions')
def estimate_paired(values,key,with_test=False):
    values=np.asarray(values,dtype=float)
    require(values.ndim==1 and len(values)>=2 and np.isfinite(values).all(),'Invalid independent-family effects.')
    rng=np.random.default_rng(int(stable_key(('statistics',key))[:16],16))
    boot=np.empty(BOOTSTRAP_REPLICATES)
    for start in range(0,BOOTSTRAP_REPLICATES,256):
        stop=min(start+256,BOOTSTRAP_REPLICATES)
        boot[start:stop]=values[rng.integers(0,len(values),size=(stop-start,len(values)))].mean(axis=1)
    result={'n_families':len(values),'estimate':float(values.mean()),'ci_low':float(np.quantile(boot,.025)),
        'ci_high':float(np.quantile(boot,.975)),'negative_families':int((values<0).sum())}
    if with_test:
        exceed=0;observed=abs(values.mean())
        for start in range(0,SIGNFLIP_REPLICATES,1024):
            count=min(1024,SIGNFLIP_REPLICATES-start)
            signs=rng.integers(0,2,size=(count,len(values)))*2-1
            exceed+=int((np.abs((signs*values).mean(axis=1))>=observed-1e-12).sum())
        result['p_two_sided']=(exceed+1)/(SIGNFLIP_REPLICATES+1)
    return result

def holm(pvalues):
    pvalues=np.asarray(pvalues,dtype=float);order=np.argsort(pvalues)
    adjusted=np.empty(len(pvalues));running=0.0
    for i,index in enumerate(order):
        running=max(running,(len(pvalues)-i)*pvalues[index]);adjusted[index]=min(1.0,running)
    return adjusted.tolist()

def family_average(frame,keys,value):
    # A family contributes four equally weighted observations: two mirrors x two maps.
    counts=frame.groupby(['family_id',*keys],dropna=False).size()
    require((counts==4).all(),'Family cell must contain both mirrors and both mappings.')
    return frame.groupby(['family_id',*keys],as_index=False,dropna=False)[value].mean()

family_loss=family_average(loss_frame,['model','route','axis','scale','clamp'],'specific_loss')
family_loss.to_csv(ANALYSIS_DIR/'family_specific_losses.csv',index=False)
primary=[];primary_values=[]
for route in ('L48_via_L49','L48_via_L55','L49_via_L55'):
    f=family_loss[(family_loss.route==route)&(family_loss.axis=='native')&(family_loss.scale=='relative')]
    wide=f.pivot(index='family_id',columns=['model','clamp'],values='specific_loss')
    require(len(wide)==N_FAMILIES and not wide.isna().any().any(),'Incomplete primary pairs.')
    changes=wide[('lora','native')]-wide[('native','native')]
    interaction=(wide[('lora','lora')]-wide[('lora','native')])-(wide[('native','lora')]-wide[('native','native')])
    for name,values in [('recipient_change_native_clamp',changes),('model_x_clamp_source',interaction)]:
        key=route+'__'+name
        primary.append({'test':key,'route':route,'contrast':name,**estimate_paired(values,key,True)})
        primary_values.extend({'test':key,'family_id':fid,'value':float(v)} for fid,v in values.items())
for row,p in zip(primary,holm([r['p_two_sided'] for r in primary])):row['p_holm']=p
primary_frame=pd.DataFrame(primary)
primary_frame.to_csv(ANALYSIS_DIR/'primary_six_tests.csv',index=False)
pd.DataFrame(primary_values).to_csv(ANALYSIS_DIR/'primary_family_contrasts.csv',index=False)

# Continuous margin avoids accuracy ceiling as the sole outcome; it does not prove linearity.
adds=patch_frame[patch_frame.kind=='add']
family_add=family_average(adds,['model','layer','axis','scale','dose'],'margin_shift')
family_add.to_csv(ANALYSIS_DIR/'family_additive_effects.csv',index=False)
curves=[]
for keys,frame in family_add.groupby(['model','layer','axis','scale','dose']):
    curves.append(dict(zip(['model','layer','axis','scale','dose'],keys),**estimate_paired(frame.margin_shift,('curve',keys))))
curve_frame=pd.DataFrame(curves)
curve_frame.to_csv(ANALYSIS_DIR/'dose_response_curves.csv',index=False)
secondary=[]
for layer in CAPTURE_LAYERS:
    f=family_add[(family_add.layer==layer)&(family_add.axis=='native')&(family_add.dose==1)]
    w=f.pivot(index='family_id',columns=['model','scale'],values='margin_shift')
    effect=(w[('lora','relative')]-w[('native','relative')])-(w[('lora','absolute')]-w[('native','absolute')])
    secondary.append({'contrast':'dose_matching_modification','layer':layer,**estimate_paired(effect,('dose_modification',layer))})
for layer in (49,55):
    f=family_add[(family_add.layer==layer)&(family_add.scale=='relative')&(family_add.dose==1)]
    w=f.pivot(index='family_id',columns=['model','axis'],values='margin_shift')
    effect=(w[('lora','lora')]-w[('lora','native')])-(w[('native','lora')]-w[('native','native')])
    secondary.append({'contrast':'model_x_injection_source','layer':layer,**estimate_paired(effect,('injection_cross',layer))})
pd.DataFrame(secondary).to_csv(ANALYSIS_DIR/'secondary_dose_and_injection_interactions.csv',index=False)

# Summarize signed downstream transport as well as magnitudes; equal absolute
# projection removal alone does not establish the same displaced state.
projection_changes=[]
transport=trace_frame[(trace_frame.kind=='add')&(trace_frame.dose==1)&(trace_frame.trace_layer>trace_frame.layer)]
for keys,sub in transport.groupby(['layer','trace_layer','axis','scale']):
    for coordinate in MODES:
        field='projection_on_'+coordinate
        family_projection=family_average(sub,['model'],field)
        w=family_projection.pivot(index='family_id',columns='model',values=field)
        projection_changes.append(dict(zip(['input_layer','read_layer','input_axis','scale'],keys),
            coordinate_source=coordinate,native_signed_mean=float(w.native.mean()),lora_signed_mean=float(w.lora.mean()),
            native_mean_abs_projection=float(sub.loc[sub.model=='native',field].abs().mean()),
            lora_mean_abs_projection=float(sub.loc[sub.model=='lora',field].abs().mean()),
            **estimate_paired(w.lora-w.native,('signed_projection',keys,coordinate))))
pd.DataFrame(projection_changes).to_csv(ANALYSIS_DIR/'paired_signed_projection_changes.csv',index=False)

family_behavior=family_average(clean_frame,['model'],'correct')
behavior=family_behavior.pivot(index='family_id',columns='model',values='correct')
behavior_result={'native_accuracy':float(behavior.native.mean()),'lora_accuracy':float(behavior.lora.mean()),
    **estimate_paired(behavior.lora-behavior.native,'clean_behavior')}
write_json(ANALYSIS_DIR/'paired_behavior.json',behavior_result)
# Predeclared limits of generalization: max-units strata and separate answer mappings.
boundary=[]
for field in ('max_units','mapping'):
    for value in sorted(clean_frame[field].unique()):
        for route in ('L48_via_L49','L48_via_L55','L49_via_L55'):
            sub=loss_frame[(loss_frame[field]==value)&(loss_frame.route==route)&(loss_frame.axis=='native')
                &(loss_frame.clamp=='native')&(loss_frame.scale=='relative')]
            w=sub.groupby(['family_id','model']).specific_loss.mean().unstack('model')
            boundary.append({'boundary':field,'value':str(value),'route':route,**estimate_paired(w.lora-w.native,('boundary',field,str(value),route))})
pd.DataFrame(boundary).to_csv(ANALYSIS_DIR/'boundary_route_changes.csv',index=False)
clean_frame.groupby(['max_units','model'],as_index=False).agg(accuracy=('correct','mean'),inputs=('correct','size')).to_csv(ANALYSIS_DIR/'behavior_by_weight_range.csv',index=False)
display(primary_frame)
print('Primary: six direct paired tests, one Holm family. Other intervals are descriptive pointwise 95%.')
```
