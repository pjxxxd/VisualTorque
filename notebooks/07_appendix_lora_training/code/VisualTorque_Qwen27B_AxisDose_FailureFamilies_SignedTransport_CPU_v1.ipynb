```python
import importlib.util, subprocess, sys
if importlib.util.find_spec('tqdm') is None:
    subprocess.check_call([sys.executable,'-m','pip','install','-q','tqdm>=4.66'])
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 15 tracked cells.')
```

```python
%%cell_progress 6/15 Verify actual doses and intervention controls
cell_stage('Verify actual doses and intervention controls')

def compare_tables(current,saved,keys,values,label,atol=1e-10,rtol=1e-9):
    unique(current,keys,label+' current');unique(saved,keys,label+' saved')
    a=current[keys+values].rename(columns={v:v+'__current' for v in values})
    b=saved[keys+values].rename(columns={v:v+'__saved' for v in values})
    joined=a.merge(b,on=keys,validate='one_to_one',how='outer',indicator=True)
    require((joined['_merge']=='both').all(),label+' keys differ.')
    diagnostics=[]
    for value in values:
        x=joined[value+'__current'].to_numpy(dtype=float);y=joined[value+'__saved'].to_numpy(dtype=float)
        ok=np.isfinite(x).all() and np.isfinite(y).all() and np.allclose(x,y,atol=atol,rtol=rtol)
        diagnostics.append({'table':label,'column':value,'max_absolute_error':float(np.max(np.abs(x-y))), 'passed':bool(ok)})
    require(all(r['passed'] for r in diagnostics),label+' numeric values differ: '+str(diagnostics))
    return diagnostics

checks=[]
axis_geometry=read_table('axis_geometry')
unique(axis_geometry,['axis'],'axis geometry');finite(axis_geometry,['native_norm','lora_norm','signed_cosine'],'axis geometry')
require(set(axis_geometry.axis)=={'semantic','A>B','B>A','A>C','C>A','B>C','C>B'},'Incomplete coordinate summary.')
require((axis_geometry.native_norm>0).all() and (axis_geometry.lora_norm>0).all(),'Zero/negative direction norm.')
require(np.allclose(axis_geometry.native_norm,axis_geometry.lora_norm,rtol=1e-5,atol=0),'Cross-axis norm matching failed.')
require((axis_geometry.signed_cosine.abs()<=1.000001).all(),'Invalid coordinate cosine.')
save(axis_geometry,'axis_geometry');show(axis_geometry)
add_records=patch[patch.kind=='add']
checks+=compare_tables(add_records,doses,KEY+['condition_id'],DOSE_FIELDS,'recorded vs planned doses')
for model in MODELS:
    inv=pd.DataFrame([dict(model=r['model'],scene_id=r['scene_id'],mapping=r['mapping'],
        **{'clean_l'+l+'_norm':n for l,n in r['norms'].items()}) for r in identity('clean_inventory.json') if r['model']==model])
    checks+=compare_tables(clean[clean.model==model],inv,KEY,['clean_l48_norm','clean_l49_norm','clean_l55_norm'],'clean norm inventory '+model)
own=clean[KEY+['clean_l48_norm','clean_l49_norm','clean_l55_norm']]
joined=doses.merge(own,on=KEY,validate='many_to_one')
native=clean[clean.model=='native'][['scene_id','mapping','clean_l48_norm','clean_l49_norm','clean_l55_norm']].rename(
    columns={f'clean_l{l}_norm':f'native_l{l}_norm' for l in (48,49,55)})
joined=joined.merge(native,on=['scene_id','mapping'],validate='many_to_one')
for layer in (48,49,55):
    sub=joined[joined.layer==layer]
    require(np.allclose(sub.recipient_clean_norm,sub[f'clean_l{layer}_norm'],atol=1e-10,rtol=1e-9),'Wrong recipient clean norm anchor.')
    require(np.allclose(sub.paired_native_clean_norm,sub[f'native_l{layer}_norm'],atol=1e-10,rtol=1e-9),'Wrong paired Native norm anchor.')
expected=joined.dose.abs()*joined.historical_native_reference_norm*np.where(joined.scale=='relative',
    joined.recipient_clean_norm/joined.paired_native_clean_norm,1.0)
require(np.allclose(expected,joined.requested_payload_norm,atol=1e-10,rtol=1e-9),'Prescribed dose formula mismatch.')
require(np.allclose(joined.bf16_payload_norm/joined.recipient_clean_norm,joined.bf16_payload_over_clean_norm,atol=1e-10,rtol=1e-9),'Relative payload formula mismatch.')
require(((joined.bf16_payload_norm-joined.requested_payload_norm).abs()/joined.requested_payload_norm<.005).all(),'BF16 payload outside source tolerance.')
for layer,field in [(48,'projection_on_native'),(49,'projection_on_native'),(55,'projection_on_native')]:
    finite(traces[traces.trace_layer==layer],['displacement_from_own_clean_norm',field],f'L{layer} traces')
finite(traces[traces.trace_layer.isin([49,55])],['projection_on_lora'],'LoRA-coordinate traces')
source_edits=traces[(traces.kind!='self')&(traces.trace_layer==traces.layer)][KEY+['condition_id','actual_local_edit_norm']]
observed=patch[patch.kind!='self'].merge(source_edits,on=KEY+['condition_id'],validate='one_to_one')
require(np.allclose(observed.actual_local_edit_norm,observed.effective_added_displacement_norm,atol=1e-4,rtol=1e-5),'Actual hook edit and BF16 realized displacement differ.')
clamp_traces=traces[(traces.mediator!=0)&(traces.trace_layer==traces.mediator)]
finite(clamp_traces,['removed_projection_signed','removed_projection_abs','clamp_residual_signed','actual_local_edit_norm'],'clamp traces')
require(np.allclose(clamp_traces.removed_projection_signed.abs(),clamp_traces.removed_projection_abs,atol=1e-10,rtol=1e-9),'Removed absolute/signed projection mismatch.')
control={'self_max_margin_change':float(patch.loc[patch.kind=='self','margin_shift'].abs().max()),
    'self_max_centered_score_change':float(patch.loc[patch.kind=='self','centered_score_change'].max()),
    'clamp_max_abs_residual':float(clamp_traces.clamp_residual_signed.abs().max())}
require(max(control.values())<=.15,'Source technical tolerance exceeded.')
native_repeat=patch[(patch.model=='native')&(patch.kind!='self')].pivot(index=['scene_id','mapping','kind','layer','axis','dose','mediator','clamp'],columns='scale',values='margin_shift')
control['native_repeat_max_error']=float((native_repeat.absolute-native_repeat.relative).abs().max())
require(control['native_repeat_max_error']<=.15,'Identical Native dose repeat differs.')
relative=doses[doses.scale=='relative'].pivot(index=['scene_id','mapping','layer','axis','dose'],columns='model',values='bf16_payload_over_clean_norm')
control['relative_payload_max_fractional_mismatch']=float(((relative.lora-relative.native)/relative.native).abs().max())
require(control['relative_payload_max_fractional_mismatch']<=.01,'Relative BF16 dose mismatch.')
save(pd.DataFrame(checks),'numeric_reconstruction_checks');write_json(OUTPUT_ROOT/'control_checks.json',control)
show(pd.DataFrame([control]))
print('Dose and trace checks passed. Clamp-axis norm has not been substituted for removed dose.')
```

```python
%%cell_progress 7/15 Reconstruct original six primary tests
cell_stage('Reconstruct original six primary tests')

loss_rows=[]
for key,group in tqdm(patch.groupby(KEY,sort=True),desc='Rebuild specific block losses',unit='input'):
    table={tuple(r[k] for k in ['kind','layer','axis','scale','dose','mediator','clamp']):r['margin_shift'] for r in group.to_dict('records')}
    require(len(table)==70,'Ambiguous intervention records.')
    meta={k:group.iloc[0][k] for k in KEY+META}
    for scale in ('absolute','relative'):
        for layer,mediator in [(48,49),(48,55),(49,55)]:
            for source in (('native',) if layer==48 else MODELS):
                for clamp in MODELS:
                    total=table[('add',layer,source,scale,1.0,0,'none')]
                    target=table[('block',layer,source,scale,1.0,mediator,clamp)]
                    random=table[('block',layer,source,scale,1.0,mediator,'random')]
                    self_target=table[('self',mediator,'none','none',0.0,mediator,clamp)]
                    self_random=table[('self',mediator,'none','none',0.0,mediator,'random')]
                    loss_rows.append({**meta,'route':f'L{layer}_via_L{mediator}','axis':source,'scale':scale,'clamp':clamp,
                        'total_effect':total,'target_clamped_response':target,'random_clamped_response':random,
                        'self_target':self_target,'self_random':self_random,
                        'target_block_loss':total-target+self_target,'random_block_loss':total-random+self_random,
                        'specific_loss':random-target+self_target-self_random})
loss=pd.DataFrame(loss_rows)
loss_keys=KEY+['route','axis','scale','clamp']
checks+=compare_tables(loss,saved_loss,loss_keys,COMPONENTS,'specific block losses')
save(loss,'reconstructed_specific_losses')
# These exact original estimators reproduce the prespecified six tests, rather
# than treating this later CPU audit as another independent confirmation.
SEED=int(protocol['seed'])
BOOTSTRAP_REPLICATES=int(protocol['bootstrap'])
SIGNFLIP_REPLICATES=int(protocol['signflip'])
def stable_key(value,seed=None):
    return hashlib.sha256(f'{SEED if seed is None else seed}:{value!r}'.encode()).hexdigest()
def estimate_paired(values,key,with_test=False):
    values=np.asarray(values,dtype=float)
    require(values.ndim==1 and len(values)>=2 and np.isfinite(values).all(),'Invalid independent-family effects.')
    rng=np.random.default_rng(int(stable_key(('statistics',key))[:16],16))
    boot=np.empty(BOOTSTRAP_REPLICATES)
    for start in range(0,BOOTSTRAP_REPLICATES,256):
        stop=min(start+256,BOOTSTRAP_REPLICATES)
        boot[start:stop]=values[rng.integers(0,len(values),size=(stop-start,len(values)))].mean(axis=1)
    result={'n_families':len(values),'estimate':float(values.mean()),'ci_low':float(np.quantile(boot,.025)),
        'ci_high':float(np.quantile(boot,.975)),'negative_families':int((values<0).sum())}
    if with_test:
        exceed=0;observed=abs(values.mean())
        for start in range(0,SIGNFLIP_REPLICATES,1024):
            count=min(1024,SIGNFLIP_REPLICATES-start)
            signs=rng.integers(0,2,size=(count,len(values)))*2-1
            exceed+=int((np.abs((signs*values).mean(axis=1))>=observed-1e-12).sum())
        result['p_two_sided']=(exceed+1)/(SIGNFLIP_REPLICATES+1)
    return result

def holm(pvalues):
    pvalues=np.asarray(pvalues,dtype=float);order=np.argsort(pvalues)
    adjusted=np.empty(len(pvalues));running=0.0
    for i,index in enumerate(order):
        running=max(running,(len(pvalues)-i)*pvalues[index]);adjusted[index]=min(1.0,running)
    return adjusted.tolist()

def family_average(frame,keys,value):
    # A family contributes four equally weighted observations: two mirrors x two maps.
    counts=frame.groupby(['family_id',*keys],dropna=False).size()
    require((counts==4).all(),'Family cell must contain both mirrors and both mappings.')
    return frame.groupby(['family_id',*keys],as_index=False,dropna=False)[value].mean()
family_loss=family_average(loss,['model','route','axis','scale','clamp'],'specific_loss')
checks+=compare_tables(family_loss,read_table('family_specific_losses'),['family_id','model','route','axis','scale','clamp'],['specific_loss'],'family losses')
primary=[];primary_values=[]
for route in ROUTES:
    f=family_loss[(family_loss.route==route)&(family_loss.axis=='native')&(family_loss.scale=='relative')]
    w=f.pivot(index='family_id',columns=['model','clamp'],values='specific_loss')
    require(len(w)==N_FAMILIES and not w.isna().any().any(),'Incomplete primary pairs.')
    changes=w[('lora','native')]-w[('native','native')]
    interaction=(w[('lora','lora')]-w[('lora','native')])-(w[('native','lora')]-w[('native','native')])
    for name,values in [('recipient_change_native_clamp',changes),('model_x_clamp_source',interaction)]:
        key=route+'__'+name
        primary.append({'test':key,'route':route,'contrast':name,**estimate_paired(values,key,True)})
        primary_values.extend({'test':key,'family_id':fid,'value':float(v)} for fid,v in values.items())
for row,p in zip(primary,holm([r['p_two_sided'] for r in primary])):row['p_holm']=p
primary=pd.DataFrame(primary);primary_values=pd.DataFrame(primary_values)
checks+=compare_tables(primary,read_table('primary_six_tests'),['test'],['estimate','ci_low','ci_high','p_two_sided','p_holm','negative_families','n_families'],'six primary tests')
checks+=compare_tables(primary_values,read_table('primary_family_contrasts'),['test','family_id'],['value'],'primary family contrasts')
save(primary,'verified_primary_six_tests');save(primary_values,'verified_primary_family_contrasts')
save(pd.DataFrame(checks),'numeric_reconstruction_checks')
show(primary,['route','contrast','n_families','estimate','ci_low','ci_high','p_holm'])
print('Original six tests reproduced, including their original multiplicity correction.')
```

```python
%%cell_progress 8/15 Identify failed families and inspect exact predictions
cell_stage('Identify failed families and inspect exact predictions')

def summarize_values(values,key):
    x=np.asarray(values,dtype=float)
    require(x.ndim==1 and len(x)>0 and np.isfinite(x).all(),'Invalid family values for '+str(key))
    result={'n_families':len(x),'estimate':float(x.mean()),'ci_low':None,'ci_high':None}
    if len(x)<2:return result
    rng=np.random.default_rng(int(hashlib.sha256(f'{AUDIT_SEED}:{key!r}'.encode()).hexdigest()[:16],16))
    means=np.empty(AUDIT_BOOTSTRAP)
    for start in range(0,AUDIT_BOOTSTRAP,256):
        end=min(start+256,AUDIT_BOOTSTRAP)
        means[start:end]=x[rng.integers(0,len(x),size=(end-start,len(x)))].mean(axis=1)
    result.update(ci_low=float(np.quantile(means,.025)),ci_high=float(np.quantile(means,.975)))
    return result

def paired_model_effect(frame,value,key):
    by_family=frame.groupby(['family_id','model'])[value].mean().unstack('model')
    require(set(by_family)==set(MODELS) and not by_family.isna().any().any(),'Unpaired model families: '+str(key))
    return {'native_mean':float(by_family.native.mean()),'lora_mean':float(by_family.lora.mean()),
        **summarize_values(by_family.lora-by_family.native,key)}

def source_interaction(frame,value,source_column,key):
    w=frame.groupby(['family_id','model',source_column])[value].mean().unstack(['model',source_column])
    require(set(w.columns)==set(itertools.product(MODELS,MODELS)) and not w.isna().any().any(),'Incomplete 2 x 2 interaction: '+str(key))
    native_gain=w[('native','lora')]-w[('native','native')]
    lora_gain=w[('lora','lora')]-w[('lora','native')]
    return {'native_recipient_native_axis':float(w[('native','native')].mean()),
        'native_recipient_lora_axis':float(w[('native','lora')].mean()),
        'lora_recipient_native_axis':float(w[('lora','native')].mean()),
        'lora_recipient_lora_axis':float(w[('lora','lora')].mean()),
        'native_axis_swap_gain':float(native_gain.mean()),'lora_axis_swap_gain':float(lora_gain.mean()),
        **summarize_values(lora_gain-native_gain,key)}

def subsets(frame):
    yield 'all','all',frame
    for field in ['max_units','side','mapping','diagnostic_cell']:
        for value in sorted(frame[field].unique()):yield field,str(value),frame[frame[field]==value]

def paired_factor_difference(frame,value,factor,left,right,key):
    w=frame.groupby(['family_id','model',factor])[value].mean().unstack(['model',factor])
    require(set(w.columns)==set(itertools.product(MODELS,(left,right))),'Unexpected within-family factor levels: '+str(key))
    require(not w.isna().any().any(),'Incomplete paired within-family factor: '+str(key))
    delta_right=w[('lora',right)]-w[('native',right)]
    delta_left=w[('lora',left)]-w[('native',left)]
    return summarize_values(delta_right-delta_left,key)

def independent_group_difference(left,right,key):
    a,b=np.asarray(left,dtype=float),np.asarray(right,dtype=float)
    require(len(a)>0 and len(b)>0 and np.isfinite(a).all() and np.isfinite(b).all(),'Invalid independent groups.')
    result={'n_left_families':len(a),'n_right_families':len(b),'estimate':float(b.mean()-a.mean()),'ci_low':None,'ci_high':None}
    if min(len(a),len(b))<2:return result
    rng=np.random.default_rng(int(hashlib.sha256(f'{AUDIT_SEED}:{key!r}'.encode()).hexdigest()[:16],16))
    differences=np.empty(AUDIT_BOOTSTRAP)
    for start in range(0,AUDIT_BOOTSTRAP,256):
        end=min(start+256,AUDIT_BOOTSTRAP);n=end-start
        differences[start:end]=b[rng.integers(0,len(b),size=(n,len(b)))].mean(1)-a[rng.integers(0,len(a),size=(n,len(a)))].mean(1)
    result.update(ci_low=float(np.quantile(differences,.025)),ci_high=float(np.quantile(differences,.975)))
    return result

def semantic_sign(values):
    values=np.asarray(values,dtype=float)
    return np.where(values>0,'left',np.where(values<0,'right','balanced'))

scene_physics=scenes.copy()
scene_physics['left_torque']=scene_physics.left_units*scene_physics.left_distance
scene_physics['right_torque']=scene_physics.right_units*scene_physics.right_distance
scene_physics['relative_torque_gap']=scene_physics.net_torque.abs()/(scene_physics.left_torque+scene_physics.right_torque)
scene_physics['weight_only_prediction']=semantic_sign(scene_physics.left_units-scene_physics.right_units)
scene_physics['distance_only_prediction']=semantic_sign(scene_physics.left_distance-scene_physics.right_distance)
w,d,t=scene_physics.weight_only_prediction,scene_physics.distance_only_prediction,scene_physics.side
recomputed_cell=np.select([(w!=t)&(d==t),(d!=t)&(w==t),(w==t)&(d==t)],
    ['weight_conflict','distance_conflict','both_cues_aligned'],default='mixed_nonbalanced')
require(np.array_equal(recomputed_cell,scene_physics.diagnostic_cell),'Physical cue group labels do not match geometry.')
for fid,sub in scene_physics.groupby('family_id'):
    require(len(sub)==2,'Expected one mirror pair per family.')
    a,b=sub.iloc[0],sub.iloc[1]
    require((a.left_units,a.left_distance,a.right_units,a.right_distance)==
        (b.right_units,b.right_distance,b.left_units,b.left_distance),'Invalid mirror geometry.')
physics_fields=['scene_id','left_units','left_distance','right_units','right_distance','net_torque',
    'left_torque','right_torque','relative_torque_gap','weight_only_prediction','distance_only_prediction','image_sha256']
behavior_inputs=clean.merge(scene_physics[physics_fields],on='scene_id',validate='many_to_one')
behavior_inputs['agrees_weight_only']=behavior_inputs.predicted_semantic==behavior_inputs.weight_only_prediction
behavior_inputs['agrees_distance_only']=behavior_inputs.predicted_semantic==behavior_inputs.distance_only_prediction
save(behavior_inputs,'all_clean_inputs_with_physics')

INPUT_KEY=['family_id','scene_id','mapping']
case_inputs=clean.loc[clean.model=='native',INPUT_KEY+META[1:]+['correct','predicted_semantic','clean_margin']].rename(
    columns={c:'native_'+c for c in ['correct','predicted_semantic','clean_margin']})
case_inputs=case_inputs.merge(clean.loc[clean.model=='lora',INPUT_KEY+['correct','predicted_semantic','clean_margin']].rename(
    columns={c:'lora_'+c for c in ['correct','predicted_semantic','clean_margin']}),on=INPUT_KEY,validate='one_to_one')
case_inputs=case_inputs.merge(scene_physics[physics_fields],on='scene_id',validate='many_to_one')
case_inputs['transition']=np.select([~case_inputs.native_correct&case_inputs.lora_correct,
    case_inputs.native_correct&~case_inputs.lora_correct,case_inputs.native_correct&case_inputs.lora_correct],
    ['repaired','regressed','both_correct'],default='both_wrong')
label_rows=[]
for item in prompts:
    label_map=dict(zip(item['semantics'],item['labels']))
    label_rows.append({'scene_id':item['scene_id'],'mapping':item['mapping'],
        **{meaning+'_label':label_map[meaning] for meaning in ('left','right','balanced')}})
case_inputs=case_inputs.merge(pd.DataFrame(label_rows),on=['scene_id','mapping'],validate='one_to_one')
for prefix,col in [('gold','side'),('native','native_predicted_semantic'),('lora','lora_predicted_semantic')]:
    case_inputs[prefix+'_label']=[r[r[col]+'_label'] for r in case_inputs.to_dict('records')]
save(case_inputs,'paired_clean_inputs_with_physics')

family_cases=case_inputs.groupby('family_id',sort=True).agg(native_correct=('native_correct','sum'),
    lora_correct=('lora_correct','sum'),n_inputs=('scene_id','size'),native_margin=('native_clean_margin','mean'),
    lora_margin=('lora_clean_margin','mean')).reset_index()
require(family_cases.n_inputs.eq(4).all(),'Incomplete behavior family.')
family_cases['native_wrong']=4-family_cases.native_correct
family_cases['lora_wrong']=4-family_cases.lora_correct
family_cases['accuracy_change']=(family_cases.lora_correct-family_cases.native_correct)/4
# Use the correct-left orientation to give every mirror family a consistent description.
canonical_geometry=scene_physics[scene_physics.side=='left'][['family_id','max_units','diagnostic_cell',
    'left_units','left_distance','right_units','right_distance','left_torque','right_torque','relative_torque_gap']]
family_cases=family_cases.merge(canonical_geometry,on='family_id',validate='one_to_one')
family_cases=family_cases.sort_values(['lora_wrong','family_id'],ascending=[False,True]).reset_index(drop=True)
failed_family_ids=family_cases.loc[family_cases.lora_wrong>0,'family_id'].tolist()
failed_cases=family_cases[family_cases.family_id.isin(failed_family_ids)].copy()
case_inputs['lora_failed_family']=case_inputs.family_id.isin(failed_family_ids)
failure_inputs=case_inputs[case_inputs.lora_failed_family].copy()
require(int(family_cases.lora_wrong.sum())==int((~case_inputs.lora_correct).sum()),'Failure counts disagree.')
save(family_cases,'all_family_behavior_geometry');save(failed_cases,'failed_family_geometry')
save(failure_inputs,'failed_families_all_four_inputs')
save(case_inputs[~case_inputs.lora_correct],'lora_wrong_inputs')
behavior=[]
for scope,value,sub in subsets(clean):
    behavior.append({'scope':scope,'value':value,**paired_model_effect(sub,'correct',('behavior',scope,value))})
behavior=pd.DataFrame(behavior);save(behavior,'behavior_boundaries')
heuristics=[]
for scope,sub in [('all',behavior_inputs),('lora_failed_families',behavior_inputs[behavior_inputs.family_id.isin(failed_family_ids)])]:
    for model,part in sub.groupby('model'):
        for name in ['agrees_weight_only','agrees_distance_only']:
            heuristics.append({'scope':scope,'model':model,'metric':name,'n_inputs':len(part),
                'n_families':part.family_id.nunique(),'fraction':float(part[name].mean()),'descriptive_only':True})
heuristics=pd.DataFrame(heuristics);save(heuristics,'single_cue_compatibility_descriptive')
print(f'LoRA wrong inputs: {int(family_cases.lora_wrong.sum())}; failing physical families: {len(failed_family_ids)} / {N_FAMILIES}')
print('Geometry tables use correct-left orientation. Cue agreement is compatibility, not a recovered algorithm.')
show(failed_cases,limit=200)
show(failure_inputs,['family_id','side','mapping','left_units','left_distance','right_units','right_distance',
    'gold_label','native_predicted_semantic','native_label','lora_predicted_semantic','lora_label',
    'native_clean_margin','lora_clean_margin','transition'],limit=200)
show(behavior);show(heuristics)
```

```python
%%cell_progress 9/15 Build paired input and family transport records
cell_stage('Build paired input and family transport records')

# Every table join names fields explicitly; no dependence on pandas suffixes.
transport_base=traces[(traces.kind=='add')&(traces.axis=='native')&(traces.trace_layer>traces.layer)].copy()
transport_base=transport_base.merge(patch[KEY+['condition_id','margin_shift']],on=KEY+['condition_id'],validate='many_to_one')
transport_base=transport_base.merge(clean[KEY+['correct','predicted_semantic','clean_margin']],on=KEY,validate='many_to_one')
transport_rows=[]
for coordinate in MODELS:
    part=transport_base.drop(columns=['projection_on_native','projection_on_lora']).copy()
    part['coordinate_source']=coordinate
    part['p']=transport_base['projection_on_'+coordinate].to_numpy(dtype=float)
    part['route']=[f'L{a}_via_L{b}' for a,b in zip(part.layer,part.trace_layer)]
    part['positive_mass']=part.p.clip(lower=0)
    part['negative_mass']=(-part.p).clip(lower=0)
    part['absolute_projection']=part.p.abs()
    part['dose_aligned_projection']=part.p*np.sign(part.dose)
    part['negative_fraction']=(part.p<0).astype(float)
    part['exact_zero_fraction']=(part.p==0).astype(float)
    transport_rows.append(part)
transport_inputs=pd.concat(transport_rows,ignore_index=True)
TREATMENT=['route','coordinate_source','scale','dose']
PAIR_KEYS=['family_id','scene_id','mapping']+TREATMENT
unique(transport_inputs,['model']+PAIR_KEYS,'transport inputs')
finite(transport_inputs,['p','positive_mass','negative_mass','displacement_from_own_clean_norm'],'transport decomposition')
require(np.allclose(transport_inputs.p,transport_inputs.positive_mass-transport_inputs.negative_mass),'Signed mass identity failed.')
require(np.allclose(transport_inputs.absolute_projection,transport_inputs.positive_mass+transport_inputs.negative_mass),'Absolute mass identity failed.')
save(transport_inputs,'transport_all_inputs_native_injection')
focus=transport_inputs[(transport_inputs.scale=='relative')&(transport_inputs.dose==1)].copy()
require(len(focus)==N_FAMILIES*4*2*3*2,'Incomplete focus transport rows.')
family_meta=family_cases[['family_id','max_units','diagnostic_cell','lora_wrong','native_wrong','relative_torque_gap']]

DESCRIPTORS=['signed','absolute','positive','negative','negative_fraction','exact_zero_fraction',
    'abs_family_signed','mean_abs_scene_signed','mean_abs_mapping_signed','state_displacement','margin_shift']
def family_descriptors(frame):
    unique(frame,['family_id','model','scene_id','mapping'],'one treatment/subgroup transport')
    grouping=['family_id','model']
    base=frame.groupby(grouping).agg(signed=('p','mean'),absolute=('absolute_projection','mean'),
        positive=('positive_mass','mean'),negative=('negative_mass','mean'),negative_fraction=('negative_fraction','mean'),
        exact_zero_fraction=('exact_zero_fraction','mean'),state_displacement=('displacement_from_own_clean_norm','mean'),
        margin_shift=('margin_shift','mean'))
    base['abs_family_signed']=base.signed.abs()
    base['mean_abs_scene_signed']=frame.groupby(grouping+['scene_id']).p.mean().abs().groupby(grouping).mean()
    base['mean_abs_mapping_signed']=frame.groupby(grouping+['mapping']).p.mean().abs().groupby(grouping).mean()
    require(np.isfinite(base[DESCRIPTORS].to_numpy()).all(),'Invalid family decomposition.')
    return base.reset_index()

family_transport=[]
for keys,sub in transport_inputs.groupby(TREATMENT,sort=True):
    block=family_descriptors(sub)
    for key,value in zip(TREATMENT,keys):block[key]=value
    family_transport.append(block)
family_transport=pd.concat(family_transport,ignore_index=True).merge(family_meta,on='family_id',validate='many_to_one')
save(family_transport,'transport_family_descriptors_all_doses')
print('Saved matched per-input and per-family transport, retaining both coordinate systems and every original Native-axis dose.')
```

```python
%%cell_progress 10/15 Decompose positive negative and cancellation mass
cell_stage('Decompose positive negative and cancellation mass')

def safe_divide(a,b):
    a,b=np.broadcast_arrays(np.asarray(a,dtype=float),np.asarray(b,dtype=float))
    out=np.full(a.shape,np.nan,dtype=float)
    np.divide(a,b,out=out,where=np.abs(b)>1e-12)
    return out

def aggregate_metrics(descriptors):
    # Last dimension contains means of family descriptors; works for bootstrap batches.
    v={k:descriptors[...,i] for i,k in enumerate(DESCRIPTORS)}
    s,a=v['signed'],v['absolute']
    result={k:v[k] for k in ['signed','absolute','positive','negative','negative_fraction',
        'exact_zero_fraction','state_displacement','margin_shift']}
    result.update(cancellation_total=a-np.abs(s),
        cancellation_mapping_first=a-v['mean_abs_scene_signed'],
        cancellation_side_after_mapping=v['mean_abs_scene_signed']-v['abs_family_signed'],
        cancellation_between_families=v['abs_family_signed']-np.abs(s),
        cancellation_side_first=a-v['mean_abs_mapping_signed'],
        cancellation_mapping_after_side=v['mean_abs_mapping_signed']-v['abs_family_signed'],
        signed_coherence=safe_divide(s,a),cancellation_fraction=safe_divide(a-np.abs(s),a))
    return result

def paired_metric_summary(family_frame,key):
    unique(family_frame,['family_id','model'],'paired family decomposition')
    w=family_frame.pivot(index='family_id',columns='model',values=DESCRIPTORS).sort_index()
    require(set(w.columns.get_level_values(1))==set(MODELS) and not w.isna().any().any(),'Unpaired decomposition.')
    arrays={m:w.xs(m,axis=1,level='model')[DESCRIPTORS].to_numpy(dtype=float) for m in MODELS}
    n=len(w);require(n>0,'No families for '+str(key))
    point={m:aggregate_metrics(arrays[m].mean(axis=0)) for m in MODELS}
    boot={m:{metric:[] for metric in point[m]} for m in MODELS}
    if n>=2:
        rng=np.random.default_rng(int(hashlib.sha256(f'{AUDIT_SEED}:decomposition:{key!r}'.encode()).hexdigest()[:16],16))
        for start in range(0,AUDIT_BOOTSTRAP,256):
            count=min(256,AUDIT_BOOTSTRAP-start)
            # Shared family weights preserve Native/LoRA pairing, including nonlinear summaries.
            weights=rng.multinomial(n,np.full(n,1/n),size=count)/n
            for m in MODELS:
                metrics=aggregate_metrics(weights@arrays[m])
                for metric,value in metrics.items():boot[m][metric].append(value)
    result=[]
    for metric in point['native']:
        nv,lv=float(point['native'][metric]),float(point['lora'][metric])
        row={'metric':metric,'n_families':n,'native_mean':nv,'lora_mean':lv,'estimate':lv-nv,
            'ci_low':np.nan,'ci_high':np.nan,'defined_bootstrap_replicates':0,'ci_status':'one_family'}
        if n>=2:
            delta=np.concatenate(boot['lora'][metric])-np.concatenate(boot['native'][metric])
            valid=np.isfinite(delta);row['defined_bootstrap_replicates']=int(valid.sum())
            # Undefined ratios remain undefined; do not condition the interval on nonzero resamples.
            if valid.all() and np.isfinite(row['estimate']):
                row.update(ci_low=float(np.quantile(delta,.025)),ci_high=float(np.quantile(delta,.975)),ci_status='pointwise')
            else:row['ci_status']='undefined_denominator'
        result.append(row)
    return pd.DataFrame(result)

def attach_fields(frame,fields):
    result=frame.copy()
    for k,v in fields.items():result[k]=v
    return result

decomposition=[]
for keys,sub in tqdm(family_transport.groupby(TREATMENT,sort=True),desc='Paired transport decomposition',unit='condition'):
    decomposition.append(attach_fields(paired_metric_summary(sub,keys),dict(zip(TREATMENT,keys))))
decomposition=pd.concat(decomposition,ignore_index=True)
focus_decomposition=decomposition[(decomposition.scale=='relative')&(decomposition.dose==1)].copy()
# These algebra checks are independent of any interpretation or chosen sign threshold.
decomposition_checks=[]
for keys,sub in decomposition.groupby(TREATMENT):
    for m in MODELS:
        v=sub.set_index('metric')[m+'_mean']
        residuals={'signed':v.signed-(v.positive-v.negative),'absolute':v.absolute-(v.positive+v.negative),
            'mapping_first':v.cancellation_total-(v.cancellation_mapping_first+v.cancellation_side_after_mapping+v.cancellation_between_families),
            'side_first':v.cancellation_total-(v.cancellation_side_first+v.cancellation_mapping_after_side+v.cancellation_between_families)}
        require(max(abs(x) for x in residuals.values())<1e-8,'Cancellation reconstruction failed.')
        require(min(v[k] for k in v.index if k.startswith('cancellation_'))>=-1e-8,'Negative cancellation component.')
        decomposition_checks.append({**dict(zip(TREATMENT,keys)),'model':m,**residuals,'passed':True})
save(decomposition,'transport_decomposition_all_doses');save(focus_decomposition,'transport_decomposition_focus')
save(pd.DataFrame(decomposition_checks),'cancellation_identity_checks')
print('All decomposition identities passed. Intervals are exploratory and pointwise.')
show(focus_decomposition[focus_decomposition.metric.isin(['signed','absolute','positive','negative','cancellation_total',
    'cancellation_mapping_first','cancellation_side_after_mapping','cancellation_between_families'])],
    ['route','coordinate_source','metric','native_mean','lora_mean','estimate','ci_low','ci_high'],limit=100)
```

```python
%%cell_progress 12/15 Audit sign flips thresholds and family influence
cell_stage('Audit sign flips thresholds and family influence')

def sign_class(values,epsilon):
    values=np.asarray(values,dtype=float)
    return np.where(values>epsilon,'positive',np.where(values < -epsilon,'negative','near_zero'))

pair_projection=focus.pivot(index=PAIR_KEYS,columns='model',values='p').reset_index()
require(not pair_projection[['native','lora']].isna().any().any(),'Unpaired input projections.')
pair_projection=pair_projection.rename(columns={'native':'native_p','lora':'lora_p'}).merge(
    case_inputs[INPUT_KEY+['side','diagnostic_cell','max_units','transition','lora_failed_family']],on=INPUT_KEY,validate='many_to_one')
sign_rows=[];sign_summaries=[];LOO=[]
for epsilon in SIGN_THRESHOLDS:
    part=pair_projection.copy();part['epsilon']=epsilon
    part['native_sign']=sign_class(part.native_p,epsilon);part['lora_sign']=sign_class(part.lora_p,epsilon)
    part['sign_transition']=part.native_sign+' -> '+part.lora_sign
    sign_rows.append(part)
    for keys,group in part.groupby(['route','coordinate_source']):
        for scope,value,sub in subsets(group):
            for a,b in itertools.product(['positive','near_zero','negative'],repeat=2):
                indicator=((sub.native_sign==a)&(sub.lora_sign==b)).astype(float)
                family_fraction=indicator.groupby(sub.family_id).mean()
                sign_summaries.append({'route':keys[0],'coordinate_source':keys[1],'scope':scope,'value':value,
                    'epsilon':epsilon,'native_sign':a,'lora_sign':b,'n_inputs':len(sub),'count_inputs':int(indicator.sum()),
                    **summarize_values(family_fraction,('sign',epsilon,keys,scope,value,a,b))})
sign_inputs=pd.concat(sign_rows,ignore_index=True);sign_summary=pd.DataFrame(sign_summaries)
save(sign_inputs,'paired_sign_transitions_by_input');save(sign_summary,'paired_sign_transition_fractions')
for keys,sub in family_transport[(family_transport.scale=='relative')&(family_transport.dose==1)].groupby(['route','coordinate_source']):
    w=sub.pivot(index='family_id',columns='model',values=['signed','absolute','positive','negative'])
    for metric in ['signed','absolute','positive','negative']:
        delta=w[(metric,'lora')]-w[(metric,'native')]
        for fid in delta.index:
            remaining=delta.drop(fid)
            LOO.append({'route':keys[0],'coordinate_source':keys[1],'metric':metric,'omitted_family':fid,
                'full_estimate':float(delta.mean()),'leave_one_out_estimate':float(remaining.mean()) if len(remaining) else np.nan})
leave_one_out=pd.DataFrame(LOO);save(leave_one_out,'leave_one_family_out_changes')
show(sign_summary[(sign_summary.scope=='all')&(((sign_summary.native_sign=='positive')&(sign_summary.lora_sign=='negative'))|
    ((sign_summary.native_sign=='negative')&(sign_summary.lora_sign=='positive')))],
    ['route','coordinate_source','epsilon','native_sign','lora_sign','count_inputs','n_inputs','estimate','ci_low','ci_high'],limit=100)
print('Near-zero thresholds are sensitivity choices, not precision guarantees. Leave-one-family-out results are influence diagnostics.')
```
