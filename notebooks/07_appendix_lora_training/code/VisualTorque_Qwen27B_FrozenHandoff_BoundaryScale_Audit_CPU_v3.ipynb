```python
print('[3/11] Configure the read-only secondary audit')
import ast, hashlib, itertools, json, math, platform, shutil, zipfile
from datetime import datetime, timezone

IMPLEMENTATION_SHA256 = 'sha256:db0602cb31ff332cba249a9eb18892ee38695d17e4f27aa3ad42adbd1e7ee33b'
RUN_VERSION='v3'
OUTPUT_ROOT=SOURCE_ROOT.parent/f'qwen27b_frozen_handoff_boundary_scale_audit_{RUN_VERSION}'
ANALYSIS_DIR,FIGURE_DIR=OUTPUT_ROOT/'analysis',OUTPUT_ROOT/'figures'
assert SOURCE_ROOT.resolve()!=OUTPUT_ROOT.resolve()
for directory in (OUTPUT_ROOT,ANALYSIS_DIR,FIGURE_DIR):directory.mkdir(parents=True,exist_ok=True)
MODELS=('native','lora')
MAPS=('forward','reverse')
COHORTS=('both_correct','right_error')
SIDES=('left','right')
KEY=['model','scene_id','mapping_variant']
META=['match_id','cohort','side','unseen_train_dev_geometry']
METRICS=('upstream_l49_specific_loss','upstream_l55_specific_loss','l49_l55_specific_loss')
LABELS={'upstream_l49_specific_loss':'L48 via L49',
        'upstream_l55_specific_loss':'L48 via L55','l49_l55_specific_loss':'L49 via L55'}
INJECTION_LAYERS={'upstream_l49_specific_loss':48,'upstream_l55_specific_loss':48,'l49_l55_specific_loss':49}
CLAMP_LAYERS={'upstream_l49_specific_loss':49,'upstream_l55_specific_loss':55,'l49_l55_specific_loss':55}
CORE=('self_block_l49','self_random_block_l49','self_block_l55','self_random_block_l55',
      'upstream_repair','upstream_repair_block_l49','upstream_repair_random_block_l49',
      'upstream_repair_block_l55','upstream_repair_random_block_l55')
RIGHT=('l49_repair','l49_repair_block_l55','l49_repair_random_block_l55','l55_repair','l55_bypass_after_l49_block')
BOOTSTRAP_REPLICATES=10000
SEED=20260906
# Sanity check for FP32 rescaling versus its nominal target, not a relaxation
# of the saved-CSV reconstruction check. One ULP = one float32 spacing here.
METADATA_NORM_MAX_FP32_ULPS=32
SCALE_CSV_ATOL,SCALE_CSV_RTOL=1e-10,1e-9

def json_default(value):
    if isinstance(value,np.generic):return value.item()
    if isinstance(value,Path):return str(value)
    raise TypeError(type(value).__name__)
def canonical(value):return json.dumps(value,sort_keys=True,ensure_ascii=False,allow_nan=False,default=json_default)
def digest(value):return 'sha256:'+hashlib.sha256(canonical(value).encode()).hexdigest()
from visualtorque_checks import sha256_file
def read_json(path):return json.loads(Path(path).read_text('utf-8'))
def write_json(path,value):
    path=Path(path);temporary=path.with_suffix(path.suffix+'.tmp')
    temporary.write_text(canonical(value)+'\n',encoding='utf-8');temporary.replace(path)
def save_table(frame,name):
    path=ANALYSIS_DIR/name;temporary=path.with_suffix('.csv.tmp')
    frame.to_csv(temporary,index=False);temporary.replace(path)
def safe_source_path(relative):
    relative=Path(relative)
    path=(SOURCE_ROOT/relative).resolve()
    assert not relative.is_absolute() and SOURCE_ROOT.resolve() in path.parents, 'Unsafe source manifest path.'
    return path
def freeze_record(path,identity):
    if path.is_file():
        assert read_json(path)['identity']==identity, 'Inputs/settings changed. Use a new RUN_VERSION; never mix audits.'
    else:write_json(path,{'identity':identity,'created_at_utc':datetime.now(timezone.utc).isoformat()})
    return digest(identity)
def require(condition,message):
    if not bool(condition):raise ValueError(message)
def strict_bool(series):
    converted=series.map(lambda x: x if isinstance(x,(bool,np.bool_)) else
        True if isinstance(x,str) and x.strip().lower()=='true' else
        False if isinstance(x,str) and x.strip().lower()=='false' else None)
    require(converted.notna().all(),f'Invalid boolean column: {series.name}')
    return converted.astype(bool)
def read_csv_checked(name):
    frame=pd.read_csv(SOURCE_ROOT/'analysis'/name,float_precision='round_trip',dtype={key:str for key in
        ('model','match_id','scene_id','mapping_variant','condition','cohort','side')})
    for key in ('model','match_id','scene_id','mapping_variant','condition','cohort','side'):
        if key in frame:
            require(frame[key].notna().all(),f'Missing identifier: {name}/{key}')
            frame[key]=frame[key].astype(str)
    for key in ('correct','patched_correct','newly_correct','newly_wrong','train_geometry_seen',
                'dev_geometry_seen','unseen_train_dev_geometry','evidence_of_change'):
        if key in frame:frame[key]=strict_bool(frame[key])
    return frame
def show(frame,columns=None):
    with pd.option_context('display.max_rows',100,'display.max_columns',20,'display.width',180):
        display((frame if columns is None else frame[columns]).round(5))
```

## 5. Validate raw scores and reconstruct the primary effects


```python
print('[5/11] Validate raw scores and reconstruct the primary effects')
def validate_scores(frame,margin_column,correct_column):
    for _,row in frame.iterrows():
        labels=ast.literal_eval(row.candidate_labels)
        semantics=ast.literal_eval(row.candidate_semantics)
        scores=np.asarray(ast.literal_eval(row.candidate_logprobs),dtype=float)
        require(len(labels)==len(set(labels))==3 and set(semantics)=={'left','balanced','right'},'Invalid answer mapping.')
        require(scores.shape==(3,) and np.isfinite(scores).all(),'Invalid candidate scores.')
        require(row.predicted_semantic==semantics[int(np.argmax(scores))],'Prediction/score mismatch.')
        exact,opposite=row.side,('right' if row.side=='left' else 'left')
        margin=scores[semantics.index(exact)]-scores[semantics.index(opposite)]
        require(abs(float(row[margin_column])-margin)<1e-9,'Saved margin differs from candidate scores.')
        require(bool(row[correct_column])==(row.predicted_semantic==exact),'Saved accuracy differs from prediction.')

def validate_raw_frames(clean,patch,overlap):
    require(len(clean)==16*N_FAMILIES and len(patch)==184*N_FAMILIES,'Incomplete raw tables.')
    require(not clean.duplicated(KEY).any() and not patch.duplicated(KEY+['condition']).any(),'Duplicate input or condition.')
    require(set(clean.model)==set(patch.model)==set(MODELS),'Missing/unexpected model.')
    require(set(clean.mapping_variant)==set(patch.mapping_variant)==set(MAPS),'Missing/unexpected mapping.')
    require(clean.match_id.nunique()==N_FAMILIES and clean.scene_id.nunique()==4*N_FAMILIES,'Wrong family/scene counts.')
    require(not overlap.duplicated('scene_id').any() and len(overlap)==4*N_FAMILIES,'Invalid overlap table.')
    require(set(overlap.scene_id)==set(clean.scene_id),'Overlap/raw scene mismatch.')
    family_ids=set(clean.match_id)
    require(set(unseen_ids)<=family_ids,'Unseen list contains an unknown family.')
    expected=set(itertools.product(MODELS,COHORTS,SIDES,MAPS))
    for family,g in clean.groupby('match_id'):
        require(set(g[['model','cohort','side','mapping_variant']].itertuples(index=False,name=None))==expected,
                f'Unbalanced family: {family}')
    for scene,g in clean.groupby('scene_id'):
        require(all(g[k].nunique()==1 for k in META),f'Input metadata changes across model/mapping: {scene}')
    merged=clean.merge(overlap[['scene_id','match_id','train_geometry_seen','dev_geometry_seen','unseen_train_dev_geometry']],
        on='scene_id',suffixes=('','_audit'),validate='many_to_one')
    for key in ('match_id','train_geometry_seen','dev_geometry_seen','unseen_train_dev_geometry'):
        require((merged[key]==merged[key+'_audit']).all(),f'Geometry metadata mismatch: {key}')
    require((clean.unseen_train_dev_geometry==clean.match_id.isin(unseen_ids)).all(),'Frozen unseen assignments changed.')
    require((clean.unseen_train_dev_geometry==~(clean.train_geometry_seen|clean.dev_geometry_seen)).all(),'Invalid unseen mask.')
    joined=patch.merge(clean[KEY+META+['clean_margin','correct']],on=KEY,suffixes=('','_clean'),
                       how='left',validate='many_to_one',indicator=True)
    require(joined._merge.eq('both').all(),'Patch references an unknown clean input.')
    for key in META:
        require((joined[key]==joined[key+'_clean']).all(),f'Patch metadata mismatch: {key}')
    require(np.allclose(joined.clean_margin,joined.clean_margin_clean,atol=1e-9,rtol=0),'Patch/clean baseline mismatch.')
    for _,g in joined.groupby(KEY):
        expected_conditions=set(CORE+(RIGHT if g.side.iloc[0]=='right' else ()))
        require(set(g.condition)==expected_conditions,'Missing/unexpected intervention condition.')
    require(np.isfinite(clean[['clean_margin','clean_l48_norm','clean_l49_norm','clean_l55_norm']]).all().all(),
            'Nonfinite clean data.')
    require((clean[['clean_l48_norm','clean_l49_norm','clean_l55_norm']]>0).all().all(),'Zero/negative clean norm.')
    require(np.isfinite(patch[['clean_margin','margin_shift']]).all().all() and patch.patched_tokens.eq(1).all(),
            'Invalid intervention score/token scope.')
    require(np.isfinite(patch[['removed_projection_l49','removed_projection_l55']]).all().all(),
            'Missing/nonfinite recorded clamp projection.')
    validate_scores(clean,'clean_margin','correct')
    scored=patch.assign(patched_margin=patch.clean_margin+patch.margin_shift)
    validate_scores(scored,'patched_margin','patched_correct')
    return joined

def derive_path_metrics(w):
    s=w.copy()
    s['upstream_total']=w['upstream_repair']
    for layer in (49,55):
        s[f'upstream_l{layer}_block_loss']=w['upstream_repair']-w[f'upstream_repair_block_l{layer}']+w[f'self_block_l{layer}']
        s[f'upstream_l{layer}_random_loss']=w['upstream_repair']-w[f'upstream_repair_random_block_l{layer}']+w[f'self_random_block_l{layer}']
        s[f'upstream_l{layer}_specific_loss']=s[f'upstream_l{layer}_block_loss']-s[f'upstream_l{layer}_random_loss']
    s['l49_total']=w['l49_repair']
    s['l49_l55_block_loss']=w['l49_repair']-w['l49_repair_block_l55']+w['self_block_l55']
    s['l49_l55_random_loss']=w['l49_repair']-w['l49_repair_random_block_l55']+w['self_random_block_l55']
    s['l49_l55_specific_loss']=s['l49_l55_block_loss']-s['l49_l55_random_loss']
    s['l55_direct']=w['l55_repair']
    s['l55_bypass_gain']=w['l55_bypass_after_l49_block']-w['upstream_repair_block_l49']
    return s
clean=read_csv_checked('clean_records.csv')
patch=read_csv_checked('patch_records.csv')
overlap=read_csv_checked('sft_geometry_overlap.csv')
joined=validate_raw_frames(clean,patch,overlap)
prompt_lookup={(row['scene_id'],row['mapping_variant']):row for row in original_prompts}
require(len(prompt_lookup)==len(original_prompts)==8*N_FAMILIES,'Invalid frozen prompt inventory.')
for row in clean.itertuples():
    frozen=prompt_lookup[(row.scene_id,row.mapping_variant)]
    require(ast.literal_eval(row.candidate_labels)==frozen['labels'] and
            ast.literal_eval(row.candidate_semantics)==frozen['semantics'],'Raw answers differ from the frozen prompt mapping.')
require(((~joined.correct.astype(bool))&joined.patched_correct).equals(joined.newly_correct),'Repair flag mismatch.')
require((joined.correct.astype(bool)&(~joined.patched_correct)).equals(joined.newly_wrong),'Regression flag mismatch.')
index=KEY+META
wide=patch.pivot(index=index,columns='condition',values='margin_shift').reset_index()
paths=derive_path_metrics(wide)
save_table(paths,'recomputed_mapping_path_metrics.csv')
target=paths[(paths.cohort=='right_error')&(paths.side=='right')].copy()
source_primary=read_csv_checked('primary_paired_changes.csv').set_index('metric')
require(set(source_primary.index)==set(METRICS) and len(source_primary)==3,'Unexpected primary result rows.')
reproduced=[]
for metric in METRICS:
    family=target.groupby(['match_id','model'])[metric].mean().unstack('model').sort_index()
    actual={'native_mean':family.native.mean(),'lora_mean':family.lora.mean(),'estimate':(family.lora-family.native).mean()}
    old=source_primary.loc[metric]
    require(int(old.n_families)==N_FAMILIES,'Primary independent unit mismatch.')
    for key,value in actual.items():require(abs(float(old[key])-value)<1e-9,f'Primary result cannot be reconstructed: {metric}/{key}')
    reproduced.append({'metric':metric,**actual,'ci_low':old.ci_low,'ci_high':old.ci_high,
        'original_p_holm':old.p_holm,'n_families':N_FAMILIES})
primary=pd.DataFrame(reproduced)
save_table(primary,'primary_reconstruction_check.csv')
show(primary)
print('All raw input/condition identities, scores, frozen geometry labels and primary estimates verified.')
```

## 6. Audit unseen geometry, answer mappings and behavior


```python
print('[6/11] Audit unseen geometry, answer mappings and behavior')
def generator(key):
    return np.random.default_rng(int(hashlib.sha256(f'{SEED}:{key}'.encode()).hexdigest()[:16],16))
def mean_ci(values,key):
    values=np.asarray(values,dtype=float)
    require(np.isfinite(values).all(),'Nonfinite bootstrap values.')
    n=len(values)
    if n==0:return {'estimate':np.nan,'ci_low':np.nan,'ci_high':np.nan,'n_families':0}
    if n==1:return {'estimate':float(values.mean()),'ci_low':np.nan,'ci_high':np.nan,'n_families':1}
    rng=generator(key)
    means=values[rng.integers(0,n,size=(BOOTSTRAP_REPLICATES,n))].mean(axis=1)
    lo,hi=np.quantile(means,[.025,.975])
    return {'estimate':float(values.mean()),'ci_low':float(lo),'ci_high':float(hi),'n_families':n}
def paired_values(frame,metric):
    # Average ONLY repeated mappings of the same family, cohort and side.
    require(frame.cohort.nunique()<=1 and frame.side.nunique()<=1,'Do not pool cohorts or sides implicitly.')
    p=frame.groupby(['match_id','model'])[metric].mean().unstack('model').reindex(columns=MODELS).sort_index()
    require(p.notna().all().all(),'Unpaired family/model data.')
    return p,p.lora-p.native
def estimate_pair(frame,metric,key):
    p,d=paired_values(frame,metric)
    return {'metric':metric,**mean_ci(d,key),
            'native_mean':float(p.native.mean()),'lora_mean':float(p.lora.mean())}
def subset(frame,scope,mapping='average'):
    require(scope in ('all','unseen_train_dev'),'Unknown scope.')
    require(mapping in ('average',*MAPS),'Unknown mapping.')
    result=frame if scope=='all' else frame[frame.match_id.isin(unseen_ids)]
    return result if mapping=='average' else result[result.mapping_variant==mapping]
def both_mapping_values(frame,metric):
    p=frame.pivot(index='match_id',columns=['model','mapping_variant'],values=metric)
    require(p.notna().all().all() and set(p.columns)==set(itertools.product(MODELS,MAPS)),'Incomplete mapping pairs.')
    return pd.DataFrame({mapping:p[('lora',mapping)]-p[('native',mapping)] for mapping in MAPS})
def correlation(x,y,rank=False):
    x=np.asarray(x,dtype=float);y=np.asarray(y,dtype=float)
    if len(x)<3 or np.std(x)<1e-12 or np.std(y)<1e-12:return np.nan
    if rank:x=pd.Series(x).rank().to_numpy();y=pd.Series(y).rank().to_numpy()
    return float(np.corrcoef(x,y)[0,1])

boundary=[]
for scope in ('all','unseen_train_dev'):
    for mapping in ('average',*MAPS):
        group=subset(target,scope,mapping)
        for metric in METRICS:
            boundary.append({'scope':scope,'mapping':mapping,**estimate_pair(group,metric,('boundary',scope,mapping,metric))})
boundary=pd.DataFrame(boundary)
save_table(boundary,'geometry_mapping_boundary.csv')
show(boundary,['scope','mapping','metric','n_families','native_mean','lora_mean','estimate','ci_low','ci_high'])
mapping_interactions=[]
for scope in ('all','unseen_train_dev'):
    group=subset(target,scope)
    for metric in METRICS:
        if len(group):p=both_mapping_values(group,metric);difference=p.forward-p.reverse
        else:difference=[]
        mapping_interactions.append({'scope':scope,'metric':metric,
            **mean_ci(difference,('mapping-interaction',scope,metric))})
mapping_interactions=pd.DataFrame(mapping_interactions)
save_table(mapping_interactions,'mapping_difference_of_model_changes.csv')
print('Mapping interaction = (LoRA-Native)_forward - (LoRA-Native)_reverse; CI crossing zero is NOT equivalence.')
show(mapping_interactions)
behavior_rows=[]
for scope in ('all','unseen_train_dev'):
    for cohort,side in itertools.product(COHORTS,SIDES):
        group=subset(clean[(clean.cohort==cohort)&(clean.side==side)],scope)
        for metric in ('correct','clean_margin'):
            behavior_rows.append({'scope':scope,'cohort':cohort,'side':side,
                **estimate_pair(group,metric,('behavior',scope,cohort,side,metric))})
behavior_boundary=pd.DataFrame(behavior_rows)
save_table(behavior_boundary,'behavior_geometry_boundary.csv')
show(behavior_boundary)
```

## 7. Decompose target, random and self-clamp responses


```python
print('[7/11] Decompose target, random and self-clamp responses')
# Each metric is an exact signed sum. The upstream total cancels in the
# target-minus-random specific contrast; it must not be called a mediation fraction.
COMPONENTS={
 'upstream_l49_specific_loss':{'clamped_target':('upstream_repair_block_l49',-1),
    'clamped_random':('upstream_repair_random_block_l49',1),'self_target':('self_block_l49',1),'self_random':('self_random_block_l49',-1)},
 'upstream_l55_specific_loss':{'clamped_target':('upstream_repair_block_l55',-1),
    'clamped_random':('upstream_repair_random_block_l55',1),'self_target':('self_block_l55',1),'self_random':('self_random_block_l55',-1)},
 'l49_l55_specific_loss':{'clamped_target':('l49_repair_block_l55',-1),
    'clamped_random':('l49_repair_random_block_l55',1),'self_target':('self_block_l55',1),'self_random':('self_random_block_l55',-1)},
}
decomposition=[]
for scope in ('all','unseen_train_dev'):
    group=subset(target,scope)
    for metric,components in COMPONENTS.items():
        total=np.zeros(len(group))
        for term,(condition,sign) in components.items():
            total+=sign*group[condition].to_numpy()
            signed=group.assign(contribution=sign*group[condition])
            decomposition.append({'scope':scope,'metric':metric,'component':term,'condition':condition,'sign':sign,
                **{k:v for k,v in estimate_pair(signed,'contribution',('components',scope,metric,term)).items() if k!='metric'}})
        require(np.allclose(total,group[metric],rtol=0,atol=1e-9),'Specific loss decomposition identity failed.')
decomposition=pd.DataFrame(decomposition)
save_table(decomposition,'specific_change_decomposition.csv')
show(decomposition[decomposition.scope=='all'],['metric','component','sign','native_mean','lora_mean','estimate','ci_low','ci_high'])
control_metrics=('upstream_total','upstream_l49_block_loss','upstream_l49_random_loss',
    'upstream_l55_block_loss','upstream_l55_random_loss','l49_total','l49_l55_block_loss',
    'l49_l55_random_loss','l55_direct','l55_bypass_gain')
control_rows=[]
for scope in ('all','unseen_train_dev'):
    for metric in control_metrics:
        control_rows.append({'scope':scope,**estimate_pair(subset(target,scope),metric,('control',scope,metric))})
controls=pd.DataFrame(control_rows)
save_table(controls,'total_and_control_responses.csv')
show(controls[controls.scope=='all'])
self_rows=[]
for keys,group in patch[patch.condition.isin(CORE[:4])].groupby(['model','cohort','side','condition']):
    self_rows.append(dict(zip(['model','cohort','side','condition'],keys),n_inputs=len(group),
        n_families=group.match_id.nunique(),mean_shift=group.margin_shift.mean(),
        p95_absolute_shift=group.margin_shift.abs().quantile(.95),max_absolute_shift=group.margin_shift.abs().max()))
self_controls=pd.DataFrame(self_rows)
save_table(self_controls,'self_clamp_numerical_check.csv')
print('Self-clamp deviations (all cohorts):')
show(self_controls.groupby('model',as_index=False).agg(max_absolute_shift=('max_absolute_shift','max')))
# Right error-versus-correct difference in model changes, paired by geometry.
context_interactions=[]
for scope in ('all','unseen_train_dev'):
    group=subset(paths[paths.side=='right'],scope)
    for metric in METRICS:
        if len(group):
            p=group.groupby(['match_id','model','cohort'])[metric].mean().unstack(['model','cohort'])
            d=(p[('lora','right_error')]-p[('native','right_error')])-(p[('lora','both_correct')]-p[('native','both_correct')])
        else:d=[]
        context_interactions.append({'scope':scope,'metric':metric,**mean_ci(d,('context',scope,metric))})
context_interactions=pd.DataFrame(context_interactions)
save_table(context_interactions,'error_vs_correct_context_interaction.csv')
show(context_interactions)
```

## 8. Audit intervention scales and natural margins


```python
print('[8/11] Audit intervention scales and natural margins')
def reconcile_scale_references(clean,patch,saved_scale,direction_meta,protocol):
    # Reproduce the parent CSV's actual formulas, not a newly defined dose.
    nominal={48:float(direction_meta['upstream_norm'])*float(direction_meta['matched_upstream_alpha']),
             49:float(direction_meta['semantic_norm']),55:float(direction_meta['code_norm'])}
    require(all(np.isfinite(x) and x>0 for x in nominal.values()),'Invalid fixed dose metadata.')
    require(abs(direction_meta['matched_upstream_alpha']-protocol['matched_upstream_alpha'])<1e-12,
            'Direction/protocol dose mismatch.')
    require('l49_axis_norm' in patch,'Missing v3 patch field l49_axis_norm; restore the complete v3 patch_records.csv.')
    norm_columns=[f'clean_l{layer}_norm' for layer in (48,49,55)]
    ratio_columns=[f'l{layer}_payload_over_clean_norm' for layer in (48,49,55)]
    # Match the parent's actual scale_frame export. Geometry membership was NOT
    # saved in that CSV; the authoritative clean labels were verified in stage 5.
    required_saved=KEY+['match_id','cohort','side']+norm_columns+ratio_columns
    require(set(required_saved)<=set(saved_scale.columns),
            'Saved scale table is missing required v3 columns: '+str(sorted(set(required_saved)-set(saved_scale.columns))))
    require(not saved_scale.duplicated(KEY).any() and len(saved_scale)==len(clean),'Invalid saved scale table.')
    columns=KEY+META+['clean_margin']+norm_columns
    shared_metadata=[key for key in META if key in saved_scale.columns]
    saved_columns=KEY+shared_metadata+norm_columns+ratio_columns
    # Explicit renaming avoids assuming merge suffixes exist for left-only data.
    right=saved_scale[saved_columns].rename(columns={key:key+'_saved' for key in shared_metadata+norm_columns})
    check=clean[columns].merge(right,on=KEY,how='outer',validate='one_to_one',indicator=True)
    require(check._merge.eq('both').all(),'Saved scale table has different inputs.')
    for key in shared_metadata:
        require((check[key]==check[key+'_saved']).all(),f'Saved scale metadata differ: {key}')

    # These two conditions use the unnormalized SEMANTIC_DIRECTION (up to sign).
    # Exclude random axes and action='add' records, whose axis-norm field is zero.
    semantic_conditions=('self_block_l49','upstream_repair_block_l49')
    anchors=patch[patch.condition.isin(semantic_conditions)]
    require(len(anchors)==2*len(clean) and not anchors.duplicated(KEY+['condition']).any(),
            'Missing/duplicate L49 target-axis norm anchors.')
    require(set(anchors[KEY].itertuples(index=False,name=None))==set(clean[KEY].itertuples(index=False,name=None)),
            'L49 norm anchors refer to different model/scene/mapping inputs.')
    measured=anchors.l49_axis_norm.to_numpy(dtype=float)
    require(np.isfinite(measured).all() and (measured>0).all(),'Invalid recorded L49 target-axis norm.')
    semantic_norm=float(np.median(measured))
    spacing=float(np.spacing(np.float32(nominal[49])))
    nominal_tolerance=METADATA_NORM_MAX_FP32_ULPS*spacing
    semantic_consistent=bool(np.allclose(measured,semantic_norm,atol=1e-12,rtol=0))
    semantic_near_nominal=bool(abs(semantic_norm-nominal[49])<=nominal_tolerance)
    payload_norms={**nominal,49:semantic_norm}
    methods={48:'direction_library.upstream_norm * matched_upstream_alpha',
             49:'patch_records.l49_axis_norm on fixed semantic target clamps',
             55:'direction_library.code_norm (parent CSV common nominal code dose)'}
    diagnostics=[]
    for layer,dose in payload_norms.items():
        clean_norm=check[f'clean_l{layer}_norm'].to_numpy(dtype=float)
        saved_norm=check[f'clean_l{layer}_norm_saved'].to_numpy(dtype=float)
        ratio=check[f'l{layer}_payload_over_clean_norm'].to_numpy(dtype=float)
        require(np.isfinite(clean_norm).all() and (clean_norm>0).all() and
                np.isfinite(saved_norm).all() and (saved_norm>0).all() and
                np.isfinite(ratio).all() and (ratio>0).all(),f'Invalid relative-scale data at L{layer}.')
        expected=dose/clean_norm
        implied=ratio*clean_norm  # Diagnostic only; not used to choose the dose.
        norms_match=bool(np.allclose(clean_norm,saved_norm,atol=SCALE_CSV_ATOL,rtol=SCALE_CSV_RTOL))
        ratios_match=bool(np.allclose(expected,ratio,atol=SCALE_CSV_ATOL,rtol=SCALE_CSV_RTOL))
        anchors_match=semantic_consistent if layer==49 else True
        nominal_match=semantic_near_nominal if layer==49 else True
        diagnostics.append({'layer':layer,'norm_source':methods[layer],
            'nominal_reference_norm':nominal[layer],'reconstruction_reference_norm':dose,
            'nominal_absolute_difference':abs(dose-nominal[layer]),
            'nominal_fp32_ulp_difference':abs(dose-nominal[layer])/spacing if layer==49 else 0.,
            'recorded_l49_norm_min':float(measured.min()) if layer==49 else None,
            'recorded_l49_norm_max':float(measured.max()) if layer==49 else None,
            'csv_implied_numerator_min':float(implied.min()),'csv_implied_numerator_max':float(implied.max()),
            'max_csv_absolute_error':float(np.max(np.abs(expected-ratio))),
            'saved_geometry_label_present':'unseen_train_dev_geometry' in saved_scale.columns,
            'clean_norms_match':norms_match,'csv_ratios_match':ratios_match,
            'fixed_axis_anchors_agree':anchors_match,'nominal_norm_compatible':nominal_match,
            'passed':norms_match and ratios_match and anchors_match and nominal_match})
    return payload_norms,pd.DataFrame(diagnostics)

scale=clean[KEY+META+['clean_margin','clean_l48_norm','clean_l49_norm','clean_l55_norm']].copy()
saved_scale=read_csv_checked('intervention_relative_scale.csv')
payload_norms,scale_reference_diagnostics=reconcile_scale_references(
    clean,patch,saved_scale,direction_meta,original_protocol)
# Save detailed evidence before stopping on an actual inconsistency.
save_table(scale_reference_diagnostics,'scale_reference_norm_diagnostics.csv')
write_json(OUTPUT_ROOT/'scale_reference_norm_diagnostics.json',{
    'source_capture':capture_digest,'csv_atol':SCALE_CSV_ATOL,'csv_rtol':SCALE_CSV_RTOL,
    'metadata_norm_max_fp32_ulps':METADATA_NORM_MAX_FP32_ULPS,
    'rows':scale_reference_diagnostics.astype(object).where(scale_reference_diagnostics.notna(),None).to_dict(orient='records')})
show(scale_reference_diagnostics,['layer','nominal_reference_norm','reconstruction_reference_norm',
    'nominal_absolute_difference','nominal_fp32_ulp_difference','max_csv_absolute_error','passed'])
failed=scale_reference_diagnostics[~scale_reference_diagnostics.passed]
require(failed.empty,'Scale references remain inconsistent at layers '+str(failed.layer.tolist())+
    '. See scale_reference_norm_diagnostics.json and analysis/scale_reference_norm_diagnostics.csv. '
    'The nominal L49 target is not substituted for the measured direction norm.')
for layer,dose in payload_norms.items():
    scale[f'l{layer}_payload_over_clean_norm']=dose/scale[f'clean_l{layer}_norm']
save_table(scale,'recomputed_intervention_relative_scale.csv')
scale_rows=[]
for scope in ('all','unseen_train_dev'):
    group=subset(scale[(scale.cohort=='right_error')&(scale.side=='right')],scope)
    for layer,dose in payload_norms.items():
        for metric in (f'clean_l{layer}_norm',f'l{layer}_payload_over_clean_norm'):
            scale_rows.append({'scope':scope,'layer':layer,'fixed_payload_norm':dose,
                **estimate_pair(group,metric,('scale',scope,metric))})
scale_summary=pd.DataFrame(scale_rows)
save_table(scale_summary,'relative_scale_paired_changes.csv')
show(scale_summary)
# clamp_selected_coordinate normalizes the axis. Its pre-rounding correction
# norm is |coefficient|, not the norm of the axis originally passed to the hook.
clamp_measurements=[]
for metric,layer in CLAMP_LAYERS.items():
    norm_column=f'clean_l{layer}_norm'
    for component in ('clamped_target','clamped_random'):
        condition=COMPONENTS[metric][component][0]
        records=patch[(patch.cohort=='right_error')&(patch.side=='right')&(patch.condition==condition)].copy()
        records=records.merge(clean[KEY+[norm_column]],on=KEY,validate='one_to_one')
        records['absolute_removed_projection']=records[f'removed_projection_l{layer}'].abs()
        records['removed_projection_over_clean_norm']=records.absolute_removed_projection/records[norm_column]
        for scope in ('all','unseen_train_dev'):
            for measure in ('absolute_removed_projection','removed_projection_over_clean_norm'):
                summary=estimate_pair(subset(records,scope),measure,('clamp-size',scope,metric,component,measure))
                clamp_measurements.append({'scope':scope,'metric':metric,'layer':layer,'component':component,
                    'measure':measure,**{k:v for k,v in summary.items() if k!='metric'}})
clamp_measurements=pd.DataFrame(clamp_measurements)
save_table(clamp_measurements,'actual_clamp_projection_changes.csv')
print('Actual removed projection magnitude (before BF16 rounding), not the axis norm:')
show(clamp_measurements[(clamp_measurements.scope=='all')&(clamp_measurements.measure=='absolute_removed_projection')])
# Descriptive association only. No post-treatment regression is used to declare
# the causal response change "explained" by margin or activation scale.
family_diagnostics=pd.DataFrame(index=sorted(clean.match_id.unique()))
family_diagnostics.index.name='match_id'
primary_clean=clean[(clean.cohort=='right_error')&(clean.side=='right')]
margin_pair,margin_delta=paired_values(primary_clean,'clean_margin')
family_diagnostics['native_clean_margin']=margin_pair.native
family_diagnostics['lora_clean_margin']=margin_pair.lora
family_diagnostics['delta_clean_margin']=margin_delta
for layer in payload_norms:
    # Average log ratios over the two mappings within each family.
    p=primary_clean.pivot(index=['match_id','mapping_variant'],columns='model',values=f'clean_l{layer}_norm')
    family_diagnostics[f'log_relative_dose_ratio_l{layer}']=np.log(p.native/p.lora).groupby('match_id').mean()
for metric in METRICS:
    _,difference=paired_values(target,metric)
    family_diagnostics[metric]=difference
family_diagnostics['unseen_train_dev_geometry']=family_diagnostics.index.isin(unseen_ids)
save_table(family_diagnostics.reset_index(),'family_scale_margin_effect_diagnostics.csv')
associations=[]
for scope in ('all','unseen_train_dev'):
    frame=family_diagnostics if scope=='all' else family_diagnostics[family_diagnostics.unseen_train_dev_geometry]
    for metric in METRICS:
        for covariate in ('delta_clean_margin','log_relative_dose_ratio_l48','log_relative_dose_ratio_l49','log_relative_dose_ratio_l55'):
            associations.append({'scope':scope,'metric':metric,'covariate':covariate,'n_families':len(frame),
                'pearson':correlation(frame[covariate],frame[metric]),'spearman':correlation(frame[covariate],frame[metric],True)})
associations=pd.DataFrame(associations)
save_table(associations,'descriptive_scale_margin_associations.csv')
show(associations[associations.scope=='all'])
print('Log relative dose ratio < 0 means an ADDITIVE payload / clean norm is smaller after LoRA.')
print('Clamp axes are normalized: changing their scalar norm is not a dose change. Removal depends on the induced displacement.')
print('These associations do not identify mediation, dose equivalence, saturation, or coordinate rotation.')
```

## 9. Check sensitivity to individual physical families


```python
print('[9/11] Check sensitivity to individual physical families')
sensitivity=[]
for scope in ('all','unseen_train_dev'):
    for metric in METRICS:
        _,d=paired_values(subset(target,scope),metric)
        values=d.to_numpy();n=len(values)
        leave_one=(values.sum()-values)/(n-1) if n>1 else np.array([])
        sensitivity.append({'scope':scope,'metric':metric,'n_families':n,
            'mean_change':float(values.mean()) if n else np.nan,
            'median_change':float(np.median(values)) if n else np.nan,
            'n_negative':int((values<0).sum()),'n_zero':int((values==0).sum()),'n_positive':int((values>0).sum()),
            'leave_one_out_min_mean':float(leave_one.min()) if n>1 else np.nan,
            'leave_one_out_max_mean':float(leave_one.max()) if n>1 else np.nan})
sensitivity=pd.DataFrame(sensitivity)
save_table(sensitivity,'family_influence_sensitivity.csv')
show(sensitivity)
print('Leave-one-family-out is an influence diagnostic, not another independent replication or p-value.')
```
