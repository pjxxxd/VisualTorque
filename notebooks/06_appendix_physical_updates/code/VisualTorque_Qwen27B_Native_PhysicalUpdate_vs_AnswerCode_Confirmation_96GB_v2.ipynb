```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 14 tracked cells.')
```

```python
%%cell_progress 4/14 Select disjoint physical families and render
cell_stage('Select disjoint physical families and render')
def physical_key(values):
    x=tuple(map(int,values));return min(x,x[2:]+x[:2])
def outcome(values):
    u,d,v,e=values;t=u*d-v*e
    return 'left' if t>0 else 'right' if t<0 else 'balanced'
def collect_geometries(obj):
    found=set()
    def visit(x):
        if isinstance(x,dict):
            keys=('left_units','left_distance','right_units','right_distance')
            if all(k in x for k in keys):found.add(physical_key([x[k] for k in keys]))
            stacks=x.get('stacks')
            if isinstance(stacks,list) and len(stacks)==2 and all(isinstance(s,dict) and set(('units','distance','side'))<=set(s) for s in stacks):
                by={s['side']:s for s in stacks}
                if set(by)=={'left','right'}:found.add(physical_key([by['left']['units'],by['left']['distance'],by['right']['units'],by['right']['distance']]))
            for v in x.values():visit(v)
        elif isinstance(x,list):
            for v in x:visit(v)
    visit(obj);return found

def family_geometries(family):return {physical_key(v) for v in family['values'].values()}

def select_families(excluded,n):
    # Solely physical enumeration: all before/after truths and disjointness are known before inference.
    neutral={u:[] for u in range(2,MAX_UNITS-DELTA_UNITS+1)};targets=[]
    for u in range(2,MAX_UNITS-DELTA_UNITS+1):
        for d,v,e in itertools.product(range(1,7),range(1,MAX_UNITS+1),range(1,7)):
            before=(u,d,v,e);after=(u+DELTA_UNITS,d,v,e)
            if any(physical_key(x) in excluded for x in (before,after)):continue
            if outcome(before)==outcome(after)=='right':neutral[u].append((before,after))
        for v,e,df,ds in itertools.product(range(1,MAX_UNITS+1),range(1,7),range(2,7),range(1,6)):
            if ds>=df:continue
            vals={'flip_before':(u,df,v,e),'flip_after':(u+DELTA_UNITS,df,v,e),
                  'stable_before':(u,ds,v,e),'stable_after':(u+DELTA_UNITS,ds,v,e)}
            if [outcome(vals[k]) for k in ('flip_before','flip_after','stable_before','stable_after')]!=['right','left','right','right']:continue
            keys={physical_key(x) for x in vals.values()}
            if len(keys)==4 and not keys&excluded:targets.append(vals)
    targets.sort(key=stable_key)
    # Preserve scarce recipient configurations when assigning neutral donors; this is a geometry-only packing rule.
    target_uses=Counter(physical_key(v) for vals in targets for v in vals.values())
    used=set(excluded);selected=[]
    for vals in targets:
        keys={physical_key(x) for x in vals.values()}
        if keys&used:continue
        u=vals['flip_before'][0];source=[];reserved=keys.copy()
        pool=sorted(neutral[u],key=lambda x:(sum(target_uses[physical_key(v)] for v in x),stable_key([vals,x])))
        for pair in pool:
            new={physical_key(x) for x in pair}
            if new&(used|reserved):continue
            source.append(pair);reserved|=new
            if len(source)==2:break
        if len(source)!=2:continue
        full=dict(vals)
        for i,pair in enumerate(source):
            full[f'source{i}_before'],full[f'source{i}_after']=pair
        fid='physical_update_'+stable_key(full)[:18]
        selected.append({'family_id':fid,'values':full,'before_units':u,'delta_units':DELTA_UNITS})
        used|=reserved
        if len(selected)==n:break
    require(len(selected)==n,f'Only {len(selected)} disjoint families available, requested {n}. No automatic relaxation. Inspect the exposure census and predeclare a new protocol if needed.')
    return selected,{'eligible_receiver_contexts':len(targets),'neutral_pairs_by_units':{str(k):len(v) for k,v in neutral.items()},'excluded_geometries':len(excluded)}

exposure=set();exposure_audit=[]
for path in EXPOSURE_MANIFESTS:
    contents=([json.loads(s) for s in path.read_text(encoding='utf-8').splitlines() if s.strip()] if path.suffix=='.jsonl' else read_json(path))
    require(bool(contents),'Empty historical geometry manifest: '+str(path))
    keys=collect_geometries(contents)
    # Depth support and the old matched cohort may contain only multi-stack scenes, which cannot duplicate this single-stack design.
    if path.name not in ('depth_support.jsonl','frozen_scene_manifest.jsonl'):
        require(keys,'No single-stack geometries parsed from '+str(path))
    exposure|=keys;exposure_audit.append({'path':str(path.relative_to(PROJECT_ROOT)),'sha256':sha(path),'single_stack_geometries':len(keys)})
families,pool_audit=select_families(exposure,N_FAMILIES)
freeze(SHARED_ROOT/'geometry_selection.json',{'families':families,'pool':pool_audit,'exposure_audit':exposure_audit,'excluded_keys':sorted(exposure)})
scene_rows=[]
for f in tqdm(families,desc='Render eight scenes and their mirrors',unit='family'):
    for role,values in f['values'].items():
        u,d,v,e=values;sid=f['family_id']+'__'+role
        ap=SHARED_ROOT/'selected_images'/(sid+'__anchor.png');mp=SHARED_ROOT/'selected_images'/(sid+'__mirror.png')
        # Render into a new temporary file; validate immutable outputs before replacing anything.
        temp=ap.with_suffix('.render.png')
        render_scene(Scene(sid,(Stack('L',Side.LEFT,d,u),Stack('R',Side.RIGHT,e,v))),temp,RenderStyle(**STYLE))
        if ap.is_file():require(sha(ap)==sha(temp),'Frozen image changed: '+ap.name);temp.unlink()
        else:temp.replace(ap)
        temp=mp.with_suffix('.render.png')
        with Image.open(ap) as im:ImageOps.mirror(im.convert('RGB')).save(temp)
        if mp.is_file():require(sha(mp)==sha(temp),'Frozen mirror changed.');temp.unlink()
        else:temp.replace(mp)
        for orientation,path,vs in [('anchor',ap,values),('mirror',mp,values[2:]+values[:2])]:
            scene_rows.append({'family_id':f['family_id'],'scene_id':sid+'__'+orientation,'role':role,'orientation':orientation,
                'values':list(vs),'outcome':outcome(vs),'changed_side':'left' if orientation=='anchor' else 'right',
                'image_name':path.name,'image_sha256':sha(path)})
require(len(scene_rows)==N_FAMILIES*16,'Wrong geometry grid.')
require(len(set().union(*(family_geometries(f) for f in families)))==N_FAMILIES*8,'Cross-family geometry overlap.')
freeze(SHARED_ROOT/'scene_manifest.json',scene_rows)
SCENES={r['scene_id']:r for r in scene_rows}
SCENE_LOOKUP={(r['family_id'],r['orientation'],r['role']):r for r in scene_rows}
print('Frozen geometry:',N_FAMILIES,'families,',len(scene_rows),'rendered scenes; pool:',pool_audit)
display(pd.DataFrame([{'family':f['family_id'],'source0':f['values']['source0_before'],'source1':f['values']['source1_before'],
                      'flip_before':f['values']['flip_before'],'stable_before':f['values']['stable_before']} for f in families]))
```

```python
%%cell_progress 5/14 Freeze neutral-source prompts and confirmatory protocol
cell_stage('Freeze neutral-source prompts and confirmatory protocol')
RULE=('Each square is one unit of weight. Distance is measured in tick intervals from the center pivot. '
      'Torque equals weight times distance. Total torque on a side is the sum over stacks on that side.')
SEMANTIC_TEXT={'left':'left side','balanced':'equal','right':'right side'}
def prompt_for(mapping):
    return '\n\n'.join(['Answer this question using only the supplied balance-scale scene.',RULE,
        'Question: Which way will the balance scale tip?',
        'Options:\n'+'\n'.join(f'{label}. {SEMANTIC_TEXT[semantic]}' for label,semantic in zip(LABELS,mapping)),
        'Do not explain your reasoning. Reply with exactly one option letter and nothing else. Valid replies are: A, B, C.'])
def source_mappings(fid,orientation,context):
    # Share this independently assigned map pair across source backgrounds so background-only contrasts preserve the prompt.
    first=int(stable_key([fid,orientation,'source_mapping'])[:8],16)%6
    p=PERMUTATIONS[first];rotated=p[1:]+p[:1]
    return (first,PERMUTATIONS.index(rotated))
def make_example(row,mapping_id):
    mapping=PERMUTATIONS[mapping_id];u,d,v,e=row['values'];gold=row['outcome']
    return BehaviorExample(example_id=row['scene_id']+f'__m{mapping_id}',question_id=row['scene_id'],scene_id=row['scene_id'],
        condition=PromptCondition.IMAGE_RULE,question_kind='outcome',variant=f'physical_update_m{mapping_id}',
        split='frozen_confirmation',problem_type='single_stack',outcome=gold,net_torque=u*d-v*e,
        prompt=prompt_for(mapping),image_path=SHARED_ROOT/'selected_images'/row['image_name'],labels=LABELS,
        option_text=tuple(SEMANTIC_TEXT[s] for s in mapping),option_semantics=mapping,
        gold_label=LABELS[mapping.index(gold)],gold_semantic=gold)
def capture_key(row,mapping_id):return row['scene_id']+f'__m{mapping_id}'
capture_inputs=[];prompt_inventory=[]
for row in scene_rows:
    ids=(source_mappings(row['family_id'],row['orientation'],int(row['role'][6])) if row['role'].startswith('source') else range(6))
    for mid in ids:
        e=make_example(row,mid);capture_inputs.append((row,mid))
        prompt_inventory.append({'capture_key':capture_key(row,mid),'scene_id':row['scene_id'],'mapping_id':mid,
            'prompt':e.prompt,'semantics':list(e.option_semantics),'labels':list(LABELS),'image_sha256':row['image_sha256']})
require(len(capture_inputs)==N_FAMILIES*64,'Wrong clean grid.')
require(len({capture_key(r,m) for r,m in capture_inputs})==len(capture_inputs),'Duplicate clean input.')
PROMPT_DIGEST=freeze(SHARED_ROOT/'prompt_inventory.json',prompt_inventory)
PLAN=[{'condition_id':'zero','kind':'zero','context':-1,'source_mapping_slot':-1,'alpha':0.0}]
for context,slot in itertools.product(range(2),range(2)):
    for kind,alpha in [('physical',1.),('physical_half',.5),('reverse',-1.),('random',1.),('background',1.),('mapping',1.)]:
        PLAN.append({'condition_id':f'{kind}__s{context}__m{slot}','kind':kind,'context':context,'source_mapping_slot':slot,'alpha':alpha})
TARGET_INPUTS=[(r,m) for r,m in capture_inputs if r['role'] in ('flip_before','stable_before')]
require(len(PLAN)==25 and len(TARGET_INPUTS)==N_FAMILIES*24,'Wrong patch plan.')
PROTOCOL={'implementation':IMPLEMENTATION,'schema':'answer-neutral-update-1.0','model_id':MODEL_ID,'revision':MODEL_REVISION,
    'seed':SEED,'families':families,'n_families':N_FAMILIES,'max_units':MAX_UNITS,'delta_units':DELTA_UNITS,'style':STYLE,'exposure_hashes':EXPOSURE_HASHES,
    'toolkit_hashes':TOOLKIT_HASHES,'prompt_digest':PROMPT_DIGEST,'capture_layers':CAPTURE_LAYERS,'injection_layer':INJECTION_LAYER,
    'token':'final_prompt_token','plan':PLAN,'primary_alpha':1.,'primary_tests':['physical_vs_clean','physical_vs_random','physical_vs_background','physical_vs_mapping'],
    'primary_scope':'all families; flip contexts; mean over paired mirrors, donor contexts/maps, six receiver mappings',
    'primary_metric':'counterfactual-side minus original-side candidate logprob gain','multiplicity':'Holm four two-sided paired family sign-flip tests',
    'resampling':'mirror-equivalent physical family','bootstraps':BOOTSTRAPS,'signflips':SIGNFLIPS,
    'stable_retention_ci_lower_bound':-STABLE_LOSS_BOUND,'target_answer_used_in_physical_payload':False,
    'behavior_selection':False,'numerics':NATIVE_NUMERICS_POLICY,'synthetic_only':SYNTHETIC_ONLY}
PROTOCOL_DIGEST=freeze(SHARED_ROOT/'frozen_protocol.json',PROTOCOL)
if MODEL_MODE=='lora':require(read_json(SHARED_ROOT/'native/RUN_COMPLETE.json')['protocol_digest']==PROTOCOL_DIGEST,'Native protocol mismatch.')
freeze(OUTPUT_ROOT/'model_identity.json',MODE_IDENTITY)
print('Protocol frozen:',PROTOCOL_DIGEST,'| clean:',len(capture_inputs),'| interventions:',len(TARGET_INPUTS)*len(PLAN))
```

```python
%%cell_progress 7/14 Define exact scoring and cached state capture
cell_stage('Define exact scoring and cached state capture')

# The v1 IMPLEMENTATION remains the frozen experiment identity, not the v2 code hash.
INJECTION_AUDIT_SCHEMA='bf16-injection-vector-audit-2.0'
NUMERICAL_AUDIT_PATCH='sha256:7e1127fb91c5ba07301eda95e12bf82324d041b8f122ebf8132b01f610c7e080'
freeze(OUTPUT_ROOT/'numerical_audit_patch_v2.json',{
    'schema':INJECTION_AUDIT_SCHEMA,'patch_code_digest':NUMERICAL_AUDIT_PATCH,
    'v1_notebook_sha256':'sha256:97b4bb979f153ad0515995660e2265471c34d231ee14577fe423580017357dcd','v1_frozen_implementation':IMPLEMENTATION,
    'run_id':RUN_ID,'protocol_digest':PROTOCOL_DIGEST,
    'scope':'CPU norm reporting plus exact BF16 vector checks; no payload, model, or scientific-design changes',
    'legacy_policy':'Retain original records after hashes and v1 scalar gate; do not claim retroactive vector checks'})

def audit_tensor_hash(t):
    t=t.detach().cpu().contiguous()
    return digest({'shape':list(t.shape),'dtype':str(t.dtype),
        'bytes_sha256':hashlib.sha256(t.view(torch.uint8).numpy().tobytes()).hexdigest()})

def audit_norm64(t):
    return float(torch.linalg.vector_norm(t.detach().cpu().double()))

def audit_injection(result,clean,payload,dose,input_key,condition_id):
    # Retain the actual pre-hook state: the post-hook state alone cannot distinguish
    # a correct addition from replay drift relative to the cached clean state.
    before=result['injection_before'].detach().cpu()
    cached=clean['states'][INJECTION_LAYER].detach().cpu()
    after=result['states'][INJECTION_LAYER].detach().cpu()
    cast=payload.detach().to(device='cpu',dtype=torch.bfloat16)
    require(before.shape==cached.shape==after.shape==cast.shape,'Injection audit: tensor shape mismatch.')
    require(before.dtype==cached.dtype==after.dtype==cast.dtype==torch.bfloat16,'Injection audit: expected BF16 tensors.')
    require(all(torch.isfinite(t).all() for t in (before,cached,after,cast)),'Injection audit: nonfinite tensor.')
    planned_after=cached+cast
    live_expected_after=before+cast
    actual_delta=after.float()-before.float()
    expected_delta=planned_after.float()-cached.float()
    actual_norm=norm(actual_delta)  # same CPU FP32 reporting convention as frozen v1 doses
    frozen_norm=float(dose['expected_actual_added_norm'])
    device_norm=float(result['edit']['actual_added_norm_device_fp32'])
    checks={'prestate_equal':torch.equal(before,cached),
        'hook_addition_equal':torch.equal(after,live_expected_after),
        'effective_delta_equal':torch.equal(actual_delta,expected_delta),
        'frozen_cpu_norm_equal':abs(norm(expected_delta)-frozen_norm)<1e-5}
    audit={'schema':INJECTION_AUDIT_SCHEMA,'patch_code_digest':NUMERICAL_AUDIT_PATCH,
        'passed':all(checks.values()),**checks,
        'cached_before_sha256':audit_tensor_hash(cached),'live_before_sha256':audit_tensor_hash(before),
        'expected_after_sha256':audit_tensor_hash(planned_after),'actual_after_sha256':audit_tensor_hash(after),
        'expected_delta_sha256':audit_tensor_hash(expected_delta),'actual_delta_sha256':audit_tensor_hash(actual_delta),
        'prestate_max_abs_error':float((before.float()-cached.float()).abs().max()),
        'hook_max_abs_error':float((after.float()-live_expected_after.float()).abs().max()),
        'delta_max_abs_error':float((actual_delta-expected_delta).abs().max()),
        'actual_norm_cpu_fp32':actual_norm,'expected_norm_cpu_fp32':norm(expected_delta),
        'actual_norm_cpu_fp64':audit_norm64(actual_delta),'expected_norm_cpu_fp64':audit_norm64(expected_delta),
        'frozen_expected_norm':frozen_norm,'device_norm_fp32':device_norm,
        'device_vs_frozen_abs_error':abs(device_norm-frozen_norm),
        'legacy_scalar_gate_would_pass':abs(device_norm-frozen_norm)<1e-5}
    if not audit['passed']:
        folder=OUTPUT_ROOT/'analysis/injection_diagnostics';folder.mkdir(parents=True,exist_ok=True)
        stem=input_key+'__'+condition_id
        tensor_path=folder/(stem+'.pt');json_path=folder/(stem+'.json')
        torch_atomic(tensor_path,{'cached_before':cached,'live_before':before,'actual_after':after,
            'payload_fp32':payload.detach().cpu(),'cast_payload':cast,'planned_after':planned_after})
        failed=[k for k,v in checks.items() if not v]
        write_json(json_path,{'run_id':RUN_ID,'input_key':input_key,'condition_id':condition_id,
            'failed_checks':failed,'audit':audit,'tensor_sha256':sha(tensor_path)})
        raise ValueError('Injection vector audit failed for '+input_key+' / '+condition_id+
            ': '+', '.join(failed)+'. Diagnostics: '+str(json_path)+
            '. Frozen doses and completed records were kept. Do not loosen the gate or delete locks.')
    result['edit']['actual_added_norm']=actual_norm
    return audit

def validate_injection_record(record):
    actual=float(record['edit']['actual_added_norm']);expected=float(record['dose']['expected_actual_added_norm'])
    require(math.isfinite(actual) and actual>=0 and math.isfinite(expected) and expected>=0,'Invalid saved injection norm.')
    require(abs(actual-expected)<1e-5,'Saved injection does not pass its frozen CPU/v1 norm gate.')
    audit=record.get('injection_audit')
    if audit is None:
        require('actual_added_norm_device_fp32' not in record['edit'],'A v2 record is missing its vector audit.')
        return 'legacy_v1_scalar_only'
    require(audit['schema']==INJECTION_AUDIT_SCHEMA and audit['patch_code_digest']==NUMERICAL_AUDIT_PATCH,'Unknown injection audit version.')
    require(all(audit[k] is True for k in ('passed','prestate_equal','hook_addition_equal','effective_delta_equal','frozen_cpu_norm_equal')),'Failed saved vector checks.')
    for a,b in (('cached_before_sha256','live_before_sha256'),('expected_after_sha256','actual_after_sha256'),('expected_delta_sha256','actual_delta_sha256')):
        require(audit[a]==audit[b],'Saved injection vector hashes disagree.')
    require(all(audit[k]==0 for k in ('prestate_max_abs_error','hook_max_abs_error','delta_max_abs_error')),'Nonzero saved vector error.')
    require(audit['actual_norm_cpu_fp32']==actual and audit['expected_norm_cpu_fp32']==actual and audit['frozen_expected_norm']==expected,'Saved audit norm mismatch.')
    require(audit['actual_norm_cpu_fp64']==audit['expected_norm_cpu_fp64'] and math.isfinite(audit['actual_norm_cpu_fp64']),'Saved FP64 vector norm mismatch.')
    device=float(record['edit']['actual_added_norm_device_fp32'])
    require(math.isfinite(device) and device>=0 and audit['device_norm_fp32']==device,'Invalid device norm diagnostic.')
    require(audit['device_vs_frozen_abs_error']==abs(device-expected) and audit['legacy_scalar_gate_would_pass']==(abs(device-expected)<1e-5),'Device diagnostic mismatch.')
    return 'v2_exact_vectors'

def prepare(example):
    plan=adapter._candidate_plan(example)
    require(plan is not None,'A/B/C must be single-token candidates.')
    prefix,token_by_label=plan
    inputs=adapter._append_text_prefix(adapter._to_device(adapter._template([example.messages()],add_generation_prompt=True)),prefix)
    require(inputs['input_ids'].shape[0]==1,'All forwards use batch size one.')
    require(inputs['attention_mask'][0,-1].item()==1,'Final token must be unpadded.')
    fingerprint=digest({'ids':inputs['input_ids'].detach().cpu().tolist(),'attention':inputs['attention_mask'].detach().cpu().tolist(),
                       'token_by_label':{k:int(v) for k,v in token_by_label.items()}})
    return inputs,token_by_label,fingerprint
def hidden_tensor(output):
    if torch.is_tensor(output):return output
    if isinstance(output,(tuple,list)) and torch.is_tensor(output[0]):return output[0]
    raise TypeError('Unsupported decoder output: '+str(type(output)))
def replace_hidden(output,value):
    if torch.is_tensor(output):return value
    if isinstance(output,tuple):return (value,)+output[1:]
    if isinstance(output,list):return [value]+output[1:]
    raise TypeError(type(output))
def centered_score_error(a,b):
    a=np.asarray(a,dtype=float);b=np.asarray(b,dtype=float)
    return float(np.max(np.abs((a-a.mean())-(b-b.mean()))))
def forward(example,payload=None,expected=None):
    assert_native_precision();inputs,token_by_label,fingerprint=prepare(example)
    if expected is not None:require(fingerprint==expected,'Prepared input or candidate token identity changed.')
    states={};edit={};handles=[];injection_before={}
    def make_hook(layer):
        def hook(module,args,output):
            h=hidden_tensor(output);require(h.dtype==torch.bfloat16 and h.ndim==3,'Expected BF16 decoder state.')
            before=h[:, -1, :].detach().clone()
            if layer==INJECTION_LAYER and payload is not None:
                add=payload.to(device=h.device,dtype=h.dtype);require(add.shape==before.shape,'Payload shape mismatch.')
                after=before+add;changed=h.clone();changed[:,-1,:]=after
                injection_before['tensor']=before.cpu()
                edit.update(bf16_payload_norm=norm(add),actual_added_norm=norm((after.float()-before.float()).cpu()),
                            actual_added_norm_device_fp32=norm(after.float()-before.float()),
                            requested_fp32_norm=norm(payload),actual_cosine_to_requested=cosine(after.float()-before.float(),add.float()))
                states[layer]=after.cpu();return replace_hidden(output,changed)
            states[layer]=before.cpu()
        return hook
    try:
        for layer in CAPTURE_LAYERS:handles.append(DECODER_LAYERS[layer].register_forward_hook(make_hook(layer)))
        with torch.inference_mode():out=adapter.model(**inputs,use_cache=False)
        scores=torch.log_softmax(out.logits[0,-1].float(),dim=-1)
        values=[float(scores[token_by_label[k]].cpu()) for k in LABELS]
    finally:
        for h in handles:h.remove()
    require(set(states)==set(CAPTURE_LAYERS) and all(torch.isfinite(t).all() for t in states.values()),'Missing/nonfinite captured states.')
    require(np.isfinite(values).all(),'Nonfinite candidate scores.')
    idx=int(np.argmax(values));record={'candidate_labels':list(LABELS),'candidate_semantics':list(example.option_semantics),
        'candidate_logprobs':values,'predicted_label':LABELS[idx],'predicted_semantic':example.option_semantics[idx]}
    result={'score':record,'states':states,'fingerprint':fingerprint,'edit':edit}
    if payload is not None:
        require('tensor' in injection_before,'Injection hook was not called.')
        result['injection_before']=injection_before['tensor']
    return result
def clean_paths(key):return OUTPUT_ROOT/'clean_cache'/(key+'.pt'),OUTPUT_ROOT/'clean_records'/(key+'.json')
def load_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key);meta=read_json(jp)
    require(meta['run_id']==RUN_ID and meta['capture_key']==key and sha(tp)==meta['tensor_sha256'],'Clean capture identity/hash mismatch: '+key)
    data=torch.load(tp,map_location='cpu',weights_only=True)
    require(data['run_id']==RUN_ID and data['capture_key']==key,'Wrong tensor identity.')
    require(set(data['states'])==set(CAPTURE_LAYERS) and all(t.dtype==torch.bfloat16 and torch.isfinite(t).all() and norm(t)>0 for t in data['states'].values()),'Invalid clean tensor.')
    require(data['score']==meta['score'] and data['fingerprint']==meta['fingerprint'],'Clean metadata mismatch.')
    return data
def get_or_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key)
    if jp.is_file():return load_capture(row,mid)
    result=forward(make_example(row,mid));result.update(run_id=RUN_ID,capture_key=key)
    torch_atomic(tp,result)
    write_json(jp,{k:result[k] for k in ('run_id','capture_key','score','fingerprint')}|{'tensor_sha256':sha(tp)})
    return result
def semantic_scores(record):return dict(zip(record['candidate_semantics'],record['candidate_logprobs']))
```

```python
%%cell_progress 9/14 Freeze physical and control payloads
cell_stage('Freeze physical and control payloads')
def donor_bundle(fid,orientation,context,slot):
    maps=source_mappings(fid,orientation,context);mid=maps[slot]
    rb=SCENE_LOOKUP[(fid,orientation,f'source{context}_before')];ra=SCENE_LOOKUP[(fid,orientation,f'source{context}_after')]
    # Background control deliberately preserves the SAME mapping, even when the other context's assigned maps differ.
    other_row=SCENE_LOOKUP[(fid,orientation,f'source{1-context}_before')]
    before=load_capture(rb,mid);after=load_capture(ra,mid)
    mapped=load_capture(rb,maps[1-slot])
    other=load_capture(other_row,mid)
    require(rb['outcome']==ra['outcome']==other_row['outcome'],'Source answer neutrality violated.')
    require(before['score']['candidate_semantics']==after['score']['candidate_semantics'],'Source mapping changed across physical update.')
    return {'before':before['states'][48].float(),'after':after['states'][48].float(),
        'background':other['states'][48].float(),'mapping':mapped['states'][48].float(),
        'source_semantic':rb['outcome'],'source_label':LABELS[PERMUTATIONS[mid].index(rb['outcome'])],
        'source_mapping':mid,'source_capture_keys':[capture_key(rb,mid),capture_key(ra,mid),capture_key(other_row,mid),capture_key(rb,maps[1-slot])]}

def weight_payload(source_before,source_after,receiver_before,alpha):
    # Scientific firewall: ONLY three tensors and a prespecified scalar. No labels, predictions, target CF state or truth.
    require(source_before.shape==source_after.shape==receiver_before.shape,'Unaligned donor/receiver dimensions.')
    return (source_after.float()-source_before.float())*(norm(receiver_before)/norm(source_before))*float(alpha)

def payload_from_tensors(kind,alpha,donor,receiver_before,random_seed):
    base=weight_payload(donor['before'],donor['after'],receiver_before,1.)
    n=norm(base);require(n>1e-8,'Zero source physical difference; candidate is not usable. Do not silently redraw.')
    if kind in ('physical','physical_half','reverse'):return base*alpha
    if kind=='background':direction=donor['background']-donor['before']
    elif kind=='mapping':direction=donor['mapping']-donor['before']
    elif kind=='random':
        generator=torch.Generator(device='cpu').manual_seed(random_seed)
        direction=torch.randn(base.shape,generator=generator,dtype=torch.float32)
    else:raise ValueError(kind)
    require(norm(direction)>1e-8,'Zero control direction. No silent redraw or substitution.')
    return direction.float()*(n/norm(direction))

DONORS={}
for fid,orientation,context,slot in tqdm(list(itertools.product([f['family_id'] for f in families],('anchor','mirror'),range(2),range(2))),desc='Freeze answer-neutral donor differences'):
    DONORS[(fid,orientation,context,slot)]=donor_bundle(fid,orientation,context,slot)
# Background controls use the same two maps already captured in both donor contexts.
ALL_CLEAN_INVENTORY={p.stem:sha(p) for p in sorted((OUTPUT_ROOT/'clean_records').glob('*.json'))}
ALL_CLEAN_DIGEST=freeze(OUTPUT_ROOT/'all_clean_inventory.json',ALL_CLEAN_INVENTORY)

def make_payload(row,mid,plan,clean):
    own=clean['states'][48].float()
    if plan['kind']=='zero':return torch.zeros_like(own),{'source_semantic':None,'source_label':None,'source_mapping':None,'source_capture_keys':[]}
    key=(row['family_id'],row['orientation'],plan['context'],plan['source_mapping_slot']);donor=DONORS[key]
    # Random vector is shared across all receiver maps and both recipient contexts for this donor.
    seed=int(stable_key(['random',*key])[:15],16)
    payload=payload_from_tensors(plan['kind'],plan['alpha'],donor,own,seed)
    return payload,{k:donor[k] for k in ('source_semantic','source_label','source_mapping','source_capture_keys')}

dose_rows=[]
for row,mid in tqdm(TARGET_INPUTS,desc='Freeze exact payload fingerprints',unit='input'):
    clean=load_capture(row,mid)
    for plan in PLAN:
        p,source=make_payload(row,mid,plan,clean);cast=p.to(torch.bfloat16)
        require(norm(p)==0 or abs(norm(cast)-norm(p))/norm(p)<.005,'BF16 cast norm mismatch.')
        effective=(clean['states'][48]+cast).float()-clean['states'][48].float()
        dose_rows.append({'input_key':capture_key(row,mid),'condition_id':plan['condition_id'],
            'payload_sha256':'sha256:'+hashlib.sha256(p.contiguous().numpy().tobytes()).hexdigest(),
            'requested_norm':norm(p),'bf16_norm':norm(cast),'relative_norm':norm(cast)/norm(clean['states'][48]),
            'expected_actual_added_norm':norm(effective),**source})
DOSE_DIGEST=freeze(OUTPUT_ROOT/'frozen_doses.json',dose_rows)
DOSES={(r['input_key'],r['condition_id']):r for r in dose_rows}
pd.DataFrame(dose_rows).to_csv(OUTPUT_ROOT/'analysis/dose_inventory.csv',index=False)
print('Doses frozen. Relative payload norms:')
display(pd.DataFrame(dose_rows).relative_norm.describe())
```

```python
%%cell_progress 10/14 Validate replay and actual BF16 edits
cell_stage('Validate replay and actual BF16 edits')
technical=[]
for row,mid in (TARGET_INPUTS[0],TARGET_INPUTS[-1]):
    clean=load_capture(row,mid);e=make_example(row,mid)
    z=forward(e,torch.zeros_like(clean['states'][48]),expected=clean['fingerprint'])
    replay=forward(e,expected=clean['fingerprint'])
    score_error=centered_score_error(z['score']['candidate_logprobs'],clean['score']['candidate_logprobs'])
    replay_error=centered_score_error(replay['score']['candidate_logprobs'],clean['score']['candidate_logprobs'])
    state_error=max(norm(z['states'][l].float()-clean['states'][l].float()) for l in CAPTURE_LAYERS)
    require(max(score_error,replay_error,state_error)<=SCORE_TOLERANCE,'Zero/replay check failed; restart with frozen precision.')
    plan=next(p for p in PLAN if p['kind']=='physical');payload,_=make_payload(row,mid,plan,clean)
    actual=forward(e,payload,expected=clean['fingerprint']);expected=(clean['states'][48]+payload.to(torch.bfloat16)).float()-clean['states'][48].float()
    require(norm(actual['states'][48].float()-clean['states'][48].float()-expected)<1e-6,'Actual hook edit differs from expected BF16 addition.')
    technical.append({'input_key':capture_key(row,mid),'zero_score_error':score_error,'replay_score_error':replay_error,'zero_state_error':state_error,'passed':True})
freeze(OUTPUT_ROOT/'technical_controls.json',technical)
print('Zero/replay checks, prepared token identity and actual BF16 hook edits passed.')
```

```python
%%cell_progress 11/14 Run resumable frozen interventions
cell_stage('Run resumable frozen interventions')
def shard_path(row,mid):return OUTPUT_ROOT/'intervention_shards'/(capture_key(row,mid)+'.json')
def shard_identity(row,mid):return {'run_id':RUN_ID,'all_clean_digest':ALL_CLEAN_DIGEST,'dose_digest':DOSE_DIGEST,'input_key':capture_key(row,mid)}
def validate_shard(shard,row,mid):
    require(shard['identity']==shard_identity(row,mid),'Shard identity mismatch.')
    ids=[r['condition']['condition_id'] for r in shard['records']]
    require(len(ids)==len(set(ids)) and set(ids)<={p['condition_id'] for p in PLAN},'Duplicate/unknown conditions.')
    plan_lookup={p['condition_id']:p for p in PLAN}
    for r in shard['records']:
        body={k:v for k,v in r.items() if k!='record_digest'}
        require(digest(body)==r['record_digest'],'Record checksum failed.')
        p=r['condition'];require(p==plan_lookup[p['condition_id']],'Condition metadata changed.')
        require(r['dose']==DOSES[(capture_key(row,mid),p['condition_id'])],'Frozen dose mismatch.')
        s=r['score'];require(s['candidate_labels']==list(LABELS) and s['candidate_semantics']==list(PERMUTATIONS[mid]),'Candidate map mismatch.')
        values=np.array(s['candidate_logprobs']);require(values.shape==(3,) and np.isfinite(values).all(),'Invalid candidate scores.')
        idx=int(values.argmax());require(s['predicted_label']==LABELS[idx] and s['predicted_semantic']==PERMUTATIONS[mid][idx],'Score/prediction mismatch.')
        require(set(r['traces'])==set(map(str,CAPTURE_LAYERS)),'Missing traces.')
        validate_injection_record(r)
    return shard

def run_one(row,mid):
    path=shard_path(row,mid)
    shard=validate_shard(read_json(path),row,mid) if path.is_file() else {'identity':shard_identity(row,mid),'records':[]}
    done={r['condition']['condition_id'] for r in shard['records']}
    if len(done)==len(PLAN):return
    clean=load_capture(row,mid);after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    natural=load_capture(after_row,mid);e=make_example(row,mid)
    for p in PLAN:
        if p['condition_id'] in done:continue
        payload,_=make_payload(row,mid,p,clean);dose=DOSES[(capture_key(row,mid),p['condition_id'])]
        require('sha256:'+hashlib.sha256(payload.contiguous().numpy().tobytes()).hexdigest()==dose['payload_sha256'],'Payload changed after freezing.')
        result=forward(e,payload,expected=clean['fingerprint'])
        injection_audit=audit_injection(result,clean,payload,dose,capture_key(row,mid),p['condition_id'])
        traces={}
        for layer in CAPTURE_LAYERS:
            delta=result['states'][layer].float()-clean['states'][layer].float()
            natural_delta=natural['states'][layer].float()-clean['states'][layer].float()
            traces[str(layer)]={'induced_norm':norm(delta),'natural_cf_norm':norm(natural_delta),
                'cosine_to_natural_cf':cosine(delta,natural_delta),'projection_on_natural_cf':float((delta*natural_delta).sum()/norm(natural_delta)) if norm(natural_delta)>1e-12 else None}
        record={'condition':p,'score':result['score'],'dose':dose,'edit':result['edit'],'traces':traces,'injection_audit':injection_audit}
        record['record_digest']=digest(record);shard['records'].append(record);write_json(path,shard)
    validate_shard(shard,row,mid)

def run_all():
    start=time.monotonic();new_done=0
    for i,(row,mid) in enumerate(tqdm(TARGET_INPUTS,desc=MODEL_MODE+': frozen transfer conditions',unit='input')):
        path=shard_path(row,mid);complete=path.is_file() and len(read_json(path).get('records',[]))==len(PLAN)
        run_one(row,mid)
        if not complete:
            new_done+=1
            if new_done in (1,8):print('Remaining ETA from completed inputs:',round((time.monotonic()-start)/new_done*(len(TARGET_INPUTS)-i-1)/3600,2),'hours')
run_all()
```

```python
%%cell_progress 12/14 Verify complete grids and assemble raw scores
cell_stage('Verify complete grids and assemble raw scores')
records=[];trace_rows=[];shard_hashes={}
for row,mid in TARGET_INPUTS:
    path=shard_path(row,mid);shard=validate_shard(read_json(path),row,mid)
    require(len(shard['records'])==len(PLAN),'Incomplete condition grid.')
    shard_hashes[path.name]=sha(path);clean=load_capture(row,mid)
    cfrow=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    cf=load_capture(cfrow,mid);base_sem=row['outcome'];cf_sem=cfrow['outcome'];cf_label=LABELS[PERMUTATIONS[mid].index(cf_sem)]
    changed=row['changed_side'];opposite='right' if changed=='left' else 'left'
    # Signed margin always points toward the physically increased side, even in nonflip contexts.
    baseline=semantic_scores(clean['score']);natural=semantic_scores(cf['score'])
    for r in shard['records']:
        s=r['score'];ss=semantic_scores(s);p=r['condition'];dose=r['dose']
        base={'family_id':row['family_id'],'scene_id':row['scene_id'],'orientation':row['orientation'],'receiver_context':row['role'].split('_')[0],
            'receiver_mapping':mid,'input_key':capture_key(row,mid),**p,'base_semantic':base_sem,'cf_semantic':cf_sem,
            'cf_label':cf_label,'predicted_semantic':s['predicted_semantic'],'predicted_label':s['predicted_label'],
            'clean_correct':clean['score']['predicted_semantic']==base_sem,'natural_cf_correct':cf['score']['predicted_semantic']==cf_sem,
            'clean_cf_success':clean['score']['predicted_semantic']==cf_sem,'cf_success':s['predicted_semantic']==cf_sem,
            'margin_shift':(ss[changed]-ss[opposite])-(baseline[changed]-baseline[opposite]),
            'natural_margin_shift':(natural[changed]-natural[opposite])-(baseline[changed]-baseline[opposite]),
            'cf_vs_best_other':ss[cf_sem]-max(v for k,v in ss.items() if k!=cf_sem),
            'source_semantic':dose['source_semantic'],'source_label':dose['source_label'],
            'source_semantic_conflict':dose['source_semantic'] is not None and dose['source_semantic']!=cf_sem,
            'source_label_conflict':dose['source_label'] is not None and dose['source_label']!=cf_label,
            'source_semantic_copy':s['predicted_semantic']==dose['source_semantic'],'source_label_copy':s['predicted_label']==dose['source_label'],
            'source_endpoints_correct':False,'relative_payload_norm':dose['relative_norm'],
            'injection_audit_level':validate_injection_record(r),**r['edit'],
            **{'logprob_'+k:v for k,v in zip(LABELS,s['candidate_logprobs'])}}
        if p['context']>=0:
            ds=DONORS[(row['family_id'],row['orientation'],p['context'],p['source_mapping_slot'])]
            endpoint_scores=[read_json(clean_paths(key)[1])['score'] for key in ds['source_capture_keys'][:2]]
            base['source_endpoints_correct']=all(x['predicted_semantic']==ds['source_semantic'] for x in endpoint_scores)
        base['competence_satisfied']=base['source_endpoints_correct'] and base['clean_correct'] and base['natural_cf_correct']
        records.append(base)
        for layer,t in r['traces'].items():trace_rows.append({k:base[k] for k in ('family_id','input_key','condition_id','kind','receiver_context','orientation','receiver_mapping')}|{'layer':int(layer),**t})
frame=pd.DataFrame(records);traces=pd.DataFrame(trace_rows)
require(len(frame)==len(TARGET_INPUTS)*len(PLAN),'Record count mismatch.')
require(not frame.duplicated(['input_key','condition_id']).any(),'Duplicate grid entries.')
zero=frame[frame.kind=='zero']
require(zero.margin_shift.abs().max()<=SCORE_TOLERANCE and zero.actual_added_norm.max()==0,'All-input zero controls failed.')
require(np.isfinite(frame[['margin_shift','relative_payload_norm','actual_added_norm']].to_numpy()).all(),'Nonfinite analysis values.')
frame.to_csv(OUTPUT_ROOT/'analysis/intervention_records.csv',index=False);traces.to_csv(OUTPUT_ROOT/'analysis/state_response_traces.csv',index=False)
freeze(OUTPUT_ROOT/'shard_inventory.json',shard_hashes)
print('Complete:',len(frame),'interventions,',len(traces),'traces; zero margin max:',zero.margin_shift.abs().max())


audit_summary={'schema':INJECTION_AUDIT_SCHEMA,'run_id':RUN_ID,'patch_code_digest':NUMERICAL_AUDIT_PATCH,
    'records':len(frame),'counts':{str(k):int(v) for k,v in frame.injection_audit_level.value_counts().items()},
    'new_records_that_would_fail_old_scalar_gate':int(sum(
        not r['injection_audit']['legacy_scalar_gate_would_pass']
        for row,mid in TARGET_INPUTS for r in read_json(shard_path(row,mid))['records'] if 'injection_audit' in r)),
    'legacy_limitation':'Legacy v1 records retain original scalar checks; no saved live prestate exists for retrospective exact vector checks.'}
frame[['input_key','condition_id','injection_audit_level','actual_added_norm']+
      (['actual_added_norm_device_fp32'] if 'actual_added_norm_device_fp32' in frame else [])].to_csv(
    OUTPUT_ROOT/'analysis/injection_audit_coverage.csv',index=False)
write_json(OUTPUT_ROOT/'analysis/injection_audit_summary.json',audit_summary)
print('Injection audit coverage:',audit_summary)
```

```python
%%cell_progress 13/14 Estimate family-paired effects and fixed primary tests
cell_stage('Estimate family-paired effects and fixed primary tests')
def estimate(values,key,test=False):
    values=pd.Series(values,dtype=float).dropna().sort_index()
    require(values.index.is_unique,'Statistics require one value per family.')
    if len(values)<2:return {'n_families':len(values),'estimate':float(values.mean()) if len(values) else None,'ci_low':None,'ci_high':None,'p_two_sided':None}
    x=values.to_numpy();require(np.isfinite(x).all(),'Invalid family effects.')
    rng=np.random.default_rng(int(stable_key(['stats',key])[:15],16));means=[]
    for start in range(0,BOOTSTRAPS,256):
        count=min(256,BOOTSTRAPS-start);means.extend(x[rng.integers(len(x),size=(count,len(x)))].mean(1))
    out={'n_families':len(x),'estimate':float(x.mean()),'ci_low':float(np.quantile(means,.025)),'ci_high':float(np.quantile(means,.975))}
    if test:
        exceed=0
        for start in range(0,SIGNFLIPS,1024):
            count=min(1024,SIGNFLIPS-start);signs=rng.integers(0,2,size=(count,len(x)))*2-1
            exceed+=int((np.abs((x*signs).mean(1))>=abs(x.mean())-1e-12).sum())
        out['p_two_sided']=(exceed+1)/(SIGNFLIPS+1)
    return out
def holm(pvalues):
    out=np.zeros(len(pvalues));running=0.
    for rank,i in enumerate(np.argsort(pvalues)):
        running=max(running,(len(pvalues)-rank)*pvalues[i]);out[i]=min(1.,running)
    return out
flip=frame[(frame.receiver_context=='flip')&(frame.kind!='zero')]
paired=flip.pivot(index=['family_id','input_key','context','source_mapping_slot'],columns='kind',values='margin_shift')
require(not paired.isna().any().any(),'Unpaired primary conditions.')
primary=[];family_values=[]
values=paired.physical.groupby('family_id').mean()
require(len(values)==N_FAMILIES,'Primary families missing.')
primary.append({'test':'physical_vs_clean',**estimate(values,'physical_vs_clean',True)})
family_values.extend({'test':'physical_vs_clean','family_id':fid,'value':float(v)} for fid,v in values.items())
for control in ('random','background','mapping'):
    values=(paired.physical-paired[control]).groupby('family_id').mean()
    require(len(values)==N_FAMILIES,'Primary families missing.')
    name='physical_vs_'+control;primary.append({'test':name,**estimate(values,name,True)})
    family_values.extend({'test':name,'family_id':fid,'value':float(v)} for fid,v in values.items())
primary=pd.DataFrame(primary);primary['p_holm']=holm(primary.p_two_sided.to_numpy())
primary['passed']=(primary.estimate>0)&(primary.p_holm<.05)
summaries=[]
for scope,sub in [('all',frame)]+[('mapping_'+str(mid),g) for mid,g in frame.groupby('receiver_mapping')]+[('side_'+s,g) for s,g in frame.groupby('orientation')]:
    for (context,kind),g in sub.groupby(['receiver_context','kind']):
        for metric in ('cf_success','margin_shift','relative_payload_norm'):
            v=g.groupby('family_id')[metric].mean();summaries.append({'scope':scope,'receiver_context':context,'kind':kind,'metric':metric,**estimate(v,[scope,context,kind,metric])})
summary=pd.DataFrame(summaries)
physical=frame[frame.kind=='physical'].copy()
behavior_effects=[]
for context,g in physical.groupby('receiver_context'):
    values=(g.cf_success.astype(float)-g.clean_cf_success.astype(float)).groupby(g.family_id).mean()
    behavior_effects.append({'context':context,'metric':'cf_accuracy_gain_over_clean',**estimate(values,['accuracy_gain',context])})
behavior_effects=pd.DataFrame(behavior_effects)
stable=behavior_effects[behavior_effects.context=='stable'].iloc[0]
competence=[]
for context,g in physical.groupby('receiver_context'):
    for group,part in [('all',g),('competence_satisfied',g[g.competence_satisfied])]:
        competence.append({'context':context,'subset':group,'inputs_with_donor_variants':len(part),'total_inputs_with_donor_variants':len(g),
            **estimate(part.groupby('family_id').cf_success.mean(),['competence',context,group])})
competence=pd.DataFrame(competence)
copy_rows=[]
for kind,g in flip.groupby('kind'):
    for name,mask,metric in [('semantic',g.source_semantic_conflict,'source_semantic_copy'),('literal_letter',g.source_label_conflict,'source_label_copy')]:
        sub=g[mask];copy_rows.append({'kind':kind,'copy_type':name,'eligible_records':len(sub),
            **estimate(sub.groupby('family_id')[metric].mean(),['copy',kind,name])})
copy_summary=pd.DataFrame(copy_rows)
dependence=[]
for metric in ('margin_shift','cf_success'):
    w=physical[physical.receiver_context=='flip'].groupby(['family_id','context'])[metric].mean().unstack('context')
    dependence.append({'contrast':'source_context_1_minus_0','metric':metric,**estimate(w[1]-w[0],['source_context',metric])})
dependence=pd.DataFrame(dependence)
flip_gain=behavior_effects[behavior_effects.context=='flip'].iloc[0]
DECISION={'primary_four_pass':bool(primary.passed.all()),'stable_retention_gate':bool(stable.ci_low>-STABLE_LOSS_BOUND),
    'flip_accuracy_gain_ci_positive':bool(flip_gain.ci_low>0),
    'scope':'frozen candidate additive update at L48; not natural necessity or multiplication proof',
    'native_only':MODEL_MODE=='native','synthetic_only':SYNTHETIC_ONLY}
DECISION['label']=('candidate_margin_transfer_with_declared_retention' if DECISION['primary_four_pass'] and DECISION['stable_retention_gate']
                   else 'candidate_margin_effect_with_retention_unresolved' if DECISION['primary_four_pass'] else 'prespecified_transfer_not_confirmed')
for name,table in [('primary_four_tests',primary),('primary_family_contrasts',pd.DataFrame(family_values)),('condition_summary',summary),
    ('behavior_effects',behavior_effects),('competence_diagnostics',competence),('source_copy_diagnostics',copy_summary),('source_context_dependence',dependence)]:
    table.to_csv(OUTPUT_ROOT/'analysis'/(name+'.csv'),index=False)
write_json(OUTPUT_ROOT/'decision.json',DECISION)
display(primary.round(5));display(behavior_effects.round(5));display(competence.round(5));display(copy_summary.round(5))
print('Frozen decision:',DECISION['label'])
```
