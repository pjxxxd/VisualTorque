```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 13 tracked cells.')
```

```python
%%cell_progress 6/13 Reconstruct all intervention scores and vector-audit summaries
cell_stage('Reconstruct all intervention scores and vector-audit summaries')
def check_vector_audit(rec):
    audit=rec.get('injection_audit');dose=rec['dose'];edit=rec['edit']
    require(audit is not None and audit['schema']=='bf16-injection-vector-audit-2.0','Missing vector audit.')
    require(audit['patch_code_digest']==PROTOCOL['implementation'],'Vector audit implementation mismatch.')
    for k in ('passed','prestate_equal','hook_addition_equal','effective_delta_equal','frozen_cpu_norm_equal'):require(audit[k] is True,'Failed saved vector check: '+k)
    for a,b in [('cached_before_sha256','live_before_sha256'),('expected_after_sha256','actual_after_sha256'),('expected_delta_sha256','actual_delta_sha256')]:require(audit[a]==audit[b],'Vector hash mismatch.')
    for k in ('prestate_max_abs_error','hook_max_abs_error','delta_max_abs_error'):require(audit[k]==0,'Nonzero saved vector error.')
    for val in (audit['actual_norm_cpu_fp32'],audit['expected_norm_cpu_fp32'],audit['frozen_expected_norm'],edit['actual_added_norm']):
        close(val,dose['expected_actual_added_norm'],'FP32 effective dose')
    close(audit['actual_norm_cpu_fp64'],audit['expected_norm_cpu_fp64'],'FP64 effective dose')
    close(audit['actual_norm_cpu_fp64'],dose['effective_induced_norm'],'FP64 norm versus dose inventory')
    return True

def reconstruct_geometry(t):
    n,m=float(t['induced_norm']),float(t['natural_cf_norm']);p=t['projection_on_natural_cf']
    require(math.isfinite(n) and math.isfinite(m) and n>=0 and m>=0,'Invalid state norm.')
    if m<=AXIS_EPS:
        return {'cf_aligned_fraction':None,'relative_distance_to_cf':None,'closeness_gain_over_clean':None,'cosine_to_natural_cf':None}
    require(p is not None and math.isfinite(p) and abs(p)<=n+1e-8*max(n,1),'Invalid state projection.')
    sq=n*n+m*m-2*p*m;require(sq>=-1e-9*max(n*n+m*m,1),'Impossible state distance.')
    dist=math.sqrt(max(sq,0))/m
    return {'cf_aligned_fraction':p/m,'relative_distance_to_cf':dist,'closeness_gain_over_clean':1-dist,
        'cosine_to_natural_cf':p/n if n>AXIS_EPS else None}

records_out=[];trace_rows=[];pair_records=[]
for key in tqdm(TARGET_KEYS,desc='Reconstruct candidate scores and saved BF16 checks',unit='input'):
    c=CLEAN[key];row=c['scene'];prompt=c['prompt'];base=c['scores'];after=CLEAN[BEFORE_AFTER[key]];cfsem=after['scene']['outcome']
    changed=row['changed_side'];opposite='right' if changed=='left' else 'left'
    shard=read_json('model/intervention_shards/'+key+'.json')
    require(shard['identity']=={'run_id':RUN_ID,'all_clean_digest':read_json('model/all_clean_inventory.json')['digest'],
        'dose_digest':read_json('model/frozen_doses.json')['digest'],'input_key':key},'Shard identity mismatch.')
    ids=[r['condition']['condition_id'] for r in shard['records']]
    require(len(ids)==len(set(ids))==39 and set(ids)==set(PLAN),'Incomplete shard.')
    for rec in shard['records']:
        require(digest({k:v for k,v in rec.items() if k!='record_digest'})==rec['record_digest'],'Record digest mismatch.')
        plan=rec['condition'];dose=rec['dose'];cid=plan['condition_id']
        require(plan==PLAN[cid] and dose==DOSES[(key,cid)],'Condition/dose changed.')
        require(check_vector_audit(rec),'Saved vector audit failed.')
        ss=scores(rec['score'],prompt)
        is_cf=rec['score']['predicted_semantic']==cfsem;base_cf=c['meta']['score']['predicted_semantic']==cfsem
        r={'family_id':row['family_id'],'input_key':key,'scene_id':row['scene_id'],'receiver_context':row['role'].split('_')[0],
            'orientation':row['orientation'],'receiver_mapping':prompt['mapping_id'],**plan,
            'predicted_semantic':rec['score']['predicted_semantic'],'predicted_label':rec['score']['predicted_label'],
            'cf_semantic':cfsem,'cf_success':is_cf,'clean_cf_success':base_cf,
            'margin_shift':ss[changed]-ss[opposite]-base[changed]+base[opposite],
            'cf_accuracy_gain':float(is_cf)-float(base_cf),
            'cf_best_other_gain':margin(ss,cfsem)-margin(base,cfsem),
            'clean_correct':c['meta']['score']['predicted_semantic']==row['outcome'],
            'natural_cf_correct':after['meta']['score']['predicted_semantic']==cfsem,
            'actual_added_norm':rec['edit']['actual_added_norm'],
            **{'logprob_'+k:v for k,v in zip(LABELS,rec['score']['candidate_logprobs'])}}
        records_out.append(r)
        require(set(rec['traces'])=={'48','49','55'},'Missing state layers.')
        for layer,t in rec['traces'].items():
            rebuilt=reconstruct_geometry(t)
            for metric,value in rebuilt.items():close(value,t[metric],'State geometry '+metric,rtol=2e-7,atol=2e-7)
            if layer=='48':
                for metric in ('induced_norm','natural_cf_norm','projection_on_natural_cf','cosine_to_natural_cf','cf_aligned_fraction','relative_distance_to_cf','closeness_gain_over_clean'):
                    close(t[metric],dose['effective_'+metric],'L48 expected effective '+metric)
            trace_rows.append({k:r[k] for k in PAIR+['condition_id','kind','receiver_context','orientation','receiver_mapping']}|{'layer':int(layer),**t})
        if plan['kind']=='full':
            source_keys=dose['source_capture_keys'];require(len(source_keys)==2,'Wrong donor endpoints.')
            s0,s1=[CLEAN[k] for k in source_keys];truth=s0['scene']['outcome']
            for endpoint,source in [('before',s0),('after',s1)]:
                require(source['scene']['family_id']==row['family_id'] and source['scene']['orientation']==row['orientation']
                    and source['scene']['role']==f"source{plan['context']}_{endpoint}" and source['prompt']['mapping_id']==dose['source_mapping'],
                    'Donor endpoint is not from the planned family/context/mapping.')
            require(truth==s1['scene']['outcome'] and s0['prompt']['prompt']==s1['prompt']['prompt'],'Source neutrality/prompt mismatch.')
            pred0=s0['meta']['score']['predicted_semantic'];pred1=s1['meta']['score']['predicted_semantic']
            stable=pred0==pred1;unique_top=not tie_top(s0['scores']) and not tie_top(s1['scores'])
            both=pred0==pred1==truth
            flags={'prediction_stable':stable,'stable_unique_top':stable and unique_top,'source_both_correct':both}
            for flag,v in flags.items():require(dose[flag] is v,'Saved source flag mismatch: '+flag)
            pair_records.append({k:r[k] for k in PAIR+['receiver_context','orientation','receiver_mapping']}|flags|
                {'stable_wrong':stable and not both,'axis_defined':dose['axis_defined'],'source_mapping':dose['source_mapping'],
                 'signed_coefficient':dose['signed_coefficient'],'source_before_key':source_keys[0],'source_after_key':source_keys[1]})
frame=pd.DataFrame(records_out);traces=pd.DataFrame(trace_rows);pair_meta=pd.DataFrame(pair_records).set_index(PAIR).sort_index()
require(len(frame)==22464 and len(traces)==67392 and len(pair_meta)==2304 and pair_meta.index.is_unique,'Reconstructed counts wrong.')
compare_table(frame,read_table('intervention_records'),['input_key','condition_id'],fields=[c for c in frame if c not in ('input_key','condition_id','cf_semantic')],label='Raw-score reconstruction')
compare_table(traces,read_table('state_response_traces'),['input_key','condition_id','layer'],label='State traces')
zero=frame[frame.kind=='zero'];require(zero.actual_added_norm.eq(0).all() and zero.margin_shift.abs().max()<=.15,'Zero replay failed.')
require(traces.groupby(['input_key','layer']).natural_cf_norm.nunique().eq(1).all(),'Natural norm differs between conditions.')
save(frame,'reconstructed_intervention_records');save(traces,'reconstructed_state_traces');save(pair_meta.reset_index(),'reconstructed_pair_metadata')
print('Rebuilt all candidate scores, source flags and state traces; saved live-vector checks all passed.')
```

```python
%%cell_progress 7/13 Reproduce the six original tests and behavior summaries
cell_stage('Reproduce the six original tests and behavior summaries')
def original_estimate(values,key,test=False):
    values=pd.Series(values,dtype=float).dropna().sort_index();require(values.index.is_unique,'One value per family required.')
    if len(values)<2:return {'n_families':len(values),'estimate':float(values.mean()) if len(values) else None,'ci_low':None,'ci_high':None,'p_two_sided':None}
    x=values.to_numpy();require(np.isfinite(x).all(),'Nonfinite original family effects.')
    h=digest([PROTOCOL['seed'],['stats',key]]).split(':')[1]
    rng=np.random.default_rng(int(h[:15],16));means=[]
    for start in range(0,PROTOCOL['bootstraps'],256):
        count=min(256,PROTOCOL['bootstraps']-start);means.extend(x[rng.integers(len(x),size=(count,len(x)))].mean(1))
    out={'n_families':len(x),'estimate':float(x.mean()),'ci_low':float(np.quantile(means,.025)),'ci_high':float(np.quantile(means,.975))}
    if test:
        exceed=0
        for start in range(0,PROTOCOL['signflips'],1024):
            count=min(1024,PROTOCOL['signflips']-start);signs=rng.integers(0,2,size=(count,len(x)))*2-1
            exceed+=int((np.abs((x*signs).mean(1))>=abs(x.mean())-1e-12).sum())
        out['p_two_sided']=(exceed+1)/(PROTOCOL['signflips']+1)
    return out

def holm(p):
    result=np.zeros(len(p));running=0.
    for rank,i in enumerate(np.argsort(p)):
        running=max(running,(len(p)-rank)*p[i]);result[i]=min(1.,running)
    return result

def estimate(values,key):
    values=pd.Series(values,dtype=float).dropna().sort_index();require(values.index.is_unique,'One value per family required.')
    x=values.to_numpy();require(np.isfinite(x).all(),'Nonfinite family values.')
    if len(x)<2:return {'n_families':len(x),'estimate':float(x.mean()) if len(x) else None,'ci_low':None,'ci_high':None}
    rng=np.random.default_rng(int(digest([AUDIT_SEED,key]).split(':')[1][:15],16));means=[]
    for start in range(0,AUDIT_BOOTSTRAPS,256):
        count=min(256,AUDIT_BOOTSTRAPS-start);means.extend(x[rng.integers(len(x),size=(count,len(x)))].mean(1))
    return {'n_families':len(x),'estimate':float(x.mean()),'ci_low':float(np.quantile(means,.025)),'ci_high':float(np.quantile(means,.975))}

def paired(context,metric):
    sub=frame[frame.receiver_context==context]
    w=sub[sub.context>=0].pivot(index=PAIR,columns='kind',values=metric).sort_index()
    require(set(w)==set(KINDS) and not w.isna().any().any(),'Incomplete paired conditions: '+metric)
    z=sub[sub.kind=='zero'].set_index('input_key')[metric]
    require(z.index.is_unique,'Duplicate zero inputs.')
    w['zero']=[z.loc[k] for k in w.index.get_level_values('input_key')]
    w['random_aligned_mean']=(w.random_aligned_0+w.random_aligned_1)/2
    w['rotated_residual_mean']=(w.rotated_residual_0+w.rotated_residual_1)/2
    return w

def all_contrasts(w):
    return pd.DataFrame({'aligned_vs_zero':w.aligned-w.zero,'residual_vs_zero':w.residual-w.zero,
        'full_minus_aligned':w.full-w.aligned,'aligned_vs_random_aligned':w.aligned-w.random_aligned_mean,
        'full_vs_rotated_residual':w.full-w.rotated_residual_mean,'factorial_interaction':w.full-w.aligned-w.residual+w.zero,
        'full_vs_zero':w.full-w.zero,'aligned_minus_full':w.aligned-w.full,
        'reduce_residual_vs_full':w.aligned_half_residual-w.full,'reduce_residual_vs_half':w.aligned_half_residual-w.half,
        'half_minus_full':w.half-w.full})

primary=[];primary_families=[]
for name,vals in all_contrasts(paired('flip','margin_shift'))[PRIMARY_NAMES].items():
    fam=vals.groupby('family_id').mean();require(len(fam)==24,'Original primary families missing.')
    primary.append({'test':name,**original_estimate(fam,name,True)})
    primary_families.extend({'test':name,'family_id':f,'value':float(v)} for f,v in fam.items())
primary=pd.DataFrame(primary);primary['p_holm']=holm(primary.p_two_sided.to_numpy())
primary['direction']=np.where(primary.estimate>0,'positive',np.where(primary.estimate<0,'negative','zero'))
primary['adjusted_evidence']=np.where(primary.p_holm<.05,primary.direction,'unresolved')
compare_table(primary,read_table('primary_six_tests'),['test'],label='Original six primary tests')
compare_table(pd.DataFrame(primary_families),read_table('primary_family_contrasts'),['test','family_id'],label='Original family contrasts')
decision=read_json('model/decision.json')
require(decision==completion['decision'] and decision['primary_evidence']==dict(zip(primary.test,primary.adjusted_evidence)),'Original evidence labels changed.')

condition_rows=[]
for (context,kind),g in frame.groupby(['receiver_context','kind']):
    for metric in ('margin_shift','cf_accuracy_gain','cf_best_other_gain'):
        condition_rows.append({'receiver_context':context,'kind':kind,'metric':metric,
            **original_estimate(g.groupby('family_id')[metric].mean(),['condition',context,kind,metric])})
condition_summary=pd.DataFrame(condition_rows)
compare_table(condition_summary,read_table('condition_summary'),['receiver_context','kind','metric'],label='Original condition statistics')
save(primary,'reproduced_primary_six_tests');save(pd.DataFrame(primary_families),'reproduced_primary_family_values');save(condition_summary,'reproduced_condition_summary')
print('Original six tests, family resampling, Holm p-values and original behavior summaries reproduced. No new confirmatory tests.')
show(primary)
```

```python
%%cell_progress 8/13 Audit family coverage and mapping boundaries
cell_stage('Audit family coverage and mapping boundaries')
FOCUS=['aligned_vs_zero','aligned_vs_random_aligned','full_vs_zero','aligned_minus_full','full_vs_rotated_residual','factorial_interaction']
family_rows=[];robust_rows=[];boundary_rows=[]
for context in ('flip','stable'):
    c=all_contrasts(paired(context,'margin_shift'))[FOCUS]
    for name,values in c.items():
        fam=values.groupby('family_id').mean();loo=[float(fam.drop(fid).mean()) for fid in fam.index]
        family_rows.extend({'receiver_context':context,'contrast':name,'family_id':fid,'value':float(v)} for fid,v in fam.items())
        robust_rows.append({'receiver_context':context,'contrast':name,'positive_families':int((fam>0).sum()),
            'negative_families':int((fam<0).sum()),'zero_families':int((fam==0).sum()),'median':float(fam.median()),
            'min_family':float(fam.min()),'max_family':float(fam.max()),'leave_one_out_min':min(loo),'leave_one_out_max':max(loo),
            **estimate(fam,['family',context,name])})
    meta=pair_meta.loc[c.index]
    for column in ('receiver_mapping','orientation','context','source_mapping_slot'):
        values=meta[column] if column in meta else pd.Series(meta.index.get_level_values(column),index=meta.index)
        for level in sorted(values.unique()):
            ids=values.index[values.eq(level)]
            for name in FOCUS[:4]:
                fam=c.loc[ids,name].groupby('family_id').mean()
                boundary_rows.append({'receiver_context':context,'split':column,'level':str(level),'contrast':name,
                    'paired_records':len(ids),**estimate(fam,['boundary',context,column,str(level),name])})
family_values=pd.DataFrame(family_rows);robustness=pd.DataFrame(robust_rows);boundaries=pd.DataFrame(boundary_rows)
save(family_values,'family_effects');save(robustness,'family_influence');save(boundaries,'mapping_mirror_source_boundaries')
show(robustness[robustness.receiver_context=='flip'])
show(boundaries[(boundaries.receiver_context=='flip')&(boundaries.split=='receiver_mapping')&(boundaries.contrast=='aligned_vs_random_aligned')])
```

```python
%%cell_progress 9/13 Compare accuracy and answer transitions on matched inputs
cell_stage('Compare accuracy and answer transitions on matched inputs')
BEHAVIOR_PAIRS=[('aligned','full'),('aligned_half_residual','full'),('aligned_half_residual','half'),('half','full'),
    ('aligned','zero'),('full','zero'),('residual','zero'),('cf_half','cf_delta'),('cf_half','zero'),('cf_delta','zero')]
accuracy_rows=[];transition_rows=[];accuracy_family=[]
def behavioral_pair(context,new_kind,reference):
    sub=frame[frame.receiver_context==context]
    a=sub[sub.kind==new_kind];b=sub[sub.kind==reference]
    # Shared CF/zero conditions are paired once per receiver, never expanded into fake independent observations.
    if a.context.lt(0).all() and b.context.lt(0).all():keys=['family_id','input_key']
    else:keys=PAIR
    columns=['predicted_semantic','cf_semantic','cf_success','margin_shift','cf_best_other_gain']
    def prepare(g,prefix):return g[[c for c in keys if c in g]+columns].rename(columns={c:prefix+c for c in columns})
    if b.context.lt(0).all() and len(keys)==4:
        right=b[['family_id','input_key']+columns].rename(columns={c:'reference_'+c for c in columns})
        joined=prepare(a,'new_').merge(right,on=['family_id','input_key'],how='left',validate='many_to_one')
    else:
        unique(a,keys,'new behavior');unique(b,keys,'reference behavior')
        joined=prepare(a,'new_').merge(prepare(b,'reference_'),on=keys,how='outer',validate='one_to_one',indicator=True)
        require(joined['_merge'].eq('both').all(),'Unpaired behavior inputs.');joined=joined.drop(columns='_merge')
    require(joined.new_cf_semantic.eq(joined.reference_cf_semantic).all() and not joined.reference_cf_success.isna().any(),'Behavior pairing failed.')
    joined['accuracy_difference']=joined.new_cf_success.astype(float)-joined.reference_cf_success.astype(float)
    joined['margin_difference']=joined.new_margin_shift-joined.reference_margin_shift
    joined['best_other_difference']=joined.new_cf_best_other_gain-joined.reference_cf_best_other_gain
    joined['repair']=joined.new_cf_success & ~joined.reference_cf_success
    joined['regression']=~joined.new_cf_success & joined.reference_cf_success
    return joined

for context,(new_kind,reference) in itertools.product(('flip','stable'),BEHAVIOR_PAIRS):
    joined=behavioral_pair(context,new_kind,reference);label=new_kind+'_minus_'+reference
    for metric in ('accuracy_difference','margin_difference','best_other_difference','repair','regression'):
        fam=joined.groupby('family_id')[metric].mean()
        accuracy_rows.append({'receiver_context':context,'contrast':label,'metric':metric,'paired_records':len(joined),
            'distinct_receiver_inputs':joined.input_key.nunique(),**estimate(fam,['behavior_pair',context,label,metric])})
        accuracy_family.extend({'receiver_context':context,'contrast':label,'metric':metric,'family_id':fid,'value':float(v)} for fid,v in fam.items())
    def category(pred,target):return 'cf_correct' if pred==target else 'balanced' if pred=='balanced' else 'opposite_side'
    joined['from_category']=[category(p,t) for p,t in zip(joined.reference_predicted_semantic,joined.reference_cf_semantic)]
    joined['to_category']=[category(p,t) for p,t in zip(joined.new_predicted_semantic,joined.new_cf_semantic)]
    for (fr,to),g in joined.groupby(['from_category','to_category']):
        transition_rows.append({'receiver_context':context,'contrast':label,'from_category':fr,'to_category':to,
            'paired_records':len(g),'distinct_receiver_inputs':g.input_key.nunique(),'families_present':g.family_id.nunique(),
            'note':'descriptive repeated-pair counts, not independent samples'})
    require(int(joined.repair.sum())-int(joined.regression.sum())==int(joined.accuracy_difference.sum()),'Repair/regression accounting failed.')
paired_behavior=pd.DataFrame(accuracy_rows);transitions=pd.DataFrame(transition_rows)
save(paired_behavior,'paired_behavior_effects');save(pd.DataFrame(accuracy_family),'paired_behavior_family_values');save(transitions,'three_class_pair_transitions')
print('Direct paired accuracy differences; positive = new condition improves over reference. CIs do not establish equivalence.')
show(paired_behavior[paired_behavior.metric=='accuracy_difference'])
```

```python
%%cell_progress 10/13 Audit BF16 tails using every fixed threshold
cell_stage('Audit BF16 tails using every fixed threshold')
match_rows=[];amplitude_rows=[];component_saved=read_table('component_inventory')
unique(component_saved,PAIR,'Saved component inventory')
component_saved=component_saved.set_index(PAIR).sort_index()
require(component_saved.index.equals(pair_meta.index),'Component inventory support differs.')
for flag in ('prediction_stable','stable_unique_top','source_both_correct','axis_defined'):component_saved[flag]=strict_bool(component_saved[flag])
for idx,meta in tqdm(pair_meta.iterrows(),total=len(pair_meta),desc='Recompute effective norm matching',unit='pair'):
    fid,key,context,slot=idx
    def d(kind):return DOSES[(key,f'{kind}__s{context}__m{slot}')]
    full,a,r=d('full'),d('aligned'),d('residual')
    for kind in KINDS:
        other=d(kind)
        for field in ('source_capture_keys','source_mapping','source_semantic','source_label','prediction_stable','stable_unique_top','source_both_correct','axis_defined','signed_coefficient'):
            require(other[field]==full[field],'Paired condition metadata differs: '+kind+' / '+field)
    for flag in ('prediction_stable','stable_unique_top','source_both_correct','axis_defined'):
        require(bool(component_saved.loc[idx,flag])==bool(meta[flag]),'Saved component flags differ: '+flag)
    close(component_saved.loc[idx,'full_norm64'],full['requested_induced_norm'],'Full requested norm')
    close(component_saved.loc[idx,'natural_norm64'],full['natural_norm64'],'Natural norm metadata')
    close(component_saved.loc[idx,'signed_coefficient'],full['signed_coefficient'],'Signed projection coefficient')
    an,rn,fn=[x['effective_induced_norm'] for x in (a,r,full)]
    base={'family_id':fid,'input_key':key,'context':context,'source_mapping_slot':slot}
    for kind,reference in [(f'random_aligned_{s}','aligned') for s in range(2)]+[(f'rotated_residual_{s}','full') for s in range(2)]:
        x,ref=d(kind),d(reference)
        requested_error=abs(x['requested_induced_norm']-ref['requested_induced_norm'])/max(ref['requested_induced_norm'],AXIS_EPS)
        effective_error=abs(x['effective_induced_norm']-ref['effective_induced_norm'])/max(ref['effective_induced_norm'],AXIS_EPS)
        requested_proj=x['requested_projection_on_natural_cf'];reference_proj=ref['requested_projection_on_natural_cf']
        projection_difference=x['effective_projection_on_natural_cf']-ref['effective_projection_on_natural_cf'] if requested_proj is not None else None
        require(requested_error<2e-6,'Requested norm match failed.')
        if kind.startswith('rotated') and requested_proj is not None:
            require(abs(requested_proj-reference_proj)/max(ref['requested_induced_norm'],AXIS_EPS)<2e-6,'Requested residual rotation projection failed.')
        match_rows.append({**base,'kind':kind,'reference':reference,'requested_norm_relative_error':requested_error,
            'effective_norm_relative_error':effective_error,'effective_projection_difference':projection_difference,
            'reference_effective_norm':ref['effective_induced_norm'],
            'effective_projection_error_over_full_norm':abs(projection_difference)/max(fn,AXIS_EPS) if projection_difference is not None else None})
    closure=float(component_saved.loc[idx,'effective_component_closure_norm'])
    close(closure/max(fn,AXIS_EPS),float(component_saved.loc[idx,'effective_component_closure_relative']),'Recorded BF16 closure ratio')
    amplitude_rows.append({**base,'effective_A_norm':an,'effective_R_norm':rn,'effective_full_norm':fn,
        'A_over_full_norm':an/fn if fn>AXIS_EPS else None,'zero_effective_A':an<=AXIS_EPS,
        'A_quantization_error_relative':a['quantization_error_norm']/max(a['requested_induced_norm'],AXIS_EPS),
        'closure_relative':closure/max(fn,AXIS_EPS)})
matching=pd.DataFrame(match_rows)
compare_table(matching,read_table('control_matching_audit'),PAIR+['kind'],fields=['reference','requested_norm_relative_error','effective_norm_relative_error','effective_projection_difference'],label='Saved norm matching')
quality=pair_meta.join(pd.DataFrame(amplitude_rows).set_index(PAIR),how='left',validate='one_to_one')
norm_w=matching.pivot(index=PAIR,columns='kind',values='effective_norm_relative_error').reindex(quality.index)
require(norm_w.notna().all().all(),'Missing norm controls.')
quality['A_control_max_error']=norm_w[['random_aligned_0','random_aligned_1']].max(axis=1)
quality['R_control_max_error']=norm_w[['rotated_residual_0','rotated_residual_1']].max(axis=1)
quality['all_control_max_error']=norm_w.max(axis=1)
quality['amplitude_bin']=pd.cut(quality.A_over_full_norm,bins=AMPLITUDE_BINS,right=False,labels=['[0,.01)','[.01,.1)','[.1,1)','[1,inf)']).astype('object')

MATCH_MASKS={'all':pd.Series(True,index=quality.index)}
for limit in NORM_LIMITS:MATCH_MASKS[f'norm_le_{limit:.2f}']=quality.all_control_max_error.le(limit)
for limit in CLOSURE_LIMITS:MATCH_MASKS[f'closure_le_{limit:.2f}']=quality.closure_relative.le(limit)
MATCH_MASKS['norm_le_0.02_and_closure_le_0.02']=quality.all_control_max_error.le(.02)&quality.closure_relative.le(.02)
MATCH_MASKS['stable_unique_top']=quality.stable_unique_top.eq(True)
MATCH_MASKS['stable_unique_top_and_norm_le_0.02']=quality.stable_unique_top.eq(True)&quality.all_control_max_error.le(.02)
quality['all']=True
for name,mask in MATCH_MASKS.items():quality[name]=mask

def evaluate_mask(w,metadata,mask,name,context,metric):
    require(mask.index.equals(w.index) and metadata.index.equals(w.index),'Mask/pair index misalignment.')
    c=all_contrasts(w)[FOCUS];ids=mask.index[mask]
    fam_ids=set(ids.get_level_values('family_id'));reference=c[c.index.get_level_values('family_id').isin(fam_ids)]
    kept_meta=metadata.loc[ids]
    supports={'receiver_context':context,'subset':name,'metric':metric,'paired_records':len(ids),
        'distinct_receiver_inputs':len(set(ids.get_level_values('input_key'))),'supported_families':len(fam_ids),
        'total_pairs':len(w),'retained_fraction':len(ids)/len(w) if len(w) else None,
        'negative_coefficient_pairs':int(kept_meta.signed_coefficient.lt(0).sum()),'zero_effective_A_pairs':int(kept_meta.zero_effective_A.sum())}
    rows=[]
    for contrast in FOCUS:
        for status,selected in [('retained',c.loc[ids,contrast]),('excluded',c.loc[~mask,contrast]),('unfiltered_same_families',reference[contrast])]:
            fam=selected.groupby('family_id').mean()
            rows.append({'receiver_context':context,'subset':name,'metric':metric,'contrast':contrast,'population':status,
                'paired_records':len(selected),**estimate(fam,['sensitivity',context,name,metric,contrast,status])})
        kept=c.loc[ids,contrast].groupby('family_id').mean();unfiltered=reference[contrast].groupby('family_id').mean()
        diff=kept-unfiltered
        rows.append({'receiver_context':context,'subset':name,'metric':metric,'contrast':contrast,'population':'retained_minus_unfiltered_same_families',
            'paired_records':len(ids),**estimate(diff,['selection_change',context,name,metric,contrast])})
    return rows,supports

sensitivity_rows=[];support_rows=[];retention_rows=[]
for context in ('flip','stable'):
    for metric in ('margin_shift','cf_accuracy_gain'):
        w=paired(context,metric);meta=quality.loc[w.index]
        for name,global_mask in MATCH_MASKS.items():
            mask=global_mask.reindex(w.index);require(not mask.isna().any(),'Missing sensitivity mask.')
            rows,support=evaluate_mask(w,meta,mask,name,context,metric);sensitivity_rows.extend(rows);support_rows.append(support)
            if metric=='margin_shift':
                for fid,part in mask.groupby('family_id'):
                    retention_rows.append({'receiver_context':context,'subset':name,'family_id':fid,'kept_pairs':int(part.sum()),'total_pairs':len(part),'retained_fraction':float(part.mean())})
sensitivity=pd.DataFrame(sensitivity_rows);support=pd.DataFrame(support_rows);family_retention=pd.DataFrame(retention_rows)
amplitude_summary=[]
for (context,bin_name),meta in quality.groupby(['receiver_context','amplitude_bin'],observed=True):
    w=paired(context,'margin_shift').loc[meta.index]
    for name,vals in all_contrasts(w)[['aligned_vs_zero','aligned_vs_random_aligned']].items():
        amplitude_summary.append({'receiver_context':context,'amplitude_bin':str(bin_name),'contrast':name,'paired_records':len(meta),
            'median_A_control_error':float(meta.A_control_max_error.median()),'max_A_control_error':float(meta.A_control_max_error.max()),
            **estimate(vals.groupby('family_id').mean(),['amplitude',context,str(bin_name),name])})
amplitude_summary=pd.DataFrame(amplitude_summary)
tail_summary=matching.groupby('kind').agg(records=('effective_norm_relative_error','size'),mean_relative_error=('effective_norm_relative_error','mean'),
    median_relative_error=('effective_norm_relative_error','median'),p95_relative_error=('effective_norm_relative_error',lambda x:x.quantile(.95)),max_relative_error=('effective_norm_relative_error','max')).reset_index()
for name,table in [('recomputed_control_matching',matching),('pair_quality_and_masks',quality.reset_index()),('sensitivity_effects',sensitivity),
    ('sensitivity_support',support),('family_retention',family_retention),('amplitude_strata',amplitude_summary),('norm_error_summary',tail_summary)]:save(table,name)
print('Fixed 1%, 2%, 5% thresholds. All outputs retained; no threshold chosen by benefit.')
show(tail_summary)
show(support[(support.receiver_context=='flip')&(support.metric=='margin_shift')])
show(sensitivity[(sensitivity.receiver_context=='flip')&(sensitivity.metric=='margin_shift')&(sensitivity.contrast=='aligned_vs_random_aligned')&(sensitivity.population=='retained')])
```

```python
%%cell_progress 11/13 Pair downstream distance changes with output changes
cell_stage('Pair downstream distance changes with output changes')
geometry_rows=[]
for (context,layer,kind),sub in traces.groupby(['receiver_context','layer','kind']):
    for metric in ('cosine_to_natural_cf','cf_aligned_fraction','relative_distance_to_cf','closeness_gain_over_clean'):
        valid=sub.dropna(subset=[metric]);fam=valid.groupby('family_id')[metric].mean()
        geometry_rows.append({'receiver_context':context,'layer':int(layer),'kind':kind,'metric':metric,
            'valid_pairs':len(valid),'eligible_pairs':len(sub),'eligible_families':sub.family_id.nunique(),
            **original_estimate(fam,['geometry',context,int(layer),kind,metric])})
geometry_summary=pd.DataFrame(geometry_rows)
compare_table(geometry_summary,read_table('state_geometry_summary'),['receiver_context','layer','kind','metric'],label='Original state geometry statistics')

def geometry_pairs(context,layer):
    g=traces[(traces.receiver_context==context)&(traces.layer==layer)]
    w=g[g.context>=0].pivot(index=PAIR,columns='kind',values='relative_distance_to_cf').sort_index()
    z=g[g.kind=='zero'].set_index('input_key').relative_distance_to_cf
    w['zero']=[z.loc[k] for k in w.index.get_level_values('input_key')]
    return w

state_pairs=[];state_families=[];joint_rows=[];joint_counts=[];association_rows=[]
STATE_COMPARISONS=[('aligned','zero'),('aligned','full'),('aligned_half_residual','full'),('half','full')]
for context,layer in itertools.product(('flip','stable'),(49,55)):
    distance=geometry_pairs(context,layer);margins=paired(context,'margin_shift');accuracy=paired(context,'cf_accuracy_gain')
    require(distance.index.equals(margins.index) and accuracy.index.equals(margins.index),'State/output pairing changed.')
    for new,reference in STATE_COMPARISONS:
        label=new+'_minus_'+reference
        data=pd.DataFrame({'distance_reduction':distance[reference]-distance[new],
            'margin_difference':margins[new]-margins[reference],'accuracy_difference':accuracy[new]-accuracy[reference]})
        valid=data.dropna();families=valid.groupby('family_id').mean()
        for metric in data.columns:
            state_pairs.append({'receiver_context':context,'layer':layer,'contrast':label,'metric':metric,'valid_pairs':len(valid),'eligible_pairs':len(data),
                **estimate(families[metric],['paired_state',context,layer,label,metric])})
            state_families.extend({'receiver_context':context,'layer':layer,'contrast':label,'metric':metric,'family_id':fid,'value':float(v)} for fid,v in families[metric].items())
        # Outcome quadrants summarize the entire measured set; never filter the primary tests.
        def direction(x,eps):return 'positive' if x>eps else 'negative' if x<-eps else 'zero'
        v=valid.copy();v['margin_direction']=[direction(x,1e-12) for x in v.margin_difference]
        v['distance_reduction_direction']=[direction(x,1e-9) for x in v.distance_reduction]
        for ms,ds in itertools.product(('positive','zero','negative'),repeat=2):
            hit=v.margin_direction.eq(ms)&v.distance_reduction_direction.eq(ds)
            rates=hit.groupby('family_id').mean()
            joint_rows.append({'receiver_context':context,'layer':layer,'contrast':label,'margin_direction':ms,'distance_reduction_direction':ds,
                'valid_pairs':len(valid),'eligible_pairs':len(data),'raw_pair_count':int(hit.sum()),
                **estimate(rates,['quadrant',context,layer,label,ms,ds])})
        rho=(float(families.distance_reduction.rank().corr(families.margin_difference.rank()))
            if len(families)>=3 and families.distance_reduction.nunique()>1 and families.margin_difference.nunique()>1 else None)
        association_rows.append({'receiver_context':context,'layer':layer,'contrast':label,'n_families':len(families),
            'spearman_descriptive':rho,'scope':'family associations, not mediation; stable signed margin is not accuracy'})
state_pairs=pd.DataFrame(state_pairs);state_families=pd.DataFrame(state_families);joint=pd.DataFrame(joint_rows);associations=pd.DataFrame(association_rows)
for name,table in [('reproduced_geometry_summary',geometry_summary),('paired_state_output_effects',state_pairs),('paired_state_family_values',state_families),
    ('state_output_joint_quadrants',joint),('state_output_family_associations',associations)]:save(table,name)
print('Paired distance reduction: positive means closer than the reference. L48 construction is not new evidence.')
show(state_pairs[(state_pairs.receiver_context=='flip')&(state_pairs.metric=='distance_reduction')])
show(joint[(joint.receiver_context=='flip')&(joint.margin_direction=='positive')&(joint.distance_reduction_direction=='positive')])
```
