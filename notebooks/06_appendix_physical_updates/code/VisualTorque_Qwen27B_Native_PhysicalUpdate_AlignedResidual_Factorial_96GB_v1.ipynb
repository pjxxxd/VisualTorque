```python
# Run once. Every later code cell reports both cell and inner-loop progress.
import re
import threading
import time
from IPython.core.magic import register_cell_magic
from tqdm.auto import tqdm

_SETUP_PROGRESS = tqdm(total=1, desc='Initialize progress display', unit='cell')
_ACTIVE_CELL_PROGRESS = None

def cell_stage(message):
    if _ACTIVE_CELL_PROGRESS is not None:
        _ACTIVE_CELL_PROGRESS['stage'] = str(message)
        elapsed = time.monotonic() - _ACTIVE_CELL_PROGRESS['started']
        _ACTIVE_CELL_PROGRESS['bar'].set_postfix_str(
            f"{message} | elapsed {elapsed:.0f}s", refresh=True
        )

@register_cell_magic
def cell_progress(line, cell):
    global _ACTIVE_CELL_PROGRESS
    specification = line.strip() or 'Cell progress'
    match = re.fullmatch(r'(\d+)/(\d+)\s+(.+)', specification)
    if match:
        current, total, label = int(match.group(1)), int(match.group(2)), match.group(3)
        initial = current - 1
    else:
        current, total, label, initial = 1, 1, specification, 0
    started = time.monotonic()
    bar = tqdm(total=total, initial=initial, desc=label, unit='cell',
               dynamic_ncols=True, leave=True)
    state = {'bar': bar, 'started': started, 'stage': 'starting'}
    _ACTIVE_CELL_PROGRESS = state
    stopped = threading.Event()

    def heartbeat():
        while not stopped.wait(1.0):
            elapsed = time.monotonic() - started
            bar.set_postfix_str(
                f"{state['stage']} | elapsed {elapsed:.0f}s", refresh=True
            )

    worker = threading.Thread(target=heartbeat, daemon=True)
    worker.start()
    succeeded = False
    try:
        cell_stage('running')
        result = get_ipython().run_cell(cell)
        result.raise_error()
        succeeded = True
        return result.result
    finally:
        stopped.set()
        worker.join(timeout=2.0)
        elapsed = time.monotonic() - started
        if succeeded:
            bar.n = current
            bar.set_postfix_str(f"complete | elapsed {elapsed:.0f}s", refresh=True)
        else:
            bar.set_postfix_str(f"failed | elapsed {elapsed:.0f}s", refresh=True)
        bar.close()
        _ACTIVE_CELL_PROGRESS = None

_SETUP_PROGRESS.update(1)
_SETUP_PROGRESS.close()
print('Cell progress helper ready: 15 tracked cells.')
```

```python
%%cell_progress 6/15 Load Native Qwen27B with the parent numerical policy
cell_stage('Load Native Qwen27B with the parent numerical policy')
assert_native_precision()
# Release layer references too: they otherwise keep the previous 27B decoder on GPU when Run all is repeated.
for stale_name in ('DECODER_LAYERS','adapter'):
    globals().pop(stale_name,None)
gc.collect();torch.cuda.empty_cache()
adapter=HuggingFaceMultimodalAdapter(MODEL_ID,revision=MODEL_REVISION,dtype='bfloat16',device_map='cuda:0')
adapter.model.eval()
def resolve_module(root,path):
    for key in path.split('.'):root=getattr(root,key)
    return root
def discover_layers(model):
    for path in ('model.language_model.layers','model.text_model.layers','language_model.model.layers','model.model.layers','model.layers'):
        try:layers=resolve_module(model,path)
        except AttributeError:continue
        if isinstance(layers,torch.nn.ModuleList) and len(layers)==64:return path,layers
    raise RuntimeError('Expected 64-layer Qwen3.5 decoder not found.')
LAYER_PATH,DECODER_LAYERS=discover_layers(adapter.model)
require(all(p.dtype==torch.bfloat16 for p in adapter.model.parameters() if p.ndim>=2),'Base weights are not all BF16; no quantization is allowed.')
implementations=sorted({f'{rule.__module__}.{rule.__name__}' for module in adapter.model.modules()
                       for rule in [getattr(module,'chunk_gated_delta_rule',None)] if rule is not None})
paths={inspect.getsourcefile(type(m)) for m in adapter.model.modules() if 'qwen3_5' in type(m).__module__}
runtime={'torch':torch.__version__,'transformers':importlib.metadata.version('transformers'),'peft':importlib.metadata.version('peft'),
    'numpy':np.__version__,'pandas':pd.__version__,'pillow':importlib.metadata.version('pillow'),'cuda':torch.version.cuda,
    'gpu':torch.cuda.get_device_name(0),'precision':assert_native_precision(),'batch_size':1,
    'decoder_path':LAYER_PATH,'attention':getattr(adapter.model.config,'_attn_implementation',None),
    'linear_attention':implementations,'model_source_hashes':{Path(p).name:sha(p) for p in paths if p and Path(p).is_file()},
    'synthetic_only':SYNTHETIC_ONLY}
RUNTIME_DIGEST=freeze(OUTPUT_ROOT/'runtime_lock.json',runtime)
RUN_ID=digest({'protocol':PROTOCOL_DIGEST,'runtime':RUNTIME_DIGEST,'model':MODE_IDENTITY})
freeze(OUTPUT_ROOT/'run_identity.json',{'run_id':RUN_ID})
print('Model loaded:',MODEL_MODE,'| run:',RUN_ID,'| linear attention:',implementations)
```

```python
%%cell_progress 7/15 Define exact scoring and BF16 vector audits
cell_stage('Define exact scoring and BF16 vector audits')
INJECTION_AUDIT_SCHEMA='bf16-injection-vector-audit-2.0'
NUMERICAL_AUDIT_PATCH=IMPLEMENTATION
def audit_tensor_hash(t):
    t=t.detach().cpu().contiguous()
    return digest({'shape':list(t.shape),'dtype':str(t.dtype),
        'bytes_sha256':hashlib.sha256(t.view(torch.uint8).numpy().tobytes()).hexdigest()})

def audit_norm64(t):
    return float(torch.linalg.vector_norm(t.detach().cpu().double()))

def audit_injection(result,clean,payload,dose,input_key,condition_id):
    # Retain the actual pre-hook state: the post-hook state alone cannot distinguish
    # a correct addition from replay drift relative to the cached clean state.
    before=result['injection_before'].detach().cpu()
    cached=clean['states'][INJECTION_LAYER].detach().cpu()
    after=result['states'][INJECTION_LAYER].detach().cpu()
    cast=payload.detach().to(device='cpu',dtype=torch.bfloat16)
    require(before.shape==cached.shape==after.shape==cast.shape,'Injection audit: tensor shape mismatch.')
    require(before.dtype==cached.dtype==after.dtype==cast.dtype==torch.bfloat16,'Injection audit: expected BF16 tensors.')
    require(all(torch.isfinite(t).all() for t in (before,cached,after,cast)),'Injection audit: nonfinite tensor.')
    planned_after=cached+cast
    live_expected_after=before+cast
    actual_delta=after.float()-before.float()
    expected_delta=planned_after.float()-cached.float()
    actual_norm=norm(actual_delta)  # same CPU FP32 reporting convention as frozen v1 doses
    frozen_norm=float(dose['expected_actual_added_norm'])
    device_norm=float(result['edit']['actual_added_norm_device_fp32'])
    checks={'prestate_equal':torch.equal(before,cached),
        'hook_addition_equal':torch.equal(after,live_expected_after),
        'effective_delta_equal':torch.equal(actual_delta,expected_delta),
        'frozen_cpu_norm_equal':abs(norm(expected_delta)-frozen_norm)<1e-5}
    audit={'schema':INJECTION_AUDIT_SCHEMA,'patch_code_digest':NUMERICAL_AUDIT_PATCH,
        'passed':all(checks.values()),**checks,
        'cached_before_sha256':audit_tensor_hash(cached),'live_before_sha256':audit_tensor_hash(before),
        'expected_after_sha256':audit_tensor_hash(planned_after),'actual_after_sha256':audit_tensor_hash(after),
        'expected_delta_sha256':audit_tensor_hash(expected_delta),'actual_delta_sha256':audit_tensor_hash(actual_delta),
        'prestate_max_abs_error':float((before.float()-cached.float()).abs().max()),
        'hook_max_abs_error':float((after.float()-live_expected_after.float()).abs().max()),
        'delta_max_abs_error':float((actual_delta-expected_delta).abs().max()),
        'actual_norm_cpu_fp32':actual_norm,'expected_norm_cpu_fp32':norm(expected_delta),
        'actual_norm_cpu_fp64':audit_norm64(actual_delta),'expected_norm_cpu_fp64':audit_norm64(expected_delta),
        'frozen_expected_norm':frozen_norm,'device_norm_fp32':device_norm,
        'device_vs_frozen_abs_error':abs(device_norm-frozen_norm),
        'legacy_scalar_gate_would_pass':abs(device_norm-frozen_norm)<1e-5}
    if not audit['passed']:
        folder=OUTPUT_ROOT/'analysis/injection_diagnostics';folder.mkdir(parents=True,exist_ok=True)
        stem=input_key+'__'+condition_id
        tensor_path=folder/(stem+'.pt');json_path=folder/(stem+'.json')
        torch_atomic(tensor_path,{'cached_before':cached,'live_before':before,'actual_after':after,
            'payload_fp32':payload.detach().cpu(),'cast_payload':cast,'planned_after':planned_after})
        failed=[k for k,v in checks.items() if not v]
        write_json(json_path,{'run_id':RUN_ID,'input_key':input_key,'condition_id':condition_id,
            'failed_checks':failed,'audit':audit,'tensor_sha256':sha(tensor_path)})
        raise ValueError('Injection vector audit failed for '+input_key+' / '+condition_id+
            ': '+', '.join(failed)+'. Diagnostics: '+str(json_path)+
            '. Frozen doses and completed records were kept. Do not loosen the gate or delete locks.')
    result['edit']['actual_added_norm']=actual_norm
    return audit

def validate_injection_record(record):
    actual=float(record['edit']['actual_added_norm']);expected=float(record['dose']['expected_actual_added_norm'])
    require(math.isfinite(actual) and actual>=0 and math.isfinite(expected) and expected>=0,'Invalid saved injection norm.')
    require(abs(actual-expected)<1e-5,'Saved injection does not pass its frozen CPU/v1 norm gate.')
    audit=record.get('injection_audit')
    require(audit is not None,'New diagnostic records require the full vector audit.')
    require(audit['schema']==INJECTION_AUDIT_SCHEMA and audit['patch_code_digest']==NUMERICAL_AUDIT_PATCH,'Unknown injection audit version.')
    require(all(audit[k] is True for k in ('passed','prestate_equal','hook_addition_equal','effective_delta_equal','frozen_cpu_norm_equal')),'Failed saved vector checks.')
    for a,b in (('cached_before_sha256','live_before_sha256'),('expected_after_sha256','actual_after_sha256'),('expected_delta_sha256','actual_delta_sha256')):
        require(audit[a]==audit[b],'Saved injection vector hashes disagree.')
    require(all(audit[k]==0 for k in ('prestate_max_abs_error','hook_max_abs_error','delta_max_abs_error')),'Nonzero saved vector error.')
    require(audit['actual_norm_cpu_fp32']==actual and audit['expected_norm_cpu_fp32']==actual and audit['frozen_expected_norm']==expected,'Saved audit norm mismatch.')
    require(audit['actual_norm_cpu_fp64']==audit['expected_norm_cpu_fp64'] and math.isfinite(audit['actual_norm_cpu_fp64']),'Saved FP64 vector norm mismatch.')
    device=float(record['edit']['actual_added_norm_device_fp32'])
    require(math.isfinite(device) and device>=0 and audit['device_norm_fp32']==device,'Invalid device norm diagnostic.')
    require(audit['device_vs_frozen_abs_error']==abs(device-expected) and audit['legacy_scalar_gate_would_pass']==(abs(device-expected)<1e-5),'Device diagnostic mismatch.')
    return 'v2_exact_vectors'

def prepare(example):
    plan=adapter._candidate_plan(example)
    require(plan is not None,'A/B/C must be single-token candidates.')
    prefix,token_by_label=plan
    inputs=adapter._append_text_prefix(adapter._to_device(adapter._template([example.messages()],add_generation_prompt=True)),prefix)
    require(inputs['input_ids'].shape[0]==1,'All forwards use batch size one.')
    require(inputs['attention_mask'][0,-1].item()==1,'Final token must be unpadded.')
    fingerprint=digest({'ids':inputs['input_ids'].detach().cpu().tolist(),'attention':inputs['attention_mask'].detach().cpu().tolist(),
                       'token_by_label':{k:int(v) for k,v in token_by_label.items()}})
    return inputs,token_by_label,fingerprint
def hidden_tensor(output):
    if torch.is_tensor(output):return output
    if isinstance(output,(tuple,list)) and torch.is_tensor(output[0]):return output[0]
    raise TypeError('Unsupported decoder output: '+str(type(output)))
def replace_hidden(output,value):
    if torch.is_tensor(output):return value
    if isinstance(output,tuple):return (value,)+output[1:]
    if isinstance(output,list):return [value]+output[1:]
    raise TypeError(type(output))
def centered_score_error(a,b):
    a=np.asarray(a,dtype=float);b=np.asarray(b,dtype=float)
    return float(np.max(np.abs((a-a.mean())-(b-b.mean()))))
def forward(example,payload=None,expected=None):
    assert_native_precision();inputs,token_by_label,fingerprint=prepare(example)
    if expected is not None:require(fingerprint==expected,'Prepared input or candidate token identity changed.')
    states={};edit={};handles=[];injection_before={}
    def make_hook(layer):
        def hook(module,args,output):
            h=hidden_tensor(output);require(h.dtype==torch.bfloat16 and h.ndim==3,'Expected BF16 decoder state.')
            before=h[:, -1, :].detach().clone()
            if layer==INJECTION_LAYER and payload is not None:
                add=payload.to(device=h.device,dtype=h.dtype);require(add.shape==before.shape,'Payload shape mismatch.')
                after=before+add;changed=h.clone();changed[:,-1,:]=after
                injection_before['tensor']=before.cpu()
                edit.update(bf16_payload_norm=norm(add),actual_added_norm=norm((after.float()-before.float()).cpu()),
                            actual_added_norm_device_fp32=norm(after.float()-before.float()),
                            requested_fp32_norm=norm(payload),actual_cosine_to_requested=cosine(after.float()-before.float(),add.float()))
                states[layer]=after.cpu();return replace_hidden(output,changed)
            states[layer]=before.cpu()
        return hook
    try:
        for layer in CAPTURE_LAYERS:handles.append(DECODER_LAYERS[layer].register_forward_hook(make_hook(layer)))
        with torch.inference_mode():out=adapter.model(**inputs,use_cache=False)
        scores=torch.log_softmax(out.logits[0,-1].float(),dim=-1)
        values=[float(scores[token_by_label[k]].cpu()) for k in LABELS]
    finally:
        for h in handles:h.remove()
    require(set(states)==set(CAPTURE_LAYERS) and all(torch.isfinite(t).all() for t in states.values()),'Missing/nonfinite captured states.')
    require(np.isfinite(values).all(),'Nonfinite candidate scores.')
    idx=int(np.argmax(values));record={'candidate_labels':list(LABELS),'candidate_semantics':list(example.option_semantics),
        'candidate_logprobs':values,'predicted_label':LABELS[idx],'predicted_semantic':example.option_semantics[idx]}
    result={'score':record,'states':states,'fingerprint':fingerprint,'edit':edit}
    if payload is not None:
        require('tensor' in injection_before,'Injection hook was not called.')
        result['injection_before']=injection_before['tensor']
    return result
def clean_paths(key):return OUTPUT_ROOT/'clean_cache'/(key+'.pt'),OUTPUT_ROOT/'clean_records'/(key+'.json')
def load_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key);meta=read_json(jp)
    require(meta['run_id']==RUN_ID and meta['capture_key']==key and sha(tp)==meta['tensor_sha256'],'Clean capture identity/hash mismatch: '+key)
    data=torch.load(tp,map_location='cpu',weights_only=True)
    require(data['run_id']==RUN_ID and data['capture_key']==key,'Wrong tensor identity.')
    require(set(data['states'])==set(CAPTURE_LAYERS) and all(t.dtype==torch.bfloat16 and torch.isfinite(t).all() and norm(t)>0 for t in data['states'].values()),'Invalid clean tensor.')
    require(data['score']==meta['score'] and data['fingerprint']==meta['fingerprint'],'Clean metadata mismatch.')
    return data
def get_or_capture(row,mid):
    key=capture_key(row,mid);tp,jp=clean_paths(key)
    if jp.is_file():return load_capture(row,mid)
    result=forward(make_example(row,mid));result.update(run_id=RUN_ID,capture_key=key)
    torch_atomic(tp,result)
    write_json(jp,{k:result[k] for k in ('run_id','capture_key','score','fingerprint')}|{'tensor_sha256':sha(tp)})
    return result
def semantic_scores(record):return dict(zip(record['candidate_semantics'],record['candidate_logprobs']))
```

```python
%%cell_progress 9/15 Freeze signed components and norm-matched controls
cell_stage('Freeze signed components and norm-matched controls')
def norm64(t):return float(torch.linalg.vector_norm(t.detach().cpu().double()))

def weight_payload(source_before,source_after,receiver_before):
    # No target after state, label, prediction or outcome is accepted here.
    require(source_before.shape==source_after.shape==receiver_before.shape,'Donor/receiver dimensions differ.')
    require(norm(source_before)>AXIS_EPS,'Zero donor baseline norm.')
    return (source_after.float()-source_before.float())*(norm(receiver_before)/norm(source_before))

def split_update(full,natural):
    v,g=full.double(),natural.double();gn=norm64(g)
    defined=gn>AXIS_EPS
    coefficient=float((v*g).sum()/(g*g).sum()) if defined else 0.
    a=coefficient*g if defined else torch.zeros_like(v);r=v-a
    scale=max(norm64(v),AXIS_EPS)
    require(norm64(v-a-r)<=1e-10*scale,'FP64 component closure failed.')
    require(not defined or abs(float((r*g).sum()))<=1e-10*max(norm64(r)*gn,1.),'FP64 residual orthogonality failed.')
    return a.float(),r.float(),{'axis_defined':defined,'signed_coefficient':coefficient,'full_norm64':norm64(v),
        'aligned_norm64':norm64(a),'residual_norm64':norm64(r),'natural_norm64':gn,
        'fp32_closure_relative_error':norm64(full-a.float()-r.float())/scale,
        'fp32_residual_projection':float((r.float().double()*g).sum()/gn) if defined else None}

def random_matched(shape,magnitude,seed,orthogonal_to=None):
    if magnitude<=AXIS_EPS:return torch.zeros(shape,dtype=torch.float32)
    generator=torch.Generator(device='cpu').manual_seed(seed)
    x=torch.randn(shape,generator=generator,dtype=torch.float64)
    if orthogonal_to is not None and norm64(orthogonal_to)>AXIS_EPS:
        q=orthogonal_to.double();x=x-(x*q).sum()/(q*q).sum()*q
    require(norm64(x)>AXIS_EPS,'Degenerate random control; no silent redraw.')
    return (x*(magnitude/norm64(x))).float()

def endpoint_flags(before,after,truth):
    a,b=before['score'],after['score']
    stable=a['predicted_semantic']==b['predicted_semantic']
    unique=all(np.sum(np.asarray(s['candidate_logprobs'])==max(s['candidate_logprobs']))==1 for s in (a,b))
    return {'prediction_stable':bool(stable),'stable_unique_top':bool(stable and unique),
        'source_both_correct':bool(a['predicted_semantic']==b['predicted_semantic']==truth),
        'source_confidence_change':float(sorted(b['candidate_logprobs'])[-1]-sorted(b['candidate_logprobs'])[-2]
                                      -sorted(a['candidate_logprobs'])[-1]+sorted(a['candidate_logprobs'])[-2])}

DONORS={}
for f,ori,context,slot in itertools.product(families,('anchor','mirror'),range(2),range(2)):
    fid=f['family_id'];mid=source_mappings(fid,ori,context)[slot]
    rb=SCENE_LOOKUP[(fid,ori,f'source{context}_before')];ra=SCENE_LOOKUP[(fid,ori,f'source{context}_after')]
    before,after=cc(rb,mid),cc(ra,mid)
    require(rb['outcome']==ra['outcome'],'Parent source is not answer-neutral.')
    DONORS[(fid,ori,context,slot)]={'before':before['states'][48],'after':after['states'][48],
        'source_semantic':rb['outcome'],'source_label':LABELS[PERMUTATIONS[mid].index(rb['outcome'])],
        'source_mapping':mid,'source_capture_keys':[capture_key(rb,mid),capture_key(ra,mid)],
        **endpoint_flags(before,after,rb['outcome'])}

@lru_cache(maxsize=4096)
def local_components(input_key,context,slot):
    row,mid=CAPTURE_LOOKUP[input_key];own=cc(row,mid)['states'][48]
    after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    natural=cc(after_row,mid)['states'][48].float()-own.float()
    donor=DONORS[(row['family_id'],row['orientation'],context,slot)]
    full=weight_payload(donor['before'],donor['after'],own)
    a,r,meta=split_update(full,natural)
    source={k:donor[k] for k in ('source_semantic','source_label','source_mapping','source_capture_keys','prediction_stable','stable_unique_top','source_both_correct','source_confidence_change')}
    return full,a,r,natural,{**source,**meta}

def make_payload(row,mid,plan,clean):
    own=clean['states'][48];key=capture_key(row,mid);kind=plan['kind']
    if plan['context']<0:
        after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
        g=cc(after_row,mid)['states'][48].float()-own.float()
        p=torch.zeros_like(own,dtype=torch.float32) if kind=='zero' else g*(.5 if kind=='cf_half' else 1.)
        return p,{'axis_defined':norm64(g)>AXIS_EPS,'natural_norm64':norm64(g),'source_capture_keys':[],
            'source_semantic':None,'source_label':None,'source_mapping':None,'prediction_stable':None,'stable_unique_top':None,
            'source_both_correct':None,'source_confidence_change':None,'signed_coefficient':None}
    v,a,r,g,meta=local_components(key,plan['context'],plan['source_mapping_slot'])
    if kind=='full':p=v
    elif kind=='half':p=.5*v
    elif kind=='aligned':p=a
    elif kind=='residual':p=r
    elif kind=='aligned_half_residual':p=a+.5*r
    elif kind.startswith('rotated_residual_'):
        seed=int(stable_key(['rotate_R',key,plan['context'],plan['source_mapping_slot'],int(kind[-1])])[:15],16)
        p=a+random_matched(r.shape,norm64(r),seed,g)
    elif kind.startswith('random_aligned_'):
        seed=int(stable_key(['random_A',key,plan['context'],plan['source_mapping_slot'],int(kind[-1])])[:15],16)
        p=random_matched(a.shape,norm64(a),seed)
    else:raise ValueError('Unknown condition '+kind)
    return p.contiguous(),meta

def displacement_geometry(delta,natural):
    d,g=delta.double(),natural.double();dn,gn=norm64(d),norm64(g)
    defined=gn>AXIS_EPS
    projection=float((d*g).sum()/gn) if defined else None
    return {'induced_norm':dn,'natural_cf_norm':gn,
        'projection_on_natural_cf':projection,'cosine_to_natural_cf':projection/dn if defined and dn>AXIS_EPS else None,
        'cf_aligned_fraction':projection/gn if defined else None,
        'relative_distance_to_cf':norm64(d-g)/gn if defined else None,
        'closeness_gain_over_clean':1-norm64(d-g)/gn if defined else None}

def tensor_fingerprint(t):return 'sha256:'+hashlib.sha256(t.contiguous().numpy().tobytes()).hexdigest()

ALL_CLEAN_INVENTORY={capture_key(r,m):sha(clean_paths(capture_key(r,m))[1]) for r,m in capture_inputs}
require(ALL_CLEAN_INVENTORY==CLEAN_INVENTORY,'Clean inventory changed.')
ALL_CLEAN_DIGEST=freeze(OUTPUT_ROOT/'all_clean_inventory.json',ALL_CLEAN_INVENTORY)
dose_rows=[];component_rows=[];matching_rows=[]
for row,mid in tqdm(TARGET_INPUTS,desc='Freeze components and effective BF16 geometry',unit='input'):
    clean=cc(row,mid);own=clean['states'][48];key=capture_key(row,mid)
    after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    g=cc(after_row,mid)['states'][48].float()-own.float()
    per_input={}
    for plan in PLAN:
        p,meta=make_payload(row,mid,plan,clean);cast=p.to(torch.bfloat16)
        require(torch.isfinite(p).all() and torch.isfinite(cast).all(),'Nonfinite payload.')
        effective=(own+cast).float()-own.float()
        requested=displacement_geometry(p,g);realized=displacement_geometry(effective,g)
        dose={'input_key':key,'condition_id':plan['condition_id'],'payload_sha256':tensor_fingerprint(p),
            'requested_norm':norm(p),'bf16_norm':norm(cast),'relative_norm':norm(cast)/norm(own),
            'expected_actual_added_norm':norm(effective),'quantization_error_norm':norm64(effective-p),
            **meta,**{'requested_'+k:v for k,v in requested.items()},**{'effective_'+k:v for k,v in realized.items()}}
        dose_rows.append(dose);per_input[plan['condition_id']]=(p,effective,dose)
    for context,slot in itertools.product(range(2),range(2)):
        def item(kind):return per_input[f'{kind}__s{context}__m{slot}']
        v,a,r,g,meta=local_components(key,context,slot)
        closure=item('full')[1]-item('aligned')[1]-item('residual')[1]
        component_rows.append({'family_id':row['family_id'],'input_key':key,'receiver_context':row['role'].split('_')[0],
            'context':context,'source_mapping_slot':slot,**meta,'effective_component_closure_norm':norm64(closure),
            'effective_component_closure_relative':norm64(closure)/max(norm64(item('full')[1]),AXIS_EPS)})
        for kind,reference in [(f'rotated_residual_{s}','full') for s in range(2)]+[(f'random_aligned_{s}','aligned') for s in range(2)]:
            p,eff,d=item(kind);ref,refeff,rd=item(reference)
            req_error=abs(norm64(p)-norm64(ref))/max(norm64(ref),AXIS_EPS)
            require(req_error<2e-6,'Requested equal-norm control failed.')
            gp=d['requested_projection_on_natural_cf'];rp=rd['requested_projection_on_natural_cf']
            if kind.startswith('rotated') and gp is not None:
                require(abs(gp-rp)/max(norm64(ref),AXIS_EPS)<2e-6,'Requested residual rotation changed projection.')
            matching_rows.append({'family_id':row['family_id'],'input_key':key,'context':context,'source_mapping_slot':slot,
                'kind':kind,'reference':reference,'requested_norm_relative_error':req_error,
                'effective_norm_relative_error':abs(norm64(eff)-norm64(refeff))/max(norm64(refeff),AXIS_EPS),
                'effective_projection_difference':(d['effective_projection_on_natural_cf']-rd['effective_projection_on_natural_cf']) if gp is not None else None})
DOSE_DIGEST=freeze(OUTPUT_ROOT/'frozen_doses.json',dose_rows)
DOSES={(r['input_key'],r['condition_id']):r for r in dose_rows}
components=pd.DataFrame(component_rows);matching=pd.DataFrame(matching_rows)
pd.DataFrame(dose_rows).to_csv(OUTPUT_ROOT/'analysis/dose_inventory.csv',index=False)
components.to_csv(OUTPUT_ROOT/'analysis/component_inventory.csv',index=False)
matching.to_csv(OUTPUT_ROOT/'analysis/control_matching_audit.csv',index=False)
require(len(DOSES)==len(TARGET_INPUTS)*39 and len(components)==len(TARGET_INPUTS)*4,'Frozen component grid wrong.')
require(components.fp32_closure_relative_error.max()<2e-6,'FP32 component reconstruction changed.')
print('Undefined target axes (donor-receiver pairs):',int((~components.axis_defined).sum()),'of',len(components))
display(components[['signed_coefficient','aligned_norm64','residual_norm64','effective_component_closure_relative']].describe())
print('Norm matching is exact before BF16; realized mismatch is recorded, never silently corrected.')
display(matching.groupby('kind').agg(requested_norm_error_max=('requested_norm_relative_error','max'),effective_norm_error_mean=('effective_norm_relative_error','mean'),effective_norm_error_max=('effective_norm_relative_error','max')))
```

```python
%%cell_progress 11/15 Run all resumable component interventions
cell_stage('Run all resumable component interventions')
def shard_path(row,mid):return OUTPUT_ROOT/'intervention_shards'/(capture_key(row,mid)+'.json')
def shard_identity(row,mid):return {'run_id':RUN_ID,'all_clean_digest':ALL_CLEAN_DIGEST,'dose_digest':DOSE_DIGEST,'input_key':capture_key(row,mid)}
def validate_shard(shard,row,mid):
    require(shard['identity']==shard_identity(row,mid),'Shard identity mismatch.')
    ids=[r['condition']['condition_id'] for r in shard['records']]
    require(len(ids)==len(set(ids)) and set(ids)<={p['condition_id'] for p in PLAN},'Duplicate/unknown conditions.')
    plan_lookup={p['condition_id']:p for p in PLAN}
    for r in shard['records']:
        body={k:v for k,v in r.items() if k!='record_digest'}
        require(digest(body)==r['record_digest'],'Record checksum failed.')
        p=r['condition'];require(p==plan_lookup[p['condition_id']],'Condition metadata changed.')
        require(r['dose']==DOSES[(capture_key(row,mid),p['condition_id'])],'Frozen dose mismatch.')
        s=r['score'];require(s['candidate_labels']==list(LABELS) and s['candidate_semantics']==list(PERMUTATIONS[mid]),'Candidate map mismatch.')
        values=np.array(s['candidate_logprobs']);require(values.shape==(3,) and np.isfinite(values).all(),'Invalid candidate scores.')
        idx=int(values.argmax());require(s['predicted_label']==LABELS[idx] and s['predicted_semantic']==PERMUTATIONS[mid][idx],'Score/prediction mismatch.')
        require(set(r['traces'])==set(map(str,CAPTURE_LAYERS)),'Missing traces.')
        validate_injection_record(r)
    return shard

def run_one(row,mid):
    path=shard_path(row,mid)
    shard=validate_shard(read_json(path),row,mid) if path.is_file() else {'identity':shard_identity(row,mid),'records':[]}
    done={r['condition']['condition_id'] for r in shard['records']}
    if len(done)==len(PLAN):return
    clean=cc(row,mid);after_row=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    natural=cc(after_row,mid);e=make_example(row,mid)
    for p in PLAN:
        if p['condition_id'] in done:continue
        payload,_=make_payload(row,mid,p,clean);dose=DOSES[(capture_key(row,mid),p['condition_id'])]
        require('sha256:'+hashlib.sha256(payload.contiguous().numpy().tobytes()).hexdigest()==dose['payload_sha256'],'Payload changed after freezing.')
        result=forward(e,payload,expected=clean['fingerprint'])
        injection_audit=audit_injection(result,clean,payload,dose,capture_key(row,mid),p['condition_id'])
        traces={}
        for layer in CAPTURE_LAYERS:
            delta=result['states'][layer].float()-clean['states'][layer].float()
            natural_delta=natural['states'][layer].float()-clean['states'][layer].float()
            traces[str(layer)]=displacement_geometry(delta,natural_delta)
        record={'condition':p,'score':result['score'],'dose':dose,'edit':result['edit'],'traces':traces,'injection_audit':injection_audit}
        record['record_digest']=digest(record);shard['records'].append(record);write_json(path,shard)
    validate_shard(shard,row,mid)

def run_all():
    start=time.monotonic();new_done=0
    for i,(row,mid) in enumerate(tqdm(TARGET_INPUTS,desc=MODEL_MODE+': frozen transfer conditions',unit='input')):
        path=shard_path(row,mid);complete=path.is_file() and len(read_json(path).get('records',[]))==len(PLAN)
        run_one(row,mid)
        if not complete:
            new_done+=1
            if new_done in (1,8):print('Remaining ETA from completed inputs:',round((time.monotonic()-start)/new_done*(len(TARGET_INPUTS)-i-1)/3600,2),'hours')
run_all()
```

```python
%%cell_progress 12/15 Verify scores and directly measured state distances
cell_stage('Verify scores and directly measured state distances')
records=[];trace_rows=[];shard_hashes={}
for row,mid in tqdm(TARGET_INPUTS,desc='Verify completed grid',unit='input'):
    path=shard_path(row,mid);shard=validate_shard(read_json(path),row,mid)
    require(len(shard['records'])==len(PLAN),'Incomplete condition grid.')
    shard_hashes[path.name]=sha(path);clean=cc(row,mid)
    cfrow=SCENE_LOOKUP[(row['family_id'],row['orientation'],row['role'].replace('_before','_after'))]
    natural=cc(cfrow,mid);baseline=semantic_scores(clean['score']);ns=semantic_scores(natural['score'])
    changed=row['changed_side'];opposite='right' if changed=='left' else 'left';cfsem=cfrow['outcome']
    for record in shard['records']:
        s=record['score'];ss=semantic_scores(s);plan=record['condition'];dose=record['dose']
        base={'family_id':row['family_id'],'input_key':capture_key(row,mid),'scene_id':row['scene_id'],
            'orientation':row['orientation'],'receiver_context':row['role'].split('_')[0],'receiver_mapping':mid,**plan,
            'predicted_semantic':s['predicted_semantic'],'predicted_label':s['predicted_label'],
            'margin_shift':ss[changed]-ss[opposite]-baseline[changed]+baseline[opposite],
            'natural_margin_shift':ns[changed]-ns[opposite]-baseline[changed]+baseline[opposite],
            'cf_accuracy_gain':float(s['predicted_semantic']==cfsem)-float(clean['score']['predicted_semantic']==cfsem),
            'cf_success':s['predicted_semantic']==cfsem,'clean_cf_success':clean['score']['predicted_semantic']==cfsem,
            'natural_cf_correct':natural['score']['predicted_semantic']==cfsem,'clean_correct':clean['score']['predicted_semantic']==row['outcome'],
            'cf_best_other_gain':(ss[cfsem]-max(v for k,v in ss.items() if k!=cfsem))-(baseline[cfsem]-max(v for k,v in baseline.items() if k!=cfsem)),
            'axis_defined':dose['axis_defined'],'prediction_stable':dose['prediction_stable'],'stable_unique_top':dose['stable_unique_top'],
            'source_both_correct':dose['source_both_correct'],'signed_coefficient':dose['signed_coefficient'],
            'relative_payload_norm':dose['relative_norm'],'actual_added_norm':record['edit']['actual_added_norm'],
            'injection_audit_level':validate_injection_record(record),**{'logprob_'+k:v for k,v in zip(LABELS,s['candidate_logprobs'])}}
        records.append(base)
        for layer,t in record['traces'].items():
            # Direct distance is captured from FP64 vectors; also reconstruct it from scalar traces.
            n,m=t['induced_norm'],t['natural_cf_norm'];p=t['projection_on_natural_cf']
            require(math.isfinite(n) and math.isfinite(m) and n>=0 and m>=0,'Invalid trace norm.')
            if m>AXIS_EPS:
                sq=n*n+m*m-2*p*m
                require(sq>=-1e-9*max(n*n+m*m,1),'Impossible trace geometry.')
                require(math.isclose(math.sqrt(max(0,sq))/m,t['relative_distance_to_cf'],rel_tol=1e-7,abs_tol=1e-7),'Distance reconstruction failed.')
            trace_rows.append({k:base[k] for k in ('family_id','input_key','condition_id','kind','context','source_mapping_slot','receiver_context','orientation','receiver_mapping')}|{'layer':int(layer),**t})
frame=pd.DataFrame(records);traces=pd.DataFrame(trace_rows)
require(len(frame)==len(TARGET_INPUTS)*len(PLAN) and not frame.duplicated(['input_key','condition_id']).any(),'Incomplete/duplicate intervention rows.')
require(len(traces)==3*len(frame),'Trace count mismatch.')
require(np.isfinite(frame[['margin_shift','cf_accuracy_gain','cf_best_other_gain','actual_added_norm']].to_numpy()).all(),'Nonfinite scores.')
zero=frame[frame.kind=='zero']
require(zero.margin_shift.abs().max()<=SCORE_TOLERANCE and zero.actual_added_norm.max()==0,'Zero control failed.')
require(set(frame.injection_audit_level)=={'v2_exact_vectors'},'Every new record must have the vector audit.')
frame.to_csv(OUTPUT_ROOT/'analysis/intervention_records.csv',index=False)
traces.to_csv(OUTPUT_ROOT/'analysis/state_response_traces.csv',index=False)
freeze(OUTPUT_ROOT/'shard_inventory.json',shard_hashes)
print('Complete:',len(frame),'interventions;',len(traces),'traces; all actual BF16 additions verified.')
```

```python
%%cell_progress 13/15 Run six frozen family tests and descriptive boundaries
cell_stage('Run six frozen family tests and descriptive boundaries')
def estimate(values,key,test=False):
    values=pd.Series(values,dtype=float).dropna().sort_index()
    require(values.index.is_unique,'Statistics require one value per family.')
    if len(values)<2:return {'n_families':len(values),'estimate':float(values.mean()) if len(values) else None,'ci_low':None,'ci_high':None,'p_two_sided':None}
    x=values.to_numpy();require(np.isfinite(x).all(),'Invalid family effects.')
    rng=np.random.default_rng(int(stable_key(['stats',key])[:15],16));means=[]
    for start in range(0,BOOTSTRAPS,256):
        count=min(256,BOOTSTRAPS-start);means.extend(x[rng.integers(len(x),size=(count,len(x)))].mean(1))
    out={'n_families':len(x),'estimate':float(x.mean()),'ci_low':float(np.quantile(means,.025)),'ci_high':float(np.quantile(means,.975))}
    if test:
        exceed=0
        for start in range(0,SIGNFLIPS,1024):
            count=min(1024,SIGNFLIPS-start);signs=rng.integers(0,2,size=(count,len(x)))*2-1
            exceed+=int((np.abs((x*signs).mean(1))>=abs(x.mean())-1e-12).sum())
        out['p_two_sided']=(exceed+1)/(SIGNFLIPS+1)
    return out

def holm(pvalues):
    out=np.zeros(len(pvalues));running=0.
    for rank,i in enumerate(np.argsort(pvalues)):
        running=max(running,(len(pvalues)-rank)*pvalues[i]);out[i]=min(1.,running)
    return out

PAIR_INDEX=['family_id','input_key','context','source_mapping_slot']
def paired_values(context,metric):
    sub=frame[frame.receiver_context==context]
    w=sub[sub.context>=0].pivot(index=PAIR_INDEX,columns='kind',values=metric)
    require(set(w.columns)==set(PAIR_KINDS) and not w.isna().any().any(),'Incomplete paired conditions for '+metric)
    z=sub[sub.kind=='zero'].set_index('input_key')[metric]
    require(z.index.is_unique,'Duplicate zero inputs.')
    w['zero']=[float(z.loc[k]) for k in w.index.get_level_values('input_key')]
    w['random_aligned_mean']=(w.random_aligned_0+w.random_aligned_1)/2
    w['rotated_residual_mean']=(w.rotated_residual_0+w.rotated_residual_1)/2
    return w

def contrasts(w):
    return {'aligned_vs_zero':w.aligned-w.zero,'residual_vs_zero':w.residual-w.zero,
        'full_minus_aligned':w.full-w.aligned,'aligned_vs_random_aligned':w.aligned-w.random_aligned_mean,
        'full_vs_rotated_residual':w.full-w.rotated_residual_mean,
        'factorial_interaction':w.full-w.aligned-w.residual+w.zero,
        'full_vs_zero':w.full-w.zero,'full_minus_residual':w.full-w.residual,
        'reduce_residual_vs_full':w.aligned_half_residual-w.full,
        'reduce_residual_vs_half':w.aligned_half_residual-w.half}

primary=[];family_primary=[]
main=contrasts(paired_values('flip','margin_shift'))
for name in PRIMARY_TESTS:
    values=main[name].groupby('family_id').mean()
    require(len(values)==N_FAMILIES and np.isfinite(values).all(),'All 24 families are required for primary inference.')
    primary.append({'test':name,**estimate(values,name,True)})
    family_primary.extend({'test':name,'family_id':fid,'value':float(v)} for fid,v in values.items())
primary=pd.DataFrame(primary);primary['p_holm']=holm(primary.p_two_sided.to_numpy())
primary['direction']=np.where(primary.estimate>0,'positive',np.where(primary.estimate<0,'negative','zero'))
primary['adjusted_evidence']=np.where(primary.p_holm<.05,primary.direction,'unresolved')

secondary=[];robustness=[]
for context,metric in itertools.product(('flip','stable'),('margin_shift','cf_accuracy_gain','cf_best_other_gain')):
    cs=contrasts(paired_values(context,metric))
    for name,values in cs.items():
        fam=values.groupby('family_id').mean()
        secondary.append({'receiver_context':context,'metric':metric,'contrast':name,**estimate(fam,['secondary',context,metric,name])})
        if metric=='margin_shift':
            loo=[float(fam.drop(fid).mean()) for fid in fam.index]
            robustness.append({'receiver_context':context,'contrast':name,'positive_families':int((fam>0).sum()),'negative_families':int((fam<0).sum()),
                'zero_families':int((fam==0).sum()),'leave_one_out_min':min(loo),'leave_one_out_max':max(loo)})
secondary=pd.DataFrame(secondary);robustness=pd.DataFrame(robustness)

condition_rows=[]
for (context,kind),sub in frame.groupby(['receiver_context','kind']):
    for metric in ('margin_shift','cf_accuracy_gain','cf_best_other_gain'):
        condition_rows.append({'receiver_context':context,'kind':kind,'metric':metric,**estimate(sub.groupby('family_id')[metric].mean(),['condition',context,kind,metric])})
condition_summary=pd.DataFrame(condition_rows)

subset_rows=[];support_rows=[]
for context in ('flip','stable'):
    full=frame[(frame.receiver_context==context)&(frame.kind=='full')].set_index(PAIR_INDEX)
    masks={'all':np.ones(len(full),dtype=bool),'prediction_stable':full.prediction_stable.eq(True).to_numpy(),
        'stable_unique_top':full.stable_unique_top.eq(True).to_numpy(),'source_both_correct':full.source_both_correct.eq(True).to_numpy(),
        'axis_defined':full.axis_defined.eq(True).to_numpy()}
    for name,mask in masks.items():
        ids=full.index[mask]
        support_rows.append({'receiver_context':context,'subset':name,'donor_receiver_pairs':len(ids),'distinct_receiver_inputs':len(set(ids.get_level_values('input_key'))),
            'n_families':len(set(ids.get_level_values('family_id'))),'total_pairs':len(full)})
        for metric in ('margin_shift','cf_accuracy_gain'):
            w=paired_values(context,metric).loc[ids]
            for label in ('full_vs_zero','aligned_vs_zero','residual_vs_zero','aligned_vs_random_aligned','full_vs_rotated_residual','factorial_interaction'):
                vals=contrasts(w)[label].groupby('family_id').mean()
                subset_rows.append({'receiver_context':context,'subset':name,'metric':metric,'contrast':label,**estimate(vals,['subset',context,name,metric,label])})
subsets=pd.DataFrame(subset_rows);supports=pd.DataFrame(support_rows)

boundary_rows=[]
for mid in range(6):
    keys=set(frame.loc[frame.receiver_mapping==mid,'input_key'])
    for name,values in main.items():
        vals=values[values.index.get_level_values('input_key').isin(keys)].groupby('family_id').mean()
        boundary_rows.append({'receiver_mapping':mid,'contrast':name,**estimate(vals,['mapping',mid,name])})
boundaries=pd.DataFrame(boundary_rows)

anchor_keys=['family_id','input_key','context','source_mapping_slot']
new=frame[frame.kind=='full'][anchor_keys+['receiver_context','margin_shift','cf_success','logprob_A','logprob_B','logprob_C']]
old=PARENT_PATCHES[anchor_keys+['margin_shift','cf_success','logprob_A','logprob_B','logprob_C']]
require(not new.duplicated(anchor_keys).any() and not old.duplicated(anchor_keys).any(),'Parent replay duplicate pair.')
parent_replay=new.merge(old,on=anchor_keys,how='outer',suffixes=('_new','_parent'),validate='one_to_one',indicator=True)
require(parent_replay['_merge'].eq('both').all(),'Parent full-update pairs changed.')
parent_replay['margin_shift_difference']=parent_replay.margin_shift_new-parent_replay.margin_shift_parent
parent_replay['centered_score_error']=[centered_score_error([r.logprob_A_new,r.logprob_B_new,r.logprob_C_new],[r.logprob_A_parent,r.logprob_B_parent,r.logprob_C_parent]) for r in parent_replay.itertuples()]
anchor_summary=[]
for context,g in parent_replay.groupby('receiver_context'):
    for metric in ('margin_shift_new','margin_shift_parent','margin_shift_difference'):
        anchor_summary.append({'receiver_context':context,'metric':metric,**estimate(g.groupby('family_id')[metric].mean(),['anchor',context,metric])})
anchor_summary=pd.DataFrame(anchor_summary)
for name,table in [('primary_six_tests',primary),('primary_family_contrasts',pd.DataFrame(family_primary)),('secondary_contrasts',secondary),
    ('condition_summary',condition_summary),('family_robustness',robustness),('subset_support',supports),('subset_effects',subsets),
    ('mapping_boundaries',boundaries),('parent_full_replay',parent_replay.drop(columns='_merge')),('parent_full_replay_summary',anchor_summary)]:
    table.to_csv(OUTPUT_ROOT/'analysis'/(name+'.csv'),index=False)
display(primary.round(5))
print('Same-cohort parent anchor, descriptive only:');display(anchor_summary.round(5))
print('Subset denominators (not independent observations):');display(supports)
```
