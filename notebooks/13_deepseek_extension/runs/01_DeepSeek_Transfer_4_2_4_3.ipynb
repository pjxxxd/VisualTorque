```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
PROJECT_ROOT=Path('/content/drive/MyDrive/VisualTorque')  # 改成你的实际项目目录
RUN_ID='main'
SOURCE_SUPPORT=None  # 通常无需修改；自动查找原干预实验 support.json
assert PROJECT_ROOT.is_dir(), PROJECT_ROOT
```

```python
import base64,io,zipfile,hashlib
PACKAGE_ROOT=Path('/content/visualtorque_deepseek_package')
payload=base64.b64decode('')
with zipfile.ZipFile(io.BytesIO(payload)) as z:
    for member in z.infolist():
        assert (PACKAGE_ROOT/member.filename).resolve().is_relative_to(PACKAGE_ROOT.resolve())
        z.extract(member,PACKAGE_ROOT)
print('Experiment sources ready:',PACKAGE_ROOT)
```

```python
import subprocess,sys
ENV=Path('/content/visualtorque_deepseek_env')
subprocess.check_call([sys.executable,str(PACKAGE_ROOT/'install.py'),'--env',str(ENV)])
PYTHON=ENV/'bin/python'
def run_stage(stage):
    command=[str(PYTHON),'-u',str(PACKAGE_ROOT/'run.py'),'--project',str(PROJECT_ROOT),'--run-id',RUN_ID,'--stage',stage]
    if SOURCE_SUPPORT: command += ['--source-support',str(SOURCE_SUPPORT)]
    subprocess.check_call(command)
run_stage('preflight')
```

运行阶段：`calibration`


```python
run_stage('calibration')
```

运行阶段：`validation`


```python
run_stage('validation')
```

运行阶段：`formats`


```python
run_stage('formats')
```

即使前面的阶段中断，也可以运行下面一格打包已有结果。缺少完成标记时，报告会显示未完成。


```python
run_stage('export')
from google.colab import files
files.download(str(PROJECT_ROOT/'results/deepseek_vl2_small'/(RUN_ID+'_results.zip')))
```
