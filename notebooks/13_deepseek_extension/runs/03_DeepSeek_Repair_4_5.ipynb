```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
PROJECT_ROOT=Path('/content/drive/MyDrive/VisualTorque')  # 改成你的实际项目目录
RUN_ID='blackwell_cu128'  # 三份 notebook 使用相同值；请勿改回旧环境的 main
SOURCE_SUPPORT=None  # 可手动指定已有 support.json
assert PROJECT_ROOT.is_dir(), PROJECT_ROOT
previous_support=PROJECT_ROOT/'results/deepseek_vl2_small/main/prediction/support.json'
if SOURCE_SUPPORT is None and previous_support.is_file():
    SOURCE_SUPPORT=previous_support
    print('Reuse frozen families/images from:',SOURCE_SUPPORT)
print('Results:',PROJECT_ROOT/'results/deepseek_vl2_small'/RUN_ID)
```

```python
import base64,io,zipfile,hashlib
PACKAGE_ROOT=Path('/content/visualtorque_deepseek_package')
payload=base64.b64decode('')
with zipfile.ZipFile(io.BytesIO(payload)) as z:
    for member in z.infolist():
        assert (PACKAGE_ROOT/member.filename).resolve().is_relative_to(PACKAGE_ROOT.resolve())
        z.extract(member,PACKAGE_ROOT)
print('Experiment sources ready:',PACKAGE_ROOT)
```

```python
import sys,shutil
sys.path.insert(0,str(PACKAGE_ROOT))
from notebook_runner import run_logged
ENV=Path('/content/visualtorque_deepseek_py311_cu128')
LOG_DIR=PROJECT_ROOT/'results/deepseek_vl2_small'/RUN_ID/'logs'
print('Colab kernel:',sys.version.split()[0],'; experiments use Python 3.11.11 / CUDA 12.8.')
try:
    run_logged([sys.executable,'-u',str(PACKAGE_ROOT/'install.py'),'--env',str(ENV)],LOG_DIR/'install.log')
finally:
    gpu_report=ENV/'gpu_check.json'
    if gpu_report.is_file():
        LOG_DIR.mkdir(parents=True,exist_ok=True)
        shutil.copy2(gpu_report,LOG_DIR/'gpu_check.json')
PYTHON=ENV/'bin/python'
def run_stage(stage):
    command=[str(PYTHON),'-u',str(PACKAGE_ROOT/'run.py'),'--project',str(PROJECT_ROOT),'--run-id',RUN_ID,'--stage',stage]
    if SOURCE_SUPPORT: command += ['--source-support',str(SOURCE_SUPPORT)]
    run_logged(command,LOG_DIR/(stage+'.log'))
run_stage('preflight')
```

运行阶段：`repair-development`


```python
run_stage('repair-development')
```

运行阶段：`repair-confirmation`


```python
run_stage('repair-confirmation')
```
